# 언어모델
## NLP (자연어 처리)
자연어-> 사람이 이해할 수 있는말. 소통 용 언어. 파이썬과 자바 등의 컴퓨터 언어랑 대비됨. 즉 사람이 쓰는 말을 컴퓨터가 이해하고 생성할 수 있도록 하는 기술 분야    
nlp: 텍스트를 숫자로 바꿔 처리하는 것이 출발점.   
*언어모델-> NLP-> t숫자벡터-> 의미 추출*, 즉 어떤 단어 다음에 어떤 단어가 올 확률이 높은지를 계산하는 모델   

- 이산적, 순차적, 중의적, 희소성, 사변성 등의 특징이 있다

## 통계 기반 모델

### SLM (Statistical Language Model)

분포가설 아이디어에 근거.  -> 조건부 확률, 카운트 기반 확률   
대량의 문장에서 단어가 함께 등장한 횟수를 세고, 그 빈도로 다음 단어의 확률을 계산하는 방식이다      
문장 전체의 확률은 각 단어가 앞 단어들 뒤에 올 조건부 확률을 차례로 곱해서 구함.  -> 조건부 확률, 그에비해 카운트 기반은 이전 단어 시퀀스의 등장 빈도를 구한다    

**분포 가설**   
비슷한 맥락에서 쓰이는 단어는 비슷한 의미를 가진다는 가정   

SLM이 동시 등장 빈도로 확률을 추정하는 것도 이 관점에 기반   


### N-gram
**등장 배경**   
*and adorable little boy is spreading?*
-> n-1개의 단어   
즉 문장 일부만 보고 계산..   

앞의 모든 단어를 조건으로 확률을 계산하면 똑같은 문장이 코퍼스에 거의 없어 확률이 0이 되기 쉬움.   
이를 해결키 위해 바로 앞 N-1개 단어만 보고 다음 단어를 예측하자는 아이디어에서 등장.    

**한계**   
- 희소성 문제: 학습 훈련 데이터에 없는 단어조합은 여전히 확률이 0이 됨 -> 해당 시퀀스 예측불가   
- 문맥제한: N을 줄이면 문맥을 거의 반영하지 못함, 앞선 몇개 단어만 고려   


---

### PPL (Perplexity)
언어모델의 성능을 평가하는 지표.   
모델이 다음 단어를 고를 때 얼마나 혼란해하는지.   
값이 낮을수록 실제 문장을 더 잘 예측한다는 의미.   

*→ PPL이 낮다 = 모델이 예측에 덜 혼란스러움 = 모델이 문맥을 잘 이해함 = 모델의 성능이 좋음*   

평균적으로 선택지가 몇 개쯤 남아 있는 상태인지로 이해가능. 신경망 기반 언어모델은 단어간 관계 학습하여 일반화 능력 뛰어남   

---

### NLM (Neural Language Model)
신경망을 이용해 다음 단어를 예측하는 모델. -> 통계적 한계 극복!     
단어를 밀집 벡터(임베딩)로 표현해 의미가 비슷한 단어는 벡터 공간에서도 가깝게 위치함.   

- **통계 모델:** '축구'와 '농구'는 그냥 완전히 다른 이름일 뿐.
- **신경망 모델:** '축구'와 '농구'를 지도 위에 찍어보니 아주 가까운 곳에 위치. 둘 다 '운동’이고, '구기 종목'. 특징 공유   

**SLM과의 차이**   
- SLM은 빈도를 세서 확률을 구하고, NLM은 학습을 통해 확률 함수를 직접 학습함

## 딥러닝 기반 모델

### TRANSFORMER
rnn 기반 모델, 인코더와 디코더가 반복 실행됨.   
-> 트랜스포머는 이에 나아가 인코더와 디코더를 여러 층으로 깊게 쌓는 방식을 사용   -> 전이학습   

### BERT
트랜스포머의 인코더만 사용한 모델임.   
문장의 앞뒤 문맥 동시 보는 양방향 구조   
image.png

기본적으로 인코더를 여러 개 쌓아올리는 구조   

문장 중간의 단어를 가리고 맞히는 방식(MLM)으로 사전학습Masked Language Model.   
문장의 의미를 이해하는 데 강해 분류, 질의응답 등 이해 중심 작업에 주로 활용됨.   

+Next Sentence Prediction(NSP)으로 논리 흐름까지 학습   

### GPT
트랜스포머의 디코더만 사용한 모델.   
앞 단어들만 보고 다음 단어를 예측하는 단방향 구   
덕분에 자연스러운 문장 생성에 강함.   
   

### LLM (Large Language Model)
수십억에서 수천억 개 이상의 파라미터를 가진 초대형 언어모델.   
방대한 데이터로 학습해 별도의 추가 학습 없이 지시문만으로 다양한 작업을 수행할 수 있음.   
규모가 일정 수준을 넘으면 추론 등 이전에 없던 능력이 나타나기도...  

- 대규모 학습
- 예측및 생성   
| 종류 | 기반 | 특화 |
| --- | --- | --- |
| BERT | encoder | 문장 이해 |
| GPT | decoder | 다음 단어 예측, 문장 생성 |    

**한계**   
- 사실이 아닌 내용을 그럴듯하게 생성하는 환각 현상   
- 학습 이후의 최신 정보를 알지 못함
- 학습 및 운영 비용이 매우 큼

---


### RAG (Retrieval-Augmented Generation)
새 지식에 대해 답변할 수 없는 문제 발생  
그래서
LLM이 답변을 생성하기전  외부 자료를 먼저 검색해 참고하도록 하는 방식.   
-> 재학습 없이  최신 정보나 내부 문서를 반영할 수 있음.   
근거 기반으로 답변하므로 환각을 줄일 수 있음(정보정확성).   


### LangChain
**아이디어**   
LLM+RAG 파이프라인 쉽게 구성하는 프레임 워크로, 구성요소를 쉽게 연결한다.      

**특징**   
- 표준화: 프롬프트, 모델 등 비슷 기능 요소를 똑같은 형식 컴포넌트로 표준화   

- 체이닝: 여러 처리 단계를 순서대로 엮어 하나의 흐름으로 구성   

한계 이전 단계 돌아가는 로직 어려움.
 -> angGraph는 비선형적인 작업 형태로, 사이클 (Cycle), 루프 (Loop)

## sLM (small Language Model)
파라미터 수를 수억에서 수십억 개 수준으로 줄인 경량 언어모델.   
소규모 언어모델   
범용성보다 효율성과 특정 분야 성능에 초점을 둠.   작교 효율적

### 모델 압축
큰 모델을 작고 가볍게 만드는 기술임.   

- 양자화(Quantization): 가중치를 더 적은 비트로 표현해 용량과 연산량을 줄임
- 가지치기(Pruning): 중요도가 낮은 연결이나 파라미터를 제거   
- 저차원분해: 큰 모델(교사)의 출력을 작은 모델(학생)이 따라 학습함
### sLM의 장단점
**장점**   
- 학습 및 운영 비용이 적음
- 응답 속도가 빠름
- 기기 내부(온디바이스)에서도 구동 가능함
->  데이터를 외부로 보내지 않아 보안 및 개인정보 보호에 유리함

**단점**   
- 편향, 일반화 제한됨, 환각   

### sLM과 LLM의 결합
두 모델의 장점을 살려 역할을 분담      
간단하거나 반복적인 작업은 sLM이 빠르고 저렴하게 처리하고, 어렵고 복잡한 작업만 LLM이 처리
LLM이 sLM 학습을 돕는 교사 역할을 하기도 함.   
이를 통해 비용, 속도, 성능 사이의 균형을 맞출 수 있음.
