# NLP (자연어처리)
>인간이 의사소통을 위해 일상생활에서 사용하는 언어를 컴퓨터가 분석하고 이해하며 생성할 수 있도록 하는 인공지능 기술

[자연어 특징]
1. 이산적 데이터
    - 연속적인 숫자 데이터와 달리 직접 구학적으로 처리하기 어려워서 단어와 문장을 숫자 또는 연속적인 벡터로 직접 변환해야 함

2. 순차적 데이터
    - 단어가 등장하는 순서에 따라 문장 의미가 바뀌기에 단어의 순서와 앞뒤 문맥을 함께 고려해 처리할 수 있는 모델구조가 필요함

3. 중의성과 모호성
    - 단어 하나가 여러 의미로 해석될 수 있어 핵심 단어가 아닌 문맥 전체를 파악해 의미를 판단해야 함

4. 희소성
    - 단어의 종류와 조합이 매우 다양하기에 다 학습할 수 없어 학습하지 않은 표현도 단어 간 유사도를 이용해 확률적으로 추론할 수 있는 능력이 필요함

5. 가변성과 진화
    - 시대에 따라 지속적으로 변하는 성질이 있어 언어와 표현을 고정 규칙이 아닌 데이터로부터 유연하게 패턴을 학습하는 머신러닝 방식이 필요함

[자연어처리 역할]
1. 형태소 분석

2. 구문 분석

3. 의미 분석

4. 담화 분석

# 언어모델
> 인간이 말하고 쓰는 방식을 모방하는 시스템

언어 모델은 학습 방식에 따라 (1) **통계적 언어모델** (2) **신경망 언어모델** 구분

## 통계적 언어 모델 (SLM)

### 분포 가설
비슷한 문맥에서 등장하는 단어들은 서로 유사한 의미를 가질 가능성이 높다는 가설으로 이 분포가설로부터 통계 기반 언어 모델이 등장

- 조건부 확률: 앞에 등장한 단어들을 조건으로 다음 단어가 등장할 확률 계산

- 카운트 기반 확률: 특정 문맥 다음에 어떤 단어가 얼마나 자주 등장했는지 계산
    - **한계**: 가능한 단어의 조합이 매우 많아 특정 표현이 학습 데이터에 거의 등장하지 않거나 전혀 등장하지 않는 희소문제가 발생 

### N-gram
> 문장이 길어질수록 발생하는 한계점 (데이터 희소성 / 계산의 복잡성)을 완화하기 위해 N-gram 등장

연속된 N개의 단어로 구성된 묶음으로, 직전 N-1개의 단어를 바탕으로 다음 단어의 등장 확률을 예측하는 통계적 언어모델

N이 커질수록 더 긴 문맥 반영 가능

[한계]
- 희소문제: N값이 커질수록 가능한 단어 조합의 수가 급격히 증가해 훈련 데이터에서 특정 단어 조합을 찾기 어려워짐

- 문맥 제한: 멀리 떨어진 단어 사이의 관계 파악 어려워서 앞 부분의 중요한 정보가 예측 과정에서 제외될 수 있음

### PPL
> 언어 모델이 주어진 문장을 예측할 때 얼마나 혼란스러워하는지 수치로 나타낸 것

- 문장이 훈련데이터에 희소하게 등장할수록 예측 어려움(=PPL이 높아짐)

- 문장에 예측하기 쉬운 단어가 추가되면 PPL 감소


## 신경망 언어모델(NLM)
> 기존 SLM의 한계를 완화하기 위해 **인공신경망**을 활용해 단어의 의미와 문맥 학습함

- 임베딩 기술
    > 이산적 데이터를 연속적인 숫자 벡터로 변환해 단어 간 의미적 관계를 표현하는 기술

    - 이를 통해 의미나 문맥이 유사한 단어들을 벡터 공간에서 서로 가까운 위치에 놓이게 학습하고 의미적 유사성 파악이 가능함

#### NLM과 SLM의 차이
- SLM은 단어 등장 빈도와 확률 계산 / NLM은 인공신경망의 가중치 학습

- SLM은 단어를 개별 기호로 취급 / NLM은 단어를 연속적인 벡터로 표현

- SLM은 유사 단어간 관계 반영 어려움 / NLM은 임베딩을 통해 의미적 유사성 학습

- SLM은 희소문제에 취약 / NLM은 학습된 표현을 통해 희소문제 완화

- SLM은 직전 단어만 고려 / NLM은 다양한 범위의 문맥 학습 가능


# 딥러닝기반 모델
RNN과 LSTM의 한계를 보완하기 위해 Transformer가 등장 

### BERT
> 트랜스포머의 인코더 구조를 기반으로, 문장의 양방향 문맥을 학습하여 언어의 의미를 이해하는 신경망 언어 모델

[특징]
- **양방향**: 단어의 의미파악시 앞에서 등장한 단어와 뒤에 등장한 단어까지 함께 고려

- **트랜스포머 기반**: 인코더 구조만 활용해 문장 내 단어 간 관계 분석

- **사전학습 + 파인튜닝**: 대량의 텍스트 데이터로 언어의 일반적 구조와 문맥 파악하고 이후 추가학습

[학습 방법]
- 1. 입력 문장 토큰화

- 2. 임베딩: 벡터변환

- 3. 트랜스포머를 통한 문맥 학습: self- attention을 활용해 각 토큰이 문장 내 다른 토큰들과 어떤 관계를 가지는지 분석

- 4. 출력 및 활용

[사전학습]
> 특정 NLP 작업 수행 전 대량의 텍스트 데이터를 활용해 언어의 일반적 구조와 문맥 미리 학습

1. **MLM**
 - 입력 문장의 일부 토큰을 가리고 주변 문맥을 활용해 가려진 원래 토큰을 예측하도록 학습
 - 동일 단어여도 문맥에 따라 달라지는 의미 학습

2. **NSP**
 - 두 문장이 주어지면 두번째 문장이 첫번째 문장 바로 다음에 등장하는 문장인지 예측하도록 학습
 - 두 문장 사이의 연결관계와 문맥적 연관성 학습 가능

### GPT
> 트랜스포머의 디코더 기반 구조를 활용해 이전 단어들을 바탕으로 다음 단어를 예측하고 텍스트를 생성

즉, 이전 토큰을 바탕으로 다음 토큰을 순차적으로 예측하는 자기회귀방식으로 텍스트 생성

최근 GPT는 별도의 파인튜닝 없이 프롬포트에 제공된 지시와 예시만으로 다양한 작업 수행하는 능력을 보여줌

### LLM
> 방대한 텍스트 데이터와 대규모 신경망을 활용해 인간의 언어를 이해하고 생성하도록 학습된 인공지능 모델

[작동방식]
1. 학습
- 단어 사이의 관계, 문맥적 의미, 문법구조, 세상의 지식까지 언어의 패턴을 스스로 터득

- 사전학습을 통해 다음 토큰을 예측하는 작업을 반복하고 파인튜닝을 통해 사전에 학습된 모델을 특정 작업이나 목적에 맞는 데이터로 추가학습해 성능을 개선

- 학습 과정에서 예측값과 정답의 차이를 계산하고, 이를 줄이도록 모델의 가중치를 업데이트

2. 예측 
- 문장을 토큰으로 분리해 벡터로 변환하고 transformer의 attention을 활용해 문맥을 분석한 후 이를 바탕으로 다음 토큰을 예측해 결과를 생성  


[한계]
- 환각현상: 사실과 다르거나 존재하지 않는 정보를 그럴듯하게 생성

- 높은 학습 및 업데이트 비용: 대규모 데이터를 활용하기에 막대한 자원과 비용 소모

### RAG의 구조
> 외부 지식 검색과 LLM의 텍스트 생성 능력을 결합해 답변의 정확성과 최신성을 높힘

[구조]
1. 질의 인코더: 사용자의 질문을 숫자 벡터로 변환해 관련성 높은 문서를 검색할 수 있도록 함

2. 지식 검색기: 벡터 DB에서 질문 벡터와 유사도가 높은 문서를 검색해 LLM 전달

3. 지식 증강 생성기: 사용자의 질문과 검색된 외부 문서를 결합해 LLM에 입력

[장점]
1. 최신정보 반영: 모델 재학습 없이도 외부 문서 검색 가능

2. 정보 정확성 향상: 모델 내부 지식보다 더 많은 외부 자료를 근거로 답변 생성 

##### CAG: 모델이 신뢰도를 스스로 판단하고 활용할 수 있도록 훈련시킴
기존 REG는 검색된 문서가 질문과 관련성이 높더라도 오래됐거나 부정확한 정보일 수 있어 

=> CAG를 통해 정보의 관련성뿐만 아니라 출처의 **신뢰성**, **최신성**을 고려한 답변 생성 

질문 입력 -> 정보 검색 -> 신뢰도 평가 -> 정보 선별 -> 최종 답변 생성

### Langchain
> LLM의 다양한 기능을 모듈화하고 연결하여 개발 효율성을 높이는 오픈 소스 프레임 워크 

[특징]
- 추상화: 간단한 인터페이스로 제공

- 표준화: 일괄된 방식으로 사용 가능

- 체이닝: 여러 작업을 순서대로 연결하여 하나의 작업 흐름으로 구성

#### LangGraph
> 여러 작업을 노드와 엣지로 연결해 조건에 따른 분기와 반복이 가능한 AI 워크플로를 구현하는 프레임워크

Langchain의 선형적 구조에만 적용하기 적합하고, 복잡한 시스템에서는 오류가 자주 발생한다는 한계를 보완하기 위해 조건부 분기, 반복 실행, 상태 관리가 가능한 프레임워크 

즉, 비선형적이고 동적인 작업흐름 구현 가능

#### LangSmith
> LLM기반 애플리케이션의 실행 과정을 추적하고, 성능을 평가하며 운영상태를 모니터링할 수 있도록 지원하는 개발 및 관리 플랫폼

[핵심 기능]
- 디버깅 및 추적

- 테스트 및 평가

- 모니터링 

- 프롬포트 허브


# sLM
> 자연어를 처리하고 이해하며 생성할 수 있는 인공지능 모델 

- LLM보다 작고 효율적이어서 메모리와 연산 자원을 비교적 적게 사용 

- 대신 복잡한 추론보다는 특정 분야나 비교적 단순한 작업에 효율적

### 모델 압축 방법
1. 가지치기
    - 신경망에서 중요도가 낮은 연결이나 뉴런, 즉 불필요한 부분을 제거하여 모델을 가볍게 만드는 기술

    - 모델의 크기와 연산량을 줄일 수 있지만, 중요한 부분까지 제거하면 성능이 저하될 수 있음

2. 양자화
    - 숫자의 표현 정밀도를 낮춰 메모리 사용량과 연산 비용을 줄이는 방법

    - 모델 크기와 메모리 사용량을 줄이고 연산 속도를 개선할 수 있지만, 정밀도가 낮아져 일부 성능 저하 가능

3. 저차원 분해
    - 크고 복잡한 가중치 행렬을 더 작은 행렬들의 곱으로 근사하여 파라미터 수와 연산량을 줄이는 방법

    - 더 적은 파라미터로 중요한 정보를 표현해 저장공간이나 연산량을 줄일 수 있음

4. 지식 증류
    - 성능이 뛰어난 큰 모델의 지식을 작은 모델에 전달하여 적은 파라미터로도 좋은 성능을 낼 수 있도록 하는 기술

    - 작은 모델의 성능을 높여 적은 자원으로 효율적인 추론 가능

### 장점
1. 성능적 측면
    - 연산량이 적어 빠른 응답 가능

    - 적은 자원으로 작동할 수 있어 제한된 환경에서도 실행 가능

    - 기업 내부 서버나 개인 기기에 실행할 수 있어 보안 및 데이터 보호 용이

2. 비용적 측면
    - 모델 실행에 필요한 연산량과 메모리가 적어 낮은 비용으로 운영 가능

    - 상대적으로 적은 에너지를 사용해 전력 소비를 줄일 수 있음

    - 하드웨어 비용 절감 가능

### 한계 

1. 편향된 답변 생성 가능

2. 제한된 일반화로 인한 낮은 범용성과 추론 능력

3. 사실과 다른 정보를 생성하는 환각 현상

### LLM과의 결합 - 결합 추론
> sLM과 LLM을 함께 사용해 각 모델의 장점을 살림

- 간단한 작업: 빠르고 효율적인 sML이 처리

- 복잡한 작업: 추론이나 전문적 분석 작업은 LLM이 처리

=>  둘을 적절히 결합해 성능과 비용 사이의 균형을 맞춤


# 분류모델과 생성모델의 차이

- 생성 모델: 주어진 학습 데이터를 학습해 해당 데이터의 분포를 따르는 유사 데이터를 생성하는 모델

- 데이터의 특징과 정답 레이블의 관계를 학습하여 분류 기준을 형성하고 분류하는 특징을 가진 분류 모델과는 달리 생성 모델은 데이터의 특징과 패턴을 학습해 새로운 데이터를 생성한다는 특징이 있음


# 생성 모델의 분류

1. 명시적 확률밀도 모델: 데이터의 확률분포를 명시적으로 정의하여 특정 데이터의 값을 직접적으로 수치화해서 출력할 수 있는 모델 

    - 다루기 쉬운 모델: 데이터의 확률밀도를 **수학적으로 직접 계산** 할 수 있는 모델
        - 대표 모델: NADE, MADE, PixelRNN, nonliner ICA

    - 근사적 밀도 모델: 데이터의 확률 밀도를 **근사적인 방법**을 통해 얻는 모델
        - 대표 모델: VAE, Boltzmann Machine

2. 암시적 확률밀도 모델: 직접 확률밀도를 계산하지 않고, 학습한 분포에서 새로운 데이터를 생성하는 모델
    - 직접 샘플링: **무작위 노이즈**를 입력해서 새로운 데이터 직접 생성
        - 대표모델: GAN

    - 간접/반복적 샘플링: 여러 단계의 **확률적인 상태 변화 규칙을 반복 적용해** 새 데이터를 생성 
        - 대표모델: GSN

### AE
> 입력 데이터의 중요한 특징을 압축하고, 이를 바탕으로 원래 데이터와 최대한 유사하게 복원하도록 학습하는 인굥신경망 모델

- 기존 인코더 디코더와 유사한 구조를 활용하지만 AE는 입력 데이터 자체를 복원하는 것이 목적!

- 작동과정
    1. 압축 단계: 중요 특징 추출 및 정보 압축

    2. 병목 통과: 핵심 정보를 잠재 벡터로 표현

    3. 복원 단계: 잠재 벡터를 활용해 원본 데이터와 유사한 데이터를 복원하고 둘을 비교해 중요 정보의 보존 정도 파악

- 왜 압축했다가 복원? : 압축하기 위해 정하기 위해 어떤 정보가 중요한지, 데이터의 핵심적 특징과 구조를 학습할 수 있음 

### VAE
> 입력 데이터의 특징을 **확률 분포로 학습**하고, 해당 분포에서 샘플링해 새로운 데이터를 생성할 수 있는 생성 모델

- 잠재 공간이 불규칙하게 형성될 수 있는 AE의 한계을 보완하기 위해 등장

- VAE는 데이터의 특징을 가우시안분포 형태로 학습하고 잠재공간이 일정한 분포를 따르도록 학습함

- VAE는 복원오차를 최소화하고, KL발산 최소화를 목표로 학습

=> AE는 입력 데이터의 압축과 복원에 집중 / VAE는 확률분포를 학습해 새 데이터까지 생성하도록 발전

[AE와 VAE 차이]

- AE는 입력 데이터를 하나의 고정된 잠재 벡터로 표햔

- VAE는 입력 데이터를 확률 분포로 표현

[구조와 작동 과정]

1. 입력데이터의 특징을 분석해 확률분포 예측

2. 예측한 확률 분포에서 잠재 벡터를 추출하는 샘플링 및 재파라미터화

3. 샘플링한 잠재 벡터를 이용해 데이터를 복원하거나 새롭게 생성

### GAN의 구조와 장단점
> 생성자와 판별자가 서로 경쟁해 실제 데이터와 유사한 새로운 데이터를 생성하도록 학습하는 생성모델

- 생성자:  실제 데이터와 유사한 가짜데이터 생성
    
- 판별자: 입력된 데이터가 진짜인지 가짜인지 구별

=> 생성자가 실제 데이터의 분포를 학습하고 생성하는 것이 최종 목표

[구조]
1. 무작위 노이즈: 생성자에 입력되는 무작위 숫자 벡터

2. 생성자

3. 판별자 

[장점]

1. 실제 존재하진 않지만 실제처럼 보이는 새로운 데이터 생성 가능

2. 이미지의 세부적 특징을 정교하게 표현해 높은 이미지 선명도를 가짐

3. 무작위 노이즈를 생성자에 한번 통과시켜 빠른 생성 속도

4. 잠재공간의 부드러움

[한계점]

1. 생성자와 판별자 경쟁 과정에서 두 신경망의 학습 균형이 무너져 학습 과정이 불안정

2. 학습 초기에는 생성자의 이미지 품질이 너무 낮아 판별자가 가짜데이터를 쉽게 구별 가능

3. 생성자가 특정 형태의 데이터만 반복적으로 생성

4. 생성 데이터의 품질을 하나의 지표만으로 평가 어려움

5. 실제 이미지에 해당하는 잠재 벡터를 찾기 어렵다는 역매핑의 어려움

### 확산모델
> 이미지에 노이즈를 추가한 데이터를 학습하고, 무작위 노이즈를 점차 제거하며 새로운 이미지 생성

- 실제 데이터의 특징 & 분포를 학습해 새로운 데이터를 자연스럽고 다양하게 생성하는 것이 목적

- 모호성 모델링: 하나의  조건에서 여러 결과가 가능하다는 불확실성을 확률적으로 표현

##### 순확산

- 원본 이미지에 노이즈를 점차 추가해 최종적으로 무작위 노이즈에 가까운 상태로 만듦

- 이후 역확산 과정에서 노이즈를 제거하는 방법을 학습할 수 있도록 함

##### 역확산 
    
- 무작위 노이즈에서 시작해 노이즈를 단계적으로 제거하면서 이미지를 생성하는 과정 

-  순확산으로 만든 잡음 데이터로 신경망이 학습한 노이즈 제거 방법을 활용해 무작위 노이즈를 실제와 유사한 자연스러운 이미지로 바꾸는 것

[장점]

1. 압도적인 샘플 품질 

2. 학습의 안정성

3. 다양성 확보

[단점]

1. 느린 생성속도

2. 높은 계산 비용

3. 구조적 제약
