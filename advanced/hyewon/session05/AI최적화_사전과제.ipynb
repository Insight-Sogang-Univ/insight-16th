# 전이학습
> 처음부터 학습하는 비용을 줄이기 위해 사전학습 모델 재사용

## 딥러닝 모델의 한계
1. 데이터 부족

2. 과적합 가능성이 높음

3. 모델 학습 시 많은 시간과 비용 소모

-> 이미 수백만장 이미지를 학습한 AI의 이미지 특징 파악 능력을 가져오자!

    
## 전이학습
> 모델이 사전 학습을 통해 배운 능력을 활용해 새로운 작업을 학습하는 것으로, 이 전이학습을 통해 모델 전체를 처음부터 학습시킬 필요 없이 새 작업 진행 가능

- Source: 기존에 학습한 문제

- Target: 새롭게 해걀할 문제

- Domain: 사용한 데이터의 특성

- Task: 수행하려는 작업

이 전이학습은 source와 target의 domain이나 task 중 적어도 하나가 다를 때, source에서 배운 지식을 target에 활용하는 방법이라고 말할 수 있음

### 분류
1. Domain과 Task에 따른 분류
    - 귀납적
        - Domain은 같아도 되고 달라도 되지만 task는 서로 다름

        - Target 정답 레이블 필수, Source 레이블은 필수 아님

        - 예시: 이미지 분류 -> 객체 탐지 / 동물 사진 분류 -> 흉부 X=ray 폐렴 분류

    - 변환적
        - Domain은 다르고 task 는 같음

        - source에만 레이블 있고, target에는 없음

        - 예시: 뉴스 감정 분석 -> 영화 리뷰 감정 분석

    - 비지도
        - domain은 같아도 되고 달라도 됨

        - source와 target 둘다 레이블 없음

        - 한 분야에서 익힌 군집화 지식 -> 다른 분야 군집화 


2. 전이 대상에 따른 분류
    - 인스턴스 기반: 기존 데이터 중 일부를 가져옴 

    - 특성 기반: 데이터에 나타나는 특징 표현

    - 파라미터 기반: 기존 모델의 학습된 가중치, 편향

    - 관계 기반: 데이터나 개체 사이의 관계 패턴 

### 전이학습의 패러다임
Pre-training -> Foundation Model -> Fine-tuning -> Task-specific Model

#### **Foundation Model**
> 여러 작업에 활용할 수 있도록, 방대한 데이타로 사전학습한 기반 모델

- 하나의 모델로 시작하여 원한느 작업에 따라 추가학습을 진행해 여러 특화 모델을 만들 수 있고, 시작점이 된 모델

- 예를 들면 언어모델을 고객문의와 모범 답변으로 추가 학습시켜 고객 상담모델로, 회의록과 요약문으로 추가학습시켜 회의록 요약모델로 만들 수 있음

이 foundation model이 여러 특화 모델로 발전시킬 때 새 데아터로 모델을 추가학습하는 것을 **fine-tuning** 한다고 한다.

#### **Fine-tuning** 
> fine-tuning이란 사전 학습된 모델을 새 작업의 데이터로 추가학습하고, 기존 모델의 가중치 일부 혹은 전체를 조정하는 것 

- 모델의 모든 가중치를 반드시 업데이트할 필요는 없음

1. Full Fine-tuning: 모든 가중치를 업데이트
    - 기존 특징 추출층과 새 분류층을 모두 학습하므로 이미지의 특징을 추출하는 방식부터 최종 분류 기준까지 전체적으로 조정 가능

    - 연산비용과 메모리 사용량이 큼

2. Partial Fine-tuning: 재사용할 층은 고정, 필요한 층만 변화 
    - 일부 층을 고정하고 나머지 층의 가중치만 업데이트
    
    - 일반적으로 앞쪽 층을 고정하고 뒤쪽 층을 학습하는 방식

3. Frozen Backbone: 새롭게 판단하는 부분만 학습
    - 새 분류층만 학습

#### **PEFT**
> 사전학습 모델의 가중치 대부분 또는 전부를 고정하고, 새 작업에 필요한 최소한의 파라미터만 학습

- 업데이트 할 파라미터가 줄어서 메모리 부담이 낮아짐

- foundation model은 한번만 저장하고, 작업별로 학습한 작은 파라미터 묶음만 따로 저장하면 됨

1. LoRA: 기존 모델에 더할 수정분을 작은 두 행렬로 학습

2. Adapter: 기존 모델 중간에 작은 신경망을 넣는 방식

3. Prompt tuning: 모델 입력에 학습 가능한 숫자 벡터를 붙이는 방식

-> 가중치 수정 같은 추가 학습 없이 입력만 바꿔서 결과를 이끌어내는 Zero-shot, Few-shot도 있음

#### **전이 학습의 한계와 활용 사례**
1. Negative Transfer: 전이학습 적용 전보다 적용 후에 새 작업에서의 성능이 낮은 경우

2. Catastrophic Forgetting: 새 데이터를 학습하면서 기존 능력이 약해지는 경우

3. Bias Transfer: 사전학습에서 얻은 편향이 새 작업에도 이어지는 경우

4. Transferability 사전학습 모델이 배운 지식이 새 작업에 얼마나 잘 활용될 수 있는지 측정의 어려움

[전이학습 활용 사례]
- CV(의료 영상): 일반 이미지로 사전학습하고 흉부 X-ray로 fine-tuning하여 정상과 폐렴을 분류

- NLP(법률 문서): 대규모 텍스트로 사전학습하고 계약서 조항으로 fine-tuning을 통해 조항 유형 분류

#### **딥러닝 모델 경량화**
> 학습이 끝난 모델을 실제로 사용할 때 드는 메모리와 연산 비용을 줄이는 방법

1. **가지치기**: 모델에서 예측에 미치는 영향이 작다고 판단한 부분을 제거

2. **양자화**: 모델이 쓰는 숫자를 덜 정밀하게 표현해서 저장공간을 줄임

3. **지식 증류**: 큰 모델을 Teacher로 모델로 두고, 작은 student모델이 정답봐 teacher 모델의 출력분포를 함께 학습하도록 하는 빙법. 

4. **모델 구조 경량화**: 모델을 설계할 때부터 같은 작업을 더 적은 계산으로 수행할 수 있는 구조를 선택
- 너무 단순하게 만들면 비슷한 경우를 구분하기 어려워짐


# AI Alignment 
## AI 정렬이 필요한 이유와 정렬 기준
아무리 성능이 좋아도, 사실과 다른 문장을 만들거나 사전 학습 데이터에서 생성된 평향이 드러나는 등의 사용자의 의도와 다르게 흘러갈 수 있음

-> 모델이 사람의 의도와 목표를 이해하도록 만들어야 함

> 사람에게 해당 답변이 바람직한지 정렬하는 기준을 정하고, 그 기준에 맞게 답하게 조정하는 것이 AI Alignment

[정렬기준]
1. Helpful: 사용자의 의도를 파악하고 실질적으로 도움이 되는 답변 제공

2. Honest: 사실에 근거해 답하며 불확실한 정보엔 불확실성을 드러냄

3. Harmless: 답변이나 행동이 사람에게 해를 끼칠 가능성 줄임

## AI 정렬의 두 가지 관점
1. Outer Alignment: 목표를 제대로 설정했는가?

- 사람이 정한 학습 목표나 보상 기준이 실제 원하는 결과를 잘 반영하는지 확인

- **보상 해킹 가능성**: 보상 기준의 허점을 이용해 점수만 높이는 현상

2. Inner Alignment: 모델이 실제로 무엇을 배웠는가?

- 학습 과정에서 의도에 맞는 목표나 전략을 익혔는지 확인

- **기만적 정렬 가능성**: 다른 목표를 추구하면서 훈련, 감독 중에만 사람의 의도에 맞게 행동할 수 있다는 가설적 가능성

## RLHF
> 사람들의 피드백을 학습에 사용하여, 사람들이 선호하는 답변에 높은 점수를 주는 방식

- SFT: 답변을 생성하도록 언어 모델 학습 

- Reward Model: 사람의 선호를 점수로 변환하는 모델 학습

- RL: 학습된 보상 최대화 

-> 부작용: 보상해킹, 아첨 등..

# 프롬프트 엔지니어링
> AI가 더 좋은 답을 내도록 입력 프롬포트를 설계하는 기술

- Fine-tuning은 모델을 바꾸고, 프롬포트 엔지니어링은 모델은 그대로 두고 입력만 바꿈

## Prompt vs Fine-tuning
- 프롬포트: 비용이 낮고 바로 시도하고 수정 가능하지만, 모델이 원래 가진 능력 안에서만 가능

- Fine-tuning: 비용이 높고, 학습시간이 필요하지만, 원하는 결과가 안나오면 모델 자체를 바꿔 특정 도메인에 맞출 수 있음

-> 보통 프롬포트부터 시도하고, 원하는 결과가 안나오면 파인 튜닝을 고려

## 주요 기법

1. Zero-shot / Few-shot
    - Zero-shot: 예시 없이 지시만 줌

    - Few-shot: 입력과 출력의 예시를 몇개 보여준 뒤 같은 형식으로 답하게 하는 방식

2. Chain-of-Thought: 중간 풀이 과정을 단계별로 쓰게 함

3. Role-Prompting: 역할을 부여해서 답의 관점, 말투, 설명수준 조절

4. Instruction Prompting: 원하는 것을 구체적으로 적어서 지시

5. Self-Consistency: 같은 문제를 여러번 풀게하고 , 가장 많이 나온 최종 답을 선택

-> 일반적으로 Instruction prompting 에서 시작

## 프롬프트 엔지니어링의 한계

1. Prompt Injection: 사용자 입력에 지시처럼 보이는 문장을 끼워 넣어 개발자의 지시 대신 따르게 만드는 것

- 모델은 어디까지가 지시이고, 어디부터가 번역할 모델인지 경계를 확실하게 알지 못함

- 완화: 구분자, 권한 최소화, 출력 검증

2. Hallucination: 사실이 아닌 내용을 그럴듯하게 만들어냄

- 사실을 찾아보는 게 아니라 학습한 내용을 바탕으로 다음 말을 이어 붙이기 때문

- 완화: 근거 자료 제공(RAG), 지시, 출처 요구하고 직접 확인

3. 모델 의존성: 같은 프롬프트도 모델이나 버전이 바뀌면 결과가 달라짐

- 완화: 모델 버전 고정, 회귀 테스트

4. 성능 한계: 모델이 원래 못하는 일은 프롬프트로 해결되지 않음 

- 완화: 외부 도구 연결, fine-tuning

-> 외부 도구 연결, 권한 최소화, 출력 검증은 프롬프트만 고쳐서는 안되고 모델을 둘러싼 시스템을 바꿔야 함..

# 하네스 엔지니어링

## 하네스의 정의와 구성 요소
> 하네스란 모델이 실제로 일을 할 수 있도록 필요한 도구와 정보, 규칙을 갖춰 주는 것

[구성 요소]
1. 지시/컨텍스트: 모델이 무엇을 보고 시작하나

2. 도구 인터페이스: 모델이 무엇을, 어떤 형식으로 할 수 있나

3. 실행 환경: 행동이 실제로 어디서 실행되나

4. 상태/메모리: 지금까지 한 일을 어떻게 기억하나

5. 관측/피드백: 행동의 결과를 어떻게 돌려받나

6. 권한/가드레일: 무엇은 하면 안 되나

## 에이전트 실패 유형별 하네스 개선 방법
> 에이전트가 실패했다고 모델의 추론력만 의심하면 안 되고, 인터페이스에 따라서도 성능과 행동이 크게 달라짐

1. 엉뚱한 도구를 부른다 -> 도구 이름, 설명, 입력 형식 수정 (도구 인터페이스)

2. 같은 실수를 반복한다 -> 도구 반환값, 오류 메시지 수정 (관측/피드백)

3. 앞부분을 잊는다 -> 넘기는 정보, 진행 기록, 요약 (지시/컨텍스트, 상태/메모리)

4. 위험한 행동을 바로 실행한다 -> 허용 범위, 사람 승인, 샌드박스 (권한/가드레일, 실행 환경)

5. 결과가 맞는지 모른다 -> 테스트, 검증, 완료 조건 (관측/피드백)

# 루프 엔지니어링
## 루프의 정의
> 루프 엔지니어링이란 모델이 행동을 제안하고, 시스템이 행동을 실행하고, 그 결과를 다시 모델에게 돌려주는 과정을 종료 조건까지 반복하도록 설계하는 것

- 루프 한바퀴: 관측 -> 판단 -> 행동 -> 검증

    - 여기서 검증 결과가 다음 바퀴의 관측이 됨

## ReAct(Reasoning + Acting)
> 생각만 하면 모르는 내용을 지어낼 수 있고, 행동만 하면 같은 행동을 반복하거나 답을 내지 못함 -> 그러면 둘을 번갈아 하게 하자!

- Thought(추론) -> Action(행동) -> Observation(관측)

- 더 좋은 성능을 보이고, 없는 내용을 지어내는 실수가 크게 줄어듦

## 루프의 구성 요소
1. 목표: 무엇을 끝내려 하나

2. 트리거: 루프가 언제 시작되나

3. 관측과 행동: 매 반복에서 무엇을 보고 무엇을 하나

4. 정상 종료 조건: 성공해서 멈추는 조건

5. 안전 상한: 위험해서 멈추는 조건

6. 실패/승인 처리: 막혔을 때, 위험할 때 어떻게 하나

- 멈추는 조건은 두 종류
    - 성공해서: 정상 종료 조건
    
    - 위험해서: 안전 상한, 실패/승인 처리 

- 루프 패턴: 재시도, 주기 실행, 이벤트 트리거, 목표 달성까지 반복 

-> 어떤 패턴이든 종료 조건과 안전 상한을 같이 적어야 함

## 통제 장치와 실행 단계
- 무한 반복, 비용 폭주, 컨텍스트 누적, 같은 부작용 반복, 되돌리기 어려운 행동과 같은 실패가 한번하면 작은 실수지만 루프 안에서 반복되고 자동실행되면 영향이 커짐

[통제 장치]
- 반복, 시간, 비용 상한 -> 무한 반복, 비용 폭주 방지

- 테스트와 완료 조건 -> 끝나지 않았는데 끝났다고 하는 것 방지

- 중복 실행 방지 -> 같은 부작용의 반복 방지

- 체크포인트와 롤백 -> 중간에 잘못된 변경이 쌓이는 것 방지

- 위험도에 따른 행동 분리 -> 자동 실행 / 사후 검토 / 사람 승인 후 실행

[학습 단계의 정렬 vs 실행 단계의 통제]
- 학습 단계의 정렬
    - 모델의 행동을 사람의 선호와 기준을 가깝게 만드는 RLHF처럼 모델 안 파라미터를 바꾸는 방법
    
    - 그래도 reward hacking, 아첨과 같은 부작용 존재

- 실행 단계의 통제
    위험한 행동은 사람 승인 후에만 실행하고, 권한은 최소화해서 주고, 결과는 테스트로 검증

-> 모델이 빈틈을 찾더라도 위험한 행동까지는 이어지지 않도록 하네스와 루프를 잘 설계해야 함
