# 추천 시스템 기초
### 추천 시스템과 데이터 유형
> 추천시스템이란 사용자의 행동 기록과 아이템 정보를 바탕으로, 사용자가 좋아할만한 아이템을 예측해 골라주는 시스템

1. 명시적 데이터 - 고객들에 의해 입력된 데이터

    - 예) 별점, 좋아요, 싫어요, 리뷰

    - 취향을 정확히 반영할 수 있음

    - 그러나 사람마다 점수를 주는 기준이 다르고 취향도 변함 

2. 암시적 데이터 - 고객들의 행동을 기반으로 간접적으로 수집된 데이터

    - 예) 조회수, 클릭, 구매, 시청 시간

    - 대량으로 데이터 수집 가능

    - 행동이 곧 선호라고 단정할 수 없음

### 추천 시스템 분류

세가지로 분류할 수 있음

#### 1. 내용 기반 필터링: 아이템 유사도를 근거로
- 사용자가 이용하거나 선택한 아이템과 특징이 비슷한 아이템 추천

#### 2. 협업 필터링: 다수 사용자의 행동 패턴을 근거로
- 두 사용자의 취향이 빗스하다면, 한 사람이 좋아한 아이템을 다른 사람도 좋아할거라는 가정
    - 사용자 기반: 특정 사용자와 유사한 취향의 사용자가 선호한 아이템 추천 

    - 아이템 기반: 특정 사용자가 선호한 아이템과 유사한 아이템 추천 

#### 3. 하이브리드: 두 시스템의 조합
- 두 알고리즘의 조합으로 단점을 보완


# 내용 기반 필터링
> 아이템을 숫자로 표현 -> 사용자 프로필 생성 -> 유사도 계산 -> 점수가 높은 아이템 추천

### 사용자 프로필과 유사도
- 사용자 프로필: 사용자가 어떤 특성을 좋아하는지 정리한 정보

    1. 직접 지정형: 사용자가 자신의 취향을 직접 알려줌

    2. 간접 지정형: 사용자의 과거 행동 이력에서 취향을 추론

    [프로필 획득 방식]
    1. 사용자가 가지고 있는 아이템 특성 가중치 평균값 활용

    2. 설문, 키워드 입력 등으로 사용자의 기호 파악

    3. 유저의 평가를 바탕으로 분류 문제로 해결

- 유사도: 두 대상이 얼마나 비슷한지 나타내는 숫자
    - 아이템 정보를 숫자벡터로 변환해서 유사도 계산

    - 사용자와 아이템을 같은 기준으로 표현해야 서로 비교 가능

    - 사용자가 선호하는 아이템과 비교해 유사도 높은 아이템을 선별해 추천

### 장단점
- 장점
    1. 아이템 콜드 스타트 문제 해결

    2. 다른 사용자의 데이터 필요 없음

    3. 추천 이유 설명 용이

- 단점
    1. 피처로 담을 수 있는 정보에 한계가 있어 정확도 떨어짐

    2. 비슷한 것만 계속 추천

    3. 사용자 콜드 스타트


# 협업 필터링
> 가정: 과거에 비슷한 선호도를 가진 사용자들은 앞으로도 비슷한 선호도를 가질 확률이 높다

### 메모리 기반 협업 필터링
- 시스템에 저장된 사용자 - 아이템 평점을 그대로 꺼내 새 아이템의 평점을 예측

- 별도의 학습 없이 비슷한 사용자나 아이템을 찾아 써서 '이웃기반 협업 필터링'이라고도 부름

1. 사용자 기반 접근: 나와 취향이 비슷한 사용자의 평점을 활용해 내 평점을 예측하고 추천
    - 나와 유사도가 높은 사용자 이웃 선정 

    - 이웃의 평점을 **유사도로 가중평균**해 내 평점 예측

    - 예측 평점이 높은 아이템 추천

2. 아이템 기반 접근: 내가 평가한 아이템들과 평점 패턴이 비슷한 아이템을 찾아 추천하는 방식
    - 아이템의 속성이 아닌 사용자들이 그 아이템을 함께 좋아하고 싫어한 패턴으로 유사도 판단

    - 아이템끼리의 관계는 상대적으로 덜 변해서 재사용 용이

### 메모리 기반 협업 필터링의 한계
1. 희소 행렬 문제: 주어진 평점, 선호도가 적을수록 성능 저하
    - 데이터가 희소하면 유사도를 믿기 어렵고 예측 자체가 어려움

    - 따라서 모델 기반 협업 필터링 이용으로 해결

2. 콜드스타트 문제: 데이터가 충분히 쌓이지 않아 적절한 추천을 못함
    - 신규사용자, 신규 아이템 상황에서 주로 발생

    - 가입할 때 장르나 작품 직접 지정

    - 새 아이템에는 내용기반 필터림 사용

3. 확장성 문제: 아이템과 사용자가 늘어날수록 계산량이 커짐
    - KNN협업 필터링 사용으로 해결 
    
### 모델 기반 협업 필터링
> 관측 평점 데이터로 예측 모델을 학습하고, 데이터의 패턴을 모델의 파라미터에 담아 관측되지 않은 평점을 예측

- 학습단계: 이미 존재하는 평점을 가장 잘 설명하는 모델 파라미터를 찾음

- 예측단계: 학습된 파라미터로 비어있는 칸의 평점을 계산 

[잠재요인]

평점에는 드러나지 않지만 사용자와 아이템 사이의 관계를 결정짓는 숨은 특성

[Latent Factor Model]

사용자와 아이템을 같은 잠재 공간 위의 벡터로 나타내고, 두 벡터의 관계로 평점을 설명하는 모델 

[행렬분해]

큰 평점 행렬을 작은 행렬 두개의 곱으로 쪼개는 방법

=> 모델기반 협업 필터링은 평점 이면의 잠재된 패턴을 학습함으로써, 직접 겹치는 평가 없이도 전체 데이터의 흐름으로 빈칸을 추정하고 추천 시점의 계산을 가볍게 만듦


# VAE 기반 협업 필터링
> 사용자의 상호작용 기록을 신경망으로 확률분포로 압축한 뒤, 다시 복원하면서 아직 보지 않은 아이템의 선호를 예측하는 모델

### MF의 한계 & VAE 구조
[MF한계]

1. 선형성: 각 요인이 점수에 따로따로 더해지고, 요인끼리 얽히는 효과는 표현하기 어려움

2. 불확실성 무시: 기록이 적은 사용자의 추정이 불안정해도 구분 어려움

=> 이 두 문제를 해결하기 위해 VAE사용

[VAE]

- 가운데 잠재벡터를 점 하나가 아닌 **분포**로 표현하는 오토인코더

- Multi-VAE는 사용자의 상호작용 기록을 입력받아 인코더로 취향을 부포로 요약하고 디코더로 전체 아이템의 선호확률을 복원해 추천

    - 입력: 아이템 수만큼 긴 0/1 벡터로, 평점데이터는 4 이상만 1로 바꿔 사용

    - 인코더: 평균, 표준편차 출력

    - 잠재 벡터: 평균과 표준편차로 정해진 정규분포에서 샘플링한 벡터

    - 디코더: 모든 아이템의 점수를 계산한 뒤 소프트맥스 확률로 변환

### 목적 함수 & KL Divergence
1. 목적함수: 데이터가 나올 로그확률 하한을 키우는 방식 

    디코더가 잠재벡터를 활용해 원래 입력을 최대한 비슷하게 복원하도록 만들고, (일부 사용자에 대한 과적합을 막고 잠재공간이 불규칙해지지 않도록 연속적이고 부드럽게 유지 -> KL Divergence항)

2. KL Divergence: 두 확률분포가 얼마나 다른지를 나타내는 값

    KL이 분포를 N(0,1)쪽으로 당겨 사용자들이 중심 근처에 모이도록 함. 즉 잠재공간을 정규분포 근처로 압축하여 부드럽게 만듦

    -> 기록이 적은 사용자들도 비슷한 취향의 사용자들 근처에 놓이게 되어 안정적인 예측 가능


# GNN 기반 사용자 선호 예측
> 관계 자체를 데이터로 보고 모델링

### GNN 구조와 프로세스
내 이웃 노드의 정보를 나에게 전달/요약!

1. 그래프 구조 생성: 노드와 엣지 정의하고 각 노드에 피처를 부여하기

2. 메시지 패싱: 각 노드가 이웃 노드로부터 정보를 받아 전달.. 레이어의 수만큼 반복하고 임베딩을 갱신하는 대상 노드를 타깃노드라 함 (한 레이어: 메세지 생성-> 집계 -> 갱신)
    - 메시지 생성: 각 노드가 이웃에세 보낼 자신의 정보 생성

    - 집계: 메세지 생성 이후, 여러 이웃 노드에게 받은 메세지를 하나의 벡터로 통합

    - 갱신: 집계된 이웃 정보와 자신의 이전 상태를 결합해 새로운 임베딩 계산


3. 최종 출력 

### LightGCN
> GNN 방식을 따르는 여러 아키텍처 중 하나인 GCN이 사용되는 NGCF에서 몇가지 구성을 삭제해 단순화한 것

레이어마다 가중치 행렬을 곱하고, 비선형 활성화 함수라는 두 연산을 제거하고, 추천의 핵심인 '사용자 - 아이템' 간 연결관게만 전달하도록 구조 단순화 

- 특징 변화 제거: 학습 파라미터 제거

- 비선형 활성화 함수 제거

=> LightGCN은 '정규회된 이웃 임베딩의 합 + 레이어 결합' 만으로 구성된 가볍고 효율적인 모델이다~

### GNN 활용 사례 및 장단점
1. 핀터레스트 
    약 30억개의 노드와 180억개의 엣지로 구성된 그래프 속에서 사용자들이 상호작용하는 핀, 보드, 카테고리를 추천해 분석

2. 소셜네트워크, 금융 보완 등등.......

- 장점
    - 고차원 협업 시그널 포착

    - 정형화되지 않은 데이터 학습 가능: 관계를 하나의 그래프에 담음

    - 콜드 스타트 문제 완화에 효과적: 신규 아이템도 임베딩 얻을 수 있음

- 단점
    - 과평탄화 문제: 레이어가 많아질수록 연결된 그래프 내의 모든 노드 간의 임베딩이 비슷해짐 

    - 과압축 문제: 노드 정보 손실 

    - 계산 복잡도 증가

    - 순서 반영 어려움

#  Transformer 기반 추천 시스템
### 기존 추천 시스템의 한계
1. CF의 순차적 한계
    - 순서 무시: 같은 아이템을 샀다면 구매 순서가 달라도 같은 사용자로 봄

    - 문맥 부족: '방금' 한 구매와 '1년 전' 구매의 중요도 구분못함

2. GNN의 순차적 한계
    - 지역적 정보 전파의 한계: 정보가 희석되고 압축됨 (과압축 문제)

    - 장기 의존성 문제: 레이어 수만큼의 단계까지만 볼 수 있어서 멀리 떨어진 행동 간의 관계 포착 어려움

### Transformer: 사용자 행동을 순서로 접근
다음 행동은 장기적인 평균 취향보다 **최근 행동의 순서와 조합(문맥)**에 더 큰 영향을 받음

1. 문맥에 따라 달라지는 동일 아이템의 의미
   -> 같은 아이템이라도 앞에 등장한 아이템에 따라 서로 다른 목저긍로 이해되고 그에 따라 다음 추천도 달라짐.

2. 시간 순서를 뛰어넘는 핵심 의도 포착 
    -> 다음 행동과 관련한 아이템이 가장 최근이 아니라 시퀀스 앞부분에 있어도 이를 반영 

### Self-Attention
> 시퀀스 안의 모든 아이템 쌍의 관련성을 거리와 상관없이 동시에 계산

[연산 과정] 
1. Attention Score 계산: Query와 각 아이템의 Key를 내적해 관련성 점수 계산

2. 가중치 변환: Softmax로 점수를 합이 1이 되는 비율로 바꿈

3. 가중합: 각 아이템의 Value에 그 가중치를 곱해 더함

=> 관련성에 따라 정보 반영 비율을 다르게 적용하는 것이 핵심!

### Transformer 기반 주요 추천 시스템 모델
두 모델의 차이는 어느 방향의 문맥을 보느냐!

[SASRec]

>단방향 어텐션으로 과거 정보만 활용해, 다음 아이템을 예측하도록 학습한 모델

- 특징: 적응력 집중과 빠른 학습 

- 문제점: 계산량이 많아 긴 행동 이력을 다루기 어렵고, 단방향성임

[BERT4Rec]

>양방향 어텐션으로 과거와 미래 정보를 모두 활용해, 가려진 아이템을 예측하도록 학습한 모델

- 특징: 앞뒤 문맥 모두 학습에 사용하고 가리는 위치를 바꿔 가며 여러 학습샘플을 만들 수 있음 

=> 성능 차이는 어텐션방향보다 학습 설정의 영향이 컸으며, 상황에 따라서 모델 선택하는 것이 좋다


# 오늘날의 추천 시스템
> 추천을 랭킹 문제가 아닌 언어 문제로 푼다 -> 서술형 추천 가능

아이템을 텍스트로 이해하면
- 대화형 추천 가능, zero-shot 추천 가능 (상호작용없이 텍스트 설명 만으로 추천)

### P5
텍스트 프롬포트 -> 텍스트 답변 형식으로 통일해 하나의 언어모델로 학습한 모델

### TalllRec
학습 비용을 줄이기 위해 범용 LLM을 좋아할까 형식의 추천 지시문으로 튜닝하는 프레임 워크 

=> LLM은 단순한 성능 향상이 아니라, 추천 시스템이 사용자와 상호작용하는 방식 자체를 근본적으로 바꾸고 있음

### 인기평향과 XAI 문제
1. 인기평향: 소수의 인기 아이템이 추첨을 독점하고, 대다수의 비인기아이템은 거의 추천되지 않음
    - 데이터 불균형 -> 모델의 편향 강화 -> rich get richer

    - 넷플릭스에서도 인기 영화는 전체의 16.5퍼센트이지만 전체 평점의 86%를 차지

    -> 비인기, 신규 아이템은 노출 기회를 얻기 어렵고, 사용자는 비슷한 콘텐츠만 반복해서 추천받게 됨

2. 설명가능성 
이걸 왜 나한체 추천했는지 고객을 설득해야하는데 , 그 설명이 실제 추천 근거와 일치한다는 보장이 없음
