# 시계열 데이터란?

### 시계열 데이터의 정의

- 시간의 흐름에 따라 일정한 간격으로 순서대로 기록한 데이터. 주가, 기온, 매출처럼 같은 대상을 시간에 따라 반복해서 기록하면 모두 시계열이다.

    $$Y_1, Y_2, \dots, Y_t, \dots, Y_T$$

    $t$는 시점, $Y_t$는 $t$ 시점의 관측값, $T$는 마지막 관측 시점.

1. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**시계열의 조건**</mark> :
    순서 자체가 정보라서 순서를 섞으면 언제 오르고 언제 떨어졌는지가 망가진다. 또 매일, 매월처럼 관측 간격이 일정해야 시점끼리 비교할 수 있다.

2. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**단변량 / 다변량**</mark> :
    - 단변량 : 변수 하나만 관측하고 자기 과거 값으로 미래를 예측한다. 
        - 삼성전자 일별 종가, 서울 일별 기온
    - 다변량 : 여러 변수를 같이 관측하고 다른 변수의 영향까지 본다. 
        - 아메리카노 판매량 ← 기온 + 강수량 + 요일
    - 다변량이 항상 좋은 건 아니다. 내일 판매량을 예측하려면 내일 기온도 알아야 해서 요일이나 공휴일처럼 미리 정해진 변수를 많이 쓴다.

- 지금까지 다룬 데이터는 한 시점에 여러 대상을 보는 횡단면 데이터였다. 학생 100명의 키는 순서를 바꿔도 결과가 같지만 식당의 월별 매출은 3월과 8월을 바꾸면 의미가 사라진다.

    => 시계열 분석의 주 목표는 미래 예측이고 과거 패턴이 미래에도 어느 정도 반복된다는 가정 위에 있다. 그래서 코로나처럼 과거에 없던 일은 예측하기 어렵고 먼 미래일수록 불확실성이 커진다.

### 시계열 데이터의 특성

1. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**시간의존성**</mark> :
    현재 값이 과거 값의 영향을 받고 보통 가까운 과거일수록 영향이 크다. 그래서 train과 test를 랜덤으로 섞어 나누면 5월 데이터로 3월을 맞히는 **데이터 누수**가 생긴다. 시계열은 반드시 과거로 학습하고 미래로 평가한다.

2. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**계절성과 추세**</mark> :
    - 계절성 : 일, 주, 월, 년 같은 일정한 주기로 비슷한 패턴이 반복된다.
    - 추세 : 장기적으로 증가하거나 감소하는 방향성이다.
    - 월별 항공기 승객 수처럼 둘이 같이 나타나는 경우가 많다.

3. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**자기상관성**</mark> :
    같은 변수의 현재와 과거가 얼마나 비슷하게 움직이는지를 나타내고 값은 -1 ~ 1이다. 오늘 판매량 열을 한 칸 내려 어제 판매량과 나란히 놓고 상관계수를 구하면 그게 자기상관이다.
    - 시차 : 몇 시점 전과 비교하는지. -> 어제와 비교하면 시차 1, 일주일 전이면 시차 7.
    - 시차를 바꿔가며 구하면 반복 주기를 찾을 수 있다. 월별 아이스크림 판매량은 12개월 전과 비교할 때 자기상관이 높게 나온다.

    자기상관이 있으면 일반 회귀를 그대로 쓸 수 없다. 회귀는 오차끼리 독립이라고 가정하는데 시계열은 오차도 어제를 따라가서 유의하지 않은 변수의 p-value가 낮게 나올 수 있다. 또 넷플릭스 가입자 수와 서울 아파트 값처럼 둘 다 시간에 따라 오르기만 해도 상관이 높게 나오는 허위 상관이 생기고 예측값으로 다음 값을 예측하다 보면 첫날 오차가 계속 불어난다.

    => 그래서 일반 회귀를 그대로 쓰지 않고 시계열 전용 모델을 쓰거나 차분, 추세 제거 같은 전처리 후 분석한다.

### 시계열 데이터의 분석요인

- 실제 시계열 그래프에는 아래 4가지 움직임이 섞여 있다.

    | 요인 | 의미 | 주기 | 예측 가능성 | 예시 |
    | --- | --- | --- | --- | --- |
    | 추세 | 장기적인 증가, 감소 방향 | 없음 | 높음 | 1인 가구 수 증가 |
    | 계절 | 1년 이내 고정 주기로 반복 | 고정 | 높음 | 여름 에어컨 판매 급증, 주말 대학가 매출 감소 |
    | 순환 | 계절보다 길고 불규칙한 주기 | 수년, 불규칙 | 낮음 | 경기 호황, 불황 |
    | 불규칙 | 위 셋을 빼고 남은 것 | 없음 | 불가능 | 폭우로 그날 매출 급감 |

- 계절과 순환은 주기가 정해져 있는지로 구분한다. 매년 8월에 오른다면 계절, 몇 년에 한 번인데 언제인지 모른다면 순환이다.
- 순환은 긴 데이터가 있어야 보이고 추세와 구분도 어려워서 보통 추세에 묶어 **추세-순환 요인**으로 본다.

# 시계열 데이터 전처리

### 결측치/이상치 처리

- 시계열은 값 하나가 앞뒤 값과 이어져 있어서 잘못된 값 하나가 주변 분석까지 망가뜨린다. 그래서 앞뒤 흐름을 고려해서 처리해야 한다.

1. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**결측치**</mark> :
    행을 지우면 관측 간격이 깨지고 전체 평균으로 채우면 시간 흐름이 무시된다. 그래서 앞뒤 관측값으로 사이 값을 추정하는 **보간**을 가장 많이 쓴다. 2일 110, 4일 130이면 직선으로 이어 3일을 120으로 채우는 식이다.

    - 선형 보간 : 두 점을 직선으로 잇는다. 빠르고 간단하지만 비선형이거나 급하게 변하는 구간에는 약하다.
    - 다항식 보간 : 모든 점을 지나는 다항식을 찾는다. 비선형에 유리하지만 차수가 높으면 과적합된다.
    - 스플라인 보간 : 구간마다 다항식을 써서 부드러운 곡선을 만든다.
    - 조각별 상수 보간 : 직전 값, 직후 값, 가장 가까운 값으로 채운다. 원본 값만 쓰지만 계단 모양이 된다.

    회귀는 오차를 허용하고 점 사이를 지나가지만 보간은 관측값을 정답으로 보고 반드시 지나간다. 잘 모르겠으면 선형 보간부터 쓴다.

2. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**이상치**</mark> :
    같은 값도 시점에 따라 정상일 수도, 이상일 수도 있다. 35도는 8월이면 정상, 1월이면 이상치이다. 그래서 추세와 계절성을 고려해서 판단한다.

    간단한 탐지 방법으로는 IQR과 Z-score가 있다. IQR은 Q1−1.5×IQR 미만이나 Q3+1.5×IQR 초과를, Z-score는 $Z=\frac{x-\mu}{\sigma}$의 절댓값이 3을 넘는 값을 이상치로 본다. 둘 다 추세나 계절성이 강하면 잘 안 맞아서 구간을 옮겨가며 계산한다. 추세와 계절성이 뚜렷하면 STL로 분해한 뒤 잔차가 큰 값을 잡는다.

    모델로 찾을 수도 있는데, CART는 예측 오차가 큰 지점을 이상치로 보고 클러스터링은 어느 군집에도 속하지 않는 데이터를, 그리고 오토인코더는 정상 데이터로만 학습시킨 뒤 복원 오차가 큰 지점을 이상치로 본다. 처음엔 IQR, Z-score로 시작하고 데이터가 크고 복잡하면 모델을 쓴다.

    찾은 다음엔 오류인지 실제 사건인지 먼저 확인한다. 그냥 지우면 간격이 깨지니 **NaN으로 바꾼 뒤 보간**하는 경우가 많다.

### 평활화

- 자잘한 노이즈를 줄여 큰 흐름을 보는 기술로, 주로 EDA 단계에서 쓴다. 대표 방법이 이동평균이다.

1. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**이동평균**</mark> :
    연속된 N개를 묶어 평균을 내고 묶음을 한 칸씩 옮기며 계산한다.

    $$MA_t = \frac{Y_t + Y_{t-1} + \cdots + Y_{t-N+1}}{N}$$

    100, 130, 110, 140, 120에 N=3을 쓰면 113.3, 126.7, 123.3으로 완만해진다.
    - N이 작으면 변화에 빨리 반응하지만 노이즈가 남고 크면 매끄럽지만 늦게 반응한다.
    - N을 계절 주기와 같게 두면 계절성이 상쇄된다. 월별 데이터에 12개월 이동평균을 쓰면 추세만 남는다.

2. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**이동평균의 종류**</mark> :
    - 단순 이동평균 : 모든 값에 같은 가중치를 준다. 간단하지만 최근 변화에 늦게 반응한다.
    - 가중 이동평균 : 최근 값일수록 큰 가중치를 준다. 오늘 3, 어제 2, 그제 1로 주는 식이고 가중치는 직접 정해야 한다.
    - 지수 이동평균 : 과거로 갈수록 가중치가 지수적으로 줄고 묶음 없이 모든 과거를 조금씩 반영한다.

    $$EMA_t = \alpha Y_t + (1-\alpha) EMA_{t-1}$$

    α가 클수록 최근 값을 더 중요하게 본다.

### 요소분해

- 섞여 있는 시계열을 각 요인으로 다시 나누는 과정이다. 순환은 따로 찾기 어려워서 추세에 포함하고 보통 **추세, 계절, 잔차** 세 개로 나눈다.

1. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**가법모형 / 승법모형**</mark> :
    요인들을 더해서 결합하는지, 곱해서 결합하는지에 따라 나뉜다.

    $$\text{가법} : Y_t = T_t + S_t + R_t \qquad \text{승법} : Y_t = T_t \times S_t \times R_t$$

    | 구분 | 가법모형 | 승법모형 |
    | --- | --- | --- |
    | 계절 변동 크기 | 추세와 상관없이 일정 | 추세에 비례해서 커짐 |
    | 계절 요인 단위 | 원래 단위 | 비율 |
    | 그래프 | 물결 높이가 일정 | 물결 높이가 점점 커짐 |
    | 예시 | 여름마다 항상 +50잔 | 여름마다 항상 ×1.2 |

    매출, 주가, 방문자 수 같은 현실 데이터는 승법에 가까운 경우가 많다. 승법모형에 로그를 취하면 $\log Y_t = \log T_t + \log S_t + \log R_t$가 돼서 가법처럼 다룰 수 있다.

2. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**seasonal_decompose / STL**</mark> :
    - seasonal_decompose : 이동평균으로 추세를 먼저 뽑는 전통적인 방법이다.
        1. 중심 이동평균으로 추세 추정
        2. 원래 값에서 추세 제거
        3. 같은 계절 위치끼리 평균 내서 계절 추정
        4. 남은 값이 잔차
    - STL : 이동평균 대신 **LOESS**를 쓴다.
        - Inner Loop : 추세 제거 → 계절 추정 → 계절 제거 → 추세 추정을 반복하며 T와 S를 다듬는다.
        - Outer Loop : 잔차가 큰 시점에 작은 가중치를 줘서 이상치 영향을 줄인다. → 강건한 분해

    => seasonal_decompose는 단순하고, STL은 시간에 따라 바뀌는 패턴과 이상치에 더 유연하다. 분해해 두면 잔차로 이상치를 찾거나 계절조정 데이터를 만들 수 있다.

# 정상성 판단

### 정상성의 정의 - 강한 정상성과 약한 정상성

- 정상성은 평균, 분산, 자기상관 같은 통계적 특성이 시간에 따라 변하지 않는 성질이다. 어느 시점에서 잘라 봐도 비슷하게 생겼으면 정상이다. 규칙이 안 바뀌어야 과거에서 배운 게 미래에도 통하기 때문에 예측 모델의 전제가 된다.

- 정상 시계열은 평균과 분산이 일정하고 벗어나도 다시 평균으로 돌아온다. 비정상 시계열은 추세 때문에 평균이 변하거나 폭이 커지고 한 방향으로 흐르거나 주기적으로 반복된다.

1. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**강한 정상성**</mark> :
    시간을 h만큼 옮겨도 확률분포 자체가 완전히 같아야 한다. 평균, 분산뿐 아니라 왜도, 첨도까지 일정해야 해서 현실 데이터에서는 거의 없다.

    $$(Y_{t_1}, \dots, Y_{t_k}) \overset{d}{=} (Y_{t_1+h}, \dots, Y_{t_k+h})$$

2. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**약한 정상성**</mark> :
    세 개의 조건만 만족하면 된다.
    - 평균이 일정하다.
    - 분산이 일정하다.
    - 자기공분산이 시점이 아니라 시차에만 의존한다. 1월 1일과 1월 2일의 관계가 7월 1일과 7월 2일의 관계와 같다.

### 정상성의 판단 - 시각 / 통계

1. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**시각적 판단**</mark> :
    - 정상 : 수평선 주위를 비슷한 폭으로 오르내리고 벗어나도 평균으로 돌아온다.
    - 비정상 : 계속 오르거나 내림, 같은 모양 반복, 폭이 점점 커지는 나팔 모양이 보인다.
    - ACF 그래프로도 보며 정상이면 막대가 몇 칸 만에 0 근처로 떨어지고 비정상이면 아주 천천히 줄어든다.

2. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**단위근**</mark> :
    그래프는 보는 사람마다 판단이 다를 수 있어서 통계적으로도 확인한다. 이때의 기준은 단위근이다.

    $$Y_t = \phi Y_{t-1} + \varepsilon_t$$

    - φ = 0.5면 충격 10이 5, 2.5로 줄어 평균으로 돌아온다. → 정상
    - φ = 1이면 충격이 그대로 남고 새 충격이 계속 쌓인다. 평균으로 안 돌아오고 분산도 커진다. → 단위근, 비정상
    - 식을 풀면 해가 z = 1/φ라서 φ = 1이면 해가 1이 되고 그래서 단위근이라고 부른다.

3. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**ADF / KPSS 검정**</mark> :
    두 검정은 귀무가설이 서로 반대라서 같이 쓴다. ADF는 단위근이 있다(비정상)에서 출발해서 p-value < 0.05면 정상으로 보고, KPSS는 정상이다에서 출발해서 p-value < 0.05면 비정상으로 본다.

    | ADF | KPSS | 진단 | 다음 단계 |
    | --- | --- | --- | --- |
    | 정상 | 정상 | 정상 시계열 | 바로 모델링 |
    | 비정상 | 비정상 | 비정상 시계열 | 차분 등 후 재검정 |
    | 비정상 | 정상 | 추세 정상 | 추세 제거 후 재검정 |
    | 정상 | 비정상 | 차분 정상 | 차분 후 재검정 |

    추세 정상은 매년 일정하게 늘어나는 매출처럼 추세선 주위로는 안정적인 경우이고 차분 정상은 주가처럼 충격이 쌓이며 랜덤하게 움직이는 경우이다.

### 정상성 확보 방법

- 정상성이 깨지는 원인은 추세, 계절성, 분산 변화 세 가지이고 원인별로 처방이 다르다. 무조건 차분하는 게 아니다.

1. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**차분**</mark> :
    값 대신 "얼마나 변했나"를 본다.

    $$\Delta Y_t = Y_t - Y_{t-1}$$

    100, 112, 119, 133, 140을 차분하면 +12, +7, +14, +7로 10 근처에서 오르내린다. 단위근 시계열을 차분하면 "오늘 새로 생긴 충격"만 남아서 정상이 된다.
    - 1차 차분으로도 추세가 남으면 2차 차분을 한다. 너무 많이 하면 없던 패턴이 생겨서 보통 2차까지만 하고, 할 때마다 검정해서 최소 횟수만 한다.
    - 차분한 데이터로 예측하면 결과도 변화량이라서 역차분으로 원래 단위로 되돌린다.

2. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**계절 차분**</mark> :
    한 주기 전 값을 뺀다. 1차 차분이 "어제와 비교"라면 계절 차분은 "작년 같은 때와 비교"이다.

    $$\Delta_m Y_t = Y_t - Y_{t-m}$$

3. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**로그 변환**</mark> :
    큰 값은 많이, 작은 값은 조금 줄여서 값이 커질수록 같이 커지던 변동 폭을 맞춘다. 10, 100, 1000이 1, 2, 3이 된다. 0이나 음수에는 못 써서 1을 더한 뒤 로그를 취하기도 하며 이 외에는 Box-Cox 변환도 쓴다.

    => 정상성 판단 → 원인에 맞는 방법으로 정상성 확보 → 재검정 → 모델링 순서로 진행한다.

# 단변량 시계열 분석

- 변수 하나의 기록만으로 자기 과거를 보고 예측한다. 돌아갈 평소가 있는 안정적 시계열이면 AR, MA, ARMA를 바로 쓰고 추세가 있으면 차분 후 ARIMA을 쓰고 계절성이 있으면 계절 차분 후 SARIMA를 쓴다.

- 값은 그날 관측된 최종 결과이고 충격은 그날 새로 생긴 예측할 수 없는 변동이다. 충격은 데이터에 안 보여서 모형으로 추정하고 예측할 때 오늘 충격은 0으로 둔다. 주말 효과나 미리 공지한 할인처럼 미리 알 수 있는 건 충격이 아니다.

### 안정적 시계열 분석 - AR / MA

1. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**AR**</mark> :
    자기상관을 그대로 활용하는 모형이고 오늘 값을 자기 과거 **값**으로 설명한다.

    $$y_t = c + \phi_1 y_{t-1} + \phi_2 y_{t-2} + \cdots + \phi_p y_{t-p} + \varepsilon_t$$

    AR은 오늘 판매량 = 어제 판매량의 일부 + 오늘 충격이다. 계수 0.8인 가게에 방송 충격 100이 들어오면 100 → 80 → 64 → 51로 줄어든다. 어제 값 안에 그저께 값이 들어 있어서 영향이 끝없이 이어지며 옅어진다.

2. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**MA**</mark> :
    오늘 값을 과거 **충격**으로 설명한다.

    $$y_t = \mu + \varepsilon_t + \theta_1 \varepsilon_{t-1} + \cdots + \theta_q \varepsilon_{t-q}$$

    MA은 오늘 판매량 = 오늘 충격 + 어제 충격의 일부이다. 같은 충격 100, 계수 0.8이면 100 → 80 → 0 → 0이다. 영향이 q일 뒤에 딱 끊긴다.

    => AR은 지나간 충격이 어제 값에 담겨 넘어와서 끝없이 옅어지고 MA는 충격 자체가 직접 넘어와서 정해진 날수 뒤에 끊긴다.

### AR, MA의 전제조건

- AR과 MA를 쓰려면 정상성과 가역성이 필요하다. AR은 가역성을 항상 만족해서 정상성만, MA는 정상성을 항상 만족해서 가역성만 계수로 확인하면 된다.

1. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**정상성**</mark> :
    충격을 받아도 시간이 지나면 평소로 돌아오는 성질이다. AR은 계수 절댓값이 1보다 작아야 한다. AR(1) 계수가 0.8이면 100 → 80 → 64로 줄지만, 1이면 그대로 남고 1.1이면 110 → 121로 불어난다. AR(p)는 특성방정식의 모든 근의 절댓값이 1보다 커야 한다. MA는 영향이 q일 뒤 끝나서 항상 정상이다.

    $$1-\phi_1 z-\phi_2 z^2-\cdots-\phi_p z^p=0$$

    ADF, KPSS는 **데이터**가 정상인지 보는 검정이고 모델의 정상성은 추정한 **계수**가 조건을 지키는지 특성방정식으로 본다. 보는 대상이 달라서 서로 대신할 수 없다.

2. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**가역성**</mark> :
    관측값으로 지난 충격을 거꾸로 계산해 낼 수 있는 성질이다. MA(1)을 뒤집으면 오늘 충격 = 오늘 값 − 계수 × 어제 충격이라 어제 충격이 또 필요하고 첫날까지 거슬러 올라가 첫날 이전 충격은 0으로 어림한다. 이 어림이 4만큼 틀렸을 때 계수 0.5면 2, 1, 0.5로 사라지지만 계수 2면 8, 16, 32로 커진다. 그래서 계수 절댓값이 1보다 작아야 한다. MA(q)는 밑에 특성방정식의 모든 근의 절댓값이 1보다 커야 한다. AR은 관측값만으로 충격을 구할 수 있어서 항상 가역이다.

    $$1+\theta_1 z+\theta_2 z^2+\cdots+\theta_q z^q=0$$

### ARMA

- AR과 MA를 합친 모형으로 오늘 값이 과거 값과 과거 충격 양쪽의 영향을 받는다. 충격은 하루에 하나이고 그 여파가 두 길로 넘어온다.

$$y_t = c+\phi_1y_{t-1}+\cdots+\phi_py_{t-p}+\varepsilon_t+\theta_1\varepsilon_{t-1}+\cdots+\theta_q\varepsilon_{t-q}$$

AR 계수 0.5, MA 계수 0.4에 충격 100이 들어오면 AR만 있을 땐 100 → 50 → 25, MA만 있을 땐 100 → 40 → 0이지만 ARMA는 100 → 90 → 45 → 22.5가 된다. 다음 날 90은 입소문 50 + 하루 늦게 온 40이고 이틀 뒤가 45인 건 방송 보고 온 40명도 입소문을 내기 때문이다.

- AR만으로 설명하면 시차를 여러 개 넣어야 하는데 ARMA는 더 적은 계수로 설명할 수 있다.
- AR 부분은 정상성, MA 부분은 가역성 조건을 각각 만족해야 한다.
- ARMA(p,0) = AR(p), ARMA(0,q) = MA(q), ARMA(0,0)은 백색잡음이다.

### 차수 결정 방법 : 자기 상관 함수 (ACF, PACF)

1. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**ACF**</mark> :
    오늘 값과 k시점 전 값의 상관계수이다. 중간 날짜를 거쳐 온 영향까지 합쳐서 잰다. **MA의 차수 q**를 정할 때 본다.

    $$\rho(k)=\frac{Cov(y_t,\ y_{t-k})}{Var(y_t)}$$

    계수 0.8인 AR(1)이면 ACF는 0.8, 0.64, 0.51로 천천히 줄어든다.

2. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**PACF**</mark> :
    중간 시점의 영향을 빼고 오늘과 k시점 전이 직접 얼마나 닮았는지 잰다. **AR의 차수 p**를 정할 때 보며 계수 0.8인 AR(1)이면 시차 1에서 0.8, 시차 2부터 0이다.

    | ACF | PACF | 모형 |
    | --- | --- | --- |
    | 천천히 감소 or 진동 감소 | p+1부터 0 근처 | AR(p) |
    | q+1부터 0 근처 | 천천히 감소 or 진동 감소 | MA(q) |
    | q+1부터 0 근처 | p+1부터 0 근처 | ARMA(p, q) |

- 점선(보통 $\pm 1.96/\sqrt{n}$) 안쪽이면 0으로 본다. 막대가 급격히 줄어들기 직전의 시차를 고르면 된다.
- 그래프로 명확하지 않으면 범위만 잡고 AIC, BIC 기준으로 그리드 서치한다. 값이 낮을수록 좋고 보통 p, q는 0~3 또는 그래프로 잡은 값의 ±1 안에서 찾는다.

### 불안정 시계열 분석 - ARIMA / SARIMA

- AR, MA는 "결국 평소로 돌아온다"는 식으로만 예측한다. 아이 키 기록을 그대로 넣으면 지난 평균을 평소로 보고 키가 줄어든다고 예측한다. 하지만 1년간 자란 길이(6, 5, 6, 7cm)에는 평소가 있다. 그래서 모형을 고치는 게 아니라 **데이터를 모형이 다룰 수 있게 바꾼다.**

1. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**ARIMA (p, d, q)**</mark> :
    ARMA에 차분을 더한 모형이다. p는 AR 차수, d는 차분 횟수, q는 MA 차수이다.
    1. 차분해서 증감으로 바꾼다. $w_t = y_t - y_{t-1}$
    2. 증감을 ARMA로 예측한다. $w_t = c+\phi_1 w_{t-1}+\varepsilon_t+\theta_1\varepsilon_{t-1}$
    3. 예측한 증감을 마지막 값에 더한다. $y_t = y_{t-1}+w_t$

    판매량이 100, 103, 104, 108, 110이고 증감 예측이 +2.5면 6일 예측은 110 + 2.5 = 112.5이다.

    ARIMA(0,0,0)은 백색잡음, ARIMA(p,0,0)은 AR(p), ARIMA(0,0,q)는 MA(q)와 같다. 상수 없는 ARIMA(0,1,0)은 랜덤워크 $Y_t = Y_{t-1} + \varepsilon_t$로, 내일의 최선 예측이 오늘 값이다(주가, 환율). 상수가 있으면 표류를 포함하는 랜덤워크 $Y_t = Y_{t-1} + c + \varepsilon_t$가 되는데, 표류 c = 2.5면 110 → 112.5 → 115 → 117.5처럼 한 방향으로 흘러간다.

2. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**SARIMA (p, d, q)(P, D, Q)m**</mark> :
    ARIMA에 계절 차분과 계절 AR, 계절 MA를 더한 모형이다. 지난달 대비 증감이 추세를 없애고, 작년 같은 달 대비 증감이 계절마다 다른 수준을 없앤다.

    $$w_t = c+\phi_1 w_{t-1}+\Phi_1 w_{t-12}+\varepsilon_t+\theta_1\varepsilon_{t-1}+\Theta_1\varepsilon_{t-12}$$

    대문자 계수가 붙은 항이 계절 AR, 계절 MA이다. 붕어빵 가게의 10월, 11월이 작년보다 50씩 더 팔렸으면 12월도 작년 900 + 50 = 950으로 예측한다.

    p, d, q는 ARIMA와 같고, P, D, Q는 같은 걸 한 주기 전 기준으로 한 번 더 하는 것이다. 새로 볼 건 m 하나로, 한 주기가 몇 시점인지(월별 12, 분기별 4, 요일 7)를 뜻한다.

    계절성이 있으면 ACF가 시차 12, 24, 36에서 다시 솟고 물결 높이가 줄지 않는다. 이 상태에선 p, q를 못 읽으니 계절 차분부터 한다.

    => d, D는 ADF로 정상이 나올 때까지 차분한 횟수로 정하고(대부분 0 또는 1), p, q, P, Q는 ACF, PACF로 범위를 잡은 뒤 AIC, BIC로 고른다.

# 다변량 시계열 분석

### VAR

- 변수 여러 개를 같이 놓고 각 변수를 자기 과거와 다른 변수의 과거로 예측한다. VAR은 AR을 변수 여러 개로 넓힌 것이다.

1. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**VAR 모형**</mark> :
    변수마다 식이 하나씩 있고 각 식에 모든 변수의 어제 값이 들어간다. X는 판매량, Y는 광고비.

    $$X_t = a_{11} X_{t-1} + a_{12} Y_{t-1} + \varepsilon_{x,t}$$
    $$Y_t = a_{21} X_{t-1} + a_{22} Y_{t-1} + \varepsilon_{y,t}$$

    $a_{11}, a_{22}$는 자기 자신의 영향이고, $a_{12}, a_{21}$이 서로 주는 영향으로 VAR에서 새로 생긴 계수이다.

    예를 들어 오늘 판매량 = 어제 판매량의 60% + 어제 광고비의 30%, 오늘 광고비 = 어제 광고비의 50% − 어제 판매량의 20%라면 0.3은 광고가 판매를 끌어올린다는 뜻이고, −0.2는 잘 팔리면 광고를 줄인다는 뜻이다.

2. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**적합한 경우와 부적합한 경우**</mark> :
    - 적합 : 판매량과 광고비처럼 서로 영향을 주고받을 때, 한 변수가 변하면 다른 변수가 어떻게 반응하는지 알고 싶을 때, 변수가 적고 데이터가 충분할 때
    - 부적합 : 기온 → 판매량처럼 한 방향뿐일 때, 변수는 많은데 데이터가 적을 때
    - 변수는 **Granger 인과관계 검정**으로 고른다. X의 과거가 Y 예측에 도움이 되는가를 보고 p-value < 0.05면 포함한다. 진짜 인과가 아니라 예측에 도움이 되는지만 의미한다.

3. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**충격 반응 함수 (IRF)**</mark> :
    한 변수에 충격을 한 번 줬을 때 다른 변수가 시간에 따라 어떻게 반응하는지 본다. 위 계수에서 1일차 광고비 충격 100이 들어오면 판매량은 0 → 30 → 33 → 25.5 → 16.2로, 이틀 뒤에 효과가 가장 크고 며칠에 걸쳐 돌아온다. 계수표만으로는 알 수 없는 부분이다.

4. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**예측오차 분산 분해**</mark> :
    예측이 빗나간 정도를 어느 쪽 충격 탓인지 비율로 나눈다. 1일 뒤 판매량 예측 오차는 100% 판매 쪽 충격 탓이지만 2일 뒤부터는 아직 모르는 광고비가 식에 들어가서 광고 쪽 몫이 6%, 12%, 17%로 늘어난다.

    => IRF는 어떻게 반응하는가, 분산 분해는 누구의 영향이 얼마나 큰가를 보여준다.

# 딥러닝을 활용한 시계열 분석

### RNN 계열, Transformer 계열, Foundation Model

- 전통 통계 모델은 데이터가 많으면 느리고 비선형 패턴을 잡기 어렵다. 그래서 딥러닝을 쓴다.

1. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**RNN 계열**</mark> :
    RNN은 간단하지만 장기 의존성을 못 배우고 LSTM은 게이트로 장기 기억을 유지하고 GRU는 LSTM을 단순화해서 더 빠르다. 아주 긴 시계열이나 병렬 처리에는 약하다.

2. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**Transformer 계열**</mark> :
    Informer는 연산을 줄여 긴 시계열에 강하고, Autoformer는 요소분해를 구조에 넣어 주기가 뚜렷한 판매량 예측에 좋다. TFT는 날씨, 휴일 같은 외부 변수를 같이 쓰고 어텐션으로 어떤 변수가 중요했는지 해석할 수 있어서 유통 수요 예측에 많이 쓴다.

3. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**Foundation Model**</mark> :
    TimeGPT, TimesFM, Chronos처럼 대규모 데이터로 미리 학습해 두고 여러 문제에 범용으로 쓰는 모델이다. LLM을 시계열에 프롬프트나 파인튜닝으로 활용하기도 한다.

    => 데이터가 적으면 ARIMA 같은 통계 모델이 더 나을 수 있고, 해석이 중요하면 TFT를 고려한다.
