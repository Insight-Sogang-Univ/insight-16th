# 딥러닝

### 픽셀과 채널

- 픽셀 : 이미지 데이터를 구성하는 가장 기본 단위로, 특정 위치의 색상 또는 밝기 정보를 가진다.

- 채널 : 색상 정보를 구성하는 각각의 데이터 층이다.

- 색공간 : 색상을 숫자로 표현하는 체계로 그레이 스케일, RGB, HSV 등이 있다.

1. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**색공간별 표현**</mark> :

    | 색공간 | 표현 방식 | 채널 수 | 저장 형태 (28 × 28 기준) |
    | --- | --- | --- | --- |
    | 그레이 스케일 | 밝기만 표현 (0 = 검정, 255 = 흰색) | 1 | (28 × 28) 또는 (28 × 28 × 1) |
    | RGB | 빨강, 초록, 파랑의 조합 | 3 | (28 × 28 × 3) |
    | HSV | 색상, 채도, 명도 기준 | 3 | (28 × 28 × 3) |

    ***흑백은 2차원, 컬러는 3차원인 이유**
    - 흑백은 픽셀마다 밝기 값 하나뿐이라 채널이 1개이고, 컬러는 흑백과 같은 구조의 판이 R, G, B 3장 겹쳐진 것이다.

    - RGB는 빨강, 초록, 파랑 세가지 색상을 조합하여 합쳐 최종적인 하나의 색상을 만드는 방식이라 (255, 255, 255)가 흰색, (0, 0, 0)이 검정이다.

    ***텐서** : 다차원 배열을 통칭하는 용어로 0차원 스칼라, 1차원 벡터, 2차원 행렬, 3차원 이상 텐서로 나뉜다.

# 합성곱 신경망 (CNN)

### CNN 개념과 등장 배경

- 컴퓨터에게 이미지는 수많은 픽셀 값으로 이루어진 숫자 데이터이다.

- 이를 1차원으로 펼쳐서 학습하면 가까운 픽셀끼리의 관계를 반영하기 어려워, 주변 픽셀이 모여 만드는 특징을 학습하기 어렵다.

- 그래서 공간 구조를 유지하면서 특징을 학습하는 CNN이 필요하다.

| 구분 | 구조 |
| --- | --- |
| MLP | FC Layer + ... + FC Layer -> 처음부터 1차원으로 펼친다. |
| CNN | (Conv + Pool) + ... + Flatten + FC Layer 특징을 추출하여 차원을 계속 줄여간 후 1차원으로 펼친다. |

- 용어정리
    - **Convolution Layer**: 필터를 이미지 위에서 이동시키며 모서리, 질감, 형태 등 다양한 특징을 추출
    <br>

    - **Pooling Layer**: 추출된 특징의 크기를 줄이고 중요한 특징을 위주로 맵을 유지
    <br>

    - **Flatten**: 여러 차원으로 구성된 맵을 FC Layer에 입력할 수 있도록 1차원 배열로 변환
    <br>

    - **Fully connected Layer** (분류): 최종적으로 추출된 특징을 합쳐 이미지가 어떤 클래스에 속하는지 판단

=> CNN은 **특징 추출(Conv + Pool)** 후 **분류(FC Layer)** 를 하는 구조이다.

### MLP의 한계와 CNN
일반적인 정형 데이터는 독립적인 값으로 나열 된 열로 구성 
반면 이미지는 픽셀들이 가로와 세로 방향으로 연결되어 있고 주변 픽셀 간의 관계와 위치 정보가 중요!!!

| 한계 | MLP의 문제 | CNN의 해결 |
| --- | --- | --- |
| 위치 변화에 취약 | 픽셀 위치마다 가중치를 학습해서 숫자가 조금만 이동해도 다른 입력이 됨 | 같은 필터를 모든 위치에 적용 + Pooling → 위치가 달라도 특징 탐지 |
| 파라미터 폭발 | 완전연결 구조라 가중치가 픽셀 수에 비례해 급증 → 과적합, 느린 학습 | 부분 연결 + 가중치 공유 → 이미지가 커져도 파라미터 수 유지 |
| 공간 정보 손실 | Flatten 하면서 2차원 구조가 사라짐 | H × W × C 구조를 유지한 채 연산 |


### 합성곱(Convolution)과 풀링(Pooling)

1. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**Convolution Layer**</mark> :
    필터로 이미지를 훑으면서 선, 모서리, 질감 같은 특징을 추출하는 계층이다.

    ***연산 과정** : 필터를 슬라이딩 → 같은 위치끼리 곱한 뒤 모두 더함 → 결과를 Feature Map에 저장

    ***필터 vs 커널**
    - 커널 : 2차원 가중치 행렬 하나 (연산의 기본 단위)
    - 필터 : 커널의 묶음. RGB 입력이면 채널별 커널 3개가 필터 하나이다.

    ***Feature Map**
    - Convolution의 결과로, 필터가 특징에 강하게 반응할수록 값이 커진다.
    - Feature Map 개수 = 필터 개수

    ***RGB 컨볼루션**
    - R, G, B 채널별로 각각 곱한 뒤 전부 더하고 편향을 더해 값 하나를 만든다.
    - 따라서 필터 1개 → 출력 채널 1개, 필터 K개 → 출력 채널 K개

2. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**Padding**</mark> :
    이미지 가장자리에 0을 채워 넣는 것으로, 출력 크기를 유지하고 가장자리 특징도 학습하게 한다.
    - 4 × 4 이미지 + 3 × 3 필터 → Padding 없으면 3 × 3, Padding = 1이면 4 × 4

3. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**Stride**</mark> :
    필터가 한 번에 이동하는 거리이다.
    - Stride 1 : 세밀한 특징 추출, Feature Map 큼
    - Stride 2 이상 : 연산량 감소, Feature Map 작음

4. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**Pooling**</mark> :
    일정 영역(보통 2 × 2)의 값을 하나로 요약해서 크기를 줄이는 방법이다. (Max Pooling = 최댓값, Average Pooling = 평균값)

    ***장단점**
    - 장점 : 계산량 감소, 작은 위치 변화에 강건, 과적합 완화
    - 단점 : 세부 정보와 정확한 위치 정보 손실

=> Conv, Pool은 특징 추출 담당 / Padding, Stride는 출력 크기를 조절하는 하이퍼파라미터


### 이미지 딥러닝 응용

| 작업 | 내용 | 방식 |
| --- | --- | --- |
| 이미지 분류 | 이미지 전체를 하나의 클래스로 분류 | CNN + FC Layer |
| 객체 탐지 | 개별 객체의 종류와 위치(Bounding Box)를 함께 예측 | CNN + Classification Head + Regression Head |
| 이미지 캡셔닝 | 객체 간 관계까지 파악해 문장으로 표현 | CNN(특징 추출) → Attention(관련 영역 집중) → RNN(문장 생성) |

***Bounding Box** : 객체를 둘러싼 사각형으로, 중심 좌표 (x, y)와 너비 w, 높이 h로 표현한다.

# 순환 신경망 (RNN)

### 순차(Sequence) 데이터

- 순서를 바꾸면 의미가 달라지는 데이터이다. ("개가 사람을 물었다" ≠ "사람이 개를 물었다")

| 비교 | 정형 데이터 | 순차 데이터 |
| --- | --- | --- |
| 순서를 섞으면 | 의미 그대로 | 의미가 달라짐 (순서의존성) |
| 샘플 간 관계 | 상호독립적 | 이전 시점이 현재에 영향 (자기상관성) |
| 예시 | 고객 정보 | 주가, 문장, 음성, 비디오 |

=> 기존 머신러닝은 정형 데이터를 전제로 설계되어서 순차 데이터를 다룰 RNN이 필요하다.

### RNN의 구조와 원리

1. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**구조와 원리**</mark> :
    은닉층에 **순환 구조**가 있어서, 매 시점 현재 입력과 이전 은닉 상태로 새 은닉 상태를 만들고 다음 시점으로 넘긴다.

    $$h_t = \tanh(U x_t + V h_{t-1} + b) \qquad o_t = f(W h_t + c)$$

    - U : 입력 가중치 / V : 순환 가중치 / W : 출력 가중치
    - tanh : -1 ~ 1로 제한해 값 폭주 방지
    - 출력 f : 이진 분류는 sigmoid, 다중 분류는 softmax

2. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**RNN 아키텍처**</mark> :

    | 구조 | 활용 |
    | --- | --- |
    | One to Many | 이미지 캡셔닝 |
    | Many to One | 스팸 분류, 감성 분류 |
    | 동기식 Many to Many | 객체 태깅 (매 시점 출력) |
    | 비동기식 Many to Many | 기계 번역, 요약 (끝까지 읽고 출력) |

### RNN의 한계와 장기 의존성 문제

RNN은 시간을 거슬러 오차를 전파하는데, chain rule에 의해 한 시점마다 같은 값이 계속 곱해진다.


| 한계 | 원인 | 결과 |
| --- | --- | --- |
| 기울기 소실 | 곱해지는 값 < 1 (tanh 미분 ≤ 1이라 자주 발생) | 먼 과거 정보가 반영 안 됨 |
| 기울기 폭주 | 곱해지는 값 > 1 | 오차가 진동하거나 NaN 발생 |
| 느린 훈련 | 시점을 순서대로만 계산 가능 (병렬 불가) | 긴 시퀀스일수록 비용 증가 |

=> 결국 관련 정보가 멀리 떨어질수록 학습하지 못하는 **장기 의존성 문제**로 귀결된다.

=> 그래서 중요한 건 기억하고 불필요한 건 잊는 LSTM, GRU가 등장했다.

# LSTM & GRU

***게이트** : 0~1 값을 정보에 곱해서 얼마나 통과시킬지 정하는 구성이다. (0 = 버림, 1 = 유지)
- 값은 sigmoid로 모델이 스스로 계산한다.

1. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**LSTM**</mark> :
    기억을 **셀 상태(장기 기억)** 와 **은닉 상태(단기 기억 & 출력)** 두 개로 나눠서 관리한다.

    | 단계 | 역할 | 수식 |
    | --- | --- | --- |
    | Forget Gate | 무엇을 잊을까 | $f_t = \sigma(W_f [h_{t-1}, x_t] + b_f)$ |
    | Input Gate | 무엇을 새로 기억할까 | $i_t = \sigma(W_i [h_{t-1}, x_t] + b_i)$, $\tilde{C}_t = \tanh(W_C [h_{t-1}, x_t] + b_C)$ |
    | Cell 업데이트 | 기억 갱신 (덧셈 구조가 핵심) | $C_t = f_t \odot C_{t-1} + i_t \odot \tilde{C}_t$ |
    | Output Gate | 무엇을 내보낼까 | $h_t = o_t \odot \tanh(C_t)$ |

    ***장단점**
    - 장점 : 덧셈 구조로 기울기 소실 완화, 장기 의존성 학습 가능
    - 단점 : 계산량이 많고 여전히 병렬 처리 불가, 매우 긴 시퀀스에서는 한계

2. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**GRU**</mark> :
    LSTM을 단순화한 모델로, 은닉 상태 하나에 기억과 출력을 모두 담는다.

    - **Update Gate** : Forget + Input을 합쳐 "과거 : 새 정보 = 몇 : 몇"을 결정


    - **Reset Gate** : 새 정보를 만들 때 과거를 얼마나 무시할지 결정
    - Output Gate 없음

3. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**LSTM vs GRU**</mark> :

    | 구분 | LSTM | GRU |
    | --- | --- | --- |
    | 상태 | 셀 + 은닉 (2개) | 은닉 (1개) |
    | 게이트 | Forget · Input · Output (3개) | Update · Reset (2개) |
    | 장점 | 정교한 기억 관리 | 단순하고 빠름 |
    | 단점 | 복잡, 계산량 많음 | 정교함 떨어짐 |

    => 우열이 아니라 **정교함(LSTM)** 과 **효율(GRU)** 사이의 선택이고, 많은 과제에서 성능은 비슷하다.

# Seq2Seq

- 입력 시퀀스를 받아 다른 시퀀스를 출력하는 모델로, **입력과 출력 길이가 달라도 된다.** ("I am a student" → "je suis étudiant")

- 활용 : 기계 번역, 챗봇, 문서 요약, 음성 인식

=> 인코더 → 컨텍스트 벡터 → 디코더

1. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**인코더**</mark> :
    입력 문장을 토큰화 → 임베딩 → LSTM에 한 단어씩 입력하고, **마지막 은닉 상태만** 컨텍스트 벡터로 디코더에 넘긴다.

2. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**컨텍스트 벡터**</mark> :
    문장 전체의 요약 벡터이자 인코더와 디코더를 잇는 **유일한 통로**이다. 크기는 은닉 상태 크기로 정하는 하이퍼파라미터이다.

3. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**디코더**</mark> :
    컨텍스트 벡터를 첫 은닉 상태로 받고, \<sos\>부터 시작해서 softmax로 다음 단어를 예측 → 예측한 단어를 다음 입력으로 넣는 과정을 \<eos\>가 나올 때까지 반복한다.

### Seq2Seq의 한계 : 병목 현상

문장 길이와 상관없이 **고정된 크기의 벡터 하나**에 전부 압축해야 해서 정보가 손실되는 현상이다.

- 짧은 문장은 충분히 담기지만, 긴 문장은 앞부분 정보가 희미해져 번역 품질이 급격히 떨어진다.

=> 이 한계를 해결한 것이 **Attention**
