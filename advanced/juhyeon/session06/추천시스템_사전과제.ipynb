# 추천 시스템 기초

### 추천 시스템과 데이터 유형

- 사용자의 행동 기록과 아이템 정보로 좋아할 만한 아이템을 예측해 골라주는 시스템이다. 검색은 아는 걸 찾게 돕고 추천은 모르는 걸 발견하게 돕는다.

1. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**명시적 데이터**</mark> :
    별점, 좋아요, 리뷰처럼 선호를 직접 표현한 데이터이다. 취향이 분명하지만 남기는 사람이 적고 사람마다 점수 기준이 다르다.

2. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**암시적 데이터**</mark> :
    클릭, 구매, 시청 시간처럼 행동에서 선호를 추측하는 데이터이다. 많이 쌓이지만 잘못 클릭했거나 선물일 수 있어서 행동이 곧 선호는 아니다.

=> 실제로는 암시적 데이터가 훨씬 많아서 둘을 섞어 쓴다.

### 추천 시스템 분류

| 구분 | 근거 | 예시 |
| --- | --- | --- |
| 내용 기반 | 아이템 속성의 유사도 | SF를 자주 봄 → 다른 SF 추천 |
| 협업 | 다수 사용자의 행동 패턴 | 나와 취향이 비슷한 사람이 본 영화 추천 |
| 하이브리드 | 둘 이상 조합 | 신작은 내용 기반, 평점이 쌓이면 협업 |

- 협업 필터링은 비슷한 **사용자**가 좋아한 걸 추천하는 User-based와, 내가 좋아한 것과 비슷한 **아이템**을 추천하는 Item-based로 나뉜다.

# 내용 기반 필터링

### 사용자 프로필과 유사도

- 아이템을 숫자로 표현 → 사용자 프로필 생성 → 유사도 계산 → 점수 높은 아이템 추천 순서이다.

1. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**사용자 프로필**</mark> :
    사용자가 좋아하는 특성을 정리한 정보이다. 가입 때 직접 고르게 하거나 본 아이템들의 특성 값을 평균 내서 만든다. 어벤져스, 인터스텔라, 노트북을 봤으면 [액션 0.4, 로맨스 0.4, SF 0.47, 공포 0.03]이 되고, 새 영화 "듄"과 코사인 유사도가 약 0.8이라 추천한다.

2. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**유사도**</mark> :
    장르, 감독 같은 feature를 원-핫 인코딩해서 벡터로 만들고 비교한다. 사용자와 아이템은 같은 feature로 표현해야 한다.
    - 내적 : 같은 위치끼리 곱해서 더한 값. feature가 많은 아이템이 유리해진다.
    - 코사인 유사도 : 내적을 길이로 나눠 방향만 본다.
    <br>
     $\cos(A,B) = \frac{A \cdot B}{\|A\|\|B\|}$
    - 그 외 자카드, 유클리드, 피어슨이 있다.

3. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**TF-IDF**</mark> :
    텍스트를 벡터로 바꿀 때 **이 문서에 자주 나오고 다른 문서에는 드문 단어**에 높은 값을 준다. "귀엽다"처럼 여러 문서에 나오면 IDF가 0에 가까워지고 "사냥한다"처럼 한 문서에만 나오면 커진다. 줄거리를 TF-IDF로 바꾸고 코사인 유사도로 비슷한 영화를 추천한다.

    $$TF\text{-}IDF = tf(d,t) \times \log\left(\frac{n}{1+df(t)}\right)$$

### 장단점

| 장점 | 단점 |
| --- | --- |
| 속성만 있으면 새 아이템도 추천 가능 | 속성 정보가 부족하면 정확도가 떨어짐 |
| 다른 사용자 데이터가 필요 없음 | 비슷한 것만 계속 추천 |
| 추천 이유 설명이 쉬움 | 기록 없는 신규 사용자는 추천 어려움 |

***Cold Start** : 데이터가 부족해 추천을 못 하는 문제로, 새 아이템은 내용 기반으로 풀리지만 새 사용자는 어렵다.

# 협업 필터링

### 메모리 기반 협업 필터링

- 저장된 평점을 그대로 꺼내 비슷한 "이웃"을 찾아 예측한다. 학습 단계가 없다.

1. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**사용자 기반**</mark> :
    - 나와 비슷한 사용자 TOP-N의 평점을 유사도로 가중평균해서 예측한다. 
    - 유사도는 주로 피어슨 상관계수를 쓴다. 
    - A와 B가 어바웃 타임, 노팅힐에 둘 다 높은 점수를 줬으면 B가 5점 준 라라랜드를 A에게 추천한다.

2. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**아이템 기반**</mark> :
    - 사람들이 함께 좋아하고 싫어한 패턴으로 아이템 유사도를 본다. 
    - 어바웃 타임을 좋아한 사람들이 노팅힐도 좋아했으면 어바웃 타임을 좋아하는 D에게 노팅힐을 추천한다.

| 기준 | User-based | Item-based |
| --- | --- | --- |
| 유리한 환경 | 아이템이 자주 바뀌는 곳 | 사용자가 많은 대형 플랫폼 |
| 계산 속도 | 추천할 때마다 이웃을 새로 찾음 | 미리 계산해두고 재사용 → 빠름 |
| 다양성 | 의외의 아이템 추천 가능 | 본 것과 비슷한 추천이 많음 |

### 메모리 기반 협업 필터링의 한계

1. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**희소 행렬**</mark> :
    평점 행렬은 대부분 비어 있다. 함께 평가한 게 적으면 유사도를 믿기 어렵기 때문에 모델 기반으로 해결한다.

2. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**콜드 스타트**</mark> :
    기록이 없는 신규 사용자나 신규 아이템은 추천이 안 되서 가입 때 취향고르기, 하이브리드, 인구통계 정보로 완화한다.

3. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**확장성**</mark> :
    사용자와 아이템이 늘수록 계산량이 커지므로 유사도 높은 k명만 쓰는 KNN 협업 필터링을 쓴다.

### 모델 기반 협업 필터링

- 평점 데이터로 모델을 미리 학습해두고 추천할 때는 그 모델로 빈칸을 빠르게 예측한다. 대표적인 게 잠재 요인 모델이고 구현 방법이 행렬 분해이다.

1. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**잠재 요인**</mark> :
    평점에 직접 드러나지 않는 숨은 특성이다. 축의 개수만 정해두고 각 축의 의미는 학습으로 정해져서 사람이 해석하기 어렵다.

2. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**행렬 분해 (MF)**</mark> :
    큰 평점 행렬을 사용자 행렬과 아이템 행렬, 작은 두 행렬의 곱으로 쪼개고 예측 평점은 두 벡터의 내적으로 계산한다.

    $$\hat{r}_{ui} = \mu + b_u + b_i + p_u^T q_i$$

    평점이 있는 칸만 보고 오차가 줄도록 벡터를 고치며 과적합을 막는 규제 항도 같이 둔다. $\mu, b_u, b_i$는 전체 평균과 사람, 아이템별 점수 경향이다.

=> 희소성과 확장성은 많이 풀었지만 **내적이라는 선형 계산**으로 예측한다는 한계가 남는다.

# VAE 기반 협업 필터링

### MF의 한계와 VAE 구조

- MF는 선형이라 "로맨스도 좋고 음악도 좋은데 뮤지컬 로맨스는 싫다" 같은 조합을 못 담는다. 또 사용자를 점 하나로 표현해서 기록 5개인 사람과 5,000개인 사람을 똑같이 확신한다.

1. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**AE**</mark> :
    입력을 작게 압축했다가 복원하는 신경망이다. 비선형이지만 사용자를 여전히 점 하나로 표현한다.

2. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**VAE**</mark> :
    사용자를 평균 μ와 표준편차 σ를 가진 **분포**로 표현하고 여기서 σ는 불확실성을 의미한다. 학습할 때 μ 근처에서 z를 뽑아 쓰기 때문에 기록이 적어도 노이즈에 덜 흔들린다.

3. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**Mult-VAE**</mark> :
    0/1 상호작용 벡터 → 인코더 → z 샘플링 → 디코더 → 확률 높은 순으로 추천한다.

### 목적 함수와 KL Divergence

$$\mathcal{L} = \underbrace{E_{q(z|x)} [ \log p(x|z) ]}_{\text{복원 항}} - \underbrace{D_{\text{KL}}(q(z|x) \,\|\, p(z))}_{\text{KL 항}}$$

- 복원 항은 입력을 잘 복원하게 하고 KL 항은 분포가 표준 정규분포에서 너무 벗어나지 않게 막는다. 예를들면, 시험으로 치면 "기출 풀기"와 "기본 개념 지키기"이다.

- KL Divergence는 두 분포가 얼마나 다른지 나타낸다. 중심에서 멀어지거나 너무 좁게 확신하거나 너무 퍼지면 값이 커진다.

=> KL 항 덕분에 기록이 적은 사용자도 비슷한 사용자 근처에 놓여서 희소한 데이터에서도 안정적으로 추천한다.

# GNN 기반 사용자 선호 예측

### GNN 구조와 프로세스

- 사용자-아이템 상호작용은 원래 그래프이다. 행렬에선 칸 하나였던 정보가 **나 → 영화 X → X를 본 사람 → 영화 Y** 같은 경로가 된다.

1. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**그래프**</mark> :
    노드와 엣지로 이루어진 자료구조로, 엣지는 방향이나 가중치를 가질 수 있다. 이미지, 텍스트와 달리 노드마다 이웃 수가 달라서 CNN, RNN 대신 GNN을 쓴다.

2. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**메시지 패싱**</mark> :
    이웃 노드의 정보를 나에게 모아 임베딩을 갱신한다.

    <br>
    [한 레이어의 3단계]
    
    - 메시지 생성 : 이웃이 자기 임베딩으로 메시지를 만든다.
    - 집계 : 받은 메시지를 평균, 합, attention 등으로 합친다.
    - 갱신 : 집계 결과와 내 이전 임베딩을 합쳐 새 임베딩을 만든다.

    K번 반복하면 K단계 떨어진 이웃 정보까지 반영된다.

### LightGCN

| 아키텍처 | 특징 |
| --- | --- |
| GCN | 이웃 임베딩을 연결 수로 정규화해 평균, 가장 기본 |
| GAT | attention으로 중요한 이웃에 가중치 |
| GraphSAGE | 이웃 일부만 샘플링, 대규모 그래프에 적용 |

- 추천에선 GCN을 적용한 NGCF가 먼저 나왔다. NGCF는 레이어마다 가중치 행렬 W와 비선형 활성화 함수를 썼는데 추천 그래프 노드엔 ID 임베딩만 있어서 두 연산이 도움이 안 되고 학습만 방해했다.

- LightGCN은 이 두 개를 빼고 **정규화된 이웃 임베딩의 합 + 레이어 결합**만 남겨 더 단순하고 NGCF보다 평균 약 16% 성능을 높혔다.

### GNN 활용 사례 및 장단점

- Pinterest가 대표 사례. 핀과 보드를 노드로 둔 약 30억 노드, 180억 엣지 그래프에서 이웃 일부만 샘플링해 학습했고 그 외 친구 추천, 분자 구조 분석, 이상 거래 탐지, 도착 시간 예측에도 쓴다.

| 장점 | 단점 |
| --- | --- |
| 이웃의 이웃까지 관계를 반영 | 레이어가 깊으면 임베딩이 다 비슷해짐 -> 과평탄화 |
| 소셜 관계, 속성 등을 한 그래프에 담음 | 먼 노드 정보가 압축되며 손실 -> 과압축 |
| 속성 노드로 콜드 스타트 완화 | 계산량이 크고 순서를 반영 못 함 |

# Transformer 기반 추천 시스템

### 기존 추천 시스템의 한계

- CF는 구매 순서를 무시하고, 방금 산 것과 1년 전에 산 것을 구분 못 한다.
- GNN은 레이어 수만큼만 볼 수 있어서 멀리 떨어진 행동 간 관계를 잡기 어렵다.

=> 둘 다 어떤 순서로, 지금 어떤 의도로 행동하는지 못 봐서 시퀀스 전체를 한 번에 보는 Transformer가 나왔다.

### Transformer과 Transformer 기반 주요 추천 시스템 모델

1. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**Self-Attention**</mark> :
    시퀀스 안 모든 아이템 쌍의 관련성을 거리와 상관없이 한 번에 계산한다. 수영복 → 비치타월 → 선크림이면 선크림은 "휴가 준비물", 스킨 → 로션 → 선크림이면 "스킨케어"로 해석된다. 프린터를 산 뒤 관련 없는 걸 5개 샀어도 프린터에 바로 높은 가중치를 줘서 토너를 추천할 수 있다.

    $$\text{Attention}(Q, K, V) = \text{softmax}\left(\frac{QK^T}{\sqrt{d}}\right) V$$

    Query와 Key를 내적해 관련성 점수를 내고 softmax로 비율로 바꾼 뒤 Value를 그 비율만큼 더한다.

2. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**SASRec / BERT4Rec**</mark> :

    | 구분 | SASRec | BERT4Rec |
    | --- | --- | --- |
    | 방향 | 단방향 | 양방향 |
    | 학습 목표 | 다음 아이템 예측 | [MASK]로 가린 아이템 예측 |
    | 학습 속도 | 빠름 | 상대적으로 느림 |
    | 잘 맞는 상황 | 피드, 커머스처럼 다음 행동 예측 | 장바구니처럼 순서가 느슨한 경우 |

    처음엔 BERT4Rec이 더 좋다고 했지만 후속 연구에서 손실 함수를 맞추니 SASRec이 비슷하거나 더 좋았다. -> 따라서 상황에 맞게 고르면 된다.

# 오늘날의 추천 시스템

### LLM, 인기 편향과 XAI, 비용 문제

1. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**LLM**</mark> :
    추천을 점수가 아닌 문장으로 낸다. "비 오는 날 혼자 보기 좋은 영화"처럼 대화형 추천이 되고 기록 없는 신상품도 설명만으로 추천할 수 있어서 콜드 스타트에 강하다. 

2. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**인기 편향**</mark> :
    인기 아이템이 더 추천되고 더 클릭돼서 격차가 계속 커진다. Spotify 실험에서 개인화 추천으로 스트리밍은 28.90% 늘었지만 소비 다양성은 11.51% 줄었다.

3. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**XAI**</mark> :
    딥러닝은 블랙박스라 추천 이유를 설명하기 어렵다. LLM이 이유를 말해줘도 실제 근거와 같다는 보장은 없다.

4. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**2단계 구조**</mark> :
    모든 아이템에 딥러닝을 돌리면 비용이 너무 커서 가벼운 모델로 수백 개 후보를 먼저 거르고 무거운 모델로 순위를 매긴다.

=> 실제 추천 시스템은 모델 하나가 아니라 지금까지 본 모델을 단계별로 섞어서 비용과 정확도를 같이 챙긴다.
