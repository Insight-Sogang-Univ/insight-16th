    1. 가지돌기로 외부 전기신호를 받음
    2. 모인 신호가 일정 크기 이상이면
    3. 축삭돌기로 신호를 전달

## 퍼셉트론
- 초기 형태의 인공신경망이자 **딥러닝의 기원**이 되는 **선형 이진 분류기**
- 흐름 : 입력 -> 가중합 -> 활성화 함수 -> 출력
  1. 입력값 x 가중치 : $x_1w_1$
  2. 총합(+편향) : $x_1w_1 + x_2w_2 + \dots + x_nw_n + b$
  3. 임계값을 넘으면 1, 아니면 0 출력 (계단함수)
- 학습 = 원하는 출럭이 나오도록 **가중치를 조정**하는 작업

<br>

### 용어정리

| 용어 | 의미 |
| --- | --- |
| **가중치** ($w$) | 입력값이 출력에 주는 영향력(중요도). 클수록 해당 입력이 중요 |
| **활성화 함수** | 가중합을 받아 출력값을 결정하는 함수. 0과 1사이를 출력.(퍼셉트론은 계단 함수) |
| **임계값** ($\theta$) | 활성화 함수에 사용되는 값으로 계단 함수에서는 가중합이 임계값을 넘으면 1, 넘지 못하면 0을 출력 |
| **편향** ($b$) | 임계값을 좌변으로 넘긴 형태. 입력값에 편향 b가 곱해진 형태. $\sum w_i x_i + b \ge 0 \rightarrow 1$ |

<br>

활성화함수는 계단 함수 외에도 다양한 활성화 함수를 사용
- Sigmoid
- Tanh : -1~1을 출력
- ReLU
- leaky ReLU

-> 임계값과 편향은 표현의 차이일 뿐이며 오늘날 딥러닝에서는 **편향**을 주로 사용

<br>

## 단층 퍼셉트론

1. 정의
  - **입력층+출력층**으로 구성된 단층 퍼셉트론
  - 값을 보내는 단계와 받아서 풀력하는 단계로만 이루어짐

2. 논리 게이트 구현
  - 논리 게이트 : n개의 입력값과 1개의 출력값으로 논리 연산을 수행하는 디지털 회로 구성요소
  - 단층 퍼셉트론으로 AND, OR, HAND는 쉽게 구현 가능

| 게이트 | 조건 | [w1, w2, b] 예시 |
| --- | --- | --- |
| **AND** | 둘 다 1일 때만 1 | `[0.5, 0.5, -0.7]` |
| **OR** | 둘 다 0일 때만 0 | `[0.6, 0.6, -0.5]` |
| **NAND** | 둘 다 1일 때만 0 (AND의 부호 반전) | `[-0.5, -0.5, 0.7]` |
| **XOR** | 서로 다를 때만 1 | **가능한 조합 없음** |

<br>

3. 단층 퍼셉트론의 한계 :
  XOR 게이트를 구현할 수 없음!
  - 퍼셉트론은 **입력과 가중치의 선형 결합**으로 이루어진 구조 -> 결정 경계가 **직선 하나**
  - 따라서 직선 하나로 나눌 수 있는 **선형 분류 문제**만 해결 가능
    - AND, OR, NAND : 직선 하나로 분리
    - XOR : 직선 하나로 분리 불가능 -> 곡선이나 비선형 영역이 필요


<br>

## 다층 퍼셉트론

- XOR은 AND, NAND, OR을 조합하면 만들 수 있음. 즉, 퍼셉트론의 층을 더 쌓자!
- 구성 : input + 하나 이상의 hidden layers + output

- 🌟 : 층만 쌓는다고 비선형이 되지 않음
  - 활성화 함수 없이 가중합만 이어 붙이면 결과적으로 하나의 선형식으로 구성됨

   $y = W_2(W_1x+b_1)+b_2 = (W_2W_1)x + (W_2b_1+b_2) = Wx+b$

   - MLP가 XOR를 구현할 수 있는 이유 : 은닉층에서 비선형 활성화 함수로 새로운 특징을 만들고, 다음층에서 그 특징을 조합하기 때문

   $h = \sigma(W_1x+b_1), \qquad y = W_2h+b_2 \quad (\sigma: \text{비선형 활성화 함수})$

   - XOR 구성 예시 (은닉 뉴런 2개짜리 은닉층 1개로 충분)
      - 은닉층: 뉴런1 = NAND, 뉴런2 = OR
      - 출력층: 두 결과에 AND -> XOR  

    - 단층도 계단 함수를 쓰지만 가중합 **한 번**에 임계값을 적용하므로 경계는 결과를 **다시 조합**할 수 있다는 점이 차이

<br>

## 딥러닝 모델의 학습 방법

1. 순전파와 역전파, 연쇄 법칙
  - 학습 알고리즘의 기본 구조
    1. 순전파 : 입력값으로부터 **예측** 수행
    2. 오차 계산 : 예측과 정답 비교 -> 손실함수
    3. 역전파 : 출력층 -> 입력층 방향으로 오차를 전달
    4. 가중치 업데이트
    5. 1~4번을 반복하며 오차가 0에 가까워지도록 함

    <br>

  - 전체 흐름
    1. 순전파 : 입력x -> 은닉층([w1] -> [w2]) -> 예측값
    2. 손실계산 : $L = ((예측값) - y)^2$
    3. 역전파 :
      - w1은 L과 직접 연결되지 않고 w1 -> w2 -> L로 순차적으로 연결
      - w1의 책임 = (w1이 w2에 준 영향)x(w2가 L에 준 영향) -> Chain Rule (연쇄법칙)
    4. 가중치 업데이트 : 계산된 값으로 가중치를 조금씩 수정하여 모델의 정확도를 높임

  - 연쇄 법칙 (Chain Rule)
    - 합성함수의 미분 = 각 함수 미분의 곱
    - 각 가중치가 최종 손실이 얼마나 기여했는지를 층별 영향을 순서대로 곱해 계산

    - 구한 기울기는 **손실을 줄이기 위해 가중치를 얼마나 조정**을 해야하나를 알려주는 방향

  - 기울기 소실 문제 (Vanishing Gradient)
    - 은닉층이 많고 각 층의 미분값이 1보다 작으면, 곱할수록 0에 수렴
    $$0.25 \times 0.25 \times 0.25 \times \cdots \approx 0$$
    <br>

    - 0.25 = 시그모이드 미분의 최댓값
    - 결국 **앞쪽 층의 가중치가 거의 업데이트가 되지 않아** 학습이 제대로 이루어지지 않음

<br>

## 손실 함수

1. 정의
  - 모델 예측이 얼마나 틀렸는지를 **수치로 표현**하는 함수
  - **실제값과 예측값의 차이**를 계산
  - 용어 정리
    - 목적 함수 : 최적화하려는 전체 대상 / 최대화, 최소화
    - 손실 함수 : 데이터 한 개에 대한 오차 / 최소화
    - 비용 함수 : 전체 데이터에 대한 평균 오차 / 최소화
    <br>

2. 종류
  - 1) 평균제곱오차(MSE) - 회귀
  $\text{MSE} = \frac{1}{N}\sum(y-\hat{y})^2$

    - 연속값 예측, 실제값과의 차이를 직접 측정하는 회귀문제
    <br>

  - 2) 이진 크로스 엔트로피 (BCE) - 이진분류
  $L = -\left[y\log\hat{y} + (1-y)\log(1-\hat{y})\right]$
    - $P(y|x) = p^y(1-p)^{1-y}$ 에 −log 적용
      - 정답이 1이면 p가 높아야 잘 예측한 것
      - 정답이 0이면 1-p가 높아야 잘 예측한 것
      - log : 확률의 곱을 합으로 바꿔줌
      - 마이너스 : 확률을 높이는 목표를 손실을 줄이는 목표로 바꿔줌
    - 정답이 1이면 $-\log\hat{y}$만 남음
      - 예측 확률이 1에 가까우면 손실 -> 0
      - 예측 확률이 0에 가까우면 손실 급증
    - 정답이 0이면 **반대로 동작**
    - MSE는 오차 크기만 보지만, BCE는 **얼마나 확신하고 예측했는지**까지 반영
    <br>

  - 3) 크로스 엔트로피 오차 (CEE) - 다중 분류
  $\text{CEE} = -\sum y\log\hat{y} \quad (y: \text{원-핫 인코딩 값}, \hat{y}: \text{예측값 확률})$
    - 다중 클래스 분류 문제에서 예측 확률이 실제 클래스와 얼마나 일치하는지 평가
    - BCE와 유사하지만 **다중 클래스**에 사용



<br>

## 활성화 함수

1. 정의
  - 각 층 계산 : 입력값 x 가중치 + 편향 -> **활성화 함수** -> 출력값 (wx + b)
  - 활성화 함수가 필요한 **두가지 이유**
    1) 비선형성 : 없으면 층을 쌓아도 선형 계산의 반복 - XOR 같은 문제 해결 불가
    2) 미분 가능성 : 역전파는 미분값을 연쇄적으로 곱하므로 미분이 되어야 함

    - 가중치 : 입력값이 결과에 얼마나 영향을 미치는지를 결정하는 값
    - 편향 : 활성화 함수의 출력값을 조정하는 값으로 기준점 역할을 수행

    | | 단층 퍼셉트론 | 다층 퍼셉트론 |
    | --- | --- | --- |
    | 활성화 함수 | 계단 함수 | Sigmoid, ReLU 등 |
    | 출력값 | 0 또는 1 | 연속값 |
    | 특징 | 선형 문제만 | 비선형 문제도 학습 가능 |

2. 종류
  - 계단 함수
  $$h(x) = \begin{cases}1 & (x \ge 0)\\ 0 & (x < 0)\end{cases}$$
    - 단층 퍼셉트론에서 사용, 출력 0 또는 1
    - 미분값이 0. -> 0에서는 미분이 불가능하기 때문에 역전파 불가 -> 다층에서 사용 불가능

  - 시그모이드
  $$f(x) = \frac{1}{1+e^{-x}}$$
    - 출력이 0~1 연속값. 얼마나 확실한지를 표현, 미분가능
    - 단점 : 미분값 최대 0.25일 떼, 층이 깊으면 **기울기가 소실**
    - 이진 분류 출력층에서 주로 사용

  - ReLU
  $$f(x) = \max(0, x)$$
    - 가장 많이 사용됨
    - 0보다 크면 그대로이고 0이하면 0으로 출력
    - 많이 쓰이는 이유
      1. 계산이 간단하고 빠름
      2. **비선형성 제공**
      3. 양수 구간 미분값=1 이면 기울기 소실 완화, 역전파 안정적

    - 단점: 음수 입력은 항상 0으로 Dead Neuron 발생. 이를 보완하기 위해 Leaky ReLU를 활용

  - Softmax
  $$f(x_i) = \frac{e^{x_i}}{\sum_j e^{x_j}}$$
    - 각 클래스 출력을 **확률로 변환** (모두 0~1, 합=1)
    - **다중 분류 출력층** 전용으로 사용
    - Softmax + CEE는 가장 표준적인 다중 분류 조합

  - Tanh
  $$f(x) = \frac{e^x - e^{-x}}{e^x + e^{-x}}$$
    - 시그모이드의 변형으로 출력이 -1~1사이
    - 중심이 0이라 학습이 더욱 안정적임
    - But, 여전히 기울기 소실이 발생. 그래서 실무에서는 잘 쓰지 않음

## 옵티마이저

1. 경사하강법 (Gradient Descent)
  - 손실을 가장 적게 만드는 가중치 W 찾는 것이 목표
  - 미분값의 부호로 방향이 정해짐
  | 현재 W 위치 | 기울기 | W − 기울기 | 결과 |
  | --- | --- | --- | --- |
  | 최솟값 오른쪽 | 양수 | W 감소 | 왼쪽으로 이동 |
  | 최솟값 왼쪽 | 음수 | W 증가 | 오른쪽으로 이동 |
  | 최솟값 | 0 | 그대로 | 멈춤 |

  $$W \leftarrow W - \alpha\frac{\partial E}{\partial W}$$

  - 편향 b도 같은 방식으로 x축을 b로 바꾸기만 하면 최적의 b도 같은 방법으로 찾을 수 있음
  <br>

2. 학습률
  - 한 번에 얼마나 움직일지를 보폭으로 비유
  - 미분값을 빼면 문제가 생기기 때문에 미분값에 α를 곱해 이동량을 조절
  | 학습률 | 결과 |
  | --- | --- |
  | 너무 작음 | 수렴하지만 **너무 오래** 걸림 |
  | 너무 큼 | 최솟값을 지나쳐 **발산** ->손실이 커짐 |
  | 적당함 | |

  - 시작값 : 0.001. 학습률로 0.01/0.001/0.0001 시도
  <br>

3. 배치, 에포크, 이터레이션
| 용어 | 의미 |
| --- | --- |
| **배치** | 가중치 업데이트에 쓰는 **데이터 묶음** = 하이퍼파라미터. 크면 느리고 메모리 부족, 작으면 불안정 |
| **에포크** | **전체 데이터**를 학습하는 횟수 |
| **이터레이션** | 한 번의 에포크를 끝내는 데 필요한 배치 수 |
   - 데이터 수 = 배치크기 x 이터레이션
   <br>


4. 경사하강법의 종류
| 종류 | 배치 크기 | 특징 | 코드 |
| --- | --- | --- | --- |
| **배치 GD** | 전체 데이터 | 오차를 구할 때 전체 데이터를 한 묶으로 묶음. 수렴이 안정적, 메모리 많이 필요. 업데이트 횟수 적음 | `batch_size=len(X_train)` |
| **확률적 GD (SGD)** | 하나의 데이터 | 자원 적은 컴퓨터에도 사용 가능, 불안정·정확도 낮을 수 있음 | `batch_size=1` |
| **미니 배치 GD** | 적절한 크기 | 빠르고 안정적, **가장 많이 사용** | `batch_size=128` | model.fit()의 bath_size 기본값 32. 보통 2^n으로 설정 |

<br>

5. 경사하강법의 한계 : Local Minima
  - MSE처럼 볼록한 손실함수는 반드시 바닥에 도달
  - 딥러닝 손실함수는 울퉁불퉁
  <br>

6. 옵티마이저 종류
  - 모멘텀 - 변수가 가던 방향을 개선
    - 기존 GD는 이전 기울기를 잊는 "기억상실증"
    - 가던 방향으로 관성을 부여 → Movement = Negative of Gradient + Momentum
    - 관성으로 Local Minima 탈출하고 완만한 구간을 빠르게 통과하는 효과를 보임

  - RMSProp — 학습률 개선
    - 기존 GD는 기울기가 큰 방향엔 너무 많이, 작은 방향엔 너무 조금 이동
    - 최근 기울기를 참고해 방향마다 학습률을 따로 조정해서 안정적이고 빠르게 학습하도록 함
      - 급한 경사에서는 속도 줄임 / 완만한 경사에서는 속도 유지·증가
  
  - Adam - RMSProp과 Momentum을 합친 것
    - 뱡향과 스텝 사이즈를 적절하게 합친 방법 -> 방향과 학습률 두가지를 모두 잡기
    - 1차 모멘트(기울기의 평균) + 2차 모멘트(기울기의 제곱 평균)를 사용하여 학습률 조정

7. 선택기준

| 상황 | 선택 | 이유 |
| --- | --- | --- |
| 잘 모르겠다 / 대부분 | **Adam** | 방향 + 보폭 모두 잡음, 사실상 기본값 |
| 학습이 한쪽으로 튐 | 모멘텀 | 방향 안정화 |
| 파라미터별 기울기 차이가 큼 | RMSProp | 방향별 학습률 조정 |
