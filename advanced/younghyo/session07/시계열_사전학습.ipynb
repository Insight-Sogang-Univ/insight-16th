## 시계열 분석

### 시계열 데이터란?
시계열 데이터: 시간 순서대로 일정한 간격으로 기록된 데이터. 주가, 기온, 매출처럼 같은 대상을 시간에 따라 반복해서 기록한 것

- 순서 자체가 정보라서 섞으면 의미가 사라짐
- 관측 간격이 일정해야 시점끼리 비교 가능
- 횡단면 데이터는 한 시점에 여러 대상을 보고, 시계열은 한 대상을 여러 시점에 걸쳐 봄

단변량 vs 다변량

- 단변량: 변수 하나만 보고 자기 과거로 예측함. ex. 일별 주가
- 다변량: 여러 변수를 같이 보고 서로의 영향까지 고려함. ex. 기온, 요일로 판매량 예측
- 다변량이 무조건 좋은 건 아님. 내일 기온처럼 미래 값을 모르는 변수는 그것도 따로 예측해야 함

시계열 분석의 목적은 과거 패턴을 찾아서 미래를 예측하는 것

- 과거 패턴이 미래에도 반복된다는 가정이 깔려 있음. 그래서 코로나처럼 처음 있는 일은 예측이 어려움
- 순서: 시계열 이해 -> 전처리 -> 정상성 판단 -> 모델링
- 일반 회귀를 그대로 못 씀. 어제 값이 오늘 값에 영향을 줘서 회귀의 가정이 깨짐

### 시계열 데이터의 특성
시간의존성: 현재 값이 과거 값의 영향을 받는 성질

- 보통 가까운 과거일수록 영향이 큼
- train/test를 랜덤으로 나누면 미래로 과거를 맞히는 데이터 누수가 생김. 반드시 과거로 학습하고 미래로 평가해야 함

계절성과 추세

- 계절성: 일정한 주기로 비슷한 패턴이 반복됨. 여름마다 에어컨이 잘 팔리는 것
- 추세: 장기적으로 오르거나 내리는 방향성. 스마트폰 보급률이 꾸준히 오르는 것
- 실제 데이터는 둘이 같이 나타나는 경우가 많음

자기상관성: 같은 변수의 현재와 과거 사이의 상관관계. 시계열에서 가장 중요한 특성

- 오늘 더우면 내일도 더울 가능성이 높은 것
- 시차를 바꿔 가며 계산하면 반복 주기를 찾을 수 있음. 월별 데이터에서 12개월 전과 자기상관이 높으면 1년 주기
- 자기상관이 있으면 일반 회귀분석을 쓸 수 없음
    - 오차끼리 독립이어야 하는데 오차도 어제를 따라가서 검정 결과가 왜곡됨
    - 둘 다 시간에 따라 오르기만 해도 상관이 높게 나오는 허위 상관이 생김
    - 예측 오차가 쌓이고 패턴이 바뀔 수도 있음
- 그래서 시계열 전용 모델을 쓰거나, 차분 같은 전처리를 하고 분석함

### 시계열 데이터의 분석요인
실제 시계열은 4가지 요인이 섞인 결과

- 추세: 장기적으로 오르거나 내리는 흐름. 1인 가구 수 증가
- 계절: 1년 이내의 정해진 주기로 반복됨. 날씨, 명절, 요일 같은 이유로 생김
- 순환: 계절보다 길고 주기가 일정하지 않은 변동. 경기 호황과 불황
    - 달력으로 말할 수 있으면 계절, 언제 올지 모르면 순환
    - 찾기 어려워서 보통 추세에 포함해서 봄
- 불규칙: 나머지를 다 빼고 남은 예측할 수 없는 변동. 잔차, 노이즈라고도 함

### 시계열 데이터 전처리
값 하나가 앞뒤와 이어져 있어서 잘못된 값 하나가 주변 분석까지 망가뜨림. 그래서 시간 흐름을 고려해서 전처리해야 함

결측치 처리

- 행을 지우면 관측 간격이 깨지고, 전체 평균으로 채우면 흐름이 무시됨
- 그래서 앞뒤 값으로 빈칸을 채우는 보간법을 가장 많이 씀
    - 선형 보간: 앞뒤를 직선으로 이음. 간단하지만 급한 변화에 약함
    - 다항식 보간: 곡선으로 이음. 차수가 높으면 과적합
    - 스플라인 보간: 구간마다 곡선을 나눠서 부드럽게 이음
    - 조각별 상수 보간: 직전 값이나 직후 값을 그대로 씀. ffill, bfill
- 회귀는 점을 안 지나도 되지만, 보간은 관측값을 그대로 믿고 반드시 지남
- 잘 모르겠으면 선형 보간부터 시작

이상치 처리

- 시계열 이상치는 언제 나타났는지가 중요함. 35도는 8월엔 정상이지만 1월엔 이상치
- 탐지 방법
    - IQR: 사분위 범위에서 너무 먼 값을 이상치로 봄. 간단하지만 추세나 계절성이 강하면 부적합
    - Z-score: 평균에서 표준편차 3배 넘게 떨어지면 이상치로 봄. 큰 이상치가 기준을 끌어올리는 문제가 있음
    - STL 분해: 추세와 계절을 빼고 남은 잔차가 크면 이상치로 봄. 진짜 이상한 값만 잡음
- 탐지 모델: 예측 오차가 큰 지점을 잡는 CART, 군집에서 벗어난 값을 잡는 클러스터링, 복원이 잘 안 되는 값을 잡는 오토인코더
- 처리: 먼저 오류인지 실제 사건인지 확인하고, 결측치로 바꾼 뒤 보간으로 채우는 경우가 많음

평활화: 자잘한 노이즈를 줄여서 큰 흐름을 보는 기술

- 이동평균이 대표적임. 일정 개수씩 묶어서 평균을 내고 한 칸씩 옮김
- 묶는 개수가 작으면 변화에 빠르게 반응하지만 노이즈가 남고, 크면 매끄럽지만 반응이 늦음
- 묶는 개수를 계절 주기와 같게 하면 계절성이 상쇄됨. 월별 데이터에 12개월 이동평균
- 종류
    - 단순 이동평균: 모든 값을 똑같이 취급함
    - 가중 이동평균: 최근 값에 더 큰 가중치를 줌
    - 지수 이동평균: 과거로 갈수록 가중치가 점점 줄어들고 모든 과거를 조금씩 반영함

요소분해: 섞여 있는 시계열을 추세, 계절, 잔차로 다시 나누는 과정

- 순환은 찾기 어려워서 추세에 포함하고 세 가지로 나눔
- 가법 모형: Y = T + S + R
    - 계절 변동 크기가 항상 일정할 때 씀. 여름마다 항상 50잔 더 팔리는 카페
- 승법 모형: Y = T × S × R
    - 추세가 커질수록 계절 변동도 같이 커질 때 씀. 여름마다 항상 20% 더 팔리는 카페
    - 현실의 매출, 주가는 승법에 가까운 경우가 많음. 로그를 취하면 가법처럼 다룰 수 있음
- 그래프 물결 높이가 일정하면 가법, 점점 커지면 승법
- seasonal_decompose: 이동평균으로 추세를 먼저 뽑고, 추세를 뺀 나머지에서 같은 계절끼리 평균 내서 계절을 구함. 남은 게 잔차
- STL: 이동평균 대신 LOESS로 구간마다 유연하게 추세와 계절을 추정함
    - 추세와 계절을 번갈아 다듬는 과정을 반복함
    - 잔차가 큰 이상치에는 작은 가중치를 줘서 이상치에 덜 흔들림
    - 시간에 따라 바뀌는 계절 패턴도 잘 잡음

### 정상성 판단
정상성: 평균, 분산, 자기상관이 시간에 따라 변하지 않는 성질. 어디서 잘라 봐도 비슷한 모습

- 정상 시계열은 일정한 평균 주변을 오르내리고, 벗어나도 다시 돌아옴. 주가 등락률
- 비정상 시계열은 추세나 계절성이 있거나 변동 폭이 바뀜. 주가 자체, GDP
- 규칙이 변하지 않아야 과거로 배운 게 미래에도 통함. 현실 데이터는 대부분 비정상이라 정상으로 만들어서 분석해야 함

강한 정상성과 약한 정상성

- 강한 정상성: 분포 전체가 시간에 따라 전혀 안 변함. 현실에선 거의 없음
- 약한 정상성: 평균이 일정하고, 분산이 일정하고, 두 시점의 관계가 날짜가 아니라 시간 간격에만 달려 있으면 됨
- 대부분의 시계열 모델은 약한 정상성을 가정함

정상성 판단 - 시각

- 정상: 수평선 주위로 비슷한 폭으로 오르내림
- 비정상: 계속 오르거나 내리고, 같은 모양이 반복되고, 폭이 점점 커짐
- ACF 그래프가 빠르게 0으로 떨어지면 정상, 아주 천천히 줄면 비정상

정상성 판단 - 통계

- 그래프는 사람마다 판단이 달라서 단위근 검정으로 객관적으로 확인함
- 단위근이 있으면 과거 충격이 사라지지 않고 계속 쌓임. 평균으로 안 돌아오고 분산도 계속 커져서 비정상
- ADF 검정: 귀무가설이 비정상. p-value가 0.05보다 작으면 정상
- KPSS 검정: 귀무가설이 정상. p-value가 0.05보다 작으면 비정상
- 둘은 출발점이 반대라 같이 써서 서로 보완함
    - 둘 다 정상 -> 바로 모델링
    - 둘 다 비정상 -> 차분 후 다시 검정
    - ADF만 비정상 -> 추세 정상 시계열이라 추세 제거
    - KPSS만 비정상 -> 차분 정상 시계열이라 차분

정상성 확보 방법

- 추세가 있으면 차분: Y(t) - Y(t-1)
    - 값 대신 얼마나 변했는지를 봄. 계속 오르는 판매량도 하루 증가량은 비슷한 값 주변에 머묾
    - 1차 차분으로 부족하면 2차까지 하지만, 너무 많이 하면 없던 패턴이 생김
    - 예측 결과가 변화량이라 원래 값으로 되돌리는 역차분이 필요함
- 계절성이 있으면 계절 차분: 한 주기 전 값을 뺌. 올해 8월과 작년 8월을 비교하는 것
- 변동 폭이 커지면 로그 변환: 큰 값을 더 많이 줄여서 폭을 맞춰 줌. 0이나 음수엔 못 씀
- 비정상이라고 무조건 차분하는 게 아니라 원인에 맞는 방법을 골라야 함

### 단변량 시계열 분석
변수 하나의 기록만 가지고 자기 과거로 예측하는 것

- 값은 그날 관측된 결과이고, 충격은 그날 새로 생긴 예측할 수 없는 변동
- 주말 효과나 미리 공지한 할인처럼 미리 알 수 있는 건 충격이 아님

안정적 시계열 분석: 정상 시계열, 즉 돌아갈 평소가 있는 시계열

- AR: 오늘 값을 자기 과거 값으로 설명함
    - 손님이 어제 손님 수를 보고 오는 가게. 입소문처럼 영향이 계속 이어지면서 점점 옅어짐
    - AR(p)는 과거 p개 시점의 값을 씀
- MA: 오늘 값을 과거의 충격으로 설명함
    - 손님이 지난 방송을 보고 오는 가게. 정해진 날수가 지나면 영향이 딱 끊김
    - MA(q)는 과거 q개 시점의 충격을 씀
- 오늘 충격은 아직 모르니까 예측할 땐 0으로 둠

AR, MA의 전제조건

- 정상성: 충격을 받아도 평소로 돌아와야 함
    - AR은 계수 조건이 필요함. AR(1)이면 계수의 절댓값이 1보다 작아야 함
    - MA는 영향이 정해진 날 뒤에 끝나서 항상 정상
- 가역성: 관측된 값으로 지난 충격을 거꾸로 계산할 수 있어야 함
    - AR은 항상 가역
    - MA는 계수 조건이 필요함. MA(1)이면 계수의 절댓값이 1보다 작아야 오차가 안 불어남
- 단위근 검정은 데이터를 보는 것이고, 모델의 정상성은 추정한 계수로 따로 확인함

ARMA: AR과 MA를 합친 모형

- 오늘 값이 과거 값과 과거 충격 양쪽의 영향을 받음
- AR만 쓸 때보다 적은 계수로 설명할 수 있음
- AR 부분은 정상성, MA 부분은 가역성을 만족해야 함

차수 결정 방법: ACF, PACF

- ACF: 오늘과 k시점 전의 상관. 중간 시점을 거쳐 온 영향까지 다 포함함 -> MA 차수 q 결정
- PACF: 중간 시점의 영향을 빼고 직접적인 관계만 봄 -> AR 차수 p 결정
- 그래프 막대가 갑자기 신뢰구간 안으로 떨어지기 직전 시차를 고름
    - ACF는 천천히 줄고 PACF가 p 뒤에 끊기면 AR(p)
    - ACF가 q 뒤에 끊기고 PACF는 천천히 줄면 MA(q)
    - 둘 다 끊기면 ARMA(p, q)

불안정 시계열 분석: 추세나 계절성 때문에 돌아갈 평소가 없는 시계열

- AR, MA는 결국 평소로 돌아온다고만 예측함. 키처럼 계속 크는 데이터를 그대로 넣으면 키가 줄어든다고 예측함
- 그래서 1년간 자란 길이처럼 차분해서 평소를 만든 뒤 모델에 넣음
- ARIMA: ARMA에 차분을 더한 모형
    - p는 AR 차수, d는 차분 횟수, q는 MA 차수
    - 차분해서 증감으로 바꾸고 -> 증감을 ARMA로 예측하고 -> 마지막 실제 값에 더해서 되돌림
    - 랜덤워크는 상수 없는 ARIMA(0,1,0). 내일 예측이 오늘 값과 같음. 주가, 환율
    - 상수가 있으면 표류가 있는 랜덤워크. 떠돌면서도 한 방향으로 흘러감
- SARIMA: ARIMA에 계절 차분과 계절 AR, MA를 더한 모형
    - 지난달 대비로 추세를 없애고, 작년 같은 달 대비로 계절 효과를 없앰
    - 대문자 P, D, Q는 ARIMA를 한 주기 전 기준으로 한 번 더 하는 것이고, m은 주기 길이. 월별 데이터면 12
    - 계절성이 있으면 ACF가 12, 24, 36에서 다시 솟음. 이 상태론 차수를 못 읽으니 계절 차분부터 함
- 하이퍼파라미터 정하기
    - d, D는 ADF로 정상이 될 때까지 차분해 보고 정함. 보통 0 아니면 1
    - p, q는 ACF, PACF로 대략 범위를 잡고, AIC, BIC가 가장 낮은 조합을 그리드 서치로 고름

### 다변량 시계열 분석
여러 변수를 같이 놓고, 각 변수를 자기 과거와 다른 변수의 과거로 예측하는 것

VAR: AR을 여러 변수로 넓힌 모형

- 변수마다 식이 하나씩 있고, 각 식에 모든 변수의 과거 값이 들어감
- 판매량과 광고비로 보면, 광고를 하면 판매가 늘고 판매가 떨어지면 광고를 늘리는 식으로 서로 영향을 주고받음
- 적합한 경우: 변수끼리 양방향으로 영향을 줄 때, 변수가 적고 데이터가 충분할 때
- 부적합한 경우: 영향이 한쪽으로만 갈 때, 변수가 많은데 데이터가 적을 때. 변수가 늘면 계수가 빠르게 늘어남
- 변수 선택은 Granger 인과관계 검정으로 함. 한 변수의 과거가 다른 변수 예측에 도움이 되는지 보는 것이고, 진짜 인과관계를 뜻하진 않음
- 분석 방법
    - 충격 반응 함수: 한 변수에 충격을 줬을 때 다른 변수가 며칠에 걸쳐 어떻게 반응하는지 봄
    - 예측오차 분산 분해: 예측이 빗나간 게 어느 변수 탓인지 비율로 나눠서 봄

### 딥러닝을 활용한 시계열 분석 (참고)
전통 모델은 데이터가 많으면 느리고 비선형 패턴을 잘 못 잡아서 딥러닝을 씀

- RNN 계열: RNN, LSTM, GRU. 순서대로 정보를 넘기는 구조라 긴 시계열이나 병렬 처리엔 약함
- Transformer 계열: 긴 시계열은 Informer, 주기가 뚜렷하면 Autoformer, 외부 변수가 많고 해석이 필요하면 TFT
- Foundation Model: 시계열을 대량으로 미리 학습한 범용 모델. TimeGPT, Chronos 등
- 데이터가 적으면 딥러닝보다 ARIMA 같은 통계 모델이 더 나을 수 있음
