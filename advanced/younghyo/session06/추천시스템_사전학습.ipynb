## 추천 시스템

### 추천 시스템 기초
추천 시스템: 사용자의 행동 기록과 아이템 정보로 사용자가 좋아할 만한 아이템을 예측해 골라 주는 시스템 (ex. 넷플릭스 추천 콘텐츠, 유튜브 다음 영상)

- 아이템: 상품, 영화, 음악, 뉴스, 앱 등 모두 포함
- 고객 입장: 찾는 시간과 노력이 줄어듦, 몰랐던 취향 발견 -> 검색은 아는 것을 찾고, 추천은 모르는 것을 발견하게 도움
- 기업 입장: 이용 시간, 구매 전환, 재방문 증가, 비인기 아이템(롱테일) 노출

데이터 유형

- 명시적 데이터: 사용자가 선호를 직접 표현 (ex. 별점, 좋아요·싫어요, 리뷰)
    - 장점: 선호와 불호가 분명함
    - 단점: 평가를 남기는 사람이 적음, 취향이 변함, 사람마다 점수 기준이 다름
- 암시적 데이터: 행동에서 선호를 간접적으로 추측 (ex. 조회수, 클릭, 구매, 시청 시간, 검색)
    - 장점: 따로 입력하지 않아도 대량으로 쌓임
    - 단점: 행동이 곧 선호는 아님 (잘못 클릭, 선물 구매) / 행동이 없다고 싫어하는 것도 아님 (아직 못 봤을 수 있음)
- 실제 서비스는 암시적 데이터가 훨씬 많아서 둘을 섞어 씀

추천 시스템 분류 (무엇을 근거로 추천하나)

- 내용 기반 필터링: 아이템 특징의 유사도 기준 (ex. SF 영화를 자주 봄 -> 다른 SF 영화 추천)
- 협업 필터링: 다수 사용자의 선호 패턴 기준 -> 취향이 비슷한 사람이 좋아한 아이템은 나도 좋아할 가능성이 높음
    - 사용자 기반: 나와 취향이 비슷한 사용자가 좋아한 아이템 추천 (ex. A와 B가 라면, 콜라를 같이 삼 -> A가 산 생수를 B에게 추천)
    - 아이템 기반: 내가 좋아한 아이템과 함께 선호되는 아이템 추천 (ex. 콜라를 산 사람들이 생수도 자주 삼 -> 콜라를 좋아하는 B에게 생수 추천)
- 하이브리드: 두 가지 이상을 조합해 서로의 단점 보완 (ex. 새 영화는 장르·감독으로 내용 기반 추천 -> 평점이 쌓이면 협업 필터링)
    - Weighted Ensemble: 여러 모델의 결과를 점수나 투표로 합침
    - Mixed: 여러 알고리즘의 결과를 함께 보여 줌
    - Switch: 상황에 따라 알고리즘을 골라 씀 (위의 새 영화 예시)
    - Feature Combination: 여러 feature를 모두 합쳐 하나의 모델로 학습
    - Meta-Level: 첫 번째 모델의 결과가 다음 모델의 입력이 됨

### 내용 기반 필터링
흐름: 아이템을 숫자로 표현 -> 사용자 프로필 생성 -> 유사도 계산 -> 점수가 높은 아이템 추천

사용자 프로필: 사용자가 어떤 특성을 좋아하는지 정리한 정보

- 직접 지정형: 사용자가 취향을 직접 알려 줌 (ex. 음악 앱 가입 시 좋아하는 장르 선택)
- 간접 지정형: 과거 행동에서 취향을 추론 (ex. 히어로 영화를 본 사람에게 스파이더맨 추천)
- 만드는 방법
    - 본 아이템들의 특성 가중치(0~1) 평균 (ex. 어벤져스, 인터스텔라, 노트북의 [액션, 로맨스, SF, 공포] 평균 -> [0.4, 0.4, 0.47, 0.03] -> 신규 영화 듄 [0.4, 0, 0.9, 0]과 코사인 유사도 약 0.8 -> 추천)
    - 설문, 키워드, 평가로 따로 파악 (명시적·암시적 피드백)
    - 기존 평가를 정답 라벨로 써서 분류 문제로 해결 (결정 트리, 랜덤 포레스트, KNN 등)

유사도: 두 대상이 얼마나 비슷한지 나타내는 숫자

- 아이템과 사용자를 같은 feature의 숫자 벡터로 표현해야 비교 가능
- 아이템을 숫자로: 장르, 감독, 배우 같은 feature를 원-핫 인코딩 (속하면 1, 아니면 0)
- 내적: 같은 위치끼리 곱해서 더한 값 -> 클수록 비슷함 / 0과 1 벡터라면 겹치는 feature 개수와 같음
- 유사도 종류: 자카드(교집합 크기), 코사인(벡터 사이 각도), 유클리드(벡터 사이 거리, 작을수록 비슷함), 피어슨(선형 상관관계)
- 코사인 유사도: cos(A, B) = A·B / (|A||B|) -> 길이는 무시하고 방향만 봄 (ex. [1, 1]과 [5, 5]는 1) / 거리로 쓸 때는 1 - 코사인 유사도
- 내적 vs 코사인: 내적은 feature가 많은 아이템이 유리함 / 코사인은 이 쏠림은 줄지만 정보가 풍부하다는 점은 반영 못함

TF-IDF: 특정 문서에는 자주 나오고 다른 문서에는 드문 단어에 높은 값을 주는 방법 -> "그리고", "있다" 같은 흔한 단어의 영향을 줄임

- TF: 문서 d에서 단어 t가 나온 횟수
- IDF: log(n / (1 + DF)) -> 단어가 나온 문서 수(DF)가 많을수록 작아짐
- TF-IDF = TF × IDF -> 높을수록 그 문서에서 중요한 단어
- ex. 문서 3개(고양이는 귀엽다 / 강아지는 귀엽다 / 고양이는 사냥한다) -> "귀엽다"는 IDF = log(3/3) = 0, "사냥한다"는 log(3/2) ≈ 0.176 -> 사냥한다가 더 중요
- 추천: 줄거리를 TF-IDF 벡터로 변환 -> 재밌게 본 영화와 코사인 유사도 비교 -> 유사도가 높은 영화 추천

장단점

- 장점: 아이템 콜드 스타트 해결 (속성만 있으면 신규 아이템 추천 가능), 다른 사용자 데이터 불필요, 추천 이유 설명 쉬움
- 단점: feature로 담을 수 있는 정보에 한계, 비슷한 것만 계속 추천 (과도한 특화), 기록이 없는 신규 사용자는 추천 어려움
- 콜드 스타트: 정보가 부족해 적절한 추천을 못하는 문제 -> 아이템 콜드 스타트는 내용 기반으로 해결 가능, 사용자 콜드 스타트는 어려움

### 협업 필터링
가정: 과거에 비슷한 선호를 가진 사용자는 앞으로도 비슷한 선호를 가질 가능성이 높음

메모리 기반 협업 필터링: 저장된 평점을 그대로 꺼내 비슷한 이웃을 찾아 예측 (학습 단계 없음, 이웃 기반 CF라고도 함)

- 사용자 기반: 나와 유사도가 높은 TOP-N명의 평점을 유사도로 가중평균해 예측 -> "당신과 비슷한 고객들이 이 상품도 구매했습니다"
    - 평점 데이터에서는 피어슨 상관계수를 많이 씀 (두 사용자가 모두 평가한 아이템 기준)
    - 예측 평점 = 내 평균 + (이웃별 유사도 × 이웃 평점과 이웃 평균의 차이)의 합 / 유사도 절댓값의 합
    - ex. A는 B와 비슷(0.97), C와 정반대(-1) -> B는 라라랜드를 평균보다 0.75점 높게, C는 0.5점 낮게 줌 -> A의 예측 약 3.96점 -> 추천
- 아이템 기반: 내가 평가한 아이템과 평점 패턴이 비슷한 아이템 추천 -> "이 상품을 선택한 고객들은 이 상품도 구매했습니다"
    - 아이템 속성이 아니라 함께 좋아하고 싫어한 패턴으로 유사도 판단
    - 아이템 관계는 잘 안 변해서 유사도를 미리 계산해 재사용하기 좋음
    - ex. 어바웃 타임을 높게 준 사람은 노팅힐도 높게 줌 -> 어바웃 타임을 좋아하는 D에게 노팅힐 추천, 낮게 준 C에게는 추천하지 않음

| 기준 | 사용자 기반 | 아이템 기반 |
|---|---|---|
| 정확성 | 아이템 수가 사용자 수보다 많을 때 유리 | 사용자 수가 아이템 수보다 많을 때 유리 |
| 공간 | 사용자 간 유사도 저장 | 아이템 간 유사도 저장 |
| 안정성 | 사용자 변화의 영향을 많이 받음 | 아이템 목록이 잘 안 변하면 안정적 |
| 계산 속도 | 추천할 때마다 이웃을 새로 찾음 | 미리 계산해 재사용 -> 빠름 |
| 다양성 | 의외성, 참신성이 강점 | 비슷한 추천이 많아 다양성 부족 |
| 활용 | 뉴스, SNS 친구 추천, 음악 | 아마존 같은 대형 플랫폼 |

- 사용자가 많으면 아이템 기반이 유리한 이유: 두 아이템을 함께 평가한 사용자는 많음, 새 사용자가 새 아이템보다 자주 추가됨

메모리 기반의 한계

- 희소 행렬: 사용자-아이템 행렬 대부분이 비어 있음 (ex. 넷플릭스 프라이즈 약 99%가 빈칸) -> 함께 평가한 아이템이 적어 유사도를 믿기 어려움 -> 모델 기반 CF로 해결
- 콜드 스타트: 기록이 없는 신규 사용자, 신규 아이템(first-rater 문제)은 추천 불가 -> 가입 시 취향 직접 선택, 내용 기반과 하이브리드, 같은 카테고리 임베딩 평균, 인구통계 정보 활용
- 확장성: 사용자와 아이템이 늘수록 유사도 계산량 증가 -> KNN CF (유사도가 높은 k명만 사용, k가 너무 작으면 흔들리고 너무 크면 흐려짐)

모델 기반 협업 필터링: 관측된 평점으로 모델을 학습하고, 학습된 파라미터로 빈칸의 평점을 예측

- 학습 단계: 기존 평점을 가장 잘 설명하는 파라미터를 미리 찾음 / 예측 단계: 학습된 파라미터로 빠르게 계산
- 평점 이면의 잠재된 패턴을 학습 -> 직접 겹치는 평가가 없어도 빈칸 추정, 추천 시 계산이 가벼움
- 잠재 요인: 평점에 직접 드러나지 않는 숨은 특성 (취향, 성향을 숫자로 표현)
- 잠재 요인 모델: 사용자와 아이템을 같은 잠재 공간의 벡터로 표현 / 축의 개수(k)는 사람이 정하지만 축의 의미는 학습으로 정해져 해석이 어려움
- 행렬 분해(MF): 큰 평점 행렬을 작은 행렬 두 개의 곱으로 쪼갬 -> R̂ = PᵀQ, 예측 평점 = 사용자 벡터와 아이템 벡터의 내적
    - ex. 사용자 (로맨스 0.9, 액션 0.2) -> 노팅힐 (0.8, 0.1) 0.74점, 존 윅 (0.1, 0.9) 0.27점 -> 노팅힐 추천
- 학습 과정: k 정하기 -> 벡터를 임의 값으로 시작 -> 평점이 있는 칸만 예측해 오차 계산 -> 오차가 줄도록 수정 반복 -> 빈칸 예측
- 손실 함수: (실제 평점 - 예측 평점)² 의 합 + λ × (벡터 크기)² -> 뒷부분은 과적합을 막는 규제
- SGD: 평점을 하나씩 보며 조금씩 수정 / ALS: 사용자와 아이템 벡터를 번갈아 고정하고 계산 (병렬 처리에 유리)
- 편향: 예측 = 전체 평균 + 아이템 편향 + 사용자 편향 + 내적 (ex. 평균 3.7 + 타이타닉 0.5 - 깐깐한 Joe 0.3 = 3.9점)
- 분해 방식: SVD(수학적 분해, 결측치 처리 필요) / 학습 기반 MF(관측된 평점만 학습) / NMF(비음수 제약, 해석력 높음) / PMF(확률적 접근)

| | 메모리 기반 | 모델 기반 |
|---|---|---|
| 예측 방식 | 비슷한 사용자, 아이템을 직접 찾음 | 학습한 벡터를 내적 |
| 확장성 | 추천할 때마다 유사도 계산 | 학습은 미리, 추천 시엔 내적만 -> 빠름 |
| 콜드 스타트 | 해결 어려움 | 해결 어려움 (상호작용이 없으면 벡터 학습 불가) |
| 장점 | 추천 이유 설명 쉬움 | 데이터를 압축해 빠르고 일반화에 유리 |

### VAE 기반 협업 필터링
VAE-CF: 사용자의 상호작용 기록을 신경망으로 확률분포로 압축한 뒤 복원하면서, 아직 보지 않은 아이템의 선호를 예측

MF의 한계

- 선형성: 내적은 요인별로 곱해서 더하기만 함 -> 요인끼리 얽히는 효과 표현 불가 (ex. 로맨스도 음악 영화도 좋지만 뮤지컬 로맨스는 싫은 사용자)
- 불확실성 무시: 사용자마다 벡터 하나(점 추정) -> 기록 5개와 5,000개인 사용자를 똑같이 확신
- 딥러닝의 해법: 비선형 활성화 함수를 가진 신경망, 평균과 분산을 함께 출력

AE vs VAE

- 오토인코더(AE): 입력 -> 인코더 -> 작은 잠재 벡터 z -> 디코더 -> 복원 / 입력 자체가 정답
    - AutoRec: AE를 CF에 적용한 초기 모델 -> 선형 RMSE 0.872, sigmoid 0.831 -> 비선형성이 성능을 높임
    - 하지만 AE도 사용자를 점 하나로 표현
- VAE: 잠재 벡터를 분포로 표현 -> 평균 μ(취향의 중심), 표준편차 σ(불확실성)
    - ex. 5편만 본 사용자 -> AE는 우연히 본 공포 영화 같은 노이즈까지 반영 / VAE는 μ 주변을 샘플링해 비슷한 위치에서 비슷한 추천이 나오게 학습
    - 안정적인 이유: μ 주변을 사용해 학습, KL 항이 기준 분포 쪽으로 당김

구조 (Mult-VAE)

- 입력: 아이템 수만큼 긴 0/1 벡터 (평점은 4점 이상만 1)
- 인코더: μ와 σ 출력 -> 샘플링: 분포에서 z를 뽑음 -> 디코더: 모든 아이템 점수를 softmax로 확률 변환 -> 확률 높은 순으로 추천

목적 함수 ELBO: 복원 항 - KL Divergence 항을 최대화

- 복원 항: 입력을 최대한 비슷하게 복원 (AE와 같은 역할)
- KL 항: 잠재 분포가 표준 정규분포 N(0, 1)에서 크게 벗어나지 않게 함 -> 과적합 방지, 잠재 공간을 연속적이고 부드럽게 유지
- 비유: 복원 항은 "기출을 많이 풀어라", KL 항은 "기본 개념에서 벗어나지 마라"

KL Divergence: 두 확률분포가 얼마나 다른지 나타내는 값

- 정규분포끼리는 KL = ½(μ² + σ² - ln σ² - 1)
- μ=0, σ=1 -> 0 / μ=2 -> 2.0 / σ=0.1 -> 약 1.81 / σ=3 -> 약 2.90 -> 중심에서 멀거나, 너무 확신하거나, 너무 퍼지면 벌점이 커짐
- AE는 벡터가 흩어지고 빈 공간이 학습되지 않음 / VAE는 분포가 중심 근처에 연속적으로 모임 -> 기록이 적은 사용자도 비슷한 사용자 근처에 놓여 희소한 데이터에 강함

장점: 불확실성 반영, 희소한 데이터에서 안정적, 비선형 패턴 학습

### GNN 기반 사용자 선호 예측
배경: CF, MF, VAE는 상호작용을 행렬의 칸 하나로 다룸 -> "A가 본 영화를 B도 봤고, B는 Y도 봤다" 같은 이어진 관계와 행동 순서를 직접 활용 못함

GNN(그래프 신경망): 사용자-아이템 상호작용을 그래프로 보고 관계 자체를 학습

- 그래프: 노드(사용자, 아이템 같은 개체)와 엣지(관계, 상호작용)로 구성 / 엣지는 방향과 가중치(평점, 구매 횟수)를 가질 수 있음
- 행렬에서는 칸 하나였던 정보가 그래프에서는 경로가 됨 (ex. 나 -> 영화 X -> X를 본 사용자 -> 영화 Y)
- 쓰는 이유: 관계 같은 추상적인 개념을 직접 학습, 격자 구조가 없는 Non-Euclidean 데이터 학습 가능 (이미지·텍스트는 이웃 구조가 고정, 그래프는 노드마다 이웃 수가 다름)

구조와 프로세스

- 그래프 구조 생성: 노드와 엣지 정의, 초기 피처(임베딩) 부여
    - ex. 사용자 3명 + 영화 3편 = 노드 6개, 평점 5개 = 엣지 5개
    - 인접 행렬: 노드 간 연결 여부 (연결되면 1) / 피처 행렬: 노드 고유의 특성 (20대, 액션, SF 등)
- 메시지 패싱: 각 노드가 이웃의 정보를 받아 자기 임베딩을 갱신 (모든 노드가 한 번씩 타깃 노드)
    - 메시지 생성: 이웃 노드가 자기 임베딩으로 메시지를 만듦
    - 집계: 받은 메시지를 하나로 합침 (sum, mean, max, attention)
    - 갱신: 집계 결과 + 자신의 이전 임베딩 -> 신경망과 활성화 함수를 거쳐 새 임베딩
- K번 반복하면 K단계 떨어진 이웃 정보까지 반영 -> 이웃 취향까지 반영된 최종 임베딩으로 선호도 예측

주요 아키텍처 (메시지를 어떻게 만들고 어떻게 합치는가)

| 아키텍처 | 작동 방식 | 특징 |
|---|---|---|
| GCN | 이웃 임베딩을 연결 수로 정규화해 평균 | 가장 기본 구조, 전체 그래프 필요 |
| GRN | RNN + GNN | 시간에 따라 변하는 그래프 포착 |
| GAT | 이웃마다 attention으로 중요도를 학습해 가중 평균 | 중요한 이웃에 집중, 계산량 증가 |
| GraphSAGE | 이웃 일부만 샘플링, 자기 정보와 concat | 대규모 그래프, 새 노드에도 적용 가능 |

- 비유: GNN은 친구 의견을 똑같이 평균 내는 사장, GAT는 취향이 비슷한 친구 의견에 더 귀 기울이는 사장
- GNN은 그래프 신경망 전체 개념, GCN은 합성곱으로 이웃을 집계하는 구체적인 모델 -> 추천에서는 GCN 계열이 가장 널리 쓰임 (NGCF 2019 -> LightGCN 2020)

LightGCN: NGCF에서 불필요한 연산을 덜어 낸 단순한 모델

- 배경: 추천 그래프의 노드에는 ID 임베딩만 있어서, NGCF의 가중치 행렬과 비선형 활성화 함수가 성능에 도움이 안 되고 오히려 학습을 방해함
- 제거한 것: 특징 변환(레이어마다 곱하던 W), 비선형 활성화 함수(ReLU 등)
- 정규화된 이웃 임베딩의 합 + 레이어 결합만으로 구성 -> NGCF보다 평균 약 16% 높은 성능

활용 사례

- Pinterest: 핀과 보드를 노드로, 약 30억 노드와 180억 엣지 -> 전체 연산이 불가능해 GraphSAGE처럼 이웃 일부만 샘플링
- 그 외: 소셜 네트워크(친구 추천), 화학·생명과학(분자 구조, 약물 상호작용), 금융·보안(이상 거래 탐지), 교통(이동 시간 예측), 코드 자동 완성

장단점

- 장점: 이웃의 이웃까지 반영하는 고차원 협업 신호 포착, 상호작용·소셜 관계·속성을 하나의 그래프로 학습, 이웃이나 속성 노드로 콜드 스타트 완화
- 단점
    - 과평탄화: 레이어가 많아질수록 모든 노드의 임베딩이 비슷해짐
    - 과압축: 기하급수적으로 늘어나는 이웃 정보를 고정 크기 벡터에 담아 먼 정보가 손실됨
    - 계산 복잡도: 수십억 노드 그래프는 자원과 메모리가 많이 필요
    - 순서 반영 어려움: 신발 -> 바지 -> 셔츠와 셔츠 -> 바지 -> 신발이 같은 그래프

### Transformer 기반 추천 시스템
기존 추천 시스템의 순차적 한계

- CF: 구매 순서 무시, 방금 한 구매와 1년 전 구매를 구분 못함, 다음 행동보다 일반적인 선호 예측에 초점
- GNN: A -> B -> C에서 C가 A의 정보를 받으려면 B를 거쳐야 해서 정보가 희석됨, 레이어 수만큼만 볼 수 있어 장기 의존성 포착 어려움
- CF는 누가 무엇을 좋아하는지, GNN은 무엇이 무엇과 연결됐는지를 잘 보지만, 둘 다 어떤 순서로 지금 어떤 의도로 행동하는지는 못 봄 -> Transformer 등장

Transformer: 사용자 행동을 순서(시퀀스)로 접근 -> 다음 행동은 장기 평균 취향보다 최근 행동의 순서와 조합(문맥)에 더 큰 영향을 받음

- 문맥에 따라 달라지는 의미: 수영복 -> 비치타월 -> 선크림은 휴가 준비물(방수팩, 래시가드 추천) / 스킨 -> 로션 -> 선크림은 스킨케어(클렌징폼 추천)
- 장기 의존성 포착: 프린터 -> 양말 -> 라면 -> 폰케이스 -> 생수 -> 샴푸 -> ? -> 거리와 상관없이 프린터에 높은 가중치 -> 토너 추천

Self-Attention: 시퀀스 안의 모든 아이템 쌍의 관련성을 거리와 상관없이 동시에 계산

- Query: 현재 아이템이 관련성을 계산하는 기준 / Key: Query와 내적되어 관련성 점수 결정 / Value: 점수에 따라 가중합되는 실제 정보
- Attention(Q, K, V) = softmax(Q·Kᵀ / √d) · V
- 과정: Q·K로 관련성 점수 (ex. 수영복 2.0, 비치타월 1.5, 양말 0.1) -> softmax로 비율 변환 (57.0%, 34.5%, 8.5%) -> 비율대로 Value 가중합 -> 선크림은 약 92%가 수영복, 비치타월 정보라 휴가 준비물로 해석

SASRec: 단방향 어텐션으로 과거만 보고 다음 아이템을 예측 (GPT와 유사)

- 구조: 아이템 임베딩 + 위치 임베딩 -> Self-Attention(과거만 참조) -> FFN -> 아이템 임베딩과 내적해 다음 아이템 점수 계산
- ex. 노트북 -> 마우스 -> 키보드 -> 다음은 노트북 거치대
- 특징: 데이터가 희소하면 최근 아이템, 풍부하면 긴 과거에 집중 / RNN과 달리 병렬 처리로 빠름
- 문제점: 계산량이 시퀀스 길이의 제곱에 비례, 단방향이라 앞부분 아이템은 쓸 정보가 적음, 장바구니처럼 순서가 엄격하지 않은 경우도 있음

BERT4Rec: 양방향 어텐션으로 앞뒤를 모두 보고 가려진 아이템을 예측 (BERT와 유사)

- 시퀀스 일부를 [MASK]로 가리고 앞뒤 문맥으로 예측 (ex. 노트북 -> 마우스 -> [MASK] -> 키보드 -> 노트북 거치대)
- 가리는 이유: 양방향으로 보면 정답 아이템을 직접 보게 되는 정보 누출이 생김
- 특징: 앞뒤 문맥 모두 사용, 가리는 위치를 바꿔 가며 학습 샘플을 여러 개 만들 수 있음

| 구분 | SASRec | BERT4Rec |
|---|---|---|
| 어텐션 방향 | 단방향 (과거만) | 양방향 (과거 + 미래) |
| 학습 목표 | 다음 아이템 예측 | 가려진 아이템 예측 |
| 학습 효율 | 빠름 | 상대적으로 느림 |

- 원 논문은 BERT4Rec이 더 좋다고 했지만, 후속 연구에서 손실 함수를 같게 맞추자 SASRec이 비슷하거나 더 좋았고 학습도 빨랐음 (ex. Beauty 데이터 NDCG@10 SASRec 0.0327 vs BERT4Rec 0.0187) -> 성능 차이는 어텐션 방향보다 학습 설정의 영향이 큼
- 선택 기준: 상호작용이 많고 다음 행동 예측이 목적 -> SASRec / 시퀀스가 짧음 -> 상황에 따라 다름 / 행동 순서가 느슨함 (장바구니) -> BERT4Rec

### 오늘날의 추천 시스템
LLM의 시대: 추천을 점수 계산이 아닌 언어 문제로 풂

- 과거: (사용자 A, 아이템 123) -> 0.8점 / LLM: "A는 아이템 1, 2, 3을 좋아했으니 다음은 아이템 4" 같은 문장으로 출력
- ID가 아닌 의미 기반 추천, 텍스트·이미지·오디오 같은 멀티모달 활용, 추천 이유를 함께 제시하는 서술형 추천 가능
- 대화형 추천: 자연어 조건 반영 (ex. "비 오는 일요일에 혼자 보기 좋은, 너무 슬프지 않은 영화")
- Zero-shot 추천: 상호작용 없이 텍스트 설명만으로 추천 -> 신상품 첫날에도 추천 가능, 콜드 스타트 해결
- P5: 여러 추천 과제를 "텍스트 프롬프트 -> 텍스트 답변"으로 통일해 하나의 언어 모델로 학습 / 처음부터 학습해 비용이 큼
- TALLRec: 범용 LLM을 "좋아할까? (Yes/No)" 지시문으로 튜닝 / LoRA로 일부만 학습해 적은 샘플로도 높은 성능

인기 편향: 소수의 인기 아이템이 추천을 독점하는 현상

- 강화 과정: 상호작용이 인기 아이템에 몰림 -> 데이터가 많은 아이템의 점수가 높아 더 자주 추천 -> 추천, 노출, 클릭, 학습이 반복되며 격차가 커짐 (Rich-get-richer)
- 사례: 넷플릭스 프라이즈에서 인기 영화 16.5%가 평점의 약 86% 차지 / Spotify 팟캐스트 개인화 추천으로 스트리밍 28.90% 증가, 소비 다양성 11.51% 감소
- 비인기·신규 아이템은 노출 기회를 얻기 어렵고, 사용자는 비슷한 콘텐츠만 반복해서 받음

설명 가능성(XAI): 왜 이걸 추천했는지 설명할 수 있는가

- GNN, Transformer 같은 딥러닝은 강력하지만 블랙박스
- 고객 설득과 비즈니스 담당자에게 근거 제시가 필요
- LLM은 이유를 자연어로 만들 수 있지만, 그 설명이 실제 추천 근거와 일치한다는 보장은 없음

비용을 고려한 도입

- 딥러닝 전면 도입의 비용: GPU 비용, 요청마다 전체 아이템을 계산하는 서버 비용, 트래픽만큼 매일 누적되는 비용
- 2단계 구조 (유튜브, 아마존 등)
    - 1단계 후보군 생성: 수백만 개 -> 수백 개 / 선형 모델 (내용 기반, 협업 필터링, 행렬 분해) -> 빠르고 저렴
    - 2단계 순위 매기기: 수백 개 -> 상위 수십 개 노출 / 딥러닝 (GNN, Transformer, VAE) -> 비선형성, 순서, 관계까지 반영해 정확도를 높임
- 실제 추천 시스템은 단일 모델이 아니라 여러 모델을 단계별로 조합해 비용과 정확도를 함께 확보
