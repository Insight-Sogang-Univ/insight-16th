# 심화세션 2차 : 퍼셉트론 & 딥러닝 기초

## 1. 인공신경망의 시작 : 퍼셉트론
### 1. 인공신경망이란?

- **인공신경망** : 우리 뇌의 신경망을 모방한 모델 <br>
   (1) 입력을 받아 가중치와 편향을 적용 <br>
   (2) 총합한 값을 활성화 함수로 넘긴다. <br>
       - 기준 미달이면 0에 가깝게 보내고 기준 통과면 유의미한 값으로 보낸다. <br>
    (3) output <br>


- **퍼셉트론** : 신경망의 기원이 되는 선형 이진 분류기<br>
    <절차><br>
    (1) 입력값 * 가중치 <br>
    (2) 곱한 값들의 총합 <br>
    (3) 총합이 임계값을 넘으면 1, 넘지 않으면 0<br>
    * 활성화 함수 : 임계값 초과 여부에 따라 출력값을 정하며 종류에 따라 출력값이 달라질 수 있다.
    * 학습 : 원하는 출력을 하도록 가중치를 조정해나가는 작업

- **용어정리** <br>
    (1) 가중치 : 입력값이 출력값에 주는 영향력 결정, 각 입력값마다 존재하며 값이 클수록 중요하다 <br>
    (2) 활성화 함수 : 퍼셉트론은 계단 함수 사용 (임계값 넘으면 1 아니면 0), 이후로 등장한 신경망들은 다양한 종류를 지님 <br>
     <br>
    (3) 임계값 : 활성화 함수에 사용되는 값으로 보통 세타로 표현한다. <br>  <br>
    (4) 편향 : <br>  <br>   <br> -> 계단함수를 구현한 (3)의 식에서 임계값을 넘기면 편향으로 표현된다 즉, 임계값과 편향은 표현의 차이이다.


### 2. 단층 퍼셉트론과 한계
- 단층 퍼셉트론 : 입력층 + 출력층

- 논리 게이트 구현 : 디지털 회로에서 기본적인 논리 연산을 수행할 때 사용되는 구성 요소/ n개의 입력 -> 하나의 출력값 <br>
     * 종류 : AND OR NAND NOR XOR 등 다양하며 단층 퍼셉트론으로 구현 가능하다. 또한 이를 조합해서 복잡한 논리 회로도 구성할 수 있다.

- 대표적인 논리 게이트  <br>
     * AND : 입력값이 모두 1인 경우에만 출력값이 1이 나오는 구조이다. 
     * OR :  둘 다 0일 떄만 0
     * NAND : 둘 다 1일 떄만 0 = AND의 부호를 뒤집은 결과
     * XOR : 서로 다를 때만 1 = 가능한 조합이 없다.

- **단층 퍼셉트론의 한계** <BR>
     * XOR 게이트를 구현할 수 없다. 
     * 퍼셉트론은 입력값과 가중치의 선형 결합으로 이루어진 구조이기 때문에 직선 하나로 영역을 나눌 수 있는 *선형 분류 문제*에 대해서만 구현이 가능하다. -> 그러나 xor 게이트는 비선형 영역으로만 구현할 수 있다. <BR>

### 3. 다층 퍼셉트론
- XOR 게이트는 기존 다른 종류의 게이트들을 조합하면 만들 수 있다 = 퍼셉트론의 층을 더 쌓으면 된다.
- 다층 퍼셉트론은 **은닉층에서 비선형 활성화 함수를 적용해서 새로운 특징을 만들고 다음 층에서 그 특징들을 조합하기 때문이다.**<BR><BR>
- 더 복잡한 문제를 풀기 위해서는 *더 많은 은닉층*을 추가하면 된다. -> 두 개 이상 은닉층을 갖는 MLP가 딥러닝의 시초 모델로 발전했다.
* 심층 신경망 : 은닉층이 2개 이상인 신경망
* 딥러닝 : 심층 신경망을 학습 (가중치를 스스로 찾아보도록)


## 2. 딥러닝 모델은 어떻게 학습할까?
### 1. 순전파와 역전파
- 학습 알고리즘의 기본 구조
     * 1. 순전파를 통해 입력값으로부터 예측
     * 2. 오차 계산으로 예측 정확도 측정
     * 3. 역전파를 통해 출력층 -> 입력층 으로 오차 전달
     * 4. 가중치 업데이트
     * 5. 1-4번을 반복...
      * 이렇게 반복하다보면 오차값은 0으로 수렴하게 된다.
      * 순전파와 역전파의 과정은 연쇄법칙으로 일어나게 된다. -> 이 과정에서 오차 줄이는 것이 필요하다.

- **연쇄 법칙** : 합성함수의 미분은 각 함수의 미분의 곱으로 나타낼 수 있다는 법칙. 딥러닝 모델 학습 시키기 위해 오차를 줄여야하는데 이때 역전파 과정에서 가중치가 *최종 손실에 얼마나 기여했는지 계산하는 데에 사용*
     * 기울기 = 손실 함수(예측값과 실제값의 차이)를 줄이기 위해 가중치를 어떻게 바꿀지를 알려준다.

- **기울기 소실 문제** : 기울기를 계속 곱했는데 결국 기울기가 0이 되는 문제 -> 앞쪽 층의 가중치가 거의 업데이트 되지 않아 학습이 제대로 이루어지지 않는 문제가 생긴다.

- 전체적 흐름 정리
    * 순전파 = 예측하기 (입력->가중치1->가중치2->예측값)
    * 손실 계산 = 얼마나 틀렸는가 (예측값과 실제 정답의 차이)
    * 역전파 = 책임 추적하기 (역전파 사용 / 각 가중치가 손실에 얼마나 기여?)
    * 가중치 업데이트 = 수정하기 (계산값을 바탕으로 가중치 수정을 통해 모델의 정확도를 높인다.)

### 2. 손실 함수
- **손실 함수** : 모델의 예측이 얼마나 틀렸는지 수치로 표현하는 방법
       * 종류 : 평균제곱오차(회귀), 크로스 엔트로피 오차(분류)
- 손실 함수의 종류 <BR>
    * 1. 평균 제곱 오차 : 예측 값이 연속적이고 예측과 실제의 차이를 측정하는  회귀 문제에 적합하다. <BR>
    * 2. 이진 크로스 엔트로피 : 이진 분류 문제 (로지스틱 회귀)에서 사용하는 손실 함수이다.<BR>
    * 3. 크로스 엔트로피 오차 : 다중 클래스 분류 문제에서 예측 확률이 실제 클래스와 얼마나 일치하는가 평가하는 데 적합하다. (2번과 유사하나 다중 분류에서 사용된다는 차이) <BR>

### 3. 활성화 함수
- **활성화 함수** 
    * 왜 활성화 함수를 거칠까? <BR>: 1) 비선형성의 필요 2) 미분이 필요 (역전파는 미분값을 연쇄적으로 곱해야하기에 활성화 함수는 미분이 가능해야한다)
    * 가중치 : 입력값이 결과에 얼마나 영향을 미치는가
    * 편향 : 활성화 함수의 출력값을 조정, 기준점

- **계단 함수** <BR>
 <BR>
   * 입력값과 가중치의 합을 모두 더하고 그 합이 임계값을 넘으면 1 아니면 0
   * 임계값은 뉴런이 얼마나 쉽게 활성화 되는지를 제어한다.
   * 미분값이 0이라 역전파가 불가능하다.

- **시그모이드 함수** <BR>
 <BR>
   * 입력 신호를 0 1 사이에서 연속적인 값으로 변환하는 비선형 함수
   * 계단 함수와 차이는 출력값의 형태이다 <BR> (연속적으로 변화 -> 얼마나 확실한지 표현할 수 있으며 미분도 가능하다.)
   * 한계 : 미분값이 최대 0.25로 기울기 소실이 발생한다.

- **ReLU 함수** <BR> 
 <BR>
   * 딥러닝에서 가장 많이 사용되는 이유 <BR> 1. 계산이 간단하고 효율적이다 <BR> 2. 비선형성을 제공한다 (음수는 0, 양수는 그대로 -> 은닉층에 비선형성을 추가한다.)<BR> 3. 기울기 소실 완화 : 양수 구간에서 미분값이 항상 1이기 때문에 아무리 곱해도 (양수에서는) 기울기 소실이 되지 않는다. <BR> * 음수에서는 0이 되지만 실무에서는 양수가 더 중요하기에 자주 사용

- **SOFTMAX 함수** <BR> <BR>
   * 각 클래스에 대한 출력값을 확률로 변환하는 함수이다. 
   * 출력값이 0~1, 전체 합이 1
   * SOFTMAX + 크로스 엔트로피 : 예측 확률과 정답 확률의 차이를 바로 계산할 수 있어 학습 효율이 높다.

### 4. 경사 하강법
- 원리 : 목표는 *손실을 가장 작게 만드는 w를 찾는 것* <br> <br> <br>
    * 우리는 이 곡선의 바닥을 찾아야한다. -> 어느 쪽에서 출발하든 자동으로 바닥에 모이게 된다.

- 학습률 : 그럼 "얼마나" 움직여야할까? <br> <br> 미분값에 a를 곱해서 이동량을 조절한다. 
    * $ W \leftarrow W - \boldsymbol{\alpha} \frac{\partial E}{\partial W} $
    * 너무 작으면 너무 오래걸린다
    * 너무 크면 수렴을 못하고 발산한다. (손실이 계속 커진다)
    * 적당해야한다. -> 0.001부터 시작하자!

- 배치 / 에포크 / 이터레이션 
    * **배치** :  $ \frac{\partial E}{\partial W} $ 를 구하려면 데이터를 넣어야하는데 **나눠서 넣어야 한다.** 
       * 매개변수를 조정하기 위해 사용하는 데이터의 묶음이며 하이퍼파라미터이다.
       * 너무 크기가 크면 학습이 느려지고 메모리가 부족해진다.
       * 너무 작으면 학습이 불안정해질 수 있다. (자주 업데이트)
    * **에포크** : 모든 데이터셋을 학습하는 횟수
    * **이터레이션 또는 스텝** : 한 번의 에포크를 **끝내기 위해 필요한 배치 수**
    * 배치 크기 X 이터레이션

- 종류
    * 1. 배치 경사하강법 : 오차를 구할 때 전체 데이터를 한 묶음으로 묶는다.
        * 업데이트 횟수는 적지만 메모리가 많이 필요하다. 수렴이 안정적이다.
    * 2. 확률적 경사하강법 : 랜덤으로 선택된 하나의 데이터에 대해서만 계산 (기존 모든 데이터 계산으로 인한 시간 문제 해결)
        * 매개변수의 변경폭이 불안정하고 정확도가 낮을 수가 있다.
    * 3. 미니 배치 경사하강법 : 배치 크기를 적절히 지정한다.
        * 전체 데이터보다 빠르고 배치 크기 1보다 안정적이다. -> 가장 널리 사용

### 5. 옵티마이저
- 경사하강법만으로 부족하다 <br> -> 딥러닝의 손실함수는 모양을 예측할 수 없다. 진짜 바닥이 아닌 곳에서도 기울기가 0일 수 있으므로 경사하강법에서는 목적지가 아닌 곳에서 멈축 수 있다. <br> -> **옵티마이저로 경사하강법을 더 똑똑하게 굴리자**

- 1. 모멘텀 : 이전에 가던 방향을 기억하도록 관성을 부여 <-> 기존 경사하강법 <br>   <br> * local minima를 관성의 힘으로 탈출 가능 & 기울기 작은 구간도 빠르게 통과

- 2. 알엠에스프롭 : 각 방향마다 학습률 따로 조절 <-> 기존 경사하강법 (무조건 현재 기울기만)

- 3. 아담 : 1+2, 방향과 스텝 사이즈를 적절하게 합친 방법 , 대부분 또는 잘 모르겠을 때 아담으로 시작한다.

### 3. 활성화 함수
- **활성화 함수** 
    * 왜 활성화 함수를 거칠까? <BR>: 1) 비선형성의 필요 2) 미분이 필요 (역전파는 미분값을 연쇄적으로 곱해야하기에 활성화 함수는 미분이 가능해야한다)
    * 가중치 : 입력값이 결과에 얼마나 영향을 미치는가
    * 편향 : 활성화 함수의 출력값을 조정, 기준점

- **계단 함수** <BR>
 <BR>
   * 입력값과 가중치의 합을 모두 더하고 그 합이 임계값을 넘으면 1 아니면 0
   * 임계값은 뉴런이 얼마나 쉽게 활성화 되는지를 제어한다.
   * 미분값이 0이라 역전파가 불가능하다.

- **시그모이드 함수** <BR>
 <BR>
   * 입력 신호를 0 1 사이에서 연속적인 값으로 변환하는 비선형 함수
   * 계단 함수와 차이는 출력값의 형태이다 <BR> (연속적으로 변화 -> 얼마나 확실한지 표현할 수 있으며 미분도 가능하다.)
   * 한계 : 미분값이 최대 0.25로 기울기 소실이 발생한다.

- **ReLU 함수** <BR> 
 <BR>
   * 딥러닝에서 가장 많이 사용되는 이유 <BR> 1. 계산이 간단하고 효율적이다 <BR> 2. 비선형성을 제공한다 (음수는 0, 양수는 그대로 -> 은닉층에 비선형성을 추가한다.)<BR> 3. 기울기 소실 완화 : 양수 구간에서 미분값이 항상 1이기 때문에 아무리 곱해도 (양수에서는) 기울기 소실이 되지 않는다. <BR> * 음수에서는 0이 되지만 실무에서는 양수가 더 중요하기에 자주 사용

- **SOFTMAX 함수** <BR> <BR>
   * 각 클래스에 대한 출력값을 확률로 변환하는 함수이다. 
   * 출력값이 0~1, 전체 합이 1
   * SOFTMAX + 크로스 엔트로피 : 예측 확률과 정답 확률의 차이를 바로 계산할 수 있어 학습 효율이 높다.

### 4. 경사 하강법
- 원리 : 목표는 *손실을 가장 작게 만드는 w를 찾는 것* <br> <br> <br>
    * 우리는 이 곡선의 바닥을 찾아야한다. -> 어느 쪽에서 출발하든 자동으로 바닥에 모이게 된다.

- 학습률 : 그럼 "얼마나" 움직여야할까? <br> <br> 미분값에 a를 곱해서 이동량을 조절한다. 
    * $ W \leftarrow W - \boldsymbol{\alpha} \frac{\partial E}{\partial W} $
    * 너무 작으면 너무 오래걸린다
    * 너무 크면 수렴을 못하고 발산한다. (손실이 계속 커진다)
    * 적당해야한다. -> 0.001부터 시작하자!

- 배치 / 에포크 / 이터레이션 
    * **배치** :  $ \frac{\partial E}{\partial W} $ 를 구하려면 데이터를 넣어야하는데 **나눠서 넣어야 한다.** 
       * 매개변수를 조정하기 위해 사용하는 데이터의 묶음이며 하이퍼파라미터이다.
       * 너무 크기가 크면 학습이 느려지고 메모리가 부족해진다.
       * 너무 작으면 학습이 불안정해질 수 있다. (자주 업데이트)
    * **에포크** : 모든 데이터셋을 학습하는 횟수
    * **이터레이션 또는 스텝** : 한 번의 에포크를 **끝내기 위해 필요한 배치 수**
    * 배치 크기 X 이터레이션

- 종류
    * 1. 배치 경사하강법 : 오차를 구할 때 전체 데이터를 한 묶음으로 묶는다.
        * 업데이트 횟수는 적지만 메모리가 많이 필요하다. 수렴이 안정적이다.
    * 2. 확률적 경사하강법 : 랜덤으로 선택된 하나의 데이터에 대해서만 계산 (기존 모든 데이터 계산으로 인한 시간 문제 해결)
        * 매개변수의 변경폭이 불안정하고 정확도가 낮을 수가 있다.
    * 3. 미니 배치 경사하강법 : 배치 크기를 적절히 지정한다.
        * 전체 데이터보다 빠르고 배치 크기 1보다 안정적이다. -> 가장 널리 사용

### 5. 옵티마이저
- 경사하강법만으로 부족하다 <br> -> 딥러닝의 손실함수는 모양을 예측할 수 없다. 진짜 바닥이 아닌 곳에서도 기울기가 0일 수 있으므로 경사하강법에서는 목적지가 아닌 곳에서 멈축 수 있다. <br> -> **옵티마이저로 경사하강법을 더 똑똑하게 굴리자**

- 1. 모멘텀 : 이전에 가던 방향을 기억하도록 관성을 부여 <-> 기존 경사하강법 <br>   <br> * local minima를 관성의 힘으로 탈출 가능 & 기울기 작은 구간도 빠르게 통과

- 2. 알엠에스프롭 : 각 방향마다 학습률 따로 조절 <-> 기존 경사하강법 (무조건 현재 기울기만)

- 3. 아담 : 1+2, 방향과 스텝 사이즈를 적절하게 합친 방법 , 대부분 또는 잘 모르겠을 때 아담으로 시작한다.
