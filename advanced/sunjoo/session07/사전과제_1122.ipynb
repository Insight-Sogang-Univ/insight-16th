# 시계열

# 1. 시계열 데이터란?
## 1) 시계열 데이터의 정의
### 시계열 데이터란?
 <br>
- 시간의 흐름에 따라 일정한 간격, 순서대로 관찰된 데이터
- 주가, 기온, 매출
### 시계열 데이터의 조건
- 시간 순서의 의미가 있다.
- 관측 간격이 일정하다. (주기가 존재한다.)
- 지금까지의 데이터 : 횡단면 데이터 (한 시점에 여러 대상 관찰) <-> 시계열: 한 대상을 여러 시점에 걸쳐 관찰

### 단변량 시계열과 다변량 시계열
- **단변량 시계열** : 시간에 따라 **하나의 변수만** 관측한 데이터 / 자기 자신의 과거 값만 보고 미래 예측
     - 삼전 종가, 서울의 평균 온도
- **다변량 시계열** : 시간에 따라 **여러 변수를 동시에** 관측한 데이터 / 자기 자신의 과거만이 아니라 *다른 변수 영향도 고려*
     - 아메리카노 판매량 <- 기온, 요일 등
     - 전력 사용 <- 기온, 습도
    
### 시계열의 분석과 목적
- 시계열 분석 : 시간 순서대로 쌓인 데이터에서 패턴을 찾고 미래 값을 예측 및 분석
- 목표 : 미래 데이터 예측 / 과거의 기록을 보고 다음 시점 값 추정
     - 가정 : 과거의 패턴이 미래에도 어느 정도 반영이 될 것이다.
     - 먼 미래일수록 예측이 어렵다.
- 분석 과정 : 시계열 이해 -> 전처리 -> 정상성 판단 -> 모델링과 예측
- 활용방안 : 거의 모든 산업!!
     - 금융
     - 유통 커머스
     - 에너지
     - 기상
     - IT 서비스

## 2) 시계열 데이터의 특징
### 1. 시간 의존성
- 데이터가 시간에 따라 변하고, 현재 시점의 값이 과거 시점 값에 영향을 받는 성질
     - 과거 값이 현재 값에 영향을 준다
     - 보통 가까운 과거일수록 영향이 크다
     - 순서를 섞으면 안된다.
- 변화의 2가지 모양
### 2. 계절성과 추세
- 계절성 : 일정한 주기로 비슷한 패턴의 반복 = 언제마다 반복되나?
- 추세 : 시간의 흐름에 따라 데이터가 장기적으로 증가하거나 감소하는 방향성 = 어느 방향으로 가나?
     - 전반적 경향성을 파악할 수 있다.
- 둘은 동시에 나타나는 경향이 있다.

### 3. 자기상관성
- '자기 자신'끼리 상관관계를 갖는 특성, 한 시계열 내에서 현재 값과 과거 값이 얼마나 비슷하게 움직이는지를 나타나는 정도 = **과거 값으로 지금 값을 짐작할 수 있는 정도**
- 일반 상관계수와 같이 : -1~1 (1에 가까우면 비례, -1에 가까우면 반비례, 0에 가까우면 과거와 지금이 관계 없음)
- 시차를 바꿔가면서 계산하면 계절성(반복주기)을 찾을 수 있음
     - 시차: 현재 값을 몇 시점 전의 값과 비교하는가
- 문제점 : 회귀 분석이 불가능하다.
     - 회귀분석 : 데이터가 *독립적*이라는 가정 위에서 변수의 영향을 검정하고, 관계를 찾고, 미래를 예측한다. 하지만 자기 상관이 있으면 이 세가지가 모두 흔들리기 때문에 그대로 쓸 수가 없다.
     - 1. 오차항 독립성 가정을 위반 -> 통계 결과가 위반된다
     - 2. 허위 상관관계 문제 -> 시계열 끼리 높은 상관관계는 추세 때문이 아닐지 
     - 3. 예측의 어려움과 모델의 복잡성의 증가
- 결론 : 시계열 데이터는 시계열 전용 모델을 사용하거나, 사전처리 (차분이나 추세 제거 등)를 진행하고 분석해야한다.

## 3) 시계열 데이터의 분석 요인

### 1. 추세요인
- 시계열이 시간 전반에 걸쳐서 증가하거나 감소하는 경향성을 나타내는 요인
- 큰 흐름만 본다
- 주로 인구 변화, 물가 상승 등 *오랜 시간 걸쳐 작용하는 원인 때문에 발생*

### 2. 계절요인
- 1년 이내의 일정한 주기로 같은 패턴이 반복되는 요인
- 주로 달력, 날씨 사회적 관습처럼 규칙적으로 돌아오는 원인 때문에 발생

### 3. 순환 요인
- 계절성보다 더 길고 불규칙한 주기로 반복되는 변동
- 1년 이상 년 단위 이상의 움직임이기에 긴 기간의 데이터를 봐야 파악할 수 있음
- 대표적 원인은 경기 변동.
- <-> 계절성 : 주기가 고정이며 1년 이내
- 주의점 : 찾아내기가 어려우며 추세와 구분하기 어려움

### 4. 불규칙 요인
- 시계열에서 추세,계절,순환요인 모두 제거하고 남은 나머지 움직임
- 우연히 예측할 수 없이 생기는 변동 - 잔차, 노이즈

=> 우리가 보는 시계열은 이 4가지 요인이 모두 섞인 결과


# 2. 시계열 데이터 전처리
- 필요성 : 시계열은 결측치, 이상치, 노이즈 때문에 잘못된 패턴을 학습하면 주변에 연결된 데이터 모두 망가지게 된다. 따라서 전처리 필요!
- 과정
     - 1. 데이터 확인
     - 2. 결측치 탐지 및 처리
     - 3. 이상치 탐지 및 처리
     - 4. 평활화
     - 5. 요소분해
     - 6. 정상성 판단 및 확보

## 1) 결측치 이상치 탐지 및 처리
### 시계열 데이터의 결측치
- 행을 지우면 관측 간격이 깨져 시점끼리 비교가 어려워지며 전체 평균으로 채우면 시간 흐름이 무시됨
- 앞 뒤 흐름을 고려해야 추세나 계절성이 왜곡되지 않음
- 결측치 처리 방법 : 삭제, 단순 대치, 보간, 이동 평균 대치, 머신러닝 기법
     - 보간법 : 주어진 데이터 포인트 값으로 두 관측값 사이의 비어있는 값을 추정
     - 과정 : 결측 위치 확인 -> 앞 뒤 관측값 확인 -> 두 점 연결하는 규칙 정하기 (직선? 곡선?) -> 값 추정

### 시계열 데이터의 이상치
- 언제 이상치가 나타났는지가 중요하다. 같은 값도 시점에 따라 정상일수도, 아닐 수도 있다. -> **추세와 계절성** 을 고려해서 이상치를 판단
- 이상치 탐지 방법 : 
     - 1. IQR 범위에서 너무 떨어지면 이상치로 판단. (주기성이 없거나 비교적 안정적 구간의 값들에서 이상치 탐지 할 때)<br>
     - <br>
          - 장: 계산이 간단하고 직관적이고 데이터가 정규분포를 따르지 않아도 되지만 단: 추세나 계절성이 강한 시계열에는 부적합하며 분포가 비정상적이면 오탐률이 높다.
    - 2. Z-score (계절성이나 추세가 약하거나 값의 변동폭이 일정한 시계열에서 효과적이다)<br>
     -  <br>
          - 장 : 계산이 간단하고 해석이 쉬우나 단: 데이터가 정규분포를 따른다고 가정해야하며 평균과 표준편차가 이상치 자체에 영향을 받는다.
     - 3. STL 분해 (추세, 계절, 잔차 로 분해하는 방법)<br>
     - <br>
          - 장: 추세와 계절성을 제거하고 진짜 이상한 값을 잡아낼 수 있으며 비정상 시계열에도 유연하게 적용이 가능하나 단: 매우 긴 시계열에는 파라미터 튜닝이 필요하며 계산 비용도 큼.

- 이상치 탐지 모델
     1. 분류 및 회귀 트리 (CART) : 데이터를 조건에 따라 계속 나눠가면서 예측하는 의사결정나무 모델 (예측 오차가 일정 임계점 이상인 지점= 이상치 판단) 
        - 시계열 적용 => 과거 구간의 값을 입력해서 다음 시점 값 예측, 예측 오차 기반으로 이상치 판단
        - 장점 : 비선형 관계 포착, 변수 간 상호작용이 있는 시계열에 적용 가능, 
        -  단점 : 노이즈가 많으면 과적합되기 쉽고, 트리 깊이에 따라 민감도가 달라짐

     2. 클러스터링 기반 이상 탐지 : 군집화 이후 어느 군집에도 잘 속하지 않는 데이터를 이상치로 판단 (kmeans, DBSCAN)
        - 시계열 적용 => 시계열 특징을 각 시점 벡터로 표현해서 클러스터링 수행, 중심으로부터의 거리나 군집 밀도를 기준으로 이상치를 판단.
        - 장점 : 비지도 학습이 가능하며 집단 이상치도 잡을 수 있다
        - 단점 : 군집 수나 거리 기주너럼 *하이퍼파라미터 설정*에 따라 결과가 크게 달라진다. 시계열의 순서 정보는 직접적으로 반영되지 않는다.

     3. 오토인코더 : 작게 압축했다가 원래대로 복원. 이상치가 들어오면 복원 오차가 커져서 이 오차가 기준보다 크면 이상치로 판단<br>
     - <br>
        - 시계열 적용 => LSTM 오토인코더를 사용하면 시간적 의존성까지 학습 가능하다. 또한 일정 길이로 잘라 정상 구간을 학습시키며 테스트 구간의 복원 오차가 큰 곳을 이상으로 탐지한다.
        - 장 : 복잡한 비선형과 다변량 시계열도 다룰 수 있다
        - 단: 많은 데이터와 계산이 필요하다. 모델 해석이 어려우며 파라미터 튜닝이 까다롭다.

- 이상치 처리 방법
    - 삭제 : 이상치 제거 (nan으로 바꾼 뒤 채워주는 경우가 더 많음)
    - 단순 대치 : 평균, 중앙값으로
    - 보간 : 이상치 -> 결측치로 바꾸고 앞뒤 값으로 보간해서 채우는 방법


## 2) 평활화
### 평활화의 개념
- 노이즈를 줄여서 데이터를 부드럽게 만드는 기술 -> 데이터의 큰 흐름이 드러남

### 이동평균
- 일정한 개수의 데이터를 묶어서 옆으로 한 칸씩 옮겨가며 평균을 계속 계산
- N은 묶음의 크기를 의미 : N이 작으면 원래 데이터와 비슷하게 움직이며, 크면 그래프가 매끄러워짐
- 이동 평균 사용의 목적 
    1. 노이즈 제거 및 추세 파악
    2. 계절성 제거
    3. 결측치 및 이상치 대치
    4. 예측의 기초 자료
- 이동 평균의 유형
    - 단순 이동 평균 : 모든 값에 같은 가중치
    - 가중 이동 평균 : 최근 값일수록 더 큰 가중치
    - 지수 이동 평균 : 최근 값에 큰 가중치를 두되 과거로 갈수록 지수적으로 줄어드는 방법
- 핵심 : 평활화는 노이즈를 줄여서 큰 흐름을 보는 기술이다!


## 3) 요소분해 / 시계열 분해

### 요소 분해
- 섞여있는 추세,계절,순환,불규칙 요인을 각각의 요인으로 나누는 과정
- 왜 할까?
    - 각각 패턴을 명확하게 이해 가능
    - 변동 원인 분석 가능
    - 더 정교한 분석 가능

### 분해 모형
1. 가법 모형 : 추세, 계절, 잔차를 더한 값
    - 각 요인들이 독립적으로 더해지고 모든 요소의 영향이 일정
2. 승법 모형 : 추세 계절 잔차를 곱한 값
    - 추세가 커질 수록 나머지들도 비례해서 커짐

### 분해 방법
1. MA 기반 분해 : 이동 평균으로 추세를 뽑아내고 분리한다.
    - 작동방식 : 추세 추출 -> 추세 제거 -> 계절성 추정 -> 잔차 산출
2. STL LOESS 기반 분해 : 이동 평균 대신 LOESS를 사용해서 추세랑 계절을 추정
    - 원리 : inner loop + outer loop 의 이중 반복 구성
    - 이너루프 : 추세와 계절을 번갈아 추정
    - 아우터루프 : 이상치의 영향 줄이기

### 요소 분해 정리
- 추세+계절+ 잔차로 나누는 과정
- 결합 방식 => 가법 / 승법
- 계산 방법 => 이동평균, 단순 / STL




# 3. 정상성 판단
## 1) 정상성이란?
- 개념 : 통계적 특성이 시간에 따라 변하지 않는 성질
- 어느 시점에서 잘라 보더라도 비슷한 모습이라면 정상성을 갖는다.

- 정상 시계열과 비정상 시계열의 구분

|구분|정상시계열|비정상시계열|
|---|---|---|
|평균| 일정| 시간에 따라 변함|
|분산| 일정| 시간에 따라 커지거나 작아짐|
|패턴 | 일정한 평균 주위를 오르내리며 다시 돌아옴| 한 방향으로 가거나 주기적으로 반복|

- 정상성이 왜 중요할까 -> 예측을 가능하기하기 위한 가정

## 2) 강한 정상성과 약한 정상성
### 강한 정상성
- 확률분포 자체가 시간에 따라 전혀 변하지 않음 : 분포가 완전히 같아야함!!
- 특징 : 평균 분산 왜도 첨도 등 모든 분포의 특성이 시간에 따라 변하지 않고 일정
- 한계 : 현실에서는 잡음 및 외부 요인 때문에 이런 상태는 거의 찾아볼 수 없음

### 약한 정상성
- 평균, 분산, 자기공분산만 시간에 따라 변하지 않으면 되는 성질
- 약한 정상성이 되기 위한 조건
    1. 평균 일정하다
    2. 분산 일정하다
    3. 자기공분산이 *시차*에만 의존한다.
- 특징 : 현실 데이터에서도 적용이 가능하다.

## 3) 정상성 판단하기
### 시각적으로 판단
- 정상 시계열
    - 수평선 중심으로 위아래 오르내림
    - 오르내리는 폭이 시간에 상관없이 비슷함
    - 평균에서 벗어나도 다시 평균으로 돌아오는 경향이 있음
    - 특정 주기로 반복되는 패턴이 없음
- 비정상 시계열
    - 추세 : 평균이 계속 변함
    - 계절성: 일정한 주기로 같은 모양 반복 
    - 분산 변함

### 통계적 검정 
- 사람마다 주관적 판단으로 달라질 수 있기에 객관적으로 확인 필요 
- **단위근**: 시계열을 표현하는 식의 해가 1이 되는 경우 = 과거 충격이 누적됨 => 단위근이 있다면 비정상계열이다
- 단위근 검정 방법 
    - ADF : 비정상을 가정하고 증거가 되면 정상 판결
    - KPSS : 정상을 가정하고 증거가 되면 비정상 판결
    - 두 모델 다 정상-> 정상 시계열
    - 두 모델 다 비정상 -> 추세 정상 시계열일 가능성 -> 추세 제거 뒤 검정

## 4) 정상성 확보를 위한 여러가지 방법들
### 1. 추세 제거
- 시간에 따라 평균이 변하는 경우에 적용해서 흐름을 걷어낸다.
- 차분을 통해 확보 (차분: 전날 대비 얼마나 변했나를 보는 방법)
    - 추세와 단위근이 사라지게 됨
    - 1차 차분 = 기본 / 2차 차분 = 1차 차분 결과를 한 번 더 차분 / 역차분 = 차분값을 원래 값으로

### 2. 계절성 제거
- 일정한 주기마다 같은 모양이 반복되는 흐름을 걷어낸다
- 오늘 값에서 한 주기 전 값을 빼는 방법

### 3. 분산 안정화
- 폭이 시간이 지날수록 커지거나 작어지는 경우 폭을 따로 맞춰 주어야 한다.
- 로그변환 : 데이터에 로그 값을 취해서 값의 크기를 줄이는 방법
    - 한계 : 0이나 음수에는 사용할 수 없으며 모든 분산 변화를 해결하지는 못한다


# 4. 단변량 시계열 분석
- 변수 *하나*만의 기록으로 자신의 과거 분석하고 예측

## 1) 안정적 시계열 분석
- = 정상 시계열
### 자기 회귀 모델
- 오늘 값을 자기 자신의 과거 값으로 설명하는 모델 (자기상관성 활용)
- 과거 p개 시점 값 사용 = AR(p)
- '손님이 어제 손님 수를 보고 온다.' => 어제 값이라는 그릇에 담겨서 충격이 넘어온다

### 이동 평균 모델
- 오늘 값을 과거의 충격으로 설명하는 모델
- 과거 q개 시점의 충격 사용 = MA(q)
- '손님이 어제 방송을 보고 온다.' => 충격 자체가 직접 넘어온다

### AR,MA 모델의 전제 조건
1. 정상성 : 충격을 받아도 시간이 지나면 평소로 돌아온다.
     - 정상성 조건 
        - AR : 계수 조건이 필요하다 (p=1일 때, 계수의 절댓값< 1 / 나머지 일 때, 특정 방정식의 모든 근의 절댓값이 1보다 커야함)
        - MA : 항상 만족한다. (충격이 q일 뒤에 끝나기 때문)
2. 가역성 : 지난 충격을 거꾸로 계산해 낼 수 있는 성질이다.
    - 가역성 조건
        - AR : 항상 만족한다. (필요한 숫자가 전부 관측이 가능한 값들이다)
        - MA : 계수 조건이 필요하다  (q=1 일 때 계수의 절댓값 < 1, / 나머지 일 때 특정 방정식의 모든 근의 절댓값이 1보다 커야함)

### ARMA 모형
= AR + MA 로 한 식에 합친 모형이다. *과거 값과 과거 충격* 양쪽의 영향을 받는다.
- 특징 : AR부분이 정상성 조건을, MA 부분이 가역성 조건을 각각 만족해야한다.

### 차수 결정 방법 : 자기 상관 함수
- 자기 상관 계수(ACF) : 오늘 값과 k 시점 전 값이 얼마나 닮았는지를 상관계수로 잰 것이다. 시차 k를 바꿔가며 구한 값을 막대로 그린다.
- 특징
    - -1~1 : 시차 0에서는 항상 1
    - 중간 날짜를 거쳐 온 영향까지 합해서 잰다.
    - 그래프 점선 안쪽이라면 보통 0으로 본다.

- 부분 자기 상관 계수 (PACF) : 중간 시점들의 영향을 제외하고 오늘 값과 k 시점 전 값이 직접 얼마나 닮았는지를 잰 것이다.
- 특징 : 시차 1에서는 중간 날짜가 없다. 따라서 ACF와 값이 같다.

## 2) 불안정 시계열 분석
- 정상성이 없는 시계열이다.(돌아갈 평소가 없는)
    - 평소가 없는 이유
        - 추세 : 계속 오르내린다 -> 처분으로 처방
        - 계절성 : 달마다 평소가 다르다 -> 계절 차분으로 처방한다.
- 핵심 : **모형을 고치는 것이 아니라 데이터를 모형이 다를 수 있는 *숫자* 로 바꿔주는 것이다**
- 과정 :
    1. 비정상 시계열 데이터
    2. 정상성 확보 (차분, 계절성 차분)
    3. 정상 시계열
    4. ARMA 를 사용해서 안정적 시계열 분석

### ARIMA
= ARMA + **차분** : 판매량을 다루던 자리에 '증감'을 넣는다.
- ARIMA (AR 차수, 차분 횟수, MA 차수)
- 모형의 흐름 
    1. 차분해서 증감을 만든다
    2. 증감을 ARMA로 예측한다.
    3. 예측한 증감을 더해서 판매량으로 되돌린다.
- 모델의 특수한 경우 : 백색잡음, 자기회귀, 이동평균, 랜덤위크, 포류를 포함하는 랜덤워크(랜덤워크에서 상수만 빼지 않고 남겨둔 것 -> 표류(매일변하는정도)만 남는다)

### SARIMA
- ARIMA 에 계절 차분과, 계절 AR, 계절 MA를 더한 모형
- SARIMA : ARIMA(지난달대비)에 작년 같은 달 대비 증감을 더한 것이다. 계절마다 다른 수준을 없앤다.
    - 계절 차분 -> 증감 예측 -> 판매량으로 되돌린다
- 계절성 데이터에서의 ACF /PACF
    - ACF가 오르내리면서 주기마다 치솟는다. / PACF도 막대가 시차까지 이어져서 끊기는 자리가 없다. 
    - 따라서 계절 차분을 먼저 수행하는 것이 중요!!
- 불안정 시계열에서 하이퍼파라미터의 결정 
    - d,D 결정 방법 : ADF로 정상성 확보 여부 판단 후 결정
    - p,q 결정 방법 : AC,PACF로 범위 측정하고 그리드 서치 후 최종 선택을 진행한다
    





# 5. 다변량 시계열 분석
- 변수 여러 개를 함께 놓고 각 변수를 자기 자신의 과거와 다른 변수의 과거로 예측

## 1) VAR
- AR 을 변수 여러 개로 넓힌 것
- 예시 ) AR은 '판매량'만 보지만 VAR은 판매량과 광고비를 함께 본다. (광고-> 판매량 증가 / 판매량 하락 -> 광고 늘림)
- 사용할 경우 : 변수가 서로 영향을 주고 받는 경우, 모든 변수가 정상이거나 차분하면 정상이 될 수 있는 경우, 변수가 적고 데이터가 충분한 경우
- 부적합한 경우 : 영향이 한 방향인 경우, 데이터가 적은 경우, 비율이 고정되지 않은 관계가 강한 경우

### 분석 방법
- 충격 반응 함수 : 한 변수에 충격 주었을 때 다른 변수가 어떻게 반응하는지 확인
- 예측 오차 분산 분해 : 빗나간 정도를 어느 쪽 충격 탓인지 나눠서 비율로 보여준다.

# 6. 딥러닝을 활용한 시계열 분석
- 전통 모델의 한계
    1. 데이터 많으면 속도가 느려짐
    2. 크기가 고정된 데이터만 가능
    3. 비선형 패턴 포착의 어려움
    - => 딥러닝으로 해결해보자

- 1. 순환 신경망 (RNN)
    - 적합한 경우 : 순차적 패턴이 중요한 시계열 데이터, 과거 정보가 예측에 영향을 줄 때, 시계열 길이와 목적에 따라 모델을 선택 가능할 때
    - 부적합한 경우 : 매우 긴 시계열에서는 학습 어려움, 병렬 처리 효율이 낮음, 학습 속도가 상대적으로 느린 편
- 2. 트랜스포머계열
    1) informer, 장기 예측 특화 -> 계산 복잡도를 낮춤
    2) autoformer, 주기성 포착 특화 -> 요소 분해를 딥러닝 구조에 결합
    3) TFT , 실무 활용 1위! -> 과거 데이터와 외부변수를 섞어서 예측 가능
- 3. foundation model
    - 특징 : 특정 문제 전용이 아니다. 범용적이며 사전학습 후 fine tuning으로 특화한다.
    - 유형 : LLM 기반, 시계열 전용
    - LLM 기반 모델 활용 방식 
        - adaptation 없이 - 이미 시계열 지식을 가지고 있다고 가정
        - adapt LLM - 시계열 데이터로 파인튜닝
        - adapt to LLM - 텍스트로 변환하고 LLM활용
- 4. 딥러닝 모델 선택 가이드

|상황|추천모델|이유|
|---|---|---|
| 짧은 시계열| 통계 모델, RNN| 데이터가 작으면 단순해야 과적합 방지됨|
| 중간 길이 | LSTM, GRU | 장기 의존성 학습 가능, 적절한 복잡도|
| 매우 긴 시계열 | 인포머, 오토포머 | 긴 시퀀스의 전역 패턴 효율적 학습|
| 장기 의존성 중요 | LSTM, 트랜스포머 | GRU 보다 먼 과거 정보 유지에 강함|
| 주기 패턴 뚜렷 | 오토포머 | 주기를 잘 포착|
| 외생 변수 많음 | TFT | 정적/동적 변수 융합 + 해석 가능 |
| 대규모 데이터 | 트랜스포머 | 병렬 연산 -> 학습 효율적|
| 범용 예측 필요 | foundation model | 사전 학습된 지식으로 빠른 적용을 함|
