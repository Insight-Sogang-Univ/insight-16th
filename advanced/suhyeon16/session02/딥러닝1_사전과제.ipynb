# 1. 합성곱 신경망

## 0. 픽셀과 채널

### 1) 픽셀의 개념
- 픽셀: 이미지 데이터를 구성하는 가장 기본적인 단위

#### 1-1) 이미지 데이터에서의 픽셀의 활용
- 이미지 크기 : 2차원 배열 = 세로 픽셀수 \* 가로 픽셀수

#### 1-2) 픽셀 값의 표현 방법
- 흑백 이미지: 하나의 픽셀 값으로 밝기를 표현하며, 일반적으로 0\~255까지의 값을 사용
- 컬러 이미지: 하나의 픽셀을 빨강, 초록, 파랑 세가지 색상으로 표현함.

### 2) 채널
- 채널: 이미지의 색상 정보를 구성하는 각각의 데이터 층을 의미함

### 3) 색공간
- 색공간: 이미지를 구성하는 색상을 숫자로 표현하는 방법이나 체계를 의미함
- 대표적인 색공간:
    - 그레이 스케일: 색상의 구분 없이 밝기 정보만으로 이미지를 표현함
    - RGB: 빨, 초, 파로 세가지 섹상의 조합으로 색을 표현함
    - HSV: 색상, 채도, 명도를 기준으로 색을 표현함

#### 3-1) 그레이 스케일: 이미지의 색상 정보 없으 픽셀의 밝기만을 이용해 표현하는 방식
- 특징:
    - 0\~255 범위의 명도 값을 사용함. 0은 검정, 255은 흰색을 의미함
    - 채널값이 1인 이유는 빨강, 초록, 파랑과 같은 여러 색상 성분을 구분할 필요 없이 하나의 밝기 정보만 저장하면 되기 때문.
- 저장방법: 흑백이미지는 3차원 텐서(스칼라, 벡터, 행렬을 아우르는 개념)로 표현 가능. Numpy에서는 2차원 배열로 저장하기도 함

#### 3-2) RGB: 빨, 초, 파 세가지 색상을 조합하여 하나의 색으로 표현하는 색공간
- 특징:
    - 하나의 픽셀은 세 가지 색깔, 삼원색의 조합으로 이루어짐. 세가지 채널의 값이 합쳐져 최종적인 하나의 색상을 만들어내는 원리이므로, 채널 값은 3으로 표현함
- 저장방법: 높이 28, 너비 28이면 3차원 텐서임. 채널은 깊이라고도 함.

## 1. CNN의 개념

### 1) 이미지 인식 방법 차이

#### 1-1) 사람이 이미지를 인식하는 방법
사람은 이미지를 구성하는 모든 픽셀 하나하나 살펴보지 않음.

#### 1-2) 컴퓨터가 이미지를 인식하는 방법
사람은 바로 강아지, 고양이 구분이 가능함. 컴퓨터는 불가능하기 때문에 이미지를 작은 픽셀 단위로 구분하고, 각 픽셀에 저장된 색상과 밝기 등의 숫자 정보를 처리함 -> 하나의 강아지 사진이 아닌 수 많은 픽셀 값이 모여 만들어진 숫자 데이터임

#### 1-3) 픽셀 정보를 그냥 학습하면 생기는 문제점
이미지를 1차원 배열로 펼쳐서 학습하면 이미지가 갖고 있던 2차원 공간 구조가 명확하게 활용되지 않음. (눈, 귀, 꼬리처럼 주변 픽셀들이 모여 만들어지는 이미지의 특징을 학습하기 어려워짐)

### 2) MLP와 CNN의 구조 비교
- MLP: 입력된 데이터를 FC Layer에 전달하여 학습하는 구조
- CNN: 이미지의 공간적 특징을 유지하면서 Conv와 Pool을 반복하여 이미지의 특징을 추출함. 이후 flatten을 통해 추출된 특징을 1차원ㅇ,로 변환, FC layer을 적용하여 이미지를 분류함

### 3) CNN
- 특징 추출:
    - Convolution Layer: 필터를 이미지 위에서 이동시키며 모서리, 질감, 형태 등 이미지의 다양한 특징을 추출
    - Pooling Layer: 추출된 특징 맵의 크기를 줄이고 중요한 특징을 유지하여 연산량을 감소시킴
- Flatten: 여러 차원으로 이루어진 특징 맵을 FC Layer에 입력할 수 있도록 1차원 배열로 변환
- 분류: Fully Connected Layer: 추출된 특징들을 종합하여 이미지가 어떤 클래스에 속하는지 최종적으로 분류

-> 이미지 인식을 위한 모델: 학습하는 부분 & 분류하는 부분

-> CNN에 새롭게 추가된 과정은 이미지의 특징을 학습하는 부분

## 2. 왜 이미지 처리에 MLP가 아닌 CNN을 사용할까?

#### 1-1) 흑백 이미지의 구조: 각 픽셀에 하나의 밝기 값을 저장하는 방식

#### 1-2) 컬러 이미지의 구조: 하나의 픽셀을 빨, 초, 파 세 가지 색상 값으로 구성됨

#### 1-3) 대표적인 이미지 데이터셋: MNIST - 0부터 9까지의 숫자를 담은 크기의 흑백 이미지 데이터셋으로, 이미지 분류와 CNN 학습의 기초 예제로 자주 활용됨.

#### 1-4) 이미지 구조를 유지하는 모델이 필요한 이유
일반적인 정형 데이터: 나이, 소득, 키와 같이 각 열에 독립적 의미를 가진 값들의 나열

이미지: 픽셀들이 가로와 세로 방향으로 연결되어 주변 픽셀 간의 관계와 위치 정보가 중요함.

-> CNN을 통해 픽셀 간의 공간적 관계와 이미지의 구조를 유지하면서 특징을 학습할 수 있는 모델

### 2) MLP의 한계

#### 2-1) 위치, 크기, 회전 등의 변화에 취약
- MLP는 이미지를 1차원 배열로 펼쳐서 입력받고, 각 픽셀 위치에 대응하는 가중치를 학습함. (즉 학습 데이터에서 보지 못한 위치나 형태의 변화가 발생하면 기존의 숫자로 인식하지 못할 가능성이 높아져 성능이 크게 저하된다는 한계가 있음)

#### 2-2) 파라미터 폭발
- MLP의 경우 완전연결 구조로 인한 파라미터 급증이 이루어지며, 가중치 수가 이미지 픽셀의 개수만큼 생성됨
- 완전 연결: 신경망에서 이전 레이어의 모든 노드가 다음 레이어의 모든 노드와 빠짐없이 연결되어 있는 구조 (과적합 가능성 상승, 이미지가 클 수록 학습이 느리다는 한계 존재)
- CNN에서는 이미지 전체의 픽셀을 다음 계층의 모든 노드에 연결하는 대신, 작은 영역에만 필터를 적용하는 부분 연결 구조를 사용함. => 이미지 크기가 커지더라도 필터의 크기와 개수가 동일하다면 파라미터 수는 크게 증가하지 않음

#### 2-3) 공간 정보 손실
- MLP는 이미지를 입력받을 때 2차원 또는 3차원으로 구성된 이미지 데이터를 1차원 벡터로 펼치는 Flatten 과정을 거침.
- MLP는 픽셀 간의 위치 관계나 주변 픽셀들이 모여 만들어내는 특징을 직접적으로 활용하기 어렵다는 한계가 있음.
- CNN에서는 이미지를 입력 받을 시 처음부터 1차원으로 펼치치 않고, H\*W\*C 형태의 다차원 구조를 유지한 상태로 처리함

## 3. 합성곱과 풀링

### 1. Convolution Layer
- 개념: 입력 이미지 전체를 일정한 크기의 필터로 훑으면서 이미지의 특징을 추출하고 학습하는 계층임
- 연산 과정:
    1) 필터가 이미지 위를 슬라이딩
    2) 각 위치에서의 요소별 곱셈 후 합계 계산
    3) 결과를 Feature Map에 저장
- 필터와 커널의 차이:
    - 커널: 이미지의 특정 패턴을 추출하기 위해 사용하는 2차원 가중치 행렬
    - 필터: 여러개의 커널이 모인 집합 묶음
- Feature Map
    - Convolution 연산을 통해 생성되는 결과 데이터
    - 필터가 이미지의 여러 ㅟ치에서 계산한 결과값을 모아 하나의 Feature Map을 구성함
    - 필터가 특정 특징에 반응할 수록 해당 위치의 Feature Map 값이 커진다는 특징이 있음
- 차원 결정 방식:
    - Feature Map의 개수는 사용하는 필터의 개수에 따라 결정됨

### 2. Padding
- Padding: Convolution 연산을 수행하기 전에 입력 이미지의 가장자리에 추가적인 픽셀을 삽입하는 방법
- 효과: 이미지의 공간적 크기를 유지하면서 가장자리의 특징까지 학습할 수 있도록 하는 역할을 함

### 3. Stride
- Stride = 1
    - 한칸씩 이동, 이미지의 여러 위치 세밀히 확인, Feature Map의 크기가 상대적으로 큼
- Stride = 2
    - 두칸씩 이동, 일부 위치를 건너뛰며 연산, Feature Map의 크기가 상대적으로 작음
- Stride는 특징을 얼마나 세밀하게 추출할지와 연산 효율을 어떻게 조절할지 결정하는 요소임

### 4. Pooling
- 개념: Pooling은 이미지의 작은 위치 변화에 대한 민감도를 낮춰 모델이 특징을 더 안정적으로 인식할 수 있도록 도와줌
- 일정한 크기의 영역을 설정하고, 해당 영역에 포함된 값들을 하나의 값으로 요약함.
- Pooling을 활용하지 않는다면: 필터는 엄격하게 특정한 픽셀 위치에서만 해당 패턴을 찾으려고 함
- Max Pooling: 지정된 영역 안에서 가장 큰 값을 선택하는 방식
- Average Pooling: 지정된 영역 안에 있는 모든 값의 평균을 계산하는 방식
- 장점:
    - 차원 축소로 계산량 감소
    - 위치 변화에 대한 강건성
    - 과적합 방지에 도움
    - 노이즈 영향 완화
- 단점:
    - 정보 소실의 위험
    - 정확한 위치 정보 감소

## 4. 대표적인 CNN 아키텍처 살펴보기

### 1) AlexNet
- 개념: 대규모 이미지 데이터를 활용해 이미지의 특징을 직접 학습하고 분류하는 딥러닝의 가능성
- 구조: Convolution Layer 5개, Fully Connected Layer 3개
- 주요 특징: ReLU 활성화 함수 도입, 다중 GPU 병렬 연산, 드롭아웃, 데이터 증강, 로컬 응답 정규화

### 2) VGGNet
- 개념: 네트워크의 깊이를 늘리면 이미지 인식 성능이 향상될 수 있다는 점을 보여준 모델
- 구조: Convolution Layer 13개, Fully Connected Layer 3개
- 주요 특징: 작은 필터의 반복 사용, 깊은 네트워크, 일관된 구조

### 3) ResNet
- 개념: 깊은 신경망을 효과적으로 학습하기 위해 개발된 CNN 모델
- 구조: 핵심 혁신 기술: 잔차 학습과 지름길

# 2. 이미지 딥러닝 응용

## 1) 이미지 분류
- 이미지 분류 모델 : ViT (이미지를 작은 패치 단위로 나누고, 이를 Transformer에 입력하여 이미지의 특징을 학습하는 모델),
- CNN의 확장 가능성
    - CNN은 이미지 전체를 하나의 클래스로 분류하는 작업뿐 아니라, 이미지 안에 존재하는 여러 객체를 찾아내는 작업에도 활용됨

## 2) 객체 탐지
- 객체 탐지는 이미지 분류에서 한 단계 확정된 작업. 이미지 안에 있는 개별 객체의 종류와 위치를 함께 예측함
- Bounding Box: 이미지에서 객체가 위치한 영역을 둘러싸는 사각형임
- 모델 적용 방식:
    - Classification Head: 객체의 종류 예측
    - Regression Head: 객체의 위치 좌표 예측

## 3) 이미지 캡셔닝
- 이미지의 내용을 자연어 문장으로 표현하는 작업
- 모델 적용 방식: CNN & RNN
    - CNN: 이미지 특징 추출
    - RNN: 문장 생성
- 이미지 캡셔닝 단계
    1) CNN을 이용한 특징 추출
    2) Attention Network를 이용하여 단어 생성
- Attention은 문장을 생성할 때 이미지의 모든 영역을 동이랗게 고려하지 않고, 현재 생성하려는 단어와 관련된 중요한 영역에 더 높은 가중치를 부여하는 방식

# 3. 과거를 기억하는 신경망

## 3-1) 순차 데이터
- 순차 데이터: 데이터의 순서가 의미를 갖는 데이터로, 순서를 바꾸면 정보가 달라지는 데이터
- 순차 데이터 세부 분류:
    1) 순서의 기준: 시간적 순차 데이터/비시간적 순차 데이터
    2) 자료형: 수치형/범주형
- 기존 정형 데이터 vs. 순차 데이터
    - 정형 데이터: 순서를 섞어도 의미 그대로, 상호 독립적임
    - 순차 데이터: 순서 바꾸면 의미가 달라짐, 상호 의존적임

## 3-2) RNN의 구조와 원리
- 구조: 기존 신경망과 비교했을 때 가장 큰 차이점은 은닉층에 순환 구조가 존재한다는 것
- 원리:
    - 핵심: 매 시점마다 현재 시점 입력값과 이전 시점 은닉상태를 이용해 현재의 은닉상태를 계산하고 그 결과를 다음 시점으로 넘겨줌
- RNN 아키텍쳐:
    - One to Many: 하나의 입력에서 시퀀스 출력
    - Many to One: 시퀀스 입력에서 하나의 값 출력
    - Many to Many: 시퀀스 입력에서 시퀀스 출력

## 3-3) RNN의 한계 및 장기 의존성 문제

### 1) RNN의 한계점:
- 시간에 따른 역전파 구조
- RNN은 순전파를 통해 시퀀스를 처리한 뒤, 시간을 거슬렁 로라가며 오차를 전파하여 가중치 기울기를 누적 계산함
- 기울기 소실: 역전파 과정에서 기울기가 과거 시점으로 갈 수록 점점 작아져 0에 가까워지는 현상
- 기울기 폭주: 역전파 과정에서 기울기가 과거 시점으로 갈 수록 점점 커져 매우 큰 값이 되는 현상
- 느린 훈련 시간: RNN은 시점을 하나씩 순서대로만 계산할 수 있어서, 시퀀스가 길 수록 학습에 시간이 오래 걸리게 됨

### 2) 장기 의존성 문제: 시퀀스에서 관련된 정보가 멀리 떨어져 있을수록, RNN이 그 관계를 기억하고 학습하지 못하는 문제 => 그래서 중요한 정보는 잊지 않고 오래 기억하고, 불필요한 정보는 잊는 메모리 구조 필요 (LSTM&GRU)

# 4. 똑똑하게 기억하고 잊는 법: LSTM&GRU

## 4-1) LSTM&GRU
- LSTM: 기억을 두 개의 상태로 나눠서 관리 (셀 상태: 장기 기억, 은닉 상태: 단기 기억 & 출력), 세 개의 게이트: Forget Gate, Input Gate, Output Gate
- GRU: 기억을 하나로 통합해서 관리, 두 개의 게이트: Update Gate, Reset Gate

## 4-2) 게이트로 정보의 흐름을 제어하다: LSTM

### 1) LSTM 구조
- 셀 상태(장기 기억); 구조도 맨 위로 가로지르는 선 -> 시점을 따라 쭉 이어짐 => 기억할 내용과 잊어버릴 내용을 선택해, 중요한 정보를 오래 가져갈 수 있음

### 2) 단게별로 알아보는 LSTM 수식
1. Forget Gate: 무엇을 잊을지
2. Input Gate: 무엇을 새로 기억할지
3. Final memory cell: 기억 업데이트
4. Output Gate: 무엇을 내보낼지

### 3) LSTM 장단점
- 장점: 기울기 소실 완화, 장기 의존성 학습 가능, 문장 번역/음성 인식 등 긴 시퀀스 과제에서 RNN보다 좋은 성능
- 단점: 계산량이 많아 학습이 느리고 메모리 많이 사용, 병렬 처리 불가, 매우 긴 시퀀스에서는 여전히 한계

## 4-3) 더 단순하고 효율적인 구조: GRU

### 1) GRU의 정의: LSTM의 핵심 아이디어는 유지하면서 구조를 단순하게 줄인 모델
- LSTM: 상태 2개, 게이트 3개
- GRU: 상태 1개, 게이트 2개

### 2) LSTM과의 차이점
- Forget Gate와 Input Gate를 Update Gate 하나로 통합
- Reset Gate 사용
- Output Gate가 없음
- Gate의 개수가 3개에서 2개로 감소

### 3) GRU의 장단점:
- 장점: 상태 1개, 게이트 2개로 단순, 파라미터가 적어 학습이 빠르고 메모리 사용 적음, 파라미터가 적어 데이터 적을 때 학습 안정적, Update Gate 하나로 기억 갱신을 간단히 처리, 많은 과제에서 LSTM과 비슷한 성등
- 단점: 장기 기억과 출력을 분리하지 못함, 데이터가 많고 복잡한 과제에서는 LSTM이 나을 수도 있음, 잊는 양과 더하는 양을 따로 조절 불가 (정교함이 떨어짐), 항상 LSTM보다 낫다고 단정 지을 수 없음

## 4-4) LSTM vs GRU: 언제 무엇을 선택할까
- 기준: 정교함과 효율 사이의 선택
- 문제 상황에 따른 적합한 모델이 다름

# 5. 문장을 입력받아 문장을 출력하다: Seq2Seq

## 5-1) Seq2Seq 의 기본 구조: 인코더와 디코더
- Seq2Seq: 입력 시퀀스를 받아 다른 시퀀스를 출력하는 모델 ( 기존 RNN과의 차이: 입력과 출력의 길이가 서로 달라도 됨)
- 특징: 인코더(입력 문장을 읽고 하나의 컨텍스트 벡터로 압축) 디코더(컨텍스트 벡터를 받아 출력 문장을 한 단어씩 생성) 두개의 모듈로 구성
    1) 인코더: 입력 문장의 모든 단어를 순서대로 읽고, 문장 전체의 정보를 하나의 벡터로 압축하는 LSTM<br>
    작동 방식: 단어 토큰화 -> 임베딩 -> LSTM 셀에 한 단어씩 입력 -> 마지막 시점의 은닉 상태 = 문장 전체 정보가 담긴 컨텍스트 벡터 -> 디코더로 전달
    2) 디코더: 컨텍스트 벡터를 받아, 출력 문장을 한 단어씩 순서대로 생성하는 LSTM<br>
    작동 방식: 컨텍스트 벡터를 디코더 LSTM의 첫 은닉 상태로 받음 -> 시작 신호\<sos\>를 첫 입력으로 넣음 -> 매 시점 softmax로 다음 단어의 확률 계산 (가장 확률이 높은 단어 선택) -> 방금 예측한 단어를 다음 시점의 입력으로 다시 넣음 -> 종료 신호 \<eos\>가 나오면 생성 종료
    3) 컨텍스트 벡터: 인코더가 입력 문장 전체를 읽고 만든 문장 요약 벡터. 인코더 LSTM의 마지막 시점 은닉 상태. 인코더와 디코더를 연결하는 유일한 통로
