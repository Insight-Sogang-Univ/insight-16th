# 퍼셉트론 & 딥러닝 기초

## 1. 퍼셉트론

### 1-1.
- 인공신경망: 우리 뇌의 신경망을 모방한 모델 -> 마치 생물학적 뉴런이 정보를 처리하는 구조와 비슷함
- 퍼셉트론 : 초기 형태의 인공신경망으로, 신경망의 기원이 되는 선형 이진 분류기임.
    - -> 구하는 절차: 1) 입력값과 가중치 곱하기, 2) 곱한 값들의 총합 구하기, 3) 총합이 임계값을 넘으면 1, 넘지 않으면 0 출력
- 용어 정리 :
    1) 가중치: 입력값이 출력값에 주는 영향력 결정, 각 입력값에는 각각의 가중치가 존재함, 가중치 값이 클 수록 입력값이 중요하다 !!
    2) 활성화 함수: 신호가 일정 이상의 값을 넘으면 1, 넘지 못하면 0을 출력하는 퍼셉트론의 성질은 활성화 함수로 계단 함수 사용
    3) 임계값: 활성화 함수에 사용되는 값으로, 세타로 표현함
    4) 편향: 계단 함수를 구현한 식에서 임계값을 좌변으로 넘기면 편향으로 표현 가능. 임계값과 편향은 표현의 차이임.

### 1-2. 단층 퍼셉트론과 그 한계
1) 단층 퍼셉트론: 입력층 & 출력층으로만 이루어진 퍼셉트론을 단층 퍼셉트론이라 함
2) 논리 게이트 구현: 단층 퍼셉트론을 이용하여 AND, NAND, OR 논리 게이트(기본적인 논리 연산)를 쉽게 구현할 수 있음
3) 대표적인 논리 게이트:
    - AND: 입력값이 모두 1인 경우에만 출력값이 1이 나오는 구조
    - OR: 둘다 0일 때만 0
    - NAND (AND의 부호 뒤집은 것): 둘다 1일 때만 0
    - XOR: 서로 다를 때만 1
4) 단층 퍼셉트론의 한계: XOR 게이트를 구현할 수 없다. 퍼셉트론은 입력값과 가중치의 선형 결합으로 이루어진 구조이므로 선형 분류 문제에 대해서만 구현 가능 -> 즉 직선이 아닌 곡선/비선형 영역으로 분리해야만 구현 가능

### 1-3. 다층 퍼셉트론:
- 퍼셉트론의 층을 더 쌓은 구조
- 단순히 층을 많이 쌓는다 해서 비선형성이 생기지 않음
- 다층 퍼셉트론이 XOR을 구현 가능한 이유는 은닉층에서 비선형 활성화 함수를 적용해 새로운 특징을 만들고, 다음 층에서 그 특징들을 조합하기 때문임.

## 2. 딥러닝 모델은 어떻게 학습하는가?

### 2-1. 순전파와 역전파
1) 학습 알고리즘의 기본 구조:
    - 순전파를 통해 입력값으로부터 예측 수행
    - 오차 계산을 통해 예측 정확도 측정
    - 역전파를 통해 출력층에서 다시 입력층 방향으로 오차 전달
    - 가중치 업데이트
    - 반복!! 오차값이 0에 가까워질 때 까지

-> 연쇄법칙을 통해 순전파/역전파가 일어남

2) 연쇄 법칙: 합성함수의 미분은 각 함수의 미분의 곱으로 나타낼 수 있다는 법칙
    - 딥러닝에서는 입력값이 최종 출력에 얼마나 영향을 미쳤는지 계산하기 위해 각 층이 다음 층에 미친 영향을 순서대로 곱해나가야 함.
    - 딥러닝 모델 학습시 오차 줄이는게 관건. 이를 위해선 역전파 과정에서 각 가중치가 최종 손실에 얼마나 기여했는지를 연쇄법칙을 통해 계산해야함.
3) 기울기 소실 문제: 기울기들을 계속 곱했지만 기울기가 0이 되는 문제
    - 기울기 소실 문제 발생시 앞쪽 층의 가중치가 거의 업데이트 되지 않아서 학습이 제대로 이뤄지지 않는 문제가 생김

### 2-2. 손실 함수
1) 손실 함수: 모델의 예측이 얼마나 틀렸는지를 수치로 표현하는 방법
    - 오차 계산법: 실제값과 예측값의 차이를 계산하는데 사용되는 함수
    - 손실 함수 종류: 회귀 - 평균제곱오차, 분류 - 크로스 엔트로피 오차
2) 손실 함수의 종류
    - 평균 제곱 오차: 예측값이 연속적이고, 회귀 문제에 적합함
    - 이진 크로스 엔트로피 : 이진 분류 문제 (로지스틱 회귀)에서 사용하는 손실 함수임
    - 크로스 엔트로피 오차: 다중 클래스 분류 문제에서 예측 확률이 실제 클래스와 얼마나 일치하는지 평가하는 데 적합함 (다중 분류에서 사용)

### 2-3

#### 1. 활성화 함수
- 신경망 과정에서 활성화 함수를 거치는 이유 :
    1) 비선형이 필요함 (XOR 같은 문제를 풀지 못함)
    2) 미분이 되어야 함 (역전파는 미분값을 연쇄적으로 곱해 나감. 즉 활성화 함수는 미분이 가능해야 함)

#### 2. 계단 함수
- 입력값과 가중치의 곱을 모두 더한 뒤, 그 합이 임계값을 넘으면 1, 아니면 0을 출력하는 함수
- 임계값: 뉴런이 얼마나 쉽게 활성화 되는지를 제어
- 가중치: 각 입력 신호가 결과에 미치는 영향력 제어

#### 3. 시그모이드 함수
- 입력 신호를 0과 1사이의 연속적인 값으로 변환하는 비선형 함수임
- 계단 함수와 출력값의 형태에서 차이점이 있음 : 계단 함수 - 0 또는 1만 출력, 시그모이드 - x값에 따라 출력값이 연속적으로 변화함

#### 4. ReLU 함수
- 입력이 0보다 크면 그대로, 0 이하면 0을 출력하는 비선형 함수임
- ReLU 많이 사용하는 이유 :
    1) 계산이 간단하고 효율적
    2) 비선형성 제공
    3) 기울기 소실 완화

#### 5. Softmax 함수
- 각 클래스에 대한 출력값을 확률로 변환하는 함수 (다중 클래스 분류 문제의 출력층에서 사용)

### 2-4. 경사 하강법

#### 1. 경사하강법의 원리
- 목표: 손실을 가장 작게 만드는 W를 찾는 것
- 손실 함수를 그래프로 그릴 때 x축을 W, y축을 E로 두면 아래로 볼록한 곡선이 생김. 곡선의 바닥값을 확인하기 위해선 미분값의 부호를 확인하면 됨.

#### 2. 학습률
- 미분값을 그대로 빼면 문제가 생김. 오히려 멀어지기 때문에 이동량을 조절함. (0.001부터 시작하면 됨)

#### 3. 배치/에포크/이터레이션
1) 배치
    - 학습 과정에서 매개변수를 조정하기 위해 사용하는 데이터의 묶음을 의미함
    - 배치 크기가 너무 클 경우 학습이 느려지며 메모리가 부족해질 수 있음
    - 배치 크기가 너무 작으면 가중치가 자주 업데이트되어 학습이 불안정해짐
2) 에포크
    - 모든 데이터셋을 학습하는 횟수
3) 이터레이션 또는 스텝
    - 한 번의 에포크를 끝내기 위해 필요한 배치의 수
    - 1Epoch = 배치 크기 * 이터레이션

#### 4. 경사하강법의 종류
1) 배치 경사 하강법:
    - 오차 구할 시 전체 데이터를 한 묶음으로 묶음 (Batch size = Data size)
    - 전체 데이터로 학습하므로 업데이트 횟수 적음
    - 한번에 처리 = 메모리가 많이 필요함
    - 수렴이 안정적
2) 확률적 경사하강법:
    - 단점: 전체 데이터를 계산하기 때문에 소요 시간이 김
    - 랜덤으로 선택된 하나의 데이터에 대해서만 계산함
    - 하나의 데이터만 메모리에 저장하면 됨 = 자원이 적은 컴퓨터에서도 가능
    - 매개변수의 변경폭이 불안정하고 정확도가 낮을 수 있음
3) 미니 배치 경사하강법:
    - 배치 크기를 조정하여 해당 개수 만큼 매개변수를 조정
    - 전체 데이터보다 빠르고, 배치 크기 1보다 안정적 -> 가장 많이 사용

### 2-5. 옵티마이저

#### 1. 모멘텀
- 기존 경사하강법이 기울기를 잊는 증상을 완화하기 위해 이전에 가던 방향을 기억하도록 관성을 부여
- Local Minima를 관성의 힘으로 탈출 가능
- 기울기가 매우 작은 구간도 빠르게 통과함

#### 2. RMSProp
- 기울기가 큰 방향에서는 너무 많이, 작은 방향에서는 너무 조금 이동함 -> 각 방향마다 학습률을 따로 조정해서 안정적이고 빠르게 학습하도록 함

#### 3. Adam
- RMSProp와 Momentum을 합친 방법 (방향과 학습률 두 가지를 모두 잡기 위한 방법)
