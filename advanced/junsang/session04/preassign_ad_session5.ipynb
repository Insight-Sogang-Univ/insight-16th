## 1. NLP  

### 1-1 NLP
> NLP(Natural Language Processing): 자연어 처리
- 자연어 : 인간이 의사소통을 위해 사용하는 언어로, 일대다 형식  
- 프로그래밍 언어 : 컴퓨터에 명령을 전달하기 위해 설계된 언어, 일대일 형식  
-> 자연어는 유현하지만 모호성이 높고, 프로그래밍언어는 상대적으로 명확하고 엄격  

#### 자연어의 특징  

1. 이산적 데이터(Discreteness)  
- 자연어는 단어, 문자 등 구분된 기호로 구성되는 이산적 데이터  
- 연속적인 숫자 데이터와 달리 컴퓨터가 직접 수학적으로 처리하기 어려움  
-> NLP는 단어와 문장을 숫자 or 연속적인 벡터로 변환(Embedding)  

2. 순차적 데이터(Seqentiality)  
- 자연어는 단어가 등장하는 순서에 따라 문장의 의미가 변화  
-> NLP는 단어의 순서와 앞뒤 문맥을 함꼐 고려해 처리할 수 있는 모델 구조(RNN, Transformer,...)가 필요  

3. 중의성과 모호성(Ambiguity)  
- 하나의 단어나 문장이 여러가지로 해석될 수 있음(*e.g* 배 : 신체 부위, 과일, 선박)  
-> NLP는 핵심 단어 하나가 아닌 문맥 전체를 파악하여 적절한 의미를 판단해야함  

4. 희소성(Sparsity)  
- 자연어는 단어의 종류와 조합이 매우 다양하여 모든 표현을 학습하기 어려움  
-> NLP는 학습하지 않은 표현도 단어 간 유사도를 이용해 확률적으로 추론할 수 있는 능력이 필요  

5. 가변성과 진화(Variability & Evolution)  
- 자연어는 시대와 사회적 환경에 따라 지속적으로 변화(신조어, 줄임말, 새로운 표현...)
-> NLP는 변화하는 언어와 새로운 표현을 고정된 규칙이 아니라 데이터로부터 유연하게 패턴을 학습하는 머신러닝 방식이 필요  


#### NLP의 개념  

> NLP란? 인간이 일상생활에서 사용하는 자연어를 컴퓨터가 분석하고 이해하며 생성할 수 있도록 하는 인공지능 기술  
활용 분야 : 기계번역, 감성분석, 텍스트 분류, 문서 요약, 질의응답 및 챗봇, etc...  


#### NLP의 역할  

  

1. 형태소 분석(Morphological Analysis)  
- 문장을 의미를 가진 가장 작은 단위인 형태소로 나누고, 각 형태소의 품사와 문법적 특징을 분석  
    - *e.g.* '나는 학교에 간다' -> '나(대명사) + 는(조사) + 학교(명사) + 에(조사) + 가-(동사 어간) + ㄴ다(어미)'  

2. 구문 분석(Syntax Analysis)  
- 문장의 문법적 구조를 분석하여 단어들이 어떤 관계로 연결되어 있는지 파악  
    - *e.g.* '철수가 사과를 먹었다' -> 철수(주어), 사과(목적어), 먹었다(서술어)  

3. 의미 분석(Semantic Analysis)  
- 단어와 문장의 의미를 분석하여 문맥에 맞는 의미를 파악  

4. 담화 분석(Discourse Analysis)  
- 여러 문장 사이의 연결 관계와 전체 대화의 맥락을 분석하여 의미를 파악


### 1-2 언어 모델  

#### 언어모델의 정의  

> 언어모델이란? 대량의 텍스트 데이터를 학습하여 단어 또는 토큰의 등장 확률을 예측하고, 문장의 자연스러움을 판단하는 모델 *i.e. 인간이 말하고 쓰는 방식을 모방하는 시스템*  

- 앞에 등장한 단어와 문맥을 바탕으로 다음에 어떤 단어가 나올 가능성이 높은지 예측
- 문장을 구성하는 단어들의 등장 확률을 계산하여 문장 전체의 자연스러움을 평가
- 일반적으로 확률이 높은 문장을 자연스러운 문장으로 판단
    - *e.g.* '나는 아침에 밥을 _____'  빈칸에 올 단어로 먹었다/마셨다/달렸다 중 먹었다가 오는 것을 판단하는 것  

언어모델은 학습 방식에 따라 통계적 언어모델(SLM)과 신경망 언어모델(NLM)으로 구분


## 2. 통계적 언어 모델(SLM)  

### 2-1 SLM(Statistics Language Model)이란? SLM과 분포가설   

SLM은 분포가설(Distribution Hypothesis)라는 아이디어에 근거  
> 분포가설이란? 비슷한 문맥에서 등장하는 단어들은 서로 유사한 의미를 가질 가능성이 높다는 가설  

#### 조건부 확률  

특정 사건이 발생했다는 조건 아래, 다른 사건이 발생할 확률  
- 앞에 등장한 단어들을 조건으로 다음 단어가 등장할 확률을 계산  

#### 카운트 기반 확률  

- 특정 문맥 다음에 어떤 단어가 얼마나 자주 등장했는지 계산
- 등장 빈도가 높은 단어일수록 높은 확률을 부여  

**카운트 기반 모델의 한계 : 희소 문제(Sparsity Problem)**  
> 희소문제란? 가능한 단어의 조합이 매우 많아 특정 표현이 학습 데이터에 거의 등장하지 않거나 전혀 등장하지 않는 문제  
카운트 기반 모델은 학습데이터에서 관측된 단어의 등장빈도에 의존하기에 실제로 자연스러운 표현이더라도 관측되지 않은 단어에 대해선 확률이 0으로 계산됨  
- *e.g.* '나는 학교에 갔다'라는 문장은 학습데이터에 자주 있는 반면, '나는 학교에 왔다'라는 문장은 없다면, 실제로 말이되는 문장이지만 카운트 기반 모델에선 확률이 0으로 계산됨  

=> 희소문제를 완화하기 위해 확률 추정 기법과 새로운 언어 모델 필요  


### 2-1 N-gram   

#### N-gram이 나오게 된 배경  

SLM은 이전에 등장한 단어들을 바탕으로 다음 단어의 등장 확률을 계산하지만, 문장이 길어질수록 한계가 발생(데이터 희소성 문제, 계산의 복잡성 문제)  
-> 이를 완화하기 위해 N-gram모델 등장

> N-gram이란? 연속된 N개의 단어로 구성된 묶음으로, 직전 N-1개의 단어를 바탕으로 다음 단어의 등장 확률을 예측하는 통계적 언어 모델  

- 이전의 모든 단어가 아닌 가장 최근의 N-1개 단어만 활용하여 다음 단어 예측   

**SLM vs N-gram**  
- 공통점  
    - 두 모델 모두 단어의 등장 빈도와 조건부 확률을 활용하여 다음 단어를 예측하는 통계적 언어 모델  
    - 학습 데이터에서 관측된 단어 조합을 바탕으로 문장의 등장 확률과 자연스러움을 평가  

- 차이점  
    - SLM : 이전의 모든 단어를 조건으로하여 다음 단어 표현, 정확도는 높을지라도 긴 문맥에 대한 추정 어려움  
    - N-gram : 직전 N-1개만 고려하므로 확률 추정이 상대적으로 단순  

#### N-gram의 한계  

1. 희소문제  
    - N값이 커질수록 가능한 단어 조합의 수가 급격히 증가하여, 훈련 데이터에서 특정 단어 조합을 찾기 어려워지는 희소 문제 발생(마찬가지로 등장하지 않으면 확률 0)  

2. 문맥 제한  
    - 직전 N-1개만 고려하므로 멀리 떨어진 단어 사이의 관계는 파악이 어려움  
    - 문장이 길어질수록 앞부분의 중요한 정보가 예측 과정에서 제외딜 수 있음  


### 2-3 PPL(Perplexity)  

> PPL이란? 언어모델이 주어진 문장을 예측할 때 얼마나 혼란스러워하는지(perplexed)를 수치로 나타낸 지표  

- PPL이 낮을수록 : 모델의 예측 불확실성이 낮아 실제 등장한 단어를 잘 예측  
- PPL이 높을수록 : 모델의 예측 불확실성이 높아 실제 등장한 단어를 예측하기 어려움   


#### PPL의 특징  

1. 문장 W의 빈도수와 PPL의 관계  
    - 문장 W가 훈련 데이터에 희소하게 등장할수록 모델이 해당 표현을 학습할 기회가 적어 예측에 어려움을 겪을 수 있음  
    -> 희소한 문장일수록 PPl이 높아지고, 자주 등장하는 문장은 상대적으로 낮음  

2. 단어개수 N과 PPL의 관계  
    - 문장에 예측하기 쉬운 단어 추가시 예측 난이도가 낮아져 PPL이 감소할 수 있음  
    - N이 증가한다고 PPL이 반드시 낮아지는것은 X, 추가된 단어의 예측 난이도에 따라 달라짐  


## 3. 신경망 언어 모델 (NLM)  

> 기존의 SLM의 한계 완화를 위해 인공신경망을 활용하여 단어의 의미와 문맥을 학습하는 신경망 언어모델(NLM)이 등장  

#### NLM의 특징  
- 단어를 연속적인 숫자 벡터로 변환하여 단어간 의미적 관계 학습
- 신경망을 통해 다양한 단어 조합의 특징을 학습하고, 관측되지 않은 표현에도 일반화할 수 있도록 설계
- 등장빈도뿐 아니라 단어 간 관계와 문맥을 반영하여 다음 단어 예측  

**신경망 언어 모델의 키워드 : 임베딩**  

  

위 그림처럼 의미나 문맥이 유사한 단어들은 벡터공간상에서 서로 가까운 위치에 놓이도록 학습하여 단어간 의미적 유사성을 파악하고, 관측되지 않은 표현에도 일반화 가능   

| 구분 | 통계적 언어 모델(SLM) | 신경망 언어 모델(NLM) |
| --- | --- | --- |
| 학습 방식 | 단어 등장 빈도와 확률 계산 | 인공 신경망의 가중치 학습 |
| 단어 표현 | 단어를 개별 기호로 취급 | 단어를 연속적인 벡터로 표현 |
| 의미적 관계 | 유사 단어 간 관계 반영이 어려움 | 임베딩을 통해 의미적 유사성 학습 |
| 희소 문제 | 관측되지 않은 조합에 취약 | 학습된 표현을 통해 희소 문제 완화 |
| 문맥 처리 | N-gram은 직전 N-1개 단어만 고려 | 신경망 구조에 따라 다양한 범위의 문맥 학습 |
| 계산 비용 | 상대적으로 낮음 | 학습에 많은 연산 자원 필요 |  



## 4. 딥러닝 기반 모델  

### 4-1 Transformer의 등장  

이전 세션에서 배운 장기의존성 문제, 기울기 소실문제, 순차처리의 비효율성(RNN, LSTM의 한계)을 해결하기 위하여 Transformer가 등장 (Review of Session04)  
Self-Attention을 통해 단어간 관계와 장거리 문맥을 효과적으로 학습하고, 병렬처리와 위치 인코딩을 활용하여 학습 효율 향상 및 단어 순서정보 반영  

### 4-2 BERT  

> BERT란? Transformer의 인코더 구조를 기반으로, 문장의 양방향 문맥을 학습하여 언어의 의미를 이해하는 신경망 언어 모델  
B(Bidirectional) + ER(Encoder Representations) + T(from Transformers)  

#### BERT의 특징  

1. 양방향(Bidirectional)  
    - 특정 단어의 의미를 파악할 때 앞에 등장한 단어뿐 아니라 뒤에 등장한 단어까지 함께 고려  

2. Transformer 기반  
    - Transformer의 인코더 구조만 활용하여 문장 내 단어 간 관계를 분석  
    - Self-Attention을 통해 각 단어가 다른 단어들과 얼마나 관련되어 있는지 학습  

3. 사전학습(Pre-training) + 파인튜닝(Fine-tuning)  
    - 사전학습 : 대량의 텍스트 데이터를 활용하여 언어의 일반적인 구조와 문맥 학습  
    - 파인튜닝 : 사전학습된 모델을 특정 작업의 데이터로 추가 학습하여 목적에 맞게 활용  


#### BERT의 학습 방법  

BERT는 여러 개의 Transformer 인코더 층을 쌓은 구조, 입력 문장을 토큰으로 분리한 후 Self-Attention을 통해 양방향 문맥을 학습  

1. 입력 문장 토큰화(Tokenization)  
    - 입력 문장을 단어 또는 서브워드 단위의 토큰으로 분리  
    - 문장 시작과 끝에 각각 'CLS', 'SEP' 등의 특수 토큰 추가  

2. 임베딩(Embedding)  
    - 각 토큰을 벡터로 변환  
    - 토큰의 의미, 위치, 문장 구분 정보를 반영하여 입력 표현 생성  

3. Transformer Encoder를 통한 문맥 학습  
    - Self-Attention을 활용하여 각 토큰이 문장 내 다른 토큰들과 어떤 관계를 가지는지 분석  
    - 여러 인코더 층을 거치며 문맥이 반영된 벡터 표현 생성  

4. 출력 및 활용  
    - 학습된 문맥 정보를 바탕으로 문장 분류, 감성 분석, 질의응답 등 문장 이해를 목적으로 하는 작업 수행  
    - 사전학습 이후 파인튜닝하여 활용  


#### BERT의 사전학습  

> 사전학습이란? 특정 NLP 작업을 수행하기 전에 대량의 텍스트 데이터를 활용하여 언어의 일반적인 구조와 문맥을 미리 학습하는 과정  

BERT는 MLM과 NSP라는 두가지 학습 방법 사용  

##### Masked Language Model(MLM)  
> MLM? 빈칸채우기 방식을 통해, 입력 문장의 일부 토큰을 가린 뒤 주변 문맥을 활용하여 가려진 원래 토큰을 예측하도록 학습하는 방법  

**학습 과정**  
1. 입력 문장에서 일부 토큰을 무작위로 선택
2. 선택된 토큰의 대부분을 'Mask'로 대체
3. BERT가 가려진 토큰의 앞뒤 문맥을 함께 분석
4. 원래 토큰을 예측하고 정답과 비교하여 모델 학습  

  


MLM학습을 통해 앞뒤 문맥을 함께 활용하는 양방향 문맥 이해 능력이 햫상되고, 동일한 단어라도 문맥에 따라 달라지는 의미 학습 가능  

##### Next Sentence Prediction(NSP)  
> NSP? 두 문장이 주어졌을 때, 두 번째 문장이 첫 번째 문장 바로 다음에 등장하는 문장인지 예측하도록 학습하는 방법  

**학습 과정**  
1. 두 개의 문장 A와 B 입력
2. 문장 B가 실제로 문장 A 다음에 등장하는 문장인지 판단
3. 실제 다음 문장이면 'IsNext', 무작위로 선택된 문장이면 'NotNext'로 분류
4. 예측 결과와 정답을 비교하여 모델 학습  

  

NSP학습을 통해 두 문장 사이의 연결관계와 문맥적 연관성을 학습하도록 설계  


### 4-3 GPT   

> GPT란? Transformer의 디코더 기반 구조를 활용하여 이전 단어들을 바탕으로 다음 단어를 예측하고 텍스트를 생성하는 언어 모델

**초기 GPT구조**  
- Decoder 기반 : Transformer의 인코더 없이 디코더 구조를 활용하며, 인코더와 연결되는 Cross-Attention은 사용하지 않음  
- Masked Self-Attention : 미래 토큰을 참고하지 않고 이전 토큰들만 활용하여 다음 토큰 예측 (디코더의 핵심!)
- Feed Forward Network : Attention 결과를 추가로 변환하여 언어적 특징 학습  

-> GPT는 이전 토큰을 바탕으로 다음 토큰을 순차적으로 예측하는 자기회귀 방식으로 텍스트 생성  


### 4-4 LLM   

> LLM(Large Language Model)이란? 방대한 텍스트 데이터와 대규모 신경망을 활용하여 인간의 언어를 이해하고 생성하도록 학습된 인공지능 모델  

#### LLM의 작동 방식  

**학습** : 방대한 데이터를 학습하며, 단어와 단어 사이의 관계, 문맥적 의미, 문법 구조, 세상의 지식등 다양한 언어의 패턴을 스스로 터득  
학습 과정에서 예측값과 정답의 차이를 계산하고, 이를 줄이도록 모델의 가중치 업데이트  

**예측** : 
- 입력 처리 : 사용자의 질문이나 문장을 토큰으로 분리하고 벡터로 변환
- 문맥 분석 : Transformer의 Attention을 활용하여 입력 토큰 사이의 관계 분석
- 다음 토큰 예측 : 이전 문맥을 바탕으로 다음에 등장할 토큰의 확률 계산
- 결과 생성 : 학습한 패턴을 기반으로 다음에 올 확률이 가장 높은 단어를 순서대로 예측하며 응답 생성


### 4-5 RAG  

> RAG? 외부 지식 검색과 LLM의 텍스트 생성 능력을 결합하여 답변의 정확성과 최신성을 높이는 기술  

#### RAG의 구조  

  

1. 질의 인코더(Query Encoder)에서 사용자의 질문을 임베딩 모델을 통해 숫자 벡터로 변환. 이때, 질문의 의미를 벡터로 표현하여 관련성이 높은 문서 검색  

2. 지식 검색기(Knowledge Retriever) : 벡터 데이터베이스(Vector DB)에서 질문 벡터와 유사도가 높은 문서를 검색하고, 답변 생성에 활용할 수 있도록 LLM에 전달  

3. 지식 증강 생성기(Knowledge-Augmented Generator) : 사용자의 질문과 외부문서를 결합하여 LLM에 입력하고 LLM이 최종 답변을 생성  

#### RAG의 장점  

1. 최신 정보 반영  
    - 모델을 재학습하지 않고도 외부 데이터베이스나 최신 문서를 검색하여 활용 가능  
    - 새로운 정보가 업데이트되면 이를 검색하여 답변에 반영 가능  

2. 정보 정확성 향상  
    - 모델 내부 지식뿐 아니라 검색된 외부 자료를 근거로 답변 생성  
    - 검증 가능한 정보를 활용하여 환각 현상을 완화하고 답변의 신뢰성 향상 가능  


#### CAG(Credibility-aware Gerneration)  

> CAG? 모델이 신뢰도를 스스로 판단하고 활용할 수 있도록 훈련시키는 기법  

기존 RAG는 검색된 문서가 질문과 관련성이 높더라도 오래되었거나 부정확한 정보일 가능성이 있었음. CAG는 검색된 정보의 관련성뿐 아니라 출처의 신뢰성, 최신성 등을 고려하여 답변 생성  

-> 신뢰도가 낮은 정보의 영향은 줄이고, 신뢰도가 높은 정보를 우선활용하여 답변의 정확성 개선  

**CAG의 구조**  

1. 질문 입력 : 사용자의 질문을 분석하여 필요한 정보 파악  
2. 정보 검색 : 외부 데이터베이스에서 질문과 관련된 문서 검색  
3. 신뢰도 평가 : 검색된 정보의 출처, 최신성, 관련성 등을 종합적으로 평가  
4. 정보 선별 : 신뢰도가 높은 정보에 더 큰 비중을 두어 답변 생성에 활용  
5. 최종 답변 생성 : 신뢰성 높은 정보를 중심으로 최종 답변 생성  

  


### 4-6 Langchain  
 > LangChain? LLM을 외부 데이터, API, 다양한 도구와 연결하여 복잡한 AI 애플리케이션을 쉽게 개발할 수 있도록 지원하는 오픈소스 프레임워크  

 기존 LLM과 RAG를 구현할때는 구현의 복잡성, 외부 시스템 연동, 유지보수의 어려움 등이 존재  
 -> 이를 해결하고자 LLM과 다양한 기능을 모듈화 하고 연결하여 개발 효율성을 높이는 LongChain등장  

#### LongChain의 특징  

1. 추상화  
    - LLM 호출, 프롬프트 관리, 문서 검색 등 복잡한 기능을 간단한 인터페이스로 제공  
        - *e.g* 복잡한 API 호출 코드를 직접 작성하지 않고, 간단한 명령어로 ChatGPT에 질문하고 답변을 받을 수 있음  

2. 표준화  
    - 서로 다른 LLM, 데이터베이스, 외부 도구 등을 일관된 방식으로 사용할 수 있도록 지원  
        - *e.g.* ChatGPT 대신 Claude를 사용하고 싶을 때, 전체 코드를 다시 작성하지 않고 모델 연결 부분만 변경하여 활용가능  

3. 체이닝  
    - 여러 작업을 순서대로 연결하여 하나의 작업 흐름으로 구성  
        - *e.g.* 영어 문장 입력 -> 한국어로 번역 -> 번역된 문장 요약 -> 최종 결과 출력  

#### LangGraph  
> LangGraph? 여러 작업을 노드(Node)와 엣지(Edge)로 연결하여, 조건에 따른 분기와 반복이 가능한 AI 워크플로를 구현하는 프레임워크  

LangChain은 복잡한 AI시스템에선 작업결과에 따라 실행 경로가 달라지거나, 이전 단계로 돌아가 재실행해야하는 상황이 발생  
-> 이를 지원하기 위해 '조건부 분기', '반복 실행', '상태관리'가 가능한 LangGraph가 활용됨  

**LangGraph 핵심구조**  

  

1. User : 사용자의 질문이나 요청 입력
2. Router : 입력된 요청을 분석하여 적절한 Agent로 전달
3. Agent 1 / Agent 2 : 각 Agent가 담당 작업을 수행하고 필요시 다른 Agent와 정보 교환
4. Tool : 외부 검색, API 호출 등의 기능을 실행하여 작업에 필요한 정보 제공
5. 반복 및 최종 응답 : 작업 결과에 따라 다른 Agent나 Tool을 다시 호출하고, 필요한 처리가 완료되면 사용자에게 답변 전달  

-> LangGraph는 노드와 엣지를 활용하여 여러 Agent와 Tool을 유연하게 연결하고, 비선형적이고 동적인 작업 흐름을 구현 가능


#### LangSmith  
> LangSmith? LLM 기반 애플리케이션의 실행 과정을 추적하고, 성능을 평가하며, 운영 상태를 모니터링할 수 있도록 지원하는 개발 및 관리 플랫폼  

- LangChain, LangGraph 등으로 구현한 AI 애플리케이션의 동작을 분석하고 개선하는 데 활용  
- 모델의 입력과 출력, 실행 과정, 오류 등을 확인하여 애플리케이션의 신뢰성과 성능 향상 지원  

**핵심 기능**  
1. 디버깅 및 추적  
    - LLM의 입력, 출력과 실행 과정을 추적하여 오류의 원인을 파악하고 문제 해결 지원  

2. 테스트 및 평가  
    - 다양한 테스트 데이터를 활용해 모델의 정확성과 성능을 평가하고, 변경 전후의 결과를 비교하여 개선 효과 확인   

3. 모니터링  
    - 모델의 응답 시간, 오류 발생률, 사용량 및 비용 등을 모니터링하여 성능 저하를 조기에 발견하고 안정적인 운영 지원  

4. 프롬프트 허브  
    - 프롬프트를 저장, 공유하고 버전을 관리하며, 수정 전후의 결과를 비교하여 프롬프트 개선 지원  



## 5. sLM  

### 5-1 sLM(small Language Model)  

> sLM이란? 소규모 언어 모델(sLM)은 자연어를 처리하고 이해하며 생성할 수 있는 인공지능(AI) 모델, LLM보다 규모와 범위가 작음   

sLM은 LLM보다 작고 효율적이어서 메모리와 연산자원을 적게 사용  
-> 모바일 앱처럼 컴퓨팅 자원이 제한된 환경, 오프라인 추론 등에 활용 가능  

#### LLM vs sLM  

| 구분 | LLM | sLM |
| --- | --- | --- |
| 모델 크기 | 상대적으로 큼 | 상대적으로 작음 |
| 성능 | 복잡한 추론과 다양한 작업에 강점 | 특정 분야나 비교적 단순한 작업에 효율적 |
| 연산 비용 | 상대적으로 높음 | 상대적으로 낮음 |
| 응답 속도 | 많은 연산이 필요할 수 있음 | 같은 환경에서 빠른 처리에 유리 |
| 활용 환경 | 주로 고성능 서버 | 서버, 스마트폰, 노트북 etc... |


### 5-2 모델 압축   

> 모델압축 : 모델의 중요한 성능은 최대한 유지하면서 크기와 연산량을 줄이는 기술. 메모리 사용량과 계산 비용을 줄이고, 더 제한된 환경에서도 모델을 효율적으로 사용 가능  

#### 가지치기(Pruning)  

> 가지치기 : 신경망에서 중요도가 낮은 연결이나 뉴런을 제거하여 모델의 크기와 연산량을 줄이는 방법

**작동원리**  
- 학습된 신경망에서 결과에 미치는 영향이 작은 연결이나 뉴런 찾기
- 중요도가 낮은 부분을 제거하고, 핵심적인 연결만 남김  
- 필요에 따라 추가 학습을 진행하여 성능 저하 보완  

-> 모델의 크기와 연산량을 줄일 수 있으나, 중요한 부분까지 제거시 성능저하 가능성도 존재  

#### 양자화(Quantization)  

> 모델의 가중치 등을 표현할때 사용하는 숫자의 정밀도를 낮춰 메모리 사용량과 연산 비용을 줄이는 방법  

**작동원라**  
- 기존 모델은 가중치를 FP32(32비트)와 같은 높은 정밀도로 저장 가능  
- 이를 FP16(16비트)이나 INT8(8비트) 등 더 낮은 정밀도로 변환
- 숫자를 저장하는 데 필요한 비트 수가 줄어들어 모델이 더 가벼워짐  

  

-> 모델 크기와 메모리 사용량을 줄이고 연산 속도 개선이 가능하나, 정밀도가 낮아지며 일부 성능 저하 가능성 존재  

#### 저차원 분해(Low-Rank Factorization)

> 저차원 분해 : 모델의 크고 복잡한 가중치 행렬을 더 작은 행렬들의 곱으로 근사하여 파라미터 수와 연산량을 줄이는 방법

**작동원리**  
- 신경망의 많은 숫자로 구성된 큰 가중치 행렬을 더 작은 두개의 행렬로 나누어 표현  
- 전체 정보를 그대로 저장하는 대신, 중요한 특징을 적은 수의 파라미터로 표현  

-> 적은 파라미터로 중요한 정보를 표현하여 저장공간이나 연산량을 줄이는데 활용 가능   

#### 지식 증류(Knowledge Distillation)  

> 지식 증류 : 성능이 뛰어난 큰 모델(Teacher Model)의 지식을 작은 모델(Student Model)에 전달하여, 작은 모델도 비슷한 성능을 낼 수 있도록 학습시키는 방법  

**작동원리**  
- 먼저 학습이 완료된 큰 모델을 교사모델로 활용하여, 교사 모델의 예측 결과나 학습된 지식을 작은 모델에게 전달  
- 작은 모델은 교사 모델의 출력을 따라하도록 학습하며 중요 지식 습득  

-> 큰 모델의 지식을 활용하여 작은 모델의 성능 향상이 가능하며, 적은 자원으로도 효율적인 추론 가능  

### 5-4 sLM의 장점  

sLM은 LLM보다 모델 크기가 작아 빠른 처리 속도와 높은 비용 효율성을 확보하는 장점 존재  

1. 성능적 측면  
    - 빠른 응답 속도 : 모델 크기가 작고 처리해야 할 연산량이 적어, LLM보다 빠르게 결과를 생성 가능  
    - 제한된 환경에서도 실행 가능 : 상대적으로 적은 메모리와 컴퓨팅 자원으로 작동할 수 있어 스마트폰이나 노트북 등에서도 활용 가능  
    - 보안 및 데이터 보호 : 기업 내부 서버나 개인 기기에서 직접 실행 가능하여, 외부 클라우드 API로 데이터를 전송하지 않고도 활용 가능  

2. 비용적 측면  
    - 낮은 운영 비용 : 모델 실행에 필요한 연산량과 메모리가 적어 서버 운영 및 AI 서비스 제공 비용 감소  
    - 적은 전력 소비 : 대규모 모델보다 상대적으로 적은 에너지를 사용하므로 전력 소비 감축에 유리  
    - 하드웨어 비용 절감 : 고성능 GPU와 같은 비싼 장비에 대한 의존도를 낮춰 AI 기술을 쉽게 도입 가능  


### 5-5 sLM의 한계(단점)  

sLM은 효율성 측면에선 강점이 있으나, 모델 크기가 작고 학습가능한 정보에 제약이 있다는 한계 존재  

1. 편향(Bias)  
    - 편향된 답변 생성 : 학습 데이터에 포함된 사회적 편견이나 잘못된 정보가 답변에 반영될 수 있음. 데이터의 다양성이 부족하면 특정 집단이나 관점에 치우친 결과를 생성하기도 함  

2. 제한된 일반화 및 복잡한 작업에서의 성능 저하  
    - 낮은 범용성과 추론 능력 : 학습하지 않은 새로운 상황이나 복잡한 추론 및 전문적인 지식이 필요한 작업의 경우 LLM에 비해 정확도가 낮을 가능성 존재  

3. 환각 현상(Hallucination)  
    - 사실과 다른 정보 생성 : sLM도 LLM과 마찬가지로 충분한 근거가 없는 질문에 대해 실제로 존재하지 않거나 사실과 다른 정보를 사실처럼 생성할 수 있음  



### 5-6 sLM과 LLM의 결합  

> 결합추론(Hybrid Inference)이란? sLM과 LLM을 함께 활용하여 각 모델의 장점을 살리는 방식  

sLM은 빠르고 비용이 적게 들지만 복잡한 작업에서는 성능이 떨어지는 반면, LLM은 복잡한 문제 해결에 유리하지만 많은 컴퓨팅 자원과 비용이 필요  
-> 이러한 차이를 고려하여 작업의 난이도에 따라 적절한 모델을 활용하는 것이 결합 추론  

**작동방식**  
1. 간단한 작업 : 문장 분류, 간단한 요약 등은 빠르고 효율적인 sLM이 처리  
2. 복잡한 작업 : 여러 단계의 추론이나 전문적인 분석이 필요한 작업은 LLM이 처리  
3. 결과 : 모든 작업에 LLM을 사용하지 않아도 되므로 비용과 처리 시간을 줄이면서 필요한 수준의 성능 확보 가능  

=> sLM의 효율성과 LLM의 높은 문제 해결 능력을 결합하여 성능과 비용의 균형을 맞추는 방식!



## 6. 생성모델

### 6-1 분류모델 vs 생성모델  

> 생성모델(Generative Model)이란? 주어진 학습 데이터를 학습하고 학습 데이터의 분포를 따르는 유사 데이터를 생성하는 모델  

   

| 구분 | 분류 모델 | 생성 모델 |
| --- | --- | --- |
| 학습 방법 | 데이터의 특징과 정답 레이블의 관계 학습 | 데이터의 특징과 확률분포 학습 |  
| 학습 목표 | $P(Y\|X)$ 모델링 : 데이터 X가 주어졌을 때 정답 Y에 속할 확률을 학습하여 분류 | $P(X)$ 모델링 : 실제 데이터 X가 나타나는 확률분포를 학습하여 새로운 데이터 생성 |
| 주요 특징 | 데이터의 차이점을 학습하여 분류 기준 형성 | 데이터의 공통적인 특징과 다양한 패턴을 학습하며, 입력 데이터의 적합성을 평가하고 비정상적인 입력을 탐지하는 데 활용 가능 |
| 출력 결과 | 범주 또는 범주별 예측 확률 | 새로운 이미지, 텍스트, 음성 등|
| 한계 | 처음 보는 이상한 데이터도 기존에 학습한 라벨 중 하나로 잘못 분류할 수 있으며, 일반적인 분류 모델은 새로운 데이터 생성 불가능 | 부자연스러운 결과 생성 가능, 생성 결과의 품질 평가 어려움 |


### 6-2 생성모델의 분류  

#### 명시적 확률밀도 모델(Explicit Density Model)  

> 명시적 확률밀도 모델이란? 데이터의 확률분포를 수학적으로 정의하고, 특정 데이터의 P(X)의 값을 직접적으로 수치화하여 출력할 수 있는 생성 모델  

1. 다루기 쉬운 밀도 모델(Tractable Density Model)  
    - 데이터의 확률밀도를 수학적으로 직접 계산할 수 있는 모델  
    - 확률분포를 계산 가능한 형태로 구성하여, 특정 데이터가 학습한 분포에서 얼마나 그럴듯한지 정확한 수치로 표현 가능  
    - 대표모델 : 
        - NADE, MADE, PixelRNN/PixelCNN : 데이터의 각 요소에 대한 조건부 확률을 순차적으로 계산하는 모델  
        - Change of Variables Model (nonlinear ICA) : 변수 변환을 활용하여 복잡한 데이터의 확률밀도를 계산하는 모델

2. 근사적 밀도 모델(Approximate Density Model)  
    - 데이터의 확률밀도를 직접 계산하기 어려워 근사적인 방법을 활용하는 모델  
    - 복잡한 확률분포를 계산할 때 발생하는 어려움을 해결하기 위해 변분 추론이나 마르코프 연쇄 등의 방법을 활용  
    - 대표모델 : 
        - VAE(Variational Autoencoder) : 변분 추론을 활용하여 계산하기 어려운 확률분포를 근사적으로 학습하는 모델  
        - Boltzmann Machine : 마르코프 연쇄 등의 방법을 활용하여 데이터의 상태를 반복적으로 변화시키며 표본을 추출하여 확률분포를 학습하는 모델  


#### 암시적 확률밀도 모델(Implicit Density Model)  

> 암시적 확률밀도 모델이란? 데이터의 확률 분포 P(X)가 무엇인지 수치로 알려주지 않고, 학습한 데이터의 분포를 바탕으로 새로운 데이터를 생성(Sampling)하는 데 초점을 맞춘 생성 모델  

명시적인 수식 계산 대신, 데이터가 가지는 특징과 패턴을 학습  
학습 데이터와 유사한 새로운 데이터는 생성 가능하나, 특정 데이터의 확률 밀도 계산은 불가  

1. 직접 샘플링(Direct Sampling)  
    - 학습된 생성 모델에 무작위 노이즈를 입력하여 새로운 데이터를 직접 생성하는 방식  
    - 여러 단계의 반복적인 상태 변화 없이 생성 모델을 한 번 통과하여 결과 생성 가능  
    - 대표 모델 :  
        - GAN (Generative Adversarial Network) : 생성자(Generator)와 판별자(Discriminator)의 경쟁을 통해 데이터의 특징을 학습하고, 무작위 노이즈로부터 새로운 데이터를 생성하는 모델  

2. 간접/반복적 샘플링(Indirec/Iterative Sampling)  
    - 여러 단계의 확률적인 상태 변화를 반복하여 새로운 데이터를 생성하는 방식  
    - 한 번에 데이터를 생성하는 것이 아니라, 여러 단계의 상태 변화를 거쳐 학습한 데이터의 분포를 따르는 표본 생성  
    - 대표 모델 :  
        - GSN (Generative Stochastic Network) : 마르코프 연쇄의 확률적 상태 전이 과정을 학습하고, 이를 반복 적용하여 새로운 데이터를 생성하는 모델  


## 7. AE, VAE  

### 7-1 AE   

> AE(Auto Encoder)란? 입력 데이터의 중요한 특징을 압축하고, 이를 바탕으로 원래 데이터와 최대한 유사하게 복원하도록 학습하는 인공신경망 모델   

- 입력 데이터를 압축하고 복원하면서 데이터의 핵심적인 특징 학습  
- 입력 데이터와 복원된 데이터의 차이를 최소화하는 것이 목표  
- 별도의 정답 레이블 없이 입력 데이터 자체를 복원 목표로 활용 가능  

#### 기존 인코더, 디코더와의 차이?  

기존 인코더, 디코더는 입력 데이터를 이해하고 이를 바탕으로 원하는 결과를 생성했다면, AE의 인코더, 디코더는 입력 데이터의 특징을 압축하고 원래 대이터로 복원  
-> 기존 인코더, 디코더와 유사한 구조를 사용하지만, AE는 입력 데이터 자체를 복원하는것이 핵심 목적!  

#### AE의 구조 및 작동 과정  

AE는 인코더(Encoder) -> 병목(Bottleneck) -> 디코더(Decoder)의 구조, 입력 데이터를 압축한 뒤 다시 복원하는 과정을 거침  

  

1. 압축 단계(Encoder)  
    - 입력 데이터에서 중요한 특징을 추출하고 정보를 압축하는 단계  
    - 인코더는 입력 데이터(x)를 더 적은 수의 값으로 표현하여 잠재 벡터(z)로 변환  
        - *e.g.* 손글씨 숫자 '4'의 이미지에서 숫자의 형태, 획 등 중요한 특징 추출    

2. 병목 통과(Bottleneck)  
    - 인코더가 추출한 핵심 정보를 잠재 벡터(Latent Vector, z)로 표현하는 단계  
    - 제한된 공간에 정보를 담기 때문에 데이터의 핵심적인 특징을 학습하도록 유도  

3. 복원 단계(Decoder)  
    - 압축된 잠재 벡터(z)를 활용하여 원본 데이터와 유사한 데이터를 복원하고, 원본과 복원 결과를 비교하여 중요한 정보가 얼마나 잘 보존되었는지 확인하는 단계  
    - 디코더는 잠재 벡터를 입력받아 복원 데이터 (x')를 생성  
        - *e.g.* 압축된 손글씨 특징을 바탕으로 숫자 '4' 이미지를 다시 복원  

**왜 데이터를 압축할까?**  
만일 데이터를 압축하지 않고 그대로 전달시, 모델은 중요한 특징을 충분히 학습하지 않고, 입력 데이터를 단순히 복사하는 방식으로 복원  
But, 잠재공간의 크기를 제한하면 모든 정보를 담기 어렵기에 모델은 어떤 정보가 원본을 복원하는데 가장 중요한지 학습해야 함!  

**왜 다시 복원할까?**  
압축된 정보만으로도 원본과 유사한 데이터를 만들어낼 수 있는지 확인하기 위해서  
모델은 원본 이미지와 복원된 이미지의 차이인 복원 오차(Reconstruction Error)를 줄이도록 학습하며, 이를 통해 인코더와 디코더가 중요한 특징을 더욱 효과적으로 표현하도록 개선됨  

=> 이러한 과정을 통해 AE는 데이터의 핵심적인 특징과 구조(Non-trivial Structure)를 학습  


### 7-2 VAE  

기존 AE는 입력 데이터를 압축하고 복원하는데 초점을 맞추어서, 잠재공간이 불규칙하게 형성된다는 한계가 존재. 이를 해결하기 위해 데이터를 고정된 값이 아닌 확률분포로 표현하고, 잠재공간을 규칙적으로 구성하는 VAE도입  
-> 기존 데이터 복원뿐아니라 확률분포에서 새로운 값을 샘플링하여 다양한 데이터 생성이 가능해짐  

> VAE(Variational Autoencoder)란? 입력 데이터의 특징을 확률분포로 학습하고, 해당 분포에서 샘플링하여 새로운 데이터를 생성할 수 있는 생성 모델  

- 기존 AE의 인코더, 디코더 구조를 기반으로 만들어진 모델로 데이터를 단순히 압축하고 복원하는 것을 넘어, 데이터의 확률적 특성을 학습하여 새로운 데이터를 생성 가능  
- 변분 추론(Variational Inference)을 활용하여 복잡한 확률분포를 근사적으로 학습  


#### VAE의 학습 목표  

1. 복원 오차(Reconstruction Loss)최소화  
    - 원본 데이터와 복원 데이터 사이의 차이를 최소화하여 원본의 특징을 최대한 잘 복원하도록 학습  

2. KL발산(KL Divergence) 최소화  
    - 인코더가 예측한 잠재분포가 표준정규분포에 가까워지도록 학습  

-> 잠재공간이 불규칙하게 형성되는 문제를 완화하고, 새로운 데이터를 생성하기 좋은 잠재공간 형성을 유도  

#### AE와 VAE의 차이  

AE와 VAE의 가장 큰 차이는 입력 데이터의 특징을 잠재공간(Latent Space)에 어떤 형태로 표현하고 배치하도록 학습하는가  

- AE : 입력 데이터를 하나의 고정된 잠재 벡터로 표현, 잠재공간에 데이터가 불규칙하게 배치  
- VAE : 입력 데이터를 확률분포로 표현, 잠재공간이 일정한 분포를 따르도록 학습하여 새로운 데이터를 생성하기 쉽게 함  

> 잠재공간(Latent Space)이란? 입력 데이터에서 추출한 핵심적인 특징을 숫자 벡터로 표현하는 공간(비슷한 데이터는 잠재공간에서 비슷한 위치에)  

- AE의 잠재공간 : AE는 입력 데이터의 특징을 하나의 고정된 잠재벡터로 표현, 동일한 입력 데이터에 대해 일반적으로 동일한 잠재 벡터 생성  
- VAE의 잠재공간 : VAE는 입력 데이터를 고정점이 아닌 확률 분포 형태로 잠재공간에 표현, 입력 데이터마다 평균과 분산을 계산하여 해당 데이터의 특징을 나타내는 확률 분포를 학습시키고, 잠재분포가 표준정규분포에 가까워지도록 학습하여 잠재공간이 연속적이고 규칙적으로 형성되도록 유도  

  

왼쪽의 AE의 경우 잠재벡터들이 불규칙하게 분포하고, 데이터 사이에 빈 영역이 발생하는 반면, 우측 그림의 VAE의 경우 잠재분포가 일정한 확률분포에 가까워지도록 학습되어 데이터 사의의 공간에서도 새로운 잠재 벡터를 샘플링하기 쉬움


#### VAE의 구조와 작동 과정  

1. 분포 예측 단계(Probabilistic Encoder) : 입력 데이터의 특징을 분석하여 확률분포를 예측하는 단계  
    - 입력 데이터 x를 확률적 인코더(Probabilistic Encoder)에 전달하면, 인코더는 데이터의 특징을 분석하여 평균과 표준편차를 예측  
    - 예측된 평균과 표준편차를 이용하여 입력 데이터의 특징을 나타내는 확률분포를 형성  

2. 샘플링 및 재파라미터와(Sampling & Reparametrization) : 예측한 확률분포에서 잠재벡터(z)를 추출하는 단계  
    - 앞서 예측한 확률분포에서 잠재 벡터(z)를 무작위로 샘플링  
    - 같은 확률분포에서도 서로 다른 값이 추출될 수 있기 때문에 다양한 형태의 데이터가 생성  

3. 생성단계(Probabilistic Decoder) : 샘플링한 잠재벡터를 이용하여 데이터를 복원하거나 새롭게 생성하는 단계  
    - 샘플링한 잠재 벡터(z)를 확률적 디코더에 전달하면, 디코더는 잠재 벡터를 바탕으로 데이터의 확률분포를 예측하고, 이를 이용해 이미지를 복원하거나 생성  
    - 학습 과정에선 원본 데이터 x와 복원 데이터 x′이 최대한 유사해지도록 학습



## 8. GAN  

### 8-1 GAN의 개념  

> GAN(Generative Adversarial Network)이란? 생성자(Generator)와 판별자(Discriminator)라는 두 신경망이 서로 경쟁하면서 실제 데이터와 유사한 새로운 데이터를 생성하도록 학습하는 생성 모델   

생성자와 판별자라는 두개의 신경망으로 구성되며, 생성자는 진짜같은 가짜 데이터를 만들고, 판별자는 입력된 데이터가 진짜인지 가짜인지 구별하는 역할 수행  
-> 두 신경망이 서로 경쟁하며 학습하기 때문에 적대적(Adversarial)생성망이라고 부름  

- 생성자(Generator, G)  
    - 무작위 노이즈를 입력받아 실제 데이터와 유사한 가짜 데이터를 생성  
    - 판별자가 가짜 데이터를 진짜라고 판단하도록 학습

- 판별자(Discriminator, D)  
    - 실제 데이터와 생성자가 만든 가짜 데이터를 입력받음  
    - 입력된 데이터가 진짜일 확률을 예측하여 두 데이터를 구별하도록 학습  


#### GAN의 목적  

1. 생성자의 목적  
    - 판별자가 진짜라고 판단할 만큼 실제 데이터와 유사한 가짜 데이터를 생성하는 것  
    - 학습이 진행될수록 실제 데이터의 특징을 더 잘 반영하도록 개선됨  

2. 판별자의 목적  
    - 실제 데이터와 생성자가 만든 데이터를 최대한 정확하게 구별하는 것  
    - 생성자의 이미지가 정교해질수록 이를 구별하는 능력도 함께 학습  

3. GAN의 궁극적인 목적  
    - 생성자와 판별자의 경쟁을 통해 생성자가 실제 데이터의 분포를 학습하도록 만드는 것  
    - 최종적으로 실제 데이터와 구별하기 어려울 만큼 자연스러운 새로운 데이터를 생성하는 것이 목표   


### 8-2 GAN의 구조  

GAN은 생성자(Generator)와 판별자(Discriminator)가 서로 반대되는 목표를 가지고 경쟁하며 학습하는 구조  = 'MiniMax Game' (게임이론)  

  

#### GAN의 구조  

1. 무작위 노이즈 (Random Noise)  
    - 생성자에 입력되는 무작위 숫자 벡터(z)를 의미  
    - 생성자는 이 노이즈를 바탕으로 새로운 강아지 이미지를 생성  

2. 생성자 (Generator)   
    - 입력 : 무작위 노이즈(z)  
    - 출력 : 생성된 가짜 강아지 이미지(G(z))  
    - 목표 : 판별자가 진짜라고 판단할 만큼 실제 강아지와 유사한 이미지 생성  

3. 판별자 (Discriminator)  
    - 입력 : 실제 강아지 이미지(x) 또는 생성자가 만든 가짜 강아지 이미지(G(z))  
    - 출력 : 입력된 이미지가 진짜일 확률(0 ~ 1)  
    - 목표 : 실제 강아지 이미지와 생성된 가짜 강아지 이미지를 정확하게 구별하는것  

-> 판별자는 실제이미지에는 높은 확률을, 가짜 이미지에는 낮은 활률을 부여하도록 학습  


### 8-3 GAN의 장단점  

#### 장점  

1. 진짜같은 가짜 생성 가능  
    - 실제 데이터의 특징과 분포를 학습하여 현실에 존재하지 않지만 실제처럼 보이는 새로운 데이터 생성 가능  

2. 높은 이미지 선명도  
    - 판별자의 평가를 바탕으로 생성자가 학습하기 때문에 이미지의 세부적인 특징을 정교하게 표현 가능  

3. 빠른 생성 속도  
    - 학습이 완료되면 무작위 노이즈를 생성자에 한 번 통과시키는 방식으로 이미지 생성 가능, 여러 단계를 반복하는 일반적인 확산 모델과 비교하여 빠르게 이미지 생성 가능  

4. 잠재공간의 부드러움 (Smooth Latent Space)  
    - 잘 학습된 GAN은 잠재 벡터를 조금씩 변화시켰을 때 생성된 이미지의 특징도 자연스럽게 변화 가능  
    **다만 모든 GAN에서 이러한 특성이 보장되는 것은 아님** 


#### 한계점(단점)  

1. 학습의 불안정성(Training Instability) : 생성자와 판별자가 서로 경쟁하는 과정에서 두 신경망의 학습 균형이 무너져 안정적인 학습이 어려워지는 문제  
    - 문제점  
        - 생성자와 판별자가 서로 경쟁하는 구조이기 때문에 학습 과정이 불안정해질 수 있음  
        - 두 신경망의 학습 속도나 성능 차이가 지나치게 커지면 균형이 무너질 수 있음  
    - 해결 방법  
        - 생성자와 판별자의 학습률 및 학습 횟수를 조절하여 균형 유지  
        - WGAN-GP와 같이 학습 안정성을 개선한 GAN 모델 활용  

2. 훈련 초기 문제(Early Training Problem) : 학습 초기에 판별자가 가짜 데이터를 너무 쉽게 구별하여 생성자가 충분한 학습 신호를 받지 못하는 문제  
    - 문제점 
        - 학습 초기에는 생성자가 만드는 이미지의 품질이 매우 낮아 판별자가 가짜 이미지를 너무 쉽게 구별 가능  
    - 해결 방법  
        - Non-saturating Loss: 생성자가 학습 초기에 더 강한 학습 신호를 받을 수 있도록 손실함수를 변경  
        - 판별자가 지나치게 빠르게 학습하지 않도록 학습률이나 업데이트 횟수 조절  

3. 모드 붕괴(Mode Collapse) : 생성자가 다양한 데이터를 생성하지 못하고 특정한 형태의 데이터만 반복적으로 생성하는 문제  
    - 문제점 : 생성자가 다양한 데이터를 생성하지 못하고 특정한 형태의 데이터만 반복적으로 생성하는 현상  
    - 해결 방법 :  
        - Minibatch Discrimination : 여러 생성 이미지를 함께 비교하여 판별자가 이미지의 다양성까지 고려하도록 학습  
        - WGAN-GP 등 학습을 개선하는 방법을 활용하여 모드 붕괴 현상을 완화 가능  

4. 평가 지표의 한계(Evaluation Metrics) : 생성된 데이터의 사실성, 선명도, 다양성 등을 하나의 지표로 완벽하게 평가하기 어려운 문제  
    - 문제점 : 생성된 이미지의 품질을 하나의 지표만으로 완벽하게 평가하기 어렵고, 이미지가 얼마나 진짜 같은지뿐만 아니라 얼마나 다양한지도 평가해야 함  
    - 해결 방법 : 
        - FID(Frechet Inception Distance) : 실제 이미지와 생성 이미지의 특징 분포가 얼마나 유사한지 평가  
        - IS(Inception Score) : 생성 이미지의 분류 확신도와 다양성을 함께 평가하는 지표  

5. 역매핑의 어려움(Inverse Mapping) : 기본 GAN은 잠재 벡터를 이미지로 변환하는 방향으로 학습하기 때문에, 반대로 실제 이미지에 해당하는 잠재 벡터를 찾기 어려운 문제  
    - 문제점 : 기본적인 GAN은 잠재 벡터(z) -> 생성 이미지(x')의 단방향 구조로 학습, 따라서 실제 이미지가 주어졌을 때, 이를 생성할 수 있는 잠재 벡터를 직접 찾아내기 어려움  
    - 해결 방법 : 
        - GAN Inversion : 실제 이미지를 생성할 수 있는 잠재 벡터를 역으로 추정하는 방법  
        - BiGAN(Bidirectional GAN) : 이미지에서 잠재 벡터를 추정하는 인코더를 추가하여 역방향 학습도 수행하는 모델  


## 9. 확산 모델  

### 9-1 확산모델(Diffusion Model)의 개념  

> 확산모델이란? 이미지에 노이즈(잡음)를 추가한 데이터를 학습하고, 무작위 노이즈를 점차 제거하면서 새로운 이미지를 생성하는 생성형 AI 모델  

**기본 원라**  
- 실제 이미지에 노이즈를 조금씩 추가하며 이미지가 흐트러지는 과정을 만듦  
- 모델은 노이즈가 포함된 이미지에서 노이즈를 제거하는 법 학습  
- 학습 완료시 무작위 노이즈에서 시작해 새로운 이미지 생성 가능  

#### 확산 모델의 목적  

확산 모델의 궁극적인 목표는 실제 데이터의 특징과 분포를 학습하여, 새로운 데이터를 자연스럽고 다양하게 생성하는 것  


### 9-2 확산 모델의 구조  

#### 순확산(Forward Diffusion)  

> 순확산이란? 원본이미지에 가우시안 노이즈(잡음, Gaussian Noise)를 조금씩 추가하여, 최종적으로 무작위 노이즈에 가까운 상태로 만드는 과정  

- 가우시안 노이즈 : 정규분포를 따르는 무작위 잡음  

**목표** 
이미지에 다양한 수준의 노이즈를 추가하여, 모델이 이후 역확산 과정에서 노이즈를 제거하는 방법을 학습하도록 하는것  


**작동과정**  

  


1. 시작단계 - 원본이미지($x_0$)  
    - 순확산은 노이즈가 없는 깨끗한 원본 데이터에서 시작  

2. 노이즈 추가 단계 - 점진적인 변화($x_1, x_2, ...$)  
    - 원본 이미지에 가우시안 노이즈를 조금씩 추가, 단계가 진행될수록 노이즈가 누적되어 이미지가 점차 흐려짐  

3. 결과단계 - 무작위 노이즈($x_T$)  
    충분한 단계를 거치면 원본 이미지의 특징과 정보가 거의 사라지고, 최종적으로 무작위 노이즈에 가까운 상태($x_T$)가 됨. 이렇게 생성된 여러 단계의 데이터는 이후 역확산 모델이 노이즈를 제거하고 새로운 이미지를 생성하는 방법 학습시에 활용  


#### 역확산(Reverse Diffusion)  

> 역확산이란? 무작위 노이즈에서 시작하여 노이즈를 단계적으로 제거하면서 새로운 이미지를 생성하는 과정  

**목표** 

순확산으로 만든 잡음 데이터로 신경망이 학습한 노이즈 제거 방법을 활용하여, 무작위 노이즈를 실제와 유사한 자연스러운 이미지로 바꾸는것  


**작동과정**  

  


1. 시작단계 - 무작위 노이즈($x_T$)  
    - 역확산은 무작위 노이즈 상태에서 시작. 새로운 이미지를 생성하기 위한 초기상태  

2. 노이즈 제거 단계 - 이미지 생성(Denosing U-Net)  
    - 학습된 신경망인 Denoising U-Net을 이용하여 현재 이미지에 포함된 노이즈를 예측하고 이를 바탕으로 노이즈를 조금씩 제거. 이 과정을 반복하며, 이미지의 색상과 윤곽이 점차 나타나고 구체적인 형태를 갖추게 됨  

3. 결과단계 - 새로운 이미지($x_0$)  
    - 반복적인 노이즈 제거 과정이 끝나면, 무작위 노이즈가 자연스러운 이미지로 변하며 최종적으로 새로운 이미지가 생성됨  


### 9-3 확산모델의 장단점  

#### 장점  
1. 압도적인 샘플 품질  
    - GAN은 노이즈에서 이미지를 한 번에 생성하는 반면, 확산 모델은 노이즈를 단계적으로 제거하면서 GAN보다 더 정교하고 다양한 이미지를 생성하는데 유리  

2. 학습의 안전성  
    - GAN처럼 생성자와 판별자가 서로 경쟁하는 방식이 아니라, 노이즈를 예측하고 제거하는 방법을 학습. 이떄, MSE를 통해 실제 노이즈와 예측 노이즈의 차이를 계산하며, 손실곡선을 통해 학습과정의 변화를 수치로 확인 가능  
    -> GAN에 비해 학습과정이 비교적 안정적  

3. 다양성 확보  
    - 무작위 노이즈에서 시작하므로 같은 조건에서도 서로 다른 이미지를 생성 가능. 이를 통해 특정 결과에 치우치지 않고 다양하고 자연스러운 결과물 생성 가능  


#### 단점  
1. 느린 생성 속도  
    - GAN은 노이즈에서 이미지를 한 번에 생성하여 속도가 빠른 반면, 확산 모델은 여러 단계에 걸쳐 노이즈를 반복적으로 제거해야 하기에 이미지 생성 시간이 오래걸리며, 빠른 응답이 필요한 환경에선 한계 존재  

2. 높은 계산 비용  
    - 학습 과정에서 많은 데이터와 연산이 필요하며, 이미지 생성 시에도 신경망을 반복적으로 실행해야 하여 GPU와 같은 고성능 하드웨어가 필요하고, 시간과 자원 소모가 큼  

3. 구조적 제약  
    - 순확산과 역확산이라는 단계적인 구조에 의존하기 때문에, 노이즈 제거 단계와 생성 과정을 자유롭게 변경하기 어려움. 특히, 노이즈 제거 단계의 수나 방식에 따라 생성 속도와 품질이 달라지므로, 이를 적절하게 설계해야 함
