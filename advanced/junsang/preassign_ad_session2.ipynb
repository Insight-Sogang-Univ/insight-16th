> 통게는 왜 필요한가?  
예를 들어, 우리나라 흡연율을 파악할때 우리나라 전 인구(모집단)를 파악하는것이 베스트이나 이는 현실적으로 쉽지 않다.  
따라서, 일부 샘플(표본집단)만 추출하여 분석하고 이를 모집단에 적용!  
  

### 자료의 종류와 척도의 차이?  
자료의 종류와 척도는 서로 관련된 개념이지만, 정보를 바라보는 관점이 다르다.  
자료의 종류는 데이터의 본질적 성격을 구분하고, 척도는 그 데이터를 측정하는 방식의 규칙을 구분한다.


## 자료의 척도  
> 자료 : 설명과 해석을 위해 수집되고 분석되며, 요약되는 사실과 수치  
 1. 양적 자료 : 수치로 표현되는 자료
    - 구간(등간)척도 : 값들 사이에 순서가 있고, 그 간격이 일정하여 덧셈이나 평균같은 연산이 가능한 척도  
    *ex) 온도(섭씨, 화씨), 시간*  
    **단, 곱셈이나 나눗셈은 의미를 지니지 못한다** ?
     

    - 비율척도 : 절대영점과 임의의 측정단위로 구성하고, 동간성이 유지되는 척도  
    *ex) 거리, 무게, 시간*
        -> 0 값은 그 변수에 대해 아무것도 존재하지 않는 상태 

 2. 질적 자료 : 수치로 표현되더라도 연산의 의미를 가지지 못하는 자료
    - 명목척도 : 레이블이나 이름으로 요소의 속성을 식별하려는 척도  
    *ex) 펀드 유형(국내주식 : 1, 국제주식 : 2, 채권 :3)*
    - 서열척도 : 명목형의 분류 속성 + 항목간의 순서 또는 순위를 나타내는 특성 척도  
    *ex) 만족도(좋음, 보통, 나쁨)*

 
## 기술 통계  
 > 기술통계 : 수치와 시각화를 통해 샘플을 요약하여 표헌하는 방법론 *ex) 시험점수 분포*
 - 기술통계의 3요소  
 1. 중심화 경향 : 데이터의 중심적인 경향을 나타내는 통계랑  
    - 평균(mean) : 데이터의 총합 / 갯수 **총합을 사용하기에 이상치(Outlier)에 민감할 수 있다**  
    -> 절사 평균 : outlier제거를 위해 데이터의 max, min value를 제외한 평균
    - 중앙값(median) : 데이터를 크기순으로 정렬했을때 가운데에 위치한 값  
    - 최빈값(mode) : 데이터에서 가장 자주 나타나는 값  

 2. 자료의 산포 : 데이터가 퍼져있는 정도를 설명하는 통계량  
    - 범위(Range) : 데이터의 Maximum과 Minimum의 차이를 나타내는 지표, 전체 변동 범위를 나타낸다 **양 극단값을 사용하기에 Outlier에 민감할 수 있다**  
    - 분산(Variance) $\frac{\sum_{i=1}^{N}(x_i-\mu)^2}{N}$ : 데이터 값들이 평균으로부터 얼마나 멀리 퍼져있는지 나타내는 통계지표 
       -> 분산이 크다 : 데이터가 많이 퍼져있다
    - 표준편차(Standard Deviation) $\sqrt{variance}$ : 분산에 제곱근을 취한값  
        **Why $\sqrt{var}$ ?** : 분산 계산이 자료의 제곱을 취하기에 단위가 sample과 맞지 않게 되고, 이를 보정해주기 위하여 루트를 취한다.  
    - 사분위수(quartile) : 데이터의 표본을 4개의 동일한 크기로 나눈값  
        - Q1(lower quartile) : 하위 25%, Q2 : 하위 50%, Q3(upper quartile) : 하위 75%(상위 25%)  
          

    - 사분위간 범위(IQR) : Q3에서 Q1을 제외한 범위 *i.e. 데이터의 중앙 50%*  
    - 변동계수(Coefficient of Variation) : 데이터의 절대적인 변동성이 아닌 상대적인 변동성을 나타내는 통계적 지표  
        $변동계수 = \frac{Std.dev}{mean}*100$

 3. 자료의 정리 및 시각화 : 히스토그램, 박스플롯, 파이차트 etc... 


## 확률분포  
 - 확률변수 : random한 세계에서 어떤 특정 사건이 발생할 때 나타날 수 있는 모든 경우의 수를 특정 값으로 대응시키는 함수  
   *ex) Coin toss에서 H=1, T=0*  
 - 확률분포(Probability Distribution) : 확률변수가 특정한 값을 가질 확률을 나타내는 함수  
 이때, 확률변수의 값이 이산적(유한하거나 셀수 있는 경우)이면 이산확률분포, 연속적(무한한 범위)이면 연속확률분포라고 한다.  
 
    - 이산확률분포의 예 : Bernoulli Distribution, Binomial Distribution, Poisson Distribution , Geometric Distribution
    - 연속확률분포의 예 : Normal Distribution, Chi-squared Distribution, T-Distribution, Gamma Distribution  
 ### 대표적인 분포 모형
  - 정규분포 : 종모양, 평균을 기준으로 대칭, 표준편차가 커질수록 데이터가 넓게 펴지는 형태  **+)Z분포 : 정규분포를 평균이0, 표준편차가 1이 되도록 표준화 한 정규분포** 
    

  - 카이제곱 분포 : 독립적인 정규분포를 따르는 변수들의 제곱의 합(이때 합한 변수의 개수 = 자유도)  
  

  - T-분포 : 모집단의 분산을 모를때 표본의 분산을 사용해 모분산을 추정하여 만든 분포(샘플수가 적을때 사용)  
    

  - F-분포 : 두 **독립**적인 카이제곱분포의 비율, 두 집단간의 분산차이를 비교할때 사용  
    



## 추론통계  
> 추론통계 : 추정과 검정을 통해 샘플로부터 모집단을 추론하는 방법 *ex) 시청률 조사*  

### 추정  
> 추정 : 모집단을 대표할 수 있는 일부를 표본으로 추출하여 분석한 통계량을 이용해서 모집단의 모수를 예측하는 과정  
1. 점 추정 : 어떤 모수를 표본에 기초하여 계산된 하나의 값으로 추정하는 것
2. 구간추정 : 모수의 참 값이 포함되리라고 기대하는 추정치를 일정한 범위와 신뢰도를 가지고 나타내는 방법  

### 검정(가설검정과 과정)  
> 가설 검정 : 모집단의 실제값이 얼마가 된다는 주장과 관련해, 표본의 정의를 사용해서 가설의 합당성 여부를 판정하는 과정  
> 과정 : 가설 설정 -> 유의 수준 설정 -> 검정통계량 산출 -> 가설 기각/채택 판단  
1. 가설 설정 : 귀무가설 $H_0$(*주로 기각하고자 하는 명제*)와 대립가설 $H_1$(*내가 주장하고자 하는 명제*)을 설정  
2. 유의 수준 설정 : **유의수준**이란? 연구자가 결과를 판단할 때 허용할 수 있는 오류의 최대치 *주로 5%*  
    1종 오류($\alpha$) : 귀무가설이 실제로 참이었는데 기각할 확률  
    2종 오류($\beta$) : 귀무가설이 실제로 거짓이었는데 채택할 확률  
    | | $H_0$가 사실 | $H_0$가 거짓 |  
    |---|---|---|
    | $H_0$채택 | 옳은 결정, 확률($1-\alpha$) | 제 2종오류, 확률($\beta)|
    | $H_0$기각 | 제 1종오류, 확률($\alpha$) | 옳은 결정, 확률($1-\beta$)|  

    유의 수준은 1종 오류의 확률을 의미, 검정력은 2종 오류의 확률을 의미한다.  
3. 검정통계량 산출 : 검정통계량이란? 가설 검정의 결과를 판단하는데 사용되는 값  
    데이터의 분포나 특징에 따라  
    - Z값(Z-score) : 표준 정규분포를 사용하여 얻은 값, 표본의 크기가 클 때(30이상) 사용  
    - T값(T-score) : T-distribution을 사용하여 얻은 값, **모집단의 평균**에 대한 가설 검정이나 **두 표본 평균간의 차이**를 검정할때 주로 사용  
    - F값(F-score) : F-distribution을 사용하여 얻은 값, 주로 두 표본 분산 사이의 차이를 비교할 때 사용  
4. 가설 기각/채택 판단 : 유의 확률(p-value) < 유의 수준($\alpha$) 이면 Reject $H_0$, p-value > $\alpha$ 이면 Do not Reject $H_0$   

### 자료형에 따른 검정 방법  
1. 수치형 - 수치형 : (피어슨)상관분석  
  > 피어슨 상관분석 : 두 변수간 피어슨 상관계수가 얼마나 유의미한지 검정하는 방법  
  - $H_0$ : 두 변수간 선형 상관관계가 존재하지 않는다  
  - $H_a$ : 두 변수간 선형 상관관계가 존재한다  
    -> 공분산과 상관계수를 계산하여 |상관계수|<0.3 이면 약한 상관관계, |상관계수|<0.5 이면 중간 상관관계, |상관계수|>0.5 이면 강한 상관관계  

2. 범주형 - 수치형 : F-test(ANOVA), T-test  
    - T-test : 범주가 2개일 때 사용, 이때 t-value = 두 평균간 차이를 표준오차로 나눈 값
          - $H_0$ : 두 그룹의 평균에는 차이가 없다  
          - $H_a$ : 두 그룹의 평균에는 차이가 있다  
        일반적으로 t-value의 절댓값이 2보다 클 때, p-value가 유의수준보다 작을때를 기준으로 한다  
    - F-test(ANOVA) : 범주가 3개 이상일 때 사용, 이때 F-value = $\frac{집단 간 분산}{집단 내 분산}$ (*i.e. 분산의 비율*)  
        일반적으로 F-value의 값이 2~3 이상이면 reject $H_0$  
    **주의!** ANOVA는 전체 평균 대비 각 그룹간 차이가 있는지만 알 수 있다. 즉, 모두가 서로 다르다가 아니라 하나가 튀고 나머지가 비슷하더라도 귀무가설을 기각하는 경우가 발생!  
    -> 사후 분석을 진행하여 어떤 범주가 튀는지 파악하는 작업 진행  

3. 범주형 - 범주형 : Chi^2-test  
    기대빈도와 관측빈도의 차이를 가지고 검정 진행  
    $ \chi^2 = \sum \frac{(O_i-E_i)^2}{E_i}$ , $O_i$ : observed value(관측빈도), $E_i$ : Expected value(기대빈도)  
    이때, $\chi^2$ 통계량의 해석은 주로 자유도의 2배를 기준으로 잡는다. 즉, 이보다 크면 차이가 존재한다로 해석   
    *범주형 자유도(df) = '범주의 수'-1 & $\chi^2$ 변수에서의 자유도 = (df of x) * (df of y)*

4. 수치형 - 범주형 : 분류세션에서 다룰 예정
