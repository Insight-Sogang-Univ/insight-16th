# 추천 시스템 기초

## 추천 시스템과 데이터 유형

### 추천 시스템이란?

<br>

_사용자의 행동 기록과 아이템 정보를 바탕으로, 사용자가 좋아할 만한 아이템을 예측해 골라 주는 시스템_

- 사용자의 행동 기록과 아이템 정보를 분석하여 개인의 취향을 파악
- 사용자가 관심을 가질 만한 상품, 영화, 음악 등의 아이템을 예측하여 추천
- 예시 : 넷플릭스에서 영화를 시청하고 평가한 기록을 바탕으로 새로운 영화 추천

### 추천 시스템이 필요한 이유

**고객 입장**
- 찾는 시간과 노력이 줄어듦
- 몰랐던 취향의 아이템도 발견할 수 있음
- 검색은 사용자가 이미 알고 있는 것을 찾게 돕고, 추천은 아직 모르는 것을 발견하게 도움

**기업 입장**
- 이용 시간, 구매 전환, 재방문을 늘릴 수 있음
- 잘 안 보이던 비인기 아이템(롱테일)을 노출할 수 있음

### 추천 시스템에 사용되는 데이터의 종류

추천을 하려면 먼저 사용자가 무엇을 좋아하는지 알아야 함. 선호 정보를 어떻게 얻느냐에 따라 데이터는 크게 두 가지로 나뉨.

#### 명시적 데이터 (Explicit Data)

_고객들에 의해 입력된 데이터로, 사용자가 선호를 직접 표현한 데이터_

- 예시 : 별점, 좋아요, 싫어요, 리뷰

**장점**
- 비교적 분명한 선호와 불호를 통해 사용자의 취향을 정확히 파악할 수 있음

**단점**
- 평가를 남기는 사람 자체가 적고, 남기더라도 본 것 중 일부에만 남기므로 명백한 데이터를 얻기 어려움
- 사람의 취향은 변할 수 있으므로 과거의 선호도가 현재와 동일하지 않을 수 있음
- 사람마다 점수를 주는 기준이 다름

#### 암시적 데이터 (Implicit Data)

_고객들의 행동을 기반으로 수집된 데이터로, 행동에서 선호를 간접적으로 추측하는 데이터_

- 예시 : 조회수, 클릭, 구매, 시청 시간, 검색

**장점**
- 사용자가 따로 입력하지 않아도 대량으로 쌓여 얻기 쉬움

**단점**
- 행동이 곧 선호라고 단정할 수 없음
- 행동이 있어도 잘못 클릭했거나 선물로 구매한 것일 수 있어 좋아한다는 보장이 없음
- 행동이 없어도 아이템을 아직 못 봤을 수 있어 싫어한다는 보장이 없음

> 실제 서비스에서는 암시적 데이터가 훨씬 많아서 두 가지를 섞어 쓰는 경우가 대부분임.

## 추천 시스템 분류

 <br>

추천 시스템은 무엇을 근거로 추천하느냐에 따라 크게 세 가지로 나눌 수 있음.

- 내용 기반 필터링 (Content-Based Filtering)
- 협업 필터링 (Collaborative Filtering)
- 하이브리드 필터링 (Hybrid Filtering)

### 내용 기반 필터링 (Content-Based Filtering)

_콘텐츠의 유사도를 기반으로 사용자에게 추천하는 방법_

- 사용자가 이용하거나 선택한 아이템과 특징이 비슷한 아이템을 추천
- 사용자가 선호하는 아이템의 특징을 분석하여 비슷한 특징을 가진 아이템을 추천
- 예시 : SF 영화를 자주 본 사용자에게 다른 SF 영화 추천

### 협업 필터링 (Collaborative Filtering)

_서비스 내에 있는 다른 사용자의 행동과 평가 데이터를 함께 활용하여 추천하는 방법_

- 두 사용자의 취향이 비슷하다면, 한 사람이 좋아한 아이템을 다른 사람도 좋아할 가능성이 높다고 보는 방식

협업 필터링은 무엇을 기준으로 비슷함을 찾느냐에 따라 두 가지로 나뉨.

- 사용자 기반 (User-Based Filtering) : 사용자의 선호와 비슷한 사용자가 선호한 아이템 추천
- 아이템 기반 (Item-Based Filtering) : 사용자가 선호한 아이템과 유사한 아이템 추천

### 하이브리드 필터링 (Hybrid Filtering)

_두 가지 이상의 다양한 추천 방식의 알고리즘을 조합하여 만들어진 방식_

- 콘텐츠 기반 필터링과 협업 필터링을 결합
- 콘텐츠 기반 필터링은 아이템의 특징을 활용하고, 협업 필터링은 사용자 간의 행동 패턴을 활용
- 각각의 장점과 단점을 보완하여 추천 품질을 개선할 수 있음

### 추천 시스템 분류 비교

| 구분 | 설명 | 핵심 기준 |
|---|---|---|
| 내용 기반 필터링 | 사용자의 행동 이력을 기준으로 추천 | 아이템의 특징과 유사도 |
| 협업 필터링 | 여러 사용자의 행동과 평가 데이터를 함께 활용 | 사용자 또는 아이템 간 유사도 |
| 하이브리드 필터링 | 여러 추천 시스템을 조합 | 다양한 추천 방식의 결합 |

# 내용 기반 필터링

## 사용자 프로필과 유사도

내용 기반 필터링이 작동하려면 사용자가 무엇을 좋아하는지(사용자 프로필), 그리고 아이템끼리 얼마나 비슷한지(유사도)가 필요함.

**핵심 흐름**

아이템을 숫자로 표현 → 사용자 프로필 생성 → 유사도 계산 → 점수가 높은 아이템 추천

### 사용자 프로필 (User Profile)

_사용자가 어떤 특성을 좋아하는지 정리한 정보_

#### 추천 시스템에 사용되는 데이터

**직접 지정형**
- 사용자가 자신의 취향을 직접 알려 주는 방식
- 예시 : 음악 앱 가입 시 좋아하는 장르와 아티스트를 고르게 하는 경우

**간접 지정형**
- 사용자의 과거 행동 이력에서 취향을 추론하는 방식
- 예시 : 히어로 영화를 좋아하는 사람에게 스파이더맨을 추천하는 경우

### 사용자 프로필을 획득하는 방식

#### 아이템의 특성 가중치 평균값 활용

- 사용자가 가지고 있는 아이템의 특성 가중치를 활용
- 각 아이템의 특성을 얼마나 가지고 있는지 나타낸 값을 활용하여 사용자 프로필 구성

#### 명시적 피드백

- 좋아요/싫어요, 0~5점 평점
- 사용자가 작성한 댓글
- 댓글은 긍정적인지 부정적인지 별도로 분석해야 함

#### 암시적 피드백

- 클릭 로그, 구매, 저장, 북마크, 방문한 페이지
- 사용자의 행동을 통해 취향을 간접적으로 파악

#### 유저의 평가를 바탕으로 분류(Classification) 문제로 해결

- 아이템에 대한 유저의 기존 평가를 정답 라벨로 사용
- 새 아이템을 보았을 때 사용자가 어떤 평가를 할지 예측
- 아이템의 특징과 유저의 평가를 결합하여 분류 모델을 적용

> 많은 사용자 프로필을 구성할수록 추천 시스템의 정확도와 만족도가 높아짐.

### 유사도 (Similarity)

_두 대상이 얼마나 비슷한지 나타내는 숫자_

- 아이템 정보(텍스트, 이미지, 영상 등)를 숫자 벡터로 변환한 뒤 유사도 계산
- 사용자와 아이템을 같은 기준으로 표현하여 서로 비교
- 사용자가 선호하는 아이템과 비교하여 유사도가 높은 아이템을 선별하고 추천
- 예시 : STEP, CF 등 유사도 방식 활용

### TF-IDF (Term Frequency - Inverse Document Frequency)

영화 줄거리나 상품 설명과 같은 텍스트는 아이템의 중요한 특징임. 텍스트를 숫자로 바꾸는 가장 쉬운 방법은 단어 빈도를 세는 것임.

_문서 내 단어의 등장 빈도와 전체 문서에서의 등장 빈도를 함께 고려하여 단어의 중요도를 계산하는 방법_

#### TF (Term Frequency)

_단어 빈도_

- 특정 문서 안에서 해당 단어가 얼마나 자주 등장하는지를 나타냄
- 문서 내에서 자주 등장하는 단어일수록 높은 값을 가짐

#### IDF (Inverse Document Frequency)

_역문서 빈도_

- 여러 문서에서 해당 단어가 얼마나 자주 등장하는지를 고려
- 여러 문서에 빈번하게 등장하는 단어일수록 낮은 값을 가짐

> TF-IDF는 특정 문서에서 자주 등장하지만 전체 문서에서는 자주 등장하지 않는 단어에 높은 값을 부여함.

## 장단점

### 장점

**아이템 콜드 스타트 문제 해결**
- 신규 아이템도 특징 정보만 있으면 추천 가능

**내 기록만으로도 프로필을 만들 수 있음**
- 다른 사용자들의 데이터가 필요하지 않음

**추천 이유 설명 가능**
- 어떤 Feature 때문에 추천했는지 보여 줄 수 있음

### 단점

**Feature로 담을 수 있는 정보에 한계**
- 속성 정보가 부족하거나 품질이 낮으면 정확도가 떨어짐

**비슷한 것만 계속 추천 (Over-specialization)**
- 과도한 특화가 발생할 수 있음
- 기존에 좋아하던 장르에서 새로운 장르를 발견하기 어려움

**기록이 없는 신규 사용자 문제**
- 사용자에 대한 기록이 없으면 내용 기반으로도 해결하기 어려움

### Cold Start Problem

_충분한 정보가 수집되지 않아 적절한 제품을 추천해 주지 못하는 문제_

- 아이템 콜드 스타트 : 새 아이템이 등장했을 때 평가가 없으면 내용 기반으로 해결 가능
- 사용자 콜드 스타트 : 새 사용자의 기록이 없으면 내용 기반으로도 해결하기 어려움

# 협업 필터링

_과거에 비슷한 선호도를 가진 사용자들은 앞으로도 비슷한 선호도를 가질 가능성이 높다는 가정에 기반한 추천 방식_

## 메모리 기반 협업 필터링

### 메모리 기반 협업 필터링 (Memory-Based Collaborative Filtering)

_시스템에 저장된 사용자·아이템 평점을 그대로 꺼내서 새 아이템의 평점을 예측하는 방법_

- 별도의 학습 단계 없이 저장된 평점을 직접 비교
- 비슷한 사용자나 아이템을 찾아 평점을 예측
- Neighborhood-based CF(이웃 기반 협업 필터링)라고도 함

### 메모리 기반 협업 필터링의 두 가지 접근법

#### 사용자 기반 (User-Based) 접근

_나와 비슷한 취향을 가진 사용자의 평점을 활용해 내 평점을 예측하고 추천_

- 나와 유사도가 가장 높은 사용자를 N명으로 선정
- 이들의 평점을 모아 비슷한 특성을 가진 사용자가 높게 평가한 아이템을 추천
- 예측 평점이 높은 아이템을 추천

#### 아이템 기반 (Item-Based) 접근

_내가 평가한 아이템들과 평점 패턴이 비슷한 아이템을 찾아 추천하는 방식_

- 아이템의 속성이 아니라 사용자들이 아이템을 함께 평가하거나 구매한 패턴으로 유사도 판단
- 사용자의 행동이나 취향이 바뀌더라도 아이템들 간의 관계는 상대적으로 변하지 않음
- 아이템 유사도를 미리 계산해 두고 재사용하기 좋음

### User-Based VS Item-Based 비교

| 구분 | 사용자 기반 (User-Based) | 아이템 기반 (Item-Based) |
|---|---|---|
| 정확성 | 사용자 간 유사도가 높으면 잘 맞음 | 아이템 간 유사도가 높으면 잘 맞음 |
| 변화 | 사용자의 변화에 따라 추천이 바뀜 | 아이템 간 관계가 안정적이면 유리 |
| 안정성 | 사용자 변화가 많으면 추천이 불안정 | 아이템 특성이 안정적이면 유리 |
| 계산 속도 | 사용자 간 유사도를 직접 계산해야 함 | 아이템 유사도를 미리 계산해 둘 수 있음 |
| 다양성 | 비슷한 사용자가 평가한 다양한 아이템까지 추천 | 기존에 본 아이템과 비슷한 아이템 중심 추천 |

### 사용자 수가 많으면 아이템 기반이 유리한 이유

- 사용자는 보통 아이템 수보다 훨씬 많음
- 사용자 수는 증가하지만 함께 평가한 아이템은 상대적으로 제한적임
- 새 사용자가 추가되어도 아이템 간 유사도는 크게 바뀌지 않음
- 아이템 간 유사도는 미리 계산하여 재사용할 수 있음

> 정답이 있는 것은 아니고 일반적인 경향임.
>
> - 새로운 추천이 필요한 플랫폼 운영 → User-based
> - 정확하고 안정적인 추천이 중요한 대형 플랫폼 → Item-based 학습

## 메모리 기반 협업 필터링의 한계

### 희소 행렬 문제 (Sparse Matrix)

_주어진 평점, 선호도 데이터가 적을 경우 성능이 저하되는 문제_

- 실제 사용자는 아이템별 평점 행렬에 대부분의 칸이 비어 있음
- 아이템별 패턴이 빈약하면 유사도를 찾기 어려움
- 두 사용자가 함께 평가한 아이템이 적으면 비슷한 이웃을 찾기 어려움

**데이터가 희소하면 생기는 문제**

- 두 사용자가 함께 평가한 아이템이 거의 없을 수 있음
- 공통 아이템이 1개뿐이어도 비슷한 유사도가 나타날 수 있음
- 비슷한 이웃을 찾지 못하면 추천이 어려워짐

→ 모델 기반 협업 필터링(Model-Based CF) 이용

### 콜드 스타트 문제

_데이터가 충분히 쌓이지 않아 적절한 추천을 하지 못하는 문제_

**신규 사용자**
- 평가 기록이 없어 비슷한 사용자를 찾을 수 없음
- 예시 : 넷플릭스에서 방금 가입한 사용자는 아직 평가한 영화가 없음

**신규 아이템**
- 아무도 평가하지 않아 기존 사용자와의 유사도를 계산하기 어려움
- 누군가 처음 평가하기 전까지 추천되지 않는다는 의미에서 First-Rater 문제라고도 부름
- 예시 : 오늘 새로 등록된 영화는 아직 평점 데이터가 없음

**해결 방법**
- 가입할 때 좋아하는 장르나 작품을 직접 지정하도록 하기
- 새 아이템에는 내용 기반 필터링을 함께 사용하는 하이브리드 방식
- 앱에 등록된 모든 모델에 골고루 노출하여 임시 평점을 부여
- 인구통계학적 정보(나이, 성별 등)를 활용해 비슷한 사용자 그룹 추천

### 확장성 문제 (Scalability)

_아이템과 사용자가 계속 늘어날수록 계산량이 커지는 문제_

- 아이템과 사용자의 평점을 직접 비교하면 계산량이 계속 증가
- 데이터가 많아질수록 모든 사용자를 비교하는 데 많은 비용이 발생

**해결 방법 : KNN 협업 필터링**

- 모든 사용자를 쓰지 않고 유사도가 높은 N명의 이웃만 골라 예측
- 모든 사용자에게 계산하지 않고 가까운 이웃만 활용

## 모델 기반 협업 필터링

### 모델 기반 협업 필터링 (Model-Based Collaborative Filtering)

_관측된 평점 데이터를 예측 모델로 학습하고, 데이터의 패턴을 모델의 파라미터에 담아 관측되지 않은 평점을 예측하는 방법_

- 메모리 기반 : 추천이 필요할 때마다 저장된 평점을 직접 비교하여 이웃을 찾음
- 모델 기반 : 평점 데이터를 먼저 학습한 모델로 요약해 두고 추천은 이 모델만으로 수행

### 모델 기반 협업 필터링의 동작

**학습 단계 (Training)**
- 이미 존재하는 평점을 가장 잘 설명하는 모델의 파라미터를 학습
- 추천과 별도로 미리 수행
- 새 데이터가 쌓이면 주기적으로 다시 학습

**예측 단계 (Inference)**
- 학습된 파라미터로 비어 있는 칸의 평점을 계산
- 추천이 필요한 시점에 빠르게 수행

### 모델 기반 협업 필터링의 특징

- 잠재 요인(Latent Factor)을 활용하여 사용자와 아이템 간의 관계를 학습
- 관측되지 않은 평점도 잠재 요인 간의 관계를 통해 예측 가능
- 평점 데이터의 패턴을 학습하여 모델에 저장

### 잠재 요인 (Latent Factor)

_평점에 직접 드러나지는 않지만 사용자와 아이템 사이의 관계를 결정하는 숨은 특성_

- 모델이 관측된 데이터를 바탕으로 숫자로 추정
- 예시 : 취향, 성향과 같이 실제 데이터에는 없는 개념을 숫자로 표현

### Latent Factor Model

- 사용자와 아이템을 같은 잠재 공간 안의 벡터로 나타냄
- 두 벡터의 관계로 평점을 설명하는 모델
- 평점 행렬을 사용자 잠재 요인 행렬과 아이템 잠재 요인 행렬로 나누어 학습

### 행렬 분해 (Matrix Factorization)

 <br>

_큰 평점 행렬을 작은 행렬 두 개의 곱으로 쪼개는 방법_

**행렬 분해의 구조**

\[
\hat{R}=P^TQ
\]

- \(\hat{R}\) : 예측 평점 행렬
- \(P\) : 사용자 잠재 요인 행렬
- \(Q\) : 아이템 잠재 요인 행렬
- \(k\) : 잠재 요인의 개수

**예측 평점 계산**

\[
\hat{r}_{ui}=p_u^Tq_i
=\sum_{k=1}^{K}p_{uk}q_{ik}
\]

- 사용자와 아이템의 잠재 벡터 내적으로 예측 평점을 계산
- 사용자 잠재 벡터와 아이템 잠재 벡터가 서로 얼마나 관련이 있는지를 활용

### 메모리 기반과 모델 기반의 차이

| 구분 | 메모리 기반 | 모델 기반 |
|---|---|---|
| 학습 단계 | 없음 | 있음 |
| 예측에 쓰는 것 | 평점 행렬 데이터 | 학습된 모델 |
| 학습소 데이터 | 함께 평가한 아이템의 개수가 부족할 수 있음 | 전체 패턴을 학습해 일반화 |
| 예측 방식 | 비슷한 사용자의 평점 직접 활용 | 학습한 벡터 내적 |

# VAE 기반 협업 필터링

<br>

_사용자의 상호작용 기록을 신경망으로 확률분포로 압축한 뒤, 다시 복원하면서 아직 보지 않은 아이템의 선호를 예측하는 모델_

## MF의 한계와 VAE 구조

### MF의 한계와 딥러닝의 등장

#### 선형성

- 행렬 분해의 예측은 내적, 즉 요인 벡터를 곱해서 더하기만 하는 구조임
- 각 요인이 점수에 따로 더해지는 형태
- 여러 요인이 복잡하게 결합되어 나타나는 사용자 취향을 표현하기 어려움
- 예시 : 로맨스도 좋고 음악도 좋아하지만, 둘이 섞인 뮤지컬 작품에는 특별한 점수를 부여하는 사용자

#### 불확실성 무시

- 일반적인 MF는 사용자마다 벡터를 하나 학습함
- 이를 점 추정(Point Estimate)이라고 함
- 기록이 적은 사용자도 벡터 하나로 확신하기 때문에 기록이 적은 사용자와 풍부한 사용자의 차이를 표현하기 어려움

### 딥러닝의 해법

| MF의 한계 | 딥러닝 해법 |
|---|---|
| 선형성 | 비선형 활성화 함수를 가진 신경망 사용 |
| 불확실성 무시 | 평균과 분산을 함께 출력 |

→ 이 두 가지 한계를 극복하기 위해 도입된 모델이 VAE임.

### 오토인코더 (AutoEncoder, AE)

<br>

_입력을 작은 벡터로 압축했다가 복원하는 신경망 모델_

\[
입력\ x \rightarrow Encoder \rightarrow 잠재\ 벡터\ z
\rightarrow Decoder \rightarrow 복원된\ \hat{x}
\]

- 입력을 복원하는 데 꼭 필요한 핵심 정보를 잠재 벡터로 압축
- 잠재 벡터를 하나의 고정된 값으로 표현

### VAE (Variational AutoEncoder)

_가운데 잠재 벡터를 점이 아닌 분포로 표현하는 오토인코더_

- 평균 \(\mu\) : 사용자 취향의 중심 위치
- 표준편차 \(\sigma\) : 그 위치를 얼마나 확신할 수 있는지 나타내는 값

#### AE vs VAE

| 구분 | AE | VAE |
|---|---|---|
| 잠재 벡터 표현 | 단일 값(Point Estimate) | 평균과 표준편차를 가진 정규분포 |
| 불확실성 반영 | 표현하지 않음 | 사용자 취향의 불확실성 표현 |
| 장점 | 구조가 단순함 | 기록이 적은 사용자에게도 안정적이고 일반화에 유리함 |

### VAE가 더 안정적인 이유

**점 하나가 아니라 구간이나 분포를 사용해 학습**
- 학습할 때마다 다른 곳을 추정할 수 있도록 분포로 학습
- 디코더는 \(\mu\)에서 조금 벗어나더라도 비슷한 추천이 나오도록 학습

**기본 분포 쪽으로 잠재 공간을 정규화**
- KL 항이 분포를 표준 정규분포 쪽으로 당김
- 기록 및 개인으로 극단적인 위치로 가기 어렵게 함
- 기록이 적은 사용자의 경우 잠재 공간에 안정적으로 놓일 수 있음

## 목적 함수와 KL Divergence

### VAE 기반 CF (Multi-VAE)의 작동 구조

**입력 (Input)**
- 사용자의 상호작용 기록을 0/1 벡터로 입력
- 평점 데이터는 4점 이상만 1로 바꾸어 사용

**인코더 (Encoder)**
- 입력된 벡터를 받아 평균과 표준편차를 출력
- 잠재 공간의 분포를 추정

**잠재 벡터 (z)**
- 평균과 표준편차를 이용해 잠재 벡터 샘플링

**디코더 (Decoder)**
- 잠재 벡터를 전체 아이템의 점수로 변환
- 모든 아이템의 점수를 계산한 뒤 Softmax를 통해 선호 확률 출력
- 확률이 높은 순서로 아직 상호작용하지 않은 아이템 추천

### VAE의 목적 함수 ELBO (Evidence Lower Bound)

_VAE는 ELBO라는 목적 함수를 최대화하며, 정확한 복원과 잠재 공간 정규화라는 두 가지 목표를 동시에 수행함._

\[
\mathcal{L}(x;\theta,\phi)
=
E_{q_\phi(z|x)}[\log p_\theta(x|z)]
-
D_{KL}(q_\phi(z|x)||p(z))
\]

#### 복원 항 (Reconstruction Loss)

- 디코더가 잠재 벡터 \(z\)를 사용하여 원래 입력 \(x\)를 최대한 비슷하게 복원하도록 함
- 사용자가 선호한 아이템을 잘 예측하도록 학습

#### KL Divergence 항

- 인코더가 출력하는 잠재 분포 \(q(z|x)\)가 표준 정규분포 \(p(z)\)에서 크게 벗어나지 않도록 제약
- 잠재 공간이 일부 사용자 데이터에 과적합되지 않도록 함
- Smooth한 잠재 공간을 형성하여 일반화에 기여

### KL Divergence란?

_두 확률분포가 얼마나 다른지를 나타내는 값_

VAE에서는 사용자 분포를 표준 정규분포에 가깝게 만든다.

\[
D_{KL}(N(\mu,\sigma^2)||N(0,1))
=
\frac{1}{2}(\mu^2+\sigma^2-\log\sigma^2-1)
\]

- 평균이 0에서 멀어지면 벌점이 커짐
- 분산이 지나치게 작거나 커져도 벌점이 커짐
- 표준 정규분포와 유사한 분포를 형성하도록 함

### KL 항이 만드는 차이: AE vs VAE

- AE : 정확하게 복원만 하면 잠재 벡터들이 제약 없이 흩어질 수 있음
- VAE : 정확하게 복원하면서도 잠재 공간을 표준 정규분포에 가깝게 유지

**VAE의 장점**
- 확률적 잠재 분포 모델링 : 기존 MF와 달리 불확실성을 반영
- 희소한 상호작용 데이터 처리 : KL Divergence를 통해 잠재 공간을 정규화하여 기록이 적은 사용자도 안정적인 예측 가능
- 비선형성 학습 : 신경망을 통해 MF가 표현하기 어려운 복잡하고 비선형적인 패턴을 학습

# GNN 기반 사용자 선호 예측

## GNN 구조와 프로세스

### GNN (Graph Neural Network, 그래프 신경망)


_사용자와 아이템의 상호작용을 그래프로 표현하고, 노드 간의 관계 정보를 학습하는 신경망_

- 사용자와 아이템의 상호작용을 관계 중심으로 표현
- 사용자는 노드, 아이템도 노드, 구매·평가·시청 등의 관계는 엣지로 표현
- 개별 아이템의 정보뿐 아니라 사용자와 아이템 간 관계를 학습

### GNN의 모티브

사용자와 아이템의 상호작용을 분석적으로 표현하기 위해 그래프를 사용함.

- 사용자는 노드
- 아이템도 노드
- 사용자가 아이템을 구매하거나 평가한 관계는 엣지

### 그래프 (Graph)

_노드(Node)와 엣지(Edge)를 모아 구성한 자료 구조_

#### 노드 (Node / Vertex)
- 개체를 의미함
- 예시 : 사용자, 아이템, 상품, 원자 등

#### 엣지 (Edge / Arc)
- 노드 사이의 관계나 상호작용을 의미함
- 방향을 가질 수 있음
  - 'A가 B를 팔로우'와 'B가 A를 팔로우'는 다름
- 가중치를 가질 수 있음
  - 평점 5점, 구매 3회 등

### 그래프를 사용하는 이유

**관계 상호작용과 같은 비정형적인 개념을 다루기에 적합**
- 취향이 비슷하다.
- 빠르게 구매한다.
- 단순한 속성만으로 표현하기 어려운 관계를 그래프 형태로 표현

**Non-Euclidean 공간도 표현하고 학습 가능**
- 그래프는 격자 구조가 없는 비유클리드 공간의 데이터를 노드와 엣지로 표현하는 자료구조
- GNN은 이러한 관계 데이터를 직접 입력으로 받아 학습

### GNN의 구조

<br>

GNN은 그래프를 구성한 뒤 각 노드가 이웃 노드로부터 정보를 전달받고, 이를 집계하여 자신의 임베딩을 업데이트함.

**그래프 구조 생성**
- 노드와 엣지를 정의
- 각 노드에 초기 피처(임베딩)를 부여

**메시지 패싱 (Message Passing)**
- 각 노드가 이웃으로부터 정보를 받아 전달
- 메시지 생성, 집계, 갱신 단계로 구성

**레이어 연결성**
- 여러 층을 거치면서 더 멀리 있는 이웃의 정보까지 반영

**그래프 증강**
- 그래프 정보를 활용하고 확장

**학습 목표 (Learning Objective)**
- 최종적으로 업데이트된 노드 임베딩을 활용해 예측 수행

### 메시지 패싱 (Message Passing)

_각 노드가 자신의 이웃 노드로부터 정보를 받아 자신의 임베딩을 업데이트하는 과정_

#### 메시지 생성 (Message)

- 각 노드가 이웃에게 보낼 자신의 정보를 생성
- 이웃 노드에게 전달할 메시지를 만듦

#### 집계 (Aggregation)

- 여러 이웃 노드에서 받은 메시지를 하나의 벡터로 통합
- Sum, Mean, Max, Attention-weighted Sum 등 다양한 집계 방법 사용

#### 갱신 (Update)

- 집계된 이웃 정보와 자신의 이전 임베딩을 결합하여 새로운 임베딩을 계산
- 이 과정을 반복하면서 더 먼 거리의 이웃 정보까지 반영

### GNN의 프로세스

**그래프 구조 생성**
- 사용자와 아이템을 노드로 정의
- 평가 기록을 통해 엣지 연결
- 노드와 엣지를 활용해 전체 그래프 구성

**메시지 패싱**
- 각 노드는 이웃 노드로부터 메시지를 생성
- 주변 노드가 보낸 메시지를 집계
- 집계한 결과와 자신의 이전 임베딩을 결합하여 갱신

**반복 학습**
- 메시지 패싱 과정을 여러 층에서 반복
- 이웃 노드의 정보가 여러 단계에 걸쳐 전달됨

**최종 출력**
- 업데이트된 노드 임베딩을 활용하여 선호도 예측 수행

## LightGCN

### LightGCN (Light Graph Convolutional Networks)

_추천 시스템에 필요한 핵심 정보인 사용자-아이템 간 연결 관계 전달에 집중하도록 기존 GNN 구조를 단순화한 모델_

### LightGCN의 탄생 배경

- NGCF는 노드 임베딩에 여러 연산을 적용하고, 레이어마다 가중치 행렬 W를 곱하고 비선형 활성화 함수를 적용
- 이러한 연산은 일반적인 특성 학습에는 유용하지만, 추천 그래프에는 사용자 ID와 아이템 ID에 해당하는 ID 임베딩만 존재
- 실제로 필요한 정보는 사용자와 아이템 간의 연결 관계

→ LightGCN은 불필요한 연산을 제거하고 추천의 핵심인 사용자-아이템 간 연결 관계 전달에 집중하도록 구조를 단순화함.

### 제거한 GNN 구성 요소

**특징 변환 (Feature Transformation) 제거**
- 레이어마다 복잡한 학습 파라미터 W를 사용하는 과정 제거

**비선형 활성화 함수 (Non-linear Activation) 제거**
- ReLU와 같은 활성화 함수의 중간 단계 제거

### LightGCN의 특징

- 이웃 정보를 단순하게 전달
- 사용자와 아이템의 연결 관계에 집중
- 불필요한 가중치 변환과 비선형 연산을 제거
- 간결하고 효율적인 모델 구성

## GNN 활용 사례 및 장단점

### GNN 활용 사례

#### Pinterest

- GNN을 추천에 적용한 대표적인 사례
- 핀, 보드, 카테고리 등을 분석
- 사용자들이 상호작용하는 핀과 보드의 관계를 그래프로 구성
- 노드와 엣지 정보를 활용하여 추천

#### Netflix 추천 시스템 예시

- 사용자와 영화의 관계를 그래프로 구성
- 사용자와 영화 각각의 노드를 연결
- 평점 기록을 엣지로 표현
- 메시지 패싱을 통해 사용자의 선호도와 아이템 정보를 반영

### 주요 GNN 아키텍처

#### GCN (Graph Convolutional Network)
<br>

- 그래프 위에서 Convolution 연산 수행
- 이웃 노드와 타깃 노드에 동일한 파라미터 적용

#### GRN (Graph Recurrent Network)
<br>

- RNN 구조를 그래프에 결합
- 시간에 따라 변화하는 특징이 임베딩에 반영

#### GAT (Graph Attention Network)
 <br>
- 이웃 노드 간 중요도를 학습
- 가중 평균을 통해 타깃 노드의 이웃 정보를 반영

#### GraphSAGE
<br>
- 대규모 그래프에 대해 샘플링 기반으로 학습
- 이웃 노드와 타깃 노드의 가중치 값을 다르게 적용

### GNN의 장점

**고차원 협업 신호로부터 구조적 특징 포착**
- 단순한 '나의 장치' 요인뿐 아니라, 서로 비슷한 성향을 가진 사용자들과 내가 본 아이템과 관련된 다른 아이템 등의 관계 정보를 학습할 수 있음

**정형화되지 않은 데이터 학습 가능**
- 상호작용, 소셜 관계, 지식 그래프, 사용자 속성을 하나의 그래프에 함께 담아 학습할 수 있음

**Cold Start 문제 완화에 효과적**
- 상호작용이 없는 신규 아이템 속성 노드나 이웃 정보를 통해 임베딩을 얻을 수 있음

### GNN의 단점

**과평활화 (Over-smoothing) 문제**
- 메시지 전달을 많이 반복하면 GNN 모델의 레이어가 깊어질수록 연결된 노드들의 임베딩이 비슷해지는 현상
- 모든 노드의 임베딩이 같은 값으로 수렴하는 현상이 발생할 수 있음

**과압축 (Oversquashing) 문제**
- 단계마다 기하급수적으로 늘어나는 이웃 정보를 고정 크기 벡터에 담으면서 노드의 정보가 손실되는 현상

**계산 복잡도 증가**
- 실제 세계에서는 수십억 개의 노드와 엣지가 포함된 그래프에 GNN을 적용해야 하므로 막대한 자원과 메모리가 요구됨

**순서 반영의 어려움**
- GNN은 모든 상호작용을 집합으로 보지만 순서를 반영하지 않음
- '산발 → 바지 → 셔츠' 순서로 구매한 것과 '셔츠 → 바지 → 산발' 순서로 구매한 것이 같은 그래프가 되므로 다음 행동을 예측하기 어려움

# Transformer 기반 추천 시스템

## 기존 추천 시스템의 한계

### CF의 순차적 한계

**순서 무시**
- 같은 아이템을 샀다면 구매 순서가 달라도 같은 사용자로 봄

**문맥 부족**
- '방금' 한 구매와 '1년 전' 구매의 중요도를 구분하지 못함

**예측의 한계**
- '다음 행동'보다는 '일반적인 선호'를 예측하는 데 초점을 맞춤

### GNN의 순차적 한계

**지역적 정보 전달 한계**
- A → B → C의 관계에서 C의 정보를 얻으려면 B를 거쳐야 함
- 거치는 동안 정보가 희석되고 약해질 수 있음

**장기 의존성 문제**
- GNN은 여러 레이어의 단계까지만 볼 수 있어 멀리 떨어진 행동 간의 관계를 포착하기 어려움
- 반면 Self-Attention은 시퀀스 안의 모든 아이템 쌍을 거리와 상관없이 한 번에 직접 비교할 수 있음

> CF는 '누가 무엇을 좋아하는가', GNN은 '무엇이 무엇과 연결되었는가'를 잘 포착하지만, 둘 다 시퀀스 전체의 맥락을 이해하는 데 한계가 있음.

## Transformer와 Transformer 기반 주요 추천 시스템 모델

### Transformer: 사용자 행동을 순서(Sequence)로 접근

사용자의 최근 행동 시퀀스에는 현재 선호와 구매 목적이 반영되어 있음.

- 같은 사용자라도 시점마다 다른 행동을 할 수 있음
- 같은 아이템이라도 함께 등장한 아이템에 따라 의미가 달라질 수 있음
- 과거 행동의 순서와 문맥을 활용하여 다음 행동을 예측

### 문맥에 따라 달라지는 동일 아이템의 의미 (Self-Attention)

같은 아이템이라도 함께 등장한 아이템에 따라 다르게 해석됨.

- 문맥 A : 수영복 → 비치타월 → 선크림
- 문맥 B : 수면복 → 로션 → 선크림

같은 선크림이라도 앞에 등장한 아이템에 따라 서로 다른 목적으로 해석되고, 이에 따라 다음 추천 결과도 달라짐.

### 시간 순서를 뛰어넘는 핵심 의도의 포착 (Long-Range Dependency)

다음 행동과 관련된 아이템이 가장 최근의 아이템이 아닐 수도 있음.

- 프린터 → 잉크 → 라면 → 폰케이스 → 생수 → 쇼핑백 → ?
- Transformer는 현재 시점과 과거 모든 아이템의 관련성을 계산
- 거리가 멀더라도 관련성이 높은 아이템에 높은 가중치를 부여할 수 있음

### Self-Attention

_시퀀스 안의 모든 아이템 간의 관련성을 거리와 상관없이 동시에 계산하는 방식_

Self-Attention은 시퀀스의 각 아이템을 세 가지 벡터로 변환하여 사용함.

#### Query (Q)

- 아이템 임베딩 × 가중치 행렬 \(W_Q\)
- 현재 아이템이 다른 아이템들과 어떤 관계를 찾는지 나타내는 벡터

#### Key (K)

- 아이템 임베딩 × \(W_K\)
- Query와 내적하여 관련성 점수를 계산하는 데 사용

#### Value (V)

- 아이템 임베딩 × \(W_V\)
- 관련성 점수에 따라 가중합되는 실제 정보

### Self-Attention 연산 과정

\[
Attention(Q,K,V)
=
softmax\left(\frac{QK^T}{\sqrt{d_k}}\right)V
\]

**Attention Score 계산 (Q · K)**
- Query와 각 아이템의 Key를 내적하여 관련성 점수를 계산

**가중치 변환 (Softmax)**
- 점수를 합이 1이 되는 비율로 변환

**가중합 (Score · V)**
- 각 아이템의 Value에 비율을 곱해 더함
- 관련성이 높은 아이템의 정보가 더 많이 반영됨

### SASRec

<br>

_단방향 어텐션으로 과거 정보만 활용해 다음 아이템을 예측하도록 학습한 모델_

**방식**
- 단방향 어텐션을 사용
- 과거의 행동 정보만 활용
- 다음 아이템 예측 (Next-item prediction)

**특징**
- 적응적 집중 : 데이터가 복잡한 최신 아이템에 집중하도록 Attention 조절
- 빠른 학습 : RNN처럼 순서대로 계산하지 않고 모든 위치를 병렬 처리

**문제점**
- 계산량 : Attention 연산량이 시퀀스 길이의 제곱에 비례
- 단방향 : 과거 정보만 참조하므로 미래 정보를 활용하지 못함
- 과거 행동만을 활용하므로 미래 정보를 함께 활용할 수 없음

### BERT4Rec

_양방향 어텐션으로 과거와 미래 정보를 모두 활용해 가려진 아이템을 예측하도록 학습한 모델_

**방식**
- 양방향 문맥 활용
- 아이템의 앞뒤 문맥을 모두 학습에 사용
- [MASK] 토큰으로 가려진 아이템을 예측 (Masked-item prediction)

**특징**
- 양방향 문맥 : 각 아이템의 앞뒤 문맥을 모두 학습에 사용
- 데이터 활용 : 한 시퀀스에서 여러 위치를 가려 여러 학습 샘플을 만들 수 있음

### SASRec VS BERT4Rec 비교

| 구분 | SASRec | BERT4Rec |
|---|---|---|
| 어텐션 방향 | 단방향 (과거만) | 양방향 (과거 + 미래) |
| 학습 목표 | 다음 아이템 예측 | 가려진 아이템 예측 |
| 학습 효율 | 빠름 | 상대적으로 느림 |

### BERT4Rec이 항상 더 좋을까?

- 원 논문에서는 BERT4Rec이 SASRec보다 높은 성능을 보였다고 보고
- 후속 연구에서는 두 모델의 성능이 항상 일치하지는 않는 것으로 나타남
- 성능 차이는 어텐션 방향보다 학습 설계의 영향을 받을 수 있음
- 상황에 따라 모델을 선택해야 함

**상황별 선택 기준**

- 상호작용이 많고 '다음 행동' 예측이 중요하면 SASRec
- 상호작용이 적고 시퀀스를 양방향으로 학습해 빈칸 예측 효과가 중요하면 BERT4Rec
- 행동 순서가 중요한 도메인에서는 BERT4Rec의 양방향 문맥이 유리할 수 있음

# 오늘날의 추천 시스템

## LLM의 시대

트랜스포머가 시퀀싱 추천의 룰을 뒤집은 것은 아니지만, 이제는 LLM이 추천의 패러다임을 바꾸고 있음.

### 관점의 전환: 추천을 랭킹 문제가 아닌 언어 문제로 본다

**과거: 전통적인 추천 시스템**
- 관점 : 추천을 랭킹 문제로 봄
- 과제 : 점수를 매기고 순서를 정함
- 예시 : 사용자 A, 아이템 123 → 0.8점(좋음)

**LLM**
- 추천을 점수 계산이 아닌 문장으로 풀 수 있음
- ID가 아닌 의미(Semantic) 기반 추천
- 텍스트, 이미지, 오디오 등 비정형 데이터를 활용

→ 점수 대신 추천 이유를 함께 제시하는 서술형 추천이 가능해짐.

### 새로운 가능성

아이템을 텍스트로 이해하면 기존 모델로는 어려웠던 두 가지가 가능해짐.

#### 대화형 추천

- 사용자가 자연스러운 입력으로 직접 요청
- 예시 : 비 오는 일요일에 보기 좋은, 너무 슬프지 않은 영화 추천해 줘.

#### Zero-shot 추천

- 상호작용 데이터 없이도 아이템 설명만으로 추천
- 신상품이 등록된 직후 구매 기록이 없어도 상품 설명만으로 추천 대상이 될 수 있음
- Cold-Start 문제의 완화

## P5 / TALLRec

LLM을 추천에 적용하는 대표적인 방식은 두 가지임.

### P5

- 여러 추천 과제를 텍스트 프롬프트 → 텍스트 답변 형식으로 통일
- 하나의 언어 모델로 학습
- 여러 추천 과제를 같은 방식으로 처리
- 언어 모델을 추천 데이터로 처음부터 학습하므로 많은 컴퓨팅 자원과 시간이 필요

### TALLRec

- 학습 비용을 줄이기 위해 범용 LLM을 '좋아할까? (Yes/No)' 형식의 추천 지시문으로 튜닝하는 프레임워크
- LoRA로 일부 파라미터만 학습
- 적은 샘플로도 기존 모델보다 높은 성능을 보였다고 보고됨

> LLM은 단순한 성능 향상이 아니라 추천 시스템이 사용자와 상호작용하는 방식 자체를 바꾸고 있음.

## 인기 편향과 XAI 문제

### 인기 편향 (Popularity Bias)

_소수의 인기 아이템이 추천을 독점하고, 대다수의 비인기 아이템은 거의 추천되지 않는 현상_

**원인**

- 데이터 불균형 : 상호작용이 인기 아이템에 몰려 있음
- 모델의 편향 강화 : 데이터가 많은 아이템일수록 예측 점수가 높아 더 자주 추천됨
- Rich-get-richer : 추천 → 노출 → 클릭 → 학습 데이터 반영이 반복되며 격차가 커짐

**사례**

- Netflix Prize 데이터 : 인기 영화는 전체의 16.5%(2,938편)지만 전체 평점의 약 86%를 차지
- Spotify : 재생 수가 높은 스트리밍이 28.90% 증가했지만, 개인의 소수 타이틀은 11.17% 감소

→ 비인기 신규 아이템은 노출 기회를 얻기 어렵고, 사용자는 비슷한 콘텐츠만 반복해서 접하게 됨.

### 설명 가능성 (XAI)

- 이걸 왜 추천했는지 설명할 수 있는가?
- GNN, Transformer 같은 복잡한 딥러닝 모델은 블랙박스 문제가 있음
- 추천 이유를 고객에게 설득력 있게 제시하고, 비즈니스 담당자에게 추천의 근거를 제공해야 함

> LLM은 추천 이유를 자연어로 생성할 수 있지만, 그 설명이 실제 추천 근거와 일치한다는 보장은 없음.

## 컴퓨팅 자원, Cold-Start를 고려한 추천 시스템 도입

### 비용 문제: 딥러닝을 전면 도입하기에는 너무 비쌈

**GPU 비용**
- 대형 모델의 학습·추론에 고가의 GPU가 필요

**실시간 스코어링 서버 비용**
- 요청마다 수천만 개의 아이템을 점수화하면 서버 비용이 급증

**지속 비용**
- 대규모 트래픽에서는 사용자 수 × 요청 수만큼의 비용이 매일 누적

→ 모든 아이템에 딥러닝을 적용하기에는 비용 부담이 큼.

### 단계별 추천 시스템 구성

#### 후보군 생성

- 목표 : 수백만 개 아이템 → 수백 개 후보로 축소
- 사용자 모델 : 선형 모델 (내용 기반 필터링, 협업 필터링, 행렬 분해)
- 압도적인 속도 : 딥러닝 모델과 달리 후보를 계산하는 데 시간이 적게 소요되어 빠르게 처리 가능
- 비용을 줄이고 학습 데이터를 확보

#### 순위 매기기

- 목표 : 수백 개 후보를 정교하게 정렬해 상위 순위 결정
- 사용자 모델 : 딥러닝 기반 (GNN, Transformer 등)
- 복잡한 패턴과 사용자·아이템의 상호작용을 결합하여 추천 순위를 계산

> 결국 실제 추천 시스템은 단일 모델이 아니라 지금까지 배운 모델들을 단계별로 조합해 비용과 정확도를 함께 고려하여 설계함.
