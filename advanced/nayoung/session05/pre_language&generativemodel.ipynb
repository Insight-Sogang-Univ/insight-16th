# 언어 모델 & 생성 모델
## NLP
**NLP** : 자연어 처리
**NLP 정의**: 인간이 의사소통을 위해 사용하는 언어, 일대다 대응
- 자연어 개념
인간이 일상생활에서 사용하는 자연어를 컴퓨터가 분석하고 이해하며 생성할 수 있도록 하는 인공지능 기술<br>

- 자연어의 특징
1. 이산적 데이터
- 단어, 문자 등 구분된 기호로 구성
- 연속적 x, 컴퓨터가 직접 수학적으로 처리하기 어려움
> NLP는 단어와 문장을 숫자 또는 연속적인 벡터로 변환(Embedding)해야 함

2. 순차적 데이터
- 단어가 등장하는 순서에 따라 문장의 의미가 달라짐
> NLP는 단어의 순서와 앞뒤 문맥을 함께 고려해 처리할 수 있는 모델 구조(RNN, Transformer 등) 가 필요함

3. 중의성과 모호성
- 하나의 단어나 문장이 여러 가지 의미로 해석될 수 있음
> NLP는 핵심 단어 하나가 아닌 문맥 전체를 파악하여 적절한 의미를 판단해야 함

4. 희소성
- 단어의 종류와 조합이 매우 다양, 모든 표현 학습 어려움
> NLP는 학습하지 않은 표현도 단어 간 유사도를 이용해 확률적으로 추론할 수 있는 능력이 필요함

5. 가변성과 진화
- 시대와 사회적 환경에 따라 지속적으로 변화
- 신조어, 줄임말, 새로운 표현 등이 등장하면서 언어 사용 방식이 달라짐
> NLP는 변화하는 언어와 새로운 표현을 고정된 규칙이 아니라 데이터로부터 유연하게 패턴을 학습하는 머신러닝 방식이 필요함

- 자연어 처리 역할
 <br>

1. 형태소 분석(Morphological Analysis)

- 문장을 의미를 가진 가장 작은 단위인 형태소로 나누고, 각 형태소의 품사와 문법적 특징을 분석
- ex. "나는 학교에 간다" → "나/대명사 + 는/조사 + 학교/명사 + 에/조사 + 가-/동사 어간 + -ㄴ다/어미"

2. 구문 분석(Syntax Analysis)

- 문장의 문법적 구조를 분석하여 단어들이 어떤 관계로 연결되어 있는지 파악
- ex. "철수가 사과를 먹었다" → 철수(주어), 사과(목적어), 먹었다(서술어)

3. 의미 분석(Semantic Analysis)

- 단어와 문장의 의미를 분석하여 문맥에 맞는 의미를 파악
- ex. "은행에서 돈을 찾았다" → '은행'을 금융기관으로 해석

4. 담화 분석(Discourse Analysis)

- 여러 문장 사이의 연결 관계와 전체 대화의 맥락을 분석하여 의미를 파악
- ex. "철수는 시험에 합격했다. 그는 매우 기뻐했다." → '그'가 철수를 가리킨다는 것을 파악

## 언어 모델이란?
대량의 텍스트 데이터를 학습하여 단어 또는 토큰의 등장 확률을 예측하고, 문장의 자연스러움을 판단하는 모델

## 통계기반 모델

### SLM과 분포 가설
#### SLM: 통계적 언어 모델
#### 분포 가설: 비슷한 문맥에서 등장하는 단어들은 서로 유사한 의미를 가질 가능성이 높다는 가설
- 예: 나는 아침에 커피를 마셨다 / 나는 아침에 차를 마셨다
**→ '커피'와 '차'는 비슷한 문맥에서 등장하므로, 의미적으로 관련된 단어일 가능성이 높다고 판단**
- 조건부 확률: 특정 사건이 발생했다는 조건 아래, 다른 사건이 발생할 확률
- 카운트 기반 확률: 특정 문맥 다음에 어떤 단어가 얼마나 자주 등장했는지 계산
    - 카운트 기반 확률의 한계: 가능한 단어의 조합이 매우 많아 특정 표현이 학습 데이터에 거의 등장하지 않거나 전혀 등장하지 않는 문제
    > 새로운 언어 모델의 필요성 대두

### N-gram이 나온 배경과 한계

#### 배경
통계적 언어 모델(SLM)은 이전에 등장한 단어들을 바탕으로 다음 단어의 등장 확률을 계산하지만, 문장이 길어질수록 한계가 발생

> 한계 극복을 위해 **N-gram 모델**이 등장

#### 한계
- **데이터 희소성**: 고려해야 할 단어 조합이 증가하면서 학습 데이터에 등장하지 않은 조합이 많아짐
- **계산의 복잡성**: 이전의 모든 단어를 고려할 경우 확률을 계산하고 저장하는 데 많은 자원이 필요

### PPL
언어 모델이 주어진 문장을 예측할 때 얼마나 혼란스러워하는지(perplexed)를 수치로 나타낸 지표
- 높을수록: 예측 어려움
- 낮을수록: 예측 쉬움

#### 특징

- 문장 W가 훈련 데이터에 **희소하게 등장할수록** 모델이 해당 표현을 학습할 기회가 적어 **예측에 어려움**을 겪을 수 있음

→ 희소한 문장일수록 PPL이 높아지는 경향, 자주 등장하는 문장은 상대적으로 PPL이 낮을 가능성이 높음

- 문장에 예측하기 쉬운 단어가 추가되면 평균 예측 난이도가 낮아져 PPL이 감소할 수 있음
- 단, N이 증가한다고 PPL이 반드시 낮아지는 것은 아니며, 추가된 단어의 예측 난이도에 따라 달라짐

### NLM과 SLM의 차이

#### NLM
> 통계적 언어 모델(SLM)의 한계를 완화하기 위해 인공 신경망을 활용하여 단어의 의미와 문맥을 학습하는 신경망 언어 모델(NLM)이 등장
- 단어를 연속적인 숫자 벡터로 변환하여 단어 간 의미적 관계를 학습
- 신경망을 통해 다양한 단어 조합의 특징을 학습하고, 관측되지 않은 표현에도 일반화할 수 있도록 설계
- 단순한 등장 빈도를 넘어 단어 간 관계와 문맥을 반영하여 다음 단어를 예측

#### NLM vs. SLM
| 구분 | 통계적 언어 모델(SLM) | 신경망 언어 모델(NLM) |
| --- | --- | --- |
| 학습 방식 | 단어 등장 빈도와 확률 계산 | 인공 신경망의 가중치 학습 |
| 단어 표현 | 단어를 개별 기호로 취급 | 단어를 연속적인 벡터로 표현 |
| 의미적 관계 | 유사 단어 간 관계 반영이 어려움 | 임베딩을 통해 의미적 유사성 학습 |
| 희소 문제 | 관측되지 않은 조합에 취약 | 학습된 표현을 통해 희소 문제 완화 |
| 문맥 처리 | N-gram은 직전 N-1개 단어만 고려 | 신경망 구조에 따라 다양한 범위의 문맥 학습 |
| 계산 비용 | 상대적으로 낮음 | 학습에 많은 연산 자원 필요 |

## 딥러닝기반 모델
### Transformer의 등장
- RNN과 LSTM 한계
    - 장기 의존성 문제

    - 기울기 소실 문제

    - 순차 처리의 비효율성
> 한계 극복을 위해 Transformer 등장

### BERT
Transformer의 인코더 구조를 기반으로, 문장의 양방향 문맥을 학습하여 언어의 의미를 이해하는 신경망 언어 모델

- BERT 각각의 의미
- **B(Bidirectional)**: 단어의 앞뒤 문맥을 모두 고려하는 양방향 학습
- **ER(Encoder Representations)**: 인코더를 활용하여 문맥이 반영된 단어의 벡터 표현을 생성
- **T(from Transformers)**: Transformer의 인코더 구조를 기반으로 설계된 모델

#### 특징
1. 양방향(Bidirectional)

- 특정 단어의 의미를 파악할 때 **앞에 등장한 단어뿐 아니라 뒤에 등장한 단어**까지 함께 고려
- 문장 전체의 문맥을 활용하여 단어의 의미를 보다 정확하게 파악할 수 있음

2. Transformer 기반

- Transformer의 **인코더 구조만 활용**하여 문장 내 단어 간 관계를 분석
- Self-Attention을 통해 각 단어가 다른 단어들과 얼마나 관련되어 있는지 학습

3. 사전학습(Pre-training) + 파인튜닝(Fine-tuning)

- **사전학습** : 대량의 텍스트 데이터를 활용하여 언어의 일반적인 구조와 문맥을 학습
- **파인튜닝** : 사전학습된 모델을 특정 작업의 데이터로 추가 학습하여 목적에 맞게 활용

#### 학습 방법
<br>

1. 입력 문장 토큰화(Tokenization)

- 입력 문장을 단어 또는 서브워드 단위의 **토큰으로 분리**
- 문장 시작과 끝에 각각 `[CLS]`, `[SEP]` 등의 특수 토큰을 추가

2. 임베딩(Embedding)

- 각 토큰을 컴퓨터가 처리할 수 있는 **벡터로 변환**
- 토큰의 의미, 위치, 문장 구분 정보를 반영하여 입력 표현 생성

3. Transformer Encoder를 통한 문맥 학습

- Self-Attention을 활용하여 **각 토큰이 문장 내 다른 토큰들과 어떤 관계**를 가지는지 분석
- 여러 인코더 층을 거치면서 문맥이 반영된 벡터 표현 생성

4. 출력 및 활용

- 학습된 문맥 정보를 바탕으로 문장 분류, 감성 분석, 질의응답 등 문장 이해를 목적으로 하는 작업을 수행
- 사전학습 이후 특정 작업에 맞게 파인튜닝하여 활용

#### 사전학습
특정 NLP 작업을 수행하기 전에 대량의 텍스트 데이터를 활용하여 언어의 일반적인 구조와 문맥을 미리 학습하는 과정

- Masked Language Model(MLM)

    1. 입력 문장에서 일부 토큰을 무작위로 선택
    2. 선택된 토큰의 대부분을 `[MASK]`로 대체
    3. BERT가 가려진 토큰의 앞뒤 문맥을 함께 분석
    4. 원래 토큰을 예측하고 정답과 비교하여 모델을 학습

- Next Sentence Prediction(NSP)
    1. 두 개의 문장 A와 B를 입력
    2. 문장 B가 실제로 문장 A 다음에 등장하는 문장인지 판단
    3. 실제 다음 문장이면 `IsNext`, 무작위로 선택된 문장이면 `NotNext`로 분류
    4. 예측 결과와 정답을 비교하여 모델 학습

### GPT
Transformer의 디코더 기반 구조를 활용하여 이전 단어들을 바탕으로 다음 단어를 예측하고 텍스트를 생성하는 언어 모델

- 초기 GPT 구조

- **Decoder 기반** : Transformer의 인코더 없이 디코더 구조를 활용하며, 인코더와 연결되는 Cross-Attention은 사용하지 않음
- **Masked Self-Attention** : 미래 토큰을 참고하지 않고 이전 토큰들만 활용하여 다음 토큰을 예측
- **Feed Forward Network** : Attention 결과를 추가로 변환하여 언어적 특징을 학습

### LLM
방대한 텍스트 데이터와 대규모 신경망을 활용하여 인간의 언어를 이해하고 생성하도록 학습된 인공지능 모델
- 대량의 데이터에서 언어의 패턴과 문맥적 관계를 학습하여 다양한 자연어 처리 작업 수행
- 대부분 Transformer 구조를 기반으로 하며, 텍스트 생성뿐 아니라 번역, 요약, 질의응답 등에도 활용

#### 작동 방식
<br>

1. 학습 : 방대한 데이터를 학습하면서 단순히 정보를 암기하는 것을 넘어, 단어와 단어 사이의 관계, 문맥적 의미, 문법 구조, 그리고 세상의 지식까지 언어의 패턴을 스스로 터득

- **사전학습(Pre-training)** : 대량의 텍스트 데이터를 활용하여 다음 토큰을 예측하는 작업을 반복하며 언어의 구조와 문맥을 학습
- **파인튜닝(Fine-tuning)** : 사전학습된 모델을 특정 작업이나 목적에 맞는 데이터로 추가 학습하여 성능을 개선
- 학습 과정에서 예측값과 정답의 차이를 계산하고, 이를 줄이도록 모델의 가중치를 업데이트

2. 예측

- **입력 처리** : 사용자의 질문이나 문장을 토큰으로 분리하고 벡터로 변환
- **문맥 분석** : Transformer의 Attention을 활용하여 입력 토큰 사이의 관계를 분석
- **다음 토큰 예측** : 이전 문맥을 바탕으로 다음에 등장할 토큰의 확률을 계산
- **결과 생성** : 학습한 패턴을 기반으로 다음에 올 확률이 가장 높은 단어를 순서대로 예측하면서 응답 생성

#### 한계
1. 환각 현상
2. 높은 학습 및 업데이트 비용

### RAG
외부 지식 검색과 LLM의 텍스트 생성 능력을 결합하여 답변의 정확성과 최신성을 높이는 기술

#### 구조
<br>

**질의 인코더(Query Encoder)**

- 사용자의 질문을 임베딩 모델을 통해 **숫자 벡터로 변환**
- 질문의 의미를 벡터로 표현하여 **관련성이 높은 문서를 검색**할 수 있도록 함

**지식 검색기(Knowledge Retriever)**

- 벡터 데이터베이스(Vector DB)에서 질문 벡터와 **유사도가 높은 문서를 검색**
- 검색된 문서를 답변 생성에 활용할 수 있도록 LLM에 전달

**지식 증강 생성기(Knowledge-Augmented Generator)**

- 사용자의 질문과 검색된 외부 문서를 결합하여 LLM(Text Model)에 입력
- LLM이 검색된 문서의 내용을 참고하여 최종 답변을 생성

#### 장점
1. 최신 정보 반영
2. 정보 정확성 향상

### CAG
모델이 신뢰도를 스스로 판단하고 활용할 수 있도록 훈련시키는 기법

#### 구조
- 질문 입력 : 사용자의 질문을 분석하여 필요한 정보 파악
- 정보 검색 : 외부 데이터베이스에서 질문과 관련된 문서 검색
- 신뢰도 평가 : 검색된 정보의 출처, 최신성, 관련성 등을 종합적으로 평가
- 정보 선별 : 신뢰도가 높은 정보에 더 큰 비중을 두어 답변 생성에 활용
- 최종 답변 생성 : 신뢰성 높은 정보를 중심으로 최종 답변 생성

### Langchain의 아이디어, 특징
LLM을 외부 데이터, API, 다양한 도구와 연결하여 복잡한 AI 애플리케이션을 쉽게 개발할 수 있도록 지원하는 오픈소스 프레임워크
- LLM과 다양한 기능을 모듈화하고 연결하여 개발 효율성을 높이는 LangChain이 등장

#### 특징

1. 추상화

- LLM 호출, 프롬프트 관리, 문서 검색 등 복잡한 기능을 간단한 인터페이스로 제공
- ex. 복잡한 API 호출 코드를 직접 작성하지 않고, 간단한 명령어로 ChatGPT에 질문하고 답변을 받을 수 있음

2. 표준화

- 서로 다른 LLM, 데이터베이스, 외부 도구 등을 일관된 방식으로 사용할 수 있도록 지원
- ex. ChatGPT 대신 Claude를 사용하고 싶을 때, 전체 코드를 다시 작성하지 않고 모델 연결 부분만 변경하여 활용할 수 있음

3. 체이닝

- 여러 작업을 순서대로 연결하여 하나의 작업 흐름으로 구성
- ex. 영어 문장 입력 → 한국어로 번역 → 번역된 문장 요약 → 최종 결과 출력


## sLM
소규모 언어 모델(sLM)은 자연어를 처리하고 이해하며 생성할 수 있는 인공지능(AI) 모델
- sLM은 일반적으로 수백만~수십억 개 수준의 매개변수를 가지며, LLM은 수십억~수천억 개 이상의 매개변수를 사용

### LLM vs. sLM
| 구분 | LLM | sLM |
| --- | --- | --- |
| 모델 크기 | 상대적으로 큼 | 상대적으로 작음 |
| 성능 | 복잡한 추론과 다양한 작업에 강점 | 특정 분야나 비교적 단순한 작업에 효율적 |
| 연산 비용 | 상대적으로 높음 | 상대적으로 낮음 |
| 응답 속도 | 많은 연산이 필요할 수 있음 | 같은 환경에서 빠른 처리에 유리 |
| 활용 환경 | 주로 고성능 서버 | 서버, 스마트폰, 노트북 등 |

### 모델 압축
모델의 중요한 성능은 최대한 유지하면서 크기와 연산량을 줄이는 기술

1. 가지치기<br>

신경망에서 불필요한 부분을 제거하여 모델을 가볍게 만드는 기술

- **개념 :** 신경망에서 중요도가 낮은 연결이나 뉴런을 제거하여 모델의 크기와 연산량을 줄이는 방법

- **작동 원리**
    1. 학습된 신경망에서 결과에 미치는 영향이 작은 연결이나 뉴런을 찾음
    2. 중요도가 낮은 부분을 제거하고, 핵심적인 연결만 남김
    3. 필요에 따라 추가 학습을 진행하여 성능 저하를 보완

 <br>
- 설명
    - Before Pruning : 뉴런과 연결이 많아 신경망 구조가 복잡한 상태이다.
    - After Pruning : 불필요한 연결과 뉴런이 제거되어 구조가 단순해진 상태
    - 연결만 제거하는 비구조적 가지치기와 뉴런·블록 등을 제거하는 구조적 가지치기로 나눌 수 있음
- **효과 :** 모델의 크기와 연산량을 줄일 수 있지만, 중요한 부분까지 제거하면 성능이 저하될 수 있음

2. 양자화 <br>
숫자의 표현 정밀도를 낮춰 모델을 더 작고 효율적으로 만드는 기술

- **작동 원리**
    - 기존 모델은 가중치를 FP32(32비트)와 같은 높은 정밀도로 저장할 수 있음.
    - 이를 FP16(16비트)이나 INT8(8비트) 등 더 낮은 정밀도로 변환
    - 숫자를 저장하는 데 필요한 비트 수가 줄어들어 모델을 더 가볍게 만들 수 있음
 <br>
- **그림 설명**
    - **FP32 :** 가중치를 높은 정밀도로 표현하는 방식
    - **INT8 :** 가중치를 8비트 정수로 근사하여 표현하는 방식
    - 그림처럼 숫자의 표현 방식을 바꾸면 필요한 저장 공간을 줄일 수 있음
- **효과 :** 모델 크기와 메모리 사용량을 줄이고 연산 속도를 개선할 수 있지만, 정밀도가 낮아지면서 일부 성능이 저하될 수 있음

3. 저차원 분해
<br>큰 행렬을 작은 행렬들의 곱으로 표현하여 파라미터를 효율적으로 사용하는 기술

- **개념 :** 모델의 크고 복잡한 가중치 행렬을 더 작은 행렬들의 곱으로 근사하여 파라미터 수와 연산량을 줄이는 방법
- **작동 원리**
    - 신경망에는 많은 숫자로 구성된 큰 가중치 행렬이 존재
    - 이 행렬을 더 작은 두 개의 행렬로 나누어 표현
    - 전체 정보를 그대로 저장하는 대신, 중요한 특징을 적은 수의 파라미터로 표현
 <br>
- **그림 설명**
    - **Original Weight Matrix W :** 기존 모델의 가중치 행렬
    - **Matrix A, B :** 적은 수의 파라미터를 가진 두 개의 작은 행렬
    - **ΔW = A × B :** 두 행렬을 곱해 가중치의 변경량을 표현
    - **Final Weight :** 기존 가중치에 학습된 변경량을 반영한 결과
- **효과 :** 적은 파라미터로 중요한 정보를 표현하여 저장 공간이나 연산량을 줄이는 데 활용할 수 있음

4. 지식 증류 <br>

큰 모델의 지식을 작은 모델에 전달하여, 적은 파라미터로도 좋은 성능을 낼 수 있도록 하는 기술
- **개념 :** 성능이 뛰어난 큰 모델(Teacher Model)의 지식을 작은 모델(Student Model)에 전달하여, 작은 모델도 비슷한 성능을 낼 수 있도록 학습시키는 방법
- **작동 원리**
    - 먼저 학습이 완료된 큰 모델을 교사 모델로 사용
    - 교사 모델의 예측 결과나 학습된 지식을 작은 모델에 전달
    - 작은 모델은 교사 모델의 출력을 따라 하도록 학습하면서 중요한 지식을 습득
<br>
- **그림 설명**
    - **Teacher Model :** 많은 파라미터를 가진 큰 모델로, 학습에 필요한 지식을 제공
    - **Knowledge Transfer :** 교사 모델의 지식을 학생 모델에 전달하는 과정
    - **Student Model :** 전달받은 지식을 학습하는 작은 모델
- **효과 :** 큰 모델의 지식을 활용하여 작은 모델의 성능을 높일 수 있으며, 적은 자원으로도 효율적인 추론이 가능해짐

### sLM의 장단점
#### 장점
1. 성능적 측면
- 빠른 응답 속도
- 제한된 환경 실행 가능
- 보안 및 데이터 보호

2. 비용적 측면
- 낮은 운영 비용
- 적은 전력 소비
- 하드웨어 비용 절감

#### 단점(한계)

1. 편향(편향된 답변 생성)
2. 제한된 일반화 및 복잡한 작업 성능 저하
3. 환각 현상(사실과 다른 정보 생성)

### sLM과 LLM의 결합
결합 추론(Hybrid Inference): sLM과 LLM을 함께 활용하여 각 모델의 장점을 살리는 방식 <br>

 <br>
- **① 간단한 작업 :** 문장 분류, 간단한 요약 등은 빠르고 효율적인 sLM이 처리
- **② 복잡한 작업 :** 여러 단계의 추론이나 전문적인 분석이 필요한 작업은 LLM이 처리
- **③ 결과 :** 모든 작업에 LLM을 사용하지 않아도 되므로 비용과 처리 시간을 줄이면서 필요한 수준의 성능을 확보 가능

## 분류모델과 생성모델의 차이
| 구분 | 분류 모델 | 생성 모델 |
| --- | --- | --- |
| 학습 방법 | 데이터의 특징과 정답 레이블의 관계 학습 | 데이터의 특징과 확률분포 학습 |
| 학습 목표 | $$P(Y|X)$$ 모델링: 데이터 X가 주어졌을 때 정답 Y에 속할 확률을 학습하여 분류 | $$P(X)$$ 모델링 : 실제 데이터 X가 나타나는 확률분포를 학습하여 새로운 데이터 생성 |
| 주요 특징 | 데이터의 차이점을 학습하여 분류 기준 형성 | 데이터의 공통적인 특징과 다양한 패턴을 학습하며, 입력 데이터의 적합성을 평가하고 비정상적인 입력을 탐지하는 데 활용 가능 |
| 출력 결과 | 범주 또는 범주별 예측 확률 | 새로운 이미지, 텍스트, 음성 등 |
| 한계 | 처음 보는 이상한 데이터도 기존에 학습한 라벨 중 하나로 잘못 분류할 수 있으며, 일반적인 분류 모델은 새로운 데이터 생성 불가능 | 부자연스러운 결과 생성 가능, 생성 결과의 품질 평가 어려움 |

## 생성모델의 분류
데이터의 확률 밀도를 직접 계산 가능한가?
- 명시적 확률밀도 모델
    - 데이터의 확률분포를 수학적으로 정의하고, 특정 데이터의 P(X)의 값을 직접적으로 수치화하여 출력할 수 있는 생성 모델
    1. 다루기 쉬운 밀도 모델
    - 데이터의 **확률밀도를 수학적으로 직접 계산**할 수 있는 모델
    - 확률분포를 계산 가능한 형태로 구성하여, 특정 데이터가 학습한 분포에서 얼마나 그럴듯한지 정확한 수치로 나타낼 수 있음
    - 대표 모델
    - **NADE, MADE, PixelRNN/PixelCNN :** 데이터의 각 요소에 대한 조건부 확률을 순차적으로 계산하는 모델
    - **Change of Variables Model (nonlinear ICA) :** 변수 변환을 활용하여 복잡한 데이터의 확률밀도를 계산하는 모델

    2. 근사적 밀도 모델
    - 데이터의 **확률밀도를 직접 계산하기 어려워 근사적인 방법을 활용**하는 모델
    - 복잡한 확률분포를 계산할 때 발생하는 어려움을 해결하기 위해 **변분 추론이나 마르코프 연쇄 등의 방법**을 활용함
    - 대표 모델
    - **VAE(Variational Autoencoder)** : 변분 추론을 활용하여 계산하기 어려운 확률분포를 근사적으로 학습하는 모델
    - **Boltzmann Machine** : 마르코프 연쇄 등의 방법을 활용하여 데이터의 상태를 반복적으로 변화시키며 표본을 추출하여 확률분포를 학습하는 모델
            
    
- 암시적 확률밀도 모델
데이터의 확률 분포 P(X)가 무엇인지 수치로 알려주지 않고, 학습한 데이터의 분포를 바탕으로 새로운 데이터를 생성(Sampling)하는 데 초점을 맞춘 생성 모델

1. 직접 샘플링
- 학습된 생성 모델에 **무작위 노이즈를 입력하여 새로운 데이터를 직접 생성**하는 방식
- 여러 단계의 반복적인 상태 변화 없이 **생성 모델을 한 번 통과하여 결과 생성** 가능
- 대표 모델
    - **GAN (Generative Adversarial Network)** : 생성자(Generator)와 판별자(Discriminator)의 경쟁을 통해 데이터의 특징을 학습하고, 무작위 노이즈로부터 새로운 데이터를 생성하는 모델

2. 간접/반복적 샘플링
- **여러 단계의 확률적인 상태 변화를 반복하여 새로운 데이터를 생성**하는 방식
- 한 번에 데이터를 생성하는 것이 아니라, **여러 단계의 상태 변화를 거쳐 학습한 데이터의 분포를 따르는 표본을 생성**함.
- 대표 모델
    - **GSN (Generative Stochastic Network)** : 마르코프 연쇄의 확률적 상태 전이 과정을 학습하고, 이를 반복 적용하여 새로운 데이터를 생성하는 모델
        - **GSN의 작동 원리**
            - 현재 상태를 바탕으로 다음 상태가 결정되는 확률적인 규칙을 학습함.
            - 학습한 상태 변화 과정을 반복하여 새로운 데이터를 생성

### AE
입력 데이터의 중요한 특징을 압축하고, 이를 바탕으로 원래 데이터와 최대한 유사하게 복원하도록 학습하는 인공신경망 모델 <br>

- 입력 데이터를 압축하고 복원하면서 데이터의 핵심적인 특징을 학습함
- 입력 데이터와 복원된 데이터의 차이를 최소화하는 것이 목표
- 별도의 **정답 레이블 없이** 입력 데이터 자체를 복원 목표로 활용할 수 있음

#### 구조 및 작동 과정
 <br>

1. 압축 단계 (Encoder)

- 입력 데이터에서 **중요한 특징을 추출하고 정보를 압축**하는 단계
- 인코더는 입력 데이터(x)를 더 적은 수의 값으로 표현하여 **잠재 벡터(z)**로 변환
    - **잠재 벡터(z)**는 원본 데이터의 중요한 특징을 압축하여 함축된 정보를 저장하고 있는 벡터
- **예시 :** 손글씨 숫자 ‘4’ 이미지에서 숫자의 형태, 획 등 **중요한 특징을 추출**

2. 병목 통과 (Bottleneck)

- 인코더가 추출한 핵심 정보를 **잠재 벡터(Latent Vector, z)**로 표현하는 단계
- 제한된 공간에 정보를 담기 때문에 **데이터의 핵심적인 특징을 학습하도록 유도**

3. 복원 단계 (Decoder)

- **압축된 잠재 벡터 (z)**를 활용하여 **원본 데이터와 유사한 데이터를 복원하고, 원본과 복원 결과를 비교하여 중요한 정보가 얼마나 잘 보존되었는지 확인**하는 단계
- 디코더는 잠재 벡터를 입력받아 **복원 데이터 (x')를 생성**
- **예시:** 압축된 손글씨 특징을 바탕으로 숫자 '4' 이미지를 다시 복원

### VAE
입력 데이터의 특징을 확률분포로 학습하고, 해당 분포에서 샘플링하여 새로운 데이터를 생성할 수 있는 생성 모델 <br>

- 기존 AE의 인코더, 디코더 구조를 기반으로 만들어진 모델
- 데이터를 단순히 압축하고 복원하는 것을 넘어, 데이터의 확률적 특성을 학습하여 새로운 데이터를 생성할 수 있음
- 변분 추론(Variational Inference)을 활용하여 복잡한 확률분포를 근사적으로 학습함

#### 학습시 고려사항
1. 복원 오차 최소화
2. KL 발산 최소화
- KL 발산은 두 확률분포가 얼마나 다른지를 수치로 나타내는 지표

#### 구조 및 작동 과정
 <br>

1. 분포 예측 단계 (Probabilistic Encoder) <br>

**입력 데이터의 특징을 분석하여 확률분포를 예측하는 단계** <br>

- 입력 데이터 x를 **확률적 인코더(Probabilistic Encoder)**에 전달
- 인코더는 데이터의 특징을 분석하여 **평균(μ)과 표준편차(σ)**를 예측
    - **평균(μ):** 확률분포의 중심으로, 데이터의 대표적인 특징을 나타냄
    - **표준편차(σ):** 확률분포가 평균을 중심으로 얼마나 퍼져 있는지를 나타냄

- 사용 수식
    
    $$
    q_φ(z|x)
    $$
    
    - 입력 데이터 x가 주어졌을 때 잠재 벡터 z가 따르는 확률분포를 추정하는 인코더

2. 샘플링 및 재파라미터화 (Sampling & Reparameterization) <br>

**예측한 확률분포에서 잠재 벡터(z)를 추출하는 단계** <br>

- 앞서 예측한 확률분포에서 **잠재 벡터(z)를 무작위로 샘플링**함.
    - 사용 수식
        
        $$
        z = μ + σ ⊙ ε
        $$
        
        - **z:** 최종적으로 생성되는 잠재 벡터
        - **μ:** 인코더가 예측한 평균
        - **σ:** 인코더가 예측한 표준편차
        - **ε:** 표준정규분포에서 추출한 무작위 노이즈
        - **⊙:** 각 요소끼리 곱하는 연산
        
        $$
        \varepsilon \sim \mathcal{N}(0, I)
        $$
        
        - 평균이 0이고 각 차원의 분산이 1인 표준정규분포에서 노이즈를 추출한다는 의미
- 같은 확률분포에서도 서로 다른 값이 추출될 수 있기 때문에 **다양한 형태의 데이터를 생성** 가능

3. 생성 단계 (Probabilistic Decoder) <br>

**샘플링한 잠재 벡터를 이용하여 데이터를 복원하거나 새롭게 생성하는 단계**

- 샘플링한 잠재 벡터 z를 **확률적 디코더(Probabilistic Decoder)**에 전달
- 디코더는 잠재 벡터를 바탕으로 **데이터의 확률분포를 예측**하고, 이를 이용해 이미지를 복원하거나 생성
    - 사용 수식
        
        $$
        z \sim \mathcal{N}(0, I) \rightarrow x' \sim p_{\theta}(x \mid z)
        $$
        
        - **잠재 벡터 추출 :** 표준정규분포에서 새로운 잠재 벡터 z를 무작위로 샘플링
        - **새로운 데이터 생성 :** 추출한 z를 학습된 디코더에 입력하고, 디코더가 예측한 확률분포에서 새로운 데이터 x′를 생성
        
        **기존 데이터를 입력하지 않아도 잠재공간에서 새로운 값을 추출하여 새로운 데이터를 생성할 수 있음**
        
- 학습 과정에서는 **원본 데이터 x와 복원 데이터 x′가 최대한 유사**해지도록 학습

### GAN의 구조와 장단점
GAN: 생성자(Generator)와 판별자(Discriminator)라는 두 신경망이 서로 경쟁하면서 실제 데이터와 유사한 새로운 데이터를 생성하도록 학습하는 생성 모델<br>
 <br>

#### 구조
1. 무작위 노이즈 (Random Noise)

- 생성자에 입력되는 **무작위 숫자 벡터(z)**를 의미
- 생성자는 이 노이즈를 바탕으로 새로운 강아지 이미지를 생성

2. 생성자 (Generator, G)

- **입력:** 무작위 노이즈(z)
- **출력:** 생성된 가짜 강아지 이미지(G(z))
- **목표:** 판별자가 진짜라고 판단할 만큼 실제 강아지와 유사한 이미지 생성

3. 판별자 (Discriminator, D)

- **입력:** 실제 강아지 이미지(x) 또는 생성자가 만든 가짜 강아지 이미지(G(z))
- **출력:** 입력된 이미지가 진짜일 확률(0~1)
- **목표:** 실제 강아지 이미지와 생성된 가짜 강아지 이미지를 정확하게 구별

#### 장점

1. 진짜 같은 가짜 생성 가능

    - 실제 데이터의 특징과 분포를 학습하여 **현실에 존재하지 않지만 실제처럼 보이는 새로운 데이터**를 생성 가능
    - 예시: 실존하지 않는 사람의 얼굴 이미지 생성

2. 높은 이미지 선명도

    - 판별자의 평가를 바탕으로 생성자가 학습하기 때문에 **이미지의 세부적인 특징을 정교하게 표현**할 수 있음.
    - 특히 이미지 생성 분야에서 기존 VAE보다 선명하고 사실적인 결과를 보여주는 경우가 많음.

3. 빠른 생성 속도

    - 학습이 완료되면 **무작위 노이즈를 생성자에 한 번 통과시키는 방식**으로 이미지를 생성 가능
    - 여러 단계를 반복하는 일반적인 확산 모델과 비교하여 빠르게 이미지를 생성할 수 있다는 장점

4. 잠재공간의 부드러움 (Smooth Latent Space)

    - 잘 학습된 GAN은 잠재 벡터를 조금씩 변화시켰을 때 **생성된 이미지의 특징도 자연스럽게 변화**할 수 있음.
    - 예시: 잠재 벡터를 변화시키면서 강아지의 표정이나 자세가 점진적으로 변하는 이미지 생성
    - 다만 모든 GAN에서 이러한 특성이 보장되는 것은 아님.

#### 단점(한계)
1. 학습의 불안정성
2. 훈련 초기 문제
3. 모드 붕괴
4. 평가 지표의 한계
5. 역매핑의 어려움

### 확산모델
이미지에 노이즈(잡음)를 추가한 데이터를 학습하고, 무작위 노이즈를 점차 제거하면서 새로운 이미지를 생성하는 생성형 AI 모델

#### 기본 원리

- 실제 이미지에 노이즈를 조금씩 추가하면서 이미지가 흐트러지는 과정을 만듦
- 모델은 노이즈가 포함된 이미지에서 노이즈를 제거하는 방법을 학습
- 학습이 완료되면 무작위 노이즈에서 시작해 새로운 이미지를 만들어낼 수 있음

#### 순확산과 역확산

##### 순확산
<br>
원본 이미지에 가우시안 노이즈(잡음, Gaussian Noise)를 조금씩 추가하여, 최종적으로 무작위 노이즈에 가까운 상태로 만드는 과정<br>

1. 시작 단계 – 원본 이미지 ($x_0$)

    순확산은 **노이즈가 없는 깨끗한 원본 데이터**에서 시작한다. 그림에서는 바다와 절벽의 색상과 형태가 선명하게 나타나는 이미지가 이에 해당한다.

2. 노이즈 추가 단계 – 점진적인 변화 ($x_1$, $x_2$, …)

    원본 이미지에 **가우시안 노이즈**를 조금씩 추가하며, 이 과정은 여러 단계에 걸쳐 반복된다. **단계가 진행될수록 노이즈가 누적되어 이미지가 점차 흐려지고, 바다와 절벽의 윤곽도 알아보기 어려워진다**.

3. 결과 단계 – 무작위 노이즈 ($x_T$)

충분한 단계를 거치면 원본 이미지의 특징과 정보가 거의 사라지고, 최종적으로 **무작위 노이즈에 가까운 상태($x_T$)**가 된다. 이렇게 만들어진 여러 단계의 데이터는 이후 **역확산 모델이 노이즈를 제거하고 새로운 이미지를 생성하는 방법을 학습하는 데 활용**된다.

##### 역확산
무작위 노이즈에서 시작하여 노이즈를 단계적으로 제거하면서 새로운 이미지를 생성하는 과정. 순확산과 반대 방향으로 진행 <br>
<br>

1. 시작 단계 – 무작위 노이즈 ($x_T$)

    역확산은 원본 이미지의 특징을 알아볼 수 없는 **무작위 노이즈 상태**에서 시작한다. 그림에서는 오른쪽의 무작위 잡음으로 이루어진 이미지가 이에 해당하며, **새로운 이미지를 생성하기 위한 초기 상태**를 의미한다.

2. 노이즈 제거 단계 – 이미지 생성 (Denoising U-Net)

    학습된 신경망인 **Denoising U-Net**을 이용하여 현재 이미지에 포함된 노이즈를 예측하고, 이를 바탕으로 **노이즈를 조금씩 제거**한다. 이 과정을 **여러 번 반복**하면서 이미지의 색상과 윤곽이 점차 나타나고, 구체적인 형태를 갖추게 된다.

3. 결과 단계 – 새로운 이미지 ($x_0$)

    반복적인 노이즈 제거 과정이 끝나면 무작위 노이즈가 자연스러운 이미지로 변하며, 최종적으로 **새로운 이미지가 생성($x_0$)**된다. 그림에서는 노이즈가 점차 제거되면서 바다와 절벽의 모습이 선명하게 나타나는 것을 확인할 수 있다.

#### 확산 모델 장단점
##### 장점
1. 압도적 샘플 품질
2. 학습 안정성
3. 다양성 확보

##### 단점
1. 느린 생성 속도
2. 높은 계산 비용
3. 구조적 제약
