## 들어가기 전에: 인공신경망과 퍼셉트론

### 인공신경망 개념


<br>
우리 뇌의 **신경망(뇌를 구성하는 신경 세포끼리 서로 연결되어 정보를 처리하는 구조)을 모방한 모델**

1. **입력(**$x_1, x_2, \dots, x_n$**)**들을 받아 각각 **가중치**($w_1, w_2, \dots, w_n$)와 **편향**($b$)을 적용
2. 종합한 값을 활성화 함수에 넘김(활성화 함수는 임계값을 기준으로)
    - 기준 미달 → **0**에 가깝게 죽여서 내보냄
    - 기준 통과 → **유의미한 값**으로 살려서 내보냄
3. output 출력

## 퍼셉트론

### 퍼셉트론 기초 개념

<br>
**퍼셉트론의 출력값을 구하는 절차**

⇒ **`입력 → 가중합 → 활성화 함수 → 출력`**

1. 입력값과 가중치를 곱한다 → $x_1 w_1$
2. 곱한 값들의 총합을 구한다 → $x_1 w_1 + x_2 w_2 + \dots + x_n w_n$
(편향을 더해줄지는 옵션)
3. 총합이 임계값($\theta$)을 넘으면 $1$, 넘지 않으면 $0$을 출력한다
<br>

### 퍼셉트론 관련 용어 정리

1. 가중치
- 입력값이 출력값에 주는 영향력(중요도)을 결정
- 각 입력값에는 각각의 가중치가 존재
- **가중치의 값이 클수록 해당 입력 값이 중요함**을 의미
<br>

2. 활성화 함수
- 신호가 일정 이상의 값(임계값 $\theta$)을 넘으면 1, 넘지 못하면 0을 출력하는 퍼셉트론의 성질은 활성화 함수로 **계단 함수**를 사용
- 이후 등장한 발전된 신경망들은 계단 함수 외에도 다양한 활성화 함수를 사용
<br>

3. 임계값(Threshold, 𝜃)
- 활성화 함수에 사용되는 값으로, 보통 세타($\theta$)로 표현
- 계단 함수에서는 가중합이 임계값을 넘으면 1, 넘지 못하면 0을 출력
<br>

4. 편향(bias)
- 위의 계단 함수를 구현한 식에서, 임계값을 좌변으로 넘기면 편향($b$)으로 표현됨

### 단층 퍼셉트론
**단순히 입력층과 출력층**으로만 이루어진 퍼셉트론 
<br>



#### 논리 게이트 구현
단층 퍼셉트론을 이용하면 **AND, NAND, OR 논리 게이트**를 쉽게 구현할 수 있음

* 논리 게이트란? <br>
디지털 회로에서 기본적인 논리 연산을 수행하는 데 사용되는 구성 요소예요.
게이트 연산에 쓰이는 것은 n개의 입력값과 하나의 출력값
<br>

| 게이트 | 조건 | 가중치·편향 예시 [w1, w2, b] |
| --- | --- | --- |
| **AND** | 둘 다 1일 때만 1 | `[0.5, 0.5, -0.7]` |
| **OR** | 둘 다 0일 때만 0 | `[0.6, 0.6, -0.5]` |
| **NAND** | 둘 다 1일 때만 0 | `[-0.5, -0.5, 0.7]` |
| **XOR** | 서로 다를 때만 1 | ❌ **가능한 조합이 없음** |


#### 한계
 XOR 게이트를 구현할 수 없다!!
퍼셉트론은 입력값과 가중치의 선형 결합으로 이루어진 구조이기 때문에, 단층 퍼셉트론은 직선 하나로 두 영역을 나눌 수 있는 선형 분류 문제에 대해서만 구현할 수 있습니다.

### 다층 퍼셉트론


<br>

기존의 AND, NAND, OR 등의 게이트를 조합하여 **층을 여러 겹 쌓아 만드는 퍼셉트론**
-> 단층 퍼셉트론이 구현할 수 없는 XOR 등 구현 가능
<br>

다층 퍼셉트론이 XOR을 구현할 수 있는 이유는 은닉층에서 비선형 활성화 함수를 적용해 새로운 특징을 만들고, 다음 층에서 그 특징들을 조합하기 때문

#### 두 개 이상의 은닉층을 갖는 MLP가 딥러닝의 시초 모델로 발전
- 심층 신경망 (Deep Neural Network, DNN) = 은닉층이 2개 이상인 신경망
- 딥러닝 (Deep Learning) = 심층 신경망을 학습(가중치를 스스로 찾도록) 시킴

3. 오답 노트를 통해 **내가 틀린 문제를 다시 학습** → ***역전파***

### 손실 함수
_모델이 정확한지 오차를 계산하는 방법_ <br>

⇒ **실제값(정답)과 예측값(출력값)의 차이를 계산**하는 데 사용되는 함수 <br>

1. 평균 제곱 오차
예측 값이 연속적이고, 예측 값과 실제 값의 차이를 직접 측정하는 회귀 문제에 적합 <br>


<br>

2. 이진 크로스 엔트로피

<br>
모델이 예측한 ‘정답이 1일 확률’을 p라고 가정

- 정답이 1이면 → p가 높아야 잘 예측한 것
- 정답이 0이면 → 1 − p가 높아야 잘 예측한 것

BCE는 **실제 정답에 부여한 확률에 −log를 적용한 손실** 정답의 확률이 높으면 손실이 작고, 낮으면 손실이 커짐

로그는 여러 데이터의 확률을 곱하는 계산을 합으로 바꿔줌. 앞의 마이너스는 ‘확률을 높이는 목표’를 ‘손실을 줄이는 목표’로 바꿔줌

$$
P(y\mid x)=p^y(1-p)^{1-y}
$$

**여기에 −log를 적용하면 BCE(Binary Cross Entropy)**

$$
L=-\log\left[p^y(1-p)^{1-y}\right]=-\left[y\log p+(1-y)\log(1-p)\right]
$$
<br>
(1) 실제 정답이 1일 때 → $L=-\log{\hat{y}}$만 남음. <br>

- 예측 확률이 1에 가까울수록 → loss가 0에 가까워짐 = 잘 맞힘
- 예측 확률이 0에 가까울수록 → loss가 급격히 증가 (확신했지만 틀림)
- ex) 실제 레이블이 고양이(1)인데 고양이일 예측 확률이 0.1이라면, 모델이 고양이가 아니라고 확신하며 틀린 것이므로 −log(0.1)로 계산되어 손실 값이 큼.

(2) 실제 정답이 0일 때 → 반대로 동작함. <br>

MSE는 단순히 loss의 크기만 보는 반면, BCE는 **얼마나 확신을 가지고 예측했는지**도 loss에 반영함.

3. 크로스 엔트로피 오차 
<br>



<br>
다중 클래스 분류 문제에서 예측 확률이 실제 클래스와 얼마나 일치하는지 평가하는 데 적합 <br>

### 손실 함수 정리

| **손실 함수** | **사용 상황** | **예시** |
| --- | --- | --- |
| **MSE** | 회귀 문제 | 주택 가격 예측, 온도 예측, 주식 가격 예측 |
| **BCE** | 이진 분류 문제 | 이메일 스팸 필터링(스팸/정상), 질병 진단(양성/음성) |
| **CEE** | 다중 클래스 분류 문제 | 손글씨 숫자 인식(MNIST), 이미지 분류(CIFAR-10) |

### 활성화 함수

#### 활성화 함수를 거치는 이유
**(1) 비선형성이 필요하다** <br>

활성화 함수가 없으면 층을 아무리 쌓아도 결국 **선형 계산의 반복**일 뿐. 1장에서 본 XOR 같은 문제를 못 풂. <br>

---

**(2) 미분이 되어야 한다** ← **2-1과 이어지는 부분** <br>

2-1에서 역전파는 **미분값을 연쇄적으로 곱해 나감** 그렇다면 활성화 함수는 **미분이 가능해야** 함. <br>

그런데 1장에서 쓴 **계단 함수는 미분이 안 됨.** 0에서 끊기고, 나머지 구간은 기울기가 0아님.
<br> 
⇒ **역전파를 할 수가 없음.**

#### 계단함수

<br>
입력값과 가중치의 곱을 모두 더한 뒤, 그 합이 임계값을 넘으면 1, 아니면 0을 출력하는 함수.<br> 주로 단층 퍼셉트론에서 사용됨. <br>

- 입력 신호: 입력값과 가중치의 곱
- 출력 신호: 0 또는 1

임계값은 뉴런이 얼마나 쉽게 활성화되는지를 제어하고, 가중치는 각 입력 신호가 결과에 미치는 영향력을 제어함. <br>
즉, 가중치는 각 입력 신호가 결과에 미치는 영향력을 제어.

#### 시그모이드 함수
 <br>

입력 신호를 0과 1 사이의 연속적인 값으로 변환하는 비선형 함수 <br>
계단 함수와의 차이점?: 출력값의 형태 <br>
계단 함수는 0 또는 1만 출력하는 반면, 시그모이드는 x값에 따라 출력값이 연속적으로 변화. 덕분에 "확실히 맞다/틀리다"가 아니라 얼마나 확실한지를 표현할 수 있고, 미분도 됨.

#### ReLU 함수
 <br>

입력이 0보다 크면 그대로, 0 이하면 0을 출력하는 비선형 함수

##### 왜 ReLU를 가장 많이 쓸까요?
계산이 간단하고 효율적 — 0보다 크냐 아니냐만 비교하므로 빠름 <br>
비선형성 제공 — 음수는 0으로, 양수는 그대로 처리하면서 은닉층에 비선형성을 추가 <br>
기울기 소실 완화  — 양수 구간에서 미분값이 항상 1이므로, 아무리 곱해도 값이 죽지 않습니다. 역전파가 안정적으로 전달됨

#### Softmax 함수

각 클래스에 대한 출력값을 확률로 변환하는 함수

##### 활성화 함수 정리

| 활성화함수 | 출력 범위 | 주요 사용처 | 장점 | 단점 |
| --- | --- | --- | --- | --- |
| 계단 함수 | 0 또는 1 | 단층 퍼셉트론 | 구조가 단순 | **미분 불가** |
| Sigmoid | 0 ~ 1 | 이진 분류 출력층 | **확률**로 해석 가능 | **기울기 소실** |
| Tanh | -1 ~ 1 | 은닉층 (요즘은 ReLU에 밀림) | 출력 중심이 0으로 학습 안정적 | **기울기 소실** |
| ReLU | 0 ~ ∞ | **은닉층 전반** | **계산 빠름, 기울기 소실 완화** | 음수 입력 시 출력이 항상 0 |
| Softmax | 0 ~ 1 (합=1) | 다중 분류 출력층 | 클래스별 확률 직접 출력 | 출력층 전용, 은닉층엔 부적합 |

### 경사하강법 종류

#### 배치 경사하강법


- 전체 데이터로 학습하므로 **업데이트 횟수가 적음** (1 Epoch 당 1 업데이트)
- 한 번에 처리하므로 **메모리가 많이 필요**
- 항상 같은 데이터에 대해 경사를 구하므로 **수렴이 안정적**

#### 확률적 경사하강법
**기존 경사하강법의 단점?** → 전체 데이터를 계산하므로 시간이 너무 오래 걸림

매개변수를 조정할 때 전체가 아니라 **랜덤으로 선택된 하나의 데이터**에 대해서만 계산합니다.



- 하나의 데이터만 메모리에 저장하면 되므로 **자원이 적은 컴퓨터에서도 사용 가능**
- 매개변수의 **변경폭이 불안정**하고 **정확도가 낮을** 수 있음

#### 미니 배치 경사하강법
배치 크기가 1이면 정확도가 낮고, 전체 배치는 메모리가 부족하다고 했죠. ⇒ 그래서 나온 게 **미니 배치**



- **배치 크기를 적절히 지정**하여 해당 개수만큼 매개변수를 조정
- 전체 데이터보다 **빠르고**, 배치 크기 1보다 **안정적** ⇒ **가장 많이 사용**

### 옵티마이저

#### 알엠에스프롭 (RMSProp)

기존 경사하강법은 무조건 현재 기울기만 보고 내려감. 그러나 RMSProp은 **최근 기울기를 참고**해서

- **급한 경사에서는 속도를 줄이고**
- **완만한 경사에서는 속도를 유지하거나 증가**

#### 아담 (Adam)

**RMSProp**과 **Momentum**을 합친, **"방향과 스텝 사이즈를 적절하게 합친 방법"**

</aside>

⇒ **방향**과 **학습률** 두 가지를 모두 잡기 위한 방법

각 파라미터의 1차 모멘트(기울기의 평균)와 2차 모멘트(기울기의 제곱 평균)를 사용하여 학습률을 조정

- 사용법?
    
    옵티마이저는 `compile()`의 `optimizer` 인자에서 지정 단순히 문자열 `'adam'`으로 써도 작동
    
    ```python
    adam = tf.keras.optimizers.Adam(learning_rate=0.001, beta_1=0.9, beta_2=0.999)
    model.compile(loss='categorical_crossentropy', optimizer=adam, metrics=['acc'])
    
    # 이렇게 짧게 써도 무방
    model.compile(loss='categorical_crossentropy', optimizer='adam', metrics=['acc'])
    ```
    
    `learning_rate=0.001` — **2-4 ②에서 본 그 α**
    
</aside>

#### 상황별 옵티마이저 사용법

| 이럴 때 | 이걸 쓰세요 | 왜 |
| --- | --- | --- |
| **잘 모르겠다 / 대부분의 경우** | **Adam** | 방향과 보폭을 둘 다 잡음. 사실상 기본값 |
| 학습이 자꾸 한쪽으로 튄다 | 모멘텀 | 방향을 안정화 |
| 파라미터마다 기울기 크기 차이가 크다 | RMSProp | 방향별로 학습률을 따로 조정 |


#### 한 줄 정리
- **모멘텀** = **방향**을 개선
- **RMSProp** = **보폭(학습률)**을 개선
- **Adam** = **둘 다** = 모멘텀 + RMSProp
