# NLP와 언어 모델

## NLP

> ### NLP란?

>> 1) 개념
- 인간의 **자연어**를 컴퓨터가 분석·이해·생성하도록 하는 AI 기술
- 언어를 **숫자**(벡터)로 변환한 뒤 단어 간 관계와 문맥을 분석
- 활용: 기계 번역(파파고), 감성 분석, 텍스트 분류(스팸 탐지), 문서 요약, 질의응답·챗봇

>> 2) 자연어 vs 프로그래밍 언어

- 자연어
    - 대응방식: **일대다**
    - 특징 유연하지만 모호성 높음

- 프로그래밍 언어
    - 대응방식: **일대일**
    - 특징: 명확하고 엄격

>> 3) 자연어의 특징과 NLP의 대응

- **이산성** : 단어·문자 등 구분된 기호 - **임베딩**으로 벡터화
- **순차성** : 순서에 따라 의미가 달라짐 - 앞뒤 문맥 고려 (RNN, Transformer)
- **중의성** : 하나의 단어에 여러 의미 (ex. 배) - **문맥 전체**를 파악 
- **희소성** : 단어·조합이 너무 다양 - 단어 유사도로 **확률적 추론** 
- **가변성** : 신조어 등 계속 변화 - 데이터로부터 패턴을 학습하는 ML 

>> 4) NLP의 역할 (분석 단계)
1. **형태소 분석**: 의미의 최소 단위로 분리 + 품사 분석
2. **구문 분석**: 문법 구조·단어 관계 파악 (주어·목적어·서술어)
3. **의미 분석**: 문맥에 맞는 의미 파악 (ex. 은행 → 금융기관)
4. **담화 분석**: 문장 간 연결·맥락 파악 (ex. '그' = 철수)

## 언어 모델

> ### 언어 모델의 정의
- 대량의 텍스트로 **단어(토큰)의 등장 확률**을 예측하고 문장의 자연스러움을 판단하는 모델
- 쉽게 말해 **인간이 말하고 쓰는 방식을 모방하는 시스템**
- 예: "나는 아침에 밥을      " → 먹었다 80% / 마셨다 15% / 달렸다 5%
- 분류: **SLM**(통계적) / **NLM**(신경망)

# 통계기반 모델

## SLM과 분포 가설

> ### 분포 가설 (Distributional Hypothesis)
- 비슷한 문맥에서 등장하는 단어는 **비슷한 의미**를 가질 가능성이 높다는 가설
- SLM의 근거이자, 이후 **임베딩 기술**의 이론적 기반
- 예: "아침에 **커피**를 마셨다" / "아침에 **차**를 마셨다" → 커피와 차는 의미적으로 관련

>> 1) 조건부 확률
- 앞선 단어들을 조건으로 다음 단어의 확률을 계산

$$
P(갔다 \mid 나는\ 학교에)
$$

>> 2) 카운트 기반 확률
- 특정 문맥 다음에 어떤 단어가 **얼마나 자주** 나왔는지로 확률 부여
- 예: "나는 학교에" 100회 → 갔다 70%, 왔다 20%, 뛰었다 10%

>> 3) 한계: 희소 문제 (Sparsity)
- 가능한 단어 조합이 많아 특정 표현이 데이터에 **거의/전혀 없는** 문제
- 한 번도 안 나온 조합은 **확률 0**으로 계산 (문법적으로 가능해도)
- 예: "나는 학교에 순간이동했다" → 0
- → 확률 추정 기법과 새로운 언어 모델이 필요

---

## N-gram이 나온 배경과 한계

> ### 등장 배경
- 문장이 길어질수록 SLM에 문제 발생
    - **데이터 희소성**: 학습에 없는 조합 증가
    - **계산 복잡성**: 모든 이전 단어를 고려하면 자원 소모 큼

> ### 정의
- 연속된 **N개 단어**의 묶음. **직전 N-1개 단어**만 보고 다음 단어를 예측
- N이 클수록 더 긴 문맥 반영
- 예(N=4): "An adorable little boy is spreading ___" → `boy`, `is`, `spreading`만 고려, 앞의 `An adorable little`은 제외

> ### SLM과의 관계
- 공통점: 빈도와 조건부 확률로 다음 단어를 예측
- 차이: **SLM**은 모든 이전 단어를 조건으로 표현 가능하나 긴 문맥 추정이 어려움 / **N-gram**은 N-1개로 제한해 단순화
- → N-gram은 별개 개념이 아니라 **SLM을 단순화한 대표 모델**

> ### 한계
- **희소 문제**: N↑ → 조합 수 급증 → 미등장 N-gram은 확률 0 → 더 많은 데이터 필요
- **문맥 제한**: 멀리 떨어진 단어 간 관계 파악 불가, 앞부분 핵심 정보 누락

---

## PPL (Perplexity)

> ### 정의
- 언어 모델이 문장을 예측할 때 **얼마나 혼란스러워하는지**를 나타내는 지표
- **낮을수록** 예측 불확실성이 낮음 = **잘 예측**
- **높을수록** 예측이 어려움

> ### 특징
- 훈련 데이터에 **희소한 문장**일수록 PPL이 높아지는 경향
- 예측하기 쉬운 단어가 추가되면 PPL 감소 가능 (단, N이 늘면 반드시 낮아지는 것은 아님)
- SLM뿐 아니라 **NLM의 성능 평가**에도 사용

---

## NLM과 SLM의 차이

> ### NLM의 등장
- SLM의 한계를 완화하기 위해 **인공신경망**으로 의미와 문맥을 학습
- 단어를 **연속적인 벡터**로 변환 → 단어 간 의미 관계 학습
- 관측되지 않은 표현도 **일반화**

> ### 핵심 키워드: 임베딩
- 이산적 언어 데이터를 **연속 벡터**로 변환, 단어를 고차원 공간의 한 점으로 표현
- 의미·문맥이 유사한 단어는 **벡터 공간에서 가깝게** 위치 (ex. moon, mars, space, nasa)

> ### 비교

>> #### 통계적 언어 모델 (SLM)
- 학습 방식: 텍스트 내 단어 등장 빈도 및 확률 계산
- 단어 표현: 단어를 독립적인 개별 기호로 취급
- 의미적 관계: 단어 간 의미적 유사성이나 관계를 반영하기 어려움
- 희소 문제: 학습 데이터에 없는 새로운 단어 조합(미관측 조합)에 취약함
- 문맥 처리: 직전 N-1개의 단어만 고려(N-gram)하여 문맥 파악 범위가 제한적임
- 계산 비용: 학습 및 연산에 드는 자원과 비용이 상대적으로 낮음

>> #### 신경망 언어 모델 (NLM)
- 학습 방식: 인공 신경망의 가중치 조절을 통한 학습
- 단어 표현: 단어를 연속적인 수치 벡터로 변환(임베딩)하여 표현
- 의미적 관계: 임베딩을 통해 단어 간의 유기적인 의미와 유사성 학습 가능
- 희소 문제: 학습된 벡터 표현을 활용하여 미관측 조합에 대한 취약성 완화
- 문맥 처리: 신경망 구조에 따라 다양하고 훨씬 넓은 범위의 문맥 파악 가능
- 계산 비용: 정교한 처리를 위해 학습 시 막대한 연산 자원과 비용 요구

# 딥러닝기반 모델

## BERT

> ### 정의
- Transformer의 **인코더** 기반, **양방향 문맥**을 학습해 의미를 이해하는 모델
- **B**idirectional / **ER** Encoder Representations / **T** from Transformers

> ### 특징
- **양방향**: 앞뒤 단어를 모두 고려
- **Transformer 인코더만** 사용 (Self-Attention)
- **사전학습 + 파인튜닝**

> ### 학습 과정
1. 토큰화 (`[CLS]`, `[SEP]` 특수 토큰 추가)
2. 임베딩 (의미·위치·문장 구분 정보 반영)
3. Transformer Encoder로 문맥 학습
4. 출력 → 분류·감성 분석·QA 등에 활용

> ### BERT 이후 모델
- **RoBERTa**: NSP 제거, Dynamic Masking, 더 많은 데이터·큰 배치
- **ALBERT**: 임베딩 차원 축소 + 계층 간 파라미터 공유로 경량화, NSP 대신 **SOP**

---

## GPT

> ### 정의
- Transformer **디코더** 기반, 이전 단어로 **다음 단어를 예측**해 텍스트를 생성하는 모델
- **Masked Self-Attention** + FFN (인코더가 없으므로 Cross-Attention 미사용)
- **자기회귀** 방식으로 순차 생성

> ### 최근 GPT: In-Context Learning
- 파인튜닝 없이 프롬프트의 지시·예시만으로 수행 (GPT-3)
- **Zero-shot**(지시만) / **One-shot**(예시 1개) / **Few-shot**(예시 여러 개)


---

## LLM

> ### 정의
- 방대한 텍스트와 대규모 신경망으로 언어를 이해·생성하도록 학습된 모델
- 대부분 Transformer 기반 (GPT 계열, LLaMA, Claude, Gemini 등)

> ### 작동 방식
- **학습**
    - **사전학습**: 다음 토큰 예측을 반복하며 언어 구조·문맥 학습
    - **파인튜닝**: 특정 목적의 데이터로 추가 학습
    - 예측값과 정답의 차이를 계산해 **가중치 업데이트**
- **예측**: 입력 토큰화·벡터화 → Attention으로 문맥 분석 → 다음 토큰 확률 계산 → 순차 생성

> ### LLM이 GPT 방식을 쓰는 이유
- BERT는 이해 중심, GPT는 **다음 토큰 순차 예측으로 생성**에 특화
- 질문에 새 문장을 생성해야 하는 대화형 LLM에는 **자기회귀 구조**가 적합

> ### 한계
- **환각(Hallucination)**: 사실과 다른 정보를 그럴듯하게 생성 (RAG 등으로 완화, 완전 해결은 아님)
- **높은 학습·업데이트 비용**: 최신 정보가 자동 반영되지 않음

---

## RAG의 구조

> ### 정의
- **외부 지식 검색 + LLM 생성**을 결합해 답변의 **정확성·최신성**을 높이는 기술
- 배경: 기존 LLM은 학습 시점 이후 지식에 답변 불가

> ### 구조

- **질의 인코더**: 질문을 임베딩해 숫자 벡터로 변환
- **지식 검색기**: Vector DB에서 질문과 **유사도 높은 문서** 검색
- **지식 증강 생성기**: 질문 + 검색 문서를 LLM에 입력해 답변 생성

> ### 장점
- **최신 정보 반영**: 재학습 없이 외부 DB·문서 활용
- **정확성 향상**: 검증 가능한 근거로 환각 완화

> ### CAG (Credibility-aware Generation)
- 모델이 **신뢰도를 스스로 판단·활용**하도록 훈련
- RAG의 한계: 관련성이 높아도 오래되거나 부정확한 정보일 수 있음
- 관련성 + **출처 신뢰성 + 최신성**을 종합 평가해 신뢰도 높은 정보를 우선 활용

## Langchain의 아이디어, 특징

> ### LangChain
- LLM을 외부 데이터·API·도구와 연결해 **복잡한 AI 앱을 쉽게 개발**하게 하는 오픈소스 프레임워크
- 배경: 구현의 복잡성, 외부 시스템 연동, 유지보수 어려움 → **모듈화·연결**

> ### LangGraph
- **노드와 엣지**로 연결해 **조건 분기·반복**이 가능한 워크플로 구현
- LangChain의 한계: 선형 구조에 적합, 결과에 따른 경로 변경·재실행 어려움
- 구성: **Node**(LLM·Agent·Tool) / **Edge**(실행 경로) / **State**(공유 정보)

> ### LangSmith
- LLM 앱의 **추적·평가·모니터링** 플랫폼
- 기능: 디버깅·추적 / 테스트·평가 / 모니터링 / 프롬프트 허브

# sLM

## 모델 압축

> ### sLM이란?
- LLM보다 규모가 작은 언어 모델 (sLM: 수백만~수십억, LLM: 수십억~수천억 이상 매개변수)
- 메모리·연산 적게 사용 → 모바일 등 제한된 환경, **오프라인 추론** 가능

> ### 모델 압축
- 성능은 최대한 유지하면서 **크기와 연산량을 줄이는** 기술

- **가지치기** (Pruning): 중요도 낮은 연결·뉴런 제거 (비구조적/구조적), 크기·연산 감소, 과도하면 성능 저하
- **양자화** (Quantization): 숫자 정밀도를 낮춤 (FP32 → FP16/INT8), 메모리↓ 속도↑, 일부 성능 저하 
- **저차원 분해** (Low-Rank Factorization): 큰 행렬을 작은 행렬들의 곱으로 근사, 파라미터·연산 감소 
- **지식 증류** (Knowledge Distillation): Teacher 모델의 지식을 Student 모델에 전달, 작은 모델로 높은 성능 

- ※ **LoRA**: 저차원 분해 원리를 활용한 미세조정 방법. 모델 전체 압축이 아니라 **가중치 변경량**만 작은 행렬로 학습

> ### sLM 예시
- **DistilBERT**(지식 증류), **Gemma**(Google), **GPT-4o mini**(OpenAI), **Llama**(Meta), **Ministral**(Mistral, 온디바이스), **Phi**(Microsoft)

## sLM의 장단점

- 장점
    - 성능: 빠른 응답이 가능하며, 자원이 제한된 환경에서도 원활하게 실행할 수 있음
    - 보안 및 데이터: 외부 통신 없이 내부 서버나 개인 기기(온디바이스)에서 독립적으로 실행되어 데이터 보호에 탁월함
    - 비용: 모델 운영 비용, 전력 소모, 하드웨어 구축 등 전반적인 유지 비용을 크게 절감할 수 있음
- 단점 (한계)
    - 성능: 복잡한 추론이나 고도의 전문 지식이 필요한 작업에서는 정확도가 떨어져 범용성이 낮음
    - 보안 및 데이터: 학습 데이터에 내재된 편견이 결과물에 그대로 반영되는 편향(Bias) 문제가 발생할 수 있음
    - 신뢰도: 잘못된 정보나 거짓을 마치 사실인 것처럼 생성해 내는 환각(Hallucination) 현상이 발생할 수 있음


## sLM과 LLM의 결합

> ### 결합 추론 (Hybrid Inference)
- sLM과 LLM을 **함께 활용**해 장점을 살리는 방식
- **간단한 작업**(분류, 간단한 요약) → sLM
- **복잡한 작업**(다단계 추론, 전문 분석) → LLM
- 예: 챗봇에서 간단한 질문은 sLM이 빠르게, 복잡한 분석은 LLM이 답변
- 핵심: **성능과 비용의 균형**

# 분류모델과 생성모델의 차이

> ### 생성 모델이란?
- 학습 데이터의 **분포를 따르는 유사 데이터**를 생성하는 모델



# 생성모델의 분류

> ### 분류 기준: 확률밀도를 직접 계산할 수 있는가?

- **명시적**: $P(X)$를 수학적으로 정의해 **수치로 출력** 가능
    - **Tractable**: 직접 계산 (조건부 확률 순차 계산 등)
    - **Approximate**: 근사 (**변분 추론**, **마르코프 연쇄**)
- **암시적**: $P(X)$ 수치는 주지 않고 **샘플링**(생성)에 집중
    - **직접 샘플링**: 노이즈를 한 번 통과시켜 생성 (GAN)
    - **간접/반복 샘플링**: 여러 단계의 상태 변화를 반복 (GSN)
- 용어
    - **변분 추론**: 계산하기 어려운 분포를 다루기 쉬운 분포(정규분포 등)로 근사
    - **마르코프 연쇄**: 다음 상태가 **현재 상태에만** 의존 (맑음→맑음 80%, 비→맑음 40%)

## AE

> ### 정의
- 입력의 중요한 특징을 **압축**하고 원본과 유사하게 **복원**하도록 학습하는 신경망
- 정답 레이블 없이 입력 자체가 목표
- 활용: 데이터 압축, 노이즈 제거, 이상 탐지

> ### 구조

- 기존 인코더·디코더는 **새 결과 생성**, AE는 **입력 자체를 복원**

> ### 왜 압축하고 복원할까?
- **압축**: 병목이 없으면 단순 복사 → 크기를 제한해 **핵심 특징**을 학습하도록 유도
- **복원**: 복원 오차(Reconstruction Error)를 줄이며 특징 표현을 개선
- 결과: 데이터의 **핵심 구조**(Non-trivial Structure) 학습

## VAE

> ### 등장 배경
- AE의 잠재 공간은 **불규칙**, 빈 영역에서 샘플링하면 부자연스러운 결과
- AE는 기존 데이터 복원에는 좋으나 **새 데이터 생성**에 한계

> ### 정의
- 입력 특징을 **확률분포**(가우시안)로 학습, 샘플링해 새 데이터를 생성하는 모델

> ### 학습 목표
1. **복원 오차 최소화**
2. **KL 발산 최소화**: 잠재분포가 **표준정규분포**에 가까워지도록 (두 분포가 같으면 0)

> ### AE vs VAE

| 구분 | AE | VAE |
| --- | --- | --- |
| 표현 | 고정된 **점** | **확률분포** (평균 $\mu$, 분산 $\sigma^2$) |
| 잠재공간 | 불규칙, 빈 영역 존재 | 연속적·규칙적 |
| 생성 | 어려움 | 샘플링으로 새 데이터 생성 |

> ### 작동 과정

>> 1) 분포 예측 (Probabilistic Encoder)
- 인코더가 $\mu$, $\sigma$를 예측

$$
q_\phi(z \mid x)
$$

>> 2) 샘플링 및 재파라미터화
- 무작위성을 노이즈 $\varepsilon$으로 분리 → **기울기 전달 가능**

$$
z = \mu + \sigma \odot \varepsilon, \quad \varepsilon \sim \mathcal{N}(0, I)
$$

- 예: $\mu=2,\ \sigma=0.5$ → $\varepsilon=1$이면 $z=2.5$, $\varepsilon=-1$이면 $z=1.5$

>> 3) 생성 (Probabilistic Decoder)

$$
z \sim \mathcal{N}(0, I) \rightarrow x' \sim p_\theta(x \mid z)
$$

- 학습 후에는 입력 없이 **표준정규분포에서 z를 뽑아** 새 데이터 생성

## GAN의 구조와 장단점

> ### 개념
- 2014년 Ian Goodfellow 제안. **생성자**(G)와 **판별자**(D)가 경쟁하며 학습
- 비유: **위조지폐범(G)** vs **경찰(D)**
- 최적점: G는 진짜와 구분 불가한 데이터를 생성, D는 확률 **0.5**(찍는 수준)

> ### 구조
- **노이즈 z** → **G**: 가짜 $G(z)$ 생성 → **D**: 진짜일 확률(0~1) 출력
- 예: $D(x)=0.9$(진짜), $D(G(z))=0.1$(가짜)

> ### 최소-최대 게임 (Minimax)
- **D**: 진짜/가짜 구별 능력을 **최대화**
- **G**: D를 속여 구별 능력을 **최소화**
- 경쟁이 반복되며 서로 성능 향상

> ### 장점
- **진짜 같은 가짜** 생성 (존재하지 않는 얼굴)
- **높은 선명도**: VAE는 픽셀 평균(MSE) 최적화로 **흐릿**, GAN은 D를 속이는 목표로 미세 질감 표현
- **빠른 생성 속도**: 노이즈를 한 번 통과 (확산 모델 대비)
- **잠재공간의 부드러움** (단, 모든 GAN이 보장되지는 않음)

> ### 한계점과 해결
- 학습 불안정성
    - 원인·결과: G(생성자)와 D(판별자)의 학습 균형 붕괴로 인한 진동 및 무한루프 발생
    - 해결 방법: 학습률 및 학습 횟수 조절, WGAN-GP 기법 도입

- 훈련 초기 문제
    - 원인·결과: D가 가짜를 너무 쉽게 구별해버려 G로 전달되는 기울기 신호가 약해짐
    - 해결 방법: Non-saturating Loss 적용, D의 학습 속도 및 비중 조절

- 모드 붕괴 (Mode Collapse)
    - 원인·결과: G가 D를 속일 수 있는 소수의 특정 패턴(유형)만 반복해서 생성하여 결과물의 다양성 감소
    - 해결 방법: Minibatch Discrimination 기법 적용, WGAN-GP 사용

- 평가 지표의 한계
    - 원인·결과: 결과물의 사실성, 선명도, 다양성을 단일 수치로 완벽하게 평가하기 어려움
    - 해결 방법: FID(Fréchet Inception Distance) 및 IS(Inception Score) 지표 활용, 사람의 정성적 평가 병행
- 역매핑 어려움
    - 원인·결과: 생성된 이미지에서 원본 잠재 벡터(Latent Vector)를 역으로 추적해 내는 구조가 기본적으로 부재함
    - 해결 방법: GAN Inversion 기법, BiGAN 모델 등 활용

> ### 적용 사례
- **NVIDIA Progressive GAN**: 저해상도→고해상도로 점진 확장, 가상 인물 얼굴 생성
- **Deep Video Portraits (cGAN)**: 한 사람의 표정·움직임을 다른 사람의 영상에 합성
- **Facebook ExGAN**: 참고 사진을 활용해 감은 눈을 뜬 눈으로 복원 (Eye In-Painting)

## 확산모델

> ### 개념
- 이미지에 노이즈를 추가한 데이터를 학습하고, **무작위 노이즈를 점차 제거**하며 새 이미지를 생성
- 활용: 이미지 생성·복원·편집, 영상 생성
- **모호성 모델링**: 하나의 조건에서 여러 결과가 가능하다는 **불확실성을 확률적으로 표현** → 같은 프롬프트로 다양한 결과 생성

>> 순확산과 역확산

>> 순확산과 역확산

- **순확산 (Forward)**
    - 방향: $x_0 \rightarrow x_T$
    - 과정: **가우시안 노이즈**를 단계적으로 추가
    - 시작: 깨끗한 원본 이미지
    - 결과: 무작위 노이즈에 가까운 상태
    - 목표: 노이즈 제거법을 학습하기 위한 데이터 생성
- **역확산 (Reverse)**
    - 방향: $x_T \rightarrow x_0$
    - 과정: **Denoising U-Net**이 노이즈를 예측해 단계적으로 제거
    - 시작: 무작위 노이즈
    - 결과: 새로운 이미지
    - 목표: 학습한 노이즈 제거법으로 이미지 생성

- 가우시안 노이즈: **정규분포를 따르는 무작위 잡음**

> ### 장점
- **압도적 샘플 품질**: 단계적 제거로 정교하고 다양한 이미지
- **학습 안정성**: 경쟁 구조 없이 노이즈를 예측, **MSE**로 손실 확인 가능
- **다양성 확보**: 무작위 노이즈에서 시작

> ### 단점
- **느린 생성 속도**: 반복적인 노이즈 제거 (GAN은 한 번에 생성)
- **높은 계산 비용**: GPU 등 고성능 자원 필요
- **구조적 제약**: 단계 수·방식에 따라 속도·품질이 달라 설계가 필요
