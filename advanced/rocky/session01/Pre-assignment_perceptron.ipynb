# 딥러닝 기초

## 1. 단층 퍼셉트론

### 1-1. 인공신경망과 퍼셉트론

> **인공신경망**: 뇌의 신경망(신경 세포끼리 연결되어 정보를 처리하는 구조)을 모방한 모델

>> 생물학적 뉴런의 처리 구조
  - 가지돌기로 외부 신호 수신
  - 모인 신호가 일정 크기 이상이면
  - 축삭돌기로 신호 전달
- 퍼셉트론의 처리 구조
  - 입력(x)에 가중치(w)와 편향(b) 적용
  - 종합한 값을 활성화 함수에 전달
  - 임계값 기준으로 출력 결정

>> **퍼셉트론**: 신경망(딥러닝)의 기원이 되는 **선형 이진 분류기**

- 출력 절차: **입력 → 가중합 → 활성화 함수 → 출력**
  - 입력값 × 가중치 계산
  - 곱한 값들의 총합 계산 (편향은 선택)
  - 총합이 임계값(θ)을 넘으면 1, 아니면 0 (계단 함수)
- **학습(Training)**: 원하는 출력이 나오도록 가중치를 조정하는 작업

### 1-2. 핵심 용어

- **가중치(w)**: 입력이 출력에 미치는 영향력(중요도). 클수록 중요한 입력
- **활성화 함수**: 임계값 초과 여부에 따라 출력을 결정하는 함수
- **임계값(θ)**: 뉴런이 얼마나 쉽게 활성화되는지 결정하는 기준
- **편향(b)**: 임계값을 좌변으로 넘긴 형태. 기준점 역할

> **임계값과 편향은 표현의 차이**
- 임계값: Σwx ≥ θ → 1
- 편향: Σwx + b ≥ 0 → 1
- 현재 딥러닝에서는 **편향** 개념을 주로 사용

### 1-3. 단층 퍼셉트론과 논리 게이트

> **단층 퍼셉트론**: 입력층과 출력층만으로 구성된 퍼셉트론

- AND, NAND, OR 게이트를 구현 가능

>> #### 1 AND 게이트

**입력이 모두 1일 때만 1 출력**

- 가중치·편향 예시 `[w1, w2, b]`: `[0.5, 0.5, -0.7]`
- 다른 조합도 가능: `[0.5, 0.5, -0.8]`, `[1.0, 1.0, -1.0]`

>> #### 2 OR 게이트

**입력이 모두 0일 때만 0 출력**

- 가중치·편향 예시: `[0.6, 0.6, -0.5]`

>> #### 3 NAND 게이트

**입력이 모두 1일 때만 0 출력**

- 가중치·편향 예시: `[-0.5, -0.5, 0.7]`
- AND의 부호를 뒤집은 형태

>> #### 4 XOR 게이트

**입력이 서로 다를 때만 1 출력**

- 가중치·편향 **조합이 존재하지 않음** 
- 단층 퍼셉트론으로 구현 불가 → 다층 퍼셉트론 필요

> NAND는 AND의 부호를 뒤집은 것

### 1-4. 단층 퍼셉트론의 한계

> **XOR 게이트를 구현할 수 없음**

- 원인: 입력과 가중치의 **선형 결합** 구조
  - 직선 하나로 두 영역을 나누는 **선형 분류 문제**만 해결 가능
  - AND → 직선 하나로 분리 가능
  - XOR → 곡선/비선형 영역이 필요하여 분리 불가
- 단층 퍼셉트론도 계단 함수(비선형)를 쓰지만, 가중합 한 번에 임계값을 적용하므로 결정 경계는 직선
- 해결책: **층을 여러 개 쌓기** → 다층 퍼셉트론

## 2. 다층 퍼셉트론 (MLP)

> **MLP**: 입력층 + 하나 이상의 은닉층 + 출력층으로 구성된 퍼셉트론

- XOR 구현 방식
  - 은닉층의 두 뉴런이 각각 **NAND**와 **OR** 계산
  - 출력층이 두 결과에 **AND** 적용
  - **뉴런 2개짜리 은닉층 하나**로 충분

>> **층을 쌓는 것만으로는 비선형성이 생기지 않음**
- 활성화 함수 없이 가중합·편향만 계산하면 여러 층이 하나의 연산으로 합쳐짐
- y = W₂(W₁x + b₁) + b₂ = (W₂W₁)x + (W₂b₁ + b₂) = Wx + b
- 결정 경계가 여전히 직선 → XOR 불가
- **핵심**: 은닉층에서 **비선형 활성화 함수**로 새로운 특징을 만들고, 다음 층에서 이를 조합함

- h = σ(W₁x + b₁), y = W₂h + b₂ (σ: 비선형 활성화 함수)

>> 용어 정리
  - **심층 신경망(DNN)**: 은닉층이 2개 이상인 신경망
  - **딥러닝**: 심층 신경망을 학습(가중치를 스스로 찾도록)시키는 것

## 3. 딥러닝 모델의 학습 방법

### 3-1. 순전파와 역전파, 연쇄 법칙

> **학습 알고리즘의 기본 구조**

 1. **순전파**: 입력으로부터 예측 수행
 2. **오차 계산**: 예측 정확도 측정
 3. **역전파**: 출력층 → 입력층 방향으로 오차 전달
 4. **가중치 업데이트**
 5. 1~4 반복 → 오차가 0에 가까워짐


#### 1. 연쇄 법칙 (Chain Rule)

> **합성함수의 미분은 각 함수의 미분의 곱**으로 나타낼 수 있음

- 역전파에서 각 가중치가 **최종 손실에 얼마나 기여했는지** 계산하는 데 사용
- ∂L/∂w₁ = ∂L/∂aₙ × … × ∂a₂/∂a₁ × ∂a₁/∂w₁
- 예: w₁ → w₂ → L 로 간접 연결된 경우
  - w₁의 책임 = (w₁이 w₂에 준 영향) × (w₂가 L에 준 영향)
- 기울기 = “손실을 줄이려면 가중치를 어떻게 바꿀까?”를 알려주는 신호

#### 2. 기울기 소실 문제 (Vanishing Gradient)

> 은닉층이 많고 각 단계의 미분값이 1보다 작으면, 곱할수록 **기울기가 0에 수렴**

- 0.25 × 0.25 × 0.25 × … ≈ 0
  - 0.25 = **시그모이드 미분의 최댓값**
- 결과: 앞쪽 층의 가중치가 거의 업데이트되지 않아 **학습이 안 됨**
- 해결: **ReLU**

---

### 3-2. 손실 함수의 정의와 종류

> **손실 함수**: 모델 예측이 얼마나 틀렸는지를 수치로 표현하는 함수
>> → 실제값(정답)과 예측값(출력값)의 차이를 계산


- **목적함수**:  최적화하려는 전체 대상 최대화/최소화
- **손실함수**: **데이터 한 개**에 대한 오차 최소화 
- **비용함수**: **전체 데이터**에 대한 평균 오차 최소화

손실함수와 비용함수는 실무에서 혼용하는 경우가 많음 (공통점: 최소화가 목적)

#### 1 평균 제곱 오차 (MSE)

- 식: (1/N) Σ(y − ŷ)²
- **연속적인 값**을 예측하는 **회귀 문제**에 적합
- 예: 집 가격 예측

#### 2 이진 크로스 엔트로피 (BCE)

- 식: L = −(y log ŷ + (1−y) log(1−ŷ))
- **이진 분류 문제**(로지스틱 회귀)에 사용
- 의미: 실제 정답에 부여한 확률에 **−log**를 적용한 손실
> 
  
- 정답이 1 → L = −log ŷ (예측 확률이 1에 가까울수록 손실 감소)
- 정답이 0 → 반대로 동작
- 확신했지만 틀리면 손실이 급격히 증가
- MSE는 오차 크기만 보지만, BCE는 **얼마나 확신하고 예측했는지**까지 반영

- 이름에 “엔트로피”가 들어가는 이유: 엔트로피 = 정보의 불확실성 → 확실성까지 평가하기 때문

#### 3 크로스 엔트로피 오차 (CEE)

- 식: −Σ y log ŷ (y: 원-핫 인코딩, ŷ: 예측 확률)
- **다중 클래스 분류**에 사용

> **(−)가 붙는 이유**:  
- 0~1 사이 값에 log를 취하면 항상 음수 → 부호를 뒤집어 손실을 양수로 만들기 위함

---

### 3-3. 활성화 함수의 정의와 종류

> **활성화 함수**: 입력 × 가중치 + 편향 값을 받아 출력값으로 변환하는 함수

> #### 왜 필요한가
1. **비선형성 필요**: 없으면 층을 아무리 쌓아도 선형 계산의 반복 → XOR 불가
2. **미분 가능해야 함**: 역전파가 미분값을 연쇄적으로 곱하기 때문

> 발전 과정
  - **계단 함수**: 미분 불가 → 역전파 자체가 불가
  - **시그모이드**: 미분 가능, 그러나 최대 0.25 → 기울기 소실
  - **ReLU**: 양수 구간 미분값 1 → 기울기 소실 완화


#### 1. 계단 함수

> **임계값을 넘으면 1, 아니면 0을 출력하는 함수**

$$
h(x) = \begin{cases} 1 & (x \geq 0) \\ 0 & (x < 0) \end{cases}
$$

- 입력 신호: 입력값과 가중치의 곱
- 출력 신호: 0 또는 1

- **임계값**은 뉴런이 얼마나 쉽게 활성화되는지를 제어
- **가중치**는 각 입력 신호가 결과에 미치는 영향력을 제어합니다.

- 즉, **가중치는 각 입력 신호가 결과에 미치는 영향력을 제어**



> 계단 함수는 **미분값이 0**이라 역전파를 할 수 없음 

- 그래서 다층 퍼셉트론에서는 쓰지 않음

#### 2. 시그모이드 함수

$$
f(x)=\frac{1}{1+e^{-x}}
$$

입력 신호를 **0과 1 사이의 연속적인 값**으로 변환하는 비선형 함수

> **계단 함수와의 차이점**: 출력값의 형태
- 계단 함수는 0 또는 1만 출력하는 반면, 시그모이드는 x값에 따라 출력값이 연속적으로 변화함 

- "확실히 맞다/틀리다"가 아니라 **얼마나 확실한지** 표현 가능, **미분** 가능


#### 3. ReLU (가장 많이 사용)


$$
f(x)=\max(0,x)
$$

> **입력이 0보다 크면 그대로, 0 이하면 0을 출력하는 비선형 함수**

- 가장 많이 쓰는 이유
  - **계산이 간단하고 효율적**: 0보다 큰지만 비교
  - **비선형성 제공**: 은닉층에 비선형성 추가
  - **기울기 소실 완화**: 양수 구간 미분값이 항상 1 → 곱해도 값이 죽지 않음
- 단점: 음수 입력 시 출력이 항상 0 (Dead Neuron)

> **Leaky ReLU**: 
- 음수 구간에 미세한 기울기를 부여한 변형
- 다만 실무에서는 양수를 살리는 것이 더 중요하여 기본 ReLU를 많이 사용

#### 4. Softmax

$$
f(x_i)=\frac{e^{x_i}}{\sum_j e^{x_j}}
$$

> **각 클래스의 출력값을 확률로 변환하는 함수**

- **다중 클래스 분류의 출력층**에서 사용
- 출력값이 모두 0~1, 전체 합 = 1 → 클래스별 확률을 직접 해석 가능
- 예: 고양이/강아지/새 분류 → [0.7, 0.2, 0.1]

> **Softmax + 크로스 엔트로피(CEE)**

- 예측 확률과 정답 확률의 차이를 바로 계산 가능 → 학습 효율 높음
- 가장 표준적인 다중 분류 조합

### 3-4. 경사하강법

- **목표**: 손실을 가장 작게 만드는 W를 찾는 것
- **갱신식**: W ← W − α · ∂E/∂W

> #### 학습률 (α)

- 한 번에 **얼마나** 움직일지 정하는 보폭 (개발자가 직접 설정하는 **하이퍼파라미터**)  

>> 학습률 -> 결과
- 너무 작음 -> 수렴은 하나 너무 오래 걸림
- 너무 큼 -> 최솟값을 지나쳐 **발산**
- 적당해야함

권장: **0.001부터 시작** (Adam 기본값)

자동 탐색: Learning Rate Scheduler, Grid/Random Search

> #### 배치 · 에포크 · 이터레이션

- **배치** : 매개변수 조정에 사용하는 데이터 묶음
- **에포크** : 전체 데이터셋을 학습하는 횟수
- **이터레이션** : 1 에포크를 끝내는 데 필요한 배치 수

- 1 에포크 데이터 수 = 배치 크기 × 이터레이션
  - 예: 데이터 2,000개, 배치 200 → 이터레이션 10
- 배치가 너무 크면 학습이 느리고 메모리 부족, 너무 작으면 학습이 불안정

> #### 경사하강법의 종류


- **배치 GD**: 전체 데이터를 한 묶음으로 계산, 수렴 안정적, 업데이트 적음
- **SGD**: 랜덤 1개 데이터로 계산, 변동 불안정, 정확도 낮을 수 있음
- **미니 배치 GD**: 적절한 크기로 묶어 계산 빠르고 안정적 → **가장 많이 사용**

### 3-5. 옵티마이저의 정의와 종류

**옵티마이저**: 경사하강법을 더 똑똑하게 굴리기 위한 방법

> #### 왜 필요한가: Local Minima

- MSE·RMSE는 볼록함수 → 경사하강법으로 반드시 바닥에 도달
- 딥러닝의 손실함수는 층이 많아 **울퉁불퉁**
  - 기울기가 0이면 멈추는데, **Local Minimum**에서도 기울기는 0 → 갇힘
  - 기울기가 완만한 구간은 빠져나오는 데 오래 걸림

작은 기울기에서는 조금씩, 큰 기울기에서는 많이 움직이는 한계를 보완해야 함

> #### 1. 모멘텀 (Momentum)

>> **이전에 가던 방향으로 계속 가도록 속도(관성)를 추가**

- 기존 경사하강법: 이전 기울기를 잊는 “기억상실증”
- 관성을 가진 공처럼 작은 언덕을 넘을 수 있음
- 효과: **Local Minima 탈출**, 기울기가 작은 구간 빠르게 통과
- 실제 이동 = 현재 기울기 + 이전 이동(관성)
- β(모멘텀 계수): 일반적으로 **0.9**

---
```python
tf.keras.optimizers.SGD(learning_rate=0.01, momentum=0.9)
```
---

> #### 2. RMSProp

>> **방향마다 학습률을 따로 조정**

- 문제: 기울기가 큰 방향은 너무 많이, 작은 방향은 너무 조금 이동
- 최근 기울기를 참고
  - **급한 경사** → 속도 감소
  - **완만한 경사** → 속도 유지/증가

> #### 3. Adam

>> **모멘텀 + RMSProp**: 방향과 보폭을 모두 개선

- 1차 모멘트(기울기 평균)와 2차 모멘트(기울기 제곱 평균)로 학습률 조정
- 옵티마이저는 `compile()`의 `optimizer` 인자에서 지정
- 단순히 문자열 `'adam'`으로 써도 작동

---
```python
adam = tf.keras.optimizers.Adam(learning_rate=0.001, beta_1=0.9, beta_2=0.999)
model.compile(loss='categorical_crossentropy', optimizer=adam, metrics=['acc'])

# 이렇게 짧게 써도 됩니다
model.compile(loss='categorical_crossentropy', optimizer='adam', metrics=['acc'])
```
---

> #### 선택 가이드

- 잘 모르겠음 -> **Adam** : 방향·보폭 모두 개선, 사실상 기본값
- 학습이 한쪽으로 튐 -> 모멘텀 : 방향 안정화
- 파라미터별 기울기 크기 차이가 큼 -> RMSProp : 방향별 학습률 조정

> **결론: 
- 일단 Adam으로 시작하고, 잘 안 되면 그때 바꾼다**
- 모멘텀 = **방향** 개선 / RMSProp = **보폭** 개선 / Adam = **둘 다**
