# AI 트렌드

## AI 에이전트란?

> ### AI 에이전트의 정의

>> 1) 정의
- **AI 에이전트**: 사용자가 정한 목표를 달성하기 위해 AI 모델을 활용해 **환경과 상호작용하는 시스템**
    - **추론 + 계획 + 도구 실행**이 하나로 맞물려 작동
- 비유 (회사 사장 기준)
    - **LLM** = 자문위원: 아는 건 많지만 말로만 답함
    - **에이전트** = 비서: 목표만 주면 계획하고 직접 움직임

>> 2) 생성형 AI vs 에이전틱 AI
- **생성형 AI**
    - 프롬프트에 맞는 결과물을 **한 번 출력**하고 끝
- **에이전틱 AI**
    - 목표를 달성할 때까지 **추론과 행동을 반복**

> ### ReAct (생각 → 행동 → 관찰)

>> 1) 개념
- **ReAct** = Reasoning(추론) + Acting(행동)
    - **생각 → 행동 → 관찰 → 생각**의 순환 반복
- **관찰 결과에 따라 다음 생각이 달라짐**이 핵심
- 예시: "주문 #1234 배송이 늦어요" (CS 에이전트)
    - 생각: 주문 상태 확인 → 행동: `get_order(1234)` → 관찰: 4일 지연
    - 생각: 지연 원인 확인 → 행동: `track_shipment()` → 관찰: 응답 없음
    - 생각: 일시적 오류일 수 있어 재시도 → 관찰: 물류센터 적체
    - 생각: 배송비 환불 대상 → 행동: 사과 + 환불안 작성
    - 생각: 돈이 나가니 확인 필요 → 행동: 담당자 승인 요청
    - 승인 후 환불 + 메시지 발송 → 종료
    - 재시도가 계속 실패하면 상담원 연결로 계획 변경

>> 2) 에이전트가 폭주하지 않게 하는 장치
- 최대 반복 횟수
- 비용 한도
- 중복 실행 방지 (환불이 두 번 나가지 않게)
- 중요한 행동 전 **사람 승인**
- 목표만 주면 알아서 성공하는 것이 아님 → **정해진 범위 안에서 순환**

>> 3) 주의
- 모델은 도구를 **직접 실행하지 않음**
    - "이 도구를 이런 값으로 불러 달라"는 **요청만** 내놓고, 실제 실행은 **앱**이 수행

> ### 다중 에이전트

>> 1) 구조: 오케스트레이터 + 서브 에이전트
- **MAS(Multi Agent System)**: 여러 에이전트가 역할을 나눠 함께 일하는 구조
- **오케스트레이터(지휘자)**
    - 요청을 작은 일로 쪼갬 → 누구에게 맡길지 결정 → 결과 통합
- **서브 에이전트(전문가)**
    - 맡은 역할만 수행 (검색봇은 검색만, 검토봇은 검토만)

>> 2) 역할을 나누는 이유
- **스스로 실수를 잡기 어려움**
    - 단일 에이전트는 계획·실행·검증을 혼자 해 처음 판단을 그대로 믿기 쉬움
    - 다른 에이전트가 처음 보는 눈으로 재확인
- **동시에 여러 갈래 조사**
    - 서브 에이전트마다 자기 컨텍스트 창 → 더 빠르고 기억 부담 감소
- 사례: Anthropic 리서치 기능
    - 단일 에이전트보다 내부 평가에서 **90.2%** 좋은 결과
    - 대신 토큰은 일반 채팅의 약 **15배**

>> 3) 비용·성능 조합
- 모든 역할에 비싼 모델 X → 역할별로 **급이 다른 모델을 혼합**
    - 오케스트레이터: 고성능 모델 (판단 능력이 전체 품질 좌우)
    - 검색·정리봇: 경량 모델
    - 코드 검증봇: 코드 특화 모델

## MCP란?

> ### 도구 호출의 원리: Function Calling

>> 1) 개념
- 모델이 "이 함수를 이 인자로 불러 줘"라는 **구조화된 요청**을 내놓는 기능
    - 실제 실행과 결과 전달은 **앱**이 담당
>> 2) 흐름
- 앱 → 모델: 질문 + 쓸 수 있는 함수 목록
- 모델: 어떤 함수를 어떤 값으로 부를지 출력 (예: `get_weather`, `{"city": "서울"}`)
- 앱: 함수를 실제 실행
- 앱 → 모델: 실행 결과 전달 → 모델이 최종 답 작성

> ### MCP가 필요한 이유: n×m → n+m

>> 1) 표준이 없을 때
- 앱마다 × 도구마다 연동 코드를 따로 작성 → **n×m개**
- 도구가 하나 늘 때마다 작업량이 크게 증가
>> 2) 표준이 생기면
- 앱은 MCP 클라이언트 1번, 도구는 MCP 서버 1번 → **n+m개**
>> 3) MCP 정의
- **MCP(Model Context Protocol)**: LLM 앱과 외부 데이터·도구(서버)가 통신하는 **개방형 표준 프로토콜**
    - 2024.11 Anthropic 공개 → OpenAI 등 채택 → 2025.12 Agentic AI Foundation에 기증
    - 비유: AI 앱과 도구 사이의 **USB-C**
>> 4) 해결된 것
- n×m → n+m
- **발견의 표준화**: 클라이언트가 도구 목록 요청 → 서버가 반환
- **원격 서버 인증 통일**: OAuth 표준
>> 5) Function Calling vs MCP
- **경쟁 관계가 아님**, 실제 시스템은 둘 다 사용
    - Function Calling: 모델이 도구 호출 요청을 내놓는 방법 (모델 API 쪽)
    - MCP: 앱이 도구를 찾고·연결하고·호출하는 방법 (연동 계층)

> ### MCP의 구조: Host / Client / Server
- **Host**
    - 사용자가 쓰는 AI 앱 (예: Claude 데스크톱, Cursor)
    - LLM을 품고 클라이언트를 관리, 어떤 서버·도구 호출을 허용할지 결정
- **Client**
    - 호스트 안의 통신 담당, **서버 1개당 1개**
    - MCP 규격으로 메시지 전달만 수행
- **Server**
    - 도구·데이터를 MCP 규격으로 제공 (예: 노션, 깃허브 서버)
    - 기능을 선언하고 요청이 오면 실제 작업 수행
- 통신: **JSON-RPC 2.0**, 로컬은 `stdio`, 원격은 `Streamable HTTP`

> ### MCP 서버의 기본 요소
- **Tools**: 실행할 수 있는 행동 (노션 페이지 만들기, 깃허브 이슈 등록)
- **Resources**: 읽기 전용 데이터 (파일 내용, 문서, DB 스키마)
- **Prompts**: 재사용 가능한 템플릿 (코드 리뷰 요청 양식, 회의록 요약 양식)

> ### 작동 흐름 (예: 노션 "이번 주 회의록 요약해 줘")
- 준비: 호스트 연결(OAuth) → `tools/list`로 도구 목록 확보
- 질문 → 호스트가 모델에 질문 + 도구 목록 전달
- 모델은 `search_pages` 호출 **요청만** 출력
- 호스트가 허용 여부 확인 → `tools/call` → 서버가 실제 검색
- 결과가 모델에 전달 → 요약 답변
- 모델의 역할 = **도구 고르기 + 답 쓰기**

## MCP의 한계

> ### 보안 책임은 누구에게 있는가
- **프로토콜이 아니라 호스트가 책임**
    - 사용자 동의, 서버에 넘길 권한 범위, 허용할 도구 선택 = 호스트 몫
    - MCP는 "연결하는 법"만 정하고 "무엇을 허락할지"는 정하지 않음
- "MCP를 썼으니 안전하다"는 성립하지 않음
    - 표준 단자라고 아무 USB나 꽂아도 안전한 것은 아님
- 연결이 쉬워진 만큼 공격받을 틈도 넓어짐

> ### 도구 오염 / 프롬프트 주입

>> 1) 신뢰할 수 없는 서버
- 출처 모를 서버 연결 = 출처 모를 프로그램 설치
- MCP 서버는 누구나 만들어 올릴 수 있음
>> 2) 도구 오염(Tool Poisoning)
- 도구 설명(`description`) 안에 **사용자에겐 안 보이고 모델에게만 보이는 지시**를 숨기는 공격
    - 모델은 description 전체를 읽고 행동
    - 2025.4 Invariant Labs 시연: 덧셈 도구 설명에 "설정 파일을 읽어 함께 보내라" 숨김
>> 3) 프롬프트 주입(Prompt Injection)
- 도구가 가져온 내용 안에 **지시가 섞여 들어오는 것**

>> 4) 안전하게 쓰는 방법
- 공식·검증된 서버만 연결
- 필요한 권한만 부여 (읽기만 필요하면 읽기 권한만)
- 되돌리기 어려운 호출은 **사람 승인**
- 처음 연결하는 서버는 도구 설명을 직접 확인

## 데이터 구조화

> ### 왜 구조화가 필요한가

>> 1) Vanilla RAG의 한계: Top-K 딜레마
- **Top-K**: 질문과 유사도가 높은 상위 k개 청크만 LLM에 전달
    - k가 작으면 필요한 정보를 놓침
    - k가 크면 관련 없는 정보가 섞임
- 원인
    - 청크를 **하나씩 따로 채점**
    - 정답에 필요한 조각이 질문과 안 비슷한 경우가 많음
    - 예: "영업팀A가 관리하는 고객들의 불만사항은?"
        - 청크1(영업팀A–제품B) → 청크2(고객C–제품B 구매) → 청크3(고객C 배송 불만)
        - 청크3에는 "영업팀A"가 없어 유사도가 낮음
- 근본 원인: **청크들 사이의 관계를 표현하지 못함**
>> 2) 한계를 극복하려는 시도
- **Hybrid Search**: 벡터 + 키워드 검색 (RRF로 결합)
    - 한계: 질문에 없는 단어로 연결된 정보는 여전히 어려움
- **Re-ranking**: 후보를 넓게 가져와 정밀 모델로 재평가
    - 한계: 1차 후보에 없으면 살릴 수 없음
- **Query Rewriting**: Rewrite, HyDE, Multi-Query, 질문 분해
    - 한계: 비용·지연 증가, 의도가 흐려질 수 있음
- **반복 검색(Multi-hop)**: 검색 → 추론 → 다음 검색
    - 한계: 비용·시간 증가, 관계를 매번 새로 찾음
- **GraphRAG**: 엔티티·관계를 미리 그래프로 저장 후 연결을 따라 검색
    - 한계: 구축 비용 큼, 데이터 변경 시 갱신 필요
>> 3) 실제 사내 환경
- 예: "Q3 매출이 왜 떨어졌나?"
    - 매출은 ERP, 이탈은 CRM, 원인 분석은 회의록, 담당자는 HR에 분산
    - 벡터 DB는 조각만 가져오고 **서로의 관계를 모름** → 맥락 없는 답변
- **GIGO**: 구조화되지 않은 데이터를 넣으면 구조화되지 않은 답변이 나옴
- 해결: **데이터 구조화**, 그 방법론이 **그래프 구조**

> ### 그래프 구조와 팔란티어 온톨로지

>> 1) 그래프 자료구조
- **노드**(사물) + **엣지**(관계)
    - 예: `봉준호 —감독→ 기생충 —수상→ 아카데미 작품상`
- 테이블은 "무엇이 있는가", 그래프는 "**서로 어떻게 연결되어 있는가**"를 저장
>> 2) 팔란티어 온톨로지
- 그래프 구조를 기업 전체에 적용해 흩어진 데이터(ERP, CRM, 센서, 문서)를 **하나의 관계망**으로 묶음
    - "온톨로지는 조직의 디지털 트윈"
- 온톨로지 자체는 일반적 방법론, **액션 타입·역할**은 팔란티어 온톨로지의 특징
>> 3) 구성요소
- **객체 타입 / 객체 / 객체 세트**
    - 객체 타입: 테이블 스키마 (예: 고객이란 이름·등급·생성일자를 가진 것)
    - 객체: 한 행 (고객A, 제품B)
    - 객체 세트: 객체의 모음 (VIP 고객 전체)
- **속성 / 속성값 / 공유 속성**
    - 속성: 열 (이름, 날짜, 금액), 속성값: `김철수`, `VIP`
    - 공유 속성: 여러 객체 타입이 공통으로 쓰는 속성
- **링크 타입 / 링크**
    - 링크 타입: 관계의 스키마 ("구매했다", "담당한다")
    - 링크: 실제 관계 ("고객A가 제품B를 구매")
- **액션 타입** (일반 DB와의 가장 큰 차이)
    - 읽기를 넘어 **실제 변경·행동까지** 온톨로지 위에서 정의
    - ReAct의 **Act**, MCP의 **Tools**와 같은 역할
- **역할(Roles)**
    - 직급·부서·지역에 따라 객체 단위로 접근 제어
    - **AI 에이전트도 사람처럼 권한을 부여**받아 범위 안에서만 조회·실행
    - "보안은 호스트가 책임"의 기업판 답
>> 4) 세 구조의 대응
- 관계형 DB: 스키마 / 행 / 열 / 값 / 조인
- 그래프 DB: 노드 / 엣지
- 온톨로지: 객체 타입 / 객체 / 속성 / 속성값 / 링크 타입 + **액션 타입**

## XAI

> ### 블랙박스가 왜 문제인가

>> 1) 문제 사례
- AI는 암 발견, 대출 심사, 채용 판단까지 하지만 "**왜 그 결정을 내렸나**"에는 답하지 못함
- 예: 정확도 99% 피부암 AI가 **반창고가 있으면 암으로 분류**
    - 반창고 사진은 대부분 의사가 이미 살핀 병변 → 반창고 자체를 신호로 학습
    - 입력·출력만 보면 발견 불가
>> 2) 규제와 실용적 이유
- 딥러닝은 수백만~수십억 파라미터가 얽혀 사람이 읽을 수 없음
- 의료: FDA가 결론뿐 아니라 **근거**를 요구
- 금융: EU AI Act가 대출·보험·채용 AI의 불리한 결정에 **이유 설명 의무** 부과
- 자율주행: 사고 시 판단 재현·설명이 가능해야 책임 소재 규명
- 디버깅: 잘못 학습한 부분을 찾으려면 내부를 봐야 함

> ### 기존 해결 시도들

>> 1) 회귀 분석 (완전한 화이트박스)
- 계수를 직접 읽을 수 있어 설명이 완벽
- 한계: 이미지·텍스트 등 복잡한 패턴에서 **성능이 낮음**
>> 2) SHAP / LIME (사후적 설명)
- 딥러닝은 그대로 두고 외부에서 **입력 특성의 기여도를 근사 계산**
- **SHAP**
    - 게임 이론의 Shapley Value 적용, feature별 기여도를 수치화
    - 장점: 이론적 근거 탄탄, 기여도 합 = 예측값 − 기준값
    - 단점: 계산 비용 높음
- **LIME**
    - 샘플 하나 주변을 무작위 샘플링해 **지역적으로 간단한 선형 모델로 근사**
    - 장점: 모델 종류 무관, 가벼움
    - 단점: 비결정적, 모델 전체의 일관성 보장 못 함
- 공통 한계: 실제 연산이 아니라 **역으로 추정**한 것 → "진짜 이유"가 아닐 수 있음
>> 3) LLM + RAG
- "이유를 GPT에게 묻는" 접근
- 한계: **블랙박스가 블랙박스를 설명**, 실제 내부 연산과 일치한다는 보장 없음
>> 4) 비교 요약
- 회귀 분석: 설명 완전 / 성능 낮음 / 신뢰도 높음
- SHAP·LIME: 근사 설명 / 성능 유지 / 신뢰도 부분적
- LLM + RAG: 생성된 설명 / 성능 유지 / 신뢰도 낮음

> ### Mechanistic Interpretability

>> 1) 개념
- 질문: "**블랙박스 자체를 열면 어떨까?**"
- 모델이 *무엇을* 출력하는가가 아니라 ***어떻게* 도달하는가**를 이해
    - 가중치·뉴런·어텐션 헤드를 직접 분석, **회로 수준까지 역공학**
>> 2) 일반 XAI와의 차이
- **일반 XAI (SHAP, LIME)**
    - 외부에서 관찰, 입력-출력 관계, 사후적(post-hoc)
    - "창문으로 들여다보기"
- **Mechanistic Interpretability**
    - 내부를 직접 해부, 가중치·회로 수준, 구조적
    - "집 구조도를 직접 그리기"
>> 3) 핵심 기법: Activation Patching
- 모델 내부 **어느 위치가 답을 결정하는지** 찾는 실험
    - 입력 A("프랑스의 수도는" → 파리), B("독일의 수도는" → 베를린) 준비
    - B 처리 중 특정 위치의 값을 A의 활성값으로 교체
    - 출력이 "파리"로 바뀌면 그 위치가 **핵심 경로**
    - 바뀌지 않으면 그 위치 **하나만으로는** 답을 결정하지 않음
- 비유: 정상 기기의 부품을 하나씩 바꿔 끼우며 원인 찾기

> ### 핵심 개념 3가지

>> 1) Features (특징)
- 모델이 입력에서 알아챈 **개념** (특정 단어~말투·상황 같은 추상 개념)
- 저장 위치
    - **개별 뉴런**: 개념 하나가 뉴런 하나에 (빨강 = R 다이얼)
    - **방향 벡터**: 개념 하나가 여러 뉴런의 조합에 (주황 = R 많이 + G 조금)
- 개념이 등장하면 해당 뉴런(조합)의 활성값이 커짐
- 현실은 뉴런 하나에 여러 개념이 섞이는 경우가 많음

>> 2) Circuits (회로)
- 여러 뉴런/어텐션 헤드가 연결되어 특정 기능을 수행하는 **서브네트워크**
- 예: 간접 목적어 인식 회로
    - 헤드 A: 주어 탐지 / 헤드 B: 동사 탐지 / 헤드 C: 조합해 간접 목적어 위치 예측
- 회로를 알면 기능 제거·이식 같은 조작이 이론적으로 가능

>> 3) Superposition (중첩)
- 표현할 개념은 수백만~수십억인데 뉴런 수는 훨씬 적음
    - → **하나의 뉴런이 여러 개념을 중첩해서 저장**
- 원리
    - 개념을 뉴런 하나가 아닌 **활성 공간의 방향**으로 표현
    - 개념이 차원보다 많으면 방향을 **조금씩 겹치게** 배치 → **간섭(interference)** 발생
    - 개념이 **드물게(sparse)** 나타나므로 간섭이 문제되지 않아 이득이 큼
    - 방향이 뉴런 축과 맞지 않아 뉴런 하나에 여러 특징이 걸쳐 보임
- 가장 큰 해석 난관 → 해결 도구: **Sparse Autoencoder(SAE)**
    - 중첩된 개념을 분리해 독립적으로 추출
