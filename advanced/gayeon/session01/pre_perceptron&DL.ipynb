# 퍼셉트론 & 딥러닝 요약본

## 1. 퍼셉트론

### 1-1. 인공신경망이란?

**1) 인공신경망**
- 우리 뇌의 신경망(뇌를 구성하는 신경 세포끼리 서로 연결되어 정보를 처리하는 구조)을 모방한 모델



   1. 입력($x_1, x_2, \dots, x_n$)들을 받아 각각 가중치($w_1, w_2, \dots, w_n$)와 편향($b$)을 적용
   2. 종합한 값을 활성화 함수에 넘김. 이때 활성화 함수는 임계값을 기준 삼음
       - 기준 미달이면? → 0에 가깝게 죽여서 내보냄
       - 기준 통과면? → 유의미한 값으로 살려서 내보냄
   3. output이 출력됨

**2) 퍼셉트론**
- 퍼셉트론(Perceptron)은 초기 형태의 인공신경망으로, 신경망(딥러닝)의 기원이 되는 선형 이진 분류기(Linear Binary Classifier)



- 퍼셉트론의 출력값을 구하는 절차: 입력 → 가중합 → 활성화 함수 → 출력

   1. 입력값과 가중치를 곱함 → $x_1w_1$
   2. 곱한 값들의 총합을 구함 → $x_1w_1 + x_2w_2 + \cdots + x_nw_n$ (편향을 더해줄지는 옵션)
   3. 총합이 임계값($\theta$)을 넘으면 1, 넘지 않으면 0을 출력 (이 경우 활성화 함수는 Step 함수)

**3) 용어 정리**

**(1) 가중치(weight)**
- 입력값이 출력값에 주는 영향력(중요도)을 결정
- 각 입력값에는 각각의 가중치가 존재
- 가중치의 값이 클수록 해당 입력 값이 중요함을 의미

**(2) 활성화 함수(Activation function)**



- 신호가 일정 이상의 값(임계값 $\theta$)을 넘으면 1, 넘지 못하면 0을 출력하는 퍼셉트론의 성질은 활성화 함수로 계단 함수를 사용
- 이후 등장한 발전된 신경망들은 계단 함수 외에도 다양한 활성화 함수를 사용

**(3) 임계값(Threshold)**
- 활성화 함수에 사용되는 값으로, 보통 세타($\theta$)로 표현

$$
\text{if} \sum_{i}^{n} w_i x_i \ge \theta \rightarrow y = 1, \quad \text{if} \sum_{i}^{n} w_i x_i < \theta \rightarrow y = 0
$$

- 계단 함수에서는 가중합이 임계값을 넘으면 1, 넘지 못하면 0을 출력

**(4) 편향(bias)**
- 위의 계단 함수를 구현한 식에서, 임계값을 좌변으로 넘기면 편향($b$)으로 표현할 수 있음

$$
\text{if} \sum_{i}^{n} w_i x_i + b \ge 0 \rightarrow y = 1, \quad \text{if} \sum_{i}^{n} w_i x_i + b < 0 \rightarrow y = 0
$$

- 그림으로 표현하면 입력값 1에 편향 b가 곱해진 형태로 표현됨



> 임계값과 편향은 표현의 차이. 오늘날 딥러닝에서는 임계값 대신 편향(bias) 개념을 주로 사용


### 1-2. 단층 퍼셉트론과 그 한계

**1) 단층 퍼셉트론**
- 단순히 입력층과 출력층으로만 이루어진 퍼셉트론을 단층 퍼셉트론이라고 함
- 값을 보내는 단계와, 값을 받아서 출력하는 단계로만 이루어짐



**2) 논리 게이트 구현**
- 단층 퍼셉트론을 이용하면 AND, NAND, OR 논리 게이트를 쉽게 구현할 수 있음
- 논리 게이트란?
   - 디지털 회로에서 기본적인 논리 연산을 수행하는 데 사용되는 구성 요소
   - 게이트 연산에 쓰이는 것은 n개의 입력값과 하나의 출력값
   - AND, OR, NOT, NAND, NOR, XOR 등 다양한 종류가 있으며, 이러한 논리 게이트를 조합하여 복잡한 논리 회로를 구성할 수 있음

**3) 대표적인 논리 게이트**
- AND 게이트: 입력값이 모두 1인 경우에만 출력값이 1이 나오는 구조
- 가중치와 편향 값을 $w_1, w_2, b$라고 할 때, AND 게이트를 만족하는 조합은 [0.5, 0.5, -0.7], [0.5, 0.5, -0.8], [1.0, 1.0, -1.0] 등 다양함
- 나머지는 원리가 똑같고 숫자만 바뀜

| 게이트 | 조건 | 가중치·편향 예시 [w1, w2, b] |
|---|---|---|
| AND | 둘 다 1일 때만 1 | [0.5, 0.5, -0.7] |
| OR | 둘 다 0일 때만 0 | [0.6, 0.6, -0.5] |
| NAND | 둘 다 1일 때만 0 | [-0.5, -0.5, 0.7] |
| XOR | 서로 다를 때만 1 | 가능한 조합 없음 |

*NAND는 AND의 부호를 뒤집은 것. XOR만 조합이 없음*

**4) 단층 퍼셉트론의 한계**

- 퍼셉트론은 입력값과 가중치의 선형 결합으로 이루어진 구조이기 때문에, 단층 퍼셉트론은 직선 하나로 두 영역을 나눌 수 있는 선형 분류 문제에 대해서만 구현할 수 있음
   - AND 게이트 → 하나의 직선으로 분리 가능
   - XOR 게이트 → 하나의 직선으로 분리 불가능
- XOR 게이트는 직선이 아닌 곡선/비선형 영역으로 분리해야만 구현할 수 있음



> 따라서 퍼셉트론 하나로는 XOR을 구현할 수 없음 → 퍼셉트론을 여러 계층으로 쓰면 가능


### 1-3. 다층 퍼셉트론

- XOR 게이트는 기존의 AND, NAND, OR 게이트를 조합하면 만들 수 있음. 즉, 퍼셉트론의 층을 더 쌓으면 가능
- 단순히 층을 많이 쌓는 것만으로 비선형성이 생기지는 않음. 활성화 함수 없이 가중합과 편향만 계산한다면, 여러 층을 연결해도 결국 하나의 연산으로 합쳐짐

$$
y = W_2(W_1x + b_1) + b_2 = (W_2W_1)x + (W_2b_1 + b_2) = Wx + b
$$

- 따라서 마지막에 임계값으로 분류해도 결정 경계는 여전히 직선이므로, XOR을 구현할 수 없음
- 다층 퍼셉트론이 XOR을 구현할 수 있는 이유는 은닉층에서 비선형 활성화 함수를 적용해 새로운 특징을 만들고, 다음 층에서 그 특징들을 조합하기 때문
   - 예를 들어 은닉층의 두 뉴런이 각각 NAND와 OR을 계산하고, 출력층이 두 결과에 AND를 적용하면 XOR이 됨

$$
h = \sigma(W_1x + b_1), \quad y = W_2h + b_2 \quad (\sigma: \text{비선형 활성화 함수})
$$



- 중간에 층을 추가하여 (입력층 + 하나 이상의 은닉층 + 출력층)으로 구성하면 됨
- XOR보다 더 복잡한 문제를 해결할 땐 중간에 수많은 은닉층을 추가할 수 있음



- 두 개 이상의 은닉층을 갖는 MLP가 딥러닝의 시초 모델로 발전
   - 심층 신경망 (Deep Neural Network, DNN) = 은닉층이 2개 이상인 신경망
   - 딥러닝 (Deep Learning) = 심층 신경망을 학습(가중치를 스스로 찾도록)시킴

## 2. 딥러닝 모델은 어떻게 학습할까?

### 2-1. 순전파와 역전파

**1) 학습 알고리즘의 기본 구조**
1. 순전파를 통해 입력값으로부터 예측 수행
2. 오차 계산을 통해 예측 정확도 측정
3. 역전파를 통해 출력층에서 다시 입력층 방향으로 오차 전달
4. 가중치 업데이트
5. 1~4를 반복

> 이 과정을 계속 반복하다 보면 오차값이 0에 가까워지게 됨



> 이러한 순전파/역전파의 과정은 연쇄법칙을 통해서 일어남

**2) 연쇄 법칙 (Chain Rule)**

- 합성함수의 미분은 각 함수의 미분의 곱으로 나타낼 수 있다는 법칙으로, 딥러닝에서는 입력값이 여러 은닉층을 거친 후 최종 출력값이 나오는데 이때 입력값이 최종 출력에 얼마나 영향을 미쳤는지 계산하려면 각 층이 다음 층에 미친 영향을 순서대로 곱해나가야 함


$$
\frac{\partial L}{\partial w_1} = \frac{\partial L}{\partial a_n} \times \cdots \times \frac{\partial a_2}{\partial a_1} \times \frac{\partial a_1}{\partial w_1}
$$

- 딥러닝 모델의 성능을 높이려면, 모델의 오차를 계산해 순전파와 역전파를 활용해서 오차를 줄이는 것이 필요
- 손실 함수 = 예측값과 실제값의 차이를 수치로 표현
- 연쇄법칙을 통해 구하는 기울기 = '손실 함수를 줄이기 위해 가중치를 어떻게 바꿀까?'를 알려주는 신호

**3) 기울기 소실 문제 (Vanishing Gradient Problem)**
- 만약 은닉층도 엄청 많고, 각 단계마다 미분값이 1보다 작은 값이라면 한없이 0에 가까워짐

$$
0.25 \times 0.25 \times 0.25 \times \cdots \approx 0
$$

- 기울기들을 계속해서 곱했는데, 결국 기울기가 0이 되는 문제가 발생
- 기울기 소실 문제가 있으면 앞쪽 층의 가중치가 거의 업데이트되지 않아서 학습이 제대로 이뤄지지 않는 문제가 생김

> ReLU가 등장한 배경

**4) 흐름 정리**

| 단계 | 내용 |
|---|---|
| 1. 순전파 = 예측하기 | 입력값(x)이 각 층(w)을 통과하면 예측값(ŷ)이 나옴 |
| 2. 손실 계산 = 얼마나 틀렸는지 측정하기 | 예측값(ŷ)과 실제 정답(y)을 비교해서 오차(손실)를 계산 |
| 3. 역전파 = 책임 추적하기 | 손실이 계산되면, 출력층부터 거꾸로 거슬러 올라가며 각 가중치가 손실에 얼마나 기여했는지 계산 (연쇄법칙 사용) |
| 4. 가중치 업데이트 = 수정하기 | 계산값을 바탕으로 가중치를 조금씩 수정. 이 과정을 계속 반복하면 모델의 정확도가 올라감 |

- w1은 손실 L에 직접 연결되어 있지 않고 w1 → w2 → L 순서로 간접적으로 연결되어 있음
- 따라서 w1의 책임은 "w1이 w2에 준 영향 × w2가 L에 준 영향"으로 곱해서 계산 (연쇄 법칙)

### 2-2. 손실 함수

**1) 목적함수/손실함수/비용함수**

| 구분 | 정의 | 방향 |
|---|---|---|
| 목적함수 | 최적화하려는 전체 대상 | 최대화/최소화 |
| 손실함수 | 데이터 한 개에 대한 오차 | 최소화 |
| 비용함수 | 전체 데이터에 대한 평균 오차 | 최소화 |

- 목적함수(Objective Function): 우리가 달성하고 싶은 목표. 최적화하려는 대상을 수학적으로 표현한 넓은 개념의 함수
   - 문제의 성격에 따라 최대화할 수도 있고, 최소화할 수도 있음 (시험점수는 최대화, 실수는 최소화)
- 손실함수(Loss Function): 개별 데이터에 대한 오차를 측정하는 함수
   - 예측값 $\hat{y}$와 실제값 $y$의 차이를 계산해서 한 개의 데이터 포인트에 대해 모델이 얼마나 틀렸는지 평가
- 비용함수(Cost Function): 전체 데이터셋에 대한 손실을 합산하거나 평균 내어 모델의 전체적인 성적을 평가하는 함수
   - 머신러닝의 목적함수는 보통 비용을 최소화하는 것이라, 학습할 때 대부분은 비용함수를 최소화하는 것이 목표

**2) 손실 함수란?**
- 손실 함수는 모델의 예측이 얼마나 틀렸는지를 수치로 표현하는 방법
   - 실제값(정답)과 예측값(출력값)의 차이를 계산하는 데 사용되는 함수
- 일반적으로 회귀에서는 평균제곱오차(MSE)를, 분류에서는 크로스 엔트로피 오차를 사용


**3) 손실 함수의 종류**

**(1) 평균 제곱 오차 (Mean Squared Error, MSE)**

$$
MSE = \frac{1}{N}\sum_{i=1}^{N}(y_i - \hat{y}_i)^2
$$

- 예측 값이 연속적이고, 예측 값과 실제 값의 차이를 직접 측정하는 회귀 문제에 적합
- ex) 모델이 예측한 집 가격과 실제 집 가격의 차이를 최소화하는 집 가격 예측

**(2) 이진 크로스 엔트로피 (Binary Cross-Entropy, BCE)**



$$
L = -(y\log\hat{y} + (1-y)\log(1-\hat{y}))
$$

- 이진 분류 문제(로지스틱 회귀)에서 사용하는 손실 함수
- 수식의 의미
   - 모델이 예측한 '정답이 1일 확률'을 p라고 하면
      - 정답이 1이면 → p가 높아야 잘 예측한 것
      - 정답이 0이면 → 1 − p가 높아야 잘 예측한 것
   - BCE는 실제 정답에 부여한 확률에 −log를 적용한 손실 정답의 확률이 높으면 손실이 작고, 낮으면 손실이 커짐
   - 로그는 여러 데이터의 확률을 곱하는 계산을 합으로 바꿔주고, 앞의 마이너스는 '확률을 높이는 목표'를 '손실을 줄이는 목표'로 바꿔줌

$$
P(y \mid x) = p^y(1-p)^{1-y}
$$

$$
L = -\log\left[p^y(1-p)^{1-y}\right] = -[y\log p + (1-y)\log(1-p)]
$$

- (1) 실제 정답이 1일 때 → $L = -\log\hat{y}$만 남음
   - 예측 확률이 1에 가까울수록 → loss가 0에 가까워짐 = 잘 맞힘
   - 예측 확률이 0에 가까울수록 → loss가 급격히 증가 (확신했지만 틀림)
   - ex) 실제 레이블이 고양이(1)인데 고양이일 예측 확률이 0.1이라면, 모델이 고양이가 아니라고 확신하며 틀린 것이므로 −log(0.1)로 계산되어 손실 값이 큼
- (2) 실제 정답이 0일 때 → 반대로 동작

> MSE는 단순히 loss의 크기만 보는 반면, BCE는 얼마나 확신을 가지고 예측했는지도 loss에 반영

**(3) 크로스 엔트로피 오차 (Cross Entropy Error, CEE)**

$$
-\sum y\log\hat{y} \quad (y: \text{원-핫인코딩 값}(1, 0), \ \hat{y}: \text{예측값(확률)})
$$

- 다중 클래스 분류 문제에서 예측 확률이 실제 클래스와 얼마나 일치하는지 평가하는 데 적합
- 2번과 거의 유사한 형태지만, 다중 분류에서 사용

**4) 손실 함수 정리**

| 손실 함수 | 사용 상황 | 예시 |
|---|---|---|
| MSE | 회귀 문제 | 주택 가격 예측, 온도 예측, 주식 가격 예측 |
| BCE | 이진 분류 문제 | 이메일 스팸 필터링(스팸/정상), 질병 진단(양성/음성) |
| CEE | 다중 클래스 분류 문제 | 손글씨 숫자 인식(MNIST), 이미지 분류(CIFAR-10) |

### 2-3. 활성화 함수 (Activation Function)
**1) 활성화 함수란?**
- 신경망의 각 층은 기본적으로 이런 계산을 함: 입력값 × 가중치 + 편향 → 활성화 함수 → 출력값(wx+b)
- 왜 굳이 활성화 함수를 거칠까?
   - (1) 비선형성이 필요하다: 활성화 함수가 없으면 층을 아무리 쌓아도 결국 선형 계산의 반복. XOR 같은 문제를 못 풂
   - (2) 미분이 되어야 한다: 2-1에서 역전파는 미분값을 연쇄적으로 곱해 나간다고 했음. 그렇다면 활성화 함수는 미분이 가능해야 함
   > 그런데 계단 함수는 미분이 안 됨. 0에서 끊기고, 나머지 구간은 기울기가 0 ⇒ 역전파를 할 수가 없음


- 활성화 함수의 발전 과정

| 함수 | 미분 | 결과 |
|---|---|---|
| 계단 함수 | 미분 불가 | 역전파 자체가 안 됨 |
| 시그모이드 | 미분 가능! 하지만 | 미분값이 최대 0.25 → 0.25를 5번 곱하면 0.00098… ⇒ 기울기 소실 |
| ReLU | 양수 구간 미분값 = 1 | 곱해도 기울기 소실 발생 X |

- 가중치와 편향
   - 가중치(Weight): 입력값이 결과에 얼마나 영향을 미치는지를 결정하는 값
   - 편향(Bias): 활성화 함수의 출력값을 조정하는 값으로, 기준점 역할을 수행

- 활성화 함수는 단층/다층 퍼셉트론에 따라 다르게 사용됨

| 구분 | 단층 퍼셉트론 | 다층 퍼셉트론 |
|---|---|---|
| 활성화 함수 | 계단 함수 | Sigmoid, ReLU 등 |
| 출력값 | 0 또는 1 (이진 출력) | 연속적인 값 (확률 또는 실수) |
| 특징 | 선형 문제만 풀 수 있음 | 복잡한 비선형 문제도 학습 가능 |

**2) 계단함수 (Heaviside Step Function)**



$$
h(x) = \begin{cases} 1 & (x \ge 0) \\ 0 & (x < 0) \end{cases}
$$

- 입력값과 가중치의 곱을 모두 더한 뒤, 그 합이 임계값을 넘으면 1, 아니면 0을 출력하는 함수. 주로 단층 퍼셉트론에서 사용됨
   - 입력 신호: 입력값과 가중치의 곱
   - 출력 신호: 0 또는 1
- 임계값은 뉴런이 얼마나 쉽게 활성화되는지를 제어하고, 가중치는 각 입력 신호가 결과에 미치는 영향력을 제어

> 계단 함수는 미분값이 0이라 역전파를 할 수 없음. 따라서 다층 퍼셉트론에서는 쓰지 않음


**3) 시그모이드 함수**



$$
f(x) = \frac{1}{1 + e^{-x}}
$$

- 입력 신호를 0과 1 사이의 연속적인 값으로 변환하는 비선형 함수
- 계단 함수와의 차이점: 출력값의 형태
   - 계단 함수는 0 또는 1만 출력하는 반면, 시그모이드는 x값에 따라 출력값이 연속적으로 변화함
   - 덕분에 "확실히 맞다/틀리다"가 아니라 얼마나 확실한지를 표현할 수 있고, 미분도 됨

> 하지만 미분값이 최대 0.25. 층을 쌓아 곱해 나가면 기울기 소실이 발생


**4) ReLU 함수**



$$
f(x) = \max(0, x)
$$

- 입력이 0보다 크면 그대로, 0 이하이면 0을 출력하는 비선형 함수. 단순하지만 딥러닝에서 가장 많이 사용됨
- 왜 ReLU를 가장 많이 쓸까?
   1. 계산이 간단하고 효율적: 0보다 크냐 아니냐만 비교하므로 빠름
   2. 비선형성 제공: 음수는 0으로, 양수는 그대로 처리하면서 은닉층에 비선형성을 추가
   3. 기울기 소실 완화 : 양수 구간에서 미분값이 항상 1이므로, 아무리 곱해도 값이 죽지 않음. 역전파가 안정적으로 전달됨


**5) Softmax 함수**



$$
f(x_i) = \frac{e^{x_i}}{\sum_j e^{x_j}}
$$

- 각 클래스에 대한 출력값을 확률로 변환하는 함수. 다중 클래스 분류 문제의 출력층에서 사용
- 출력값이 모두 0~1 사이이고 전체 합이 1이 되기 때문에, 각 클래스에 속할 확률을 직접 읽을 수 있음

> Softmax + 크로스 엔트로피 조합: 예측 확률(softmax)과 정답 확률의 차이(크로스 엔트로피)를 바로 계산할 수 있어 학습 효율이 높음 ⇒ 가장 표준적인 다중 분류 조합

**6) 활성화 함수 정리**

| 활성화함수 | 출력 범위 | 주요 사용처 | 장점 | 단점 |
|---|---|---|---|---|
| 계단 함수 | 0 또는 1 | 단층 퍼셉트론 | 구조가 단순 | 미분 불가 |
| Sigmoid | 0 ~ 1 | 이진 분류 출력층 | 확률로 해석 가능 | 기울기 소실 |
| Tanh | -1 ~ 1 | 은닉층 (요즘은 ReLU에 밀림) | 출력 중심이 0으로 학습 안정적 | 기울기 소실 |
| ReLU | 0 ~ ∞ | 은닉층 전반 | 계산 빠름, 기울기 소실 완화 | 음수 입력 시 출력이 항상 0 |
| Softmax | 0 ~ 1 (합=1) | 다중 분류 출력층 | 클래스별 확률 직접 출력 | 출력층 전용, 은닉층 부적합 |


### 2-4. 경사하강법 (Gradient Descent)

**1) 경사하강법의 원리**
- 목표: 손실을 가장 작게 만드는 W를 찾는 것
- 손실함수를 W에 대한 함수로 보면 (편의상 b는 고정)

$$
E(W) = \frac{1}{n}\sum_{i=1}^{n}[t_i - (Wx_i + b)]^2
$$

- x축을 W, y축을 E로 두면 아래로 볼록한 곡선이 그려짐. 우리가 찾는 건 이 곡선의 바닥



- 어떻게 바닥으로 가나 → 미분값의 부호만 보면 됨

| 지금 W의 위치 | 접선 기울기 | 현재값 − 기울기 | 결과 |
|---|---|---|---|
| 최솟값 오른쪽 | 양수 | 현재값 − 양수 | W가 작아짐 → 왼쪽으로 |
| 최솟값 왼쪽 | 음수 | 현재값 − 음수 | W가 커짐 → 오른쪽으로 |
| 최솟값 도달 | 0 | 현재값 − 0 | 멈춤 |

<br>

> 어느 쪽에서 출발하든 자동으로 바닥에 모임

**2) 학습률 (Learning Rate, α)**
- 얼마나 움직일까?
- 미분값을 그대로 빼면 문제가 생김. 한 번에 너무 많이 움직여서 최솟값을 지나치고, 반대편에서 또 지나치고... 오히려 멀어짐



- 미분값에 α를 곱해 이동량을 조절

| 학습률 | 무슨 일이 벌어지나 |
|---|---|
| 너무 작음 | 수렴은 하지만 너무 오래 걸림 |
| 너무 큼 | 수렴 못 하고 발산 (손실이 계속 커짐) |
| 적당함 | Good |



- 0.001부터 시작. Adam의 기본값이기도 함

**3) 배치 · 에포크 · 이터레이션**
- 위 식의 ∂E/∂W를 구하려면 데이터를 넣어야 함
- 데이터가 100만 개라면 한 번에 다 넣을까? 메모리가 못 버팀. 그래서 나눠서 넣음 ⇒ 이를 배치라고 함

**(1) 배치 (Batch)**



- 학습 과정에서 매개변수를 조정하기 위해 사용하는 데이터의 묶음을 의미하며, 사용자가 임의로 정할 수 있는 하이퍼파라미터
- 배치 크기가 너무 크면 학습이 느려지고 메모리가 부족해질 수 있음
- 배치 크기가 너무 작으면 가중치가 자주 업데이트되어 학습이 불안정해질 수 있음

**(2) 에포크 (Epoch)**



- 모든 데이터셋을 학습하는 횟수를 의미

**(3) 이터레이션 (Iteration) 또는 스텝 (Step)**
- 한 번의 에포크를 끝내기 위해 필요한 배치의 수
- 1 Epoch에서 다루는 데이터의 수 = 배치 크기 × 이터레이션
   - 전체 데이터가 2,000일 때 배치 크기를 200으로 하면 이터레이션은 총 10


**4) 경사하강법의 종류**

**(1) 배치 경사하강법 (Batch Gradient Descent)**
- 오차를 구할 때 전체 데이터를 한 묶음으로 묶음 (Batch size = Data size)



- 전체 데이터로 학습하므로 업데이트 횟수가 적음 (1 Epoch 당 1 업데이트)
- 한 번에 처리하므로 메모리가 많이 필요
- 항상 같은 데이터에 대해 경사를 구하므로 수렴이 안정적

**(2) 확률적 경사하강법 (Stochastic Gradient Descent, SGD)**
- 기존 경사하강법의 단점: 전체 데이터를 계산하므로 시간이 너무 오래 걸림
- 매개변수를 조정할 때 전체가 아니라 랜덤으로 선택된 하나의 데이터에 대해서만 계산



- 하나의 데이터만 메모리에 저장하면 되므로 자원이 적은 컴퓨터에서도 사용 가능
- 매개변수의 변경폭이 불안정하고 정확도가 낮을 수 있음

**(3) 미니 배치 경사하강법 (Mini-Batch Gradient Descent)**
- 배치 크기가 1이면 정확도가 낮고, 전체 배치는 메모리가 부족 ⇒ 미니 배치



- 배치 크기를 적절히 지정하여 해당 개수만큼 매개변수를 조정
- 전체 데이터보다 빠르고, 배치 크기 1보다 안정적 ⇒ 가장 많이 사용

### 2-5. 옵티마이저 (Optimizer)

**0) 경사하강법만으로는 부족 - Local Minima**
- 앞에서 본 손실함수 그래프는 예쁘게 아래로 볼록한 모양이었음. 그런 함수라면 경사하강법으로 반드시 바닥에 도달함
- 문제는 딥러닝의 손실함수는 그렇게 안 생겼다는 것. 층이 여러 개라 울퉁불퉁하고, 어떤 모양인지 예측할 수도 없음
- 진짜 바닥(Global Minimum)이 아닌 작은 웅덩이(Local Minimum)에서도 기울기는 0

> 경사하강법을 더 똑똑하게 굴릴 방법이 필요함 ⇒ 옵티마이저

**1) 모멘텀 (Momentum)**
- 기존 경사하강법은 매번 현재 기울기만 보고 움직임
- 이전에 가던 방향을 기억하도록 관성을 부여
- 관성을 가진 공처럼 이전 방향을 참고하면, 더 부드럽고 빠르게 작은 언덕도 넘어갈 수 있음



- 빨간색 화살표: 경사하강법이 이동하려는 방향 (현재 기울기)
- 초록색 화살표: 이전 단계의 이동을 반영한 모멘텀 방향 (관성)
- 파란색 화살표: 실제 이동 방향 (둘의 합)

> Local Minima를 관성의 힘으로 탈출할 수 있음
> 기울기가 매우 작은 구간도 빠르게 통과함

**2) 알엠에스프롭 (RMSProp)**
- 기존 경사하강법의 또 다른 문제: 기울기가 큰 방향에서는 너무 많이, 작은 방향에서는 너무 조금 이동함
- 각 방향마다 학습률을 따로 조정해서 안정적이고 빠르게 학습하도록 함
- 기존 경사하강법은 무조건 현재 기울기만 보고 내려가나, RMSProp은 최근 기울기를 참고해서 급한 경사에서는 속도를 줄이고, 완만한 경사에서는 속도를 유지하거나 증가

**3) 아담 (Adam)**
- RMSProp과 Momentum을 합친, 방향과 스텝 사이즈를 적절하게 합친 방법
- 방향과 학습률 두 가지를 모두 잡기 위한 방법
- 각 파라미터의 1차 모멘트(기울기의 평균)와 2차 모멘트(기울기의 제곱 평균)를 사용하여 학습률을 조정

**4) 그래서 뭘 쓰나요?**

| 이럴 때 | 이걸 쓰세요 | 왜 |
|---|---|---|
| 잘 모르겠다 / 대부분의 경우 | Adam | 방향과 보폭을 둘 다 잡음. 사실상 기본값 |
| 학습이 자꾸 한쪽으로 튄다 | 모멘텀 | 방향을 안정화 |
| 파라미터마다 기울기 크기 차이가 크다 | RMSProp | 방향별로 학습률을 따로 조정 |
