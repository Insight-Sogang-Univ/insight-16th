# AI 에이전트

### AI 에이전트란
- **사용자가 정한 목표를 이루려고 AI 모델로 환경과 상호작용하는 시스템**
- 추론 + 계획 + 도구 사용 실행이 하나로 맞물려 돌아감
- 알프레드 비유: "커피 한잔 다오"
  - 주방 가기 → 커피 머신 쓰기 → 내리기 → 들고 오기로 계획
  - 갖고 있는 도구로 실제 실행

### 생성형 AI vs 에이전틱 AI
- 생성형: 프롬프트에 맞는 결과 한 번 출력하고 끝
- 에이전틱: 목표 이룰 때까지 추론이랑 행동을 반복

### ReAct
- Reasoning + Acting
- **생각 → 행동 → 관찰 → 다시 생각** 반복
- 배송 지연 CS 에이전트 예시
  - 주문 상태 확인 → 4일 지연 중
  - 원인 확인 → 배송사 서버 응답 x → 재시도 → 물류센터 적체
  - 환불안 작성 → 돈 나가니까 담당자 승인 → 환불, 메시지 보내고 끝
  - 재시도가 계속 실패하면 상담원 연결로 계획 바꿈
- 핵심: 관찰 결과에 따라 다음 생각이 달라짐
- 폭주 막으려고 앱이 정해 두는 것
  - 최대 반복 횟수, 비용 한도
  - 중복 실행 방지: 환불 두 번 나가면 안 됨
  - 중요한 행동 전엔 사람 승인
- 목표만 주면 알아서 성공 x → 정해진 범위 안에서 반복하는 것
- 주의: 모델이 도구를 직접 실행하는 게 아님
  - "이 도구 이 값으로 불러줘" 요청만 내고 실행은 앱이 함

### 다중 에이전트
- 여러 에이전트가 역할 나눠서 같이 일하는 구조 = MAS
- 제일 흔한 형태: **오케스트레이터 + 서브 에이전트**
  - 오케스트레이터 = 지휘자: 일 쪼개기 → 누구한테 맡길지 정하기 → 결과 합치기
  - 서브 에이전트 = 전문가: 맡은 일만 함. 검색봇은 검색만, 검토봇은 검토만
- 왜 나눔?
  - 혼자 계획, 실행, 검증 다 하면 자기 실수를 잘 못 잡음 → 검토는 다른 눈으로
  - 서브 에이전트마다 자기 컨텍스트 창 → 여러 갈래 동시에 조사, 빠름
- Anthropic 리서치 기능: 단일 에이전트보다 90.2% 좋은 결과, 대신 토큰은 일반 채팅의 약 15배
- 무조건 다중이 답은 x
  - 순서 정해진 반복 작업 → 고정 워크플로우면 충분
  - 도구 몇 개로 해결 → 단일 에이전트
  - 나눠서 조사하고 합칠 큰 작업 → 다중 에이전트
- 역할마다 모델 급도 다르게: 오케스트레이터는 고성능 모델, 검색·정리봇은 가벼운 모델


# MCP

### 도구 호출의 원리: Function Calling
- **모델이 "이 함수를 이 값으로 불러줘"라는 구조화된 요청을 내놓는 기능**
- 흐름
  1. 앱 → 모델: 질문 + 쓸 수 있는 함수 목록
  2. 모델: 부를 함수랑 값을 적어서 냄 ex. get_weather, city = 서울
  3. 앱이 실제로 실행
  4. 결과를 다시 모델한테 → 최종 답
- 실행이랑 결과 전달은 전부 앱 담당

### MCP가 필요한 이유: n×m → n+m
- 표준 없을 때: 앱마다 × 도구마다 연동 코드를 따로 짬 → n×m개
  - 도구 하나 늘 때마다 작업량 크게 ↑
- MCP 있으면: 앱은 클라이언트 한 번, 도구는 서버 한 번만 → n+m개
- **MCP = LLM 앱이랑 외부 데이터·도구가 통신하는 개방형 표준 프로토콜**
  - 2024년 Anthropic이 공개 → OpenAI 등도 채택 → 리눅스 재단 쪽에 기증해서 업계 공동 표준
- USB-C 비유: 단자 하나로 노트북, 폰 다 충전
- 해결된 것
  - 서버 한 번 만들면 MCP 지원하는 모든 앱에서 동작
  - 도구 목록 확인 방식 통일 → 앱에 도구를 미리 적어 둘 필요 x
  - 로그인 방식 통일 → OAuth
- Function Calling vs MCP: 경쟁 관계 x, 보통 둘 다 씀
  - MCP로 도구 목록 가져옴 → 모델이 Function Calling으로 하나 고름 → 앱이 MCP로 실행
- 도구 정의에서 description이 중요
  - 모델이 description 읽고 그 도구를 쓸지 말지 정함 → 한 줄만 바꿔도 행동이 달라짐
  - inputSchema = 모델이 엉뚱한 값 못 넣게 막는 울타리

### MCP의 구조: Host / Client / Server
- Host: 사용자가 쓰는 AI 앱. LLM 품고 클라이언트 관리, 어떤 도구 호출 허용할지 결정 ex. Claude, Cursor
- Client: Host 안에서 통신 담당. 서버 1개당 1개, 메시지 전달만 함
- Server: 도구랑 데이터를 MCP 규격으로 내놓는 프로그램 ex. 노션 서버, 깃허브 서버
- "이번 주 회의록 요약해 줘" 흐름
  - 미리: 서버 연결, 로그인 → 도구 목록 받아 둠
  - 질문 → 모델은 search_pages 호출 요청만 냄 → 호스트가 허용 확인 → 서버가 노션에서 실제 검색 → 결과를 모델한테 → 요약
- 모델이 하는 일은 도구 고르기 + 답 쓰기뿐, 연결·호출·실행은 Host, Client, Server 몫

### MCP 서버의 기본 요소
- Tools: 실행할 수 있는 행동 → 노션 페이지 만들기, 깃허브 이슈 등록
- Resources: 읽기 전용 데이터 → 파일 내용, 문서, DB 스키마
  - RAG가 참고할 문서도 Resources로 넘길 수 있음
- Prompts: 재사용하는 템플릿 → 코드 리뷰 요청 양식, 회의록 요약 양식


# MCP의 한계

### 보안 책임은 누구에게?
- **프로토콜 x, 호스트가 책임짐**
- MCP는 "연결하는 법"만 정함, "무엇을 허락할지"는 안 정함
  - 사용자 동의, 권한 범위, 허용할 도구 = 다 호스트 몫
- USB-C가 표준이라고 아무 USB나 꽂아도 안전한 건 x
- 연결이 쉬워진 만큼 공격받을 틈도 넓어짐
  - MCP 서버는 누구나 만들 수 있음 → 출처 모를 서버 연결 = 출처 모를 프로그램 설치

### 도구 오염 / 프롬프트 주입
- 도구 오염 = tool poisoning
  - 도구 description 안에 사용자한텐 안 보이고 모델한테만 보이는 지시를 숨김
  - 덧셈 도구 설명에 "설정 파일 읽어서 같이 보내라, 사용자한텐 알리지 마라"를 숨긴 사례
  - 사용자 화면엔 덧셈 도구로만 보이는데 모델은 숨긴 지시까지 다 읽음
- 프롬프트 주입 = prompt injection
  - 도구가 가져온 내용 안에 지시가 섞여 들어오는 것
  - GitHub MCP 사례: 공개 저장소 이슈에 지시 숨김 → 에이전트가 읽고 따름 → 비공개 저장소 내용을 공개 PR에 올려버림
  - 서버 코드 문제 x, 공개·비공개 저장소를 다 읽고 쓰는 넓은 권한이 문제
  - 권한 넓을수록 피해 ↑
- 안전하게 쓰려면
  - 검증된 서버만 연결
  - 필요한 권한만 주기 → 읽기만 필요하면 읽기만
  - 글 올리기, 지우기처럼 되돌리기 어려운 건 사람 승인
  - 처음 연결하는 서버는 도구 설명 직접 확인


# 데이터 구조화

### 왜 구조화가 필요한가
- 에이전트가 일하려면 결국 구조화된 데이터가 필요
- 사내 데이터 줄 때 제일 흔한 방법이 RAG인데, 문서를 벡터 DB에 넣기만 하면 한계 있음
- Top-K 딜레마
  - 질문이랑 비슷한 상위 k개 조각만 가져옴
  - k 작음 → 필요한 정보 놓침 / k 큼 → 상관없는 정보까지 섞임
- 왜 생기나
  - 조각을 하나씩 따로 채점함 → 조각끼리 연결은 모름
  - "영업팀A가 관리하는 고객들의 불만은?"
    - 영업팀A는 제품B 담당 → 고객C가 제품B 구매 → 고객C가 배송 지연 불만
    - 정작 답이 있는 마지막 조각엔 "영업팀A"가 없어서 유사도가 낮음
- 결국 바닐라 RAG는 **조각 사이의 관계를 표현 못 함**
- 보완 시도: 하이브리드 검색, 리랭킹, 질문 재작성, 반복 검색, GraphRAG 등
- 실제 회사는 더 복잡: 매출은 ERP, 고객 이탈은 CRM, 원인 분석은 회의록 → 서로 무슨 관계인지 모름

### 그래프 구조
- **노드 + 엣지로 데이터 표현**
  - 노드: 존재하는 것 ex. 봉준호, 기생충
  - 엣지: 노드 사이 관계 ex. 봉준호 —감독→ 기생충
- 테이블은 "무엇이 있는가", 그래프는 "어떻게 연결돼 있는가"를 저장
- GraphRAG: 문서에서 엔티티랑 관계를 미리 뽑아 그래프로 저장 → 검색할 때 연결 따라감
  - k 크기가 아니라 연결 구조로 필요한 정보를 가져옴
  - 대신 그래프 만드는 비용이 큼

### 팔란티어 온톨로지
- 온톨로지 = 데이터를 개념과 관계로 구조화하는 방법론
- 팔란티어는 그래프 구조를 기업 전체에 적용 → 흩어진 사내 데이터를 하나의 관계망으로 묶음
  - "The Ontology is the digital twin of an organization."
- 구성요소
  - 객체 타입 = 테이블 스키마 / 객체 = 행 / 객체 세트 = 객체 모음 ex. VIP 고객 전체
  - 속성 = 열 / 속성값 = 실제 값
  - 링크 타입 = 관계의 틀 ex. "구매했다" / 링크 = 실제 관계 ex. 고객A가 제품B를 구매
  - **액션 타입**: 읽기를 넘어서 실제 변경이랑 행동까지 정의 ← 일반 DB랑 제일 큰 차이
    - ex. 불량 나면 담당자한테 알림 보내기
    - ReAct의 행동, MCP의 Tools랑 같은 역할
  - 역할: 직급, 부서별로 객체 단위까지 권한 제어
    - 에이전트도 사람처럼 권한 받음 → "보안은 호스트 책임"에 대한 기업판 답
- 의료 기기 제조사 사례: 수백 개 시스템 데이터를 하나로 묶음
  - "이 납품 왜 지연됐나?" 한 번에 공장 → 작업지시서 → 물류 → 고객까지 따라가며 추론


# XAI

- 이제 AI가 권한 받아서 직접 실행까지 함 → "왜 그렇게 했는가"에 답할 수 있어야 함
- 피부암 진단 AI가 반창고 붙은 사진을 암으로 분류
  - 반창고 사진 = 대부분 의사가 이미 본 병변 → 반창고 자체를 암 신호로 배워버림
  - 입력이랑 출력만 보면 절대 못 찾는 문제

### 블랙박스가 왜 문제인가
- 딥러닝은 수십억 개 파라미터가 얽혀 있어서 사람이 못 읽음
- 규제 때문에 "알면 좋은" 수준이 아님
  - 의료: FDA는 진단 AI에 결론뿐 아니라 근거도 요구
  - 금융: EU AI Act → 대출, 보험, 채용 AI가 불리한 결정하면 이유를 설명할 의무
  - 자율주행: 사고 나면 왜 그렇게 판단했는지 재현할 수 있어야 책임을 따짐
- 실용적 이유: 잘못 배웠을 때 내부를 봐야 디버깅 ㄱㄴ

### 기존 해결 시도들
- 회귀 분석: 계수를 직접 읽음 → 설명은 완벽한데 복잡한 패턴엔 성능이 너무 낮음
- SHAP / LIME: 모델은 그대로 두고 바깥에서 입력 특성의 기여도를 근사
  - SHAP: 게임 이론의 Shapley Value로 feature별 기여도 계산
  - LIME: 샘플 하나 주변을 간단한 선형 모델로 근사
  - 한계: 입력 바꿔보면서 역으로 추정 → "진짜 이유"가 아니라 "그렇게 보이는 것"일 수도
- LLM + RAG: GPT한테 이유 물어보기 → 블랙박스로 블랙박스 설명하는 셈, 실제 연산이랑 같다는 보장 x

### Mechanistic Interpretability
- "블랙박스 자체를 열어버리면?"
- **모델이 뭘 출력하는지가 아니라 어떻게 그 답에 도달하는지를 이해하기**
- 가중치, 뉴런, 어텐션 헤드를 직접 분석해서 회로 수준까지 역공학
- 일반 XAI = 창문으로 집 들여다보기 / Mech Int = 집 구조도를 직접 그리기
- Activation Patching
  - "프랑스의 수도는" → 파리 / "독일의 수도는" → 베를린
  - 독일 문장 처리하는 중에 특정 위치 값을 프랑스 문장 값으로 바꿔 끼움
  - 답이 파리로 바뀌면 → 그 위치가 답을 정하는 핵심 경로
  - 정상 기기 부품 하나씩 바꿔 끼워보면서 고장 원인 찾는 느낌

### 핵심 개념 3가지
- Features: 모델 안에 뭐가 있나
  - 모델이 입력에서 알아챈 개념. 단어일 수도, 말투나 상황 같은 추상적인 개념일 수도
  - 뉴런 하나에 담기거나 여러 뉴런 조합, 즉 방향에 담김
  - 그 개념이 나오면 해당 뉴런이 켜짐
- Circuits: 그게 어떻게 연결돼서 일하나
  - 여러 뉴런, 어텐션 헤드가 연결돼서 특정 기능을 하는 서브네트워크
  - Induction Heads: 예시 보고 패턴 따라 하는 능력이 어텐션 헤드 단 2개로 구현돼 있었음
- Superposition: 왜 해석이 어렵나
  - 표현할 개념은 엄청 많은데 뉴런은 적음 → 뉴런 하나가 여러 개념을 겹쳐서 저장
  - 개념 방향을 조금씩 겹쳐 배치 → 간섭 생기지만 개념들이 동시에 잘 안 켜져서 감수함
  - 그래서 뉴런 하나만 봐서는 무슨 개념인지 바로 모름
  - 해결 도구: Sparse Autoencoder → 겹친 개념을 분리
    - Claude 3 Sonnet에서 "골든 게이트 브릿지" feature 찾아서 강제로 켰더니 자기를 골든 게이트 브릿지라고 소개함
- 정리: XAI가 "설명을 어떻게 만들까"에서 "AI 내부를 어떻게 이해할까"로 넘어가는 중
