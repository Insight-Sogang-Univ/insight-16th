# 퍼셉트론

### 인공신경망  
- 우리 뇌의 신경망을 모방한 모델
### 퍼셉트론
- 딥러닝의 기원이 되는 선형 이진 분류기  


퍼셉트론의 출력값을 구하는 절차  
입력 -> 가중합 -> 활성화 함수 -> 출력  
* 활성화 함수: 임계값 초과 여부에 따라 출력값을 결정.

training: 원하는 출력값을 내보내도록 **가중치를 조정**해나가는 작업

### 용어 정리

1. 가중치: 입력값이 출력값에 주는 영향력을 결정
2. 활성화 함수: 계단 함수 외에도 다양한 활성화 함수 존재
3. 임계값: 보통 세타(θ)로 표현 / 계단 함수에서는 가중합이 임계값을 넘으면 1, 넘지 못하면 0을 출력
4. 편향: 위의 계단 함수를 구현한 식에서, 임계값을 좌변으로 넘기면 편향(b)으로 표현

### 단층 퍼셉트론의 한계

- 입력층, 출력층만 있는 퍼셉트론
- 단층 퍼셉트론을 이용하면 AND, NAND, OR 논리 게이트를 쉽게 구현할 수 있음

- **단층 퍼셉트론은 직선 하나로 두 영역을 나눌 수 있는 선형 분류 문제에 대해서만 구현할 수 있음**




- XOR은 다층 퍼센트론으로 구현가능

### 다층 퍼셉트론 

- 다층 퍼셉트론이 XOR을 구현할 수 있는 이유: 은닉층에서 비선형 활성화 함수를 적용해 새로운 특징을 만들고, 다음 층에서 그 특징들을 조합하기 때문






# 딥러닝 모델의 학습법

### 순전파, 역전파

- 학습 알고리즘은 다음의 반복으로 이루어짐
- 순전파를 통해 입력값으로부터 예측 수행 -> 오차 계산을 통해 예측 정확도 측정 ->  역전파를 통해 출력층에서 다시 입력층 방향으로 오차 전달 -> 가중치 업데이트

반복하다보면 오차값이 0에 가까워짐

이는 연쇄법칙을 통해 일어남

### 연쇄 법칙 (Chain Rule)
- 합성함수의 미분은 각 함수의 미분의 곱으로 나타낼 수 있다는 법칙  


### 기울기 소실 문제
- 기울기들을 계속해서 곱했는데, 결국 기울기가 0이 되는 문제가 발생

1. 순전파 = 예측하기
2. 손실 계산 = 얼마나 틀렸는지 측정하기
3. 역전파 = 책임 추적하기  ← 여기서 연쇄법칙이 사용
4. 가중치 업데이트 = 수정하기

### 계단함수
- 입력값과 가중치의 곱을 모두 더한 뒤, 그 합이 임계값을 넘으면 1, 아니면 0을 출력하는 함수
