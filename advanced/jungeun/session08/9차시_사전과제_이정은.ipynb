### AI 트렌드

### AI 에이전트
- 사용자가 정한 목표를 달성하기 위해 AI 모델을 활용해 환경과 상호작용하는 시스템
- 추론 + 계획 + 도구를 사용한 실행이 하나로 맞물려 돌아감

### 에이전틱 AI의 작동 방식
- 생성형 AI
    - 프롬프트에 맞는 결과물을 한 번 출력하고 끝
- 에이전틱 AI
    - 목표를 달성할 때까지 추론과 행동을 반복
    - 생각 -> 행동 -> 관찰 -> 생각 순환 반복
    - 관찰 결과에 따라 다음 생각이 달라짐
- 폭주 방지
    - 최대 반복 횟수
    - 비용 한도
    - 중복 실행 방지
    - 중요한 행동 전 사람 승인
### 다중 에이전트
- MAS 구조
    - 여러 에이전트가 역할을 나눠 함께 일하는 구조
    - 오케스트레이터 : 지휘자
    - 서브 에이전트 : 전문가
- 역할을 나누는 이유?
    - 스스로 실수를 잡기 어려움
    - 동시에 여러 갈래를 조사
    - 실제로 이렇게 리서치 진행하면 90.2% 좋은 결과 발생
    - 모든 역할에 가장 비싼 모델 사용 X
    - 역할마다 필요한 능력에 맞춰 급이 다른 모델 섞어 사용

### MCP
### MCP 이전의 도구 호출
- function calling, tool caling
    - 모델이 구조화된 요청을 내놓는 기능
    - 실제 실행과 결과전달은 앱이 담당
- 과정
    - 앱 -> 모델 : 질문 + 사용 가능 함수 목록
    - 모델 : 어떤 함수를 어떤 값으로 부를지 적음
    - 앱 : 함수 실행
    - 앱 -> 모델 : 실행 결과 전달
    - 모델 : 최종 답 작성
- 표준 없으면?
    - 도구 하나 늘 때 작업량도 크게 증가
- 표준 생기면?
    - 모두 같은 규칙을 쓰니 어떤 앱이든 어떤 도구와 바로 연결되어, 필요한 작업이 줄어듦
- 주로 function calling, MCP 둘 다 사용
### MCP의 등장
- MCP
    - LLM 앱과 외부 데이터, 도구가 통신하는 개방형 표준 프로토콜
    - AI 앱과 도구 사이의 USB-C
- 해결점
    - 서버를 한 번 만들면 MCP 지원하는 모든 앱에서 동작
    - 발견의 표준화
    - 원격 서버 인증 통일
### MCP의 구조
- Host 
    - 사용자가 쓰는 AI 앱
- Client
    - 호스트 안의 통신 담당
- Server
    - 도구, 데이터를 MCP 규격으로 내놓는 프로그램
### MCP 서버의 기본 요소
- Tools
- Resources
- Prompts
### MCP의 작동 흐름
- AI가 외부 도구를 쓰는 과정
- 질문 전
    - 연결 : 호스트 -> 노션 서버
    - 도구 목록 요청 : 클라이언트 -> 서버
    - 도구 목록 응답 : 서버 -> 클라이언트
- 질문 후
    - 질문 : 사용자 -> 호스트
    - 전달 : 호스트 -> 모델
    - 도구 선택 : 모델 -> 호스트
    - 호출 : 호스트 -> 클라이언트 -> 서버
    - 실행 : 서버
    - 결과 전달 : 서버 -> 클라이언트 -> 호스트 -> 모델
    - 답변 : 모델 -> 사용자
- 모델
    - 도구 고르기
    - 답 쓰기
- 연결, 호출, 실행
    - 호스트, 클라이언트, 서버
### MCP의 한계
- 보안 : 호스트 책임
    - 사용자 동의
    - 서버에 넘길 권한 범위
    - 서용할 도구 선택
- 위험요소
    - 신뢰할 수 없는 서버를 붙이는 위험
    - 도구 설명을 통한 주입
    - 너무 넓은 권한 연동의 위험
- 해결법
    - 공식, 검증된 서버만 연결
    - 필요한 권한만 주기
    - 되돌리기 어려운 호출은 사람이 승인
    - 처음 연결하는 서버는 도구 설명 직접 확인

- 에이전틱 AI, 피지컬 AI가 일하려면 구조화된 데이터 필요

    - 정답에 필요한 조각이 질문과 안 비슷할 때가 많음
    - 근본적으로 청크들 사이의 관계를 바닐라 RAG가 표현하지 못함

### 블랙박스
- 딥러닝 모델은 입력 -> 출력의 과정이 블랙박스
- 과정을 알아야 하는 이유는 규제 떄문
    - 의료 : 진단에 대한 근거 요구
    - 금융 : 고객에게 이유 설명 요구
    - 자율 주행 : 사고 발생 이유 설명 요구 
    - 실용 : 어디가 문제인지 디버깅 위해
### 기존 해결 시도
- 회귀 분석 
    - 완전한 화이트 박스
    - 가장 고전적인 방법
    - 선형 모델이기에 충분한 정확도 불가
- SHAP/LIME
    - 사후적 설명
    - 어떤 입력이 이 결과에 얼마나 기여했는지 근사해서 보여줌
    - 역추정 방법이기에 진짜 이유가 아닐 수도 있음
- LLM + RAG
    - 블랙박스로 블랙박스 설명
    - 설명을 요구한 GPT의 과정도 블랙박스이기에 그럴듯하지만 일치한다는 보장 없음
### 기계적 해결 가능성
- 블랙박스 자체를 열자
- 모델이 무엇을 출력하는가가 아닌, 모델이 어떻게 그 답에 도달하는가 이해하고자 함
- 회로 수준까지 역공학 하는 연구 분야
- 핵심 기법 : 활성값 패칭
    - 모델 내부의 어느 위치가 답을 결정하는지 찾음
    - 1. 입력 두 개 준비
    - 2. B 처리 도중, 특정 위치의 값을 A로 바꿈
    - 3. 출력이 어떻게 바뀌는지 봄
### 핵심개념
- 1. 모델 안에는 무엇이 있는가?
    - 특징
    - 모델이 입력에서 알아챈 개념
    - 저장 위치 : 개별 뉴런, 방향 벡터
- 2. 그것들은 어떻게 연결되어 일하는가?
    - 회로
    - 여러 뉴런, 어텐션 헤드가 연결되어 특정 기능 수행하는 서브네트워크
- 3. 그런데 왜 해석하기 어렵나?
    - 중첩
    - 하나의 뉴런이 여러 개념을 중첩해서 저장
    - 이를 해결하기 위해 SAE 등장
    - 중첩된 개념들을 수학적으로 분리해서 독립적으로 추출하는 방법
### 실제 연구 사례
- Inducion Heads
    - GPT 모델은 예시 몇 개 보여주면 그 패턴 따라함
    - 어텐션 헤드는 단 2개
    - 1. 현재 토큰과 비슷한 토큰이 앞에 나왔는지 찾음
    - 2. 헤드1이 찾은 위치의 다음 토큰 복사 출력
- Sparce Autoencoders
    - 중첩 문제 해결 위해 도입
    - 특정 feature가 모델의 자기 인식에까지 영향 줄 수 있음
