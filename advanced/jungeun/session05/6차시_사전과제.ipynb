## AI 최적화 & 엔지니어링

### 전이학습

#### 딥러닝 모델의 한계
- 데이터의 부족
- 데이터가 부족하기 때문에 과적합 발생 가능성 높음
- 모델 학습 시 많은 시간과 비용 소모

#### 전이학습
- 모델이 사전 학습을 통해 배운 능력을 활용해 새로운 작업을 학습
- 예시 : 흉부 X-ray를 통해 정상, 폐렴 분류 AI
- 1. 동물 사진으로 모델 사전 학습
- 2. 정상, 폐렵 구분에 쓸 부분 가져오기
- 3. 정답이 있는 X-ray로 학습
- 4. 필요 시 파인 튜닝

#### 전이학습의 4가지 용어
- Source : 기존에 학습한 문제
- Target : 새롭게 해결할 문제
- Domain : 사용한 데이터의 특성
- Task : 수행하려는 작업
- 결론 : 전이학습은 Source와 Target의 Domain이나 Task 중 적어도 하나가 다를 때, Source에서 배운 지식을 Target에 활용하는 방법

#### 전이학습의 분류
- 귀납적 전이 학습 : 기존에 학습했던 모델의 능력을 바탕으로 정답 레이블이 존재하는 사진 여러 장을 학습하면서 새로운 이미지를 보더라도 분류할 수 있도록 하는 것
- 변환적 전이 학습
- 비지도 전이 학습

#### 전이 대상에 따른 분류
- 전이 대상 : 새 작업을 배울 때 기존 모델에서 가져오는 대상
- 1. 인스턴스 기반 : 기본 데이터의 일부
- 2. 특성 기반 : 데이터 나타내는 특징 표현
- 3. 파라미터 기반 : 기존 모델의 학습된 가중치, 편향
- 4. 관계 기반 : 데이터나 개체 사이의 관계 패턴


### pre-training & fine-tuning
#### 전통적인 전이학습 : 가중치와 단어 벡터의 재사용
- 이미지 분야 : CNN은 사진의 종류를 맞히는 과정에서 윤곽, 질감, 모양 등 특징을 찾는 법을 학습하고, 그 능력은 가중치에 담김. 이때 새로운 이미지 분류 작업에서는 해당 가중치를 가져와 특징을 뽑고, 마지막 분류층은 새 작업에 맞체 교체
- 언어 분야 : 정적 단어 임베딩, 즉 미리 만든 단어 벡터를 재사용함. 이때 벡터를 감정 분석이나 스팸 분류 모델의 입력으로 재사용하는데, 같은 단어에는 같은 벡터가 부여되기 때문에 문맥에 따른 의미 차이는 담기지 않음

#### 트랜스포머 이후의 전이학습
- 이미지 분야 : 대규모 이미지로 사전 학습한 모델을 여러 작업의 출발점으로 사용함. 
- 1. 자기지도학습 : 이미지 일부를 가리고 복원하는 과정에서 모델이 이미지의 특징을 파악하는 법 배움
- 2. 이미지, 텍스트 공동 학습 : 사진과 그 사진을 설명하는 문장을 함께 보여주며, 이미지와 문장이 얼마나 잘 맞는지 판단
- 언어 분야 : 대규모 말뭉치로 모델을 사전 학습해 문맥을 반영하는 표현 얻음
- BERT : 문장의 특징을 받아 분류층을 붙임. 해당 분류층으로 학습하고, 필요에 따라 BERT 자체도 파인 튜닝
- GPT : 기존의 문장 생성 구조 활용. 기존의 답변과 모범 답변 짝지은 구조를 학습하여 활용

#### Foundation Model 
- 여러 작업에 활용할 수 있도록, 방대한 데이터로 사전 학습한 기반 모델
- 하나의 모델에서 시작하여 원하는 작업에 따라 추가 학습을 진행해 여러 특화 모델 생성

#### Fine-tuning
- 사전 학습된 모델을 새 작업의 데이터로 추가 학습하고, 기존 모델의 가중치 일부 또는 전체 조정
- 특징 추출 -> 예측 -> 오차 계싼 -> 가중치 업데이트

#### Fine-tuning 방법
##### Full Fine tuning 
- 기존 특징 추출층과 새로운 분류층을 모두 학습하여, 이미지의 특징을 추출하는 방식부터 최종 분류 기준까지 전체적으로 조정 가능
- 모든 가중치에 대한 기울기 계산과 업데이트가 이루어져서 메모리 사용량 큼
- 새로운 데이터 부족하면 기존 학습 지식 망각 문제 발생

##### Partial Fine tuning
- 일부 층을 고정하고, 나머지 층의 가중치만 업데이트
- 이미지 분류 모델 : 각 층이 학습하는 특징의 성격이 다르기에 일반적으로 앞쪽 층 고정하고 뒤쪽 층 학습
- 가중치가 적어 학습 부담 줄일 수 있음
- 어떤 층을 고정하고, 어떤 층을 학습할 건지 결정 필요

##### Frozen Backbone
- 새 분류층만 학습
- 세 방식 중 학습 부담이 가장 적음
- 기존 모델이 새 작업에 필요한 특징 잘 못 뽑으면 분류층의 변경만으로 새로운 도메인에 적응하거나 성능 높이는 데에는 한계 존재

#### Fine - tuning 한계 극복
##### PEFT
- 사전 학습 모델의 가중치 대부분 고정하고, 새 작업에 필요한 최소한의 파라미터만 학습하는 방법
- 학습 이후에 Foundation Model 1번만 저장하고, 작업 별로 학습한 작은 파라미터 묶음만 저장
- 구현 방법 : LoRA, 기존 모델에 더할 수정분을 작은 두 행렬로 학습
- LoRA의 Low Rank : 수정 행렬을 작업 행렬 A와 B의 곱으로 표현
- LoRA의 Adaption : 이 보정으로 모델을 새 작업에 맞춤
- r의 크기 줄이면 : 학습할 숫자가 줄어들어 메모리와 연산 비용 감소하지만, 스정분 표현하는 것에는 한계
- r의 크기 늘리면 : 다양한 수정분 표현이 가능하지만 학습해야할 숫자들이 늘어나 메모리와 연산 비용 증가

##### Adapter
- 기존 모델 중간에 작은 신경망 넣는 방식

##### Prompt tuning
- 모델 입력에 학습 가능한 숫자 벡터를 붙이는 방식


##### Few shot
- 몇 개의 예시를 프롬프트에 제공하는 방식
- 모델은 예시를 참고해 기준과 응답 형식을 파악해 작업을 수행하기 때문에 알맞은 예시와 입력 설계 능력 필요

##### Zero shot
- 원하는 작업을 예시 없이 지시만 주는 방식
- 모델은 지시와 사전학습 시 익힌 지식을 바탕으로 작업 수행

#### 전이학습의 한계
##### Negative Transfer
- 전이학습 적용 전보다 적용 후에 새 작업에서의 성능이 낮은 경우
- 기존 작업에 유용했던 특징에 과도하게 의존하여, 새 작업에 필요한 특징을 학습하는 데에 방해

##### Catastrophic Forgetting
- 새 데이터를 학습하며 기존 능력이 약해짐
- 새 작업에서의 성능은 좋아졌지만 기존 작업의 성능이 떨어진 경우

##### Bias Transfer
- 사전 학습에서 얻은 편향이 새 작업에도 이어지는 경우
- 사회적 편향의 경우 특정 집단에 불이익을 준다는 점에서 모델 성능 뿐만 아니라 사회적 책임의 문제로 이어짐

##### Transferablility 측정의 어려움
- Transferability : 사전 학습 모델이 배운 지식이 새 작업에 얼마나 잘 활용될 수 있는지

#### 전이학습 사용 사례
- CV 의료 영향 -> X-ray에서 질환 분류
- NLP 법률 문서 -> 계약서 조항 분류

#### 딥러닝 모델 경량화
- 가지치기 : 모델에서 예측에 미치는 영향이 작다고 판단한 부분 제거
- 양자화 : 모델이 쓰는 숫자를 덜 정밀하게 표현하여 저장 공간 줄임
- 지식 증류 : 큰 모델을 Teacher 모델로 두고, 작은 Student 모델이 정답과 Teacher 모델의 출력 분포를 함께 학습하도록 하는 방법
- 모델 구조 경량화 : 모델을 설계 시 같은 작업을 더 적은 계산으로 수행할 수 있는 수조 선택

#### AI Alignment
- 기존 한계 : LLM은 사실과 다른 문장을 만들거나 사전 학습 데이터에서 생성된 편향이 드러나는 등의 사용자의 의도와 다르게 흘러갈 수 있음
- AI Alignment : 어떤 답변이 사람에게 바람직한 지 기준을 정하고, 그 기준에 맞게 답하도록 조정하는 과정
- Helpful, Honest, Harmless
- Outer Alignment : 목표를 제대로 설정했는가?
- 보상 해킹 : 보상 기준의 허점을 이용해 점수만 높으는 현상
- Inner Alignment : 모델이 실제로 무엇을 배웠는가?
- 기만적 정렬 : 모델이 다른 목표를 추구하며, 평가만 통과하기 위해 훈련 및 감독 중에만 사람의 의도에 맞게 행동할 수 있다는 가설적 가능성

##### 가치 정렬을 어렵게 만드는 요인
- 가치 정렬의 모호성 : 사람마다 가치관이 달라서 모두가 동의하는 기준 만들기 어려움
- 확장 가능한 감독 : 사람이 직접 확인할 수 있는 범위에는 한계 존재

#### RLHF
- 언어 모델은 사전 학습을 통해 문맥에 맞게 이어질 말 예측
- 어떤 답변이 더 유용하고 안정한지는 사람의 피드백을 통해 학습
- RLHF은 사람들이 선호하는 답변에 높은 점수 부여

##### InstructGPT식 고전적 RLHF 파이프라인
- SFT : 답변을 생성하도록 언어 모델 학습
- Reward Model : 사람의 선호를 점수로 변환하는 모델 학습
- RL : 학습된 보상 최대화 
- RL 과정에서의 InstructGPT는 PPO라는 강화학습 알고리즘 사용함. 
- PPO : 보상 높이는 방향으로 모델 업데이트 하되, 답변이 생성하는 방식이 급격히 바뀌는 것을 억제

##### 부작용
- 보상 해킹 : 좋은 답변 보단 높은 보상 점수 받도록 학습
- 아첨 : 모델이 사실, 근거보단 사용자 생각에 무조건적으로 동의

#### 프롬프트 엔지니어링
- AI가 더 좋은 답을 내도록 입력을 설계하는 기술

##### 기법
- Zero shot : 예시 없이 지시만 주는 방식
- Few shot : 입력과 출력의 예시를 몇 개 보여준 뒤 같은 형식으로 답하게 하는 형식
- Chain of Though : 모델에게 중간 풀이 과정을 단계별로 쓰게 하는 방식
- Role Prompting : 모델에게 역할을 부여해서 답의 관점, 말투 설명 수준을 조절
- Instruction Prompting : 원하는 것을 구체적으로 적어서 지시하는 방식
- 이때 답의 형식, 길이, 누구를 위한 답, 꼭 들어가야 할 내용 정함
- Self Consistency : 같은 문제를 여러 번 풀게 하고, 가장 많이 나온 최종 답을 선택하는 방식

#### 한계
##### Prompt injection
- 사용자가 입력 자리에 지시처럼 보이는 문장을 끼워 넣어서, 모델이 개발자의 지시 대신 그 문장을 따르게 만드는 것
- 번역 모델 : Prompt Injection 발생 시 모델은 입력을 번역할 문장이 아닌 새로운 지시로 읽을 것. 따라서 번역 수행 X

##### Hallucination 
- 모델이 사실이 아닌 내용을 그럴듯하게 만들어내는 문제

##### 모델 의존성
- 같은 프롬프트도 모델이나 버전이 바뀌면 결과가 달라지는 문제
- 완화 방법
- 1. 모델 버전 고정
- 2. 회귀 테스트

##### 성능 한계
- 모델이 원래 못하는 일은 프롬프트로 해결 X
- 완화 방법
- 1. 외부 도구 연결
- 2. fine tuning


#### 하네스 엔지니어링
##### 프롬프트만으로 안 되는 일
- 파일을 읽고, 도구를 쓰고, 결과를 확인하고, 실패하면 복구해야 하는 일이라면 입력 1번 -> 출력 1번으로 끝나지 않을 것

##### 하네스
- 고객센터 AI처럼 도구를 쓰며 여러 단계를 스스로 진행하는 AI를 에이전트라고 함
- 모델 본체 : LLM
- 모델이 일하도록 둘러싼 시스템 : 하네스
- 하네스는 모델이 실제로 일을 할 수 있도록 필요한 도구, 정보, 규칙 갖춰주는 것

##### 구성
- 지시, 컨텍스트 : 모델이 무엇을 보고 시작하나
- 도구 인터페이스 : 모델이 무엇을, 어떤 형식으로 할 수 있나
- 실행 환경 : 행동이 실제로 어디서 실행되냐
- 상태/메모리 : 지금까지 한 일을 어떻게 기억하나
- 관측/피드백 : 행동의 결과를 어떻게 돌려받나
- 권한/가드레일 : 무엇은 하면 안되나

##### 실패는 어디서 오는가
- 인터페이스에 따라 어떤 행동을 어떤 형식으로 할 수 있는지에 따라 성능과 행동이 크게 달라짐
- 증상과 수정할 곳
- 1. 엉뚱한 도구 부름 : 도구 인터페이스
- 2. 같은 실수 반복 : 관측, 피드백
- 3. 앞부분 잊음 : 지시, 컨텍스트, 상태, 메모리
- 4. 위험한 행동을 바로 실행 : 권한, 가드레일, 실행 환경
- 5. 결과가 맞는지 모름 : 관측, 피드백

#### 루프 엔지니어링
- 루프 엔지니어링 : 모델이 행동을 제안하고, 시스템이 행동을 실행하고, 그 결과를 다시 모델에게 돌려주는 과정을 종료 조건까지 반복하도록 설계하는 것
- 루프 : AI가 답 읽기 -> 틀린 곳 찾기 -> 고치기의 과정 반복을 사람 대신 시스템이 하도록 만든 것
- 루프의 4단계 : 관측 - 판단 - 행동 - 검증
- 하네스는 모델이 일하는 공간과 조건 설계
- 루프는 그 요소들이 시간에 따라 어떤 순서로 반복되는지 설계
- 결과를 보고 다음 행동을 정하는 시스템 설계

##### 루프의 뿌리 : ReAct
- ReAct 이전에는 생각만 하는 방식과 행동만 하는 방식이 따로 연구
- 이때 생각만 하면 모르는 내용 지어냄, 행동만 하면 답 내지 못함
- ReAct은 이 둘을 반복하여 문제 완화
- 추론 -> 행동 -> 관측

##### 루프의 구성 요소
- 목표 : 무엇을 끝내려 하나
- 트리거 : 루프가 언제 시작되나
- 관측과 행동 : 매 반복에서 무엇을 보고 무엇을 하나
- 정상 종료 조건 : 성공해서 멈추는 조건
- 안전 상한 : 위험해서 멈추는 조건
- 실패, 승인 처리 : 막혔을 때, 위험할 때 어떻게 하나

##### 멈추는 조건
- 1. 성공해서 멈추는 조건 : 정상 종료
- 2. 위험해서 멈추는 조건 : 안전 상한, 실패, 승인 처리

##### 통제
- 루프 안에서 반복되고 자동으로 실행되면서 영향 커짐
- 실패마다 막는 장치 존재
- 마막 장치인 위험도에 따른 행동 분리는 행동을 자동 실행, 사후 검토, 사람 승인 후 실행 3단계로 나눔
- 학습 단계의 정렬과 실행 단계의 통제

#### 정리
- 상황별 선택 기준 : AI가 더 잘 일하게 하려면 무엇을 손볼까
- 도메인 특화 성능이 필요하고 데이터 충분 : 파인 튜닝, PEFT
- 모바일에 올려야 함 : 경량화
- 사람이 원하는 답을 못 냄 : 정렬(RLHF)
- API만 쓸 수 있고 출력 형식이 문제 : 프롬프트
- 도구를 쓰거나 파일을 읽어야 함 : 하네스
- 사람이 매번 시키는 게 문제 : 루프
