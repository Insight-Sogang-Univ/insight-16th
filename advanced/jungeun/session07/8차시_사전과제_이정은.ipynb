### 8차시 사전과제
### 시계열
#### 시계열 데이터
- 시간의 흐름에 따라 일정한 간격, 순서대로 관측 및 기록된 데이터
- 주가, 기온, 매출
#### 시계열 데이터 조건
- 시간 순서가 의미 가짐
    - 순서 자체가 정보
    - 순서 섞으면 데이터 의미 사라짐
- 관측 간격이 일정
    - 이 간격을 주기라고 부름
#### 단변량 시계열
- 시간에 따라 하나의 변수만 관측
- 오로지 자기 자신의 과거 값만 보고 미래 예측
- 사례
    - 삼성전자의 일별 종가
    - 서울의 일별 평균 온도
#### 다변량 시계열
- 시간에 따라 여러 변수를 동시에 관측
- 자기 자신의 과거 + 다른 변수들의 영향까지 고려
- 사례
    - 이메리카노의 판매량 : 기온 + 강수량 + 요인
#### 시계열 분석
- 시간 순서대로 쌓인 데이터에서 패턴 찾아냄
- 이를 바탕으로 미래 값 예측
- 시계열 분석 목표
    - 미래 데이터 예측
    - 지금까지의 기록으로 다음 시점의 값 추정
    - 기본 가정 : 과거의 패턴이 미래에도 어느정도 반복될 것이다.
- 활용 방안
    - 시간에 따라 기록되는 숫자는 다 활용 가능
    - 금융, 유통 및 커머스, 에너지, 기상, IT 서비스
#### 특징
- 시간 의존성
    - 데이터가 시간에 따라 변하고, 현재 시점의 값이 과거 시점의 값에 영향을 받는 성질
    - 가까운 과거일수록 영향 큼
    - 순서 섞으면 안됨
- 계절성 
    - 일정하게 반복되는 모양
    - 언제마나 반복되나?
- 추세 
    - 전반적으로 증가하거나 감소하는 모양
    - 어느 방향으로 가나?
- 자기 상관성
    - 시간 의존성은 과거가 현재에 영향을 준다는 성질
    - 이 영향은 자기 상관성으로 잴 수 있음
    - 자기 상관성 : 자기 자신끼리 상관관계를 갖는 특성
    - 1 : 비례
    - -1 : 반비례
    - 시차를 바꿔가며 계산하면 반복주기(계절성)을 찾을 수 있음
- 자기 상관성이 있는 데이터는 회귀분석이 불가함
    - 회귀분석은 데이터가 서로 독립적이라는 가정 위에서 이루어지기 때문
    - 오차항 독립성 가정 위반 : 통계 결과 왜곡
    - 허위 상관관계 문제
    - 예측의 어려움, 모델 복잡성 증가
- 시계열 데이터는 시계열 전용 모델 사용
- 혹은 사전처리하여 분석
#### 분석 요인
- 추세 요인
    - 시계열이 시간 전반에 걸쳐 증가하거나 감소하는 경향성을 나타내는 요인
- 계절 요인
    - 1년 이내의 일정한 주기로 같은 패턴이 반복되는 요인
- 순환 요인
    - 계절성보다 더 길고 불규칙한 주기로 반복되는 변동
    - 경기 변동 : 호황 -> 후퇴 -> 불황 -> 회복
- 불규칙 요인
    - 시계열에서 추세, 계절, 순환요인을 모두 제거하고 남은 나머지 움직임
    - 잔차, 노이즈


#### 시계열 데이터 전처리
- 시계열은 값 하나가 앞뒤 값과 이어져 있기에, 잘못된 값 하나는 그 주변의 분석까지 망가뜨림
- 시계열 특성을 반영한 전처리 필요
#### 데이터 확인
#### 결측치/이상치 탐지 및 처리
- 결측치
    - 특정 시점의 값이 수집되지 않았거나 누락된 것
    - 시계열은 행을 지우면 관측 간격이 깨져 시점끼리 비교 불가
    - 흐름을 고려해서 채워야 함
- 결측치 처리 방법
    - 삭제
    - 단순 대치
    - 보간
    - 이동 평균 대치
    - 머신러닝 기법
- 보간법
    - 주어진 데이터 포인트 값을 이용해, 두 관측값 사이의 비어있는 값을 추정하는 방법
- 이상치
    - 데이터의 일반적인 패턴에서 크게 벗어난 극단적인 값
    - 언제 나타났는지가 중요
    - 추세와 계절성 고려해서 이상치 판단
- 이상치 탐지 방법 : IQR
    - 주기성이 없거나 안정적인 구가느이 값들에서 이상치 탐지
    - 장점 : 계산이 간단, 데이터가 정규분포 따르지 않아도 사용 가능
    - 단점 : 추세, 계절성이 강한 시계열에는 부적합
- 이상치 탐지 방법 : Z-score
    - 각 값이 평균에서 몇 표준편차만큼 떨어져 있는지를 계산하는 방법 
    - 3보다 크면 이상치로 판단
    - 계절성,추세가 약하거나 값의 변동폭이 일정할 때 효과적
    - 장점 : 계산이 간단 + 해석 쉬움
    - 단점 : 데이터가 정규분포 따른다고 가정
- 이상치 탐지 방법 : STL 분해
    - 시계열을 추세, 계절, 잔차로 분해하는 방법
    - 잔차가 특정 임곗값을 초과하면 이상치 탐지
    - 계절성, 추세가 뚜렷한 시계열에 적합
    - 장점 : 진짜 이상한 값만 잡아낼 수 있음
    - 단점 : 긴 시계열에는 파라미터 튜닝 필요
#### 이상치 탐지 모델
- 분류 및 회귀 트리
    - CART는 데이터를 조건에 따라 계속 나눠 가며 예측하는 의사결정나무 모델
    - 예측 오차가 일정 임계값 이상인 지점을 이상치로 판단
    - 시계열 : 과거 구간의 값을 입력으로 하여 다음 시점의 값 예측
    - 예측오차 기반으로 이상치 판단
- 클러스터링 기반 이상 탐지
    - 군집화를 한 뒤, 어느 군집에도 잘 속하지 않는 데이터를 이상치로 판단
    - 시계열 : 특징 추출 후, 각 시점을 벡터로 표현해 클러스터링 수행
    - 중심으로부터의 거리나 군집 밀도를 기준으로 이상치 판단
- 오토 인코더
    - 입력 데이터를 작게 압축했다가 다시 원래대로 복원하도록 학습하는 신경망
    - 시계열 : LSTM Autoencoder을 사용하여 시계열의 시간적 의존성까지 학습 가능
    - 시계열을 일정 길이로 잘라 정상 구간을 학습시키고, 테스트 구간의 복원오차가 큰 곳을 이상으로 탐지
#### 이상치 처리 방법
- 삭제
    - 이상치 제거
    - 시계열에서는 주로 값을 비워서 결측치로 바꾼 뒤 다시 채워줌
- 단순 대치
    - 평균, 중앙값 등으로 바꿈
    - 간단하지만 추세, 계절성 무시
- 보간
    - 이상치를 결측치로 바꾸고, 앞뒤 값으로 보간해서 채움
    - 시간의 흐름 반영해 많이 사용
#### 평활화
- 시계열의 자잘한 노이즈를 줄여 데이터를 부드럽게 만드는 기술
- 이동평균
    - 일정한 개수의 연속된 데이터를 묶어 평균 내고, 그 묶음을 한 칸씩 옮겨가며 평균 계속 계산하는 방법
    - N이 클수록 더 긴 기간의 데이터를 평균에 포함 시킴
    - 대신 갑자기 변화가 생기면 평균에 반영되기에 오랜 시간
    - 즉 N은 보고싶은 흐름의 길이에 맞게 정해야 함
- 이동평균 사용 목적
    - 노이즈 제거 및 추세 파악
    - 계절성 제거
    - 결측치 및 이상치 대치
    - 예측의 기초 자료
- 이동 평균의 유형
    - 단순 이동평균
    - 가중 이동평균
    - 지수 이동평균 
#### 요소 분해
- 추세, 계절, 순환, 불규칙으로 섞여 있는 시계열을 각각의 요인으로 다시 나누는 과정
- 이유
    - 각각의 패턴 명확히 이해 가능
    - 변동 원인 분석 가능
    - 더 정교한 분석 가능
#### 분해 모형
- 가법 모형
    - 시계열 = 추세, 계절, 잔차를 더한 값
    - 시계열의 관측 값은 변동 요인들의 가법으로 설명될 수 있다 가정
    - 각 변동 요인들이 독립적으로 더해지며, 모든 요소의 영향이 일정
    - 시계열 전체에서 계절성의 진폭이 일정할 때 적합
- 승법 모형
    - 시계열 = 추세, 게절, 잔차 곱한 값
    - 시계열의 관측 값은 변동 요인들의 승법으로 설명될 수 있다 가정
    - 추세가 커질수록 계절성이나 불규칙성의 효과도 비례해서 커짐
    - 진폭이 시간에 따라 달라질 때 적합
#### 분해 방법
- Seasonal Decompse = MA 기반 분해
    - 이동평균으로 추세를 먼저 뽑고 나머지에서 계절과 잔차로 분리
    - 1. 추세 추출
    - 2. 추세 제거
    - 3. 계절성 추정
    - 4. 잔차 산출
- STL-LOESS 기반 분해
    - 이동평균 대신 LOESS 사용해 추세와 계절 추정
    - 데이터 전체에 선 하나를 긋는 대신, 각 지점마다 주변 데이터만 사용해서 작은 회귀선 그리고 이를 이어 붙여 부드러운 곡산 만드는 방법
    - 구간마다 유연하게 맞추기에 시간에 따라 바뀌는 추세, 계절 패턴 잘 잡아냄
- Inner Loop
    - 추세와 계절을 번갈아 추정
    - 1. 원래 데이터에서 현재 추세를 빼서 추세 제거
    - 2. 계절 요인 추정
    - 3. 계절 요인 빼서 계절성 제거
    - 4. LOESS 적용해서 추세 요인 추정
    - 5. 새로 구한 추세로 다시 1번부터 반복하며 T,S 정확하게 다듬음
- Outer Loop
    - 이상치 영향 줄이기
    - 1. Inner Loop 끝나면 잔차 계산
    - 2. 잔차가 유독 큰 시점에 작은 가중치 부여
    - 3. 가중치 반영해서 Inner Loop 다시 수행
    - 강건한 분해


#### 정상성

- 시계열의 통계적 특성이 시간에 따라 변하지 않는 성질
- 통계적 특성은 주로 평균, 분산, 자기상관 구조를 의미
- 정상성이 중요한 이유
  - 과거의 패턴과 성질이 미래에도 똑같이 유지될 것이라 가정
  - 비정상 시계열 데이터는 예측이 어려워, 수학적인 처리를 통해 정상성을 띄도록 변환한 뒤 분석해야 함

#### 강한 정상성과 약한 정상성
- 강한 정상성
  - 시계열의 확률분포 자체가 시간에 따라 전혀 변하지 않는 성질
  - 평균, 분산, 왜도, 첨도 등 분포의 모든 특성이 시간에 따라 변하지 않고 일정
  - 현실 데이터는 잡음과 외부요인이 항상 존재하기 때문에 이런 상태는 거의 찾아볼 수 없음
- 약한 정상성
  - 평균, 분산, 자기공분산만 시간에 따라 변하지 않으면 되는 성질
  - 평균이 일정하다 
  - 분산이 일정하다 
  - 자기공분산이 시차에만 의존한다
  - 현실 데이터에도 적용 가능
  - AR, MA, ARIMA 등 대부분의 시계열 모델이 가정하는 정상성이 바로 약한 정상성

#### 정상성 판단
- 시각적으로 파악
  - 정상 시계열
    - 일정한 수평선 중심으로 위아래로 오르내림
    - 오르내리는 폭이 시간에 상관없이 비슷함
    - 평균 회귀
    - 특정 주기로 반복되는 패턴이 없음
  - 비정상 시계열
    - 추세 : 그래프가 계속 오르거나 내림
    - 계절성 : 일정한 주기로 같은 모양이 반복됨
    - 분산 변화 : 시간이 지날수록 오르내리는 폭이 커지거나 작아짐

#### 단위근
- 시계열을 표현하는 식의 근(해)이 1이 되는 경우
- 과거 충격이 영구적으로 누적되는 특성
- 데이터가 평균으로 돌아오지 않고 무작위로 떠돌게 됨
- 단위근이 있으면 비정상 시계열
- 충격이 쌓이면 생기는 문제
  - 평균으로 돌아오지 않음
  - 분산이 계속 커짐

#### 단위근 검정
- 가설검정으로 판단
- ADF 검정과 KPSS 검정
- 두 검정은 귀무가설이 서로 반대
- ADF
  - 일단 비정상이라고 가정하고, 증거가 충분하면 정상으로 판결
- KPSS
  - 일단 정상이라고 가정하고, 증거가 충분하면 비정상으로 판결

#### 정상성 확보 방법
- 추세 제거
  - 차분
  - 오늘 값에서 어제를 빼서, 값 자체 대신 얼마나 변했는지를 보는 방법
  - 추세가 사라짐
  - 단위근이 사라짐
- 계절성 제거
  - 계절 차분
  - 오늘 값에서 한 주기 전의 값을 빼는 방법
  - 1차 차분이 어제와 비교라면 계절 차분은 작년 같은 때와 비교

#### 분산 안정화
- 로그 변환
  - 데이터에 로그를 취해서 값의 크기를 줄이는 방법
  - 큰 값은 많이, 작은 값은 조금 줄이기 때문에 값이 커질수록 함께 커지던 변동 폭이 비슷한 크기로 정리됨
- 한계
  - 0이나 음수에는 쓸 수 없음
  - 모든 분산 변화를 해결하지는 못함

#### 안정적 시계열
#### 자기 회귀 모델 (AR)
- 오늘 값을 자기 자신의 과거 값으로 설명하는 모형
- 시계열 데이터의 자기상관성 활용
- AR(p) : 과거 p개 시점의 값 사용
- AR(1) : 바로 전 시점의 값 하나만 사용
#### 이동 평균 모델 (MA)
- 오늘 값을 과거의 충격으로 설명하는 모형
- MA(q) : 과거 q개 시점의 충격 사용
- MA(1) : 바로 전 시점의 충격 하나만 사용
#### AR, MA 모델의 전제 조건
- AR과 MA를 쓰려면 정상성과 가역성이 필요
- 정상성
    - 충격을 받아도 시간이 지나면 평소로 돌아옴
    - AR(1) : 계수의 절댓값이 1보다 작아야 함
    - MA : 항상 정상
- 정상성 조건에 단위근 검정을 사용하면 안 되는 이유
    - ADF와 KPSS는 데이터가 정상인지 보는 검정
    - 모델의 정상성은 추정한 계수가 조건을 지키는지 확인
- 가역성
    - 판매량 기록에서 지난 충격을 거꾸로 계산해 낼 수 있는 성질
- 가역성을 확인하는 이유
    - MA 모형은 과거의 충격을 알아야 예측할 수 있지만, 데이터에는 최종 결과만 적혀 있어 충격이 직접 보이지 않음
    - 과거의 기록으로 충격을 왜곡 없이 안전하게 되돌려 계산할 수 있는지 확인
    - AR : 항상 가역 
    - MA(1) : 계수의 절댓값이 1보다 작아야 함
#### ARMA 모형
- AR과 MA를 한 식에 합친 모형
- 오늘 값이 과거 값과 과거 충격 양쪽의 영향을 받음
- 특징
  - AR만으로 설명하려면 여러 시차를 넣어야 하지만, ARMA는 더 적은 계수로 설명 가능
  - AR 부분이 정상성 조건을, MA 부분이 가역성 조건을 각각 만족해야 함
#### 자기 상관 함수 (ACF)
- 오늘 값과 k시점 전 값이 얼마나 닮았는지를 상관계수로 잰 것
- ARMA 모형에서 q 결정에 활용
- 시차 k를 바꿔 가며 구한 값을 막대로 그림
- 특징
  - -1에서 1 사이의 값을 가짐
  - 시차 0에서는 항상 1
  - 중간 날짜를 거쳐 온 영향까지 합쳐서 잼
  - 그래프의 점선 안쪽이면 0으로 봄
#### 부분 자기 상관 함수 (PACF)
- ARMA 모형에서 p 결정에 활용
- 중간 시점들의 영향을 빼고, 오늘 값과 k시점 전 값이 직접 얼마나 닮았는지를 잰 것
- 시차 1에서는 중간 날짜가 없으므로 ACF와 값이 같음
#### 불안정 시계열 분석
- 돌아갈 평소가 없는, 정상성이 없는 시계열
- 추세 : 계속 오르거나 내림
- 계절성 : 달마다 평소가 다름 
- 핵심 : 데이터를 모형이 다룰 수 있는 숫자로 바꿔 주는 것
- 불안정한 데이터의 시계열 분석 프로세스
  - 비정상 시계열 데이터 (
  - 정상성 확보 
  - 정상 시계열
  - ARMA를 사용해 안정적 시계열 분석
#### ARIMA 모형
- ARMA에 차분을 더한 모형
- ARIMA 모형의 흐름
  1. 차분을 진행해 판매량을 어제 대비  증감으로 바꿈
  2. 증감을 ARMA로 에측
  3. 예측한 증감을 더해 판매량으로 되돌림
#### SARIMA 모형
- ARIMA에 계절 차분과 계절 AR, 계절 MA를 더한 모형
- SARIMA는 여기에 작년 같은 달 대비 증감을 더한 것
- 지난달 대비 증감은 꾸준히 오르는 추세를 없애고
- 작년 같은 달 대비 증감은 계절마다 다른 수준을 없앰
- SARIMA 모형의 흐름
  - 계절 차분
  - 증감 예측
  - 판매량으로 되돌림

#### 다변량 시계열 분석
#### VAR
- AR을 변수 여러 개로 넓힌 모형
- 변수마다 식이 하나씩 있고, 각 식에 모든 변수의 과거 값이 들어감
#### VAR 모형이 적합한 경우
- 변수들이 서로 영향을 주고받는 경우
- 한 변수가 변하면 다른 변수가 며칠에 걸쳐 어떻게 반응하는지 알고 싶을 때
- 모든 변수가 정상이거나, 차분하면 정상이 되는 경우
- 변수의 수가 적고 데이터가 충분한 경우
#### VAR 모형이 부적합한 경우
- 영향이 한 방향뿐인 경우
- 변수는 많은데 데이터가 적은 경우
- 비율이 고정되지 않은 관계가 강한 경우
#### VAR 모델을 이용한 분석 방법
- 충격 반응 함수 (IRF)
  - 한 변수에 충격을 한 번 주었을 때 다른 변수가 시간에 따라 어떻게 반응하는지를 보는 도구
- 예측오차 분산 분해 
  - 예측이 빗나간 정도를 어느 쪽 충격 탓인지 나누어 비율로 보여 줌
#### 딥러닝을 활용한 시계열 분석
- 전통 통계 모델의 한계
  - 데이터가 많으면 처리 속도가 느려짐
  - 크기가 고정된 데이터만 가능
  - 비선형 패턴 포착이 어려움
#### 순환 신경망 (RNN 계열)
- RNN, LSTM, GRU
- 적합한 경우
  - 순차적 패턴이 중요한 시계열 데이터
  - 과거 정보가 미래 예측에 영향을 주는 문제
  - 시계열 길이와 목적에 따라 모델 선택 가능
- 부적합한 경우
  - 매우 긴 시계열에서는 학습이 어려울 수 있음
  - 병렬 처리 효율이 낮음
  - Transformer 계열 대비 학습 속도가 느린 편
#### Transformer 계열
- Informer (장기 예측 특화)
  - 불필요한 연산을 줄여 계산 복잡도를 대폭 낮춘 모델
  - 매우 긴 시계열 데이터 처리에 강력
- Autoformer (주기성 포착 특화)
  - 요소 분해(추세/계절성) 개념을 딥러닝 구조에 직접 결합한 모델
  - 연간 계절성 등 주기 패턴이 뚜렷하게 나타나는 판매량 예측에 강점
- TFT 
  - 과거 데이터뿐만 아니라 다양한 외부 변수를 섞어서 예측할 수 있는 다목적 아키텍처
  - 어떤 변수가 예측에 가장 중요했는지 어텐션 스코어를 통한 해석 가능
#### Foundation Model
- 거대한 데이터로 사전 학습하여 다양한 문제를 해결할 수 있는 범용 모델
- 특징
  - 범용적 모델
  - 사전 학습 후 Fine-tuning으로 특화
#### LLM 기반 모델 활용 방식
- Without Adaptation
  - LLM이 이미 시계열 지식을 가지고 있다고 가정
  - 프롬프트 엔지니어링에 집중
- Adapt LLM
  - LLM을 시계열 데이터로 Fine-tuning
- Adapt to LLM
  - 시계열을 텍스트로 변환 후 LLM 활용
#### 딥러닝 모델 선택 가이드
- 핵심 트레이드오프
- 단순함 vs 성능 : RNN < LSTM/GRU < Transformer
- 해석 가능 vs 정확도 : 통계 모델 < 딥러닝 
- 학습 시간 vs 예측 성능 : 간단한 모델은 빠르고, 복잡한 모델은 정확함
- 데이터 크기 : 적으면 통계 모델/RNN, 많으면 Transformer가 적합
