### 7차시 사전과제 - 이정은

### 추천 시스템
- 사용자의 행동 기록과 아이템 정보를 바탕으로, 사용자가 좋아할 만한 아이템을 예측해 골라 주는 시스템

##### 필요한 이유
- 고객 : 검색으로 이미 알고 있는 것을 찾고, 추천으로 아직 모르는 것을 발견
- 기업 : 잘 안보이던 비인기 아이템을 노출 가능

##### 사용되는 데이터의 종류와 특징
- 명시적 데이터
    - 고객들에 의해 입력된 데이터
    - 사용자가 선호를 직접 표현
    - 장점 : 사용자 취향 정확히 반영
    - 단점 : 명백한 데이터 X, 선호의 변화, 기준의 모호성
- 암시적 데이터
    - 고객들의 행동을 기반으로 수집된 데이터
    - 장점 : 사용자가 따로 입력 X해도 얻기 쉬움
    - 단점 : 행동 = 선호는 아닐 수도 있음

##### 추천 시스템의 분류
- 무엇을 근거로 추천하느냐?
- 1. 내용 기반 필터링
    - 콘텐츠의 유사도 기반
- 2. 협업 필터링
    - 서비스 내에 있는 다수의 사용자로부터 얻은 선호도 기반으로 사용자에게 추천
    - 두 사용자의 취향 비슷하면, 한 사람이 좋아한 아이템 -> 다른 사람도 좋아할 것
    - 사용자 기반 : 유사한 취향의 사용자
    - 아이템 기반 : 사용자가 선호한 아이템과 유사한 아이템 
- 3. 하이브리드 필터링
    - 추천 시스템 알고리즘을 조합하여 만들어진 방식
    - 콘텐츠 기반 필터링과 협업 필터링의 단점 보완

##### 내용 기반 필터링
- 사용자가 무엇을 좋아하는지, 아이템끼리 얼마나 비슷한지 알아야함
- 1. 사용자 필터링
    - 사용자가 어떤 특성 좋아하는지 정리
    - 직접 지정형 : 사용자가 자신의 취향 직접 알려줌
    - 간접 지정형 : 사용자의 과거 행동 이력에서 취향 추론 
- 사용자 프로필 획득하는 법!
    - 사용자가 가진 아이템 특성 가중치의 평균 값
    - 설문, 키워드 입력, 평가
    - 유저의 평가 바탕으로 분류 문제로 해결
- 좋은 사용자 프로필 구성 -> 추천 시스템의 정확도와 만족도 상승
- 2. 유사도
    - 아이템 정보를 숫자 벡터로 변환 후 유사도 계산
    - 사용자와 아이템을 같은 기준으로 표현해야 함
    - 사용자가 선호하는 아이템과 비교해 유사도 높은 아이템 선별 추천
- 3. TF-IDF
    - 특정 문서에는 자주 나오고 다른 문서에는 드문 단어에 높은 값을 주는 방법
    - TF : 단어 빈도
    - IDF : 역문서 빈도 
- 장점 
    - 아이템 콜드 스타트 문제 해결
    - 다른 사용자의 데이터 불필요
    - 추천 이유 설명 용이
- 단점
    - 속성 정보 부족, 품질 낮으면 정확도 감소
    - 비슷한 것만 계속 추천
    - 기록 없는 신규 사용자에게 부적합

##### 협업 필터링
- 과거에 비슷한 선호도를 가진 사용자들은 앞으로도 비슷한 선호도 가질 가능성 높음

##### 메모리 기반 협업 필터링 
    - 시스템 저장 사용자 - 아이템 평점 그대로 꺼내서 새 아이템의 평점 예측
    - 별도의 학습 없이 비슷한 이웃을 찾아 쓰기에 이웃 기반 협업 필터팅이라 부름
- 사용자 기반 
    - 나와 취향이 비슷한 사용자의 평점 활용하여 내 평점 예측 + 추천
    - Top-N 명을 이웃으로 선정
    - 이웃들의 평점 유사도로 가중평균해서 아이템 평점 예측
    - 예측 평점 높은 아이템 추천
- 아이템 기반
    - 내가 평가한 아이템들과 평점 패턴이 비슷한 아이템을 찾아 추천
    - 사용자들이 그 아이템을 함께 좋아하고 싫어한 패턴으로 유사도 판단
    - 아이템 유사도 미리 계산 후 재사용 가능
- 사용자 수가 많으면 아이템 기반이 유리함!
- 새로운 추천 필요한 플랫폼 : 사용자 기반
- 정확하고 안정적 추천 중요한 대형 플랫폼 : 아이템 기반

##### 메모리 기반 협업 필터링 한계
- 희소 행렬 문제
    - 주어진 평점, 선호도 데이터가 적을 경우 성능 저하
    - 모델 기반 협업 필터링 이용
- 콜드 스타트 문제
    - 데이터가 충분히 쌓이지 않아 적절한 추천 불가
    - 신규 사용자
    - 신규 아이템 : first rater 문제
    - 가입 시 선호 장르나 작품 직접 지정
    - 새 아이템에는 내용 기반 필터링 함께 사용
    - 임베팅 평균 사용
    - 인구 통계학적 정보 사용
- 확장성 문제
    - 아이템과 사용자가 계속 늘어날수록 계산량 커짐
    - KNN 협업 필터링 이용 : 모든 사용자 쓰는 것이 아닌, 유사도 높은 k명의 이웃만 골라 예측

##### 모델 기반 협업 필터링
- 관측된 평점 데이터로 예측 모델 학습
- 데이터의 패턴을 모델 파라미터에 담아 관측되지 않은 평점 예측
    - 모델 기반
    - 메모리 기반
- 동작 단계
    - 1. 학습 단계
    - 2. 예측 단계
- 모델 기반 협업 필터링은 평점 이면의 잠재된 패턴을 학습하여, 직접 겹치는 평가 없어도 전체 데이터의 흐름으로 빈칸 추정 + 추천 시점의 계산 가볍게 만들기 가능
- 잠재 요인
    - 평점에 직접 드러나지는 않지만, 사용자와 아이템 사이의 관계를 결정짓는 숨은 특성
- Latent Factor Model
    - 사용자와 아이템을 같은 잠재 공간 위의 벡터로 표현하고, 두 벡터의 관계로 평점 설명
- 행렬 분해
    - 큰 평점 행렬을 작은 행렬 두 개의 곱으로 쪼갬
    - 예측 평점은 두 벡터의 내적으로 계산
    - 희소성과 확장성 문제 해결
    - 내적이라는 선형 계산으로 예측한다는 한계 존재



#### 딥러닝 기반 추천 시스템
##### VAE 기반
- 사용자의 상호작용 기록을 신경망으로 확률 분포로 압축 -> 다시 복원하며 아직 보지 않은 아이템의 선호 예측
- 인코더 -> 샘플링 -> 디코더
- 장점
    - 불확실성 반영
    - 희소한 상호작용 데이터에서 강건

##### MF 한계와 VAE 극복
- 선형성 : 행렬 분해의 예측은 내적 사용
    - 비선형 활성화 함수 가진 신경망 사용
- 불확실성 무시 : 점 추정
    - 평균과 분산 함께 출력

##### AE 아닌 VAE
- AE 
    - 입력을 작은 벡터로 압축했다가 복원
    - 사용자를 점 하나로 표현
- VAE
    - 가운데 잠재 벡터를 점 하나가 아닌 분포로 표현
    - 점 하나가 아니라 그 주변을 사용해 학습
    - 기준 분포 쪽으로 당겨짐
    - = VAE가 더 안정적임

##### VAE 기반 CF의 작동 구조
- Mult-VAE는 사용자의 상호작용 기록을 입력 받음
- 인코더로 취향 분포 요약
- 디코더로 전체 아이템의 선호 확률 복원해 추천

##### VAE의 목적함수와 KL Divergence
- 목적 함수 ELBO
    - 데이터가 나올 로그 확률 하한을 키움
    - 1. 복원 항 : 원래 입력과 최대한 비슷하게
    - KL Divergence 항 : 연속적으고 부드럽게
    - 중심에서 멀어지거나, 지나치게 확신하거나, 지나치게 퍼지면 벌점 커짐
- VAE-CF 장점
    - 확률적 잠재 분포 모델링으로 불확실성 반영 + 노이즈의 영향 줄임
    - 희소한 상호작용 데이터에서도 안정적
    - 비선형성 학습




#### GNN
- 관계 자체를 데이터로 모고 모델링

##### GNN의 모티브
- 그래프
    - 노드와 엣지로 모아 구성한 자료 구조
    - 관계, 상호작용과 같이 추상적 개념 다루기에 적합
    - 비유클리드 공간의 데이터도 표현, 학습 가능

##### GNN 구조
- 그래프 구조 생성
- 메시지 패싱
    - 메시지 생성
    - 집계
    - 갱신
- 최종 출력
- 내 이웃 노드의 정보를 나에게 전달 및 요약이 핵심

##### GNN 프로세스
- GNN은 그래프 입력 받아 메시지 패싱 수행
- 각 노드는 자신의 이웃 노드들의 정보를 수집하여 자신의 은닉 상태 업데이트
- 과정 반복하여 이웃들의 취향까지 반영한 최종 임베딩
- 1. 그래프 구조 생성
- 2. 메시지 패싱
    - 메시지 생성 : 각 노드가 이웃에게 보낼 자기 정보 생성
    - 집계 : 여러 이웃 노드에서 받은 메시지 하나의 벡터로 통합
    - 갱신 : 집계된 이웃 정보와 자신의 이전 상태 결합해 새로운 임베딩 계산

##### 주요 GNN 아키텍처
- GCN 
    - 그래프 위에서 Convolution 연산 수행
    - 이웃 노드와 타겟 노드에 동일한 파라미터 적용
- GRN
    - RNN + GNN 결합
    - 시간에 따라 노드 특성 업데이트 되어 그래프의 동적 변화 포착
- GAT
    - 이웃 노드 간 중요도 학습
    - 중요 노드에 가중치
- GraphSAGE
    - 대규모 그래프에 대해 샘플링 기반 학습
    - 이웃 노드와 다겟 노드의 가중치 값을 더하지 않고 concat

##### LightGCN
- 두 연산 제거, 추천의 핵심인 연결관계만 전달하도록 구조 단순화
- 1. 특징 변환 제거
- 2. 비선형 활성화 함수 제거
- 높은 성능 가져옴

##### GNN 사례
- 핀터레스트
    - 그래프 구성 : 사용자가 저장한 이미지와 핀을 모아 둔 보드를 노드, 핀이 보드에 저장됨을 엣지로
    - 규모 억 단위
    - 전체 그래프 연산 불가능
    - 각 노드의 이웃 중 중요한 일부만 샘플링해 학습

##### 장단점
- 장점
    - 고차원 협업 시그널
    - 정형화되지 않은 데이터 학습 가능
    - 콜드 스타트 문제 완화에 효과적
- 단점
    - 과평탄화 문제
    - 과압축 문제
    - 계산 복잡도 증가
    - 순서 반영의 어려움

##### CF의 순차적 한계
- 순서 무시, 문맥 부족, 예측의 한계
##### GNN의 순차적 한계
- 지역적 정보 전파의 한계, 장기 의존성 문제
##### Transformer 기반 추천 시스템
- 사용자 행동을 순서로 접근
    - 문맥에 따라 달라지는 동일 아이템의 의미
    - 시간 순서를 뛰어넘는 핵심 의도의 포착
- Self-Attention
    - 현재 아이템과 다른 아이템 간의 관련성을 계산해, 관련성이 높은 아이템의 정보를 더 많이 반영
    - Query, Key, Value
- 연산 과정
    - 관련성을 계산하여 가중치 만듦
    - 1. Attention Score 계산
    - 2. 가중치 변환
    - 3. 가중합
##### 주요 추천 시스템 모델

    - 빠른 학습 : 병렬 처리
    - 문제 1. 계산량

    - 시퀀스의 일부 mask 토큰으로 가림
    - 과거, 미래 모두 참조하여 예측
    - 양방향 문맥을 학습에 모두 사용
    - 데이터 활용 : 여러 학습 샘플
- 상황에 따라 모델 선택!

##### 오늘날의 추천 시스템
##### 추천을 랭킹 문제가 아닌 언어 문제로 해결
    - 과거 : 조합별로 미리 하급된 공식에 따라 점수 출력
    - LLM : 추천을 점수 계산이 아닌 문장으로 출력
    - 즉 서술형 추천이 가능해짐
- 새로운 가능성
    - 대화형 추천
    - zero-shot 추천 : cold-start 문제 해결
- P5
    - 여러 추천 과제를 텍스트 프롬프트 -> 답변 형식으로 통일
    - 많은 컴퓨팅 자원과 시간 필요
- TALLRec
    - 학습 비용 줄이기 위해 yes, no 형식의 추천 지시문으로 튜닝
    - 적은 샘플로 높은 성능 가능
