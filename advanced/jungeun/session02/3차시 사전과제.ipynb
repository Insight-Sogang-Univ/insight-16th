### 3차시 사전과제

- 합성곱 신경망

#### 픽셀
- 이미지 데이터를 구성하는 가장 기본적인 단위
- 이미지 크기 : 세로 픽셀 수 X 가로 픽셀 수
- 흑백 : 0(검은색)~255(흰색)
- 컬러 : 빨강, 초록, 파랑 3가지 색상 값으로 표현, 각 채널은 0~255까지의 값을 가지며, 세 값의 조합으로 하나의 색상 완성

#### 색공간
- 이미지를 구성하는 색상을 숫자로 표현하는 방법이나 체계
##### 그레이 스케일
-  색상의 구분 없이 밝기 정보만으로 이미지 표현
- 0(검정) ~ 255(흰색) 범위이 명도 값 사용
- 3차원 텐서로 표현 가능하지만, 일반적으로 2차원 배열로 저장

##### RGB 
- 빨강, 초록, 파랑의 조합으로 색 표현
- 채널 값은 3으로 표현
- 3차원 텐서로 표현하며, 채널은 깊이라고 부르기도 함

##### HSV 
- 색상, 채도, 명도를 기준으로 색 표현

##### CNN
- 인간은 이미지 전체를 볼 때 물체를 구분하는 중요 영역과 특징에 집중하여 이미지를 파악하고, 각 부분 간의 관계를 파악
- 컴퓨터는 이미지를 작은 픽셀 단위로 구분하고, 픽셀에 저장된 밝기 등의 숫자 정보를 처리함
- 픽셀 정보를 그냥 학습하면 이미지가 본래 가지고 있던 공간 구조가 명확하게 활용되지 않아서 이미지의 특징 학습이 어려워짐

##### MLP VS CNN
- MLP는 2차원 이미지를 1차원 배열로 펼친 뒤 사용
- CNN은 이미지의 공간적인 특징을 유지하며 Conv와 Pool을 반복하여 이미지의 특징 추출 -> 이후 Flatten 통해 추출된 특징 1차원 편환 -> FC Layer 적용하여 이미지 분류

##### CNN
- 입력 이미지
- 특징 추출 : Convolution과 Pooling layor 통해 이미지의 특징 추출
- Flatten : 여러 차원으로 이루어진 특징을 FC Layer에 입력하기 위해 1차원 배열로 변환
- 분류 : 추출된 특징들을 종합하여 이미지가 어떤 클래스에 속하는지 최종 분류
- 최종 분류 결과

##### MLP와 CNN 중 CNN 활용 이유 1. 위치, 크기, 회전 등 변화에 취약
- 이미지 데이터는 단순한 숫자가 아닌, 픽셀 간의 공간적 관계와 이미지 구조를 유지하면서 특징 학습이 가능한 모델이 필요하기 때문에 CNN을 활용
- MLP는 이미지를 1차원 배열로 펼치기 때문에 학습 데이터에서 보지 못한 위치나 형태의 변화가 발생하면 기존의 숫자로 인식하지 못해 성능 저하 가능성이 존재함
- CNN은 동일한 필터를 이미지의 여러 위치에서 반복 적용하기에 특징이 이미지의 어느 위치에 나타나더라도 동일한 필터로 탐지 가능



##### 2. 파라미터 폭발
- MLP는 완전 연결 구조로, 가중치의 수가 이미지 픽셀의 개수만큼 생성 => 과적합 가능성 상승, 학습 느림
- CNN은 부분 연결 구조를 활용하기 때문에 이미지 크기가 커지더라도 필터의 크기와 개수가 동일하다면 파라미터 수는 크게 증가하지 않음.

##### 3. 공간 정보 손실
- MLP는 고차원의 데이터를 1차원으로 Flatten 하면서 공간 정보가 손실됨
- CNN은 이미지의 공간적 특징을 먼저 학습한 뒤 Flatten을 수행하기 때문에 픽셀의 위치 관계와 주변 영역의 특징 활용에 유리

##### 합성곱과 풀링
##### Convolution Layer
- 입력 이미지 전체를 일정한 크기의 필터로 훑으며 이미지의 특징 추출 + 학습
- 필터로 수직선, 수평선, 모서리, 질감 등 이미지의 패턴이나 특징 감지
- 연산 과정 : 필터가 이미지 위를 슬라이딩 -> 각 위치에서 요소별 곱셈 후 합계 계산 -> 결과를 Feature Map에 저장

##### Feature Map
- 결과 데이터
- 필터가 특정 특징에 반응할수록 해당 위치의 Feature Map 값이 커짐

##### Padding
- Convolution 연산 수행 전에 입력 이미지의 가장자리에 추가적인 픽셀 삽입
- Zero Padding 방식 가장 많이 사용
- 가장자리 영역은 필터가 적용되는 횟수가 상대적으로 적어지는 문제 해결 위해
- 이미지의 공간적 크기를 유지하며 가장자리의 특징까지 학습할 수 있도록 하는 역할

##### Stride
- 커질수록 필터가 한 번에 이동하는 거리가 늘어나므로 입력 이미지에서 연산하는 위치가 줄어, 처리해야 할 데이터의 양도 감소
- 특징을 얼마나 세밀하게 추출할지와 연산 효율을 어떻게 조절할지 결정하는 요소

##### Pooling
- 이미지의 작은 위치 변화에 대한 민감도를 낮춰 모델이 특징을 더 안정적으로 인식하게 함
- Max pooling : 지정된 영역 안에서 가장 큰 값 선택
- Average pooling : 지정된 영역 안에서 모든 값의 평균 계산

- 장점 : 차원 축소로 계산량 감소, 위치 변화에 대한 강건성, 과적합 방지에 도움, 노이즈 현상 오나화
- 단점 : 정보 소실의 위험, 정확한 위치 정보 감소

##### 이미지 분류
- 이미지를 수치 데이터로 인식 -> 사전 정의된 클래스 중 하나로 분류

- Vision Transformer : 패치 기반 입력 처리

##### ViT
- 이미지를 작은 패치 단위로 나누고, 이를 Transformer에 입력하여 이미지의 특징을 학습하는 모델
- 이미지 패치 간의 관계 학습

##### CNN의 확장 가능성
- YOLO : 이미지 안에 존재하는 사람, 자동차, 자전거 등을 각각 탐지하고, 여러 객체를 찾아내는 작업에 활용

##### 객체 탐지
- 이미지 안에 있는 개별 객체의 종류와 위치를 함께 예측
- Bounding Box는 이미지에서 객체가 위치한 영역을 둘러싸는 사각형
- 객체 탐지 모델은 사각형의 위치와 크기를 예측하여 이미지 안에서 객체가 어디에 있는지 표시
- 일반적으로 중심 좌표 (x,y), 너비(w), 높이(h) 이용하여 위치 표현
- 모델 적용
- Classification Head(객체의 종류 예측)
- Regression Head(객체의 위치 좌표 예측)

##### 이미지 캡셔닝
- 이미지의 내용을 자연어 문장으로 표현하는 작업
- 이미지 특징을 추출하는 CNN과 문장을 생성하는 RNN이 함꼐 사용
- 객체에 대한 판단과 관계를 파악하고 자연어 형태로 표현하는 작업
- CNN : 이미지 특징 추출
- RNN : 문장 생성
- 단계 : CNN을 이용한 특징 추출 -> Attention Network를 이용하여 단어 생성 -> RNN을 이용하여 문장 완성

##### RNN
- 순차 데이터 : 데이터의 순서가 의미를 갖는 데이터로, 순서를 바꾸면 정보가 달리지는 데이터
- 시간적 순차 데이터 : 시간 순
- 비시간적 순차 데이터 : 시간 외 기준
- 수치형 : 값의 크기와 거리에 의미 O
- 범주형 : 값의 크기와 거리에 의미 X
- 정형 데이터 : 샘플 간 관계는 상호 독립적으로 순서를 섞어도 의미 그대로 유지
- 순차 데이터 : 샘플 간 관계는 상호 의존적으로 순서를 섞으면 의미 달라짐

##### RNN
- 순차 데이터를 처리할 별도의 모델
- 은닉층에 순환 구조가 존재함
- 구성 : 입력층, 은닉층, 출력층, 입력 가중치, 순환 가중치, 출력 가중치
- 원리 : 매 시점마다 현재 시험 입력값과 이전 시점 은닉 상태를 이용해 현재의 은닉 상태를 계싼하고 그 결과를 다음 시점으로 넘겨줌
- 은닉 상태 계산 : 현재 입력 X + 이전 은닉 상태 -> 새로운 은닉 상태
- 출력 계산 : 은닉 상태로 현재 시점의 출력을 만듦

##### RNN 아키텍처 
###### one to many : 하나의 입력에서 시퀀스 출력
- ex. 이미지 캡셔닝
###### many to one : 시퀀스 입력에서 하나의 값 출력
- ex. 스팸 메일 분류, 감성 분류,
- 전체 문장을 읽은 후 최종 판별을 함
###### many to many : 시퀀스 입력에서 시퀀스 출력
- 동기식 many to many : 매 시점마다 즉각 출력
- ex. 객체 태깅
- 비동기식 many to many : 전체 입력을 끝까지 읽은 뒤 한 번에 출력
- ex. 기계 번역, 요약

##### RNN의 한계
- 시간에 따른 역전파 구조 : RNN은 순전파를 통해 시퀀스를 처리한 뒤, 시간을 거슬러 올라가며 오차를 전파하여 가중치 기울기를 누적 계산
- 기울기 소실 : 역전파 과정에서 기울기가 과거 시점으로 갈수록 0에 가까워지는 현상
- 기울기 폭주 : 역전파 과정에서 기울기가 과거 시점으로 갈수록 점점 커져 매우 큰 값이 되는 현상
- 느린 훈련 시간 : RNN은 시점을 하나씩 순서대로만 계산할 수 있어서, 시퀀스가 길수록 학습에 시간이 오래 걸림
- CNN은 필터의 위치가 독립적이라서 여러 위치를 동시에 처리 가능
- RNN은 순차적이라서 앞 시점의 결과가 있어야 다음 계산 가능
- 장기 의존성 문제 : 시퀀스에서 관련된 정보가 멀리 떨어져 있을수록, RNN이 그 관계를 기억하고 학습하지 못하는 문제


##### LSTM & GRU
- 중요한 정보는 오래 기억하고, 불필요한 정보는 잊는 메모리 구조 필요

##### LSTM
- 기억을 2개의 상태로 나눠서 관리
- 셀 상태 : 장기 기억
- 은닉 상태 : 단기 기억 & 출력
- Forget Gate, Input Gate, Output Gate로 구성
- 장점 : 장기 기억과 단기 기억을 분리해서 관리하여 정보를 정교하게 다룸
- 딘점 : 구조가 복잡하고 계산량이 많음

##### GPU 
- 기억을 하나로 통합해서 관리
- 은닉 상태에 기억과 출력 모두 포함
- Update Gate, Reset Gate 로 구성
- 장점 : LSTM보다 단순하고 계산이 빠름
- 단점 : 정교함은 상대적으로 떨어짐

##### LSTM 구조
- 정보가 큰 변화 없이 다음 시점으로 흘러감
- 게이트가 정보를 빼거나 더해서 조절
- Forget gate : 무엇을 잊을까?
- Input gate : 무엇을 새로 기억할까?
- Final memory cell : 기억 업데이트
- Output gate : 무엇을 내보낼까?
- 장점 : 기울기 소실 완화, 장기 의존성 학습 가능, 문장 번역, 음성 인식 등 시퀀스 과제에서 RNN보다 좋은 성능
- 단점 : 계산량이 많아 학습이 느리고, 메모리 많이 사용, 병렬 처리 불가, 시퀀스가 길면 기울기 소실 여전히 존재

##### GPU
- LSTM의 핵심 아이디어는 유지하면서, 구조는 단순하게 줄인 모델
- Forget gate + Input gate = Update gate
- Reset gate 으로 과거 기억을 얼마나 무시할지 정함
- Output gate 존재하지 않음
- Gate의 개수가 3개에서 2개로 감소
- 장점 : 구조가 단순하여 이해와 구현이 쉬움, 파라미터가 적어 학습이 빠르고 시간 절약, 데이터가 적어 학습 안정적, 기억 관리 간단, LSTM과 비슷한 성능

##### Seq2Seq
- 입력 시퀀스를 받아 다른 시퀀스를 출력하는 모델
- 입력과 출력의 길이가 서로 달라도 됨
- 인코더와 디코더로 구성

##### 인코더
- 입력 문장의 모든 단어를 순서대로 읽고, 문장 전체의 정보를 하나의 벡터로 압축
- 단어 토큰화
- 임베딩
- LSTM 셀에 한 단어씩 입력
- 마지막 시점의 은닉 상태 = 문장 전체 정보가 담긴 컨텍스트 벡터

##### 디코더 
- 컨텍스트 백터를 받아, 출력 문장을 한 단어씩 순서대로 생성하는 LSTM

##### 컨텍스트 벡터
- 인코더가 입력 문장 전체를 읽고 만든 문장 요약 벡터 
- 인코더 LSTM의 마지막 시점 은닉 상태
- 인코더와 디코더를 연결하는 유일한 통로

##### 한계
- 병목 현상 : 인코더가 입력 문장 전체를 고정된 크기의 컨텍스트 백터 하나에 압축하면서 정보가 좁은 통로에 막혀 손실되는 현상
