> # **1.CNn**

## 1-1. 픽셀, 채널, 색공간

### **픽셀**
- 이미지 데이터를 구성하는 가장 기본적인 단위
- 이미지의 특정 위치에 해당하는 밝기 또는 색상 정보를 숫자로 표현
- 이미지 크기: 세로 픽셀 수 × 가로 픽셀 수
- 흑백 이미지의 경우 각 픽셀은 일반적으로 0~255의 밝기 값으로 표현
  - 0: 검은색 / 255: 흰색

### **채널**
- 이미지의 색상이나 밝기 정보를 구분하여 저장하는 각각의 데이터 층
- 이미지의 색공간에 따라 채널 수가 달라짐
  - Grayscale: 1개
  - RGB: 3개

### **색공간**
- 이미지의 색상 정보를 숫자로 표현하는 방법이나 체계
- 대표적인 색공간: Grayscale, RGB, HSV

1. **Grayscale**
   - 색상 정보 없이 픽셀의 밝기만으로 이미지를 표현
   - 각 픽셀은 하나의 명도 값으로 구성
   - 이미지 크기: H × W × 1 (일반적으로 H × W로 저장)

2. **RGB**
   - Red(빨강), Green(초록), Blue(파랑) 세 가지 색상의 조합으로 색을 표현
   - 각 채널은 일반적으로 0~255의 값을 가짐
   - 이미지 크기: H × W × 3

3. **HSV**
   - Hue(색상), Saturation(채도), Value(명도)를 기준으로 색을 표현
   - RGB와 달리 색상, 채도, 명도 정보를 구분하여 표현

<br>

## 1-2. CNN의 개념과 구조

### **CNN (Convolutional Neural Network)**
- 이미지의 공간적 특징을 유지하면서 학습하는 신경망
- 이미지에 포함된 픽셀 간의 위치 관계와 주변 영역의 특징을 활용
- 필터를 이용해 이미지의 특징을 추출하고, 이를 바탕으로 이미지를 분류

### **CNN의 전체 구조**

**(Convolution + Pooling) 반복 → Flatten → Fully Connected Layer**

1. **특징 추출**
   - Convolution Layer: 필터를 이용해 이미지의 모서리, 질감, 형태 등의 특징 추출
   - Pooling Layer: 특징 맵의 크기를 줄이고 중요한 특징을 유지

2. **Flatten**
   - 여러 차원으로 이루어진 특징 맵을 1차원 배열로 변환

3. **Classification (분류)**
   - Fully Connected Layer: 추출한 특징들을 종합하여 이미지가 어떤 클래스에 속하는지 최종적으로 분류

=> CNN은 이미지의 특징을 먼저 학습한 후, 이를 바탕으로 분류한다는 점이 핵심

> # **2. MLP vs. CNN**

## 2-1. MLP의 한계와 CNN을 사용하는 이유

### **MLP (Multi-Layer Perceptron)**
- 입력 데이터를 Fully Connected Layer에 전달하여 학습하는 구조
- 이미지 데이터를 입력받을 때 일반적으로 2차원 이미지를 1차원 배열로 Flatten하여 사용
- 이미지의 공간적 정보가 명시적으로 유지되지 않는다는 한계

### **MLP의 한계**

1. **위치 변화에 취약**
   - MLP는 각 픽셀의 위치에 대응하는 가중치를 학습
   - 이미지 속 객체의 위치가 조금만 바뀌어도 입력 데이터가 크게 달라짐
   - 학습하지 않은 위치나 형태의 변화에 취약

   => CNN은 동일한 필터를 이미지의 여러 위치에 반복 적용하고 Pooling을 통해 작은 위치 변화에도 비슷한 특징을 유지

2. **파라미터 폭발**
   - MLP는 완전연결 구조이므로 입력 노드와 다음 층의 모든 노드가 연결됨
   - 이미지가 커질수록 가중치 수가 급격히 증가
   - 과적합 가능성 상승 및 학습 속도 저하

   => CNN은 작은 영역에 필터를 적용하는 부분 연결 구조와 가중치 공유를 사용하여 파라미터 수를 줄임

3. **공간 정보 손실**
   - MLP: 이미지를 처음부터 1차원 배열로 변환
   - 픽셀 값 자체는 유지되지만, 주변 픽셀 간의 공간적 관계가 명시적으로 유지되지 않음

   => CNN은 이미지의 다차원 구조를 유지한 상태에서 Convolution과 Pooling을 통해 특징을 먼저 추출한 후 Flatten 수행

### **MLP vs. CNN 핵심 차이**

- MLP: 이미지의 픽셀 값을 1차원으로 펼쳐서 학습
- CNN: 이미지의 공간적 구조를 유지하면서 특징을 추출한 후 학습

=> 이미지의 위치 관계와 주변 픽셀의 특징을 효과적으로 학습하기 위해 CNN 사용

> # **3. 합성곱과 풀링**

## 3-1. Convolution Layer

### **개념**
- 입력 이미지에 일정한 크기의 필터(Filter)를 적용하여 이미지의 특징을 추출하는 계층
- 필터는 수직선, 수평선, 모서리, 질감 등 이미지의 특정한 패턴을 감지

### **연산 과정**

1. 필터가 이미지 위를 일정한 간격으로 이동 (Sliding)
2. 필터가 위치한 영역의 픽셀 값과 필터의 가중치를 같은 위치끼리 곱한 후 모두 더함
3. 계산 결과를 Feature Map의 해당 위치에 저장

### **Feature Map**
- Convolution 연산을 통해 생성되는 결과 데이터
- 필터가 특정 특징에 반응할수록 해당 위치의 Feature Map 값이 커짐
- 사용하는 필터의 개수에 따라 Feature Map의 개수가 결정됨

=> 여러 개의 필터를 사용하면 이미지의 다양한 특징을 추출할 수 있음

## 3-2. Padding

### **개념**
- Convolution 연산을 수행하기 전 입력 이미지의 가장자리에 추가적인 픽셀을 삽입하는 방법
- 일반적으로 이미지 바깥에 0을 추가하는 Zero Padding 사용

### **효과**
- Padding을 적용하지 않으면 Convolution을 반복할수록 이미지의 크기가 감소
- 가장자리 픽셀은 필터가 적용되는 횟수가 상대적으로 적음
- Padding을 통해 Feature Map의 크기를 조절하고 가장자리의 특징까지 학습 가능

=> 이미지의 공간적 크기를 유지하고 가장자리 정보의 손실을 줄이는 역할

## 3-3. Stride

### **개념**
- 필터가 이미지 위에서 한 번 이동할 때 이동하는 간격 또는 보폭

**Stride = 1**
- 한 칸씩 이동
- 이미지의 여러 위치를 세밀하게 확인
- Feature Map의 크기가 상대적으로 큼

**Stride = 2**
- 두 칸씩 이동
- 일부 위치를 건너뛰며 연산
- Feature Map의 크기가 상대적으로 작음

=> Stride가 커질수록 연산 위치가 줄어들어 Feature Map의 크기와 연산량 감소

## 3-4. Pooling

### **개념**
- 일정한 영역을 설정하고 해당 영역에 포함된 값들을 하나의 값으로 요약하는 과정
- Feature Map의 공간적 크기를 줄이고 작은 위치 변화에 대한 민감도를 낮춤

### **종류**

1. **Max Pooling**
   - 지정된 영역에서 가장 큰 값을 선택

2. **Average Pooling**
   - 지정된 영역에 포함된 값들의 평균을 계산

### **장점**
- Feature Map의 차원을 축소하여 계산량 감소
- 작은 위치 변화에도 비슷한 특징을 유지
- 불필요한 세부 정보와 공간적 복잡성을 줄여 과적합 완화에 도움

### **단점**
- 크기를 줄이는 과정에서 일부 세부 정보 손실
- 객체의 정확한 위치나 경계 정보가 중요한 작업에서는 불리할 수 있음

=> Convolution Layer와 Pooling은 특징 추출을 담당하며, Padding과 Stride는 출력 크기와 정보 보존 정도를 조절

> # **4. 이미지 딥러닝 응용**

## 4-1. 이미지 분류

- 이미지를 픽셀 값으로 구성된 수치 데이터로 인식
- 사전에 정의된 클래스 중 하나로 이미지를 분류하는 작업

**대표적인 모델**
- AlexNet: 초기 CNN 모델로, ReLU와 Dropout 등 다양한 기술을 도입하여 딥러닝 이미지 인식의 발전에 기여
- VGGNet: 작은 3 × 3 필터를 반복적으로 적용하고 네트워크를 깊게 구성하여 복잡한 이미지 특징 학습
- ResNet: Skip Connection(잔차 연결)을 도입하여 깊은 신경망의 학습을 효과적으로 수행

## 4-2. 객체 탐지

### **개념**
- 이미지 안에 존재하는 개별 객체의 종류와 위치를 함께 예측하는 작업
- 이미지 전체의 클래스를 분류하는 Image Classification에서 확장된 개념

### **Bounding Box**
- 이미지에서 객체가 위치한 영역을 둘러싸는 사각형
- 중심 좌표(x, y), 너비(w), 높이(h)를 이용하여 위치와 크기 표현

### **모델 적용 방식**
- CNN을 통해 이미지의 Feature Map 추출
- 이후 두 가지 Head를 통해 객체의 종류와 위치를 각각 예측

1. **Classification Head**
   - 객체가 사람, 자동차, 강아지 등 어떤 종류인지 예측

2. **Regression Head**
   - 객체를 둘러싸는 Bounding Box의 위치와 크기 예측

- 대표 모델: YOLO, Faster R-CNN, SSD

## 4-3. 이미지 캡셔닝 
### **개념**
- 이미지 속 객체의 종류뿐만 아니라 객체 간의 관계와 행동까지 파악하여 자연어 문장으로 표현하는 작업

### **모델 적용 방식: CNN + RNN**

1. **CNN: 이미지 특징 추출**
   - 이미지에서 사람, 자전거, 도로 등 객체와 시각적 특징 추출

2. **RNN: 문장 생성**
   - CNN이 추출한 이미지 특징을 바탕으로 단어를 순서대로 생성
   - 이전에 생성한 단어 정보를 활용하여 다음에 올 단어를 예측

3. **Attention**
   - 문장을 생성할 때 현재 단어와 관련된 이미지 영역에 더 높은 가중치를 부여
   - 예를 들어 '사람'을 생성할 때는 사람 영역에, '자전거'를 생성할 때는 자전거 영역에 집중

=> 이미지의 특징을 추출하는 CNN과 자연어를 순차적으로 생성하는 RNN을 결합한 응용 분야

> # **5. 순차 데이터**

## 5-1. 순차 데이터란?

- 데이터의 순서가 의미를 가지며, 순서를 바꾸면 데이터의 의미가 달라지는 데이터
- 순서에 따라 각 데이터 간의 관계가 달라지는 것이 특징

### **순차 데이터의 예시**
1. 주가 및 기온: 시간에 따라 변화하는 수치 데이터
2. 자연어: 문장 속 단어의 순서에 따라 의미 변화
3. DNA 염기서열: 염기 배열 순서에 따라 정보 변화
4. 음성 및 오디오: 시간에 따른 소리의 진폭 변화
5. 비디오: 프레임의 순서와 시간에 따른 객체의 움직임

## 5-2. 정형 데이터 vs. 순차 데이터

**정형 데이터**
- 각 데이터의 순서를 바꾸어도 의미가 달라지지 않음
- 샘플 간의 관계가 상대적으로 독립적
- 예시: 고객 정보 데이터

**순차 데이터**
- 데이터의 순서를 바꾸면 의미가 달라짐
- 이전 시점이나 순번의 정보가 현재 데이터에 영향을 미침
- 예시: 주가, 문장, DNA 염기서열

### **순차 데이터의 주요 특징**

1. **순서 의존성**
   - 데이터를 구성하는 요소들의 순서 자체가 중요한 정보
   - 예: '개가 사람을 물었다'와 '사람이 개를 물었다'는 단어는 같지만 의미가 달라짐

2. **자기상관성**
   - 이전 시점의 데이터가 다음 시점의 데이터에 영향을 미치는 특성
   - 과거의 정보가 현재의 데이터와 연관됨

=> 기존의 회귀, 결정트리, SVM 등 일반적인 머신러닝 모델만으로는 순차 데이터의 순서와 의존성을 효과적으로 학습하기 어려움. 따라서 순차 데이터를 처리하기 위한 RNN과 같은 모델이 필요

> # **6. RNN**

## 6-1. 구조와 원리

### **RNN (순환 신경망)**
- 순차 데이터를 입력받아 순서 정보를 유지하며 처리하는 신경망
- 은닉층에 순환 구조가 존재하여 이전 시점의 정보를 다음 시점으로 전달
- 현재 입력과 이전 시점의 은닉 상태를 함께 활용하여 새로운 은닉 상태를 계산

### **RNN의 주요 구성 요소**

- 입력층(x): 각 시점에 들어오는 입력 데이터
- 은닉층(h): 현재 입력과 이전 은닉 상태를 바탕으로 새로운 은닉 상태 생성
- 출력층(o): 은닉 상태를 바탕으로 현재 시점의 결과 출력
- 입력 가중치(U): 입력 데이터를 은닉층으로 전달
- 순환 가중치(V): 이전 은닉 상태를 현재 은닉층으로 전달
- 출력 가중치(W): 은닉 상태를 최종 출력으로 변환

### **작동 원리**
1. 현재 시점의 입력값 xₜ와 이전 시점의 은닉 상태 hₜ₋₁를 입력받음
2. 두 정보를 결합하여 새로운 은닉 상태 hₜ를 계산
3. 현재 은닉 상태를 다음 시점으로 전달
4. 은닉 상태를 바탕으로 각 시점의 출력 생성

- **Tanh 활성화 함수**
  - 값을 -1~1 사이로 제한하여 은닉 상태가 지나치게 커지는 것을 방지
  - 비선형성을 추가하여 복잡한 패턴 학습 가능

=> 이전 시점의 정보를 현재 시점에 전달하는 순환 구조가 RNN의 핵심

## 6-2. RNN 아키텍처 종류와 활용 예시

### **1. One to Many**
- 하나의 입력을 받아 여러 개의 순차적인 출력을 생성
- 예시: 이미지 캡셔닝
  - 이미지 정보를 입력받아 단어를 순서대로 생성하여 문장 완성

### **2. Many to One**
- 여러 개의 순차적인 입력을 받아 하나의 결과 출력
- 예시: 스팸 메일 분류, 감성 분석
  - 문장을 순서대로 읽은 후 전체 문장이 스팸인지, 긍정인지 부정인지 판단

### **3. Many to Many**
- 여러 개의 순차적인 입력을 받아 여러 개의 순차적인 출력 생성

- 동기식 Many to Many
  - 각 입력 시점에 대응하는 출력을 즉시 생성
  - 예시: 객체 태깅

- 비동기식 Many to Many
  - 전체 입력을 읽은 후 출력 시퀀스 생성
  - 예시: 기계 번역, 문서 요약

## 6-3. RNN의 한계점

### **1. 기울기 소실 **
- 역전파 과정에서 기울기가 과거 시점으로 갈수록 점점 작아져 0에 가까워지는 현상
- 과거 시점의 정보가 학습에 거의 반영되지 않음
- 결과적으로 먼 과거의 중요한 정보를 기억하기 어려움

### **2. 기울기 폭주 **
- 역전파 과정에서 기울기가 과거 시점으로 갈수록 지나치게 커지는 현상
- 가중치가 크게 변하면서 오차가 급격히 증가하거나 진동
- 심한 경우 NaN이 발생하여 학습 중단 가능

### **3. 느린 훈련 시간**
- RNN은 이전 시점의 계산이 완료되어야 다음 시점의 계산 가능
- 시퀀스를 순서대로 처리하기 때문에 병렬 처리에 한계
- 시퀀스가 길어질수록 계산 시간과 메모리 부담 증가

### **장기 의존성 문제 **

- 시퀀스에서 관련된 정보가 멀리 떨어져 있을수록 RNN이 해당 관계를 기억하고 학습하지 못하는 문제
- 기울기 소실과 폭주, 긴 시퀀스의 높은 계산 비용 등이 주요 원인

=> 중요한 정보는 오래 기억하고 불필요한 정보는 잊을 수 있는 새로운 메모리 구조의 필요성 대두

> # **7. LSTM & GRU**

## 7-1. LSTM & GRU의 개념

- 기본 RNN의 장기 의존성 문제를 완화하기 위해 등장한 모델
- Gate를 활용하여 어떤 정보를 기억하고 어떤 정보를 잊을지 선택적으로 조절
- Gate는 일반적으로 Sigmoid 함수를 사용하여 0~1 사이의 값으로 정보의 통과 정도 결정
  - 0에 가까울수록 정보 차단
  - 1에 가까울수록 정보 유지

## 7-2. LSTM (Long Short-Term Memory)

### **개념**
- 장기 기억과 단기 기억을 분리하여 관리하는 신경망
- 두 가지 상태를 활용

1. **Cell State**
   - 장기 기억을 담당
   - 중요한 정보를 다음 시점까지 전달

2. **은닉 상태**
   - 단기 기억 및 현재 출력 담당

### **LSTM의 세 가지 Gate**

1. **Forget Gate**
   - 이전 셀 상태에서 더 이상 필요하지 않은 정보를 삭제

2. **Input Gate**
   - 새롭게 입력된 정보 중 중요한 내용을 선택하여 장기 기억에 저장

3. **Output Gate**
   - 장기 기억 중 현재 필요한 정보만 선택하여 은닉 상태로 출력

### **장단점**

- 장점
  - 장기 기억을 별도로 관리하여 기울기 소실 문제 완화
  - 중요한 정보는 유지하고 불필요한 정보는 선택적으로 제거
  - 긴 시퀀스의 정보와 문맥을 학습하는 데 유리

- 단점
  - Gate 3개와 상태 2개로 구조가 복잡
  - 계산량과 메모리 사용량이 많음
  - 여전히 시점별 순차 계산이 필요하여 병렬 처리에 한계

## 7-3. GRU 

### **개념**
- LSTM의 핵심 아이디어인 선택적 정보 기억과 삭제는 유지하면서 구조를 단순화한 모델
- Cell State와 Hidden State를 구분하지 않고 하나의 Hidden State로 기억과 출력을 관리

### **GRU의 두 가지 Gate**

1. **Update Gate**
   - 과거의 기억과 새로운 정보를 어느 정도 비율로 반영할지 결정
   - LSTM의 Forget Gate와 Input Gate의 역할을 통합

2. **Reset Gate**
   - 새로운 정보를 생성할 때 과거의 기억을 얼마나 무시할지 결정

### **장단점**

- 장점
  - Gate 2개와 상태 1개로 구조가 단순
  - LSTM보다 파라미터와 계산량이 적어 학습 속도가 빠름
  - 적은 데이터에서도 비교적 안정적으로 학습 가능

- 단점
  - 장기 기억과 출력을 별도로 관리하지 못함
  - 정보 관리의 정교함이 상대적으로 떨어질 수 있음
  - 데이터가 많고 복잡한 과제에서는 LSTM이 더 좋은 성능을 보일 수 있음

## 7-4. LSTM vs. GRU: 상황별 모델 선택

**LSTM이 적합한 경우**
- 긴 시퀀스에서 문맥과 장기 의존성을 정교하게 학습해야 하는 경우
- 데이터가 충분하고 복잡한 패턴을 학습해야 하는 경우
- 계산 효율보다 모델 성능이 중요한 경우
- 예시: 기계 번역, 언어 모델링, 장기 시계열 예측

**GRU가 적합한 경우**
- 학습 속도와 계산 효율이 중요한 경우
- 컴퓨팅 자원이 제한적인 경우
- 데이터 양이 적어 과적합 위험이 있는 경우
- 실시간 예측이 필요한 경우

=> LSTM과 GRU는 우열 관계라기보다 **기억 관리의 정교함(LSTM)과 계산 효율성(GRU)** 사이에서 문제 상황에 맞게 선택하는 모델

> # **8. Seq2Seq**

## 8-1. Seq2Seq의 구조와 원리

### **Seq2Seq (Sequence-to-Sequence)**
- 하나의 입력 시퀀스를 받아 다른 시퀀스를 출력하는 모델
- 입력과 출력의 길이가 서로 달라도 된다는 특징
- 기존 Many to Many 구조를 Encoder와 Decoder라는 두 개의 모듈로 나누어 구현
- 내부 셀로 주로 LSTM 또는 GRU 사용

**대표적인 활용 분야**
- 기계 번역
- 챗봇
- 문서 요약
- 음성 인식

### **1. Encoder**

- 입력 문장의 모든 단어를 순서대로 읽고 문장 전체의 정보를 하나의 Context Vector로 압축
- 작동 과정
  1. 입력 문장을 단어 단위로 토큰화
  2. 각 단어를 숫자 벡터로 변환 (Embedding)
  3. LSTM에 단어를 순서대로 입력하며 은닉 상태 업데이트
  4. 마지막 시점의 은닉 상태를 Context Vector로 전달

### **2. Context Vector**

- 인코더가 입력 문장 전체를 읽고 만든 문장 요약 벡터
- 인코더와 디코더를 연결하는 역할
- 마지막 시점의 은닉 상태에 문장 전체의 정보가 압축되어 있음
- LSTM을 사용하는 경우 셀 상태도 함께 전달

### **3. Decoder**

- Context Vector를 바탕으로 출력 문장을 한 단어씩 생성
- 작동 과정
  1. Context Vector를 초기 은닉 상태로 전달
  2. 시작 신호 `<sos>`를 첫 입력으로 사용
  3. Softmax를 통해 다음 단어의 확률 계산
  4. 예측한 단어를 다음 시점의 입력으로 다시 사용
  5. 종료 신호 `<eos>`가 나오면 문장 생성 종료

=> Encoder가 입력 문장을 읽고 요약하면, Decoder가 해당 정보를 바탕으로 새로운 출력 문장을 생성

## 8-2.  병목 

### **개념**
- 인코더가 입력 문장 전체의 정보를 고정된 크기의 Context Vector 하나에 압축하면서 정보가 손실되는 현상

### **발생 원인**
- 입력 문장의 길이와 관계없이 동일한 크기의 Context Vector에 모든 정보를 담아야 함
- 짧은 문장은 상대적으로 정보 손실이 적지만, 긴 문장은 모든 정보를 충분히 담기 어려움
- 문장 앞부분의 정보가 여러 시점을 거치며 희미해져 장기 의존성 문제가 발생할 수 있음

### **결과**
- 긴 입력 시퀀스일수록 정보 손실 증가
- 문맥 파악이 어려워지고 번역이나 문장 생성의 정확도 저하

=> 고정된 크기의 Context Vector에 모든 정보를 압축하는 것이 기본 Seq2Seq의 한계

**이를 해결하기 위해 다음 단계에서는 Attention Mechanism을 활용하여 입력 문장의 중요한 부분에 선택적으로 집중하는 방법을 학습함**
