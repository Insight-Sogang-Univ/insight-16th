
> # **1. 시계열 데이터란?**

## 1-1. 시계열 데이터의 정의

### **시계열 데이터(Time Series Data)란?**
: 시간의 흐름에 따라 일정한 간격으로 관측하고 기록한 데이터

- 주가, 기온, 매출액, 전력 사용량 등이 대표적인 시계열 데이터
- 관측값의 시간적 순서가 중요한 정보를 가짐
- 과거 관측값이 현재 및 미래의 관측값에 영향을 미칠 수 있음
- 주로 과거의 패턴을 분석하고 미래 값을 예측(Forecasting)하는 데 활용

$$
Y_1,Y_2,\dots,Y_t,\dots,Y_T
$$

- $Y_t$: 시점 $t$에서 관측된 값
- $t$: 시간 인덱스
- $T$: 마지막 관측 시점

### **시계열 데이터의 주요 분류**

| 구분 | 단변량 시계열 | 다변량 시계열 |
| --- | --- | --- |
| 정의 | 하나의 변수를 시간에 따라 관측 | 여러 변수를 시간에 따라 동시에 관측 |
| 예측에 활용하는 정보 | 자기 자신의 과거 값 | 자기 자신과 다른 변수의 과거 값 |
| 예시 | 삼성전자 일별 종가 | 주가, 환율, 금리, 거래량 |

**시계열 데이터와 횡단면 데이터의 차이**

- **횡단면 데이터**: 특정 시점에 여러 대상을 관측한 데이터
- **시계열 데이터**: 하나 또는 여러 대상을 시간의 흐름에 따라 반복 관측한 데이터
- 시계열 데이터는 관측 순서를 임의로 변경하면 시간적 의존성과 패턴이 훼손될 수 있음

## 1-2. 시계열 데이터의 특성

### **1) 시간의존성(Temporal Dependence)**

- 현재 시점의 관측값이 과거 시점의 관측값에 영향을 받는 성질
- 일반적으로 가까운 과거의 관측값일수록 현재 값과 강한 관계를 가짐
- 시간 순서 자체가 중요한 정보이므로 데이터를 무작위로 섞어 분석하면 안 됨
- 학습 데이터에 미래 정보가 포함되면 데이터 누수(Data Leakage)가 발생할 수 있음

### **2) 추세(Trend)와 계절성(Seasonality)**

- **추세**: 시간에 따라 데이터가 장기적으로 증가하거나 감소하는 방향성
- **계절성**: 일정한 주기로 유사한 패턴이 반복되는 성질
- 실제 시계열에서는 추세와 계절성이 동시에 나타나는 경우가 많음

### **3) 자기상관성(Autocorrelation)**

: 동일한 시계열에서 현재 값과 과거 값 사이에 존재하는 상관관계

- 자기상관계수는 -1에서 1 사이의 값을 가짐
- 1에 가까움: 과거 값과 현재 값이 강한 양의 상관관계를 가짐
- -1에 가까움: 강한 음의 상관관계를 가짐
- 0에 가까움: 선형적인 상관관계가 약함
- 시차(Lag)를 바꾸면서 자기상관을 계산하면 시간적 의존성과 반복 패턴을 분석할 수 있음

**시차(Lag)란?**
- 현재 값과 비교하는 과거 시점까지의 시간 간격
- 일별 데이터에서 어제와 비교하면 Lag 1, 일주일 전과 비교하면 Lag 7

=> 시계열 분석에서는 관측값 사이의 **시간의존성과 자기상관성**을 고려하는 것이 중요

## 1-3. 시계열 데이터의 분석요인

### **시계열 데이터의 4가지 분석요인**

| 요인 | 정의 | 특징 | 예시 |
| --- | --- | --- | --- |
| 추세(Trend) | 장기적인 증가·감소 경향 | 장기적 방향성 | 인구 증가, 출생아 수 감소 |
| 계절(Seasonal) | 일정한 주기의 반복 변동 | 주기가 비교적 일정 | 여름철 에어컨 판매 증가 |
| 순환(Cycle) | 장기간에 걸친 불규칙한 반복 변동 | 주기가 일정하지 않음 | 경기 호황과 불황 |
| 불규칙(Irregular) | 다른 요인으로 설명되지 않는 변동 | 우연적·비예측적 | 자연재해로 인한 매출 급감 |

- 실제 시계열 데이터는 여러 분석요인이 결합된 형태
- 순환 요인은 추세와 구분하기 어려워 추세-순환 요인으로 함께 다루기도 함
- 일반적인 요소분해에서는 추세, 계절성, 잔차의 세 가지 성분으로 분리

> # **2. 시계열 데이터 전처리**

## 2-1. 결측치 및 이상치 처리

### **1) 결측치(Missing Value)**

: 특정 시점의 관측값이 누락되거나 수집되지 않은 경우

- 일반 데이터와 달리 결측값을 단순 삭제하면 시계열의 관측 간격이 깨질 수 있음
- 평균값으로 일괄 대체하면 시간적 추세와 계절성이 왜곡될 수 있음
- 따라서 앞뒤 시점의 관계를 고려한 보간법(Interpolation)을 주로 활용

**결측치 처리 방법**

| 방법 | 설명 | 특징 |
| --- | --- | --- |
| 삭제 | 결측값이 존재하는 행 제거 | 시간 간격이 깨질 수 있음 |
| 단순 대치 | 평균·중앙값 등으로 대체 | 추세와 계절성 반영 어려움 |
| 선형 보간 | 앞뒤 값을 직선으로 연결 | 변화가 완만한 데이터에 적합 |
| 다항식 보간 | 다항식으로 결측값 추정 | 비선형 데이터에 활용 |
| 스플라인 보간 | 구간별 다항식으로 추정 | 부드러운 곡선 형태에 적합 |
| Forward Fill | 직전 관측값으로 대체 | 계단형 데이터에 적합 |
| Backward Fill | 직후 관측값으로 대체 | 미래 관측값을 활용 |
| 이동평균 대치 | 주변 관측값의 평균으로 대체 | 단기적인 노이즈 완화 |

**보간법(Interpolation)이란?**
- 이미 관측된 데이터 포인트를 활용하여 두 관측값 사이의 누락된 값을 추정하는 방법
- 기존 관측값을 유지하면서 시간 흐름을 반영해 결측값을 보완
- 예측 시점 이후의 값을 사용할 수 없는 상황에서는 미래 데이터 누수에 주의해야 함

### **2) 이상치(Outlier)**

: 시계열 데이터의 일반적인 패턴에서 크게 벗어난 관측값

- 동일한 수치라도 관측 시점에 따라 정상값 또는 이상치가 될 수 있음
- 이상치를 판단할 때 값의 크기뿐 아니라 추세와 계절성을 함께 고려해야 함

**이상치 탐지 방법**

**① IQR(Interquartile Range)**

- 데이터의 제1사분위수(Q1)와 제3사분위수(Q3)를 활용
- 일반적으로 아래 범위를 벗어나면 이상치로 판단

$$
IQR=Q_3-Q_1
$$

$$
X<Q_1-1.5IQR \quad \text{or} \quad X>Q_3+1.5IQR
$$

**② Z-score**

- 관측값이 평균으로부터 몇 표준편차만큼 떨어져 있는지 측정
- 일반적으로 절댓값이 3을 초과하면 이상치 후보로 판단

$$
Z=\frac{X-\mu}{\sigma}
$$

**③ STL 분해**

- 시계열을 추세, 계절성, 잔차로 분해
- 잔차에 IQR 또는 Z-score 등의 기준을 적용하여 이상치 탐지
- 계절성이 뚜렷한 시계열에 적합

**④ 모델 기반 탐지**

- CART: 예측값과 실제값의 오차를 활용
- 클러스터링: 군집에서 크게 벗어난 관측값을 탐지
- 오토인코더: 입력값과 복원값 사이의 오차를 활용

**이상치 처리 방법**

- **삭제**: 이상값을 제거하되, 시간 간격 유지를 위해 결측치로 변환 후 처리 가능
- **단순 대치**: 평균 또는 중앙값 등으로 대체
- **보간**: 이상치를 결측치로 변경한 뒤 주변 관측값으로 추정

=> 이상치가 실제 사건에 의해 발생한 유의미한 변화인지, 측정 오류인지 먼저 확인한 후 처리 방법을 결정해야 함

## 2-2. 평활화(Smoothing)

### **평활화란?**
: 시계열 데이터의 단기적인 불규칙 변동과 노이즈를 줄여 장기적인 흐름을 파악하는 기법

- 불규칙한 변동을 완화하여 추세를 쉽게 파악
- 이동평균(Moving Average)이 대표적인 평활화 기법
- 탐색적 데이터 분석(EDA), 추세 파악, 결측치 대치 등에 활용

### **이동평균(Moving Average)**

: 연속된 일정 구간의 관측값을 평균하고, 구간을 이동하면서 평균을 반복 계산하는 방법

$$
MA_t=\frac{Y_t+Y_{t-1}+\cdots+Y_{t-N+1}}{N}
$$

- $N$: 윈도우 크기(Window Size)
- $N$이 작을수록 최근 변화에 민감하지만 노이즈가 많이 남음
- $N$이 클수록 그래프가 매끄러워지지만 변화에 대한 반응이 느려짐

### **이동평균의 주요 유형**

| 구분 | 특징 | 장점 | 단점 |
| --- | --- | --- | --- |
| SMA(단순 이동평균) | 모든 관측값에 동일 가중치 | 계산이 간단함 | 최근 변화에 둔감 |
| WMA(가중 이동평균) | 최근 값에 더 높은 가중치 | 최근 변화 반영 | 가중치 설정 필요 |
| EMA(지수 이동평균) | 과거로 갈수록 가중치가 지수적으로 감소 | 최근 변화에 빠르게 반응 | 평활화 계수 설정 필요 |

**지수 이동평균(EMA)**

$$
EMA_t=\alpha Y_t+(1-\alpha)EMA_{t-1}
$$

- $\alpha$: 평활화 계수
- $\alpha$가 클수록 최근 관측값의 영향이 커짐

## 2-3. 요소분해(Decomposition)

### **요소분해란?**
: 여러 변동 요인이 결합된 시계열 데이터를 추세(Trend), 계절성(Seasonality), 잔차(Residual)로 분리하는 과정

- 시계열 내부의 패턴을 각각 파악할 수 있음
- 추세 및 계절성의 영향을 분석할 수 있음
- 잔차를 활용한 이상치 탐지와 예측 모델링에 활용

### **1) 가법모형(Additive Model)**

$$
Y_t=T_t+S_t+R_t
$$

- 추세, 계절성, 잔차가 더해지는 구조
- 추세의 크기와 관계없이 계절 변동의 절대적인 크기가 일정할 때 적합
- 각 성분의 영향이 원본 데이터와 같은 단위로 표현됨

### **2) 승법모형(Multiplicative Model)**

$$
Y_t=T_t\times S_t\times R_t
$$

- 추세, 계절성, 잔차가 곱해지는 구조
- 추세가 증가할수록 계절 변동 폭도 비례하여 증가하는 경우 적합
- 계절 효과가 절대적인 크기보다 비율로 나타남

**가법모형과 승법모형 비교**

| 구분 | 가법모형 | 승법모형 |
| --- | --- | --- |
| 결합 방식 | 덧셈 | 곱셈 |
| 계절 변동 크기 | 일정 | 추세에 비례 |
| 계절 요인 | 절대적인 변화량 | 비율 |
| 적용 상황 | 진폭이 일정한 데이터 | 진폭이 증가·감소하는 데이터 |

승법모형에 로그를 적용하면 곱셈 구조를 덧셈 구조로 변환할 수 있음

$$
\log Y_t=\log T_t+\log S_t+\log R_t
$$

### **3) Seasonal Decompose**

: 이동평균을 활용하여 추세를 추정하고 계절성과 잔차를 분리하는 전통적인 요소분해 방법

**작동 과정**
1. 중심 이동평균을 이용해 추세 추정
2. 원본 데이터에서 추세 제거
3. 동일한 계절 위치의 평균을 이용해 계절성 추정
4. 추세와 계절성을 제거한 나머지를 잔차로 계산

- 구현이 간단하고 직관적임
- 고정적인 계절 패턴을 추정하는 데 적합
- 급격한 변화와 이상치에 상대적으로 취약

### **4) STL(Seasonal-Trend Decomposition using LOESS)**

: LOESS(국소 가중 회귀)를 활용하여 추세와 계절성을 유연하게 추정하는 요소분해 기법

**작동 과정**

- **Inner Loop**: 추세와 계절성을 번갈아 추정하면서 반복적으로 개선
- **Outer Loop**: 이상치에 낮은 가중치를 부여하여 추정 결과의 왜곡을 완화
- 시간에 따라 변화하는 추세와 계절 패턴을 유연하게 반영 가능
- Robust 옵션을 활용하면 이상치의 영향을 줄일 수 있음

**Seasonal Decompose와 STL 비교**

| 구분 | Seasonal Decompose | STL |
| --- | --- | --- |
| 추정 방법 | 이동평균 | LOESS |
| 추세 추정 | 상대적으로 단순 | 유연한 곡선 추정 |
| 계절성 | 비교적 고정된 패턴 | 시간에 따른 변화 반영 가능 |
| 이상치 대응 | 상대적으로 취약 | Robust 설정 가능 |
| 특징 | 간단하고 직관적 | 유연하고 정교함 |

> # **3. 정상성 판단**

## 3-1. 정상성의 정의

### **정상성(Stationarity)이란?**
: 시계열의 통계적 특성이 시간에 따라 변하지 않는 성질

- 평균, 분산, 자기공분산 구조 등이 안정적으로 유지됨
- 과거의 패턴이 미래에도 유지될 것이라는 시계열 예측의 기본 가정과 관련
- AR, MA, ARMA 등 전통적인 시계열 모형에서는 정상성이 중요한 전제조건

### **1) 강한 정상성(Strict Stationarity)**

: 시계열의 결합확률분포가 시간 이동에 관계없이 동일하게 유지되는 성질

$$
(Y_{t_1},\dots,Y_{t_k})
\overset{d}{=}
(Y_{t_1+h},\dots,Y_{t_k+h})
$$

- 모든 시점의 결합확률분포가 시간 이동에 대해 동일
- 평균, 분산뿐 아니라 왜도, 첨도 등 분포의 전체 특성이 유지됨
- 현실 데이터에서는 이를 직접 검증하기 어려움

### **2) 약한 정상성(Weak Stationarity)**

: 평균과 분산이 일정하고 자기공분산이 시차에만 의존하는 성질

**약한 정상성의 3가지 조건**

1. 평균이 시간에 관계없이 일정함
2. 분산이 시간에 관계없이 일정하고 유한함
3. 자기공분산이 특정 시점이 아닌 시차에만 의존함

| 구분 | 강한 정상성 | 약한 정상성 |
| --- | --- | --- |
| 기준 | 결합확률분포 전체 | 평균, 분산, 자기공분산 |
| 조건 | 상대적으로 엄격 | 상대적으로 완화 |
| 특징 | 분포 전체의 시간적 불변성 | 주요 통계적 특성의 안정성 |
| 활용 | 이론적 정의에 중요 | 일반적인 통계적 시계열 모델링 |

## 3-2. 정상성 판단 방법

### **1) 시각적 판단**

**시계열 그래프 활용**

- 일정한 평균을 중심으로 변동하는지 확인
- 시간이 지날수록 변동 폭이 증가하거나 감소하는지 확인
- 뚜렷한 추세나 계절성이 존재하는지 확인

**ACF 그래프 활용**

- 정상적인 ARMA 계열 시계열은 일반적으로 자기상관이 시차에 따라 감소
- 단위근이 존재하는 비정상 시계열은 자기상관이 매우 천천히 감소하는 경향

### **2) 통계적 판단**

**① ADF(Augmented Dickey-Fuller) 검정**

: 시계열에 단위근(Unit Root)이 존재하는지 검정하는 방법

- **귀무가설(H0)**: 단위근이 존재한다 → 비정상
- **대립가설(H1)**: 단위근이 존재하지 않는다
- p-value < 0.05이면 귀무가설 기각 → 단위근이 없다는 통계적 근거

**② KPSS 검정**

: 시계열의 정상성을 귀무가설로 설정하여 검정하는 방법

- **귀무가설(H0)**: 시계열이 정상성을 가진다
- **대립가설(H1)**: 시계열이 정상성을 가지지 않는다
- p-value < 0.05이면 귀무가설 기각 → 비정상이라는 통계적 근거

**ADF와 KPSS 비교**

| 구분 | ADF | KPSS |
| --- | --- | --- |
| 검정 대상 | 단위근 존재 여부 | 정상성 |
| 귀무가설 | 단위근 존재 | 정상성 |
| p-value < 0.05 | 단위근 가설 기각 | 정상성 가설 기각 |
| 활용 | 단위근에 의한 비정상성 판단 | 정상성 판단 보완 |

- 두 검정은 귀무가설이 반대이므로 함께 활용 가능
- 검정 결과가 불일치하면 추세 포함 여부, 검정 설정 및 데이터 특성을 추가 확인

## 3-3. 정상성 확보 방법

### **1) 차분(Differencing)**

: 현재 값에서 직전 시점의 값을 빼서 추세 또는 단위근에 의한 비정상성을 완화하는 방법

$$
\Delta Y_t=Y_t-Y_{t-1}
$$

- 1차 차분: 현재 값에서 직전 값을 뺌
- 2차 차분: 1차 차분 결과를 다시 차분
- 단위근에 의한 비정상 시계열을 정상화하는 데 활용
- 과도하게 차분하면 정보 손실 및 불필요한 변동이 발생할 수 있음

### **2) 계절 차분(Seasonal Differencing)**

: 현재 값에서 한 계절 주기 이전의 값을 빼서 계절 효과를 제거하는 방법

$$
\Delta_mY_t=Y_t-Y_{t-m}
$$

- $m$: 계절 주기
- 월별 데이터에서 연간 계절성이 존재하면 $m=12$
- 반복적인 계절 패턴으로 인한 비정상성을 완화

### **3) 로그 변환(Log Transformation)**

: 데이터에 로그를 적용하여 값이 커질수록 변동 폭이 증가하는 현상을 완화하는 방법

$$
Y'_t=\log(Y_t)
$$

- 큰 값의 범위를 압축하여 분산 안정화
- 승법적 변동 구조를 가법적 구조로 변환
- 양수 데이터에 적용 가능
- 로그 변환만으로 모든 비정상성이 해결되는 것은 아님

### **4) 추세 제거(Detrending)**

- 시계열의 장기적 증가·감소 추세를 추정한 뒤 원본 데이터에서 제거
- 결정적 추세가 존재하는 시계열의 정상성 확보에 활용

**비정상성 유형별 처리 방법**

| 비정상성 원인 | 처리 방법 |
| --- | --- |
| 추세 및 단위근 | 차분 |
| 결정적 추세 | 추세 제거 |
| 계절성 | 계절 차분, 계절성 제거 |
| 분산 변화 | 로그 변환, Box-Cox 변환 |

=> 정상성 확보 후에는 시계열 그래프와 ADF, KPSS 검정 등을 통해 다시 확인해야 함

> # **4. 단변량 시계열 분석**

## 4-1. 안정적 시계열 분석

### **안정적 시계열 분석이란?**

- 평균, 분산, 자기공분산 구조가 안정적인 정상 시계열을 분석하는 방법
- 자기 자신의 과거 값과 과거 오차를 활용해 현재 및 미래 값을 설명
- 대표적인 모델로 AR, MA, ARMA가 있음

### **1) AR(Autoregressive Model)**

: 현재 값을 자기 자신의 과거 관측값으로 설명하는 자기회귀 모형

$$
Y_t=c+\phi_1Y_{t-1}+\cdots+\phi_pY_{t-p}+\varepsilon_t
$$

- $p$: 자기회귀 차수
- $\phi_i$: 각 시차의 자기회귀계수
- $\varepsilon_t$: 현재 시점의 충격 또는 오차
- 과거 관측값이 현재 관측값에 미치는 영향을 반영
- 정상성 조건을 만족하면 충격의 영향이 시간이 지나면서 감소

**AR(1) 모델**

$$
Y_t=c+\phi_1Y_{t-1}+\varepsilon_t
$$

- 직전 시점의 관측값 하나만 활용
- 정상성 조건: $|\phi_1|<1$

### **2) MA(Moving Average Model)**

: 현재 값을 현재 및 과거의 오차(충격)로 설명하는 이동평균 모형

$$
Y_t=\mu+\varepsilon_t+\theta_1\varepsilon_{t-1}+\cdots+\theta_q\varepsilon_{t-q}
$$

- $q$: 이동평균 차수
- $\theta_i$: 이동평균계수
- $\varepsilon_t$: 현재 시점의 충격
- 과거 충격이 현재 관측값에 미치는 영향을 반영
- 특정 충격의 영향은 유한한 q시점 이후 사라짐

**MA(1) 모델**

$$
Y_t=\mu+\varepsilon_t+\theta_1\varepsilon_{t-1}
$$

- 현재 충격과 직전 시점의 충격을 활용
- 가역성 조건: $|\theta_1|<1$

**AR과 MA 비교**

| 구분 | AR | MA |
| --- | --- | --- |
| 설명 변수 | 과거 관측값 | 과거 오차 |
| 활용 정보 | 자기상관성 | 과거 충격 |
| 충격의 영향 | 점진적으로 감소 | 일정 시차 이후 사라짐 |
| 차수 | p | q |

※ 평활화에서의 이동평균은 관측값을 평균하는 기법이고, MA 시계열 모형은 과거 오차를 이용한다는 차이가 있음

### **3) AR, MA의 전제조건**

**① 정상성(Stationarity)**

- 충격의 영향이 지속적으로 증가하지 않고 통계적 특성이 안정적으로 유지되는 성질
- AR 모형은 정상성 유지를 위한 계수 조건이 필요
- 유한 차수 MA 모형은 백색잡음 오차를 가정하면 정상성을 만족

**② 가역성(Invertibility)**

- 관측값으로부터 과거 충격을 안정적으로 복원할 수 있는 성질
- MA 모형에서 과거 오차를 추정하기 위해 중요
- MA(1)의 가역성 조건은 $|\theta_1|<1$

**AR(p)의 정상성 조건**

$$
1-\phi_1z-\phi_2z^2-\cdots-\phi_pz^p=0
$$

- 특성방정식의 모든 근이 단위원 밖에 존재해야 함

**MA(q)의 가역성 조건**

$$
1+\theta_1z+\theta_2z^2+\cdots+\theta_qz^q=0
$$

- 특성방정식의 모든 근이 단위원 밖에 존재해야 함

### **4) ARMA(Autoregressive Moving Average)**

: AR과 MA를 결합하여 과거 관측값과 과거 오차를 동시에 활용하는 모형

$$
Y_t=c+\sum_{i=1}^{p}\phi_iY_{t-i}
+\varepsilon_t+\sum_{j=1}^{q}\theta_j\varepsilon_{t-j}
$$

- AR(p): 과거 관측값 반영
- MA(q): 과거 오차 반영
- 정상적인 시계열을 분석하는 데 활용
- AR 부분의 정상성과 MA 부분의 가역성을 고려해야 함

**ARMA의 특수한 경우**

| 모델 | 의미 |
| --- | --- |
| ARMA(p,0) | AR(p) |
| ARMA(0,q) | MA(q) |
| ARMA(0,0) | 상수와 백색잡음 |

### **5) 차수 결정 방법: ACF와 PACF**

**ACF(Autocorrelation Function, 자기상관함수)**

: 현재 값과 k시점 이전 값 사이의 자기상관을 측정하는 함수

$$
\rho(k)=\frac{Cov(Y_t,Y_{t-k})}{Var(Y_t)}
$$

- 직접적인 영향과 중간 시점을 거치는 간접적인 영향을 모두 포함
- 시차별 자기상관을 계산하여 반복 패턴을 파악
- 주로 MA 차수 q를 추정하는 데 활용

**PACF(Partial Autocorrelation Function, 편자기상관함수)**

: 중간 시점의 영향을 제거하고 현재 값과 k시점 이전 값의 직접적인 상관관계를 측정하는 함수

- 중간 시점의 간접적 영향을 제거
- 주로 AR 차수 p를 추정하는 데 활용

**ACF, PACF를 활용한 모형 식별**

| 모형 | ACF | PACF |
| --- | --- | --- |
| AR(p) | 점진적 감소 | p시차 이후 절단 |
| MA(q) | q시차 이후 절단 | 점진적 감소 |
| ARMA(p,q) | 일반적으로 점진적 감소 | 일반적으로 점진적 감소 |

- **절단(Cut-off)**: 특정 시차 이후 자기상관이 통계적으로 유의하지 않게 되는 형태
- **점진적 감소(Tailing-off)**: 시차가 증가할수록 자기상관이 점차 줄어드는 형태
- 실제 데이터에서는 그래프가 명확하지 않을 수 있으므로 AIC, BIC 등의 정보 기준을 함께 활용

## 4-2. 불안정 시계열 분석

### **불안정 시계열이란?**

- 추세, 계절성 등으로 인해 정상성을 만족하지 않는 시계열
- 정상성을 확보하기 위한 차분 과정이 필요할 수 있음
- 대표적인 모형으로 ARIMA, SARIMA가 있음

### **1) ARIMA(Autoregressive Integrated Moving Average)**

: ARMA 모형에 차분(Integrated) 과정을 추가한 모형

$$
ARIMA(p,d,q)
$$

- $p$: AR 차수
- $d$: 차분 횟수
- $q$: MA 차수

**ARIMA 분석 과정**

1. 시계열을 차분하여 정상성을 확보
2. 차분된 데이터에 ARMA 모형 적용
3. 미래의 차분값 예측
4. 역차분을 통해 원래 단위의 예측값 복원

**ARIMA(1,1,1) 예시**

$$
W_t=Y_t-Y_{t-1}
$$

$$
W_t=c+\phi_1W_{t-1}+\varepsilon_t+\theta_1\varepsilon_{t-1}
$$

- 원본 시계열을 1차 차분
- 차분된 시계열에 AR(1)과 MA(1)을 적용

**ARIMA의 특수한 경우**

| 모형 | 의미 |
| --- | --- |
| ARIMA(p,0,0) | AR(p) |
| ARIMA(0,0,q) | MA(q) |
| ARIMA(p,0,q) | ARMA(p,q) |
| ARIMA(0,1,0) | 상수가 없는 경우 랜덤워크 |

### **2) SARIMA(Seasonal ARIMA)**

: ARIMA에 계절 차분과 계절 AR, MA 구조를 추가한 모형

$$
SARIMA(p,d,q)(P,D,Q)_m
$$

| 파라미터 | 의미 |
| --- | --- |
| p | 비계절 AR 차수 |
| d | 비계절 차분 횟수 |
| q | 비계절 MA 차수 |
| P | 계절 AR 차수 |
| D | 계절 차분 횟수 |
| Q | 계절 MA 차수 |
| m | 계절 주기 |

- 월별 데이터의 연간 계절성: $m=12$
- 분기별 데이터의 연간 계절성: $m=4$
- 일별 데이터의 주간 계절성: $m=7$

**SARIMA 분석 과정**

1. 시계열의 추세와 계절성 확인
2. 필요한 비계절 차분(d)과 계절 차분(D) 수행
3. ADF 등 정상성 검정으로 결과 확인
4. ACF, PACF를 활용해 차수 후보 추정
5. AIC, BIC 등을 활용해 최적 모델 선택
6. 잔차 진단 및 미래 값 예측

**ARIMA와 SARIMA 비교**

| 구분 | ARIMA | SARIMA |
| --- | --- | --- |
| 분석 대상 | 비계절적 비정상 시계열 | 계절성이 존재하는 시계열 |
| 차분 | 비계절 차분 | 비계절 차분 + 계절 차분 |
| 주요 파라미터 | p, d, q | p, d, q, P, D, Q, m |
| 특징 | 추세 및 자기상관 구조 반영 | 추세, 자기상관 및 계절성 반영 |

> # **5. 다변량 시계열 분석**

## 5-1. VAR(Vector Autoregression)

### **VAR이란?**

: 여러 시계열 변수를 각각 자기 자신의 과거 값과 다른 변수의 과거 값으로 설명하는 벡터 자기회귀 모형

- 단변량 AR 모형을 여러 변수로 확장한 형태
- 변수 간 시간적 상호작용을 동시에 고려
- 경제·금융 데이터 등 여러 변수가 상호 영향을 주는 상황에 활용

### **VAR(1) 모델**

두 변수 $X_t$, $Y_t$를 사용하는 VAR(1) 모형은 다음과 같음

$$
X_t=a_{11}X_{t-1}+a_{12}Y_{t-1}+\varepsilon_{X,t}
$$

$$
Y_t=a_{21}X_{t-1}+a_{22}Y_{t-1}+\varepsilon_{Y,t}
$$

- $a_{11},a_{22}$: 자기 자신의 과거 값이 현재 값에 미치는 영향
- $a_{12},a_{21}$: 다른 변수의 과거 값이 현재 값에 미치는 영향
- $\varepsilon_t$: 각 변수의 충격 또는 오차

**VAR 모델의 주요 특징**

- 여러 변수의 상호작용을 분석할 수 있음
- 변수 간 동태적인 관계와 충격의 파급효과 분석에 활용
- 기본적으로 정상성을 만족하는 변수들을 대상으로 모델링
- 변수 수가 많아질수록 추정해야 하는 계수가 빠르게 증가

### **VAR 모델의 주요 분석 방법**

**1) Granger Causality Test(그레인저 인과관계 검정)**

: 한 변수의 과거 정보가 다른 변수의 미래 값을 예측하는 데 도움이 되는지 검정하는 방법

- 귀무가설(H0): X의 과거 정보는 Y의 예측에 도움이 되지 않음
- 대립가설(H1): X의 과거 정보는 Y의 예측에 도움이 됨
- p-value < 0.05이면 귀무가설 기각

※ 그레인저 인과관계는 통계적 예측 가능성을 의미하며 실제 인과관계를 증명하는 것은 아님

**2) IRF(Impulse Response Function, 충격반응함수)**

: 한 변수에 충격이 발생했을 때 다른 변수가 시간에 따라 어떻게 반응하는지 분석하는 방법

- 충격의 영향 방향과 지속 기간을 분석
- 변수 간 동태적 상호작용을 확인
- 경제적 충격의 파급효과 분석에 활용

**3) FEVD(Forecast Error Variance Decomposition, 예측오차 분산분해)**

: 특정 변수의 예측오차가 자기 자신의 충격과 다른 변수의 충격에 의해 얼마나 설명되는지 분석하는 방법

- 예측오차에 대한 변수별 기여도를 분석
- 변수 간 상대적 영향력을 파악
- IRF와 함께 VAR 모형의 주요 해석 도구로 활용

**VAR 모델의 주요 분석 방법 비교**

| 구분 | 목적 | 핵심 질문 |
| --- | --- | --- |
| Granger Causality | 변수 간 예측 관계 확인 | X의 과거가 Y의 예측에 도움이 되는가? |
| IRF | 충격의 시간적 파급효과 분석 | X에 충격이 발생하면 Y는 어떻게 반응하는가? |
| FEVD | 예측오차의 원인별 기여도 분석 | Y의 예측오차 중 X의 충격이 차지하는 비중은 얼마인가? |

> # **6. 전체 내용 정리**

## 6-1. 시계열 분석의 기본 과정

1. **시계열 데이터 이해**: 시간의존성, 추세, 계절성, 자기상관성 파악
2. **데이터 전처리**: 결측치·이상치 처리, 평활화, 요소분해
3. **정상성 판단**: 시각적 분석 및 ADF, KPSS 검정
4. **정상성 확보**: 차분, 계절 차분, 로그 변환, 추세 제거
5. **모델 선택 및 학습**: 데이터 특성에 맞는 AR, MA, ARMA, ARIMA, SARIMA, VAR 적용
6. **예측 및 평가**: 잔차 진단과 예측 결과 평가

## 6-2. 핵심 모델 정리

| 모델 | 주요 특징 | 적용 대상 |
| --- | --- | --- |
| AR | 과거 관측값을 활용 | 정상 단변량 시계열 |
| MA | 과거 오차를 활용 | 정상 단변량 시계열 |
| ARMA | AR과 MA 결합 | 정상 단변량 시계열 |
| ARIMA | 차분과 ARMA 결합 | 비계절적 비정상 시계열 |
| SARIMA | ARIMA에 계절성 반영 | 계절성이 존재하는 시계열 |
| VAR | 여러 변수의 과거 값을 함께 활용 | 다변량 시계열 |

## 6-3. 핵심 개념 정리

- **시계열 데이터**: 시간 순서에 따라 관측된 데이터
- **시간의존성**: 현재 값이 과거 값의 영향을 받는 성질
- **자기상관성**: 동일한 변수의 현재와 과거 값 사이의 상관관계
- **추세**: 시계열의 장기적인 증가·감소 방향
- **계절성**: 일정한 주기로 반복되는 패턴
- **평활화**: 노이즈를 완화하여 전체적인 흐름을 파악하는 기법
- **요소분해**: 추세, 계절성, 잔차를 분리하는 과정
- **정상성**: 시계열의 통계적 특성이 시간에 따라 유지되는 성질
- **차분**: 인접한 시점 간 차이를 계산해 비정상성을 완화하는 방법
- **ACF**: 현재 값과 과거 값의 자기상관을 나타내는 함수
- **PACF**: 중간 시점의 영향을 제거한 편자기상관을 나타내는 함수
- **ARIMA**: 차분을 통해 비정상 시계열을 분석하는 모형
- **SARIMA**: 계절성을 고려한 ARIMA 모형
- **VAR**: 여러 변수 간 시간적 상호작용을 고려하는 다변량 모형
