
> # **1. AI 에이전트란?**

## 1-1. AI 에이전트의 정의

### **AI 에이전트(AI Agent)란?**
: 사용자가 설정한 목표를 달성하기 위해 AI 모델을 활용하여 추론하고, 계획을 수립하며, 외부 환경과 상호작용하는 시스템

- 일반적인 LLM은 사용자의 입력에 대해 텍스트를 생성하는 역할을 수행
- AI 에이전트는 LLM의 추론 능력에 외부 도구와 실행 기능을 결합
- 목표 달성을 위해 필요한 작업을 판단하고 실행 결과에 따라 다음 행동을 결정
- 검색, 데이터 조회, 코드 실행, 이메일 발송 등 다양한 외부 도구를 활용할 수 있음

### **AI 에이전트의 주요 구성요소**

| 구성요소 | 설명 |
| --- | --- |
| 추론(Reasoning) | 사용자의 요청과 현재 상황을 분석 |
| 계획(Planning) | 목표 달성에 필요한 작업과 수행 순서를 결정 |
| 도구(Tools) | 외부 데이터 조회 및 실제 작업 수행에 필요한 기능 |
| 행동(Action) | 선택한 도구를 활용하여 환경과 상호작용 |
| 관찰(Observation) | 실행 결과를 확인하고 다음 행동에 반영 |

### **생성형 AI(Generative AI)와 에이전틱 AI(Agentic AI)의 차이**

| 구분 | 생성형 AI | 에이전틱 AI |
| --- | --- | --- |
| 주요 목적 | 콘텐츠 및 답변 생성 | 목표 달성 및 업무 수행 |
| 작동 방식 | 입력에 대한 결과물 생성 | 추론과 행동의 반복 |
| 외부 환경과의 상호작용 | 필수적이지 않음 | 외부 도구를 적극적으로 활용 |
| 실행 결과 반영 | 추가 요청에 따라 반영 | 실행 결과를 바탕으로 다음 행동 결정 |
| 예시 | 이메일 초안 작성 | 이메일 작성, 발송 및 결과 확인 |

- 생성형 AI와 에이전틱 AI는 완전히 독립된 기술이 아님
- 에이전틱 AI는 생성형 AI의 추론 및 생성 능력을 기반으로 실행 기능을 확장한 시스템

=> AI 에이전트의 핵심은 단순한 답변 생성을 넘어 **목표 달성을 위해 환경과 상호작용하고 행동한다는 것**

## 1-2. 에이전틱 AI의 작동 방식: ReAct

### **ReAct(Reasoning + Acting)란?**
: AI 모델의 추론(Reasoning)과 행동(Acting)을 결합하여, 행동의 결과를 바탕으로 다음 추론을 수행하는 방식

**ReAct의 기본 작동 과정**

1. **생각(Thought)**: 현재 상황을 분석하고 다음에 수행할 행동을 결정
2. **행동(Action)**: 필요한 외부 도구를 호출하여 작업 수행
3. **관찰(Observation)**: 도구의 실행 결과 및 환경의 변화 확인
4. 관찰 결과를 바탕으로 다시 추론하고 다음 행동을 결정
5. 목표를 달성하거나 설정된 종료 조건에 도달하면 작업 종료

$$
\text{Thought} \rightarrow \text{Action} \rightarrow \text{Observation} \rightarrow \text{Thought} \rightarrow \cdots
$$

### **ReAct의 활용 예시: 고객 서비스 에이전트**

사용자 요청: "주문 #1234의 배송이 늦어요."

| 단계 | 생각(Thought) | 행동(Action) | 관찰(Observation) |
| --- | --- | --- | --- |
| 1 | 주문 상태 확인 필요 | 주문 조회 | 배송이 4일 지연됨 |
| 2 | 배송 지연 원인 확인 필요 | 배송 추적 | 배송사 서버 응답 오류 |
| 3 | 일시적 오류일 가능성 판단 | 배송 추적 재시도 | 물류센터 적체 확인 |
| 4 | 배송비 환불 대상인지 판단 | 환불안 및 사과 메시지 작성 | 환불안 작성 완료 |
| 5 | 금전적 거래에 대한 승인 필요 | 담당자에게 승인 요청 | 담당자 승인 |
| 6 | 승인된 작업 실행 | 환불 및 메시지 발송 | 작업 완료 |

- 이전 행동의 결과에 따라 다음 행동이 달라질 수 있음
- 도구 호출에 실패하면 재시도하거나 다른 해결 방법을 탐색
- 모든 작업이 성공하는 것은 아니므로 종료 조건을 설정해야 함

**에이전트의 실행을 통제하기 위한 요소**

- 최대 반복 횟수 설정
- 도구 호출 및 모델 사용에 대한 비용 한도 설정
- 동일한 작업의 중복 실행 방지
- 중요한 행동을 수행하기 전 사람의 승인 요청

**에이전트의 상태 및 메모리 관리**

| 구분 | 설명 | 예시 |
| --- | --- | --- |
| 단기 상태(Short-term State) | 현재 작업의 진행 상황 | 주문번호, 환불 승인 여부 |
| 장기 메모리(Long-term Memory) | 여러 세션에 걸쳐 유지되는 정보 | 사용자의 선호 언어 |
| 컨텍스트(Context) | 현재 모델 호출에 실제로 전달된 정보 | 주문 상태, 환불 규정 |

- 장기 메모리에 저장된 정보가 항상 모델에 전달되는 것은 아님
- 실제 도구의 실행 결과를 확인하고 작업 상태를 갱신해야 함

=> ReAct의 핵심은 정해진 순서대로 작업을 수행하는 것이 아니라 **관찰 결과에 따라 다음 행동을 유연하게 변경하는 것**

## 1-3. 다중 에이전트(Multi-Agent System)

### **다중 에이전트 시스템(MAS)이란?**
: 여러 AI 에이전트가 각자의 역할을 분담하고 협력하여 하나의 목표를 달성하는 시스템

- 복잡한 작업을 여러 개의 하위 작업으로 분해
- 각 에이전트는 자신의 역할에 필요한 작업을 독립적으로 수행
- 대표적인 구조는 오케스트레이터(Orchestrator)와 서브 에이전트(Sub-agent)의 결합

### **1) 오케스트레이터(Orchestrator)**

: 전체 작업을 계획하고 서브 에이전트의 실행을 관리하는 에이전트

- 사용자의 요청을 분석하고 세부 작업으로 분해
- 각 작업에 적합한 서브 에이전트를 선택
- 서브 에이전트의 작업 결과를 수집하고 통합
- 최종 결과물 생성 및 전체 작업 관리

### **2) 서브 에이전트(Sub-agent)**

: 오케스트레이터로부터 할당받은 특정 작업을 수행하는 에이전트

- 검색, 분석, 코드 작성, 결과 검증 등 특정 역할을 담당
- 각자의 독립적인 컨텍스트를 활용하여 작업 수행
- 작업 결과를 오케스트레이터에 전달

**다중 에이전트의 기본 구조**

```text
                    사용자 요청
                         ↓
                  오케스트레이터
                  ↙      ↓      ↘
            검색 담당  분석 담당  검증 담당
             에이전트   에이전트   에이전트
                  ↘      ↓      ↙
                    결과 통합
                         ↓
                     최종 답변
```

### **다중 에이전트의 장점과 한계**

| 장점 | 한계 |
| --- | --- |
| 복잡한 작업의 분할 및 병렬 처리 가능 | 에이전트 간 작업 조율 필요 |
| 역할별 전문화 가능 | 모델 호출 및 토큰 사용량 증가 |
| 독립적인 결과 검증 가능 | 시스템 구조의 복잡성 증가 |
| 개별 에이전트의 컨텍스트 부담 감소 | 오케스트레이터의 판단 오류가 전체 작업에 영향을 줄 수 있음 |

### **작업에 따른 에이전트 구조 선택**

| 구분 | 적합한 작업 | 예시 |
| --- | --- | --- |
| 고정 워크플로우 | 수행 순서가 명확한 반복 작업 | 매일 신규 주문 정리 및 메일 발송 |
| 단일 에이전트 | 소수의 도구로 해결 가능한 작업 | 고객의 배송 지연 문의 처리 |
| 다중 에이전트 | 여러 하위 작업으로 분할 가능한 복잡한 작업 | 다수의 기업 및 산업에 대한 리서치 |

**실제 활용 사례: Anthropic의 다중 에이전트 리서치 시스템**

- 오케스트레이터가 전체 조사 계획을 수립하고 여러 서브 에이전트에 조사 업무를 분배
- 각 에이전트가 독립적으로 조사한 결과를 통합하여 최종 답변 생성
- 첨부 세션 자료에 따르면 내부 평가에서 단일 에이전트 대비 90.2% 높은 성과를 기록
- 일반적인 채팅보다 약 15배 많은 토큰을 사용하는 등 비용 증가가 발생

=> 다중 에이전트는 모든 작업에 적합한 것이 아니며, **작업의 복잡성과 성능 개선 효과, 실행 비용을 함께 고려하여 선택해야 함**

> # **2. MCP(Model Context Protocol)**

## 2-1. MCP의 정의

### **MCP(Model Context Protocol)란?**
: LLM 기반 애플리케이션이 외부 데이터 및 도구와 통신할 수 있도록 연결 방식을 표준화한 개방형 프로토콜

- 2024년 11월 Anthropic이 오픈소스로 공개
- AI 애플리케이션과 외부 서비스 사이의 공통 연결 규격을 제공
- 외부 도구의 탐색, 호출 및 결과 전달 방식을 표준화
- 서로 다른 AI 애플리케이션에서 동일한 MCP 서버를 재사용할 수 있음

**MCP의 주요 특징**

- 외부 도구와 데이터를 공통된 방식으로 연결
- 서버가 제공하는 도구의 목록과 사용 방법을 동적으로 확인 가능
- 인증이 필요한 원격 서버에서 표준화된 인증 방식 활용 가능

=> MCP는 다양한 AI 애플리케이션과 외부 도구를 연결하는 **공통 인터페이스**의 역할을 수행

## 2-2. 도구 호출의 원리: Function Calling

### **Function Calling(함수 호출)이란?**
: LLM이 사용자의 요청을 처리하기 위해 필요한 함수와 입력값을 선택하고, 구조화된 호출 요청을 생성하는 기능

- LLM 자체가 외부 함수를 직접 실행하는 것은 아님
- 모델은 어떤 함수를 어떤 인자로 호출할지 결정
- 실제 함수 실행 및 결과 전달은 애플리케이션이 담당

**Function Calling의 작동 과정**

1. 애플리케이션이 사용자의 질문과 사용 가능한 함수 목록을 모델에 전달
2. 모델이 적절한 함수와 입력값을 선택
3. 모델이 구조화된 함수 호출 요청을 생성
4. 애플리케이션이 실제 함수를 실행
5. 실행 결과를 모델에 전달
6. 모델이 실행 결과를 바탕으로 최종 답변을 생성하거나 다음 행동을 결정

**함수 호출 요청 예시**

사용자 요청: "서울의 날씨를 알려줘."

```json
{
    "name": "get_weather",
    "arguments": {
        "city": "서울"
    }
}
```

- `name`: 호출할 함수의 이름
- `arguments`: 함수 실행에 필요한 입력값
- 실제 날씨 데이터 조회는 애플리케이션이 해당 함수를 실행하여 수행

=> Function Calling은 모델이 외부 도구를 **직접 실행하는 기능이 아니라, 실행에 필요한 요청을 생성하는 기능**

## 2-3. MCP가 필요한 이유

### **n×m 연동 문제**

- MCP와 같은 공통 표준이 없으면 각 애플리케이션과 외부 도구 사이의 연동 코드를 개별적으로 개발해야 함
- 애플리케이션의 개수가 증가하거나 새로운 도구가 추가될수록 개발 및 유지보수 비용이 증가

애플리케이션의 개수를 $n$, 외부 도구의 개수를 $m$이라고 가정하면,

**MCP 도입 이전의 연동 작업량**

$$
C_{\text{Before}} = n \times m
$$

**MCP 도입 이후의 연동 작업량**

$$
C_{\text{After}} = n + m
$$

- $n$: MCP를 지원하도록 구현해야 하는 애플리케이션의 수
- $m$: MCP 서버로 구현해야 하는 외부 도구의 수
- MCP를 활용하면 모든 애플리케이션과 도구의 조합에 대해 개별적인 연동 코드를 작성할 필요가 줄어듦

**애플리케이션 3개, 외부 도구 4개의 경우**

| 구분 | 연동 방식 | 필요한 연동 작업 수 |
| --- | --- | --- |
| MCP 도입 이전 | 앱과 도구를 개별적으로 연동 | 3 × 4 = 12 |
| MCP 도입 이후 | 공통 MCP 규격을 활용 | 3 + 4 = 7 |

- 위의 수식은 연동 구조의 단순화 효과를 나타내는 개념적 모델
- 실제 업무 로직, 인증 및 외부 API 구현에 필요한 작업까지 사라지는 것은 아님

### **Function Calling과 MCP의 차이**

| 구분 | Function Calling | MCP |
| --- | --- | --- |
| 주요 위치 | 모델 API | 애플리케이션과 외부 서비스 사이 |
| 주요 역할 | 모델의 도구 호출 요청 생성 | 외부 도구의 연결 및 호출 표준화 |
| 핵심 기능 | 함수 선택 및 입력값 결정 | 도구 탐색, 통신 및 결과 전달 |
| 해결하는 문제 | 모델이 어떤 도구를 어떻게 호출할 것인가 | 다양한 도구를 어떻게 공통된 방식으로 연결할 것인가 |

- Function Calling과 MCP는 경쟁 관계가 아닌 상호보완적인 관계
- MCP를 통해 도구 목록을 가져오고, 모델이 Function Calling으로 필요한 도구를 선택한 후 MCP를 통해 실제 호출을 수행

=> MCP는 개별적인 연동 구조를 표준화하여 연동 복잡도를 **$n \times m$에서 $n + m$으로 감소시키는 효과**를 가짐

## 2-4. MCP의 구조

### **MCP의 3가지 구성요소: Host / Client / Server**

| 구성요소 | 정의 | 주요 역할 | 예시 |
| --- | --- | --- | --- |
| Host | 사용자가 이용하는 AI 애플리케이션 | LLM 및 Client 관리, 도구 사용 권한 결정 | Claude Desktop, Cursor |
| Client | Host 내부의 통신 담당 구성요소 | MCP Server와 메시지 송수신 | Notion MCP Client |
| Server | 외부 기능과 데이터를 제공하는 프로그램 | 도구 목록 제공, 요청 처리 및 실제 작업 수행 | Notion MCP Server, GitHub MCP Server |

**MCP의 기본 연결 구조**

```text
                         사용자
                            ↓
                          Host
                    ┌────────────┐
                    │    LLM     │
                    └────────────┘
                     ↙          ↘
                MCP Client    MCP Client
                     ↕          ↕
                MCP Server    MCP Server
                     ↕          ↕
                   Notion      GitHub
```

- 하나의 Host는 여러 MCP Client를 관리할 수 있음
- 각 MCP Client는 개별 MCP Server와 연결되어 통신을 담당
- MCP Server는 외부 API, 데이터베이스 또는 파일 시스템 등에 접근하여 작업을 수행

### **MCP의 작동 과정**

사용자 요청: "이번 주 노션 회의록을 요약해 줘."

| 단계 | 수행 주체 | 수행 내용 |
| --- | --- | --- |
| 1 | Host | 사용자 인증 및 서버 연결 |
| 2 | Client | 서버에 사용 가능한 도구 목록 요청(`tools/list`) |
| 3 | Server | 도구의 이름, 설명, 입력 형식 반환 |
| 4 | Host | 사용자의 질문과 도구 목록을 모델에 전달 |
| 5 | LLM | 회의록 검색에 필요한 도구 호출 요청 생성 |
| 6 | Client | 서버에 도구 호출 요청 전달(`tools/call`) |
| 7 | Server | 실제 회의록 검색 수행 |
| 8 | Host | 검색 결과를 모델에 전달 |
| 9 | LLM | 회의록 요약 및 최종 답변 생성 |

=> 모델은 도구의 선택과 답변 생성을 담당하며, **연결과 실제 실행은 Host, Client, Server가 담당**

## 2-5. MCP 서버의 기본 요소

### **MCP 서버의 3가지 기본 요소(Primitives)**

| 요소 | 정의 | 주요 역할 | 예시 |
| --- | --- | --- | --- |
| Tools | 실행 가능한 기능 | 외부 작업 수행 | 노션 페이지 생성, GitHub 이슈 등록 |
| Resources | 참조 가능한 데이터 | 모델에 필요한 정보 제공 | 문서, 파일, DB 스키마 |
| Prompts | 재사용 가능한 프롬프트 템플릿 | 정형화된 작업 방식 제공 | 코드 리뷰 양식, 회의록 요약 양식 |

### **1) Tools**

- AI 에이전트가 수행할 수 있는 행동을 정의
- 외부 API 호출, 데이터 조회 및 수정 등의 기능을 제공
- 도구의 이름, 기능 설명 및 입력값의 형식 등을 명시

**Tool의 정의 예시**

```json
{
    "name": "search_pages",
    "description": "노션에서 키워드로 페이지를 검색한다.",
    "inputSchema": {
        "type": "object",
        "properties": {
            "query": {
                "type": "string"
            },
            "limit": {
                "type": "integer",
                "minimum": 1,
                "maximum": 20
            }
        },
        "required": ["query"]
    }
}
```

| 항목 | 설명 |
| --- | --- |
| name | 도구의 이름 |
| description | 모델이 도구의 용도와 사용 여부를 판단하는 근거 |
| inputSchema | 도구에 전달할 입력값의 형식 |
| properties | 입력 가능한 값의 목록 및 제약 조건 |
| required | 반드시 입력해야 하는 값 |

### **2) Resources**

- 모델이나 애플리케이션이 참고할 수 있는 데이터를 제공
- 파일 내용, 문서, 데이터베이스 스키마 등이 해당
- RAG에서 활용하는 참고 문서를 표준화된 방식으로 전달하는 데 활용 가능

### **3) Prompts**

- 특정 작업에 반복적으로 활용할 수 있는 프롬프트 템플릿
- 코드 리뷰, 회의록 요약, 데이터 분석 등 정형화된 업무에 활용
- 작업에 필요한 지시와 컨텍스트를 일관된 방식으로 제공할 수 있음

## 2-6. MCP의 한계와 보안 문제

### **1) 보안 책임은 누구에게 있는가?**

- MCP는 외부 도구와의 연결 및 통신 방식을 정의하는 프로토콜
- 어떤 도구를 허용할지, 어떤 권한을 부여할지에 대한 정책은 Host가 결정
- 실제 사용자 인증, 입력 검증 및 접근 권한 확인은 서버와 업무 시스템에서도 수행되어야 함
- MCP를 사용한다는 사실만으로 시스템의 안전성이 보장되는 것은 아님

### **2) 도구 오염(Tool Poisoning)**

: 외부 도구의 설명(description) 등에 악의적인 지시를 삽입하여 모델이 의도하지 않은 행동을 수행하도록 유도하는 공격

- 모델은 도구의 설명을 읽고 해당 도구를 사용할지 판단
- 공격자는 정상적인 기능 설명에 악의적인 지시를 포함시킬 수 있음
- 사용자가 해당 지시를 직접 확인하지 못하는 경우 위험성이 증가

**도구 오염 예시**

- 정상적인 도구: 두 수를 더하는 계산 도구
- 오염된 도구: 계산과 무관한 외부 정보에 접근하도록 유도하는 지시가 설명에 포함된 도구
- 모델이 악의적인 지시를 신뢰하면 사용자가 허용하지 않은 행동을 요청할 수 있음

### **3) 프롬프트 주입(Prompt Injection)**

: 모델이 읽는 외부 데이터에 악의적인 지시를 삽입하여 기존의 작업 목적이나 행동을 변경하도록 유도하는 공격

- 웹페이지, 이메일, 문서, GitHub 이슈 등 외부 데이터에 악의적인 지시가 포함될 수 있음
- 모델이 외부 데이터와 실제 사용자 지시를 명확하게 구분하지 못하면 발생
- 에이전트에 부여된 권한이 넓을수록 정보 유출 및 비인가 작업의 위험이 커짐

**도구 오염과 프롬프트 주입의 차이**

| 구분 | 도구 오염(Tool Poisoning) | 프롬프트 주입(Prompt Injection) |
| --- | --- | --- |
| 주요 공격 위치 | 도구 설명 및 메타데이터 | 모델이 읽는 입력 및 외부 콘텐츠 |
| 공격 방식 | 도구에 악의적인 지시 삽입 | 문서 등에 악의적인 지시 삽입 |
| 주요 위험 | 의도하지 않은 도구 호출 및 행동 | 작업 목적 변경 및 정보 유출 |
| 공통점 | 모델이 신뢰해서는 안 되는 지시를 따르도록 유도 | 모델이 신뢰해서는 안 되는 지시를 따르도록 유도 |

### **MCP 활용 시 보안 관리 방안**

- 공식적이고 검증된 MCP 서버 연결
- 업무 수행에 필요한 최소한의 권한만 부여
- 도구 설명 및 입력 형식 검토
- 데이터 조회와 수정에 대한 접근 권한 분리
- 데이터 삭제, 외부 발송 등 중요한 작업 전 사용자 승인

=> MCP는 연결 방식의 표준이지 보안 자체를 보장하는 기술이 아니므로, **Host의 실행 통제와 Server의 접근 권한 검증이 함께 필요함**

> # **3. 데이터 구조화: 그래프와 온톨로지**

## 3-1. 데이터 구조화의 필요성

### **데이터 구조화란?**
: 분산되어 있는 데이터를 일정한 형식과 의미에 따라 정리하고, 데이터 간의 관계를 표현하는 과정

- 실제 기업의 데이터는 ERP, CRM, 회의록, 문서 등 다양한 시스템에 분산되어 있음
- AI 에이전트가 복잡한 업무를 수행하려면 개별 데이터뿐 아니라 데이터 사이의 관계도 이해해야 함
- 단순히 데이터를 수집하여 모델에 전달하는 것만으로는 복잡한 관계에 대한 추론이 어려울 수 있음

### **Vanilla RAG의 한계**

**RAG(Retrieval-Augmented Generation)란?**
- 사용자의 질문과 관련된 외부 데이터를 검색하여 LLM의 답변 생성에 활용하는 방식
- 일반적인 Vanilla RAG는 문서를 청크(Chunk) 단위로 분할하고, 질문과 유사도가 높은 청크를 검색

**Top-K 검색이란?**
: 질문과의 유사도가 가장 높은 상위 $k$개의 문서 조각을 선택하여 모델에 전달하는 검색 방식

| 구분 | 장점 | 한계 |
| --- | --- | --- |
| k가 작은 경우 | 관련 없는 정보의 유입 감소 | 필요한 정보가 누락될 수 있음 |
| k가 큰 경우 | 필요한 정보가 포함될 가능성 증가 | 관련 없는 정보 및 처리 비용 증가 |

- 적절한 k의 값을 설정하기 어려운 문제를 Top-K 딜레마(Top-K Dilemma)라고 함
- 일반적인 벡터 검색은 개별 청크와 질문의 유사도를 중심으로 검색하기 때문에 청크 간의 관계를 직접 반영하기 어려움

**Vanilla RAG의 한계 예시**

사용자 질문: "영업팀 A가 관리하는 고객들의 불만사항은 무엇인가?"

```text
청크 1: 영업팀 A는 제품 B를 담당한다.
청크 2: 고객 C는 제품 B를 구매했다.
청크 3: 고객 C는 배송 지연에 대한 불만을 제기했다.
```

- 실제 답변을 생성하려면 청크 1, 청크 2, 청크 3의 관계를 모두 파악해야 함
- 청크 3에는 질문에 포함된 '영업팀 A'라는 표현이 존재하지 않음
- 질문과의 유사도만으로 검색하면 정답에 필요한 청크 3이 누락될 수 있음

### **Vanilla RAG의 한계를 개선하기 위한 방법**

| 방법 | 설명 | 한계 |
| --- | --- | --- |
| Hybrid Search | 키워드 검색과 벡터 검색을 결합 | 직접적으로 연결되지 않은 정보 탐색에 한계 |
| Re-ranking | 검색 결과를 정밀한 모델로 재평가 | 초기 검색에서 누락된 정보는 복구하기 어려움 |
| Query Rewriting | 질문을 검색에 적합한 형태로 재작성 | 추가 모델 호출 및 질문 의도 왜곡 가능성 |
| Multi-hop Retrieval | 검색과 추론을 반복하여 정보를 탐색 | 검색 횟수에 따른 비용 및 시간 증가 |
| GraphRAG | 엔티티와 관계를 그래프로 구조화하여 검색 | 그래프 구축 및 갱신 비용 발생 |

=> Vanilla RAG의 근본적인 한계는 개별 문서의 유사도뿐 아니라 **문서 및 데이터 사이의 관계를 고려할 필요가 있다는 것**

## 3-2. 그래프 구조(Graph Structure)

### **그래프 자료구조란?**
: 데이터의 개체와 개체 간의 관계를 노드(Node)와 엣지(Edge)로 표현하는 자료구조

$$
G = (V, E)
$$

- $G$: 그래프(Graph)
- $V$: 노드(Vertex, Node)의 집합
- $E$: 엣지(Edge)의 집합

### **그래프의 주요 구성요소**

| 구성요소 | 정의 | 예시 |
| --- | --- | --- |
| 노드(Node) | 그래프를 구성하는 개별 개체 | 고객, 제품, 영업팀 |
| 엣지(Edge) | 노드 사이의 관계 | 구매, 담당 |
| 속성(Property) | 노드 또는 관계가 가지는 정보 | 이름, 구매일자, 구매금액 |

**그래프 구조의 예시**

```text
[영업팀 A] ──담당──> [제품 B]
                        ↑
                       구매
                        │
                     [고객 C] ──제기──> [배송 지연 불만]
```

- 개별 데이터뿐 아니라 데이터 사이의 연결 관계를 표현
- 특정 노드에서 연결된 다른 노드로 이동하면서 관련 정보 탐색 가능
- 질문과 직접적으로 유사하지 않은 데이터도 연결 관계를 통해 검색할 수 있음

**Multi-hop 추론이란?**
- 하나의 관계가 아닌 여러 단계의 관계를 연결하여 정보를 도출하는 방식
- 영업팀 → 담당 제품 → 구매 고객 → 고객 불만과 같이 여러 관계를 탐색

### **GraphRAG란?**

: 문서에서 개체(Entity)와 관계(Relationship)를 추출하여 지식 그래프를 구축하고, 이를 검색 및 답변 생성에 활용하는 RAG 방식

| 검색 방식 | 설명 | 적합한 질문 |
| --- | --- | --- |
| Local Search | 특정 엔티티를 중심으로 연결된 정보를 탐색 | 고객 A와 제품 B의 관계는 무엇인가? |
| Global Search | 그래프의 커뮤니티별 요약을 활용하여 전체적인 주제를 파악 | 전체 고객 불만의 주요 원인은 무엇인가? |

- Vanilla RAG가 질문과 문서의 유사도를 중심으로 검색한다면, GraphRAG는 데이터 사이의 관계까지 활용
- 여러 데이터에 걸친 복잡한 관계를 추론하는 데 유리
- 그래프를 구축하는 초기 비용과 지속적인 데이터 갱신 비용이 필요

## 3-3. 팔란티어 온톨로지(Palantir Ontology)

### **온톨로지(Ontology)란?**
: 특정 영역의 개념, 속성 및 개념 간의 관계를 체계적으로 정의하는 데이터 구조화 방법론

### **팔란티어 온톨로지(Palantir Ontology)란?**
: 기업 내부의 다양한 데이터와 업무 관계를 통합하고, 실제 업무에서 수행할 수 있는 행동 및 접근 권한까지 정의하는 의미 기반의 데이터 구조

- ERP, CRM, 물류 시스템 등 분산된 데이터를 하나의 의미 있는 관계망으로 통합
- 실제 기업의 객체와 관계를 디지털 환경에 구현
- 단순한 데이터 조회를 넘어 업무 수행과 의사결정에 활용
- 일반적인 온톨로지가 개념과 관계의 정의에 중점을 둔다면, 팔란티어 온톨로지는 실제 행동과 권한 관리까지 포함하는 운영 체계를 제공

### **팔란티어 온톨로지의 주요 구성요소**

| 구성요소 | 정의 | 예시 |
| --- | --- | --- |
| 객체 타입(Object Type) | 객체의 종류 및 구조를 정의 | 고객, 제품, 계약 |
| 객체(Object) | 객체 타입에 해당하는 실제 개체 | 고객 A, 제품 B |
| 객체 세트(Object Set) | 여러 객체의 집합 | VIP 고객 전체 |
| 속성(Property) | 객체가 가지는 특징 | 이름, 등급, 계약일 |
| 속성값(Property Value) | 객체의 속성에 해당하는 실제 값 | VIP, 2024-01-03 |
| 공유 속성(Shared Property) | 여러 객체 타입이 공통으로 사용하는 속성 | 생성일자 |
| 링크 타입(Link Type) | 객체 사이의 관계 유형 | 구매했다, 담당한다 |
| 링크(Link) | 객체 사이에 존재하는 실제 관계 | 고객 A가 제품 B를 구매 |
| 액션 타입(Action Type) | 객체에 대해 수행할 수 있는 행동의 정의 | 주문 상태 변경, 알림 발송 |
| 역할 및 권한(Roles) | 데이터 조회 및 행동 수행 권한 관리 | 담당 직원만 고객 정보 조회 |

### **1) 액션 타입(Action Type)**

- 데이터를 조회하는 것뿐 아니라 실제 변경 및 행동까지 정의
- 에이전트가 온톨로지의 객체와 관계를 바탕으로 판단하고, 정의된 액션을 실행할 수 있음
- ReAct의 행동(Action) 및 MCP의 Tools와 연결되는 개념

예시: 생산설비에서 불량이 발생하면 담당자에게 알림 발송

### **2) 역할 및 권한(Roles)**

- 직급, 부서 및 업무에 따라 데이터 접근 권한을 관리
- AI 에이전트에도 업무 수행에 필요한 권한을 부여할 수 있음
- 허용된 범위 안에서 데이터를 조회하고 액션을 실행하도록 통제

### **관계형 DB, 그래프 DB, 팔란티어 온톨로지 비교**

| 구분 | 관계형 DB | 그래프 DB | 팔란티어 온톨로지 |
| --- | --- | --- | --- |
| 기본 구조 | 테이블 | 노드 및 엣지 | 객체 및 링크 |
| 개체 표현 | 행(Row) | 노드(Node) | 객체(Object) |
| 특성 표현 | 열(Column) | 속성(Property) | 속성(Property) |
| 관계 표현 | 외래키 및 조인(Join) | 엣지(Edge) | 링크(Link) |
| 주요 목적 | 정형 데이터 저장 및 관리 | 데이터 간 관계 탐색 | 기업 데이터와 업무의 의미적 통합 |
| 행동 및 권한 | 별도의 로직 및 권한 체계 | 별도의 로직 및 권한 체계 | 온톨로지와 연계된 액션 및 권한 관리 |

### **팔란티어 온톨로지의 활용 예시**

의료기기 제조기업의 납품 지연 원인을 분석하는 상황을 가정

- 생산 데이터는 제조 관리 시스템에 저장
- 재고 데이터는 ERP 및 창고 시스템에 저장
- 고객 계약 정보는 CRM에 저장
- 납품 및 운송 정보는 물류 시스템에 저장

```text
[제조 공장] ──생산──> [작업 지시서] ──출하──> [고객]
      │                    │
  [기계 장비]          [입고 패키지]
                           │
                       [출고 화물]
```

- 분산된 데이터를 온톨로지의 객체와 링크로 통합
- 납품 지연이 발생하면 공장, 작업 지시서, 물류, 고객 사이의 관계를 탐색
- 원인 분석 이후 정의된 액션을 통해 담당자에게 알림을 보내는 등 실제 업무 수행 가능

=> 팔란티어 온톨로지의 핵심은 데이터를 단순히 연결하는 것을 넘어 **의미, 관계, 행동 및 권한을 하나의 구조 안에서 관리하는 것**

> # **4. XAI(Explainable Artificial Intelligence)**

## 4-1. XAI의 정의 및 필요성

### **XAI(Explainable AI)란?**
: 인공지능 모델의 예측 및 의사결정에 대한 근거를 사람이 이해할 수 있는 형태로 제공하기 위한 기술 및 연구 분야

- 딥러닝 모델의 예측 정확도가 높더라도 판단 근거가 적절하다는 보장은 없음
- 모델의 의사결정 과정과 결과에 영향을 미친 요인을 파악할 필요가 있음
- 의료, 금융, 자율주행 등 모델의 판단이 사람에게 직접적인 영향을 미치는 분야에서 중요성이 높음

### **블랙박스(Black Box) 문제란?**
: 모델의 입력과 출력은 확인할 수 있지만, 내부에서 어떤 연산과 판단 과정을 거쳐 결과를 도출했는지 파악하기 어려운 문제

- 딥러닝 모델은 수많은 파라미터와 비선형 연산으로 구성
- 복잡한 내부 연산을 사람이 직접 이해하기 어려움
- 모델의 높은 정확도가 올바른 판단 과정에서 비롯된 것인지 확인하기 어려움

**블랙박스 문제의 예시**

피부암 진단 AI가 피부 사진을 분석하여 암 발생 여부를 판단하는 경우

- 모델이 실제 피부 병변의 특징을 학습하는 것이 이상적
- 그러나 학습 데이터에서 반창고가 부착된 사진과 암 진단 결과가 자주 함께 등장하면, 모델이 반창고 자체를 암의 신호로 학습할 수 있음
- 테스트 정확도가 높더라도 실제로는 잘못된 특징을 활용하여 판단하고 있을 수 있음

### **블랙박스 문제를 해결해야 하는 이유**

| 구분 | 설명 |
| --- | --- |
| 신뢰성 | 모델이 적절한 근거를 바탕으로 판단하는지 확인 |
| 오류 탐지 | 잘못된 학습 및 판단 과정의 원인 파악 |
| 공정성 | 특정 집단에 대한 편향이나 차별적 판단 확인 |
| 규제 대응 | 의료, 금융 등 고위험 분야의 설명 및 책임 요구에 대응 |
| 안전성 | 모델의 의사결정으로 발생할 수 있는 위험을 사전에 검토 |

=> 모델의 성능뿐만 아니라 **모델이 어떤 근거로 판단했는지 이해하고 검증하는 과정**이 중요

## 4-2. 블랙박스 문제를 해결하기 위한 기존 시도

### **1) 회귀 분석(Regression)**

: 독립변수와 종속변수 사이의 관계를 수학적으로 표현하는 분석 방법

**선형 회귀의 기본 형태**

$$
\hat{y} = \beta_0 + \beta_1x_1 + \beta_2x_2 + \cdots + \beta_nx_n
$$

- $\hat{y}$: 예측값
- $\beta_0$: 절편
- $\beta_i$: 각 독립변수에 대한 회귀계수
- $x_i$: 독립변수

**장점**

- 모델의 구조와 계수를 직접 확인할 수 있음
- 다른 변수들을 고정했을 때 개별 변수의 변화가 예측값에 미치는 영향을 해석할 수 있음
- 대표적인 화이트박스(White Box) 모델에 해당

**한계**

- 복잡한 비선형 관계를 단순한 선형 모델로 표현하기 어려움
- 이미지, 텍스트 등 복잡한 데이터에서는 충분한 예측 성능을 확보하지 못할 수 있음

### **2) SHAP / LIME: 사후적 설명(Post-hoc Explanation)**

: 학습된 모델은 유지하면서, 모델의 예측 결과에 영향을 미친 입력 특성을 분석하여 설명하는 방법

**① SHAP(SHapley Additive exPlanations)**

: 게임 이론의 Shapley Value를 머신러닝 모델 해석에 적용하여 각 입력 특성의 예측 기여도를 계산하는 방법

- 입력 특성이 포함된 경우와 포함되지 않은 경우의 예측값 차이를 고려
- 여러 특성 조합에서의 기여도를 종합하여 개별 특성의 기여도를 산출
- 양수의 SHAP Value는 예측값을 증가시키는 방향으로 기여
- 음수의 SHAP Value는 예측값을 감소시키는 방향으로 기여

$$
f(x) = \phi_0 + \sum_{i=1}^{n}\phi_i
$$

- $f(x)$: 모델의 예측값
- $\phi_0$: 기준 예측값
- $\phi_i$: i번째 특성의 SHAP Value

**SHAP의 작동 과정**

1. 기준 예측값 설정
2. 입력 특성의 다양한 조합 고려
3. 각 특성의 포함 여부에 따른 예측값 변화 계산
4. 특성별 Shapley Value 산출
5. 기여도를 시각화하여 예측 결과 해석

**장점**
- 게임 이론에 기반한 수학적 설명 제공
- 개별 예측과 전체 모델의 특성 중요도 분석 가능

**한계**
- 고려해야 하는 특성의 조합이 많아지면 계산 비용이 증가
- 입력 특성의 기여도를 설명할 수 있으나 모델 내부의 실제 연산 구조까지 직접 설명하지는 못함

**② LIME(Local Interpretable Model-Agnostic Explanations)**

: 특정 데이터 주변에서 복잡한 블랙박스 모델을 해석 가능한 단순 모델로 근사하는 방법

- **Local**: 모델 전체가 아닌 특정 예측 결과를 중심으로 분석
- **Interpretable**: 사람이 이해할 수 있는 형태의 설명 제공
- **Model-Agnostic**: 특정 모델 구조에 종속되지 않음

**LIME의 작동 과정**

1. 해석하고자 하는 입력 데이터 선택
2. 기존 입력 데이터를 조금씩 변형하여 주변 샘플 생성
3. 생성된 샘플에 대한 블랙박스 모델의 예측값 계산
4. 원본 데이터와의 유사도를 고려하여 단순한 모델 학습
5. 단순 모델의 계수를 활용하여 예측 결과 해석

**장점**
- 다양한 블랙박스 모델에 적용 가능
- 특정 예측 결과를 직관적으로 설명할 수 있음

**한계**
- 주변 샘플의 생성 방식에 따라 설명 결과가 달라질 수 있음
- 개별 데이터에 대한 국소적 설명이므로 모델 전체의 판단 구조를 설명하기 어려움

### **3) LLM + RAG를 활용한 설명**

: 모델의 입력 데이터, 예측 결과 및 관련 외부 정보를 LLM에 전달하여 자연어 형태의 설명을 생성하는 방식

- 복잡한 예측 결과를 사람이 이해하기 쉬운 문장으로 변환 가능
- RAG를 활용하여 관련 문서와 참고 근거를 함께 제공할 수 있음

**한계**

- 설명을 생성하는 LLM 역시 블랙박스 모델에 해당
- 생성된 설명이 실제 모델 내부의 판단 및 연산 과정과 일치한다는 보장이 없음
- 그럴듯한 설명과 실제 의사결정 근거를 구분해야 함

### **기존 설명 방법 비교**

| 방법 | 설명 방식 | 장점 | 한계 |
| --- | --- | --- | --- |
| 회귀 분석 | 모델 구조 및 계수 직접 해석 | 높은 해석 가능성 | 복잡한 패턴 표현에 한계 |
| SHAP | 입력 특성별 기여도 계산 | 정량적인 설명 가능 | 계산 비용 및 내부 연산 설명의 한계 |
| LIME | 개별 예측 주변에서 단순 모델로 근사 | 다양한 모델에 적용 가능 | 국소적 설명 및 결과의 불안정성 |
| LLM + RAG | 자연어 설명 생성 | 사용자 친화적인 설명 | 실제 판단 과정과의 일치 여부 검증 필요 |

=> 기존의 사후적 설명 방식은 예측에 영향을 미친 요인을 파악하는 데 유용하지만, **모델 내부에서 실제로 어떤 연산이 이루어졌는지 직접 설명하는 데에는 한계가 있음**

## 4-3. Mechanistic Interpretability

### **Mechanistic Interpretability(기계적 해석 가능성)란?**
: 딥러닝 모델의 내부 구조와 연산 과정을 직접 분석하여, 특정 기능이 모델 내부에서 어떻게 구현되는지를 이해하는 연구 분야

- 가중치(Weight), 뉴런(Neuron), 활성값(Activation), 어텐션 헤드(Attention Head) 등을 직접 분석
- 특정 기능을 수행하는 모델 내부의 구성요소와 연결 관계를 파악
- 모델 내부의 연산 과정을 회로(Circuit) 수준에서 역공학(Reverse Engineering)하는 것을 목표로 함

### **기존 XAI와 Mechanistic Interpretability의 차이**

| 구분 | 기존 XAI(SHAP, LIME 등) | Mechanistic Interpretability |
| --- | --- | --- |
| 접근 방식 | 모델 외부에서 입력과 출력 분석 | 모델 내부 구조 및 연산 분석 |
| 핵심 질문 | 어떤 입력 특성이 예측에 영향을 미쳤는가? | 어떤 내부 연산을 통해 예측 결과가 생성되었는가? |
| 분석 대상 | 입력 변수 및 예측값 | 뉴런, 가중치, 활성값, 어텐션 헤드 |
| 분석 수준 | 입력과 출력 사이의 관계 | 모델 내부의 기능 및 회로 |
| 주요 목적 | 예측 결과에 대한 설명 | 실제 판단 메커니즘에 대한 이해 |

### **핵심 기법: Activation Patching(활성값 패칭)**

: 서로 다른 입력을 처리할 때 나타나는 모델 내부의 활성값을 교체하고, 출력의 변화를 관찰하여 특정 내부 구성요소의 인과적 영향을 분석하는 방법

**Activation Patching의 작동 과정**

1. 서로 다른 출력이 생성되는 두 개의 입력을 준비
2. 각 입력에 대한 모델 내부의 활성값을 확인
3. 비교 입력을 처리하는 과정에서 특정 위치의 활성값을 기준 입력의 활성값으로 교체
4. 활성값 교체에 따른 출력 변화를 확인
5. 해당 위치가 특정 예측 결과에 미치는 영향을 분석

**Activation Patching의 예시**

| 구분 | 입력 | 모델의 출력 |
| --- | --- | --- |
| 기준 입력 A | 프랑스의 수도는 | 파리 |
| 비교 입력 B | 독일의 수도는 | 베를린 |

```text
입력 B → [레이어 1] → [레이어 2] → [레이어 3] → 출력
                           ↑
                입력 A의 활성값으로 교체
```

- 활성값을 교체한 이후 출력이 '베를린'에서 '파리'로 변경된다면, 해당 위치가 수도를 예측하는 연산에 인과적으로 기여한다는 근거가 될 수 있음
- 출력이 변경되지 않는다면 해당 위치의 활성값 교체만으로는 예측 결과를 변경하기 어려움을 의미
- 하나의 실험만으로 전체 판단 경로를 확정할 수는 없으며, 여러 위치와 입력에 대한 추가적인 검증이 필요

## 4-4. Mechanistic Interpretability의 핵심 개념 3가지

### **1) Features(특징)**

: 모델 내부에서 표현되는 특정 개념이나 의미

- 단어, 사물, 언어, 문체, 상황 등 다양한 수준의 개념을 표현
- 특정 개념이 입력에 등장하면 관련된 내부 활성값이 변화
- Feature는 개별 뉴런에 대응하거나 여러 뉴런의 활성값 조합으로 표현될 수 있음

**Feature의 표현 방식**

| 구분 | 설명 | 예시 |
| --- | --- | --- |
| 개별 뉴런 | 특정 뉴런이 하나의 개념에 주로 반응 | 왕족 관련 표현에 반응하는 뉴런 |
| 방향 벡터 | 여러 뉴런의 활성값 조합이 하나의 개념을 표현 | 여러 활성값의 조합으로 표현되는 특정 의미 |

- 실제 모델에서는 하나의 뉴런이 여러 개념에 반응하는 경우가 많음
- 따라서 Feature와 Neuron이 항상 일대일로 대응하는 것은 아님

### **2) Circuits(회로)**

: 여러 뉴런 및 어텐션 헤드가 연결되어 특정 기능을 수행하는 모델 내부의 부분적인 네트워크

- 개별 구성요소가 어떤 기능을 담당하는지 분석
- 각 구성요소가 정보를 어떻게 전달하고 결합하는지 파악
- 특정 기능이 모델 내부에서 구현되는 방식을 설명하는 데 활용

**이미지 분류 모델의 회로 예시**

```text
기초 특징 탐지
    ↓
선, 곡선 등의 패턴 인식
    ↓
눈, 코, 귀 등의 세부 특징 인식
    ↓
여러 특징의 결합
    ↓
얼굴 인식
```

- 개별 뉴런을 분석하는 것만으로는 복잡한 기능을 충분히 설명하기 어려움
- 여러 뉴런의 연결과 정보 전달 과정을 분석해야 특정 기능의 작동 원리를 이해할 수 있음

### **3) Superposition(중첩)**

: 모델이 제한된 활성 공간에서 여러 Feature를 서로 일부 겹치는 방향으로 표현하는 현상

- 모델이 표현해야 하는 개념의 수는 활성 공간의 차원보다 많을 수 있음
- 모든 Feature에 독립적인 차원을 할당하기 어려우므로 여러 개념을 중첩하여 표현
- 이 과정에서 여러 Feature가 서로 영향을 미치는 간섭(Interference)이 발생할 수 있음

**Superposition이 발생하는 원리**

1. 모델은 Feature를 활성 공간 내의 방향 벡터로 표현
2. 제한된 차원에서는 모든 Feature를 서로 직교하는 방향으로 배치할 수 없음
3. 더 많은 Feature를 표현하기 위해 서로 일부 겹치는 방향을 활용
4. 하나의 Feature가 활성화될 때 다른 Feature의 정보가 일부 섞이는 간섭 발생
5. 여러 Feature가 동시에 활성화되는 빈도가 낮은 희소성(Sparsity)을 활용하여 간섭을 줄임

**Superposition으로 인한 문제**

- 하나의 뉴런에 여러 개념이 혼재될 수 있음
- 개별 뉴런을 분석하는 것만으로 특정 개념을 식별하기 어려움
- 모델 내부의 기능 및 회로를 해석하는 과정이 복잡해짐

### **Sparse Autoencoder(SAE)**

: 모델 내부의 활성값에 중첩되어 있는 Feature를 분리하여 해석 가능한 희소 표현을 추출하는 방법

- 모델 내부의 활성값을 더 많은 차원의 희소한 표현으로 변환
- 동시에 활성화되는 Feature의 수를 제한하여 개별 Feature를 분리하도록 학습
- 개별 뉴런만으로 파악하기 어려운 개념을 발견하는 데 활용

### **핵심 개념 비교**

| 개념 | 핵심 질문 | 설명 |
| --- | --- | --- |
| Features | 모델 안에는 무엇이 표현되어 있는가? | 모델이 학습하고 표현하는 개념 및 특징 |
| Circuits | 이러한 특징은 어떻게 연결되어 작동하는가? | 특정 기능을 수행하는 내부 연산 구조 |
| Superposition | 모델 내부의 특징을 왜 해석하기 어려운가? | 제한된 활성 공간에 여러 특징이 중첩되는 현상 |

=> Mechanistic Interpretability는 Feature와 Circuit을 분석하고 Superposition으로 인한 해석의 어려움을 해결함으로써 **모델 내부의 실제 작동 원리를 이해하는 것을 목표로 함**

> # **5. 전체 내용 정리**

## 5-1. AI 시스템의 구성 과정

1. **AI 에이전트**: LLM에 추론, 계획 및 외부 도구를 활용한 실행 기능을 부여
2. **MCP**: AI 애플리케이션과 외부 도구 및 데이터의 연결 방식을 표준화
3. **그래프 및 온톨로지**: 데이터 사이의 의미와 관계를 구조화하고 행동 및 권한을 관리
4. **XAI**: 모델의 판단 근거를 설명하고 내부 작동 원리를 분석

## 5-2. 핵심 기술 정리

| 구분 | 해결하고자 하는 문제 | 핵심 기술 및 개념 |
| --- | --- | --- |
| AI 에이전트 | LLM이 답변 생성에 그치고 실제 업무를 수행하지 못하는 문제 | ReAct, Multi-Agent System |
| MCP | 애플리케이션과 외부 도구마다 개별적인 연동이 필요한 문제 | Function Calling, Host, Client, Server |
| 데이터 구조화 | 분산된 데이터 사이의 의미 및 관계를 파악하기 어려운 문제 | GraphRAG, Graph DB, Palantir Ontology |
| XAI | 모델의 판단 근거와 내부 연산을 이해하기 어려운 문제 | SHAP, LIME, Mechanistic Interpretability |

## 5-3. 핵심 개념 정리

- **AI 에이전트**: 목표 달성을 위해 추론하고 외부 환경과 상호작용하는 시스템
- **ReAct**: 추론, 행동, 관찰을 반복하여 다음 행동을 결정하는 방식
- **다중 에이전트**: 여러 에이전트가 역할을 분담하여 하나의 목표를 달성하는 구조
- **Function Calling**: 모델이 외부 함수의 구조화된 호출 요청을 생성하는 기능
- **MCP**: AI 애플리케이션과 외부 도구의 연결 및 통신을 표준화하는 프로토콜
- **Tool Poisoning**: 도구의 설명 등에 악의적인 지시를 삽입하는 공격
- **Prompt Injection**: 모델이 읽는 입력에 악의적인 지시를 삽입하는 공격
- **GraphRAG**: 데이터 사이의 관계를 그래프로 구조화하여 검색에 활용하는 방식
- **Palantir Ontology**: 기업 데이터의 의미와 관계, 행동 및 권한을 통합하는 구조
- **XAI**: AI 모델의 예측과 의사결정 근거를 설명하기 위한 기술
- **Mechanistic Interpretability**: 모델 내부의 실제 연산 구조 및 기능을 분석하는 연구 분야
- **Features**: 모델 내부에 표현되는 개념 및 특징
- **Circuits**: 특정 기능을 수행하는 모델 내부의 연산 구조
- **Superposition**: 제한된 활성 공간에 여러 Feature가 중첩되어 표현되는 현상

=> 실제 업무에서 AI를 활용하기 위해서는 모델 자체의 성능뿐만 아니라 **실행 능력, 외부 도구와의 연결, 데이터 구조화 및 의사결정의 설명 가능성**을 종합적으로 고려해야 함
