## AI 에이전트란?<br>
### AI 에이전트의 정의 / 생성형 AI vs 에이전틱 AI <br>
AI 에이전트 : 사용자의 목표 달성하기 위해 AI 모델로 환경이랑 상호작용 -> 추론+ 계획+도구<br>

생성형 AI : 프롬프트 -> 출력<br>
에이전틱 AI : 목표 달성까지 추론+행동 반복<br>
= ReAct<br>

### ReAct (생각 → 행동 → 관찰)<br>
생각->행동->관찰->생각..<br>
모델이 도구를 직접 실행하진 x 요청만 o<br>
실제 실행은 앱이 함<br>

### 다중 에이전트 : 오케스트레이터 + 서브 에이전트<br>
- MAS 구조 : 멀티 에이전트 = 여러 에이전트가 역할을 나눠 함께 일함<br>
= 오케스트레이터(일 쪼개줄게) + 서브 에이전트 (역할 해낼게)<br>

why? 역할 왜 나눠<br>
- 스스로 실수 잡기 어려움 -> 검토를 다른 에이전트가 담당해서 다시 확인!<br>
- 동시에 여러개 조사 가능 ! = 각자의 컨텍스트 창 사용 -> fast, 효율 up<br>

역할 나누기<br>
- 고정 워크플로우 = 단순 + 순서<br>
- 단일 에이전트 = 도구 몇개면 ok<br>
- 다중 에이전트 = 따로 조사해야함 = 큰 작업<br>

비용/성능 조합 <br>
: 각 역할별로 필요한 능력에 맞춰 급이 다른 모델 섞기<br>
- 오케스트레이터 : 고성능 모델<br>
- 검색/정리 : 경량 모델<br>
- 코드 검증 : 코드 특화모델<br>


## MCP란? <br>
### 도구 호출의 원리 : Function Calling<br>
- function callng<br>
: 모델이 도구 호출 요청을 내놓는 방법<br>
앱>모델 : 질문+ 쓸 수 있는 함수 목록 내놔<br>
모델 : 어떤 함수 + 어떤 값 부를까? 정하기<br>
앱 : 실행<br>
앱>모델 : 실행 결과 전달<br>
모델 : 최종 답 작성<br>


### MCP가 필요한 이유 : n×m → n+m<br>
MCP 없이는 n*m, MCP 있으면 n+m <br>
표준이 생겨서 각 도구별로 연결할 필요 없이 표준에 각각 연결만 하면됨<br>

MCP : 앱이 도구 찾기/연결/호출 하는 법 <br>

### MCP의 구조 : Host / Client / Server<br>
host : 사용자가 쓰는 AI 앱 (클로드) <br>
-> 어떤 서버/도구 호출하지?<br>
client : 호스트 내의 통신 담당 <br>
-> 1 서버 1 클라언트 <br>
-> MCP 규격으로 메시지 전달<br>
server : MCP 규격으로 도구/데이터 내놓는 프로그램 (노션 서버)<br>
-> 가진 기능 선언 + 요청오면 작업 수행<br>

### MCP 서버의 기본 요소 : Tools / Resources / Prompts<br>
= 서버가 내놓을 수 잇는 기능의 종류<br>
- Tools : 실행할 수 있는 행동 ex 노션 페이지 만들기<br>
- Resources : 읽기 전용 데이터 ex 파일 내용<br>
  RAG가 참고할 문서를 resources로 넘길 수 있음<br>
- Prompts : 재사용가능 템플릿 ex  회의록 요약 양식<br>

점점 요청 하나가 독립적으로 처리되는 중<br>

## 데이터 구조화<br>
### 왜 구조화가 필요한가<br>
RAG로 끌어오는데엔 한계가 있음<br>
Vanila RAG 한계 )<br>
RAG에서 질문과의 유사도가 높은 상위k 문서만 선택 -> LLM에 전달 <br>
but, k를 몇으로 설정하지 = top-k 딜레마<br>

왜 생김? <br>
- 바닐라 RAG는 각 조각을 독립적으로 채점<br>
- 정답에 필요한 거랑 질문이 안비슷할 때도 많음<br>
= 청크들 사이의 관계를 바닐라 RAG가 표현 못함<br>
-> 데이터 구조화가 필요 !!<br>

### 그래프 구조와 팔란티어 온톨로지<br>
그래프 자료구조 : 데이터 = 노드+엣지<br>
-> 서로 어떻게 연결되어 있는 지 저장<br>

팔란티어 온톨로지 : 흩어진 데이터를 하나의 의미있는 관계망으로 묶기<br>

객체 타입 = 테이블의 틀<br>
객체 = 테이블 한 행<br>
객체 세트 = 여러 객체 모음<br>

속성 = 객체의 특징, 테이블 열<br>
속성값 = 실제 값<br>

공유 속성 = 여러 객체 타입에서의 공통 속성<br>

링크 타입 = 관계 스키마<br>
링크 = 실제 관계 인스턴스<br>

액션 타입 = 실제 변경+행동을 온톨로지 위에서 정의 <br>

역할 = AI에이전트도 권한 부여 받음 + 세밀하게 제어<br>
