## NLP (자연어 처리) <br>

자연어 : 의사소통용 : 1:n<br>
프로그래밍 언어 : 컴퓨터에게 명령용 : 1:1<br>

<자연어><br>
- 이산적 -> 숫자 또는 연속적 벡터로 변환하는 임베딩 필요<br>

- 순차적이어서 순서에 따라 문장 의미가 변함 -> 문맥 고려 가능한 모델 필요<br>

- 중의성, 모호성 -> 문맥 전체 파악 필요<br>

- 희소성 -> 표현을 다양하게 학습하기 어려움<br>

- 가변성, 진화 -> 계속 학습해줘야 함 (ML)<br>

<NLP 개념><br>
- 자연어를 컴퓨터가 분석, 이해, 생성할 수 있는 기술 (다만, 컴퓨터는 숫자 형태로 진행)<br>

<NLP 역할><br>
- 형태소 분석 : 문장 -> 형태소 -> 품사 & 문법 분석<br>
- 구문 분석 : 단어들이 어떤 관계 가짐?<br>
-의미 분석 : 단어와 문장 간의 관계 -> 문맥 파악<br>
- 담화 분석 : 문장 간의 관계 -> 맥락 분석<br>

<언어모델><br>
대량의 텍스트 학습 -> 단어 등장 확률 예측 -> 문장이 자연스러워짐<br>

- 기준) 학습 방식 -> SLM / NLM<br>
<SLM> : 통계적 언어 모델 : 비슷한 문맥에 등장하면 서로 유사할 거다~ (분포가설)<br>

- 조건부 확률 : A 발생 하에 B가 일어날 확률
  -> 문맥에 적합한 단어 예측<br>
- 카운트기반 확률 : 특정 문맥 다음에 어떤 단어가 얼마나 자주 등장하는 지 -> 빈도 높으면 확률도 높아짐<br>

  but, 한번도 등장하지 않은 단어 조합은 확률을 0으로 계산하는 희소문제 발생<br>

-> <N-gram> : 데이터 희소성&계산 복잡성 완화<br>
직전 N-1개 단어 바탕으로 다음 단어 등장 확률 예측<br>
N이 커질 수록 더 긴 문맥 반영<br>
  but, N이 커져서 특정 단어 조합 찾기 어려워지는 희소 문제 발생 & 거리가 먼 단어 간 관계 파악 어려움<br>

-> <PPL> : 모델이 문장 예측 시 얼마나 혼란스러워 하는 지 : 낮을 수록 좋음<br>
  -> 모델의 예측 성능 평가에 사용<br>

<NLM> : 신경망 언어 모델 : 단어 의미와 문맥 학습 + 관측되지 않은 표현도 일반화 가능<br>

- 임베딩 : 단어 -> 연속적 숫자 벡터<br>
  유사하면 벡터끼리 가까움, 의미적 유사성 파악, 일반화 가능<br>

## 딥러닝 기반 언어 모델<br>
< RNN& LSTM 한계> <br>
- 장기 의존성<br>
- 기울기 소실<br>
- 순차 처리의 비효율성<br>

<Transformer 등장><br>
- Self-Attention -> 단어 관계, 문맥 학습<br>
- 병렬 처리 & 위치 인코딩 -> 순서 정보 반영+ 효율 up<br>

입력 문장 -> 토큰 -> 벡터 + 위치 인코딩
-> 인코더 (Self-Attention : 단어간 관련도 계산, FFN : 정보->신경망->특징 학습)
+ 인코더... (쌓을 수록 정교) -> 디코더 (Masked Self-Attention : 미래 단어 보지 말고 다음 단어 예측, Cross-Attention : 입출력 문장 간 관계 학습, FFN : Attention 결과 변환 -> 다음 단어 예측에 활용) -> 디코더... -> 단어 순차적 생성<br>

<BERT><br>
Transformer 인코더 기반 + 양방향 문맥 학습<br>

- 양방향: 앞에 등장한 단어 + 뒤에 등장한 단어 고려<br>

- 인코더 구조만 활용 : 문장 내 단어 간 관계 분석(Self-Attention)<br>

- 사전학습+fine-tuning : 일반적 구조&문맥 학습 + 특정 작업의 데이터 추가 학습 -> 목적에 맞게끔<br>

학습 방법<br>
입력 문장 -> 토큰 + 특수 토큰 추가 -> 임베딩(토큰 -> 벡터) -> Transformer <br>Encoder 문맥학습 (토큰끼리 어떤 관계? -> 문맥 반영된 벡터 표현 생성) -> 출력 및 활용 (ex: 감성 분석, 질의 응답..)<br>

사전 학습<br>
특정 NLP 작업을 수행하기 전 대량의 텍스트 데이터로 일반적인 구조&문맥 미리 학습<br>
- BERT : 정답없이도 학습 가능 : MLM, NSP<br>

<MLM> : 일부 토큰만 가리고 문맥 활용해서 가려진 토큰 예측하도록 학습<br>

일부 토큰 무작위로 선택 -> [MASK]로 대체 -> 가려진 부분 근처의 문맥 분석 -> 예측 및 정답 확인 -> 가중치 업뎃<br>

-> 양방향 문맥 이해 up ! 여러 문맥에 맞는 동일 단어 의미 학습<br>

<NSP> : 두 문장이 있을 때 두 번째 문장이 첫 번째 문장 다음에 등장하는 문장인지 예측하도록 학습 -> 문맥적 연관성<br>

+) RoBERTa : BERT - NSP + Dynamic Masking
ALBERT : BERT 경량화 모델<br>

<GPT><br>
Transformer의 디코더 기반 구조 -> 이전 단어 바탕으로 다음 단어 예측 -> 텍스트 생성<br>

초기 GPT ) Decoder 기반 (Masked Self-Attention, FFN)<br>

최근 GPT ) 별도 파인튜닝 없이 프롬프트만으로 작업 수행<br>


<LLM> : 대구모 텍스트 + 신경망 -> 인간의 언어 이해+생성<br>

단어-단어 관계+문맥+문법+지식+언어 패턴을 스스로 터득<br>

한계) 환각 현상 : 없는 정보를 그럴 듯하게 생성<br>

## 생성 모델 <br>

분류 모델 = 강아지 사진 -> '강아지'<br>
생성 모델 = 강아지 사진 -> 새로운 강아지 사진<br>

생성 모델 분류 ) 데이터의 확률 밀도 직접 계산 가능 여부<br>
- 명시적 확률밀도 모델<br>
  학습한 데이터 분포에서 특정 데이터가 얼마나 그럴 듯한지 나타내는 값<br>
  유사=확률밀도 높음<br>
  - 다루기 쉬운 밀도 모델 : 수학적 계산 가능<br>
  - 근사적 밀도 모델 : 직접 X 근사 O<br>
  예) VAE, Boltzmann Machine<br>

- 암시적 확률밀도 모델<br>
P(X)를 수치로 알려주지 않고, 데이터 생성하는 데에 초점 -> 특징&패턴 학습<br>

  - 직접 샘플링 -> 생성<br>
  : 무작위 노이즈 입력 -> 생성 모델 한번 통과<br>
  - 간접 샘플링 -> 생성<br>
  : 확률 상태 반복 -> 새 데이터 생성<br>
  여러 단계 상태 변화 거침<br>



## AE, VAE<br>
<AE> : 입력 데이터의 중요 특징 압축 -> 원래 데이터와 유사하게 복원 -> 차이 최소화<br>

기존) 입력 이해 -> 결과 생성<br>
AE) 입력 특징 압축 -> 원래 데이터로 복원<br>

구조) 인코더 -> 병목 -> 디코더<br>

인코더 : 중요 특징 추출 & 정보 압축<br>
병목 : 잠재 벡터 & 핵심 특징 학습 유도<br>
디코더 : 원본과 유사하게 복원 & 비교<br>

왜 압축 후 복원? 데이터를 표현하는 데 가장 중요한 특징을 학습하도록 유도 : 얼마나 중요한지 파악 가능<br>
-> 데이터의 핵심 특징 & 구조 학습<br>

<VAE><br>
기존 한계) 잠재 공간 불규칙하게 형성<br>

VAE : 입력 데이터 특징 BY 확률 분포 -> 샘플링 -> 새로운 데이터 (변분 추론 활용)
- 복원 오차 최소화 : 원본 특징 최대한 복원<br>
- KL 발산 최소화 : 새로운 데이터 생성하기 좋은 잠재공간 유도<br>

활용) 이미지 생성, 데이터 증강, 이상 탐지, 특징 추출..<br>

AE VS VAE : 입력 데이터 특징을 잠재공간에 어떤 형태로 표현 & 배치하도록 학습했는 지<br>
AE: 1개, 고정된, 불규칙<br>
VAE: 확률 분포, 일정<br>

구조) 확률 분포 예측 -> 샘플링&재파라미터화 -> 생성<br>

<GAN> : 생성자와 판별자가 경쟁 - 실제 데이터와 유사한 새 데이터 생성하도록 학습<br>

생성자 ) 진짜 같은 가짜 데이터 생성<br>
판별자 ) 입력된 데이터가 진짠지 가짠지 구별<br>
-> 적대적 생성 신경망<br>
-> 생성자가 실제 데이터의 분포 학습하도록 만들기<br>
-> 자연스러운 새 데이터 생성이 목표<br>

구조)<br>
무작위 노이즈 -> 생성자 -> 가짜 이미지 생성 -> 판별자 -> 구별<br>

장점) 진짜 같은 새 데이터 + 선명도+  생성 속도+ 잠재공간 부드럽<br>

단점) 학습 불안정 + 가짜같은 거 쉽게 구별 -> 학습 덜됨 + 비슷한 이미지만 반복 + 평가 지표 한계 + 역매핑 어렵<br>
