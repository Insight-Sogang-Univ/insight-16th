## Attention
Seq2Seq의 한계 ) <br>
- 한 개의 context vector만 사용해서 출력 문장 생성 -> 정보 손실의 문제<br>
- context vector = 인코더의 마지막 은닉상태<br>
  -> 앞쪽 정보 희미함<br>

Attention 도입 )<br>
- 디코더가 단어 생성할 때마다 입력 문장 전체 고려<br>
- 현재 단어와 관련 있는 부분 집중 -> 출력 단어마다 새로운 context vector 만듦<br>

Attention 정의 )<br>
- 가장 중요한 부분에 높은 가중치 부여하기<br>
  = 어떤 단어에 집중할 지 정도 결정<br>

- 인코더 : 단어 -> LSTM -> 단어마다 은닉 상태 생성<br>
- 디코더 : <sos> 토큰 ~ 단어 차례대로 생성<br>
          -> 다음 단어 예측<br>

순서 ) <br>
인코더 -> 현재vs은닉: 관련도(유사정도) 점수 계산 -> softmax -> context vector -> decoder&context vector -> dense -> softmax -> 단어 출력<br>

* 관련도 점수+softmax = attention weight
-> a 단어에 80% 집중할거야!
** 어텐션 가중치 만큼 은닉 상태 정보 섞음 -> context vector

장점 )<br>
- 동시 계산 <br>
- 장기 의존성 덜함 (vs RNN)<br>
- 한번에 모든 위치 직접 연결 (vs CNN)<br>
- 관련도 계산을 독립적으로 동시 계산 (vs RNN)<br>

## self-attention vs cross-attention <br>

cross-attention : = 남에게 묻기<br>
- Q from decoder<br>
- K, V from encoder <br>
-> 서로의 시퀀스를 참고해서 정보 만듦<br>
- encoder가 각 단어 별로 (key, value) 만듦<br>
-> q랑 유사도 계산할 때 k 사용 & 그걸 바탕으로 v와 가중합<br>
but, RNN 기반<br>

self-attention: = 나에게 묻기<br>
- Q,K,V from hidden state<br>
- 단어 -> 벡터 (벡터 임베딩)<br>
- Q와 K 비교 -> 관련도 점수 계산<br>
- softmax&attention weight -> 단어의 중요도 판단<br>
- 비율&value -> 새 벡터 (전체 문맥 반영)
병렬 처리 가능<br>

인코더(self) -cross- 디코더(self) <br>

## scaled dot-product Attention<br>

Q & K -> 관련도 계산 <br>
Q*K = attention score<br>
: (질문하는 단어, 참고대상 단어) <br>
-> softmax 적용<br>
-> Attention Weight 산출<br>
    = attention score * value<br>

문제 ) 벡터 차원 커지면 내적 값이 지나치게 커질 수 있음 -> 기울기 소실..<br>

해결 ) 내적-**스케일링**-softmax : <br>스케일링 추가<br>
attention score -> root(d_k) -> softmax -> Attention Weight -> Attention Weight*value -> 최종 결과<br>

## multi-head attention<br>

head = attention 계산 한 번하는 독립 단위<br>
- single head : 헤드 1개 = 가중치 행렬 세트 1개 <br>
  -> 한 단어를 이해하려면 가중치를 나눠 줘야됨 -> 정보 흐릿 (1명이 다 커버)<br>
- multi head : attention 작게 쪼개서 여러번 동시에<br>
  -> 각 헤드들이 독립적으로 질문 하나에 집중 & 결과 이어붙이기 -> 정보 또렷 <br>(8명이 분배해서 커버)<br>

흐름 ) 분할 -> 병렬 어텐션 -> 결합 -> 최종 투영<br>
(헤드 개수만큼 동시에 독립적으로 진행) <br>
- 분할 : 헤드마다 다른 가중치 행렬존재 <br>
        -> 단어*가중치(Q,K,V)<br>
- 병렬 어텐션 계산 ; 독립적으로 계산<br>
  - 점수 매기기 단어 1개(Q) & 단어 N개(K) 내적 -> 관련도 점수 N개 <br>
  - 스케일링<br>
  - SOFTMAX -> Attention weight<br>
  - 가중합 : V * 가중치 -> 헤드결과<br>

- 결합 : 헤드 개수만큼 이어 붙임 -> 한줄<br>
- 최종 투영 : 단순 나열 -> W 곱 -> 하나의 의미<br>

## 트랜스포머 전처리 <br>

흐름) input -> 토큰화 -> 임베딩 -> positional encoding<br>

- 토큰화 : 문장 -> 토큰 : I love you -> I,love, you<br>
- 임베딩 : 토큰 -> 벡터 : 뜻이 비슷한 <br>단어가 가까운 좌표에 놓임 (무작위 -> 자리 잡기)<br>
- positional encoding : 단어 순서 정보 + 임베딩 벡터<br>
  모든 단어 한꺼번에 봄 -> 순서 알려주기<br>

## encoder  : 정보 이해<br>
= multi-head -> feed-forward network<br>

multi-head self attention <br>
= 원문 단어끼리 서로 보며 문맥 파악<br>

feed-forward network <br>
= 단어 하나하나 따로 가공<br>
- 같은 가중치, 단어별 독립 처리<br>

- Norm : 층 정규화 평균 0 , 분산 1 근처로<br>
