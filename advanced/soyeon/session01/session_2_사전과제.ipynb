## 인공신경망

<br>

## 퍼셉트론
= **선형** 이진 분류기
입력*가중치 -> 활성화 함수 -> 출력
<br>
총합 > a -> 1, < a -> 0
<br>

- 가중치:<br>
  입력값의 중요도 -> 가중치 값 크면 중요한 거
- 활성화 함수: <br>
  ex) 계단함수, sigmoid, Tanh, ReLU,..
- 임계값: <br>
  theta =  기준 값
- 편향:<br>
  b=-theta

## 단층 퍼셉트론
= only 입력층&출력층

- 논리게이트(Logic Gate) 구현<br>
n개 입력값 -Gate-> 1개 출력값<br>
ex) AND <br>
y= 0.5x_1+0.5x_2 <br>
AND니까 둘다 1일 때만 y=1 <br>

- 한계: XOR 게이트 구현 X <br>
퍼셉트론 = 선형결합, XOR 게이트 = 비선형 -> 구현 X <br>
계단함수(비선형)써도 임계값 한번 적용해서 직선임 <br>

## 다층 퍼셉트론 (MLP)
= 단층 + 은닉층 -> 새로운 특징 조합 <br>
-> XOR 구현 가능 <br>


# 딥러닝 학습
## 순전파 & 역전파

### 학습 알고리즘
- 입력값 -> 예측 by 순전파<br>
  ㄴ x->w1->w2->y(예측)
- 예측 정확도 측정 by 오차계산<br>
  ㄴ L=(y_exp - y)^2
- 출력층->입력층 오차 전달 by 역전파<br>
  ㄴ 연쇄법칙
- 가중치 업뎃<br>
  ㄴ 정확도 up<br>
  ㄴ 경사하강법
- 반복 <br>

--> 오차값 ~~ 0

### 연쇄법칙
합성함수 미분 = 각 함수 미분의 곱 <br>
각 가중치가 최종 손실에 얼마나 기여했는 지 <br>
-> 손실 함수를 줄이기 위해 가중치를 어떻게 바꿀까?<br>

### 기울기 소실 문제
은닉층 많음 + 미분값<1 -> 곱할 수록 0에 가까워짐<br>
-> 앞쪽 측 가중치 업뎃 x = 학습 제대로 x

- 목적함수 <br>
= 달성하고 싶은 목표 (maxi/mini)<br>

- 손실함수<br>
= **개별** 데이터에 대한 오차 측정 + 최소화<br>
예측값-실제값 -> 모델이 얼마나 틀렸는 지<br>
ㄴ *MSE* : 1/N sum(y-y_pred)**2 <br>
  연속적, 회귀에 적합, 단순히 loss 크기<br>
ㄴ *BCE* : L=-(ylog(y_pred)+(1-y)log(1-y_pred))<br>
  이진분류, 실제 정답에 -ㅣ=log -> 정답일 수록 작음, 확신 정도->loss<br>
  (log : 확률 높이기 -> 손실 줄이기)<br>
  (- : 0~1일 땐 log에 넣으면 음수 나와서)<br>
ㄴ *CEE* : - sum(y log (y_pred))<br>
  다중 클래스 분류(예측 확률이 실제 클래스와 얼마나 일치하는 지)<br>

- 비용함수<br>
= 모델의 **전체**적인 성적 평가 + 최소화<br>
-> 비용함수로 현재 성적 계산<br>
-> 비용 줄어드는 방향으로 모델 수정<br>
-> 반복<br>


## 활성화 함수 <br>
(wx+b) -> **활성화함수** -> 출력값 <br>

- 비선형성<br>
- 미분가능<br>
  ㄴ =역전파 가능<br>
  계단함수 : 미불 : 역전파 x<br>
  시그모이드: 미가 : 기울기 소실<br>
  ReLU : 미가+ 양수구간=1 : 기울기 소실 x<br>

가중치 = 입력값이 결과에 얼마나 영향<br>
편향 = 출력값 조정, 기준점 역할<br>

- 계단함수<br>
: y = 1 (x>=0), 0(x<0)<br>
미분 값 0 -> 역전파 x<br>
- 시그모이드<br>
: y = 1/(1+e**(-x))<br>
y= 0~1 : 연속적으로 변화 -> 얼마나 확실한지<br>
미가 but, 미분값 최대 0.25 -> 기울기 소실<br>

- ReLU<br>
:y=max(0,x)<br>
계산 간단, 비선형성 제공, 기울기 소실 완화<br>(양수의 미분값 항상 1) but, deadneuron(음수에선 0)<br>

- softmax<br>
: y_i=e**x_i / sum(e**x_i)<br>
출력값을 확률로 변환, 다중 크래스 분류 <br>문제의 출력층, 전체 합=1 <br>
softmax + CEE -> 다중분류<br>

- tanh<br>
: y= (e**x-e**(-x))/(e**x+e**(-x))<br>
y: -1~1 : 중심이 0이라서 업뎃 안정적 , 기울기 소실<br>



## 경사하강법 <br>
-> 손실 작게 만드는 W 찾기<br>

접선 기울기 > 0 : 왼쪽으로 이동<br>
접선 기울기 = 0 : 멈춤<br>
접선 기울기 < 0 : 오른쪽으로 이동<br>

W=W-alpha(aE/aW)<br>

- 학습률 (alpha)<br>
= 보폭<br>
작으면 오래 걸림, 크면 발산<br>
보통 0.001<br>

- 배치<br>
= 데이터 묶음 (매개변수 조정용)<br>
크면 학습 느려짐 + 메모리 부족<br>
작으면 가중치 업뎃 잦아서 학습 불안정<br>

- 에포크<br>
= 데이터셋 학습하는 횟수<br>

- 이터레이션/스텝<br>
= 한 번의 에포크를 끝내기 위해 필요한 배치 수<br>
-> 데이터 = 배치크기*이터레이션<br>


- 배치 하강법<br>
: Batch size=Data size<br>
업뎃 횟수 적음 , 메모리 많이 필요, 수렴 안정적<br>
단점) 시간 오래 걸림<br>
- 확률적 경사하강법 (SGD)<br>
랜덤으로 선택된 하나의 데이터만 계산<br>
-> 자원 적은 컴도 가능, but, 보폭 불안정+정확도 낮을 수도<br>
- 미니 배치 경사하강법<br>
배치 크기 지정 -> 안정적+빠름<br>

## 옵티마이저 <br>

기존 한계) 경사하강법 -> 기울기 0이면 멈춤<br>
but, local minimum인지 global minimum인지 모름<br>

- RMSProp<br>
기존 한계) 큰 방향에선 마니 작은 방향에선 조금 이동<br>
-> 각 방향마다 학습률 따로 조정<br>
ex) 급한 경사에선 속도 줄이기..<br>

- Adam <br>
= 방향+학습률<br>
