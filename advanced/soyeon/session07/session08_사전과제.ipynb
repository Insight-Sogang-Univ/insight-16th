## 시계열 데이터 <br>
시간의 흐름에 따라 일정 간격, 순서대로 관측하고 기록된 데이터 <br>

조건)  <br>
- 시간 순서에 의미 존재 <br>
- 관측 간격 일정 (주기) <br>

시간에 따라 몇개의 변수 관찰하냐 <br>

- 단변량 시계열 <br>
: 시간에 따라 하나의 변수만 관측 <br>
-> 자기 자신의 과거 값만 보고 미래 예측 <br>
ex) 주식 <br>

- 다변량 시계열 <br>
: 시간에 따라 여러 변수를 동시에 관측 <br>
-> 자기 자신의 과거 + 다른 변수들의 영향 함께 고려 <br>
ex) 아메리카노의 판매량 (기온+강수량+요일) <br>
-> 변수가 다다익선은 아니고 미래 시점에서도 그 값을 알 수 있는 지를 따져야 함 <br>

시계열 분석) <br>
시간 순서대로 쌓인 데이터에서 패턴 찾기 + 미래 값 예측하기 <br>

시계열 주 목표)  <br>
미래 데이터 예측 : 지금까지의 기록을 보고 다음 시점의 값 추정 <br>
- 기본 가정 : 과거의 패턴이 미래에도 어느정도 반복될 것이다. <br>
- 순서 : 시계열 이해 -> 전처리 -> 정상성 판단 -> 모델링&예측 <br>

활용 방안) <br>
금융, 에너지, 기상...  -> 예측을 바탕으로 미리 결정하기 위해 사용 <br>

## 시계열 데이터의 특징 <br>
- 시간 의존성 : 과거 -영향-> 현재 <br>
: 데이터가 시간에 따라 변하고, 현재 시점의 값이 과거 시점의 값에 영향을 받음 <br>

특징) 과거 -영향-> 현재, 가까운 과거일 수록 영향 up, 순서 섞으면 안됨 + 데이터 누수 주의 <br>

-> 계절성(일정하게 반복) / 추세 (전반적으로 증/감) <br>

-- 계절성 : 언제마다 반복? <br>
일정한 주기로 비슷한 패턴이 반복 <br>

-- 추세 : 어느 방향으로 감? <br>
시간의 흐름에 따라 데이터가 장기적으로 증/감 하는 방향성 (멀리서 본 거) <br>

- 자기상관성 <br>
: 과거 -영향-> 현재 : 영향이 얼마나 강한 지 숫자로 나타냄 -> 과거값으로 지금 값 짐작 가능할 정도 <br>
-> 자기 자신끼리 상관관계를 갖는 특성 <br>
상관관계) 서로 다른 변수 사이 재기 <br>
자기상관) 같은 변수의 현재와 과거 비교 <br>
(상관관계 != 자기 상관) <br>

특징)  <br>
-- 한 시계열 내에서 현재랑 과거가 얼마나 비슷하게 움직이는 지의 정도 <br>
-- 일정한 패턴이나 반복이 존재함 <br>
-- -1(반비례) ~ 1 (비례) , 0에 가까우면 과거랑 지금이 관계없는 거 <br>
-- 시차(현재값을 몇 시점 전의 값이랑 비교) 바꿔가면서 계산 -> 반복주기 찾을 수 있음 <br>

문제점) <br>
-- 회귀 분석 불가 : 데이터끼리 묶여있어서.. <br>
   ㄴ 회귀는 데이터가 서로 독립적이라는 가정 하에 진행됨 <br>
   - 오차항 독립성 가정 위반 -> 시계열은 오늘의 값이 어제의 값을 따라가서 오차도 어제를 따라감 -> 통계 결과 왜곡 <br>
   - 허위 상관관계 문제 -> 시간에 따라 증감하면 아무 관계 없어도 상관관계가 높게 나옴 -> 추세때문이 아닐까 고려하기 <br>
   - 예측 어려움 & 모델 복잡성 증가 -> 과거 관계까지 모델에 넣음 -> 예측 오차 쌓임 -> 패턴 바뀔 수도 있음 <br>

## 시계열 데이터 분석 요인 <br>
- 추세 <br>
  : 시계열이 시간 전반에 걸쳐 증/감하는 경향성 나타내는 요인 (큰 흐름) <br>

- 계절 <br>
  : 1년 이내의 일정한 주기로 같은 패턴이 반복되는 요인 (규칙적으로 돌아옴)  <br>
  연, 월, 주 주기 <br>

- 순환 <br>
  : 계절성보다 길고 불규칙한 주기로 반복되는 변동 <br>
  일정하지 x , 언제 올지 모름 -> 계절과의 차이점 <br>

- 불규칙 <br>
  : 시계열에서 추세, 계절, 순환요인 모두 제거하고 남은 나머지 움직임(잔차 , 노이즈) <br>
  ex) 유명인이 다녀온 후 카페의 매출 급증 <br>

실제 시계열엔 4가지 요인이 모두 섞여 있음 <br>

## 시계열 전처리  <br>
필요성) <br>
- 결측치, 이상치, 노이즈 -> 큰 흐름 보는 거 방해 <br>
- 이런 데이터를 모델에 넣으면 잘못된 패턴 학습하게 됨 <br>
-> 시계열 특성을 반영한 전처리 필요 <br>

전처리 프로세스) <br>
데이터 확인 -> 결측치 탐지, 처리 -> 이상치 탐지, 처리 -> 평활화 -> 요소 분해 -> 정상성 판단 & 확보 <br>

- 결측치 <br>
: 특정 시점의 값이 수집되지 않았거나 누락
-> 지우면 관측 간격 깨짐 + 시간 흐름 무시됨 -> 채워야 함! <br>

-- 처리법) <br>
삭제/단순 대치/보간/이동평균 대치/머신러닝 기법 <br>

---- 보간법) <br>
두 관측값 사이의 비어있는 값 추정하기 <br>
= 두 점 연결하는 규칙 정해서 값 추정하기 <br>
- 선형 보간법 <br>
  선형적인 데이터일 때 적합 <br>
- 다항식 보간법 <br>
  다항식 함수로 근사 -> 비선형 데이터에 유용  <br>but, 차수 높아질 수록 과적합 up <br>
- 스플라인 보간법 <br>
  구간별 다항식 -> 곡선형 데이터에 유용 <br>
- 조각별 상수 보간 <br>
  각 구간에서 동일한 상수 값 유지 -> 결측치 채움 <br>
  -- 이전 값 유지 : 직전 관측치 값으로 결측치 대체 <br>
  -- 다음 값 유지 : 직후 관측치 값으로 결측치 대체 <br>
  -- 최근접 이웃 보간 : 앞/뒤 관측값 중 거리상 가까운 값 선택 <br>

보간법 vs 회귀) <br>
보간법 : 선이 점 지나감 , 회귀 : 점들 사이를 잘 지나가는 선이라서 지날수도 안지날수도 <br>

## 시계열 이상치<br>
: 데이터의 일반적인 패턴에 크게 벗어난 극단적인 값<br>

- 값의 크기 + 추세 + 계절성 (언제 나타났는 지)<br>

탐지법)<br>
- IQR : 이 범위에서 멀리 떨어진 값을 이상치로 판단<br>
  비교적 안정적인 구간 값에서 이상치 탐지 시 사용<br>

- Z-score : 평균에서 얼만큼의 표준편차만큼 떨어져 있는 지 계산<br>
  계절성 약한 or 값의 변동폭이 일정한 시계열에서 효과적임<br>

- STL : 시계열을 계절(S), 추세(T), 잔차(R)로 분해<br>
  잔차가 특정 임곗값 초과하면 이상치 탐지 + 계절성/추세가 뚜렷한 시계열에 적합<br>


이상치 탐지 모델)<br>
- CART : 데이터를 조건에 따라 계속 나눠가면서 예측하는 의사결정나무 모델<br>
  과거 구간의 값 = 입력 -> 다음 시점의 값 예측 <br>-> 예측 오차 기반으로 이상치 판단  <br>
  비선형 관계 포착 잘함 but,,,, 노이즈 많은 건 과적합되기 쉬움, 트리 깊이에 따라 탐지 민감도 달라짐<br>

- 클러스터링 기반 : 군집화 후 어느 군집에도 속하지 않는 데이터를 이상치로 판단<br>
  K-means, DBSCAN<br>
  시계열의 특징 추출 -> 각 시점을 벡터로 표현 -> 클러스터링<br>
  비지도 학습 + 패턴 전체가 이상한 경우도 잡을 수 있음 but, 하.파 설정에 따라 결과 마니 달라짐<br>


- AE: 작게 압축했다가 원래대로 복원<br>
  -> 이상치가 들어오면 복원 제대로 안되고, 오차 커짐 -> 이상치 판단<br>
  복잡한 비선형 패턴, 다변량 시계열 다룰 수 있음<br>

처리방법)<br>
- 삭제 : 이상치 제거 / 근데 보통 결측치를 채우는 편<br>
- 단순 대치 : 평균, 중앙값 으로 대체 but, 추세, 계절성 무시<br>
- 보간 : 이상치 -> 결측치 , 앞 뒤로 값 보간, 시간의 흐름 반영할 수 있음<br>

but, 노이즈 남아있음<br>

# 평활화<br>
시계열 데이터의 자잘한 노이즈 줄여서 데이터 부드럽게 만듦<br>
-> 데이터의 큰 흐름이 드러남 (EDA에 주로 사용)<br>

- 이동평균)<br>
일정한 개수의 연속된 데이터를 묶어 평균 냄 -> 묶음을 한 칸씩 옮기기 -> 평균 계속 계산<br>
N: 한번에 평균 낼 데이터 개수 (묶음의 크기)<br>
N이 클수록 더 긴 기간의 데이터를 평균에 포함시킴<br>
-> N이 크다고 좋은 건 x , 보고 싶은 흐름의 길이에 맞게 정하기<br>

--- 목적)<br>
노이즈 제거&추세 파악, 계절성 제거, 결측치/이상치 대치, 예측의 기초 자료<br>

--- 유형)<br>
- 단순 이동평균 : 묶음 안의 모든 값에 똑같은 가중치 주고 평균 내기<br>
-- 계산 직관적 but, 변화에 늦게 반응함<br>

- 가중 이동평균 : 최근 값일수록 더 큰 가중치 줌<br>
-- 변화에 빠르게 반응 but, 가중치를 어떻게 줄지 직접 정해야 <br>

- 지수 이동평균 : 모든 과거 값을 조금씩 반영<br>
-- 변화에 빠르게 반응, 과거 정보 모두 반영 but, 알파 값 정해야 하고, 직관적은 x<br>

=> 평활화 = 노이즈 줄여서 큰 흐름 보는 기술, N이 클수록 매끄러워짐 but, 변화에 늦게 반응, 최근 값에 가중치 줄수록 변화에 빠르게 반응<br>

## 요소 분해 <br>
시계열 = 추세+계절+순환+불규칙 <br>
-> 섞인 걸 각 요인으로 나누기<br>

-> 각 패턴 명확한 이해 가능 + 변동의 원인 분석 가능 + 정교한 분석 가능<br>

## 분해 모형 (결합 방식)<br>
= 어떤 식으로 결합되어 있는가<br>
- 가법 모형:<br>
  시계열 = 추세+계절+잔차<br>
  -> 각 변동 요인들이 독립적으로 더해짐+모든 요소의 영향 일정<br>
  진폭 일정할 때 적합<br>

- 승법 모형:<br>
  시계열 = 추세x계절x잔차<br>
  추세 up 계절성/불규칙 up<br>
  진폭 시간에 따라 달라질 때 적합<br>

## 분해 방법 (구현 알고리즘) (계산법)<br>
= 실제로 각 요인을 어떻게 계산해 낼 것인가?<br>

- MA 기반 분해 : Seasonal Decompose <br>
이동평균(MA) -> 추세 -> 계절 -> 잔차 분리<br>
추세 추출 후 제거 후 계절성 추정 순<br>

- LOESS 기반 분해: STL <br>
각 지점마다 주변 데이터만 사용해서 작은 회귀선 <br>그리고 부드러운 곡선 만들기<br>
-> 구간마다 유연하게 맞추기<br>

Inner Loop : 추세와 계절 번갈아 추정<br>
추세 제거 -> 계절 요인 추정 -> 계절 요인 제거 -> 추세요인 추정 -> ...<br>

Outer Loop : 이상치 영향 줄이기<br>
Inner Loop 끝나면 잔차 계산 -> 잔차 크면 작은 가중치 부여 -> Inner Loop 수행<br>


## 정상성 <br>
= 통계적 특성이 시간에 따라 변하지 않는 성질<br>
-> 어느 시점에서 잘라 보더라도 비슷한 모습<br>

통계적 특성 : 평균, 분산, 자기상관 구조..<br>

과거의 패턴이 미래에도 반복될 것을 가정 -> 예측 가능<br>
-> 모델링 전에 정상성 판단하고 확보하는 과정이 필요함<br>

- 강한 정상성)<br>
시계열의 확률 분포 = 시간에 따라 전혀 변하지 않는 성질<br>
-> 시점 이동이 일어나도 분포랑 같아야 함<br>

- 특징) 왜도, 첨도 등 분포의 모든 특성이 시간에 따라 변치 않고 일정함 <br>
- 한계) 현실엔 강한 정상성의 데이터는 없음<br>

- 약한 정상성)<br>
평균, 분산, 자기공분산만 시간에 따라 변하지 않으면 됨<br>
- 특징) 평균 일정+분산 일정+자기공분산이 시차에만 의존<br>
-> 보통 현실에선 약한 정상성의 데이터 존재<br>

- 정상성 판단하기:<br>
-- 시각적 : 그래프 (평균 중심 오르내림 + 폭은 시간에 상관없이 비슷 + 평균 회귀 + 특정 주기로 반복 x)<br>

-- 통계적 검정 : 객관적 판단<br>
단위근 : 시계열을 표현하는 식의 해가 1이 됨 = 과거 충격이 영구적으로 누적 <br>
-> 단위근 있으면 비정상 시계열<br>

단위근의 시계열적 의미 : 충격이 쌓이면(갑자기 큰 값으로 튀거나..) 평균으로 돌아오지 않음 + 분산이 계속 커짐<br>
-> 평균과 분산이 시간에 따라 변해서 약한 정상성의 조건이 깨짐<br>

단위근 검정 방법:<br>
단위근이 있는 지 여부 판단 <br>
- ADF 검정<br>
  시계열이 과거 값에 얼마나 의존하는 지 측정<br>
- KPSS 검정<br>
  시계열이 일정한 평균 수준을 유지하는 지 측정<br>

-> 두 모델 사용시<br>
A 정상 K 정상 : 정상 시계열<br>
A 비정상 K 비정상 : 비정상 시계열<br>
A 정상 K 비정상 : 차분 정상 시계열 -> 차분 후 재검정<br>
A 비정상 K 정상 : 추세 정상 시계열 -> 추세 제거 후 재검정<br>

- 추세 제거: <br>
시계열이 장기적으로 한 방향으로 계속 오르 내림 <br>-> 시간에 따라 평균이 변하는 경우에 적용<br>
차분 ) <br>
오늘-어제 -> 얼마나 변했냐 (1차)<br>
1차 차분에 또 차분 (2차)<br>
차분한 값 되돌리기 (역차분)<br>
-> 추세 사라짐 , 단위근 사라짐 <br>
-> 정상 시계<br>

- 계절성 제거:<br>
일정한 주기마다 같은 모양을 반복 -> 시점에 따라 평균이 달라짐 <br>
계절 차분 : 오늘 - 한 주기 전의 값 (작년의 같은 때와 비교)<br>

- 분산 안정화:<br>
시계열의 오르 내리는 폭이 시간이 지날수록 커지거나 작아짐 -> 분산 일정하지 않은 경우 적용 <br>

-- 로그 변환 : 데이터 + 로그 -> 값의 크기 줄이기<br>
but, 0 이나 음수 사용 x , 모든 분산 변화 해결 x<br>


## 단변량 시계열 분석 <br>

- 안정적 시계열 분석 (정상) : 돌아갈 평소가 있는 시계<br>

- 자기 회귀 모델 : 오늘 값을 과거의 값으로 설명 -> 자기상관성 활용하기<br>

AR(p) 과거 p개 시점의 값 사용<br>
AR(1) 바로 전 시점의 값 하나만 사용<br>
-> 오늘 판매량 = 어제 판매량 일부 + 오늘의 충격<br>

- 이동 평균 모델 : 오늘 값을 과거의 충격으로 설명<br>
MA(q) 과거 q개 시점의 충격 사용<br>
MA(1) 바로 전 시점의 충격 하나만 사용<br>

AR, MA 둘다 정상성, 가역성이 필요함<br>
AR : 정상성 계수 조건 필요<br>
MA : 가역성 계수 조건 필요<br>
가역성 : 지난 충격을 거꾸로 계산해낼 수 있는 성질<br>
오늘 충격 = 오늘 값 - 계수 x 어제 충격<br>
-> 과거의 기록으로 충격을 왜곡 없이 <br>안전하게 되돌려 계산할 수 있을 지 확인<br>

- ARMA 모형<br>
오늘 값 : 과거 값 & 과거 충격 영향<br>
AR과MA를 한식에 합침 (ARMA(p,q))<br>
특징) AR과 비교했을 때 더 적은 계수만으로 장부 설명 가능 , AR부분이 정상성 조건을, MA부분이 가역성 조건을 만족해야 함<br>

- 차수 결정 법 : 자기 상관 함수 (ACF)<br>
-- 자기 상관 계수 : 오늘 값과 k 시점 전 값이 얼마나 닮았는 지를 상관계수로 잰 것 (ARMA에서 q 결정)<br>
-- 부분 자기 상관 계수 (PACF): 중간 시점들의 영향 빼고, 오늘 값과 k 시점 전 값이 얼마나 닮았는 지 잰 것 (ARMA의 p 결정)<br>

막대가 급격히 줄어들기 직전의 시차 고름 -> PACF에서 고르면 AR의 차수 p, ACF에서 고르먄 MA의 차수 q<br>

불안정 시계열 분석 = 정상성 없는 시계열<br>
비정상 시계열 데이터 -> 정상성 확보 -> 정상 시계열 -> ARMA 사용 안정적 시계열 분석<br>

- ARIMA : ARMA + 차분 <br>
차분 -> 증감 만들기 -> 증감을 ARMA 예측하기 -> 예측한 증감을 어제 판매량에 더하기 <br>

ARIMA 모형이 특수한 경우<br>
- 랜덤워크 : 상수가 없는 ARIMA(0,1,0)<br>
내일 값의 최선의 예측 = 오늘 값 <br>

-- 표류를 포함하는 랜덤워크 : <br>
  표류 : 매일 보통 얼마씩 변하는 가<br>
  -> 오늘 값 = 어제 값 + c + 오늘 충격<br>

SARIMA (seasonal ARIMA)<br>
ARIMA + 계절 차분 + 계절 AR, 계절 MA<br>
-> ARIMA 에 작년 같은 달 대비의 증감을 더한거임<br>

불안정 시계열에서 하파 결정하기 <br>
- 차분, 계절 차분 : ADF 로 정상성 확보 여부 판단 후 결정<br>
- AR, MA 차수 : ACF, PACF 로 범위 추정 -> 그리드 서치 -> 최종 선택<br>

## 딥러닝을 활용한 시계열 분석<br>
전통 통계 모델의 한계 )<br>
- 데이터 많으면 처리 속도 느려짐 <br>
- 크기 고정된 데이터만 가능<br>
- 비선형 패턴 포착 어려움<br>

-> RNN 계열<br>
+: 순차적 패턴 중요한 시계열 데이터 , 과거 정보가 미래에 영향 줌, 시계열 길이와 목적에 따라 모델 선택 가능 <br>
-: 매우 긴 시계열엔 학습 어렵, 병렬처리 효율 낮음<br>

-> Transformer 계열<br>
+: 장기 예측 특화, 주기성 포착 특화, TFT (과거 데이터+외부 변수 섞어서 예측)<br>

-> Foundation Model <br>
: 거대한 데이터로 사전 학습 -> 다양한 문제로 해결할 수 있는 범용 모델<br>
