## 추천 근거1) 내용기반필터링 <br>

필요한 거)<br>
사용자가 뭘 좋아하는 지 + 아이템끼리 얼마나 비슷한지 <br>

흐름) <br>
아이템 -> 숫자 & 사용자 프로필 생성 -> 유사도 계산 -> 점수 높은 거 추천 <br>

- 사용자 프로필 <br>
: 사용자가 어떤 특성 좋아하는 지 정리<br>

각 사용자의 아이템별 특성 가중치의 평균 값 -> 피드백(명시/암시)으로 기호 파악 -> 평가 예측 분류 문제 -> 사용자 프로필 획득<br>

- 유사도 <br>
: 둘이 얼마나 비슷한 지<br>

상품 -> 숫자 벡터 -> 유사도 계산 (벡터 내적)<br>

- TF-IDF <br>
: 한 문서에는 자주 나오고 다른 문서에는 드물게 나오는 단어에 높은 값 줌<br>
-> '그리고' 같은 단어가 중요하다고 고려되지 않도록<br>

-- TF : 단어 빈도up TF up<br>
-- IDF : 해당 단어가 등장한 문서 up IDF down<br>
-> TF-IDF up = 중요도 up<br>


+) 속성만 존재하면 추천 가능, 다른 사용자 데이터 필요 x , 추천 근거 설명 가능<br>

-) 속성 정보만으론 정확도 등 한계 존재, 새로운 장르 발견 어려움, 기록 없으면 추천 <br>

cold start problem : 정보 x -> 추천 시작 못함<br>
- 아이템) 평가 없어도 내용 기반으로 해결 가능<br>
- 사용자) 기록 없으면 내용 기반으로도 해결 어려움<br>

## 딥러닝 기반 추천 시스템<br>

- VAE 기반 협업 필터링<br>
사용자 상호작용 기록 -> 확률분포로 압축 -> 복원 : 아직 보지 않은 아이템의 선호 예측<br>

입력 -> 인코더 (확률 분포 생성) -> z 분포 -> 가우시안 샘플링(무작위 z택) -> 디코더(z로 부터 data 생성) -> y출력<br>

+) 불확실성 표현, 희소한 상호작용 데이터에서 안정적<br>

MF 한계)<br>
- 선형성 : 행렬 분해의 예측 -> 내적 -> 요인끼리 얽히는 효과 표현 어렵<br>
- 불확실성 무시 : MF : 점 추정 (벡터 하나 학습) ->  기록 적은 사용자의 추정이 불안정해도 구분 x..<br>

딥러닝) <br>
- 선형성 -> 비선형 활성화 함수를 가진 신경망 사용 <br>
- 불확실성 무시 -> 평균과 분산 함께 출력<br>

=> VAE 사용<br>
why? AE는 점추정임<br>
VAE는 주변 학습(분포) & 기준 분포쪽으로 당김 -> 기록 몇개만으로 극단적인 위치에 가지 x<br>

- Mult-VAE<br>
사용자의 상호작용 기록 입력 -> 인코더(취향을 분포로 요약) -> 디코더 (아이템 선호확률 복원) -> 추천<br>

-- 목적함수 ELBO<br>
: 데이터가 나올 로그 확률 하한 키우기<br>
1. 복원항 : z로 x와 비슷하게 복원<br>
2. KL divergence 항 : 과적합 막고 잠재공간 불규칙해지지 않게 연속적이고 부드럽게 유지<br>
(두 확률 분포가 얼마나 다른 지 -> 지나치거나 중심에서 멀어지면 벌점 up)<br>
--> 잠재 공간을 정규 분포 근처로 압축 -> 부드럽 -> 희소한 데이터에 강함<br>

=> VAE : 비선형 관계 학습 + 사용자를 분포로 표현 -> 기록 적은 사용자에게도 안정적 추천 가능 !<br>



## GNN 기반 사용자 선호 예측<br>
기존 한계)<br>
CF, MF,VAE는 사용자-아이템 상호작용을 행렬 칸 하나하나로 다룸 -> 여러 상호작용으로 이어진 관계는 개별 칸만으로는 직접 활용 x,,, 행동 순서도 고려 x<br>

GNN) <br>
그래프 -> 경로 : 사용자-아이템 상호작용 <br>
= 서로 연결된 거대한 그래프<br>

그래프 : node --w-- node <br>
- 추상적 개념 다루기 적합 : 관계 추측 x<br> 관계 자체를 직접 학습 가능<br>
- 비유클리드 공간 표현 & 학습 <br>
그래프 = 격자 구조 x -> 노드&엣지<br>

작동구조)<br>
그래프 구조 생성(노드, 엣지 정의 & 초기 피처 부여) -> 메시지 패싱(타겟 노드 선정& 메시지 생성&집계(임베딩 요약본)&갱신(원래 임베딩&집계결과 -> 새정보)) -> 최종 출력
: 내 이웃 노드의 정보를 나에게 전달/요약<br>

프로세스)<br>
노드 = 자신의 이웃 노드들의 정보 수집 -> 은닉 상태 업데이트
-> k번 반복 -> 더 멀리 있는 이웃 정보까지 반영 -> 이웃들의 취향 반영한 최종 임베딩 얻기 -> 두 노드 간 선호도 예측<br>

주요 GNN 아키텍쳐)<br>
- GCN : 그래프 위에서 convolution 연산 -> 파라미터 : 이웃노드=타겟노드<br>

- GRN : RNN+GNN -> 시간에 따라 노드 특성 업뎃 -> 그래프의 동적 변화 포착<br>

- GAT : 이웃 노드간 중요도 학습 + 가중 평균<br>

- GraphSAGE : 대규모 그래프 -> 샘플링 기반으로 학습 (concat)<br>

- LightGCN <br>
-- 메시지 어떻게 만드는 지 : 가중치 행렬 곱/attention 쓰기<br>
-- 메시지 어떻게 합치는 지: 평균/가중 평균/샘플링<br>

등장 배경)<br>
NGCF : 노드 분류용 GCN 연산 -> 레이어마다 가중치 곱 + 비선형 활성화 함수 적용<br>
=> 노드에 ID임베딩만 존재 -> 학습 방해<br>
=> 두 연산 제거 + 사용자-아이템 간 연결 관계만 전달하도록 구조 단순화 (특징변환, 비선형 활성화 함수 제거)<br>
==> LightGCN
