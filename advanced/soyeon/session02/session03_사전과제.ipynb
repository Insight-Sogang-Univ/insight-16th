## CNN (합성곱 신경망)<br>
컴: 이미지=수많은 픽셀에 저장된 숫자 데이터<br>
이미지 -> 1차원 배열 : 이미지 특징 학습 어려움<br>

MLP : FClayer+..+FClayer<br>
 ㄴ 입력 : 2차원->1차원<br>
 ㄴ FCLayer 거듭하면서 학습<br>

한계 ) <br>
- 위치, 크기, 회전 등 변화 취약 : 인지x -> 성능 down<br>
- 파라미터 폭발 : 'FC'라서 과적합, 느린 학습 등의 단점도 있음<br>
- 공간정보 손실 : 2차원 구조 명시적 유지 x<br>

이미지 구조 유지 필요:<br>
  픽셀 간 공간관계 + 위치 정보<br>

//<br>
커널 = 특정 패턴 추출용 matrix<br>
필터 = 커널 묶음<br>
피처맵 = convolution 연산 후 얻은 결과 데이터 , 차원은 필터의 개수 영향 받음<br>
패딩= cov 연산 전 가장자리에 추가 픽셀 두르기 -> 가장자리에 필터 덜가는 거 완화<br>
stride = 보폭, 연산 효율 조절<br>
  1이면 세부적 특징 추출 유리<br>
  2이상이면 연산량 줄임->피처맵 크기 줄임<br>

CNN : Conv+Pool+ ..+flatten+FClayer..<br>
  ㄴ 특징 추출 : Conv+Pool<br>
    ㄴ convolution layer <br>
      이미지에서 필터 이동 -> 이미지 특징 추출<br>
      작은 영역에만 필터 적용 = 부분 연결 구조<br>
      -> 이미지 크기 커져도 필터의 크기&개수만 동일하면 됨<br>
      
      방법) 필터가 이미지 위에서 움직임 <br>
      -> 각 위치에서 요소별 곱셈+계산<br>
      -> 피처맵에 저장<br>

    ㄴ Pooling Layer<br>
      특징 맵 크기 줄임, 중요 특징 유지, 연산량 감소<br>
      작은 위치 변화에 대한 민감도 낮춤 -> 특징 안정적 인식 도움<br>
        ㄴ max pooling : 영역 내 가장 큰 값<br>
        ㄴ average pooling : 영역 내 평균<br>

        장점) 차원 축소, 위치변화 노상관, 과적합 방지, 노이즈 영향 적어짐<br>
        단점) 정보 소실<br>
        
  ㄴ 1차원 변환 : Flatten<br>
    특징맵을 1차원으로 변환<br>
    특징 먼저 학습 후 flatten해서 1차원이여도 됨<br>
  ㄴ 이미지 분류 : FClayer<br>
    특징 종합 -> 이미지가 어떤 클래스에 속하는 지 분류<br>

## RNN<br>
순차 데이터 : 순서에 의미 존재<br>
시간vs비시간, 수치vs범주<br>

정형 vs 순차<br>
정형 : 순서에 의미 x <br>
순차 : 순서에 의미 o<br>

기존 한계) 정형 데이터가 전제엿음 -> 순차 데이터 처리에 한계 존재<br>
-> RNN!!! <br>

입력층 -> 은닉층 -> 출력층<br>
현재 시점 입력값&이전 시점 은닉 상태 -> 현재 은닉 상태<br>

은닉 상태 계산 시 tanh 사용<br>
출력 계산시 <br>
  ㄴ 이진 분류 : sigmoid<br>
  ㄴ 다중 분류 : softmax<br>

RNN 아키텍처<br>
  ㄴ one to many : 1->n : 이미지 캡셔닝<br>
  ㄴ many to one : n->1 : 스팸 메일 분류, 감성분류<br>
  ㄴ many to many : n->n : 객체 태깅, 기계번역 , 요약...<br>

RNN 한계)<br>
- 역전파 과정에서 기울기 소실 -> 학습 정체<br>
- 시퀀스가 너무 길면 기울기 폭주<br>
- 순차적으로 진행 -> 훈련 시간 up<br>

= 장기의존성 문제<br>
: 시퀀스에서 관련 정보 멀면 그 관계 기억 잘 못함 + 학습 못함<br>

기존 한계 ) 장기의존성 문제 <br>
필요한 부분) 중요 정보 오래 기억+ 불필요한 정보 잊기<br>
-> gate (정보 얼마나 통과 시킬지) 도입<br>
## LSTM<br>
사용 gate:<br>
  ㄴ forget : 장기기억에서 필요없는 내용 삭제<br>
  ㄴ input: 중요한 거 장기기억에 저장<br>
  ㄴ output: 장기기억에서 필요한거 단기기억에 내보내기<br>

상태 2개 <br>
  ㄴ 장기기억: 셀<br>
    ㄴ 기억 업뎃 : 이전의 기억+새정보<br>
  ㄴ 단기기억+출력 : 은닉<br>

정보 정교하게 관리 가능 but, 복잡<br>
+ :장기의존성 학습 가능, 기울기 소실 완화<br>
- : 메모리 마니,,, 병렬 x<br>

## GRU<br>
사용 gate: <br>
  ㄴ update: 과거:새=n;m<br>
    = LSTM의 forget+input<br>
  ㄴ reset :: 과거 기억 얼마나 무시or참고<br>

게이트랑 상태가 줄어서 단순 but, 정교함은 떨어짐<br>

+ : gate 줄었음 : output gate없음<br>
-: gate를 합치다 보니 장기기억이랑 출력 분리 못함<br>

사실 둘은 비슷해서 해결 문제에 따라 모델 선택하기<br>
정교함 : LSTM, 효율 : GRU<br>
