

## 1. NLP와 언어모델

### NLP


- 사람이 쓰는 말(=자연어)을 컴퓨터가 분석하고 이해하고 만들어낼 수 있게 하는 기술
- 자연어는 말하나에 뜻이 여러 개라 유연 but 모호함. 프로그래밍 언어는 일대일이라 명확
- 컴퓨터는 말을 바로 이해 못 해서 결국 숫자(벡터)로 바꿔서 처리함

### 자연어 특징 → NLP에 필요한 것
| 특징 | 내용 | 그래서 필요한 것 |
| --- | --- | --- |
| 이산적 | 단어, 문자 같은 기호라 계산이 안 됨 | 벡터로 바꾸기 (=임베딩) |
| 순차적 | 순서 바뀌면 뜻이 바뀜 | 순서, 문맥을 보는 구조 (RNN, Transformer) |
| 중의성 | 배 = 신체, 과일, 선박 | 문맥 전체로 판단 |
| 희소성 | 조합이 너무 많아서 다 못 배움 | 안 본 표현도 확률적으로 추론 |
| 가변성 | 신조어, 줄임말이 계속 생김 | 규칙 말고 데이터로 학습 |

### NLP 처리 단계


1. 형태소 분석 : 나/대명사 + 는/조사 + 학교/명사 ...
2. 구문 분석 : 철수(주어), 사과(목적어), 먹었다(서술어)
3. 의미 분석 : "은행에서 돈을 찾았다" → 은행 = 금융기관
4. 담화 분석 : "철수는 합격했다. 그는 기뻐했다" → 그 = 철수

활용 : 번역(파파고), 감성 분석(리뷰 긍부정), 스팸 분류, 요약, 챗봇 등에 쓰인다!

### 언어 모델
- 많은 텍스트를 학습해서 다음 단어가 나올 확률을 예측하고 문장이 자연스러운지 판단하는 모델
- 사람이 말하는 방식을 흉내 내는 시스템
- "나는 아침에 밥을 ___" → 먹었다 80%, 마셨다 15%, 달렸다 5%
- 학습 방식에 따라 통계적 언어 모델(SLM)이랑 신경망 언어 모델(NLM)로 나뉨

## 2. 통계적 언어 모델 (SLM)

### 기본 아이디어?
- 분포 가설 : 비슷한 문맥에 나오는 단어는 뜻도 비슷함
  - "아침에 커피를 마셨다" / "아침에 차를 마셨다" → 커피랑 차는 비슷한 단어
  - 나중에 임베딩의 이론적 기반
- 조건부 확률 : 앞 문맥이 주어졌을 때 다음 단어 확률, P(갔다 | 나는 학교에)
- 카운트 기반 : "나는 학교에"가 100번 나왔는데 그 뒤에 갔다 70번 왔다 20번 뛰었다 10번 → 갔다 70% 인거임

### 한계 : 희소 문제
- 가능한 단어의 조합이 매우 많아 특정 표현이 학습 데이터에 거의 등장하지 않거나 전혀 등장하지 않는 문제
- "나는 학교에 순간이동했다"는 문법적으로 말이 되는데 데이터에 없으면 0 처리되는 예시

### N-gram
- 데이터의 희소성 & 계산의 복잡성 때문에 등장
- 직전 N-1개만 보고 다음 단어 예측
- 앞 단어가 많아질수록 희소해지고 계산도 복잡해져서 나온다
- 4-gram이란? : "An adorable little boy is spreading ___" → boy, is, spreading 3개만 본다는거임
- SLM이랑 다른 게 아니라, SLM을 계산하기 쉽게 문맥 범위를 줄인 버전
- 한계
  - N 키우면 조합이 확 늘어나서 희소 문제가 심해짐
  - N 작으면 멀리 있는 단어 관계를 못 봄

### PPL 
- 모델이 문장을 예측할 때 ***얼마나 헷갈려하는지*** 나타내는 지표
- 낮을수록 잘 맞힘 높을수록 못 맞힘
- 데이터에 드물게 나오는 문장일수록 PPL이 높아지는 경향
- SLM, NLM 둘 다 평가에 씀. 예시로 SLM 80, NLM 25면 NLM이 훨씬 잘 맞히는 거임

## 3. 신경망 언어 모델 (NLM)
- SLM 한계를 줄이기위해 신경망으로 단어 뜻이랑 문맥을 학습함
- 핵심은 임베딩! : 단어를 벡터 공간의 점으로 표현해서 비슷한 단어끼리 가까이 놓임


  - moon, mars, nasa는 우주 동네 / bible, god, church는 종교 동네

| 구분 | SLM | NLM |
| --- | --- | --- |
| 학습 | 빈도 세서 확률 계산 | 신경망 가중치 학습 |
| 단어 표현 | 각각 따로인 기호 | 연속적인 벡터 |
| 비슷한 단어 | 관계 반영 어려움 | 임베딩으로 학습 |
| 희소 문제 | 안 본 조합에 약함 | 많이 완화됨 |
| 문맥 | N-gram은 N-1개만 | 구조에 따라 길게 가능 |
| 계산 비용 | 낮음 | 높음 |

## 4. 딥러닝 기반 언어 모델

### 저번주 배운 transformer



### BERT
- 트랜스포머 인코더구조를 기반으로 양방향이라 단어 앞뒤 문맥을 다 봄 → 문장 이해에 강함
- 입력 : 토큰화하고 [CLS], [SEP] 붙임 → 임베딩 → 인코더 여러 층 거침
- 사전학습 + 파인튜닝 구조
  - MLM : 단어 일부를 [MASK]로 가리고 앞뒤 보고 맞히기
    - "How are [MASK] doing today" → you
    - 같은 단어라도 문맥에 따라 뜻이 달라지는 걸 배움

    
  - NSP : 문장 B가 문장 A 바로 다음 문장인지 맞히기 
    - 문장끼리의 관계를 배움

    
- BERT 이후
  - RoBERTa 
  - ALBERT 

### GPT
- transformer의 디코더 기반 구조를 활용하여 이전 단어들을 바탕으로 다음 단어를 예측하고 텍스트를 생성하는 언어 모델
- masked self-attention으로 앞 단어만 보고 다음 단어를 예측하는 자기회귀 방식 
- GPT-3부터 파인튜닝 없이 프롬프트만 보고 하는 in-context learning
  - zero-shot : 예시 없이 지시만
  - one-shot : 예시 1개
  - few-shot : 예시 여러 개

| 구분 | BERT | GPT |
| --- | --- | --- |
| 구조 | 인코더 | 디코더 |
| 방향 | 양방향 | 단방향(자기회귀) |
| 학습 | MLM, NSP | 다음 토큰 예측 |
| 목적 | 이해 | 생성 |
| 주로 하는 것 | 감성 분석, 분류, 개체명 인식 | 챗봇, 생성, 번역, 요약 |

### LLM
- 어엄청 많은 데이터와 큰 신경망으로 학습한 언어 모델. 대부분 트랜스포머 기반 (GPT, LLaMA, Claude, Gemini등등)
- 학습 : 사전학습(다음 토큰 예측 반복) → 파인튜닝(목적에 맞게 추가 학습)
- 예측 : 토큰화하고 벡터로 변환 → 어텐션으로 관계 분석 → 다음 토큰 확률 계산 → 하나씩 생성
- 대화형 LLM이 GPT 방식을 쓰는 이유 : 질문을 이해하는 것보다 새 문장을 만들어야 해서 자기회귀가 맞음
- 한계
  - 환각(할루시네이션) : 추론하는거처럼 보이지만 사실은 그냥확률로 다음 단어를 고르는 거에 불과함;;
  - 비용 : 사전학습, 추가 학습 다 엄청 진짜 비쌈. 



### RAG
- 외부 지식 검색 + LLM 생성을 합쳐서 정확도랑 최신성을 높이는 기술
- 구조


  1. 질의 인코더 : 질문을 벡터로 바꿈
  2. 지식 검색기 : 벡터 DB에서 비슷한 문서 찾기
  3. 지식 증강 생성기 : 질문 + 찾은 문서를 같이 LLM에 넣어서 답변
- 장점 : 재학습 없이 최신 정보 반영, 근거 기반이라 환각이 줄어듦

### CAG 
- RAG는 관련만 있으면 오래되거나 틀린 문서도 가져올 수 있음
- CAG는 관련성 + 출처 신뢰도 + 최신성까지 평가해서 믿을 만한 정보 위주로 답변
- 예시 : "David Cameron 직책?"
  - RAG : 신뢰도 낮은 문서를 보고 "현재 공직 없음"이라고 틀림
  - CAG : Reuters(최신, 믿을 만함) 위주로 보고 "외무장관"이라고 맞힘


### LangChain
- LangChain : LLM이랑 외부 데이터, API, 도구를 연결해서 앱 만들기 쉽게 하는 오픈소스 프레임워크
  - 추상화 : 복잡한 API 호출을 간단한 명령어로
  - 표준화 : ChatGPT → Claude로 바꿀 때 모델 연결 부분만 바꾸면 됨
  - 체이닝 : 번역 → 요약 → 출력처럼 작업을 순서대로 연결
- LangGraph : LangChain은 일자 흐름이라 분기나 반복이 어려움 → 노드랑 엣지로 연결해서 조건 분기, 반복 가능
  - 노드 : LLM, Agent, Tool 같은 작업 단위
  - 엣지 : 다음에 어디로 갈지
  - 상태 : 노드끼리 공유하는 정보(이전 결과, 대화 문맥)


- LangSmith : 실행 과정 추적, 디버깅, 성능 평가, 모니터링, 프롬프트 버전 관리
- 외부 API : OpenAI 키 발급 → 키는 처음 한 번만 전체가 보이고 유출되면 내 돈으로 결제되니까 메모장 env에 넣기~


## 5. sLM 
- LLM보다 작은 모델. sLM은 수백만~수십억, LLM은 수십억~수천억 파라미터
- 메모리랑 연산을 적게 써서 폰, 노트북에서도 됨. 인터넷 없이 기기 안에서 돌리는 오프라인 추론도 가능

| 구분 | LLM | sLM |
| --- | --- | --- |
| 크기 | 큼 | 작음 |
| 성능 | 복잡한 추론에 강함 | 특정 분야, 단순한 작업에 효율적 |
| 비용 | 높음 | 낮음 |
| 속도 | 느릴 수 있음 | 빠름 |
| 환경 | 고성능 서버 | 서버, 폰, 노트북 |



### 모델 압축 방법
- 가지치기(Pruning) : 영향이 작은 연결이나 뉴런을 잘라냄. 너무 자르면 성능이 떨어짐
- 양자화(Quantization) : FP32 → FP16, INT8로 정밀도를 낮춰서 용량을 줄임. 대신 성능이 조금 떨어질 수 있음
- 저차원 분해 : 큰 가중치 행렬을 작은 행렬 A × B로 근사함. LoRA가 이 원리로 변경량만 학습
- 지식 증류 : 큰 teacher 모델의 출력을 작은 student 모델이 따라 하게 학습함 (DistilBERT)

### 예시
- DistilBERT(BERT 증류), Gemma(구글), GPT-4o mini(OpenAI), Llama 소형(메타), Ministral(온디바이스), Phi(MS)

### 장단점
- 장점 : 빠름, 적은 자원으로 돌아감, 기기 안에서 돌려서 데이터가 밖으로 안 나감, 운영비랑 전력이 적게 들고 비싼 GPU 의존도가 낮음
- 단점 : 데이터 편향이 그대로 나옴, 복잡한 추론이나 전문 지식은 LLM보다 약함, 환각도 있음

### 결합 추론


- 쉬운 작업(분류, 간단 요약)은 sLM, 어려운 작업(여러 단계 추론, 전문 분석)은 LLM
- 챗봇 예시 : 간단한 질문은 sLM이 빠르게 복잡한 분석은 LLM이
- 비용이랑 성능의 균형을 맞추는 방식

## 6. 생성 모델

### 분류 모델 vs 생성 모델
| 구분 | 분류 모델 | 생성 모델 |
| --- | --- | --- |
| 학습 목표 | P(Y\|X) : X 주면 Y 맞히기 | P(X) : 데이터 분포 자체 학습 |
| 배우는 것 | 데이터끼리 차이점 | 공통 특징이랑 다양한 패턴 |
| 출력 | 라벨, 확률 | 새 이미지, 텍스트, 음성 |
| 한계 | 이상한 데이터도 아는 라벨 중 하나로 분류함, 생성 불가 | 부자연스러운 결과가 나올 수 있음, 품질 평가가 어려움 |

  

### 생성 모델 분류 
- 명시적 밀도 모델 : P(X)를 수치로 계산하거나 근사
  - 직접 계산 가능 : NADE, MADE, PixelRNN/PixelCNN, 변수 변환 모델
  - 근사 : VAE, 볼츠만 머신
  - 마르코프 연쇄 : 다음 상태가 현재 상태로만 정해짐. 오늘 맑으면 내일 맑음 80%, 비 20%

- 암시적 밀도 모델 : P(X)는 안 알려주고 샘플만 뽑음
  - 직접 샘플링 : GAN. 노이즈 넣으면 한 번 통과로 바로 생성
  - 간접 반복 샘플링 : GSN. 상태 변화 규칙을 반복 적용해서 생성

## 7. AE, VAE

- 입력을 압축했다가 다시 원래대로 복원하도록 학습하는 모델. 정답 라벨이 필요x
- 일반 인코더-디코더는 한국어 → 영어처럼 다른 걸 만들지만, AE는 입력 자체를 복원하는 게 목적
- 구조 : 인코더 → 병목 → 디코더


- 왜 압축했다가 복원함?
  - 압축 안 하면 그냥 복사해버림. 공간을 제한해야 뭐가 중요한지 배움 
  - 복원은 압축한 정보만으로 원본을 만들 수 있는지 확인하는 용도. 복원 오차를 줄이도록 학습
- 활용 : 데이터 압축, 노이즈 제거, 이상 탐지

### VAE 
- 등장 배경 : AE는 잠재공간이 불규칙해서 빈 영역에서 값을 뽑으면 이상한 결과가 나옴 → 새 데이터 생성이 어려움
- VAE는 확률분포로 표현
- 학습 목표 두 개
  - 복원 오차 최소화 : 원본이랑 비슷하게 복원
  - KL 발산 최소화 : 잠재분포가 표준정규분포에 가까워지게 (KL 발산 = 두 분포가 얼마나 다른지, 같으면 0)
- AE vs VAE 잠재공간
  - AE : 고정된 점, 불규칙하고 빈 곳이 생김
  - VAE : 분포(곡선), 연속적이고 규칙적이라 사이 공간에서 뽑아도 자연스러움
  - 웃는 정도 예시 : AE는 -1이나 1 같은 점 하나, VAE는 중심이랑 퍼짐을 가진 곡선
  
- 작동 단계
  1. 분포 예측 : 인코더가 μ, σ 예측
  2. 샘플링 + 재파라미터화
     - 그냥 랜덤으로 뽑으면 기울기가 안 흘러서 랜덤을 ε로 따로 빼는 것
     - μ=2, σ=0.5일 때 ε=1이면 2.5, ε=-1이면 1.5
  3. 생성 : 디코더가 z로 x' 만듦. 학습 끝나면 표준정규분포에서 z만 뽑아서 넣어도 새 데이터가 나옴
- 활용 : 이미지 생성, 데이터 증강, 이상 탐지, 특징 추출

## 8. GAN

### 개념
- 2014년 Ian Goodfellow. 생성자 G랑 판별자 D가 경쟁하면서 학습
- 위조지폐범(G)은 진짜 같은 위조지폐를 만들고, 경찰(D)은 진짜인지 가짜인지 가려냄
- 서로 실력이 늘다가 끝에는 D가 0.5, 걍 찍는 수준

### 구조 
- 노이즈 z → 생성자 → 가짜 이미지 G(z)
- 판별자는 진짜 x 또는 가짜 G(z)를 받아서 진짜일 확률(0~1)을 냄
  - D(x) = 0.9, D(G(z)) = 0.1 이런 식으로 학습
- 판별자는 구별 능력을 최대화, 생성자는 판별자가 구별 못 하게 최소화
- 학습 끝나면 생성자만 꺼내서 씀



### 장점
- 실제처럼 보이는 새 데이터 
- VAE보다 선명함. VAE는 MSE로 평균을 맞추다 보니 흐릿해지고, GAN은 판별자를 속여야 해서 질감까지 살림
- 노이즈를 한 번만 통과시키면 돼서 빠름
- 잘 학습되면 z를 조금씩 바꿀 때 이미지도 자연스럽게 변함

### 한계랑 해결법
| 문제 | 내용 | 해결 |
| --- | --- | --- |
| 학습 불안정 | 한쪽이 너무 세지면 균형이 무너짐, 진동, 수렴 안 됨 | 학습률 조절, WGAN-GP |
| 초기 학습 문제 | 초반에 D가 너무 쉽게 맞혀서 G한테 기울기가 거의 안 감 | non-saturating loss, D 학습 속도 조절 |
| 모드 붕괴 | 속이기 성공한 비슷한 것만 반복 생성 | minibatch discrimination, WGAN-GP |
| 평가 어려움 | 사실성, 선명도, 다양성을 지표 하나로 못 봄 | FID, IS, 사람 평가 |
| 역매핑 어려움 | z → 이미지 방향만 있어서 이미지의 z를 못 찾음 | GAN inversion, BiGAN |

### 사례
- NVIDIA Progressive GAN : 저해상도부터 고해상도로 키우면서 없는 사람 얼굴 생성
- Deep Video Portraits (cGAN) : A의 표정, 움직임을 B 영상에 입힘
- Facebook ExGAN : 눈 감은 사진을 참고 사진 보고 눈 뜬 사진으로 복원

## 9. 확산 모델 (Diffusion)

### 개념
- 이미지에 노이즈를 더한 데이터로 학습하고, 노이즈에서 시작해서 조금씩 노이즈를 빼면서 새 이미지를 만듦
- 목적 : 실제 분포를 배워서 자연스럽고 다양한 데이터 생성
- 모호성 모델링 : 노이즈가 많으면 원래 뭐였는지 하나로 못 정하니까 가능한 결과들의 분포를 배움
  - "공원에 앉은 강아지" 하나로 품종, 털색, 자세가 다른 여러 결과가 나옴

### 구조
- 순확산 (forward) : 원본 x0에 가우시안 노이즈(잡음)를 조금씩 더함 → x1, x2 ... → 완전 노이즈 xT. 학습 데이터를 만드는 과정
- 역확산 (reverse) : xT에서 시작해서 Denoising U-Net이 노이즈를 예측하고 조금씩 빼는 걸 반복 → 새 이미지 x0



### 장점?단점?
- 장점
  - 품질이 제일 좋음. 한 번에 만드는 GAN보다 단계별로 다듬어서 정교
  - 경쟁 구조가 아니라 MSE로 노이즈 예측만 학습해서 GAN보다 안정적이고 손실 곡선으로 확인도 됨
  - 노이즈에서 시작해서 같은 조건이어도 결과가 다양함
- 단점
  - 여러 단계를 반복해서 GAN보다 느림
  - 학습이랑 생성 모두 계산이 많아서 GPU가 필요함
  - 단계 수나 방식에 따라 속도랑 품질이 달라져서 설계가 까다로움
