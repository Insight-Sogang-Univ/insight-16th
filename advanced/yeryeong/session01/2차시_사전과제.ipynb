### 1. 인공신경망의 시작: 퍼셉트론

#### 1-1. 인공신경망이란?

- **인공신경망**  
: 우리 뇌의 신경망을 모방한 모델
    - 가지돌기 통해 외부 전기신호 받아들임
    - 모인 신호가 일정 이상의 크기를 가지면 축삭돌기 통해 신호 전달

  1. 입력($x_1, x_2, …, x_n$)에 각각 가중치($w_1, w_2, …, w_n$)와 편향($b$) 적용
  2. 종합한 값 활성화 함수에 넘김 (임계값 기준)
     - 기준 미달 → 0에 가깝게
     - 기준 통과 → 유의미한 값으로
  3. output 출력

- **퍼셉트론 (Perceptron)**  
: 초기 형태의 인공신경망, 신경망 기원이 되는 **선형 이진 분류기**
    - 흐름: `입력 → 가중합 → 활성화 함수 → 출력`
    - 절차
        1. 입력값과 가중치 곱함
        2. 곱한 값들의 총합 구함  (편향은 옵션)
        3. 총합이 임계값($θ$)을 넘으면 1, 넘지 않으면 0 출력 (Step 함수)
    - **Training(학습)**: 원하는 출력값을 내보내도록 가중치를 조정해가는 작업

- **용어 정리**
    - **가중치 (weight)**: 중요도 결정
        - 각 입력값에는 각각 가중치 존재
        - 값이 클수록 중요
    - **활성화 함수 (Activation function)**
        - 퍼셉트론은 계단 함수 사용
    - **임계값 (Threshold, $θ$)**: 활성화 함수에 사용되는 값
        - \begin{aligned}&\text{if } \sum_{i}^{n} w_i x_i \ge \theta \rightarrow y=1 \\&\text{if } \sum_{i}^{n} w_i x_i < \theta \rightarrow y=0\end{aligned}
    - **편향 (bias)**
        - \begin{aligned}&\text{if } \sum_{i}^{n} w_i x_i + b \ge 0\rightarrow y=1 \\&\text{if } \sum_{i}^{n} w_i x_i + b < 0\rightarrow y=0\end{aligned}
        - 임계값과 표현의 차이
        - 오늘날 딥러닝에서 주로 사용

#### 1-2. 단층 퍼셉트론과 그 한계

- **단층 퍼셉트론**  
: 입력층과 출력층으로만 이루어진 퍼셉트론

- **논리 게이트 (Logic Gate)**  
: 디지털 회로에서 기본적인 논리 연산을 수행하는 구성 요소
    - n개의 입력값과 하나의 출력값
    - AND, NAND, OR 구현 가능

| 게이트 | 조건 | 가중치·편향 예시 `[w1, w2, b]` |
|---|---|---|
| AND | 둘 다 1일 때만 1 | `[0.5, 0.5, -0.7]` |
| OR | 둘 다 0일 때만 0 | `[0.6, 0.6, -0.5]` |
| NAND | 둘 다 1일 때만 0 | `[-0.5, -0.5, 0.7]` |
| XOR | 서로 다를 때만 1 | 가능한 조합이 없음 |

- **단층 퍼셉트론의 한계**: XOR 게이트 구현 X
    - 입력값과 가중치의 선형 결합 → **선형 분류 문제**만 구현 가능
    - AND 게이트 → 하나의 직선으로 분리 가능
    - XOR 게이트 → 하나의 직선으로 분리 불가능 (곡선/비선형 영역으로 분리)

#### 1-3. 다층 퍼셉트론

- **MLP가 XOR을 구현할 수 있는 이유**
    - 은닉층에서 **비선형 활성화 함수** 적용 → 새로운 특징을 만들고, 다음 층에서 그 특징들 조합
    - ex) 은닉층 두 뉴런이 각각 NAND와 OR 계산 → 출력층이 AND 적용 = XOR
    - $h = σ(W_1x + b_1), \quad y = W_2h + b_2$   
     ($σ$: 비선형 활성화 함수)
    - 단층 퍼셉트론도 결정 경계 직선

- **구성**: 입력층 + 하나 이상의 은닉층 + 출력층
- **심층 신경망 (DNN)** = 은닉층이 2개 이상인 신경망
- **딥러닝 (Deep Learning)** = 심층 신경망을 학습시킴

### 2. 딥러닝 모델 학습

#### 2-1. 순전파와 역전파 (Forward & Backward Propagation)

- **학습 알고리즘의 기본 구조**
    1. **순전파**를 통해 입력값으로부터 예측 수행
    2. **오차 계산**을 통해 예측 정확도 측정
    3. **역전파**를 통해 출력층에서 입력층 방향으로 오차 전달
    4. **가중치 업데이트**
    5. 1~4 반복 → 오차값이 0에 가까워짐

- **연쇄 법칙 (Chain Rule)**  
: 합성함수의 미분은 각 함수의 미분의 곱으로 나타낼 수 있다는 법칙
    - 역전파에서 각 가중치가 최종 손실에 얼마나 기여했는지 계산할 때 사용
    - **손실 함수** = 예측값과 실제값의 차이를 수치로 표현
    - **기울기** = 손실 함수를 줄이기 위한 가중치 설정 신호

- **기울기 소실 문제 (Vanishing Gradient Problem)**
    - → 학습 문제 → ReLU 등장

- **전체 흐름 정리**
    1. **순전파** = 예측  
    : `입력 x → [w1] → [w2] → 예측값 ŷ`
    2. **손실 계산** = 얼마나 틀렸는지 측정  
    : $L = (\hat{y} - y)^2$
    3. **역전파** = 책임 추적 (연쇄법칙)  
    4. **가중치 업데이트** = 수정

#### 2-2. 손실 함수 (Loss Function)

- **손실 함수**  
: 모델의 예측이 얼마나 틀렸는지를 수치로 표현하는 방법
    - 실제값과 예측값의 차이를 계산
    - 회귀 → MSE, 분류 → 크로스 엔트로피

- **평균 제곱 오차 (MSE)**
    - $MSE = \dfrac{1}{n}\sum_{i=1}^{n}(y_i - \hat{y}_i)^2$
    - 회귀 문제에 적합 (ex. 집 가격 예측)

- **이진 크로스 엔트로피 (BCE)**
    - $BCE = -[\,y\log\hat{y} + (1-y)\log(1-\hat{y})\,]$
    - 이진 분류 문제 (로지스틱 회귀)
    - 실제 정답에 부여한 확률에 $-\log$ 적용한 손실
        - 정답 1일 때 → $L = -\log\hat{y}$
            - 예측 확률이 1에 가까울수록 loss → 0 (잘 맞힘)
            - 예측 확률이 0에 가까울수록 loss 급격히 증가 (틀림)
        - 정답이 0일 때 → 반대로 동작
    - 로그: 확률의 곱 → 합으로
    - MSE는 loss의 크기만, BCE는 **얼마나 확신을 가지고 예측했는지** 반영

- **크로스 엔트로피 오차 (CEE)**
    - $CEE = -\sum_{k} y_k \log\hat{y}_k$
    - 다중 클래스 분류 문제에서 사용

| 손실 함수 | 사용 상황 |
|---|---|
| MSE | 회귀 |
| BCE | 이진 분류 |
| CEE | 다중 클래스 분류 |

#### 2-3. 활성화 함수 (Activation Function)

- 각 층의 계산: `입력값×가중치 + 편향 → 활성화 함수 → 출력값`

- **활성화 함수를 거치는 이유**
    1. **비선형성 필요** → 없으면 선형 계산의 반복
    2. **미분이 되어야 함** → 역전파는 미분값을 연쇄적으로 곱하기 때문

- **발전 과정**
    - 계단 함수: 미분 불가
    - 시그모이드: 미분 가능 → 기울기 소실
    - ReLU: 양수 구간 미분값 = 1 → 기울기 소실 발생 X

| | 단층 퍼셉트론 | 다층 퍼셉트론 |
|---|---|---|
| 활성화 함수 | 계단 함수 | Sigmoid, ReLU 등 |
| 출력값 | 0 또는 1 | 연속적인 값 (확률 또는 실수) |
| 특징 | 선형 문제만 | 비선형 문제도 학습 가능 |

- **계단 함수 (Step Function)**
    - 가중합이 임계값을 넘으면 1, 아니면 0
    - 미분값이 0이라 역전파 불가 → 다층 퍼셉트론에서는 쓰지 않음

- **시그모이드 함수**
    - $σ(x) = \dfrac{1}{1+e^{-x}}$
    - 입력 신호를 0과 1 사이의 연속적인 값으로 변환
    - 얼마나 확실한지 표현 가능, 미분 가능
    - 미분값 최대 0.25 → 기울기 소실

- **ReLU 함수 (주로 사용)**
    - $ReLU(x) = \max(0, x)$
    - 입력이 0보다 크면 그대로, 0 이하면 0
    - 계산 간단, 효율적 / 비선형성 제공 / 양수 구간 미분값 1 → 기울기 소실 완화

- **Softmax 함수**
    - $softmax(z_i) = \dfrac{e^{z_i}}{\sum_j e^{z_j}}$
    - 각 클래스 출력값을 확률로 변환 (0~1, 전체 합 1)
    - 다중 클래스 분류 출력층에서 사용
    - **Softmax + 크로스 엔트로피** = 가장 표준적인 조합

#### 2-4. 경사하강법 (Gradient Descent)

- 손실을 가장 작게 만드는 W를 찾음
    - x축 W, y축 E → 아래로 볼록한 곡선의 **바닥** 찾음
    - $W \leftarrow W - α\dfrac{∂E}{∂W}$

| 지금 W의 위치 | 접선 기울기 | 결과 |
|---|---|---|
| 최솟값 오른쪽 | 양수 | W 작아짐 → 왼쪽으로 |
| 최솟값 왼쪽 | 음수 | W 커짐 → 오른쪽으로 |
| 최솟값 도달 | 0 | 멈춤 |

- b도 똑같은 방법으로 찾을 수 있음
- 비유: 안개 낀 산에서 가장 가파른 쪽으로 한 걸음씩 내려가기

- **학습률 (α)**  
: 미분값에 곱해 이동량 조절
    - 너무 작음 → 수렴은 하지만 오래 걸림
    - 너무 큼 → 수렴 못 하고 발산
    - 0.001부터 시작 (Adam의 기본값)

- **배치 · 에포크 · 이터레이션**
    - **배치 (Batch)**: 매개변수 조정을 위해 사용하는 데이터의 묶음 (하이퍼파라미터)
        - 너무 크면 학습 느려지고 메모리 부족
        - 너무 작으면 학습 불안정
    - **에포크 (Epoch)**: 모든 데이터셋을 학습하는 횟수
    - **이터레이션 (Iteration)**: 한 번의 에포크를 끝내기 위해 필요한 배치의 수
        - 1 Epoch 데이터 수 = 배치 크기 × 이터레이션

- **경사하강법의 종류**
    - **배치 경사하강법**: 전체 데이터를 한 묶음으로 (`batch_size=len(X_train)`)
        - 업데이트 횟수 적음, 메모리 많이 필요, 수렴 안정적
    - **확률적 경사하강법 (SGD)**: 랜덤으로 선택된 하나의 데이터만 계산 (`batch_size=1`)
        - 자원 적어도 사용 가능, 변경폭 불안정·정확도 낮을 수 있음
    - **미니 배치 경사하강법**: 배치 크기를 적절히 지정 (`batch_size=128`)
        - 전체보다 빠르고 배치 1보다 안정적 → **가장 많이 사용**
        - `batch_size` 기본값 32, 보통 $2^n$으로 선택

#### 2-5. 옵티마이저 (Optimizer)

- **경사하강법의 한계: Local Minima**
    - 딥러닝 손실함수 → 작은 웅덩이(Local Minimum)에서도 기울기 0

- **모멘텀 (Momentum)**  
: 변수 가던 방향으로 속도 추가
    - **관성** 부여
    - Local Minima 탈출, 기울기 작은 구간 빠르게 통과
    - `tf.keras.optimizers.SGD(learning_rate=0.01, momentum=0.9)`

- **RMSProp**  
: 각 방향마다 학습률을 따로 조정
    - 급한 경사 → 속도 줄임 / 완만한 경사 → 속도 유지 또는 증가

- **Adam**  
: RMSProp + Momentum
    - 1차 모멘트(기울기의 평균)와 2차 모멘트(기울기의 제곱 평균)로 학습률 조정
