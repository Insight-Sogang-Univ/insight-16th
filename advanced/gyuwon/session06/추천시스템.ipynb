### 1. 추천 시스템이란?

> 사용자의 정보 데이터를 분석, 개인의 취향에 맞는 아이템을 추천하는 알고리즘

* 아이템 : 상품, 영화, 음악, 뉴스, 앱 등 추천 대상 전부

* 필요한 이유

    - 고객 입장 : 찾는 시간과 노력이 줄어듦, 몰랐던 취향의 아이템도 발견 가능

        - 검색은 이미 알고 있는 것을 찾게 돕고, 추천은 아직 모르는 것을 발견하게 도움

    - 기업 입장 : 이용 시간, 구매 전환, 재방문이 늘어남 / 잘 안 보이던 비인기 아이템(롱테일)도 노출 가능

#### 1-1. 사용되는 데이터의 종류와 특징

* 명시적 데이터(Explicit Data)

    - **고객들에 의해 입력된 데이터. 사용자가 선호도를 직접적으로 표현한 데이터.**
    
        - Ex) 평점, 좋아요, 투표, 스케일 점수
    
    - 명백한 데이터를 얻기 쉽지 않음. 모든 사용자가 리뷰나 평점, 별점 등을 남기지 않기 때문
    
    - 사람의 취향은 변함 -> 과거의 선호도가 현재와도 동일한지 알 수 없음

    - 사람마다 점수를 주는 기준이 다름

    → 당시의 사용자의 **취향이 정확하게 반영**된다는 장점이 있으나, **데이터 자체를 얻기 
    힘들다**는 단점

* 암시적 데이터 (Implicit Data)

    - **고객들의 행동을 기반으로 수집된 데이터. 사용자의 선호도가 간접적으로 표현된 데이터**

        - Ex) 조회수, 영상시청 기간, 구매 기록, 페이지 체류 기간

    - 사용자가 남기는 발자취 -> 사용자가 서비스를 이용하면 로그데이터나 검색/클릭 기록은 무조건 남음
    
    - 행동이 곧 선호라고 단정할 수 없음

        - 행동이 있어도 잘못 클릭했거나 선물로 산 것일 수 있음 -> 좋아한다는 보장 X

        - 행동이 없어도 아직 못 봤을 수 있음 -> 싫어한다는 보장 X

    - 데이터를 얻기 굉장히 쉽지만, 사용자의 취향이나 선호도를 직접적으로 알 수 없다는 단점

-> 실제 서비스에서는 암시적 데이터가 훨씬 많아서 두 가지를 섞어 쓰는 경우가 대부분

#### 1-2. 추천시스템의 분류

* 내용 기반 필터링(Content-Based Filtering)

    - 콘텐츠의 유사도를 기반으로 사용자에게 추천하는 방법

    - 사용자가 이용하거나 선택한 아이템과 유사한 아이템을 추천하는 방식

        - Ex) SF 영화를 자주 봄 -> 다른 SF 영화 추천

* 협업 필터링(Collaborative Filtering)

    - 서비스 내에 있는 **다수의 사용자로부터 얻은 선호도 정보**를 기반으로 사용자에게 추천하는 방법

        - 두 명의 사용자가 비슷한 관심사를 가지고 있다면 한 사용자의 데이터를 바탕으로 다른 사용자에게 추천해주는 방식, 사용자 간의 선호도를 서로 고려하여 많은 선택 사항들로부터 아이템을 걸러내거나 선택

        - 사용자 기반(User-Based Filtering): 특정 사용자와 유사한 취향의 사용자가 선호한 아이템 추천

        - 아이템 기반 (Item-Based Filtering): 사용자가 선호한 아이템과 유사한 아이템을 추천

* 하이브리드 필터링(Hybrid Filtering)

    - 2가지 이상의 다양한 종류의 추천시스템 알고리즘을 조합하여 만들어진 알고리즘

    - 콘텐츠 기반 필터링과 협업 필터링 각각의 단점들을 보완하기 위해 제시된 방식

        - Ex) 오늘 공개된 영화는 평점이 하나도 없어서 협업 필터링으로는 추천 불가 -> 장르, 감독 정보로 내용 기반 추천을 하다가 평점이 쌓이면 협업 필터링으로 넘어감

| 하이브리드 방식 | 설명 |
| --- | --- |
| Weighted Ensemble | 여러 모델의 추천 결과를 점수(Soft Ensemble)나 투표(Hard Vote)로 합쳐서 최종 추천 아이템을 정함 |
| Mixed | 특징이 다른 여러 추천 알고리즘의 결과를 플랫폼에 함께 보여 줌 |
| Switch | 사용자나 서비스 상태 같은 상황에 따라 알고리즘의 추천 결과를 골라서 보여 줌 |
| Feature Combination | 가지고 있는 데이터에서 얻을 수 있는 feature를 모두 조합해서 알고리즘을 학습 |
| Meta-Level | 첫 번째 모델의 결과가 다음 모델의 input이 되는 방식 |

-> 위의 새 영화 예시(내용 기반으로 시작해서 평점이 쌓이면 협업 필터링)는 Switch에 해당

### 2. 내용 기반 필터링(Content-Based Filtering)

> 내용 기반 필터링에 필요한 것 : 사용자가 무엇을 좋아하는지(사용자 프로필) + 아이템끼리 얼마나 비슷한지(유사도)

* 흐름 : 아이템을 숫자로 표현 -> 사용자 프로필 생성 -> 유사도 계산 -> 점수가 높은 아이템 추천

#### 2-1. 사용자 프로필

* 추천 시스템에 사용되는 데이터    
    
    * 직접 지정형 : 사용자에게 자신이 선호하는 아이템의 특징을 명시적으로 지정하도록 하는 방법

        - OTT 서비스 가입시, 제시하는 몇 개 영화들에 대한 선호도를 체크해야 하는 경우

    * 간접 지정형 : 사용자의 과거 행동 이력에 기반하여 사용자의 프로필을 작성하는 방법

        - SF 영화를 즐겨본 이력이 존재하는 유저에게 인터스텔라를 추천해주는 경우

* 사용자의 프로필을 획득하는 방식

    - 특성 가중치는 사용자가 가지고 있는 아이템 특성 가중치의 평균 값을 활용

        - 특성 가중치 : 아이템이 각 특성을 얼마나 가지고 있는지 나타낸 값 (0~1), 아래 예시 참고

    - 사용자의 기호를 따로 파악하거나 설문조사 또는 자발적인 키워드 입력 등으로 프로필 획득

        - Like/Dislike, Rating from 0 to 5, 유저가 직접 작성한 댓글

        - 검색 로그, 아이템 선택 후 구매 혹은 선택, 특정 아이템을 위해 비슷한 아이템을 tag한 이력

    - 아이템에 대한 유저의 기존 평가를 바탕으로 분류 문제를 해결하는 것과 같음

        - 주어진 User profile 정보를 바탕으로 새로운 아이템을 유저가 평가한 metric 중 하나를 선택
        
        - 아이템에 대한 유저의 기존 평가 $= \{r_1,r_2, …r_n\}$

        - 아이템 특징이 입력, 유저 평가가 정답인 구조라서 결정 트리, 랜덤 포레스트, KNN 같은 분류 모델을 쓸 수 있음

* 좋은 사용자 프로필을 구성할수록, 추천 시스템의 정확도와 만족도가 높아짐 -> 다양한 방식(명시적+암묵적)을 혼합하여 사용하는 것이 일반적

* 예시) 특성 가중치 평균으로 사용자 프로필 만들기

| 인터스텔라 | 0.3 | 0.2 | 0.9 | 0.0 |
| 노트북 | 0.0 | 0.9 | 0.0 | 0.1 |

$\text{User Profile} = \frac{1}{3}([0.9,0.1,0.5,0.0] + [0.3,0.2,0.9,0.0] + [0.0,0.9,0.0,0.1]) = [0.4,\ 0.4,\ 0.47,\ 0.03]$

- SF 선호가 제일 높고, 액션 / 로맨스는 중간, 공포는 거의 안 좋아하는 사용자

- 새 영화 "듄" = [0.4, 0.0, 0.9, 0.0] -> 프로필과 코사인 유사도 약 0.8 -> 유사도가 높으니까 추천

#### 2-2. 유사도(Similarity)

> 추천시스템을 이해하는데 중요한 수학적 개념

* 영상, 이미지, 텍스트 등의 형태로 존재하는 아이템 정보 데이터를 벡터화한 이후 아이템 벡터간의 유사도를 측정 -> 측정된 유사도를 바탕으로 고객이 선호하는 아이템과 비교하여 유사도가 높은 아이템 선별

* step

    1. 인풋 데이터 만들기

        - 아이템 정보 활용

        - 테이블 형태로 데이터 정리

        - 숫자로 변환하기(원-핫 인코딩)

    2. 유저 정보 활용

        - 유저의 취향 파악

        - 유저를 n개의 feature로 표현 -> 아이템과 같은 feature로 표현해야 서로 비교 가능

    3. 벡터의 내적과 유사성

        - 벡터 : n개의 feature로 만들어진 데이터 -> n차원의 벡터로 표현 가능

        - 벡터 내적 : 두 벡터의 동일 위치 값들끼리 곱한 후 모두 더한 값

            - 유사성 : 내적 값이 클수록 두 벡터의 유사성이 높음

            - 0과 1로만 된 벡터라면 내적 = 겹치는 feature의 개수

* 유사도 방식의 종류

| 기반 | 명칭 | 설명 |
| --- | --- | --- |
| 집합 | 자카드 유사도 | 아이템 집합 간의 교집합 크기로 측정 |
| 각도 | 코사인 유사도 | 벡터 사이의 각도로 측정 |
| 거리 | 유클리디안 거리 | 벡터 사이의 거리로 측정, 값이 작을수록 유사함 |
| 상관관계 | 피어슨 상관계수 | 벡터 간 선형 상관관계로 측정 |

* 코사인 유사도 : 두 벡터 사이의 각도가 작을수록 1에 가까워짐 (클수록 비슷)

    - $\cos(A, B) = \cos\theta = \frac{A \cdot B}{\|A\| \, \|B\|}$ -> 내적을 두 벡터의 길이로 나눈 값

    - 길이는 무시하고 방향이 얼마나 같은지만 봄 -> Ex) [1, 1]과 [5, 5]는 길이가 달라도 방향이 같아서 코사인 유사도 = 1

    - 거리처럼 쓰고 싶으면 1에서 빼서 코사인 거리(1 - 코사인 유사도)로 씀

* 내적 vs 코사인

    - 원-핫 벡터에서는 feature를 많이 가진 아이템일수록 1이 많아서 벡터가 길어짐

    - 내적 : 길이까지 반영 -> 장르가 여러 개인 영화처럼 feature가 많은 아이템이 사용자와 겹치는 feature도 많아져서 유리해짐

    - 코사인 : 길이를 무시 -> feature 개수 때문에 생기는 쏠림은 줄지만, 아이템 정보가 풍부하다는 점은 반영 못 함

    -> 어떤 유사도를 쓰냐에 따라 추천 결과의 성격이 달라짐

#### 2-3. TF-IDF (Term Frequency - Inverse Document Frequency)

> 영화 줄거리, 상품 설명 같은 텍스트도 아이템의 중요한 특징 -> 단어가 몇 번 나왔는지만 세면 "그리고", "있다"처럼 어느 문서에나 나오는 단어가 가장 중요한 단어처럼 취급되는 문제

* TF-IDF(Term Frequency - Inverse Document Frequency) : 특정 문서 내에 특정 단어가 얼마나 자주 등장하는 지를 의미하는 단어 빈도(TF) 와 전체 문서에서 특정 단어가 얼마나 자주 등장하는지를 의미하는 역문서 빈도(IDF)를 통해서 "다른 문서에서는 등장하지 않지만 특정 문서에서만 자주 등장하는 단어"를 찾아서 문서 내 단어의 가중치를 계산하는 방법

    - TF(d,t) : 특정 문서 d에서의 특정 단어 t의 등장 횟수

    - DF(t) : 특정 단어 t가 등장한 문서의 수

    - n : 전체 문서의 수

    - $IDF(d,t) = log(\frac{n}{1+DF(t)})$ -> 그 단어가 등장한 문서 수가 많을수록 작아짐

    - $TF\text{-}IDF(d,t) = TF(d,t) \times IDF(d,t)$ -> 높을수록 다른 문서들보다 해당 문서에서 더 중요한 단어

* 예시) 문서 A "고양이는 귀엽다", 문서 B "강아지는 귀엽다", 문서 C "고양이는 사냥한다" (n = 3, log는 밑이 10)

    - "귀엽다" : A, B에 등장 -> DF = 2 -> $IDF = log(\frac{3}{1+2}) = log(1) = 0$ -> A에서 TF-IDF = 1 × 0 = 0 -> 여러 문서에 나와서 중요도 낮음

    - "사냥한다" : C에만 등장 -> DF = 1 -> $IDF = log(\frac{3}{1+1}) = log(1.5) ≈ 0.176$ -> C에서 TF-IDF = 1 × 0.176 = 0.176 -> 다른 문서에 덜 나와서 더 중요한 단어

* 추천에서 쓰는 방법

    1. 영화마다 줄거리를 TF-IDF 벡터로 변환

    2. 사용자가 재밌게 본 영화의 벡터와 다른 영화들의 벡터를 코사인 유사도로 비교

    3. 유사도가 높은 영화를 추천 -> "우주", "탐사" 같은 단어의 TF-IDF가 높은 영화를 좋아했으면, 이 단어들이 비슷하게 중요하게 나오는 다른 SF 영화가 추천됨

#### 2-4. 내용 기반 필터링의 장단점

| **장점** | **단점** |
| --- | --- |
| 아이템이 가지고 있는 **feature 자체**에만 집중하기 때문에 다른 유저의 평점 등에 영향을 받지 않음 (내 기록만으로 프로필을 만들어서 다른 사용자의 데이터가 필요 없음) | item의 feature를 추출해야 하고 이를 통해서 추천하기 때문에 정확한 feature를 추출하지 못하면 정확도가 떨어짐 |
| item의 feature를 통해서 추천을 하기 때문에 추천이 된 이유를 설명하기 용이함 | 기존의 item과 유사한 item 위주로만 추천하기에 **새로운 장르의 item을 추천하기 어려움** (과도한 특화, Over-specialization) |
| 속성만 있으면 신규 아이템도 추천 가능 -> **아이템 콜드 스타트 해결** | 기록이 없는 **신규 사용자**는 추천하기 어려움 (사용자 콜드 스타트) |

> Cold start problem : 충분한 정보가 수집되지 않아서 적절한 제품을 추천해주지 못하는 문제

* 아이템 콜드 스타트 : 새 아이템에 평가가 없음 -> 내용 기반으로 해결 가능

* 사용자 콜드 스타트 : 새 사용자의 기록이 없음 -> 내용 기반으로도 해결 어려움

### 3. 협업 필터링 (Collaborative Filtering)

> 협업 필터링의 가정 : 과거에 비슷한 선호도를 가진 사용자들은 앞으로도 비슷한 선호도를 가질 가능성이 높음

#### 3-1. 메모리 기반 협업 필터링 (Memory-Based Collaborative Filtering)

* 사용자 또는 아이템 간 유사도를 기반으로 함

* 시스템에 저장된 사용자-아이템 평점을 그대로 꺼내서 새 아이템의 평점을 예측 -> 별도의 학습 단계 없이 비슷한 사용자나 아이템, 즉 "이웃"을 찾아 쓰기 때문에 이웃 기반 협업 필터링(Neighborhood-based CF, NBCF)이라고도 부름

* 비교적 초기에 개발된 알고리즘

* 사용자 A와 비슷한 취향을 가진 사용자가 선호하는 아이템을 추천

* 접근법 2가지

    - 사용자 기반 : 특정 사용자와 유사한 다른 사용자를 TOP-N으로 선정해 이 TOP-N 사용자가 좋아하는 아이템을 추천

        - 특정 사용자와 타 사용자 간의 유사도를 측정한 뒤 가장 유사도가 높은 TOP-N 사용자를 추출해 그들이 선호하는 아이템을 추천

        - 유저 간의 유사도가 높을수록 가중치를 부여하는 방식 -> 이웃들의 평점을 유사도로 가중평균해서 내가 아직 평가하지 않은 아이템의 평점을 예측

        - "당신과 비슷한 고객들이 이 상품도 구매했습니다"

    - 아이템 기반 : 사용자들이 그 아이템을 좋아하는지/싫어하는지의 평가 척도가 유사한 아이템을 추천하는 기준이 되는 알고리즘(”아이템간의 속성”이 얼마나 유사한가 X)

        - 추천하고자 하는 target item과 유사한 아이템의 평점을 활용

        - 사용자는 계속 늘고 취향도 바뀌지만 아이템끼리의 관계는 상대적으로 덜 변함 -> 아이템 유사도를 미리 계산해 두고 재사용하기 좋음

        - "이 상품을 선택한 고객들은 이 상품도 구매했습니다"

* 사용자 기반 예시) A는 라라랜드를 좋아할까?

|  | 어바웃타임 | 노팅힐 | 라라랜드 | 아이언맨 |
| --- | --- | --- | --- | --- |
| **A** | 5 | 4 | **?** | 1 |
| **B** | 5 | 5 | 5 | 2 |
| **C** | 1 | 2 | 2 | 5 |

- 사용자 간 유사도 : 평점 데이터에서는 피어슨 상관계수를 많이 씀 ($r_{ui}$ : 사용자 u가 아이템 i에 준 평점, $\bar{r}_u$ : 사용자 u의 평균 평점, i : 두 사용자가 모두 평가한 아이템)

$$
sim(u, v) = \frac{\sum_{i} (r_{ui} - \bar{r}_u)(r_{vi} - \bar{r}_v)}{\sqrt{\sum_{i} (r_{ui} - \bar{r}_u)^2} \sqrt{\sum_{i} (r_{vi} - \bar{r}_v)^2}}
$$

- 예측 평점 : 이웃들이 그 아이템을 자기 평균보다 얼마나 높게(낮게) 줬는지를 유사도만큼 반영해서 내 평균에 더함

$$
\hat{r}_{ui} = \bar{r}_u + \frac{\sum_{v} sim(u, v)\,(r_{vi} - \bar{r}_v)}{\sum_{v} |sim(u, v)|}
$$

- 계산

    1. 유사도 : A와 B는 패턴이 거의 같음 -> $sim(A, B) \approx 0.97$ / A와 C는 정반대 -> $sim(A, C) = -1$

    2. 평균 : A ≈ 3.33, B = 4.25, C = 2.5

    3. 평균과의 차이 : B는 라라랜드에 자기 평균보다 0.75점 높게, C는 0.5점 낮게 줌

    4. $\hat{r}_{A,\text{라라랜드}} = 3.33 + \frac{0.97 \times 0.75 + (-1) \times (-0.5)}{0.97 + 1} \approx 3.96$ -> 약 4점이라 A에게 라라랜드 추천

* 아이템 기반 예시)

    - A : 어바웃 타임 5점, 노팅힐 4점, 아이언맨 3점 / B : 어바웃 타임 5점, 반지의 제왕 2점, 노팅힐 4점 / C : 아이언맨 5점, 어벤져스 4점, 어바웃 타임 2점 / D : 어바웃 타임 5점

    - 어바웃 타임을 높게 준 A, B가 노팅힐도 높게 줌 -> 평가가 같이 움직이니까 어바웃 타임과 노팅힐은 비슷한 아이템

    - C는 어바웃 타임을 낮게 줬으니 노팅힐도 낮게 줄 가능성이 높음 -> 노팅힐 추천 X

    - D는 어바웃 타임을 좋아하니까 노팅힐 추천

| 기준 | **사용자 기반 (User-based)** | **아이템 기반 (Item-based)** |
| --- | --- | --- |
| **정확성** | **아이템 수**가 사용자 수보다 많을 때 유리 | **사용자 수**가 아이템 수보다 많을 때 유리 |
| **공간, 효율성** | 사용자 간 유사도를 저장 -> 사용자가 많으면 부담 | 아이템 간 유사도를 저장 -> 아이템이 많으면 부담 |
| **안정성** | 사용자 변화의 영향을 많이 받음 -> 아이템이 자주 바뀌고 사용자 유입이 적은 환경에 적합 | 아이템 목록이 잘 변하지 않는 환경에서 안정적 |
| **계산 속도** | 추천할 때마다 비슷한 사용자를 새로 찾아야 함 | 아이템 유사도를 미리 계산해 두고 재사용 -> 빠름 |
| **다양성** | 비슷한 사용자가 평가한 독특한 아이템까지 추천 -> 의외성, 참신성이 강점 | 기존에 본 아이템과 비슷한 추천이 많아서 다양성은 상대적으로 부족 |
| **활용 예** | 뉴스, 소셜 미디어(알 수도 있는 친구 추천), 음악 추천 서비스 | 아마존 등 사용자가 많은 대형 플랫폼 |

* 사용자 수가 많으면 아이템 기반이 유리한 이유

    - 보통 사용자 수가 아이템 수보다 훨씬 많음 -> 두 **사용자**가 함께 평가한 아이템은 적지만, 두 **아이템**을 함께 평가한 사용자는 많을 가능성이 높음

    - 새 사용자가 새 아이템보다 자주 추가됨 -> 아이템 간 유사도는 가끔만 다시 계산하면 됨

-> 정답이 있는 건 아니고 일반적인 경향! 새로운 추천이 필요한 플랫폼이면 User-based, 정확하고 안정적인 추천이 중요한 대형 플랫폼이면 Item-based

#### 3-2. 메모리 기반 협업 필터링의 한계

* 희소 행렬 문제

    - 실제로 유저-아이템 행렬 **대부분의 원소는 비어 있음 (sparse matrix, 희소 행렬)**

    - Ex) 넷플릭스 프라이즈 데이터 : 사용자 약 48만 명 × 영화 약 1만 8천 편 = 약 85억 칸인데 실제 평점은 약 1억 개 -> 전체 칸의 약 99%가 비어 있음

    - 두 사용자가 함께 평가한 아이템이 적어서 유사도를 믿기 어려움 (공통 아이템이 1개뿐이어도 코사인 유사도가 1로 나올 수 있음)

    - 비슷한 이웃을 찾지 못하면 예측 자체가 어려움

    **⇒ model-based CF 이용!** (이웃을 직접 비교하는 대신, 관측된 평점으로 사용자와 아이템의 특징을 학습해서 빈칸을 채움 -> 3-3)

* 콜드 스타트 문제

    - 협업 필터링은 **기존의 평점/행동 데이터를 기반**으로 유사도를 계산하는데, 데이터가 아예 없는 경우 추천이 불가능함

| 유형 | 상황 | 설명 |
| --- | --- | --- |
| **User Cold Start** | 신규 유저 유입 | 평가 이력이 없어 유사 유저 탐색 불가 (Ex. 넷플릭스에 방금 가입한 사용자) |
| **Item Cold Start** | 신규 아이템 등록 | 아무도 평가하지 않아 추천 후보에 못 오름 -> 누군가 처음 평가하기 전까지 추천이 안 돼서 first-rater 문제라고도 함 |

* 해결 방법

    - **직접 지정형 프로필** 활용 (가입 시 선호 장르/아이템 직접 선택)

    - **Content-Based Filtering** 과 혼합하여 아이템 특성 기반으로 보완

    - 임베딩을 쓰는 모델이라면 같은 카테고리 아이템들의 임베딩 평균으로 새 아이템을 임시로 표현

    - **인구통계학적 정보** (나이, 성별 등) 활용

* 확장성 문제

    - 아이템/유저가 계속 늘어날 경우, **확장성**이 떨어짐 **(Scalability)**

    - 아이템 i에 대한 평점을 예측하기 위해서는, **아이템 i에 대해 평가를 한 모든 유저와의** 유사도를 구해야 함. **BUT! 유저가 많아지면?** 연산은 계속 늘어나고, 오히려 성능이 떨어지기도 함

    - 데이터를 샘플링하거나 줄이면 빨라지긴 하지만 그만큼 추천 품질도 떨어짐

    **⇒ 따라서 KNN CF 이용!**

    - KNN 협업 필터링 : 모든 사용자를 쓰지 않고, 유사도가 높은 k명의 이웃만 골라서 예측

        - 이웃은 예측하려는 아이템을 실제로 평가한 사용자 중에서만 고를 수 있음

        - k가 너무 작으면 소수의 의견에 예측이 많이 흔들리고, 너무 크면 덜 비슷한 사용자까지 섞여서 예측이 흐려짐

#### 3-3. 모델 기반 협업 필터링 (Model-Based Collaborative Filtering)

> 메모리 기반 CF(NBCF)의 한계(Sparsity, Scalability)를 보완하기 위해 등장한 방식으로, 단순히 유사도를 계산하는 것이 아니라, 데이터로부터 패턴을 학습한 모델을 만들어 평점을 예측함 -> 관측된 평점 데이터를 학습(Training) 하여, 관측되지 않은 평점을 예측(Inference) 하는 방식

* 두 단계로 동작

    - 학습 단계 (Training) : 이미 있는 평점을 가장 잘 설명하는 모델의 파라미터를 찾음 -> 추천과 별도로 미리 해 두고, 새 데이터가 쌓이면 주기적으로 다시 학습

    - 예측 단계 (Inference) : 학습된 파라미터로 비어 있는 칸의 평점을 계산 -> 추천이 필요한 시점에 빠르게 수행

* 메모리 기반 협업 필터링은 관측된 평점 정보만으로 유사도를 계산하고 추천을 수행하지만, 이는 숨겨진 선호 요인이나 아이템의 특성을 반영하지 못하고, 단순히 유사도에 의존하기 때문에 결측이 많은 환경이나 복잡한 상황에서 정확도가 떨어짐

* 잠재 요인(Latent Factor) : 사용자와 아이템 사이의 관계를 설명해주는, 눈에 보이지 않는 숨겨진 특성(요인)

    - Latent Factor Model : 주어진 데이터로는 알 수 없는 사용자-아이템 간의 숨겨진 특성(feature)을 찾기 위해 주어진 Matrix를 사용자-잠재요인 아이템-잠재요인으로 각각 분해하여 학습하는 모델(유저와 아이템을 구분 없이, 동일한 차원의 벡터 공간에 좌표로 표현)

        - 유저와 아이템이 같은 공간에 놓이기 때문에, **서로의 거리나 각도로 유사도 파악 가능**

        - 사용자와 아이템을 보이지 않는 축(=잠재적인 차원)들을 사용해 나타낼 수 있고, 이를 통해 평점을 예측할 수 있다고 가정하는 모델

    - Factor Model vs Latent Factor Model : 축의 개수는 둘 다 사람이 정하지만(Latent는 하이퍼파라미터 k), Factor Model은 축의 의미(Ex. 액션, 로맨스)를 사람이 미리 정하고 Latent Factor Model은 축의 의미가 학습 과정에서 정해짐 -> 사람이 해석하기 어려움

* 행렬 분해 (Matrix Factorization)

    - User-Item Matrix (유저-아이템 행렬)에는 유저의 취향+아이템의 특성이 녹아 있지만, 그 결과값만 볼 수 있음
    
    - 이 행렬을 분해하여 평점을 유저의 잠재 요인과 아이템의 잠재요인의 내적으로 설명할 수 있다면, 이 방법을 통해 평점을 예측할 수 있음

* 행렬 : 숫자를 행과 열에 맞춰 직사각형으로 배열한 것 -> 추천에서는 사용자-아이템 평점 표를 그대로 행렬로 봄 (행 : 사용자, 열 : 아이템, 칸 : 평점)

    - 행렬의 곱 : (3 × 2) 행렬과 (2 × 4) 행렬을 곱하면 (3 × 4) 행렬, 결과의 각 칸은 앞 행렬의 한 행과 뒤 행렬의 한 열을 **내적**한 값

    - 이 성질을 거꾸로 이용해서 큰 평점 행렬을 작은 두 행렬의 곱으로 나타내는 것 = 행렬 분해

$$
\hat{R} = P^T Q
$$

- $\hat{R}$ (m × n) : 예측 평점 행렬
- $P$ (k × m) : 사용자 잠재 요인 행렬, 각 열이 사용자 벡터 $p_u$
- $Q$ (k × n) : 아이템 잠재 요인 행렬, 각 열이 아이템 벡터 $q_i$
- $k$ : 잠재 요인의 개수 (하이퍼파라미터)

$$
\hat{r}_{ui} = p_u^T q_i = \sum_{s=1}^{k} p_{su}\, q_{si}
$$

-> 사용자 u의 아이템 i에 대한 예측 평점은 두 벡터의 내적

* 예시) 잠재 요인이 2개일 때 (첫 번째 요인 = 로맨스, 두 번째 요인 = 액션이라고 가정)

    - 사용자 p = (0.9, 0.2) -> 로맨스를 많이 좋아하고 액션은 조금 좋아함

    - 노팅힐 q = (0.8, 0.1) -> 0.9 × 0.8 + 0.2 × 0.1 = 0.74

    - 존 윅 q = (0.1, 0.9) -> 0.9 × 0.1 + 0.2 × 0.9 = 0.27

    -> 노팅힐 점수가 더 높아서 노팅힐 추천. 요인별로 "사용자 취향 × 영화 특성"을 곱해서 더한 값이라 취향과 영화 특징이 같은 방향으로 클수록 점수가 커짐

    - 로맨스, 액션은 설명하려고 붙인 이름이고, 실제 잠재 요인은 이렇게 깔끔하게 해석되지 않는 경우가 대부분

* 행렬 분해 학습 과정

    1. 잠재 요인 개수 k 정하기 (Ex. k = 2)

    2. 모든 사용자와 아이템의 벡터를 임의의 값으로 시작

    3. **평점이 있는 칸만** 예측하고, 실제 평점과의 오차 계산

    4. 오차가 줄어들도록 벡터를 조금씩 수정 (SGD, ALS 등)

    5. 3~4를 반복 -> 학습된 벡터로 **빈칸의 평점을 예측**

* 손실 함수 : 모델이 얼마나 틀렸는지를 숫자 하나로 나타낸 점수 -> 이 값을 최대한 작게 만드는 게 학습 목표

$$
\sum_{\text{평점이 있는 칸}} (\text{실제 평점} - \text{예측 평점})^2 + \lambda \times (\text{벡터 크기})^2
$$

- 앞부분 (오차) : 평점이 있는 모든 칸에서 예측이 실제 평점에 가까워지게 함
- 뒷부분 (규제) : 벡터 값이 지나치게 커지는 걸 막아서 과적합을 줄임 -> $\lambda$가 클수록 강한 규제

* 벡터를 고치는 방법

    - SGD (확률적 경사 하강법) : 평점을 하나씩 보면서 그때마다 오차가 줄어드는 방향으로 조금씩 수정

    - ALS (교대 최소제곱법) : 사용자 벡터와 아이템 벡터를 번갈아 고정하고 나머지 하나를 계산 -> 병렬 처리에 유리

* 편향 (Bias) : 사람마다 점수를 주는 기준이 다르고, 작품마다 평이 좋은 영화, 낮은 영화가 있음 -> 이런 성향을 예측 식에 편향 항으로 따로 더해 줌

    - $\hat{r}_{ui} = \mu + b_i + b_u + p_u^T q_i$ ($\mu$ : 전체 평균 평점, $b_i$ : 아이템이 평균보다 얼마나 높게(낮게) 평가받는지, $b_u$ : 사용자가 평균보다 얼마나 높게(낮게) 평가하는지)

    - Ex) 전체 평균 3.7점, 타이타닉은 평균보다 +0.5, 깐깐한 사용자 Joe는 평균보다 -0.3 -> Joe의 타이타닉 기본 예측 = 3.7 + 0.5 - 0.3 = 3.9점

    - 편향이 경향을 먼저 설명해 주면, 잠재 요인은 이 사용자와 이 아이템이 얼마나 잘 맞는지에만 집중해서 학습할 수 있음

* 행렬 분해 방식 (위의 학습 과정은 학습 기반 MF)

| 구분 | 접근 방식 | 특징 |
| --- | --- | --- |
| **SVD** | 수학적 분해 기법 | 행렬 대수학에 기반해서 분해, 희소 행렬에는 바로 적용이 안 돼서 결측치 처리가 필요 |
| **학습 기반 MF** | 최적화 알고리즘 (SGD, ALS) | 관측된 평점에 대해서만 학습 -> 희소성 문제를 우회적으로 해결 |
| **NMF** | 제약 조건 기반 모델 | 분해된 행렬에 비음수(Non-Negative) 제약을 추가 -> 잠재 요인의 해석력이 높아짐 |
| **PMF** | 확률적 모델 | 평점이 사용자 벡터와 아이템 벡터의 내적을 평균으로 하는 정규분포를 따른다고 봄 |

* 메모리 기반 vs 모델 기반

| 구분 | 메모리 기반 | 모델 기반 |
| --- | --- | --- |
| 학습 단계 | 없음 | 있음 |
| 예측 방식 | 원본 평점 데이터에서 비슷한 사용자, 아이템을 직접 찾음 | 학습한 벡터를 내적 |
| 희소한 데이터 | 함께 평가한 아이템이 적어서 유사도 근거가 부족 | 전체 패턴을 학습해서 빈칸을 일반화 |
| 확장성 | 추천할 때마다 많은 유사도 계산이 필요 | 학습은 미리 하고 추천할 땐 벡터 내적만 계산 -> 빠름 |
| 콜드 스타트 | 해결 어려움 | 해결 어려움 (상호작용이 없으면 벡터를 학습할 수 없음) |
| 장점 | 추천 이유를 설명하기 쉬움 | 데이터를 압축해서 예측이 빠르고 일반화에 유리 |

-> 행렬 분해로 희소성과 확장성 문제는 많이 해결했지만, 결국 **내적이라는 선형 계산**으로 예측한다는 한계가 남음 -> 딥러닝 기반 추천으로 넘어감

### 4. VAE 기반 협업 필터링 (Variational Autoencoder CF)

> 사용자의 아이템 상호작용 벡터(예: 구매/시청 이력)를 입력으로 받아, 잠재 공간에서 재구성하도록 VAE를 학습하는 방식

* 구조

    - 인코더: 사용자 이력을 잠재변수 z 분포로 매핑 -> 상호작용 벡터 $x_u$를 받아서 평균 μ와 표준편차 σ로 변환 (취향을 고정된 좌표 하나 대신 분포 $q(z \mid x_u)$로 추정)

    - 샘플링: 인코더가 추정한 분포(μ, σ)에서 잠재 벡터 z를 무작위로 뽑아 디코더에 전달

    - 디코더: z로부터 아이템 분포를 복원 -> 아이템별 선호 확률을 출력하고, 확률이 높은 순서로 아직 상호작용하지 않은 아이템 추천

* 장점

    - 기존 MF처럼 단일 포인트 벡터가 아니라 확률적 잠재 분포로 사용자 선호를 모델링 하여 불확실성을 반영

    - 희소한 상호작용 데이터에서도 강건 (기록이 적은 사용자도 과적합을 줄이고 비교적 안정적으로 표현)

* 대표 모델 : Mult-VAE

#### 4-1. MF 의 한계와 딥러닝의 등장

* 선형성 : MF의 예측은 내적, 즉 요인별로 곱해서 더하기만 하는 구조 -> 각 요인이 점수에 따로따로 더해지고, 요인끼리 얽히는 효과는 표현하기 어려움

    - Ex) 로맨스도 좋고 음악 영화도 좋은데 둘이 섞인 뮤지컬 로맨스는 별로인 사용자 -> 내적은 로맨스 점수와 음악 점수를 각각 더하니까 둘 다 높은 영화가 오히려 가장 높은 점수를 받음

* 불확실성 무시 : 일반적인 MF는 사용자마다 벡터를 딱 하나 학습 (점 추정, Point Estimate)

    - 기록이 5개뿐인 사용자와 5,000개인 사용자의 벡터를 똑같이 확신하고 사용 -> 기록이 적은 사용자의 추정이 불안정해도 구분 못 함

| MF의 근본적 한계 | 내용 | 딥러닝의 해법 |
| --- | --- | --- |
| **선형성의 한계** | 평점 예측($\hat{r}_{u, i}$)이 단순한 잠재 벡터의 내적(Dot Product)으로만 이루어집니다. | 비선형 활성화 함수(ReLU 등)를 가진 신경망을 사용하여 복잡한 패턴을 학습합니다. |
| **불확실성 무시** | 사용자 취향을 단 하나의 벡터값(Point Estimate)으로만 표현합니다. | 평균과 분산을 함께 출력해서 **확률 분포**로 취향을 모델링하여 데이터의 불확실성을 반영합니다. |

=> 이 한계를 극복하기 위해 VAE 등장

---

#### 4-2. 왜 AutoEncoder(AE)가 아닌 VAE(Variational AE)인가?

> 딥러닝의 비선형성을 추천 시스템에 도입한 초기 시도: 일반적인 AutoEncoder(AE)
→ VAE(Variational AutoEncoder)는 불확실성을 모델링함으로써 추천 시스템의 고질적인 문제인 희소성(Sparsity)에 대한 해답을 제시

* 오토인코더 (AutoEncoder, AE) : 입력을 작은 벡터로 압축했다가 복원하는 신경망 모델

    - 입력 x (큼) -> 인코더 -> 잠재 벡터 z (작음) -> 디코더 -> 복원된 x̂ (큼)

    - 입력을 복원하는 데 꼭 필요한 핵심 정보만 남기도록 학습, 정답이 없고 입력 자체가 정답

    - AutoRec : 오토인코더를 협업 필터링에 적용한 초기 모델 (아이템 하나의 평점 벡터를 입력해서 복원)

        - 은닉층을 선형으로만 쓰면 RMSE 0.872, sigmoid를 쓰면 0.831 (RMSE는 낮을수록 좋음) -> 신경망의 비선형성이 추천 성능을 실제로 높일 수 있음

    - AE도 사용자를 점 하나로 표현한다는 점은 MF와 같음

* VAE (Variational AE) : 가운데 잠재 벡터를 점 하나 대신 분포로 표현하는 오토인코더

    - 평균 μ : 사용자 취향의 중심 위치

    - 표준편차 σ : 그 위치를 얼마나 확신할 수 없는지 (불확실성)

| 구분 | AutoEncoder (AE) | Variational AutoEncoder (VAE) |
| --- | --- | --- |
| **잠재 벡터 표현** | 단일 값 (Point Estimate) | **확률 분포** (평균 $\mu$, 분산 $\sigma$) |
| **데이터 불확실성** | 반영 불가 (단순 복원에 집중) | **데이터의 불확실성을 모델링**하여 안정성 확보 |
| **적용 시 장점** | 비선형성 학습 | 희소한 데이터에서 더욱 안정적이며, 일반화 성능 우수 |

* 예시) 본 영화가 5편뿐인 사용자 (로맨스 3편, SF 1편, 공포 1편)

| 방식 | 사용자 표현 | 결과 |
| --- | --- | --- |
| AE (점 추정) | z = [1.2, -0.7] | 5편만으로 위치를 하나로 확정 -> 우연히 본 공포 영화 같은 노이즈까지 그대로 반영될 수 있음 |
| VAE (분포 추정) | μ = [1.2, -0.7], σ = [0.8, 0.9] | 학습 때 주변 z를 샘플링해서 비슷한 위치에서는 비슷한 추천이 나오도록 학습됨 |

* VAE가 더 안정적인 이유

    1. 점 하나 대신 그 주변을 사용해 학습 : 학습할 때마다 μ 근처에서 z를 무작위로 뽑아 디코더에 넣음 -> μ에서 조금 벗어난 위치에서도 비슷한 추천이 나오도록 학습됨

    2. 기준 분포 쪽으로 당겨짐 : KL 항이 분포를 표준 정규분포 쪽으로 당겨서, 기록 몇 개만으로 극단적인 위치로 가기 어려움 (4-4)

#### 4-3. VAE 기반 CF (Mult-VAE)의 작동 구조

> 사용자의 아이템 상호작용 벡터($x$, 예: 구매/시청 이력)를 입력으로 받아, 잠재 공간에서 재구성하도록 학습하는 방식

| 구성 요소 | 역할 |
| --- | --- |
| **입력** ($x$) | 사용자의 아이템 상호작용 벡터 -> 아이템 수만큼 긴 0/1 벡터 (평점 데이터는 4점 이상만 1로 바꿔서 사용) |
| **인코더** | 사용자 이력을 잠재변수 $z$ 분포로 매핑 (평균 μ와 표준편차 σ 출력) |
| **잠재 벡터** ($z$) | μ, σ로 정해진 정규분포에서 샘플링된 최종 취향 벡터 |
| **디코더** | $z$로부터 모든 아이템의 점수를 계산한 뒤 softmax로 확률로 변환 |

---

#### 4-4. VAE 의 비밀 병기 : 목적 함수와 KL Divergence

> 목적 함수인 ELBO(Evidence Lower Bound)에 있으며, 이를 통해 두 가지 목표(정확한 복원, 잠재 공간 정규화)를 동시에 최적화

* ELBO : 데이터가 나올 로그 확률의 하한(lower bound)을 키우는 방식 -> VAE는 이 값을 최대화

$\mathcal{L}(x; \theta, \phi) = \underbrace{E_{q_\phi(z|x)} [ \log p_\theta(x|z) ]}_{\text{재구성 오차 최소화 (Reconstruction Loss)}} - \underbrace{D_{\text{KL}}(q_\phi(z|x) || p(z))}_{\text{잠재 공간 정규화 (KL Divergence)}}$

- 재구성 오차 최소화 (첫 번째 항):

    - 목표: 디코더가 잠재 벡터 $z$를 사용하여 원래 입력 $x$를 최대한 비슷하게 복원하도록 만듦 (AE의 역할과 동일)

- KL Divergence 항 (두 번째 항):

    - 목표: 인코더가 출력하는 잠재 분포 $q(z|x)$가 사전 분포 $p(z)$ (표준 정규 분포)와 유사하도록 강제
    - 역할: 이 항은 모델이 특정 데이터에 과적합하여 잠재 공간을 불규칙하게 만드는 것을 방지하고, 잠재 공간을 넓고 부드럽게(Smooth) 유지하도록 함

        - AE : 퍼즐 조각들 : 유저 벡터들이 의미 없는 공간에 제멋대로 흩어져 있음

        - VAE : 맞춰진 퍼즐: KL Divergence가 유저 벡터들을 표준 정규 분포 주변에 부드럽고 연결된 상태로 유지하도록 강제

- 시험 공부로 비유하면 첫 번째 항은 "기출 문제를 많이 풀어라", 두 번째 항은 "기본 개념에서 벗어나지 마라" -> 첫 번째만 있으면 기출 답만 외워서 처음 보는 문제를 못 풀고, 두 번째만 있으면 모든 문제에 같은 답을 쓰게 됨

* KL Divergence : 두 확률분포가 얼마나 다른지를 나타내는 값

    - VAE에서는 사용자의 분포가 N(0, 1)과 같으면 0, 평균이 중심에서 멀어지거나 분산이 너무 작거나 크면 값이 커짐

    - 두 분포가 모두 정규분포면 식으로 바로 계산 가능

$$
D_{KL}\big(\mathcal{N}(\mu, \sigma^2) \,\|\, \mathcal{N}(0, 1)\big) = \frac{1}{2}\left(\mu^2 + \sigma^2 - \ln \sigma^2 - 1\right)
$$

| 사용자 분포 | KL | 의미 |
| --- | --- | --- |
| μ = 0, σ = 1 | 0 | 표준 정규분포와 같음, 벌점 없음 |
| μ = 2, σ = 1 | 2.0 | 중심에서 멀리 떨어짐 |
| μ = 0, σ = 0.1 | 약 1.81 | 너무 좁게 확신함 |
| μ = 0, σ = 3 | 약 2.90 | 너무 넓게 퍼짐 |

-> 중심에서 멀어지거나, 지나치게 확신하거나, 지나치게 퍼지면 벌점이 커짐

* KL 항이 만드는 잠재 공간 차이

| 구분 | AE | VAE |
| --- | --- | --- |
| 잠재 공간 모양 | 정확하게 복원만 하면 돼서 사용자 벡터가 듬성듬성 흩어지고, 벡터 사이의 빈 공간은 학습되지 않음 | KL 항이 분포를 N(0, 1) 쪽으로 당겨서 사용자들이 중심 근처에 연속적으로 모임 |
| 기록이 적은 사용자 | 몇 개 안 되는 기록만으로 위치가 확정 -> 우연한 기록(노이즈) 때문에 엉뚱한 곳에 놓이기 쉬움 | 근거가 부족한 만큼 중심 쪽으로 당겨져서 비슷한 사용자들 근처에 안정적으로 놓임 |

-> 기록이 적은 사용자도 비슷한 취향의 사용자들 근처에 놓여서, 기록이 많은 사용자들로부터 학습된 패턴을 같이 활용할 수 있음 -> VAE 기반 CF가 희소한 데이터에 강한 이유

- VAE-CF의 주요 장점

    -  확률적 잠재 분포 모델링 (불확실성 반영)

        - 기존 MF처럼 단일 포인트 벡터가 아니라 확률적 잠재 분포로 사용자 선호를 모델링하여, 예측 시의 불확실성을 반영하고 노이즈의 영향을 줄임

    - 희소한 상호작용 데이터에서도 안정적

        - KL Divergence를 통해 잠재 공간을 정규화하여, 기록이 적은 사용자도 비슷한 사용자들 근처에 안정적으로 표현되며 희소성 문제에 효과적으로 대응

    - 비선형성 학습

        - 신경망 구조를 통해 MF의 한계였던 복잡하고 비선형적인 패턴을 학습하여 추천 성능을 높임

### 5. GNN 기반 사용자 선호 예측

> CF, MF, VAE는 사용자와 아이템의 상호작용을 행렬의 칸 하나하나로 다룸 -> "A가 본 영화를 B도 봤고, B는 영화 Y도 봤다"처럼 여러 상호작용이 이어진 관계는 칸 하나만으로 직접 활용할 수 없고, 행동의 순서도 고려하지 않음

-> 이 '관계' 자체를 데이터로 보고 모델링하자

#### 5-1. GNN(Graph Neural Network, 그래프 신경망)

> 사용자와 아이템의 상호작용을 ‘누가(User) 무엇을(Item) 좋아했다’의 집합으로 보는 것을 넘어, 이들이 서로 연결된 거대한 관계망(Graph)을 형성

* 사용자-아이템 상호작용은 원래 그래프 : 사용자와 아이템은 점, "봤다", "샀다"는 행동은 둘을 잇는 선

    - 행렬에서 칸 하나였던 정보가 그래프에서는 경로가 됨 : 나 - 영화 X - (X를 본) 다른 사용자 - 영화 Y

    - 이 경로를 따라가면 나와 비슷한 사람이 좋아한 다른 영화를 자연스럽게 찾을 수 있음

* 그래프(Graph) : 노드(Node)와 그 노드를 잇는 선(Edge)을 모아 구성한 자료 구조

    - 노드 : 현실 세계의 개체 (사용자, 아이템, 사람, 원자 등)

    - 엣지 : 개체 간 관계나 상호작용

        - 방향을 가질 수 있음 : 'A가 B를 팔로우'와 'B가 A를 팔로우'는 다름

        - 가중치를 가질 수 있음 : 평점 5점, 구매 3회 등

        - Ex) 소셜 네트워크 : 노드 = 사용자, 엣지 = 친구 관계(방향 없음) 또는 팔로우(방향 있음), 가중치 = 메시지를 주고받은 횟수

    - 그래프 사용하는 이유

        - 관계, 상호작용과 같이 다소 추상적인 개념을 다루기에 적합 -> "취향이 비슷하다", "함께 구매한다" 같은 관계가 [노드 - 엣지 - 노드] 형태로 저장돼서 관계 자체를 직접 학습 가능

        - Non-Euclidean space도 표현하고 학습할 수 있음

| 데이터 | 구조 | 이웃의 형태 |
| --- | --- | --- |
| 이미지 | 픽셀이 바둑판처럼 배열된 격자 | 모든 칸이 상하좌우 이웃을 가지고 순서가 고정됨 |
| 텍스트 | 단어가 일렬로 놓인 시퀀스 | 앞뒤 이웃이 정해져 있음 |
| SNS, 분자, 추천 | 그래프 | 노드마다 이웃 수가 제각각이고 정해진 순서나 좌표가 없음 |

- Euclidean space : 좌표와 거리가 정의된 공간 -> 이미지, 텍스트처럼 규칙적인 격자 위에 놓인 데이터, CNN이나 RNN 적용 가능
- Non-Euclidean space : 고정된 좌표나 격자가 없는 공간 -> 노드 번호를 바꿔도 같은 그래프
- 그래프를 행렬, 벡터로 억지로 펼치면 정보가 손실되거나 비효율이 생김 -> 노드와 엣지 구조를 그대로 쓰는 GNN으로 학습

#### 5-2. GNN 의 구조

* 그래프 구조 생성 : 노드는 개별 객체, 엣지는 객체 간 관계나 연결성을 의미 -> 각 노드에 초기 피처(임베딩) 부여

* 메시지 패싱 : 각 노드가 이웃 노드로부터 정보를 받아 전달

    - 메시지 생성 : 이웃 노드에 전달할 메시지 생성

    - 집계 : 받은 여러 이웃의 메시지를 하나로 합침 (평균, 합, 최대값, attention 등)

    - 갱신 : 자신의 노드 상태를 업데이트 함 (집계 결과 + 자신의 이전 임베딩 -> 새 임베딩)

* 최종 출력

> 내 이웃 노드(나와 비슷한 사용자, 내가 본 아이템)의 정보를 나에게 전달하고 요약하는 게 GNN의 기본 아이디어

#### 5-3. GNN 의 프로세스

> GNN은 그래프를 입력 받아 메시지 패싱(Message Passing)을 수행. 이때, 각 노드는 자신의 이웃 노드들의 정보(잠재 요인)를 수집하여 자신의 은닉 상태(임베딩)를 업데이트

* 이 과정을 K번 반복하여 더 멀리 있는 이웃의 정보까지 반영 -> 이웃들의 취향까지 반영한 최종 임베딩을 얻고, 두 노드 간 선호도를 예측할 수 있음

* 그래프 구조 생성 : 노드와 엣지를 정의한 후, 각각의 피처를 담아 표현

* 메시지 패싱 : 그래프에서 노드 간 정보를 전달하고 상호작용하는 방법

    - 보통 하나의 GNN 레이어에서 ‘메시지 생성 - 집계 - 갱신’ 의 세 단계로 이루어짐(k개의 레이어로 구성되어 있다면 k번의 과정 반복됨)

    - 임베딩을 갱신하는 대상 노드 = 타깃 노드, 모든 노드가 한 번씩 돌아가면서 타깃 노드가 됨

    - 1번 하면 한 단계 떨어진 이웃의 정보가, l번 반복하면 l단계만큼 떨어진 이웃의 정보까지 반영됨

    - 메시지 생성 : 각 노드가 이웃에게 보낼 자신의 정보를 생성

    - 집계 : 메시지 생성 이후, 여러 이웃 노드에서 받은 메시지를 하나의 벡터로 통합

    - 갱신 : 집계된 이웃 정보와 자신의 이전 상태를 결합해 새로운 임베딩을 계산

* 예시) 넷플릭스 평점 데이터로 그래프 만들기

| 사용자 | 사용자 특징 | 영화 | 영화 특징 | 평점 |
| --- | --- | --- | --- | --- |
| U1 | 20대 | A | 액션, 평균 평점 4.3점 | 5 |
| U1 | . | B | 스릴러, 평균 평점 4.1점 | 4 |
| U2 | 30대 | A |  | 3 |
| U2 | . | C | SF, 평균 평점 4.8점 | 5 |
| U3 | 20대 | C |  | 4 |

- 노드 : 사용자 3개(U1, U2, U3) + 영화 3개(A, B, C) = 6개
- 엣지 : 평점 기록 5개 (Ex. U1이 A에 5점 -> U1 -(5점)- A)
- 전체는 B - U1 - A - U2 - C - U3 로 이어진 하나의 그래프

* 인접 행렬 (Adjacency matrix) : 노드 간 관계 정보, 노드가 6개라서 6 × 6

|  | U1 | U2 | U3 | A | B | C |
| --- | --- | --- | --- | --- | --- | --- |
| U1 | 0 | 0 | 0 | 1 | 1 | 0 |
| U2 | 0 | 0 | 0 | 1 | 0 | 1 |
| U3 | 0 | 0 | 0 | 0 | 0 | 1 |
| A | 1 | 1 | 0 | 0 | 0 | 0 |
| B | 1 | 0 | 0 | 0 | 0 | 0 |
| C | 0 | 1 | 1 | 0 | 0 | 0 |

* 피처 행렬 (Feature matrix) : 관계 정보를 뺀 노드 고유의 feature, 6개 노드 × 5개 피처 (쓸 피처는 임의로 고를 수 있음)

|  | 20대 | 30대 | 액션 | SF | 평균평점 4.5 이상 |
| --- | --- | --- | --- | --- | --- |
| U1 | 1 | 0 | 0 | 0 | 0 |
| U2 | 0 | 1 | 0 | 0 | 0 |
| U3 | 1 | 0 | 0 | 0 | 0 |
| A | 0 | 0 | 1 | 0 | 0 |
| B | 0 | 0 | 0 | 0 | 0 |
| C | 0 | 0 | 0 | 1 | 1 |

* U1이 타깃 노드일 때 메시지 패싱

| 단계 | U1의 경우 |
| --- | --- |
| 메시지 생성 | 이웃인 영화 A, B가 각자의 임베딩으로 메시지를 만듦 |
| 집계 | A, B의 메시지를 평균 내서 하나의 벡터로 요약 |
| 갱신 | U1의 이전 임베딩과 요약 벡터를 결합해서 U1의 임베딩 갱신 |

-> 같은 과정이 U2, U3, A, B, C 모든 노드에서 동시에 수행됨

* 수식

    - 메시지 생성 : $m_u^{(l)} = MSG^{(l)}(h_u^{(l-1)})$ ($h_u^{(l-1)}$ : 이전 레이어 결과인 u 노드의 임베딩, $m_u^{(l)}$ : 현재 레이어에서 계산된 u 노드의 메시지)

    - 집계 : $h_v^{(l)} = AGG^{(l)}(\{m_u^{(l)} \mid u \in N(v)\})$ (N(v) : v의 이웃 노드들)

    - 갱신 : $h_v^{(l)} = Combine(AGG^{(l)}(\{m_u^{(l)} \mid u \in N(v)\}),\ m_v^{(l)})$ -> 이웃 요약과 자기 정보를 하나의 벡터로 결합한 뒤, $h_v^{(l+1)} = \sigma(W^{(l)} \cdot h_v^{(l)})$처럼 신경망 레이어와 활성화 함수를 통과하면서 갱신 완료

#### 5-4. LightGCN (Light Graph Convolutional Networks)

* GNN 아키텍처들의 차이는 두 가지로 결정됨 : 메시지를 어떻게 만드는가(가중치 행렬을 곱하는가, attention을 쓰는가), 메시지를 어떻게 합치는가(평균, 가중 평균, 샘플링)

| 아키텍처 | 작동 방식 | 특징 |
| --- | --- | --- |
| **GCN** (Graph Convolutional Network) | 이웃 임베딩을 연결 수(차수)로 정규화해서 평균 | 가장 기본적인 구조, 학습 시 전체 그래프가 필요 |
| **GRN** (Graph Recurrent Network) | RNN 구조를 GNN에 결합 | 시간에 따라 변하는 그래프(동적 관계)를 포착 |
| **GAT** (Graph Attention Network) | 이웃마다 attention으로 중요도를 학습해서 가중 평균 | 중요한 이웃에 집중하지만 계산량이 늘어남 |
| **GraphSAGE** | 이웃을 일부만 샘플링하고, 이웃 정보를 더하지 않고 자기 정보와 concat | 대규모 그래프, 학습 때 없던 새 노드에도 적용 가능 |

- 비유) 카페 사장이 단골 민수가 신메뉴를 주문할지 민수 친구들 의견으로 예측할 때, GNN은 친구 5명 의견을 똑같이 평균 내고 GAT는 평소 민수와 주문이 많이 겹치는 친구 의견에 더 큰 비중을 둠
- GNN은 그래프 데이터를 학습하는 신경망 전체를 가리키는 개념이고, GCN은 그중 합성곱(Convolution) 연산으로 이웃 정보를 집계하는 구체적인 모델
- 추천에서 가장 널리 쓰이는 건 GCN 계열 -> GCN을 사용자-아이템 그래프에 적용한 NGCF(2019)가 먼저 나오고, 이를 단순화한 LightGCN(2020)이 뒤이어 등장

* LightGCN : GNN 의 한종류로 추천 시스템에 특화되어 복잡성을 제거하고 최적화한 버전

    - 탄생 배경 : NGCF는 노드 분류용 GCN의 연산을 그대로 써서 레이어마다 가중치 행렬 W를 곱하고 비선형 활성화 함수를 적용함 -> 이 두 연산은 노드에 텍스트나 속성 같은 풍부한 피처가 있을 때 유용한데, 추천 그래프의 노드에는 사용자 ID, 아이템 ID 임베딩만 있음 -> 실험해 보니 추천 성능에 도움이 안 되고 오히려 학습을 방해함

    - 특징 변환(Feature Transformation) 제거

        - 레이어마다 복잡했던 학습 파라미터 ($W^{(l)}$) 삭제

    - 비선형 활성화 함수 (Non-linear Activation) 제거

        - ReLU와 같은 활성화 함수를 사용하지 않고, 이웃 정보를 합침 → 주로 정규화된 합 혹은 평균 사용

        - 제거한 이유 : 임베딩 전파 과정에서 이미 비선형성이 충분히 반영되고, 과적합도 막을 수 있음

    - 정규화된(Normalized) 이웃 임베딩의 합(Aggregation) + 레이어 결합만으로 구성 -> CF의 본질에 충실하게 ‘그래프 구조’와 ’전파’라는 뼈대만 남긴, 매우 가볍고 효율이 좋은 모델

    - 같은 실험 조건에서 NGCF보다 평균 약 16% 높은 성능 (He et al., SIGIR 2020)

#### 5-5. GNN 활용 사례

* SNS(Pinterest)

    - 사용자가 저장한 이미지(핀)와 핀을 모아 둔 보드를 노드로, '핀이 보드에 저장됨'을 엣지로 하는 그래프

* 핀테크, 금융 및 보안 : 사기 및 자금 세탁 방지, 이상 거래 탐지, 악성 사용자 분류

* Traffic : 한 장소에서 다른 장소까지 시간이 얼마나 걸리는지 예측

* 프로그램 분석 : 프로그램 안에서 코드 auto suggestion

#### 5-6. GNN 의 장단점

* 장점

    - 고차원 협업 시그널 (High-order Collaborative Signals) 포착 : 단순히 '나'의 잠재 요인(MF, VAE) 분석을 넘어, 나와 취향이 비슷한 사용자들(이웃)과 내가 본 아이템과 연관된 다른 아이템들의 정보를 직접 반복해서 반영할 수 있음

    - 정형화되지 않은 데이터 학습 가능 : GNN은 ‘관계’를 학습하기 때문에, 상호작용, 소셜 관계, 지식 그래프, 사용자 속성을 하나의 그래프에 함께 담아 학습할 수 있음

    - Cold Start 문제 완화에 효과적 : 상호작용이 없는 신규 아이템도 속성 노드나 이웃 정보를 통해 임베딩을 얻을 수 있음

* 단점

    - 과평탄화 (Over-smoothing) 문제 : 모든 노드의 임베딩이 같은 값으로 수렴하는 현상 -> 메시지 전달을 많이 반복하여 GNN 모델의 레이어가 많아질수록, 연결된 그래프 내의 모든 노드 간의 임베딩이 비슷해지는 문제가 발생

    - 과압축 (Oversquashing) 문제 : GNN의 층이 깊어질수록 집계하는 이웃의 범위가 기하급수적으로 커지는데, 이를 고정된 크기의 벡터 하나에 모두 압축하려고 하면 먼 노드의 정보가 손실됨

    - 계산 복잡도 ↑ : 실제 세계에서는 수십억 개의 노드와 엣지가 포함된 그래프에 GNN을 적용해야 하므로 막대한 자원과 메모리가 요구

    - 순서 반영의 어려움 : GNN은 모든 상호작용을 집합으로 보지만, 순서를 반영하지는 못함 -> '신발 → 바지 → 셔츠' 순서로 산 것과 '셔츠 → 바지 → 신발' 순서로 산 것이 같은 그래프가 돼서 다음 행동을 예측하기 어려움

### 6. Transformer 기반 추천 시스템

#### 6-1. 기존 추천 시스템의 순차적 한계

* CF의 순차적 한계 

    - **순서 무시:** [신발 → 바지 → 셔츠]를 구매한 사용자와 [셔츠 → 바지 → 신발]을 구매한 사용자를 동일하게 간주

    - **문맥 부족:** '방금' 구매한 아이템과 '1년 전'에 구매한 아이템의 중요도를 구분하기 어려움

    - **예측의 한계:** 사용자의 '다음' 행동을 예측하기보다는 '일반적인' 선호도를 찾음

* GNN 의 순차적 한계

    - **지역적 정보 전파의 한계**

        - GNN은 기본적으로 메시지 패싱 통해 관계를 학습
        
        - 따라서, 세션 기반 GNN이 [A → B → C] 순서를 방향성 있는 그래프로 표현하더라도, 'C'가 'A'의 정보를 받으려면 'B'를 거쳐야 함 (레이어 2개 필요)

        - 이 과정에서 정보가 여러 단계를 거치며 희석되고 압축될 수 있음 (과압축 문제)

    - **장기 의존성 문제**

        - GNN은 레이어 수만큼의 단계까지만 볼 수 있어서 멀리 떨어진 행동 간의 관계를 포착하기 어려움

        - 이 방식은 Transformer의 Self-Attention처럼 '3일 전 클릭(A)'과 '방금 전 클릭(E)'의 관계를 동적으로 직접 계산하는 데 한계가 있음
    
        - GNN은 정의된 그래프 구조(이웃)에 의존하는 반면, 어텐션은 시퀀스 내 모든 아이템 간의 관계를 거리에 상관없이 한 번에 유연하게 계산

-> CF는 '누가 무엇을 좋아하는가', GNN은 '무엇이 무엇과 연결되었는가'를 잘 포착하지만, 둘 다 사용자가 '어떤 순서로, 지금 어떤 의도로' 행동하는지는 보지 못함 -> 시퀀스 전체를 한 번에 보면서 지금 맥락에 중요한 과거 행동에 집중하는 Transformer가 대안으로 등장

---

#### 6-2. Transformer: 사용자 행동을 순서(Sequence)로 접근

> 행동 순서(문맥)가 중요한 이유: 의도의 파악 -> Transformer는 시퀀스 내의 문맥과 핵심 의도를 파악하는 데 강력

* 사용자의 최근 행동 시퀀스에는 지금 세션의 탐색, 구매 목적이 반영되어 있음 -> 같은 사용자라도 세션마다 보는 상품군이 달라서, 다음 행동은 장기적인 평균 취향보다 **최근 행동의 순서와 조합(문맥)**에 더 큰 영향을 받음

* 예시 1) 문맥에 따라 달라지는 같은 아이템의 의미 (Self-Attention)

|  | 문맥 ① | 문맥 ② |
| --- | --- | --- |
| 행동 순서 (클릭 시퀀스) | 수영복 → 비치타월 → **선크림** | 스킨 → 로션 → **선크림** |
| Transformer의 해석 | 선크림이 수영복, 비치타월과 강하게 attention 됨 | 선크림이 스킨, 로션과 강하게 attention 됨 |
| 해석된 선크림의 의미 | 휴가 준비물 | 데일리 스킨케어 단계 |
| 다음 추천 결과 | 방수팩, 래시가드 | 클렌징폼, 선크림 리필 |

* 예시 2) 시간 순서를 뛰어넘는 핵심 의도 포착 (Long-Range Dependency)

    - **프린터** → 양말 → 라면 → 폰케이스 → 생수 → 샴푸 → ( ? )

    - 프린터와 지금 시점 사이에 관련 없는 행동이 5개 있음 -> Transformer는 지금 시점과 과거 모든 아이템의 관련성을 한 번에 계산해서, 거리와 관계없이 프린터에 높은 가중치를 줄 수 있음 -> 토너처럼 프린터와 관련된 상품 추천

    - 순서대로 정보를 전달하는 RNN, GNN은 5단계를 거치면서 프린터의 정보가 희석됨

-> 두 예시 모두 현재 아이템과 다른 아이템 간의 관련성을 계산해서, 관련성이 높은 아이템의 정보를 더 많이 반영하는 방식 = Self-Attention

---

#### 6-3. Transformer의 핵심: Self-Attention

> 시퀀스 안의 모든 아이템 쌍의 관련성을 거리와 상관없이 동시에 계산 -> 두 아이템 사이에 행동이 많이 끼어 있어도 직접 연결할 수 있어서 장기 의존성 문제 해결

| 개념 | 정의 | 역할 |
| --- | --- | --- |
| Query (Q) | 아이템 임베딩 × 가중치 행렬 $W_Q$ | 현재 아이템이 다른 아이템과의 관련성을 계산하는 기준 |
| Key (K) | 아이템 임베딩 × $W_K$ | Query와 내적되어 관련성 점수를 결정 |
| Value (V) | 아이템 임베딩 × $W_V$ | 관련성 점수에 따라 가중합되는 실제 정보 |

$$
Attention(Q, K, V) = softmax\left(\frac{Q K^T}{\sqrt{d}}\right) V
$$

* 예시) 수영복, 비치타월, 양말이 있는 시퀀스에서 선크림 기준으로 계산

    1. Attention Score (Q · K) : 선크림의 Query와 각 아이템의 Key를 내적 -> 수영복 2.0 / 비치타월 1.5 / 양말 0.1

    2. 가중치 변환 (Softmax) : 합이 1이 되는 비율로 바꿈 -> 수영복 57.0% / 비치타월 34.5% / 양말 8.5%

    3. 가중합 (Score · V) : 선크림의 출력 = 0.570 × V(수영복) + 0.345 × V(비치타월) + 0.085 × V(양말)

    -> 수영복과 비치타월 정보가 약 92%를 차지해서 '휴가 준비물'로 해석됨. 관련성에 따라 정보를 반영하는 비율을 다르게 주는 게 Self-Attention

#### 6-4. Transformer 기반 주요 추천 시스템 모델

* SASRec : 단방향 어텐션으로 과거 정보만 활용하여 미래 행동을 추측하는 방식으로 훈련을 받은 추천 시스템 모델

    - **방식:** 단방향(Uni-directional) 어텐션 (GPT와 유사)
    
    - **어텐션 계산:** 현재 아이템은 '오직 과거의 아이템'들만 참조할 수 있음 (미래 정보 차단)
    
    - **목표:** 다음 아이템 예측 (Next-item prediction)

        - `[신발, 바지, 셔츠]` 시퀀스가 주어지면, `[??]` 다음에 올 아이템을 예측

    - **구조:** 시간순 행동 시퀀스 -> Embedding Layer (아이템 임베딩 + 위치 임베딩) -> Self Attention Layer (과거 아이템만 참조) -> FFN (위치별 비선형 변환) -> Prediction Layer (출력 벡터와 아이템 임베딩을 내적해서 다음 아이템 점수 계산)

    - **특징:**
        
        - 과거 정보만 활용하여 시간의 흐름을 명확하게 반영
        
        - 학습 과정에서 사용자의 '다음 행동'을 예측하는 데 초점이 맞춰져 있음

        - 적응적 집중 : 데이터가 희소하면 최근 아이템에, 풍부하면 긴 과거 맥락에 집중하도록 attention을 조절

        - 빠른 학습 : RNN처럼 순서대로 계산하지 않고 모든 위치를 병렬로 처리
    
    - **문제점 :**
        
        - **계산량 :** attention 연산량이 시퀀스 길이의 제곱(n²)에 비례해서 긴 행동 이력을 다루기 부담스러움
        
        - **단방향성 :** 이전에 등장한 값만 참고할 수 있고 미래의 값은 참고할 수 없음 -> 시퀀스 앞부분의 아이템은 활용할 정보가 적고, 장바구니에 동시에 담은 물건처럼 실제 행동 순서가 엄격하지 않은 경우도 있음

**→ 이 중 단방향성 문제를 보완하기 위해 등장한 알고리즘이 - BERT4rec**

* BERT4Rec : 양방향 어텐션으로 과거 정보와 미래 정보를 활용하여 예측하는 방법을 훈련한 모델

    - **방식:** 양방향(Bi-directional) 어텐션 (BERT와 유사)
    
    - **어텐션 계산:** 시퀀스의 일부를 `[MASK]` 토큰으로 일부러 가리고, 가려진 아이템이 '과거와 미래의 모든 아이템'을 참조할 수 있음 

        - `[MASK]`로 가리는 이유 : 양방향으로 학습하면 모델이 target 아이템의 정보를 직접 보게 되는 정보 누출(information leakage)이 생길 수 있기 때문
        
    - **목표:** 시퀀스 내 가려진 아이템 예측 (Masked-item prediction)

        - `[노트북, 마우스, [MASK], 키보드]` -> 앞뒤 문맥으로 `[MASK]` = 노트북 거치대를 예측

    - **특징:**

        - 아이템의 양방향 문맥(과거와 미래)을 모두 고려
        
        - 학습 과정에서 시퀀스 전체의 선호도를 파악하는 데 더 초점이 맞춰져 있음

        - 한 시퀀스에서 가리는 위치를 바꿔 가며 학습 샘플을 여러 개 만들 수 있음

* SASRec vs. BERT4Rec

    - SASRec: 단방향 어텐션 → 최대한 현실과 유사한 방식으로 과거 데이터만 보고 학습

    - BERT4Rec: 양방향 어텐션 → 아이템 간의 숨겨진 복잡한 맥락을 훨씬 더 깊고 풍부하게 학습

| 구분 | SASRec | BERT4Rec |
| --- | --- | --- |
| 어텐션 방향 | 단방향 (과거만) | 양방향 (과거 + 미래) |
| 학습 목표 | 다음 아이템 예측 | 가려진 아이템 예측 |
| 학습 효율 | 빠름 | 상대적으로 느림 |

* BERT4Rec이 항상 더 좋을까?

    - 원 논문 (Sun et al., CIKM 2019) : BERT4Rec이 SASRec보다 높은 성능을 보였다고 보고

    - 후속 연구 (Klenitskiy & Vasilev, RecSys 2023) : 두 모델은 손실 함수가 달랐음 -> 같은 손실 함수로 맞추니 SASRec이 비슷하거나 더 높은 성능을 보였고 학습도 더 빠름 (Beauty 데이터셋 NDCG@10 기준 개선된 SASRec 0.0327, BERT4Rec 0.0187)

    -> 성능 차이는 어텐션 방향보다 학습 설정의 영향이 컸음, 상황에 따라 모델을 골라야 함

| 상황 | 유리한 모델 | 이유 |
| --- | --- | --- |
| 상호작용이 많고 '다음 행동' 예측이 목적 (실시간 피드, 커머스) | SASRec | 학습, 추론 효율이 높고 손실 함수를 맞추면 성능도 동등 이상 |
| 상호작용이 적어서 시퀀스가 짧은 경우 | 상황에 따라 다름 | BERT4Rec은 한 시퀀스에서 마스킹 샘플을 여러 개 만들 수 있어 데이터 증강 효과가 있지만, 손실 함수를 맞추면 차이가 줄어든다는 반론도 있음 |
| 행동 순서가 느슨한 도메인 (장바구니에 동시에 담기 등) | BERT4Rec | 양방향 문맥이 의미 있음 |

### 7. 제언: 오늘날의 추천 시스템

#### 7-1. LLM의 시대

> 트랜스포머가 시퀀셜 추천의 문을 열었다면, 지금은 LLM이 추천의 패러다임을 바꾸는 중 -> 추천을 랭킹 문제가 아닌 언어 문제로 풂

* 과거 : 조합별로 미리 학습된 공식에 따라 점수 출력 -> Ex) (사용자 A, 아이템 123) -> 0.8점

* LLM : 점수 계산 대신 문장으로 출력

    - ID 대신 의미(Semantic) 기반 추천, 텍스트, 이미지, 오디오 같은 멀티모달 데이터 활용

    - Ex) "사용자 A는 [아이템 1, 2, 3]을 좋아했습니다. 이 사용자가 좋아할 만한 다음 아이템은 [아이템 4]입니다."

    - 점수 대신 추천 이유를 같이 제시하는 서술형 추천이 가능해짐

* 새로운 가능성

    - 대화형 추천 : 사용자가 자연어로 말한 조건을 반영해서 추천 (Ex. "비 오는 일요일에 혼자 보기 좋은, 너무 슬프지 않은 영화 추천해줘")

    - Zero-shot 추천 : 상호작용 데이터 없이 텍스트 설명만으로 추천 -> 신상품이 등록된 첫날 구매 기록이 0건이어도 상품 설명만으로 추천 대상이 될 수 있음 -> Cold-Start 문제의 해결책

* LLM을 추천에 적용하는 대표 방식

    - P5 (Geng et al., RecSys 2022) : 여러 추천 과제를 "텍스트 프롬프트 -> 텍스트 답변" 형식으로 통일해서 하나의 언어 모델로 학습 / 추천 데이터로 처음부터 학습해서 컴퓨팅 자원과 시간이 많이 필요함

    - TALLRec (Bao et al., RecSys 2023) : 범용 LLM을 "좋아할까? (Yes/No)" 형식의 추천 지시문으로 튜닝 / LoRA로 일부 파라미터만 학습해서 비용을 줄이고, 적은 샘플로도 기존 모델보다 높은 성능을 보고

#### 7-2. 인기 편향과 XAI 문제

> 추천 모델이 정교해져도 무엇을 추천하는가(인기 편향)와 왜 추천하는가(설명 가능성)의 문제는 남음

* 인기 편향 (Popularity Bias) : 소수의 인기 아이템이 추천을 독점하고, 대다수의 비인기 아이템은 거의 추천되지 않는 현상

    - 강화되는 3단계 : 데이터 불균형 (상호작용이 인기 아이템에 몰림) -> 모델의 편향 강화 (데이터가 많은 아이템일수록 예측 점수가 높아 더 자주 추천) -> Rich-get-richer (추천 -> 노출 -> 클릭 -> 학습 데이터 반영이 반복되면서 격차가 커짐)

    - Netflix Prize 데이터 : 인기 영화는 전체의 16.5%(2,938편)인데 전체 평점의 약 86%를 차지 -> 학습 데이터부터 소수에 쏠려 있음

    - Spotify 팟캐스트 추천 실험 : 개인화 추천을 적용하니 스트리밍은 28.90% 늘었지만 개인의 소비 다양성은 11.51% 줄어듦

    -> 비인기, 신규 아이템은 노출 기회를 얻기 어렵고, 사용자는 비슷한 콘텐츠만 반복해서 추천받게 됨

* 설명가능성 (XAI) : 이걸 왜 추천했는지 설명할 수 있는가?

    - GNN, Transformer 같은 딥러닝 모델은 강력하지만 블랙박스 문제가 있음

    - "왜 이걸 나에게 추천했는지" 고객을 설득해야 하고, 비즈니스 담당자에게도 추천의 근거를 제시해야 해서 필요함

    - LLM은 추천 이유를 자연어로 만들 수 있지만, 그 설명이 실제 추천 근거와 일치한다는 보장은 없음

#### 7-3. 컴퓨팅 자본, Cold-Start를 고려한 추천시스템 도입

* 딥러닝은 전면 도입하기엔 너무 비쌈

    - GPU 비용 : 대형 모델의 학습, 추론에 고가의 GPU가 필요

    - 실시간 스코어링 서버비 : 요청마다 전체 아이템에 딥러닝 연산을 하면 서버 비용이 급증

    - 지속 비용 : 대규모 트래픽에서는 사용자 수 × 요청 수만큼의 비용이 매일 누적

* 선형 모델은 정확도는 낮지만 빠르고 저렴하고 해석이 쉬움, 딥러닝과 LLM은 정확도가 높지만 무거움 -> 유튜브, 아마존 같은 현업에서는 선형 모델로 후보를 먼저 거르고 딥러닝으로 정렬하는 2단계 구조를 사용

| 단계 | 목표 | 사용 모델 | 이유 |
| --- | --- | --- | --- |
| 1단계 : 후보군 생성 | 수백만 개 아이템 -> 수백 개 후보로 축소 | 선형 모델 (내용 기반 필터링, 협업 필터링, 행렬 분해) | 계산이 단순해서 압도적으로 빠르고, 학습 비용도 훨씬 적음 |
| 2단계 : 순위 매기기 | 수백 개 후보를 정교하게 정렬해서 상위 수십 개를 노출 | 딥러닝 모델 (GNN, Transformer, VAE) | 추려진 후보에만 쓰니까 비선형성, 순서, 관계 같은 문맥 정보를 모두 써서 정확도를 높일 수 있음 |

-> 실제 추천 시스템은 모델 하나만 쓰지 않고, 지금까지 배운 모델들을 단계별로 조합해서 비용과 정확도를 함께 확보
