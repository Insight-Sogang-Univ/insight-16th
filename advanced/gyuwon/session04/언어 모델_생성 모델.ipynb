### 1. NLP 와 언어 모델

#### (1) NLP(Natural Language Processing)

##### **자연어**

> 인간이 소통을 위해 자연스럽게 발생시켜 사용하는 언어(영어, 한국어) = 우리가 일상에서 사용하는 언어

* 프로그래밍 언어 (C, Python): 엄격한 규칙, 논리적, 일대일 대응. 에러가 나면 돌아가지 않음

* 자연어 (Natural Language): 인간이 소통을 위해 자연발생적으로 사용하는 언어. 모호함, 유연함, 일대다 대응

##### **자연어의 특징**

1. ***이산적 데이터(Discreteness)***

* 이미지나 소리는 연속적인 신호지만, 언어는 '단어'라는 독립적이고 딱 끊어지는 단위로 구성됨

* 컴퓨터가 이해할 수 있도록 이산적인 단어를 연속적인 벡터로 변환하는 과정(임베딩)이 필요

2. ***순차적 데이터(Sequentiality)***

* 단어가 나열되는 순서에 따라 문장의 의미가 완전히 결정됨

    - "사자가 사람을 먹다" vs "사람이 사자를 먹다"

* 데이터의 앞뒤 관계와 위치 정보를 유지하며 처리할 수 있는 모델 구조(RNN, Transformer 등)가 필요함

3. ***중의성과 모호성(Ambiguity)***

* 동일한 형태의 단어라도 문맥에 따라 뜻이 천차만별

    - 배(Fruit/ship/stomach)

* 단어 하나가 아닌 문맥을 전체적으로 파악하는 기술이 핵심

4. ***희소성(Sparsity)***

* 단어의 조합은 무한에 가깝지만 우리가 가진 데이터에 나타나는 조합은 극히 일부(Zipf 의 법칙)

* 한 번도 본 적 없는 문장이 들어와도 단어 간 유사도를 이용해 확률적으로 추론할 수 있는 일반화 능력 중요

5. ***가변성과 진화(Variability & Evolution)***

* 언어는 고정되어 있지 않으며 신조어, 은어, 오타, 이모지 등이 수시로 등장하며 문법 파괴가 잦음

* 고정된 규칙이 아니라 데이터로부터 유연하게 패턴을 학습하는 머신러닝 방식이 필수적

##### **NLP 개념**

> 모호한 자연어를 컴퓨터가 이해할 수 있는 '숫자(데이터)'로 변환하여 유의미한 가치를 창출하는 인공지능 기술 분야

* 컴퓨터는 자연어의 의미를 바로 이해 못하기 때문에 숫자(벡터)로 바꾼 다음 단어 간 관계와 문맥을 분석함

* 기계 번역(파파고, 구글 번역), 감성 분석(리뷰 긍정/부정), 텍스트 분류(스팸 메일), 문서 요약, 질의응답 및 챗봇 등에 활용

##### **NLP 역할**

* 형태소 분석 : 문장을 의미를 가진 가장 작은 단위(형태소)로 나누고 품사, 문법적 특징 분석

    - 나는 학교에 간다 -> 나/대명사 + 는/조사 + 학교/명사 + 에/조사 + 가-/동사 어간 + -ㄴ다/어미

* 구문 분석 : 문장의 문법 구조를 분석해서 단어끼리 어떤 관계인지(주어, 목적어, 서술어) 파악

* 의미 분석 : 단어와 문장의 의미를 문맥에 맞게 파악

    - 은행에서 돈을 찾았다 -> 여기서 은행은 금융기관

* 담화 분석 : 여러 문장 사이의 연결 관계, 대화 전체의 맥락 파악

    - 철수는 시험에 합격했다. 그는 매우 기뻐했다. -> '그' = 철수

#### (2) 언어 모델

> 인간이 말하고 쓰는 방식을 모방하는 시스템

* 단어 시퀀스(문장)에 확률을 할당하여 가장 자연스러운 단어 시퀀스를 찾아내는 모델

* 언어 모델이 문장의 자연스러움을 판단하게 하는 가장 보편적인 방법은 '이전 단어들'이 주어졌을 때 '다음 단어'를 예측하도록 훈련시키는 것

    - 나는 아침에 밥을 ___ -> 먹었다(80%), 마셨다(15%), 달렸다(5%) 중 확률이 제일 높은 '먹었다' 예측

    - 문장 전체 확률도 계산 가능 -> '밥을 먹었다'가 '밥을 달렸다'보다 확률이 높음 = 더 자연스러운 문장

* 학습 방식에 따라 통계적 언어 모델(SLM)/신경망 언어 모델(NLM)로 구분됨

### 2. 통계적 언어 모델(SLM)

#### (1) Statistical Language Model

* 분포 가설(비슷한 문맥에서 함께 나타나는 단어들은 비슷한 의미를 가진다는 가설)에 근거함

    - '나는 아침에 커피를 마셨다', '나는 아침에 차를 마셨다' -> 커피와 차는 비슷한 문맥에 나오니까 의미도 관련 있을 가능성 높음

    - 이후 임베딩 기술의 이론적 기반이 됨

* ***조건부 확률***

    - 지금까지의 단어 다음에 올 단어로 가장 확률이 높은 단어 선택

    - P(갔다 | 나는 학교에) : '나는 학교에'가 주어졌을 때 '갔다'가 나올 확률

* ***카운트 기반***

    - 이전 단어 시퀀스의 등장 빈도를 통해 다음 단어의 확률 계산

    - '나는 학교에'가 100번 나왔는데 그 다음이 갔다 70번, 왔다 20번, 뛰었다 10번이면 '갔다' 확률 70%

    - 카운트 기반의 통계적 언어 모델은 직관적이지만 희소 문제(Sparsity Problem) 을 가짐

    - 이전에 경험한 적 있는 조합에 대해서만 잘 예측함 -> 문법적으로 가능한 문장이어도 학습 데이터에 없으면 확률 0 (ex. 나는 학교에 순간이동했다)

#### (2) N-gram

* 희소하게 등장한 것에 대해서 잘 예측을 못하고 문장이 길어질 수록 다음 단어를 계산하기 어려운 SLM 의 한계 극복 위해 탄생

* 전체 문장 대신에 문장의 일부만 봄 -> 연속된 N 개 단어 묶음 기준으로, 직전 N-1 개 단어만 보고 다음 단어 예측

    - N=4 : 'An adorable little boy is spreading ___' 에서 'boy is spreading' 만 보고 예측하고 'An adorable little' 은 버림

    - 학습 데이터에서 'boy is spreading' 다음에 insults 가 smiles 보다 많이 나왔으면 insults 예측

    - N 이 클수록 더 긴 문맥 반영

* N-gram 도 SLM 의 한 종류 -> SLM 은 이전 단어 전부를 조건으로 두는데 N-gram 은 직전 N-1 개로 범위를 줄여서 계산과 확률 추정을 단순하게 만든 것

* 희소 문제

    - 데이터 부족 : 훈련 데이터에서 나타나지 않은 단어 조합은 확률이 0이 되어 해당 단어 시퀀스를 예측하거나 평가할 수 없음

    - N 의 크기가 커질수록 훈련 데이터에서 해당 조합을 찾기 어려워져 희소 문제가 심해짐

* 문맥 제한

    - 앞선 N-1 개의 단어만 고려하므로, 문장 앞부분에 나온 정보를 뒷부분 결정에 반영하지 못함

#### (3) Perplexity(PPL)

> 언어 모델이 문장을 얼마나 혼란스러워하는지를 수치로 나타낸 것

* PPL 이 낮을수록 예측 불확실성이 낮음 = 실제로 나온 단어를 잘 예측함, 높을수록 예측을 어려워함

* 문장 W가 희소하게 등장하면 모델이 훈련 데이터에서 더 적게 봤으므로 예측할 때 더 혼란스럽기에 PPL 이 커짐

* 문장에 예측하기 쉬운 단어가 추가되면 평균 예측 난이도가 낮아져서 PPL 이 낮아질 수 있음 -> N 이 커진다고 항상 낮아지는 건 아니고 추가된 단어가 얼마나 예측하기 쉬운지에 따라 다름

* SLM 뿐만 아니라 NLM 의 성능 평가에도 사용

* 신경망 기반 언어 모델은 단어와 문맥 간의 관계를 학습하여 일반화 하는 능력이 뛰어남 -> 훈련 데이터에서 직접 보지 못한 문장이 등장하더라도 훨씬 유연하게 대처 가능 -> 같은 평가 데이터에서 SLM 보다 낮은 PPL 을 낼 수 있음 (ex. SLM 80, NLM 25)

### 3 신경망 언어 모델(NLM)

* 통계적 언어 모델의 한계를 극복하기 위해 등장 -> SLM 은 단어를 단순히 이름으로만 기억했지만 NLM 은 모든 단어를 거대한 다차원 지도 위의 한 점(좌표)로 보냄(임베딩)

    - 의미나 쓰이는 문맥이 비슷한 단어끼리 가까이 위치함 (moon, mars, space, nasa 끼리 / bible, god, church 끼리)

* 워드 임베딩을 통해 NLM 은 데이터에 없는 문맥도 어느 정도 예측할 수 있는 일반화 능력을 가짐

* 이후 RNN, LSTM, Transformer 로 발전하면서 긴 문맥도 처리할 수 있게 됨

| **특징** | **통계적 모델 (SLM)** | **신경망 모델 (NLM)** |
| --- | --- | --- |
| **학습 방식** | 단어 등장 빈도로 확률 계산 | 신경망 가중치 학습 |
| **단어 표현** | 단어를 각각 다른 기호로 취급 | 연속적인 벡터 |
| **미학습 단어 조합** | 확률 0 (예측 불가) | 유사 단어를 통해 확률 추론 가능 |
| **의미 파악** | 불가능 (단순 빈도수) | 가능 (임베딩 벡터 간 거리 계산) |
| **문맥** | N-gram 은 직전 N-1 개 단어만 | 신경망 구조에 따라 더 넓은 문맥 |
| **계산 비용** | 상대적으로 낮음 | 학습에 연산 자원이 많이 필요 |

### 4. 딥러닝 기반 언어모델

#### (1) Transformer 의 등장

* RNN, LSTM 의 한계

    - 장기 의존성 문제 : 문장이 길어질수록 앞부분 정보를 유지하기 어려움

    - 기울기 소실 : 역전파 과정에서 기울기가 점점 작아져서 앞부분이 충분히 학습되지 않음

    - 순차 처리 : 단어를 순서대로 처리해야 해서 병렬 연산이 어렵고 학습이 느림

* 2017년 Google 의 Attention Is All You Need 논문에서 Transformer 제안 -> 이후 BERT, GPT 등 현대 언어 모델의 기반

    - Self-Attention 으로 단어 간 관계와 장거리 문맥 학습

    - 병렬 처리 + 위치 인코딩으로 학습 효율을 높이고 순서 정보 반영

* 구조

    - 임베딩 + 위치 인코딩 : 토큰을 벡터로 바꾸고, 순환 구조가 없으니 위치 정보를 더해서 순서 구분

    - 인코더 : Self-Attention(문장 내 단어끼리 얼마나 관련 있는지 계산) -> Feed Forward Network, 이걸 여러 층 쌓음

    - 디코더 : Masked Self-Attention(미래 단어는 안 보고 이전 단어만 참고) -> Cross-Attention(인코더 정보 참고) -> Feed Forward Network

    - 번역 예시 : 'I am a student' 를 인코더가 문맥 분석 -> 디코더가 'Je' -> 'suis' -> 'étudiant' 순서로 하나씩 생성

#### (2) BERT(Bidirectional Encoder Representations from Transformers)

Bidirectional : 텍스트의 앞뒤 문맥을 동시에 고려해서


Encoder Representations : 인코더로 문맥이 반영된 단어 벡터를 만들어주는


from Transformers : 트랜스포머 인코더 구조 기반 모델

=> 한 단어를 이해할 때 그 단어의 좌우 문맥을 동시에 보고 이해함

* 양방향 : 문장을 읽을 때 앞뒤를 동시에 고려 -> 한 단어를 이해할 때 그 단어의 좌우 문맥을 동시에 보고 이해

* 트랜스포머 기반 : 트랜스포머의 encoder 부분만 사용 -> Self-Attention 으로 단어끼리 얼마나 관련 있는지 학습

* 사전 학습 + 파인 튜닝 : 큰 데이터로 먼저 일반적인 언어 능력을 배우고, 나중에 특정 과제에 맞게 살짝 조정

    - 사전학습은 Masked Language Model(MLM) + Next Sentence Prediction(NSP) 를 통해 진행됨 -> 정답 레이블 없이 텍스트만으로 학습 가능

    - MLM : 빈 칸 채우기 방식 -> 문장 일부 토큰을 [MASK] 로 가리고 앞뒤 문맥으로 원래 토큰을 맞힘 (How are [MASK] doing today -> you) -> 자연스럽게 문맥을 이해하고 추론하는 능력 향상, 같은 단어도 문맥에 따라 다른 의미로 학습

    - NSP : 두 문장에 대해 이 문장들이 서로 앞뒤 관계인지 그냥 무작위 위치에 있는 두 개의 문장인지 학습 -> 진짜 다음 문장이면 IsNext, 아니면 NotNext

* input → Embedding → 트랜스포머 인코더 12개(or 24개) → Output

    - 입력 문장을 토큰으로 나누고 맨 앞에 [CLS], 문장 끝에 [SEP] 추가

    - 임베딩할 때 토큰 의미 + 위치 + 문장 구분 정보를 같이 넣음

    - 입력 문장을 임베딩 한 뒤 여러 개의 encoder 를 통과하면서 각 단어의 의미를 문맥적으로 풍부하게 만듦

    - Transformer 는 decoder 까지 있기에 문장 생성을 목적으로 하지만 BERT 는 encoder 만 있기에 문장 이해를 목적으로 함 (문장 분류, 감성 분석, 질의응답 등)

* 이후 모델로 RoBERTa, ALBERT (A Lite BERT) 등이 있음

    - RoBERTa: BERT의 사전 훈련 프로세스를 개선하여 성능을 향상시킨 실전형 모델 -> NSP 제거, 학습할 때마다 마스킹 위치를 바꾸는 Dynamic Masking, 더 많은 데이터 + 큰 배치 + 더 긴 학습

    - ALBERT (A Lite BERT): BERT의 경량화 버전으로, 가볍지만 성능은 유지 -> 임베딩 차원을 줄이고 층끼리 파라미터를 공유해서 파라미터 수 감소, NSP 대신 SOP(문장 순서 예측) 사용

#### (3) GPT

* encoder 기반인 BERT 와 달리 decoder 기반으로 지금까지의 단어를 바탕으로 다음 단어를 예측 -> 자기회귀 방식으로 텍스트 생성

    - 인코더가 없으니 Cross-Attention 은 안 씀

    - Masked Self-Attention 으로 미래 토큰은 안 보고 이전 토큰만 보고 예측

* few shot : 몇개의 예시만 보고 task 수행(GPT 3 부터 도입)

    - GPT-3 는 파인튜닝 없이 프롬프트에 있는 지시와 예시만으로 여러 task 수행 = In-Context Learning

    - zero shot (예시 없이 지시만) / one shot (예시 1개) / few shot (예시 여러 개)

| **구분** | **BERT** | **GPT** |
| --- | --- | --- |
| **구조** | Transformer Encoder | Transformer Decoder |
| **방향** | 양방향 | 단방향(자기회귀) |
| **학습 방법** | MLM, NSP | 다음 토큰 예측 |
| **목적** | 언어 이해 | 텍스트 생성 |
| **task** | 감성 분석, 텍스트 분류, 개체명 인식 | 챗봇, 텍스트 생성, 번역, 요약 |

---

#### (4) LLM(Large Language Model)

> 방대한 양의 데이터를 학습하여 자연어 및 기타 유형의 콘텐츠를 이해하고 생성하여 광범위한 작업을 수행할 수 있는 기초 모델의 범주

* ex. GPT 계열(ChatGPT), LLaMA, Claude, Gemini

* 학습과 예측이라는 두가지 핵심 단계를 통해 작동

    - 대규모 학습(training) : 인터넷의 웹페이지, 책, 기사 등 수십억 개의 문장으로 이루어진 방대한 데이터 학습 -> 단순히 정보 암기가 아닌 단어와 단어 사이 관계, 문맥적 의미, 문법 구조, 세상의 지식 등의 언어 패턴을 스스로 터득

    - 사전학습(다음 토큰 예측을 반복하면서 언어 구조, 문맥 학습) + 파인튜닝(특정 목적 데이터로 추가 학습) -> 예측값과 정답의 차이를 줄이도록 가중치 업데이트

    - 예측 및 생성 : 학습이 완료되고 사용자의 질문이나 프롬프트를 받으면 학습한 패턴을 기반으로 다음에 올 가장 확률 높은 단어를 순서대로 예측하며 응답 생성 (입력 토큰화, 벡터 변환 -> Attention 으로 문맥 분석 -> 다음 토큰 확률 계산 -> 생성)

* LLM 은 BERT 가 아닌 GPT 사용 -> decoder 기반

    - LLM 은 단순히 이해를 위한 게 아니라 텍스트 생성을 위한 것이기에 GPT 사용

* Hallucination : 환각처럼 그럴 듯하지만 틀린 정보를 말하는 것

    - 다음 토큰 확률로 답을 만들기 때문에 사실인지는 보장 X

    - RAG, 외부 검색, 검증 과정으로 줄일 수 있지만 완전히 해결된 문제는 아님

* 업데이트 비용 : LLM 은 파라미터 수가 엄청나게 많기에 새로운 정보를 반영할 때 계산량이 많음

    - 사전학습 자체에 연산 자원, 시간, 비용이 많이 들고, 학습 이후에 바뀐 최신 정보는 자동으로 반영 안 됨 -> 외부 지식을 활용하는 기술 필요(RAG)

---

#### (5) RAG(Retrieval Augmented Generation)

> 외부에서 최신 정보를 가져와서 답변하게 하기 위한 것 -> 질문에 대해 모델이 모르면 검색해서 답하는 구조

* RAG 구조

    * 질의 인코더(Query encoder) : 사용자의 질문을 임베딩 모델로 벡터로 바꿈

    * 지식 검색기(Knowledge Retriever) : 인코딩된 질문을 바탕으로 벡터 DB 에서 유사도가 높은 문서를 검색

    * 지식 증강 생성기(Knowledge Augmented Generator) : 질문 + 검색한 문서를 같이 LLM 에 넣어서 답변 생성

* RAG 장점

    - 최신 정보 반영 가능 -> 모델을 다시 학습하지 않아도 외부 DB 나 최신 문서를 검색해서 활용

    - 모르는 것에 대해 지식 보완이 가능해 hallucination 문제 감소

* CAG(Credibility-Aware Generation)

    - RAG 한계 보완위해 제안된 프레임 워크

    - 검색된 문서 중 일부가 신뢰성이 낮거나 잘못된 정보일 경우 생성된 답변의 품질과 정확도가 크게 하락하기에 모델이 문서를 단순히 활용하는 수준을 넘어 문서의 신뢰도를 스스로 판단하고 활용할 수 있도록 훈련시키는 기법

    - 질문 입력 → 정보 검색 → 각 정보의 신뢰도 평가 → 높은 신뢰도 정보 우선 사용 → 최종 답변 생성

    - 신뢰도는 관련성, 최신성, 출처의 신뢰성을 같이 보고 평가

    - ex. 'David Cameron 은 어떤 직책?' -> RAG 는 신뢰도 낮은 문서를 참고해서 '공직 없음'이라고 틀림, CAG 는 최신 Reuters 기사를 우선해서 '외무장관'으로 답함

---

#### (6) LangChain

> LLM + RAG 직접 구현은 어렵기에 pipeline 쉽게 구성할 수 있도록 도와주는 역할

* 직접 구현할 때 어려운 점 : 구현이 복잡함, 외부 시스템 연동, 유지보수

* LLM 기반으로 앱 제작 시 필요한 구성 요소(외부 데이터, API, 도구) 쉽게 연결

* 파이썬과 JavaScript 라이브러리 제공 -> 다양하게 활용 가능

* LangChain 특징

    - 추상화 : LLM 서비스를 만들 때 필요한 각종 작업을 간결하게 표현(간소화) -> 복잡한 API 호출 코드 없이 간단한 명령어로 LLM 호출

    - 표준화: 비슷한 기능을 갖추고 있는 요소들을 똑같은 형식을 갖춘 컴포넌트로 표준화 -> ChatGPT 대신 Claude 를 쓰고 싶으면 모델 연결 부분만 바꾸면 됨

    - 체이닝: 자주 활용하는 주요 컴포넌트를 쉽게 연결해서 LLM 서비스의 로직을 쉽게 파악 (Input → 컴포넌트 → Output), ex. 영어 문장 → 한국어 번역 → 요약 → 출력

    - LangChain 은 중간에 조건에 따라 다른 작업을 수행하거나 결과가 좋지 않을 때 이전 단계로 돌아가 다시 실행하는 등의 복잡한 로직을 구현하기 어려움

* LangGraph : 비선형적인 작업 형태로 사이클, 루프를 만들 수 있음 -> 사이클을 통해 검색과 검증을 반복하며 완벽한 답을 찾아냄

    - 노드 (Node): 특정 작업을 수행하는 단위로, 랭체인의 컴포넌트를 그대로 가져와 노드로 사용 할 수 있음 (LLM, Agent, Tool 등)

    - 엣지 (Edge): 노드와 노드를 연결하는 선으로, 어떤 노드에서 다음 노드로 이동할지를 결정함

        - 조건부 엣지(conditional edge)를 통해 분기 로직(if-else문)을 쉽게 구현할 수 있음

    - 상태 (State): 노드끼리 공유하는 정보 -> 이전 작업 결과, 대화 문맥 등을 저장

    - ex. 사용자 질문 -> Router 가 알맞은 Agent 로 보냄 -> Agent 가 필요하면 Tool(검색, API) 호출 -> 결과에 따라 다시 호출하거나 최종 답변

* LangSmith : 대규모 언어 모델(LLM) 기반 애플리케이션을 위한 통합 개발 및 운영 플랫폼

    - 디버깅 및 추적: LLM 애플리케이션의 모든 실행 과정을 단계별로 시각화

    - 테스트 및 평가: 미리 준비된 질문과 정답 데이터셋을 만들어, 개발 중인 LLM 애플리케이션의 성능을 자동으로 평가

    - 모니터링: 실제 사용자들이 애플리케이션을 어떻게 사용하는지 모니터링 (API 호출 횟수, 지연시간, 토큰 사용량 등을 추적해 최적화)

    - 프롬프트 허브: 다양한 프롬프트를 저장하고, 버전을 관리하며, 쉽게 테스트할 수 있는 공간 제공

* 외부 API 활용 : LLM 내부 지식만을 이용해서 해결할 수 없는 질문에 대해 외부 API 와 연동해서 답변할 수 있음

    - OpenAI API 키 발급 : 회원가입 -> 결제 정보 등록 -> 키 생성 -> 복사해서 보관

    - 키는 처음 만들 때 한 번만 전체가 보이고, 유출되면 다른 사람이 내 키로 GPT 를 쓰고 결제는 내 쪽으로 됨 -> 공유 X

### 5. sLM(small Language Model)

#### (1) sLM 정의

> 자연어 콘텐츠를 처리, 이해 및 생성할 수 있는 AI 모델 -> 대규모 언어 모델(LLM) 보다 규모와 범위가 작음

* 수백만~수십억 개 수준의 매개변수를 가지며 LLM 은 수십억~수천억 개 이상의 매개변수 가짐

    - 매개변수 : 모델이 학습을 통해 얻는 내부 값 -> 모델의 동작과 수행 방식에 영향을 줌

* sLM 은 LLM 보다 더 작고 효율적이기에 메모리와 연산 자원을 적게 사용 -> 모바일 앱처럼 리소스가 제한된 환경, 또는 데이터 네트워크 없이 사용자 쿼리에 대한 오프라인 추론을 통해 응답을 생성해야 하는 상황에 적합

* LLM 은 복잡한 추론이나 다양한 작업에 강하고 주로 고성능 서버에서 돌림, sLM 은 특정 분야나 비교적 단순한 작업에 효율적이고 연산 비용이 낮아서 응답이 빠름 (서버, 스마트폰, 노트북)

---

#### (2) 모델 압축

* 모델의 중요한 성능은 최대한 유지하면서 크기와 연산량을 줄이는 기술

* 가지치기 : 신경망에서 중요도가 더 낮거나 중복되거나 불필요한 매개변수를 제거함

    - 가지치기 되는 매개변수에는 뉴런 간의 연결에 해당하는 숫자 가중치(이 경우 가중치는 0으로 설정됨), 뉴런 자체 또는 신경망의 계층이 포함됨

    - 연결만 제거하면 비구조적 가지치기, 뉴런이나 블록 단위로 제거하면 구조적 가지치기

    - 제거한 뒤 필요하면 추가 학습으로 성능 보완, 중요한 부분까지 지우면 성능이 떨어짐

* 양자화 : 고정밀 데이터를 저정밀 데이터로 변환 -> 계산 부하를 줄이고 추론 속도 높일 수 있음

    - FP32(32비트)로 저장하던 가중치를 FP16, INT8(8비트)처럼 비트 수를 줄여서 저장 -> 메모리 ↓, 대신 정밀도가 낮아져 성능이 조금 떨어질 수 있음

* 저차원 분해 : 큰 가중치 행렬을 더 작은 저차원 행렬로 분해하는 것 -> 매개변수가 줄어들고 계산 횟수가 줄어들며 복잡한 행렬 연산을 단순화 할 수 있음

    - 큰 행렬 W 를 작은 행렬 A, B 의 곱으로 근사

    - LoRA 도 이 원리를 사용 -> 모델 전체를 압축한다기보다 가중치 변경량(ΔW = A × B)만 작은 행렬로 학습해서 학습 비용을 줄이는 미세조정 방법

* 지식 증류 : 사전 학습된 **교사 모델**의 학습 내용을 **학생 모델**로 이전하는 것 -> 학생 모델이 교사 모델의 예측 결과를 따라 하도록 학습해서 작은 모델로도 좋은 성능

---

#### (3) sLM 의 예

* DistilBERT : BERT 에 지식 증류를 적용해서 작고 빠르게 만든 모델, 문장 분류나 텍스트 분석에 사용

* Gemma : Google 의 경량 언어 모델

* GPT-4o mini : OpenAI 의 소형 모델, 빠른 응답 속도와 비용 효율

* Llama : Meta, 작은 버전은 모바일 같은 제한된 환경에서도 사용 가능

* Ministral : Mistral AI, 스마트폰이나 기기 내부에서 돌리는 온디바이스 환경용

* Phi : Microsoft, 작은 크기에 비해 높은 성능이 목표

---

#### (4) sLM 장점

- 성능적 측면

    - **빠른 응답 속도**: 모델 크기가 작기 때문에 계산량 ↓ → LLM 보다 결과가 빨리 나옴

    - **제한된 환경에서도 실행 가능**: 메모리, 연산 자원을 적게 써서 스마트폰이나 노트북에서도 돌릴 수 있음

    - **보안:** 작은 모델이라서 기업 내부 서버나 개인 기기에서 직접 실행하기가 상대적으로 쉬움 → 민감한 데이터를 외부 클라우드 API로 보내지 않아도 됨

- 비용적 측면

    - **낮은 운영 비용**: 연산량과 메모리가 적어서 서버 운영, AI 서비스 제공 비용 ↓

    - **에너지 효율**: 계산량이 적으므로 전력 소비 ↓

    - **하드웨어 비용 절감**: 고성능 GPU 서버가 없어도 사용 가능 → 진입 장벽 ↓

---

#### (5) sLM 한계

- **편향**: 학습 데이터에 들어 있는 사회적 편견이나 잘못된 정보가 답변에 반영될 수 있음 → 데이터가 다양하지 않으면 특정 집단이나 관점에 치우친 결과

- **제한된 일반화 & 복잡한 작업에서의 성능 저하**: 학습하지 않은 새로운 상황, 복잡한 추론, 전문 지식이 필요한 작업에서는 LLM 보다 정확도가 떨어짐

- **할루시네이션(환각)**: LLM 과 마찬가지로 근거가 부족한 질문에 사실이 아닌 정보를 사실처럼 만들어낼 수 있음

---

#### (6) LLM 과 sLM 의 결합(결합 추론)

> 복잡하고 어려운 작업에는 LLM, 가볍고 간단한 작업에는 SLM을 적절히 배분해서 사용 -> 모든 작업에 LLM 을 쓰지 않아도 돼서 비용과 처리 시간을 줄이면서 필요한 성능은 확보

* 간단한 작업(문장 분류, 간단한 요약) -> sLM

* 복잡한 작업(여러 단계 추론, 전문적인 분석) -> LLM

* ex. AI 챗봇에서 간단한 질문은 sLM 이 빠르게 답하고, 복잡한 분석 요청은 LLM 이 답함

### 6. 생성 모델의 개념

#### (1) 분류 모델 vs 생성 모델

> 학습 데이터의 분포를 학습해서 그 분포를 따르는 비슷한 데이터를 새로 만들어내는 모델

| **구분** | **분류 모델** | **생성 모델** |
| --- | --- | --- |
| **학습 목표** | P(Y\|X) : 데이터 X 가 주어졌을 때 정답 Y 일 확률 | P(X) : 데이터 X 자체가 나타나는 확률분포 |
| **학습 내용** | 데이터 간 차이점 -> 분류 기준 | 데이터의 공통적인 특징, 다양한 패턴 |
| **출력** | 범주 or 범주별 확률 | 새로운 이미지, 텍스트, 음성 등 |
| **한계** | 처음 보는 이상한 데이터도 아는 라벨 중 하나로 분류해버림, 새 데이터 생성 불가 | 부자연스러운 결과가 나올 수 있고 품질 평가가 어려움 |

* 생성 모델은 입력이 학습한 분포에 맞는지 평가할 수 있어서 비정상적인 입력을 탐지하는 데도 쓸 수 있음

---

#### (2) 생성 모델의 분류

* 데이터의 확률밀도를 직접 계산할 수 있는지에 따라 나뉨

    - 확률밀도 : 학습한 분포에서 특정 데이터가 얼마나 그럴듯한지 나타내는 값 -> 강아지 사진으로 학습했으면 강아지 사진은 높고 자동차 사진은 낮음

1. ***명시적 확률밀도 모델(Explicit Density Model)***

* 데이터 분포를 수학적으로 정의해서 P(X) 값을 직접 계산하거나 근사할 수 있음

* 다루기 쉬운 밀도 모델(Tractable Density Model) : 확률밀도를 수학적으로 정확히 계산 가능

    - NADE, MADE, PixelRNN/PixelCNN (각 요소의 조건부 확률을 순서대로 계산), Change of Variables Model (변수 변환 이용)

* 근사적 밀도 모델(Approximate Density Model) : 직접 계산이 어려워서 근사하는 방법 사용

    - VAE : 변분 추론 사용 -> 복잡한 분포를 정규분포처럼 계산하기 쉬운 분포로 근사

    - Boltzmann Machine : 마르코프 연쇄 사용 -> 다음 상태가 현재 상태에 의해서만 확률적으로 결정되는 과정을 반복하면서 샘플 추출 (ex. 오늘 맑으면 내일 맑을 확률 80%, 비 올 확률 20%)

2. ***암시적 확률밀도 모델(Implicit Density Model)***

* P(X) 값을 수치로 알려주지는 않고 학습한 분포에서 새로운 데이터를 뽑는(샘플링) 것에 집중

* 직접 샘플링 : 무작위 노이즈를 생성 모델에 한 번 통과시켜서 바로 생성 -> GAN

* 간접/반복적 샘플링 : 확률적인 상태 변화를 여러 번 반복해서 생성 -> GSN (마르코프 연쇄의 상태 전이 규칙을 학습해서 반복 적용)

### 7. AE, VAE

#### (1) AE(Auto Encoder)

> 입력 데이터의 중요한 특징을 압축했다가, 원래 데이터와 최대한 비슷하게 복원하도록 학습하는 신경망

* 정답 레이블 없이 입력 데이터 자체를 복원 목표로 씀

* 기존 인코더-디코더는 입력을 이해해서 원하는 결과를 만듦(한국어 문장 -> 영어 번역), AE 는 입력 자체를 복원하는 게 목적(숫자 4 이미지 -> 숫자 4 이미지)

* 구조 : 인코더 → 병목(Bottleneck) → 디코더

    - 인코더 : 입력 x 에서 중요한 특징을 뽑아서 잠재 벡터 z 로 압축

    - 병목 : 작은 공간에 정보를 담아야 해서 핵심 특징만 남게 됨

    - 디코더 : z 로 복원 데이터 x' 를 만들고 원본이랑 비교

* 왜 압축했다가 복원하나

    - 압축을 안 하면 입력을 그냥 복사해버릴 수 있음 -> 잠재공간 크기를 제한해서 어떤 정보가 복원에 중요한지 배우게 함

    - 복원 오차(Reconstruction Error)를 줄이면서 압축된 정보만으로 원본을 만들 수 있는지 확인

    - 결국 데이터를 통째로 외우는 대신 핵심 특징과 구조를 학습하게 됨

* 활용 : 데이터 압축, 노이즈 제거, 이상 탐지(정상 데이터 특징을 학습해두고 벗어나는 데이터 탐지)

---

#### (2) VAE(Variational Auto Encoder)

> 입력 데이터의 특징을 확률분포로 학습하고, 그 분포에서 샘플링해서 새로운 데이터를 생성할 수 있는 모델

* 등장 배경 : AE 는 잠재공간이 불규칙하게 만들어지고 데이터 사이에 빈 영역이 생김 -> 거기서 값을 뽑아 디코더에 넣으면 부자연스러운 결과가 나옴 -> 복원은 잘하는데 새 데이터 생성은 어려움

* AE 는 입력을 잠재공간의 점 하나(고정된 벡터)로 표현하고, VAE 는 평균(μ)과 분산(σ²)을 가진 분포(곡선)로 표현

    - 잠재공간이 연속적이고 규칙적으로 만들어짐 -> 값을 조금씩 바꾸면 특징도 자연스럽게 바뀌고, 데이터 사이 공간에서 뽑아도 자연스러운 데이터가 나옴

* 학습 목표

    - 복원 오차 최소화 : 원본과 복원 결과의 차이를 줄임

    - KL 발산 최소화 : 인코더가 예측한 잠재분포를 표준정규분포에 가깝게 만듦 (KL 발산 = 두 확률분포가 얼마나 다른지, 같으면 0)

* 작동 과정

    1. 분포 예측(Probabilistic Encoder) : 입력 x 로 평균 μ 와 표준편차 σ 를 예측 -> $q_\phi(z|x)$

    2. 샘플링 + 재파라미터화 : $z = \mu + \sigma \odot \epsilon$, $\epsilon \sim N(0, I)$

        - 그냥 무작위로 뽑으면 기울기가 전달이 안 돼서 학습이 안 됨 -> 무작위성을 노이즈 ε 로 따로 빼서 학습할 수 있게 함

        - ex. μ = 2, σ = 0.5 일 때 ε = 1 이면 z = 2.5, ε = -1 이면 z = 1.5

    3. 생성(Probabilistic Decoder) : z 를 디코더에 넣어서 x' 생성 -> 학습이 끝나면 표준정규분포에서 z 만 뽑아서 학습 데이터에 없던 새로운 데이터도 생성 가능

* 활용 : 이미지 생성, 데이터 증강, 이상 탐지, 특징 추출

### 8. GAN

#### (1) GAN(Generative Adversarial Network) 개념

> 생성자(Generator)와 판별자(Discriminator) 두 신경망이 서로 경쟁하면서 실제 데이터와 비슷한 새로운 데이터를 생성하도록 학습하는 모델

* 2014년 Ian Goodfellow 등이 제안

* 위조지폐범(생성자)과 경찰(판별자)

    - 생성자 : 진짜 같은 위조지폐를 만들어서 경찰을 속이려고 함 -> 처음엔 어설프다가 점점 정교해짐

    - 판별자 : 지폐가 진짜인지 가짜인지 구별 -> 위조 기술이 늘수록 감별 능력도 같이 높아짐

    - 최적점에서는 생성자가 진짜와 구분 안 되는 데이터를 만들고, 판별자는 찍어서 맞히는 수준(확률 0.5)이 됨

* 목적 : 경쟁을 통해 생성자가 실제 데이터의 분포를 학습하게 하는 것 -> 학습이 끝나면 보통 생성자만 써서 데이터 생성

* 활용 : 이미지 생성(가상 인물 얼굴), 이미지 복원, 데이터 증강

---

#### (2) GAN 구조

* 무작위 노이즈 z → 생성자 G → 가짜 이미지 G(z)

* 판별자 D 는 실제 이미지 x 또는 G(z) 를 받아서 진짜일 확률(0~1) 출력

    - D(x) = 0.9 : 실제 이미지를 90% 확률로 진짜라고 판단

    - D(G(z)) = 0.1 : 가짜 이미지를 10% 확률로 진짜라고 판단

* 최소-최대 게임(Minimax Game)

    - 판별자 : 진짜는 1, 가짜는 0 에 가깝게 출력하도록 -> 구별 능력 최대화

    - 생성자 : 판별자가 진짜와 가짜를 구별하지 못하도록 -> 판별자를 속이는 방향으로 학습(최소화)

    - 이 과정을 반복하면서 생성자가 점점 진짜 같은 이미지를 만듦

---

#### (3) GAN 장단점

* 장점

    - 진짜 같은 가짜 생성 : 실제로 없는 사람의 얼굴 이미지 같은 것

    - 선명한 이미지 : VAE 는 픽셀 간 평균(MSE)을 최적화해서 흐릿해지는 경향이 있는데, GAN 은 판별자를 속여야 하니까 미세한 질감까지 살아남

    - 빠른 생성 속도 : 학습 후에는 노이즈를 생성자에 한 번만 통과시키면 됨 (확산 모델은 여러 단계 반복)

    - 부드러운 잠재공간 : 잘 학습되면 잠재 벡터를 조금씩 바꿀 때 이미지도 자연스럽게 바뀜

* 한계

    - 학습 불안정 : 서로 반대 목표로 동시에 학습해서 한쪽이 너무 강해지면 균형이 무너짐 -> 성능이 계속 오르내리거나(진동) 수렴을 못함 -> 학습률, 학습 횟수 조절, WGAN-GP

    - 훈련 초기 문제 : 초반에는 생성자가 너무 못 만들어서 판별자가 쉽게 구별 -> 생성자한테 가는 기울기가 너무 작아짐 -> Non-saturating Loss, 판별자 학습 속도 조절

    - 모드 붕괴(Mode Collapse) : 판별자를 잘 속이는 특정 형태만 계속 생성해서 다양성 ↓ (비슷하게 생긴 강아지만 계속 나옴) -> Minibatch Discrimination, WGAN-GP

    - 평가 지표 한계 : 사실성, 선명도, 다양성을 지표 하나로 평가하기 어려움 -> FID(실제/생성 이미지 특징 분포 비교), IS(분류 확신도 + 다양성), 사람의 시각적 평가를 같이 씀

    - 역매핑 어려움 : z → 이미지 방향으로만 학습해서 실제 이미지에 해당하는 z 를 찾기 어려움 -> GAN Inversion, BiGAN(이미지 → z 인코더 추가)

---

#### (4) GAN 적용 사례

* NVIDIA Progressive GAN : 저해상도에서 고해상도로 점점 키우면서 학습 -> 실존하지 않는 사람의 고해상도 얼굴 생성

* Deep Video Portraits (Conditional GAN) : A 의 표정, 머리 움직임, 시선을 B 의 영상에 적용 -> 영화 특수효과, 더빙, VR

* Facebook Eye In-Painting (Exemplar GAN) : 같은 사람의 눈 뜬 참고 사진을 이용해서 눈 감은 사진을 눈 뜬 사진으로 복원

### 9. 확산 모델

#### (1) 확산 모델(Diffusion Model) 개념

> 이미지에 노이즈를 추가한 데이터로 학습하고, 무작위 노이즈에서 노이즈를 조금씩 제거하면서 새로운 이미지를 생성하는 모델

* 순확산 : 원본 이미지에 노이즈를 점점 추가 / 역확산 : 노이즈를 단계적으로 제거하면서 이미지 생성

* 목적 : 실제 데이터의 특징과 분포를 학습해서 새로운 데이터를 자연스럽고 다양하게 생성하는 것

* 모호성 모델링(Ambiguity Modeling) : 하나의 조건에서도 여러 결과가 가능하다는 불확실성을 확률적으로 표현

    - '공원에 앉아 있는 강아지' 하나로도 품종, 털 색, 자세, 배경이 다른 이미지들이 나올 수 있음

* 활용 : 이미지 생성, 이미지 복원, 이미지 편집, 영상 생성

---

#### (2) 순확산과 역확산

* 순확산(Forward Diffusion)

    - 원본 이미지 $x_0$ 에 가우시안 노이즈(정규분포를 따르는 잡음)를 여러 단계에 걸쳐 조금씩 추가

    - $x_0$ → $x_1$ → $x_2$ → ... → $x_T$ 로 갈수록 이미지가 흐려지고, 마지막에는 거의 무작위 노이즈가 됨

    - 이렇게 만든 단계별 데이터로 역확산 모델이 노이즈 제거하는 방법을 학습함

* 역확산(Reverse Diffusion)

    - 무작위 노이즈 $x_T$ 에서 시작

    - Denoising U-Net 이 현재 이미지에 섞인 노이즈를 예측해서 조금씩 제거 -> 여러 번 반복하면서 색과 윤곽이 생김

    - 반복이 끝나면 새로운 이미지 $x_0$ 생성

---

#### (3) 확산 모델 장단점

* 장점

    - 샘플 품질 : 노이즈를 단계적으로 제거해서 GAN 보다 정교하고 다양한 이미지를 만드는 데 유리

    - 학습 안정성 : GAN 처럼 경쟁하는 대신 실제 노이즈와 예측한 노이즈의 차이(MSE)를 줄이는 방식 -> 손실 곡선으로 학습 상태를 확인할 수 있고 비교적 안정적

    - 다양성 : 무작위 노이즈에서 시작하니까 같은 조건에서도 다른 이미지가 나옴

* 단점

    - 느린 생성 속도 : GAN 은 한 번에 생성하는데 확산 모델은 노이즈 제거를 여러 단계 반복해야 함

    - 높은 계산 비용 : 학습할 때도 생성할 때도 신경망을 반복 실행해야 해서 GPU 가 필요하고 시간, 자원이 많이 듦

    - 구조적 제약 : 노이즈 제거 단계 수나 방식에 따라 속도와 품질이 달라져서 설계를 잘 해야 함
