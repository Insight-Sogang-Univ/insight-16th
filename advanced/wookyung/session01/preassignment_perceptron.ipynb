## 1. 인공신경망의 시작: 퍼셉트론
### 1-1. 인공신경망이란?
#### [인공신경망]
- 뇌의 신경망(신경 세포끼리 연결되어 정보 처리하는 구조)을 모방한 모델
- 생물학적 뉴런의 정보 처리

    <img src="https://upload.wikimedia.org/wikipedia/commons/1/10/Blausen_0657_MultipolarNeuron.png" width="450" style="background-color:white; padding:10px;">

    1. 가지돌기로 외부 전기신호 받음
    2. 모인 신호가 일정 크기 이상이면
    3. 축삭돌기로 신호 전달
- 인공신경망 처리 과정
    1. 입력($x_1, \dots, x_n$)에 가중치($w_1, \dots, w_n$)와 편향($b$) 적용
    2. 종합한 값을 활성화 함수에 넘김 -> 임계값 기준으로 미달이면 0에 가깝게, 통과하면 유의미한 값으로 내보냄
    3. output 출력

#### [퍼셉트론]
- 초기 형태의 인공신경망, 딥러닝의 기원
- 선형 이진 분류기(Linear Binary Classifier)

    <img src="https://upload.wikimedia.org/wikipedia/commons/6/6a/Perceptron-unit.svg" width="350" style="background-color:white; padding:10px;">
- 출력 절차 : 입력 -> 가중합 -> 활성화 함수 -> 출력
    1. 입력값 × 가중치
    2. 곱한 값들의 총합 (편향 더하는 건 옵션)
    3. 총합이 임계값($\theta$) 넘으면 1, 아니면 0 (활성화 함수 = 계단 함수)
- Training(학습) : 원하는 출력값 내보내도록 가중치 조정하는 작업

#### [용어 정리]
- 가중치(weight)
    - 입력값이 출력값에 주는 영향력(중요도)
    - 클수록 해당 입력값이 중요
- 활성화 함수
    - 퍼셉트론은 계단 함수 사용, 이후 신경망은 다양한 함수 사용
    - Sigmoid, Tanh -> 미분 불가 문제 해결
    - ReLU -> 기울기 소실 문제 해결
    - Leaky ReLU -> Dying ReLU 해결, but 실무에서는 ReLU 많이 씀
- 임계값(Threshold, $\theta$)
    - $\sum w_ix_i \ge \theta$ 이면 1, $\sum w_ix_i < \theta$ 이면 0
- 편향(bias)
    - 임계값을 좌변으로 넘긴 것 -> $\sum w_ix_i + b \ge 0$ 이면 1, 아니면 0
    - 임계값과 편향은 표현의 차이
    - 요즘 딥러닝은 편향 개념을 주로 사용


### 1-2. 단층 퍼셉트론과 그 한계
#### [단층 퍼셉트론]
- 입력층 + 출력층으로만 구성
- 값을 보내는 단계, 받아서 출력하는 단계로만 이루어짐

#### [논리 게이트]
- 디지털 회로에서 기본 논리 연산 수행하는 구성 요소
- n개의 입력값, 하나의 출력값
- AND, OR, NOT, NAND, NOR, XOR 등 -> 조합해서 복잡한 논리 회로 구성
- 단층 퍼셉트론으로 AND, NAND, OR 쉽게 구현 가능
    - AND : 둘 다 1일 때만 1 -> ex. $[w_1, w_2, b] = [0.5, 0.5, -0.7]$
    - OR : 둘 다 0일 때만 0 -> ex. $[0.6, 0.6, -0.5]$
    - NAND : 둘 다 1일 때만 0 -> ex. $[-0.5, -0.5, 0.7]$ (AND 부호 뒤집은 것)
    - XOR : 서로 다를 때만 1 -> 가능한 조합 없음

#### [단층 퍼셉트론의 한계]
- XOR 구현 불가
- 입력값과 가중치의 선형 결합 -> 직선 하나로 나눌 수 있는 선형 분류 문제만 가능
    - AND -> 직선 하나로 분리 가능
    - XOR -> 직선 하나로 분리 불가능, 곡선/비선형 영역으로 나눠야 함
- 가중치를 아무리 바꿔도 4개 중 최대 3개까지만 맞음 -> 구조상 불가능


### 1-3. 다층 퍼셉트론(MLP)
- XOR은 AND, NAND, OR 조합하면 만들 수 있음 -> 층을 더 쌓으면 가능
- 단, 층만 쌓는다고 비선형성 생기지 X
    - 활성화 함수 없으면 여러 층이 결국 하나의 연산으로 합쳐짐 -> $Wx + b$
    - 결정 경계 여전히 직선 -> XOR 불가
- MLP가 XOR 가능한 이유
    - 은닉층에서 비선형 활성화 함수로 새로운 특징 만들고, 다음 층에서 조합
    - ex. 은닉층 두 뉴런이 NAND, OR 계산 -> 출력층이 AND 적용 -> XOR
    - 뉴런 두 개 은닉층 하나로 충분
- 단층도 계단 함수(비선형) 쓰지만 가중합 한 번에 임계값 적용 -> 결정 경계 직선
- 구성 : 입력층 + 하나 이상의 은닉층 + 출력층

    <img src="https://upload.wikimedia.org/wikipedia/commons/4/46/Colored_neural_network.svg" width="250" style="background-color:white; padding:10px;">
- 심층 신경망(DNN) = 은닉층 2개 이상인 신경망
- 딥러닝 = 심층 신경망을 학습(가중치 스스로 찾도록)시킴


## 2. 딥러닝 모델의 학습
### 2-1. 순전파와 역전파
#### [학습 알고리즘 기본 구조]
1. 순전파 -> 입력값으로 예측
2. 오차 계산 -> 예측 정확도 측정
3. 역전파 -> 출력층에서 입력층 방향으로 오차 전달
4. 가중치 업데이트
5. 1~4 반복 -> 오차 0에 가까워짐

- 문제 풀기(순전파) -> 채점(오차 계산) -> 오답노트(역전파) -> 향상된 머리(가중치 업데이트)

#### [연쇄 법칙(Chain Rule)]
- 합성함수의 미분 = 각 함수 미분의 곱
- 입력이 출력에 준 영향 계산하려면 각 층이 다음 층에 준 영향을 순서대로 곱해야 함
- 역전파에서 각 가중치가 최종 손실에 얼마나 기여했는지 계산할 때 사용

$$ \frac{\partial L}{\partial w_1} = \frac{\partial L}{\partial a_n} \times \cdots \times \frac{\partial a_2}{\partial a_1} \times \frac{\partial a_1}{\partial w_1} $$

- 연쇄 법칙으로 구한 기울기 = 손실 줄이려면 가중치를 어떻게 바꿀지 알려주는 신호

#### [기울기 소실 문제]
- 은닉층 많고 각 단계 미분값이 1보다 작으면 -> 곱할수록 0에 가까워짐 ($0.25 \times 0.25 \times \cdots \approx 0$)
- 0.25 = 시그모이드 미분의 최댓값
- 앞쪽 층 가중치 거의 업데이트 X -> 학습 제대로 안 됨
- ReLU가 등장한 이유

#### [전체 흐름]
1. 순전파 = 예측하기
2. 손실 계산 = 얼마나 틀렸는지 측정
3. 역전파 = 책임 추적하기
    - $w_1$은 L에 직접 연결 X, $w_1$ -> $w_2$ -> L 간접 연결
    - $w_1$의 책임 = $w_1$이 $w_2$에 준 영향 × $w_2$가 L에 준 영향 -> 연쇄 법칙
4. 가중치 업데이트 = 수정하기 -> 얼마나 수정할지는 경사하강법에서 정함


### 2-2. 손실 함수
#### [목적함수 · 손실함수 · 비용함수]
- 목적함수 : 최적화하려는 전체 대상, 최대화/최소화 둘 다 가능
- 손실함수 : 데이터 하나에 대한 오차 -> ex. 1번 문제 틀려서 깎인 점수
- 비용함수 : 전체 데이터에 대한 손실 합/평균 -> ex. 시험 감점 합
- 손실함수와 비용함수는 혼용 많이 함 -> 둘 다 최소화가 목적

#### [손실 함수란?]
- 모델 예측이 얼마나 틀렸는지 수치로 표현
- 실제값과 예측값의 차이 계산
- 회귀 -> MSE, 분류 -> 크로스 엔트로피

#### [손실 함수 종류]
- 평균 제곱 오차(MSE) : $\frac{1}{N}\sum(y - \hat{y})^2$
    - 회귀 문제에 적합
    - ex. 집 가격 예측
- 이진 크로스 엔트로피(BCE) : $-\left(y\log\hat{y} + (1-y)\log(1-\hat{y})\right)$
    - 이진 분류 문제(로지스틱 회귀)
    - 정답 1이면 $p$가, 정답 0이면 $1-p$가 높아야 잘 예측한 것
    - 실제 정답에 부여한 확률에 $-\log$ 적용 -> 정답 확률 높으면 손실 작고, 낮으면 손실 큼
    - 정답 1인데 예측 확률 0에 가까우면 -> loss 급격히 증가 (확신했는데 틀림)
    - MSE는 loss 크기만 보고, BCE는 얼마나 확신했는지도 반영
    - 엔트로피 = 정보의 불확실성 -> 얼마나 확실성 갖고 맞췄는지도 봐서 이름에 들어감
- 크로스 엔트로피 오차(CEE) : $-\sum y\log\hat{y}$
    - 다중 클래스 분류
    - BCE와 거의 같은 형태
- BCE, CEE에 (-) 붙는 이유
    - 확률 0~1을 log에 넣으면 항상 음수 -> 손실은 양수여야 하니까 마이너스 붙임

| 손실 함수 | 사용 상황 | 예시 |
|:---:|:---:|:---:|
| MSE | 회귀 | 주택 가격, 온도, 주식 가격 예측 |
| BCE | 이진 분류 | 스팸 필터링, 질병 진단 |
| CEE | 다중 분류 | 손글씨 숫자 인식(MNIST), 이미지 분류(CIFAR-10) |


### 2-3. 활성화 함수
#### [활성화 함수란?]
- 입력값 × 가중치 + 편향 -> 활성화 함수 -> 출력값
- 필요한 이유
    1. 비선형성 -> 없으면 층 쌓아도 선형 계산 반복, XOR 못 품
    2. 미분 가능해야 함 -> 역전파는 미분값을 연쇄적으로 곱하기 때문
- 발전 과정
    - 계단 함수 -> 미분 불가, 역전파 안 됨
    - 시그모이드 -> 미분 가능, but 미분값 최대 0.25 -> 기울기 소실
    - ReLU -> 양수 구간 미분값 1 -> 기울기 소실 X

#### [계단 함수]
<img src="https://upload.wikimedia.org/wikipedia/commons/d/d9/Dirac_distribution_CDF.svg" width="300" style="background-color:white; padding:10px;">

- 가중합이 임계값 넘으면 1, 아니면 0
- 주로 단층 퍼셉트론에서 사용
- 미분값 0 -> 역전파 불가 -> 다층에서는 안 씀

#### [시그모이드 함수]
<img src="https://upload.wikimedia.org/wikipedia/commons/8/88/Logistic-curve.svg" width="300" style="background-color:white; padding:10px;">

- $f(x) = \frac{1}{1+e^{-x}}$
- 0과 1 사이 연속적인 값으로 변환하는 비선형 함수
- 계단 함수와 차이 -> 출력이 연속적 -> 얼마나 확실한지 표현 가능, 미분도 됨
- 미분값 최대 0.25 -> 층 쌓으면 기울기 소실

#### [ReLU 함수]
- $f(x) = \max(0, x)$
- 0보다 크면 그대로, 0 이하면 0
- 딥러닝에서 가장 많이 사용
    - 계산이 간단하고 빠름
    - 은닉층에 비선형성 추가
    - 양수 구간 미분값 1 -> 기울기 소실 완화
- 음수에서는 0이 돼서 뉴런이 죽음(Dead Neuron) -> leaky ReLU 같은 변형 있음, but 실무에선 기본 ReLU 많이 씀

#### [Softmax 함수]
- $f(x_i) = \frac{e^{x_i}}{\sum_j e^{x_j}}$
- 각 클래스 출력값을 확률로 변환
- 다중 분류 출력층에서 사용
- 0~1 사이, 전체 합 1
- ex. [고양이 0.7, 강아지 0.2, 새 0.1]
- Softmax + 크로스 엔트로피 -> 가장 표준적인 다중 분류 조합

#### [Tanh 함수]
<img src="https://upload.wikimedia.org/wikipedia/commons/8/87/Hyperbolic_Tangent.svg" width="300" style="background-color:white; padding:10px;">

- 시그모이드 변형, 출력 -1 ~ 1
- 중심이 0이라 가중치 업데이트 더 안정적
- 기울기 소실 여전히 발생 -> 실무에서 잘 안 씀

| 활성화 함수 | 출력 범위 | 주요 사용처 | 단점 |
|:---:|:---:|:---:|:---:|
| 계단 함수 | 0 또는 1 | 단층 퍼셉트론 | 미분 불가 |
| Sigmoid | 0 ~ 1 | 이진 분류 출력층 | 기울기 소실 |
| Tanh | -1 ~ 1 | 은닉층 (요즘은 ReLU에 밀림) | 기울기 소실 |
| ReLU | 0 ~ ∞ | 은닉층 전반 | 음수 입력 시 항상 0 |
| Softmax | 0 ~ 1 (합=1) | 다중 분류 출력층 | 은닉층엔 부적합 |


### 2-4. 경사하강법
#### [경사하강법 원리]
- 목표 : 손실을 가장 작게 만드는 W 찾기
- 손실함수를 W에 대한 함수로 보면 아래로 볼록한 곡선 -> 이 곡선의 바닥 찾기
- 미분값의 부호만 보면 됨
    - 최솟값 오른쪽 -> 기울기 양수 -> W 작아짐
    - 최솟값 왼쪽 -> 기울기 음수 -> W 커짐
    - 최솟값 도달 -> 기울기 0 -> 멈춤
- $W \leftarrow W - \alpha\frac{\partial E}{\partial W}$
- b도 똑같은 방법으로 찾음
- 안개 낀 산에서 발밑 기울기만 느끼면서 가장 가파른 쪽으로 한 걸음씩 내려가는 것

    <img src="https://upload.wikimedia.org/wikipedia/commons/f/ff/Gradient_descent.svg" width="250" style="background-color:white; padding:10px;">

#### [학습률(Learning Rate, $\alpha$)]
- 얼마나 움직일지 = 보폭
- 미분값 그대로 빼면 최솟값 지나치고 반대편에서 또 지나침 -> $\alpha$ 곱해서 이동량 조절
    - 너무 작음 -> 수렴은 하지만 오래 걸림
    - 너무 큼 -> 발산 (손실 계속 커짐)
- 0.001부터 시작 (Adam 기본값)
- 개발자가 직접 설정하는 하이퍼파라미터
    - 0.01, 0.001, 0.0001 등 작은 값부터 시도
    - 느리면 늘리고, 발산하면 줄임
    - 자동 : Learning Rate Scheduler, Grid Search, Random Search

#### [배치 · 에포크 · 이터레이션]
- 배치 : 매개변수 조정에 쓰는 데이터 묶음
    - 너무 크면 -> 느려지고 메모리 부족
    - 너무 작으면 -> 학습 불안정
- 에포크 : 전체 데이터셋을 학습하는 횟수
- 이터레이션 : 1 에포크 끝내는 데 필요한 배치 수
    - ex. 데이터 2,000개, 배치 크기 200 -> 이터레이션 10

#### [경사하강법 종류]
- 배치 경사하강법
    - 전체 데이터를 한 묶음으로
    - 1 Epoch 당 1번 업데이트, 메모리 많이 필요, 수렴 안정적
- 확률적 경사하강법(SGD)
    - 랜덤으로 선택한 데이터 하나로만 계산
    - 자원 적어도 가능, but 불안정하고 정확도 낮을 수 있음
- 미니 배치 경사하강법
    - 배치 크기 적절히 지정
    - 전체보다 빠르고 SGD보다 안정적 -> 가장 많이 사용
- 실무에서는 미니 배치, 배치 크기 32부터 시작, 보통 $2^n$으로


### 2-5. 옵티마이저
#### [경사하강법의 한계 - Local Minima]
- MSE, RMSE는 볼록함수 -> 경사하강법으로 반드시 바닥 도달
- 딥러닝 손실함수는 울퉁불퉁, 모양 예측 불가
- 진짜 바닥(Global Minimum)이 아닌 작은 웅덩이(Local Minimum)에서도 기울기 0 -> 갇힘
- 기울기 완만한 구간 -> 빠져나오는 데 오래 걸림
- 경사하강법을 더 똑똑하게 굴리는 방법 = 옵티마이저

#### [모멘텀(Momentum)]
- 가던 방향으로 계속 가도록 속도 추가
- 기존 경사하강법은 이전 기울기를 다 잊음 -> 관성 부여
- 실제 이동 방향 = 현재 기울기 + 이전 이동(관성)
- Local Minima 탈출 가능, 완만한 구간 빠르게 통과
- 모멘텀 계수 $\beta$ 일반적으로 0.9

#### [RMSProp]
- 기울기 큰 방향은 너무 많이, 작은 방향은 너무 조금 이동하는 문제
- 방향마다 학습률 따로 조정
- 급한 경사에서는 속도 줄이고, 완만한 경사에서는 유지하거나 증가

#### [Adam]
- RMSProp + Momentum -> 방향과 스텝 사이즈 둘 다 잡음
- 1차 모멘트(기울기 평균), 2차 모멘트(기울기 제곱 평균)로 학습률 조정

#### [어떤 걸 쓸까]
- 잘 모르겠다 / 대부분의 경우 -> Adam
- 학습이 자꾸 한쪽으로 튄다 -> 모멘텀
- 파라미터마다 기울기 크기 차이가 크다 -> RMSProp
- 일단 Adam으로 시작, 안 되면 바꾸기
- 모멘텀 = 방향 개선 / RMSProp = 보폭 개선 / Adam = 둘 다


### 2-6. 정리
- 모델 짤 때 정해야 하는 것 : activation / loss / optimizer / batch_size
    - 은닉층 activation -> ReLU (기울기 소실 완화)
    - 출력층 activation -> 다중 분류면 softmax
    - loss -> 다중 분류면 CEE(categorical_crossentropy)
    - optimizer -> Adam
    - batch_size -> 미니 배치, 32
- 출력 3개 + softmax + CEE -> 3개 클래스 중 하나 고르는 다중 분류 문제
- 출력 1개에 softmax 쓰면 무조건 1 나옴 -> 학습 안 됨
    - 이진 분류 -> sigmoid + BCE
    - 회귀 -> 활성화 함수 없이 + MSE
