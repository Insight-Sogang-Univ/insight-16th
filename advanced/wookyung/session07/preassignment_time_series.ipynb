# 시계열

## 1. 시계열 데이터란?
### 1-1. 시계열 데이터의 정의
#### 시계열 데이터란?
- 시간의 흐름에 따라 일정한 간격, 순서대로 관측·기록된 데이터 -> ex. 주가, 기온, 매출

$$ Y_1, Y_2, Y_3, \dots, Y_t, \dots, Y_T $$

- $t$ : 시점, $Y_t$ : t 시점에 관측된 값, $T$ : 마지막 관측 시점

#### 시계열 데이터의 조건
- 시간 순서가 의미를 가짐 -> 순서를 섞으면 데이터의 의미가 사라짐
- 관측 간격이 일정함 -> 이 간격을 주기라고 함
- 횡단면 데이터 vs. 시계열 데이터
    - 횡단면 : 한 시점에 여러 대상을 관찰 -> 행 순서를 섞어도 문제 X
    - 시계열 : 한 대상을 여러 시점에 걸쳐 관찰 -> 행 순서를 섞으면 의미가 사라짐

#### 단변량 시계열과 다변량 시계열
- 단변량 시계열 : 시간에 따라 하나의 변수만 관측한 데이터 -> 자기 자신의 과거 값만 보고 미래 예측
- 다변량 시계열 : 시간에 따라 여러 변수를 동시에 관측한 데이터 -> 다른 변수들의 영향까지 함께 고려
- 다변량이 항상 좋은 건 X
    - 미래 시점에도 그 값을 알 수 있는 변수인지 따져봐야 함
    - 그래서 요일, 공휴일처럼 미리 정해진 변수가 자주 쓰임

#### 시계열 분석과 목적
- 시계열 분석 : 시간 순서대로 쌓인 데이터에서 패턴을 찾아내고, 이를 바탕으로 미래 값을 예측하는 분석 방법
- 주 목표 : 미래 데이터 예측 (Forecasting)
- 기본 가정 : 과거의 패턴이 미래에도 어느 정도 반복될 것
    - 과거에 없던 일은 예측 어려움
    - 먼 미래일수록 불확실성이 커짐
- 분석 순서
    1. 시계열 이해 - 시계열의 특징 파악 
    2. 전처리 - 결측치·이상치 처리, 평활화, 요소분해 
    3. 정상성 판단 - 모델에 넣을 수 있는 상태인지 확인 & 만들기 
    4. 모델링 & 예측 - 단변량 / 다변량 / 딥러닝 
- 예측 자체가 목적이 아니라, 예측을 바탕으로 미리 결정하기 위해 씀


### 1-2. 시계열 데이터의 특징
#### 시간의존성 (Temporal Dependence)
- 현재 시점의 값이 과거 시점의 값에 영향을 받는 성질
    - 과거의 값이 현재 값에 영향을 미침
    - 보통 가까운 과거일수록 영향이 큼
    - 순서를 섞으면 안 됨
- 데이터 누수 (Data Leakage) 주의
    - 랜덤하게 섞어서 train/test를 나누면 미래로 학습해서 과거를 맞히는 상황 발생 -> 성능이 실제보다 좋게 나옴
    - 시계열은 반드시 과거로 학습하고, 미래로 평가해야 함

#### 계절성 (Seasonality) & 추세 (Trend)
- 계절성 : 일정한 주기(일, 주, 월, 년 등)로 비슷한 패턴이 반복되는 성질
- 추세 : 시간의 흐름에 따라 데이터가 장기적으로 증가하거나 감소하는 방향성
- 실제 데이터에서는 둘이 동시에 나타나는 경우가 많음 -> ex. 월별 항공기 승객 수 (증가 추세 + 여름 계절성)

#### 자기상관성 (Autocorrelation)
- 자기상관성 : 같은 변수의 현재 값과 과거 값이 얼마나 비슷하게 움직이는지를 나타내는 정도
    - 시계열의 가장 중요한 특성
    - 값의 범위는 -1 ~ 1
        - 1에 가까움 : 과거에 높았으면 지금도 높음 (비례)
        - -1에 가까움 : 과거에 높았으면 지금은 낮음 (반비례)
        - 0에 가까움 : 관계 X
- 시차(Lag) : 현재 값을 몇 시점 전의 값과 비교하는지
    - 시차를 바꿔가며 계산하면 반복주기(계절성)를 찾을 수 있음
    - 보통 가까운 과거일수록 자기상관이 높음
- 자기상관성의 문제점 -> 일반 회귀분석 불가
    1. 오차항 독립성 가정 위반 -> 통계적 추론(검정, 신뢰구간) 왜곡

    $$ \text{회귀 가정 : } Cov(\varepsilon_t, \varepsilon_s) = 0 \ (t \neq s) \quad \rightarrow \quad \text{자기상관 존재 : } Cov(\varepsilon_t, \varepsilon_{t-1}) \neq 0 $$

    2. 허위 상관관계 - 둘 다 시간에 따라 증가/감소하면 관계가 없어도 상관관계가 높게 나옴
    3. 예측의 어려움 & 모델 복잡성 증가 - 과거와의 관계까지 모델에 넣어야 함, 예측 오차가 쌓임, 패턴이 바뀔 수 있음
- 결론 : 시계열 전용 모델(4~6장)을 사용하거나 사전처리(차분, 추세 제거 등) 후 분석해야 함


### 1-3. 시계열 데이터의 분석요인
- 실제 시계열 그래프에는 여러 움직임이 겹쳐 있음 -> 보통 추세, 계절, 순환, 불규칙 4가지로 나눔

    <img src="https://f0nzie.github.io/hyndman-bookdown-rsuite/fpp_files/figure-html/6-decomp1-1.png" width="500" style="background-color:white; padding:10px;">

#### 추세 요인 (Trend Factor)
- 시계열이 시간 전반에 걸쳐 증가하거나 감소하는 경향성
- 짧은 오르내림은 무시하고 큰 흐름만 본 것

#### 계절 요인 (Seasonal Factor)
- 1년 이내의 일정한 주기로 같은 패턴이 반복되는 요인
- 날씨, 달력(요일, 공휴일, 명절), 사회적 관습 때문에 생김

#### 순환 요인 (Cycle Factor)
- 계절성보다 더 길고 불규칙한 주기로 반복되는 변동 -> 대표적인 원인은 경기 변동
- 계절성 vs. 순환성 : 주기가 정해져 있는지로 구분
- 주의점
    - 주기가 길어서 짧은 데이터로는 찾아내기 어려움
    - 장기 추세인지 긴 순환의 상승 구간인지 구분하기 어려움
    - 그래서 추세에 포함해 추세-순환 요인(Trend-Cycle)으로 묶기도 함

#### 불규칙 요인 (Irregular Factor)
- 추세, 계절, 순환 요인을 모두 제거하고 남은 나머지 움직임 -> 잔차(Residual) 또는 노이즈(Noise)
- 이상적으로는 위아래로 랜덤하게 흩어져 있어야 함 -> 유독 크게 튀는 값은 이상치일 수 있음

#### 추세성, 계절성, 순환성, 불규칙성 정리

| 요인 | 의미 | 주기 | 예측 가능성 |
|:---:|:---:|:---:|:---:|
| 추세 (T) | 장기적인 방향 | 없음 | 높음 |
| 계절 (S) | 정해진 주기의 반복 | 고정 (1년 이내) | 높음 |
| 순환 (C) | 불규칙한 주기의 오르내림 | 불규칙 (수년) | 낮음 |
| 불규칙 (I) | 설명되지 않는 나머지 | 없음 | 불가능 |


## 2. 시계열 데이터 전처리
- 현실 데이터는 결측치, 이상치, 노이즈가 섞여 있음
- 시계열은 값 하나가 앞뒤 값과 이어져 있어서 잘못된 값 하나가 주변 분석까지 망가뜨림 -> 시계열 특성을 반영한 전처리 필요
- 전처리 프로세스 : 데이터 확인 -> 결측치 처리 -> 이상치 처리 -> 평활화 -> 요소분해 -> 정상성 판단 & 확보

### 2-1. 결측치/이상치 탐지 및 처리
#### 시계열 데이터의 결측치
- 결측치 : 특정 시점의 값이 수집되지 않았거나 누락된 것
- 행을 지우면 관측 간격이 깨지고, 전체 평균으로 채우면 시간 흐름이 무시됨
- 추세와 계절성이 왜곡되지 않게 앞뒤 흐름을 고려해서 채워야 함

#### 결측치 처리 방법
- 삭제 : 관측 간격이 깨짐 -> 결측이 매우 적고 맨 앞이나 맨 뒤일 때만 고려
- 단순 대치 : 평균, 중앙값, 최빈값으로 채움 -> 추세와 계절성 무시
- 보간 : 앞뒤 관측값을 이용해 사이의 값을 추정 -> 시계열에서 가장 많이 사용
- 이동평균 대치 : 결측 시점 주변 일정 구간의 평균으로 채움
- 머신러닝 기법 : KNN, 회귀 모델 등으로 예측해서 채움 -> 정교하지만 복잡하고 계산량이 많음
- 보간법 : 주어진 데이터 포인트 값을 이용해 두 관측값 사이의 비어있는 값을 추정하는 방법
    - 선형 보간법 : 두 점 사이를 직선으로 연결
        - 장점 : 빠르고 간단, 직관적
        - 단점 : 비선형 관계나 급격한 변화가 있는 데이터에서 정확도가 떨어짐
    - 다항식 보간법 : 데이터 포인트를 모두 통과하는 다항식으로 근사
        - 장점 : 복잡한 비선형 관계를 잘 모델링
        - 단점 : 고차 다항식은 과적합 초래, 계산이 복잡함
    - 스플라인 보간법 : 구간별 다항식으로 보간 -> 3차 다항식을 쓰는 큐빅 스플라인(cubic spline)이 많이 사용됨
        - 장점 : 부드럽고 연속적, 고차 다항식보다 안정적
        - 단점 : 선형 보간보다 계산량이 많음
    - 조각별 상수 보간법 : 구간마다 동일한 상수 값을 유지해서 채움
        - forward fill (직전 값), backward fill (직후 값), 최근접 이웃 보간 (가장 가까운 값)
        - 장점 : 단순하고 원본 값만 사용
        - 단점 : 계단형이라 급격히 변하는 영역에서 정확도가 낮음
- 보간법 vs. 회귀
    - 공통점 : 관측값을 바탕으로 관측되지 않은 값을 유추
    - 회귀는 오차를 가장 적게 만드는 선을 찾음 (오차 허용) / 보간은 관측값을 완전히 신뢰해서 반드시 지나가도록 연결
    - 결측치 처리에서는 이미 있는 값을 바꾸면 안 되기 때문에 보간 사용
- 잘 모르겠으면 선형 보간부터 시작 / 곡선 형태면 스플라인 / 계단처럼 유지되는 값이면 ffill

#### 시계열 데이터의 이상치
- 이상치 : 데이터의 일반적인 패턴에서 크게 벗어난 극단적인 값
- 시계열에서는 언제 나타났는지가 중요 -> 같은 값이라도 시점에 따라 정상일 수도, 이상일 수도 있음 (ex. 기온 35도는 8월이면 정상, 1월이면 이상치)
- 추세와 계절성을 고려해서 판단해야 함

#### 이상치 탐지 방법
- IQR : Q1 − 1.5×IQR 미만, Q3 + 1.5×IQR 초과 값을 이상치로 판단
    - 일정 구간을 이동하면서 구간마다 계산하거나, 추세·계절성을 제거한 잔차에 적용
    - 장점 : 간단, 정규분포 가정 필요 X
    - 단점 : 추세나 계절성이 강한 시계열에는 부적합
- Z-score : 평균에서 몇 표준편차만큼 떨어져 있는지 계산 -> 보통 |Z| > 3이면 이상치

    $$ Z = \frac{x - \mu}{\sigma} $$

    - 이동 평균과 이동 표준편차로 구간마다 계산하면 추세 영향을 줄일 수 있음 (Rolling Z-score)
    - 단점 : 정규분포 가정, 평균과 표준편차가 이상치 자체에 영향을 받음
- STL 분해 : 추세(T), 계절(S), 잔차(R)로 분해한 뒤 잔차가 임곗값을 초과하면 이상치
    - 장점 : 추세와 계절성을 제거하고 진짜 이상한 값만 잡아냄
    - 단점 : 긴 시계열은 파라미터 튜닝 필요, 계산비용이 큼

#### 이상치 탐지 모델
- CART : 과거 구간 값으로 다음 시점을 예측하고, 예측 오차가 임곗값 이상인 지점을 이상치로 판단
    - 비선형 관계 포착 O, 노이즈가 많으면 과적합되기 쉬움
    - (참고) Isolation Forest : 무작위로 계속 나누면 이상치는 몇 번 안 나눠도 혼자 분리된다는 아이디어
- 클러스터링 기반 : 군집화 후 어느 군집에도 잘 속하지 않는 데이터를 이상치로 판단
    - K-means : 군집 중심에서 너무 먼 데이터 / DBSCAN : 밀집 영역에 속하지 못한 데이터
    - 라벨 없이 사용 가능, 패턴 전체가 이상한 경우도 잡음 / 하이퍼파라미터에 민감, 순서 정보 직접 반영 X
- 오토인코더 : 정상 데이터만으로 압축(Encoder) -> 복원(Decoder)을 학습 -> 복원 오차가 크면 이상치
    - LSTM Autoencoder를 쓰면 시간적 의존성까지 학습 가능
    - 복잡한 비선형·다변량 패턴 처리 O / 데이터와 계산량이 많이 필요, 해석이 어려움
- 처음에는 IQR, Z-score로 시작 -> 계절성이 뚜렷하면 STL -> 데이터가 크고 복잡하면 모델 기반

#### 이상치 처리 방법
- 먼저 오류인지 실제 사건인지 확인한 뒤 처리
- 삭제 : 그냥 지우면 관측 간격이 깨지므로 NaN으로 바꿔 결측치로 만든 뒤 다시 채움
- 단순 대치 : 평균, 중앙값 등으로 대체 -> 추세와 계절성 무시
- 보간 : 이상치를 결측치로 바꾼 뒤 앞뒤 값으로 보간 -> 시간 흐름을 반영할 수 있어 많이 사용


### 2-2. 평활화
#### 평활화의 개념
- 시계열의 자잘한 오르내림(노이즈)을 줄여 데이터를 부드럽게 만드는 기술 -> 큰 흐름 파악
- 주로 모델링 전 EDA 과정에서 사용

#### 이동평균
- 일정한 개수(N, 윈도우 크기)의 연속된 데이터를 묶어 평균을 내고, 한 칸씩 옮겨가며 계속 계산하는 방법

$$ MA_t = \frac{Y_t + Y_{t-1} + \cdots + Y_{t-N+1}}{N} $$

- N의 크기
    - 작을 때 : 최근 변화에 빠르게 반응 / 노이즈가 많이 남음
    - 클 때 : 그래프가 매끄러워져 큰 흐름이 잘 보임 / 변화에 늦게 반응, 세부 변화 사라짐
    - 보고 싶은 흐름의 길이에 맞게 정해야 함

    <img src="https://f0nzie.github.io/hyndman-bookdown-rsuite/fpp_files/figure-html/ressales3-1.png" width="500" style="background-color:white; padding:10px;">

- 사용 목적
    1. 노이즈 제거 & 추세 파악
    2. 계절성 제거 - N을 계절 주기와 똑같이 맞추면 한 주기 안의 오르내림이 상쇄됨 (ex. 월별 데이터에 12개월 이동평균)
    3. 결측치·이상치 대치
    4. 예측의 기초 자료 (ARIMA)
- 이동평균의 유형
    - 단순 이동평균 (SMA) : 모든 값에 같은 가중치 -> 간단하지만 최근 변화에 늦게 반응
    - 가중 이동평균 (WMA) : 최근 값일수록 큰 가중치 -> 가중치를 직접 정해야 함

    $$ WMA_t = \frac{3Y_t + 2Y_{t-1} + 1Y_{t-2}}{3+2+1} $$

    - 지수 이동평균 (EMA) : 과거로 갈수록 가중치가 지수적으로 줄어듦, 묶음 크기 없이 모든 과거 값 반영 -> α를 정해야 함

    $$ EMA_t = \alpha Y_t + (1-\alpha) EMA_{t-1} \quad (0 < \alpha < 1) $$

    - α가 클수록 최근 값을 더 중요하게 봄


### 2-3. 요소분해 (Decomposition, 시계열 분해)
#### 요소 분해
- 섞여 있는 시계열을 각각의 요인으로 다시 나누는 과정

$$ Y_t = f(T_t, S_t, C_t, R_t) $$

- 순환 요인은 따로 찾기 어려워서 보통 추세에 포함 -> 추세(T), 계절(S), 잔차(R) 3가지로 분해
- 하는 이유 : 각 패턴을 명확히 이해, 변동의 원인 분석, 더 정교한 분석

#### 분해 모형 (수학적 형태로의 분해)
- 가법 모형 (Additive Model) : $Y_t = T_t + S_t + R_t$
    - 각 요인이 독립적으로 더해짐
    - 계절성의 진폭이 일정할 때 적합
- 승법 모형 (Multiplicative Model) : $Y_t = T_t \times S_t \times R_t$
    - 추세가 커질수록 계절성, 불규칙성 효과도 비례해서 커짐
    - 진폭이 시간에 따라 달라질 때 적합 -> 현실의 매출, 주가, 방문자 수는 승법에 가까운 경우가 많음
    - 로그를 취하면 가법 모형처럼 다룰 수 있음 : $\log Y_t = \log T_t + \log S_t + \log R_t$
- 그래프의 물결 높이가 일정하면 가법, 점점 커지면 승법

#### 분해 방법 (구현 알고리즘)
- Seasonal Decompose - MA 기반 분해
    1. 추세 추출 - 중심 이동평균으로 추세(T) 추정
    2. Detrending - 가법 : Y − T / 승법 : Y ÷ T
    3. 계절성 추정 - 같은 계절 인덱스끼리 평균 -> 계절 요인(S), 연간 평균이 0이 되도록 정규화
    4. 잔차 산출 - 가법 : Y − T − S / 승법 : Y ÷ (T × S)
- STL - LOESS 기반 분해
    - Seasonal Decompose의 한계 보완, 이동평균 대신 LOESS 사용
    - LOESS : 각 지점마다 주변 데이터만으로 작은 회귀선을 그려 이어 붙임 (가까울수록 큰 가중치) -> 시간에 따라 바뀌는 추세와 계절도 잘 잡음
    - Inner Loop : 추세 제거 -> 계절 요인(S) 추정 -> 계절성 제거 -> 추세 요인(T) 추정 -> 반복하며 T와 S를 다듬음
    - Outer Loop : 잔차가 큰 시점에 작은 가중치 부여 후 Inner Loop 재수행 -> 이상치 영향 감소 (강건(robust)한 분해)

    <img src="https://daluchkin.github.io/fpp3_notes/figure_html/ch03-ts-decomposition-10-1.png" width="500" style="background-color:white; padding:10px;">

#### 요소분해 정리
- 결합 방식에 따라 -> 가법 모형(더하기) / 승법 모형(곱하기)
- 계산 방법에 따라 -> Seasonal Decompose(이동평균, 단순) / STL(LOESS, 유연·강건)
- 요소별로 따로 분석·모델링 가능, 잔차로 이상치 탐지, 계절조정 데이터 생성 등에 활용


## 3. 정상성 판단
### 3-1. 정상성이란?
- 정상성(Stationarity) : 시계열의 통계적 특성(평균, 분산, 자기상관 구조)이 시간에 따라 변하지 않는 성질
- 정상 vs. 비정상 시계열
    - 정상 : 평균, 분산 일정 / 일정한 평균 주위를 오르내리며 벗어나도 다시 돌아옴
    - 비정상 : 평균이 변함(추세), 분산이 커지거나 작아짐, 주기적으로 반복(계절성)
- 중요한 이유 : 과거의 패턴이 미래에도 유지된다고 가정할 수 있어야 예측 가능
- 현실 데이터는 대부분 비정상 -> 정상성을 판단(3-3)하고 확보(3-4)하는 과정 필요


### 3-2. 강한 정상성과 약한 정상성
#### 강한 정상성
- 시계열의 확률분포 자체가 시간에 따라 전혀 변하지 않는 성질

$$ (Y_{t_1}, Y_{t_2}, \dots, Y_{t_k}) \overset{d}{=} (Y_{t_1+h}, Y_{t_2+h}, \dots, Y_{t_k+h}) $$

- 모든 k, h에 대해 결합분포가 같아야 함 -> 평균, 분산뿐 아니라 왜도, 첨도 등 모든 특성이 일정
- 현실에서는 거의 찾아볼 수 없음

#### 약한 정상성
- 평균, 분산, 자기공분산만 시간에 따라 변하지 않으면 되는 성질
    1. 평균이 일정 (μ)
    2. 분산이 일정 (σ², 무한대 X)
    3. 자기공분산이 시점(t)이 아니라 시차(h)에만 의존
- AR, MA, ARIMA 등 대부분의 시계열 모델이 가정하는 정상성 = 약한 정상성


### 3-3. 정상성 판단하기
#### 시각적으로 정상성 파악하기
- 정상 : 일정한 수평선(평균) 중심으로 오르내림, 폭이 비슷, 평균 회귀, 반복 패턴 X
- 비정상 : 추세(평균 변화), 계절성(주기 반복), 분산 변화(나팔 모양)

    <img src="https://f0nzie.github.io/hyndman-bookdown-rsuite/fpp_files/figure-html/stationary-1.png" width="500" style="background-color:white; padding:10px;">

- ACF 그래프로도 확인 가능
    - 정상 -> 막대가 빠르게 0 근처로 떨어짐
    - 비정상 -> 막대가 아주 천천히 줄어듦

    <img src="https://f0nzie.github.io/hyndman-bookdown-rsuite/fpp_files/figure-html/acfstationary-1.png" width="500" style="background-color:white; padding:10px;">

#### 통계적 검정
- 그래프는 보는 사람마다 판단이 달라질 수 있음 -> 통계적 검정으로 객관적으로 확인
- 단위근 : 시계열을 표현하는 식의 근(해)이 1이 되는 경우 -> 과거 충격이 영구적으로 누적되는 특성
    - 단위근이 있으면 평균으로 돌아오지 않고 무작위로 떠돌게 됨 -> 비정상 시계열

    $$ Y_t = \phi Y_{t-1} + \varepsilon_t $$

    - φ < 1 : 충격이 점점 작아지다 사라짐 -> 평균으로 돌아옴 (정상)
    - φ = 1 (단위근) : 충격이 사라지지 않고 계속 쌓임 -> 평균으로 돌아오지 않고 분산이 계속 커짐 (비정상)

#### 단위근 검정 방법
- 두 검정은 귀무가설이 서로 반대라는 점이 핵심
- ADF (Augmented Dickey-Fuller) 검정
    - 과거 값의 영향이 사라지면 정상, 누적되어 사라지지 않으면 비정상
    - H₀ : 단위근이 있다 (비정상) / H₁ : 단위근이 없다 (정상)
    - p-value < 0.05 -> 귀무가설 기각 -> 정상
- KPSS (Kwiatkowski-Phillips-Schmidt-Shin) 검정
    - 일정한 평균 수준을 유지하는지 측정
    - H₀ : 정상이다 / H₁ : 비정상이다
    - p-value < 0.05 -> 귀무가설 기각 -> 비정상
- 두 검정 결과 조합

| ADF | KPSS | 진단 | 다음 액션 |
|:---:|:---:|:---:|:---:|
| 정상 | 정상 | 정상 시계열 | 바로 모델링 |
| 비정상 | 비정상 | 비정상 시계열 | 차분 등으로 정상성 확보 후 재검정 |
| 비정상 | 정상 | 추세 정상 시계열 | 추세 제거 후 재검정 |
| 정상 | 비정상 | 차분 정상 시계열 | 차분 후 재검정 |

- 추세 정상 : 결정적 추세가 있지만 추세선 주위로는 안정적 -> ADF는 단위근과 구분 못해서 비정상으로 판단
- 차분 정상 : 단위근이 있어 충격이 누적됨 -> KPSS가 무작위로 떠도는 부분을 감지해서 비정상으로 판단
- 실무에서는 ADF와 KPSS를 함께 사용해서 추세 제거와 차분 중 무엇을 할지 결정


### 3-4. 정상성 확보를 위한 여러가지 방법들
- 정상성이 깨지는 원인 : 추세, 계절성, 분산 변화 -> 원인을 하나씩 제거

#### 추세 제거
- 차분 : 값 자체 대신 얼마나 변했는지를 봄

$$ \Delta Y_t = Y_t - Y_{t-1} $$

- 차분으로 정상성이 확보되는 이유
    - 추세가 사라짐 -> 변화량은 비슷한 값 주위에 머물러서 평균이 일정해짐
    - 단위근이 사라짐 -> 계속 쌓이던 충격이 풀리고 새로 생긴 충격만 남음
- 차분의 종류
    - 1차 차분 : 기본 차분
    - 2차 차분 : 1차 차분한 결과를 한 번 더 차분 -> 곡선 형태의 추세 제거
        - 많이 하면 오히려 없던 패턴이 생길 수 있어서 보통 2차까지, 정상이 되는 최소 횟수만
    - 역차분 : 차분한 값을 다시 원래 값으로 되돌림 -> 예측 결과를 원래 단위로 해석하려면 필수

#### 계절성 제거
- 계절 차분 : 한 주기(m) 전의 값을 빼는 방법 -> 작년 같은 때와 비교

$$ \Delta_m Y_t = Y_t - Y_{t-m} $$

#### 분산 안정화
- 로그 변환 : 큰 값은 많이, 작은 값은 조금 줄여서 변동 폭을 비슷하게 정리
    - 승법 모형 데이터를 가법 모형처럼 다룰 수 있음
    - 한계 : 0이나 음수에는 사용 X (모든 값에 1을 더한 뒤 취하기도 함), 값에 비례해 커지는 변동에만 효과적

    <img src="https://f0nzie.github.io/hyndman-bookdown-rsuite/fpp_files/figure-html/a10diff-1.png" width="500" style="background-color:white; padding:10px;">

- 정리 : 단위근 검정으로 판단 -> 비정상 유형에 따라 처리
    - 추세 -> 차분, 추세 제거
    - 계절성 -> 계절 차분 or Deseasoning
    - 분산 변화 -> 로그 변환, Box-Cox 변환


## 4. 단변량 시계열 분석
- 변수 하나의 기록만 가지고 자기 자신의 과거로 분석하고 예측하는 것

### 4-1. 안정적 시계열 분석
- 안정적 시계열 = 정상 시계열 -> 돌아갈 평소가 있는 시계열
- 값(y) vs. 충격(ε)
    - 값 : 그날 관측된 최종 결과, 과거가 섞여 있음
    - 충격 : 그날 새로 생긴 예측할 수 없는 변동, 과거와 무관, 모형을 적용해야 추정 가능
    - 미리 알 수 있는 것(주말 효과, 공지한 할인 행사 등)은 충격 X

#### 자기 회귀 모델 (Autoregressive Model, AR)
- 오늘 값을 자기 자신의 과거 값으로 설명하는 모형 -> 자기상관성 활용

$$ AR(p) : y_t = c + \phi_1 y_{t-1} + \phi_2 y_{t-2} + \cdots + \phi_p y_{t-p} + \varepsilon_t $$

- AR(1) : $y_t = c + \phi_1 y_{t-1} + \varepsilon_t$ -> 오늘 값 = 어제 값의 일부 + 오늘 충격
- 예측할 때 오늘 충격은 아직 모르니 0으로 둠
- 어제 값 안에 그저께 값이 들어 있음 -> 영향이 끝없이 이어지며 서서히 옅어짐

#### 이동 평균 모델 (Moving Average, MA)
- 오늘 값을 과거의 충격으로 설명하는 모형

$$ MA(q) : y_t = \mu + \varepsilon_t + \theta_1 \varepsilon_{t-1} + \theta_2 \varepsilon_{t-2} + \cdots + \theta_q \varepsilon_{t-q} $$

- MA(1) : $y_t = \mu + \varepsilon_t + \theta_1 \varepsilon_{t-1}$ -> 오늘 값 = 오늘 충격 + 어제 충격의 일부
- 충격의 영향이 q 시점 뒤까지만 남고 그 다음부터 0 -> 영향이 딱 끊김

#### AR과 MA 한눈에 비교하기
- 지나간 충격 : AR은 어제 값에 담겨 넘어옴 / MA는 충격 자체가 직접 넘어옴
- 영향 기간 : AR은 끝없이 이어지며 옅어짐 / MA는 정해진 시점 뒤에 끊김

#### AR, MA 모델의 전제 조건

| 구분 | 정상성 | 가역성 |
|:---:|:---:|:---:|
| 뜻 | 충격을 받아도 평소로 돌아옴 | 관측값으로 지난 충격을 계산해 낼 수 있음 |
| AR | 계수 조건 필요 | 항상 만족 |
| MA | 항상 만족 | 계수 조건 필요 |

- 정상성
    - AR(1) : |φ| < 1 이어야 함 -> 1이면 충격이 그대로 남고, 1보다 크면 불어남
    - AR(p) : 특성방정식의 모든 근의 절댓값이 1보다 커야 함

    $$ 1 - \phi_1 z - \phi_2 z^2 - \cdots - \phi_p z^p = 0 $$

    - MA : 충격의 영향이 q 시점 뒤에 끝나므로 항상 정상
    - ADF, KPSS는 데이터를 보는 검정(모형 만들기 전), 모델의 정상성은 추정한 계수를 보는 것(모형 만든 뒤) -> 서로 대신할 수 없음
- 가역성
    - 데이터에는 최종 결과(y)만 있으므로, 충격을 관측 가능한 과거 값으로 거꾸로 표현할 수 있어야 함
    - AR : 오늘 충격 = 오늘 값 − 계수 × 어제 값 -> 전부 관측 가능하므로 항상 가역
    - MA(1) : |θ| < 1 이어야 함 -> 1보다 크면 첫날 어림한 오차가 계속 불어남
    - MA(q) : 특성방정식의 모든 근의 절댓값이 1보다 커야 함

    $$ 1 + \theta_1 z + \theta_2 z^2 + \cdots + \theta_q z^q = 0 $$

#### ARMA (Autoregressive Moving Average) 모형
- AR과 MA를 한 식에 합친 모형 -> 과거 값과 과거 충격 양쪽의 영향을 받음

$$ y_t = c + \phi_1 y_{t-1} + \cdots + \phi_p y_{t-p} + \varepsilon_t + \theta_1 \varepsilon_{t-1} + \cdots + \theta_q \varepsilon_{t-q} $$

- AR만으로 설명할 때보다 더 적은 계수로 설명 가능
- AR 부분은 정상성 조건, MA 부분은 가역성 조건을 각각 만족해야 함
- 특수한 경우 : ARMA(p,0) = AR(p), ARMA(0,q) = MA(q), ARMA(0,0) = 백색잡음

#### 차수 결정 방법: 자기 상관 함수
- ACF (Auto Correlation Function) -> q 결정
    - 오늘 값과 k시점 전 값의 상관계수, 중간 시점을 거쳐 온 영향까지 합쳐서 잼
    - 시차 0에서는 항상 1, 점선(보통 ±1.96/√n) 안쪽이면 0으로 봄

$$ \rho(k) = \frac{Cov(y_t, y_{t-k})}{Var(y_t)} $$

- PACF (Partial Autocorrelation Function) -> p 결정
    - 중간 시점들의 영향을 빼고, 오늘 값과 k시점 전 값의 직접적인 관계를 잼
    - 시차 1에서는 ACF와 값이 같음

$$ \phi_{kk} = Corr(y_t, y_{t-k} \mid y_{t-1}, \dots, y_{t-k+1}) $$

- ACF와 PACF로 차수 및 모형 결정하기

| ACF | PACF | 모형 |
|:---:|:---:|:---:|
| 천천히 감소 or 진동 감소 | p+1 시점부터 0에 근접 | AR(p) |
| q+1 시점부터 0에 근접 | 천천히 감소 or 진동 감소 | MA(q) |
| q+1 시점부터 0에 근접 | p+1 시점부터 0에 근접 | ARMA(p, q) |

- 막대가 급격히 줄어들기 직전의 시차를 고름 -> PACF에서 고르면 p, ACF에서 고르면 q


### 4-2. 불안정 시계열 분석
- 불안정 시계열 : 돌아갈 평소가 없는 시계열 (정상성 X)
    - 추세 -> 차분 (d) -> ARIMA
    - 계절성 -> 계절 차분 (D) -> SARIMA
- AR, MA는 결국 평소 근처로 돌아온다고 예측하기 때문에 평소가 없는 데이터에는 맞지 않음
- 핵심 : 모형을 고치는 게 아니라, 데이터를 모형이 다룰 수 있는 숫자로 바꿔주는 것
- 프로세스 : 비정상 시계열 -> 차분 or 계절 차분으로 정상성 확보 -> ARMA로 분석

#### ARIMA (Autoregressive Integrated Moving Average)
- ARMA에 차분을 더한 모형 -> ARIMA(p, d, q)
    - p : AR 차수, d : 차분 횟수, q : MA 차수
- 흐름
    1. 차분해서 증감을 만듦 - $w_t = y_t - y_{t-1}$
    2. 증감을 ARMA로 예측 - $w_t = c + \phi_1 w_{t-1} + \varepsilon_t + \theta_1 \varepsilon_{t-1}$ (ARIMA(1,1,1))
    3. 예측한 증감을 어제 값에 더해 되돌림 - $y_t = y_{t-1} + w_t$
- 특수한 경우
    - 백색잡음 : ARIMA(0,0,0)
    - AR(p) : ARIMA(p,0,0) / MA(q) : ARIMA(0,0,q)
    - 랜덤워크 : 상수가 없는 ARIMA(0,1,0) -> $Y_t = Y_{t-1} + \varepsilon_t$
        - 내일 값의 최선의 예측은 오늘 값 -> ex. 주가, 환율
    - 표류를 포함하는 랜덤워크 : 상수가 있는 ARIMA(0,1,0) -> $Y_t = Y_{t-1} + \varepsilon_t + c$
        - c(표류)만큼 매일 한 방향으로 흘러감

#### SARIMA (Seasonal ARIMA)
- ARIMA에 계절 차분과 계절 AR, 계절 MA를 더한 모형
    - 지난 시점 대비 증감 -> 추세 제거 / 한 주기 전 대비 증감 -> 계절마다 다른 수준 제거
- 흐름 (월별 데이터)
    1. 계절 차분 - $w_t = y_t - y_{t-12}$
    2. 증감 예측 - 지난달 항과 작년 같은 달 항을 함께 넣음 (대문자 계수가 계절 AR, 계절 MA)

    $$ w_t = c + \phi_1 w_{t-1} + \Phi_1 w_{t-12} + \varepsilon_t + \theta_1 \varepsilon_{t-1} + \Theta_1 \varepsilon_{t-12} $$

    3. 되돌리기 - $y_t = y_{t-12} + w_t$
- 하이퍼파라미터 : $SARIMA(p, d, q)(P, D, Q)_m$
    - p, d, q : 비계절 AR 차수, 차분 횟수, MA 차수 (ARIMA 그대로)
    - P, D, Q : 계절 AR 차수, 계절 차분 횟수, 계절 MA 차수 (한 주기 전 기준으로 한 번 더)
    - m : 계절 주기 -> 월별 12, 분기별 4, 요일 효과 7
- 계절성 데이터의 ACF는 시차 12, 24, 36에서 다시 솟고 물결 높이가 줄지 않음 -> p, q를 읽을 수 없으므로 계절 차분을 먼저 수행

#### 불안정 시계열에서의 하이퍼파라미터 결정 방법
- ACF, PACF는 시각적 도구일 뿐 명확한 결과를 주는 경우가 많지 않음
- d, D 결정 : ADF로 정상성 판단 -> 정상이면 0, 아니면 정상이 나올 때까지 차분 반복 (대부분 0 또는 1)
    - 추세 때문이면 차분(d), 계절성 때문이면 계절 차분(D)
- p, q 결정
    1. ACF, PACF로 범위 추정
    2. AIC, BIC 같은 정보 기준으로 그리드 서치 후 최종 선택 -> 값이 낮을수록 좋음
        - 보통 0~3 범위, 또는 그래프로 정한 값에서 ±1 범위에서 계산


## 5. 다변량 시계열 분석
- 여러 변수를 함께 놓고, 각 변수를 자기 자신의 과거와 다른 변수의 과거로 예측

### 5-1. VAR (Vector Autoregression)
- AR을 변수 여러 개로 넓힌 것 -> 변수마다 식이 하나씩 있고, 각 식에 모든 변수의 과거 값이 들어감

$$ X_t = a_{11} X_{t-1} + a_{12} Y_{t-1} + \varepsilon_{x,t} $$
$$ Y_t = a_{21} X_{t-1} + a_{22} Y_{t-1} + \varepsilon_{y,t} $$

- $a_{11}, a_{22}$ : 자기 자신에게서 넘어오는 영향 (AR 계수와 같은 역할)
- $a_{12}, a_{21}$ : 서로에게 주는 영향 (VAR에서 새로 생긴 계수)
- 적합한 경우
    - 변수들이 서로 영향을 주고받는 경우 (양방향)
    - 한 변수가 변하면 다른 변수가 어떻게 반응하는지 알고 싶을 때
    - 모든 변수가 정상이거나 차분하면 정상이 되는 경우
    - 변수 수가 적고 데이터가 충분한 경우
- 부적합한 경우
    - 영향이 한 방향뿐인 경우 -> 단변량 모형에 설명변수로 넣으면 됨
    - 변수는 많은데 데이터가 적은 경우 -> 변수가 2개면 계수 4개, 3개면 9개로 빠르게 늘어남
    - 비율이 고정되지 않은 관계가 강한 경우 -> 딥러닝 고려
- Granger Causality Test (그레인저 인과관계 검정) : VAR에 사용할 변수 선택
    - X의 과거 정보가 Y의 미래 예측에 도움이 되는지 검정
    - 제한 모델(Y의 과거값만) vs. 비제한 모델(Y의 과거값 + X의 과거값) 비교
    - H₀ : X는 Y 예측에 도움 X / H₁ : X는 Y 예측에 도움 O
    - p-value < 0.05 -> X를 VAR 모델에 포함
    - 양방향으로 검정
    - Granger 인과 ≠ 진짜 인과 -> 예측 가능성(시간적 선후관계)만 의미

#### VAR 모델을 이용한 분석 방법
- 충격 반응 함수 (IRF, Impulse-Response Function)
    - 한 변수에 충격을 한 번 줬을 때 다른 변수가 시간에 따라 어떻게 반응하는지 보는 도구
    - 계수표만으로는 알 수 없는 반응의 시점과 크기를 확인 가능
- 예측오차 분산 분해 (Variance Decomposition)
    - 예측이 빗나간 정도를 어느 변수의 충격 탓인지 비율로 나눠서 보여줌
    - 먼 미래를 예측할수록 다른 변수 충격의 몫이 커짐
- IRF는 어떻게 반응하는가 / 분산 분해는 누구의 영향이 얼마나 큰가


## 6. 딥러닝을 활용한 시계열 분석
- 등장 배경 (전통 통계 모델의 한계)
    1. 데이터가 많으면 처리 속도가 느려짐
    2. 크기가 고정된 데이터만 가능
    3. 비선형 패턴 포착이 어려움

### ① 순환 신경망 (RNN 계열)
- RNN, LSTM, GRU (이전 세션 내용)
- 적합 : 순차적 패턴이 중요하고, 과거 정보가 미래 예측에 영향을 주는 문제
- 부적합 : 매우 긴 시계열(수천~수만 step)은 학습 어려움, 병렬 처리 효율이 낮고 Transformer 대비 느림

### ② Transformer 계열
- Informer (장기 예측 특화) : 불필요한 연산을 줄여 계산 복잡도를 낮춤 -> 매우 긴 시계열에 강함 / 짧은 시계열은 과적합
- Autoformer (주기성 포착 특화) : 요소 분해(추세/계절성) 개념을 딥러닝 구조에 결합 -> 주기 패턴이 뚜렷한 데이터에 강함 / 구조 복잡
- TFT (Temporal Fusion Transformer) : 다양한 외부 변수를 섞어서 예측 가능, 어텐션 스코어로 어떤 변수가 중요했는지 해석 가능 -> 실무 활용도 1위 / 학습 시간 오래 걸림

### ③ Foundation Model
- 거대한 데이터로 사전 학습해서 다양한 문제를 해결할 수 있는 범용 모델 -> 사전 학습 후 Fine-tuning으로 특화
- 유형
    - LLM 기반 모델 : LLM을 시계열 분석에 활용
        - Without Adaptation : LLM이 이미 시계열 지식을 가지고 있다고 가정, 프롬프트 엔지니어링에 집중
        - Adapt LLM : LLM을 시계열 데이터로 Fine-tuning
        - Adapt to LLM : 시계열을 텍스트로 변환 후 LLM 활용
    - 시계열 전용 Foundation Model : 시계열 데이터로 사전 학습 -> TimeGPT, TimesFM, Moirai, Chronos

### ④ 딥러닝 모델 선택 가이드
- 핵심 트레이드오프
    - 단순함 vs. 성능 : RNN < LSTM/GRU < Transformer
    - 해석 가능 vs. 정확도 : 통계 모델 < 딥러닝 (단, TFT는 어텐션으로 해석 가능)
    - 학습 시간 vs. 예측 성능 : 간단한 모델은 빠르고, 복잡한 모델은 정확함
    - 데이터 크기 : 적으면 통계 모델/RNN, 많으면 Transformer
- 상황별 추천 모델
    - 짧은 시계열 (< 100) -> 통계 모델, RNN
    - 중간 길이 (100~1000) -> LSTM, GRU
    - 매우 긴 시계열 (1000+) -> Informer, Autoformer
    - 장기 의존성 중요 -> LSTM, Transformer
    - 주기 패턴 뚜렷 -> Autoformer
    - 외생 변수 많음 -> TFT
    - 대규모 데이터 -> Transformer 계열
    - 범용 예측 필요 -> Foundation Model
- 데이터가 적을 때는 딥러닝보다 통계 모델(ARIMA 등)이 더 효과적일 수 있음
