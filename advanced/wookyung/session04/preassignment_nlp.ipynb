# 언어모델&생성모델

## 1. NLP와 언어모델
### 1) NLP
- 자연어의 특징
    - 이산적 데이터 -> 숫자나 벡터로 변환(Embedding) 필요
    - 순차적 데이터 -> 순서와 문맥을 고려하는 모델 구조 필요
    - 중의성과 모호성 -> 문맥 전체를 파악해야 함
    - 희소성 -> 모든 표현을 학습하기 어려움
    - 가변성과 진화 -> 데이터로부터 패턴을 학습하는 방식 필요
- NLP : 자연어를 컴퓨터가 분석하고 이해하며 생성할 수 있도록 하는 인공지능 기술
- NLP의 역할
    1. 형태소 분석 - 문장을 형태소로 나누고 품사 분석
    2. 구문 분석 - 문장의 문법적 구조 분석
    3. 의미 분석 - 문맥에 맞는 의미 파악
    4. 담화 분석 - 여러 문장 사이의 연결 관계와 맥락 분석

### 2) 언어 모델
- 대량의 텍스트를 학습해서 단어(토큰)의 등장 확률을 예측하고, 문장의 자연스러움을 판단하는 모델
- 앞에 나온 단어와 문맥으로 다음 단어를 예측
- 확률이 높은 문장 = 자연스러운 문장
- 학습 방식에 따라 통계적 언어 모델(SLM) / 신경망 언어 모델(NLM)로 구분


## 2. 통계적 언어 모델(SLM)
### 1) SLM이란?
- 분포 가설에 근거 : 비슷한 문맥에서 등장하는 단어들은 유사한 의미를 가질 가능성이 높다
- 조건부 확률 : 앞에 나온 단어들을 조건으로 다음 단어의 확률 계산
    - ex. $P(갔다 \mid 나는\ 학교에)$
- 카운트 기반 확률 : 특정 문맥 다음에 어떤 단어가 얼마나 자주 등장했는지로 확률 계산
- 한계 : 희소 문제(Sparsity Problem)
    - 학습 데이터에 한 번도 등장하지 않은 조합은 확률을 0으로 계산

### 2) N-gram
- 직전 N-1개의 단어만 보고 다음 단어를 예측하는 통계적 언어 모델
- 문장이 길어질수록 생기는 희소성, 계산 복잡성 문제를 완화하기 위해 등장
- SLM과 별개가 아니라 SLM의 계산을 단순화한 것
- 한계
    - 희소 문제 : N이 커질수록 가능한 조합 수가 급격히 증가
    - 문맥 제한 : 멀리 떨어진 단어 사이의 관계 파악 어려움

### 3) Perplexity
- 언어 모델이 문장을 예측할 때 얼마나 "혼란스러워하는지"를 나타낸 지표
- PPL 낮음 -> 잘 예측
- PPL 높음 -> 예측 어려움
- SLM, NLM 성능 평가에 모두 사용 -> NLM이 SLM보다 낮은 PPL 달성 가능


## 3. 신경망 언어 모델(NLM)
- SLM의 한계를 완화하기 위해 인공 신경망으로 단어의 의미와 문맥을 학습하는 모델
- 핵심 키워드 : 임베딩
    - 단어를 연속적인 숫자 벡터로 변환해서 단어 간 의미적 관계를 표현
    - 의미나 문맥이 비슷한 단어들은 벡터 공간에서 가까이 위치

| 구분 | SLM | NLM |
|:---:|:---:|:---:|
| 학습 방식 | 단어 등장 빈도와 확률 계산 | 신경망의 가중치 학습 |
| 단어 표현 | 개별 기호 | 연속적인 벡터 |
| 희소 문제 | 관측되지 않은 조합에 취약 | 완화 |
| 계산 비용 | 낮음 | 높음 |


## 4. 딥러닝 기반 언어모델
### 1) Transformer의 등장
- RNN, LSTM의 한계 : 장기 의존성 문제, 기울기 소실 문제, 순차 처리의 비효율성
- 2017년 "Attention Is All You Need" 논문에서 제안
- Self-Attention + 병렬 처리 + 위치 인코딩
- BERT, GPT 등 현대 언어 모델의 기반

    <img src="https://nlpinkorean.github.io/images/transformer/transformer_resideual_layer_norm_3.png" width="450" style="background-color:white; padding:10px;">

- 인코더 : Self-Attention + Feed Forward Network
- 디코더 : Masked Self-Attention + Cross-Attention + Feed Forward Network

### 2) BERT
- Transformer의 인코더 구조 기반, 문장의 양방향 문맥을 학습해 언어의 의미를 이해하는 모델
- 특징
    - 양방향 : 앞뒤 문맥을 모두 고려
    - 사전학습(Pre-training) + 파인튜닝(Fine-tuning)

    <img src="https://nlpinkorean.github.io/images/bert/bert-transfer-learning.png" width="450" style="background-color:white; padding:10px;">

- 사전학습 방법
    - MLM(Masked Language Model) : 일부 토큰을 [MASK]로 가리고 앞뒤 문맥으로 원래 토큰 예측

        <img src="https://nlpinkorean.github.io/images/bert/BERT-language-modeling-masked-lm.png" width="400" style="background-color:white; padding:10px;">

    - NSP(Next Sentence Prediction) : 문장 B가 문장 A 바로 다음 문장인지 예측
- BERT 이후 모델
    - RoBERTa : NSP 제거, Dynamic Masking 적용
    - ALBERT : 파라미터 수를 줄인 경량화 모델, NSP 대신 SOP

### 3) GPT
- Transformer의 디코더 기반, 이전 단어들로 다음 단어를 예측하고 텍스트를 생성하는 모델
- Masked Self-Attention : 이전 토큰들만 보고 다음 토큰 예측
- 자기회귀 방식
- In-Context Learning : 파인튜닝 없이 프롬프트의 지시와 예시만으로 작업 수행
    - Zero-shot / One-shot / Few-shot

| 구분 | BERT | GPT |
|:---:|:---:|:---:|
| 기반 구조 | Transformer Encoder | Transformer Decoder |
| 방향성 | 양방향 | 단방향(자기회귀) |
| 학습 방법 | MLM, NSP | 다음 토큰 예측 |
| 주요 목적 | 언어 이해 | 텍스트 생성 |

### 4) LLM
- 방대한 텍스트 데이터와 대규모 신경망으로 언어를 이해하고 생성하도록 학습된 모델
- 사전학습 -> 파인튜닝
- 새로운 문장을 생성해야 해서 GPT 계열(자기회귀) 구조가 널리 사용됨
- 한계
    - 환각 현상(Hallucination) : 사실과 다른 정보를 그럴듯하게 생성
    - 높은 학습 및 업데이트 비용, 최신 정보 반영 X

### 5) RAG
- 외부 지식 검색 + LLM의 생성 능력을 결합해서 답변의 정확성과 최신성을 높이는 기술

- 구조
    - 질의 인코더 : 질문을 벡터로 변환
    - 지식 검색기 : Vector DB에서 유사도가 높은 문서 검색
    - 지식 증강 생성기 : 질문 + 검색된 문서를 LLM에 넣어 답변 생성
- 장점 : 재학습 없이 최신 정보 반영, 환각 현상 완화

### 6) LangChain
- LLM을 외부 데이터, API, 도구와 연결해서 AI 애플리케이션을 쉽게 개발하게 해주는 오픈소스 프레임워크
- 특징 : 추상화, 표준화, 체이닝
- LangGraph
    - 노드와 엣지로 연결해서 조건 분기와 반복이 가능한 워크플로 구현
    - LangChain의 선형적 구조 한계를 보완
- LangSmith
    - 실행 과정 추적, 성능 평가, 모니터링을 지원하는 플랫폼
- 외부 API 활용 시 API 키는 타인에게 공유 X


## 5. sLM(small Language Model)
### 1) sLM이란?
- LLM보다 규모와 범위가 작은 언어 모델
- 메모리와 연산 자원을 적게 사용 -> 자원이 제한된 환경에 적합
- 오프라인 추론 가능

| 구분 | LLM | sLM |
|:---:|:---:|:---:|
| 성능 | 복잡한 추론과 다양한 작업에 강점 | 특정 분야나 단순한 작업에 효율적 |
| 연산 비용 | 높음 | 낮음 |
| 활용 환경 | 고성능 서버 | 서버, 스마트폰, 노트북 등 |

### 2) 모델 압축 방법
- 성능은 최대한 유지하면서 크기와 연산량을 줄이는 기술
- 가지치기(Pruning) : 중요도가 낮은 연결이나 뉴런 제거
- 양자화(Quantization) : 숫자의 표현 정밀도를 낮춤 (FP32 -> INT8)
- 저차원 분해(Low-Rank Factorization) : 큰 가중치 행렬을 작은 행렬들의 곱으로 근사
- 지식 증류(Knowledge Distillation) : 큰 모델(Teacher)의 지식을 작은 모델(Student)에 전달

### 3) sLM의 예시
- DistilBERT, Gemma, GPT-4o mini, Llama, Ministral, Phi

### 4) sLM의 장점
- 성능적 측면 : 빠른 응답 속도, 제한된 환경에서도 실행 가능, 보안 및 데이터 보호
- 비용적 측면 : 낮은 운영 비용, 적은 전력 소비, 하드웨어 비용 절감

### 5) sLM의 한계
- 편향(Bias)
- 복잡한 작업에서의 성능 저하
- 환각 현상

### 6) LLM과의 결합-결합 추론
- 결합 추론(Hybrid Inference) : sLM과 LLM을 함께 활용하는 방식
- 간단한 작업 -> sLM, 복잡한 작업 -> LLM
- 성능과 비용 사이의 균형


## 6. 생성모델의 개념
### 1) 분류모델 vs. 생성모델
- 생성 모델 : 학습 데이터의 분포를 따르는 유사 데이터를 생성하는 모델
- 분류 모델 : $P(Y \mid X)$ 모델링 -> 데이터의 차이점을 학습해서 분류
- 생성 모델 : $P(X)$ 모델링 -> 데이터의 확률분포를 학습해서 새로운 데이터 생성

### 2) 생성모델의 분류
- 확률밀도를 직접 계산할 수 있는지에 따라 나뉨

    <img src="https://lilianweng.github.io/posts/2021-07-11-diffusion-models/generative-overview.png" width="450" style="background-color:white; padding:10px;">

- 명시적 확률밀도 모델 : $P(X)$ 값을 수치로 출력 가능
    - 다루기 쉬운 밀도 모델 : 확률밀도 직접 계산 -> PixelRNN/PixelCNN 등
    - 근사적 밀도 모델 : 근사적인 방법 활용 -> VAE(변분 추론), Boltzmann Machine(마르코프 연쇄)
- 암시적 확률밀도 모델 : $P(X)$를 계산하지 않고 새로운 데이터 생성에 초점
    - 직접 샘플링 -> GAN
    - 간접/반복적 샘플링 -> GSN


## 7. AE,VAE
### 1) AE
- 입력 데이터의 중요한 특징을 압축하고, 원래 데이터와 최대한 유사하게 복원하도록 학습하는 모델
- 정답 레이블 필요 X

    <img src="https://lilianweng.github.io/posts/2018-08-12-vae/autoencoder-architecture.png" width="450" style="background-color:white; padding:10px;">

- 구조 : 인코더 -> 병목 -> 디코더
    - 인코더 : 입력 데이터(x)를 잠재 벡터(z)로 압축
    - 병목 : 제한된 공간에 정보를 담음 -> 핵심 특징만 학습하게 됨
    - 디코더 : 잠재 벡터(z)로 복원 데이터(x') 생성
- 복원 오차(Reconstruction Error)를 줄이도록 학습
- 압축하는 이유 : 그대로 전달하면 입력을 그냥 복사해버릴 수 있음

### 2) VAE
- AE의 한계 : 잠재 공간이 불규칙해서 새로운 데이터 생성이 어려움
- VAE : 입력 데이터의 특징을 확률분포로 학습하고, 그 분포에서 샘플링해 새로운 데이터를 생성하는 모델

    <img src="https://lilianweng.github.io/posts/2018-08-12-vae/vae-gaussian.png" width="450" style="background-color:white; padding:10px;">

- 학습 목표
    - 복원 오차 최소화
    - KL 발산 최소화 -> 잠재분포가 표준정규분포에 가까워지도록
- AE vs. VAE
    - AE : 하나의 고정된 점으로 표현
    - VAE : 확률분포로 표현 -> 잠재공간이 연속적이고 규칙적
- 작동 과정
    1. 분포 예측 - 인코더가 평균($\mu$)과 표준편차($\sigma$) 예측
    2. 샘플링 및 재파라미터화 - $z = \mu + \sigma \odot \varepsilon$,  $\varepsilon \sim N(0, I)$
    3. 생성 - z를 디코더에 넣어 데이터 복원 또는 생성
- 재파라미터화 : 무작위성을 노이즈 변수($\varepsilon$)로 분리해서 기울기 전달이 가능하게 함


## 8. GAN
### 1) GAN 개념
- 생성자와 판별자 두 신경망이 서로 경쟁하면서 실제와 유사한 데이터를 생성하도록 학습하는 생성 모델
- 생성자(Generator) : 무작위 노이즈로 진짜 같은 가짜 데이터 생성
- 판별자(Discriminator) : 진짜인지 가짜인지 구별
- 위조지폐범(생성자)과 경찰(판별자)의 관계

### 2) GAN 구조

<img src="https://developers.google.com/static/machine-learning/gan/images/gan_diagram.svg" width="450" style="background-color:white; padding:10px;">

- 생성자 : 노이즈(z) -> 가짜 이미지 $G(z)$
- 판별자 : 실제 이미지(x) 또는 $G(z)$ -> 진짜일 확률(0~1)
- 최소-최대 게임(Minimax Game)
    - 판별자 : 구별 능력을 최대화
    - 생성자 : 판별자가 구별하지 못하도록 속임

### 3) GAN 장단점
- 장점
    - 진짜 같은 가짜 생성 가능
    - 높은 이미지 선명도 (VAE보다 선명)
    - 빠른 생성 속도
- 단점
    - 학습의 불안정성 : 두 신경망의 학습 균형이 무너짐
    - 훈련 초기 문제 : 생성자가 충분한 학습 신호를 못 받음
    - 모드 붕괴(Mode Collapse) : 특정한 형태의 데이터만 반복 생성
    - 평가 지표의 한계
    - 역매핑의 어려움

### 4) GAN 적용 사례
- NVIDIA : 가상 인물 이미지 생성 (Progressive GAN)
- Deep Video Portraits : 인물 영상 합성 (Conditional GAN)
- Facebook : Eye In-Painting (Exemplar GAN)


## 9. 확산 모델
### 1) 확산 모델의 개념
- 이미지에 노이즈를 추가한 데이터를 학습하고, 무작위 노이즈를 점차 제거하면서 새로운 이미지를 생성하는 모델
- 모호성 모델링 : 하나의 조건에서 여러 결과가 가능하다는 불확실성을 확률적으로 표현 -> 다양한 결과물 생성

### 2) 확산 모델의 구조: 순확산과 역확산

<img src="https://lilianweng.github.io/posts/2021-07-11-diffusion-models/DDPM.png" width="500" style="background-color:white; padding:10px;">

- 순확산(Forward Diffusion)
    - 원본 이미지($x_0$)에 가우시안 노이즈를 조금씩 추가해서 무작위 노이즈($x_T$)로 만드는 과정
- 역확산(Reverse Diffusion)
    - 무작위 노이즈($x_T$)에서 노이즈를 단계적으로 제거하면서 새로운 이미지($x_0$)를 생성하는 과정
    - Denoising U-Net으로 노이즈를 예측하고 제거

### 3) 확산 모델의 장단점
- 장점
    - 압도적인 샘플 품질
    - 학습의 안정성
    - 다양성 확보
- 단점
    - 느린 생성 속도
    - 높은 계산 비용
    - 구조적 제약
