# 1. 자연어 처리(NLP)
- 자연어란?: 인간이 소통을 위해 자연스럽게 발생시켜 사용하는 언어(한국어, 영어 등)
- NLP의 역할: 자연어를 컴퓨터가 이해할 수 있는 '숫자(데이터)'로 변환하고, 그 안에서 의미를 추출하는 모든 과정
- NLP 과정:
	1. 데이터 수집: 웹 크롤링, API 호출 등
	2. 텍스트 전처리: 모델이 이해할 수 있는 형태로 텍스트 정제하는 작업
	3. 텍스트 표현: 텍스트를 수치로 표현 → 머신러닝/딥러닝 입력 가능하게 변환
	4. 모델 학습 및 예측: 텍스트 분류, 문장 생성, 질의응답 등
	5. 후처리 및 응용: 출력 다듬기, 챗봇, 정확도 측정 등

- **특징**: 
	1. 이산적 데이터
	- 이미지나 소리는 연속적인 신호지만, 언어는 '단어'라는 독립적이고 딱 끊어지는 단위로 구성됨. <br>
	- NLP의 숙제: 컴퓨터(숫자)가 이해할 수 있도록 **이산적인 단어를 연속적인 벡터(Embedding)로 변환**하는 과정이 반드시 필요

	2. 순차적 데이터 (Sequentiality)
	- 단어가 나열되는 순서에 따라 문장의 의미가 완전히 결정됨
	<br>예: "사자가 사람을 먹다" vs "사람이 사자를 먹다"
	- NLP의 숙제: 데이터의 **앞뒤 관계**와 **위치 정보를 유지**하며 처리할 수 있는 모델 구조(RNN, Transformer 등)가 필요함

	3. 중의성과 모호성 (Ambiguity)
	- 특징: 동일한 형태의 단어라도 문맥에 따라 뜻이 다름
	<br>예: "배(Fruit/Ship/Stomach)"가 무엇인지 알려면 주변 단어를 봐야 함.
	- NLP의 숙제: 단어 하나가 아닌 **문맥**을 전체적으로 파악하는 기술이 핵심

	4. 희소성 (Sparsity)
	- 특징: 단어의 조합은 무한에 가깝지만, 우리가 가진 데이터에 나타나는 조합은 극히 일부임 (Zipf의 법칙)
	- NLP의 숙제: 한 번도 본 적 없는 문장이 들어와도 단어 간 유사도를 이용해 확률적으로 추론할 수 있는 **일반화 능력**이 중요함

	5. 가변성과 진화 (Variability & Evolution)
	- 특징: 언어는 고정되어 있지 않음. 신조어, 은어, 오타, 이모지 등이 수시로 등장하며 문법 파괴도 잦다.
	- NLP의 숙제: 고정된 규칙이 아니라 데이터로부터 **유연하게 패턴을 학습**하는 머신러닝 방식이 필수적.

# 2. 언어모델
- 개념: 단어 시퀀스(문장)에 확률을 할당하여, 가장 자연스러운 단어 시퀀스를 찾아내는 모델
- 언어 모델이 문장의 자연스러움을 판단하게 하는 가장 보편적인 방법은 **'이전 단어들'이 주어졌을 때 '다음 단어'를 예측하도록 훈력시키는 것**
- 통계기반 vs 인공 신경망 기반으로 구분

## 2.1 통계적 언어모델(SLM : Statistic Language Model)
- 개념: 단어들의 출현 빈도(확률)를 계산하여, 가장 자연스러운 단어 시퀀스(문장)를 예측하는 모델
- 아이디어: 분포 가설 
	- "비슷한 문맥에서 함께 나타나는 단어들은 비슷한 의미를 가진다"는 가설. <br>즉, <u>어떤 단어의 주변에 자주 등장하는 단어들을 보면 그 단어의 의미를 짐작할 수 있다는 뜻</u>
	- 예시: "I play soccer", "I play basketball", "I play tennis" -> soccer, basketball, tennis 단어 모두 "I play__" 라는 문맥에서 등장 <br>-> 분포 가설에 따라 모델은 이 단어들이 '스포츠'라는 카테고리에 속할것이라고 학습. 
	- 이런 분포가설을 설명해주는 기본 개념은 바로 **'조건부 확률’** 과 ‘**카운트기반 확률’**

- **조건부 확률**: 어떤 단어가 앞 단어가 주어졌을 때 나올 확률
	- ex) “나는 밥을”이라는 문맥이 있을 때 “먹었다”가 나올 확률
	- 앞 단어를 보고, 뒤에 더 높은 확률로 올 수 있는 것을 선택
- **카운트 기반**: 이전 단어 시퀀스의 등장 빈도를 통해 다음 단어의 확률을 계산
	- 한계: 희소 문제
	- 이전에 경험한 적이 있는 조합에 대해서만 잘 예측 할 수 있어서, 한번도 안 접해본 조합을 보면 예측 어려움
	<br>**-> 새로운 문장 조합에도 적용할 수 있는 신경망 모델이 필요**

### 2.1.1 N-gram
- 등장배경: SLM은 희소하게 등장한 것에 대해서는 잘 예측을 못하고, 문장이 길어질수록 다음 단어를 계산하기 어렵다
	<br>-> 전체 문장을 N개로 쪼개서 문장의 일부만 봄 = N-gram
- 예시: "Boy is riding ???" -> N=4 (예측 단어까지 합해서 4임) -> N-1개의 단어를 보고 계산!
- 한계: 희소문제, 문맥제한
	- 문맥제한 예시: trigram 모델이면 직전 2단어만 봄 -> 문장 앞부분의 중요한 정보를 반영하지 못함

### 2.1.2 Perplexity(PPL)
- 개념: 언어 모델이 문장을 얼마나 "혼란스러워하는지(perplexed)"를 수치로 나타낸 것
- 특징: 문장 W가 희소하게 등장할수록 PPL이 커지고, 단어 개수 N이 커질수록 PPL이 낮아진다. 

신경망 기반 언어 모델은 단어와 문맥 간의 관계를 학습하여 일반화하는 능력이 뛰어남. 
<br>덕분에 훈련 데이터에서 직접 보지 못한 문장이 등장하더라도 훨씬 유연하게 대처
<br>-> 신경망 언어 모델 쪽에 관심 쏠림

# 3. 신경망 언어 모델(NNLM)
- 키워드: 임베딩 = **인간의 언어(단어)를 컴퓨터의 언어(숫자)로 번역하는 역할**
- 신경망 모델: 모든 단어를 거대한 다차원 지도 위의 한 점(좌표)로 보냄. 
	- 특징: 서로 밀접한 단어일 수록 가깝게 위치. ex) '축구'와 '농구'
- 통계 모델과의 차이점
	- 통계모델: 훈련 데이터에 "재밌는 축구를 한다"라는 문장이 없으면 확률을 0으로 처리
	- NNLM: 
		1. 학습 데이터에서 "재밌는 농구를 한다"를 학습함.
		2. "축구"와 "농구"가 유사한 벡터임을 인지함.
		3. "재밌는 축구를 한다"라는 문장을 처음 봐도 높은 확률을 부여함.
- NNLM은 단어를 '기호'에서 의미를 가진 숫자(벡터)'로 바꿈으로써, **데이터에 없는 문맥도 어느 정도 예측할 수 있는 일반화 능력**을 얻음.
# 4. 딥러닝 기반 언어모델
- 트랜스포머의 등장으로 신경망을 깊게 쌓기 가능(인코더 디코더가 여러 Layer로 구성) <br> ->**언어 모델의 표현력(capacity)이 높아져 현재의 딥러닝 기반 언어모델을 만들 수 있어짐**
- 트랜스포머의 등장으로 **사전 학습**과 **미세 조정**로 구성된 전이학습은 NLP의 표준 학습 방법론이 됨

## 4.1 BERT(Bidirectional Encoder Representations from Transformers)
- 개념: 
	- Bidirectional : 텍스트의 앞뒤 문맥을 동시에 고려해서
	- Encoder : 입력값을 벡터로 바꿔주는
	- Transformers : 딥러닝 모델이다.

-> 한 단어를 이해할 때 그 단어의 좌우 문맥을 동시에 보고 이해함.
- 특징: 
	- 트랜스포머 구조의 **encoder** 부분만 사용
	- 사전학습+파인튜닝: 큰 데이터로 먼저 일반적인 언어 능력을 배우고, 나중에 특정 과제에 맞게 살짝 조정

- 구조: transformer의 encoder를 여러 개 쌓아 올린 구조 -> 따라서 BERT는 **문장 이해** 까지만 목적으로 함.
- 처리과정: input -> Embedding -> 트랜스포머 인코더 12개(or 24개) -> output

- 사전 학습
	- Masked Language Model(MLM) : "I go to school" 을 "I __ to school"이라고 만들고, 이 빈칸에 어떤 단어가 들어갈지 예측하면서 학습. 마스킹 단어는 Input 텍스트의 15%만 랜덤으로 선택
	<br> -> 자연스럽게 문맥을 이해하고 추론하는 능력 향상
	- Next Sentence Prediction(NSP): 두 문장이 서로 앞뒤 관계인지, 무작위 위치에 있는 지 학습
	<br> -> **두 문장이 이어지는 논리 흐름** 학습

## 4.2 GPT
- **decoder 기반**
- decoder는 미래의 단어를 볼 수 없게 mask로 가림 -> **GPT는 왼쪽 단어들"만" 보고 다음 단어를 예측**
- few-shot learning: 몇 개의 예시를 보고 task를 수행
	- zero-shot : 예시 없이 문제 지시만 주고 task 수행
	- one-shot: 1개의 예시만 보고 task 수행
## 4.3 LLM(Large Language Model)
- 개념: 방대한 양의 데이터를 학습하여 자연어 및 기타 유형의 콘텐츠를 이해하고 생성하여 광범위한 작업을 수행할 수 있는 기초 모델의 범주
- 특징: 
	1. Scaling Law: 모델의 크기, 데이터의 양, 계산량이 일정 임계치를 넘으면 **성능**이 선형적이 아니라 **지수적으로 폭발**
	2. 창발적 능력: "배운 적 없는 일을 하다"
	3. RLHF(Reinforcement Learning from Human Feedback (**인간 피드백 기반 강화학습**))
		- "똑똑한 AI를 '착한 AI'로 길들이기"
		- AI가 내놓은 여러 답변 중 사람이 가장 좋아하는 답변에 '점수'를 주어, 모델이 인간의 가치관과 대화 예절에 맞게 행동하도록 정렬
	4. 패러다임의 전환: "튜닝(Tuning)에서 프롬프트(Prompt)로"
		- 데이터 사이언티스트의 업무 방식이 Fine-tuning에서 Prompt Engineering 중심으로 변함
- 작동 방식
	- 대규모 학습: 수십억 개의 문장으로 이루어진 방대한 데이터 학습. 언어의 패턴을 스스로 터득함.
	- 예측 및 생성: 학습이 완료된 LLM은 사용자의 질문이나 지시(프롬프트)를 받으면, **학습한 패턴을 기반으로 다음에 올 가장 확률 높은 단어를 순서대로 예측**하며 응답을 생성
- decoder 기반
- 한계: hallucination, 업데이트 비용

## 4.4 RAG
- 필요성: LLM은 새롭게 나타난 지식에 대해서는 답변을 할 수 없음
- RAG: 외부에서 최신 정보를 가져와서 답변. 즉, **질문에 대해 모델이 모르면 검색해서 답함**
- 구조
	- 질의 인코더(Query encoder): 사용자의 질문을 벡터로 바꿈
	- 지식 검색기(Knowledge Retriever): 인코딩된 질문을 바탕으로 외부 지식에서 관련 내용을 검색
	- 지식 증강 생성기(Knowledge Augmented Generator): 검색한 내용을 참고해서 답변 생성
- 장점: **최신 정보 반영 가능**, **hallucination 감소** -> **정보 정확성 증가**
- 단점: 검색된 문서 중 일부가 신뢰성이 낮거나 잘못된 정보일 경우 생성된 답변의 품질과 정확도가 크게 하락

- CAG(Credibility-Aware Generation): RAG 한계를 보완하기 위해 제안된 프레임워크
	- 문서의 신뢰도를 스스로 판단하고 활용할 수 있도록 훈련시키는 기법
	- 구조: 질문 입력 → 정보 검색 → **각 정보의 신뢰도 평가** → 높은 신뢰도 정보 우선 사용 → 최종 답변 생성

## 4.5 LangChain
- LLM과 RAG를 직접 구현하는 것이 복잡하니까, 이러한 모델 pipeline을 쉽게 구성할 수 있도록 도와주는 역할
- LLM을 기반으로 앱 제작 시 필요한 구성 요소를 쉽게 연결
- 파이썬과 JavaScript 라이브러리 제공 → 다양하게 활용 가능
<br> -> **LangChain: LLM 앱을 빠르게 조립할 수 있는 레고 블럭**

- 특징
	- 추상화: LLM 서비스를 만들 때 필요한 각종 작업을 간결하게 표현(간소화)
	- 표준화: 비슷한 기능을 갖추고 있는 요소들을 똑같은 형식을 갖춘 컴포넌트로 표준화
		- ex) gpt, claude, gemini 등등 -> 'chat model'이라는 하나의 형식으로 통일
	- 체이닝: 자주 활용하는 주요 컴포넌트를 쉽게 연결해서 LLM 서비스의 로직을 쉽게 파악 (Input → 컴포넌트 → Output)
- 한계: 중간에 조건에 따라 다른 작업을 수행하거나, 결과가 마음에 들지 않을 때 <u>이전 단계로 돌아가 다시 실행하는 등의 복잡한 로직을 구현하기 어려움</u>
<br> -> LangGraph 등장

- LangGraph: AI가 여러 단계를 거쳐 생각하고 행동하게 만드는 흐름도 도구
- LangGraph 구조: (질문받기) ───→ (검색하기) ───→ (답변하기)
	- 노드: 작업하나. ex) def search() : 검색 작업
	- 엣지: 노드와 노드를 연결하는 선으로, **다음 이동 경로** 결정
		- 조건부 엣지: **갈림길**
		<br> ->"만약 A 조건이면 B 노드로, C 조건이면 D 노드로 가라" 와 같은 분기 로직(if-else문)을 쉽게 구현
	- **루프와 사이클**: 사이클을 통해 검색과 검증을 반복하며 완벽한 답을 찾아냄

- LangSmith: 대규모 언어 모델(LLM) 기반 애플리케이션을 위한 통합 개발 및 운영 플랫폼
	- 디버깅 및 추적, LLM어플 성능 평가, 사용자의 사용 과정 모니터링, 프롬프트 허브 제공

- 외부 API 활용하기
	- LLM 내부 지식만을 이용해서 해결할 수 없는 질문에 대해서는 외부 API와 연동해서 답변가능
# 6. sLM(Small Language Model)
- 개념: 자연어 콘텐츠를 처리, 이해 및 생성할 수 있는 AI모델
- sLM은 LLM보다 더 작고 효율적이어서 메모리와 연산 자원을 적게 사용
- **모바일 앱**처럼 리소스가 제한된 환경, 또는 사용자 쿼리에 대한 **오프라인 추론**을 통해 응답을 생성해야 하는 상황에 적합
- 모델 압축
	- 1. 가지치기: 중요도가 더 낮거나 중복되거나 불필요한 매개변수를 제거
	- 2. 양자화: 고정밀 데이터를 저정밀 데이터로 변환
	- 3. 저차원 분해: 큰 가중치 행렬을 더 작은 저차원 행렬로 분해하는 것 -> 복잡한 행렬 연산 단순화
	- 4. 지식 증류: 사전 학습된 "교사 모델"의 학습 내용을 "학생 모델"로 이전하는 것

- sLM 예시: DistilBERT, Gemma, GPT-4o mini, Llama, Ministral, Phi 등
- sLM 장점
	- 성능적 측면: 효과적인 성능, 지연 시간 감소, 보안
	- 비용적 측면: 접근성, 효율성, 에너지 효율, 비용 절감

- 한계
	- 더 큰 모델에 존재하는 편향까지 함께 전달가능
	- 제한된 일반화 & 복잡한 작업에서의 성능 저하
	- 할루시네이션

- LLM과 sLM의 결합(결합추론)
	- 복잡하고 어려운 작업에는 LLM, 가볍고 간단한 작업에는 SLM을 적절히 배분해서 사용

# 7. 생성모델 vs 분류 모델
- 생성모델: **주어진 데이터를 학습하고 학습 데이터의 분포를 따르는 유사 데이터를 생성하는 모델**

- 분류 모델과의 차이

| 구분 | 분류 모델 | 생성 모델 |
|---|---|---|
| 학습 방법 | 데이터의 특징과 정답 레이블의 관계 학습 | 데이터의 특징과 확률분포 학습 |
| 학습 목표 | P(Y\|X) 모델링: 데이터 X가 주어졌을 때 정답 Y에 속할 확률을 학습하여 분류 | P(X) 모델링: 실제 데이터 X가 나타나는 확률분포를 학습하여 새로운 데이터 생성 |
| 주요 특징 | 데이터의 차이점을 학습하여 분류 기준 형성 | 데이터의 공통적인 특징과 다양한 패턴을 학습하며, 입력 데이터의 적합성을 평가하고 비정상적인 입력을 탐지하는 데 활용 가능 |
| 출력 결과 | 범주 또는 범주별 예측 확률 | 새로운 이미지, 텍스트, 음성 등 |
| 한계 | 처음 보는 이상한 데이터도 기존에 학습한 라벨 중 하나로 잘못 분류할 수 있으며, 일반적인 분류 모델은 새로운 데이터 생성 불가능 | 부자연스러운 결과 생성 가능, 생성 결과의 품질 평가 어려움 |

# 8. 생성모델의 분류

- 분류 기준: 데이터의 확률밀도를 직접 계산할 수 있는가?

1. 명시적 확률 밀도 모델 : **데이터의 확률분포를 명시적으로 정의**하여 확률밀도를 계산하거나 근사하는 모델
    - 특정 데이터의 P(X)의 값을 직접적으로 수치화하여 출력할 수 있음
    - **확률밀도를 계산하는 방식**에 따라 다루기 쉬운 밀도 모델, 근사적 밀도 모델로 구분됨
    - 다루기 쉬운 밀도 모델: 데이터의 확률밀도를 수학적으로 직접 계산할 수 있는 모델
        - 확률 분포를 계산 가능한 형태로 구성하여, 특정 데이터가 학습한 분포에서 얼마나 그럴듯한지 정확한 수치로 나타낼 수 있음

    - 근사적 밀도 모델: 데이터의 확률밀도를 직접 계산하기 어려워 근사적인 방법을 활용하는 모델
        - 복잡한 확률분포를 계산할 때 발생하는 어려움을 해결하기 위해 변분 추론이나 마르코프 연쇄 등의 방법을 활용함
        - VAE가 여기 해당함

2. 암시적 확률 밀도 모델 : **확률 밀도를 직접 계산하지 않고,** 학습한 분포에서 새로운 데이터를 생성하는 모델
    - 데이터의 특징과 패턴을 학습함
    - **데이터를 생성하는 방식**에 따라 직접 샘플링, 간접/반복적 샘플링으로 구분됨
    - 직접 샘플링: 학습된 생성 모델에 **무작위 노이즈를 입력**하여 새로운 데이터를 직접 생성하는 방식
        - 생성모델을 한 번 통과하여 결과 생성 가능

    - 간점/반복적 샘플링: **여러 단계의 확률적인 상태 변화를 반복**하여 새로운 데이터를 생성하는 방식
        - 여러 단계의 상태 변화를 거쳐 학습한 데이터의 분포를 따르는 표본을 생성함
        - GAN이 여기 해당함

# 9. AE, VAE

- **AE란?** : 입력 데이터의 중요한 특징을 압축하고, 이를 바탕으로 원래 데이터와 최대한 유사하게 복원하도록 하는 인공 신경망 모델
- 핵심 
    - 입력 데이터를 압축하고 복원하면서 **데이터의 핵심적인 특징을 학습함**

    - 입력 데이터와 복원된 데이터의 차이를 최소화하는 것이 목표

    - **별도의 정답 레이블 없이** 입력 데이터 자체를 복원 목표로 활용할 수 있음

- 작동과정 : 인코더 -> 병목(bottle neck) -> 디코더 구조
    1. 압축단계(Encoder) : 입력 데이터의 **중요한 특징을 추출하고 정보를 압축**
        - 인코더는 입력 데이터(x)를 더 적은 수의 값으로 표현하여 잠재 벡터(z)로 변환함
        - 잠재 벡터(z)는 원본 데이터의 중요한 특징을 압축하여 함축된 정보를 저장하고 있는 벡터

    2. 병목 통과 : 인코더가 추출한 핵심 정보를 **잠재 벡터**로 표현하는 단계
        - 제한된 공간에 정보를 담기 때문에 **데이터의 핵심적인 특징을 학습하도록 유도**
    3. 복원단계(Decoder) : **압축된 잠재 벡터**를 활용하여 **원본 데이터와 유사하게 복원**하고, 원본과 복원 결과를 비교하여 **중요한 정보가 얼마나 잘 복원됐는지 확인**하는 단계
        - 디코더는 잠재 벡터를 입력받아 복원 데이터 (x')를 생성함.

- AE가 굳이 압축했다가 복원하는 이유: 압축과 복원 과정을 통해 데이터를 표현하는데 중요한 특징을 학습하도록 유도한다.

- **VAE란?** : 입력 데이터의 특징을 확률분포로 학습하고, 해당 분포에서 샘플링하여 새로운 데이터를 생성할 수 있는 생성 모델

- VAE 등장배경 : AE의 경우 입력 데이터를 압축하고 복원하는데 집중하여 **잠재 공간이 불규칙하게 형성 될 수 있다는 한계**가 있다.
<br>-> VAE는 데이터를 고정된 값이 아닌 확률분포로 표현하여 잠재공간이 규칙적으로 구성되도록 학습함.

- VAE의 학습 목표
    1. 복원 오차 최소화 : 원본 데이터와 복원 데이터 사이의 차이를 최소화하여 **원본의 특징을 최대한 잘 복원하도록 학습함.**
    2. KL 발산 최소화
        - KL 발산: 두 확률분포가 얼마나 다른지를 수치로 나타내는 지표
        - 인코더가 예측한 잠재분포가 **표준정규분포에 가까워지도록 학습함**<br>잠재공간이 불규칙하게 형성되는 문제를 완화하고, 새로운 데이터를 생성하기 좋은 잠재공간을 형성

- VAE 활용 분야
    - 이미지 증강
    - 데이터 증강
    - 이상탐지
    - 특징 추출

- VAE 작동 구조

    1. 분포 예측 단계(Encoder):  입력 데이터를 특정 점이 아닌, 데이터가 존재할 확률이 높은 '구역'으로 매핑
	2. 샘플링 및 재파라미터화: 예측한 분포에서 실제 디코더에 전달할 잠재 벡터(z)를 하나 뽑기
	3. 생성 단계(Decoder): 샘플링된 잠재 벡터(z)를 단서로 사용하여 새로운 데이터를 생성

	<요약><br>
	-인코더( 𝑞 𝜙): 입력 데이터(𝑥﻿)를 보고 데이터가 있을법한 '범위'( 𝜇 , 𝜎)를 찾습니다. 
	- 샘플링(𝑧): 그 범위 안에서 무작위로 점 하나를 콕 찍습니다. 
	- 디코더( 𝑝 𝜃): 뽑힌 점(𝑧)을 단서로 원래 이미지를 다시 그려냅니다( 𝑥 ′)

- 굳이 잠재 벡터를 확률 분포로 만드는 이유: AE와 달리 VAE는 단순 복원이 아닌 **생성**이 목적이기 때문<br>-> 잠재 공간이 **표준 정규 분포**를 따라서 빈틈이 매워지고 구조화 됨. 그 결과 어떤 지점을 선택하더라도 디코더가 그럴듯한 데이터 생성


# 3. GAN (Generative Adversarial Network)
- 개념: **생성자 신경망**과 **판별자 신경망**이 서로 **적대적으로 경쟁**하면서, **훈련**을 통해 자신의 작업을 **점점 더 정교하게 수행**하는 신경망 모델
- 예시: 생성자 - 지폐 위조범, 판별자 - 경찰
	- 생성자는 **진짜와 구분이 불가능한 완벽한 데이터** 생성
	- 판별자는 진짜와 가짜를 판별하다가 점점 찍어서 맞히는 수준(확률 0.5)에 도달

- 생성자 신경망과 판별자 신경망의 목적
	- 생성자: 진짜 분포에 가까운 가짜분포를 생성하는 것
	- 판별자: 표본이 가짜 분포에 속하는지 진짜 분포에 속하는지를 결정

-> GAN의 궁극적 목적: 실제 데이터의 분포에 가까운 데이터를 생성하는 것이며, <br>판별자가 진짜/가짜를 한 쪽으로 판단하지 못하는 경계를 최적 솔루션으로 간주

- 활용: 이미지 생성, 영상 합성

- GAN 구조: 두 개의 신경망이 최소-최대 게임(MinMax Game)을 수행하는 구조
	- 생성자
		- 입력: 우리가 직접 제어할 수 있는 **무작위 노이즈(잠재 변수 Z)**를 입력받음
		- 출력: 실제 데이터와 유사한 **가짜 데이터(G(Z))**를 생성
		- 목표: 판별자를 속여서 가짜를 진짜(1)로 믿게 만드는 것
	- 판별자
		- 입력: **실제 데이터(X)**와 생성자가 만든 **가짜 데이터(G(Z))**를 모두 입력받음
		- 출력: 입력 데이터가 진짜일 확률(0~1)을 출력
		- 목표: 진짜는 1로, 가짜는 0으로 정확히 분류하는 분류 모델의 역할을 수행

- 장점
	1. 진짜같은 가짜를 생성할 수 있음
	2. 압도적인 선명도: VAE가 흐릿한 이미지를 만드는 것과 달리, GAN은 판별자를 통과하며 **매우 선명하고 깨끗한 샘플** 생성
	3. 빠른 생성 속도: 확산 모델처럼 여러 단계를 거치지 않고, 신경망을 한번만 통과하면 결과물이 바로 나옴
	4. 잠재 공간의 부드러움: 잠재 공간 지도에서 조금씩 이동하면, 이미지가 부드럽게 변하는 보간이 가능 -> 데이터 구조를 잘 이해했다는 증거!


- 한계
	1. 학습의 불안정성: 진동현상, 무한루프
	2. 훈련 초기 문제: 그래디언트 소멸
	3. 모드 붕괴: 생성자가 판별자를 속이기 가장 쉬운 특정 패턴(예: 숫자 1) 하나만 주구장창 만드는 문제
	4. 지표의 부재: "얼마나 잘 학습되고 있는지" 보여주는 객관적인 손실 곡선이 없음
	5. 역매핑 불가능: 잠재 변수 Z에서 데이터 X를 만드는 법은 배우지만, 이미지를 넣었을 때 특징 Z를 추출하는 과정(인코더 역할)이 기본 구조에는 없음 <br> -> 이미지를 넣었을 때 그 이미지의 특징 벡터를 바로 찾아낼 수 없음

# 4. 확산 모델
- 확산 모델이란
	1. 입력 이미지에 <u>노이즈를</u> 조금씩 추가하여 <u>완전한 노이즈</u>로 만든 뒤,
	2. 이를 **다시 거꾸로 복원**하는 과정을 학습하여(노이즈 제거)
	3. **입력 이미지와 유사한 확률 분포**를 가진 **결과 이미지**를 생성하는 모델
- 확산 모델의 흐름: 원본 데이터→ 완전한 노이즈 → 원본 데이터

- 목표: 순확산과 역확산 단계를 거쳐 생성된 '결과 이미지'의 확률 분포를 '원래 데이터'의 확률 분포와 최대한 유사하게 만드는 것
- 모호성 모델링: "용 그려줘"라는 모호한 요청에 대해, 모델은 완전한 노이즈 상태에서 시작해 데이터 구조의 아주 작은 부분을 조금씩 **환각(=상상)**하며 구체적인 용 이미지를 완성해나감.

## 4.1 확산 모델의 구조: 순확산과 역확산
- **순확산: 데이터에 점진적으로 (정규분포를 가진)노이즈를 추가하는 과정**
- 순확산 과정의 목표: 원래 데이터를 점차적으로 무작위 노이즈로 변형시키는 것!
- 단계: 
	1. 깨끗한 원본 데이터 준비
	2. 원본 데이터에 점진적으로 노이즈 추가
	3. 결과: 원본 데이터가 완전한 무작위 노이즈로 변함

- **역확산: 노이즈 데이터에서 원본 데이터를 제구성하는 과정**
- 목표: 순확산 과정에서 추가된 노이즈를 제거하여 원래의 데이터를 복원하는 것
- 단계:
	1. 완전한 노이즈 데이터에서 시작
	2. 각 단계에서 "어떤 노이즈가 추가되었을까?"를 예측하며 노이즈 제거. 순확산 과정의 반대 방향으로 진행됨
	3. 결과: 수십 번의 **반복적인 디노이징** 과정을 거쳐 점진적으로 이미지 복원

- 확산모델 장점
	- 압도적인 샘플 품질 : GAN보다 더 정교하고 다양한 이미지 생성, 고해상도 확장에 유리
	- 학습의 안정성: GAN과 달리 **손실 곡선**을 확인 가능
	- 다양성 확보

- 확산모델의 단점
	- 느린 생성 속도
	- 높은 계산 비용
	- 구조적 제약: 노이즈 분포 Z와 데이터 Z의 형태(Shape)가 동일해야 한다는 제약
		- 쉽게 말해, 고양이 이미지를 생성하려면 노이즈도 이미지와 똑같이 가로 * 세로 크기가 맞아야 함. <br>-> 텍스트처럼 구조가 다른 데이터에 바로 적용하기 어려움
