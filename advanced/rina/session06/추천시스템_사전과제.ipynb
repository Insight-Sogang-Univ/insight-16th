# 1. 추천시스템과 데이터의 유형
추천 시스템이란? : 사용자의 행동 기록과 아이템 정보를 바탕으로, 사용자가 **좋아할 만한 아이템을 예측해 골라 주는 시스템**

## 1.1 사용되는 데이터의 종류와 특징
- 사용자의 선호 정보를 어떻게 얻느냐에 따라 두 가지로 나뉨

- **명시적 데이터 : 고객들에 의해 입력된 데이터.** 사용자가 선호를 **직접 표현**한 데이터
	- 예시: 별점, 좋아요 · 싫어요, 리뷰
	- 장점: 사용자의 취향 정확히 반영
	- 단점
		- 평가를 남기는 사람 자체가 적고, 남기더라도 본 것 중 일부에만 남김
		- 사람의 취향은 변함
		- 사람마다 점수를 주는 기준이 다름

- **암시적 데이터 : 고객들의 행동을 기반으로 수집된 데이터.** 행동에서 선호를 **간접적으로 추측**하는 데이터
	- 예시: 조회수, 클릭, 구매, 시청시간, 검색
	- 장점: 사용자가 따로 입력하지 않아도 대량으로 쌓여 얻기 쉬움
	- 단점: 행동이 곧 선호라고 단정할 수 없음. 행동이 있어도 잘못 클릭했거나 선물로 구매한 것일 수 있음

## 1.2 추천 시스템의 분류
- 무엇을 근거로 추천하느냐에 따라 세 가지로 분류됨
- 내용(콘텐츠) 기반 필터링, 협업 필터링, 하이브리드

- **내용 기반 필터링** : **콘텐츠의 유사도**를 기반으로 사용자에게 추천하는 방법
	- 사용자가 이용하거나 선택한 아이템과 특징이 비슷한 아이템을 추천하는 방식
	- 예시: SF 영화 좋아하니까 SF 영화 추천해야지

- **협업 필터링** : 서비스 내에 있는 **다수의 사용자로부터 얻은 선호도 정보**를 기반으로 사용자에게 추천하는 방법
	- 두 사용자의 취향이 비슷하다면, **한 사람이 좋아한 아이템을 다른 사람도 좋아할 가능성이 높다**고 봄

- 협업 필터링은 무엇을 기준으로 비슷하느냐에 따라 다시 두 가지로 분류됨
1. 사용자 기반 : 특정 사용자와 **유사한 취향의 사용자**가 선호한 아이템 추천
2. 아이템 기반 : 사용자가 선호한 아이템과 **유사한 아이템**을 추천

- 내용기반 필터링과 아이템 기반 협업 필터링의 차이: 
	- 내용 기반 필터링은 **아이템의 속성**을 기준으로, 아이템 기반 협업 필터링은 **"A를 많이 본 사람은 B도 많이 본다"와 같은 사람들의 행동**을 기준으로 비슷하다를 판단함

- **하이브리드 필터링** : 2가지 이상의 **다양한 종류의 추천 시스템 알고리즘을 조합**하여 만들어진 방식
	- **콘텐츠 기반 필터링과 협업 필터링 각각의 단점들을 보완**하기 위해 제시
	- 예시: 오늘 처음 등록된 영화는 평점도 없어서 협업 필터링으로 추천할 수 없음. 이 경우 내용 기반 추천을 쓰고, 평점이 쌓이면 협업 필터링으로 넘어가는 식으로 조합함.

# 2. 내용 기반 필터링
- 핵심흐름: 아이템을 숫자로 표현 → 사용자 프로필 생성 → 유사도 계산 → 점수가 높은 아이템 추천

## 2.1 사용자 프로필
- 개념: 사용자가 어떤 특성을 좋아하는지 정리한 정보
- 사용자의 취향을 추론하는 방식
	- 직접 지정형: 사용자가 **자신의 취향을 직접 알려주는** 방식
	- 간접 지정형: 사용자의 **과거 행동 이력에서 취향을 추론**하는 방식

- 사용자 프로필을 획득하는 방식
	1. 사용자가 가지고 있는 아이템 특성 가중치의 평균 값 활용
	2. 설문, 키워드 입력, 평가 같은 방법으로 사용자의 기호를 따로 파악
	3. 유저의 평가를 바탕으로 분류(Classification) 문제로 해결 -> 새 아이템에 유저가 어떤 평가를 줄지(좋아요/싫어요, 5점 척도) 예측하는 **분류 문제**

## 2.2 유사도
- 개념: 두 대상이 얼마나 비슷한지 나타내는 숫자
- 아이템 정보(텍스트, 이미지, 영상 등)를 **숫자 벡터로 변환**한 뒤 유사도를 계산
- 사용자와 아이템을 **같은 기준(feature)**으로 표현해야 서로 비교가능
- 사용자가 선호하는 아이템과 비교해 **유사도가 높은 아이템을 선별**해 추천

- 유사도 계산 과정
1. 아이템을 숫자로 변환
	- 아이템 정보 활용 - 영화 장르, 감독, 등급 등 아이템의 속성을 feature라 부름
	- 테이블 형태로 데이터 정리 - 행: 영화 제목, 열: 아이템의 feature, 셀: feature 값
	- 숫자로 변환하기(원-핫 인코딩) - 텍스트 데이터를 숫자 벡터로 변환

2. 사용자를 숫자로 변환
	- 사용자의 취향 파악 
	- 사용자도 아이템과 동일한 n개의 feature로 표현 -> 벡터 계산 가능하게

3. 유사도 계산
	- 벡터 내적 (dot product)
		- 유사성 : 내적값이 클수록 두 벡터의 유사성이 높음
		- 0과 1로만 된 벡터라면 내적은 **겹치는 feature의 개수**와 같음

	- 유사도 방식 종류: 자카드 유사도, 코사인 유사도, 유클리디안 거리, 피어슨 상관계수


## 2.3 TF-IDF 
- 배경: 상품설명, 줄거리 같은 텍스트도 아이템의 중요한 특징임. 텍스트를 숫자로 바꾸는 가장 쉬운 방법은 단어가 몇 번 나왔는지 세는 것인데, 이렇게 되면 조사나 어미가 가장 중요한 단어처럼 취급됨

- **TF-IDF** : **특정 문서에는 자주 나오고 다른 문서에는 드문 단어에 높은 값**을 주는 방법
- TF-IDF 값이 클 수록, 그 단어가 다른 문서들보다 해당 문서에서 더 중요한 역할을 한다는 뜼

- 추천에서 사용되는 방식
	1. 영화마다 줄거리를 TF-IDF 벡터로 변환
	2. 사용자가 재밌게 본 영화의 벡터와 다른 영화들의 벡터를 코사인 유사도로 비교
	3. 유사도가 높은 영화를 추천

## 2.4 내용 기반 필터링의 장단점
- 장점
	- **아이템 콜드 스타트 문제 해결** : 속성만 있으면 신규 아이템 추천 가능
	- 내 기록만으로 프로필을 만들기 때문에 다른 사용자의 데이터가 필요 없음
	- 어떤 feature 때문에 추천했는지 보여 줄 수 있어서 추천 이유를 설명하기 용이함

- 단점
	- feature로 담을 수 있는 정보에 한계가 있어, 속성 정보가 부족하거나 품질이 낮으면 정확도가 떨어짐
	- **비슷한 것만 계속 추천**(과도한 특화) -> 새로운 장르 발견 어려움
	- 기록이 없는 신규 사용자는 추천하기 어려움 **(사용자 콜드 스타트)**

# 3. 협업 필터링
## 3.1 메모리 기반 협업 필터링
- 시스템에 저장된 사용자 - 아이템 평점을 그대로 꺼내서 새 아이템의 평점을 예측하는 방법
- 별도의 학습 단계 없이 비슷한 사용자나 아이템, 즉 **"이웃"** 을 찾아 쓰기 때문에 **이웃 기반 협업 필터링**이라고도 부름

- 두 가지 형태의 접근 법: 사용자 기반, 아이템 기반

- **사용자 기반 접근** : 나와 취향이 비슷한 사용자(이웃)의 평점을 활용해 내 평점을 예측하고 추천
	1. 나와 유사도가 가장 높은 사용자 TOP-N명을 이웃으로 선정
	2. 이웃들의 평점을 유사도로 가중평균해 내가 아직 평가하지 않은 아이템의 평점을 예측 (유사도가 높은 이웃일수록 큰 영향)
	3. 예측 평점이 높은 아이템을 추천

	- 예시: 사용자 A: "어바웃 타임" 5점, "노팅힐" 4점, "아이언맨" 3점<br>사용자 B: "어바웃 타임" 5점, "노팅힐" 5점, "반지의 제왕" 2점, "라라랜드" 5점<br>사용자 C: "아이언맨" 5점, "어벤져스" 5점, "어바웃 타임" 2점, "라라랜드" 2점

	- 사용자 A와 사용자 B는 어바웃 타임과 노팅힐에 높은 점수를 준 것으로 보아 비슷한 취향을 가진다고 볼 수 있음<br> 사용자 A는 B가 높게 평가한 "라라랜드"를 아직 보지 않았으므로 A에게 추천

	- 사용자 간 유사도 계산 시 피어슨 상관계수를 많이 사용함
	- 예측 평점: 이웃들이 그 아이템을 자기 평균보다 얼마나 높게(낮게) 줬는지를 유사도만큼 반영해서, 내 평균에 더함

- **아이템 기반 접근** : 내가 평가한 아이템들과 평점 패턴이 비슷한 아이템을 찾아 추천하는 방식
	- 아이템의 속성이 아닌 **사용자들이 그 아이템을 함께 좋아하고 싫어한 패턴**으로 유사도 판단
	- 사용자는 계속 늘어나고 취향도 바뀌지만 아이템끼리의 관계는 상대적으로 덜 변함 → 아이템 유사도를 **미리 계산해 두고 재사용**하기 좋음

	- 예시
	-- **사용자 A**: "어바웃 타임" 5점, "노팅힐" 4점, "아이언맨" 3점
	- **사용자 B**: "어바웃 타임" 5점, "반지의 제왕" 2점, "노팅힐" 4점
	- **사용자 C**: "아이언맨" 5점, "어벤져스" 4점, "어바웃 타임" 2점
	- **사용자 D**: "어바웃 타임" 5점<br>->어바웃 타임을 높게 준 A와 B는 노팅힐도 높게 줬음 → 사람들의 평가가 함께 움직이므로 **어바웃 타임과 노팅힐은 비슷한 아이템**
	- 사용자 C: 어바웃 타임을 낮게 평가했기 때문에 어바웃 타임과 패턴이 유사한 노팅힐도 낮게 평가할 가능성이 높음 → 노팅힐은 추천하지 않음

## 3.2 메모리 기반 협업 필터링의 한계
1. 희소 행렬 문제
	- 주어진 평점, 선호도 데이터가 적을 경우 성능이 저하되는 문제

- 데이터가 희소하면 생기는 문제  
	- 두 사용자가 함께 평가한 아이템이 적어 **유사도를 믿기 어려움**
		- 공통 아이템이 1개뿐이어도 코사인 유사도가 1로 나올 수 있음
	- 비슷한 이웃을 찾지 못하면 **예측 자체가 어려움**

	⇒ **모델 기반 협업 필터링(Model-based CF) 이용**<br>이웃을 직접 비교하는 대신, 관측된 평점으로 사용자와 아이템의 특징을 학습해 빈칸을 채우는 방법

2. 콜드 스타트 문제
	- 데이터가 충분히 쌓이지 않아 적절한 추천을 못하는 문제
	- 협업 필터링은 행동 기록에만 의존하기 때문에, 기록이 없는 대상에게는 추천이 불가능함
	- 크게 **신규 사용자**(평가 기록이 없어 비슷한 이웃을 찾을 수 없음)와 **신규 아이템**(아무도 평가하지 않아 추천 후보에 오르지 못함) 상황에서 발생

	- 해결 방법: 가입할 때 직접 선호를 지정하도록 하기, 새 아이템에는 내용 기반 필터링도 함께 쓰기, 임베딩 쓰는 모델이면 임베딩 평균으로 새 아이템을 임시로 표현하기

3. 확장성 문제
	- 아이템과 사용자가 계속 늘어날수록 계산량이 커지는 문제 
	- 해결방법: **KNN 협업 필터링 이용**
		- 모든 사용자를 쓰지 않고, 유사도가 높은 k명의 이웃만 골라 예측하는 방법

## 3.3 모델 기반 협업 필터링
- 관측된 평점 데이터로 **예측 모델을 학습**하고, 데이터의 패턴을 모델의 파라미터에 담아 관측되지 않은 평점을 예측하는 방법
- 메모리 기반 vs 모델 기반
	- 메모리 기반은 추천이 필요할 때마다 저장된 평점을 직접 비교해 이웃을 찾음
	- 모델 기반은 평점 데이터를 먼저 **압축된 모델로 요약**해두고, 추천은 이 모델만으로 수행

- 두 단계로 동작함
	1. 학습 단계: 이미 존재하는 평점을 가장 잘 설명하는 모델의 파라미터를 찾음 → 추천과 별도로 미리 수행하며, 새 데이터가 쌓이면 주기적으로 다시 학습

	2. 예측 단계: 학습된 파라미터로 비어 있는 칸의 평점을 계산 → 추천이 필요한 시점에 빠르게 수행

- 등장 배경: 메모리 기반 협업 필터링은 평점 사이의 유사도에만 의존하여 평점에 드러나지 않는 취향의 구조를 포착하지 못하고, **희소성과 확장성 문제**에 취약함<br>모델 기반 협업 필터링은 평점 외의 **잠재된 패턴**을 학슴함으로써, 직접 겹치는 평가가 없어도 전체 데이터의 흐름으로 빈칸을 추정하고 추천 시점의 계산을 가볍게 만듦

- 모델 기반의 방법 중 가장 대표적인 것이 **잠재 요인 모델**이며, 이를 구현하는 핵심 기법이 **행렬 분해**이다.

#### 잠재 요인
- 평점에 직접 드러나지는 않지만, 사용자와 아이템 사이의 관계를 결정짓는 **숨은 특성**
- “취향”, ”성향” 과 같이 **실제 데이터에는 없는 개념을 숫자로 표현**

- **잠재 요인 모델** : 사용자와 아이템을 **같은 잠재 공간 위의 벡터**로 나타내고, 두 벡터의 관계로 평점을 설명하는 모델
	- 평점 행렬을 **사용자-잠재 요인** 행렬과 **아이템-잠재 요인** 행렬로 나누어 학습

#### 행렬 분해
- 큰 평점 행렬을 작은 행렬 두 개의 곱으로 쪼개는 방법
- 예측 평점 행렬 R (m x n) = 사용자 잠재 요인 행렬 P (k x m) · 아이템 잠재 요인 행렬 Q (k x n)
	- k는 잠재 벡트 크기
	- m은 user 수
	- n은 item 수



# 4. 딥러닝 기반 추천 시스템
## 4.1. VAE 기반 협업 필터링
- 사용자의 아이템 상호작용 벡터(예: 구매/시청 이력)를 입력으로 받아, 잠재 공간에서 재구성하도록 VAE를 학습하는 방법
- 등장배경: MF의 한계
	1. 선형성의 한계
	2. 불확실성 무시
		- 모델이 자신의 예측을 얼마나 확신하는지 모름
- 구조
	- 입력(x): 사용자의 아이템 상호작용 벡터
	- 인코더: 사용자 이력을 잠재변수 z 분포로 매핑
	- 잠재 벡터(z): 샘플링된 최종 취향 벡터
	- 디코더: z로부터 아이템 분포를 복원
- 장점
	- 확률적 잠재 분포로 사용자 선호를 모델링 하여 **불확실성을 반영함**
	- 희소한 상호작용 데이터에서도 강건함
	- 비선형성 학습

- AE가 아니라 VAE를 선택한 이유
	1. AE의 접근 방식: "A의 취향은 (액션 0.9, 로맨스 0.1) 벡터 하나일 것이다!" (단일 값 추정)
		- 문제점: 데이터가 부족하면 이 단일 추정치가 불안정하고 틀리기 쉬워 **노이즈에 취약**

	2. VAE의 접근 방식: "A의 취향은 (평균 액션 0.8, 표준편차 0.2)를 따르는 분포일 것이다." (분포 추정)	
		- 그냥 "latent 공간 자체를 **정규분포처럼 부드럽게**만들자" 라고 이해하면 됨
		- 강점: 취향의 불확실성(데이터 부족으로 인한 오차 범위)까지 모델링에 포함시켜 안정성을 높임

### 4.1.1 VAE의 핵심: 목적 함수와 KL Divergence
- VAE의 목표: Reconstruction Loss(재구성 오차) 최소화, KL Divergence
- Reconstruction Loss(재구성 오차)
	- 목표: 잠재벡터 z를 이용하여 원래 입력(x)를 최대한 잘 복원하기

- KL Divergence: $$DKL(q(z∣x)∥p(z))$$
	- q(z∣x): 인코더가 만든 분포
	- p(z): 우리가 원하는 이상적인 분포 -> 보통 **표준정규분포** 사용
	- **즉, latent 공간을 전체적으로 평균0, 분산1 근처로 잘 정리해라는 뜻**

- KL Divergence의 역할: latent 공간이 정규분포를 따르면서 공간이 **부드럽고 연결된 구조**가 됨
<br>->그 결과 중간값도 의미를 가지면서 "고양이 같은 강아지 상" 이라는 **희소한 데이터에도 안정적인 예측** 가능


# 5. GNN 기반 사용자 선호 예측
- 등장배경: VAE는 **관계나 순서** 라는 풍부한 정보를 활용하지 못함
- GNN은 사용자와 아이템의 상호작용을 ‘누가(User) 무엇을(Item) 좋아했다’의 집합으로 보는 것을 넘어, <br>이들이 **서로 연결된 거대한 관계망(Graph)**을 형성한다는 점에서 모티브를 얻음

-  즉, 사용자-아이템 상호작용은 본질적으로 그래프(Graph)임
	- 그래프: **노드와, 노드를 잇는 선(엣지)**를 모아 구성한 자료
	- 노드: 현실 세계의 개체
	- 엣지: 개체 간 관계나 상호작용

- 그래프를 사용하는 이유
	1. 관계, 상호작용 같이 다소 추상적인 개념을 다루기에 적합
	2. Non-Euclidean space도 표현하고 학습할 수 있음

## 5.1 GNN의 프로세스
1. 그래프 구조 생성
사용자와 영화 정보가 노드로 생성됨
사용자가 영화를 보고 평점을 준 것은 상호작용이니까 평점은 엣지가 됨

2. 메시지 패싱: 각 노드가 이웃 노드로부터 정보를 받아 전달하고 상호작용 하는 것
- 보통 하나의 GNN 레이어에서 ‘메시지 생성 - 집계 - 갱신’ 의 세 단계로 이루어짐
- **메시지 생성**: 각 노드가 이웃에게 보낼 자신의 정보를 생성. 이때, 이웃에게 보내는 메시지느 임베딩 되어 있음
- **집계**: 메시지 생성 이후, 여러 이웃 노드에서 받은 메시지를 하나의 벡터로 통합하는 단계
- **갱신**: 집계된 이웃정보와 자신의 이전 상태를 결합해 새로운 임베딩을 계산
3. 최종 출력

## 5.2 Light GCN
- GNN은 메시지 계산 방법이나 aggregation 방식에 따라 아키텍처가 결정됨
- 주요 GNN 아키텍처로는 GCN, GRN, GAT, GraphSAGE가 존재
- LightGCN: GCN의 한 종류이며, 크게는 GNN의 한 종류
- 등장배경: 추천 시스템의 핵심은 사용자와 아이템의 상호작용, 즉 관계 그 자체인데 <br>복잡한 연산이 오히려 순수한 ‘관계’ 정보를 왜곡하고 학습을 방해할 수 있다는 아이디어에서 등장
- LightGCN 특징
    - 특징 변환 제거
    - 비선형 활성화 함수 제거: 임베딩 전파 과정에서 이미 비선형성이 충분히 반영됨 + 과적합 방지
- LightGCN은 오직 정규화된(Normalized) 이웃 임베딩의 합(Aggregation)만으로 구성됨 -> 매우 가볍고 효율이 좋음

## 5.3 GNN의 장단점
- 장점: 
    - 고차원 협업 시그널 포착
    - 정형화되지 않은 데이터 학습 가능
    - 콜드 스타트 문제 완화에 효과적
- 단점
    - 과평탄화 문제
    - 과압축 문제
    - 계산 복잡도 증가
    - 해석의 어려움


# 6. Transformer 기반 추천 시스템
- 등장배경
	1. CF의 순차적 한계
	- 순서 무시, 문맥 부족, 예측의 한계

	2. GNN의 순차적 한계
	- 지역적 정보 전파의 한계
	- 장기 의존성 문제

- Transformer: 사용자 행동을 순서 (Sequence)로 접근

- Transformer는 **시퀀스 내의 문맥과 핵심 의도를 파악**하는 데 강력!

- Transformer의 문맥 파악 예시
	- 문맥 A (캠핑 준비):
	- 행동 순서: [클릭: "텐트"] → [클릭: "침낭"] → [클릭: "등산화"]
	- Transformer의 해석: "등산화"가 "텐트", "침낭"과 강하게 연관(Attention)됩니다. 이때 "등산화"는 '캠핑 장비'로 해석됩니다.
	- 다음 추천: "캠핑용 코펠", "백패킹 배낭", "랜턴"

- Transformer의 핵심의도 파악 예시
	- 트랜스포머는 사용자의 행동이 길어질 때, 가장 최근의 행동이 아니라 **멀리 떨어진 과거의 행동**도 참고
	- 행동 순서 (시간 순):
    - **[구매: "레이저 프린터"]**
    - [검색: "가습기"] (계절성 탐색)
    - [클릭: "강아지 사료"] (반려동물 용품)
    - ... (약 20개의 서로 다른 행동 발생) ...
    - **[검색: "토너"]**
	- 사용자가 "토너"를 검색했을 때, 모델은 1~4번의 모든 과거 행동을 동시에 참조함

#### Transformer 기반 주요 추천 시스템 모델

- SASRec : 단방향 어텐션으로 과거 정보만 활용해, 다음 아이템을 예측하도록 학습한 모델
	- 특징
		1. 적응적 집중: 데이터가 희소하면 최근 아이템에, 풍부하면 긴 과거 맥락에 집중하도록 Attention을 조절함
		2. 빠른 학습: RNN처럼 순서대로 계산하지 않고 모든 위치를 병렬로 처리함
	- 문제점
		1. 계산량: Attention 연산량이 시퀀스 길이의 제곱(n²)에 비례해, 긴 행동 이력을 다루기 부담스럽다.
		2. 단방향성: 과거 방향만 참조하므로 시퀀스 앞부분의 아이템은 활용할 정보가 적습니다. 또한 장바구니에 동시에 담은 물건처럼, 실제 행동 순서가 엄격하지 않은 경우도 있음

- BERT4Rec : 양방향 어텐션으로 과거와 미래 정보를 모두 활용해, 가려진 아이템을 예측하도록 학습한 모델
	- 특징
		1. 양방향 문맥: 각 아이템의 앞뒤 문맥을 모두 학습에 사용함
		2. 데이터 활용: 한 시퀀스에서 가리는 위치를 바꿔 가며 여러 학습 샘플을 만들 수 있음
		

# 7. 오늘날의 추천 시스템
## 7.1 LLM의 시대
- 추천을 랭킹 문제가 아닌 **언어 문제**로 푼다

- 과거에는 조합별로 미리 학습된 공식에 따라 점수를 출력했다면, LLM은 추천을 점수가 아닌 **문장**으로 출력
	- ID가 아닌 의미 기반 추천
	- 텍스트, 이미지, 오디오 등 멀티 모달 데이터 활용
	- 추천 이유를 함께 제시하는 **서술형 추천**이 가능해짐

- 새로운 가능성
	- **대화형 추천이 가능해짐**: 사용자가 자연어로 입력한 조건을 반영해 추천함
	- **Zero-shot 추천** : 상호작용 데이터 없이 텍스트 설명만으로 추천 -> **콜드 스타트 문제의 강력한 해결책!**

- LLM을 추천에 적용하는 방식: P5/TALLREC
- P5: 여러 추천 과제를 "텍스트 프롬프트 → 텍스트 답변" 형식으로 통일해, 하나의 언어 모델로 학습한 모델
	- 언어모델을 추천 데이터로 처음부터 학습하므로 시간과 비용이 많이 듦

- TALLREC : 학습 비용을 줄이기 위해, 범용 LLM을 "좋아할까? (Yes/No)" 형식의 추천 지시문으로 튜닝하는 프레임워크

## 7.2 인기 편향과 XAI 문제
- 인기 편향: 소수의 인기 아이템만 추천되고, 비인기 아이템은 거의 추천되지 않는 현상
	- 원인
	1. 모델 불균형: 상호작용이 인기 아이템에 몰려 있음
	2. 모델의 편향 강화: 데이터가 많은 아이템일수록 예측 점수가 높아 더 자주 추천됨
	3. Rich-get-richer: 추천 → 노출 → 클릭 → 학습 데이터 반영이 반복되며 격차가 커짐

- 설명가능성(XAI)
	- 딥러닝 모델의 블랙박스 문제로 인해 모델이 왜 추천했는지 설명하기 어려움

## 7.3 컴퓨팅 자본, Cold-start를 고려한 추천 시스템 도입
- 딥러닝은 전면 도입하기에는 비싸기 때문에 현업에서는 선형 모델로 후보를 먼저 거르고 딥러닝으로 정렬하는 2단계 구조를 사용함

1. 후보군 생성
	- 선형 모델로(내용 기반 필터링, 협업 필터링, 행렬 분해) 수백만 개 아이템을 수백 개 후보로 축소

2. 순위 매기기
	- 딥러닝 모델 (GNN, Transformer, VAE)로 수백 개 후보를 정교하게 정렬해 상위 수십 개를 노출함
