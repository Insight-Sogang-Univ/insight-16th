# 1. 전이학습과 딥러닝 경량화

## 1.1 전이학습
- 딥러닝의 한계: 데이터가 부족하면 다양한 패턴 학습이 어렵고, 과적합이 일어날 가능성이 높음. 또한 모델 학습시 많은 시간과 비용이 소모됨
<br> **그래서 등장한 것이 전이학습**

- **전이 학습이란?** : 모델이 사전 학습을 통해 배운 능력을 활용해 새로운 작업을 학습하는 것
- 예시    
    **<동물 사진을 통해 정상·폐렴 사진 구분하기>**  
    1. 동물 사진으로 모델 사전 학습   
    - 고양이와 강아지 사전으로 사전 학습된 모델은 윤곽, 질감, 밝기와 같은 특징을 파악하는 법을 익힘    
    - 이렇게 특징을 찾는 방법은 **모델의 가중치**에 담기게 됨   
    2. 정상·폐렴 구분에 쓸 부분 가져오기   
    - 이렇게 사전 학습된 모델은, X-ray에서 갈비뼈의 윤곽이나 폐 영역의 밝기 차이와 같은 특징 파악에 활용할 수 있음.    
    - **기존 모델의 특징 파악 능력은 그대로 사용하고**, 기존 모델 분류층을 **정상과 폐렴을 구분하는 새 분류층으로 교체** 해야 함.   
    3. 정답이 있는 X-ray로 학습하기    
    - 이후 **정답 레이블이 존재하는 X-ray 이미지로 새 분류층을 학습** 시키면 됨 
    4. fine-tuning(필요시)   
    - 필요하다면 모델의 **가중치**도 X-ray에 맞게 **조정**할 수 있음

- 용어 정리
    - Source: 기존에 학습한 문제. ex) 고양이 분류 문제
    - Target: 새롭게 해결할 문제. ex) 정상, 폐렴 분류 문제
    - Domain: 사용한 데이터 특성. ex) 일반 동물 사진/x-ray 
    - Task: 수행하려는 작업. ex) 고양이,강아지 분류 / 정상,질병 분류

## 1.2 전이학습 분류

1. Domain과 Task에 따른 분류

    | 종류 | Domain | Task | 정답 레이블 | 예시 |
    |---|---|---|---|---|
    | 귀납적<br>(Inductive) | Source = Target<br>or<br>Source ≠ Target | Source ≠ Target | Source : O or X<br>Target : O | 이미지   분류<br>→ 객체 탐지 |
    | 변환적<br>(Transductive) | Source ≠ Target | Source = Target | Source : O<br>Target : X | 뉴스 감정 분석<br>→ 영화 리뷰 감정  분석 |
    | 비지도<br>(Unsupervised) | Source = Target<br>or<br>Source ≠ Target | Source, Target 둘 다 비지도 | Source : X<br>Target :    X | 한 분야의 군집화 지식<br>→ 다른 분야의 군집화에 활용 |

- 귀납적 전이학습으로 예시 설명
    - 귀납적 전이학습은 기존에 학습했던 모델의 능력을 바탕으로 정답 레이블이 존재하는 사진 여러 장을 학습하면서 새로운 이미지를 보더라도 분류할 수 있도록 하는 것.

    <br>이를 전이학습의 4가지 용어로 풀이하면, 
    - Source와 Target의 Domain은 같거나 다르며, Source와 Target의 Task는 달라야 함.
    - Target의 정답 레이블은 존재해야만 하며, Source의 정답 레이블은 있어도 없어도 됨.
	- Target 정답 레이블이 필요한 이유: 새롭게 해결할 문제 '분류'는 Source에 없는 것이기 때문에 알 수가 없기 때문

- Source와 Target의 Domain은 같은데 Task가 다른 경우
	- Source의 Task: 사진 속 고양이의 존재 여부 파악
	- Target의 Task: 사진 속 고양이의 위치 분류

- Source와 Target의 Domain도 다르고 Task도 다른 경우
	- Source의 Task: 고양이와 강아지 분류하기
	- Target의 Task: 정상과 폐렴 분류하기

2. 전이 대상에 따른 분류
- 전이 대상:  새 작업을 배울 때 기존 모델에서 가져오는 대상

    | 방식 | 전이 대상 |
    |---|---|
    | 인스턴스 기반 | 기존 데이터 일부 |
    | 특성 기반 | 데이터를 나타내는 특징 표현 |
    | 파라미터 기반 | 기존 모델의 학습된 가중치, 편향 |
    | 관계 기반 | 데이터나 개체 사이의 관계 패턴 |

## 1.3 전이 학습의 현재 패러다임
- 전통적인 전이학습: 이미지 분야에서는 기존 모델의 가중치를 재사용하고 분류층만 새 작업에 맞게 교체하고, 언어분야에서는 미리 만든 단어 벡터를 재사용하는 방식을 사용함. 

- Transformer 이후의 전이학습: 
	- 이미지 분야: 자기지도 학습, 이미지·텍스트 공동 학습
		- 자기지도 학습: 이미지 일부를 가리고 복원하는 과정에서 모델이 이미지의 특징을 파악하는 법을 학습
		- 이미지·텍스트 공동 학습: 사진과 그 사진을 설명하는 문장을 함께 보여주며, 이미지와 문장이 얼마나 잘 맞는 지 판단하는 법을 학습<br>이러한 능력은 전이 학습을 거쳐 텍스트로 이미지 검색, 처음 보는 범주 분류에 활용될 수 있음

	- 언어분야: 대규모 말뭉치로 모델을 사전 학습해 문맥을 반영하는 표현을 얻음
		- 예를 들어 BERT로 가려진 단어를 예측하고, GPT 계열 모델로 문맥을 활용하는 법을 배워 문맥을 반영하는 표현을 얻음
		- 사전 학습된 언어 모델을 새 작업에 맞추는 방식은 원하는 출력 형태에 따라 달라짐.


## 1.4 Foundation Model
- **여러 작업에 활용할 수 있도록**, 방대한 데이터로 사전 학습한 기반 모델

## 1.5 Fine Tuning & PEFT
- 정의: 사전 학습된 모델을 **새 작업의 데이터로 추가 학습**하고, **기존 모델의 가중치 일부 또는 전체를 조정**

- 예시: 고양이와 강아지를 분류하도록 학습된 이미지 모델을 정상과 폐렴 구분하는 모델에 쓰기
	- 먼저, 기존 모델의 마지막 분류층을 정상과 폐렴을 구분하는 새로운 분류층으로 교체해야함
	- 이후 정답이 표시된 X-ray 이미지를 모델에 입력하면 밑의 과정이 진행됨
	1. 특징 추출
	2. 예측
	3. 오차계산
	4. 가중치 업데이트

- 중요한 점: **모델의 모든 가중치를 반드시 업데이트할 필요는 없음**

**Fine-tuning 방법**
- Full Fine-tuning: 사전 학습된 모델의 **모든 가중치를 업데이트**하는 방법
	- 장점: 기존 특징 추출층과 새로운 분류층을 모두 학습함 -> 이미지의 특징을 추출하는 방식부터 최종 분류 기준까지 전체적으로 조정할 수 있음
	- 단점: 메모리 사용량과 연산 비용이 큼. 새로운 데이터가 부족하면 과적합이 발생하거나, 기존에 학습한 지식을 망각할 수 있음

- Partial Fine-tuning: 사전학습된 모델의 **일부 층을 고정하고 나머지 층의 가중치만 업데이트**
	- 각 층이 학습하는 특징의 성격이 다르기 때문에 일반적으로 앞쪽 층을 고정하고 뒤쪽 층을 학습하는 방식을 고려
		- 앞쪽 층: 다양한 이미지에 공통적으로 활용할 수 있는 비교적 단순한 특징을 추출
	 	- 뒤쪽 층: 앞쪽에서 추출한 특징을 조합하여 특정 작업에 필요한 복잡한 특징을 학습

	- 장점: 업데이트 하는 가중치가 적어 학습 부담이 줄어듦
	- 단점: 어떤 층을 고정하고 어떤 층을 학습할 건지 결정해야 함

- Frozen Backbone : **새 분류층만 학습**
	- 예시: X-ray를 넣는다면 기존 모델이 뽑는 특징은 그대로 사용하고, 그 특징 중 어떤 것에 더 큰 비중을 둬야 정상과 폐렴이 구분될 지를 새 분류층이 배우게 됨

	- 장점: 세 방식 중 학습 부담이 가장 작음
	- 단점: 기존 모델이 새 작업에 필요한 특징을 잘 뽑아내지 못한다면 분류층만 바꿔서는 새로운 도메인에 적응하거나 성능을 높이는 데에는 한계가 존재

**Foundation Model은 유지하되, 작업마다 필요한 부분만 <u>최소한의 파라미터로 학습</u>시키는 방법은 없을까?**

**PEFT(Parameter-Efficient Fine-Tuning)**

- 정의: 사전 학습 모델의 가중치 대부분 또는 전부를 고정하고, 새 작업에 필요한 최소한의 파라미터만 학습하는 방법

- 앞서 3가지 fine-tuning에서는 용도에 따른 모델을 각각 저장해야 하여 메모리가 많이 필요함
- PEFT의 경우 학습 후 언어 모델을 한번만 저장하고, 작업 별로 학습된 파라미터 묶음만 따로 저장 -> 메모리 부담이 낮아짐

- PEFT 구현 방법 : **LoRA (Low-Rank Adaptation) - 기존 모델에 더할 수정분을 작은 두 행렬로 학습하기**

	**기존**
	- 모델 내부의 한 층의 가중치를 *W* 라고 하고, 흉부 X-ray 분류에 맞게 더할 가중치 변화량을 *ΔW* 라고 하자<br>따라서 새 작업에서 사용할 	가중치는 *W + ΔW* 가 됨. 
	- 출력 숫자들은 입력 숫자와 가중치의 곱이기 때문에, 입력 특징 벡터가 x라면 기존 층의 출력은 *Wx* 임. 
	- 여기에 수정분 *ΔW*이 만드는 출력 보정값 *ΔWx*를 더하면 새 출력값을 구할 수 있음
	- 이때 *W*가 1000*1000 일때 *ΔW*도 같은 크기이므로, 이 층의 가중치 전체를 업데이트 하려면 100만개의 값을 학습해야 함
	
	**LoRA**
	- LoRA의 Low-Rank는 수정행렬 *ΔW*를 작은 행렬 A와 B의 곱으로 표현할 수 있다는 뜻
	- 먼저 A가 입력 특징값 1000개를 조합해 r개의 중간 값을 만듦 (A: 1000*r)
	- 학습 과정에서 A의 가중치가 조정되면서, 입력 특징을 어떤 비율로 섞어서 보정헤 사용할 지를 배움
	- 이후 B는 중간 값 r개를 조합하여, 기존 출력에 더할 보정값 1000개를 만듦 (B: r*1000)
	- 입력 x에 A를 먼저 적용한 후 B를 적용하므로 B(Ax)가 보정값이 되고, B(Ax)=(BA)x이므로 가중치 수정 행렬은 ΔW=BA가 됨
	- LoRA의 Adaptation은 이 보정으로 모델을 새 작업에 맞춘다는 뜻. 기존 가중치 *W*는 고정하고, 출력에 더할 보정값 A,B를 학습
	
	
		$\
		W_{\text{new}} = W + \Delta W = W + BA$

## 1.6 전이학습의 한계
1. Negative Transfer : 전이학습 적용 전보다 **적용 후에 새 작업에서의 성능이 낮은 경우**
	- 기존 모델에서 학습한 특징이 새 작업에서는 도움되지 않는 사례
	- MMD: Negative Transfer의 판단기준. Source와 Target 데이터의 분포가 얼마나 다른지 살펴볼 때 사용

2. Catastrophic Forgetting : 새 데이터를 학습하면서 **기존 능력이 약해지는** 경우
	- 새로운 모델에 맞게 가중치가 업데이트 되어 기존 모델에서 유용했던 특징을 전만큼 잘 뽑지 못하는 사례
	- EWC: Catastrophic Forgetting을 줄이는 방법 중 하나. 
		- 기존 작업에서 중요했던 가중치를 파악한 뒤, 새 작업을 학습할 때 중요한 가중치가 크게 변화할 수록 더 큰 패널티를 주는 방식으로 기존의 가중치 보존

3. Bias Transfer : 사전 학습에서 얻은 **편향이 새 작업에도 이어지는** 경우
	- 사전학습에서 배운 특징 및 패턴들에는 데이터에 들어있는 다양한 편향이 포함될 수 있음
	- 모델이 사람의 의도와 가치에 맞게 행동할 수 있게 하려면? -> **AI Alignment로 이어짐**

4. Transferability 측정의 어려움 : **사전 학습 모델이 배운 지식**이 **새 작업에 얼마나 잘 활용**될 수 있는 지 측정하기 어려움
	- 어떤 지식을 갖고 있는 모델이 새 작업에 잘 맞을 지 미리 알기 어렵다는 문제
	- LEEP · LogME : Target 데이터와 정답을 사용하여 Fine-tuning 이전에 사전학습 모델의 Transferability 를 추정하는 방법

## 1.7 전이학습 사용 사례
- CV - 의료 영상 | X-ray에서 질환 분류 
	- 일반 이미지로 사전학습 → 라벨이 있는 흉부 X-ray로 Fine-tuning → 정상·폐렴 분류

- NLP - 법률 문서 | 계약서 조항 분류 
	- 대규모 텍스트로 사전학습 → 유형 라벨이 붙은 계약서와 조항으로 fine-tuning → 계약서 조항 유형 분류

## 1.8 딥러닝 모델 경량화
- 학습을 마친 모델이 실제로 답을 내놓으려면, 규모가 큰 Foundation Model을 메모리에 올리고 입력마다 계산을 수행해야 함
<br> 메모리와 연산 비용을 줄이는 방법은 없을까?

- 경량화 4가지 방법: **가지치기, 양자화, 지식 증류, 모델 구조 경량화**
- 가지치기: 모델에서 예측에 미치는 영향이 작다고 판단한 부분을 제거하는 방법
	- 비구조적 가지치기: 개별 가중치의 중요도를 평가한 후, 중요도가 낮은 가중치를 0으로 만듦
	- 구조적 가지치기: 영향이 작은 계산 묶음을 제거

- 양자화: 모델이 쓰는 숫자를 덜 정밀하게 표현하여 저장 공간을 줄이는 방법

- 지식 증류: 큰 모델을 Teacher 모델로 두고, 작은 Student 모델이 정답과 Teacher 모델의 출력 분포를 함께 학습하는 방법
	- 학습을 마친 뒤에 실제 서비스에선  Student만 실행하여 메모리와 계산량을 줄일 수 있지만 Teacher의 능력을 모두 구현하는 데에는 한계가 존재

- 모델 구조 경량화: 모델을 설계할 때부터 같은 작업을 더 적은 계산으로 수행할 수 있는 구조를 선택하는 방법

# 2. AI Alignment
- 등장 배경: 모델이 사실과 다른 문장을 만들거나 사전 학습 데이터에서 생성된 편향이 드러나는 등의 사용자의 의도와 다르게 흘러갈 수 잇음
	**따라서 해당 모델이 사람의의도와 목표를 이해하고, 사람이 중요하게 여기는 가치에 맞게 행동하도록 만들어야함.**

- AI Alignment 정의: 어떤 답변이 **사람에게 바람직한 지 정렬하는 기준**을 정한 후, 모델이 **그 기준에 맞게끔 답하도록 조정**하는 과정

- 정렬 기준: **Helpful, Honest, Harmless**
	- Helpful: 사용자의 의도를 파악하고 실질적으로 도움이 되는 답변을 제공
	- Honest: 사실에 근거해 답하며 불확실한 정보에 불확실성을 드러냄
	- Harmless: 답변이나 행동이 사람에게 해를 끼칠 가능성을 줄임

- AI Alignment의 두 가지 관점
	- **Outer Alignment : 목표를 제대로 설정했는가?**
	- 사람이 정한 학습 목표나 보상 기준이 실제로 원하는 결과를 잘 반영하는 지 살펴봄
	- 보상 해킹: 보상 기준의 허점을 이용해 점수만 높이는 현상

	- **Inner Alignment: 모델이 실제로 무엇을 배웠는가?**
	- Deceptive Alignment(기만적 정렬): 모델이 다른 목표를 추구하면서, 평가를 통과하기 위해 훈련이나 감독 중에만 사람의 의도에 맞게 행동할 수 있다는 가설적 가능성
		- 이 경우 겉으로 드러난 행동만 보면 모델이 어떤 이유로 그런 행동을 했는지 알기 어려움
		- 따라서 모델이 **어떤 정보를 사용하고, 어떤 내부 계산을 거쳐 결정을 내리는 지 이해하는 Interpertability(해석 가능성) 연구가 중요함**

- 정렬을 어렵게 만드는 요인
	- 가치 정렬의 모호성: 사람마다 가치관이 달라서 모두가 동의하는 기준을 만들기 어려움
	- 확장 가능한 감독: 사람이 직접 확인할 수 있는 범위에는 한계가 존재

## 2.1 RLHF
- RLHF란?: 사람들의 피드백을 학습에 사용하여, 사람들이 선호하는 답변에 높은 점수를 주는 방식

- InsturctGPT식 고전적 RLHF 파이프라인
	1. SFT(Supervised Fine-Tuning) : 답변을 생성하도록 언어 모델 학습
	- 사람이 작성한 모범 답변을 언어 모델에 보여주고 학습하는 단계

	2. Reward Model : 사람의 선호를 점수로 변환하는 모델 학습
	- 사람이 선호한 답변에 더 높은 점수를 주는 보상 모델을 학습하는 단계

	3. RL (Reinforcement Learning) : 학습된 보상 최대화
	- SFT를 마친 언어 모델이 생성한 답변을 학습된 Reward Model이 채점하고, 높은 보상을 받는 답변을 더 자주 생성하도록 조정하는 단계
	- InsturctGPT의 PPO : 보상을 높이는 방향으로 모델을 업데이트 하되, 답변을 생성하는 방식이 급격히 바뀌는 것을 억제
	- KL 패널티: SFT를 마친 직후의 기준 모델과 RL 중인 모델을 비교하여, 처음의 SFT 모델에서 얼마나 멀어지는 지를 적용

- 부작용 : Reward Hacking(보상해킹) 과 Sycophancy(아첨)
	- 보상해킹: 좋은 답변보단 높은 보상 점수 받도록 학습함
	- 아첨: 모델이 사실이나 근거보다는 사용자 생각에 무조건적으로 동의

# 3. 프롬프트 엔지니어링
- 개념: 파라미터를 건드리지 않고, AI가 더 좋은 답을 내도록 **입력(프롬프트)를 설계하는 기술**
- 프롬프트 vs Fine-tuning, 언제 무엇을 쓸까?

    | 구분 | 프롬프트 엔지니어링 | Fine-tuning |
    |---|---|---|
    | 비용 | 낮음 (API 호출 비용 정도) | 높음 (데이터 준비, GPU 학습) |
    | 속도 | 바로 시도, 바로 수정 | 학습 시간 필요 |
    | 성능 한계 | 모델이 원래 가진 능력 안에서만 | 모델 자체를 바꿔 특정 도메인에 맞춤 |
    | 사용 상황 | API만 쓸 수 있을 때, 빠른 실험,<br>요구사항이 자주 바뀔 때 | 도메인 특화 데이터가 충분하고,<br>일관된 말투나 형식이 계속 필요할 때 |

## 3.1 프롬프트 엔지니얼이 주요 기법
1. Zero-shot / Few-shot
	- Zero-shot: **예시 없이** 지시만 주는 방식
		- 예시: 영화 리뷰의 감정을 '긍정', '부정', '중립' 중 하나로만 답해줘.
		<br>리뷰: 영상미는 훌륭했지만 결말이 아쉬웠다
	- Few-shot: **입력과 출력의 예시를 몇 개 보여준 뒤** 같은 형식으로 답하게 하는 방식
		- 예시: 영화 리뷰의 감정을 '긍정', '부정', '중립' 중 하나로만 답해줘.
		<br>리뷰: 배우들 연기가 살아있다 → 긍정
		<br>리뷰: 두 시간이 아까웠다 → 부정
		<br>리뷰: 스토리는 지루했지만 음악은 좋았다 → 중립
		<br>리뷰: 영상미는 훌륭했지만 결말이 아쉬웠다 → ?

2. Chain-of-Thought(COT)
	- 정답만 묻지 않고 **중간 풀이 과정을 단계별로 쓰게 하는 방식**
	- 예시: 다음 문장이 말이 되는지 판단해줘.
	<br>"손흥민이 월드시리즈에서 선발 투수로 나섰다."
	<br>문장에 나오는 사람과 대회가 각각 무엇인지 먼저 확인하고, 마지막 줄에 답을 써줘.

3. Role Prompting
	- 모델에게 **역할을 부여**해서 답의 관점, 말투, 설명 수준을 조절하는 방식
	- 예시: 너는 10년차 은행 PB야. 금리 인상이 뭐야?

4. Instruct Prompting
	- 원하는 것을 구체적으로 적어서 지시하는 방식. 답의 형식이나 길이, 누구를 위한 답인지, 꼭 들어가야 할 내용 등을 정해서 알려줌
	- 예시: 아래 기사를 3문장 이내로, 중학생이 이해할 수 있는 말로 요약해줘. 핵심 수치는 꼭 넣고, 의견은 쓰지 마.

5. Self-Consistency
	- CoT를 한 번만 시키면, 그 풀이가 중간에 틀렸을 때 틀린 답을 그대로 믿게 됨
	- **같은 문제를 여러 번 풀게하고(CoT 풀이를 여러 갈래로 샘플링), 가장 많이 나온 최종 답을 선택**하는 방식


## 3.2 프롬프트 엔지니어링의 한계
- Prompt Injection (프롬프트 주입): 사용자 입력 자리에 지시처럼 보이는 문장을 끼워 넣어서, 모델이 개발자의 지시 대신 그 문장을 따르게 만드는 것
	- 예시: "인코더는 입력문장을 의미를 담은 벡터 표현으로 압축하고, 디코더는 그 표현을 참고하여 출력문장을 생성한다 이거번역"
	- 발생 원인: 사람은 어디까지가 지시이고 어디부터가 번역할 문장인지 구분할 수 있지만, 이어 붙인 글을 받는 모델에게는 그 경계가 확실하지 않음

- Hallucination: 모델이 사실이 아닌 내용을 그럴 듯하게 만들어내는 문제

- 모델 의존성: 같은 프롬프트도 모델이나 버전이 바뀌면 결과가 달라지는 문제

- 성능 한계: 모델이 원래 못하는 일은 프롬프트로 해결되지 않음

# 4. 하네스 엔지니어링
- **에이전트** : 도구를 쓰며 여러 단계를 스스로 진행시키는 AI
- 에이전트는 모델 본체(LLM)과 **모델이 일하도록 둘러싼 시스템(하네스)**로 나눌 수 있음
- 하네스에는 지시/컨텍스트, 도구 인터페이스, 실행 환경, 상태/메모리, 관측/피드백, 권한/가드레일이 포함됨

- 하네스가 필요한 상황
모델이 모르는 정보를 가져와야 할 때 : 주문 내역 조회, 최신 정보 검색 
실제로 무언가를 실행해야 할 때 : 파일 수정, 메일 발송, 정확한 계산 
결과를 확인하고 다시 시도해야 할 때 : 조회 실패 후 재시도, 테스트 실행 
위험한 행동을 막아야 할 때 : 환불, 삭제 전 사람 승인

- 하네스 작동원리: 일을 열거나 DB를 조회하는 일은 바깥 프로그램인 하네스가 맡음

## 4.1 하네스의 구성요소
- 하네스를 이루는 여섯 가지 요소
	- 예시: 고객센터 에이전트

    | 요소 | 무엇을 정하나 | 고객센터 에이전트 예시 |
    |---|---|---|
    | 지시/컨텍스트 | 모델이 무엇을 보고 시작하나 | 시스템 프롬프트, 고객 정보, 이전 대화 (3장의 프롬프트가 여기 해당) |
    | 도구 인터페이스 | 모델이 무엇을, 어떤 형식으로 할 수 있나 | 주문 조회 도구 `search_orders`의 정의 |
    | 실행 환경 | 행동이 실제로 어디서 실행되나 | 주문 DB에 접근하는 서버, 격리된 실행 공간(샌드박스) |
    | 상태/메모리 | 지금까지 한 일을 어떻게 기억하나 | 이미 조회한 결과, 진행 단계 |
    | 관측/피드백 | 행동의 결과를 어떻게 돌려받나 | 도구의 반환값, 오류 메시지 |
    | 권한/가드레일 | 무엇은 하면 안 되나 | 조회만 허용, 환불은 사람 승인 후 |

- 도구 스키마: 모델에게 "주문 조회" 기능을 주려면 **도구 사용 설명서**가 필요함. <br>모델은 도구의 코드를 직접 보지 않고, 이 설명서만 읽고 언제 쓸지, 무엇을 넣을지 정함

- 성공과 실패의 반환 계약: 도구가 결과를 어떤 형태로 돌려줄지 미리 정해두는 것. 성공일 시 ok:true 실패일 시 ok:false

## 4.2 에이전트의 실패는 어디서 오는걸까?
- 인터페이스만 바꿔도 결과가 달라질 수 있음
- 증상을 통해 수정할 부분을 파악할 수 있음

# 5. 루프 엔지니어링
- 루프 엔지니어링: 모델이 행동을 제안하고, 시스템이 행동을 실행하고, 그 결과를 다시 모델에게 돌려주는 과정을 **종료 조건까지 반복**하도록 설계하는 것

- 루프 한 바퀴의 구성요소

    1. 관측 : 지금 상황을 봄 → 고객 문의와 지금까지의 기록
    2. 판단 : 다음 행동을 정함 → ‘주문 조회가 필요하다’
    3. 행동 : 도구를 실행함 → search_orders 호출
    4. 검증 : 결과가 목표에 맞는지 확인함 → 조회 성공? 실패?

이때 검증결과는 **다음 바퀴의 관측**이 됨

- 하네스는 모델이 일하는 공간과 조건을 설계한다면, 루프는 그 요소들이 시간에 따라 어떤 순서로 반복되는지를 설계함
- 프롬프트를 한 번 쓰는 데서 멈추지 않고, **결과를 보고 다음 행동을 정하는 시스템을 설계함**

- ReAct = Reasoning + Acting
	- 등장배경: 이전에는 생각만 하는 방식과 행동만 하는 방식이 따로 연구됨. <br>그 결과 생각만 하면 모르는 내용을 지어낼 수 있고(Hallucination), 행동만 하면 결과를 보고 다음에 무엇을 할지 따지지 않아 같은 행동을 반복하거나 답을 내지 못할 수 있음
	- ReAct는 **Thought(추론), Action(행동), Observation(관측)**을 통해 이런 문제를 완화함

## 5.1 루프의 구성요소
- 예시: **오래된 API를 쓰는 코드를 새 API로 전부 바꾸는 마이그레이션 상황**
    | 구성 요소 | 무엇을 정하나 | 예시 |
    |---|---|---|
    | 목표 | 무엇을 끝내려 하나 | 레거시 API를 쓰는 파일을 전부 새 API로 |
    | 트리거 | 루프가 언제 시작되나 | 사람이 요청할 때, 매일 정해진 시간, 코드가 올라올 때 |
    | 관측과 행동 | 매 반복에서 무엇을 보고 무엇을 하나 | 파일 검색, 코드 수정, 테스트 실행 |
    | 정상 종료 조건 | 성공해서 멈추는 조건 | 레거시 API 사용 0건 & 전체 테스트 통과 |
    | 안전 상한 | 위험해서 멈추는 조건 | 반복 200회, 60분, 20달러 |
    | 실패/승인 처리 | 막혔을 때, 위험할 때 어떻게 하나 | 한 번 재시도 후 멈추고 보고, 삭제/배포는 사람 승인 |

## 5.2 통제
- 무한반복, 토큰/비용 폭주같은 실패는 루프 안에서 반복되고 자동으로 실행되면서 영향이 커짐

- 통제 장치
    || 통제 장치 | 막는 실패 |
    ||---|---|
    || 반복, 시간, 비용 상한 | 무한 반복, 비용 폭주 |
    || 테스트와 완료 조건 | 끝나지 않았는데 끝났다고 하는 것 |
    || 중복 실행 방지 | 같은 부작용의 반복 |
    || 체크포인트와 롤백 | 중간에 잘못된 변경이 쌓이는 것 (중간 저장 시점을 남겨 두고, 문제가 생기면 그때로 되돌림) |
    || 위험도에 따른 행동 분리 | 되돌리기 어려운 행동의 자동 실행 |

- 마지막 장치인 위험도에 따른 행동 분리는 행동을 **자동 실행/ 사후 검토/ 사람 승인 후 실행** 세 단계로 나눔
