# 언어모델 & 생성모델

## 1. NLP와 언어 모델  
  
### (1) NLP (Natural Language Processing)  
  
#### 자연어 처리란?  
  
**자연언 vs 프로그래밍 언어**  
- 자연어: 인간이 의사소통을 위해 사용하는 언어 (일대다 대응)
- 프로그래밍 언어: 컴퓨터에 명령을 전달하기 위해 설계된 언어 (일대일 대응)  
  
**자연어 특징**  
- 이산적 데이터: 단어, 문자 등 구분된 기호로 구성, 컴퓨터가 직접 수학적으로 처리하기 어려움
    - *숫자 또는 연속적인 벡터로 변환해야 함 (=Embedding)*
- 순차적 데이터: 단어 등장 순서에 따라 의미가 달라짐
    - *앞뒤 문맥을 고려해 처리할 수 있는 모델 구조 필요*
- 중의성과 모호성: 하나의 단어가 여러 의미로 해석될 수 있음
    - *문맥 전체를 파악하여 의미 판단해야 함*
- 희소성: 단어의 종류와 조합이 다양해 모든 표현을 학습하기 어려움
    - *단어 간 유사도를 활용해 학습하지 않은 표현도 확률적 추론 필요*
- 가변성과 진화: 시대적 환경에 따라 지속적으로 변화
    - *고정된 규칙이 아니라 데이터로부터 유연하기 패턴을 학습하는 머신러닝 방식*
  
**NLP의 개념**  
> 인간이 일상생활에서 사용하는 자연어를 컴퓨터가 분석하고 이해하며 생성할 수 있도록 하는 인공지능 기술  
  
**NLP의 역할**  
  
  
  
1. 형태소 분석: 문장을 형태소로 나누고, 각 형태소의 품사와 문법적 특징 분석
2. 구문 분석: 문장의 문법적 구조를 분석하여 단어들이 서로 어떤 관계로 연결되어 있는지 파악
3. 의미 분석: 단어와 문장의 의미를 분석하여 문맥에 맞는 의미 파악
4. 담화 분석: 여러 문장 사이의 연결 관계와 전체 대화의 맥락을 분석하여 의미 파악  
  
  
### (2) 언어 모델  
  
#### 언어 모델의 정의  
  
> 인간이 말하고 쓰는 방식을 모방하는 시스템  
  
- 앞에 등장한 단어와 문맥을 바탕으로 다음에 나올 가능성이 높은 단어를 뱉어냄
- 학습 방식에 따라 통계적 언어 모델(SLM)과 신경망 언어 모델(NLM)로 구분

## 2. 통계적 언어 모델 (SLM)  
  
### (1) Statistical Language Model  
  
> SLM은 분포 가설 아이디어에 근거  
  
- 분포 가설: 비슷한 문맥에서 등장하는 단어들은 유사한 의미를 가질 가능성 높음
  

**조건부 확률**  
- 특정 사건이 발생했다는 조건 하에 다른 사건 발생할 확률 
  
**카운트기반 확률**  
- 특정 문맥 다음에 어떤 단어가 자주 등장했는지 계산
- 등장 빈도 높은 단어일수록 높은 확률 부여
- 문제: 희소 문제. 특정 표현이 학습 데이터에 거의 등장하지 않거나 전혀 등장하지 않는 문제
  
  
### (2) N-gram  
  
**배경**  
- 문장이 길어질수록 한계 발생
    - 데이터 희소성: 단어 조합이 학습 데이터에 등장하지 않는 경우 많음
    - 계산의 복잡성: 이전의 모든 단어를 고려할 경우 확률을 계산, 저장하는 데 많은 자원 필요
  
**정의**  
- 연속된 N개의 단어로 구성된 묶음으로, 직전 N-1개의 단어를 바탕으로 다음 단어의 등장 확률을 예측하는 통계적 언어 모델
    - 이전 모든 단어 고려하지 않고 최근 N-1개의 단어만 활용
  
**한계**  
- 희소 문제: N값이 커질수록 가능한 단어 조합 수가 증가하여 훈련 데이터에서 특정 단어 조합 찾기 어려워짐
- 문맥 제한: 멀리 떨어진 단어 사이의 관계를 파악하기 어려움
  
  
### (3) Perplexity  
  
**정의**  
- 언어 모델이 주어진 문장을 예측할 때 얼마나 혼란스러워하는지를 수치로 나타낸 지표
    - PPL 낮으면 불확실성 낮음 -> 단어 잘 예측
    - PPL 높으면 불확실성 높음 -> 단어 예측 어려움
  
**특징**  
- 문장 W의 빈도수 - PPL 관계
    - 문장 W가 희소하게 등장할수록 PPL 높아짐
- 단어 개수 N - PPL 관계
    - 문장에 예측하기 쉬운 단어가 추가되면 PPL 감소

## 3. 신경망 언어 모델 (NLM)  
  
**등장**  
- 기존 SLM 한계 완화하기 위해 인공 신경망 활용하여 단어의 의미, 문맥을 학습하는 신경망 언어 모델 등장
- 단어를 연속적인 숫자 벡터로 변환하여 단어 간 의미적 관계 학습
- 신경망을 통해 다양한 단어 조합 특징 학습, 관측되지 않은 표현에도 일반화할 수 있또록 설계
  
**임베딩**  
- 단어를 연속적인 숫자 벡터로 변환해 단어 간 의미적 관계를 표현하는 기술
  
  
  
- 의미나 문맥이 유사한 단어들은 벡터 공간에서 서로 가까운 위치에 놓임
  
**특징**  
- 단어의 의미적 관계 학습
- 희소 문제 완화, 일반화 능력 향상
- 복잡한 문맥 학습, 성능 향상

## 4. 딥러닝 기반 언어모델  
  
### (1) Transformer의 등장  
  
**RNN과 LSTM의 한계**  
- 장기 의존성 문제: 문장이 길어질수록 앞부분의 정보를 유지하기 어려움
- 기울기 소실 문제: 역전파 과정에서 기울기가 작아져 앞부분의 정보가 충분히 학습되지 않을 수 있음
- 순차 처리의 비효율성: 순차적으로 처리하므로 병렬 연산이 제한되고 학습 속도가 느림
  
**Transformer의 등장**  
- 2017년 Google 연구진이 논문 Attention Is All You Need를 통해 제안
- Self-Attention으로 단어 간 관계와 장거리 문맥을 효과적으로 학습
- 병렬 처리와 위치 인코딩을 활용하여 학습 효율과 단어 순서 정보 개선
  
→ RNN과 LSTM의 한계를 완화하고 BERT, GPT 등 현대 언어 모델의 기반이 됨  
  
**Transformer의 구조**  
임베딩 및 위치 인코딩  
- 입력 문장을 토큰으로 나누고 각 토큰을 벡터로 변환
- 위치 인코딩을 추가하여 단어의 순서 정보 반영

인코더  
- Self-Attention: 문장 내 단어 간 관련성을 계산하여 문맥 파악
- Feed Forward Network: Attention 정보를 변환하여 풍부한 특징 학습
- 여러 인코더 층을 쌓아 문장의 의미와 문맥을 정교하게 학습
  
디코더  
- Masked Self-Attention: 미래 단어를 보지 않고 이전에 생성된 단어만 참고
- Cross-Attention: 인코더의 정보를 참고하여 입력과 출력 문장 간 관계 학습
- Feed Forward Network: Attention 결과를 변환하여 다음 단어 예측
- 여러 디코더 층을 거쳐 단어를 순차적으로 생성

### (2) BERT

**BERT란**  
- Transformer의 인코더 구조를 기반으로 양방향 문맥을 학습하는 언어 모델
- Bidirectional: 단어의 앞뒤 문맥을 모두 고려
- Encoder Representations: 문맥이 반영된 단어의 벡터 표현 생성
- from Transformers: Transformer 인코더 구조 기반
  
**특징**  
  
양방향(Bidirectional)  
- 단어의 의미를 파악할 때 앞뒤에 등장하는 단어를 모두 고려
- 문장 전체의 문맥을 활용하여 단어의 의미를 정확하게 파악
  
Transformer 기반  
- Transformer의 인코더 구조만 사용
- Self-Attention으로 문장 내 단어 간 관계 분석
  
사전학습 + 파인튜닝  
- 사전학습: 대량의 텍스트로 언어의 일반적인 구조와 문맥 학습
- 파인튜닝 : 특정 작업의 데이터로 추가 학습하여 목적에 맞게 활용

**학습 방법**  
- 여러 Transformer 인코더 층을 쌓은 구조
- 토큰화 → 임베딩 → Self-Attention을 통한 양방향 문맥 학습 → 출력
  
입력 문장 토큰화  
- 문장을 단어 또는 서브워드 단위의 토큰으로 분리
- 문장 시작과 끝에 `[CLS]`, `[SEP]` 등 특수 토큰 추가
  
임베딩  
- 토큰을 벡터로 변환
- 토큰의 의미, 위치, 문장 구분 정보를 반영

Transformer Encoder  
- Self-Attention으로 각 토큰과 다른 토큰 간 관계 분석
- 여러 인코더 층을 거쳐 문맥이 반영된 벡터 생성
  
출력 및 활용  
- 문장 분류, 감성 분석, 질의응답 등 언어 이해 작업 수행
- 사전학습 후 특정 작업에 맞게 파인튜닝
  
#### 사전학습  
- 대량의 텍스트로 언어의 일반적인 구조와 문맥을 미리 학습하는 과정
- 별도로 정답이 표시된 데이터 없이 텍스트 자체를 활용
- MLM과 NSP 두 가지 학습 방법 사용

**Masked Language Model(MLM)**  
  
> 입력 문장의 일부 토큰을 가린 뒤 주변 문맥으로 원래 토큰을 예측하는 빈칸 채우기 방식  
  
학습 과정  
1. 입력 문장에서 일부 토큰을 무작위로 선택
2. 선택된 토큰 대부분을 `[MASK]`로 대체
3. 앞뒤 문맥을 함께 분석
4. 가려진 원래 토큰을 예측하고 정답과 비교하여 학습
  
학습 효과  
- 양방향 문맥 이해 능력 향상
- 동일한 단어도 문맥에 따라 의미가 달라짐을 학습
- 문장 분류, 개체명 인식, 질의응답 등의 기반 형성

**Next Sentence Prediction(NSP)**  
  
> 두 번째 문장이 첫 번째 문장 바로 다음에 등장하는 문장인지 예측하는 방법  
  
학습 과정  
1. 두 문장 A와 B 입력
2. B가 A 다음에 실제로 등장하는 문장인지 판단
3. 실제 다음 문장이면 `IsNext`, 무작위 문장이면 `NotNext`로 분류
4. 예측 결과와 정답을 비교하여 학습
  
학습 효과  
- 두 문장 사이의 연결 관계와 문맥적 연관성 학습
- 문장 쌍 분류, 자연어 추론, 질의응답 등에 활용 가능한 표현 학습

- ALBERT
  
### (3) GPT

**GPT란?**  
  
> Transformer의 디코더 기반 구조를 활용하여 이전 단어를 바탕으로 다음 단어를 예측하고 텍스트를 생성하는 언어 모델  
  
초기 GPT의 구조  
- Decoder 기반: 인코더 없이 디코더 구조만 활용하며 Cross-Attention은 사용하지 않음
- Masked Self-Attention: 미래 토큰을 참고하지 않고 이전 토큰만 활용하여 다음 토큰 예측
- Feed Forward Network: Attention 결과를 변환하여 언어적 특징 학습
  
→ 이전 토큰을 바탕으로 다음 토큰을 순차적으로 예측하는 자기회귀(Autoregressive) 방식으로 텍스트 생성
  
최근 GPT  
> GPT-3는 별도의 파인튜닝 없이 프롬프트의 지시와 예시만으로 다양한 작업을 수행하는 In-Context Learning 능력을 보여줌  
  
- Zero-shot: 예시 없이 지시만 제공
- One-shot: 하나의 예시 제공
- Few-shot: 소수의 여러 예시 제공
  
  
### (4) LLM

**Large Language Model(LLM)**  
- 방대한 텍스트 데이터와 대규모 신경망으로 인간의 언어를 이해하고 생성하는 AI 모델
- 대량의 데이터에서 언어 패턴과 문맥적 관계를 학습하여 다양한 NLP 작업 수행
- 대부분 Transformer 기반이며 텍스트 생성, 번역, 요약, 질의응답 등에 활용

**작동 방식**  
  
학습  
- 대량의 데이터를 통해 단어 간 관계, 문맥, 문법 구조, 세상의 지식 등 언어 패턴 학습
- 사전학습: 다음 토큰 예측을 반복하며 언어 구조와 문맥 학습
- 파인튜닝: 특정 작업이나 목적에 맞는 데이터로 추가 학습
- 예측값과 정답의 차이를 계산하고 이를 줄이도록 모델 가중치 업데이트  
  
예측  
1. 입력 처리: 질문이나 문장을 토큰으로 분리하고 벡터로 변환
2. 문맥 분석: Attention으로 입력 토큰 간 관계 분석
3. 다음 토큰 예측: 이전 문맥을 바탕으로 다음 토큰의 확률 계산
4. 결과 생성: 다음 토큰을 순서대로 예측하여 응답 생성
  
  
**한계**  
  
환각  
- 실제 사실과 다르거나 존재하지 않는 정보를 그럴듯하게 생성
- 다음 토큰의 확률을 기반으로 답변하므로 사실 여부가 항상 보장되지는 않음
- RAG, 외부 정보 검색, 검증 등을 통해 완화할 수 있지만 완전히 해결되지는 않음

높은 학습 및 업데이트 비용  
- 대규모 사전학습에 막대한 연산 자원, 시간, 비용 필요
- 새로운 정보를 반영하기 위한 추가 학습에도 비용 발생
- 학습 이후의 최신 정보가 자동으로 반영되지 않음 → 외부 지식 활용 기술 필요
  
### (5) RAG  
  
**RAG란?**  
- 외부 지식 검색과 LLM의 텍스트 생성 능력을 결합하여 답변의 정확성과 최신성을 높이는 기술
- 기존 LLM은 학습 당시의 지식을 기반으로 하므로 이후 새롭게 등장한 정보 반영에 한계
  
→ 외부 정보를 검색하여 답변 생성에 활용하는 RAG 등장  
  
**구조**  
  
  
  
질의 인코더  
- 사용자의 질문을 임베딩 모델로 숫자 벡터로 변환
- 질문의 의미를 벡터로 표현하여 관련 문서 검색
  
지식 검색기  
- 벡터 DB에서 질문 벡터와 유사도가 높은 문서 검색
- 검색된 문서를 LLM에 전달
  
지식 증강 생성기  
- 사용자 질문과 검색된 외부 문서를 결합하여 LLM에 입력
- 검색 문서를 참고하여 최종 답변 생성
  
**장점**  
  
최신 정보 반영  
- 모델 재학습 없이 외부 데이터베이스나 최신 문서 활용 가능
- 새로운 정보가 업데이트되면 검색하여 답변에 반영
  
정보 정확성 향상  
- 모델 내부 지식뿐 아니라 검색된 외부 자료를 근거로 답변 생성
- 검증 가능한 정보를 활용하여 환각 현상 완화 및 신뢰성 향상

**CAG**  
  
- 모델이 정보의 신뢰도를 스스로 판단하고 활용하도록 훈련하는 기법
- 기존 RAG의 한계: 검색된 문서가 질문과 관련성이 높아도 오래되거나 부정확할 수 있음
- CAG: 정보의 관련성뿐 아니라 출처의 신뢰성, 최신성 등을 고려하여 답변 생성
  
→ 신뢰도가 낮은 정보의 영향을 줄이고 신뢰도가 높은 정보를 우선 활용  
  
CAG의 구조  
  
1. 질문 입력: 사용자의 질문 분석
2. 정보 검색: 외부 데이터베이스에서 관련 문서 검색
3. 신뢰도 평가: 출처, 최신성, 관련성 등을 종합적으로 평가
4. 정보 선별: 신뢰도가 높은 정보에 더 큰 비중 부여
5. 최종 답변 생성: 신뢰성 높은 정보를 중심으로 답변 생성
  
### (6) LangChain  
  
**LangChain**  
  
> LLM을 외부 데이터, API, 다양한 도구와 연결하여 복잡한 AI 애플리케이션을 쉽게 개발하도록 지원하는 오픈소스 프레임워크  
  
→ LLM과 다양한 기능을 모듈화하고 연결하여 개발 효율성을 높이기 위해 등장  
  
**특징**  
  
추상화  
- LLM 호출, 프롬프트 관리, 문서 검색 등의 복잡한 기능을 간단한 인터페이스로 제공
  
표준화  
- 서로 다른 LLM, 데이터베이스, 외부 도구 등을 일관된 방식으로 사용하도록 지원
- 모델 연결 부분을 변경하여 다른 LLM으로 교체 가능
  
체이닝  
- 여러 작업을 순서대로 연결하여 하나의 작업 흐름으로 구성
  
**LangGraph**  
  
> 여러 작업을 노드와 엣지로 연결하여 조건부 분기와 반복이 가능한 AI 워크플로를 구현하는 프레임워크  
  
  
  
LangChain의 한계  
- 단순 체이닝은 정해진 순서대로 실행하는 선형적 구조에 적합
- 복잡한 시스템에서는 결과에 따라 실행 경로가 달라지거나 이전 단계로 돌아가야 하는 상황 발생
  
→ 조건부 분기, 반복 실행, 상태 관리가 가능한 LangGraph 활용  
  
LangGraph의 핵심 구조  
- 노드: LLM, Agent, Tool 등 특정 작업을 수행하는 단위
- 엣지: 노드 사이의 연결 관계로 다음 작업과 이동 경로 결정
- 상태: 노드 간 공유되는 정보로 이전 작업 결과와 대화 문맥 저장
  
**LangSmith**  
- LLM 기반 애플리케이션의 실행 과정 추적, 성능 평가, 운영 상태 모니터링을 지원하는 개발 및 관리 플랫폼
- LangChain, LangGraph로 구현한 AI 애플리케이션의 동작을 분석하고 개선하는 데 활용
  
핵심 기능  
- 디버깅 및 추적: 입력, 출력, 실행 과정을 추적하여 오류 원인 파악
- 테스트 및 평가: 테스트 데이터로 모델의 정확성과 성능 평가
- 모니터링: 응답 시간, 오류 발생률, 사용량, 비용 등을 모니터링
- 프롬프트 허브: 프롬프트 저장, 공유, 버전 관리
  
**외부 API 활용법**  
  
> LLM 내부 지식만으로 해결하기 어려운 질문은 외부 API와 연동하여 답변 가능  

## 5. sLM (Small Language Model)  
  
### (1) sLM이란?

- 소규모 언어 모델(sLM): 자연어를 처리하고 이해하며 생성하는 AI 모델로, LLM보다 규모와 범위가 작은 모델
- 일반적으로 수백만\~수십억 개의 매개변수 사용
- 매개변수(Parameter): 모델이 학습을 통해 얻는 내부 값으로, 모델의 동작과 수행 방식에 영향
- LLM보다 메모리와 연산 자원을 적게 사용하여 효율적
- 모바일 앱 등 컴퓨팅 자원이 제한된 환경에 적합
- 기기 내부에서 실행하면 인터넷 연결 없이 응답하는 오프라인 추론(Offline Inference) 가능

#### LLM과 sLM의 차이

- LLM
  - 모델 크기가 크고 복잡한 추론 및 다양한 작업에 강점
  - 연산 비용이 높고 고성능 서버에서 주로 활용
- sLM
  - 모델 크기가 작고 특정 분야나 단순한 작업에 효율적
  - 연산 비용이 낮고 응답 속도가 빠름
  - 서버, 스마트폰, 노트북 등 다양한 환경에서 활용 가능

### (2) 모델 압축 방법

- 모델 압축: 중요한 성능은 최대한 유지하면서 모델 크기와 연산량을 줄이는 기술
- 메모리 사용량과 계산 비용을 줄여 제한된 환경에서도 효율적으로 실행 가능

#### 1. 가지치기

> 신경망에서 중요도가 낮은 연결이나 뉴런을 제거하여 모델 크기와 연산량을 줄이는 기술

작동 원리

1. 학습된 신경망에서 결과에 미치는 영향이 작은 연결이나 뉴런 탐색
2. 중요도가 낮은 부분 제거
3. 필요하면 추가 학습으로 성능 저하 보완

- 비구조적 가지치기: 연결만 제거
- 구조적 가지치기: 뉴런이나 블록 등을 제거
- 효과: 모델 크기와 연산량 감소
- 중요한 부분까지 제거하면 성능이 저하될 수 있음

#### 2. 양자화

> 가중치 등의 숫자 표현 정밀도를 낮춰 메모리 사용량과 연산 비용을 줄이는 기술

작동 원리

1. 기존 가중치를 FP32(32비트)와 같은 높은 정밀도로 저장
2. FP16(16비트), INT8(8비트) 등 낮은 정밀도로 변환
3. 저장에 필요한 비트 수를 줄여 모델 경량화

- FP32: 높은 정밀도로 가중치 표현
- INT8: 가중치를 8비트 정수로 근사하여 표현
- 효과: 모델 크기와 메모리 사용량 감소, 연산 속도 개선
- 정밀도가 낮아지면 일부 성능이 저하될 수 있음

#### 3. 저차원 분해

> 큰 가중치 행렬을 작은 행렬들의 곱으로 근사하여 파라미터 수와 연산량을 줄이는 기술

작동 원리

1. 신경망의 큰 가중치 행렬을 작은 두 행렬로 분해
2. 전체 정보를 그대로 저장하는 대신 중요한 특징을 적은 파라미터로 표현

- Original Weight Matrix W: 기존 모델의 가중치 행렬
- Matrix A, B: 적은 파라미터를 가진 작은 행렬
- ΔW = A × B: 두 행렬의 곱으로 가중치 변경량 표현
- Final Weight: 기존 가중치에 학습된 변경량 반영
- 효과: 저장 공간과 연산량 감소

LoRA

- 저차원 분해 원리를 활용한 효율적인 미세조정 방법
- 기존 모델 전체를 압축하는 것이 아니라, 학습할 가중치의 변경량만 작은 행렬로 표현하여 학습 비용 감소

#### 4. 지식 증류(Knowledge Distillation)

> 큰 모델(Teacher Model)의 지식을 작은 모델(Student Model)에 전달하여 적은 파라미터로도 좋은 성능을 내도록 학습하는 기술

작동 원리

1. 학습이 완료된 큰 모델을 교사 모델로 사용
2. 교사 모델의 예측 결과나 학습된 지식을 작은 모델에 전달
3. 학생 모델이 교사 모델의 출력을 따라 하도록 학습

- Teacher Model: 많은 파라미터를 가진 큰 모델로 지식 제공
- Knowledge Transfer: 교사 모델의 지식을 학생 모델에 전달
- Student Model: 전달받은 지식을 학습하는 작은 모델
- 효과: 작은 모델의 성능을 높이고 적은 자원으로 효율적인 추론 가능

### (3) sLM의 예시

- DistilBERT: BERT에 지식 증류를 적용하여 크기를 줄이고 처리 속도를 높인 모델
  - 문장 분류, 텍스트 분석 등에 활용
- Gemma: Google에서 개발한 경량 언어 모델 계열
  - 적은 자원으로 다양한 언어 처리 작업 수행
- GPT-4o mini: OpenAI에서 개발한 소형 모델
  - 빠른 응답 속도와 비용 효율성에 초점
- Llama: Meta에서 개발한 언어 모델 계열
  - 소형 버전은 모바일 등 제한된 환경에서 활용 가능
- Ministral: Mistral AI의 경량 모델 계열
  - 스마트폰 등 온디바이스(On-device) 환경에 적합
- Phi: Microsoft에서 개발한 소형 언어 모델 계열
  - 작은 모델 크기에서도 높은 언어 처리 성능을 목표로 함

### (4) sLM의 장점

> LLM보다 모델 크기가 작아 빠른 처리 속도와 높은 비용 효율성 확보

**성능적 측면**

- 빠른 응답 속도: 연산량이 적어 빠르게 결과 생성
- 제한된 환경에서도 실행 가능: 적은 메모리와 컴퓨팅 자원으로 스마트폰, 노트북 등에서 실행
- 보안 및 데이터 보호: 개인 기기나 기업 내부 서버에서 직접 실행하여 외부 클라우드로 데이터를 전송하지 않고 활용 가능

**비용적 측면**

- 낮은 운영 비용: 연산량과 메모리 사용량이 적어 서버 운영 및 서비스 비용 감소
- 적은 전력 소비: 대규모 모델보다 에너지 사용량이 적음
- 하드웨어 비용 절감: 고성능 GPU 등 고가 장비에 대한 의존도 감소

### (5) sLM의 한계

> 모델 크기가 작고 학습할 수 있는 정보에 제약이 있음

**편향**

- 학습 데이터의 사회적 편견이나 잘못된 정보가 답변에 반영될 수 있음
- 데이터 다양성이 부족하면 특정 집단이나 관점에 치우친 결과 생성

**제한된 일반화 및 복잡한 작업에서의 성능 저하**

- 학습하지 않은 새로운 상황에서 대응 능력이 부족할 수 있음
- 복잡한 추론이나 전문 지식이 필요한 작업에서 LLM보다 정확도가 낮을 수 있음

**환각**

- 근거가 부족한 질문에 대해 사실과 다르거나 존재하지 않는 정보를 생성할 수 있음
- LLM과 마찬가지로 환각 현상 발생 가능

### (6) LLM과의 결합 – 결합 추론

- 결합 추론: sLM과 LLM을 함께 활용하여 각 모델의 장점을 살리는 방식
- sLM: 빠르고 비용이 적게 들지만 복잡한 작업에서 성능이 떨어질 수 있음
- LLM: 복잡한 문제 해결에 유리하지만 연산 자원과 비용이 많이 필요

**작동 방식**

1. 간단한 작업: 문장 분류, 간단한 요약 등은 sLM이 처리
2. 복잡한 작업: 여러 단계의 추론이나 전문적인 분석은 LLM이 처리
3. 결과: 모든 작업에 LLM을 사용하지 않아 비용과 처리 시간을 줄이면서 필요한 성능 확보

→ sLM의 효율성 + LLM의 높은 문제 해결 능력을 결합하여 성능과 비용 사이의 균형을 맞추는 방식

## 6. 생성 모델의 개념

### (1) 분류 모델 vs 생성 모델

#### 생성 모델이란?

> 주어진 학습 데이터를 학습하고, 학습 데이터의 분포를 따르는 유사 데이터를 생성하는 모델

#### 분류 모델과 생성 모델의 차이

**분류 모델**

- 데이터의 특징과 정답 레이블 사이의 관계 학습
- P(Y|X) 모델링: 데이터 X가 주어졌을 때 정답 Y에 속할 확률 학습
- 데이터의 차이점을 학습하여 분류 기준 형성
- 출력: 범주 또는 범주별 예측 확률
- 처음 보는 이상한 데이터도 기존에 학습한 레이블 중 하나로 잘못 분류할 수 있음
- 일반적인 분류 모델은 새로운 데이터 생성 불가능

**생성 모델**

- 데이터의 특징과 확률분포 학습
- 데이터의 공통적인 특징과 다양한 패턴 학습
- 입력 데이터의 적합성 평가 및 비정상적인 입력 탐지에도 활용 가능
- 출력: 새로운 이미지, 텍스트, 음성 등
- 부자연스러운 결과가 생성될 수 있으며 생성 결과의 품질 평가가 어려움

### (2) 생성 모델의 분류

> 생성 모델은 '데이터의 확률밀도를 직접 계산할 수 있는가?'에 따라 구분

#### 명시적 확률밀도 모델

> 데이터의 확률분포를 명시적으로 정의하고, 특정 데이터의 P(X) 값을 직접 계산하거나 근사하는 생성 모델

확률밀도(Probability Density)

- 학습한 데이터 분포에서 특정 데이터가 얼마나 그럴듯한지 나타내는 값
- 학습 데이터와 유사한 데이터 → 확률밀도 높음
- 학습 데이터와 다른 데이터 → 확률밀도 낮음

분류

> 확률밀도를 계산하는 방식에 따라 다루기 쉬운 밀도 모델과 근사적 밀도 모델로 구분

**다루기 쉬운 밀도 모델**

- 데이터의 확률밀도를 수학적으로 직접 계산할 수 있는 모델
- 확률분포를 계산 가능한 형태로 구성하여 특정 데이터의 확률밀도를 정확한 수치로 표현

대표 모델

- NADE, MADE, PixelRNN/PixelCNN
  - 데이터의 각 요소에 대한 조건부 확률을 순차적으로 계산
- Change of Variables Model (nonlinear ICA)
  - 변수 변환을 활용하여 복잡한 데이터의 확률밀도 계산

**근사적 밀도 모델**

- 데이터의 확률밀도를 직접 계산하기 어려워 근사적인 방법을 활용하는 모델
- 변분 추론이나 마르코프 연쇄 등을 활용

대표 모델

- VAE(Variational Autoencoder)
  - 변분 추론을 활용하여 계산하기 어려운 확률분포를 근사적으로 학습
- Boltzmann Machine
  - 마르코프 연쇄 등을 활용하여 데이터의 상태를 반복적으로 변화시키고 표본을 추출하여 확률분포 학습

변분 추론

- 직접 계산하기 어려운 복잡한 확률분포를 계산하기 쉬운 확률분포로 근사하는 방법
- 정규분포처럼 다루기 쉬운 분포를 활용하여 원래 분포와 최대한 비슷해지도록 학습

→ 복잡한 확률분포를 계산하기 쉬운 분포로 근사하는 방법

마르코프 연쇄

- 다음 상태가 현재 상태에 의해서만 확률적으로 결정되는 과정
- 현재 상태를 바탕으로 다음 상태로 이동하는 과정을 반복

→ 현재 상태를 기준으로 다음 상태가 확률적으로 결정됨

#### 암시적 확률밀도 모델

- 데이터의 확률분포 P(X)를 직접 수치로 계산하지 않고, 학습한 데이터 분포를 바탕으로 새로운 데이터를 생성하는 데 초점을 둔 모델
- 데이터의 특징과 패턴을 학습하여 학습 데이터와 유사한 새로운 데이터 생성
- 특정 데이터의 확률밀도를 직접 계산하기 어려움

분류

> 데이터를 생성하는 방식에 따라 직접 샘플링과 간접·반복적 샘플링으로 구분

직접 샘플링

- 학습된 생성 모델에 무작위 노이즈를 입력하여 새로운 데이터를 직접 생성하는 방식
- 여러 단계의 반복적인 상태 변화 없이 모델을 한 번 통과하여 결과 생성

대표 모델: GAN(Generative Adversarial Network)

- 생성자(Generator)와 판별자(Discriminator)의 경쟁을 통해 데이터의 특징 학습
- 무작위 노이즈로부터 새로운 데이터 생성

작동 방식

> 무작위 노이즈 입력 → GAN 생성자 → 새로운 이미지 생성

→ 무작위 입력값을 생성 모델에 전달하여 새로운 데이터를 직접 생성하는 방식

간접/반복적 샘플링

- 여러 단계의 확률적인 상태 변화를 반복하여 새로운 데이터 생성
- 반복적인 상태 변화를 거쳐 학습한 데이터 분포를 따르는 표본 생성

대표 모델: GSN(Generative Stochastic Network)

> 마르코프 연쇄의 확률적 상태 전이 과정을 학습하고 반복 적용하여 새로운 데이터 생성

작동 원리

- 현재 상태를 바탕으로 다음 상태가 결정되는 확률적 규칙 학습
- 학습한 상태 변화 과정을 반복하여 새로운 데이터 생성
- 예: 손글씨 숫자 이미지 생성

생성 과정

> 초기 무작위 상태 → 확률적인 상태 변화 → 반복적인 변화 → 새로운 이미지 생성

→ 학습된 확률적 상태 전이 규칙을 반복 적용하여 새로운 데이터를 생성하는 방식

## 7. AE, VAE

### 7.1 AE (Auto Encoder)

#### (1) AE란?

- 입력 데이터의 중요한 특징을 압축하고, 원래 데이터와 유사하게 복원하도록 학습하는 인공신경망 모델
- 입력 데이터 자체를 복원 목표로 사용하므로 정답 레이블이 필요 없음
- 입력 데이터와 복원 데이터의 차이를 최소화하는 것이 목표

활용 사례

- 데이터 압축: 중요한 특징을 추출하여 저차원으로 표현
- 노이즈 제거: 잡음이 포함된 데이터를 깨끗하게 복원
- 이상 탐지: 정상 데이터의 특징을 학습하여 비정상 데이터 탐지

기존 인코더·디코더와의 차이

- 기존 인코더/디코더: 입력을 이해하고 원하는 결과 생성
- AE의 인코더/디코더: 입력의 특징을 압축하고 입력 데이터 자체를 복원

→ AE의 핵심 목적은 입력 데이터 자체를 복원하는 것!

#### (2) AE의 구조와 작동 과정

- 인코더 → 병목 → 디코더 구조
- 입력 데이터를 압축한 뒤 다시 복원

**1. 압축 단계**

- 입력 데이터(x)에서 중요한 특징을 추출하고 정보를 압축
- 입력 데이터를 더 적은 수의 값으로 표현하여 잠재 벡터(z)로 변환
- 잠재 벡터(z): 원본 데이터의 핵심 특징을 압축하여 저장한 벡터

**2. 병목 단계**

- 인코더가 추출한 핵심 정보를 잠재 벡터(z)로 표현
- 제한된 공간에 정보를 담도록 하여 데이터의 핵심 특징을 학습하도록 유도

**3. 복원 단계**

- 압축된 잠재 벡터(z)를 입력받아 복원 데이터(x′) 생성
- 원본 데이터와 복원 데이터의 차이를 비교하여 정보가 얼마나 잘 보존되었는지 확인

#### (3) AE는 왜 압축했다가 복원할까?

**왜 데이터를 압축할까?**

- 데이터를 압축하지 않으면 중요한 특징을 학습하지 않고 입력 데이터를 단순히 복사할 수 있음
- 잠재공간의 크기를 제한하면 모든 정보를 그대로 담을 수 없으므로 원본 복원에 중요한 정보만 학습하도록 유도
- 예: 숫자 4 이미지에서 모든 픽셀을 기억하는 대신 숫자의 획과 형태를 핵심 특징으로 압축

**왜 다시 복원할까?**

- 압축된 정보만으로 원본과 유사한 데이터를 만들 수 있는지 확인하기 위해
- 복원 오차를 줄이도록 학습하여 인코더와 디코더의 표현 능력을 개선

**결과적으로 무엇을 학습할까?**

> 단순히 입력 데이터를 복사하는 것이 아니라 데이터의 핵심 특징과 구조를 학습

### 7.2 VAE (Variational Auto Encoder)

#### (1) VAE의 등장 배경

- 기존 AE는 입력 데이터를 압축하고 복원하는 데 초점을 맞추므로 잠재공간이 불규칙하게 형성될 수 있음
- 잠재공간에 데이터가 없는 빈 영역이 생길 수 있으며, 해당 영역의 값을 디코더에 입력하면 부자연스러운 결과가 생성될 수 있음
- AE는 기존 데이터 복원에는 효과적이지만, 임의의 잠재 벡터를 이용한 새로운 데이터 생성에는 한계가 있음

VAE의 개선

- 데이터를 고정된 값이 아닌 확률분포로 표현
- 잠재공간이 일정한 분포를 따르도록 학습
- 잠재공간에서 새로운 값을 샘플링하여 다양한 데이터 생성 가능

#### (2) VAE란?

- 입력 데이터의 특징을 확률분포로 학습하고, 해당 분포에서 샘플링하여 새로운 데이터를 생성하는 모델
- 기존 AE의 인코더·디코더 구조를 기반으로 발전한 생성 모델
- 변분 추론을 활용하여 복잡한 확률분포를 근사적으로 학습

VAE의 학습 목표

1. 복원 오차 최소화

> 원본 데이터와 복원 데이터의 차이를 최소화하여 원본의 특징을 잘 복원

2. KL 발산(KL Divergence) 최소화

- 두 확률분포가 얼마나 다른지를 나타내는 지표
- 두 분포가 같으면 0이며, 차이가 클수록 일반적으로 값이 커짐
- VAE에서는 인코더가 예측한 잠재분포가 표준정규분포에 가까워지도록 학습

→ 잠재공간을 규칙적으로 형성하여 새로운 데이터 생성에 적합한 공간을 만듦

VAE의 활용 분야

- 이미지 생성: 학습한 이미지의 특징을 바탕으로 새로운 이미지 생성
- 데이터 증강: 기존 데이터와 유사한 새로운 학습 데이터 생성
- 이상 탐지: 정상 데이터의 분포에서 크게 벗어난 데이터 탐지
- 특징 추출: 복잡한 데이터를 저차원 잠재공간으로 표현

#### (3) AE와 VAE의 차이

가장 큰 차이는 입력 데이터의 특징을 잠재공간에 어떤 형태로 표현하는가에 있음.

- AE: 입력 데이터를 하나의 고정된 잠재 벡터(점)로 표현
  - 잠재공간이 불규칙하게 형성될 수 있음
  - 데이터가 없는 영역에서 값을 추출하면 의미 없는 데이터가 생성될 가능성
- VAE: 입력 데이터를 확률분포로 표현
  - 입력 데이터마다 평균(μ)과 분산(σ²)을 계산
  - 잠재분포가 표준정규분포에 가까워지도록 학습
  - 연속적이고 규칙적인 잠재공간을 형성

잠재공간(Latent Space)이란?

- 입력 데이터에서 추출한 핵심 특징을 숫자 벡터로 표현하는 공간
- 비슷한 특징을 가진 데이터는 잠재공간에서도 가까운 위치에 배치될 수 있음

VAE 잠재공간의 장점

- 연속성: 잠재공간의 값을 조금씩 바꾸어 데이터의 특징을 자연스럽게 변화시킬 수 있음
- 새로운 데이터 생성: 학습한 분포에서 새로운 값을 샘플링하여 기존 데이터와 유사한 데이터 생성 가능

#### (4) VAE의 구조와 작동 과정

VAE는 분포 예측 → 샘플링 및 재파라미터화 → 생성의 3단계로 작동함.

**1. 분포 예측 단계**

- 입력 데이터(x)를 확률적 인코더에 전달
- 인코더가 데이터의 특징을 분석하여 평균(μ)과 표준편차(σ)를 예측
  - 평균(μ): 확률분포의 중심, 데이터의 대표적인 특징
  - 표준편차(σ): 확률분포가 평균을 중심으로 퍼진 정도
- 예측한 평균과 표준편차를 이용해 입력 데이터의 특징을 나타내는 확률분포 형성

사용 수식:

\\[ q\_\phi(z|x) \\]

> 입력 데이터 x가 주어졌을 때 잠재 벡터 z가 따르는 확률분포를 추정하는 인코더

→ 하나의 입력 데이터에서도 다양한 잠재 벡터를 추출할 수 있도록 확률분포를 학습

**2. 샘플링 및 재파라미터화**

> 예측한 확률분포에서 잠재 벡터(z)를 무작위로 샘플링

사용 수식:

\\[ z = \mu + \sigma \odot \varepsilon \\]

- z: 최종적으로 생성되는 잠재 벡터
- μ: 인코더가 예측한 평균
- σ: 인코더가 예측한 표준편차
- ε: 표준정규분포에서 추출한 무작위 노이즈
- ⊙: 각 요소끼리 곱하는 연산

\\[ \varepsilon \sim \mathcal{N}(0,I) \\]

- 평균 0, 각 차원의 분산 1인 표준정규분포에서 노이즈 추출
- 같은 확률분포에서도 서로 다른 잠재 벡터가 추출될 수 있어 다양한 데이터 생성 가능

재파라미터화

- 무작위 샘플링을 그대로 사용하면 신경망 학습에 필요한 기울기를 전달하기 어려움
- 무작위성을 별도의 노이즈 변수(ε)로 분리하는 방법
- 평균과 표준편차에 노이즈를 결합하여 잠재 벡터를 계산
- 샘플링 과정에서도 신경망이 학습할 수 있도록 함


→ 같은 평균과 표준편차에서도 무작위 노이즈에 따라 서로 다른 잠재 벡터 생성

**3. 생성 단계**

- 샘플링한 잠재 벡터(z)를 확률적 디코더에 전달
- 디코더는 잠재 벡터를 바탕으로 데이터의 확률분포를 예측하고, 이를 이용해 데이터 복원 또는 생성

사용 수식:

\\[ z \sim \mathcal{N}(0,I) \rightarrow x' \sim p\_\theta(x|z) \\]

- 잠재 벡터 추출: 표준정규분포에서 새로운 잠재 벡터 z를 무작위로 샘플링
- 새로운 데이터 생성: 추출한 z를 디코더에 입력하고, 디코더가 예측한 확률분포에서 새로운 데이터 x′ 생성

복원과 새로운 데이터 생성의 차이

- 데이터 복원: 원본 숫자 4 이미지 입력 → 잠재 벡터 추출 → 원본과 유사한 숫자 4 이미지 복원
- 새로운 데이터 생성: 학습 완료 후 표준정규분포에서 새로운 잠재 벡터 추출 → 디코더 입력 → 학습 데이터에 없던 새로운 형태의 숫자 이미지 생성

→ AE는 입력 데이터의 압축과 복원에 집중하지만, VAE는 확률분포를 학습하여 새로운 데이터까지 생성할 수 있음.

## 8. GAN (Generative Adversarial Network)

### 8.1 GAN의 개념

#### (1) GAN이란?

- 생성자와 판별자라는 두 신경망이 서로 경쟁하며 실제 데이터와 유사한 새로운 데이터를 생성하는 모델

생성자와 판별자의 역할

- 생성자(Generator, G)
  - 무작위 노이즈를 입력받아 실제 데이터와 유사한 가짜 데이터 생성
  - 판별자가 가짜 데이터를 진짜라고 판단하도록 학습
- 판별자(Discriminator, D)
  - 실제 데이터와 생성자가 만든 가짜 데이터를 입력받음
  - 입력 데이터가 진짜일 확률을 예측하여 구별


#### (2) GAN의 목적

1. 생성자의 목적

- 판별자가 진짜라고 판단할 만큼 실제 데이터와 유사한 가짜 데이터 생성
- 실제 데이터의 특징을 더 잘 반영하도록 개선

2. 판별자의 목적

- 실제 데이터와 생성된 데이터를 최대한 정확하게 구별
- 실제 데이터에는 높은 확률, 가짜 데이터에는 낮은 확률 부여

3. GAN의 궁극적인 목적

- 생성자와 판별자의 경쟁을 통해 생성자가 실제 데이터의 분포를 학습하도록 하는 것
- 실제 데이터와 구별하기 어려울 만큼 자연스러운 새로운 데이터 생성
- 학습 완료 후에는 일반적으로 생성자를 이용하여 새로운 데이터 생성

활용 분야

- 이미지 생성: 실제 인물처럼 보이는 가상 얼굴 생성
- 이미지 복원: 손상되거나 일부가 누락된 이미지 복원
- 데이터 증강: 기존 학습 데이터와 유사한 새로운 데이터 생성

### 8.2 GAN 구조

- 생성자(G)와 판별자(D)가 서로 반대되는 목표를 가지고 경쟁하며 학습하는 구조
- 이러한 경쟁 관계를 최소-최대 게임이라고 함

#### (1) GAN의 구조

1. 무작위 노이즈

- 생성자에 입력되는 무작위 숫자 벡터(z)
- 생성자는 노이즈를 바탕으로 새로운 이미지 생성

2. 생성자(Generator, G)

- 입력: 무작위 노이즈(z)
- 출력: 생성된 가짜 이미지 G(z)
- 목표: 판별자가 진짜라고 판단할 만큼 실제 데이터와 유사한 이미지 생성

3. 판별자(Discriminator, D)

- 입력: 실제 이미지(x) 또는 생성된 가짜 이미지 G(z)
- 출력: 입력 이미지가 진짜일 확률(0\~1)
- 목표: 실제 이미지와 가짜 이미지를 정확하게 구별

→ 판별자는 실제 이미지에는 높은 확률, 가짜 이미지에는 낮은 확률을 부여하도록 학습

#### (2) 최소-최대 게임

> 생성자와 판별자가 서로 반대되는 목표를 가지고 경쟁하며 성능을 개선하는 학습 방식

1. 판별자: 최대화

- 실제 이미지는 진짜로, 생성된 이미지는 가짜로 정확하게 구별하는 능력을 최대화
- 실제 이미지에는 1에 가까운 값, 가짜 이미지에는 0에 가까운 값 출력

2. 생성자: 최소화

- 판별자가 실제 이미지와 가짜 이미지를 구별하지 못하도록 학습
- 생성된 이미지가 진짜라고 판단되도록 판별자를 속이는 것이 목표

3. 경쟁을 통한 반복 학습

1. 생성자가 무작위 노이즈로 가짜 이미지 생성
2. 판별자가 실제 이미지와 가짜 이미지 구별
3. 생성자가 판별자의 판단을 바탕으로 이미지 개선
4. 판별자도 다시 학습하여 구별 능력 개선


### 8.3 GAN의 장단점

#### (1) GAN의 장점

**진짜 같은 가짜 생성 가능**

- 실제 데이터의 특징과 분포를 학습하여 현실에 존재하지 않지만 실제처럼 보이는 새로운 데이터 생성

**높은 이미지 선명도**

- 판별자의 평가를 바탕으로 생성자가 학습하여 이미지의 세부 특징을 정교하게 표현
- 이미지 생성 분야에서 기존 VAE보다 선명하고 사실적인 결과를 보여주는 경우가 많음

VAE와 GAN의 차이

- VAE: 픽셀 간의 수학적 평균값(MSE 등)을 최적화하기 때문에 이미지가 흐릿해지는 경향
- GAN: 판별자를 속이는 목표를 통해 실제 데이터와 유사한 미세한 질감까지 표현하여 선명한 결과 생성

**빠른 생성 속도**

- 학습 완료 후 무작위 노이즈를 생성자에 한 번 통과시켜 이미지 생성
- 여러 단계를 반복하는 일반적인 확산 모델보다 빠르게 생성 가능

**잠재공간의 부드러움**

- 잘 학습된 GAN은 잠재 벡터를 조금씩 변화시키면 생성 이미지의 특징도 자연스럽게 변화


#### (2) GAN의 한계점

**학습의 불안정성**

> 생성자와 판별자의 학습 균형이 무너져 안정적인 학습이 어려워지는 문제

원인

- 두 신경망이 서로 반대되는 목표를 가지고 동시에 학습
- 한쪽의 성능이나 학습 속도가 지나치게 앞서면 다른 쪽이 효과적으로 학습하기 어려움

결과

- 생성 이미지의 품질이 갑자기 나빠지거나 학습 성능이 오르내리는 진동 현상 발생
- 안정적인 균형점에 도달하지 못해 학습이 수렴하지 않을 수 있음

해결 방법

- 생성자와 판별자의 학습률 및 학습 횟수 조절
- WGAN-GP 등 학습 안정성을 개선한 모델 활용

**훈련 초기 문제**

> 학습 초기에 판별자가 가짜 데이터를 너무 쉽게 구별하여 생성자가 충분한 학습 신호를 받지 못하는 문제

원인

- 생성자가 실제 데이터의 특징을 충분히 학습하지 못한 상태
- 판별자가 지나치게 정확해지면 생성자에게 전달되는 기울기가 매우 작아질 수 있음

결과

- 생성자가 개선 방향에 대한 학습 신호를 충분히 얻지 못함
- 생성자의 학습 속도가 느려지거나 학습이 제대로 진행되지 않을 수 있음

해결 방법

- Non-saturating Loss: 생성자가 더 강한 학습 신호를 받을 수 있도록 손실함수 변경
- 판별자의 학습률이나 업데이트 횟수 조절

**모드 붕괴**

> 생성자가 다양한 데이터를 생성하지 못하고 특정한 형태의 데이터만 반복적으로 생성하는 현상

원인

- 판별자를 속이는 데 성공한 특정 유형의 데이터만 계속 생성하려 하기 때문

결과

- 생성 데이터의 다양성 감소

해결 방법

- Minibatch Discrimination: 여러 생성 이미지를 함께 비교하여 판별자가 이미지의 다양성까지 고려하도록 학습
- WGAN-GP 등 학습 개선 방법 활용

**평가 지표의 한계**

> 생성 데이터의 사실성, 선명도, 다양성을 하나의 지표만으로 완벽하게 평가하기 어려운 문제

원인

- 생성 모델의 성능에는 사실성, 선명도, 다양성 등 여러 요소가 포함됨
- 특정 지표가 높더라도 모든 측면에서 우수하다고 볼 수 없음

결과

- GAN 모델 간 객관적인 성능 비교가 어려울 수 있음
- 이미지가 선명해도 모드 붕괴가 발생할 수 있음

해결 방법

- FID: 실제 이미지와 생성 이미지의 특징 분포가 얼마나 유사한지 평가
- IS: 생성 이미지의 분류 확신도와 다양성을 함께 평가
- 여러 정량적 지표와 사람의 시각적 평가를 함께 활용

**역매핑의 어려움**

> 기본 GAN은 잠재 벡터(z) → 생성 이미지(x′) 방향으로 학습하기 때문에, 실제 이미지에 대응하는 잠재 벡터를 찾기 어려움

원인

- 이미지를 잠재 벡터로 변환하는 인코더가 기본 구조에 포함되어 있지 않음

결과

- 특정 이미지의 잠재 벡터를 찾아 수정하거나 복원하는 작업이 어려움

해결 방법

- GAN Inversion: 실제 이미지를 생성할 수 있는 잠재 벡터를 역으로 추정
- BiGAN: 이미지에서 잠재 벡터를 추정하는 인코더를 추가하여 역방향 학습 수행

### 8.4 GAN 적용 사례

#### NVIDIA – 가상 인물 이미지 생성(Progressive GAN)

> GAN을 활용하여 실제로 존재하지 않는 사람의 얼굴을 실제 사진처럼 생성하는 기술

작동 원리

- 실제 인물의 얼굴 사진을 학습하여 얼굴 특징과 패턴 습득
- 생성자와 판별자를 낮은 해상도에서 높은 해상도로 점진적으로 확장하며 학습
- 학습한 특징을 바탕으로 새로운 가상 인물 이미지 생성

결과 및 활용

- 피부, 머리카락, 눈, 표정 등을 사실적으로 표현한 고해상도 이미지 생성
- 가상 인물 제작, 이미지 생성 기술 연구

#### Deep Video Portraits – 인물 영상 합성(Conditional GAN)

> 한 사람의 표정과 움직임을 다른 사람의 영상에 자연스럽게 반영하는 기술

작동 원리

- 원본 영상에서 표정, 머리 움직임, 시선 등의 정보 추출
- 추출한 움직임 정보를 다른 인물의 영상에 적용
- 조건부 GAN(cGAN)을 활용하여 자연스러운 영상 생성

결과 및 활용

- 실제 촬영하지 않은 표정이나 움직임을 자연스럽게 합성
- 영화 특수효과, 영상 편집, 더빙, VR 등

#### Facebook – Eye In-Painting(Exemplar GAN)

> GAN을 활용하여 눈을 감고 찍힌 사진을 눈을 뜬 모습으로 자연스럽게 복원하는 기술

작동 원리

- 눈을 감은 사진과 동일 인물의 눈을 뜬 참고 사진 활용
- Exemplar GAN이 참고 이미지에서 눈의 모양과 색상 등의 특징 반영
- 감겨 있는 눈 부분을 해당 인물에게 어울리는 눈으로 생성하여 채움

결과 및 활용

- 인물의 고유한 특징을 유지하면서 사진을 자연스럽게 보정
- 인물 사진 보정, 이미지 복원 및 편집

## 9. 확산 모델

### 9.1 확산 모델의 개념

#### 1. 확산 모델이란?

- 정의: 이미지에 노이즈를 추가한 데이터를 학습하고, 무작위 노이즈를 점차 제거하면서 새로운 이미지를 생성하는 생성형 AI 모델
- 기본 원리
  - 실제 이미지에 노이즈를 조금씩 추가하는 과정을 만든다.
  - 모델은 노이즈가 포함된 이미지에서 노이즈를 제거하는 방법을 학습한다.
  - 학습 후 무작위 노이즈에서 시작해 새로운 이미지를 생성한다.

두 가지 핵심 과정

- 순확산(Forward Diffusion): 원본 이미지에 노이즈를 점차 추가
- 역확산(Reverse Diffusion): 노이즈를 단계적으로 제거하며 이미지 생성

활용 분야: 이미지 생성, 복원, 편집, 영상 생성 등

#### 2. 확산 모델의 목적

- 목표: 실제 데이터의 특징과 분포를 학습하여 자연스럽고 다양한 데이터를 생성

- 무작위 노이즈에서 실제와 유사한 이미지 생성
- 기존 이미지 복원이 아닌, 학습한 특징을 바탕으로 새로운 이미지 생성
- 하나의 조건에서도 다양한 결과물 생성

**모호성 모델링**

- 하나의 조건에서 여러 결과가 가능하다는 불확실성을 확률적으로 표현하는 것
- 노이즈가 많은 이미지에서는 원래 모습을 정확히 알기 어려움
- 가능한 이미지들의 확률분포를 학습하여 자연스러운 결과 생성
- 하나의 정답에 의존하지 않고 다양한 결과물 생성
- 예: '공원에 앉아 있는 강아지'라는 동일한 프롬프트에서도 품종, 털 색깔, 자세, 배경 등이 다른 이미지 생성 가능

### 9.2 확산 모델의 구조: 순확산과 역확산

#### 1. 순확산(Forward Diffusion)

- 정의: 원본 이미지에 가우시안 노이즈를 조금씩 추가하여 무작위 노이즈에 가까운 상태로 만드는 과정
  - 가우시안 노이즈: 정규분포를 따르는 무작위 잡음
- 목표: 다양한 수준의 노이즈를 추가하여 역확산 과정에서 노이즈 제거 방법을 학습할 수 있도록 함

**작동 과정**

1. 시작 단계(\\(x_0\\))
   - 노이즈가 없는 깨끗한 원본 이미지에서 시작
2. 노이즈 추가 단계(\\(x_1, x_2, \ldots\\))
   - 가우시안 노이즈를 조금씩 반복해서 추가
   - 단계가 진행될수록 노이즈가 누적되어 이미지의 윤곽과 특징이 흐려짐
3. 결과 단계(\\(x_T\\))
   - 원본 이미지의 정보가 거의 사라지고 무작위 노이즈에 가까워짐
   - 역확산 모델의 노이즈 제거 학습에 활용

#### 2. 역확산(Reverse Diffusion)

- 정의: 무작위 노이즈에서 시작해 노이즈를 단계적으로 제거하면서 새로운 이미지를 생성하는 과정
- 목표: 학습된 노이즈 제거 방법을 활용해 무작위 노이즈를 자연스러운 이미지로 변환

**작동 과정**

1. 시작 단계(\\(x_T\\))
   - 원본의 특징을 알아볼 수 없는 무작위 노이즈에서 시작
2. 노이즈 제거 단계(Denoising U-Net)
   - 학습된 Denoising U-Net이 현재 이미지의 노이즈를 예측
   - 예측한 노이즈를 바탕으로 노이즈를 조금씩 제거
   - 반복할수록 이미지의 색상과 윤곽, 구체적인 형태가 나타남
3. 결과 단계(\\(x_0\\))
   - 반복적인 노이즈 제거가 끝나면 자연스러운 새로운 이미지 생성

**핵심 비교**

- 순확산: 원본 이미지 \\(x_0\\) → 노이즈 추가 → \\(x_T\\)
- 역확산: 무작위 노이즈 \\(x_T\\) → 노이즈 제거 → 새로운 이미지 \\(x_0\\)

### 9.3 확산 모델의 장단점

#### 장점

1. 압도적인 샘플 품질

- GAN은 노이즈에서 이미지를 한 번에 생성하지만, 확산 모델은 단계적으로 노이즈를 제거함
- GAN보다 정교하고 다양한 이미지 생성에 유리
- 피부, 머리카락, 배경 등 복잡한 요소도 사실적으로 표현 가능

2. 학습의 안정성

- GAN처럼 생성자와 판별자가 경쟁하는 방식이 아니라 노이즈 예측 및 제거를 학습
- 평균제곱오차(MSE)로 실제 노이즈와 예측 노이즈의 차이를 계산
- 손실곡선으로 학습 변화를 수치적으로 확인 가능
- GAN에 비해 학습 과정이 비교적 안정적

3. 다양성 확보

- 무작위 노이즈에서 시작하므로 같은 조건에서도 서로 다른 이미지 생성 가능
- 특정 결과에 치우치지 않고 다양하고 자연스러운 결과물 생성

#### 단점

1. 느린 생성 속도

- GAN은 한 번에 이미지를 생성하지만, 확산 모델은 여러 단계의 노이즈 제거가 필요
- GAN보다 생성 시간이 오래 걸려 빠른 응답이 필요한 환경에 한계

2. 높은 계산 비용

- 학습에 많은 데이터와 연산 필요
- 이미지 생성 시에도 신경망을 반복 실행해야 함
- GPU 등 고성능 하드웨어가 필요하며 시간과 자원 소모가 큼

3. 구조적 제약

- 순확산과 역확산의 단계적 구조에 의존
- 노이즈 제거 단계의 수와 방식에 따라 생성 속도와 품질이 달라짐
- 단계와 방식을 적절하게 설계해야 함
