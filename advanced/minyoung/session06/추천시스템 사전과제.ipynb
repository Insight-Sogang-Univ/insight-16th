# 추천 시스템 


# Part 1. 추천 시스템 기초

# 1. 추천 시스템이란?

> **추천 시스템**: 사용자의 행동 기록과 아이템 정보를 바탕으로 사용자가 좋아할 만한 아이템을 예측해 골라 주는 시스템

## 1-1. 추천 시스템이 필요한 이유

- **고객 입장**
  - 원하는 아이템을 찾는 시간과 노력 감소
  - 몰랐던 취향의 아이템 발견
- **기업 입장**
  - 이용 시간, 구매 전환, 재방문 증가
  - 비인기 아이템의 노출 기회 확대

> 검색은 사용자가 이미 알고 있는 것을 찾도록 돕고, 추천은 아직 모르는 것을 발견하도록 도움

## 1-2. 사용되는 데이터의 종류와 특징

### 1. 명시적 데이터 

> 사용자가 선호를 **직접 표현**한 데이터

- 예: 별점, 좋아요·싫어요, 리뷰
- 장점
  - 선호와 불호가 비교적 분명해 취향을 정확히 반영
- 단점
  - 평가를 남기는 사용자가 적고, 본 아이템 중 일부에만 평가
  - 취향이 변하므로 과거 선호가 현재에도 유효한지 불확실
  - 사용자마다 점수 기준이 다름

### 2. 암시적 데이터

> 사용자의 행동을 바탕으로 선호를 **간접적으로 추측**하는 데이터

- 예: 조회수, 클릭, 구매, 시청 시간, 검색
- 장점
  - 별도 입력 없이 대량으로 수집 가능
- 단점
  - 행동이 반드시 선호를 의미하지는 않음
  - 클릭은 실수일 수 있고 구매는 선물일 수 있음
  - 행동이 없다고 싫어한다는 뜻은 아님. 아직 아이템을 보지 못했을 수 있음

→ 실제 서비스에는 암시적 데이터가 많아 두 종류를 함께 사용하는 경우가 많음.

## 1-3. 추천 시스템의 분류

추천 시스템은 **무엇을 근거로 추천하는지**에 따라 구분

### 1. 내용 기반 필터링

- 아이템의 **속성 및 유사도**를 기반으로 추천
- 사용자가 이용하거나 선택한 아이템과 특징이 비슷한 아이템을 추천
- 예: SF 영화를 자주 본 사용자에게 다른 SF 영화 추천

### 2. 협업 필터링

- 여러 사용자의 행동·평가 데이터를 함께 활용
- 취향이 비슷한 사용자는 앞으로도 비슷한 아이템을 좋아할 가능성이 높다는 가정

**두 가지 방식**

1. **사용자 기반(User-Based)**: 특정 사용자와 취향이 비슷한 사람들이 선호한 아이템 추천
2. **아이템 기반(Item-Based)**: 사용자가 선호한 아이템과 선호 패턴이 비슷한 아이템 추천

### 3. 하이브리드 필터링

- 두 가지 이상의 추천 알고리즘을 조합
- 내용 기반과 협업 필터링의 단점을 보완

- Weighted Ensemble: 여러 모델의 추천 점수 또는 투표를 합쳐 최종 아이템 결정
- Mixed: 여러 알고리즘의 추천 결과를 함께 보여 줌 
- Switch: 사용자나 서비스 상태 등 상황에 따라 알고리즘을 선택 
- Feature Combination: 여러 데이터의 feature를 조합해 하나의 추천 알고리즘 학습 
- Meta-Level: 첫 번째 모델의 결과가 다음 모델의 입력이 되는 방식 


# 2. 내용 기반 필터링

핵심 흐름: **아이템을 숫자로 표현 → 사용자 프로필 생성 → 유사도 계산 → 점수가 높은 아이템 추천**  
  
## 2-1. 사용자 프로필

> 사용자가 어떤 특성을 좋아하는지 정리한 정보

### 사용자 취향 정보를 얻는 방식

- **직접 지정형**: 사용자가 취향을 직접 입력
- **간접 지정형**: 과거 행동 이력에서 취향 추론

### 사용자 프로필 구성 방법

1. **사용자가 소비한 아이템의 특성 가중치 평균**
   - 특성 가중치: 아이템이 각 특성을 얼마나 가지고 있는지 나타내는 값(0~1)
   - 본 영화들의 특성 벡터를 평균 내 사용자 프로필 생성

2. **설문·키워드 입력·평가로 취향 파악**
   - 명시적 피드백: 좋아요/싫어요, 평점, 댓글
   - 암시적 피드백: 검색 로그, 구매, 저장, 북마크, 방문 페이지

3. **분류 문제로 해결**
   - 기존 아이템의 특징을 입력값, 사용자 평가를 정답 라벨로 사용
   - 새 아이템에 대해 사용자가 어떤 평가를 할지 예측
   - 결정 트리, 랜덤 포레스트, KNN 등 적용 가능

→ 사용자 프로필이 정확할수록 추천 정확도와 만족도가 높아짐.

## 2-2. 유사도 

> 두 대상이 얼마나 비슷한지 나타내는 숫자

- 아이템 정보를 벡터로 변환해 유사도 계산
- 사용자와 아이템은 같은 feature 기준으로 표현해야 비교 가능
- 사용자가 선호한 아이템과 유사도가 높은 아이템을 추천

### 유사도 계산 과정

**1) 아이템을 숫자로 변환**

- 아이템의 장르, 제작 국가, 감독, 배우, 등급 등을 feature로 정의
- 행은 아이템, 열은 feature인 테이블로 정리
- 텍스트 속성은 원핫 인코딩 등으로 숫자 변환
  - 해당 feature에 속하면 1, 아니면 0

**2) 사용자를 숫자로 변환**

- 직접 입력한 프로필이나 시청 기록으로 취향 파악
- 아이템과 동일한 feature 기준으로 벡터 표현

**3) 유사도 계산**

- 벡터: n개의 feature로 구성된 n차원 데이터
- **벡터 내적**: 같은 위치의 값끼리 곱한 뒤 모두 더한 값
  - 내적이 클수록 유사성이 높다고 볼 수 있음
  - 0과 1로 된 벡터라면 겹치는 feature 개수와 같음

### 대표적인 유사도 방식

**코사인 유사도**

- 두 벡터 사이 각도를 측정
- 각도가 작을수록 값이 1에 가까우며, 값이 클수록 유사
- 벡터 길이를 나누므로 크기보다 방향의 유사성에 집중
- 예: [1, 1]과 [5, 5]는 방향이 같으므로 코사인 유사도 1
- 거리로 사용하려면 `1 - 코사인 유사도`로 변환해 **코사인 거리**로 사용

**유클리디안 거리**

- 두 점 사이의 거리
- 값이 작을수록 두 데이터가 가까워 유사하다는 의미

**내적과 코사인 중 무엇을 사용할까?**

- 내적: 벡터의 방향뿐 아니라 길이도 반영
  - feature가 많은 아이템이 겹치는 feature도 많아져 유리할 수 있음
- 코사인: 벡터 길이를 무시하고 방향을 비교
  - feature 개수에 따른 쏠림은 줄지만, 정보가 풍부한 정도는 반영하지 못함

## 2-3. TF-IDF (Term Frequency - Inverse Document Frequency)

텍스트를 단순히 단어 출현 횟수로만 표현하면 '그리고', '있다'처럼 여러 문서에 흔한 단어가 중요하게 평가될 수 있음

> **TF-IDF**: 특정 문서에는 자주 등장하지만 다른 문서에는 드문 단어에 높은 가중치를 부여하는 방법

- **TF(Term Frequency)**: 특정 문서에서 단어가 등장한 빈도. 자주 등장할수록 커짐
- **IDF(Inverse Document Frequency)**: 역문서 빈도. 해당 단어가 등장한 문서 수가 많을수록 작아짐

→ TF-IDF가 높을수록 해당 단어가 다른 문서보다 그 문서에서 중요한 역할을 함

**추천 적용 예시**

1. 영화별 줄거리를 TF-IDF 벡터로 변환
2. 사용자가 좋아한 영화와 다른 영화의 벡터를 코사인 유사도로 비교
3. 유사도가 높은 영화 추천

## 2-4. 내용 기반 필터링의 장단점

- 아이템 속성만 있으면 신규 아이템 추천 가능 <-> feature로 표현할 수 있는 정보에 한계가 있어 속성이 부족하면 정확도 저하
- 다른 사용자의 데이터가 필요하지 않음 <-> 비슷한 아이템만 반복 추천하는 과도한 특화
- 어떤 feature 때문에 추천했는지 설명하기 쉬움 <-> 기록이 없는 신규 사용자는 추천하기 어려움

> **Cold Start Problem**: 충분한 정보가 수집되지 않아 적절한 추천을 하지 못하는 문제

- **아이템 콜드 스타트**: 새 아이템의 평가가 없음 → 내용 기반으로 해결 가능
- **사용자 콜드 스타트**: 새 사용자의 기록이 없음 → 내용 기반으로도 해결하기 어려움

# 3. 협업 필터링 

> 과거에 비슷한 선호도를 가진 사용자들은 앞으로도 비슷한 선호도를 가질 가능성이 높다는 가정에 기반한 추천 방식

## 3-1. 메모리 기반 협업 필터링

- 저장된 사용자-아이템 평점을 직접 사용해 새로운 아이템의 평점을 예측
- 별도 학습 없이 비슷한 사용자나 아이템인 이웃을 찾음
- **이웃 기반 협업 필터링**이라고도 함


### 1. 사용자 기반 

> 나와 취향이 비슷한 사용자(이웃)의 평점을 활용해 내 평점을 예측하고 추천

1. 나와 유사도가 높은 사용자 TOP-N명을 이웃으로 선정
2. 이웃들의 평점을 유사도로 가중평균해 미평가 아이템의 평점 예측
3. 예측 평점이 높은 아이템 추천

**사용자 간 유사도: 피어슨 상관계수**
- $r_{ui}$: 사용자가 아이템에 준 평점
- $\bar{r_u}$: 사용자 \(u\)의 평균 평점
- $i$: 두 사용자가 모두 평가한 아이템

**예측 평점**

- 이웃이 해당 아이템에 자기 평균보다 얼마나 높거나 낮은 점수를 줬는지 유사도로 가중해 반영
- 유사도가 높은 이웃일수록 예측에 큰 영향

### 2. 아이템 기반 

> 내가 평가한 아이템들과 평점 패턴이 비슷한 아이템을 찾아 추천

- 아이템 자체의 속성이 아니라, 사용자들이 아이템을 함께 좋아하거나 싫어한 **평가 패턴**으로 유사도 판단
- 사용자 수와 취향은 계속 변하지만 아이템 간 관계는 상대적으로 덜 변함
- 아이템 유사도를 미리 계산해 재사용하기 좋음


### User-Based vs Item-Based 비교

- 사용자 수는 보통 아이템 수보다 많아, 두 아이템을 함께 평가한 사용자는 많을 가능성이 있음
- 새 사용자는 새 아이템보다 자주 추가되므로 아이템 유사도를 미리 계산하는 방식이 효율적일 수 있음
- 일반적인 경향이며 정답이 정해진 것은 아님

## 3-2. 메모리 기반 협업 필터링의 한계

### 1. 희소 행렬 문제

> 사용자-아이템 평점 행렬에서 대부분의 칸이 비어 있어 유사도와 예측이 어려워지는 문제

- 실제 평점 행렬은 대부분의 사용자가 일부 아이템만 평가하므로 희소함
- 문제
  - 두 사용자가 함께 평가한 아이템이 적어 유사도를 신뢰하기 어려움
  - 공통 아이템이 1개뿐이어도 코사인 유사도가 1로 나올 수 있음
  - 비슷한 이웃을 찾지 못하면 평점 예측이 어려움
- 대응: **모델 기반 협업 필터링**
  - 이웃을 직접 비교하는 대신 관측 평점으로 사용자·아이템의 특징을 학습해 빈칸 예측

### 2. 콜드 스타트 문제

> 행동 기록이 충분하지 않아 적절한 추천을 하지 못하는 문제

- 협업 필터링은 행동 기록에 의존하므로 기록이 없는 대상에게 추천하기 어려움
- **신규 사용자**: 평가 기록이 없어 비슷한 이웃을 찾을 수 없음
- **신규 아이템**: 아무도 평가하지 않아 추천 후보에 오르지 못함
  - 첫 평가가 이뤄지기 전까지 추천되지 않는 문제를 **First-rater Problem**이라고도 함

**해결 방법**

- 가입 시 좋아하는 장르나 작품을 직접 지정하도록 함
- 신규 아이템에는 내용 기반 필터링을 함께 쓰는 하이브리드 방식 적용
- 임베딩 모델에서는 같은 카테고리 아이템 임베딩의 평균으로 신규 아이템을 임시 표현
- 인구통계학적 정보 등을 이용해 비슷한 사용자 그룹 추정

### 3. 확장성 문제

> 사용자와 아이템이 증가할수록 유사도 계산량이 커지는 문제

- 아이템 평점을 예측하려면 해당 아이템을 평가한 사용자들과 유사도를 계산해야 함
- 데이터 샘플링으로 계산량을 줄이면 추천 품질이 떨어질 수 있음
- 대응: **KNN 협업 필터링**
  - 모든 사용자를 사용하는 대신 유사도가 높은 k명의 이웃만 선택
  - 이웃은 예측하려는 아이템을 실제로 평가한 사용자 중에서 선택
  - k가 너무 작으면 소수 의견에 흔들리고, 너무 크면 덜 유사한 사용자까지 포함되어 예측이 흐려짐

## 3-3. 모델 기반 협업 필터링

> 관측된 평점 데이터로 예측 모델을 학습하고, 데이터의 패턴을 모델 파라미터에 담아 관측되지 않은 평점을 예측하는 방법


### 동작 단계

1. **학습**: 기존 평점을 잘 설명하는 모델 파라미터 탐색
   - 새 데이터가 쌓이면 주기적으로 재학습
2. **예측**: 학습된 파라미터로 비어 있는 평점 계산
   - 추천 시점에는 학습된 모델을 이용해 빠르게 수행


### 잠재 요인 (Latent Factor)

> 평점에 직접 드러나지는 않지만 사용자와 아이템 사이의 관계를 결정짓는 숨은 특성

- 관측 데이터를 바탕으로 모델이 수치적으로 추정
- 취향·성향처럼 데이터에 직접 존재하지 않는 개념을 숫자로 표현

**Latent Factor Model**

- 사용자와 아이템을 같은 잠재 공간의 벡터로 표현
- 두 벡터의 관계로 평점을 설명
- 평점 행렬을 사용자-잠재 요인 행렬과 아이템-잠재 요인 행렬로 분해


### 행렬 분해

> 큰 평점 행렬을 작은 행렬 두 개의 곱으로 나타내는 방법

**예측 평점**

- 사용자 벡터와 아이템 벡터의 내적으로 예측
- 요인별 사용자 취향과 아이템 특성이 같은 방향으로 클수록 예측 점수가 커짐
- 예시에서 '로맨스', '액션'은 설명을 위한 임의의 이름이며 실제 잠재 요인은 명확히 해석되지 않을 수 있음

### 행렬 분해 학습 과정

1. 잠재 요인 개수 k 설정
2. 사용자·아이템 벡터를 임의의 값으로 초기화
3. 평점이 있는 칸만 예측하고 실제 평점과 오차 계산
4. 오차가 줄어들도록 벡터 수정
5. 3~4 반복
6. 학습된 벡터로 빈칸의 평점 예측

**손실 함수**

- 오차 항: 예측 평점이 실제 평점에 가까워지도록 함
- 규제 항: 벡터 값이 지나치게 커지는 것을 막아 과적합 완화
- lambda: 규제 강도. 클수록 강한 규제

**최적화 방법**

- **SGD(확률적 경사 하강법)**: 평점을 하나씩 보며 오차가 줄어드는 방향으로 수정
- **ALS(교대 최소제곱법)**: 사용자 벡터와 아이템 벡터를 번갈아 고정하고 나머지를 계산. 병렬 처리에 유리

### 편향

사용자마다 점수 기준이 다르고 아이템마다 기본 평가 수준이 다르므로 예측식에 편향 항을 추가


### 행렬 분해 방식

- SVD: 수학적 분해 기법
- 학습 기반 MF: 최적화 알고리즘(SGD, ALS) 
- NMF*: 제약 조건 기반 모델
- PMF: 확률적 모델

### 메모리 기반과 모델 기반 비교

- 메모리 기반: 추천 이유를 설명하기 쉬움
- 모델 기반: 데이터를 압축해 예측이 빠르고 일반화에 유리
- 두 방식 모두 상호작용이 전혀 없는 신규 사용자·아이템의 콜드 스타트는 해결하기 어려움
- 행렬 분해는 희소성과 확장성을 완화하지만, 내적이라는 선형 계산에 의존한다는 한계가 있음

# Part 2. 딥러닝 기반 추천 시스템

# 1. VAE 기반 협업 필터링

> **VAE 기반 협업 필터링**: 사용자의 상호작용 기록을 신경망으로 확률분포로 압축한 뒤 다시 복원하면서, 아직 보지 않은 아이템의 선호를 예측하는 모델

## 1-1. MF의 한계와 딥러닝의 등장

### MF의 한계

**1. 선형성**

- 행렬 분해는 내적으로 요인별 값을 곱해 더하는 구조
- 요인들이 서로 얽히는 복잡한 조합 효과를 표현하기 어려움
- 예: 로맨스와 음악 영화를 각각 좋아하지만 두 요소가 결합된 뮤지컬 로맨스는 싫어하는 취향을 내적만으로 표현하기 어려움

**2. 불확실성 무시**

- 일반적인 MF는 사용자마다 하나의 벡터를 학습하는 점 추정
- 기록이 적은 사용자와 많은 사용자를 같은 확신도로 취급
- 기록이 적은 사용자의 불안정한 추정치를 구분하지 못함

### 딥러닝의 해법

→ 비선형 패턴과 사용자 표현의 불확실성을 다루기 위해 VAE 도입

## 1-2. 왜 AutoEncoder(AE)가 아닌 VAE인가?

### 오토인코더 (AutoEncoder, AE)

> 입력을 작은 벡터로 압축했다가 다시 복원하는 신경망 모델

**입력 x → 인코더 → 잠재 벡터 z → 디코더 → 복원된 x̂**

- 입력을 복원하는 데 필요한 핵심 정보만 남기도록 학습
- 정답이 따로 없고 입력 자체가 정답
- 추천 적용 초기 모델인 **AutoRec**은 아이템의 평점 벡터를 입력해 복원
- 은닉층에 비선형 함수를 적용했을 때 선형 구조보다 RMSE가 낮아지는 결과를 통해 비선형성이 추천 성능을 높일 수 있음을 보임
- AE도 잠재 표현을 점 하나로 나타낸다는 점은 MF와 유사

### VAE (Variational AutoEncoder)

> 잠재 벡터를 점 하나가 아닌 확률분포로 표현하는 오토인코더

- **평균 $\mu$**: 사용자 취향의 중심 위치
- **표준편차 $\sigma$**: 불확실성의 정도

**VAE가 더 안정적인 이유**

1. 학습할 때마다 $\mu$ 주변에서 z를 샘플링하므로 주변 위치에서도 비슷한 추천이 나오도록 학습
2. KL 항이 분포를 표준 정규분포 쪽으로 당겨 극단적인 위치로 치우치는 것을 억제


## 1-3. VAE 기반 CF (Mult-VAE)의 작동 구조

**Mult-VAE**는 사용자의 상호작용 기록을 입력받아 취향을 분포로 요약하고, 전체 아이템의 선호 확률을 복원

- 확률이 높은 순서로 사용자가 아직 상호작용하지 않은 아이템 추천

## 1-4. VAE의 목적 함수와 KL Divergence

VAE는 **ELBO**를 최대화하며, 정확한 복원과 잠재 공간 정규화를 함께 최적화

### 목적 함수 ELBO (Evidence Lower Bound)

**1. 복원 항**

- 디코더가 잠재 벡터 z로 원래 입력 x를 최대한 비슷하게 복원하도록 함

**2. KL Divergence 항**

- 인코더가 출력한 잠재 분포가 표준 정규분포에서 크게 벗어나지 않도록 함
- 과적합을 막고 잠재 공간을 연속적이고 부드럽게 유지

### KL Divergence

> 두 확률분포가 얼마나 다른지 나타내는 값

VAE에서는 사용자 분포가 표준 정규분포와 같으면 KL 값이 0
- 평균이 중심에서 멀어지거나 분산이 지나치게 작거나 커지면 KL 값이 증가
- KL 항은 잠재 분포가 표준 정규분포에서 과도하게 벗어나지 않도록 제약

# 2. GNN 기반 사용자 선호 예측

## 2-1. GNN (Graph Neural Network, 그래프 신경망)

> 그래프 구조로 표현된 데이터에서 노드와 이웃 노드의 관계를 학습하는 신경망

### GNN의 모티브와 그래프

- **그래프**: 노드와 엣지로 구성된 관계 표현
- 추천 시스템에서 사용자와 아이템을 노드로, 상호작용을 엣지로 표현 가능
- 그래프는 사용자-아이템 사이의 복잡한 연결 관계를 나타내는 데 적합

### 그래프를 사용하는 이유

- 사용자와 아이템의 관계는 단순한 개별 속성만으로 설명하기 어려움
- 직접 상호작용하지 않은 사용자·아이템도 연결 경로를 통해 관계를 파악할 수 있음
- 이웃의 정보를 모아 노드 표현을 갱신함으로써 고차 관계 학습 가능

## 2-2. GNN의 구조

- 노드의 초기 feature 또는 임베딩을 입력으로 사용
- 이웃 노드의 정보를 모으고 자신의 정보와 결합해 노드 표현 갱신
- 여러 층을 거치며 더 넓은 이웃의 정보를 반영

## 2-3. GNN의 프로세스

### 1. 그래프 구조 생성

- 사용자와 아이템을 노드로 구성
- 사용자-아이템 상호작용을 엣지로 연결
- 상호작용 기록을 바탕으로 이분 그래프 구성

### 2. 메시지 패싱

> 이웃 노드의 정보를 전달받아 자신의 표현을 갱신하는 과정

- 이웃 노드의 임베딩을 수집
- 수집한 정보를 합산·평균 등으로 집계
- 자신의 기존 표현과 결합해 새로운 임베딩 생성
- 여러 층을 쌓으면 더 먼 이웃의 정보까지 반영 가능

## 2-4. LightGCN 

### 주요 GNN 아키텍처

- GNN은 다양한 구조로 구현되며 추천에서는 사용자-아이템 그래프의 연결 정보를 효과적으로 학습하는 것이 중요

### LightGCN

- 협업 필터링을 위해 GCN 구조를 단순화한 모델
- 추천에 불필요한 feature transformation과 nonlinear activation을 제거
- 이웃 임베딩을 정규화해 전파하고, 여러 층에서 얻은 임베딩을 결합
- 사용자와 아이템의 상호작용 구조를 학습하는 데 집중

## 2-5. GNN 활용 사례

### SNS: Pinterest

- 사용자와 아이템의 상호작용 관계를 그래프로 표현
- 그래프 이웃 정보를 이용해 사용자 선호와 아이템 관계를 학습
- 학습한 표현을 추천에 활용

## 2-6. GNN의 장단점

**장점**

- 사용자-아이템의 복잡한 관계와 고차 연결 정보 학습
- 이웃 정보를 통해 직접 관측되지 않은 관계까지 활용 가능
- 그래프 구조에 담긴 협업 신호를 효과적으로 반영

**단점**

- 그래프 구성과 갱신에 비용이 발생
- 노드·엣지가 많아질수록 계산 부담 증가
- 여러 층의 메시지 패싱에서 정보가 과도하게 섞이거나 희석될 수 있음

# 3. Transformer 기반 추천 시스템

## 3-1. 기존 추천 시스템의 순차적 한계

### CF의 순차적 한계

- 사용자 행동의 순서보다 사용자-아이템 평점이나 상호작용 관계를 중심으로 학습
- 어떤 행동이 먼저·나중에 일어났는지, 행동 간 문맥을 충분히 반영하기 어려움

### GNN의 순차적 한계

- 그래프의 연결 관계와 이웃 정보를 잘 학습하지만, 행동의 시간적 순서를 직접 표현하는 데 한계
- 연결되어 있다는 사실만으로 행동의 선후 관계나 순서에 따른 의도 변화를 충분히 파악하기 어려움

## 3-2. Transformer: 사용자 행동을 시퀀스로 접근

- 사용자의 행동을 시간 순서가 있는 시퀀스로 표현
- 같은 아이템도 앞뒤 행동 문맥에 따라 의미가 달라질 수 있음
- Self-Attention으로 현재 행동과 관련 있는 과거 행동에 집중
- 멀리 떨어진 행동 사이의 관계인 Long-Range Dependency도 포착 가능

## 3-3. Transformer의 핵심: Self-Attention

> 시퀀스 안의 각 아이템이 다른 아이템들과 얼마나 관련 있는지 계산해, 관련도에 따라 정보를 반영하는 방식

### Query, Key, Value

- **Query(Q)**: 현재 위치에서 찾고자 하는 정보
- **Key(K)**: 각 위치가 가진 정보의 기준
- **Value(V)**: 실제로 전달할 정보

### 연산 과정

1. 각 입력을 Q, K, V 벡터로 변환
2. Query와 Key의 유사도를 계산해 관련성 점수 생성
3. 점수를 softmax로 정규화해 Attention 가중치 계산
4. 가중치와 Value를 곱해 합산
5. 관련성이 높은 아이템의 정보가 더 크게 반영된 표현 생성

## 3-4. Transformer 기반 주요 추천 시스템 모델

### 1. SASRec

- **단방향** 구조: 현재 위치에서 과거 아이템만 참조
- 학습 목표: 다음 아이템 예측
- 사용자의 과거 행동을 바탕으로 다음에 선택할 아이템을 예측
- 순차적 행동 예측에 적합

### 2. BERT4Rec

- **양방향** 구조: 과거와 미래 아이템 문맥을 모두 참조
- 학습 목표: 가려진 아이템 예측
- 시퀀스 일부를 `[MASK]`로 가리고, 가려진 위치의 아이템을 예측
- 정보 누출을 방지하기 위해 예측 대상 아이템을 마스킹
- 한 시퀀스에서 마스킹 위치를 바꿔 여러 학습 샘플 생성 가능

**BERT4Rec이 항상 더 좋은 것은 아님**

- 원 논문은 BERT4Rec이 SASRec보다 높은 성능을 보고
- 후속 연구에서는 손실 함수를 동일하게 맞추자 SASRec이 비슷하거나 더 높은 성능을 보이고 학습도 더 빠른 결과를 보고
- 성능 차이는 어텐션 방향뿐 아니라 학습 설정의 영향도 받으므로 상황에 따라 선택

# 4. 제언: 오늘날의 추천 시스템

## 4-1. LLM의 시대

### 관점의 전환: 추천을 랭킹 문제가 아닌 언어 문제로 접근

- **기존 추천**: 사용자-아이템 조합에 대한 점수를 계산
- **LLM 기반 추천**
  - 아이템 ID보다 의미 기반으로 추천
  - 텍스트·이미지·오디오 등 멀티모달 데이터 활용
  - 점수뿐 아니라 추천 이유를 문장으로 설명하는 서술형 추천 가능

### 새로운 가능성

1. **대화형 추천**
   - 사용자가 자연어로 입력한 조건을 반영
2. **Zero-shot 추천**
   - 상호작용 데이터 없이 아이템의 텍스트 설명만으로 추천
   - 신규 상품도 구매 기록이 없을 때 설명만으로 추천 후보가 될 수 있어 콜드 스타트 완화 가능

### P5 / TALLRec

**P5 (Geng et al., RecSys 2022)**

- 여러 추천 과제를 `텍스트 프롬프트 → 텍스트 답변` 형식으로 통일
- 하나의 언어 모델로 다양한 추천 과제를 학습
- 단점: 언어 모델을 추천 데이터로 학습하므로 많은 컴퓨팅 자원과 시간이 필요

**TALLRec (Bao et al., RecSys 2023)**

- 범용 LLM을 추천 지시문 형식으로 튜닝
- "좋아할까? (Yes/No)" 형식의 추천 지시문 활용
- LoRA로 일부 파라미터만 학습해 비용 절감
- 적은 샘플로도 기존 모델보다 높은 성능을 보고

→ LLM은 추천 성능뿐 아니라 사용자가 추천 시스템과 상호작용하는 방식도 변화시킴.

## 4-2. 인기 편향과 XAI 문제

### 1. 인기 편향 

> 소수의 인기 아이템이 추천을 독점하고, 비인기 아이템은 거의 추천되지 않는 현상

**강화 과정**

1. **데이터 불균형**: 상호작용이 인기 아이템에 집중
2. **모델의 편향 강화**: 데이터가 많은 아이템의 예측 점수가 높아져 더 자주 추천
3. **Rich-get-richer**: 추천 → 노출 → 클릭 → 학습 데이터 반영이 반복되며 격차 확대

### 2. 설명가능성 (XAI)

- 핵심 질문: *왜 이 아이템을 추천했는가?*
- GNN, Transformer 등 딥러닝 모델은 성능이 강력하지만 내부 판단 과정을 설명하기 어려운 **블랙박스** 문제가 있음
- 고객 설득과 비즈니스 의사결정을 위해 추천 근거를 제시할 필요가 있음

> LLM이 자연어로 추천 이유를 생성할 수는 있지만, 설명이 실제 추천 근거와 일치한다는 보장 없음

## 4-3. 컴퓨팅 비용과 Cold-Start를 고려한 추천 시스템 도입

### 비용 문제

- **GPU 비용**: 대형 모델의 학습·추론에 고가 GPU 필요
- **실시간 스코어링 서버 비용**: 요청마다 전체 아이템에 딥러닝 연산을 수행하면 서버 비용 증가
- **지속 비용**: 대규모 트래픽에서는 사용자 수와 요청 수에 비례해 비용 누적


### 1단계: 후보군 생성

- 목표: 수백만 개 아이템에서 수백 개 후보로 축소
- 모델: 선형 모델
  - 내용 기반 필터링
  - 협업 필터링
  - 행렬 분해
- 장점: 빠른 계산과 낮은 학습 비용

### 2단계: 순위 매기기

- 목표: 수백 개 후보를 정교하게 정렬해 상위 수십 개 노출
- 모델: 딥러닝 모델(GNN, Transformer, VAE 등)
- 장점
  - 사용자 맥락·아이템 속성·상호작용 이력을 결합
  - 비선형성, 순서, 관계 등 복잡한 패턴 학습
  - 후보군 안에서 정교한 순위 산출
