# NLP
---



### NLP (Natural Langauge Processing)

NLP란 사람과 컴퓨터가 의사소통이 가능하게 도와주는 도구입니다. 사람이 쓰는 영어나 한글과 같은 언어(자연어)를 컴퓨터가 처리할 수 있는 숫자로 바꿔주는 역할을 합니다.



자연어의 특징으로는 다음 5가지가 있습니다.

이산적

- 언어는 이미지나 소리처럼 연속적으로 이어지는 데이터가 아니라, ‘단어’처럼 서로 구분되는 독립적인 단위로 구성됩니다.

순차적

- 언어는 단어의 순서에 따라 의미가 달라집니다.

중의성과 모호성

- 같은 단어라도 문맥에 따라 의미가 달라질 수 있습니다.

예를 들어



"배를 탔다” → 선박



“배가 나왔다” → 신체 부위

희소성

- 단어의 조합은 무한하지만, 우리가 가진 데이터에서의 조합은 극히 일부입니다.

가변성

- 언어는 고정되어 있지 않고, 새로운 단어들이 등장하며, 문법파괴도 잦습니다.

### SLM

Statistic Language Model이란?

통계 기반 언어 모델(SLM)은 분포 가설에 근거합니다.

분포 가설이란
어떤 단어의 주변에 자주 등장하는 단어들을 보면 그 단어의 의미를 짐작할 수 있다는 뜻입니다.

예시로는

영현이는 수학과를 복수전공할 예정이다.
규원이는 수학과를 복수전공할 예정이다.
태희는 수학과를 복수전공할 예정이다.

여기서 바뀌는 부분은 사람 이름뿐이고, 나머지 문장 구조는 거의 동일합니다. 이처럼 공통점 차이점을 찾아 학습합니다.

그리고 이 분포가설은 조건부확률과 카운트 기반으로 학습합니다.
---

### N-gram



- 등장배경
SLM은 희소하게 등장한 것에 대해서는 잘 예측을 못하고, 문장이 길어질수록 다음 단어를 계산하기 어렵습니다. 그래서 전체대신에 문장의 일부만 보는 방식을 개발했습니다.

 

하지만 문장의 일부만 봐서 문장 전체의 맥락을 고려하지 못하고 N의 크기가 커질수록 희소문제가 심해집니다.
### PPL



PPL은 문장이 얼마나 혼란스러워 하는지 나타낸 것입니다.

문장 W가 희소하게 등장할수록 PPL이 커지고, 예측하기 쉬운 단어가 추가되면 낮아질 수 있습니다. 


### NLM

SLM은 단어를 단순한 기호로 처리했지만, 신경망 언어모델(NNLM)은 단어를 의미를 가진 벡터로 표현하여 문맥과 유사성을 학습하고 새로운 문장까지 일반화할 수 있게 만들었습니다

# 딥러닝기반 모델
---

### BERT  



BERT는 택스트의 앞 뒤 문맥을 동시에 고려헤 입력값을 백터로 바꿔주는 딥러닝 모델입니다.
BERT는 인코더 기반으로 트랜스포머의 인코더를 여러가지 쌓아 올린 구조입니다.
사전에 학습 할때는 두가지 방식으로 학습을 합니다.
- Masked Language Model(MLM)

빈 칸 채우기 방식으로 문장에서 무작위로 빈칸을 뚫고 이 빈칸에 어떤 단어가 들어갈 지 모델이 예측하며 학습합니다.





두번째로는두 문장의 선후관계가 있는지 아니면 무작위 위치에 있는지 확인합니다.

### GPT
 
GPT는 BERT와 다르게 인코더기반이 아니라 디코더기반으로 이전까지의 단어들만을 보고 다음 단어를 예측합니다.



### LLM

방대한 양의 데이터를 학습하여 자연어 및 기타 유형의 콘텐츠를 이해하고 생성하여 광범위한 작업을 수행할 수 있는 기초 모델의 범주입니다. AI시대에 자주 언급되는 챗지피티와 같은것들을 예시로 들 수 있습니다.
수십억 개의 문장으로 이루어진 방대한 데이터를 학습하고, 이를 토대로 언어의 패턴을 스스로 학습해 다음에 올 가장 확률 높은 단어를 예측합니다.

### RAG

LLM의 단점인 업데이트 비용을 보강한 질문에 대해 모델이 모르면 검색해서 답하는 구조인 RAG입니다.

질의 인코더가 질문을 백터로 바꾸면 지식 검색기가 외부에서 관련 내용을 검색해 지식 증강 생성기가 답변을 생성해주는 구조입니다.
뿐만 아니라 문서의 신뢰도까지 반영해 답변을 생성하는 CAG로 진화하기도 합니다.

### LangChain

LLM+RAG를 직접 구현하는 것은 복잡하니까, 이러한 모델 pipeline을 쉽게 구성할 수 있도록 도와주는 역할을 하는 것이 LangChain 프레임워크입니다.
특징들은 다음과 같습니다
추상화: LLM 서비스를 만들 때 필요한 여러 작업들을 복잡한 구현 대신 간단한 형태로 표현하여 개발 과정을 단순화합니다
표준화: 유사한 기능을 수행하는 요소들을 동일한 형식의 컴포넌트로 통일하여 일관성 있게 관리합니다
체이닝: 자주 사용하는 컴포넌트들을 순서대로 연결하여 Input → 컴포넌트 → Output 형태의 흐름을 쉽게 구성 및 이해 가능하게 합니다.

이런 LangChain의 문제점중 하나인 이전단계로 돌아가는 등의 문제를 해결해주는 LangGraph가 등장하기도 했습니다.

# sLM
---

### 모델 압축

- 가지치기  
중요도가 낮거나 불필요한 가중치, 뉴런 등을 제거하여 모델 크기와 계산량을 줄이는 방법입니다.

- 양자화  
고정밀 데이터를 저정밀 데이터로 변환하여 연산 속도와 메모리 효율을 높이는 방법입니다.


- 저차원 분해  
큰 가중치 행렬을 더 작은 행렬로 분해하여 계산량을 줄이는 방법입니다.

- 지식 증류  
큰 교사 모델의 지식을 작은 학생 모델에 전달하여 경량 모델을 만드는 방법입니다.

### sLM의 장점

작은 모델 크기로 인해 빠른 추론 속도와 높은 자원 효율성을 가지며 비용 절감과 로컬 실행 측면에서 유리합니다.

### sLM의 한계

복잡한 작업에서 일반화 성능이 낮아질 수 있으며 편향과 할루시네이션 문제가 발생할 수 있습니다.

### sLM과 LLM의 결합

복잡한 작업은 LLM이, 간단한 작업은 sLM이 처리하여 정확도와 효율성을 함께 높이는 방식입니다.

### 분류모델과 생성모델

이전세션에서 해왔던 것들이 분류 모델입니다. 정답 레이블이 있고, 입력된 데이터가 어떤 레이블에 속할지 결정하는 모델로 알면 됩니다. 반면에 생성 모델은 샘플에 레이블 지정에 관여하지 않고, 데이터 자체의 확률 분포를 학습해 학습 데이터와 유사한 새로운 데이터를 만들어 냅니다.

둘의 차이를 예시를 들어 설명하자면 



생성모델은 메뚜기 월드를 만드는 박명수라고 생각하시면 됩니다. 유재석이라는 사람을 학습해서 이에 맞는 새로운 노래를 만드는 것. 그것이 생성모델이라고 이해하시면 되고,



분류모델은 이와달리 정답을 찾아주는 것으로 수능시험 보듯이 가장 정답에 가까운 것을 맞추는 모델이라고 생각해주시면 됩니다.


---

### 생성모델의 분류

- AE

오토인코더(AE) 는 입력과 동일한 출력을 만드는 것을 목적으로 합니다.

구조는 정보를 압축시켜 잠재 백터로 만들고 이를 다시 디코더로 재복원 시킵니다.

저는 이부분을 궁금하면서 제일 이해가 안되는 부분이 그냥 원본 데이터를 쓰면 되지 왜 압축시켜서 데이터를 일부분 남기고 이를 왜 다시 복원시키는지 이해가 안됬습니다. 찾아본 결과로는 원본 그대로 쓰는거는 핵심 특징을 배우지 못해서 몇가지의 정보를 제거하고 쓰는 것 입니다. 예를 들어 고양이 두마리를 학습 시켰을때





색깔은 쓸데없는 정보일것입니다. 그래서 중요한 정보인 귀가 2개이고 4발로 걷는다 이런 정보들만 남기는 것이 오토인코더입니다.

---

- VAE

Variational Auto Encoder(VAE)는 데이터의 잠재적인 구조를 학습해 원본과 비슷한 새로운 데이터를 생성합니다. AE와 차이점은 AE는 점으로 학습하고 VAE는 분포로 학습한다는 점입니다. 따라서 잠재공간이 AE와는 달리 빈 공간이 거의 없습니다.

---

- GAN

GAN(Generative Adversarial Network)은 생성자 신경망과 판별자 신경망이 서로 경쟁하고 훈련하며 점점 더 정교하게 수행합니다.

예시를 들어보겠습니다.

투수(생성자) 오타니와 타자(판별자) 쇼헤이가 있습니다.





투수는 타자를 속이기 위해 직구와 최대한 비슷한 변화구를 던지려 노력합니다.
타자는 반대로 이런 직구와 변화구를 구분하려고 최대한 노력합니다.
이렇게 서로 경쟁하며 점점 더 정교하게 수행하는 신경망 모델이 GAN입니다.
생성자는 실제와 가까운 데이터를 생성하고 판별자는 진짜와 가짜를 점점 더 정확히 판별해나가려 합니다.

이를 통해 빠르게 진짜같은 가짜를 만들고 학습된 잠재 공간에서 보간이 가능하다는 장점이 있지만, 학습이 불안전하고, 좋은지 나쁜지 판별할수 있는 지표가 없다는 단점이 있습니다.

- 확산모델

확산 모델(Diffusion Model)은 입력 이미지에 노이즈를 조금씩 추가해 완전한 노이즈로 만든 뒤 이를 다시 거꾸로 복원해 나갑니다. 어떤 모호한 개념을 하나하나 구체화 해가며 결과를 출력합니다.

이때 순확산과 역확산을 사용하는데 순확산은 데이터에 노이즐 추가하는 과정, 역확산은 노이즈를 제거하는 과정입니다.
