# 시계열 데이터



### 시계열데이터란?


순차적인 시간의 흐름에 따라 수집된 데이터이며, 순서가 중요한 시퀀스 데이터로 간주됩니다.



예령이가 드럼을 배운다고 가정을 해봅시다.
첫째 주에는 한 곡을 템포 80으로, 둘째 주에는 90, 셋째 주에는 100, 넷째 주에는 110으로 칠 수 있게 됐다면, 이렇게 시간의 흐름에 따라 실력이 변화하는 데이터를 시계열 데이터라고 합니다.

이런 시계열 데이터는 크게 단변량과 다변량으로도 나눌 수 있습니다.
단변량은 시간에 따라 연속적으로 측정된 단일 변수로 이루어진 시계열 데이터이고
다변량은 같은 시간에 기록된 둘 이상의 변수로 이루어진 시계열 데이터입니다.

---

### 시계열 데이터의 특징

시계열 데이터는 크게 3가지 특징이 있습니다
1. 시간의존성



시계열 데이터는 시간 순서에 따라 관측되서 순서를 바꾸면 의미가 바뀝니다.
예를들어 제 키는 지금 182cm 이고 중2때 175cm입니다. 이를 보면 제가 중2때 비해 신체적으로 성장했다고 해석 할 수 있지만, 반대로 지금 175cm 이고 중2때 182cm 라고 가정한다면 늙어서 키가 줄었구나라고 해석됩니다. 

그래서 시계열은 데이터를 학습용이랑 테스트용으로 나눌 때 섞으면 안됩니다. 미래로 과거를 맞히면 당연히 성능이 좋을 수 밖에 없기때문입니다

---

2. 계절성과 추세



일정 주기에 따라 반복적으로 변동되는 계절성을 띄며, 시간의 흐름에 따라 나타나는 장기적인 증가/감소 경향인 추세를 볼 수 있습니다.

아리아나 그란데의 Santa Tell Me는 매년 12월이 될때마다 차트에 다시 올라오는 계절성을 볼 수 있습니다.

그리고 해가 갈수록 12월에 올라오는 순위가 점점 높아지면서 올라가는 추세를 보입니다.

---

3. 자기상관성

이전 시점의 값이 이후 시점의 값에 영향을 미칩니다. 시간의존성에서 예시를 들었던 키처럼 10초전의 저의키가 지금 제 키와 엄청난 영향을 미친다고 볼 수 있습니다.
이러한 특성때문에 데이터끼리 독립적이지 못해 일반 회귀에서 많은 문제를 야기하기도 합니다.

---

### 시계열 데이터의 분석요인

이러한 특성을 기반으로 분석요인들을 크게 4가지로 정의할 수 있습니다.

---

1. 추세 요인



추세 요인은 데이터 전반에 걸쳐 증가하거나 감소하는 경향을 말합니다.

---

2. 계절 요인

계절요인은 일정한 주기 반복되는 패턴을 봅니다.

---

3. 순환 요인

계절성보다 더 길고 불규칙한 주기로 반복되는 변동을 순환요인이라 합니다.

계절요인은 주기를 통해서 언제 올지 알 수 있지만, 순환요인은 주기가 일정치 않아 언제올지 몰라서 추세랑 같이 보기도 합니다.

---

4. 불규칙 요인



배달관련 결제횟수나 금액이 증가하지만 2019년 초반에 급증하는걸 볼 수 있습니다. 코로나 때문에 그런것인데요. 이처럼 팬대믹 혹은 자연재해로 인해 발생하는 예측 불가능한 변동을 나타냅니다.

### 시계열 데이터 전처리

- 결측치/이상치 처리

시계열 데이터에서도 결측치나 이상치가 발생할 수 있습니다. 하지만 아까 자기상관성으로 인해 일반회귀를 사용하지 못햇던 것처럼 시계열 데이터에서의 결측치 처리방법도 다양합니다.

먼저 기초세션때 배운것처럼 삭제하거나 평균값으로 채울수 있습니다.

또 추가로 시계열데이터에서는 주어진 데이터 포인트의 값을 사용하여 두 개 이상의 데이터 포인트 사이에서 누락된 값을 추정하는 보간법을 사용할 수 있습니다.



위 사진에서 점 c를 (4.5,2.5)겠구나 예측하는것과 같습니다.

보간법은 점을 어떻게 잇느냐에 따라 나뉩니다.
* 선형 보간: 앞뒤 값을 직선으로 이어 채웁니다.
* 다항식 보간: 여러 점을 지나는 곡선으로 채웁니다.
* 스플라인 보간: 구간마다 부드러운 곡선을 이어 붙여 채웁니다.
* 조각별 상수 보간: 바로 앞이나 뒤의 값을 그대로 가져와 채웁니다.

그다음에 이상치 탐지를 어떻게 하는지 알아보겠습니다.
IQR Rule-based Anomaly Detection 은 25% 75% 분위수를 기준으로 특정값을 더 벗어나면 이상치로 간주하는것입니다.

Z-score는 수시때 지원할때 많이 들어봐서 아실거 같은데 표준편차를 사용해 평균으로부터 얼마나 벗어나는지 알아보는것 입니다.

STL 분해는 추세와 계절성을 제거하고 알아보는 방법입니다.

이 방법외에도 지금까지 배운 모델들을 활용해 이상치를 알아볼 수 있으며, 처리는 결측치와 같은 방법으로 할 수 있습니다.

* CART: 예측값과 실제값의 차이가 큰 곳을 이상치로 봅니다.
* 클러스터링: 어느 군집에도 속하지 않는 값을 이상치로 봅니다.
* 오토인코더: 원래대로 잘 복원하지 못하는 값을 이상치로 봅니다.

이상치는 지우면 시간간격이 깨져, 위에서 말한 보간법들을 사용해 채우기도 합니다

- EDA

EDA를 하기전에 먼저 평활화를 진행합니다.

평활화란시계열 데이터에서 노이즈를 줄이고, 전체적인 흐름이나 추세를  더 명확하게 파악하기 위해 데이터를 부드럽게 만드는 기술입니다.

이때 이동평균을 사용하는데, 이동평균이란 부분집합으로 평균을 구하는 것을 말합니다.



수능국어점수를 예측할때 초등학교 받아쓰기 점수까지 반영하는것보다 6월모의고사 9월 모의고사 점수만 반영하는것이 더 정확도가 높다고 생각하는것으로 이해할 수 있습니다.

한 번에 평균을 내는 개수가 크면 그래프가 매끄러워지지만 변화에 늦게 반응하고, 작으면 변화에 빨리 반응하지만 노이즈가 남습니다. 개수를 계절 주기와 같게 맞추면 계절성이 사라져 추세만 볼 수도 있습니다.

이동평균은 가중치를 주는 방식에 따라 나뉩니다.
* 단순 이동평균: 모든 값에 같은 가중치를 줍니다.
* 가중 이동평균: 최근 값일수록 큰 가중치를 줍니다. 6월 모의고사보다 9월 모의고사를 더 크게 반영하는 것과 같습니다.
* 지수 이동평균: 과거로 갈수록 가중치가 점점 줄어들면서 모든 과거 값을 조금씩 반영합니다.

- 요소분해

시계열 데이터를 위에서 배운 4가지의 개별성분으로 분해하는 과정입니다.

---

- 가법모형과 승법모형

이때 두가지 방법으로 분해할 수 있는데 첫번재는 가법모형으로 각 변동 요인들을 독립적으로 더하는 방법과 요인들을 곱하는 승법모형이 있습니다.



가법모형 - 노홍철의 자신이 작년에 비해 1억원 감소했다

승법모형 -노홍철의 자신이 작년에 비해 몇 프로 감소했다.

계절에 따른 오르내림의 폭이 일정하면 가법모형, 추세가 커질수록 그 폭도 같이 커지면 승법모형을 씁니다. 승법모형에 로그를 씌우면 곱하기가 더하기로 바뀌어 가법모형처럼 다룰 수 있습니다.

---

- seasonal_decompose / STL

seasonal_decompose는 이동평균을 적용해 고정된 계절성을 반영하며, STL은 시간에 따라 변하는 계절성 반영합니다.

이때 고정된 계절성은 매년 똑같은 계절에 같은 양이 증가한다. 변하는 계절성은 같은 계절에 증가하는 양이 점점 증가한다. 



엑소의 첫 눈이 매년 겨울마다 멜론차트에서 30계단씩 오른다면 고정된 계절성, 오르는 폭이 해마다 커진다면 변하는 계절성으로 이해하시면 됩니다.

seasonal_decompose는 이동평균으로 추세를 먼저 구하고, 원래 데이터에서 추세를 뺀 뒤 같은 계절끼리 평균을 내 계절 요인을 구합니다. 마지막에 남은 것이 잔차입니다.

STL은 이동평균 대신, 각 지점의 주변 데이터로만 작은 회귀선을 그려 잇는 LOESS를 씁니다. 추세와 계절을 번갈아 다시 구하며 다듬고, 잔차가 유독 큰 시점은 가중치를 줄여 이상치에 덜 흔들리게 합니다.


### 정상성 판단

- 정상성이란?



시간의 상관없이 일정한 성질을 띄는 것을 말합니다. 평일의 저의 기상시각과 같습니다. 저는 매일 어제 7:30에 기상했고, 담주 화요일에도,수요일도,목요일도 똑같을 겁니다.

- 강한 정상성과 약한 정상성

강한 정상성은 시간을 언제로 잡든, 데이터의 확률 분포 자체가 똑같은 상태로 현실에서 일어나기 힘든 상태를 말합니다.

약한 정상성은 실무에서 쓰이는 현실적 기준으로 평균과 분산이 일정하지만, 자기공분산이 시차에만 의존하는 상태입니다.

---

- 정상성의 판단 - 시각과 통계



시각적으로 봤을때 평균을 중심으로 진동하는 형태를 보이면 정상성을 띈다고 볼 수 있습니다. 위 그림과 같이 0을 중심으로 진동하는 형태를 띄며 정상성을 띈다고 해석합니다.

통계적으로는 단위근(Unit Root)이 있는지를 확인하는 과정을 거칩니다.

오늘 값이 어제 값의 몇 배만큼 남는지를 φ라고 하면, φ가 1보다 작을 때는 충격이 점점 줄어 평균으로 돌아옵니다. 그런데 φ가 1이면 어제 값이 그대로 남아서 충격이 사라지지 않고 계속 쌓입니다. 이 상태를 단위근이 있다고 하고, 단위근이 있으면 비정상 시계열입니다.

---

또 이를 검정하는 방법은 두가지로 나뉩니다.

ADF 검정은 데이터가 비정상 시계열이라고 가정하고 시작하며, 과거 값의 영향이 계속 누적되는지 확인하여 정상성을 판단합니다. 일반적으로 p-value가 작으면 정상 시계열로 판단합니다.

반대로 KPSS 검정은 데이터가 정상 시계열이라고 가정하고 시작하며, 평균과 구조가 시간에 따라 일정하게 유지되는지 확인합니다. 일반적으로 p-value가 작으면 비정상 시계열로 판단합니다.

두 검정은 출발점이 반대라서 함께 씁니다. 둘 다 정상이면 정상, 둘 다 비정상이면 차분이 필요합니다. ADF만 비정상이면 추세를 빼면 되는 추세 정상, KPSS만 비정상이면 차분하면 되는 차분 정상으로 봅니다.

정상성을 확보하기 위해서 추세와 계절성을 빼고 계산하거나 분산을 안정화 시키는 방법을 사용합니다.

### 단변량 시계열 분석

- AR / MA

자기 회귀 모델(AR)은 자기 자신 데이터가 미래 값을 설명해줄 것이라고 가정하고 회귀모델을 사용하는것입니다.



가연님이 응원하는 한화인데요 작년에는 2등, 올해는 9등이니까 내년에는 더 떨어지지 않을까 라고 예측합니다.

---

이동 평균 모델(MA)

과거 시점의 잔차를 독립변수로 삼아 현재 값을 예측하는 모델로 과거의 잘못 예측한 것을 현재 시점에 반영하는 모델입니다.

- AR,MA의 전제조건

둘다 정상성과 가역성 조건들을 만족해야 모델을 사용할 수 있습니다.

AR의 특성방정식을 통해서 정상성을 확인하고, MA는 특성방정식을 통해서 가역성을 확인한 후 모델을 사용합니다.

정상성은 충격을 받아도 시간이 지나면 평소로 돌아오는 성질이고, 가역성은 관측한 값으로 지난 충격을 거꾸로 계산해 낼 수 있는 성질입니다. AR은 계수의 절댓값이 1보다 작아야 정상성을 만족하고 가역성은 항상 만족합니다. 반대로 MA는 정상성은 항상 만족하고, 계수의 절댓값이 1보다 작아야 가역성을 만족합니다.

---

- ARMA

위애서의 정상성과 가역성을 만족한다면, ARMA (Autoregressive Moving Average) 모형을 사용할 수 있습니다.

이 모델은 AR(p) 모형과 𝑀𝐴(𝑞)모형을 결합한 모형
과거 p개의 관측값 + 과거 q개의 오차를  활용해서 예측하는 방법을 사용합니다.

---

- 차수 결정 방법 : 자기 상관 함수 (ACF, PACF)

자기상관계수(ACF, Auto Correlation Function)

ACF는 현재 데이터가 과거 데이터와 얼마나 관련 있는지를 확인하는 지표입니다.

PACF는 중간 시점들의 영향을 제거한 뒤, 순수한 직접 관계만 측정하는 지표입니다. 즉 현재 값과 특정 시차 값 사이의 순수한 영향력만 측정하는 방식입니다.

그래프에서 막대가 갑자기 0 근처로 떨어지기 직전의 시점을 고릅니다. PACF에서 고르면 AR의 차수 p, ACF에서 고르면 MA의 차수 q입니다. ACF는 천천히 줄고 PACF가 1 뒤에 끊기면 AR(1), 반대로 ACF가 1 뒤에 끊기고 PACF가 천천히 줄면 MA(1)로 봅니다.

### 불안정 시계열 분석

불안정 시계열 분석은 크게 아래 2가지로 볼 수 있습니다.

- ARIMA 

AR(Autoregressive) : 자기회귀, I(Integrated) : 차분, MA(Moving Average) : 이동평균을 활용하는 모델입니다.

즉,과거 값 + 과거 오차 + 차분을 이용하여 미래를 예측하는 모델입니다

---

- SARIMA



하지만 ARIMA는 계절성을 처리할 수가 없습니다시계열 데이터의 계절성을 처리할 수 있는 모델로 ARIMA에서 계절성도 처리할 수 있게끔 발전한 모델입니다.

---

### 다변량 시계열 분석

- VAR



AR 모형은 자기 자신의 과거 데이터만 사용하여 미래를 예측하는 방식입니다. 예를 들어 이번 주 프로미스나인 유튜브 조회수는 지난주 프로미스나인 조회수 흐름만 보고 예측하는 방식입니다.

반면 VAR 모형은 여러 변수의 과거 데이터를 함께 사용하여 미래를 예측하는 방식입니다. 예를 들어 이번 주 프로미스나인 화제성은 지난주 화제성뿐 아니라 팬 커뮤니티 언급량, 음원 순위, 쇼츠 조회수 등 다른 변수들의 과거 데이터까지 함께 사용하여 예측합니다.

즉 AR은 자기 자신의 과거만 사용하는 단방향 구조이고, VAR은 여러 변수들이 서로 영향을 주고받는 관계까지 함께 고려하는 모델입니다.

### 딥러닝을 활용한 시계열 분석

RNN



RNN 계열은 이전 시점의 정보를 현재 시점으로 전달하는 순환 구조 기반 시계열 딥러닝 모델입니다. 대표적으로 RNN, LSTM, GRU가 있으며, 과거 정보의 흐름을 학습하는 데 사용됩니다. LSTM과 GRU는 장기 의존성 문제를 개선한 구조이지만 긴 시계열에서는 학습 효율이 떨어질 수 있습니다.

---

Transformer



Transformer 계열은 Attention 메커니즘을 활용하여 긴 시계열의 전역 패턴과 복잡한 관계를 학습하는 모델입니다. Informer, Autoformer, TFT 등이 대표적이며, 긴 시계열·주기성·외생 변수 처리에 강점을 가집니다. 병렬 연산이 가능해 대규모 데이터 처리에 효율적이지만 구조가 복잡하고 학습 비용이 큽니다.

---

Foundation Model

Foundation Model은 대규모 데이터로 사전 학습된 범용 시계열 모델입니다. Fine-tuning이나 Prompting을 통해 다양한 시계열 문제에 활용할 수 있으며, LLM 기반 방식과 시계열 전용 모델 방식으로 발전하고 있습니다. 대표적으로 TimeGPT, TimesFM, Chronos 등이 존재합니다.
