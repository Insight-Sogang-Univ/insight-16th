# 실습 1) 논리 게이트를 직접 만들어 보기

`1-2 단층 퍼셉트론과 그 한계`를 마친 뒤에 진행합니다.

강의에서 **AND, OR, NAND는 되는데 XOR만 안 된다**고 했죠.
정말 그런지, 직접 숫자를 넣어가며 확인해 봅니다.

## 1-A. 퍼셉트론 한 개를 함수로 만들기

강의에서 본 퍼셉트론의 계산 절차 그대로입니다.

$$\text{출력} = \begin{cases} 1 & (x_1 w_1 + x_2 w_2 + b > 0) \\ 0 & (\text{그 외}) \end{cases}$$

`가중치(w1, w2)`와 `편향(b)`만 정해주면 게이트 하나가 완성됩니다.

````python
def perceptron(x1, x2, w1, w2, b):
    """퍼셉트론 한 개. 가중합이 0보다 크면 1, 아니면 0을 출력한다."""
    total = x1 * w1 + x2 * w2 + b
    return 1 if total > 0 else 0


# 입력으로 가능한 조합은 이 4가지가 전부입니다
INPUTS = [(0, 0), (0, 1), (1, 0), (1, 1)]


def print_table(name, w1, w2, b, answer):
    """게이트를 만들어 진리표를 출력하고, 정답과 맞는지 채점한다."""
    print(f"[{name}]  w1={w1}, w2={w2}, b={b}")
    correct = 0
    for (x1, x2), ans in zip(INPUTS, answer):
        out = perceptron(x1, x2, w1, w2, b)
        mark = "O" if out == ans else "X  <-- 틀림"
        correct += (out == ans)
        print(f"   ({x1}, {x2}) -> {out}   (정답 {ans})  {mark}")
    print(f"   => {correct}/4 맞음\n")
    return correct
````

## 1-B. AND · OR · NAND 만들어 보기

강의자료 `1-2 ③`의 표에 있던 값들을 그대로 넣어 봅니다.

````python
# 정답(진리표)을 먼저 적어둡니다. 순서는 (0,0), (0,1), (1,0), (1,1)
AND_ANSWER  = [0, 0, 0, 1]   # 둘 다 1일 때만 1
OR_ANSWER   = [0, 1, 1, 1]   # 둘 다 0일 때만 0
NAND_ANSWER = [1, 1, 1, 0]   # 둘 다 1일 때만 0

print_table("AND",  0.5,  0.5, -0.7, AND_ANSWER)
print_table("OR",   0.6,  0.6, -0.5, OR_ANSWER)
print_table("NAND", -0.5, -0.5, 0.7, NAND_ANSWER)
````

셋 다 `4/4`가 나왔을 겁니다.

**NAND를 보세요.** AND의 `[0.5, 0.5, -0.7]`에서 **부호만 전부 뒤집은 값**입니다.
출력을 반대로 뒤집고 싶으면 가중치의 부호를 뒤집으면 되는 거예요.

값이 저것 하나만 되는 것도 아닙니다. 아래 칸에서 **다른 숫자**로도 AND가 되는지 직접 확인해 보세요.

````python
# 여기를 채우세요 — AND가 되는 다른 조합을 찾아보세요
#    (힌트: w1 + w2 > -b > w1  이면 됩니다)

print_table("AND (내가 찾은 값)", 1.0, 1.0, -1.5, AND_ANSWER)
````

## 1-C. 이제 XOR을 만들어 보세요

XOR은 **두 입력이 서로 다를 때만 1**입니다.

$$(0,0) \to 0 \qquad (0,1) \to 1 \qquad (1,0) \to 1 \qquad (1,1) \to 0$$

아래 칸의 `w1, w2, b`를 **직접 바꿔가며** 4개를 다 맞춰 보세요.
**여러 번 시도해 보시는 게 이 실습의 핵심입니다.**

````python
XOR_ANSWER = [0, 1, 1, 0]   # 서로 다를 때만 1

# 여기를 채우세요 — 숫자를 바꿔가며 4/4를 만들어 보세요
w1 = 0.5
w2 = 0.5
b  = -0.2

print_table("XOR 도전", w1, w2, b, XOR_ANSWER)
````

아마 **3/4에서 계속 멈췄을** 거예요. 하나를 맞추면 다른 하나가 틀어집니다.

내가 숫자를 잘못 골라서일까요? 그럼 **가능한 숫자를 전부 넣어 보면** 되겠네요.

````python
import numpy as np

# -2.0부터 2.0까지 0.1 간격 = 41개 값. 세 개의 조합이니 41 x 41 x 41
grid = np.round(np.arange(-2, 2.001, 0.1), 2)

best_score = 0
best_params = None
tried = 0

for w1 in grid:
    for w2 in grid:
        for b in grid:
            tried += 1
            score = sum(perceptron(x1, x2, w1, w2, b) == ans
                        for (x1, x2), ans in zip(INPUTS, XOR_ANSWER))
            if score > best_score:
                best_score, best_params = score, (w1, w2, b)

print(f"시도한 조합의 수 : {tried:,}개")
print(f"가장 잘 맞춘 개수 : {best_score}/4")
print(f"그때의 값        : w1={best_params[0]}, w2={best_params[1]}, b={best_params[2]}")
````

**68,921개를 전부 넣어봐도 최대 3/4입니다.**

우리가 숫자를 잘못 고른 게 아니라, **구조상 불가능**한 것이었어요.

## 1-D. 왜 안 되는지 눈으로 보기

퍼셉트론이 하는 일은 결국 $x_1 w_1 + x_2 w_2 + b = 0$ 이라는 **직선 하나를 긋는 것**입니다.
직선의 어느 쪽에 있느냐로 0과 1을 가르는 거예요.

AND와 XOR의 네 점을 찍어놓고, 직선을 그어 보면 차이가 보입니다.

````python
import platform
import matplotlib.pyplot as plt

# 한글 폰트 설정 (OS 자동 감지: Mac → AppleGothic, Windows → Malgun Gothic, 그 외 → NanumGothic)
plt.rcParams['font.family'] = {'Darwin': 'AppleGothic', 'Windows': 'Malgun Gothic'}.get(platform.system(), 'NanumGothic')
plt.rcParams['axes.unicode_minus'] = False    # 마이너스 기호 깨짐 방지

fig, axes = plt.subplots(1, 2, figsize=(11, 4.5))

for ax, (title, answer, line) in zip(axes, [
        ("AND  (직선으로 나눌 수 있음)",    AND_ANSWER, (0.5, 0.5, -0.7)),
        ("XOR  (어떤 직선으로도 못 나눔)", XOR_ANSWER, None)]):

    for (x1, x2), ans in zip(INPUTS, answer):
        ax.scatter(x1, x2, s=320, marker='o' if ans == 1 else 'X',
                   c='crimson' if ans == 1 else 'steelblue',
                   edgecolors='k', zorder=3)
        ax.annotate(f"{ans}", (x1, x2), fontsize=13, fontweight='bold',
                    xytext=(10, 8), textcoords='offset points')

    if line:                                  # w1*x1 + w2*x2 + b = 0  ->  x2 = -(w1*x1+b)/w2
        w1, w2, b = line
        xs = np.linspace(-0.5, 1.5, 10)
        ax.plot(xs, -(w1 * xs + b) / w2, 'k--', lw=2, label='결정 경계')
        ax.legend(loc='upper left')

    ax.set_title(title, fontsize=13)
    ax.set_xlim(-0.5, 1.5); ax.set_ylim(-0.5, 1.5)
    ax.set_xlabel('x1'); ax.set_ylabel('x2'); ax.grid(alpha=0.3)

plt.tight_layout(); plt.show()
````

오른쪽 그림에서 **빨간 동그라미 둘**과 **파란 X 둘**은 서로 **대각선으로 마주보고** 있습니다.

직선 하나로는 대각선끼리 묶을 수가 없어요. 자를 대고 아무리 그어봐도 안 됩니다.

# 실습 2) 학습률을 바꿔가며 돌려보기

`2-4 경사하강법`을 마친 뒤에 진행합니다.

## 2-0. 데이터 준비

실습 2와 실습 3에서 **같은 데이터**를 씁니다.

`3개의 클래스`가 나선 모양으로 꼬여 있는 데이터예요.
**직선으로는 절대 못 나누는** 데이터라, 실습 1에서 본 XOR의 확장판이라고 보시면 됩니다.

````python
import importlib.util
import subprocess
import sys

# Colab에 TensorFlow가 없는 경우에만 설치합니다.
if importlib.util.find_spec('tensorflow') is None:
    subprocess.check_call([sys.executable, '-m', 'pip', 'install', '-q', 'tensorflow'])

import platform
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Input

# 한글 폰트 설정 (실습 2부터 시작하는 경우를 위해 다시 설정합니다)
plt.rcParams['font.family'] = {'Darwin': 'AppleGothic', 'Windows': 'Malgun Gothic'}.get(platform.system(), 'NanumGothic')
plt.rcParams['axes.unicode_minus'] = False    # 마이너스 기호 깨짐 방지


def make_spiral(n_per_class=100, n_class=3, noise=0.2, seed=42):
    """나선 모양으로 꼬인 다중 분류용 데이터를 만든다."""
    rng = np.random.RandomState(seed)
    X = np.zeros((n_per_class * n_class, 2))
    y = np.zeros(n_per_class * n_class, dtype=int)
    for c in range(n_class):
        idx = range(n_per_class * c, n_per_class * (c + 1))
        r = np.linspace(0.0, 1, n_per_class)
        t = np.linspace(c * 4, (c + 1) * 4, n_per_class) + rng.randn(n_per_class) * noise
        X[idx] = np.c_[r * np.sin(t), r * np.cos(t)]
        y[idx] = c
    return X, y


X, y = make_spiral()
Y = tf.keras.utils.to_categorical(y, 3)   # 정답을 [1,0,0] 형태로 변환 (CEE에 필요)

print("입력 X의 모양 :", X.shape, "  (샘플 300개, 특성 2개)")
print("정답 y의 모양 :", y.shape, "  클래스별 개수:", np.bincount(y))
print("변환된 Y 예시 :", Y[0], "<- 0번 클래스라는 뜻")

plt.figure(figsize=(5.5, 5.5))
plt.scatter(X[:, 0], X[:, 1], c=y, cmap='brg', s=25, edgecolors='k', linewidths=0.3)
plt.title('나선 데이터 - 클래스 3개'); plt.xlabel('x1'); plt.ylabel('x2')
plt.grid(alpha=0.3); plt.show()
````

## 2-A. 학습률 세 가지로 돌려보기

`2-4 ②`에서 학습률이 **너무 작으면 느리고, 너무 크면 발산**한다고 했죠.

**똑같은 모델**을 학습률만 바꿔 세 번 학습시켜 봅니다. 나머지 조건은 전부 동일합니다.

> `50`이 아니라 `50.0`으로 써야 합니다. Keras는 학습률을 실수로 설정하는 것이 안전합니다.

````python
def train_with_lr(lr, epochs=100, seed=0):
    """학습률만 바꿔서 같은 모델을 학습시키고 손실 기록을 돌려준다."""
    tf.keras.utils.set_random_seed(seed)          # 매번 같은 초기 가중치에서 출발
    model = Sequential([
        Input((2,)),
        Dense(16, activation='relu'),
        Dense(16, activation='relu'),
        Dense(3,  activation='softmax'),
    ])
    model.compile(optimizer=tf.keras.optimizers.SGD(learning_rate=lr),
                  loss='categorical_crossentropy',
                  metrics=['acc'])
    hist = model.fit(X, Y, epochs=epochs, batch_size=32, verbose=0)
    return hist.history


results = {}
for lr in [0.01, 0.5, 50.0]:
    results[lr] = train_with_lr(lr)
    h = results[lr]
    print(f"학습률 {lr:<6} | 첫 손실 {h['loss'][0]:>12,.2f} | 마지막 손실 {h['loss'][-1]:>8.4f} "
          f"| 정확도 {h['acc'][-1]:.3f}")
````

````python
plt.figure(figsize=(7.5, 5))
for lr, style in zip([0.01, 0.5, 50.0], ['-', '-', '--']):
    plt.plot(results[lr]['loss'], style, lw=2, label=f'학습률 = {lr}')
plt.yscale('log')                      # 값의 차이가 너무 커서 로그 눈금으로 봅니다
plt.xlabel('에포크'); plt.ylabel('손실 (로그 눈금)')
plt.title('같은 모델, 학습률만 바꾼 결과')
plt.legend(); plt.grid(alpha=0.3); plt.show()
````

세 곡선이 이렇게 다릅니다.

| 학습률 | 결과 | 무슨 일이 벌어졌나 |
|---|---|---|
| `0.01` | 손실이 **아주 천천히** 줄어듦 | 보폭이 너무 작아 100 에포크로는 부족 |
| `0.5` | 손실이 **잘 줄어듦** | 이 문제에는 적당한 보폭 |
| `50.0` | 손실이 크게 튀거나 높은 값에 멈춤 | 최솟값을 지나쳐 반대편으로 튐 = **발산** |

`50.0`의 정확도와 손실이 에포크에 따라 개선되는지 확인하세요. 난수 추측 수준인 `0.33` 근처에 머물거나 손실이 개선되지 않는다면 학습에 실패한 것입니다.
TensorFlow 버전과 실행 환경에 따라 세부 숫자는 달라질 수 있으므로 **곡선의 모양과 개선 여부**를 보는 것이 핵심입니다.

## 2-B. 시그모이드 vs ReLU — 층별 기울기 직접 출력하기

`2-1 ③`에서 미분값이 1보다 작으면 곱할수록 0에 가까워진다고 했죠.

$$0.25 \times 0.25 \times 0.25 \times \cdots \approx 0$$

말로만 들으면 잘 안 와닿습니다. **은닉층을 6개** 쌓고, 각 층의 기울기가 실제로 얼마인지 찍어 봅니다.

````python
def layer_gradients(activation, n_hidden=6, seed=0):
    """은닉층을 깊게 쌓고, 각 층의 가중치에 전달되는 기울기 크기를 구한다."""
    tf.keras.utils.set_random_seed(seed)
    model = Sequential(
        [Input((2,))]
        + [Dense(16, activation=activation) for _ in range(n_hidden)]
        + [Dense(3, activation='softmax')]
    )
    with tf.GradientTape() as tape:                       # 역전파를 한 번만 수행
        pred = model(X, training=True)
        loss = tf.reduce_mean(tf.keras.losses.categorical_crossentropy(Y, pred))
    grads = tape.gradient(loss, model.trainable_weights)
    return [float(tf.reduce_mean(tf.abs(g))) for g in grads[::2]]   # 가중치만 (편향 제외)


for act in ['sigmoid', 'relu']:
    g = layer_gradients(act)
    print(f"=== {act} ===")
    for i, v in enumerate(g, 1):
        where = "  <- 입력층 쪽" if i == 1 else ("  <- 출력층 쪽" if i == len(g) else "")
        safe_v = max(v, 1e-12)
        bar = "#" * max(1, int(40 + np.log10(safe_v) * 5))
        print(f"  {i}번째 층  {v:.3e}  {bar}{where}")
    print(f"  출력층 기울기가 입력층의 {g[-1] / g[0]:,.1f} 배\n")
````

숫자를 비교해 보세요.

**시그모이드** — 출력층에서 입력층으로 갈수록 **한 층씩 계속 작아지는 경향**을 확인하세요.
출력된 입력층·출력층 기울기와 배율을 직접 비교하면 차이를 확인할 수 있습니다.

기울기가 이만큼 작다는 건, 그 층의 가중치가 **거의 업데이트되지 않는다**는 뜻이에요.
앞쪽 층들은 학습에 사실상 참여하지 못합니다. → **기울기 소실**

**ReLU** — 시그모이드와 비교해 입력층 쪽 기울기가 덜 줄어드는지 확인하세요. ReLU는 양수 구간의 미분값이 1이어서 기울기 소실을 **완화**합니다.

- 시그모이드로 깊게 쌓으면 **입력층 쪽 기울기가 소멸**한다 — 이것이 ReLU가 등장한 이유

# 실습 3) 코드 세 덩어리로 조립하기

`2-6 정리`를 마친 뒤에 진행합니다.

강의에서 오늘 배운 게 **코드 세 덩어리**에 다 들어 있다고 했죠.
그 세 덩어리를 직접 돌려보고, 하나씩 바꿔가며 차이를 확인합니다.

## 3-A. 세 덩어리 그대로 돌려보기

강의자료 `2-6`의 코드와 같은 구조입니다.
입력이 5개가 아니라 **2개**(x1, x2)라는 점만 우리 데이터에 맞췄어요.

````python
tf.keras.utils.set_random_seed(0)

# ① 모델 구조 -- 어떤 층을 몇 개나 쌓을 것인가
model = Sequential()
model.add(Input((2,)))                              # 입력 2개 (x1, x2)
model.add(Dense(16, activation='relu'))             # 은닉층: ReLU  (2-3 ④)
model.add(Dense(16, activation='relu'))
model.add(Dense(3,  activation='softmax'))          # 출력층: 3개 클래스의 확률  (2-3 ⑤)

# ② 학습 방법 설정 -- 무엇으로 재고, 어떻게 고칠 것인가
model.compile(optimizer='adam',                     # 옵티마이저      (2-5 ③)
              loss='categorical_crossentropy',      # 다중 분류라서 CEE (2-2 ②)
              metrics=['acc'])

# ③ 학습 실행 -- 몇 개씩, 몇 번 반복할 것인가
history = model.fit(X, Y, batch_size=32, epochs=100, verbose=0)   # (2-4 ③)

print(f"최종 손실   : {history.history['loss'][-1]:.4f}")
print(f"최종 정확도 : {history.history['acc'][-1]:.3f}")
model.summary()
````

정확도가 **0.9를 넘게** 나왔을 겁니다.

실습 1에서 직선 하나로 XOR도 못 갈랐던 걸 생각하면, **나선 3개를 갈라낸 것**이죠.
층을 쌓고 가중치를 학습시킨 결과입니다.

모델이 실제로 어떻게 나눴는지 그려 봅니다.

````python
def plot_boundary(model, title):
    x_min, x_max = X[:, 0].min() - 0.2, X[:, 0].max() + 0.2
    y_min, y_max = X[:, 1].min() - 0.2, X[:, 1].max() + 0.2
    xx, yy = np.meshgrid(np.linspace(x_min, x_max, 200),
                         np.linspace(y_min, y_max, 200))
    grid = np.c_[xx.ravel(), yy.ravel()]
    Z = model.predict(grid, verbose=0).argmax(axis=1).reshape(xx.shape)

    plt.figure(figsize=(5.5, 5.5))
    plt.contourf(xx, yy, Z, alpha=0.25, cmap='brg')
    plt.scatter(X[:, 0], X[:, 1], c=y, cmap='brg', s=25, edgecolors='k', linewidths=0.3)
    plt.title(title); plt.xlabel('x1'); plt.ylabel('x2'); plt.show()


plot_boundary(model, '모델이 학습한 결정 경계')
````

## 3-B. 옵티마이저만 바꿔보기

`2-5`에서 모멘텀 · RMSProp · Adam을 배웠죠.

**나머지 조건은 전부 고정**하고 **옵티마이저만** 바꿉니다.
학습률도 셋 다 `0.01`로 똑같이 맞췄어요. 그래야 옵티마이저의 차이만 볼 수 있습니다.

````python
optimizers = {
    'SGD (그냥 경사하강법)': tf.keras.optimizers.SGD(learning_rate=0.01),
    'SGD + 모멘텀':          tf.keras.optimizers.SGD(learning_rate=0.01, momentum=0.9),
    'Adam':                  tf.keras.optimizers.Adam(learning_rate=0.01),
}

curves = {}
for name, opt in optimizers.items():
    tf.keras.utils.set_random_seed(0)                      # 같은 출발점
    m = Sequential([Input((2,)),
                    Dense(16, activation='relu'),
                    Dense(16, activation='relu'),
                    Dense(3,  activation='softmax')])
    m.compile(optimizer=opt, loss='categorical_crossentropy', metrics=['acc'])
    h = m.fit(X, Y, epochs=100, batch_size=32, verbose=0)
    curves[name] = h.history

    below = [i + 1 for i, v in enumerate(h.history['loss']) if v < 0.5]
    reach = f"{below[0]:>3} 에포크" if below else "  끝까지 못 감"
    print(f"{name:<22} | 최종 손실 {h.history['loss'][-1]:.4f} "
          f"| 정확도 {h.history['acc'][-1]:.3f} | 손실 0.5 도달 {reach}")

plt.figure(figsize=(7.5, 5))
for name, h in curves.items():
    plt.plot(h['loss'], lw=2, label=name)
plt.axhline(0.5, color='gray', ls=':', lw=1)
plt.xlabel('에포크'); plt.ylabel('손실')
plt.title('같은 모델·같은 학습률, 옵티마이저만 바꾼 결과')
plt.legend(); plt.grid(alpha=0.3); plt.show()
````

같은 모델, 같은 학습률인데 결과가 이렇게 갈립니다.

- **SGD** — 출력에서 손실 0.5에 도달했는지와 속도를 확인합니다
- **SGD + 모멘텀** — 관성을 추가했을 때 SGD보다 빨라졌는지 봅니다
- **Adam** — 손실 0.5에 도달한 에포크를 다른 두 방법과 비교합니다

**옵티마이저를 왜 배웠는지가 여기서 보입니다.** 모델과 데이터가 같아도 옵티마이저에 따라 수렴 속도가 달라질 수 있습니다.

## 3-C. 강의의 '오류 찾기' 코드를 직접 돌려보기

`2-6`의 '생각해봅시다'에서 본 코드입니다.

```python
model.add(Dense(1, activation='softmax'))
model.compile(optimizer='adam', loss='mse')
```

출력 1개에 softmax를 쓰면 **무조건 1이 나온다**고 했죠. 진짜 그런지 돌려 봅니다.

여기서는 이진 분류로 바꿔서 봅니다 — **0번 클래스인가, 아닌가**.

````python
y_binary = (y == 0).astype(float)      # 0번 클래스면 1, 아니면 0

tf.keras.utils.set_random_seed(0)
wrong = Sequential([Input((2,)),
                    Dense(16, activation='relu'),
                    Dense(16, activation='relu'),
                    Dense(1,  activation='softmax')])   # <-- 문제의 그 줄
wrong.compile(optimizer='adam', loss='mse', metrics=['acc'])

h = wrong.fit(X, y_binary, epochs=100, batch_size=32, verbose=0)

print("예측값 10개 :", wrong.predict(X[:10], verbose=0).ravel().round(4))
print(f"손실  : {h.history['loss'][0]:.4f} (처음)  ->  {h.history['loss'][-1]:.4f} (100 에포크 후)")
print(f"정확도: {h.history['acc'][-1]:.3f}")
````

**예측값이 전부 정확히 `1.0`** 입니다. 무엇을 넣어도 1이에요.

손실도 `0.6667`에서 **한 번도 안 움직였습니다.** 100 에포크를 돌렸는데요.

Keras 버전에 따라 *"softmax over axis of size 1 will always return 1. Did you mean sigmoid?"*와 같은 경고 메시지가 나올 수 있습니다.

**이유**는 softmax의 정의 때문입니다.

$$f(x_i)=\frac{e^{x_i}}{\sum_j e^{x_j}}$$

원소가 하나뿐이면 분자와 분모가 같아져서 **결과가 항상 1**입니다.
출력이 입력과 무관하니 기울기도 0이고, 그래서 학습이 **시작조차 안 됩니다.**

이제 직접 고쳐 보세요.

이진 분류를 하려던 것이니 → **활성화 함수**와 **손실 함수**를 바꿔야 합니다. (`2-2 ②`, `2-3 ③`)

````python
tf.keras.utils.set_random_seed(0)
fixed = Sequential([Input((2,)),
                    Dense(16, activation='relu'),
                    Dense(16, activation='relu'),
                    # 여기를 채우세요 — 출력층의 활성화 함수는?
                    Dense(1, activation='sigmoid')])

fixed.compile(optimizer='adam',
              # 여기를 채우세요 — 이진 분류의 손실 함수는?
              loss='binary_crossentropy',
              metrics=['acc'])

h = fixed.fit(X, y_binary, epochs=100, batch_size=32, verbose=0)

print("예측값 10개 :", fixed.predict(X[:10], verbose=0).ravel().round(4))
print(f"손실  : {h.history['loss'][0]:.4f} (처음)  ->  {h.history['loss'][-1]:.4f} (100 에포크 후)")
print(f"정확도: {h.history['acc'][-1]:.3f}")
````

이번엔 예측값이 **0과 1 사이에 흩어져** 있습니다. 확률이 나온 거예요.

위 출력에서 손실이 줄고 정확도가 올랐는지 확인하세요. 실행 환경에 따라 세부 숫자는 달라질 수 있습니다.

바꾼 것은 **단어 두 개**뿐입니다. `softmax` → `sigmoid`, `mse` → `binary_crossentropy`.
