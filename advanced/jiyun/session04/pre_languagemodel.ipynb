## <span style="background-color:#fae0f0; color:#c0156d; padding:4px; border-radius:5px;"> 1. 자연어 처리(NLP)와 언어 모델 </span>

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 1.1 자연어 처리의 특징 </span>
 
- <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;"> 자연어 </span> : 사람이 일상에서 쓰는 언어 (↔ 프로그래밍 언어는 규칙이 엄격하고 해석이 하나로 정해짐)
- 자연어의 어려운 점
    - 이산성 : 단어가 서로 떨어진 기호로 표현됨
    - 순차성 : 단어의 순서에 따라 의미가 달라짐
    - 중의성&모호성 : 같은 문장도 여러 의미로 해석될 수 있음
    - 희소성 : 자주 쓰이지 않는 단어나 표현이 매우 많음
    - 가변성, 진화 : 같은 의미도 다양한 방식으로 표현됨

- NLP(자연어 처리)란?
    - 인간이 일상생활에서 사용하는 자연어를 컴퓨터가 분석하고 이해하며 생성할 수 있도록 하는 인공지능 기술
    - 컴퓨터는 인간과 달리 자연어의 의미를 직접 이해 X -> NLP를 통해 언어를 컴퓨터가 처리할 수 있는 벡터 형태로 변환 & 단어 간 관계와 문맥을 분석

- 자연어 처리 분석 단계
    - 형태소 분석 → 구문 분석 → 의미 분석 → 담화 분석

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 1.2 언어 모델(Language Model) </span>

- <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;"> 언어 모델 </span>
    - 앞서 등장한 단어, 문맥을 바탕으로 다음에 올 토큰의 확률을 예측
    - 문장이 얼마나 자연스러운지 판단하는 모델
    - <span style="color:green">ex. "나는 아침에 밥을 ___" → 먹었다 80%</span>
- 학습 방식에 따른 분류
    - 통계적 언어 모델(SLM)
    - 신경망 언어 모델(NLM)


## <span style="background-color:#fae0f0; color:#c0156d; padding:4px; border-radius:5px;"> 2. 통계적 언어 모델(SLM)과 N-gram </span>
 
### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 2.1 SLM (Statistical Language Model) </span>
 
- 분포 가설 : 비슷한 문맥에서 쓰이는 단어는 비슷한 의미를 가진다
- 조건부 확률로 다음 단어를 예측하며, 확률은 말뭉치에서 횟수를 세어 구함
    - <span style="color:green">ex. "그는 학교에" 뒤에 갔다 70% / 왔다 20% / 뛰었다 10%</span>
- <span style="color:blue">한계 : 학습 데이터에 없는 표현은 확률이 0이 되는 희소성(sparsity) 문제</span>

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 2.2 N-gram </span>
 
-  N-gram
    - 다음 단어를 예측할 때 <span style="color:blue">바로 앞 N-1개 단어만</span> 사용하는 SLM의 단순화 버전
    - <span style="color:green">ex. 4-gram이면 앞 3개 단어만 봄 ("An adorable little boy is spreading ___")</span>
- 한계
    - 희소성 문제가 여전히 존재함
    - 앞의 N-1개만 보기 때문에 멀리 떨어진 문맥을 반영하지 못함

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 2.3 Perplexity (PPL) </span>
 
- 언어 모델의 성능 평가 지표
- <span style="color:blue">값이 낮을수록 좋은 모델</span> (다음 단어를 덜 헷갈려 한다는 뜻)
    - <span style="color:green">ex. SLM의 PPL이 80, NLM의 PPL이 25라면 NLM이 더 좋은 모델</span>

## <span style="background-color:#fae0f0; color:#c0156d; padding:4px; border-radius:5px;"> 3. 신경망 언어 모델(NLM) </span>
 
- <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;"> NLM </span> : 신경망으로 다음 단어를 예측하는 언어 모델
- 핵심은 <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;"> 임베딩(Embedding) </span> : 단어를 연속적인 벡터 공간의 한 점으로 표현
    - 비슷한 단어는 가까운 위치에 놓이므로, 처음 보는 조합도 비슷한 단어의 정보로 예측 가능
    - <span style="color:blue">SLM의 희소성 문제를 완화</span>

| 구분 | SLM | NLM |
|:---:|:---:|:---:|
| 단어 표현 | 이산적인 기호 | 연속 벡터(임베딩) |
| 확률 계산 | 횟수 기반 | 신경망 학습 |
| 희소성 문제 | 심함 | 완화됨 |

## <span style="background-color:#fae0f0; color:#c0156d; padding:4px; border-radius:5px;"> 4. 딥러닝 기반 언어모델 </span>
 
### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 4.1 Transformer 복습 </span>
 
- RNN/LSTM의 한계 : 장기 의존성 문제, 기울기 소실, 병렬 처리 불가
- <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;"> Transformer </span> 구조
    - 입력 : 임베딩 + 위치 인코딩(Positional Encoding)
    - 인코더 : Self-Attention + FFN
    - 디코더 : Masked Self-Attention + Cross-Attention + FFN
    - <span style="color:green">ex. 번역 : "I am a student" → "Je suis étudiant"</span>
### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 4.2 BERT </span>
 
- <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;"> BERT </span> : Transformer의 <span style="color:blue">인코더</span>만 쌓은 <span style="color:blue">양방향</span> 모델
- 사전학습(Pre-training) 후 미세조정(Fine-tuning)으로 목적에 맞게 활용
- 입력 구성
    - 특수 토큰 : [CLS] (문장 시작), [SEP] (문장 구분)
    - 세 가지 임베딩 : 토큰(의미) + 위치 + 문장 구분(Segment)
- 사전학습 방법
    - MLM (Masked Language Model) : 일부 토큰을 [MASK]로 가리고, <span style="color:blue">양쪽 문맥</span>을 보고 맞히기
        - <span style="color:green">ex. "How are [MASK] doing today" → you</span>
    - NSP (Next Sentence Prediction): 두 문장이 이어지는 문장인지(IsNext / NotNext) 판별
- 후속 모델
    - RoBERTa : NSP 제거, 동적 마스킹, 더 많은 데이터와 큰 배치
    - ALBERT : 임베딩 크기 축소, 층 간 파라미터 공유, NSP 대신 SOP 사용

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 4.3 GPT </span>
 
- <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;"> GPT </span> : Transformer의 <span style="color:blue">디코더</span>만 쌓은 모델 (Cross-Attention 없음)
- Masked Self-Attention을 사용하여 <span style="color:blue">앞의 단어만 보고 다음 단어를 예측</span> (자기회귀, Autoregressive)
- GPT-3의 <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;"> In-context Learning </span> : 파라미터를 바꾸지 않고 프롬프트의 예시만으로 과제를 수행
    - Zero-shot (예시 0개) / One-shot (1개) / Few-shot (여러 개)
### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 4.4 BERT vs GPT </span>
 
| 구분 | BERT | GPT |
|:---:|:---:|:---:|
| 구조 | 인코더 | 디코더 |
| 방향 | 양방향 | 단방향 |
| 학습 방식 | MLM, NSP | 다음 토큰 예측 |
| 강점 | 문장 이해 | 문장 생성 |

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 4.5 LLM (Large Language Model) </span>
 
- <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;"> LLM </span> : 방대한 데이터로 학습한 매우 큰 규모의 언어 모델
    - <span style="color:green">ex. GPT, LLaMA, Claude, Gemini</span>
- 동작 방식
    - 학습 : 사전학습(다음 토큰 예측) → 미세조정
    - 추론 : 토큰 → 벡터 → 어텐션 → 다음 토큰 확률 → 생성
- 챗봇형 LLM이 GPT 계열(디코더 기반)인 이유는 <span style="color:blue">다음 토큰을 이어서 생성하는 데 적합한 구조</span>이기 때문
- 한계
    - <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;"> 환각(Hallucination) </span> : 사실이 아닌 내용을 그럴듯하게 생성
    - 학습과 업데이트 비용이 매우 큼 (최신 정보를 반영하기 어려움)

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 4.6 RAG (Retrieval-Augmented Generation) </span>
 
- <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;"> RAG </span> : 질문과 관련된 외부 지식을 <span style="color:blue">검색해서 LLM에 함께 넣어주고</span> 답변을 생성하는 방식
- 구조
    - Query Encoder : 질문을 벡터로 변환
    - Knowledge Retriever : 벡터 DB에서 관련 문서 검색
    - Knowledge-Augmented Generator : 검색된 문서를 참고하여 답변 생성
- 장점 : 최신 정보 반영, 답변의 정확도 향상

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 4.7 CAG (Credibility-aware Generation) </span>
 
- <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;"> CAG </span> : 검색된 문서의 <span style="color:blue">신뢰도(관련성, 최신성, 출처의 신뢰성)</span>를 평가한 뒤 답변을 생성하는 방식
- 과정 : 질문 → 검색 → 신뢰도 평가 → 선별 → 최종 답변
- <span style="color:blue">RAG는 검색된 문서를 그대로 믿지만, CAG는 믿을 만한 문서를 골라 사용</span>
- <span style="color:green">ex. "David Cameron은 어떤 직책을 맡고 있는가?"</span>
    - Reuters (2023.11.13) : 관련성, 최신성, 출처 모두 좋음 → 신뢰도 높음
    - GOV.UK (2023.04.01) : 질문과의 관련성 부족 → 낮음
    - Financial Times (2022.12.17) : 오래된 정보 → 낮음
    - AIGC (2023.12.31) : 출처의 신뢰성 부족 → 낮음
    - RAG는 "현재 공직을 맡고 있지 않다"는 잘못된 답, CAG는 "영국 외무장관"이라는 올바른 답을 생성

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 4.8 LangChain, LangGraph, LangSmith </span>
 
- <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;"> LangChain </span> : LLM을 외부 데이터, API, 도구와 연결해 AI 애플리케이션을 쉽게 개발하게 해주는 오픈소스 프레임워크
    - 등장 배경 : LLM과 RAG를 구현할 때의 구현 복잡성, 외부 시스템 연동, 유지보수의 어려움
    - 추상화 : 복잡한 기능(LLM 호출, 프롬프트 관리, 문서 검색 등)을 간단한 인터페이스로 제공
    - 표준화 : 서로 다른 LLM, DB, 도구를 일관된 방식으로 사용
        - <span style="color:green">ex. ChatGPT에서 Claude로 바꿀 때 모델 연결 부분만 변경</span>
    - 체이닝 : 여러 작업을 순서대로 연결
        - <span style="color:green">ex. 영어 문장 입력 → 한국어 번역 → 요약 → 결과 출력</span>
- <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;"> LangGraph </span> : 작업을 노드(Node)와 엣지(Edge)로 연결해 <span style="color:blue">조건 분기와 반복</span>이 가능한 워크플로를 만드는 프레임워크
    - 필요한 이유 : LangChain의 체이닝은 정해진 순서로만 실행하는 선형 구조라서, 결과에 따라 경로가 달라지거나 이전 단계로 돌아가는 상황을 처리하기 어려움
    - 노드(Node) : LLM, Agent, Tool 등 작업을 수행하는 단위
    - 엣지(Edge) : 노드 사이의 연결로, 다음에 실행할 작업과 경로를 결정
    - 상태(State) : 노드 간에 공유되는 정보 (이전 결과, 대화 문맥 등)
    - 흐름 : User → Router → Agent 1 / Agent 2 ↔ Tool → 반복 후 최종 응답
- <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;"> LangSmith </span> : LLM 애플리케이션의 실행을 추적, 평가, 모니터링하는 개발 및 관리 플랫폼
    - 디버깅 및 추적 / 테스트 및 평가 / 모니터링 / 프롬프트 허브
<span style="color:blue">외부 API(예: OpenAI)를 쓸 때 API 키는 최초 생성 시에만 전체를 볼 수 있고, 유출되면 타인이 내 결제로 사용하게 되므로 절대 공유하면 안 됨</span>

## <span style="background-color:#fae0f0; color:#c0156d; padding:4px; border-radius:5px;"> 5. sLM (small Language Model) </span>
 
### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 5.1 sLM이란? </span>
 
- <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;"> sLM </span> : LLM보다 규모가 작은 언어 모델 (파라미터가 수백만~수십억 개 수준)
- 메모리와 연산 자원을 적게 써서 모바일 등 제한된 환경에 적합, 인터넷 없이 기기 안에서 추론하는 <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;"> 오프라인 추론 </span>도 가능

| 구분 | LLM | sLM |
|:---:|:---:|:---:|
| 모델 크기 | 큼 | 작음 |
| 성능 | 복잡한 추론, 다양한 작업에 강점 | 특정 분야, 단순한 작업에 효율적 |
| 연산 비용 | 높음 | 낮음 |
| 활용 환경 | 주로 고성능 서버 | 서버, 스마트폰, 노트북 등 |
 
### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 5.2 모델 압축 방법 </span>
 
모델 압축 : 성능은 최대한 유지하면서 크기와 연산량을 줄이는 기술
 
- 가지치기(Pruning): 중요도가 낮은 연결이나 뉴런을 제거
    - 연결만 제거하는 비구조적 가지치기, 뉴런·블록 단위로 제거하는 구조적 가지치기
    - 중요한 부분까지 제거하면 성능이 떨어질 수 있어 추가 학습으로 보완
- 양자화(Quantization) : 가중치를 표현하는 숫자의 정밀도를 낮춤
    - <span style="color:green">ex. FP32(32비트) → FP16(16비트) → INT8(8비트)</span>
    - 메모리와 연산량이 줄지만 정밀도 저하로 성능이 약간 떨어질 수 있음
- 저차원 분해(Low-Rank Factorization) : 큰 가중치 행렬을 작은 두 행렬의 곱으로 근사
    - $\Delta W = A \times B$ 형태로 파라미터 수를 줄임
    - <span style="color:blue">LoRA는 저차원 분해 원리를 쓰지만, 모델 전체를 압축하는 것이 아니라 "학습할 가중치의 변경량"만 작은 행렬로 표현해 학습 비용을 줄이는 미세조정 방법</span>
- 지식 증류(Knowledge Distillation) : 큰 Teacher 모델의 지식을 작은 Student 모델에 전달
    - Student가 Teacher의 출력을 따라 하도록 학습

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 6.3 sLM의 예시 </span>
 
- DistilBERT : BERT에 지식 증류를 적용해 크기를 줄인 모델
- Gemma (Google), GPT-4o mini (OpenAI), Llama 소형 버전 (Meta)
- Ministral (Mistral AI) : 온디바이스(On-device) 환경에 적합
- Phi (Microsoft) : 작은 크기에서도 높은 성능을 목표

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 6.4 sLM의 장점과 한계 </span>
 
- **장점**
    - 성능 측면 : 빠른 응답 속도, 제한된 환경에서도 실행 가능, <span style="color:blue">외부 클라우드로 데이터를 보내지 않아 보안에 유리</span>
    - 비용 측면 : 낮은 운영 비용, 적은 전력 소비, 고성능 GPU에 대한 의존도 감소
- **한계**
    - 편향(Bias) : 학습 데이터의 편견이 답변에 반영될 수 있음
    - 복잡한 작업에서 성능 저하 : 새로운 상황, 복잡한 추론, 전문 지식이 필요한 작업에 약함
    - 환각 현상 : LLM과 마찬가지로 발생

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 6.5 결합 추론(Hybrid Inference) </span>
 
- <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;"> 결합 추론 </span> : 작업 난이도에 따라 sLM과 LLM을 나눠 쓰는 방식
    - 간단한 작업(문장 분류, 간단한 요약) → sLM
    - 복잡한 작업(여러 단계 추론, 전문 분석) → LLM
- <span style="color:blue">성능과 비용 사이의 균형</span>을 맞추는 것이 핵심


## <span style="background-color:#fae0f0; color:#c0156d; padding:4px; border-radius:5px;"> 7. 생성 모델의 개념 </span>
 
### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 7.1 분류 모델 vs 생성 모델 </span>
 
- <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;"> 생성 모델 </span> : 학습 데이터의 분포를 학습하고, 그 분포를 따르는 유사한 새로운 데이터를 생성하는 모델

| 구분 | 분류 모델 | 생성 모델 |
|:---:|:---:|:---:|
| 학습 목표 | $P(Y\|X)$ : 데이터 X가 주어졌을 때 정답 Y일 확률 | $P(X)$ : 실제 데이터 X가 나타나는 확률분포 |
| 출력 | 범주 또는 범주별 예측 확률 | 새로운 이미지, 텍스트, 음성 등 |
| 한계 | 처음 보는 이상한 데이터도 기존 라벨 중 하나로 분류함, 새 데이터 생성 불가 | 부자연스러운 결과가 나올 수 있고 품질 평가가 어려움 |
 
### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 7.2 생성 모델의 분류 </span>
 
<span style="color:blue">기준 : 데이터의 확률밀도를 직접 계산할 수 있는가?</span>
 
- 확률밀도(Probability Density) : 학습한 분포에서 특정 데이터가 얼마나 그럴듯한지 나타내는 값
    - <span style="color:green">ex. 강아지를 학습한 모델에서 강아지는 확률밀도가 높고, 자동차는 낮음</span>

**1) 명시적 확률밀도 모델 (Explicit Density Model)** 
- 확률분포를 수학적으로 정의하여 $P(X)$ 값을 수치로 출력할 수 있음
- 다루기 쉬운 밀도 모델 : 확률밀도를 직접 계산 가능
    - NADE, MADE, PixelRNN/PixelCNN : 요소별 조건부 확률을 순차적으로 계산
    - Change of Variables Model (nonlinear ICA) : 변수 변환으로 계산
- 근사적 밀도 모델 : 직접 계산이 어려워 근사 방법을 사용
    - <span style="color:blue">VAE</span> : 변분 추론(Variational Inference)을 활용
        - 변분 추론 : 계산하기 어려운 복잡한 분포를 정규분포 같은 쉬운 분포로 근사하는 방법
    - Boltzmann Machine : 마르코프 연쇄 등을 활용
        - 마르코프 연쇄 : 다음 상태가 현재 상태에만 의존하여 확률적으로 결정되는 과정
        - <span style="color:green">ex. 오늘 맑음 → 내일 맑을 확률 80%, 비 20% / 오늘 비 → 내일 맑을 확률 40%, 비 60%</span>

**2) 암시적 확률밀도 모델 (Implicit Density Model)**
 
- $P(X)$를 수치로 알려주지 않고, 학습한 분포를 바탕으로 <span style="color:blue">새로운 데이터를 샘플링하는 데 집중</span>
- <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;"> 직접 샘플링 (Direct Sampling) </span> : 무작위 노이즈를 모델에 한 번 통과시켜 바로 생성
    - 대표 모델 : <span style="color:blue">GAN</span>
- <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;"> 간접/반복적 샘플링 (Indirect/Iterative) </span> : 여러 단계의 확률적 상태 변화를 반복하여 생성
    - 대표 모델 : GSN (마르코프 연쇄의 상태 전이 규칙을 학습해 반복 적용)

## <span style="background-color:#fae0f0; color:#c0156d; padding:4px; border-radius:5px;"> 8. AE와 VAE </span>
 
### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 8.1 AE (Auto Encoder) </span>
 
- <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;"> AE </span> : 입력 데이터의 핵심 특징을 압축한 뒤, 원본과 최대한 비슷하게 복원하도록 학습하는 신경망
    - 정답 레이블 없이 입력 자체가 복원 목표
    - 활용 : 데이터 압축, 노이즈 제거, 이상 탐지
- <span style="color:blue">기존 인코더-디코더(번역 등)와의 차이 : 입력을 이해해서 다른 결과를 만드는 것이 아니라, 입력 자체를 복원하는 것이 목적</span>
    - <span style="color:green">ex. 숫자 4 이미지 입력 → 숫자 4 이미지 복원</span>
- 구조 : 인코더(Encoder) → 병목(Bottleneck) → 디코더(Decoder)
    - 인코더 : 입력 $x$를 잠재 벡터 $z$로 압축
    - 병목 : 제한된 크기의 잠재공간에 핵심 정보만 담음
    - 디코더 : $z$로부터 복원 데이터 $x'$를 생성
- 압축했다가 복원하는 이유
    - 압축하지 않으면 입력을 그대로 복사하게 되므로, 잠재공간의 크기를 제한하여 <span style="color:blue">가장 중요한 특징만 학습하도록 유도</span>
    - 복원 오차(Reconstruction Error)를 줄이면서 인코더와 디코더가 핵심 구조(Non-trivial Structure)를 학습
### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 8.2 VAE (Variational Auto Encoder) </span>
 
- 등장 배경 : AE는 잠재공간이 불규칙하여 <span style="color:blue">데이터가 없는 빈 영역의 값을 디코더에 넣으면 부자연스러운 결과</span>가 나옴 (새 데이터 생성에 한계)
- <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;"> VAE </span> : 입력의 특징을 <span style="color:blue">고정된 점이 아닌 확률분포(가우시안)</span>로 학습하고, 그 분포에서 샘플링해 새로운 데이터를 생성하는 모델
- 학습 목표 (두 가지를 함께 최소화)
    - 복원 오차(Reconstruction Loss) : 원본과 복원 데이터의 차이
    - KL 발산(KL Divergence) : 두 분포가 얼마나 다른지 나타내는 값 (같으면 0). 인코더가 만든 잠재분포가 <span style="color:blue">표준정규분포에 가까워지도록</span> 유도
- 활용 : 이미지 생성, 데이터 증강, 이상 탐지, 특징 추출
**AE vs VAE**
 
- AE : 입력을 하나의 고정된 잠재 벡터(점)로 표현, 같은 입력은 같은 벡터, 잠재공간에 빈 영역이 생길 수 있음
- VAE : 입력마다 평균($\mu$)과 분산($\sigma^2$)을 구해 확률분포로 표현, 잠재공간이 연속적이고 규칙적이라 값을 조금씩 바꾸면 특징이 자연스럽게 변하고 새 데이터를 만들기 쉬움
- 잠재공간(Latent Space) : 핵심 특징을 숫자 벡터로 표현하는 공간. 비슷한 특징은 가까이 배치될 수 있음

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 8.3 VAE의 구조와 작동 과정 </span>
 
분포 예측 → 샘플링 및 재파라미터화 → 생성의 3단계
 
1. 분포 예측 (Probabilistic Encoder)
    - 입력 $x$로부터 평균 $\mu$와 표준편차 $\sigma$를 예측 ($q_\phi(z|x)$)
    - 하나의 데이터에서 다양한 잠재 벡터를 뽑을 수 있게 하기 위함
2. 샘플링 및 재파라미터화 (Reparameterization Trick)
$$z = \mu + \sigma \odot \epsilon, \quad \epsilon \sim N(0, I)$$
 
   - $\epsilon$ : 표준정규분포에서 뽑은 무작위 노이즈, $\odot$ : 원소별 곱
   - 단순 무작위 샘플링은 기울기(Gradient)를 전달하기 어려우므로, <span style="color:blue">무작위성을 $\epsilon$로 분리</span>해 샘플링 과정에서도 학습이 가능하게 만드는 기법
   - <span style="color:green">ex. $\mu=2$, $\sigma=0.5$일 때 $\epsilon=1$이면 $z=2.5$, $\epsilon=-1$이면 $z=1.5$</span>

1. 생성 (Probabilistic Decoder)
    - $z$를 디코더에 넣어 데이터의 확률분포 $p_\theta(x|z)$를 예측하고 이미지를 복원 또는 생성
    - <span style="color:blue">학습이 끝나면 입력 데이터 없이도 $z \sim N(0, I)$에서 새로 뽑아 디코더에 넣으면 새로운 데이터를 생성</span>

## <span style="background-color:#fae0f0; color:#c0156d; padding:4px; border-radius:5px;"> 9. GAN </span>
 
### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 9.1 GAN의 개념 </span>
 
- <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;"> GAN (Generative Adversarial Network) </span> : 생성자와 판별자 두 신경망이 서로 경쟁하며 실제와 유사한 데이터를 만들도록 학습하는 모델 (2014, Ian Goodfellow)
    - 생성자(Generator, G) : 무작위 노이즈를 입력받아 가짜 데이터를 생성, 판별자를 속이는 것이 목표
    - 판별자(Discriminator, D) : 진짜와 가짜를 입력받아 진짜일 확률을 예측, 둘을 구별하는 것이 목표
- <span style="color:green">ex. 위조지폐범(생성자)과 경찰(판별자). 서로 실력이 늘다가 최적점에서는 생성자의 결과가 진짜와 구분 불가능해지고, 판별자는 찍어서 맞히는 수준(확률 0.5)이 됨</span>
- 학습 후에는 일반적으로 <span style="color:blue">생성자만</span> 사용해 새로운 데이터를 생성
- 활용 : 가상 얼굴 이미지 생성, 이미지 복원, 데이터 증강
### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 9.2 GAN의 구조 </span>
 
- 무작위 노이즈 $z$ → 생성자 $G(z)$ → 가짜 이미지
- 판별자는 실제 이미지 $x$ 또는 가짜 이미지 $G(z)$를 받아 진짜일 확률(0~1)을 출력
    - <span style="color:green">ex. $D(x)=0.9$ (진짜를 진짜로 판단), $D(G(z))=0.1$ (가짜를 가짜로 판단)</span>
- <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;"> 최소-최대 게임 (Minimax Game) </span>
    - 판별자 : 구별 능력을 <span style="color:blue">최대화</span> (진짜는 1, 가짜는 0에 가깝게 출력)
    - 생성자 : 판별자가 구별하지 못하도록 <span style="color:blue">최소화</span> (가짜를 진짜로 판단하게 만듦)
    - 이 경쟁이 반복되며 둘 다 성능이 향상됨

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 9.3 GAN의 장단점 </span>
 
- **장점**
    - 진짜 같은 가짜 생성 : 존재하지 않는 사람의 얼굴 등
    - 높은 이미지 선명도 : <span style="color:blue">VAE는 픽셀 평균값(MSE 등)을 최적화해 이미지가 흐릿해지지만, GAN은 판별자를 속여야 하므로 미세한 질감까지 살려 더 선명함</span>
    - 빠른 생성 속도 : 노이즈를 생성자에 한 번 통과시키면 됨 (확산 모델 대비 빠름)
    - 부드러운 잠재공간 : 잠재 벡터를 조금씩 바꾸면 특징도 자연스럽게 변할 수 있음 (모든 GAN에서 보장되지는 않음)

- **한계**
    - 학습의 불안정성
      - 두 신경망의 균형이 깨지면 품질이 오르내리는 진동현상이 생기고 수렴하지 못할 수 있음(무한루프)
      - 해결 : 학습률·횟수 조절, WGAN-GP
    - 훈련 초기 문제 
      - 초기에 판별자가 가짜를 너무 쉽게 구별하면 생성자에게 전달되는 기울기가 매우 작아짐
      - 해결 : Non-saturating Loss, 판별자 학습 속도 조절
    - 모드 붕괴(Mode Collapse) 
      - 생성자가 판별자를 잘 속이는 특정 형태의 데이터만 반복 생성하여 다양성이 줄어드는 현상
      - 해결 : Minibatch Discrimination, WGAN-GP
    - 평가 지표의 한계
      - 사실성, 선명도, 다양성을 하나의 지표로 평가하기 어려움. 
      - 해결 : FID(실제와 생성 이미지의 특징 분포 유사도), IS(분류 확신도와 다양성), 사람의 시각적 평가 병행
    - 역매핑의 어려움 
      - 기본 GAN은 $z \to$ 이미지 단방향이라 실제 이미지에 해당하는 $z$를 찾기 어려움
      - 해결 : GAN Inversion, BiGAN(인코더 추가)

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 9.4 GAN 적용 사례 </span>
 
- NVIDIA Progressive GAN
    - 해상도를 점진적으로 키우며 학습하여 실존하지 않는 고해상도 가상 인물 얼굴 생성
- Deep Video Portraits (Conditional GAN)
    - 한 사람의 표정·머리 움직임을 다른 사람의 영상에 자연스럽게 반영 (영화 특수효과, 더빙, VR 등)
- Facebook Eye In-Painting (Exemplar GAN)
    - 참고 사진을 활용해 눈을 감은 사진을 눈 뜬 모습으로 복원

## <span style="background-color:#fae0f0; color:#c0156d; padding:4px; border-radius:5px;"> 10. 확산 모델(Diffusion Model) </span>
 
### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 10.1 확산 모델의 개념 </span>
 
- <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;"> 확산 모델이란? </span>
    - 이미지에 노이즈를 추가하는 과정을 학습하고, <span style="color:blue">무작위 노이즈에서 노이즈를 점차 제거하며 새로운 이미지를 생성</span>하는 모델
- 두 핵심 과정
    - 순확산(Forward Diffusion) : 원본에 노이즈를 점차 추가
    - 역확산(Reverse Diffusion) : 노이즈를 단계적으로 제거하며 이미지 생성
    - $x_0$ : 이미지 데이터, $x_T$ : 노이즈가 충분히 추가된 상태
- 목적 : 데이터의 특징과 분포를 학습하여 새로운 데이터를 자연스럽고 다양하게 생성
- <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;"> 모호성 모델링 (Ambiguity Modeling) </span> : 하나의 조건에서도 여러 결과가 가능하다는 불확실성을 확률적으로 표현
    - <span style="color:green">ex. "공원에 앉아 있는 강아지"라는 같은 조건이어도 품종, 털 색, 자세, 배경이 다른 여러 이미지가 생성됨</span>

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 10.2 순확산과 역확산 </span>
 
- **순확산** : $x_0 \to x_1 \to x_2 \to \cdots \to x_T$
    - 가우시안 노이즈(정규분포를 따르는 무작위 잡음)를 조금씩 추가하여 최종적으로 무작위 노이즈 상태로 만듦
    - 목적 : 다양한 수준의 노이즈 데이터를 만들어 이후 노이즈 제거 방법을 학습시키기 위함
- **역확산** : $x_T \to \cdots \to x_0$
    - 무작위 노이즈에서 시작하여 <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;"> Denoising U-Net </span>이 현재 이미지의 노이즈를 예측하고 조금씩 제거, 이를 반복하여 새로운 이미지 생성

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 10.3 확산 모델의 장단점 </span>
 
- **장점**
    - 노이즈를 단계적으로 제거하여 GAN보다 정교하고 다양한 이미지 생성에 유리
    - 생성자-판별자 경쟁이 아니라 실제 노이즈와 예측 노이즈의 차이(MSE)를 줄이는 방식이라 비교적 학습이 안정적
    - 다양성 확보 : 무작위 노이즈에서 시작하므로 같은 조건에서도 다른 결과 생성
- **단점**
    - 여러 단계의 반복이 필요해 느림 (GAN은 한 번에 생성)
    - 높은 계산 비용 : 많은 데이터와 연산, 고성능 GPU 필요
    - 순확산/역확산의 단계적 구조에 의존하며, 노이즈 제거 단계 수와 방식에 따라 속도와 품질이 달라져 적절한 설계가 필요

<span style="color:blue">GAN은 빠르지만 학습이 불안정하고, 확산 모델은 품질과 안정성이 좋지만 느리다</span>
