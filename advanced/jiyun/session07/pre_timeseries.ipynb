## <span style="background-color:#fae0f0; color:#c0156d; padding:4px; border-radius:5px;"> O1.시계열 데이터 개념 </span>

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 1. 시계열 데이터 정의 </span>

1. <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;"> 시계열 데이터 (Time Series Data) </span>
   - 시간 순서대로 일정한 간격으로 관측된 데이터
2. **시계열 데이터의 조건**
   1. 시간 순서가 의미를 가짐 
   2. 일정한 관측 간격
- 횡단면 데이터와 구분: 횡단면은 한 시점에서 여러 대상을 관측한 것

3. **단변량 vs 다변량 시계열**
- <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;"> 단변량 시계열 </span>: 하나의 변수만 시간에 따라 기록 
    - 예: 서울의 일별 기온, 한 카페의 일별 아메리카노 판매량
- <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;"> 다변량 시계열 </span>: 여러 변수를 동시에 기록 
    - 예: 전력 사용량 : 기온 + 습도 + 공휴일 여부 / 주가 : 거래량, 환율, 금리 등..
- 다변량이라고 항상 좋은 것은 X -> 미래 시점에도 그 값을 알 수 있는 변수인지 따져보고 넣어야함!

4. **시계열 분석과 목적**
- 시계열 분석: 시계열 데이터에서 패턴을 찾아내고 이를 바탕으로 미래 값을 예측 => 주 목표 : 미래 데이터 예측
- 회귀분석과의 차이: 순서가 의미를 갖는 데이터라는 점에서 차이 발생
    - 어제의 값이 오늘의 값에 영향 O -> 일반 모델이 갖는 가정 깨짐

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 2. 시계열 데이터 특징 </span>

1. **시간 의존성 (Temporal Dependence)**: 현재 값이 과거 값에 의존
    - 가까운 과거일수록 영향이 큼ㅇ
2. **계절성(Seasonality) & 추세(Trend)**: 반복 패턴과 장기적 방향성 존재
3. **자기상관성 (Autocorrelation)**: 가까운 시점끼리 상관이 높음
4. <span style="color:blue">데이터 누수(Data Leakage) 주의</span>: 미래 정보가 학습에 섞이면 안 됨. 시계열은 반드시 시간 순서대로 Train/Test 분할

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 3. 시계열 데이터 분석요인 </span>

- <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;"> 추세 (Trend) </span>: 장기적인 증가/감소 경향성을 나타내는 요인
- <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;"> 계절성 (Seasonal) </span>: 규칙적인 주기로 반복되는 패턴 (예: 매년 여름 아이스크림 판매 증가)
    - 주기가 고정 (12개월, 7일 등)
- <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;"> 순환 (Cycle) </span>: 고정 주기 없이 나타나는 장기 변동 (예: 경기순환)
    - 주기가 불규칙하고 긺 (2~10년 등)
- <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;"> 불규칙 (Irregular) </span>: 추세·계절·순환으로 설명되지 않는 무작위 변동
    - 이상적으로는 위아래 랜덤하게 흩어져 있어야함. 유독 크게 튀는 값은 이상치일수도!!

## <span style="background-color:#fae0f0; color:#c0156d; padding:4px; border-radius:5px;"> O2.시계열 데이터 전처리 </span>

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 1. 결측치/이상치 탐지 및 처리 </span>

> 시계열 특성이 왜곡되지 않도록, 시계열 특성을 반영한 전처리가 필요함

- **전처리 프로세스**
    - 데이터확인 -> 결측치탐지&처리 -> 이상치탐지&처리 -> 평활화 -> 요소분해 -> 정상성판단&확보

#### 1) 결측치 처리

> 시계열에서는 결측치 말못 지우면 시점 비교가 불가능해지고 시간 흐름도 무시됨 ㅜㅜ

1. **기존 결측치 처리 방법**
- **삭제**: 결측이 적고 무작위일 때
- **단순 대치**: 평균, 중앙값, 최빈값 등으로 채움
- **이동평균 대치**: 주변 N개의 평균으로 채움
- **머신러닝 기법**: KNN, 회귀 모델 등으로 결측값 예측하여 채움

2. **시계열에서 중요한 결측치 처리 방법**
- **보간법**: 시간 흐름을 고려하여 채움
  - 선형 보간: 두 점을 직선으로 연결
      - 변화완만, 일정한 데이터에 사용 / 잘 모르겠을 땐 요걸루
  - 다항식 보간: 곡선으로 연결 (과적합 주의)
      - 곡선형태로 부드럽게 변하는 데이터에 사용
  - 스플라인 보간: 구간별 저차 다항식을 매끄럽게 연결
      - 곡선형태로 부드럽게 변하는 데이터에 사용
  - 조각별 상수 보간 (Piecewise Constant): 가장 가까운 점의 값 사용
      - 값이 계단처럼 유지되는 데이터에 사용 (가격표, 설정값 등)

#### 2) 이상치 탐지

> 값의 크기만 참고 X -> 추세, 계절성을 고려해 판단 필요!

1) **이상치 탐지 방법**

| 방법 | 핵심 |
|------|------|
| IQR (사분위수 범위) | Q1 - 1.5×IQR ~ Q3 + 1.5×IQR 범위 밖이면 이상치 |
| Z-score | 평균에서 표준편차 2~3배 이상 벗어나면 이상치 |
| STL 분해 | 추세·계절 제거 후 나머지(Residual)에서 탐지 |
| 분류 및 회귀 트리 (CART) | 트리 모델로 이상 패턴 분류 |
| 클러스터링 기반 이상 탐지 | 군집에서 멀리 떨어진 점을 이상치로 판단 |
| 오토인코더 | 복원 오차가 큰 데이터를 이상치로 판단 |

2) **이상치 처리**
- **삭제**
    - 그냥 지우면 관측 간격이 깨지므로 값을 비워서(NaN) 결측치로 바꾼 뒤 다시 채워줌
- **단순 대치**
    - 평균, 중앙값 등으로 바꾸기
    - 간단하지만 추세와 계절성을 무시ㅜㅠ
- **보간**
    - 이상치를 결측치로 바꾼 뒤, 앞뒤 값으로 보간해서 채우기
    - 시간 흐름 반영 가능해 시계열에서 많이 사용!


### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 2.2 평활화 (Smoothing) </span>

> 노이즈를 줄이고 추세·패턴을 명확히 하는 기법 -> EDA에서 주로 사용!

1. **이동평균 (Moving Average)**
> 일정한 개수의 연속된 데이터를 묶어 평균 내고, 그 묶음을 한 칸씩 옮겨가며 평균을 계속 계산하는 방법 => 젤 대표적인 평활화 방법
   - <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;"> SMA (단순 이동평균) </span>: N개 값의 단순 평균. 모든 값에 동일 가중치
   - <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;"> WMA (가중 이동평균) </span>: 최근 값에 더 큰 가중치 부여
   - <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;"> EMA (지수 이동평균) </span>: 최근 값에 지수적으로 큰 가중치. 과거로 갈수록 급격히 감소

1. **N에 따른 차이**
> N = 한번에 평균을 낼 데이터의 개수 (묶음의 크기)
   - N이 크면: 더 매끄럽지만 추세 변화 반영이 느림
   - N이 작으면: 반응은 빠르지만 노이즈가 많이 남음

1. **사용 목적**
   - 노이즈 제거, 추세 파악
   - 계절성 제거
   - 결측치, 이상치 대치
   - 예측의 기초자료

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 2.3 요소 분해 (Decomposition) </span>

> 여러 요인들이 섞여있는 시계열을 각각의 요인으로 다시 나누는 과정!
> 시계열을 추세(T) + 계절(S) + 순환(C, 찾아내기 어려워서 보통은 추세에 포함시킴) + 잔차(R)로 분리

1. **사용 목적**
   - 각 패턴을 명확히 이해
   - 변동의 원인 분석
   - 더 정교한 분석을 위해!

2. **분해 모형**
  - **가법 모형 (Additive)**: Y = T + S + R
    - 계절 변동의 크기가 일정할 때 사용

  - **승법 모형 (Multiplicative)**: Y = T × S × R
    - 계절 변동이 추세에 비례하여 커질 때 사용

3. **분해 방법**
   - <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;"> Seasonal Decompose </span>: 이동평균(MA) 기반
       - 단순하지만 양 끝 결측 발생
       - 이상치에 민감
   - <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;"> STL (Seasonal and Trend decomposition using LOESS) </span>: LOESS(국소 회귀) 기반
     - Inner Loop: 추세와 계절을 반복 추출
     - Outer Loop: 이상치에 낮은 가중치 부여 -> 이상치에 강건

## <span style="background-color:#fae0f0; color:#c0156d; padding:4px; border-radius:5px;"> 3. 정상성 판단 </span>

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 3.1 정상성 개념 </span>

> **정상성** : 시간이 지나도 통계적 성질(평균, 분산 등)이 변하지 않는 성질
- <span style="color:blue">대부분의 시계열 모형(AR, MA, ARMA)은 정상성을 가정</span>

1. **정상 vs 비정상 시계열**

|  | 정상 시계열 | 비정상 시계열 |
|--|-----------|-------------|
| 평균 | 일정 | 시간에 따라 변함 |
| 분산 | 일정 | 시간에 따라 변함 |
| 패턴 | 평균 회귀 | 추세/계절성 존재 |
| 예시 | 등락률, 진동 값 | 주가, GDP, 항공 승객 수 |

2. **강한 정상성 vs 약한 정상성**
- <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;"> 강한 정상성 </span>: 모든 시점 조합의 결합분포가 시간 이동에 불변. 현실에서 거의 없음
- <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;"> 약한 정상성 </span>: 실무에서 사용하는 정상성. 3가지 조건:
  - ① 평균이 일정 (μ)
  - ② 분산이 일정 (σ², 유한)
  - ③ 자기공분산이 시차(h)에만 의존 (어느 시점이냐가 아니라 "몇 칸 떨어져 있는가"에만 의존)

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 3.2 정상성 판단 (시각적) </span>

1. **그래프로 판단**
- 정상 시계열: 수평 평균 중심으로 오르내림, 폭(분산)이 일정, 평균 회귀 성질, 뚜렷한 주기 패턴 없음
- 비정상 시계열: 추세(우상향/우하향), 계절성(반복 패턴), 분산 변화(나팔 모양으로 벌어짐)

2. **ACF 그래프로 확인**
- 정상 -> 막대가 빠르게 0에 근접
- 비정상 -> 막대가 천천히 감소 (과거의 영향이 오래 남음)

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 3.3 단위근 (Unit Root) </span>

1. **수학적 의미**: $x^n = 1$의 근 (n제곱해서 1이 되는 수)

2. **시계열에서의 의미**: $Y_t = \phi Y_{t-1} + \varepsilon_t$ 에서
- φ = 1이면 -> 충격이 100% 그대로 남아 쌓임 -> 평균으로 안 돌아옴 + 분산이 계속 커짐 -> **비정상**
- |φ| < 1이면 -> 충격이 점점 줄어 사라짐 -> **정상**

3. **φ = 0.5 vs φ = 1 비교** (충격 100이 들어왔을 때)
- φ = 0.5: 100 -> 50 -> 25 -> 12.5 -> ... (절반씩 줄어 사라짐)
- φ = 1: 100 -> 100 -> 100 -> 100 -> ... (줄어들지 않고 그대로 쌓임)
  - *왜 φ = 1을 "근이 1"이라 부르나?*
    - 특성방정식에서 z = 1/φ 이므로, φ = 1이면 z = 1 -> "근이 1"

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 3.4 정상성 검정 </span>

#### 1) ADF 검정 (Augmented Dickey-Fuller)
- H₀: 단위근 있음 (비정상) / H₁: 단위근 없음 (정상)
- p < 0.05 -> 귀무가설 기각 -> **정상**
- 관점: "비정상이 아님을 증명" -> 과거 값의 영향이 사라지면 정상

#### 2) KPSS 검정
- H₀: 정상 / H₁: 비정상
- p < 0.05 -> 귀무가설 기각 -> **비정상**
- 관점: "정상임을 반증" -> 평균이 흔들리면 비정상

1. **재판 비유**
- ADF: "비정상이다"를 기본 가정 -> 증거가 충분하면 "정상" 판결
- KPSS: "정상이다"를 기본 가정 -> 증거가 충분하면 "비정상" 판결
- <span style="color:green">두 검정은 귀무가설이 반대이므로 함께 사용하여 판단</span>

2. **두 검정 조합 해석**

| ADF | KPSS | 결론 |
|-----|------|------|
| 정상 | 정상 | 정상 시계열 -> 바로 모델링 |
| 비정상 | 비정상 | 비정상 시계열 -> 차분 후 재검정 |
| 비정상 | 정상 | 추세 정상 -> 추세 제거 필요 |
| 정상 | 비정상 | 차분 정상 -> 차분 필요 |

3. **엇갈리는 이유**
- 추세 정상 시계열 (예: 매년 일정하게 늘어나는 매출)
  - ADF -> 비정상 (단위근과 구분 못함)
  - KPSS -> 정상 (추세 고려 시 안정적)
- 차분 정상 시계열 (예: 주가처럼 충격이 누적)
  - ADF -> 정상으로 나올 수 있음 (기각하는 경우 있음)
  - KPSS -> 비정상 (무작위로 떠도는 부분 감지)

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 3.5 정상성 확보 방법 </span>

#### 1) 추세 제거: 차분 (Differencing)
- <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;"> 차분 </span>: $\Delta Y_t = Y_t - Y_{t-1}$ (오늘 값 - 어제 값 = 오늘 새로 생긴 변화만 남음)
- 예: 판매량 100 -> 112 -> 119 -> 133 -> 140 -> 차분 결과: +12, +7, +14, +7 (평균 일정해짐)
- 단위근이 사라지는 원리: 오늘 값에서 어제 값을 빼면 오늘 새로 생긴 일만 남음
- **1차 차분**: 기본적인 추세 제거
- **2차 차분**: 오르는 속도 자체가 빨라질 때 (보통 2차까지만 하면 충분)
- <span style="color:blue">역차분: 예측 후 반드시 원래 단위로 복원해야 함</span>

#### 2) 계절성 제거: 계절 차분
- <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;"> 계절 차분 </span>: $\Delta_m Y_t = Y_t - Y_{t-m}$ (m: 계절 주기)
- "작년 같은 때와 비교"로 계절 효과 상쇄

#### 3) 분산 안정화: 로그 변환
- <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;"> 로그 변환 </span>: 큰 값은 많이, 작은 값은 조금 줄임
- 10, 100, 1000 -> 1, 2, 3
- 곱하기를 더하기로 바꿔서 승법 -> 가법으로 변환
- 한계: 0이나 음수에는 적용 불가 (1을 더해서 해결), 모든 분산 변화를 해결하지는 못함 (비례 형태에만 효과적)

## <span style="background-color:#fae0f0; color:#c0156d; padding:4px; border-radius:5px;"> 4. 단변량 시계열 분석 </span>

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 4.1 값(y)과 충격(ε)의 구분 </span>

- <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;"> 값 (y) </span>: 그날 관측된 최종 결과 (과거의 영향이 섞여 있음)
- <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;"> 충격 (ε) </span>: 그날 새로 생긴 예측 불가능한 변동 (매일 하나씩 들어오고, 예측 시 0으로 둠)
- 충격이 아닌 것: 주말에 많이 팔림 / 꾸준한 성장 / 공지된 할인 (미리 알 수 있으므로)

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 4.2 안정적 시계열 모형 </span>

#### 1) AR(p) 자기회귀 모형
- 과거 p개의 **값(y)**으로 현재를 설명
- $y_t = c + \phi_1 y_{t-1} + \cdots + \phi_p y_{t-p} + \varepsilon_t$

**AR(1) 상세**
- $y_t = c + \phi_1 y_{t-1} + \varepsilon_t$
- φ₁ = 어제 판매량의 몇 %가 오늘에 반영되는가
- c = 상수 (평소 수준 = c ÷ (1 - φ₁))
- εₜ = 오늘의 충격

**입소문 가게 예시** (방송 출연 -> 충격 100, 계수 φ = 0.8)
- 1일차: 100 (방송 직후)
- 2일차: 80 (어제 손님의 80%)
- 3일차: 64 -> 4일차: 51 -> 5일차: 33 ...
- 어제 값 안에 그저께 값이 포함 -> 영향이 끝없이 옅어지며 이어짐

#### 2) MA(q) 이동평균 모형
- 과거 q개의 **충격(ε)**으로 현재를 설명
- $y_t = \mu + \varepsilon_t + \theta_1 \varepsilon_{t-1} + \cdots + \theta_q \varepsilon_{t-q}$

**MA(1) 상세**
- $y_t = \mu + \varepsilon_t + \theta_1 \varepsilon_{t-1}$
- θ₁ = 어제 충격의 몇 %가 오늘에 반영되는가

**방송효과 가게 예시** (방송 출연 -> 충격 100, 계수 θ = 0.8)
- 1일차: 100 (방송 본 사람)
- 2일차: 80 (방송 본 사람 중 하루 늦게 온 80)
- 3일차: 0 -> 4일차: 0 (정해진 날수 뒤 끊김!)

**AR vs MA 비교**

| 구분 | AR | MA |
|------|-----|-----|
| 비유 | 손님이 어제 손님 수를 보고 옴 | 손님이 어제 방송을 보고 옴 |
| 영향 지속 | 끝없이 옅어지며 이어짐 | 정해진 날수(q일) 뒤 끊김 |

#### 3) 정상성 조건
- **AR**: 계수 조건 필요
  - AR(1): |φ| < 1
  - AR(p): 특성방정식 $1 - \phi_1 z - \cdots - \phi_p z^p = 0$의 모든 근 |z| > 1
- **MA**: <span style="color:blue">항상 정상</span> (충격은 q일 뒤 끝나므로)

#### 4) 가역성 조건
- **AR**: 항상 가역 (충격 = 값 - 계수 × 어제 값, 모두 관측 가능)
- **MA**: 계수 조건 필요
  - MA(1): |θ| < 1
  - MA(q): 특성방정식 $1 + \theta_1 z + \cdots + \theta_q z^q = 0$의 모든 근 |z| > 1
- 가역성이 필요한 이유: MA는 보이지 않는 과거 충격(ε)이 필요 -> 값(y)으로 역산해야 -> 계수 ≥ 1이면 오차가 눈덩이처럼 불어남

**특성방정식**: 모형 식의 계수를 옮겨 적어 만든 방정식 (오늘 -> 1, 어제 -> z, 그저께 -> z²)

<span style="color:green">데이터 정상성 검정(ADF/KPSS) ≠ 모델 정상성 조건(특성방정식): 보는 대상이 다름!</span>

#### 5) ARMA(p, q) 모형
- AR + MA 결합: $y_t = c + \phi_1 y_{t-1} + \cdots + \varepsilon_t + \theta_1 \varepsilon_{t-1} + \cdots$
- 충격은 하루에 하나, 여파가 두 길(값 경로 + 충격 경로)로 넘어옴

**ARMA 예시** (AR 계수 0.5, MA 계수 0.4, 충격 100)
- AR만: 100 -> 50 -> 25 -> 12.5
- MA만: 100 -> 40 -> 0 -> 0
- ARMA: 100 -> 90 -> 45 -> 22.5
  - 2일차 90 = 입소문 50(100×0.5) + 방송 늦게 온 40(100×0.4)
  - 3일차 45 ≠ 25 -> 방송 본 40명도 입소문을 내므로 추가 영향 발생
- 적은 계수로 복잡한 패턴을 설명 가능

**ARMA 특수 경우**: ARMA(p,0) = AR(p), ARMA(0,q) = MA(q), ARMA(0,0) = 백색잡음

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 4.3 ACF / PACF로 차수 결정 </span>

- <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;"> ACF (자기상관함수) </span>: $\rho(k) = Cov(y_t, y_{t-k}) / Var(y_t)$
  - 중간 날짜를 거쳐 온 영향까지 합쳐서 측정
  - 점선(±1.96/√n) 안쪽이면 0으로 봄 -> **q** 결정에 사용
- <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;"> PACF (부분자기상관함수) </span>: $\phi_{kk} = Corr(y_t, y_{t-k} | y_{t-1}, \ldots, y_{t-k+1})$
  - 중간 시점의 영향을 빼고 직접 상관만 측정 -> **p** 결정에 사용

1. **패턴 정리**

| 모형 | ACF | PACF |
|------|-----|------|
| AR(p) | 서서히 감소 (지수적/진동) | 시차 p 이후 절단 |
| MA(q) | 시차 q 이후 절단 | 서서히 감소 |
| ARMA(p,q) | 서서히 감소 | 서서히 감소 |

- 규칙: 막대가 급격히 줄어들기 직전의 시차를 고름

2. **실제 그래프 해석 예시**
1. ACF가 천천히 감소 + PACF가 시점 1 이후 신뢰구간 안 -> **AR(1)**
2. ACF가 시점 1 이후 신뢰구간 안 + PACF가 천천히 진동 감소 -> **MA(1)**
3. ACF가 시점 1 이후 안 + PACF가 시점 1 이후 안 -> **ARMA(1,1)**

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 4.4 불안정 시계열 모형 </span>

- 돌아갈 "평소"가 없는 시계열
- 추세 -> 차분(d) -> ARIMA / 계절성 -> 계절차분(D) -> SARIMA
- 체온 vs 키 비유: 체온은 평소(36.5℃)가 있어서 돌아옴, 키는 평소가 없음 (하지만 "1년간 자란 길이"에는 평소가 있음)
- 프로세스: 비정상 -> 정상성 확보(차분/계절차분) -> ARMA 적용

#### 1) ARIMA(p, d, q)
- ARMA에 차분(d)을 추가한 모형

1. **3단계 흐름**
   1. 차분으로 증감 계산
   2. 증감을 ARMA로 예측
   3. 예측 증감을 마지막 실제 값에 더해 복원               

2. **ARIMA(1,1,1) 예시**
   - 판매량: 100 -> 103 -> 104 -> 108 -> 110
   - 증감(차분): +3, +1, +4, +2
   - ARMA가 증감 +2.5 예측 -> 110 + 2.5 = **112.5**

3. **ARIMA 특수 경우 5가지**
   1. ARIMA(0,0,0) = 백색잡음
   2. ARIMA(p,0,0) = AR(p)
   3. ARIMA(0,0,q) = MA(q)
   4. ARIMA(0,1,0) 상수 없음 = <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;"> 랜덤워크 </span>
      - $Y_t = Y_{t-1} + \varepsilon_t$ (최선의 예측 = 오늘 값)
      - 도출: 2단계 식에서 상수/AR/MA 모두 빠지면 $w_t = \varepsilon_t$ -> $y_t = y_{t-1} + \varepsilon_t$
      - 주가, 환율 등
   5. ARIMA(0,1,0) 상수 있음 = <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;"> 표류 포함 랜덤워크 </span>
      - $Y_t = Y_{t-1} + c + \varepsilon_t$ (한 방향으로 흘러감)
      - 예: 표류 2.5 -> 오늘 110 -> 내일 112.5 -> 모레 115 -> 글피 117.5

#### 2) SARIMA(p, d, q)(P, D, Q)m
- ARIMA에 계절 차분 + 계절 AR + 계절 MA를 추가

1. **3단계 흐름**
   1. 계절 차분 (작년 같은 달 대비 증감)
   2. ARMA + 계절 항으로 예측
   3. 작년 같은 달 값에 증감을 더해 복원

2. **붕어빵 가게 예시**
   - 작년 12월 판매량: 900
   - 10월, 11월 각각 작년 대비 +50씩 늘었음
   - 12월도 +50 예측 -> 900 + 50 = **950**

3. **하이퍼파라미터 7개**
   - p: 비계절 AR 차수 (지난달 값 참고)
   - d: 비계절 차분 횟수 (전월 대비 증감)
   - q: 비계절 MA 차수 (지난달 충격 여파)
   - P: 계절 AR 차수 (작년 같은 달 값 참고)
   - D: 계절 차분 횟수 (작년 같은 달 대비 증감)
   - Q: 계절 MA 차수 (작년 같은 달 충격 여파)
   - m: 계절 주기 (월별=12, 분기별=4, 요일=7)

4. **SARIMA(1,1,1)(1,1,1)12 해석**
   - p=1: 지난달 값 참고
   - d=1: 전월 대비 증감으로 변환
   - q=1: 지난달 충격 여파 반영
   - P=1: 작년 같은 달 값 참고
   - D=1: 작년 같은 달 대비 증감으로 변환
   - Q=1: 작년 같은 달 충격 여파 반영
   - m=12: 12개월 주기

#### 3) 계절성 ACF/PACF
- ACF가 물결 모양으로 오르내리며 시차 12, 24, 36에서 솟으면 -> 주기 12인 계절성 신호
- 물결 높이가 안 줄면 "평소"가 없다는 뜻
- PACF도 시차 12 근처까지 띠 밖 막대 -> p, q를 읽을 수 없음 -> <span style="color:blue">계절 차분을 먼저 해야 함!</span>

#### 4) 하이퍼파라미터 결정법
- **d, D 결정**: ADF 검정 -> 비정상이면 차분 반복 (대부분 0 또는 1)
- **p, q 결정**: ACF/PACF로 범위 추정 -> AIC/BIC 그리드 서치 (0~3 범위)
- 안정적 시계열에서도 동일 방법 사용 가능

## <span style="background-color:#fae0f0; color:#c0156d; padding:4px; border-radius:5px;"> 5. 다변량 시계열 분석 </span>

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 5.1 VAR (Vector Autoregression) </span>

> AR을 변수 여러 개로 확장한 모형<br>
> 각 변수를 자기 자신의 과거 + 다른 변수의 과거로 예측

1. **VAR(1) 수식**
   - $X_t = a_{11}X_{t-1} + a_{12}Y_{t-1} + \varepsilon_{x,t}$
   - $Y_t = a_{21}X_{t-1} + a_{22}Y_{t-1} + \varepsilon_{y,t}$
   - 대각 계수($a_{11}, a_{22}$): 자기 영향 (AR과 동일)
   - 비대각 계수($a_{12}, a_{21}$): 상호 영향 (VAR에서 새로 추가)

2. **판매량 + 광고비 예시**
   - 판매량 = 어제 판매량의 60% + 어제 광고비의 30% + 충격
   - 광고비 = 어제 광고비의 50% - 어제 판매량의 20% + 충격
   - 계수 행렬: 0.6, 0.3 / -0.2, 0.5
   - 0.3 = "광고가 판매를 끌어올림"
   - -0.2 = "잘 팔리면 광고를 줄임"

3. **적합/부적합 상황**
   - 적합
     - 양방향 영향
     - 반응 분석 필요
     - 정상(또는 차분하면 정상)
     - 변수 적고 데이터 충분
   - 부적합
     - 한 방향 영향만 (기온->판매량 같은 경우 단변량 + 외생변수)
     - 변수 많고 데이터 적음 (변수 2개->계수 4개, 3개->9개로 급증)
     - 비선형 강함 (->딥러닝)

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 5.2 Granger 인과관계 검정 </span>

- "X의 과거 -> Y의 미래 예측에 도움이 되는가?"를 통계적으로 검정
- <span style="color:green">Granger 인과 ≠ 진짜 인과! 예측 가능성만 의미</span> (닭이 울면 해가 뜬다 비유)
- 제한 모델(Y 과거만) vs 비제한 모델(Y 과거 + X 과거) 비교
- H₀: X는 Y 예측에 도움 안 됨 -> p < 0.05이면 VAR에 포함
- 양방향으로 각각 검정해야 함

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 5.3 충격 반응 함수 (IRF) </span>

> 한 변수에 충격 1회 -> 다른 변수가 시간에 따라 어떻게 반응하는지 추적

1. **광고비 충격 100 -> 판매량 반응 예시** (AR=0.6, 비대각=0.3)
  - 2일차: 30 (어제 광고비 100 × 0.3)
  - 3일차: 33 (어제 판매량 30 × 0.6 = 18 + 어제 광고비 50 × 0.3 = 15)
  - **가장 큰 반응은 3일차** (다음 날이 아님!)
  - 이후 점차 소멸
  - 계수표만으로는 알 수 없는 정보를 IRF가 보여줌

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 5.4 예측오차 분산 분해 </span>

> 예측 오차를 "어느 변수의 충격 탓인지" 비율로 분해

1. **예시**
  - 1일 뒤: 판매량 100% / 광고비 0% (내일의 충격만 모름)
  - 2일 뒤: 94% / 6% (내일의 광고비를 모르므로)
  - 5일 뒤: 83% / 17% (예측 기간이 길수록 타변수 몫이 증가)

2. **IRF vs 분산 분해**
  - IRF = "어떻게 반응하는가"
  - 분산 분해 = "누구의 영향이 얼마나 큰가"

## <span style="background-color:#fae0f0; color:#c0156d; padding:4px; border-radius:5px;"> 6. 딥러닝을 활용한 시계열 분석 </span>

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 6.1 등장 배경 </span>

**전통 통계 모델의 한계**
- 데이터가 많으면 처리 속도 느림
- 고정 크기 데이터만 가능
- 비선형 패턴 포착 어려움

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 6.2 RNN 계열 </span>

| 모델 | 특징 | 장점 | 단점 | 적합 상황 |
|------|------|------|------|-----------|
| RNN | 이전 시점 정보를 현재로 전달 | 간단, 빠름 | 장기 의존성 학습 불가 | 짧은 시계열 |
| LSTM | 게이트로 장기 기억 유지 | 장기 의존성 학습 | 복잡, 느림 | 분기별 재무 |
| GRU | LSTM 단순화 | LSTM보다 빠름 | 장기 의존성 LSTM보다 약함 | 실시간 센서 |

- **적합**: 순차적 패턴, 과거->미래 의존성 (주가, 센서, 재무)
- **부적합**: 매우 긴 시계열, 병렬 처리 필요 시

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 6.3 Transformer 계열 </span>

- <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;"> Informer </span>: 불필요한 연산을 줄여 긴 시계열 처리에 특화
  - 적합: 전력 수요 등 긴 시계열 / 부적합: 짧은 시계열 (과적합 위험)
- <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;"> Autoformer </span>: 요소 분해(추세/계절성)를 딥러닝에 결합
  - 적합: 월별 판매량 등 주기 패턴 포착 우수 / 부적합: 구조가 복잡
- <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;"> TFT (Temporal Fusion Transformer) </span>: 외부 변수(날씨, 휴일 등) 융합 + 어텐션 기반 해석 가능
  - 적합: 유통 수요 예측, <span style="color:blue">실무 활용도 1위</span> / 부적합: 학습 시간이 오래 걸림

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 6.4 Foundation Model </span>

> 거대 데이터로 사전 학습하여 다양한 시계열 문제에 범용 적용

1. **LLM 기반 접근 3가지**
   - Without Adaptation: 프롬프팅으로 LLM 직접 활용
   - Adapt LLM: 파인튜닝으로 시계열에 맞게 조정
   - Adapt to LLM: 시계열 -> 텍스트 변환 후 LLM에 입력

2. **시계열 전용 Foundation Model**: TimeGPT, TimesFM, Moirai, Chronos 등

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 6.5 모델 선택 가이드 </span>

| 상황 | 추천 모델 |
|------|-----------|
| 짧은 시계열 (< 100) | 통계 모델, RNN |
| 중간 길이 (100~1000) | LSTM, GRU |
| 매우 긴 시계열 (1000+) | Informer, Autoformer |
| 주기 패턴 뚜렷 | Autoformer |
| 외생 변수 많음 | TFT |
| 대규모 데이터 | Transformer 계열 |
| 범용 예측 | Foundation Model |

**핵심 트레이드오프**
- 단순함 vs 성능: RNN < LSTM/GRU < Transformer
- 해석 가능 vs 정확도: 통계 모델 > TFT > 기타 딥러닝
- 데이터 적으면 통계 모델/RNN, 많으면 Transformer 계열
