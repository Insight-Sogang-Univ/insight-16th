## <span style="background-color:#fae0f0; color:#c0156d; padding:4px; border-radius:5px;"> 1. 합성곱 신경망 (CNN, Convolutional NN) </span>
 
### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 1.1 픽셀과 채널 </span>
 
#### 1. **픽셀**
- 이미지 데이터를 구성하는 **가장 기본적인 단위**, 이미지의 특정 위치의 색상 또는 밝기 정보를 가짐
- **이미지 크기** : 흑백은 보통 2차원 배열(NumPy ndarray) = (세로 픽셀 수 × 가로 픽셀 수)
- **픽셀 값 표현**
  - **흑백(Grayscale)** : 하나의 값으로 밝기 표현, 0 ~ 255 (0 = 검은색, 255 = 흰색)
  - **컬러(RGB)** : Red, Green, Blue 세 값의 조합, 각 채널 0 ~ 255
#### 2. **채널**
- 이미지의 색상 정보를 구성하는 **각각의 데이터 층**
#### 3. **색공간**
- 이미지를 구성하는 색상을 **숫자로 표현하는 방법/체계**
- 대표 : 그레이 스케일, RGB, HSV
  - **그레이 스케일** : 색상 구분 없이 밝기 정보만 사용
  - **RGB** : 빨강, 초록, 파랑 세 색상의 조합
  - **HSV** : 색상(Hue), 채도(Saturation), 명도(Value) 기준
  - 
    |      구분      |  채널 수  |      텐서 크기     |                    비고                    |
    |:--------------:|:---------:|:------------------:|:------------------------------------------:|
    | 그레이 스케일  | 1         | (28 × 28 × 1)      | NumPy에서는 채널 생략해 (28 × 28)로 저장도 함 |
    | RGB            | 3         | (28 × 28 × 3)      | 높이 × 너비 × 채널                         |
 
- 흑백의 채널이 1인 이유 : 색상 성분 구분 없이 **밝기(명도) 하나만** 저장하면 되기 때문
- RGB 예 : 빨강, 초록 값이 높고 파랑 값이 낮으면 노란색 계열
#### 4. **텐서 (Tensor)**
- **다차원 데이터 배열**을 통칭, 스칼라, 벡터, 행렬을 모두 아우르는 포괄적 개념
  - **0차원 텐서 (스칼라):** 하나의 숫자
    - *예시: 3.14*
  - **1차원 텐서 (벡터):** 숫자가 일렬로 나열된 형태 (축 1개)
    - *예시: [1.0, 2.5, 3.1]*
  - **2차원 텐서 (행렬):** 행과 열로 이루어진 2차원 표 형태 (축 2개)
    - *예시: 앞서 살펴본 (28 x 28) 크기의 흑백 이미지 배열*
  - **3차원 이상 텐서 (Tensor):** 행렬이 여러 개 쌓이거나, 채널·배치(Batch) 등의 추가 축이 포함된 다차원 구조
    - *예시: (28 x 28 x 1) = 흑백 이미지 텐서 / (28 x 28x 3)  = RGB 컬러 이미지 텐서*

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 1.2 CNN의 개념 </span>
 
#### 1. **이미지 인식 방법의 차이**
- **사람** : 모든 픽셀을 하나하나 보지 않고, 눈, 귀, 코, 꼬리처럼 **중요한 영역과 특징**에 집중 -> 종합해서 판단
- **컴퓨터** : 이미지를 픽셀 단위로 쪼개 색상, 밝기 **숫자 정보**를 처리 (강아지 사진이 아니라 수많은 픽셀 값의 덩어리)
- **픽셀을 그냥 1차원으로 펼쳐 학습하면?**
  - 원래 이미지의 **2차원 공간 구조**가 명확히 활용되지 않음
  - 가까이 있는 픽셀들의 관계가 모델에 직접 반영되지 않아, 눈, 귀, 꼬리 같은 **특징 학습이 어려움**
#### 2. **MLP와 CNN의 구조 비교**
- **MLP** = FC Layer + ... + FC Layer
  - 2차원 이미지를 **1차원 배열로 펼친 뒤** 입력
- **CNN** = (Conv + Pool) + ... + Flatten + FC Layer + ... + FC Layer
  - 이미지의 **공간적 특징을 유지**하면서 Conv와 Pool을 반복해 특징 추출 -> Flatten -> FC로 분류
- <span style="color:blue">**MLP와 CNN의 가장 큰 차이** : CNN은 FC Layer 전에 Conv와 Pool로 **이미지의 특징을 학습하는 과정**이 추가됨</span>
#### 3. **CNN 모델의 전체 구조**
1. **특징 추출 (Feature Learning)**
   - **Convolution Layer** : 필터(Kernel)를 이미지 위에서 이동시키며 모서리, 질감, 형태 등 다양한 특징 추출
   - **Pooling Layer** : 특징 맵의 크기를 줄이고 중요한 특징은 유지 -> 연산량 감소
2. **Flatten** : 여러 차원의 특징 맵을 FC Layer에 넣을 수 있도록 **1차원 배열로 변환**
3. **분류 (Classification)**
   - **Fully Connected Layer** : 추출된 특징을 종합해 최종 클래스 분류
- 분류 부분은 앞서 배운 회귀/분류 모델과 유사, **CNN에서 새로 추가된 핵심은 Feature Learning!!!**

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 1.3 왜 이미지 처리에 MLP가 아닌 CNN을 사용할까? </span>
 
#### 1. **이미지 데이터의 구조**
- **흑백** : 픽셀당 값 하나(0~255) -> H * W 2차원 배열 (H : 세로 픽셀 수, W : 가로 픽셀 수)
- **컬러(RGB)** : 픽셀당 값 3개 -> H * W * 3 3차원 배열
- 대표 데이터셋 : **MNIST** (0~9 손글씨 숫자, 28 × 28 흑백 이미지, 이미지 분류/CNN 기초 예제)
- **이미지 구조를 유지하는 모델이 필요한 이유**
  - 정형 데이터(나이, 소득, 키 등) : 각 열이 **독립적인 의미**
  - 이미지 : 픽셀이 가로, 세로로 연결, **주변 픽셀 간 관계와 위치 정보**가 중요

#### 2. **MLP의 한계 3가지**
1. <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;">**위치, 크기, 회전 등의 변화에 취약**</span>
   - MLP는 **각 픽셀 위치마다 가중치**를 학습 -> 숫자 1이 조금만 이동해도 기존 픽셀은 0, 새 위치에 값이 들어가 입력이 크게 달라짐
   - 학습 때 못 본 위치/형태로 바뀌면 인식 성능이 크게 저하
   - **CNN은?** 동일한 필터를 여러 위치에 반복 적용 -> 특징이 **어디에 나타나도 탐지** 가능, Pooling으로 작은 위치 변화에도 비슷한 특징 유지
     - 즉, 특징이 정확히 어느 픽셀에 있는지가 아니라 **존재하는지**를 학습
2. <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;">**파라미터 폭발**</span> 
   - **완전연결(Fully Connected, Dense) 구조** : 이전 층의 모든 노드가 다음 층의 모든 노드와 빠짐없이 연결
   - 가중치 수가 이미지 픽셀 수만큼 생성됨 -> 과적합 가능성 상승, 이미지가 클수록 학습이 느림
   - **CNN은?** 작은 영역에만 필터를 적용(**부분 연결**) + 같은 필터의 가중치를 여러 위치에서 **공유**
     - 이미지가 커져도 필터 크기와 개수가 같으면 파라미터 수는 크게 안 늘어남
3. <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;">**공간 정보 손실**</span>
   - MLP는 Flatten으로 1차원 벡터로 펼침 (28 × 28 -> 784개)
   - 숫자 값 자체가 사라지는 건 아니지만, **어떤 픽셀들이 가까이 있었는지**의 2차원 구조가 명시적으로 유지되지 않음
   - **CNN은?** 처음부터 펼치지 않고 H * W * C 구조 유지, 특징을 **먼저 학습한 뒤 Flatten!!**

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 1.4 합성곱(Convolution)과 풀링(Pooling) </span>
 
#### 1. **Convolution Layer**
- **개념** : 입력 이미지 전체를 일정 크기의 **필터(Filter)로 훑으며** 특징을 추출하고 학습하는 계층
  - 필터는 수직선, 수평선, 모서리, 질감 같은 **특정 패턴**을 감지
- **연산 과정**
  1. 필터가 이미지 위를 **슬라이딩** (왼쪽 위에서 시작 -> 정해진 간격만큼 오른쪽 이동 -> 행이 끝나면 다음 행)
  2. 각 위치에서 **요소별 곱셈 후 합계** (필터 영역의 픽셀 값 × 필터 가중치를 같은 위치끼리 곱하고 모두 더해 숫자 하나로)
  3. 결과를 **Feature Map**의 해당 위치에 저장 (모든 위치를 마치면 하나의 Feature Map 완성)
- **필터(Filter) vs 커널(Kernel)** (거의 혼용하지만 엄밀히는 다름)
  - <span style="color:blue">**커널** : 패턴 추출에 쓰는 **2차원 가중치 행렬** (ex. 3 × 3 커널 = 가중치 9개)</span>
  - <span style="color:blue">**필터** : **여러 개의 커널이 모인 묶음**, 다채널 입력일 때 차이가 명확</span>
    - <span style="color:green">ex. RGB(3채널)에 필터 하나 적용 = 채널별 커널 3개 필요 -> 각 채널 연산 후 **모두 더해** 하나의 Feature Map 생성</span>
- **Feature Map**
  - Convolution 연산으로 생성되는 **결과 데이터**, 필터가 특정 특징에 반응할수록 해당 위치의 값이 커짐
  - <span style="color:blue">**Feature Map의 개수 = 사용한 필터의 개수**</span>
    - ex
      - RGB에 3×5×5 필터 1개 -> Feature Map 1개 (3 : 입력 채널 수, 5×5 : 필터의 공간 크기)
      - 필터 6개 -> Feature Map 6개</span>
  - 필터마다 다른 특징 추출 (ex. 세로 경계 감지 필터, 가로 경계 감지 필터)

#### 2. **(번외) 3채널(RGB) 컬러 이미지의 컨볼루션**
1. R, G, B 채널별로 필터의 해당 채널과 **원소별 곱셈**
2. 세 채널에서 나온 값(3×3이면 각 9개)을 **모두 합산**
3. **편향(Bias)** 을 더해(선택) 출력 맵의 한 픽셀(스칼라 값) 완성
- <span style="color:blue">3채널 입력에 3채널 필터 **1개**를 통과시키면 출력 채널은 **1개**로 줄어듦</span>
  - 출력 채널을 늘리려면 이런 3차원 필터를 **여러 개(K개)** 사용

#### 3. **Padding**
- **개념** : Convolution 전에 입력 이미지 **가장자리에 픽셀을 추가**하는 방법 (보통 0을 넣는 **Zero Padding**)
- **필요한 이유** : 패딩이 없으면 가장자리는 필터가 적용되는 횟수가 적고, 연산할수록 이미지가 작아짐
- **효과** : 가장자리 특징까지 학습 + Feature Map의 크기 조절
- **예시** 
  - 4×4 이미지에 3×3 필터 -> 패딩 없으면 Feature Map **3×3**
  - Padding 1(상하좌우 0을 한 줄씩 추가, 6×6로 확장) -> **4×4**

#### 4. **Stride (보폭)**
- **개념** : 필터가 한 번에 이동하는 **간격**
- Stride가 커질수록 연산 위치가 줄어 **Feature Map 크기와 데이터 양 감소**
- 특징을 얼마나 세밀하게 추출할지 vs 연산 효율의 조절 요소

#### 5. **Pooling**
- **개념** : 일정 크기 영역(보통 2×2)의 값을 **하나의 값으로 요약**, 공간 크기 축소 + 작은 위치 변화에 대한 민감도 감소
- **종류**
  - **Max Pooling** : 영역 안에서 **가장 큰 값** 선택
  - **Average Pooling** : 영역 안 모든 값의 **평균** 계산
- **풀링을 안 쓰면?** : 필터가 엄격하게 특정 픽셀 위치에서만 패턴을 찾아, 객체나 내부 요소(ex. 집의 창문)가 미세하게 이동하면 같은 특징으로 인식하지 못함
- **장점**
  - 차원 축소로 **계산량 감소**
  - 위치 변화에 대한 **강건성**
  - 과적합 방지에 도움 (불필요한 세부 정보와 공간적 복잡성 감소)
  - 노이즈 영향 완화 (단, 모든 노이즈를 제거하는 건 아님)
- **단점**
  - **정보 소실** 위험 (세부 정보가 사라질 수 있음)
  - **정확한 위치 정보 감소** (경계/세밀한 위치가 중요한 작업에서 문제)

#### 6. **CNN 구조 정리**
- **Convolution + Pooling** : 특징 추출 담당
- **Padding, Stride** : 출력 크기와 정보 보존 정도를 조절하는 **하이퍼파라미터**

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 1.5 대표적인 CNN 아키텍처 </span>
 
#### 1. **AlexNet (2012)**
- **개념** : 2012년 ImageNet 대회(ILSVRC)에서 뛰어난 성과를 거둔 CNN, 대규모 데이터로 특징을 직접 학습해 분류하는 딥러닝의 가능성을 보여줌
- **구조** : 총 8개 학습 계층 = **Conv 5개 + FC 3개** (중간에 ReLU와 Max Pooling)
- **딥러닝 역사에 기여한 5가지 혁신 기술**
  1. **ReLU 활성화 함수 도입** : 기존 Tanh 대신 사용해 **기울기 소실** 문제 해결, 학습 속도 6배 이상 향상
  2. **다중 GPU 병렬 연산** : 3GB GPU(GTX 580) 2개를 병렬로 연결해 구조를 상/하단으로 분할 학습
  3. **드롭아웃(Dropout)** : 과적합 방지를 위해 FC 계층 뉴런 일부(ex. 50%)를 **무작위로 꺼서** 학습
  4. **데이터 증강(Data Augmentation)** : 무작위 크롭, 좌우 반전, RGB 채널 값 변경(PCA)으로 학습 데이터를 인위적으로 증가
  5. **로컬 응답 정규화(LRN)** : 측면 억제 개념으로 강하게 활성화된 뉴런 주변을 감쇠 (최근에는 **배치 정규화**로 대체)
#### 2. **VGGNet (2014)**
- **개념** : 옥스퍼드대 연구팀 개발, **네트워크 깊이를 늘리면 성능이 향상**될 수 있음을 보여줌 (VGG16, VGG19의 숫자 = 학습 가능한 계층 수)
- **구조** : VGG16 = **Conv 13개 + FC 3개**, 작은 **3 × 3 필터**를 반복 적용 + 중간중간 Max Pooling
- **주요 특징**
  - 작은 필터의 반복 사용
  - AlexNet보다 깊은 네트워크로 복잡하고 추상적인 특징 학습
  - 대부분 같은 3 × 3 필터를 써서 **구조가 단순하고 이해하기 쉬움**
- **AlexNet 대비 발전** : AlexNet은 비교적 큰 필터 + 적은 Conv 층 -> VGGNet은 **작은 필터 + 더 깊은 구성**
#### 3. **ResNet (2015)**
- **개념** : **깊은 신경망을 효과적으로 학습**하기 위해 개발, **Residual Connection** 구조 도입
- **핵심 혁신** : 잔차 학습, 지름길
  - 여러 Conv Layer 사이에 **Skip Connection** 추가
  - <span style="color:blue">Skip Connection : 입력 정보를 중간 계층을 거치지 않고 뒤쪽 계층으로 **직접 전달**, 계층을 통과한 결과와 처음 입력을 **더해서** 다음 계층으로 전달</span>
### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 1.6 이미지 딥러닝 응용 </span>
 
#### 1. **이미지 분류 (Image Classification)**
- 이미지를 수치 데이터(픽셀 값)로 인식 -> **사전 정의된 클래스 중 하나**로 분류
- 
  |      모델명      |                 특징 요약                |                주의사항                |
  |:----------------:|:----------------------------------------:|:--------------------------------------:|
  | AlexNet          | 비교적 단순한 구조, 초기 CNN 모델        | 작은 데이터셋이나 학습 개념 설명용에 적절 |
  | VGG              | 더 깊은 구조, 복잡한 블록 구조           | 계산량 증대, 메모리 요구 높음          |
  | ResNet           | Skip connection -> 매우 깊은 네트워크 가능 | 과적합 주의, batch size/regularization 조절 필요 |
  | ViT (Vision Transformer) | Transformer 구조 응용, **패치 기반** 입력 처리 | 이미지와 순차 처리 모델의 결합, 대규모 데이터 필요 |  
 
- **ViT** : 이미지를 작은 **패치(Patch)** 로 나눠 Transformer에 입력, 필터 대신 **Attention**으로 패치 간 관계를 학습
#### 2. **객체 탐지 (Object Detection)**
- 이미지 분류에서 한 단계 확장 : 분류는 이미지 **전체**의 클래스를 예측, 탐지는 개별 객체의 **종류와 위치를 함께** 예측
- **Bounding Box** : 객체가 위치한 영역을 둘러싸는 **사각형**, 보통 중심 좌표 $(x, y)$, 너비 $w$, 높이 $h$로 표현
- **모델 적용 방식** : CNN으로 Feature Map을 얻고 그 위에 **두 개의 head**를 올림
  - **Classification Head** : 객체의 **종류** 예측 (사람, 자동차, 강아지 등)
  - **Regression Head** : 객체의 **위치 좌표** 예측 (Bounding Box의 위치와 크기)
- 대표 모델 : **YOLO**(You Only Look Once), Faster R-CNN, SSD
#### 3. **이미지 캡셔닝 (Image Captioning)**
- 이미지의 내용을 **자연어 문장**으로 표현, 객체 판단을 넘어 **객체 간의 관계와 행동**까지 파악
  - <span style="color:green">ex. 사람, 자전거, 도로 -> "한 사람이 도로에서 자전거를 타고 있습니다."</span>
- **CNN + RNN 사용**
  - **CNN** : 이미지 특징 추출
  - **RNN** : 추출된 특징을 바탕으로 단어를 순서대로 생성 (이전 단어 정보로 다음 단어 예측)
- **단계** : CNN 특징 추출 -> **Attention Network로 단어 생성** -> RNN으로 문장 완성
  - **Attention** : 모든 영역을 동일하게 보지 않고, 현재 생성하려는 단어와 관련된 **중요한 영역에 더 높은 가중치** 부여
    - <span style="color:green">ex. '사람' 생성 시 사람 영역, '자전거' 생성 시 자전거 영역에 집중</span>
#### 4. **CNN 세 줄 요약**
1. (Conv + Pool)의 반복으로 **특징을 학습**
2. FC Layer를 적용해 이미지를 **분류**
3. (Conv + Pool)에 어떤 방법을 적용하느냐에 따라 **성능이 달라짐**

## <span style="background-color:#fae0f0; color:#c0156d; padding:4px; border-radius:5px;"> 2. 과거를 기억하는 신경망: RNN (Recurrent Neural Network) </span>
 
### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 2.1 순차(Sequence) 데이터 </span>
 
#### 1. **순차 데이터란?**
- 데이터의 **순서가 의미를 갖는** 데이터, 순서를 바꾸면 정보가 달라짐
#### 2. **순차 데이터 세부 분류**
1. **순서의 기준**
    |  구분  |    시간적 순차 데이터   |       비시간적 순차 데이터      |
    |:------:|:-----------------------:|:-------------------------------:|
    | 기준   | 시간 순                 | 시간 외 기준 (위치, 순번)       |
    | 예시   | 주가, 기온              | 문장 내 단어, DNA 염기서열      |

2. **자료형**
    |  구분  |         수치형         |            범주형            |
    |:------:|:----------------------:|:----------------------------:|
    | 특징   | 값의 크기와 거리에 의미 O | 값의 크기와 거리에 의미 X    |
    | 예시   | 음성, 비디오           | 웹 클릭 로그, 구매 이력      |
   - **음성** : 소리(시간에 따라 변하는 공기 진동)를 짧은 간격으로 잘라 **진폭 값**을 기록 (ex. 16kHz = 1초에 16,000번 기록), 순서가 바뀌면 전혀 다른 데이터
   - **비디오** : 이미지(프레임)를 연속으로 넘긴 것 (ex. 30프레임 = 1초에 30장), 프레임 순서가 뒤섞이면 의미가 완전히 달라짐, 시점 $t$의 프레임은 $t-1$의 객체 위치/움직임에 영향을 받음

#### 3. **기존 정형 데이터 vs 순차 데이터**
 
|     비교 관점    |       정형 데이터       |                    순차 데이터                    |
|:----------------:|:-----------------------:|:-------------------------------------------------:|
| 순서를 섞으면?   | 의미 그대로 O           | **의미가 달라짐** (요소의 **순서의존성**)         |
| 샘플 간 관계     | 상호독립적              | **상호의존적** : 이전 시점/순번이 현재에 영향 (**자기상관성**) |
| 예시             | 고객 정보 데이터        | 주가, 문장                                        |
 
- 정형 데이터 : 고객 A, B, C 행 순서를 바꿔도 의미 변화 X
- 순차 데이터 : "개가 사람을 물었다" -> "사람이 개를 물었다" = 완전히 다른 의미
- 기존 머신러닝(회귀, 결정트리, SVM)은 **정형 데이터를 전제**로 설계 -> 순차 데이터 처리에 한계 -> <span style="color:blue">**순차 데이터를 처리할 별도의 모델 = RNN**</span>

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 2.2 RNN의 구조와 원리 </span>
 
#### 1. **구조**
- 기존 신경망과 비교했을 때 가장 큰 차이 : <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;"> 은닉층에 **순환 구조**가 존재 </span>
- **입력층 ($x_t$)** : 각 시점 $t$에 들어오는 입력 데이터
- **은닉층 ($h_t$)** : 현재 입력 $x_t$와 **이전 시점의 은닉 상태 $h_{t-1}$** 를 함께 받아 새 은닉 상태 $h_t$ 생성
- **출력층 ($o_t$)** : 은닉 상태 $h_t$를 바탕으로 시점 $t$의 결과 출력
- **가중치 3종**
  - **입력 가중치 ($U$)** : $x_t$를 은닉층으로 보낼 때
  - **순환 가중치 ($V$)** : $h_{t-1}$을 현재 은닉층으로 보낼 때
  - **출력 가중치 ($W$)** : $h_t$를 출력 $o_t$로 바꿀 때

#### 2. **원리**
- **핵심** : 매 시점마다 **현재 입력 + 이전 은닉 상태**로 현재 은닉 상태를 계산하고, 그 결과를 다음 시점으로 넘김
- **은닉 상태 계산**
$$h_t = \tanh(Ux_t + Vh_{t-1} + b)$$
  - **tanh** : 값을 -1 ~ 1로 제한해 은닉 상태값이 **폭주하는 것을 방지**, **비선형성** 추가로 복잡한 패턴 학습 가능
- **출력 계산** : 은닉 상태 $h_t$로 현재 시점 출력 생성 ($c$ : 편향)
$$o_t = f(Wh_t + c)$$
  - $f$ (활성화 함수) : **이진 분류**면 시그모이드(0~1로 압축, 특정값 이상이면 1, 이하면 0), **다중 분류**면 소프트맥스(각 점수를 지수함수로 변환 후 전체 합으로 나눔, 출력 0~1, 합은 항상 1)
- <span style="color:green">ex. "나는 / 밥을 / 먹었다" : t=1 "나는" 처리 후 은닉 상태를 다음으로 전달 -> t=2 "나는"+"밥을" 정보 저장 -> t=3 문장 전체 맥락이 $h_3$에 요약, 활성화 함수 거쳐 최종 출력</span>

#### 3. **RNN 아키텍처** (시퀀스 = 순서가 있는 데이터의 나열)
1. **One to Many**: 하나의 입력 -> 시퀀스 출력 
   - ex. 이미지 캡셔닝 (사진을 첫 시점에 한 번만 넣으면 제목 출력)
2. **Many to One**: 시퀀스 입력 -> 하나의 값 출력
   - ex. 스팸 메일 분류, 감성 분류 (긍정/부정, 기쁨/분노/슬픔/놀람)
   - 전체 문장을 읽은 후 **최종 판별**하므로 사용
3. **Many to Many**: 시퀀스 입력 -> 시퀀스 출력
    - **동기식** : 매 시점마다 즉각 출력 (객체 태깅 등)
    - **비동기식** : 전체 입력을 끝까지 읽은 뒤 한 번에 출력(기계 번역, 요약 등)`


### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 2.3 RNN의 한계 및 장기 의존성 문제 </span>
 
#### 1. **BPTT (시간에 따른 역전파 구조)**
- RNN은 순전파로 시퀀스를 처리한 뒤, **시간을 거슬러 올라가며(BPTT)** 오차를 전파해 가중치 기울기를 **누적 계산**
- (참고) **가중치 기울기 계산** : 가중치를 살짝 늘렸을 때 오차가 얼마나 변하는지를 측정해, 오차를 줄이려면 가중치를 어느 방향으로 얼마나 이동할지 결정하는 과정
- **chain rule** : 여러 단계로 연결된 관계에서 각 단계의 변화율을 **곱하면** 전체 변화율
  - <span style="color:grey">비유 : 톱니바퀴 A가 1바퀴 돌면 B는 3바퀴, B가 1바퀴 돌면 C는 2바퀴 -> A가 1바퀴 돌 때 C는 3 × 2 = 6바퀴</span>
  - RNN : $h_1 \to h_2 \to h_3 \to$ 출력층 $\to$ 오차 $L$
$$\frac{\partial L}{\partial h_1} = \frac{\partial L}{\partial h_3} \times \frac{\partial h_3}{\partial h_2} \times \frac{\partial h_2}{\partial h_1}$$
  - 한 시점을 거슬러 갈 때마다 곱해지는 값
  - **같은 순환 가중치 $V$를 모든 시점이 공유** -> 같은 값이 계속 곱해지는 **등비수열** 형태, $k$시점 전까지 가면 $V^k$ 꼴

#### 2. **RNN의 한계 3가지**
1. <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;">**기울기 소실** (Vanishing Gradients)</span>
   - 역전파 시 기울기가 과거 시점으로 갈수록 점점 작아져 **0에 가까워짐**
   - **원인** : (활성화 함수 미분값) × (순환 가중치 $V$) **< 1**
     - tanh 미분값은 최대 1, 대부분 1보다 작음 / $V$가 1보다 작으면 반드시 소실, 조금 커도 소실될 수 있음
   - <span style="color:blue">기본 RNN에서는 기울기 폭주보다 **훨씬 자주** 발생 (tanh 미분값 ≤ 1이기 때문)</span>
   - 결과 : 가까운 시점은 학습되지만 **먼 과거 시점의 정보는 학습에 거의 반영 안 됨**

2. <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;">**기울기 폭주** (Exploding Gradients)</span>
   - 역전파 시 기울기가 과거 시점으로 갈수록 점점 커져 **매우 큰 값**이 됨
   - **원인** : (활성화 함수 미분값) × (순환 가중치 $V$) **> 1**
   - 가중치가 크게 튀며 오차가 갑자기 치솟거나 **진동**
     - 정상 : 10 -> 8 -> 6 -> 5 -> 4 -> 3.5 ... (점점 줄어듦)
     - 진동 : 10 -> 4 -> 12 -> 3 -> 15 -> 5 ... (오르락내리락, 안 줄어듦)
   - 심하면 오차가 너무 커져 **NaN**(계산 불가)이 발생하고 학습 중단
   - 결과 : 긴 시퀀스일수록 학습이 불안정해져 먼 의존 관계를 안정적으로 배우기 어려움

3. <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;">**느린 훈련 시간**</span>
   - 시점을 **하나씩 순서대로만** 계산 가능 -> 시퀀스 길이에 비례해 계산 시간 증가
   - 역전파를 위해 모든 시점의 은닉 상태를 저장해야 해서 **메모리 부담**도 큼
   - **CNN과의 차이** : CNN은 필터가 각 위치를 독립적으로 계산해 **병렬 처리** 가능 / RNN은 앞 시점 결과가 있어야 다음 계산 가능
   - 결과 : 앞부분의 맥락을 뒷부분이 알게 하는 데 많은 학습 비용 필요

#### 3. **장기 의존성 문제 (Long-Term Dependency Problem)**
- 세 가지 한계의 공통점이 하나의 문제로 귀결
- <span style="color:blue">**시퀀스에서 관련된 정보가 멀리 떨어져 있을수록, RNN이 그 관계를 기억하고 학습하지 못하는 문제**</span>
    |     한계     |            무슨 일이 일어나나           |                 결과                 |
    |:------------:|:---------------------------------------:|:------------------------------------:|
    | 기울기 소실  | 먼 과거로 갈수록 학습 신호가 0에 가까워짐 | 먼 과거 정보가 반영 안 됨            |
    | 기울기 폭주  | 먼 과거로 갈수록 신호가 폭발해 학습 불안정 | 먼 관계를 안정적으로 학습하지 못함   |
    | 느린 훈련    | 시퀀스가 길어질수록 학습 비용 증가      | 먼 관계 학습에 많은 비용 필요        |
- 필요한 것 : **중요한 정보는 오래 기억하고 불필요한 정보는 잊는 메모리 구조** -> **LSTM & GRU**


## <span style="background-color:#fae0f0; color:#c0156d; padding:4px; border-radius:5px;"> 3. LSTM & GRU </span>
 
### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 3.1 개요 </span>
 
#### 1. **게이트(Gate)**
- 두 모델 모두 **게이트**로 무엇을 기억하고 무엇을 잊을지 결정
- **게이트** : 정보를 **얼마나 통과시킬지** 정하는 구성
  - 0~1 사이 값을 정보에 곱함 -> 0이면 버림, 1이면 그대로, 0.5면 절반만 반영
  - <span style="color:green">ex. [5, 3, 8] ⊙ [1, 0, 0.5] = [5, 0, 4] (그대로 남김 / 지움 / 절반만 남김)</span>
  - 값은 현재 입력과 이전 은닉 상태를 보고 모델이 **스스로 계산** (sigmoid 사용)

#### 2. LSTM vs GRU
  |     구분     |             LSTM              |              GRU             |
  |:------------:|:-----------------------------:|:----------------------------:|
  | 상태         | 셀 상태 + 은닉 상태 (2개)     | 은닉 상태 (1개)              |
  | 게이트       | Forget, Input, Output (3개)   | Update, Reset (2개)          |
  | 핵심 아이디어 | 장기/단기 기억을 **분리**     | 기억과 출력을 **통합**       |
  | 장점         | 정교한 기억 관리              | 단순하고 빠름                |
  | 단점         | 복잡하고 계산량 많음          | 정교함이 떨어짐              |
 
### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 3.2 LSTM (Long Short-Term Memory) </span>
 
#### 1. **구조**
- **기억을 두 개의 상태로 나눠 관리**
    |             구분             |        역할        |
    |:----------------------------:|:------------------:|
    | **셀 상태** (Cell State, $c$) | **장기 기억**      |
    | **은닉 상태** (Hidden State, $h$) | **단기 기억 & 출력** |
  - 기존 RNN은 은닉 상태 $h$ 하나에 모든 걸 **덮어썼지만**, LSTM은 오래 남길 기억(셀 상태)을 따로 둠

- **세 개의 게이트**
  1. **Forget Gate** : 장기 기억에서 더 이상 필요 없는 내용을 **삭제**
  2. **Input Gate** : 새로 들어온 정보 중 중요한 것만 장기 기억에 **저장**
  3. **Output Gate** : 장기 기억 중 지금 필요한 부분만 꺼내 단기 기억으로 **출력**

- **핵심 : 셀 상태** = 구조도 맨 위를 가로지르는 선, 정보가 큰 변화 없이 시점을 따라 쭉 이어지며 게이트가 빼거나(잊기) 더해서(기억하기) 조절
    |      구분      |             RNN             |                 LSTM                 |
    |:--------------:|:---------------------------:|:------------------------------------:|
    | 순환되는 상태  | 은닉 상태 $h$ (1개)         | 셀 상태 $C$ + 은닉 상태 $h$ (2개)    |
    | 역할           | $h$ 하나가 기억과 출력 모두 | $c$ = 장기 기억, $h$ = 단기 기억 & 출력 |
    | 기억 갱신      | 매 시점 **통째로 덮어씀**   | 게이트로 **선택적으로 지우고 더함**  |

#### 2. **단계별 수식**
- **공통 표기**
  - $[h_{t-1}, x_t]$ : 이전 은닉 상태와 현재 입력을 이어 붙인 벡터
  - $\sigma$ : sigmoid (출력 0~1, 1에 가까울수록 많이 기억/통과, 0에 가까울수록 많이 잊음/차단)
  - $\odot$ : 같은 위치끼리 곱하기

1. **Step 1. Forget gate** : "무엇을 잊을까?"
   - 이전 셀 상태 $C_{t-1}$의 각 칸에 곱해질 0~1 값 ($f_t \approx 1$ : 그대로 유지 / $f_t \approx 0$ : 완전히 잊음)

2. **Step 2. Input gate** : "무엇을 새로 기억할까?"
   - $i_t$ : 새 정보를 얼마나 반영할지 (0~1) / $\tilde{C}_t$ : 추가할 **후보 정보** (-1~1)
   - 실제로 더해지는 양 = $i_t \odot \tilde{C}_t$

3. **Step 3. Final memory cell** : "기억 업데이트"
   - Step 1에서 남긴 이전 기억 + Step 2에서 고른 새 정보
   - <span style="color:blue">**곱셈이 아닌 덧셈**으로 이어지는 핵심 구간</span>

4. **Step 4. Output gate** : "무엇을 내보낼까?"
   - $o_t$ : 셀 상태 중 어느 부분을 내보낼지 (0~1) / 셀 상태를 -1~1로 정리한 뒤 필요한 부분만 출력

#### 3. **LSTM의 장단점**
- **장점**
  - 셀 상태의 **덧셈 구조** 덕분에 기울기가 먼 과거까지 전달 -> **기울기 소실 완화**
  - 게이트로 중요한 정보는 남기고 불필요한 정보는 지움 -> **장기 의존성 학습** 가능
  - 문장 번역, 음성 인식 등 긴 시퀀스 과제에서 RNN보다 좋은 성능
- **단점**
  - 계산량이 많아 학습이 느리고 메모리도 많이 사용
  - 여전히 시점을 순서대로 계산 -> **병렬 처리 불가** (느린 훈련 시간 문제는 그대로)
  - <span style="color:blue">기울기 소실을 **"완화"** 한 것이지 완전히 없앤 것은 아님 -> 매우 긴 시퀀스에서는 여전히 한계</span>
- -> 더 단순한 구조인 GRU 등장의 배경

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 3.3 GRU (Gated Recurrent Unit) </span>
 
#### 1. **정의**
- LSTM의 핵심 아이디어(게이트로 정보를 선택적으로 기억/삭제)는 **유지**하면서 구조를 **단순하게 줄인 모델** (LSTM을 더 가볍게 만든 버전)
  - LSTM : 상태 2개, 게이트 3개 / GRU : 상태 **1개**(은닉 상태), 게이트 **2개**
- (TMI) 2014년 **조경현 박사**가 몬트리올대(요슈아 벤지오 연구실)에서 기계 번역 연구 중 처음 제안, LSTM(1997) 17년 뒤 등장

#### 2. **LSTM과의 차이점**
1. **Forget gate와 Input gate를 Update gate 하나로 통합**
   - Update gate $z_t$ 하나로 '과거 기억 : 새 정보 = 몇 : 몇'을 한 번에 결정
      $$h_t = (1 - z_t) \odot h_{t-1} + z_t \odot \tilde{h}_t$$
   - ex. $z_t$ = 0.3 -> 과거 기억 70% + 새 정보 30%<

1. **Reset gate 사용**
   - "과거 기억을 얼마나 **무시**할지"를 먼저 정한 뒤 후보 정보 $\tilde{h}_t$ 생성
   - $r_t \approx 0$ : 과거를 거의 무시하고 현재 입력 위주 (<span style="color:green">ex. 새 문장이 시작될 때 이전 문맥을 끊음</span>)
2. **Output gate가 없음** : 출력을 거르는 단계 없이 은닉 상태 $h_t$를 **그대로 출력**
3. **Gate의 개수 3개 -> 2개**
   - 파라미터가 **LSTM의 약 3/4** -> 계산량 감소, 학습 시간 단축
   - 그런데도 성능은 LSTM과 비슷하거나 과제에 따라 더 좋은 경우도 있음
   - 
      |      구분      |             LSTM             |                 GRU                |
      |:--------------:|:----------------------------:|:----------------------------------:|
      | 기억 상태      | 셀 상태 + 은닉 상태 (2개)    | 은닉 상태 (1개)                    |
      | 잊기/기억하기  | Forget gate, Input gate 따로 | **Update gate 하나**로 (몇 대 몇)  |
      | 과거 무시 여부 | 별도 게이트 없음             | **Reset gate**                     |
      | 출력 조절      | Output gate                  | 없음 (은닉 상태 그대로 출력)       |
      | 게이트 수      | 3개                          | 2개                                |
      | 파라미터       | 많음                         | LSTM의 약 3/4                      |
   
#### 3. **GRU의 장단점 (LSTM과 비교)**
 
|   구분   |                  장점                  |                    단점                    |
|:--------:|:--------------------------------------:|:------------------------------------------:|
| 구조     | 상태 1개, 게이트 2개로 단순, 이해/구현 쉬움 | 장기 기억과 출력을 분리하지 못함          |
| 속도     | 파라미터가 적어 학습 빠름, 메모리 적게 사용 | (해당 없음)                               |
| 데이터   | 파라미터가 적어 데이터가 적을 때도 학습 안정적 | 데이터가 많고 복잡한 과제에서는 LSTM이 더 나을 수 있음 |
| 기억 관리 | Update gate 하나로 간단히 갱신        | 잊는 양과 더하는 양을 따로 조절 못해 정교함 떨어짐 |
| 성능     | 많은 과제에서 LSTM과 비슷             | 항상 LSTM보다 낫다고 단정할 수 없음        |
 
### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 3.4 LSTM vs GRU: 언제 무엇을 선택할까? </span>
- <span style="color:blue">두 모델은 **우열 관계가 아니라** 정교함(LSTM)과 효율(GRU) 사이의 **선택**</span>
- 실제 비교 연구에서도 많은 과제에서 두 모델의 성능은 비슷함
- 결론 : **문제 상황에 따라 적합한 모델이 다름**

## <span style="background-color:#fae0f0; color:#c0156d; padding:4px; border-radius:5px;"> 4. Seq2Seq </span>
 
### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 4.1 Seq2Seq의 기본 구조: 인코더와 디코더 </span>
 
#### 1. **정의**
- **Sequence-to-Sequence** : <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;"> 입력 시퀀스를 받아 **다른 시퀀스**를 출력하는 모델 </span>
- <span style="color:blue">**입력과 출력의 길이가 서로 달라도 됨** -> 기존 RNN과의 가장 큰 차이</span>
  - <span style="color:green">ex. "I am a student"(4단어) -> "je suis étudiant"(3단어), 번역은 언어마다 어순과 단어 수가 달라 Seq2Seq 사용</span>
- 앞서 본 **Many to Many**를 **두 개의 RNN(LSTM)으로 나눠** 구현한 구조
- **활용 분야** : 기계 번역, 챗봇(질문 -> 답변), 문서 요약, 음성 인식

#### 2. **구성**
- **인코더** : 입력 문장을 읽고 하나의 **컨텍스트 벡터**로 압축
- **컨텍스트 벡터** : 인코더와 디코더를 잇는 다리, 인코더를 거치며 압축된 정보
- **디코더** : 컨텍스트 벡터를 받아 출력 문장을 **한 단어씩 생성**
- 내부 셀은 주로 LSTM 또는 GRU (장기 의존성 문제 때문)
- 한 줄 요약 : 인코더(읽고 요약해서 넘기고) -> 컨텍스트 벡터 -> 디코더(다시 써냄)

#### 3. **인코더**
- 입력 문장의 모든 단어를 순서대로 읽고, 문장 전체 정보를 **하나의 벡터로 압축하는 LSTM**
- **작동 원리**
  1. **단어 토큰화** : 문장을 단어 단위로 나눔 ("I", "am", "a", "student")
  2. **임베딩(embedding)** : 각 단어를 계산 가능한 **숫자 벡터**로 변환
  3. LSTM 셀에 **한 단어씩 입력** : 매 시점 은닉 상태가 다음 시점으로 전달되며 정보 누적
  4. **마지막 시점의 은닉 상태 = 컨텍스트 벡터** -> 디코더로 전달
- <span style="color:blue">인코더의 **중간 시점 출력은 사용하지 않음**, 마지막 은닉 상태(컨텍스트 벡터)만 디코더로 넘김</span>

#### 4. **디코더**
- 컨텍스트 벡터를 받아 출력 문장을 **한 단어씩 순서대로 생성하는 LSTM**
- **작동 원리**
  1. 컨텍스트 벡터를 디코더 LSTM의 **첫 은닉 상태**로 받음
  2. 시작 신호 **`<sos>`**(start of sequence)를 첫 입력으로 넣음
  3. 매 시점 **softmax**로 다음 단어의 확률 계산 -> 확률이 가장 높은 단어 선택(argmax)
  4. **방금 예측한 단어를 다음 시점의 입력으로** 다시 넣음
     - <span style="color:green">ex. `<sos>` -> "je" -> "suis" -> "étudiant"</span>
  5. 종료 신호 **`<eos>`**(end of sequence)가 나오면 생성 종료

#### 5. **컨텍스트 벡터**
- 인코더가 입력 문장 전체를 읽고 만든 **문장 요약 벡터**
  - 인코더 LSTM의 마지막 시점 은닉 상태 (LSTM이면 **셀 상태도 함께** 넘김)
  - 인코더와 디코더를 연결하는 **유일한 통로**
- **차원 설정**
  - 컨텍스트 벡터의 크기 = 인코더 LSTM의 은닉 상태 크기 (사람이 정하는 **하이퍼파라미터**, 보통 256, 512, 1024 등 수백 차원)
  - **디코더 LSTM의 은닉 상태 크기도 같은 값**으로 맞춰야 바로 이어받을 수 있음
  - 차원이 크면 : 더 많은 정보를 담지만 계산량 증가, 과적합 위험
  - 차원이 작으면 : 빠르지만 긴 문장의 정보를 다 담지 못함

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 4.2 Seq2Seq의 한계: 병목 현상 (Bottleneck) </span>
 
#### 1. **병목 현상**
- <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;"> 인코더가 입력 문장 전체를 **고정된 크기의 컨텍스트 벡터 하나**에 압축하면서, 정보가 좁은 통로에 막혀 손실되는 현상 </span>
  - 5단어든 50단어든 같은 크기(ex. 512차원)에 담아야 함 -> **긴 문장일수록 정보 손실**
  - 앞 단어는 여러 시점을 거치며 희미해짐 -> **장기 의존성 문제가 여전히 존재**
    - ex. 짧은 문장 "I am a student" : 정보가 적어 충분히 담김 -> 번역 정확
    - 긴 문장(수십 단어) : 앞부분("I am a student")의 정보가 손실 -> 주어나 앞 내용을 틀리게 번역
- 실제 연구에서도 입력 문장이 길어질수록 기본 Seq2Seq의 번역 품질이 **급격히 하락**
#### 2. **결론**
- 고정 크기 컨텍스트 벡터에 모든 정보를 압축하기 때문에 병목 발생, 특히 **긴 입력 시퀀스일수록** 정보 손실이 심각하고 성능 하락
- 이 한계를 해결한 것이 다음 시간의 **Attention**!
