## 분류   
  


### 회귀와 분류의 차이  

1. 지도학습 : 회귀 vs 분류  

> 분류란? 기존 데이터가 어떤 레이블에 속하는지 패턴을 알고리즘으로 인지한 뒤에 새롭게 관측된 데이터에 대한 레이블을 판별하는 것  

**회귀와 분류의 차이?**  
- 회귀: 데이터가 연속형 변수를 예측하기 위해 사용될 때  
- 분류: 데이터가 범주형 변수를 예측하기 위해 사용될 때  
즉, 종속 변수가 연속형이면 회귀, 범주형이면 분류 사용!

2. 이진분류와 다중분류  
 - 이진 분류(Binary Classification): 예측하고자 하는 변수 어떤 기준에 대하여 참(True) 또는 거짓(False)의 값만을 가질 때  
 - 다중 분류(Multiclass Classification): 예측하고자 하는 변수가 가질 수 있는 값이 3개 이상일 때  

   



### 분류모델의 4가지 종류와 각 모델이 무엇인지  

1. >로지스틱 회귀(Logistic Regression) : 독립변수와 종속변수의 선형 관계성에 기반하여 분류  
 로지스틱 회귀란? 독립 변수의 선형 조합 (선형 회귀)에 로지스틱 함수를 적용하여 출력값을 0에서 1 사이로 변환해주는 것을 의미한다  

- 로지스틱 회귀의 필요성  
- 이진 분류를 단순 선형 회귀로 풀려고 할때 문제가 발생하는 경우 발생(Why?)  
    1. 이진 분류에선 x와 y의 관계를 표현하기 위해 직선이 아닌 S자 형태 함수가 필요! 직선의 경우 분류 작업이 제대로 동작하지 않는다.  
    2. 예측값은 0에서 1사이의 값을 가져야하는데, 직선으로 분류작업을 수행하면 값이 음의 무한대부터 양의 무한대까지 범위를 벗어나는 값까지 도출이 된다.

- 시그모이드 함수(Sigmoid fuction) : 출력이 0과 1사이의 값을 가지면서 S자 형태로 그려지는 함수  
    특징 : 입력값이 커지면 1에, 작아지면 0에 수렴, 입력값($x$)이 $(-\inf,\inf)$일 때, 출력값($y$)은 0~1의 값을 가진다.  

      

- 시그모이드 함수의 가중치  
    시그모이드 함수가 $ H(x) = \frac{1}{1+e^{-(wx+b)}} = sigmoid(wx+b) = \sigma(wx+b) $ 일때, 주어진 데이터에 적합한 가중치 $w$와 $b$를 계산해서 찾는것이 목표  

     1. 가중치 w의 변화에 따른 그래프 변화(기울기 변화)  
       
     w가 클수록 중심축에 가까운 형태

     2. 가중치 b의 변화에 따른 그래프 변화(그래프의 위치 변화)  
       
     b가 클수록 위로 뜨는 형태

2. 의사결정 나무(Decision Tree) : 데이터 균일도에 따른 규칙 기반의 분류  
 > 의사결정나무 : 조건에 따라 데이터를 분류하며, 최종적으로 데이터가 순수한 label의 집합으로 구성 될때까지 분류를 방복하는 분석방법  

-   1. 의사결정나무의 구성요소   
    - Root Node: 가장 상위에 위치한 노드  
    - Parent Node: 주어진 노드의 상위에 있는 노드  
    - Child Node: 하나의 노드로부터 분리된 2개 이상의 노드들  
    - Binary Tree(이진 트리): Tree 중에서 children이 최대 2개인 tree  
    - Leaf Nodes: 자식마디가 없는 노드, 모델에서 label에 해당  
    - Edge: Parent node와 child node를 연결하는 연결선  
    - Height: 특정 노드에서 가장 멀리 있는 leaf node까지의 경로에 있는 edge 개수  
    - Depth: Root node에서 특정 노드에 도달하기 위한 edge의 수  

    2. CART(이진분할; Classification and Regression Tree) 알고리즘 : 가장 널리 사용되는 의사결정나무 알고리즘으로, 데이터셋을 임계값을 기준으로 두 child로 나누는 알고리즘  
     -> 임계값은 어떻게 나눠야 하나요? : 불순도(지니 계수)가 낮아지는 방향으로!  
     *불순도란?  분류하려는 데이터 집합에서 서로 다른 클래스(범주)가 섞여 있는 정도*  
     *지니 지수(Gini Index)란? 불순도를 나타내며, 통계적 분산 정도를 정량화해서 표현한 값(0~1)*  
     $Gini=1-\sum_{i=1}^n{P_i}^2,\space P_i\text{는 class i의 비율}$  
    
    -    **CART 알고리즘의 주요 단계**  
        1. 임계값 설정 : 임계값을 기준으로 데이터를 그룹으로 나눈다  
          

        2. 불순도 감소 알고리즘 : 데이터를 분류했을때 동일한 객체들끼리 잘 모이게 하는 기준을 찾는다  
        


    
    3. 실제 학습시 고려해야 할것들 : 모수 설정, 차이점 시각화, Pruning  
     - Hyperparameter 설정 
        - min_samples_split : 분할되기 위해 노드가 가져야 하는 최소 샘플(데이터) 수  
        - min_samples_leaf : 리프 노드가 가지고 있어야 하는 최소 샘플 수  
        - min_weight_fraction_leaf : min_samples_leaf와 같지만 가중치가 부여된 전체 샘플 수에서의 비율  
        - max_leaf_nodes : 리프 노드의 최대 개수  
        - max_features : 각 노드에서 분할에 사용할 특성의 최대 수  

     - 시각화 : 시각화를 수행함으로써 분류가 잘 이루어졌는지 확인  
          
    
     - Pruning(가지치기) = 불필요한 노드 지우기  
      -> 노드가 너무 많아지면 2번 왼쪽 그림처럼 과적합이 될 확률이 높다. 이럴 때 Pruning을 통해 하부 트리를 제거하여 일반화 성능을 높일 수 있다. pruning을 하게 되면 깊이가 줄어들고 결과의 개수가 줄어듭니다.

3. SVM(Support Vector Machone) : 개별 클래스간의 촤대 마진을 효과적으로 찾아 분류  
    - SVM? 클래스를 분류할 수 있는 다양한 경계선 중 최적의 라인을 찾아내는 알고리즘
    - SVM의 구성  
        - Support vector (서포트 벡터): 데이터 중에서 결정 경계와 가장 가까이에 있는 데이터들의 집합  
        - Decision Boundary (결정 경계): 데이터 분류의 기준이 되는 경계  
        - Margin (마진): 결정 경계에서 서포트 벡터까지의 거리(여유 공간)  
        - Hyperplane (초평면): $n$차원 공간의 $(n-1)$차원  
        - Slack Variables (슬랙 변수/여유변수): 완벽한 분리가 불가능할 때 분류를 위해 허용된 오차에 대한 변수  
      
    

    - #### 최적의 선을 찾는법
    > Q. 다음중 어떤 선(결정 경계)이 가장 좋은선인가?   
    A. 정답은 F, 결정 경계는 데이터로부터 가장 멀리 떨어져 있는게 좋다. 따라서 SVM은 Margin (거리)이 가장 큰 경우를 선택함으로써 최적의 선을 찾는다  


    - SVM의 장단점  
        | 장점 | 단점 |  
        |---|---|  
        | 서포트 벡터만을 이용해서 결정 경계를 생성하므로 데이터가 적을 때 효과적 | 데이터 전처리 과정이 중요 |  
        | 새로운 데이터가 입력되면 전체 데이터 포인트와의 거리를 계산하지 않고 서포트 벡터와의 거리만 계산하면 되기 때문에 연산량 최소화 | 데이터가 많아질수록 최적화를 위한 평가 과정이 많아져서 속도가 느림 |  
        | 정확성이 뛰어나며, 비선형 모델 분류 가능 (커널 사용시) | |  
        | 마진을 최대화하는 방식이 규제 역할을 하고 노이즈의 영향이 적음 | |  

4. > KNN(K-Nearest Neighborhood) : 근접 거리 기준으로 분류  
    KNN : 데이터로부터 거리가 가까운 k개의 다른 데이터 레이블을 참조하여 분류하는 알고리즘
    - K-NN 알고리즘 예시  
       
    
    1. 그림과 같이 6개의 기존 데이터 A~F와 1개의 신규 데이터 N이 있고, 주변에 있는 이웃의 개수를 K라고 한다.  
    2. K=1이라면, 거리가 첫 번째로 가까운 C를 보고 신규 데이터인 N을 동그라미 그룹으로 분류한다.  
    3. 하지만, 만약 K=3이라면 신규 데이터인 N과 거리가 세 번째로 가까운 C, D, E를 보고, N과 가장 가까운 이웃들이 주로 속해있는 클래스로 N을 분류한다. 이때 그룹이 나눠지면 다수결의 원칙에 따른다. 여기서는 1:2가 되어 N은 세모로 분류된다.

    - KNN의 장단점  
        | 장점 | 단점 |
        |---|---|
        | 범주를 나눈 기준을 알지 못해도 데이터를 분류할 수 있음 | 다른 분류 모형과 달리 테스트 데이터의 개수에 따라 시간이 오래 걸림 |
        | 입력 데이터만 주어지면 바로 예측값을 구할 수 있어서 다른 분류 모형과는 달리 학습 과정이 필요 없음 | 학습 데이터 모두를 거리 계산에 사용하기 때문에 학습 데이터의 양도 계산 시간에 영향을 미침 |
        | 이해가 쉽고, 구현이 간단 | K의 값 결정이 어려움 |
        | 추가된 데이터의 처리가 용이 | 수치형 데이터가 아니면 유사도를 정의하기 어려움 | 
        | 훈련 데이터를 요약하거나 일반화된 모델로 만들지 않고, 원본 데이터 전체를 그대로 저장 | 데이터 내 이상치가 존재하면 분류 성능에 큰 영향을 받음 |


### 분류 평가 지표   

1. > 혼동 행렬 : 분류 모델의 예측 결과를 정확한 예측과 잘못된 예측으로 구분하여 나타낸 표(행렬)  
  
-   - 혼동 행렬 : 

    - 혼동 행렬을 이용한 분류 모델 평가 지표  
        - 정확도 (Accuracy): 모든 가능한 예측 중 참인 비율  
            (TP+TN)/(TP+TN+FP+Fn)  
              

        - 정밀도 (precision): 참이라고 예측한 경우 실제 참의 비율  
            TP/(TP+FP), 거짓을 참으로 판단한 정도를 알 수 있다  
        - 재현도 (recall): 실제로 참인 경우 중 참으로 예측하는 비율  
            TP/(TP+FN), 참을 거짓으로 판단한 정도를 알 수 있음  
        *정밀도와 재현도는 서로 Trade-off 관계*

    - 정밀도와 재현도 예시  
    1. 암 환자 판단 : 암 환자인데 암 환자가 아니라고 예측하는것이 치명적. 즉, FN을 줄여야 하므로 Recall을 높여야 한다.  

    2. 스팸 메일 분류기 : 스펨이 아닌 메일을 스팸으로 분류하는 현상(FP)이 치명적. Precision을 높여아한다. 

    **그렇다면 Recall과 Precision을 어떻게 조절하는가?**  
    -> 경곗값(Threshold)를 조절하여, 이를 낮추면 Recall이 증가, 높이면 Precision이 증가!  
    정밀도&재현도 그래프가 Threshold 결정에 도움을 줄 수 있다  
      

2. F1-Score  
    - F1-Score : 정밀도(precision)와 재현율(recall)의 조화 평균  
    $\text{F1 Score} = \frac{2}{\frac{1}{\text{Precision}}+\frac{1}{\text{Recall}}}$
                     = $\frac{2*\text{Precision}*\text{Recall}}{\text{Precision+Recall}}$  
    F1 Score는 머신러닝 모델의 성능을 평가하는 주요지표. Precision(정밀도)과 Recall(재현율) 간의 균형을 효과적으로 평가하기 위해서 조화평균을 사용한다.

3. ROC / AUC : 분류 모델의 성능 평가 도구  
    - ROC Curve : 얼마나 분류가 잘 되었는가를 보여주는 그래프  
        - True Positive Rate (TPR) : 참인 것들 중에 참이라고 예측한 비율 (=Recall)  
        - False Positive Rate (FPR) : 거짓인 것들 중에 참이라고 잘못 예측한 비율   
               
            우측 그림의 빨간선처럼 그래프가 최상단에 붙을수록 좋은 분류!  

    - AUC Curve : ROC와 x축 사이의 면적(적분값)  
        -> 모델의 성능을 숫자로 나타낼 수 있다. 0.5에서 1사이의 값을 가지며, 1에 가까울수록 분류 성능이 좋은 모델!   
            

4. 다중분류평가지표 (참고) : 이진 분류 평가지표를 사용해서 클래스별 점수를 구한 뒤, 이를 적절히 평균낸 것  
    - Macro Average : 모든 클래스에 동등한 가중치 부여  
    - Weighted Average : 각 클래스의 샘플 수로 가중치 부여  
    - Micro Average : 모든 클래스의 예측 결과를 더하여 전체적인 성능 평가(샘플 하나하나에 동등한 가중치 부여)  


### 하이퍼파라미터 최적화   
1. 하이퍼파라미터 최적화 : 적절한 하이퍼파라미터를 찾아 모델 성능을 향상시키는 것
    - 하이퍼파라미터(Hyperparameter) : 모델 학습 과정에 반영되는 값으로, 학습 시작 전에 사용자가 직접 설정하는 변수  
    **하이퍼파라미터 최적화 과정**  
    1. Hyperparameter 탐색 범위 설정 (최적 값을 찾고 싶은 하이퍼파라미터의 범위를 설정)  
    2. 평가 지표 계산 함수(성능 평가 함수) 정의(탐색하려는 Hyperparameter를 인수로 받아 평가지표 값을 계산해주는 함수를 정의)  
    3. 설정한 범위에서 값을 샘플링한 Hyperparameter 값을 사용하여 검증 데이터로 정확도 평가  
    4. 위 단계를 특정 횟수 반복하며, 정확도 결과를 보고 하이퍼파라미터의 범위 설정  

2. 하이퍼파라미터 최적화 방법  
    - Grid Search: 정해진 범위에서 Hyperparameter를 모두 순회하며 가장 좋은 성능을 내는 값을 찾는 기법  
        - 장점 : 범위가 넓고 step이 작을수록 꼼꼼하게 전 범위를 탐색하니 최적해를 정확히 찾을 수 있다.  
        - 단점 : 시간이 너무 오래 걸린다.  
        - 적용 : 넓은 범위, 큰 step을 활용해 범위를 좁힌다.  

    - Random Search: 정해진 범위에서 Hyperparameter를 무작위로 탐색  
        - 장점 : 속도가 Grid Search보다 빠르다.  
        - 단점 : 무작위라는 한계 때문에 정확도가 떨어진다. 따라서 Grid Search나 Bayesian Optimization에 비해 사용 빈도가 적다.  

    - Bayesian Optimization: 사전 정보를 바탕으로 Hyperparameter 값을 확률적으로 추정하며 탐색  
        - 특징 : "Gaussian Process"라는 통계학을 기반으로 만들어진 모델, 여러 개의 하이퍼 파라미터들에 대해서 "Acquisition Function”을 적용했을 때, "가장 큰 값"이 나올 확률이 높은 지점을 찾아낸다.  
        - 방법  
        1. 초기 값 몇 개를 무작위로 실험해본다.  
        2. 그 결과를 바탕으로 어떤 파라미터의 조합이 좋을 지를 확률 모델로 예측한다  
        3. 다음 실험할 조합을 가장 좋을 것 같은 위치로 선택한다  
        4. 결과를 반영하고 다시 반복
