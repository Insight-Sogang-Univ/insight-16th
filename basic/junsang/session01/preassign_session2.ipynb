## EDA 개념
> EDA(Exploratory Data Analysis) : 데이터 분석(모델링)을 위해 데이터를 여러가지 방식으로 파악하는 모든 과정
 - 시각화 같은 도구를 통해 패턴을 발견
 - 데이터의 특이성 확인
 - 통계와 그래프(혹은 시각적 표현)으로 가설 검정하는 과정  

1. EDA 대상 : 일변량 or 다변량?
 - Univariate(일변량) : 데이터를 설명하고 그 안에 존재하는 패턴을 찾는것이 주 목적
 - Multivariate(다변량) : 여러 변수들간의 관계를 보는것이 주 목적  
 **여러 변수를 확인하기 전 개별 데이터를 파악하는 것이 오류 대처에 용이**  

2. EDA 종류 : 시각화 or 비시각화
 - Graphic(시각화) : 차트 혹은 그림을 이용하여 데이터를 확인하는 방법    
 - Non-Graphic(비시각화) : 그래프적인 요소 없이 주로 Summary Statistics를 통해 데이터를 확인하는 방법    
대략적인 형태가 필요하다면? Graphic이, 정확한 값이 필요하다면? Non-Graphic이 유리  

3. EDA 유형 : EDA 대상과 종류에 따라 EDA유형이 구분된다. (변수가 몇개인지/시각화여부)
 - Uni-Non Graphic(일변량 비시각화) : 주어진 데이터의 Distribution을 확인하는 것이 주목적.
 - Uni-Graphic(일변량 시각화) : 주어진 데이터를 전체적으로 살펴보는 것이 주목적.
 - Multi-Non Graphic(다변량 비시각화) : 주어진 둘 이상의 변수 간 관계를 확인하는 것이 주목적.
 - Multi-Graphic(다변량 시각화) : 주어진 둘 이상의 변수 간 관계를 전체적으로 살펴보는 것이 주목적.  

## 데이터 확인
1. 절대경로 : 처음(root 파일)부터 시작하여 목적지까지 전체적인 경로(URL), 해당 파일까지의 모든 경로
 - ex) C:\Users\사용자 이름\OneDrive\Desktop  

2. 상대경로 : 현재 작업하고 있는 디렉터리를 기준으로 상하위 디렉터리와 같이 상대적으로 표현되는 경로
 - **./ : 현재 디렉터리**
- ../ : 상위 디렉터리
- / : 최상위(루트) 디렉터리  
  


## 데이터셋 파악하기  

1. 데이터 프레임 보기  
2. 데이터 변수 확인 : 변수 = 데이터 프레임의 column  
    - 변수 정의 확인 : 어떤 정보를 가지는 변수인지 확인
    - 변수 유형 확인 : 질적·범주형(Categorical Data)과 양적·수치형(Numerical Data)으로 구분
    - 변수 데이터형식(자료형) 확인 : 날짜, 수치, 텍스트, 이미지 등의 구분

3. 데이터 분포 확인  
 - 단변수 분석은 원시 데이터(raw data)의 평균값, 최빈값, 중간값 등 변수들의 분포를 산포도, 박스 플롯, 히스토그램 등의 그래프를 통해 단변수, 즉 하나의 데이터 분포를 확인해 분석할 수 있는 방법  
 

## 전처리   
> 데이터 전처리 : 데이터를 분석에 적합한 형태로 만드는 과정  
 - 전처리의 여러 기술 : 데이터 정제, 데이터 변환, 데이터 필터링, 데이터 통합
### 결측값 처리  
 - 결측값 = 데이터 수집 과정에서 측정되지 않거나 누락된 데이터  
    1. 결측값 확인 : 데이터 프레임에서 **Nan, ?, 0** 등으로 나타난 결측값 확인 - 진짜 0의 값인지? 결측되어 0으로 표기된것 뿐인지?  
    2. 결측값 처리 : 삭제 - dropna() : DataFrame내의 결측값이 포함된 레이블을 제거 / 대체 - fillna() : 결측값 대체
### 이상치 처리  
 - 이상치 : 관측된 데이터의 범위에서 많이 벗어난 값  
   

1. 이상치 확인 : 통계를 통해 확인 - describe() / 시각화를 통해 확인 - BoxPlot / Z-score를 통해 확인 / Tukey Fences를 통해 확인  
    - Tukey Fences? 이상치를 식별하는 통계적인 방법 중 하나로, 데이터의 사분위 범위(IQR, Interquartile Range)를 기반으로 합니다. IQR은 데이터의 중간 50% 범위를 측정하며, 이는 데이터의 상위 25%와 하위 25%를 제외한 중간 범위를 의미합니다.  
      
2. 이상치 처리 
    - **전체삭제** : 이상값이 Human error에 의해 발생한 경우 : 해당 관측치 삭제
    - **다른값으로 대체** : 관측치 표본이 너무 작은경우 관측치 삭제시 신뢰성 문제 발생 가능. 이 경우, 삭제하지 않고 평균 등 다른값으로 대체하거나, 예측 모델을 만들어 이상값으로 예측 후 대체 
    - **변수화** : 이상값이 자연발생한 경우, 해당 값이 발생한 사유를 파악한 후 변수화 하여 모델에 포함 가능
    - **리샘플링** : 해당 이상값을 포함하는 경우와 포함하지 않는 경우로 구분하여 모델을 모두 만들고 각각의 모델에 대해 분석

## 피처 엔지니어링(변수가공) 
> 해결하고자 하는 문제를 컴퓨터가 잘 이해할 수 있도록 피처(변수)들의 형태를 변형하거나 적절하게 처리하는 과정  *cf) 피처(feature) = 입력 변수 = column*
  
1. 레이블 인코딩(Label Encoding) vs 원핫인코딩(One-hot Encoding)  
기계는 숫자만 이해하고 텍스트는 이해하지 못하기에 텍스트로 주어지는 값들을 숫자로 바꾸는 작업을 수행해야 합니다.  
  
 - Label Encoding : 범주형 변수를 0부터 N-1까지 숫자로 변환
 - One-hot Encoding : 범주형 변수를 이진 벡터로 변환(표현하고 싶은 단어의 인덱스는 1, 다른 인덱스는 0)
2. 구간화(Bining)  
연속 데이터를 그대로 사용하기 보다는 일정한 구간으로 나눠서 분석하는 것이 효율적인 경우에 사용, 분석 목적과 방법에 따라 분석가의 도메인을 최대한 활용해서 수행하는것이 좋습니다.  
3. 파생변수 생성  
기존의 피처를 다른 피처로 변환(재정의)하여 변수를 추가, 기존 데이터를 다른 관점에서 분석해 새로운 특성과 정보를 얻기 위한 과정
Ex) '검수일자' feature를 변환하여 '검수요일' feature' 추가 가능  
  

4. 스케일링  
서로 다른 변수의 값 범위를 일정한 수준으로 맞추는 작업, 숫자의 상대적인 크기로 인한 왜곡 방지를 위해 정규화(normalize) 작업 수행  
ex) StandardScaler(표준화) - 흔히 말하는 정규화 , MinMaxScaler - 데이터를 0과 1사이의 분포로 표현


## 데이터 시각화
> 데이터 시각화(Data Visualization)는 데이터에 대한 이해를 돕기 위해 그림, 도형 등 그래픽 요소들을 이용해 데이터를 묘사하고 표현하는 과정입니다    

시각화를 통해 숫자와 텍스트로 표현되는 정보를 그래프, 차트, 그림, 도표 등의 시각적 요소로 변환하고, 이를 통해 데이터 패턴, 관계, 추세 등을 쉽게 파악 가능
- **How?**
Matplotlib : Python 프로그래밍 언어 및 수학적 확장 NumPy 라이브러리를 활용한 플로팅(그래프를 그리기 위한) 라이브러리
Seaborn : matplotlib을 기반으로 만들어져 통계 데이터 시각화에 최적화된 인기 라이브러리
