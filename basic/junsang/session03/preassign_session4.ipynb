## - 머신러닝이란?  
> 머신러닝(Machine Learning)이란? 인공지능의 한 분야로, 모델이라는 소프트웨어를 학습하여 새로운 데이터를 예측하거나 결정을 내릴 수 있도록 하는 기술이다. 데이터에서 패턴을 발견하는 것과 이 패턴을 바탕으로 새로운 데이터에 대해 예측하는 것을 가능하게 한다.

머신러닝의 종류  

||정답여부|방식|예시|  
|---|---|---|---|
|지도학습|정답(label)이 있는 데이터 사용|기계의 예측값이 데이터의 정답과 같아지도록 학습시키는 것|회귀, 분류|  
|비지도학습|정답(label)이 없는 데이터 사용|데이터의 패턴 혹은 유사성을 학습시키는 것|군집화|  
|강화학습|보상|시행착오를 반복하여 보상을 최대화 하는 행동 학습|알파고, 자율 주행|  

### 모델링(Modeling)
> 모델이란? : 학습의 결과물 | 학습을 마친 인공지능 그 자체(예측기)  
데이터를 학습하며 찾아낸 패턴이 저장된 상태이며, 새로운 데이터를 넣었을 때 답을 주는 예측기 역할  
*ex) 공부시간 $x$를 넣었을 때 시험 점수 $y$가 나오는 공식 $y=ax+b$*  

> 모델링이란? : 예측기(공식)를 만들어가는 전체적인 과정  
기계가 주어진 데이터 속에서 스스로 패턴을 찾아내, 오차가 작은 최적의 공식을 완성할때까지 학습하는 과정  

> 모델링의 과정
> 1. 설계 : 데이터의 특성에 맞추어 어떤 알고리즘으로 학습할지 결정하는 단계  
> 2. 훈련 : 데이터를 기계에게 계속 보여주며 정답을 맞히도록 수정하는 단계  
> 3. 검증 : 훈련시킨 모델이 새로운 데이터에서도 정답을 잘 맞추는지 테스트하는 단계  
  

### 지도학습 
> 지도학습이란? 정답과 레이블을 함께주고 학습시키는 방법. 에측값(prediction)과 정답(label)이 같아지는 최대한 같아지는 방향으로 학습  
> 지도학습의 과정
 - 지도학습을 위해선 data가 labeled data form이어야 함 *labeled data -> 독립변수(입력)와 종속변수(출력)가 모두 존재하는 데이터*  
1. 데이터 준비 : 독립변수(입력)와 종속변수(출력)를 갖춘 데이터를 준비
2. 예측 : 컴퓨터가 독립변수를 보고 종속변수를 예측  
3. 오차계산 : $예측값과 실제값의 차이=잔차$ 잔차를 모아 만든 지표 -> 손실(loss)  
4. 가중치 수정 : 오차가 줄어드는 방향으로 내부 수치(가중치)를 스스로 수정
5. 반복 : 2->3->4 를 반복하며 오차를 줄여나간다  
6. 학습완료 = 함수 획득 : 데이터 사이의 통계적 관계가 하나의 함수 형태로 남는다, 이후 새로운 데이터가 들어와도 규칙에 따라 결론 도출 가능(일반화)  
7. 성능평가 : 학습때 사용되지 않은 별도의 테스트 데이터로 얼마나 잘 맞는지 측정  


## - 선형회귀(단순, 다중)  
> 회귀(Regression) : 데이터에서 패턴을 찾아내어 미래값을 예측하는것, 종속변수가 연속적일때 사용 *ex)집값, 주식 가격...*  
 - 선형회귀, 비선형회귀, 규제회귀, 다항회귀 등 다양한 기법 존재!

> 선형회귀 : 데이터를 가장 잘 설명하는 회귀선(직선)을 찾는 과정  
  

- 가중치 $w$ (Weight) : 독립변수 $x$가 종속변수 $y$에 주는 영향력의 크기, (기울기 or 회귀 계수)  
- 편향 $b$ (Bias) : 독립변수 $x$가 0일 때 나오는 기본값, (y절편) 
1. 단순회귀(독립변수 $x$가 한 개)  
    데이터를 직선 형태로 표시 $y=wx+b$ , 잔차를 최소로 하는 $w,b$값을 찾는것이 목표
    **How?** 최소제곱법(Least Square Method), 경사하강법(Gradient Descent)  
    - 최소제곱법 = 최소자승법(OLS, Ordinary Least Square) : 잔차(Residual)의 제곱의 합 (RSS)이 최소가 되는 $w,b$값을 구함  
    - 경사하강법 : To be continued...  
    
  

2. 다중회귀(독립변수 $x$가 두 개 이상)  
 다중 선형회귀는 직선이 아닌 평면형태! 
  

 수식적 이해 $y_i = \beta_0+\beta_1X_{1i}+\beta_2X_{2i}+...+\beta_kX_{ki}+\varepsilon_i $  
 If 상수항과 오차항($\beta_0, \varepsilon_i$) 제거시, $y=w_0x_0+w_1x_1+\cdot\cdot\cdot+w_mx_m=\sum_{i=0}^mw_ix_i=w^Tx$ form  
 i.e. 독립변수들의 가중평균!  

**선형회귀를 모든 상황에서 쓸 수 있는것은 아니다!**  
선형회귀 사용을 위해 다음의 6가지 가정이 필요  
1. 선형성(Linearity) : 종속변수와 독랍변수 사이의 관계가 선형적이어야 한다  
2. 독립성(Independence) : 각각의 독립변수가 서로 linearly independent이어여 한다. If not -> 다중공선성이 있다!  
3. 오차항의 평균은 0이다 : $E(\varepsilon_i|X_i)=0$
4. 등분산성 (Homoscedasticity) : 오차항의 분산이 일정해야 한다 $Var(\varepsilon_i|X_i)=\sigma^2$  If not -> 이분산성(heteroscedasticity)이 있다~  
5. 오차항은 자기상관되어 있지 않다(Non-auto correlation) : 오차항의 공분산은 항상 0이어야 한다 $Cov(\varepsilon_i,\varepsilon_j|X_i)=0$  
    *시계열 자료에서 주로 발견됨!*  
6. 정규성(Normality) : 오차항이 정규분포를 따른다 $\varepsilon_i ~ N(0,\sigma^2)$  

**6가지의 가정들은 모두 회귀분석 전에 검증하는가?**  
-> 회귀 분석 후에만 검증 가능한 가정들도 있다!  

| 기본가정 | 회귀 분석 전 검증 가능? | 회귀 분석 후 검증 가능? | 검증 방법 |
|---|---|---|---|
| 선형성 | O | O | scattor plot 찍어보기 |
| 독립성(다중공선성) | O | O | VIF 지수, 상관계수 |
| 오차항 평균이 0이다 | X | O | np.mean(residuals) |
| 등분산성 | X | O | 잔차의 도표화 |
| 오차항 자기상관 없다 | (회귀 분석 전에는 간접적으로 검증 가능, 분석 후에 최종 검증) | O | Durbin-Watson 검정(분석 후) |
| 정규성 | X | O | Shapiro-Wilk 검증, Q-Q plot |  

오차항에 대한 가정들 -> 모델 학습 후에만 검증 가능(오차항 자체가 모델 선정 후 정해진다)  
만일 모델 학습 후 검증을 진행하였는데 가정에 위배될 수 있으므로 데이터 분석은 한 번에 끝나는것이 아닌 수정과 검증을 거듭 진행해야한다!

## - 다중공선성(Multicollinearity)  

> 다중공선성 : 독립 변수들간의 강한 상관관계로 인해 회귀 모델이 불안정해지는 문제 *ex) '주택의 크기'와 '방의 개수'라는 두 독립 변수 사이에 강한 상관관계가 있다면 다중 공선성 발생!  
-> 다중공선성이 높은 경우 어떤 독립 변수가 종속 변수에 얼마나 영향을 미치는지 정확하게 구분하기 어렵기 때문에 확인하고 처리하는 과정이 필요!

- **HOW?**  

1. 상관계수(correlation coefficient, $r$)  
> 상관계수 : 변수간의 통계적인 상관관계의 정도를 수치로 나타낸 것  
  

- 상관계수에 따른 상관관계  $-1 \le r \le 1 $ 사이 값을 가진다  
    - $ r>0 $ : 양의 상관관계, 한 변수가 증가하면 다른 변수도 증가하는 경향  
    - $ r<0 $ : 음의 상관관계, 한 변수가 증가하면 다른 변수는 감소하는 경향  
    - $ r=0 $ : 두 변수 사이에 직접적인 연관성이 없다
- 히트맵 or corr()함수를 통해 상관계수 확인 가능  
  

- pairplot 같은 시각화로 산점도를 찍어서도 확인 가능  
  

2. VIF지수(분산 팽창 인수) : 회귀 모델의 결정계수 R-square를 사용하여 계산한다  
- VIF가 높으면 다중공선성이 존재한다고 판단!  
    - VIF = 1 : 해당 독립 변수는 다른 변수와 상관관계가 전혀 없다  
    - VIF < 5 : 일반적으로 다중공선성 문제가 없다고 간주한다  
    - VIF > 5 : 다중공선성의 징후가 있을 수 있으며, 주의가 필요하다  
    - VIF > 10 : 다중공선성이 심각하다고 간주되며, 이 변수를 포함한 회귀 모델은 불안정할 가능성이 크다. 이 경우 해당 변수를 제거하거나, 변수 간의 상관관계를 줄이기 위한 조치가 필요할 수 있다  

### 그렇다면 다중 공선성 발생시 대처법은??  
1. 변수 제거  
2. 변수 변환  
 - 변수들을 더하거나 빼서 새로운 변수 생성(독립변수를 더하거나 빼더라도 문제가 없는 경우)
3. 규제선형모델 활용  
 - 릿지, 라쏘, 엘라스틱넷 등의 방법을 통해 모델의 복잡도를 줄일 수 있다  
4. PCA(주성분 분석)  
 - 데이터의 차원을 축소하는 통계적 기법, 고차원 데이터에서 중요한 정보를 최대한 보존하며 데이터의 복잡성을 줄이는것이 목적  (*i.e. 상관있는 변수들을 묶어 더 적은 수의 대표축으로 변환*)  
   


## - 규제선형모델  
  
 각각 underfitting, good fitting, over fitting -> 두번째 모델이 가장 좋음!  
 -> 과적합(Overfitting)모델은 학습 데이터(train-set)에는 완벽히 맞지만, 새로운 데이터(data-set)에서는 성능이 크게 떨어진다  
> 규제선형모델이란? 모델이 학습 데이터에 과적합(overfitting) 되지 않도록 규제를 가하고자 등장한 모델 
    → 선형회귀 모델에서는 특성에 곱해지는 계수(기울기)의 크기를 조정하는 것   

 학습 데이터의 잔차오류를 최소화하는 RSS랑 과적합 방지를 위해 회귀 계수 값이 커지지 않도록 하는 방법간의 균형 유지가 필요!  
 -> 오류는 최소화하면서, 특정 변수에 집착하지 않도록 균형을 맞춰 과적합 방지가 목적  

 **종류는?** 
   1. L2규제 : W(회귀계수)의 제곱에 대해 페널티를 부여하는 방식, 릿지(Ridge)회귀  
     예측 변수가 많을 때, 다중공선성이 필요할 때 사용
    회귀계수의 제곱을 페넡티로 부여하기에 회귀계수가 클수록 더 큰 페널티를 받는다. 모든 변수를 유지하며, 크기를 적절히 줄여 과적합 방지(모든 계수가 여전히 남아 있다!)

   2. L1규제 : W(회귀계수)의 절댓값에 대해 페널티를 부여하는 방식, 라쏘(Lasso)회귀  
    예측 변수 수가 많고, 그 중 일부만이 실제로 중요할 때, 모델의 해석을 간단히 유지하고자 할 때 사용  
    회귀계수의 절댓값을 페널티로 부여하기에 회귀계수의 크기와 상관없이 같은 크기의 페널티를 받는다. 만약 계수가 작았다면, 페널티로인해 변수가 0이되어 사라질 수도 있다  

   3. L2 + L1 : 엘라스틱넥(Elastic Net)규제  
    예측 변수 수가 많고, 그 중 중요한 변수를 선택하면서도 다중공선성을 관리할 때 사용  
    L1의 변수 선택 기능(일부 계수 삭제)과 L2의 규제 기능(큰 계수를 줄임)을 동시에 활용하는 모델.  
    L1은 상관관계가 높은 변수 중 일부만 선택하는 경향이 있으나, L2를 함께 적용하면 상관관계가 높은 변수를 함께 고려하면서 규제 가능!

  


## - 모델평가방법  
> Q. 열심히 만든 회귀 모델의 성능과 정합성은 어떻게 평가해야 좋은가?  
  A. 회귀분석의 결과가 얼마나 유효한지 평가하기 위해서는, 회귀선을 데이터와 함께 시각화하거나 통계지표를 이용한다!  

1. 시각화   
   

 회귀선이 데이터를 전반적으로 잘 요약하는지 한눈에 파악가능하다. 그러나, 둘 중 어느 회귀선이 해당 데이터를 더 잘 요약하는지 객관적인 비교가 어렵다.  
 따라서, 통계지표를 확인하여 두 집단의 차이의 정도를 객관적으로 비교하여 검정해야한다  

 2. 통계지표  
  - 모델 유의성 검정  
    1. F-statistic, Prob(F-statistic) : 회귀 모델이 유의미한지 검정하는 지표(종속변수와 독립변수간의 관계가 통계적으로 유의미한지)  
        - $H_0$ : 모든 회귀계수가 0이다.(독립변수들이 종속변수들과 관계가 없다)  
        - $H_1$ : 적어도 하나의 회귀계수는 0이 아니다.(독립 변수 중 하나 이상이 종속 변수에 영향을 준다)  
        -> F-statistics의 값이 클수록 통계적으로 유의미, Prob(F-statistics)=p-value  
    
    2. T-statistic, Prob(T-statistic) :각 독립변수가 종속변수에 유의미한 영향을 미치는지 판단하는 지표  
        - $H_0$ : 해당 변수의 회귀계수가 0이다.(해당 독립변수는 종속변수에 영향을 미치지 않는다)  
        - $H_1$ : 해당 변수의 회귀계수가 0이 아니다.(해당 독립변수는 종속변수에 유의미한 영향을 준다)  
        -> p-value > |T-statistics| 을 기준으로 가설 검정 수행, p-value가 더 큰 경우 Reject $H_0$  

  - 모델의 성능 평가   
    1. 결정계수(R-square, $R^2$) : 회귀 모델이 종속 변수의 전체 변동성을 얼마나 잘 설명하는지 나타내는 지표  
      $ R^2 = \frac{SSR}{SST} = 1 - \frac{SSE}{SST} $  
      SST : 평균($\bar{Y}$)으로 예측했을때 총 오차, SSE : 모델($\hat{Y}$)이 예측하고도 여전히 남은 오차, SSR : 평균에 비해 Regression 모델이 얼마나 더 잘 맞혔는지  
      $ 0\le R^2 \le 1$ , $R^2$가 1에 가까울수록 모델이 데이터를 잘 설명한다  

    2. 조정된 결정계수(Adjusted R-squared, Adjusted $R^2$)  
     $R^2$는 독립변수의 개수가 증가할수록 자연스럽게 증가한다. 즉, 데이터와 큰 관련 없는 변수를 추가해도 $R^2$은 증가  
     변수의 개수에 대한 영향을 조절한 지표가 Adjusted $R^2$, $\bar{R}^2 = 1- \frac{SSE/(n-k)}{SST/(n-1)}$ (n:샘플수, k:독립변수의 갯수)  

    3. AIC, BIC(SC)  : 두 지표 모두 정보기준(information criteria)라고도 불리며, 값이 낮을수록 좋다.  
        AIC(Akaike information criterion) : BIC에 비해 복잡성에 대한 페널티가 비교적 작다.  
        BIC(SC) :  SC = Schwarz Criterion, BIC = Bayesian Information Criterion : AIC보다 엄격한 기준으로, 데이터의 양에 따라 더 강한 페널티를 부과  

3. 성능 평가 지표  
    - 평균 제곱 오차(Mean Square Error, MSE) : 예측값과 실제값 사이의 오차의 제곱의 평균  $\text{MSE} = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2 $
    - 평균 절대 오차(Mean Absolute Error, MAE) : 모델의 예측값과 실제값 사이의 절대적 오차의 평균 $\text{MAE} = \frac{1}{n} \sum_{i=1}^{n} |y_i - \hat{y}_i|$  
    - MSE는 큰 오차에 민감하게 반응하기에 이상치에 민감하고, MAE는 모든 오차를 동일하게 취급하기에 이상치가 존재하더라도 전반적인 예측 성능 유지 가능  
    -> 모델의 특성, 데이터의 특성, 평가목적에 따라 선택!  

4. 변수 유의성 평가  
    - T-검정 : 독립변수의 회귀계수가 유의미한지 판정  
        앞선 모델 유의성 검정과 유사한 방식으로 진행, 이떄, $t = \frac{\hat{\beta} - 0}{SE(\hat{\beta})}$  
        마찬가지로 p-value와 유의수준간의 비교를 진행하여, 귀무가설 기각 여부 판단  
        - 회귀계수가 0이면 해당 독립변수는 의미 없는 변수  
        - p-value가 작을수록 해당변수는 종속변수에 더큰 영향을 줄 확률이 높아진다.  
        - t-값이 크면 귀무가설을 기각할 가능성이 높아진다.

3. 로그 회귀 모델(Log Regression Model)  
 - $ Y = a + blog(X)의 형태  

4. 스플라인 회귀(Spline Regression) : 데이터를 구간별로 나누고, 각 구간에서 다른 선형 또는 비선형 함수를 적용하여 예측하는 방식  
 - $ Y = \begin{cases} a_1 + b_1X & \text{if } X < c_1 \\ a_2 + b_2X & \text{if } c_1 \le X < c_2 \\ \dots \\ a_n + b_nX & \text{if } X \ge c_{n-1} \end{cases} $  
    이때, 구간 경계에서 **연속성 유지**가 중요!
