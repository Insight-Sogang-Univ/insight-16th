##데이터를 확인, 추출하는 방법 간단 요약

1. head(n) - 앞, tail(n) -뒤에서 n개의 데이터 확인  
    shape-(행, 열 개수), info() - 정보, describe() - column의 통계값, isnull() - 결측치 여부, dtypes - colimn의 data type  
    count() : 결측치 제외 행 개수, value counts() - 특정 column의 값의 빈도수, unique() - column별 value 중복없이 반환
2. indexing- 지목 , slicing - 범위, loc-인덱스 이름 기준(행, 열, 값 가능), iloc- 정수형 위치 기준 ###loc은 끝 범위 포함!
3. Boolean indexing - 조건문을 통한 데이터 추출, 조건이 True인 행만 추출


##집계함수 사용법  
1. 집계함수란? 여러 행으로부터 하나의 결과값 반환하는 함수. eg. mean, std dev, sum... / NaN 무시
    axis=0 행 방향 연산, axis=1 열 방향 연산
2. 그룹화 : 일부 행을 대상으로 집계함수 연산  
df.groupby('column1'),['column2'].집계함수() -> col1으로 그룹화, 그룹별로 col2에 대해 연산 ###대괄호 1개면 Series, 2개면 DataFrame form
