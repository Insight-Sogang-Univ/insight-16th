## 5차시 사전과제 - 이정은

### 분류
- 기존 데이터가 어떤 레이블에 속하는지 패턴을 알고리즘으로 인지하고, 새로운 데이터가 어떤 레이블에 속하는지 판별하는 것
- 데이터가 범주형 변수를 예측하기 위해 사용됨
- 이진 분류 : 예측하고자 하는 변수가 '참' 또는 '거짓'만을 가질 떄
- 다중 분류 : 예측하고자 하는 변수가 가질 수 있는 값이 3개 이상일 떄

### 로지스틱 회귀
- 이진 분류의 대표 알고리즘
- 샘플이 특정 클래스에 속할 확률 추정하여 기준치에 따라 분류함
- 이진 분류는 출력이 0과 1 사이의 값을 가지면서 S자 형태로 그려지는 함수를 이용해야 함

-> 시그모이드 함수
- 입력값이 커지면 1에 수렴, 입력값이 작아지면 0에 수렴
- 입력값이 음의 무한대 ~ 양의 무한대일 때, 출력값은 0 ~ 1의 값을 가짐
- 출력값이 특정값 이상이면 1, 특정값 이하면 0으로 정해서 이진 분류 문제 풂

### 의사결정나무
- 조건에 따라 데이터 분류하며, 최종적으로 데이터가 순수하면 label의 집합으로 구성될 때까지 분류 반복

-> 구성
- Root Node: 가장 상위에 위치한 노드
- Parent Node: 주어진 노드의 상위에 있는 노드
- Child Node: 하나의 노드로부터 분리된 2개 이상의 노드들
- Binary Tree(이진 트리): Tree 중에서 children이 최대 2개인 tree
- Leaf Nodes: 자식마디가 없는 노드, 모델에서 label에 해당
- Edge: Parent node와 child node를 연결하는 연결선
- Height: 특정 노드에서 가장 멀리 있는 leaf node까지의 경로에 있는 edge 개수
- Depth: Root node에서 특정 노드에 도달하기 위한 edge의 수 

1. CART
- 데이터셋을 임계값을 기준으로 두 child로 나누는 알고리즘
- 임계값은 불순도(지니 계수)가 낮아지는 방향으로 나누기
- 데이터를 분류했을 때 동일한 객체들로만 잘 모아지게 하는 기준 형성이 중요

-> 불순도
- 분류하려는 데이터 집합에서 서로 다른 범주가 섞여 있는 정도
- 지니 계수 활용 (작아질수록 순수)

-> 실제 학습 시 고려할 사항
- Hyperparameter 설정 
- 시각화 : 분류가 잘 이루어졌는지 확인
- 가지치기 : 노드가 너무 많아지면 과적합이 도리 수 있으므로 불필요한 노드를 지워야 함

### SVM
- 클래스를 분류할 수 있는 다양한 경계선(초평면) 중 최적의 라인을 찾아내는 알고리즘
- 명확하게 분류할 수 있거나, 고차원 공간에서 효과적으로 사용 가능

-> 초평면 : n차원 공간을 두 부분으로 나누는 한 차원 낮은 부분 공간

-> 구성
- Support vector (서포트 벡터): 데이터 중에서 결정 경계와 가장 가까이에 있는 데이터들의 집합
- Decision Boundary (결정 경계): 데이터 분류의 기준이 되는 경계
- Margin (마진): 결정 경계에서 서포트 벡터까지의 거리(여유 공간)
- Hyperplane (초평면): n차원 공간의 (n-1)차원 평면
- Slack Variables (슬랙 변수/여유변수): 완벽한 분리가 불가능할 때 분류를 위해 허용된 오차에 대한 변수

-> 장점
- 데이터가 적을 떄 효과적
- 새로운 데이터 입력 시 서포트 백터와의 거리만 계산하기에 연산량 최소화
- 정확성 높음
- 마진 최대화 방식이 규제 역할을 하고 노이즈의 영향 적음

-> 단점
- 데이터 전처리 과정 중요
- 데이터 많을수록 최적화 과정 길어져서 속도 느려짐

### KNN
- 비슷한 특성을 가진 데이터끼리 서로 가까이 있다는 점 이용하는 분류 알고리즘
- 가까운 K개의 데이터 기반으로 분류

-> 계산 순서
1. 데이터 준비 : 미리 학습 과정 X
2. K 값 설정 : K는 가장 가까운 이웃의 개수이며, 홀수로 설정
3. 거리 계산 : 새로운 데이터가 주어지면, 해당 값과 기존 모든 데이터 간의 거리 계산
4. 가장 가까운 K개의 이웃 선택 : 가장 가까운 이웃
5. 분류하기 : K개의 이웃 중 가장 많이 등장하는 클래스

-> K 선택 방법
- 학습용 데이터 개수의 제곱근
- K가 너무 크면 과소 적합
- K가 너무 작으면 이상값, 잡음 데이터와 이웃 가능

-> 장점
- 범주 분류 기준 몰라도 데이터 분류 가능
- 학습 과정 X
- 이해 쉽고, 구현 간단
- 추가 데이터 처리 용이
- 훈련 데이터 일반화 또는 요약 없이, 원본 그대로 저장 가능

-> 단점
- 테스트 데이터 개수 따라 시간 오래 걸림
- 학습 데이터 모두의 거리를 계산하기에 학습 데이터의 양도 계산 시간에 영향
- K 값 설정 어려움
- 유사도 정의 어려움
- 이상치에 영향 크게 받음

### 분류 평가 지표

### 혼동 행렬
- 분류 모델의 예측 결과를 정확한 예측과 잘못된 예측으로 구분하여 나타낸 표
- TP, TN, FP, FN으로 구성

-> 정확도
- 모든 가능한 예측 중 참인 비율
- (TP+TN)/(TP+TN+FP+FN)
-> 정밀도 : 참이라고 예측한 경우 실제 참의 비율
-> 재현도 : 실제로 참인 경우 중 참으로 예측하는 비율
- 정밀도와 재현도는 tradeoff 관계

### F1-Score
- 정밀도와 재현율의 조화 평균
- 정밀도와 재현율 간의 균형을 효과적으로 평가 가능

### ROC / AUC
- 분류 성능 평가하는 도구

-> ROC Curve
- 얼마나 분류가 잘 되었는가를 보여주는 그래프
- TPR : 참인 것들 중에 참이라고 예측한 비율 
- FPR : 거짓인 것들 중에 참이라고 잘못 예측한 비율
- 어떤 분류를 해도 만족할 수 없는 부분이 나오기에, 현실적인 한계 고려하여 적절한 선 찾아야함

-> AUC Curve
- ROC와 x축 사이의 면적(적분값)
- 0.5 ~ 1 사이의 값을 가지며, 1에 가까울수록 분류 성능이 좋은 것

### 하이퍼파라마터 최적화
- 하이퍼파라미터 : 학습 시작 전 사용자가 직접 설정하는 변수
- 최적화 : 적절한 하이퍼파라마터 찾아 모델 성능 향상시키는 것

-> 과정
- 하이퍼파라미터 탐색 범위 설정
- 평가 지표 계산 함수 정의
- 설정한 범위에서 값을 샘플링한 하이퍼파라미터 값을 사용하여 검증 데이터로 정확도 평가
- 윗 단계 반복하며, 정확도 결과 보고 하이퍼파라미터 범위 좁힘

-> 최적화 방법
1. Grid Search 
- 정해진 범위에서 하이퍼파라미터를 모두 순회
- 장점 : 꼼꼼하게 전 범위 탐색해서 최적해 찾기 가능
- 단점 : 시간이 오래 걸림
2. Random Search 
- 정해진 범위에서 하이퍼파라미터 무작위로 탐색
- 장점 : 속도 빠름
- 단점 : 무작위이기에 정확도 떨어짐
3. Bayesian Optimization 
- 사전 정보 바탕으로 최적 하이퍼파라미터 값 확률적으로 추정
