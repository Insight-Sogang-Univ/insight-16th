판다스 
- 데이터프래임을 기본 틀로 사용
- 데이터 분석에 특화된 파이썬 라이브러리
- numpy의 확장판

판다스 불러오기 
- import pandas as pd
- 데이터 불러오기 : df=pd.read_csv('파일 경로')

데이터프레임
- 판다스의 주요 데이터 구조
- 컴퓨터로 정보를 정리하고 저장하는 도구

1. index, column
- 데이터의 위치를 나타내는 역할
- (x,y) 좌표
2. series
- 1차원 자료구조
- 값과 인덱스로 구성
- 시리즈를 구성하는 값은 모두 같은 데이터타입
- pd.Series()
if 데이터타입 다르면 -> object 표시
3. dataframe
- 2차원 배열구조
- 여러 개의 시리즈가 모여 구성
- 시리즈 단위로는 데이터 타입 동일, 시리즈 간에는 데이터 타입 동일 무관
-pd.DataFrame()


데이터 확인
- 데이터의 특징 미리 파악
- df.head[], df.tail[] : 앞, 뒤 데이터 미리 확인
- 데이터에 대한 정보 확인 : df.shape, df.info, df.describe, df.isnull, df.dtypes
- 고유값 및 개수 확인 : df.count, df.value_counts, df.unique()

데이터 추출
- 많은 양의 데이터 중 분석에 필요한 부분만 골라내어 가공
- 인덱싱(단일항목), 슬라이싱(여러항목)
- 행단위 추출 : loc(인덱스 이름 기준), iloc(정수형 위치 기준)
- 열단위 추출 : 열 이름으로 인덱싱
- 불리언 인덱싱 : 특정 조건에 만족하는 데이터 추출

데이터 조작
1. 데이터 추가 : df['새로운 열 이름'] = 리스트
2. 데이터 삭제 : df.drop(labels, axis)
3. 데이터 정렬 : df.sort_index(), df.sort_values(by="열 이름")
4. 데이터 변환 : 열들의 데이터타입 변환, 데이터의 값 변환, 행열 전치 가능
5. 데이터프레임 병합 : concat(두 데이터프레임 이어붙임), merge(공통항목 기준)

집계함수
- 여러 행으로부터 하나의 결괏값 반환
- 결측치는 무시하고 계산
- ex. 평균, 중앙값, 표준편차, 합계, 최대값, 최소값
- 그룹화 : 일부 행을 대상으로 집계 (groupby + 집계함수)
