### 군집화 사전과제 - 이정은

비지도 학습 
- 정답이 없는 데이터 사용
- 데이터 속의 패턴과 유사도를 활용하여 학습
- 군집화, 연관분석
- 정답 레이블이 존재하지 않아 모델의 성능 명확히 측정 어려움
- 목표가 명확하지 않고, 데이터에서 새로운 특성을 찾을 떄 적합

-> 패턴 모를 때 <br>
-> 패턴 변할 때 <br>
-> 열린 문제 해결과 일반화 필요할 때 <br>
ex. 고객 세그먼트 분석, 이상 거래 탐지, 자연어 처리

군집화
- 데이터를 비슷한 특성을 가진 그룹으로 나누는 비지도 학습 기법
- 군집 : 유사한 데이터들의 집합

- vs 지도학습
- 지도학습은 종속변수를 예측하고자 하며, 독립변수와 종속변수를 설정하여 그 사이의 관계 파악
- 군집화는 종속변수를 설정하지 않고, 데이터 내부의 패턴을 인식해 유사성을 가진 그룹으로 나눔
- 사전에 정해진 목표 X

- 군집화 목표
1. 응집도 최대화 : 같은 그룹끼리 최대한 비슷하게
2. 분리도 최대화 : 다른 그룹끼리 최대한 분리

- 군집화 과정
1. 피처 선택 or 추출 : 군집화에 사용할 데이터의 특징 선택
2. 군집화 알고리즘 선택 : 데이터의 특성과 목표에 따라 선택
3. 군집 유효성 검증 : 군집화 평가
4. 결과 해석 : 각 군집의 특성 해석 <br>
-> 군집화는 1~3 과정 이후에 1로 다시 돌아감

### 군집화 과정 1번. 피처 선택

- 군집화에서 주요 고려 사항
1. 변수 유형 이해 
- 군집화에 사용될 피처의 종류와 특성 명확하게 이해
- ex. 연속형인지, 범주형인지
2. 거리 / 유사도 정의와 특성
- 군집화는 데이터 간의 거리 or 유사도를 기반으로 그룹 형성
- 변수 유형에 따라 적합한 거리 측정 방식 상이
3. 차원 축소
- 변수 많아질수록 모델 복잡도 증가, 효율성과 성능 감소
- so 유사한 변수 통합 / 필요없는 변수 제거를 통해 변수 개수 줄임

-  변수 유형 이해
1. 연속형 변수
- 거리 측정: 유클리디안 거리, 맨하탄 거리
- 알고리즘: K-means, 계층적 군집화 
- 주의사항: 단위가 다르면 큰 값의 변수의 영향이 커지므로 스케일링 필수
2. 명목형 변수
- 거리 측정: 해밍 거리(Hamming distance), 자카드 거리 
- 알고리즘: K-modes, 계층적 군집화
- 처리 방법: 원-핫 인코딩, 더미 변수 변환
3. 혼합형 변수
- 알고리즘: K-prototypes, Gower distance 기반 계층적 군집화
- 처리 방법: 각 변수 유형에 맞는 거리 함수를 조합하여 사용<br>

- 거리, 유사도 정의와 측정
- 거리 측정 : 군집화에서는 거리가 가까울수록 유사하고, 멀수록 다르다고 판단하여 군집 간이 유사성과 비유사성을 수치로 계산하는 방법 사용
- 아래 예시 : 점과 점 사이의 거리
1. 유클라디안 거리
- 가장 일반적인 거리 측정 방법
- 직선 거리 의미
- K-means에서 주로 사용
2. 맨하탄 거리
- 각 차원 차이의 절댓값의 합
- 이상치에 덜 민감
3. 코사인 유사도
- 백터 간의 각도 측정
- 텍스트 or 고차원 데이터에 효과적

- 차원 축소
- 차원의 저주<br> 
-> 피처 공간이 너무 커서 알고리즘이 데이터를 효과적으로 훈련할 수 없는 현상<br>
-> 고차원일수록 모든 점이 비슷하게 멀어지기에, 거리 기반 알고리즘의 분별력 감소
- 차원 축소란 변수의 개수를 줄이는 것
- 고차원 -> 저차원으로 줄이면서도 핵심 정보는 지켜내는 것
- 선형 투영 : 고차원에서 저차원으로 선형적으로 데이터 투영 <br>
-> ex. 주성분 분석(PCA), 특잇값 분해(SVD), 랜덤 투영
- 매니폴드 학습(비선형 차원 축소) : 유클리드 거리가 아닌 데이터 포인트들 사이의 곡선 거리를 고려하며 학습 <br>
-> ex. t-SNE, UMAP, Isomap

- PCA
- 다차원 데이터에서 가장 분산이 큰 방향을 찾고, 그 방향을 기준으로 데이터를 투영하여 차원 줄이는 것
- 데이터의 분산 최대한 보존하면서 차원 줄임
- 상관관계 높은 피처 결합하여 더 적은 수의 피처로 데이터 표현
- 고차원 데이터에서 최대 분산 방향 찾아 상관관계 감소시키고, 저차원 공간에 이들 투영 -> 주성분

- PCA 주의 사항
1. 표준화 필수 : 데이터 표준화
2. 결측값 처리 : 결측값 있으면 사용 불가
3. 해석의 어려움
4. 정보 손실 : 차원 축소 과정에서 정보 손실



### 군집화 과정 2번. 알고리즘 선택
1. 계층적 군집화
- 가장 가까운 것부터 둘씩 차례로 합쳐 가면서, 무엇이 언제 뭉쳤는지 덴드로그램으로 기록하는 방법
- 군집의 개수 사전에 설정 X, 클러스터링 종료 이후에 원하는 군집의 개수 선택
- 트리 구조를 상향식, 하향식으로 하냐에 따라 응집형과 분리형으로 구분
- 상향식 - 응집형 <br>
-> 각 샘플이 독립적인 클러스터로 시작<br>
-> 하나만 남을 때까지 가장 가까운 클러스터 병합<br>
- 하향식 - 분리형<br>
-> 하나의 클러스터에서 시작<br>
-> 유사성이 낮은 데이터를 더 작은 클러스터로 나눔

- 응집형 군집화 과정
1. 거리 행렬 계산 : 모든 데이터 포인트 간의 거리
2. 단일 클러스터 시작 
3. 가장 가까운 클러스터 병합 : 기준에 따라 병합
4. 거리 행렬 업데이트 : 새 클러스터와 나머지 간 거리 다시 계산
5. 반복 (3번, 4번)

- 군집과 군집 간의 거리 계산법
1. Single linkage
- 군집 간 가장 가까운 두 점 간 거리로 거리 정의
- 고립된 군집 찾는데 중점
- 이상치에 취약
2. Complete linkage
- 군집 간 가깡 먼 두 점 간 거리로 거리 정의
- 군집들의 내부 응집성에 중점
- 이상치에 취약
3. Average linkage
- 군집의 모든 점 쌍 사이 거리의 평균
- 모든 항목에 대한 거리 평균 구함
- 이상치에 덜 민감
- 계산량 증가
4. Centroid Method
- 군집의 중심 간의 거리 측정
- 중심 계산해야해서 계산 시간 증거
- 덴드로그램에서 거리 축 감소
5. Median
- 군집 중심점 간 산술적 중간 지점을 새로운 대표값으로 설정
- 극단값에 덜 민감
- 기하학적 구조 파악 어려움
6. Ward 연결법
- 군집 내 제곱합이 가장 적게 증가하도록 병합
- 실무에서 가장 많이 사용

- 장점
- k 미리 X 정함
- 데이터의 전체 구조 시각화 가능
- 거리만 정의되면 모든 유형 데이터에 적용 가능
- 결과 항상 동일

- 단점
- 느리고 메모리 많이 필요
- 한 번 합치면 되돌릴 수 없음
- 이상치에 민감
- 자르는 높이 사람이 정해야 함

2. k - means
- X_scaled 사용
- k-means는 유클라디안 거리를 쓰기 때문에 스케일 다르면 큰 수치를 가진 피처가 지배
- 표준화를 통한 스케일 맞추기 필요

- 장점
- 직관적이고 구현 쉬움
- 대용량 데이터에 적용 가능

- 단점
- 초기 centroid에 민감 
- 군집 수 k 결정 어려움
- 이상치에 민감 (평균 기반)
- 기하학적인 모양의 군집 파악 어려움

- 단점 보완 위한 방안
1. 엘보우 방법
- 클러스터 내 SSE 바탕으로 그래프 이용해 최적 클러스터 개수 k 추정
- k 값 바꿔가며 k-means 돌리고, 각 경우 SSE로 그래프 그림
- 급격히 기울기 변하는 지점이 최적의 k<br>
-> 장점 : 직관적이고 구현이 간단
-> 단점 : 팔꿈치 어디인지 주관 개입

3. DBSCAN
- 밀도가 높은 지역의 데이터를 하나의 군집으로 묶고, 밀도 기준 만족하지 못하는 점은 군집에 포함 X
- k 안정해도되고, 이상치 따로 빼고, 모양 X 가림
- 명시적으로 이상치 지정하여 제외
- 다른 알고리즘 대비 이상치로부터 자유
- 어떤 데이터가 여러 군집의 특정 거리 내에 있으면 가장 밀도가 높은 군집과 그룹화

- 단계
1. 샘플 분류
- 핵심 샘플 : 특정 반경 안
- 경계 샘플 : 특정 반경 안은 아니지만 다른 핵심 샘플 안에 존재
- 잡음 샘플 : 핵심 X, 경계 X
2. 클러스터 생성
3. 경계 샘플 할당 : 핵심 샘플의 클러스터에 경계 샘플 할당

- DBSCAN의 군집 수는 사람이 정하지 X
- 군집 수 대신 밀도 기준 정해

- 장점
- 클러스터 모양을 임의의 기하학적 분포도 잘 처리
- 이상치에 민감 X
- 모든 샘플 클러스터 할당 X

- 단점
- eps와 min_samples 설정에 많은 영향 받음
- 연산량 많아 속도 느림
- 밀도가 높은 곳에 집중하여, 낮은 곳은 통째로 이상치 처리

4. Gaussian Mixture Model
- 단일 분포로 표현하기 어려운 복잡한 확률 분포를 여러 개의 가우시안 분포를 합쳐서 표현
- 모델 기반 군집화
- 각 군집을 확률 분포로 간주, 전체를 여러 확률 분포의 혼합
- 데이터의 생성 메커니즘까지 모델링

- 과정
1. 주어진 전체 데이터셋의 분포 확인
2. 전체 데이터셋은 서로 다른 정규분포 형태의 확률 분포 곡선으로 구성되어 있다고 가정
3. 여러 개의 정규 분포 곡선 추출하고, 개별 데이터가 어떤 분포에 속하는지 결정<br>
-> 각각의 분포는 하나의 군집

5. 알고리즘 선택
1) 몇 개로 나눠야 할지 모름
- 계층적 군집화
2) 군집이 대체로 동글동글하고 데이터가 큼
- k-means
3) 군집 모양이 길쭉하거나 서로 겹침
- GMM
4) 이상치 골라내는 것이 목적
- DBSCAN
5) 군집 모양이 초승달, 소용돌이처럼 이상
- DBSCAN
6) 군집마다 밀도 제각각
- HDBSCAN
7) 경계에 걸친 데이터의 애매함까지 알고 싶음
- GMM

- 보통 알고리즘 2~3개 돌려보고 평가 지표로 비교해서 선택

### 군집화 과정 3번. 유효성 검증
- 비지도 학습의 경우 정답 레이블이 존재 X

1. 외부 평가
- 데이터에 정답 레이블 존재
- 군집화 결과와 실제 레이블 비교하여 유사도 펴가

2. 내부 평가
- 데이터에 정답 레이블 없는 경우
- 군집 내부 응집도와 군집 간 분리도 기반으로 품질 정량화
- 실루엣 계수, 던 지수

#### 실루엣 계수
- 클러스터 내 샘플들이 얼마나 조밀하게 모여 있는지 측정
- 군집의 품질 확인
- 분리도에서 응집도를 뺀 값을 둘 중 큰 값으로 나눔
- -1~+1 사이의 값 가짐

1) 분리도 > 응집도 : 1에 가까움
- 자신이 클러스터에 잘 속해 있고, 다른 클러스터와는 멀리 떨어져 있음
- 군집화 잘 됨
2) 분리도 = 응집도 : 0에 가까움
- 클러스터 경계에 위치
- 분리 명확도 떨어짐
3) 분리도 < 응집도 : -1에 가까움
- 잘못된 클러스터에 할당되어 군집이 섞여 있음
- 군집화 오류 발생

- 보통 0.5 이상 : 쓸만함, 0.7 이상 : 잘 됐음
- 실루엣 계수는 그래프도 함께 봄
- 각 군집의 분포가 고른지 여부도 성능 평가의 기준

#### Dunn Index
- 클러스터 간 최소 거리와 클러스터 내 최대 거리의 비율로 클러스터링 품질 평가
- 값이 클수록 좋은 군집화
- 0 이상의 값 가짐
- 실루엣 계수와 함께 종합적 판단에 활용

- 한계
- 군집 수가 많아질수록 계산 비용 증가
- 극단적인 값에 큰 영향 
- 이상치에 민감

#### 정리
- 군집화는 비지도 학습으로 정답이 없는 데이터에서 구조 찾음
- 응집도와 분리도를 둘 다 높이는 것이 목표
