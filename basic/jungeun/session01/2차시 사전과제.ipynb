2차시 사전과제 이정은

EDA & 전처리

데이터 분석 프로세스
데이터 수집 -> 탐색 -> 전처리 -> 모델링

1) 데이터 수집 : 정형, 비정형, 반정형 등 다양한 형식의 데이터 수집
2) 데이터 탐색(EDA) : 수집한 데이터의 특징 파악
3) 데이터 전처리 : 데이터를 분석에 적합한 형태로 손질
ex. 이상치 제거, 결측치 처리, 노이즈 제거 등
4) 데이터 모델링 : 데이터로부터 유용한 정보 추출을 위해 모델 구축

EDA(탐색적 데이터 분석 과정) : 주어진 데이터에 대해 알아보는 것 = '탐색과 이해'

1. 대상
- 일변량 : 변수 1개 -> 데이터 설명 + 패턴 찾기
- 다변량 : 변수 n개 -> 여러 변수 간의 관계 찾기
** 다변량에서는 개별 데이터 먼저 파악하고 여러 변수 동시에 확인하기
2. 종류
- 시각화 : 차트, 그림 등으로 데이터 확인 = 대략적인 형태 파악
- 비시각화 : summary statistics 로 데이터 확인 = 정확한 값 파악
3. 유형
- 일변량 비시각화
- 일변량 시각화
- 다변량 비시각화
- 다변량 시각화

데이터 확인

읽기 : 작업 공간에 데이터 불러오기
쓰기 : 새로운 데이터 만들기

1. 데이터 읽기
- pd.read_csv("파일경로/파일이름.csv")
- 절대 경로 : 처음 ~ 목적지까지 전체적인 경로
- 상대 경로 : 현재 작업하고 있는 디렉터리 기준으로 상하위 디렉터리와 같이 상대적으로 표현되는 경로 <br>
-> ./ : 현재 디렉터리 <br>
-> ../ : 상위 디렉터리 <br>
-> / : 최상위(루트) 디렉터리 <br>

2. 데이터 입출력
- 데이터 파일의 형식에 따라 데이터를 읽고 쓰는 함수 달라짐

(1) CSV(comma separated value)
- 쉼표로 구분하고 있는 덱스트 파일
- 데이터 크기 작고 압축이 용이해서 가장 널리 사용되는 데이터 형식

(2) EXCEL
- 데이터프레임 형태로 행과 열로 구성
- 여러 개의 시트로 구성된 데이터 읽을 때 특정 시트 설정 가능

3. 데이터셋 파악하기

(1) 데이터 프레임 보기
- head(n) : 데이터 프레임 상위 n개 데이터 보여줌

(2) 데이터 변수 확인
- 데이터 프레임의 열
- 변수 정의, 유형, 데이터 형식 확인
** 변수 유형에서 데이터의 종류
- 범주형 -> 명목형 : 순서와 무관하게 이름만 의미 부여
- 범주형 -> 순서형 : 어떤 기준에 따라 순서에 의미 부여 가능
- 수치형 -> 이산형 : 정수 단위로 셀 수 있는 데이터
- 수치형 -> 연속형 : 연속적인 값을 가지는 데이터

(3) 데이터 분포 확인
- 단변수 분석 : 하나의 데이터 분포를 확인해 분석할 수 있는 방법
- 원시 데이터 분포의 확인을 통해 전처리 아이디어 얻기 가능

전처리 
- 데이터를 분석해 적합한 형태로 만드는 과정
- 데이터의 품질을 높이는 과정
- 데이터 확인 -> 결측값 처리 -> 이상값 처리 -> 변수 가공

1. 결측값 처리
- 결측값 : 데이터 수집 과정에서 측정 X or 누락된 데이터

(1) 결측값 확인
- NaN, ?, 0
- info(), value_counts(dropna=False), isnull(), notnull(), replace() 등 사용해서 확인 가능

(2) 결측값 처리
- 삭제 : dropna() = 데이터프레임 내의 결측값 포함된 모든 레이블 제거
- 주의 : 결측값이 있는 행 or 열을 삭제하면 정보가 손실됨 
- 대체 : 평균값, 최빈값 등의 대푯값을 활용하여 대체
- fillna() = 일괄 대체, 유사 유형 대체 

2. 이상치 처리
- 이상치 : 관측된 데이터의 범위에서 많이 벗어난 값

(1) 이상치 확인
 - describe() 통해 확인
 - Boxplot으로 시각화하여 확인
 - Z-score 통해 확인 : 일반적으로 2~3 넘어가는 데이터를 이상치로 간주
 - Tukey Fences : 사분위 범위(IQR) 기반으로, Q1-1.5*IQR 미만과 Q3+1.5*IQR 초과일 때 이상치라고 판단

 (2) 이상치 제거 및 처리
 - 전체 삭제 : 이상값이 human error에 의해 발생한 경우
 - 다른 값으로 대체 : 절대적인 관측치의 숫자가 적은 경우에 삭제하지 않고 대체
 - 변수화 : 이상값이 자연 발생한 경우 이상값에 대해 파악하기 위해 해당 이상값을 변수화
 - 리샘플링 : 이상값이 자연 발생한 경우 해당 이상값을 분리해서 모델 형성
 ** 자연발생한 이상값에 별다른 특이점이 없을 땐 케이스 분리해서 분석 추천

 3. 피처 엔지니어링 (변수 가공)
 - 기존에 데이터에 정보를 추가하는 과정
 - 새로운 데이터나 변수의 추가 없이 기존의 데이터를 보다 유용하게 만드는 방법

 (1) 레이블 인코딩, 원핫인코딩
 - 기계는 텍스트를 이해하지 못하기에 텍스트 값들을 기계가 이해할 수 있도록 숫자로 변환하는 과정이 인코딩
 - 레이블 인코딩 : 범주형 변수를 숫자로 변환
 - 원핫 인코딩 : 범주형 변수를 이진 백테 1,0으로 변환

 (2) 구간화
 - 데이터를 일정한 구간으로 나눠서 분석
 - 분석 목적과 방법 등 필요한 영역에 따라 나누기

 (3) 파생변수 생성
 - 기존의 피처를 다른 피처로 재정의하여 변수 추가
 - 정해진 원칙이 없기에 분석가에 따라 다양한 파생변수 생성 가능
 
 (4) 스케일링
 - 서로 다른 변수의 값 범위를 일정한 수준으로 맞추는 작업으로 정규화 수행
 - 이때 스케일러는 이상치에 민감하기에 이상치 제거 선행 필요
 - 데이터를 같은 비율로 속상을 맞추며, 연산 속도 높이고 알고리즘 최적화에 효과적





데이터 시각화
- 데이터에 대한 이해를 돕기 위해 시각적으로 데이터 묘사하고 표현하는 과정
- 주로 matplotlib (파이썬의 그래프 그리기 위한 라이브러리)와 Seaborn 사용
- 데이터 범주형 vs 수치형 파악 -> 결측값&이상치 파악 -> 어떤 그래프 선택할지 고민

파라미터 : 프로그래밍된 함수의 입력값
- hue 파라미터 : 범주형 변수를 넣어 해당 변수 기준으로 데이터 구분하여 표시

다양한 그래프
1. boxplot : 사분위수와 이상치 보여줌
2. countplot : 범주형 변수의 빈도수 
3. histplot : 도수분포표, 수치형 변수의 구간별 빈도수
4. displot, kdeplot : 히스토그램 곡선으로 연결한 그래프
5. barplot, pointplot : 범주형 데이터 값 x에 따른 수치형 데이터 값 y의 평균
6. scatterplot, regplot : 두 연속형 변수 간의 관계 시각화
7. catplot : 수치형 데이터와 범주형 데이터의 관계
8. pieplot : 데이터의 부분과 전체 간의 비율 표현
9. heatmap : 변수간 상관계수 직관적으로 표현
** 상관계수 : 두 수치형 변수의 상관관계 정도 (-1~+1)
10. violinplot : 박스 플롯 + 커널밀도추정 함수 그래프 
11. pairplot : 여러 그래프 한번에 설정
