### 4차시 사전과제 - 이정은

### 머신러닝
- 데이터에서 패턴을 발견하는 것
- 데이터의 패턴을 바탕으로 새로운 데이터에 대해 예측하는 것

학습 방법
- 지도학습 : 기계의 예측값이 데이터의 정답과 같아지도록 학습
- 비지도학습 : 데이터의 패턴 혹은 유사성을 학습
- 강화 학습 : 시행착오를 반복하여 보상을 최대화 하는 행동 학습

### 모델링
모델
- 학습의 결과물 (예측기)
- ex y = ax + b

모델링
- 예측기를 마들어가는 전체적인 과정
- 기계가 데이터 속에서 스스로 패턴을 찾아내, 오차가 적은 최적의 공식 완성할 때까지 학습하는 과정

모델링의 과정
- 설계 -> 훈련 -> 검증
- 설계 : 데이터의 특성에 따라 알고리즘 결정
- 훈련 : 데이터를 기계에게 계속 보여주며 정답을 맞히도록 수정
- 검증 : 훈련시킨 모델이 정답을 잘 맞히는지 테스트


### 지도학습
- 정답 레이블을 주고 학습 시키는 방법
- 데이터 = labeled data = train set + test set
- train set으로 모델을 학습시키고, 학습된 모델은 test set으로 검증
- 과정 : 데이터 준비 -> 예측 -> 오차 계산 -> 가중치 수정 -> 2,3,4 반복 -> 학습 완료 = 함수 획득 -> 성능 평가

1. 회귀 : 연속적인 숫자값 예측
2. 분류 : 입력된 데이터를 주어진 항목들로 나눔

### 회귀
- 데이터에서 패턴을 찾아내어 미래 값 예측하는 것
- 종속변수가 연속적일 때 사용

1. 선형회귀
- 데이터를 가장 잘 설명하는 회귀선을 찾는 과정
- 회귀선 : y = wx + b
- 회귀선은 현재 데이터 요약 + 미래 값 예측
- w : 가중치 = 기울기 = 회귀계수
- b : 편향 

1) 단순선형회귀
- 독립변수가 1개만 존재
- 데이터를 직선 형태로 표현
- 최적의 회귀선을 찾기 위해선 실제값과 예측값 간 차이인 잔차를 줄여나가야 함
-> 방법 : 최소제곱법 (OLS)
- 최적의 회귀선을 찾는 방식 중 하나
- 잔차의 제곱합 (RSS) 가 최소가 되는 지점으로 최적의 회귀선 구하는 방식

2) 다중선형회귀
- 독립변수가 2개 이상 존재
- 독립변수가 2개면 3차원 공간이 되고, 회귀선은 평면이 됨
- 종속변수 y에 대한 독립변수들의 가중평균
-> 사용 위한 기본 가정 6개
1) 선형성
2) 독립성 : 각각의 설명변수가 서로 선형독립적
3) 오차항의 평균은 0이거나 0에 가까워야 함
4) 등분산성 : 오차항의 분산이 일정해야함
5) 오차항은 자기상관되어 있지 않아서, 공분산은 항상 0이어야함
6) 오차항이 정규분포를 따라야함
= 분석 전에는 1,2에 해당하는 데이터의 관계를 보고, 분석 후에는 3,4,5,6에 해당하는 오차항의 성질을 보면 된다

### 다중공선성
- 독립 변수들 간의 강한 상관관계로 인해 회귀 모델 불안정해짐

확인 방법
1) 상관계수 계산을 통해 파악
2) pairplot으로 시각화하여 산점도 찍어서 파악
3) VIF 지수가 높으면 다중공선성 존재, 특히 10을 넘어갈 때

대처 방법
1) 변수 제거
2) 변수 변환 : 변수들을 더하거나 빼서 새로운 변수 생성
3) 규제선형모델 활용 : 릿지, 라쏘, 엘라스틱넷
4) 주성분 분석(PCA) : 데이터의 차원 축소

규제선형모델
- 모델이 학습 데이터에 과적합되지 않도록 규제를 가하고자 등장한 모델
- 오류는 최대한 줄이되, 특정 변수에 집착하지 않도록 균형을 맞추어 과적합 방지

1) L2규제 : W의 제곱에 대해 페널티 부여 = 릿지 회귀
- 기둥에 따라 다른 힘으로 누르는 망치로, 비율이 균등하지 않을 수는 있지만 전체적으로 작아짐
- 큰 회귀계수일 수록 훨씬 강한 패널티 받음
- 규제를 통해 변수가 제거되지는 X
2) L1규제 : W의 절댓값에 대해 패널티 부여 = 라쏘 회귀
- 모든 기둥을 같은 힘으로 누르는 망치
- 작은 계수는 0이 되기도 하기 때문에 변수가 사라질 수도 있음
3) L2규제 + L1규제 : 엘라스틱넷 회귀
- 불필요한 변수는 제거하고, 남은 변수들은 적절한 크기로 유지
- 안정적인 예측과 희소성을 동시에 달성 가능



### 모델 평가방법
<모델 유의성 검정>
1. 시각화
2. 통계 지표
- F-statistic, Prob(F-statistic) : 회귀모델이 유의미한지 검정
- T-statistic, Prob(T-statistic) : 각 독립변수가 종속변수 y에 유의미한 영향을 미치는지 판단

<모델의 성능 평가>
1. 결정 계수
- R^2 = SSR/SST = 1 - SSE/SST
2. 조정된 결정 계수
- 결정 계수는 독립 변수의 개수가 증가하면 증가하기 때문에, 변수의 개수 증가에 덜 민감하도록 조정한 지표가 조정된 결정계수
3. AIC, BIC(SC)
- 정보 기준이며, 값이 낮을수록 좋음

<성능 평가 지표>
1. 평균 제곱 오차(MSE)
- 예측값과 실제값 사이의 오차 제곱의 평균
2. 평균 절대 오차(MAE)
- 모델의 예측값과 실제값 사이의 절대값 오차의 평균

<변수 유의성 평가>
- T-검정 : 독립 변수의 회귀 계수가 유의미한지 검정


### 비선형 회귀
1. 다항식 회귀 모델
- y = a + bx + cx^2 + dx^3 ...
- 독립 변수에 거듭제곱 항을 추가하여 데이터의 곡선적인 특성 모델링
- 성장률, 감소율 등이 시간에 따라 변화하는 속도가 다른 경우에 사용하면 좋음

2. 지수 회귀 모델
- 종속 변수가 지수적으로 변화하는 관계 모델링
- 값이 시간에 따라 지수적으로 증가하거나 감소하는 경우 적합

3. 로그 회귀 모델
- 종속 변수와 독립 변수 간의 관계가 로그 함수를 통해 더 잘 표현될 때 상용
- 데이터가 초기에 빠르게 증가하고 점차 증가율이 줄어드는 패턴일 때 효과적

4. 스플라인 회귀
- 데이터를 구간별로 나누고 각 구간에서 다른 선형 또는 비선형 함수를 적용하여 예측하는 방식
- 구간 경계에서의 연속성 유지가 중요
- 데이터의 패턴이 여러 구간에서 서로 다르게 나타날 때 유용
