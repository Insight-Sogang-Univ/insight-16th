## 7차시 사전과제 - 이정은

### 연관 분석
- 거래나 사건 안에 존재하는 항목 간의 일정한 연관 규칙을 발견하는 분석
- 장바구니 분석이라고 부르기도 함
- 정답값이 존재하지 않기에 비지도 학습
- Apriori, FP-Growth

### 연관 규칙
- 조건절 : 만일 ~라면
- 결과절 : 그 뒷부분의 내용
- 아이템 집합 : 조건절 또는 결과절을 구성하는 아이템들의 집합<br>
-> 아이템 집합은 상호 배반
- 연관규칙으로 상관관계 파악 O, 인과관계 파악 X<br>
-> 관찰 데이터이기 때문<br>
-> 방향성은 수학적 비율일 뿐<br>
-> 상관관계랑 인과관계는 다름!!

### 연관규칙의 조건
- 일정 이상의 지지도 : 두 품목이 함께 구매한 경우의 수가 일정 수준 이상
- 일정 이상의 신뢰도 : 품목 A를 포함하는 거래 중 품목 B를 구입하는 경우의 수가 일정 수준 이상

- 지지도: A,B 동시 거래 비율
- 신뢰도: 품목 A를 초함하는 거래 수 중 품목 A와 B가 동시에 포함되는 거래의 비율

<br>
-> 신뢰도가 높다고 무조건 좋은 규칙 아님! 그렇기에 A,B의 판매량에 진짜 유의미한 상관관계가 있는지 확인하는 향상도 측정 필요

- 향상도 : 품목 A가 주어지지 않았을 때의 품목 B의 확률 대비, 품목 A가 주어졌을 떄의 품목 B의 확률의 증가 비율
- 사건 A,B 독립 : A를 샀다는 사실이 B를 살 확률에 영향을 주지 않기에 향상도는 1
- 독립 X : A를 샀다는 사실이 B를 살 확률에 영향을 주거나, B를 샀다는 사실이 A를 살 확률에 영향을 줄 것, 즉 향상도는 1보다 크거나, 1보다 작음<br>
-> 향상도>1 : 양의 상관관계<br>
-> 향상도=1 : 독립<br>
-> 향상도<1 : 음의 상관관계<br>

- 규칙의 효용성은 지지도, 신뢰도, 향상도를 모두 반영하여 세 지표 모두 클 경우에만 효과적이라는 결론 내릴 수 있음


### 연관규칙 예시
- 월마트의 딸기맛 팝타르트 사례
- 월마트는 전체 기간을 통으로 분석하는 것이 아닌 허리케인 경보 기간 중 발생한 전체 거래와 가장 많이 팔린 품목을 분석함. 이때, 딸기맛 팝타르트가 정전 상황에서 가장 적합한 음식이기에 판매량이 크게 증가했다는 것을 알고, 허리케인 경보에 맞춰 약 35만 상자를 사전 배송함. 
- 이때 왜 딸기맛인가에 대한 인과는 설명하지 못했지만, 허리케인과 딸기맛 팝타르트의 상관관계 분석에 성공하여 높은 판매량 달성


### 연관분석 알고리즘 1번. Apriori 알고리즘
- 배경 : 연관규칙을 찾기 위해 모든 가능 조합을 시도한 것은 매우 비효율적임
- 개념 : 상위 조합에서부터 차례로 스캔하며 특정 조합이 자주 나타나지 않으면 연관된 후속 조합까지 모두 후보에서 배제하는 방식의 알고리즘
- 장점 : 하나의 조합의 검사해도 파생 후보들까지 배제할 수 있다는 점에서 시간과 연산량 줄일 수 있음
- 프로세스
1. 모든 개별 상품에 대해 지지도 계산하고, 최소 지지도 넘는 항목만 통과
2. 선별된 상품들로 2개 항목의 조합 만듦<br>
-> 이때 각 조합의 지지도 계산하고, 최소 지지도 넘지 않으면 제거
3. 남은 조합을 바탕으로 3개 이상의 항목 조합 만듦<br>
-> 이때 각 조합의 지지도 계산하고, 최소 지지도 넘지 않으면 제거<br>
-> 과정 반복
4. 남은 빈발 항목 집합들을 바탕으로 연관규칙 생성<br>
-> 각 규칙의 지지도, 신뢰도, 향상도 계산

- 장점
- 수많은 상품의 연관 구매 패턴 발견 
- 원리 간단, 이해분석 용이

- 단점
- 데이터 커질수록 속도 느려짐

### 연관분석 알고리즘 2번. FP Growth 알고리즘
- 배경 : Apriori의 데이터가 커질수록 느려지는 연산 속도 문제 해결 위한 알고리즘
- 특징<br>
-> 빈발 항목 집합을 효율적으로 찾아냄<br>
-> FP-Tree라는 구조 이용하여 구현<br>
-> 모든 거래 확인하여 아이템마다 지지도 계산하고, 최소 지지도 이상의 아이템만 선택

- 프로세스
1. 데이터셋 확인
2. 빈도순 재정렬 : 남은 항목들을 각 거래별로 빈도순 재정렬, 이떄 자주 구매된 항목이 먼저 오도록 함 = f-list
3. FP-트리 생성 : 재정렬된 데이터 바탕으로 FP-트리 생성, 이 트리는 각 거래를 노드로 표현하며, 동일한 항목은 기존 노드에 연결해 빈도 증가시키고, 새로운 항목은 새로운 노드로 추가
4. 빈발 항목 집합 추출 : 부분 트리 만들어 추출
5. 연관 규칙 도출 : 각 규칙에 대해 지지도, 신뢰도, 향상도 등의 지표 사용해 가장 유용한 규칙 선택

- 장점
- 데이터 2번만 스캔하는 트리 구조이기에 빠름
- 후보 itemset 없이, tree만 구성하면 끝

- 단점
- 대용량 데이터셋에서 메모리 효율적으로 사용 안함
- Apriori에 비해 설계 어렵고, 지지도 계산은 무조건 FP-Tree가 만들어져야 가능
