## 2차시 사전과제 - 이정은

### 통계 
: 우리가 다양한 비즈니스와 경제 상황을 이해하는데 도움을 주는 숫자로 표현

- 기술통계 : 수치와 시각화를 통해 샘플을 요약하여 표현하는 방법론 
- 추론통계 : 추정과 검정을 통해 샘플로부터 모집단을 추론하는 방법론

### 자료의 척도
자료 : 설명과 해석을 위해 수집, 분석, 요약되는 사실과 수치 <br>

질적자료 : 수치로 표현되더라도 연산의 의미 X
- 명목척도 : 레이블이나 이름으로 요소의 속성 식별
- 서열척도 : 분류 + 순서 또는 순위를 나타내는 특성 <br>

양적자료 : 수치로 표현
- 구간(등간)척도 : 순서 + 간격 일정하여 연상 가능 (곱셈, 나눗셈은 불가능)
- 비율척도 : 동간성 + 절대영점 <br>
** 절대영점 : 그 변수에 대해 아무것도 존재하지 않는 상태

-> 자료의 종류 vs 척도
- 종류 : 데이터의 대략적인 성격을 파악하는 큰 그림
- 척도 : 데이터를 다루기 위한 더욱 구체적인 규칙과 기준

### 기술통계 
- 데이터를 있는 그대로 설명하고 이해하는 데 중점

1. 중심화 경향 : 데이터의 중심적인 경향 나타냄
- 평균 : 데이터의 총합 / 데이터의 개수 = 총합을 사용하기에 이상치에 민감
-> 절사 평균 : 이상치에 민감함을 보완하기 위해 등장했으며, 데이터의 양끝을 일정 비율 잘라낸 후 평균을 계산하여 이상치에 더 강함
- 중앙값 : 데이터의 크기를 순서대로 정렬했을 때, 한가운데에 위치하는 값
- 최빈값 : 데이터에서 가장 자구 나타나는 값, 주로 이산형 데이터에 사용

2. 자료의 산포 : 데이터가 퍼져있는 정도 설명
- 범위 : 최대값 - 최소값 / 데이터의 전체 변동 범위 
- 분산 : 데이터의 값들이 평균으로부터 얼마나 멀리 퍼져있는지 나타냄
- 표준편차 : 분산에 제곱근을 취한 값
-> 표준편차는 제곱근으로 데이터와 단위를 일치시켜 주기 위해 사용
- 사분위수 : 데이터의 표본을 4개의 동일한 크기로 나눈 값
- IQR(사분위간 범위) : Q3 - Q1 / 데이터의 중앙 50%가 퍼져있는 범위
-> 이상치 : Q1 - 1.5 x IQR, Q3 + 1.5 * IQR
- 변동계수 : (표준편차 / 평균) x 100 / 데이터의 상대적인 변동성
- 첨도 : 확률분포의 꼬리가 두꺼운 정도
- 왜도 : 비대칭도, 실수 값 확률 변수의 확률 분포 비대칭성을 나타내는 지표

### 확률과 분포
- 추론통계 : 표본에서 얻은 정보를 바탕으로, 모집단의 특성을 확률적으로 추정하거나 가설을 검정하는 통계 분야

##### 확률분포
- 확률변수 : 어떤 특정 사건이 발생할 때 나타날 수 있는 모든 경우의 수
- 확률분포 : 확률변수가 특정한 값을 가질 확률
-> 이산확률분포 : 확률변수가 이상적인 값을 가지는 경우
-> 연속확률분포 : 확률변수가 연속적인 값을 가지는 경우

##### 대표적인 분포 모형
1. 정규분포 : 종 모양, 평균 기준 대칭, 표준편차 클수록 평평
- 중심극한정리 : 동일한 확률 분포를 가진 독립 확률 변수 n개의 평균의 분포는 n이 적당히 크다면 정규분포에 가까워짐
- 큰 수의 법칙 : 표본의 크기가 커질수록 표본 평균이 모평균에 근사할 것

2. Z 분포 (표준정규분포)
- 평균이 0, 표준편차가 1이 되도록 맞춘 표준적인 정규 분포

-> 표본 분포 : 정규 분포를 따르는 모집단에서 추출한 표본 통계량들이 따르는 분포 <br>
ㄴ 이를 위해선 추출한 표본이 정규분포를 따르는 모집단으로부터 왔다는 가정이 중요 <br>
= 정규성 검정

3. T 분포
- 종 모양, 정규분포보다 꼬리 두꺼움, 평균 기준 대칭, 자유도 커짐에 따라 표준정규분포에 수렴
- 표본으로 모평균을 추정할 때 사용되는 분포
- 표본 크기가 작을 때 정규분포보다 더 정확한 결과를 제공
- 표본의 크기가 30미만이면 t분포, 30이상이면 표본정규분포 사용

4. 카이제곱분포
- 0에서 시작하여 오른쪽으로 긴 꼬리를 가진 비대칭 분포
- 자유도가 증가함에 따라 중심이 오른쪽으로 이동하고 모양이 더 대칭적으로 변화하며, 정규분포에 가까워짐
- 독립적인 표준 정규 분포를 따르는 변수들의 제곱 합

5. F 분포
- 두 독립적인 카이제곱분포의 비율
- 두 집단 간의 분산 차이를 비교하는 데 사용됨

#### 추론통계
- 수집한 데이터를 기반으로 미래의 정보를 추론하고 결론을 도출하는 방법으로, 통계적 가설검정과 신뢰구간 등이 여기에 속함

##### 추정
- 모집단을 대표할 수 있는 일부를 표본으로 추출하여 분석한 통계량으로 모집단의 모수 예측
- 오차 발생 가능성을 생각하여 어느 정도 오차를 포함하는 신뢰구간 추정 방법이 일반적으로 널리 사용됨
1. 점추정 : 어떤 모수를 표본에 기초하여 계산된 하나의 값으로 추정
2. 구간추정 : 모수의 참 값이 포함되리라고 기대하는 추정치를 일정한 범위와 신뢸도를 가지고 나타내는 방법

##### 검정
- 가설 검정 : 모집단의 주장과 관련해 표본의 정보를 사용해서 가설의 합당성 여부 판정하는 과정

1. 가설 설정 
- 귀무가설 H0과 대립가설 H1 설정
- 귀무가설은 주로 기각하고자 하는 명제, 대립가설은 내가 주장하고자 하는 명제

2. 유의 수준 설정
- 유의수준이란 연구자가 결과를 판단할 때 허용할 수 있는 오류의 최대치 (1종 오류 확률)
- 1종 오류 : 귀무가설이 실제로 참인데 기각할 확률
- 2종 오휴 : 귀무가설이 실제로 거짓인데 기각하지 못할 확률

3. 검정통계량 산출
- 검정통계량은 가설 검정의 결과를 판단하는 데 사용되는 값
- Z값 : 특정 데이터가 평균으로부터 얼마나 떨어져 있는지를 표준편차 단위로 나타냄
- T값 : 정규분포와 유사하지만, 작은 표본에서 발생할 수 있는 불확실성 잘 반영
- F값 : 2개의 독립적인 카이제곱 분포된 변수들의 비율로 정의, 1에 가까우면 두 표본의 분산이 서로 비슷하다는 것 의미

4. 가설 기각, 채택 판단
- 유의 확률이 유의 수준보다 작으면 귀무가설 기각, 크면 기각 x


#### 데이터의 자료형에 따른 검정 방법
1. 피어슨 상관 분석 
- 수치형 - 수치형
- 두 변수 간 피어슨 상관계수가 얼마나 유의미한지 검정

2. F-test(ANOVA), T-test
- 범주형 - 수치형
- 범주가 2개일 때 T-test 사용, 이때 t값은 두 평균 간 차이를 표준 오차로 나눈 값
- 범주가 3개 이상일 때 F-teest 사용, 이때 F값은 집단 간 분산 / 집단 내 분산

3. Chi2-test
- 범주형 - 범주형 
- 같은 위치에 있는 기대 빈도와 관측 빈도의 차이 비교
-> 이때 서로 차이가 크다면, 둘이 독립적이지 않을 가능성 높아짐.
