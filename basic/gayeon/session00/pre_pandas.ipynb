## 2. DataFrame 구성

- Pandas에서 사용되는 대표적인 데이터 오브젝트는 **Series**와 **DataFrame**이 있음

   - **Series:** 1차원 배열의 형태, Index라는 한 가지 기준에 의해 데이터가 저장됨
   - **DataFrame:** 2차원 배열의 형태, Index와 Column이라는 두 가지 기준에 의해 데이터가 저장됨

### Series

- 값(value)과 인덱스(index)로 구성된 1차원 자료 구조
- 인덱스 레이블을 따로 지정하지 않으면 0부터 시작되는 디폴트 정수형 인덱스가 자동으로 부여됨

**Series 만들기**
1. 원하는 값들이 포함된 리스트 생성
```
series = ['LG', '애플', '애플', '삼성', 'ASUS', '삼성']
```

2. pd.Series() 함수를 사용하여 시리즈 객체 생성

```
sr = pd.Series(series)
print(sr)
```

3. 인덱스 직접 지정하는 경우, pd.Series() 함수의 인자에 인덱스를 추가
```
index = [1,2,3,4,5,6]
sr = pd.Series(series, index = index)
print(sr)
```

### DataFrame

- 행(index)과 열(column)로 구성된 2차원 배열 구조
- 여러 개의 Series가 모여 하나의 DataFrame을 구성
- 컬럼 단위로는 값들의 데이터 타입이 동일해야 하지만, 컬럼 간의 데이터타입은 달라도 무방함

**DataFrame 만들기**
1. 데이터와 각 카테고리별 네이밍을 컬럼으로 하는 딕셔너리 생성
```
data = {
    '노트북 브랜드' : ['LG', '애플', '애플', '삼성', 'ASUS', '삼성'],
    '노트북 화면 크기(인치)' : [14.0, 14.2, 15.3, 13.3, 14.0, 15.6],
    '노트북 무게(kg)' : [0.99, 1.63, 1.51, 0.87, 1.28, 1.80],
    '사용자 이름' : ['김진하', '방서영', '김민아', '조대흠', '류시현', '권유철']
}
```

2. pd.DataFrame() 함수를 사용하여 데이터프레임 구성

```
df = pd.DataFrame(data)
```

3. 인덱스, 컬럼 직접 지정
```
index = [1,2,3,4,5,6]
df = pd.DataFrame(data, index=index)
```

## 3. 데이터 확인 및 추출

- 데이터 분석 전, 주어진 DataFrame에 대한 확인이 필요함 (ex. 데이터 크기, 타입, 값의 분포, 경향 등)
- 분석할 데이터는 양이 매우 방대하므로 분석에 필요한 부분만 골라내어 가공하는 능력, 즉 데이터 추출이 매우 중요함

### 인덱싱(Indexing) & 슬라이싱(Slicing)
1. 인덱싱: 단일 항목을 지목할 때 사용
```
df['노트북 브랜드']
```

2. 슬라이싱: 범위를 지정해 여러 항목을 추출할 때 사용

```
df[start:end] 
```

### loc & iloc
1. loc: 행과 열의 이름(label) 기준
```
df.loc[행 이름]
df.loc[행 이름, 열 이름]
df.loc[:, 열 이름]
```
*끝 인덱스를 포함하여 반환*


2. iloc: 정수형 인덱스 번호 기준

```
df.iloc[행 순서, 열 순서]
```
*끝 인덱스를 포함하지 않고 반환, 즉 (시작 인덱스 ~ 끝 인덱스 - 1)*

### 불리언 인덱싱(Boolean Indexing)
- 인덱스나 컬럼명이 아니라, 특정 조건에 만족하는 데이터를 추출하고 싶을 때 사용

1. True 또는 False의 배열을 통해 데이터를 필터링
```
df[조건문]
```

2. 조건문 여러 개를 논리연산자(and, or)로 묶어 사용

```
df[(조건문1) 논리연산자 (조건문2)]
```
*loc에서도 불리언 인덱싱 사용 가능, iloc은 정수형 순서 인덱스만 받을 수 있으므로 bool 타입인 불리언 인덱싱 사용 불가능*

### 인덱스 설정
1. 기존의 열 중 하나를 인덱스로 설정
```
df.set_index(열 이름)
```
2. 인덱스를 리셋하고 기존 인덱스를 열에 추가
```
df.reset_index()
```
3. 인덱스의 재배열
```
df.reindex(인덱스 배열)
```

## 5. 집계함수

- 여러 행으로부터 하나의 결괏값을 반환하는 함수
- 각 컬럼의 평균, 표준편차, 합계 등.. 연산을 통해 대표값을 구할 때 주로 사용

### 그룹화(Grouping)

- groupby 메소드: 특정 기준으로 그룹을 묶고, 연산 처리
- groupby 자체만으로는 아무 일도 일어나지 않고 보통 집계함수와 함께 사용

  -  column1로 그룹화를 하고, 그룹 별로 column2에 대해 집계함수를 적용해 결과를 시리즈 형태로 반환
  ```
  df.groupby('column명1')[’column명2'].집계함수()
  ```
  - 그룹 별로 column2에 대해 집계함수를 적용해 결과를 데이터프레임 형태로 반환
  ```
  df.groupby('column명1')[[’column명2']].집계함수()
  ```
