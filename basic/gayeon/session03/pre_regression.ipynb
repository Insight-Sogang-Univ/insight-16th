# 회귀 요약본

## 1. 머신러닝과 모델링

### 머신러닝의 종류

|유형|정답 여부|방식|예시|
|---|---|---|---|
|지도 학습|정답(label)이 있는 데이터 사용|기계의 예측값이 데이터의 정답과 같아지도록 학습시키는 것|회귀, 분류|
|비지도 학습|정답(label)이 없는 데이터 사용|데이터의 패턴 혹은 유사성을 학습시키는 것|군집화|
|강화 학습|보상|시행착오를 반복하여 보상을 최대화하는 행동 학습|알파고, 자율 주행|

### 모델링

**1. 모델이란?**
- 데이터를 학습하며 찾아낸 패턴이 저장된 상태이며, 새로운 데이터를 넣었을 때 답을 내놓는 예측기 역할

**2. 모델링이란?**
- 예측기(공식)를 만들어가는 전체적인 과정
- 아무것도 모르는 기계가 주어진 데이터 속에서 스스로 패턴을 찾아내, 오차가 작은 최적의 공식을 완성할 때까지 학습하는 과정

**3. 모델링 과정**
- 설계: 데이터의 특성에 맞추어 어떤 알고리즘으로 학습할지 결정하는 단계
- 훈련: 데이터를 기계에게 계속 보여주며 정답을 맞히도록 수정하는 단계
- 검증: 훈련시킨 모델이 새로운 데이터에서도 정답을 잘 맞히는지 테스트하는 단계

### 지도 학습

- 정답 레이블을 함께 주고 학습을 시키는 방법
- 예측값(prediction)과 정답(label)이 최대한 같아지는 방향으로 학습

**1. Labeled Data**
- Labeled data: 독립변수(Feature)와 종속변수(Target, Label)가 모두 존재하는 데이터
    - Feature(독립변수): 모델의 입력값
    - Target(Label, 종속변수): 모델이 예측해야 하는 정답
    > Feature를 이용해 Target을 예측하는 규칙(함수)을 학습

**2. Train Set / Test Set**
- Train Set: 모델을 학습(훈련)시키는 데이터
- Test Set: 학습이 끝난 모델의 성능을 평가하는 데이터

*새로운 데이터에서도 잘 예측하는지 성능을 확인하기 위해 분리*

**3. 지도학습 과정**
1) 데이터 준비
- Feature(입력)와 Label(정답)이 있는 데이터를 준비

2) 예측
- 모델이 Feature를 입력받아 Target 예측

3) 오차 계산
- 예측값과 실제 정답의 차이 = 오차(잔차)
- 오차를 수치화한 값 = 손실(Loss)

4) 가중치 수정
- 오차가 감소하는 방향으로 내부 파라미터(가중치) 수정

5) 반복
- 예측 → 오차 계산 → 가중치 수정을 반복
- 반복할수록 Loss 감소

6) 학습 완료
- 데이터의 관계를 하나의 함수로 학습
- 새로운 데이터에도 규칙을 적용하여 예측 가능(일반화)

7) 성능 평가
- Test Set으로 모델 성능 평가
- 학습 데이터가 아닌 새로운 데이터에서 얼마나 잘 예측하는지 확인

### 회귀와 분류


- (회귀) 연속적인 숫자값을 예측
    - 주어진 데이터(X)를 기반으로 정답(Y)를 잘 맞히는 함수를 찾음


- (분류) 입력된 데이터를 주어진 항목들로 나눔
    - 기존 데이터가 어떤 레이블에 속하는지 패턴을 알고리즘으로 인지한 뒤에 새롭게 관측된 데이터에 대한 레이블을 판별



## 2. 회귀

**회귀**
- 데이터에서 패턴을 찾아 미래 값을 예측하는 기법
- 종속변수가 연속형일 때 사용

**회귀의 종류**
- 선형회귀
- 비선형회귀
- 규제회귀
- 다항회귀

### 선형회귀

**1. 선형회귀란?**

**정의**
- 데이터를 가장 잘 설명하는 회귀선을 찾는 과정
- 회귀선은 현재 데이터를 요약하면서, 관측되지 않은 미래 값을 예측하는 도구

**종류**
- 단순선형회귀: 독립변수 1개
- 다중선형회귀: 독립변수 2개 이상

**2. 가중치와 편향**

선형회귀식
$$y = wx+b$$
- 가중치(Weight, w)
  - 독립변수 x가 종속변수 y에 주는 영향력의 크기
  - 그래프에서는 기울기
  - 회귀계수(Regression Coefficient) 라고도 부름

- 편향(Bias, b)
  - x=0일 때의 기본값
  - 그래프에서는 y절편
  - 모델이 데이터에 더 유연하게 최적화되도록 위아래로 이동시키는 역할

**3. 선형회귀의 종류**

- 단순선형회귀(Simple Linear Regression)
  - 독립변수 1개
  - 데이터를 직선 형태로 표현
- 다중선형회귀(Multiple Linear Regression)
  - 독립변수 2개 이상
  - 데이터를 평면(또는 고차원 초평면)으로 표현

**4. 단순선형회귀**

$$y=wx+b$$
- 하나의 독립변수만 존재
- 데이터를 가장 잘 설명하는 y=wx+b를 찾는 과정
- 회귀계수(w)와 편향(b)을 잔차가 최소가 되도록 결정

**5. 잔차**
$$\text{잔차}=\text{실제값}-\text{예측값}$$
- 회귀선은 잔차를 줄여가는 과정
- 잔차가 작을수록 데이터를 잘 설명하는 회귀선

### 최적의 w와 b를 찾는 방법

**1. 최소제곱법**

OLS (Ordinary Least Squares)
- 최적의 회귀선을 찾는 대표적인 방법
- 회귀선과 실제 데이터의 차이를 최소화



잔차
$$\text{Residual}=\hat y-y$$

목적
$$RSS=\sum(\hat y-y)^2$$
- 잔차제곱합(RSS)이 최소가 되는 회귀선을 찾음

### 다중선형회귀



정의) 
- 독립변수가 2개 이상인 선형회귀

일반식)
$$y=W_1x_1+W_2x_2+\cdots+W_nx_n+b$$
$$y=\beta_0+\beta_1X_1+\beta_2X_2+\cdots+\beta_kX_k+\epsilon$$

특징)
- 독립변수 각각에 가중치를 곱해 종속변수를 설명
- 독립변수가 2개이면 회귀선은 평면
- 변수 수가 증가하면 차원이 확장

학습 방법)
- 최소제곱법(OLS)
- 경사하강법(Gradient Descent)

### 다중선형회귀의 기본 가정

**1. 선형성(Linearity)**
- 종속변수와 설명변수의 관계가 선형적이어야 함

**2. 독립성(Independence)**
- 설명변수들이 서로 선형독립(linearly independent) 이어야 한다.
- 선형독립이 아니면 다중공선성(Multicollinearity) 발생

**3. 오차항 평균 = 0**
$$E(\epsilon|X)=0$$
- 오차항의 평균은 0(또는 0에 가까워야 함)
- 오차항은 실제값과 예측값의 차이
- White Noise(백색잡음)의 성질 중 하나

**4. 등분산성(Homoscedasticity)**
$$Var(\epsilon|X)=\sigma^2$$
- 오차항의 분산이 일정해야 한다.
- 일정하지 않으면 이분산성(Heteroscedasticity)

**5. 자기상관 없음(Non-autocorrelation)**
$$Cov(\epsilon_i,\epsilon_j|X)=0 $$
- 오차항끼리 서로 상관이 없어야 한다.
- 자기상관이 존재하면 이전 오차의 영향을 받아 반복적인 오류가 발생
- 시계열 자료에서 자주 나타남

**6. 정규성(Normality) (선택)**
$$\epsilon_i\sim N(0,\sigma^2) $$
- 오차항이 정규분포를 따른다는 가정
- 다른 가정들에 비해 중요도가 낮으며, 위배되어도 회귀분석 결과에 큰 영향을 주지 않는 경우가 많음


### 다중공선성

1. 다중공선성이란?

개념)
- 독립 변수들 간의 강한 상관관계로 인해 회귀 모델이 불안정해지는 문제
- 다중공선성이 심하면 각 변수의 회귀계수 추정이 부정확, 해석력이 감소

문제점)
- 다중공선성이 높으면 어떤 독립 변수가 종속 변수에 얼마나 영향을 미치는지 정확하게 구분하기 어려움

2. 다중공선성 확인 방법

(1) 상관계수 (Correlation coefficient)



- 변수 간의 통계적 관계(상관관계)의 정도를 수치로 나타낸 것
- -1 ≤ r ≤ 1

| 값 | 의미|
| ----- | ----------- |
| r > 0 | 양의 상관관계 |
| r < 0 | 음의 상관관계 |
| r = 0 | 직접적인 연관성 없음 |


(2) VIF (분산 팽창 인수)

- 회귀모델의 결정계수를 이용하여 계산

| VIF | 해석              |
| --- | --------------- |
| 1   | 다른 변수와 상관관계 없음  |
| <5  | 일반적으로 문제 없음     |
| >5  | 다중공선성 징후, 주의 필요 |
| >10 | 다중공선성이 심각함      |


### 다중공선성 대처 방법
1) 변수 제거
- 상관관계가 높은 변수를 제거

2) 변수 변환
- 변수들을 더하거나 빼서 새로운 변수 생성

3) 규제선형모델 활용
- Ridge
- Lasso
- Elastic Net

4) PCA(주성분 분석)
- 데이터의 차원을 축소하는 통계적 기법
- 데이터의 방향성을 찾아 변환하고 중요한 정보만 남겨 차원을 축소

### 규제선형모델
- 모델이 학습 데이터에 과적합(overfitting)되지 않도록 규제를 가하는 모델
- 선형회귀에서는 특성에 곱해지는 계수(기울기)의 크기를 조절

**1. Ridge 회귀 (L2 규제)**

- 개념
   - 회귀계수의 제곱에 대해 페널티를 부여

- 특징
   - 큰 회귀계수일수록 더 강한 페널티
   - 모든 계수의 크기를 줄임
   - 변수는 제거되지 않음 (회귀계수가 0이 되지 않음)

**2. Lasso 회귀 (L1 규제)**
- 개념
   - 회귀계수의 절댓값에 대해 페널티를 부여

- 특징
   - 회귀계수의 크기와 상관없이 같은 페널티
   - 작은 계수는 0이 될 수 있음
   - 변수 자체가 제거될 수 있음

**3.Elastic Net (L1 + L2)**
- 개념
   - L1과 L2를 동시에 사용하는 규제모델

- 특징
   - L1의 변수 선택 기능, L2의 규제 기능을 동시에 활용
   - L1은 상관관계가 높은 변수 중 일부만 선택하는 경향이 있는데, L2를 함께 적용하면 상관관계가 높은 변수들도 함께 고려하면서 규제가 가능

## 3. 모델 평가방법
- 회귀선을 데이터와 함께 시각화
- 통계지표 활용

### 시각화


- 목적
   - 회귀선이 데이터 전반을 잘 요약하는지 확인
   - 데이터와 회귀선의 적합 정도를 직관적으로 확인

- 한계 
   - 어떤 회귀선이 더 좋은 모델인지 객관적으로 비교하기는 어려움
   - 따라서 통계지표를 함께 활용

### 통계지표



**1. 모델 유의성 검정**

(1) F-statistic, Prob(F-statistic)
   - 모델 전체의 유의성 평가
   - F-statistic이 클수록 통계적으로 유의
   - Prob(F-statistic) < 0.05

(2) T-statistic, Prob(T-statistic)
   - 개별 독립변수의 유의성 평가
   - P> |t| < 0.05

**2. 모델 성능 평가**

(1) 결정계수 (R-squared)
  - 모델이 데이터를 얼마나 잘 설명하는지 나타내는 지표
  - 값이 클수록 설명력이 높음

(2) 조정된 결정계수 (Adjusted R-squared)
  - 변수 개수를 고려하여 보정한 결정계수
  - 변수 추가에 따른 과대평가를 보완

(3) AIC, BIC
  - 모델을 비교하기 위한 지표
  - 값이 작을수록 좋은 모델

### 성능 평가 지표

(1) MSE (Mean Squared Error)

- 예측값과 실제값 사이 오차 제곱의 평균

$$ MSE=\frac1n\sum (y_i-\hat y_i)^2 $$

(2) MAE (Mean Absolute Error)

- 예측값과 실제값 사이 절대값 오차의 평균
$$ MAE=\frac1n\sum |y_i-\hat y_i| $$

### 변수 유의성 평가 (T-검정)
**1. 목적**
- 독립변수의 회귀계수가 유의미한지 검정

**2. 가설**
- 귀무가설(H0)
  - 회귀계수 = 0
  - 독립변수가 종속변수에 영향을 주지 않음

- 대립가설(H1)
  - 회귀계수 ≠ 0
  - 독립변수가 종속변수에 영향을 줌

**3. t값 계산**
$$t=\frac{\hat\beta-0}{SE(\hat\beta)}$$

**4. p-value 판단**

- p-value < 0.05
  - H0 기각
  - 해당 변수는 유의미
- p-value ≥ 0.05
  - H₀ 기각 불가
  - 해당 변수는 유의미하지 않음
