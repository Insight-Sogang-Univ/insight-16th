# 통계 요약본

## 1. Intro

- 전 인구(모집단)을 대상으로 질문하는 것이 가장 정확하나, 시간/비용적 측면에서 비현실적
- 일반적으로 모집단에 대해 일부 샘플(표본집단)만 추출하여 분석
   - 기술통계: 수치와 시각화를 통해 샘플을 요약하여 표현하는 방법론
   - 추론통계: 추정과 검정을 통해 샘플로부터 모집단을 추론하는 방법론

### 자료의 척도

1. 질적 자료
- 명목척도: 레이블이나 이름으로 요소의 속성을 식별하려고 하는 척도
- 서열척도: 명목형의 분류 속성과 더불어 항목 간의 순서 또는 순위를 나타내는 특성을 지니는 척도

2. 양적 자료
- 구간(등간)척도: 값들 사이에 순서가 있고, 그 간격이 일정하여 덧셈이나 평균과 같은 연산이 가능한 척도
- 비율척도: 절대영점과 임의의 측정단위로 구성하고, 동간성이 유지되는 척도

## 2. 기술통계

- 데이터를 있는 그대로 설명하고 이해하는 데 중점을 둔 통계 방법

### 중심화 경향

- **평균(mean)**
   - (개념) 데이터 값의 총합을 데이터의 개수로 나눈 값
   - (특징) 데이터의 중심 경향성을 나타내는 가장 대표적인 지표이나, 총합을 사용하기에 이상치(outlier)에 민감   
   -> 이를 보완하기 위해 **절사평균**이 등장

- **중앙값(median)**
   - (개념) 데이터를 크기 순서대로 일렬로 정렬했을 때 한가운데에 위치하는 값
   - (특징) 평균에 비해 이상치에 덜 민감하며, 데이터의 중심 경향성을 나타내는 대안적인 지표

- **최빈값(mode)**
   - (개념) 데이터에서 가장 자주 나타나는 값
   - (특징) 주로 이산형 데이터(Discrete Data), 데이터의 분포를 이해하는데 도움

### 자료의 산포

- **범위(range)**
   - (개념) 데이터의 최대값과 최소값의 차이를 나타내는 지표로, 데이터의 전체 변동 범위를 나타냄
   - (특징) 양쪽 극단값을 이용하기에 이상치(outlier)에 매우 민감

- **분산(variance)**
   - (개념) 데이터 값들이 평균으로부터 얼마나 멀리 퍼져있는지를 나타냄

- **표준편차(standard deviation)**
   - (개념) 분산에 제곱근을 취한 값

- **사분위수(quartile)**
   - (개념) 데이터의 표본을 네 개의 동일한 크기로 나눈 값
      - Q1 = 하위 25퍼센트에 위치한 값, 하위 사분위수
      - Q2 = 하위 50퍼센트에 위치한 값, 중앙값 
      - Q3 = 하위 75퍼센트(상위 25%)에 위치한 값, 상위 사분위수

- **사분위간 범위(IQR)**
   - (개념) Q3에서 Q1 값을 제외한 범위, 즉 데이터의 중앙 50%가 퍼져 있는 범위
   - (특징) 이상치를 탐지할 때 유용
      - 아래쪽 이상치: Q1−1.5×IQR 보다 작은 값
      - 위쪽 이상치: Q3+1.5×IQR 보다 큰 값

- **변동계수(coefficient of variation)**
   - (개념) 데이터의 절대적인 변동성이 아닌 상대적인 변동성을 나타내는 통계적 지표로, '변동계수 = (표준편차/평균) * 100

***첨도:*** *확률분포의 꼬리가 두꺼운 정도를 나타내는 척도*   
***왜도:*** *비대칭도, 실수 값 확률 변수의 확률 분포 비대칭성을 나타내는 지표*

## 3. 확률과 분포

- 추론 통계(통계적 추론)란 표본에서 얻은 정보를 바탕으로, 모집단의 특성(모수)을 확률적으로 추정하거나 가설을 검정하는 통계 분야

### 확률분포
**1. 이산확률분포:** 확률 변수가 이산적인(discrete) 값을 가지는 경우
- 유한하거나 셀 수 있다는 뜻이며 특정 값을 선택 가능

**2. 연속확률분포:** 확률 변수가 연속적인(continuous) 값을 가지는 경우
- 연속적인 값은 무한한 범위를 가지며, 특정 값을 선택하는 것이 불가능

### 대표적인 분포 모형



1. 정규분포 (Normal Distribution)

|형태|- 종(bell) 모양 <br>- 평균을 기준으로 대칭 <br>- 표준편차가 클수록 곡선이 평평해짐 (데이터가 넓게 퍼짐)|
|---|---|
|매개변수|평균(μ), 표준편차(σ)|
|특징|- 가장 잘 알려진 (연속)확률분포 <br>- 많은 자연 현상과 사회 현상이 정규분포를 따르는 것으로 알려져 있음|

2. Z분포 (표준정규분포)
- 평균이 0이고 표준편차가 1이 되도록 맞춘 표준적인 정규분포

3. T분포 (T-Distribution)

|형태| - 종(bell) 모양, 정규분포보다 꼬리가 두꺼움 <br>- 평균을 기준으로 대칭 <br>- 자유도가 커짐에 따라 표준정규분포 N(0,1)에 수렴|
|---|---|
|매개변수|자유도(df)|
|특징| - 표본으로 모평균(혹은 모평균 간의 차이)을 추정할 때 사용되는 분포 <br>- 표본 크기가 작을 때 정규분포보다 더 정확한 결과를 제공함|

- 자유도: 주어진 조건을 만족하면서 독립적으로 변할 수 있는 정보의 개수
- 과적합(Overfitting): 모델이 학습 데이터에 지나치게 맞춰져 새로운 데이터에 대한 예측 성능이 떨어지는 현상



4. 카이제곱분포 (Chi-Square Distribution)

|형태| - 0에서 시작하여 오른쪽으로 긴 꼬리를 가진 비대칭 분포<br>- 자유도가 증가함에 따라 중심이 오른쪽으로 이동하고 모양이 더 대칭적으로 변화하며, 정규분포에 가까워짐
|---|---|
|매개변수|자유도(df)|
|특징| - 독립적인 표준 정규 분포를 따르는 변수들의 제곱 합 <br>-즉, 표본분산의 분포와 관련된 지표<br>- 주로 독립성 검정, 적합도 검정 등에 사용됨|

5. F분포 (F-Distribution)

|형태| - 0에서 시작하여 오른쪽으로 긴 꼬리를 가진 비대칭 분포<br>- 자유도가 증가함에 따라 중심이 오른쪽으로 이동하고 모양이 더 대칭적으로 변화하며, 정규분포에 가까워짐
|---|---|
|매개변수|자유도(df1, df2)|
|특징| - 두 독립적인 카이 제곱 분포의 비율<br>- 두 집단 간의 분산 차이를 비교하는 데 사용됨|

## 4. 추론 통계 - 추정

### 추론통계란?
추론통계: 표본 데이터를 바탕으로 조사하지 못한 모집단의 특성이나 미래 결과를 추론하는 분야
- **추정:** 표본의 통계량을 이용해 모집단의 모수를 예측하는 과정
   - 점추정: 모수를 하나의 값으로 추정
   - 구간추정: 모수가 포함될 것으로 예상되는 범위로 추정
*모수: 모집단의 특성을 나타내는 값*
*통계량: 표본에서 계산한 값*
- **검정:** 표본 통계량을 이용해 모집단에 대한 가설이 타당한지 판단하는 과정

### 점추정
- 어떤 모수를 표본에 기초하여 계산된 하나의 값으로 추정하는 것

### 구간추정
- 모수의 참 값이 포함되리라고 기대하는 추정치를 일정한 범위와 신뢰도를 가지고 나타내는 방법



*신뢰구간은 표본에서 계산한 구간이 모집단의 실제 평균을 얼마나 잘 포함하는지를 나타냄*

## 5. 추론 통계 - 검정

### 가설 검정 절차

**1. 가설 설정**
- 귀무가설(H0): 기각하고자 하는 명제, 보통 '차이가 없다'
- 대립가설(H1): 내가 주장하고자 하는 명제, 보통 '차이가 있다'

**2. 유의수준(α) 설정**
- 연구자가 허용하는 1종 오류의 최대 확률
- 보통 0.05(5%)
  - 1종 오류(α): 참인 귀무가설을 잘못 기각
  - 2종 오류(β): 거짓인 귀무가설을 기각하지 못함

**3. 검정통계량 산출**
- 가설 검정의 결과를 판단하는 데 사용되는 값
- 대표적으로 Z값, T값, F값 등을 사용

**4. 가설 기각/채택 판단**
- p-value < α → 귀무가설 기각
- p-value > α → 귀무가설을 기각하지 못함

### 데이터의 자료형에 따른 검정 방법

**1. [수치형 - 수치형] 피어슨 상관분석**
- 두 변수 간 피어슨 상관계수가 얼마나 유의미한지 검정하는 방법
   - H0: 두 변수 간 선형 상관관계가 존재하지 않는다.
   - H1: 두 변수 간 선형 상관관계가 존재한다.
- 결과는 보통 (상관계수 r, p-value) 형태로 확인
   - 상관계수의 절댓값 |r|이 클수록 두 변수 간 선형 상관관계가 강하다고 해석
      - |r| < 0.3 → 약한 상관관계
      - 0.3 > |r| < 0.5 → 중간 정도의 상관관계
      - |r| > 0.5 → 강한 상관관계

**2. [범주형 - 수치형] T-test**
- 범주가 2개일 때, 두 집단의 평균에 차이가 있는지 검정
- T-test에서 t값은 두 평균 간 차이를 표준오차로 나눈 값
   - H0: 두 그룹의 평균에는 차이가 없다.
   - H1: 두 그룹의 평균에는 차이가 있다.
- 결과는 보통 t값, p-value, 자유도(df) 형태로 확인
   - 일반적으로 |t|가 클수록 두 집단의 평균 차이가 크다고 볼 수 있음
   - 최종적으로는 p-value를 유의수준과 비교하여 판단
      - p < 0.05 → 귀무가설 기각
      - p > 0.05 → 귀무가설을 기각하지 못함

**3. [범주형 - 수치형] F-test(ANOVA)**
- 범주가 3개 이상일 때, 여러 집단의 평균에 차이가 있는지 검정
- 여러 집단을 2개씩 묶어서 T-test를 반복하는 대신 분산분석(ANOVA)을 사용
   - H0: 모든 집단의 평균에는 차이가 없다.
   - H1: 적어도 하나의 집단 평균에는 차이가 있다.
- F값이 클수록 집단 간 차이가 집단 내 차이에 비해 크다고 해석
- 최종적으로는 p-value를 통해 유의성을 판단

**4. [범주형 - 범주형] Chi2-test**
- 두 범주형 변수 사이에 관련성이 있는지 검정
   - H0: 두 변수는 서로 관련이 없다.
   - H1: 두 변수는 서로 관련이 있다.

- 두 변수가 서로 관련이 없다고 가정했을 때의 값을 기대빈도, 실제로 관측된 값을 관측빈도라고 함
   - 기대빈도와 관측빈도의 차이가 클수록 두 변수가 독립적이지 않을 가능성이 높음
   - 결과는 보통 카이제곱 통계량, p-value, 자유도, 기대빈도 형태로 확인
   - 최종적으로 p-value < 0.05이면 귀무가설을 기각하여 두 범주형 변수 간 관련성이 있다고 판단

**5. [수치형 - 범주형]**
- 수치형 변수를 그대로 둔 상태에서 범주형 결과와의 관계를 검정하는 단순한 하나의 검정 방법은 마땅하지 않음
- 일반적으로는 범주형 변수를 target, 수치형 변수를 feature로 두고 분류모델을 만든 뒤, 해당 변수의 계수가 유의한지 검정하는 방식으로 접근
