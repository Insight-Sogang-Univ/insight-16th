# EDA & 전처리 요약본

## 0. 데이터 분석 Process

- (데이터 수집) 정형/비정형/반정형의 다양한 형식의 데이터 수집
- (데이터 탐색, EDA) 수집한 데이터의 특징을 다양한 각도에서 파악
- (데이터 전처리) 데이터를 분석(모델링)에 적합한 형태로 손질
- (데이터 모델링) 데이터로부터 유용한 정보를 추출하기 위한 모델 구축

## 1. EDA란?

- 시각화 등을 통해 패턴을 발견하거나, 데이터의 특이성을 확인하거나, 통계, 그래프 등으로 가설을 검정하는 과정을 통해 주어진 데이터에 대해 알아보는 것

### EDA 대상

**Univariate(일변량)**
- 한 번에 파악하려는 변수가 한 개로, 데이터를 설명하고 존재하는 패턴을 찾는 것이 주목적

**Multi-variate(다변량)**
- 한 번에 파악하려는 변수가 여러 개로, 여러 변수 간의 관계를 보는 것이 주목적

### EDA 종류

**시각화**
- 차트 혹은 그림 등을 이용하여 데이터를 확인

**비시각화**
- 그래픽적인 요소를 사용하지 않고 주로 Summary Statistics를 통해 데이터를 확인

### EDA 유형

**Uni-Non Graphic(일변량 비시각화)**
- 주어진 데이터의 Distribution을 확인하는 것이 주목적

**Uni-Graphic(일변량 시각화)**
- 주어진 데이터를 전체적으로 살펴보는 것이 주목적

**Multi-Non Graphic(다변량 비시각화)** 
- 주어진 둘 이상의 변수 간 관계를 확인하는 것이 주목적

**Multi-Graphic(다변량 시각화)**
- 주어진 둘 이상의 변수 간 관계를 전체적으로 살펴보는 것이 주목적

## 3. 전처리

- 데이터를 분석에 적합한 형태로 만드는 과정   
   
   - **데이터 정제(Cleansing):** 데이터에서 누락된 결측값을 보완하고 튀는 값을 이상값으로 제거해 데이터를 깨끗하게 만드는 기술
   - **데이터 변환(Transformation):** 데이터를 변환해 일관성을 확보하고 데이터의 중복을 최소화해 데이터 분석 시간을 절약하는 기술
   - **데이터 필터링(Filtering):** 데이터의 오류를 발견하고 삭제 및 보정을 통해 데이터의 품질 향상시키는 기술
   - **데이터 통합(Integration):** 유사한 성질의 데이터를 연계하는 등 데이터를 통합하는 기술

### 결측값 처리
**1. 결측값 확인**
- 데이터프레임 상에서 NaN, ?, 0 등의 값으로 나타나 있는 결측값을 확인
```
info()
value_counts(dropna=False)
isnull()
notnull()
replace()
```

**2. 결측값 처리**

(1) 삭제: 데이터가 있는 행 또는 열을 삭제
```
dropna()
```
*axis = 0 : 결측값이 있는 행 삭제*   
*axis = 1 : 결측값이 있는 열 삭제*   
*how = ‘any’ : 하나라도 결측값 있으면 삭제*   
*how = ‘all’ : 모든 값이 결측값일때만 삭제*   
*inplace = False : 반영한 내용을 새로운 DataFrame으로 반환*   
*inplace = True : 원본 DataFrame 자체를 변경하여 반영*   

(2) 대체: 결측값을 다른 값으로 대체
- 일괄 대체: 모든 변수들 일괄적으로 같은 값으로 대체
- 유사 유형 대체: 범주형 변수들을 활용해 유사한 범주에 따라 다른 값으로 대체
```
fillna()
```

### 이상치 처리
**1. 이상치 확인**

(1) 통계를 통해 확인
```
describe()
```
(2) 시각화를 통해 확인: BoxPlot
```
for i in df.describe().columns:
    df[[i]].boxplot()
    plt.show()
```
(3) Z-score을 통해 확인
- 데이터를 평균(0)과 표준편차(1)로 정규화하여, 평균으로부터 얼마나 떨어져있는지 나타냄   

(4) Tukey Fences을 통해 확인

   - Q1-(1.5*IQR) 미만
   - Q3+(1.5*IQR) 초과


**2. 이상치 제거 및 처리**

(1) 전체 삭제   
- 이상값이 Human error에 의해서 발생한 경우, 해당 관측치를 삭제   

(2) 다른 값으로 대체   
- 절대적 관측치의 숫자가 작은 경우, 이상값이 Human error에 의해 발생했더라도 관측치를 삭제하는 대신 다른 값(평균 등)으로 대체하거나, 결측값과 유사하게 다른 변수들을 사용해서 예측 모델을 만들고, 이상값을 예측한 후 해당 값으로 대체   

(3) 변수화   
- 자연발생적인 이상값의 경우, 바로 삭제하지 말고 좀 더 찬찬히 이상값에 대해 파악
- 이상값을 변수화 하여 이상값을 삭제하지 않고 모델에 포함   

(4) 리샘플링   
- 자연발생한 이상값을 분리해서 모델을 만드는 방법

### 피처 엔지니어링 (변수 가공)
- 해결하고자 하는 문제를 컴퓨터가 잘 이해할 수 있도록 피처(변수)들의 형태를 변형하거나 적절하게 처리하는 과정

**1. 레이블인코딩(Label Encoding) vs 원핫인코딩(One-hot Encoding)**
- Label Encoding: 범주형 변수를 0부터 N-1까지의 숫자로 변환
- One-Hot Encoding: 범주형 변수를 이진 벡터(0과 1)로 변환
   ```
   pd.get_dummies(데이터프레임, columns=[컬럼 리스트])
   ```


**2. 구간화 (Binning)**
- 이산적인 값으로 나타내어 구간별 차이를 드러내는 것
```
pd.cut(데이터프레임['컬럼명'], bins=[나누는 기준 리스트], labels=[지정할 label])
```

**3. 파생변수 생성**
- 기존의 피처를 다른 피처로 변환(재정의)하여 변수 추가

**4. 스케일링**
- 서로 다른 변수의 값 범위를 일정한 수준으로 맞추는 작업
- 각 열(변수)에 속하는 데이터 값을 동일한 크기 기준으로 나눈 비율로 나타내는 정규화(normalization) 작업을 수행

(1) StandardScaler (표준화)   
- 각 feature의 평균을 0, 분산을 1로 변경   

(2) MinMaxScaler   
- 모든 feature가 0과 1사이에 위치하게 만듦   
```
StandardScaler() 
MinMaxScaler()
```

## 4. 데이터 시각화

- 데이터에 대한 이해를 돕기 위해 그림, 도형 등 그래픽 요소들을 이용해 데이터를 묘사하고 표현하는 과정
- 시각적 분석기법을 활용할 때, 일반적으로 Matplotlib과 Seaborn을 사용

### 파라미터
- 매개변수(parameter, 파라미터)
- 프로그래밍된 함수의 입력값을 의미

### 그래프 종류
**1. boxplot(상자 수염 그림)**
- 사분위수와 이상치를 보여주는 그래프

**2. countplot, histplot**
- countplot: 범주형 변수의 빈도수를 확인하는 그래프
- histplot: 도수분포표를 그래프로 나타낸 것으로, 수치형 변수의 구간별 빈도수

**3. displot, kdeplot (커널밀도추정 그래프)**
- 히스토그램을 연속적으로 곡선으로 연결한 그래프

**4. barplot, pointplot**
- 범주형 데이터 값 x에 따른 수치형 데이터 값 y의 평균 값을 제공

**5. scatterplot(산점도 그래프), regplot(회귀선이 추가된 산점도 그래프)**
- scatterplot: 두 연속형 변수 간의 관계를 시각화하기 위해 사용되는 그래프
- regplot: 두개의 연속 변수 사이의 산점도를 그리고 회귀선을 함께 나타내는 그래프

**6. catplot**
- 수치형 데이터와 범주형 데이터의 관계를 볼 때 주로 사용

**7. pieplot**
- 데이터의 부분과 전체 간의 비율을 표현하는 그래프

**8. heatmap**
- 변수간 상관계수를 직관적으로 볼 수 있는 그래프, corr()메서드로 변수간의 상관계수를 구하고 이를 히트맵에 표현

**9. violinplot**
- 박스 플롯과 커널밀도추정 함수 그래프를 합쳐 놓은 그래프

**10. pairplot**
- 여러 변수간의 산점도를 한번에 보여주는 그래프
