## 2. 데이터프레임의 구성

### 2-1. Index와 Column

- 가로 방향이 **행(Index)**, 세로 방향이 **열(Column)**
- 데이터프레임 내에서 **데이터의 위치**를 나타내는 역할

### 2-2. Series

- **값(value)**과 **인덱스(index)**로 구성된 1차원 자료 구조

- **시리즈** 만드는 법
    1. 원하는 값들이 포함된 리스트 생성
    2. pd.Series() 함수를 사용하여 시리즈 객체 생성
    3. 인덱스를 직접 지정해주고 싶으면 pd.Series() 함수의 인자에 인덱스를 추가

### 2-2. DataFrame

- 행과 열로 이루어진 2차원 배열 구조
- 여러 개의 시리즈가 모여 하나의 데이터프레임 구성

- **데이터프레임** 만드는 법
    1. 데이터와 각 카테고리 별 네이밍을 컬럼으로 하는 딕셔너리 생성
    2. pd.DataFrame() 함수를 사용하여 데이터프레임 구성
    3. 인덱스, 컬럼 직접 지정 가능 (기본값은 둘 다 0,1,2…)

## 3. 데이터 확인

### 3-1. head(), tail()

- df.head(n): 앞 n개의 행 반환 (기본값 n=5)
- df.tail(n): 뒤 n개의 행 반환

### 3-2. 데이터에 대한 정보 확인

- df.shape: (행 개수, 열 개수) 반환
- df.info(): 여러 정보들을 종합적으로 반환
- df.describe(): 수치형 열들의 기본적인 통계값 반환
- df.isnull() : 데이터에 결측치(NaN)가 있는지의 여부를 반환
- df.dtypes : 열들의 데이터타입 반환

 ### 3-3. 고유값 및 개수 확인

- df.count() : 결측치의 개수를 제외한 데이터의 행 수 반환
- df[열 이름].value_counts() : 특정 열에 포함된 값들의 빈도수를 세어 반환
- df[열 이름].unique() : 특정 열에 포함된 값들을 종류별로 1개씩만 반환

## 4. 데이터 추출

### 4-1. 인덱싱(Indexing) & 슬라이싱(Slicing)

- 인덱싱(Indexing): 단일 항목을 **지목**할 때 사용
    → 예: `df['노트북 브랜드']`
- 슬라이싱(Slicing): **범위**를 지정해 여러 항목을 추출할 때 사용
    → 예: `df[start:end]`

### 4-2. 행단위 추출

- 인덱스 이름(index label) → `loc` 이용
    - 끝 인덱스를 포함하여 반환
    - `df.loc[행 이름]` ⇒ 행단위 추출
    - `df.loc[행 이름, 열 이름]` ⇒ 값단위 추출 
    - `df.loc[:, 열 이름]` ⇒ 열단위 추출

- 정수형 위치(integer position) → `iloc` 이용
    - 끝 인덱스 포함하지 않고 반환 → 범위 (시작 인덱스 ~ 끝 인덱스 - 1) 
    - `df.iloc[행 순서, 열 순서]`

### 4-3. 열단위 추출 : 컬럼명

- **열 추출 방법**
    - `df[열 이름]` 
    - `df[[열 이름]]` 
    - `df[[열 이름1', '열 이름2...]]`
    - `df[열 이름][행 이름]` 

> * **값 읽기** → `.loc`, `.iloc`, 대괄호 지정 모두 가능
> * **값 수정** → `.loc`만 가능

### 4-4. 불리언 인덱싱 (Boolean Indexing)

- **특정 조건**을 만족하는 데이터 추출
- True 또는 False의 배열을 통해 데이터를 필터링

- **추출 방법**
    - 단일 조건: `df[조건문]`
    - 다중 조건: `df[(조건문1) 논리연산자 (조건문2)]` (and, or 등을 사용)

> * **loc (가능)** → `df.loc[조건문, 열 이름]`
> * **iloc (불가능)**

### 4-5. 인덱스 설정

- **인덱스 변경 함수**
    - `df.set_index(열 이름)` : 기존의 열 중 하나를 인덱스로 설정
    - `df.reset_index()` : 인덱스를 리셋하고 기존 인덱스를 열에 추가
    - `df.reindex(인덱스 배열)` : 인덱스의 재배열에 따라 기존 행데이터도 인덱스를 따라감

## 5. 데이터 조작

### 5-1. 데이터 추가, 삭제

- **열 추가**
    - `df['새로운 열 이름'] = 리스트` : 특정 리스트를 데이터프레임에 추가
- **열 삭제**
    - `df.drop(labels, axis)`

### 5-2. 데이터 정렬

- **데이터 정렬 방법**
    - `df.sort_index()` : 인덱스를 기준으로 정렬
    - `df.sort_values(by="열 이름")` : 특정 열의 값을 기준으로 정렬

### 5-3. 데이터 변환

- **자료형 변환**
    - 현재 데이터타입 확인 : `df.dtypes`
    - `df[열 이름] = df[열 이름].astype(바꿀 자료형)`
- **데이터 값 변환**
    - `df.replace(바꾸고 싶은 값, 바꿀 값)`
- **전치** (행과 열 맞바꾸기)
    - `df.transpose()` 또는 `df.T` 사용 (같은 기능 수행)

### 5-4. 데이터프레임 병합

- concat
    - 단순히 두 데이터 프레임을 이어붙이는 방식
    - 행(가로) 또는 열(세로) 방향을 지정하여 결합
    - `df = pd.concat([df1,df2], axis=0)` ⇒ 행방향 결합
    - `df = pd.concat([df1,df2], axis=1)` ⇒ 열방향 결합

> * **NaN(Not a Number)** : 값이 비어있거나(결측치) 존재하지 않음

- merge
    - **공통 항목(열이나 인덱스)**을 기준으로 병합하는 방식
    - `on` 인자 : 기준이 될 열을 지정
    - `how` 인자 : 집합 연산처럼 어떤 방식으로 결합할지 지정
    - `df = pd.merge(df1, df2, how = 'inner', on = '~~')` ⇒ 교집합(and) 연산
    - `df = pd.merge(df1, df2, how = 'outer', on = '~~')` ⇒ 합집합(or) 연산

## 6. 집계함수

### 6-1. 집계함수란?

- 여러 행으로부터 하나의 결괏값을 반환하는 함수
- 각 컬럼의 평균, 표준편차, 합계 등 연산을 통해 대표값을 구할 때 주로 사용
- 결측치가 있을 경우 무시

- **대표적인 집계함수**
    - `df['컬럼 명'].mean()` : 평균
    - `df['컬럼 명'].median()` : 중앙값
    - `df['컬럼 명'].std()` : 표준편차
    - `df['컬럼 명'].sum()` : 합계
    - `df['컬럼 명'].max()` : 최대값
    - `df['컬럼 명'].min()` : 최소값

> * 각각의 열들에 대한 결과값 ⇒ axis = 0 (행방향 연산)
> * 각각의 행들에 대한 결과값 ⇒ axis = 1 (열방향 연산)

### 6-2. 그룹화(Grouping)

- **`groupby` 메소드** 
    - 특정 기준으로 그룹을 묶어 연산 처리를 할 수 있게 해주는 기능
    - **집계함수**와 함께 사용
- **사용법**
    - `df.groupby('column명1')[’column명2'].집계함수()` : column1로 그룹화를 하고, 그룹 별로 column2에 대해 집계함수를 적용해 결과를 시리즈 형태로 반환
    - `df.groupby('column명1')[[’column명2']].집계함수()` : 그룹 별로 column2에 대해 집계함수를 적용해 결과를 데이터프레임 형태로 반환
