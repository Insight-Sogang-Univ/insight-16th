### 1. 머신러닝과 모델링

1) **머신러닝**  
: 컴퓨터가 스스로 학습해 새로운 데이터를 예측하거나 결정을 내릴 수 있도록 하는 기술

- 머신러닝의 종류
    - 지도 학습: 정답이 있는 데이터 사용  
     
    - 비지도 학습: 정답이 없는 데이터 사용  
    
    - 강화 학습: 보상을 최대화하는 방향으로 행동 학습

2) **모델링**  
: 예측기를 만드는 전체적인 과정  

- **모델**  
: 학습의 결과물

- 모델링 과정
    1) 설계: 어떤 알고리즘으로 학습할지 결정
    2) 훈련: 정답을 맞히도록 수정
    3) 검증: 훈련시킨 모델 테스트  
    

3) **지도 학습**  
: 정답 레이블을 함께 주고 학습시키는 방법  


- 지도 학습 과정
    - **labeled data**: 독립변수와 종속변수가 모두 존재하는 데이터
        1) labeled data를 train set과 test set으로 나누기
        2) train set을 통해 모델 학습
    - 학습 단계
        1) 데이터 준비: 독립변수와 종속변수 갖춘 데이터 마련
        2) 예측: 독립변수 보고 종속변수 맞히기
        3) 오차 계산: 잔차와 손실
        4) 가중치 수정: 오차가 줄어드는 방향으로 가중치 조정
        5) 반복: 2 → 3 → 4 반복
        6) 학습 완료: 함수 획득
        7) 성능 평가: 별도의 테스트 데이터로 측정

- 회귀와 분류
    - **회귀**: 연속적인 숫자값 예측
    - **분류**: 입력된 데이터를 분류
        - 종속 변수가 연속형이면 회귀, 범주형이면 분류 사용

### 2. 회귀

1) **회귀**  
: 데이터에서 패턴을 찾아내 미래 값을 예측하는 것

2) **선형회귀**  
: 회귀선을 찾는 과정

- $y = wx + b$
    - **가중치**: 기울기
    - **편향**: y절편

- 선형회귀분석 종류
    - 단순선형회귀분석: 하나의 독립변수 존재
    - 다중선형회귀분석: 두 개 이상의 독립변수 존재

- **단순선형회귀**
    - 데이터를 직선 형태로 표현 (하나의 독립 변수)
    - $y = wx + b$를 찾는 과정

- **잔차**  
: 실제값과 예측값 간 차이

3) 최적의 $w$와 $b$를 찾는 방법

- **최소제곱법**
    - 잔차 = 회귀선($\hat{y}$)(예측값) - 관측된 데이터값($y$)
    - 잔차의 제곱 합이 최소가 되는 지점으로 최적 회귀선을 구하는 방식

- 경사하강법

4) **다중선형회귀**

- $ y = W_1 x_1 + W_2 x_2 + ... + W_n x_n + b $
    - 독립 변수가 두 개 이상인 경우

- 다중선형회귀의 기본 가정 6가지
    1) 선형성: 종속변수와 설명변수 간의 관계가 선형적이어야 한다  
    $E(y|X) = \beta_1 + \beta_2X_2 + \cdots + \beta_nX_n$
    2) 독립성: 각각의 설명변수가 서로 선형독립적이어야 한다  
    $X_i, X_j \text{ are linearly independent for all } i,j$
    3) 오차항의 평균은 0이거나 0에 가까워야 한다  
    $E(\varepsilon_i|X_i) = 0$
    4) 등분산성: 오차항의 분산이 일정해야 한다  
    $Var(\varepsilon_i|X_i) = \sigma^2$
    5) 오차항의 공분산은 항상 0이어야 한다  
    $Cov(\varepsilon_i, \varepsilon_j|X_i) = 0$
    6) 정규성: 오차항이 정규분포를 따른다  
    $\varepsilon_i \sim N(0, \sigma^2)$

5) **다중공선성**  
: 독립 변수들 간의 강한 상관관계로 인해 회귀 모델이 불안정해지는 문제

- 다중공선성 확인 방법
    1) 상관계수: 변수 간의 통계적 관계인 상관관계의 정도를 수치로 나타낸 것
    2) VIF지수 (분산 팽창 인수): 높으면 다중공선성이 존재한다고 판단

- 다중공선성 대처 방법
    1) 변수 제거
    2) 변수 변환: 변수들을 더하거나 빼서 새로운 변수 생성
    3) 규제선형모델 활용 (릿지, 라쏘, 엘라스틱넷)
    4) PCA (주성분 분석): 데이터의 차원을 축소

6) **규제선형모델**  
: 학습 데이터에 과적합되지 않도록 규제를 가하고자 등장한 모델

- 규제선형모델 종류
    1) L2규제: 회귀계수의 제곱에 대해 페널티를 부여하는 방식, 릿지 회귀
    2) L1규제: 회귀계수의 절댓값에 대해 페널티를 부여하는 방식, 라쏘 회귀
    3) L2규제 + L1규제: 엘라스틱넷 회귀

- **릿지 회귀**
    - 회귀계수가 클수록 강한 패널티
    - 계수 크기 조절
$$
\text{Cost Function} = \text{RSS} + \alpha \sum_{j=1}^{p} w_j^2
$$

- **라쏘 회귀**
    - 회귀계수 크기와 상관없이 같은 패널티
    - 변수 선택 기능
$$
\text{Cost Function} = \text{RSS} + \alpha \sum_{j=1}^{p} |w_j|
$$

- **엘라스틱넷 회귀**
    - L1의 변수 선택 기능과 L2의 규제 기능 동시에 활용
$$
\text{Cost Function} = \text{RSS} + \alpha_1 \sum_{j=1}^{p} |w_j| + \alpha_2 \sum_{j=1}^{p} w_j^2
$$

### 3. 모델 평가 방법

1) **시각화**  


2) **통계 지표**
    1) 모델 유의성 검정
        - F-statistic: 모델 전체에 대한 평가
        - T-statistic: 별 변수의 유의성에 대한 평가
    2) 모델 성능 평가
        - 결정 계수
        - 조정된 결정 계수
        - AIC, BIC

3) **성능 평가 지표**
    1) 평균 제곱 오차: 예측값과 실제값 사이의 오차 제곱의 평균  
    $\text{MSE} = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2$ 
    2) 평균 절대 오차: 예측값과 실제값 사이의 절대값 오차의 평균  
    $\text{MAE} = \frac{1}{n} \sum_{i=1}^{n} |y_i - \hat{y}_i|$

4) **변수 유의성 평가**
    - T-검정: 독립 변수의 회귀계수가 유의미한지 검정

### 4. 비선형회귀

1) **다항식 회귀 모델**  

    - $Y = a + bX + cX² + dX³$
    - 독립 변수에 거듭제곱 항을 추가
    - 곡선 형태의 관계를 모델링할 때 사용

2) **지수 회귀 모델**  

    - $Y = ae^(bX)$
    - Y에 로그를 취한 후 선형회귀 적용
    - 예측값에 지수함수를 적용하여 원래 Y로 복구
    - 지수적으로 증가하거나 감소하는 데이터에 사용

3) **로그 회귀 모델**  

    - $Y = a + b log(X)$
    - X에 로그를 취한 후 선형회귀 적용
    - 초기에 빠르게 증가하고 점진적 감소를 보이는 데이터에 사용

4) **스플라인 회귀**  

    - 데이터를 여러 구간으로 나누어 각 구간에 회귀식 적용
    - 구간별로 서로 다른 데이터 패턴을 보일 때 사용
    - 구간 경계에서 연속성을 유지하는 것이 중요
