### 1. EDA 개념

`EDA: 데이터 분석을 위해 데이터를 여러 가지 방식으로 파악하는 모든 과정`

1) EDA 대상
- Univariate (일변량): 변수가 한 개
- Multi-variate (다변량): 변수가 여러 개

2) EDA 종류
- Graphic (시각화)
- Non-Graphic (비시각화)

3) EDA 유형
- Uni-Non Graphic (일변량 비시각화): 주어진 데이터의 Distribution 확인
- Uni-Graphic (일변량 시각화): 주어진 데이터를 전체적으로 살펴봄
- Multi-Non Graphic (다변량 비시각화): 주어진 둘 이상의 변수 간 관계 확인
- Multi-Graphic (다변량 시각화) : 주어진 둘 이상의 변수 간 관계를 전체적으로 살펴봄

### 2. 데이터 확인

#### `데이터 읽기 / 쓰기`

1) 절대 경로와 상대 경로
- 절대 경로: 전체적인 경로 (URL)
- 상대 경로
    - ./ : 현재 디렉터리
    - ../ : 상위 디렉터리
    - / : 최상위(루트) 디렉터리

2) 데이터 입출력

3) CSV와 EXCEL
- CSV 파일: 데이터를 쉼표로 구분하고 있는 텍스트 파일
- EXCEL 파일: 행과 열이 데이터프레임의 행, 열로 일대일 대응

#### `데이터셋 파악하기`

1) 데이터 프레임 보기
- head(n)

2) 데이터 변수 확인
- 변수 정의, 유형, 형식

3) 데이터 분포 확인

### 3. 전처리

#### `결측값 처리`

1) 결측값 확인

2) 결측값 처리

    a) 삭제: 데이터가 있는 행 또는 열을 삭제
    - dropna()
        
    > <삭제 기준> 
    > - axis = 0 : 결측값이 있는 행 삭제 
    > - axis = 1 : 결측값이 있는 열 삭제 

    > <삭제 조건> 
    > - how = ‘any’ : 하나라도 결측값 있으면 삭제 
    > - how = ‘all’ : 모든 값이 결측값일 때만 삭제 

    > <특정 column 기준으로 삭제> 
    > - subset = [’column명’]
    > - axis = 0일 때만 적용 가능 

    > <원본 데이터 변경 여부> 
    > - inplace = False : 반영한 내용을 새로운 DataFrame으로 반환 
    > - inplace = True : 원본 DataFrame 자체를 변경하여 반영

    `주의할 점: 결측값이 무작위로 발생한 경우에 사용`

    b) 대체: 결측값을 다른 값으로 대체
    - fillna()

    - 일괄 대체: 모든 변수들 일괄적으로 같은 값으로 대체
    - 유사 유형 대체: 범주형 변수들을 활용해 유사한 범주에 따라 다른 값으로 대체

#### `이상치 처리`

1) 이상치 확인

    a) 통계 통해 확인
    - describe()

    b) 시각화 통해 확인 
    -  BoxPlot

    c) Z-score 통해 확인
    - 데이터를 평균(0)과 표준편차(1)로 정규화하여, 평균으로부터 얼마나 떨어져 있는지 나타냄

    d) Tukey Fences 통해 확인
    - IQR 기반

2) 이상치 제거 및 처리

    a) 전체 삭제
    b) 다른 값으로 대체
    c) 변수화
    d) 리샘플링

#### `피처 엔지니어링 (변수 가공)`

1) 레이블인코딩, 원핫인코딩
    - Label Encoding: 범주형 변수를 0부터 N-1까지의 숫자로 변환
    - One-Hot Encoding: 범주형 변수를 이진 벡터로 변환
    - pd.get_dummies(데이터프레임, columns=[컬럼 리스트])

2) 구간화 (Binning): 연속적인 값을 일정한 구간으로 구분 
    - pd.cut(데이터프레임['컬럼명'], bins=[나누는 기준 리스트], labels=[지정할 label])

3) 파생변수 생성: 기존의 피처를 변환해 새로운 피처 생성

4) 스케일링: 서로 다른 변수의 값 범위를 일정한 수준으로 맞추는 작업
    - StandardScaler (표준화): 각 feature의 평균을 0, 분산을 1로 변경
    - MinMaxScaler: 모든 feature가 0과 1사이에 위치
    - StandardScaler(), MinMaxScaler()

### 4. 데이터 시각화



- Matplotlib 라이브러리
- Seaborn 라이브러리

#### `파라미터`
: 프로그래밍된 함수의 입력값

- hue 파라미터: 범주형 변수를 넣어, 내가 원하는 변수를 기준으로 데이터를 구분하여 그래프에 표시

#### `다양한 그래프`

- boxplot: 사분위수, 이상치 확인
- countplot: 범주형 변수의 빈도수 확인
- histplot: 수치형 변수의 구간별 빈도수 확인
- displot, kdeplot: 히스토그램을 연속적으로 곡선으로 연결한 그래프
- barplot, pointplot: 범주형 데이터 값 x에 따른 수치형 데이터 값 y의 평균 값 제공
- scatterplot: 두 연속형 변수 간의 관계를 시각화 (산점도 그래프)
- regplot: 회귀선이 추가된 산점도 그래프
- catplot: 수치형 데이터와 범주형 데이터의 관계 확인
- pieplot: 데이터의 부분과 전체 간의 비율을 표현하는 그래프 
- heatmap: 변수간 상관계수를 직관적으로 볼 수 있는 그래프
- violinplot: 박스 플롯 + 커널밀도추정 함수 그래프
- 여러 그래프 한번에 찍기
- pairplot: 여러 변수간의 산점도를 한번에 보여주는 그래프
