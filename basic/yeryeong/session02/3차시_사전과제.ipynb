### 0. 자료의 척도

**자료**

1) 양적 자료
- 명목척도: 레이블이나 이름으로 요소의 속성을 식별하려고 하는 척도
- 서열척도: 항목 간의 순서 또는 순위를 나타내는 특성을 지니는 척도

2) 질적 자료
- 구간(등간)척도: 값들 사이에 순서가 있고, 그 간격이 일정하여 덧셈이나 평균과 같은 연산이 가능한 척도
- 비율척도: 절대영점과 임의의 측정단위로 구성하고, 동간성이 유지되는 척도

### 1. 기술통계

**기술통계**  
: 수치와 시각화를 통해 샘플을 요약하여 표현하는 방법론

1) 중심화 경향
- 평균: 데이터 값의 총합을 데이터의 개수로 나눈 값  
`series.mean()`  
- 중앙값: 데이터를 크기 순서대로 일렬로 정렬했을 때, 한가운데에 위치하는 값
`series.median()`  
- 최빈값: 데이터에서 가장 자주 나타나는 값  
`series.mode()`  

2) 자료의 산포
- 범위: 데이터의 최대값과 최소값의 차이를 나타내는 지표  
`series.max() - series.min()`  
- 분산: 데이터 값들이 평균으로부터 얼마나 멀리 퍼져있는지  
`series.var()`  
$$
\frac{\sum_{i=1}^{N}(x_i - μ)^2}{N}
$$ 
- 표준편차: 분산에 제곱근을 취한 값  
`series.std()`  
$$
\sqrt{\frac{\sum_{i=1}^{N}(x_i - μ)^2}{N}}
$$
- 사분위수: 데이터의 표본을 네 개의 동일한 크기로 나눈 값
    - Q1 = 하위 25퍼센트, 하위 사분위수
    - Q2 = 하위 50퍼센트, 중앙값 
    - Q3 = 하위 75퍼센트, 상위 사분위수  
`series.quantile(0.25)` # 25th percentile  
- 사분위간 범위 (IQR): Q3에서 Q1 값을 제외한 범위
    - 아래쪽 이상치: Q1 − 1.5×IQR 보다 작은 값
    - 위쪽 이상치: Q3 + 1.5×IQR 보다 큰 값
- 변동계수: 상대적인 변동성을 나타내는 통계적 지표
$$
변동계수 = (표준편차/평균) * 100
$$  
- 첨도 : 확률분포의 꼬리가 두꺼운 정도를 나타내는 척도  
`series.kurt()`  
- 왜도 : 비대칭도, 실수 값 확률 변수의 확률 분포 비대칭성을 나타내는 지표  
`series.skew()`  

3) 자료의 정리 및 시각화

### 2. 확률과 분포

**확률분포**  
: 확률변수가 특정한 값을 가질 확률을 나타내는 함수
- 이산확률분포: 확률 변수가 이산적인 값을 가지는 경우
- 연속확률분포: 확률 변수가 연속적인 값을 가지는 경우

- 정규분포
- Z분포
- T분포
- 카이제곱분포
- F분포

### 3. 추론 통계 - 추정

**추론통계**   
: 추정과 검정을 통해 샘플로부터 모집단을 추론하는 방법론
- 추정
    - 점추정: 추정하고자 하는 모수를 하나의 수치로 추정
    - 구간추정: 추정하고자 하는 모수가 존재하리라 예상하는 구간을 추정
- 검정

- 추정량: 표본정보에 의존하는 확률변수
- 추정치: 표본자료를 이용하여 계산한 통계량의 특정한 값

---

### 4. 추론 통계 - 검정

**가설 검정**  
: 표본의 정보를 사용해서 가설의 합당성 여부를 판정하는 과정

1. 가설 검정 절차

1) 가설 설정
- 귀무가설: 기각하고자 하는 명제
- 대립가설: 주장하고자 하는 명제 

2) 유의 수준 설정
- 유의수준: 연구자가 결과를 판단할 때 허용할 수 있는 오류의 최대치 (0.05)
- 1종 오류(𝛼) : 귀무가설이 실제로 참이었는데 기각할 확률
- 2종 오류(𝛽) : 귀무가설이 실제로 거짓인데 기각하지 못할 확률

3) 검정통계량 산출
- 검정통계량: 가설 검정의 결과를 판단하는 데 사용되는 값
- Z값
- T값
- F값

4) 가설 기각 / 채택 판단
- 유의확률(p-value): 유의 확률(p)가 유의 수준(α)보다 작을 시에는 귀무가설을 기각

2. 데이터 자료형에 따른 검정 방법

1) [수치형 - 수치형] 
- (피어슨)상관분석

2) [범주형 - 수치형] 
- F-test(ANOVA)
- T-test

3) [범주형 - 범주형] 
- Chi2-test

4) [수치형 - 범주형]
