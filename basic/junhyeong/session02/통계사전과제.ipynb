# 자료의 척도

자료는 크게 **양적 자료**(수치로 표현되고 연산이 의미를 가짐)와 **질적 자료**(수치로 표현되어도 연산 의미가 없음)로 나뉘고,
양적 자료는 구간척도·비율척도, 질적 자료는 명목척도·서열척도로 세분된다.

## 질적 자료

**명목척도**
- 레이블이나 이름으로 요소의 속성을 구분하는 척도
- 예: 펀드 유형(국내주식=1, 국제주식=2, 채권=3), 성별(남=0, 여=1)

**서열척도**
- 명목척도처럼 분류하되, 항목 간 순서/순위까지 나타내는 척도
- 예: 만족도(좋음/보통/나쁨), 직급(사원<대리<과장<차장<부장)

## 양적 자료

**구간(등간)척도**
- 값 사이에 순서가 있고 간격이 일정해서 덧셈·평균은 가능하지만, 곱셈·나눗셈은 의미 없음 -> ***온도30도인 지역이 온도5도인 지역보다 6배 더운건 아님!!***
- 예: 온도(섭씨/화씨), IQ, 시간

**비율척도**
- 절대영점이 존재하고 동간성도 유지되는 척도
- 0은 "그 변수가 아예 없음"을 의미 (예: 가격 0 = 무료,무게 0 = 없는거)
- 곱셈 나눗셈 덧셈 뺄셈 다 가능.
- 예: 거리, 높이, 무게, 시간

> 척도에 따라 가장 적합한 요약·분석 방법이 달라진다.

## 빠르게 구분하는 법
1. 덧셈이 가능한가? → 불가능하면 명목척도
2. 0이 의미를 가지는가? → 안 가지면 구간척도, 가지면 비율척도

## 척도 vs 자료의 종류

둘은 관련 있지만 보는 관점이 다르다.
- **자료의 종류** → 데이터의 본질적 성격 (예: 연속형/이산형)
- **척도** → 그 데이터를 측정하는 규칙

예를 들어 온도(섭씨/화씨)와 무게(kg)는 둘 다 무한히 세분 가능한 연속형 이지만 
- 온도는 0℃가 "온도 없음"을 뜻하지 않아 비율 계산이 불가능 → **구간척도**
- 무게는 0kg이 절대적인 영점 → **비율척도**

즉 같은 연속형 자료라도 척도는 다를 수 있다.
정리하면, 자료의 종류는 데이터의 큰 그림, 척도는 그 데이터를 다루는 구체적 규칙이다.

**기술통계가 답하는 질문 예시**

- 제품 A의 불량률은?

## 3가지 주요 도구

- <span style="color:#e0475f">**중심화 경향**</span> — 평균, 중앙값, 최빈값
- <span style="color:#3b82c4">**자료의 산포**</span> — 분산, 표준편차, 범위
- <span style="color:#2e9e5b">**자료의 정리·시각화**</span> — 히스토그램, 박스플롯, 파이차트

---

### <span style="color:#e0475f">중심화 경향</span>

**평균 (mean)**
- 데이터 총합 ÷ 개수. 가장 대표적인 중심 지표
- `2,2,4,5,27` → 합 40 ÷ 5 = **8**
- 총합 기반이라 **이상치에 민감** (ex. 평균 연봉)
- `series.mean()`
- 절사평균(Trimmed Mean): 양끝 일정 비율을 잘라낸 뒤 평균 → 이상치에 강함 (올림픽 채점 방식이 대표 예)

**중앙값 (median)**
- 정렬 후 정가운데 값 (짝수개면 가운데 두 값의 평균)
- `2,2,4,5,27` → **4**
- 평균보다 이상치에 덜 민감
- `series.median()`

**최빈값 (mode)**
- 가장 자주 나타나는 값
- `2,2,4,5,27` → **2**
- 이산형(Discrete) 데이터 분포 파악에 유용
- `series.mode()`

---

### <span style="color:#3b82c4">자료의 산포</span>

**범위 (range)**
- 최대값 − 최소값. 이상치에 **매우** 민감
- `series.max() - series.min()`

**분산 (variance)**
- 평균으로부터 얼마나 퍼져있는지. 값이 클수록 많이 퍼짐
- `series.var()`

**표준편차 (standard deviation)**
- 분산에 제곱근을 취한 값
- 분산은 단위²(예: cm²)라 원 데이터와 안 맞음 → 제곱근 취해서 단위를 다시 맞춤(cm)
- `series.std()`

**사분위수 (quartile)**
- 데이터를 4등분한 값
  - Q1: 하위 25% 지점
  - Q2: 하위 50% = 중앙값
  - Q3: 하위 75% 지점
- `series.quantile(0.25 / 0.5 / 0.75)`, `np.percentile(series, 75)`


**사분위간 범위 (IQR)**
- Q3 − Q1, 데이터 중앙 50%가 퍼진 범위. 이상치 탐지에 사용 (Tukey Fence)
  - 아래쪽 이상치: Q1 − 1.5×IQR 미만
  - 위쪽 이상치: Q3 + 1.5×IQR 초과
  

**변동계수 (CV, coefficient of variation)**
- (표준편차 ÷ 평균) × 100 — **상대적** 변동성 비교
- 코끼리(평균 5000kg, 표준편차 100kg) vs 사람(평균 65kg, 표준편차 10kg)
  - 표준편차만 보면 코끼리가 더 변동 커 보임
  - 변동계수는 코끼리 2%, 사람 약 15% → **상대적으로는 사람의 몸무게 변동이 더 큼**

---

###  첨도와 왜도

- **첨도**: 분포 꼬리가 두꺼운 정도 -> ***얼마나 뾰족한가, 3이면 정규분포 3보다 크면 뾰족***

- **왜도**: 비대칭도
  - 양수 → 오른쪽 꼬리가 길고, 자료는 왼쪽에 몰림
  - 음수 → 왼쪽 꼬리가 길고, 자료는 오른쪽에 몰림
  - 평균 = 중앙값이면 왜도 0

  -***꼬리가 긴쪽 방향을 자료의 값(양수.음수)으로 삼으면 편함, 부호를 기준으로 최빈값과 평균값이 서로 바뀌는데(중앙값은 항상 셋 중 중앙 위치) 그림으로 판단하면 편함.***
  
- `series.kurt()` (첨도), `series.skew()` (왜도)

# 확률과 분포

- **확률변수**: 랜덤한 사건의 결과를 특정 값으로 대응시키는 함수 (ex. 동전 앞면=1, 뒷면=0)
- **확률분포**: 확률변수가 특정 값을 가질 확률을 나타내는 함수
- **이산확률분포**: 확률변수가 이산적(유한/셀 수 있음)인 값을 가짐
  - 베르누이, 이항, 기하, 음이항, 초기하, 포아송분포 ***TIP) 외울때는 누포항하로 외우면 됨***
- **연속확률분포**: 확률변수가 연속적(무한 범위, 특정값 선택 불가능)인 값을 가짐
  - 정규, T, F, 감마, 지수, 카이제곱, 균등분포

##  대표적인 분포 모형 (연속확률분포)

### 정규분포 (Normal Distribution)
| 형태 | 종모양, 평균 기준 대칭, 표준편차 클수록 평평 |
|---|---|
| 매개변수 | 평균(μ), 표준편차(σ) |
| 특징 | 가장 잘 알려진 연속확률분포. 많은 자연/사회현상이 따름 |

- 모집단이 정규분포면 표본평균의 분포도 정규분포
- **중심극한정리**: 표본크기(n)가 커지면(n≥30) 모집단 분포와 무관하게 표본평균의 분포가 정규분포에 근사
- **큰 수의 법칙**: 표본 크기가 커질수록 표본평균이 모평균에 근사 → 데이터가 많을수록 통계적 정확도↑

**Z분포(표준정규분포)**: 평균 0, 표준편차 1로 맞춘 표준 정규분포

> T분포·카이제곱분포·F분포는 정규분포와 성격이 다름 — 이들은 **정규분포를 따르는 모집단에서 추출한 표본 통계량**이 따르는 '표본분포'로, 가설검정·구간추정 같은 통계적 추론을 위한 도구. 올바르게 쓰려면 표본이 정규분포 모집단에서 왔다는 전제가 필요하며, 이를 확인하는 게 **정규성 검정**.


### T분포
| 형태 | 종모양, 정규분포보다 꼬리 두꺼움, 대칭, 자유도↑일수록 N(0,1)에 수렴 |
|---|---|
| 매개변수 | 자유도(df) |
| 특징 | ***표본으로 모평균 추정 시 사용*** 표본이 작을 때 정규분포보다 정확 |

- **자유도**: 독립적으로 변할 수 있는 정보의 수 (표본크기 n에서 -1). 자유도↑ → T분포가 정규분포에 가까워짐
- 모분산을 모를 때 표본분산으로 추정하기 때문에 T분포가 표준정규분포보다 더 퍼져있음 (불확실성 반영)
- 표본크기 30 미만 → T분포, 30 이상 → 표준정규분포 (중심극한정리 근거)

### 카이제곱분포 (Chi-Square Distribution)
| 형태 | 0에서 시작, 오른쪽 긴 꼬리(비대칭), 자유도↑일수록 대칭적/정규분포에 근접 |
|---|---|
| 매개변수 | 자유도(df) |
| 특징 | 독립적인 표준정규분포 변수들의 제곱합. 표본분산 분포와 관련. 독립성/적합도 검정에 사용 |

### F분포 (F-Distribution)
| 형태 | 0에서 시작, 오른쪽 긴 꼬리(비대칭), 자유도↑일수록 대칭적/정규분포에 근접 |
|---|---|
| 매개변수 | 자유도(df1, df2) |
| 특징 | 두 독립적인 카이제곱분포의 비율. ***두 집단 간 분산 차이 비교에 사용*** |

## 분포모형을 알아야 하는 이유
1. 가설검정의 검정통계량 산출에 사용
2. 표본추출 과정의 인위적 조작(non-random sampling) 여부 확인 가능
   - 예: 스모 선수 승수 분포가 '8승' 기준에 비정상적으로 몰림 → 승부조작 가능성 시사



# 추론통계
<bn>? 이 샘플의 모집단에 대한 관심 -> 샘플로부터 모집단을 추론하는 방법

추론통계 논리의 흐름:
- 표본의 불확실성 → **표본분포**
- 모수 추정 → **점추정**
- 추정의 현실성 → **구간추정**
- 구간추정 오류관리 → **신뢰수준**
- 주장 검증 → **가설검정**
- 가설검정 오류관리 → **유의수준(제1종 오류)**

**용어 정리**
- **모수(Parameter)**: 모집단의 특성 (평균, 비율, 표준편차 등)
- **통계량(Statistic)**: 표본으로 계산한 수치 (표본평균, 사분위수, 분산 등)
- **추정량**: 모수 추정에 쓰이는 표본통계량(확률변수)
- **추정치**: 표본자료로 실제 계산해낸 값
- 점추정은 오차를 고려하지 않기 때문에, 오차를 포함하는 **구간추정**이 널리 쓰임

##  추정

**점추정**: 모수를 하나의 값으로 추정

**구간추정**: 점추정에 오차개념을 더함.

##  검정

### 가설검정 절차
**가설 설정 → 유의수준 설정 → 검정통계량 산출 → 가설 기각/채택 판단**

**1) 가설 설정**
| 귀무가설(H0) | 대립가설(H1) |
|---|---|
| 현재 통용되는 가설, 기각하고자 하는 명제 ("차이 없다") | 새롭게 주장하는 명제 ("차이 있다") |
| 지지될 때 위험성 낮음 (ex. 무죄추정) | 지지될 때 위험성 높음 |

**2) 유의수준**
- 연구자가 허용하는 오류의 최대치, 보통 0.05(5%)
- 귀무가설이 참인데도 잘못 기각할 확률을 감수하겠다는 의미
- **1종 오류(α)**: 귀무가설이 참인데 기각할 확률 (유의수준 = 1종 오류 확률)
- **2종 오류(β)**: 귀무가설이 거짓인데 기각 못 할 확률
- 가설검정은 1종 오류를 더 치명적으로 봄

**3) 검정통계량**
- **Z값**: `(X-μ)/σ`. Z값 또는 Z-점수는 표준 정규 분포(평균이 0이고 표준편차가 1인 분포)를 사용하여 얻은 값입니다.
특정 데이터가 평균으로부터 얼마나 떨어져 있는지를 표준편차 단위로 나타냅니다.

- **T값**: `(X-μ)/(s/√n)`. 작은 표본, 모평균 검정/두 표본평균 차이 검정에 사용. 같은 유의수준에서 t값이 z값보다 크고, 자유도 커지면 z값에 수렴

- **F값**: 두 독립 카이제곱분포의 비율. 두 표본 분산 비교, ANOVA에 사용. 1에 가까우면 두 분산이 비슷함을 의미

**4) 가설 기각/채택 판단 — p-value**
- p값 < 유의수준 → 귀무가설 기각 / p값 ≥ 유의수준 → 기각 못 함
- 예시(교육 프로그램 비교): t값 2.05, p값 0.02 < 유의수준 0.05 → 귀무가설 기각(새 프로그램이 효과 있다고 결론)

### 자료형에 따른 검정 방법

**① [수치형-수치형] 피어슨 상관분석**
- H0: 선형 상관관계 없음 / Ha: 선형 상관관계 있음
- 공분산: `Cov(X,Y)`, 두 변수가 함께 변하는 방향과 크기. 양수=양의관계, 음수=음의관계, 0=선형관계 없음(비선형은 있을 수 있음) → `np.cov()`
- 상관계수: 공분산을 표준편차로 나눠 -1~1로 표준화. **상관관계 ≠ 인과관계** → `df[['col1','col2']].corr()`, `spst.pearsonr()` (상관계수, p값 튜플)
  - |r|<0.3 약함, <0.5 중간, >0.5 강함 (참고용 기준)
- 스피어만 상관계수: 데이터를 ***순위화(서열화)***한 뒤 피어슨 상관계수를 구한 것

**② [범주형-수치형] T-test / F-test(ANOVA)**
- 범주 2개 → **T-test**
  - H0: 두 그룹 평균 차이 없음 / Ha: 차이 있음 ("차이 없음"은 대립가설에 못 넣음 — 귀무가설을 기각 못했다는 식으로 주장)
  - `spst.ttest_ind(그룹1, 그룹2)` → (t값, p값, df). |t|>2 기준으로 차이 판단, p<0.05 기준
  - `ind`=independent(독립 집단), `rel`=같은 집단의 전/후 비교(`ttest_rel`)
  - 단측검정: 양측검정 결과의 p값을 2로 나눠서 사용
- 범주 3개 이상 → **ANOVA(F-test)**
  - H0: 각 그룹 평균 차이 없음 / Ha: 차이 있음
  - F값 = 집단 간 분산 / 집단 내 분산
  - `spst.f_oneway(그룹1, 그룹2, 그룹3)` → (f값, p값). f값 2~3 이상이면 차이 있다고 봄
 

**③ [범주형-범주형] 카이제곱검정(Chi2-test)**
- H0: 두 범주형 변수는 관련 없음 / Ha: 관련 있음
- **기대빈도**(독립이라 가정했을 때 예상되는 값) vs **관측빈도**(실제 값)를 비교
- 카이제곱 통계량 = Σ (관측빈도-기대빈도)²/기대빈도
- 범주가 많아질수록 통계량이 커지므로, 특정 숫자가 아닌 **자유도의 약 2배**를 기준으로 판단
- 자유도 = (x변수 범주수-1) × (y변수 범주수-1)
- `table = pd.crosstab(df['col1'], df['col2'])` → `spst.chi2_contingency(table)` → (카이제곱값, p값, 자유도, 기대빈도)
