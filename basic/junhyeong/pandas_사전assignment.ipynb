# 2-1 데이터프레임이란?

판다스에서 사용하는 주요 데이터 구조이자 정보를 정리하고 저장하는 도구.
행과 열로 이루어져 있고, 행과 열은 데이터프레임에서 위치를 나타낼 수 있음 (index, column).
여러 개의 시리즈(Series)가 모여 하나의 데이터프레임을 구성함.

> **시리즈란?** 1차원 자료구조이자 특정 하나의 컬럼일 수 있음. 시리즈를 구성하는 값들은 모두 같은 데이터타입이어야 함.

# 2-2 데이터프레임 만드는 법

```python
data = {
    '1열': ['a', 'b'],
    '2열': ['c', 'd'],
    '3열': ['e', 'i'],
    '4열': ['f', 'g']
}  # <- 딕셔너리 생성

df = pd.DataFrame(data)  # <- 데이터프레임 구성
```

**결과:**

| | 1열 | 2열 | 3열 | 4열 |
|---|---|---|---|---|
| 0 | a | c | e | f |
| 1 | b | d | i | g |

기본 인덱스는 0, 1, 2 ... 이지만 직접 지정 가능!

```python
index = [1, 2]
df = pd.DataFrame(data, index=index)
```

> 분석 전 데이터를 먼저 살펴보는 이유?
> 실제 다루게 될 데이터는 크기가 방대해서 일일이 눈으로 확인하기 어렵기때문에 함수들을 활용해 **데이터의 일부나 전체적인 특징을 요약해서 빠르게 파악**하는 습관이 필요.

### 앞/뒤 일부 데이터 보기
데이터의 대략적인 구조와 들어있는 값들을 개괄적으로 파악할 때 사용합니다.
 `df.head(n)` : 앞 n개의 행 반환 (괄호를 비우면 기본값 5개 반환)
 
 `df.tail(n)` : 뒤 n개의 행 반환

### 데이터 전반적인 정보 확인
데이터프레임의 크기, 결측치 여부, 데이터 타입 등을 파악합니다.
df.shape : (행 개수, 열 개수) 반환 
df.info() : 컬럼명, 결측치, 데이터 타입 등 종합적인 정보 반환
df.describe() : 수치형 열의 기본 통계값(평균, 최대/최소, 표준편차 등) 반환
df.isnull() : 데이터의 결측치 여부를 반환
df.dtypes : 각 열의 데이터 타입 반환
 
> | 판다스 자료형 | 파이썬 자료형 | 설명 |

> | **int64** | `int` | 정수형 데이터 |

> | **float64** | `float` | 실수형 데이터 |

> | **object** | `string` | 문자열 데이터 |

> | **datetime64** | - (datetime 라이브러리) | 시간 데이터 |

###  고유값 및 데이터 개수 확인
 `df.count()` : 결측치를 제외한 실제 데이터의 개수 반환

 `df['열 이름'].value_counts()` : 특정 열에 포함된 값들의 **빈도수** 반환

 `df['열 이름'].unique()` : 특정 열에 포함된 값들을 **종류별로 1개씩(중복x)** 반환

# 3-2 데이터 추출방법

> **데이터 추출이 중요한 이유**
> 실무에서 다루는 데이터는 방대하기 때문에, 분석에 필요한 알맹이만 골라내어 가공하는 능력이 매우 필요함.

### 인덱싱(Indexing) & 슬라이싱(Slicing)

인덱스와 컬럼을 사용해 데이터를 추출하는 기본 방식입니다.

- **인덱싱(Indexing):** 단일 항목 지목 (예: `df['브랜드']`)
- **슬라이싱(Slicing):** 범위를 지정해 여러 항목 추출 (예: `df[start:end]`)

```python
data[:1]   # 0번째 데이터
data[1:3]  # 1~2번째 데이터
```

### 행단위 추출 : loc과 iloc

- 인덱스 이름을 기준으로 행을 선택할 때는 `loc`, 정수형 위치를 사용할 때는 `iloc`
- `df.loc[행 이름, 열 이름]` vs `df.iloc[행 순서, 열 순서]`
- `loc['a':'c']` → 'a', 'b', 'c' (끝 포함)
- `iloc[3:7]` → 3, 4, 5, 6 (끝 제외)

### 열단위 추출

- 값을 읽기만 할 때는 열 이름으로 인덱싱해도 되지만, **값을 수정할 때는 `.loc` 사용**

### 불리언 인덱싱

인덱스나 컬럼명이 아닌, 특정 조건에 맞는 데이터만 걸러낼 때 True/False 배열을 활용합니다.

```python
# 단일 조건
df[조건문]

# 다중 조건 (각 조건문은 반드시 괄호로 감싸야 함)
df[(조건문1) & (조건문2)]  # AND 연산
df[(조건문1) | (조건문2)]  # OR 연산

# loc을 활용한 불리언 인덱싱 (조건 만족하는 행 + 특정 열 추출)
df.loc[조건문, '열 이름']
```

###  데이터 추가/삭제
- 열 추가: `df['새 열'] = 리스트`
- 열 삭제: `df.drop(labels, axis)`

**데이터프레임 산술연산**
- Column ↔ Column: 같은 인덱스끼리 연산, 인덱스 불일치 시 NaN
- Column ↔ 숫자: 브로드캐스팅으로 전체 값에 동일 연산 적용
- NaN 포함 연산: 결과도 NaN → `fill_value` 옵션으로 대체 후 연산 가능
```python
  df['국어'].add(df['영어'], fill_value=0)
```


**axis 개념**
- `axis=0` : 행 방향 (책을 위로 쌓는 방향)
- `axis=1` : 열 방향 (책을 옆으로 쌓는 방향)
- 삭제/병합/평균 등 연산 방향 지정 시 사용

**inplace=True**
- 기본은 비파괴적(원본 유지, 새 객체 반환)
- 원본을 바로 바꾸려면 `inplace=True` 또는 `df = df.drop(~)`처럼 재할당


###  데이터 정렬
- `df.sort_index()` : 인덱스 기준 정렬
- `df.sort_values(by="열 이름")` : 특정 열 값 기준 정렬



###  데이터 변환
- 자료형 확인/변환: `df.dtypes`, `df[열] = df[열].astype(자료형)`
- 값 변환: `df.replace(기존값, 새값)`
- 전치: `df.transpose()` = `df.T`



### 데이터프레임 병합

| 구분 | 방식 | 특징 |
|---|---|---|
| concat | 단순 이어붙이기 | 동일 구조 데이터프레임에 사용, axis=0(행)/1(열) 방향 지정 |
| merge | 공통 항목 기준 결합 | `on`으로 기준 열 지정, `how`(inner=교집합/outer=합집합)로 결합 방식 지정 |



# 4-1 집계함수
여러 행으로부터 하나의 결과값을 반환하는 함수. 각 컬럼의 평균, 표준편차, 합계 등 연산을 통해 대표값을 구할 때 주로 사용. (결측치가 있을 경우 무시하고 계산.)

**[대표적인 집계함수]**

 `df['컬럼 명'].mean()` : 평균
 `df['컬럼 명'].median()` : 중앙값
 `df['컬럼 명'].std()` : 표준편차
 `df['컬럼 명'].sum()` : 합계
 `df['컬럼 명'].max()` : 최대값
 `df['컬럼 명'].min()` : 최소값
 >  **`axis=0` (행 방향 연산):** 각각의 열(Column)들에 대한 결과값을 얻고 싶을 때
>  **`axis=1` (열 방향 연산):** 각각의 행(Row)들에 대한 결과값을 얻고 싶을 때
