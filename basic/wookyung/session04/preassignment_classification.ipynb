# 분류

## 1. 분류란
- 지도학습의 일종
- 범주형 변수 예측하기 위해 사용
- 대표적으로 사용되는 머신러닝 알고리즘
    - 로지스틱 회귀
    - 결정 트리
    - 서포트 벡터 머신
    - 최소 근접
- 이진 분류 vs. 다중 분류
    - 이진 분류 : 참, 거짓
    - 다중 분류 : 값이 3개 이상
## 2. 분류 모델
- 로지스틱 회귀
    - 출력값 0에서 1사이
    - 이진 분류에 적합
    - 시그모이드 함수(로지스틱 함수) 사용 - S자 곡선
- 의사결정 나무
    - 조건에 따라 데이터 분류, 분류 반복
    - Root Node, Parent Node, Child Node 등등으로 구성
    - CART 알고리즘이 대표적
- SVM
    - 클래스를 분류하는 최적의 라인을 찾는 것
- KNN
    - 비슷한 데이터 가까이 이쓴ㄴ 것 이용

## 3. 분류 평가 지표
- 혼동 행렬
    - 예측을 정확한 예측/잘못된 예측으로 구분하여 표로 나타냄
    - 정확도 : (TP+TN) / (TP+TN+FP+FN)
    - 정밀도
    - 재현도
- F1-Score
    - 정밀도와 재현율의 조화 평균
- ROC/AUC
    - 얼마나 분류가 잘되었는거
    - TPR, FPR
## 4. 하이퍼파라미터 최적화
- 하이퍼파라미터: 학습 시작 전에 사용자가 직접 설정하는 변수
- Grid Search
- Random Search
- Bayesian Optimization
