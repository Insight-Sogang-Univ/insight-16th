### < EDA >

#### 1. EDA란?
- Exploratory Data Analysis
- 수집한 데이터를 이리저리 탐색해보면서 데이터의 패턴과 특성을 파악해보는 것
- 시각화를 많이 활용

#### 2. EDA 대상
- 일변량
    - 특정 변수 1개에 주목
    - 그 변수의 패턴과 특성 도출이 목적
    - 다변량 EDA 전에 하면 좋음
- 다변량
    - 여러 변수에 주목
    - 여러 변수들 간의 관계를 보는 것이 주목적

#### 3. EDA 종류
- 시각화
    - 그래프나 차트 등으로 데이터를 시각화해서 특성/관계 파악
    - 전체 그림을 파악하는데 유용
- 비시각화
    - Summary와 Statistics등을 위주로 데이터의 특성/관계 파악
    - 정확한 값 파악에 유용

#### 4. EDA 유형

EDA 대상과 EDA 종류에 따라 크게 4가지로 구분됨
- 일변량 비시각화
- 일변량 시각화
- 다변량 비시각화
- 다변량 시각화

### <데이터 >

#### 1. 데이터 읽기
- 절대경로
    - 처음 시작부터 주소를 쭉 쓰는 것
    - ex. C:\Users\사용자 이름\OneDrive\Desktop 
- 상대경로
    - 현재 작업위치 기준으로 주소를 쓰는 것
    - 위치 표현
        - ./ : 현재 디렉터리
        - ../ : 상위 디렉터리
        - /: : 최상위 디렉터리

#### 2. 데이터 입출력
- 데이터 입력(Reader)
    - 'read_파일형태' 함수 사용
- 데이터 출력(Writer)
    - 'to_파일형태' gkatn tkdyd
- CSV, EXCEL 파일형태를 많이 사용


#### 3. CSV & EXCEL
- CSV
    - 데이터가 ,로 구분되어 있음
    - 크기가 작고 압축이 용이함
- Excel
    - 행열이 DataFrame의 행열에 일대일 대응
    - 여러 개의 시트가 있는 경우 특정 시트만 불러올 수 있음
    - 여러 시트를 불러오고 싶으면 list로 불러오면 됨
        - sheet_name = ['Sheet1', 'Sheet2']

#### 4. 데이터셋 파악
- 데이터 프레임 파악
    - head(n) -> 데이터 구조 파악
- 데이터 변수 파악
    - 변수 정의 : 어떤 정보와 관련된 변수인지
    - 변수 유형 : 양적 변수 / 질적 변수인지
        - 양적 변수(수치형)
            - 이산형: 딱딱 나눠지는 데이터 -> ex. 형제 수
            - 연속형: 연속적인 범위의 데이터 -> ex. 키, 몸무게
        - 질적 변수(범주형)
            - 명목형: 이름을 붙이는 데이터 -> ex. 성별, 이름
            - 순서형: 순서가 있는 데이터 -> ex. 승객 등급 
    - 변수 데이터 유형 : 수치 / 텍스트 / 날짜인지
- 데이터 분포 확인
    - 산포도, 박스 플롯 히스트그램 등을 활용
    - 데이터의 평균값, 최빈값, 중간값 등을 파악


### <전처리>

#### 1. 전처리란?
- 데이터를 분석에 적합한 형태로 처리하는 것
- 데이터의 품질을 높이는 작업
- 전처리 기술
    - 데이터 정제 : 결측값 보완, 이상치 제거
    - 데이터 변환 : 데이터 일관성 확보 & 중복 최소화
    - 데이터 필터링 : 데이터 오류 삭제 및 보정
    - 데이터 통합 : 유사한 성질 데이터 연계 및 통합

#### 2. 결측값 처리

1] 결측값 확인
- 결측값이 존재하면 정확성 감소, 왜곡 발생 가능
- NaN, ?, 0 값 확인
    - 0 : 의미있는 수치의 0인지, 결측되어 0인지
- 결측값 확인용 pandas function
    - info() : 각 열별 유효값 개수 보여줌
    - isnull() : 누락값이면 True, 유효값이면 False 반환
        - df.isnull().sum() 의 형식으로 자주 사용
    - value_counts(dropna=False) : 결측값 포함 전체 데이터 확인


2] 결측값 처리
- 삭제
    - dropna()
        - df.dropna(axis=0/1, how='any'/'all', subset=[col1, col2..], inplace=True/False)
            - axis=0/1 : 0이면 결측값 있는 행 삭제, 1이면 열 삭제
            - how='any'/'all' 
                - any : 하나라도 결측값 있으면 삭제, default
                - all : 모든 값이 결측값일때만 삭제
            - subset=[col1, col2, ...]
                - axis=0일 때만 적용 가능
                - 결측치가 있는 행의 특정 열들 삭제
            - inplace=True/False
                - True : 원본 df자체 변경
                - False : 변경한 df를 새로운 df로 반환
    - 삭제는 결측값이 무작위로 발생된 경우에만
    - 그렇지 않은 경우에 삭제하면 데이터 왜곡이 발생됨
- 대체
    - 결측값을 다른 값으로 대체
    - 평균값, 최빈값 등 대푯값 활용
    - 대체 종류
        - 일괄 대체 : 일괄적으로 같은 값으로 대체
        - 유사 유형 대체 : 유사한 범주에 따라 다른 값으로 대체
    - fillna()
        - 일괄 대체 : fillna(특정 값, inplace = True)
        - 유사 유형 대체
            - 평균값 계산
            - fillna(x.mean())
    - 이웃한 값으로 대체
        - ffill() : 바로 직전행 값으로 대체
        - bfill() : 바로 다음행 값으로 대체

#### 3. 이상치 처리

1] 이상치(Outlier) 확인
- 통계를 통해 확인 : describe()
- 시각화를 통해 확인 : boxplot()
- Z-score를 통해 확인 : z-score 일반적으로 2~3 넘어가면 이상치 간주
- Tukey Fences : [(X < Q1 - 1.5*IQR) | (X > Q3 + 1.5*IQR)] 이면 이상치

2] 이상치 제거 및 처리
- 전체 삭제
    - Human error에 의해서 발생한 경우
- 다른 값으로 대체
    - 관측치 수가 절대적으로 적은 경우
    - 이상값을 다른 값으로 대체
- 변수화
    - 자연발생한 이상치인 경우
    - 좀 더 상세히 관찰 후 변수화해서 이상치 설명
- 리샘플링
    - 자연발생한 이상치인 경우
    - 이상치 포함/불포함 모델을 모두 만들기

#### 4. 피처 엔지니어링
1] 피처 엔지니어링이란?
- 데이터 전처리의 마지막 단계
- 기존의 데이터에 정보를 추가하여 데이터를 보다 유용하게 만드는 일련의 과정
- 도메인 지식 영향 높음

2] 피처 엔지니어링 방식
- 레이블 인코딩, 원핫인코딩
    - 레이블 인코딩: 범주형 변수를 0부터 n-1까지의 숫자로 변환
    - 원핫 인코딩: 범주형 변수를 이진 벡터(0/1)로 변환
        - pd.get_dummies(데이터프레임, columns=[컬럼 리스트]) 
- 구간화
    - 연속 데이터를 일정한 구간으로 나눠서 분석
    - pd.cut(데이터프레임['컬럼명'], bins=[나누는 기준 리스트], labels=[지정할 label])
- 파생변수 생성
    - 기존의 데이터를 다른 관점에서 분석하기 위해 하는 과정
    - 기존 피처 -> 다른 피처로 재정의
- 스케일링
    - 정규화 작업 -> 열별 숫자 단위 차이에 따른 왜곡 방지 목족
    - 표준화 : 각 feature 평균 0 , 분산 1로 변경
        - StandardScaler() -> from sklearn.preprocessing
    - MinMaxScaler : 모든 feature가 0과 1사이에 존재하게 변경
        - MinMaxScale() -> from sklearn.preprocessing


### <데이터 시각화>

#### 1. 데이터 시각화란?
데이터에 대한 이해를 돕기 위해 그래픽 요소로 데이터를 묘사하는 것.

#### 2. 데이터 시각화 하는 방법
- Matplotlib
    - Numpy를 활용한 플로팅 라이브러리
- Seaborn
    - Matplotlib 기반으로 만들어진 통계 데이터 시각화에 최적화된 라이브러리


#### 3. 파라미터
- 파라미터(parameter) : 프로그래밍된 함수의 입력값으로 요구되는 것
- 인자(argument) : 파라미터로 들어갈 값


#### 4. 다양한 그래프
- boxplot(상자 수염 그림)
    - 사분위수, 이상치 보여줌
- countplot, hisplot
    - countplot: 범주형 변수의 빈도수 보여줌
    - hisplot: 수치형 변수의 구간별 빈도수 보여줌
- displot, kdeplot
    - 히스토그램을 곡선으로 연속적으로 연결한 그래프
- barplot, pointplot
    - barplot: 범주형 데이터 값 x에 따른 수치형 데이터 값 y의 평균 값 제공, 막대그래프 형식으로
    - pointplot : barplot와 동일, 꺾은선그래프 형식으로
- scatterplot, regplot
    - scatterplot : 두 연속형 변수 간의 관계 시각화 -> 변수 간 패턴/상관관계/분포 파악에 유리
    - regplot: 두 연속형 변수 사이 산점도 시각화 & 회귀선
- catplot: 수치형 데이터와 범주형 데이터 관계 파악시 사용
- pieplot: 모든 데이터 sum이 전체인 경우, 각 데이터 비율 파악시 사용
- heatmap: 변수간 상관계수 직접 확인하고 싶을시 사용
- violinplot: boxplot + kdeplot
- pairplot: 여러 변수간의 산점도 한번에 보여주는 그래프
