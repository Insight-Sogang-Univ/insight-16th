# 연관분석

## 1. 연관분석이란?
### 1-1. 연관 분석의 정의
- 연관 분석 : 일련의 거래나 사건 안에 존재하는 항목 간의 일정한 '연관 규칙'을 발견하는 분석

- 정답값이 존재하지 않는 비지도 학습에 해당함
- 대표적인 연관분석 알고리즘 : Apiriori, FP-Growth

### 1-2. 연관규칙 분석
- 연관 규칙의 형태 : 조건절(A) + 결과절(B) , A -> B로 표현
- 아이템 집합 : 조건절 또는 결과절을 구성하는 아이템들의 집합
    - 상호배타적이어야 함
    - ex. 라면->[계란, 파]
- 연관규칙으로 인과관계를 알 수는 없다. 


## 2. 주요 개념
### 2-1. 적절한 연관규칙의 조건

1. 두 품목이 함께 구매된 경우의 수가 일정 수준 이상이어야 함(일정 이상 지지도)
2. 품목 A를 포함하는 거래 중 품목 B를 구입하는 경우의 수가 일정 수준 이상이어야 함(일정 이상 신뢰도)

### 2-2. 연관 분석의 평가 측도

- 지지도 : $ 지지도 = P(A \cap B) $
    - 지지도가 낮다면? -> 우연일 수 있다. -> 분석할지 말지 결정하는 척도

- 신뢰도 : $ 신뢰도 = \frac{P(A \cap B)}{P(A)} = \frac{지지도}{P(A)} $
    - 신뢰도가 낮다면? -> 연관 규칙이 무의미할 수 있다 -> 규칙의 정확도 파악 척도
- 향상도 : $ 향상도 = \frac{P(A \cap B)}{P(A)P(B)} = \frac{신뢰도}{P(B)} $
    - 향상도 > 1 -> 우연적 기회보다 높은 확률
    - 향상도 = 1 -> 독립
    - 향상도 < 1 -> 우연적 기회보다 낮은 확률

- 규칙의 효용성은 위 3가지를 **모두** 반영해 평가해야 함

## 3. 활용 사례

### 월마트 팝타르트 사례
- 지지도 관점 : $ P(재난대비 용품 \cap 딸기 팝타르트) $ 을 확인
    - 포인트는 허리케인 경보 기간 중으로 기간을 한정했다는 것
- 신뢰도 관점 : $ P(딸기 팝타르트 \mid 재난대비 용품 ) $ 을 확인
    - 값이 유의미하게 높았다 -> 분석 정확하다
- 향상도 관점 : A를 허리케인 경보 상황 B를 딸기 팝타르트 구매라고 했을 때 값이 **7**
    - 평소보다 허리케인 경보 상황에 딸기 팝타르트를 7배 더 많이 구매한다

## 4. 연관 분석 알고리즘

### 4-1. Apriori 알고리즘
- 상위 조합에서부터 차례로 스캔 -> 특정 조합 자주 발생 X -> 하위 조합도 모두 배제
    ![](https://assets.insightmediagroup.io/media/wp-content/uploads/2020/10/1_22UGOKJJ6C8B3Zhk_IAiQ.png)

- 연산량 효과적으로 감소시킬 수 있음
- 최소 지지도를 만족 못하는 조합은 모두 제거하면서 점점 집합의 크기를 키워가고 더 이상 조합을 생성할 수 없을 때까지 이 과정 반복
- 마지막으로 남은 집합들로 연관 규칙 생성
- 장점 : 원리가 간단, 이해분석이 비교적 용이
- 단점 : 여전히 데이터 크기가 커질수록 속도가 느려짐

### 4-2. FP Growth 알고리즘
- Apriori의 연산 속도 묵제를 개선한 알고리즘
    ![](https://encrypted-tbn0.gstatic.com/images?q=tbn:ANd9GcTKyi0XxeM-Y4U1ZImvJkTgvf4WjZgzwN5Kqc_pfgLuKw&s=10)
- 과정
    1. 전체 거래에서 각 상품이 몇 번 등장했는지 세고 최소 지지도 미달인 상품은 배제
    2. 빈도수 기준으로 내림차순 정렬
    3. FP-Tree 만들기
    4. FP-Tree에서 특정 항목을 기준으로 서브 tree를 만들어서 빈발 항목 집합을 추출
    5. 추출된 빈발 항목 집합을 바탕으로 연관규칙 생성
    6. 지지도/신뢰도/향상도 지표를 활용해 가장 유용한 연관규칙 도출
- 장점 : 딱 두번만 데이터 스캔하므로 대용량 데이터 처리시 훨씬 빠름
- 단점 : 설계가 어려움, 메모리 많이 소모
