# 통계

## 1. Intro

### 통계란?:  자료를 수집/분석/표현/해석하는 기술이자 과학.

### 통계학의 구분?
- 기술통계 
    - 모집단에서 추출한 샘플 자체에 대해 관심
    - 샘플 정보 탐구 주목적
    - EDA
- 추론통계 
    - 샘플로부터 모집단을 추론하는 것에 대해 관심
    - 회귀, 분류

### 자료의 척도?
- 질적 자료(연산X)
    - 명목척도 : Label, 이름으로 속성 식별
    - 서열척도 : 명목 + 순서 존재
- 양적 자료
    - 구간척도 : 순서 존재 & 간격 일정 -> 덧셈 or 평균 가능
        - 곱셈, 나눗셈 의미 X
        - ex. 온도
    - 비율척도 : 절대영점 & 임의의 측정단위
        - 0값 -> 무의 상태 의미
        - ex. 무게

## 2. 기술통계

### 3가지 주요한 도구 집합
- 중심화 경향
    - 평균 -> 이상치에 민감
    - 중앙값 -> 이상치에 덜 민감
    - 최빈값

- 자료의 산포
    - 범위 -> 이상치에 민감
    - 분산
    - 표준편차 : $\sqrt{\text{분산}}$ -> 기존 데이터 단위로 맞춰주기 위함
    - 사분위수 -> 이상치 탐지에 유용
    - 변동계수 -> $변동계수=(표준편차/평균)*100$ -> 상대적 변동성 파악에 유리

## 3. 확률과 분포

### 확률분포
- 확률변수 : 특성 사건이 나타날 수 있는 경우의 수를 특정값에 대응시키는 것
- 확률분포 : 확률변수가 특정한 값을 가질 확률 나타냄
    - 이산확률분포 : 이항분포, 포아송분포,베르누이분포
    - 연속확률분포 : 정규분포, 카이제곱분포, T분포, F분포

### 대표적인 분포 모형
- 정규분포
    - 종모양, 연속확률분포
    - 중심극한정리 -> 표본의 크기가 커질수록 표본평균의 분포 정규분포와 가까워짐
    - z분포 : 평균이 0이고 표준편차가 1인 정규분포

- T분포
    - 종모양, 연속확률분포
    - 자유도가 커질수록(더 많은 독립적인 정보가 있을수록) z분포에 수렴
    - 자유도가 지나치게 커지면 과적합 문제 발생 가능

- 카이제곱분포
    - 0에서 시작오른쪽으로 긴 꼬리 비대칭 모양, 연속확률분포
    - 자유도가 커질수록 중심이 오른쪽으로 이동, 정규분포에 가까워짐
    - 독립적인 표준정규 분포 따르는 변수들의 제곱합
    - 독립성 검정, 적합도 검정에 사용

- F분포
    - 카이제곱분포 모양과 유사
    - 두 독립적인 카이 제곱 분포의 비율
    - 두 집단 간 분산 차이 비교에 사용

## 4. 추론 통계 - 추정

### 추정 & 검정
- 추정 : 표본 -> 모수 예측
    - 점추정 : 모수를 하나의 수치로 추정
    - 구간추정 : 모수를 구간으로 추정
- 검정 : 모수에 대한 예상, 가설 등이 올바른지 확인하는 과정

### 추정량 & 추정치
- 추정량 : 모수를 추정하는데 사용되는 표본통계량
- 추정치 : 표본자료를 이용하여 계산한 통계량

## 5. 추론 통계 - 검정

### 가설 검정의 절차
1) 가설 설정 : 귀무가설 H0과 대립가설 H1을 설정
    - 귀무가설 : 기각하고자 하는 명제
    - 대립가설 : 주장하고자 하는 명제
2) 유의 수준 설정 : 오차범위 설정
    - 1종 오류 : 귀무가설이 실제로 참이었는데 기각할 확률
    - 2종 오류 : 귀무가설이 실제로 거짓인데 기각하지 못할 확률
    - 가설 검정에서는 1종 오류가 더 worse
    - 유의 수준 = 1종 오류의 확률
3) 검정통계량 산출 : 가설 검정 결과 판단 시 사용되는 값
    - Z값
    - T값
    - F값
4) 가설 기각/채택 판단
    - 유의확률(p-value) : 검정통계량에서 유도됨
    - 기각 : $유의 확률 < 유의 수준$
    - 채택 : $유의 확률 > 유의 수준$

### 자료형에 따른 검정 방법
- [수치형 - 수치형] -> (피어슨)상관분석
    - 공분산, 상관계수 구함
    - p-value랑 비교
- [범주형 - 수치형] -> F-test(ANOVA), T-test
    - T-test : 범주가 2개로 나뉠 때
        - T값 : 두 평균 간 차이 / 표준 오차
        - 2를 기준으로 크면 차이 O, 작으면 차이 X
    - F-test(ANOVA) : 범주가 3개 이상일 때
        - Analysis Of Variance
        - F값: 집단 간 분산 / 집단 내 분산
        - 마찬가지로 2~3이 기준
- [범주형 - 범주형] -> Chi2-test
    - Chi2-test
        - 카이제곱통계량 구하기
        - 자유도*2가 기준
        - 자유도는 (x변수의 자유도) * (y변수의 자유도)
- [수치형 - 범주형] -> 분류 세션
