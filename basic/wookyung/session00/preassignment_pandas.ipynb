# Pandas
Pandas는 모듈들과 함수들이 모여있는 라이브러리로 Numpy라이브러리에 있는 기능은 물론 구조화된 데이터 처리를 할 때 유용하게 사용할 수 있는 라이브러리입니다.
<br>
Pandas 라이브러리는 import pandas as pd로 불러옵니다.
<br>
Pandas에는 크게 두가지 데이터 구조가 있는데 그중 데이터 분석하면서 주로 많이 쓰게 될 구조는 DataFrame구조입니다. 
<br>
DataFrame구조는 Series구조와 다르게 인덱스를 제외하고도 열이 여러개 존재하는 구조이고 열별로 서로 다른 이름과 데이터 타입을 가질 수 있습니다. 
<br>
DataFrame구조에서 데이터를 접근할 때는 크게 head/tail 방식, loc 방식, 그리고 iloc방식을 사용할 수 있습니다.
<br>
head/tail은 각각 위에서부터/밑에서부터 데이터를 행별로 출력하는 방식입니다.
<br>
loc 방식은 행명과 열명으로 데이터에 접근하는 방식으로 head/tail 접근 방식과 다르게 슬라이싱과 개별적 접근이 가능합니다.
<br>
iloc 방식은 loc방식과 거의 유사하나 행명/열명이 아닌 행번호/열번호로 접근하는 방식입니다.
<br>
각 데이터에 접근해서 수정을 할 수 있고 삭제할 때는 drop함수나 del함수를 사용하면 됩니다. 
<br>
집계함수에는 mean, median, std, sum, max, min등이 있고 각 열별로 여러 행에 걸쳐 값을 얻을 수 있습니다.
