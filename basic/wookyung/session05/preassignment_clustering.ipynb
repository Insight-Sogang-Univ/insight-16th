# 군집화

## 1. 군집화
### 1-1. 비지도학습
#### [비지도학습 특징]
- 정답 X
- 데이터 내재된 구조 파악
- 목표가 명확 X

#### [비지도학습이 적합한 케이스]
- 패턴이 아직 알려지지 않은 경우

- 대표적인 활용 예시 ->  고객 세그먼트 분석, 이상 거래 탐지, 자연어 처리 등

    - 분리도 최대화 -> 다른 군집에 속하는 데이터는 최대한 다르게

## 2. 군집화를 위한 데이터 준비
### 2-1. 군집화 주요 고려사항
- 변수 유형 이해 -> 변수 유형에 따라 적합한 군집화 알고리즘 다름.
- 거리/유사도 정의와 측정 -> 변수 유형에 따라 적합한 거리 측정 방식이 다름.
- 차원 축소 -> 변수가 많아질수록 효율성과 성능이 감소 -> 유사한 변수 통합, 불필요 변수 제거

### 2-2. 변수 유형
- 연속형 변수 
    - 특징: 수치형 변수, 연속적
    - 거리 측정 방식: 유클리디안 거리, 맨하탄 거리 등 사용
    - 적합 알고리즘: K-means, 계층적 군집화
    - 주의사항: 스케일링 꼭 해줘야 함
- 명목형 변수
    - 특징: 범주형 변수
    - 거리 측정 방식: 해밍 거리, 자카드 거리
    - 적합 알고리즘: K-modes, 계층적 군집화(특별한 거리 함수)
    - 처리 방법: 원-핫 인코딩, 더미 변수 변환
- 혼합형 변수
    - 특징: 연속형+명목형
    - 적합 알고리즘: K-prototypes, Gower distance기반 계층적 군집화
    - 처리 방법: 각 변수 유형에 맞는 거리 함수 조합해서 사용

### 2-4. 차원 축소
- 차원의 저주 : 피처 공간이 커질수록 알고리즘이 데이터를 효과적/효율적으로 훈련할 수 없는 현상
    - 차원이 커질수록 멀리 떨어지게 되고 이게 커지면 분별력 상실
    - 피처는 많을수록 좋은 게 아님 -> 핵심 피처에 집중하는 게 군집을 더 선명하게
- 차원 축소 
    - 변수의 개수를 줄이는 것
    - 핵심 정보는 가능한 지켜내면서
    - PCA(Principal Component Analysis)
        - 상관관계가 높은 피처들 결합해 상관관계가 없는 더 적은 수의 피처로 데이터 표현
        - 최대 분산 방향을 찾아 상관관계 감소시킴
        - 주의사항 
            - 표준화 필수
            - 결측값 없게 처리해줘야함
            - 해석의 어려움
            - 정보 손실


## 3. 군집화 알고리즘
### 3-1. 계층적 군집화
- 가장 가까운 것부터 둘씩 차례로 합쳐 나가면서 그 과정을 나무 모양으로 기록하는 군집화 방법
- 기록 = 덴드로그램
- 계층적 군집화의 특징
    - 군집의 개수 사전 설정X
    - 트리 구조 구성방식(상향식/하향식)에 따라 응집형/분리형으로 나뉨
        - 상향식(응집형) - 작은 것부터 큰 것으로
        - 하향식(분리형) - 큰 것에서 작은 것으로
- 6가지 연결방법
    - 연결방법? -> 두 군집 간의 거리 계산하는 방식
    - Single Linkage(단일 연결법) - 가장 가까운 두 점 거리 = 군집 간 거리 정의
    - Complete Linkage(완전 연결법) - 가장 먼 두 점 거리 = 군집 간 거리 정의
    - Average Linkage(평균 연결법) - 모든 점 쌍 사이 거리의 평균 = 군집 간 거리
    - Centroid Method(중심 연결법) - 각 군집의 중심점 간의 거리 = 군집 간 거리
    - Median(중앙 연결법) - 두 군집 중앙값 사이 거리 = = 군집 간 거리
    - Ward's Procedure - 군집 내 제곱합이 가장 적게 증가하도록 
- 장점
    - k를 미리 정하지 않아도 됨
    - 데이터 구조 전체가 눈에 보임
    - 어떤 유형의 데이터에도 적용 가능
    - 결과가 항상 같음
- 단점
    - 느림
    - 되돌릴 수 없음
    - 이상치에 민감
    - 사람이 결국 자르는 높이 결정해야 함


### 3-2. k-means
- 데이터를 정해진 개수의 그룹으로 나누고 각 그룹의 중심점 과의 거리가 가장 가까운 데이터끼리 묶는 알고리즘
- 가장 일반적으로 사용
- 중심점 하나가 군집을 대표
- 주요 단계
    1. 랜덤하게 k개의 중심점을 초기 클러스터 중심으로 선택
    2. 각 표본들을 가장 가까운 중심점에 할당
    3. 각 클러스터에 할당된 표본드르이 평균 계산해서 중심점 이동
    4. 클러스터 할당이 변하지 않거나 최대 반복 횟수에 도달할 때까지 2,3 반복
- 장점
    - 직관적이고 구현이 쉬움
    - 대용량 데이터에도 적용 가능
- 단점
    - 초기 centroid에 민감
    - 군집 수 결정이 어려움
    - 아웃라이어에 민감
    - 기하학적이 모양의 군집은 파악하기 어려움
### 3-3. DBSCAN
- 밀도가 높은 지역의 데이터를 하나의 군집으로 묶고, 밀도 기준을 만족하지 못하는 점은 군집에 포함시키지 않는 군집화 알고리즘
- 장점
    - 클러스터 모양을 원형으로 가정 X
    - 이상치에 민감하지 않음
    - 모든 샘플을 클러스터에 할당 X
- 단점
    - eps와 min_samples 설정에 많은 영향을 받음
    - 연산량이 많아 속도가 느림
    - 밀도가 서로 다른 군집들은 잘 못 잡음

### 3-4. Gaussian Mixture Model(GMM)
- 데이터가 여러 다른모양의 가우시안 분포로 구성되었다고 가정하고 각 분포를 클러스터로 인식하는 군집화 방법
- 각 군집을 확률 분포로 간주, 전체를 여러 확률 분포의 혼합으로 모델링
- GWM 진행 과정
    1. 전체 데이터셋 분포 확인
    2. 전체 데이터셋은 서로 다른 정규분포 형태의 확률 분포 곡선으로 구성되어 있다고 가정.
    3. 여러 개 정규분포 곡선 추출, 개별 데이터가 어떤 분포에 속하는지 결정
### 3-5. 어떤 알고리즘을 고를까
- 상황 -> 알고리즘
    - 몇 개로 나눠야 할지 감이 안온다 -> 계층적 군집화
    - 군집이 대체로 동글동글하고 데이터가 크다 -> k-means
    - 군집 모양이 길쭉하거나 서로 겹친다 -> GMM
    - 이상치를 골라내는게 목적이다 -> DBSCAN
    - 군집 모양이 초승달, 소용돌이처럼 이상하다 -> DBSCAN
    - 군집마다 밀도가 제각각 -> HDBSCAN
    - 경계에 걸친 데이터의 애매함까지 알고 싶다 -> GMM
- 이럴 땐 쓰지 말기
    - k-means
        - 이상치가 많은 데이터
        - 군집 크기가 심하게 불균형한 데이터
    - 계층적 군집화 
        - 데이터가 수만건 이상 -> 계산량 감당 x
    - DBSCAN
        - 피처가 수십 개 이상인 경우
    - GMM
        - 데이터가 정규분포와 거리가 멀 때
    - 전부
        - 스케일링 안 한 데이터

## 4. 군집화 평가 방법
### 4-1. 외부 평가와 내부 평가
- 외부 평가
    - 데이터에 정답 레이블이 있는 경우 사용
    - 군집화 결과와 정답 레이블을 비교
- 내부 평가
    - 정답 레이블이 없는 경우에 사용
    - 군집 내부 응집도와 군집 간 분리도를 기반으로 품질 정량화

### 4-2. 실루엣 계수
- 클러스터 내 샘플들이 얼마나 조밀하게 모여 있는지를 측정하는 도구
    ![](https://patentimages.storage.googleapis.com/67/25/ef/073fa98e6b0053/pat00001.png)
- 1에 가까움 -> 분리도 > 응집도 -> 군집화 good
- 0에 가까움 -> 분리도 = 응집도 -> 분리가 명확 X
- -1에 가까움 -> 분리도 < 응집도 -> 군집화 잘못됨
- 0.5이상이면 쓸만함, 0.7이상이면 잘됨
### 4-3. Dunn Index
- 클러스터 간 최소 거리(분리도)와 클러스터 내 최대 거리(응집도)의 비율로 클러스터링 품질 평가
- 값이 양수이고 클수록 좋은 군집화
- 실루엣 계수와 함께 종합적으로 판단하면 좋음
- 군집 수가 커질 수록 계산 비용 커짐
- 이상치에 큰 영향받음
