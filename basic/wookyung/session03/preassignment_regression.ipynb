# 회귀

## 1. 머신러닝과 모델링
### 1) 머신러닝
- 모델 학습 -> 패턴 발견 + 데이터 예측
- 머신러닝 종류
    - 지도 학습 : 문제와 정답 모두 제공
    - 비지도 학습 : 답을 가르쳐 추지 않음
    - 강화 학습 : 보상&벌을 통해 보상은 최대화, 벌은 최소화하는 방향으로 학습

### 2) 모델링
- 모델: 학습의 결과물
- 모델링: 모델을 만들어나가는 전체적인 과정
    - 설계 -> 훈련 -> 검증

### 3) 지도 학습
- labeled 데이터 활용
- train data와 test data로 나누어 사용
- 독립변수를 통해 추정한 예측값이 종속변수와 같아지도록 모델을 훈련

### 4) 회귀 vs. 분류
- 회귀 : 예측하고자 하는 값이 연속형
- 분류 : 예측하고자 하는 값이 범주형

## 2. 회귀
### 1) 회귀란?
- 데이터 패턴 찾아 미래 값 예측하는 것
- 종속변수가 연속형


### 2) 선형회귀
- 데이터를 가장 잘 설명하는 회귀선을 찾아내는 과정
- $ y= wx+b $
    - w: 가중치(회귀계수) -> 독립변수 x가 종속변수 y에 미치는 영향의 크기
    - b: 편향 -> 기본값
- 단순선형회귀분석: 독립변수가 1개
- 다중선형회귀분석: 독립변수가 2개 이상

### 3) 최적의 W와 b를 찾는 두 가지 방법
- 최소제곱법
    - OLS(Ordinary Least Squre)이라고도 함
    - 잔차제곱의 합이 최소가 되게 하는 회귀선을 구하는 것
- 경사하강법(-> 딥러닝 심화 세션)

### 4)다중선형회귀
- 독립변수가 2개 이상
- 회귀선이 평면
- 기본 가정 6가지
    - 선형성: 종속변수와 설명변수 간의 관계가 선형적으로 표현되어야 함.
    - 독립성 : 각각의 설명변수는 서로 선형독립적이어야 함.
    - 오차항 평균 = 0
    - 등분산성 : 오차항의 분산이 일정해야 함.
    - 오차항은 자기상관X : 오차항의 공분산이 항상 0이어야 함.
    - 정규성 : 오차항이 정규분포를 따라야 함. 

### 5) 다중공선성
- 독립 변수들 간의 강한 상관관계가 있는 경우 -> 독립성 가정 위배
- 상관계수로 확인 가능
- VIF지수로 확인가능 
    - VIF>5 : 의심
    - VIF>10 : 심각한 다중공선성
- 제거, 변환, 규제선형모델 활용, PCA를 통해 해결 가능

### 6) 규제선형모델
- 모델이 overfitting되지 않도록 규제를 가하는 모델
- 종류
    - Ridge 회귀 : 회귀계수의 제곱에 대해 페널티 부여
    - Lasso 회귀 : 회귀계수의 절댓갑에 대해 페널티 부여
    - Elastic Net 회귀 : Ridge + Lasso

## 3. 모델 평가방법

### 1) 시각화
- 그래프로 확인
- 객관적인 비교는 어려움

### 2) 통계지표
- 모델 유의성 검정
    - F-statistic : 값이 클수록 모델이 통계적으로 유의미
    - T-statistic : 독립변수가 유의미한지 검정
- 모델 성능 평가
    - 결정계수(R-squared) : SSR/SST, $ 0 <= R^2 <= 1 $
    - 1에 가까울수록 모델이 데이터를 잘 설명
    - Adjusted 결정계수: 변수의 개수 영향을 감소시킨 것

### 3) 성능평가지표
- MSE: 오차 제곱의 평균
- MAE: 오차 절대값의 평균 -> MSE보다 이상치에 덜 민감


### 4) 변수 유의성 평가
- T-검정
    - 회귀계수가 유의미한지 검정

## 4. 비선형회귀

### 1) 다항식 회귀모델
- $ Y = a + bX + cX^2 + dX^3 + ... $
- 모델을 확장시키고 그 상태에서 선형 회귀 적용

### 2) 지수 회귀모델
- $Y = ae^{bX}$
- log취하고 선형 회귀 적용
- Y값 예측시에는 지수함수 적용

### 3) 로그 회귀모델
- $Y=a + blog(X)$

### 4) 스플라인 회귀
- 데이터를 구간별로 나누고 구간별로 다른 선형 또는 비선형 함수를 적용하는 것
- 구간 경계에서의 연속성 유지가 중요
