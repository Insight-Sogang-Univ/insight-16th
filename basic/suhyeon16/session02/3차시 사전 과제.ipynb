# 1. 통계학

- **기술통계**: 수치와 시각화를 통해 샘플을 요약하여 표현하는 방법 (ex. 시험점수 분포)
- **추론통계**: 추정과 검정을 통해 샘플로부터 모집단을 추론하는 방법 (ex. 시청률 조사)


# 2. 자료의 척도

- **자료**: 설명과 해석을 위하여 수집되고 분석되며, 요약되는 사실과 수치를 의미함.

- **질적 자료** :
    1) **명목척도**:
        레이블이나 이름으로 요소의 속성을 식별하는 척도

    2) **서열척도**:
        명목형의 분류 속성과 더불어, 항목간의 순서 또는 순위를 나타내는 특성을 지닌 척도

- **양적 자료**:
    1) **구간(등간)척도**: 
        값들 사이에 순서가 있고, 그 간격이 일정하여 덧셈/평균과 같은 연상이 가능한 척도 (단. 곱셈과 나눗셈은 의미 X)

    2) **비율척도**:
        절대영점과 임의의 측정단위를 구성하고, 동간성이 유지되는 척도
        (0 = 그 변수에 대해 아무것도 존재하지 않는 상태를 나타냄, 가격이 0인 경우 공짜의 의미)

> => 덧셈이 불가능하다 = 명목척도  
> => 0 값의 의미를 가지면 비율척도, 갖지 않으면 비율척도


# 3. 기술척도

**기술통계**: 데이터를 있는 그대로 설명하고 이해하는 데 중점을 둔 통계 방법

## 1) 중심화 경향

- **평균**: 데이터 값의 총합을 데이터 개수로 나눈 값 (총합을 사용하기 때문에 이상치에 민감할 수 있음)
- **절사평균**: 이상치에 대한 민감성을 보완하기 위함, 양끝 데이터를 일정 비율로 잘라내어 평균 계산함
- **중앙값**: 데이터를 크기순으로 정렬시, 한가운데에 위치하는 값 (평균에 비해 이상치에 덜 민감, 데이터의 중심 경향성을 나타내는 대안적인 지표) => `series.median()`
- **최빈값**: 데이터에서 가장 자주 나타나는 값 (주로 이산형 데이터의 분포를 이해하는데 도움 줌)

## 2) 자료의 산포

- **범위**: 데이터의 최댓값과 최솟값의 차이를 나타내는 지표, 데이터의 전체 변동 범위를 나타냄 (양쪽 극단값을 사용하므로 이상치에 매우 민감) => `series.max() - series.min()`
- **분산**: 데이터 값들이 평균으로부터 얼마나 멀리 퍼져있는지를 나타냄 => `series.var()`
- **표준편차**: 분산에 제곱근을 취한 값 => `series.std()`
- **사분위수**: 데이터의 표본을 네개의 동일한 크기로 나눈 값 (Q1 = 하위 25%에 위치한 값 (`series.quantile(0.25)`), Q2 = 하위 50%에 위치한 값 (`series.quantile(0.5)`), Q3 = 하위 75%에 위치한 값 `series.quantile(0.75)`)
- **사분위간 범위(IQR)**: Q3에서 Q1 값을 제외한 범위, 즉 데이터의 중앙 50%가 퍼져 있는 범위, 이상치 탐지시 유용.
- **변동계수**: 데이터의 절대적인 변동성이 아닌 상대적인 변동성을 나타내는 통계적 지표 (변동계수 = (표준편차/평균) * 100))

## 3) 자료의 정리 및 시각화

히스토그램, 박스플롯, 파이차트 등


# 4. 확률과 분포

- **확률변수**: random한 세계에서 어떤 특정 사건이 발생할 때 나타날 수 있는 모든 경우의 수를 특정 값으로 대응시키는 함수
- **확률분포**: 확률변수가 특정한 값을 가질 확률을 나타내는 함수
    1) **이산확률분포**: 확률 변수가 이산적인 값 가질 때
    2) **연속확률분포**: 확률 변수가 연속적인 값을 가질 때

- **대표적인 분포 모형**:
    1) **정규분포(연속확률분포)** : 종 모양, 평균을 기준으로 대칭, 표준편차가 클 수록 곡선이 평평해짐
        - **큰 수의 법칙**: 표본의 크기 클 수록, 표본 평균이 모평균에 근사하는 현상, 즉 데이터가 많을 수록 통계적 정확도 상승
        - **중심극한정리**: 동일한 확률분포를 가진 독립확룔 변수
        - **Z분포(표준정규분포)**: 평균이 0이고 표준편차가 1이 되도록 맞춘 표준적인 정규분포
        - **T분포**: 표보느로 모평균을 추정할 때 사용되는 분포 (표본 크기가 작을 때 정규분포보다 더 정확한 결과 제공)
        - **카이제곱분포**: 독립적인 표준 정규 분포를 따르는 변수들의 제곱 합
        - **F분포**: 두 독립적인 카이 제곱 분포의 비율, 두 집단 간의 분산 차이를 비교하는 데 사용됨


# 5. 추론 통계

- **추정**: 모집단을 대표할 수 있는 일부를 표본으로 추출하여 분석한 통계량(statistic)을 이용하여 모집단의 모수(=parameter)를 예측하는 과정
- **검정**: 표본을 추출하여 얻은 표본통계량으로 모집단 모수에 대한 새로운 예상/가설 등이 옳다고 판단할 수 있는지를 평가하는 과정
- **추정량**: 표본정보에 의존하는 확률변수, 모수를 추정하는 데 사용되는 표본통계량을 의미함
- **추정치**: 표본이 추출된 후 표본자료를 이용하여 계산한 통계량의 특정
함 값

1) **점추정**: 어떤 모수를 표본에 기초하여 계산된 하나의 값으로 추정하는 것

2) **구간추정**: 모수의 참 값이 포함되리라고 기대하는 추정치를 일정한 범위와 신뢰도를 가지고 나타내는 방법


# 6. 추론 통계 검정

- **가설 검정의 절차**: 가설 설정 -> 유의 수준 설정 ->검정통계량 산출 -> 가설 기각/ 채택

## 1) 가설 설정

귀무가설 H0(주로 기각하고자 하는 명제)와 대립가설 H1(내가 주장하고자 하는 명제)를 설정함

## 2) 유의 수준 설정

연구작 결과를 판단할 때 허용할 수 있는 오류의 최대치로, 주로 5%로 설정

- **1종 오류**: 귀무가설이 실제로 참이었는데 기각할 확률
- **2종 오류**: 귀무가설이 실제로 거짓인데 기각하지 못할 확률

## 3) 검정통계량 산출

가설 검정의 결과를 판단하는데 사용되는 값

- **Z값**: 표준정규분포를 사용하여 얻은 값
- **T값**: T분포를 사용하여 얻은 값, 작은 표본 크기에 대한 가설 검정에 주로 사용됨
- **F값**: F분포를 사용하여 얻은 값, 주로 두 표본 분산 사이의 차이를 비교하는 데 사용됨

## 4) 가설 기각/ 채택 판단

가설 검정에 쓰일 수 있는 또 다른 척도

- 유의 확률 < 유의 수준일 시 귀무가설 기각
- 유의확률 > 유의 수준일 시 귀무가설 기각 불가


# 7. 데이터의 자료형에 따른 검정 방법

1) **수치형 - 수치형 (피어슨 상관분석)**: 두 변수간 피어슨 상관계수가 얼마나 유의미한지를 검정하는 방법

2) **범주형 - 수치형 (F-test(ANOVA): 범주가 3개 이상, T-test: 범주가 2개)**

3) **범주형 -범주형 (Chi2-test)**: 두 개의 범주형 변수가 서로 관련이 있는지 확인하는 방법

4) **수치형 - 범주형**
