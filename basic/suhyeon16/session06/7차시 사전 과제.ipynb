# 연관 분석 

## 1. 연관 분석이란?

### 연관 분석의 정의

연관 분석은 상품이나 서비스를 구매하는 등의 거래나 사건 안에서 항목 간에 존재하는 일정한 연관 규칙을 발견하는 분석 방법*이다.

연관 분석은 흔히 장바구니 분석이라고도 불리지만, 엄밀히 말하면 장바구니 분석은 유통업에서 연관 분석을 활용하는 하나의 분야이다. 따라서 연관 분석은 상품 거래뿐만 아니라 질병 진단, 고객 행동 분석, 콘텐츠 추천 등 다양한 분야에 활용할 수 있다.

연관 분석은 별도의 정답값이 존재하지 않는 데이터에서 항목 간의 패턴을 찾아내므로 비지도 학습에 해당한다.

대표적인 연관 분석 알고리즘으로는 다음과 같은 것들이 있다.

- **Apriori**
- **FP-Growth**

---

### 1-2. 연관규칙 분석

연관규칙은 일반적으로 If-Then 형식으로 나타내며 다음과 같이 표현한다.

A → B

- **조건절**: `만일 ~라면`에 해당하는 부분으로 A에 해당
- **결과절**: 조건에 따라 함께 나타나는 결과로 B에 해당
- **아이템 집합**: 조건절 또는 결과절을 구성하는 아이템들의 집합

Itemset에는 하나뿐만 아니라 여러 개의 아이템이 포함될 수 있다. 단, 조건절과 결과절을 구성하는 Itemset은 서로 겹치지 않아야 한다.

#### 연관관계와 인과관계

연관 분석에서 주의해야 할 점은 연관관계가 인과관계를 의미하지 않는다는 것이다.

예를 들어 `A → B`의 신뢰도가 0.8이라면, 이는 A가 나타난 경우 중 80%에서 B도 함께 나타났다는 의미이다. 하지만 A가 B를 발생시킨 원인이라는 의미는 아니다.

연관 분석은 주로 관찰 데이터를 기반으로 하기 때문에 두 사건에 동시에 영향을 주는 다른 요인이 존재할 수 있다. 따라서 연관 분석은 항목들이 함께 나타나는 패턴을 발견하는 분석이지 원인과 결과를 밝히는 분석은 아니다.

# 2. 주요 개념

## 2-1. 적절한 연관규칙의 조건

유의미한 연관규칙을 찾기 위해서는 다음과 같은 조건을 고려한다.

1. 두 항목 A와 B가 함께 발생하는 경우가 일정 수준 이상이어야 한다.  
   → 일정 수준 이상의 지지도

2. A가 발생한 경우 중 B도 함께 발생하는 경우가 일정 수준 이상이어야 한다.  
   → 일정 수준 이상의 신뢰도

하지만 신뢰도가 높다고 해서 반드시 유용한 규칙이라고 할 수는 없다. 결과 항목 B가 원래부터 매우 자주 발생하는 항목일 수도 있기 때문이다.

따라서 A와 B 사이에 실제로 어느 정도의 연관성이 존재하는지를 판단하기 위해 향상도를 함께 확인한다.

---

## 2-2. 연관 분석의 평가 측도

### ① 지지도 

전체 거래 중에서 **A와 B가 동시에 포함된 거래의 비율**을 의미한다.

Support(A \rightarrow B)=P(A \cap B)


Support=
\frac{A와\ B가\ 함께\ 발생한\ 거래수}{전체\ 거래수}


지지도는 해당 규칙이 전체 데이터에서 얼마나 자주 발생하는지를 나타낸다.

지지도가 지나치게 낮다면 A와 B가 함께 나타나는 사례 자체가 적다는 의미이므로 실제로 활용할 가치가 낮을 수 있다.

---

### ② 신뢰도

A가 포함된 거래 중에서 A와 B가 동시에 포함된 거래의 비율을 의미한다.


Confidence(A \rightarrow B)=P(B|A)


또는,


Confidence(A \rightarrow B)
=
\frac{P(A \cap B)}{P(A)}


따라서 신뢰도는 A가 발생했을 때 B도 함께 발생할 가능성이 얼마나 높은지를 나타낸다.

- 지지도: 전체 거래를 기준으로 계산
- 신뢰도: A가 발생한 거래를 기준으로 계산

---

### ③ 향상도

신뢰도가 높더라도 B 자체가 원래 매우 자주 발생한다면 A와 B 사이에 특별한 관계가 있다고 보기 어렵다.

이를 보완하기 위해 사용하는 지표가 향상도이다.

향상도는 A가 주어졌을 때 B가 발생할 확률을 A와 관계없이 B가 발생할 확률과 비교한 값이다.


Lift(A \rightarrow B)
=
\frac{P(B|A)}{P(B)}


또는,


Lift(A \rightarrow B)
=
\frac{P(A \cap B)}
{P(A)P(B)}


A와 B가 독립이라면,


P(A \cap B)=P(A)P(B)


이므로 향상도는 1이 된다.


따라서 연관규칙의 유용성을 평가할 때는 **지지도, 신뢰도, 향상도를 종합적으로 고려해야 한다.**

# 3. 연관 분석의 활용

연관 분석을 통해 발견한 항목 간 관계는 실제 비즈니스 전략 및 의사결정에 활용할 수 있다.

대표적인 활용 분야는 다음과 같다.

- 함께 구매되는 상품을 파악하여 상품 배치 및 진열 전략 수립
- 연관성이 높은 상품을 활용한 교차 판매
- 고객의 구매 패턴을 활용한 상품 추천
- 특정 상황에서 수요가 증가하는 상품을 분석하여 재고 및 물류 관리
- 고객 행동 및 서비스 이용 패턴 분석
- 의료 분야에서 증상이나 질병 간의 연관 패턴 분석

연관 분석의 중요한 특징은 반드시 원인을 설명하지 못하더라도, 반복적으로 함께 발생하는 패턴을 발견하여 실제 의사결정에 활용할 수 있다는 점이다.

# 4. 연관 분석 알고리즘

## 4-1. Apriori 알고리즘

### Apriori란?

**Apriori 알고리즘**은 일정한 최소 지지도 이상으로 자주 등장하는 빈발 항목 집합을 찾아 연관규칙을 생성하는 대표적인 알고리즘이다.

상품의 수가 증가하면 가능한 Itemset의 조합도 기하급수적으로 증가하기 때문에 모든 조합을 검사하는 것은 비효율적이다.

Apriori는 이러한 문제를 해결하기 위해 다음과 같은 원리를 사용한다.

> 어떤 Itemset이 빈발하지 않다면, 해당 Itemset을 포함하는 더 큰 Itemset 역시 빈발할 수 없다.

예를 들어 `{A, B}`가 최소 지지도를 만족하지 않는다면 `{A, B, C}`, `{A, B, D}` 등 `{A, B}`를 포함하는 더 큰 Itemset 역시 후보에서 제거할 수 있다.

이를 통해 불필요한 후보를 미리 제거하여 연산량을 줄인다.

---

### Apriori 알고리즘 과정

**1. 단일 항목 집합 생성**

모든 개별 Item의 지지도를 계산하고 최소 지지도 이상인 항목만 남긴다.

**2. 2개 항목 집합 생성**

1단계에서 남은 항목을 조합하여 2개 Item으로 구성된 Itemset을 만들고 지지도를 계산한다. 최소 지지도를 만족하지 않는 조합은 제거한다.

**3. 더 큰 항목 집합 생성**

이전 단계에서 남은 Itemset을 이용하여 3개 이상의 Item으로 구성된 조합을 생성하고 동일한 과정을 반복한다.

**4. 빈발 항목 집합 선정**

더 이상 새로운 조합을 생성할 수 없을 때까지 반복하여 최종 Frequent Itemset을 선정한다.

**5. 연관규칙 생성**

선정된 Frequent Itemset을 이용하여 연관규칙을 생성하고 Support, Confidence, Lift를 계산하여 유용한 규칙을 찾는다.

---

### Apriori 알고리즘의 장단점

**장점**

- 다양한 상품 간의 연관 패턴을 발견할 수 있다.
- 알고리즘의 원리가 비교적 단순하다.
- 결과를 이해하고 해석하기 쉽다.
- 불필요한 후보 Itemset을 제거하여 전체 조합을 탐색하는 것보다 효율적이다.

**단점**

- 데이터가 커질수록 연산량이 증가한다.
- 후보 Itemset을 반복적으로 생성해야 한다.
- 여러 단계에서 데이터셋을 다시 탐색하기 때문에 대규모 데이터에서는 속도가 느려질 수 있다.


## 4-1. FP-Growth 알고리즘

### FP-Growth란?

**FP-Growth(Frequent Pattern Growth)**는 Apriori의 연산 속도 문제를 개선하기 위해 개발된 알고리즘이다.

Apriori는 후보 Itemset을 단계별로 생성하고 데이터를 반복적으로 확인해야 한다. 반면 FP-Growth는 **FP-Tree(Frequent Pattern Tree)**라는 자료구조를 사용하여 거래 데이터를 압축하고 빈발 패턴을 탐색한다.

따라서 후보 Itemset을 반복적으로 생성하지 않고도 Frequent Itemset을 찾을 수 있다.

---

### FP-Growth 알고리즘 과정

**1. 항목의 빈도 계산**

전체 거래에서 각 Item이 등장하는 빈도를 계산하고 최소 지지도 이상인 항목만 남긴다.

**2. 빈도순 재정렬**

각 거래에 포함된 Item을 전체 데이터에서 등장 빈도가 높은 순서로 정렬한다.

**3. FP-Tree 생성**

정렬된 거래를 Tree 구조에 추가한다. 공통된 Item이 존재하면 기존 경로를 공유하여 데이터를 압축한다.

**4. 빈발 항목 집합 추출**

완성된 FP-Tree를 이용하여 최소 지지도를 만족하는 Frequent Itemset을 탐색한다.

**5. 연관규칙 생성**

추출된 Frequent Itemset을 이용하여 연관규칙을 생성하고 Support, Confidence, Lift 등의 지표를 계산한다.

---

### FP-Growth 알고리즘의 장단점

**장점**

- 후보 Itemset을 반복적으로 생성할 필요가 없다.
- FP-Tree를 통해 반복되는 거래 패턴을 압축할 수 있다.
- 데이터베이스 전체를 반복적으로 탐색하는 부담이 적다.
- 대규모 데이터에서는 일반적으로 Apriori보다 빠르다.

**단점**

- Apriori보다 알고리즘의 구조가 복잡하다.
- FP-Tree를 별도로 생성하고 관리해야 한다.
- 데이터의 특성에 따라 Tree가 충분히 압축되지 않으면 메모리 사용량이 증가할 수 있다.

---

### Apriori와 FP-Growth 비교

| 구분 | Apriori | FP-Growth |
|---|---|---|
| 기본 방식 | 후보 Itemset을 단계적으로 생성 | FP-Tree를 이용하여 데이터 압축 |
| 후보 Itemset 생성 | 필요 | 필요하지 않음 |
| 데이터 탐색 | 반복적으로 수행 | 상대적으로 적음 |
| 속도 | 대규모 데이터에서 느려질 수 있음 | 일반적으로 더 빠름 |
| 구조 | 비교적 단순 | 상대적으로 복잡 |




# 데이터 분석가를 위한 경영 기초

데이터 분석에서는 통계, 머신러닝, 프로그래밍 등의 기술뿐만 아니라 무엇을 분석해야 하는지 정의하고, 적절한 지표를 설정하며, 분석 결과를 실제 의사결정으로 연결하는 과정이 중요하다.

실무 데이터 분석의 과정은 크게 다음과 같이 정리할 수 있다.

1. **문제 정의**: 무엇을 해결해야 하는가?
2. **문제 구조화**: 문제를 어떻게 나눌 것인가?
3. **지표 설정**: 무엇을 측정할 것인가?
4. **분석**: 지표가 왜 변화했는가?
5. **효과 검증**: 분석 및 개선안의 효과를 어떻게 확인할 것인가?
6. **전달**: 분석 결과를 어떻게 의사결정으로 연결할 것인가?

---

# 1. 문제 정의

## 1-1. 현상과 문제

데이터 분석을 시작하기 위해서는 먼저 현상과 문제를 구분해야 한다.

> **문제 = 현재 상태와 바라는 상태 사이의 격차**

| 구분 | 의미 |
|---|---|
| **현상** | 현재 관찰되고 있는 사실 |
| **문제** | 현상을 바탕으로 해결해야 하는 구체적인 질문 |


따라서 단순히 데이터를 관찰하는 데서 끝나는 것이 아니라, 이를 의사결정자가 판단할 수 있는 구체적인 질문으로 변환하는 과정이 필요하다.

---

## 1-2. Problem Statement

분석을 시작하기 전에는 분석의 목적과 활용 방법을 명확하게 정의해야 한다. 이를 위해 Problem Statement를 작성할 수 있다.

| 항목 | 핵심 질문 |
|---|---|
| **Situation** | 현재까지 어떤 상황이었는가? |
| **Complication** | 기존 상황에서 무엇이 달라졌는가? |
| **Question** | 분석을 통해 어떤 질문에 답해야 하는가? |
| **의사결정** | 분석 결과를 이용해 누가 무엇을 결정하는가? |
| **판단 기준** | 어떤 결과가 나왔을 때 어떤 행동을 할 것인가? |

특히 분석을 시작하기 전에 **분석 결과에 따라 어떤 의사결정을 할 것인지**를 정하는 것이 중요하다.

---

## 1-3. SMART 원칙

목표를 구체적으로 정의하기 위해 **SMART 원칙**을 사용할 수 있다.

| 원칙 | 의미 |
|---|---|
| **Specific** | 목표가 구체적이어야 함 |
| **Measurable** | 수치로 측정할 수 있어야 함 |
| **Action-oriented** | 실제 행동으로 연결되어야 함 |
| **Relevant** | 조직의 목표와 관련되어야 함 |
| **Time-bound** | 명확한 기한이 존재해야 함 |

SMART 원칙을 이용하면 모호한 목표를 **측정 가능하고 실행 가능한 목표**로 구체화할 수 있다.

---

# 2. 문제 구조화

복잡한 문제는 한 번에 해결하기 어렵기 때문에 **해결 가능한 작은 문제들로 분해**해야 한다.

## 2-1. MECE

**MECE**는 문제를 구조화할 때 사용하는 대표적인 원칙이다.

- **Mutually Exclusive (상호 배타적)**  
  → 각 항목이 서로 겹치지 않아야 한다.

- **Collectively Exhaustive (전체 포괄적)**  
  → 필요한 항목이 빠짐없이 포함되어야 한다.

즉, 문제를 나눌 때 중복도 없고 누락도 없도록 분류하는 것

좋은 MECE 구조는 다음과 같은 조건을 갖는다.

1. **논리적 일관성**: 같은 단계에서는 하나의 기준으로 분류
2. **통찰력**: 문제 해결에 도움이 되는 기준으로 분류
3. **행동 가능성**: 분류된 결과가 실제 대응으로 연결될 수 있어야 함

---

## 2-2. 로직트리(Logic Tree)

MECE를 이용하여 문제를 분해한 뒤에는 **로직트리(Logic Tree)**를 활용해 문제의 원인과 분석 대상을 구체화할 수 있다.


```text
문제 정의
   ↓
이슈 트리 작성
   ↓
가설 수립
   ↓
분석 설계
   ↓
검증
   ↓
결론
```

---

## 2-3. 분해 등식

실무에서 자주 사용하는 문제 구조화 방법 중 하나는 **지표를 구성 요소로 분해하는 것**이다.

대표적으로 매출은 다음과 같이 분해할 수 있다.

$$
매출 = 고객수 \times 구매빈도 \times 객단가
$$

객단가는 다시 다음과 같이 나눌 수 있다.

$$
객단가 = 구매건당수량 \times 평균단가
$$

커머스에서는 다음과 같은 형태도 사용할 수 있다.

$$
매출 = 방문자수 \times 구매전환율 \times 객단가
$$

이처럼 하나의 결과 지표를 구성 요소로 분해하면 어떤 요소의 변화가 전체 결과를 변화시켰는지 파악하기 쉬워진다.

---

# 3. 지표

## 3-1. 지표의 정의

지표란 측정하고 싶은 대상을 숫자로 표현한 것으로, 조직이나 서비스의 상태와 목표 달성 정도를 판단하는 기준이다.

대표적인 지표 유형은 다음과 같다.

| 유형 | 구조 | 의미 |
|---|---|---|
| **효율성** | 산출 ÷ 투입 | 투입 대비 얼마나 많은 결과를 냈는가 |
| **효과성** | 달성 ÷ 목표 | 목표를 얼마나 달성했는가 |
| **수익성** | 이익 ÷ 자원 | 자원을 이용해 얼마나 많은 이익을 냈는가 |
| **성장성** | 증가분 ÷ 기존 규모 | 이전보다 얼마나 성장했는가 |
| **안정성** | 보유 자원 ÷ 부채 등 | 기업의 재무적 안정성을 측정 |

이외에도 매출액과 같은 **절대치 지표**, NPS와 같은 **지수형 지표**, 여러 지표를 결합한 **복합 지표**가 존재한다.

---

## 3-2. 좋은 지표를 선택하는 기준

### ① 선행 지표와 후행 지표

| 구분 | 의미 |
|---|---|
| **Leading Metric(선행 지표)** | 최종 결과보다 먼저 변화하며 사전에 개입할 수 있는 지표 |
| **Lagging Metric(후행 지표)** | 이미 발생한 결과를 보여주는 지표 |

매출이나 이익은 대표적인 후행 지표이다. 반면 방문자 수, 사용 빈도 등은 최종 결과보다 먼저 움직일 수 있는 선행 지표이다.

따라서 최종 결과만 관찰하기보다는 **결과에 영향을 주는 선행 지표를 함께 관리하는 것이 중요하다.**

---

### ② Input과 Output 지표

- **Output Metric**: 조직이 최종적으로 얻고 싶은 결과  
  → 매출, 가입자 수 등

- **Input Metric**: 조직이 직접 통제하거나 변화시킬 수 있는 행동  
  → 콘텐츠 발행 수, 응대 건수, 실험 횟수 등

Output은 최종 목표를 보여주지만 직접 통제하기 어려운 경우가 많다. 따라서 실제 업무에서는 **통제 가능한 Input 지표를 함께 설정하는 것**이 중요하다.

---

### ③ 가드레일 지표

특정 지표 하나만 극단적으로 개선하면 다른 중요한 지표가 악화될 수 있다.

이를 방지하기 위해 사용하는 것이 **가드레일 지표**이다.

> **Guardrail Metric = 목표 지표를 개선하는 과정에서 악화되어서는 안 되는 지표**

예를 들어 신규 가입자를 증가시키는 것이 목표라면 단순히 가입자 수만 확인하는 것이 아니라 **가입 후 잔존율**도 함께 확인해야 한다.

따라서 목표 지표와 함께 가드레일 지표를 설정하면 **한 지표의 개선이 전체 서비스에 부정적인 영향을 미치는 것을 방지할 수 있다.**

---

## 3-3. Proxy Metric

**Proxy Metric(대리 지표)**은 직접 측정하기 어려운 개념을 측정하기 위해 사용하는 대체 지표이다.

만족도, 활성화 정도, 충성도 등은 직접적으로 관찰하기 어려울 수 있다. 따라서 이러한 개념과 관련된 **관찰 가능한 행동을 대신 측정**한다.

Proxy Metric을 설정하는 과정은 다음과 같다.

1. 측정하려는 추상적인 개념을 정의한다.
2. 해당 개념과 관련된 사용자의 행동을 찾는다.
3. 데이터로 측정할 수 있는 행동을 선택한다.
4. 이를 구체적인 지표로 정의한다.

Proxy Metric은 완벽하게 실제 개념을 측정하는 것은 아니기 때문에 **대리 지표가 실제 목표를 적절하게 반영하는지 지속적으로 검토해야 한다.**

---

## 3-4. L1 / L2 지표

지표는 수준에 따라 **L1과 L2**로 구분할 수 있다.

| 구분 | 의미 | 관점 |
|---|---|---|
| **L1** | 조직의 전체적인 성과를 나타내는 상위 지표 | 전략 |
| **L2** | L1을 구성하는 세부 지표 | 전술 |

핵심은 다음과 같이 **L1이 L2의 조합으로 설명될 수 있도록 설계하는 것**이다.

$$
L1 = L2_1 \times L2_2 \times L2_3 \cdots
$$

이를 통해 L1 지표가 변화했을 때 **어떤 L2 지표가 변화의 원인인지 추적할 수 있으며**, 각 팀이 관리해야 할 지표도 명확하게 설정할 수 있다.

---

## 3-5. KPI

**KPI(Key Performance Indicator)**는 **핵심성과지표**로, 조직이나 팀이 목표를 얼마나 잘 달성하고 있는지를 판단하는 핵심 지표이다.

KPI의 주요 역할은 다음과 같다.

- 성공의 의미를 구체적으로 정의
- 조직의 우선순위를 명확하게 설정
- 현재 목표 달성 수준을 확인
- 결과에 따라 전략과 행동을 조정

즉, KPI는 조직이 목표를 향해 제대로 이동하고 있는지 판단하는 **객관적인 기준**의 역할을 한다.

---

## 3-6. KSF

**KSF(Key Success Factor)**는 핵심성공요인으로, 특정 목표를 달성하기 위해 반드시 성공시켜야 하는 중요한 요인을 의미한다.

KPI가 **성과를 측정하는 지표**라면, KSF는 **목표 달성을 위해 무엇에 집중해야 하는지를 결정하는 요인**이라고 볼 수 있다.

### KSF의 조건

| 조건 | 의미 |
|---|---|
| **소수성** | 핵심적인 몇 개의 요인으로 제한 |
| **결정성** | 해당 요인이 실패하면 목표 달성이 어려워야 함 |
| **통제 가능성** | 조직이 자원을 투입하여 변화시킬 수 있어야 함 |
| **상황 의존성** | 산업, 시장, 시기에 따라 달라질 수 있음 |

따라서 KSF는 모든 요소를 관리하는 것이 아니라 **목표 달성에 결정적인 요소를 선별하여 자원을 집중하는 것**에 목적이 있다.

---

# 4. 지표 분석 방법

## 4-1. 변동 분해

실무에서는 특정 지표가 이전보다 증가하거나 감소했을 때 **왜 변화했는지를 분석하는 것**이 중요하다.

이를 위해 전체 지표를 구성 요소로 분해한다.

예를 들어,

$$
매출 = 고객수 \times 구매빈도 \times 객단가
$$

라면 각 요소의 변화율을 비교하여 매출 변화의 원인을 확인할 수 있다.

즉, 단순히 **"매출이 감소했다"**에서 끝나는 것이 아니라,

- 고객 수가 감소했는가?
- 구매 빈도가 감소했는가?
- 객단가가 감소했는가?

와 같이 문제를 구체화할 수 있다.

---

### 단가 효과와 믹스 효과

평균값이 변화했을 때는 **단가 효과와 믹스 효과**를 구분해야 한다.

| 구분 | 의미 |
|---|---|
| **단가 효과** | 개별 상품이나 항목의 값 자체가 변화한 효과 |
| **믹스 효과** | 서로 다른 항목의 구성 비중이 변화한 효과 |

전체 평균 가격이나 이익률이 감소했다고 해서 반드시 개별 상품의 가격이나 수익성이 감소한 것은 아니다.

가격이 낮거나 마진이 낮은 상품의 판매 비중이 증가하면 개별 상품의 값이 동일하더라도 **전체 평균이 감소할 수 있다.**

따라서 평균 지표가 변화했다면 **개별 항목의 변화와 구성 비율의 변화를 함께 확인해야 한다.**

---

## 4-2. RFM 분석

**RFM 분석**은 고객을 세 가지 기준으로 분류하여 고객의 가치와 행동을 분석하는 고객 세분화 방법이다.

| 요소 | 의미 | 핵심 질문 |
|---|---|---|
| **R: Recency** | 최근성 | 마지막 구매가 얼마나 최근인가? |
| **F: Frequency** | 빈도 | 얼마나 자주 구매했는가? |
| **M: Monetary** | 금액 | 얼마나 많은 금액을 소비했는가? |

RFM을 이용하면 고객을 행동 특성에 따라 분류하고 각 집단에 서로 다른 전략을 적용할 수 있다.

| 고객 유형 | 특징 | 활용 |
|---|---|---|
| **충성 고객** | R, F, M이 모두 높음 | 리워드 및 특별 혜택 |
| **이탈 위험 고객** | 과거 구매 빈도와 금액은 높지만 최근 구매가 적음 | 재구매 유도 |
| **신규 고객** | 최근 구매했지만 빈도가 낮음 | 온보딩 및 재구매 유도 |
| **휴면 고객** | R, F, M이 모두 낮음 | 저비용 재활성화 전략 |

RFM의 장점은 단순한 고객 분류에서 끝나지 않고 **각 고객 집단에 맞는 마케팅 전략으로 바로 연결할 수 있다는 점**이다.

---

# 5. 효과 검증: A/B 테스트

분석을 통해 개선안을 도출한 뒤에는 실제로 해당 개선안이 효과가 있었는지 검증해야 한다.

개선안을 적용한 뒤 지표가 상승했다고 해서 반드시 개선안 때문에 상승했다고 볼 수는 없다. 계절성, 마케팅 활동, 시장 변화 등 다른 요인이 동시에 영향을 미칠 수 있기 때문이다.

따라서 **A/B 테스트** 등을 이용하여 효과를 검증한다.

---

## 5-1. A/B 테스트의 기본 원칙

| 원칙 | 설명 |
|---|---|
| **랜덤 배정** | 실험군과 대조군을 무작위로 배정 |
| **사전 지표 합의** | 실험 전에 성공 판단 기준을 결정 |
| **충분한 표본** | 필요한 표본 크기를 확보 |
| **피킹(Peeking) 방지** | 결과가 좋아 보이는 순간 임의로 실험을 종료하지 않음 |
| **가드레일 확인** | 목표 지표 외의 다른 중요 지표가 악화되지 않았는지 확인 |

A/B 테스트에서는 **실험군과 대조군의 차이를 최대한 실험 요소 하나로 제한하는 것**이 중요하다.

---

## 5-2. A/B 테스트가 어려운 경우

모든 상황에서 무작위 실험이 가능한 것은 아니다. 오프라인 매장, 정책 변경, B2B 등의 상황에서는 랜덤 배정이 어려울 수 있다.

이 경우 다음과 같은 방법을 활용할 수 있다.

- **전후 비교 + 대조군**: 비슷한 특성을 가진 다른 집단과 비교
- **단계적 적용**: 일부 지역이나 집단에 먼저 적용한 후 비교
- **시계열 비교**: 계절성을 고려하여 이전의 동일한 기간과 비교

완벽한 실험이 어렵더라도 단순히 적용 전후의 수치만 비교하기보다 **적절한 비교 대상을 설정하는 것이 중요하다.**

---

## 5-3. 기대효과 추정

분석 결과를 바탕으로 개선안을 제안할 때는 예상되는 효과를 가능한 한 수치로 표현해야 한다.

기본적인 기대효과는 다음과 같이 계산할 수 있다.

$$
기대효과
=
영향받는\ 대상\ 규모
\times
개선폭
\times
단위당\ 가치
$$

기대효과를 추정할 때는 단순히 하나의 숫자를 제시하기보다 **가정의 근거와 가능한 범위를 함께 제시하는 것**이 중요하다.

특히 지나치게 낙관적인 값을 사용하는 것보다 근거가 있는 보수적인 가정을 사용하는 것이 분석 결과의 신뢰성을 높일 수 있다.

---

# 6. 전체 분석 흐름

데이터 분석의 핵심 흐름은 다음과 같이 정리할 수 있다.

```text
문제 정의
   ↓
문제 구조화
   ↓
지표 설정
   ↓
데이터 분석
   ↓
효과 검증
   ↓
의사결정 및 실행
```

데이터 분석은 단순히 모델을 만들거나 수치를 계산하는 작업이 아니라, **문제를 정의하고 적절한 지표를 통해 원인을 파악한 뒤 분석 결과를 실제 의사결정으로 연결하는 과정**이라고 볼 수 있다.
