# 회귀

## 1. 머신러닝과 모델링

- **머신러닝**: 인공지능의 한 분야로, 모델이라는 소프트웨어를 학습하여 새로운 데이터를 예측하거나 결정을 내릴 수 있도록 하는 기술
    1) **지도 학습**: 정답이 있는 데이터 사용
        - 예측값이 정답과 같아지도록 학습
        - ex. 회귀, 분류
    2) **비지도 학습**: 정답이 없는 데이터 사용
        - 데이터의 패턴이나 유사성을 학습
        - ex. 군집화
        - **군집화**
            - 정답이 없는 데이터이기 때문에 주어진 데이터로만 학습
            - 비슷한 데이터를 하나의 그룹으로 묶는 방법
            - 알고리즘이 식별해야 할 클러스터 수를 지정 또는 수정할 수 있음
            - 이를 통해 그룹을 어디까지 세분화할 것인지 제어 가능
    3) **강화 학습**: 시행착오를 반복하며 보상을 최대화하는 행동을 학습
        - ex. 알파고, 자율주행

## 2. 모델링

- **모델**: 학습의 결과물
    - 학습을 마친 인공지능 그 자체, 즉 예측기를 의미
    - 새로운 데이터를 넣었을 때 예측값을 출력

- **모델링**: 예측기를 만들어가는 전체적인 과정

- **모델링의 과정**
    1) **설계**: 데이터의 특성에 맞추어 어떤 알고리즘으로 학습할지 결정
    2) **훈련**: 데이터를 기계에게 계속 보여주며 정답을 맞히도록 수정
    3) **검증**: 훈련시킨 모델이 새로운 데이터에서도 정답을 잘 맞히는지 테스트

## 3. 지도 학습

- **지도 학습**: 정답 레이블을 함께 주고 학습시키는 방법
    - 예측값이 정답과 최대한 같아지도록 학습

- **Labeled Data**: 독립변수와 종속변수가 모두 존재하는 데이터
    - **독립변수(feature, X)**: 예측에 사용되는 입력 변수
    - **종속변수(target/label, Y)**: 예측하고자 하는 대상 변수

- 데이터를 **Train Set**과 **Test Set**으로 나눔
    - **Train Set**: 모델을 학습시키는 데이터
    - **Test Set**: 학습된 모델을 검증하는 데이터
    - Test Set을 따로 두는 이유
        → 새로운 데이터에서도 모델이 예측을 잘하는지 평가하기 위해서

- **지도 학습의 과정**
    1) 데이터 준비
    2) 예측
    3) 오차 계산
    4) 가중치 수정
    5) 반복
    6) 학습 완료
    7) Test Set으로 성능 평가

## 4. 회귀와 분류

- **회귀**: 종속변수가 연속형일 때 사용
    - 연속적인 숫자값을 예측

- **분류**: 종속변수가 범주형일 때 사용
    - 데이터가 어떤 범주에 속하는지 예측

→ **종속변수가 연속형이면 회귀, 범주형이면 분류**

## 5. 회귀

- **회귀**: 데이터에서 패턴을 찾아 미래의 값을 예측하는 방법
- 종속변수가 연속형일 때 사용
- 종류
    - 선형회귀
    - 비선형회귀
    - 규제회귀
    - 다항회귀

## 6. 선형회귀

- **선형회귀**: 데이터를 가장 잘 설명하는 회귀선을 찾는 과정
    - **x**: 독립변수
    - **y**: 종속변수
    - **w (가중치/회귀계수)**: x가 y에 미치는 영향력의 크기 = 기울기
    - **b (편향)**: x=0일 때의 기본값 = y절편

### 선형회귀의 종류

1) **단순선형회귀**
- 독립변수가 1개
- `y = wx + b`

2) **다중선형회귀**
- 독립변수가 2개 이상
- 여러 독립변수를 이용해 하나의 종속변수를 예측

## 7. 잔차

- **잔차**: 실제값과 예측값의 차이
- 회귀의 목적
    → 잔차를 최대한 줄이는 회귀선을 찾는 것

- **잔차와 오차**
    - 오차: 실제값 - 실제 평균값
    - 잔차: 실제값 - 예측값

## 8. 최소제곱법

- **최소제곱법**: 최적의 회귀선을 찾는 방법
- **잔차의 제곱합이 최소**가 되는 회귀선을 선택

- 잔차를 제곱하는 이유
    1) 양수와 음수가 서로 상쇄되는 것을 방지
    2) 큰 잔차에 더 큰 페널티를 부여
    3) 미분 가능한 형태로 만들기 위해

→ **잔차 제곱의 합이 가장 작은 선 = 최적의 회귀선**

## 9. 다중선형회귀

- **다중선형회귀**: 독립변수가 2개 이상인 선형회귀

`y = W1X1 + W2X2 + ... + WnXn + b`

- **회귀계수(W)**: 각 독립변수가 종속변수에 미치는 영향력
- **절편(b)**: 모든 독립변수가 0일 때 y값

## 10. 다중선형회귀의 기본 가정

1) **선형성**
- 종속변수와 독립변수의 관계가 선형적이어야 함

2) **독립성**
- 독립변수들이 서로 강한 상관관계를 가지면 안 됨
- 강한 상관관계가 존재하면 → **다중공선성**

3) **오차항 평균 = 0**
- 오차가 특정 방향으로 치우치지 않아야 함

4) **등분산성**
- 오차항의 분산이 일정해야 함
- 일정하지 않으면 → **이분산성**

5) **비자기상관**
- 오차항끼리 서로 상관관계를 가지면 안 됨
- 상관관계가 있으면 → **자기상관**

6) **정규성**
- 오차항이 정규분포를 따라야 함

→ 분석 전에는 주로 **데이터의 관계(선형성, 독립성)**를 확인하고,
분석 후에는 주로 **오차항의 성질**을 확인

## 11. 다중공선성

- **다중공선성**: 독립변수들끼리 강한 상관관계를 가지는 문제
- 어떤 독립변수가 종속변수에 영향을 주는지 정확하게 구분하기 어려워짐

### 확인 방법

1) **상관계수(r)**
- `-1 ≤ r ≤ 1`
- r > 0 → 양의 상관관계
- r < 0 → 음의 상관관계
- r = 0 → 직접적인 연관성 없음

2) **VIF**
- VIF = 1 → 다른 변수와 상관관계 없음
- VIF < 5 → 일반적으로 문제 없음
- VIF > 5 → 다중공선성 주의
- VIF > 10 → 심각한 다중공선성

### 해결 방법

1) 변수 제거
2) 변수 변환
3) 규제선형모델 사용
4) PCA 사용

## 12. 규제선형모델

- **규제선형모델**: 모델의 과적합을 방지하기 위해 회귀계수의 크기를 조절하는 모델

- **과적합**
    - Train Set에는 매우 잘 맞음
    - 새로운 Test Set에서는 성능이 떨어짐

→ 회귀계수에 **페널티**를 주어 특정 변수에 지나치게 의존하지 않도록 함

### 종류

1) **릿지 = L2 규제**
- 회귀계수의 **제곱합**에 페널티
- 모든 계수의 크기를 줄임
- 계수를 0으로 만들지는 않음
- 모든 변수를 유지하면서 과적합 방지

2) **라쏘 = L1 규제**
- 회귀계수의 **절댓값 합**에 페널티
- 일부 작은 계수를 0으로 만들 수 있음
- 불필요한 변수를 제거하는 **변수 선택 기능**

3) **엘라스틱넷 = L1 + L2**
- 릿지 + 라쏘
- 불필요한 변수는 제거하면서 남은 변수의 계수 크기도 조절

→ **Ridge = 계수 줄이기**
→ **Lasso = 계수 줄이기 + 변수 제거**
→ **Elastic Net = Ridge + Lasso**

## 13. 모델 평가

### 1) F-statistic

- **전체 회귀 모델이 유의미한지** 평가

- H0: 모든 회귀계수가 0
- H1: 적어도 하나의 회귀계수가 0이 아님

- **Prob(F-statistic) < 0.05**
    → H0 기각
    → 회귀 모델이 통계적으로 유의미함

### 2) T-statistic

- **각각의 독립변수가 유의미한지** 평가

- H0: 해당 변수의 회귀계수 = 0
- H1: 해당 변수의 회귀계수 ≠ 0

- **p-value < 0.05**
    → H0 기각
    → 해당 독립변수가 종속변수에 유의미한 영향을 줌

→ **F-test = 전체 모델**
→ **T-test = 개별 독립변수**

## 14. R-squared

- **R 제곱**: 회귀 모델이 종속변수의 변동을 얼마나 잘 설명하는지 나타내는 지표
- `0 ≤ R 제곱 ≤ 1`
- 1에 가까울수록 모델의 설명력이 높음
- 0에 가까울수록 모델의 설명력이 낮음

- **Adjusted R제곱**
    - 독립변수 개수가 늘어나면 R²가 자연스럽게 증가하는 문제를 보완
    - 불필요한 변수를 추가하는 것에 페널티를 부여

→ **F-statistic = 모델이 통계적으로 유의한가?**
→ **R 제곱 = 모델이 데이터를 얼마나 잘 설명하는가?**

## 15. AIC / BIC

- 모델을 평가하는 정보 기준
- **값이 낮을수록 좋은 모델**

- **AIC**: 모델 복잡성에 대한 페널티가 비교적 작음
- **BIC**: AIC보다 복잡성에 더 강한 페널티 부여

## 16. MSE와 MAE

- **MSE**: 예측값과 실제값 사이의 오차를 제곱한 후 평균
    - 큰 오차에 민감
    - 이상치에 민감

- **MAE**: 예측값과 실제값 사이의 절대 오차의 평균
    - 모든 오차를 동일하게 취급
    - MSE보다 이상치에 덜 민감

→ **MSE = 큰 오차를 더 강하게 벌점**
→ **MAE = 오차를 동일하게 취급**

## 17. 비선형회귀

- 데이터의 관계가 직선이 아닐 때 사용하는 회귀

### 1) 다항식 회귀
- X의 제곱, 세제곱 등의 항을 추가
- `Y = a + bX + cX² + dX³ + ...`
- 곡선 형태의 데이터에 사용

### 2) 지수 회귀
- 값이 지수적으로 증가하거나 감소할 때 사용
- Y에 로그를 취한 후 선형회귀 적용
- 예측 후 지수함수를 적용하여 원래 형태로 복구

### 3) 로그 회귀
- X에 로그를 취한 후 선형회귀 적용
- 초기에 빠르게 증가하다가 증가율이 점점 감소하는 데이터에 적합

### 4) 스플라인 회귀
- 데이터를 여러 구간으로 나누어 각 구간에 회귀모델 적용
- 구간마다 데이터 패턴이 다를 때 유용
