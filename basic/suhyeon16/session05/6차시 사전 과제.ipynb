# 군집화

## 1. 군집화

### 1) 머신러닝
인공지능의 한 분야. 컴퓨터가 스스로 학습하게 끔 도와주는 알고리즘이나 기술을 개발하는 분야

- 지도 학습 : 정답이 있는 데이터 사용 (회귀, 분류)
- 비지도 학습 : 정답이 없는 데이터 사용 (군집화, 연관분석)
    - 비지도 학습의 특징 :
        1. 데이터 자체에 내재된 구조를 파악하여 학습함
        2. 정답이 없기 때문에 에이전트가 해야하는 작업이 명확히 정의되어 있지 않음 (즉 모델의 성능을 명확히 측정할 수 없음)
        3. 표현 학습을 통해 데이터셋의 고유 패턴을 식별할 수 있음
        4. 패턴이 아직 알려지지 않았거나, 패턴이 계속 변하거나, 열린 문제를 해결하고 지식을 일반화해야할 시 비지도 학습이 적합함.

### 2) 군집화
데이터를 비슷한 특성을 가진 그룹으로 나누는 비지도 학습 기법임. 각 그룹이 어떻게 구성되어있는지를 파악하면서 전체 데이터 구조를 이해할 수 있음.

- 군집화의 목표: 응집도 최대화 - 같은 군집에 속하는 데이터끼리는 최대한 비슷하도록 함, 분리도 최대화 - 서로 다른 군집은 최대한 분리되도록 함

### 3) 군집화 과정

1. 피처 선택 또는 추출 - 군집화에 사용할 데이터의 피처를 선택
2. 군집화 알고리즘 선택 - 데이터의 특성과 목표에 따라 적합한 알고리즘을 선택
3. 군집 유효성 검증 - 군집화가 얼마나 잘 되었는지 평가
4. 결과 해석 - 각 군집이 지닌 특성 분석 및 해석

## 2. 군집화를 위한 데이터 준비

### 1) 군집화의 주요 고려사항

- 군집화는 여러번의 반복 시도를 통해 최적의 결과를 찾아내는 과정임.
- 변수 유형의 이해: 군집화에 사용될 피처에 대한 정보를 명확히 이해해야함. 또 변수 종류가 연속형/명목형인지에 따라 적합한 군집화 알고리즘이 달라짐
- 거리/유사도 정의와 측정: 군집화는 데이터 간의 거리 또는 유사도를 기반으로 그룹을 형성함. 또한 변수 유형에 따라 적합한 거리 측정 방식이 다르기 때문에 정의가 중요함
- 차원 축소: 변수가 많아질 수록 모델이 복잡해지고 성능이 떨어지기 때문에 불필요한 변수 제거 및 변수 통합을 통해 변수 개수를 줄여야 함.

### 2) 변수 유형 이해

1. 연속형 변수: 특징 (수치로 측정 가능), 거리 측정(유클리디안 거리, 맨하탄 거리 이용), 적합한 알고리즘(K-means, 계층적 군집화 등), 스케일링 필수
2. 명목형 변수: 특징(범주로 구분되는 변수), 거리 측정(해밍 거리, 자카드 거리 등), 적합한 알고리즘(K-modes, 계층적 군집화), 원-핫 인코딩 또는 더미 변수 변환을 통해 처리함
3. 혼합형 변수: 특징(연속형과 명목형이 섞인 데이터), 적합한 알고리즘 (K-prototypes, Gower distance 기반 계층적 군집화), 각 변수 유형에 맞는 거리 함수를 조합하여 사용함

### 3) 거리/유사도 정의와 측정
군집화에서 데이터 포인트나 군집 간의 유사성/비유사성을 수치로 계산하는 방법들. 거리가 가까울 수록 유사하고, 멀수록 다르다고 판단함.

- 유클리디안 거리: 가장 일반적인 거리 측정 방식이며 직선거리를 의미함. K-means에서 기본적으로 사용함.
- 맨하탄 거리: 각 차원의 차이의 절댓값의 합이므로, 이상치에 덜 민감하다는 특징이 있음
- 코사인 유사도: 벡터 간의 각도를 측정함. 텍스트 데이터나 고차원 데이터에 효과적임

### 4) 차원 축소

- 거리 기반의 알고리즘에선 데이터의 차원이 높을 수록 차원의 저주(피처 공간이 너무 커서 알고리즘이 데이터를 효과적으로 훈련하지 못하는 현상)가 발생함.
- 차원의 변수의 개수를 줄이므로써 노이즈를 줄여 패턴 학습을 유리하게 만듬.
- 선형 투영: 고차원 공간에서 저차원 공간으로 선형적으로 데이터를 투영함
- 매니폴드 학습(비선형 차원 축소): 유클리드 거리가 아닌 데이터 포인트들 사이의 곡선 거리를 고려하며 학습
- PCA: 데이터의 분산을 최대한 보존하면서 차원을 줄임
- 상관관계가 높은 피처들을 결합해, 상관 관계가 없는 더 적은 수의 피처로 데이터를 표현
- 원본의 고차우너 데이터에서 최대 분산 방향을 찾아 상관관계를 감소 시키고, 저차원에 공간에 이들을 투영하는 것 -> 이렇게 파생된 성분을 주성분이라고 함.
- 주의사항: 표준화 필수, 결측값 처리, 해석의 어려움, 정보 손실

## 3. 군집화 알고리즘

### 1) 계층적 군집화
가장 가까운 것부터 둘씩 차례로 합쳐 나가면서, 무엇이 언제 뭉쳤는지를 나무 모양으로 기록하는 방식

- 특징:
    - 데이터셋의 관측치를 사용해 덴드로그램 (트리 구조)을 만듬.
    - 군집의 개수를 사전에 설정하지 않고 클러스터링이 종료된 후에 원하는 군집의 개수를 선택할 수 있음.
    - 트리 구조를 상향식(응집형 계층적 군집화)로 구성하냐, 하향식(분리형 계층적 군집화)로 구성하냐에 따라 응집형과 분리형으로 구분
    - 응집형 클러스터링: 각 샘플을 독립적인 클러스터로 시작하여 하나만 남을 때까지 가장 가까운 클러스터 병함함 (과정: 거리 행렬 계산, 단일 클러스터 시작, 가장 가까운 클러스터 시작, 거리 행렬 업데이트, 반복)
    - 분할형 클러스터링: 전체를 포함하는 하나의 클로스터에서 시작해, 유사성이 낮은 데이터를 더 작은 클러스터로 나눔

- 6가지 연결 방법:
    1. 단일 연결법/최단 연결법: 군집 A와 B 사이의 가장 가까운 두 점 간 거리로 군집 간 거리를 정의
    2. 완전 연결법/최장 연결법: 군집 A와 B사이의 가장 먼 두 점 간 거리로 군집 간 거리 정의
    3. 평균 연결법: 군집 A와 B의 모든 점 쌍 사이 거리의 평균
    4. 중심 연결법: 각 군집의 중심점 간의 거리
    5. 중앙 연결법: 두 군집의 중앙값 사이의 거리
    6. Ward 연결법: 군집 내 제곱합이 가장 적게 증가하도록 병합

- 덴드로그램 읽는 법:
    1. 세로축 = 합쳐진 높이 = 그때의 군집 간 거리
    2. 낮게 붙은 것 = 가까운것
    3. 긴 세로 막대 = 그만큼 멀어질 때까지 합칠 상대가 없었다는 뜻

- 계층적 군집화의 장단점:

    **1) 장점:**
    - K를 미리 정하지 않아도 됨
    - 데이터의 구조 전체가 눈에 보임
    - 거리만 정의되면 어떤 유형의 데이터에도 적용 가능
    - 결과가 항상 같음 - 랜덤 초기화가 없음 = random_state 필요 없음

    **2) 단점:**
    - 느리고 메모리를 많이 먹음
    - 한 번 합치면 되돌릴 수 없음
    - 이상치에 민감함
    - 자르는 높이를 결국 사람이 눈으로 정해야함

### 2) K-means
데이터를 정해진 개수의 그룹으로 나누되, 각 그룹의 중심점과의 거리가 가장 가까운 데이터끼리 묶는 군집화 알고리즘

- 군집화에서 가장 일반적으로 사용되며, 중심점을 택하여 가장 가까운 데이터들을 묶어나가는 기법임.

- 주요 단계:
    1. 데이터 중 랜덤하게 K개의 중심점을 초기 클러스터 중심으로 선택
    2. 각 표본들을 가장 가까운 중심점 (주변)값들에 할당
    3. 각 클러스터에 할당된 표본들의 평균을 계산하여 중심점을 이동
    4. 클러스터 할당이 변하지 않거나 최대 반복 횟수에 도달할때까지 반복함.

- K-means의 장단점:

    **1. 장점:**
    - 직관적이고 구현이 쉬움
    - 대용량 데이터에도 적용 가능

    **2. 단점:**
    - 초기 centroid에 민감
    - 군집 수 k 결정이 어려움
    - 아웃라이어에 민감 (이상치 하나가 중심을 왜곡할 수 있음)
    - 기하학적인 모양의 군집은 파악하기 어려움

- 엘보우 방법: 클러스터 내 SSE를 바탕으로 그래프를 활용해 최적 클러스터 개수 k를 추정하는 방법

### 3) DBSCAN
밀도가 높은 지역의 데이터를 하나의 군집으로 묶고, 밀도 기준을 만족하지 못하는 점은 군집에 포함시키지 않는 군집화 알고리즘 (DBSCAN은 k를 정하지 않아도 되고, 이상치를 따로 빼며, 모양을 안가림)

- 특징:
    - 명시적으로 이상치를 지정하여 클러스터링에서 제외
    - 다른 알고리즘 대비 이상치로부터 자유로움
    - 어떤 데이터가 여러 군집의 특정 거리 내에 있으면 가장 밀도가 높은 군집과 그룹화

- DBSCAN 알고리즘의 주요 단계:
    1. 샘플 분류 (핵심 샘플, 경계 샘플, 잡음 샘플)
    2. 클러스터 생성
    3. 경계 샘플 할당

- DBSCAN의 장단점:

    **1) 장점:**
    - 임의의 기하학적 분포도 잘 처리함
    - 이상치에 민감하지 않음
    - 모든 샘플을 클러스터에 할당하지 않음

    **2) 단점:**
    - eps와 min_samples 설정에 많은 영향을 받음
    - 연산량이 많아 속도가 느림
    - 밀도가 서로 다른 군집들을 잘 잡지 못함

### 4) GMM
데이터가 여러 다른 모양의 가우시안 분포로 구성되었다고 가정하고, 각 분포를 클러스터로 인식하는 군집화 방법

- 특징:
    - 모델 기반 군집화
    - 각 군집을 확률 분포로 간주, 전체를 여러 확률 분포의 혼합으로 모델링
    - 데이터의 생성 메커니즘까지 모델링 가능

- GMM진행 과정:
    1. 주어진 전체 데이터셋의 분포를 확인함
    2. 전체 데이터셋은 서로 다른 정규분포 형태의 확률 분포 곡선으로 구성되어 있다고 가정함
    3. 여러 개의 정규 분포 곡선을 추출하고, 개별 데이터가 어떤 분포에 속하는지 결정함. 이때 각각의 분포가 하나의 군집을 이룸.

### 5) 어떤 알고리즘을 고를까

- 몇개로 나눠야 할지 감이 안온다 => 계층적 군집화 (덴드로그램을 그려놓고 나중에 자를 위치 고르기)
- 군집이 대체로 동글동글하고 데이터가 크다 =>k-means (빠르고 직관적. 대용량에서도 무리 없음)
- 군집 모양이 길쭉하거나 서로 겹친다 =>GMM (타원형/중첩 구조를 확률로 처리)
- 이상치를 골라내는 게 목적이다 =>DBSCAN(이상치에 -1을 붙여 명시적으로 분리)
- 군집 모양이 초승달/소용돌이 처럼 이상하다 => DBSCAN(모양을 가정하지 않고 밀도로만 판단)
- 군집마다 밀도가 제각각이다 => HDBSCAN (DBSCAN의 단일 밀도 기준 한계를 보완)
- 경계에 걸친 데이터의 애매함까지 알고 싶다 => GMM (predict_proba로 소속 확률을 변환)

## 4. 군집화 평가 방법

### 1) 외부 평가와 내부 평가

- 외부 평가:
    - 데이터에 정답 레이블이 존재하는 경우에 사용
    - 군집화 결과와 실제 레이블을 비교하여 얼마나 유사한지 평가

- 내부 평가:
    - 정답 레이블이 없는 경우에 사용
    - 군집 내부 응집도와 군집 간 분리도를 기반으로 품질을 정량화 -> 데이터 자체의 구조를 기반으로 판단
    - 알고리즘별 내부 지표 : k-means는 SSE 최소화, GMM은 로그 우도 최대화
    - 범용 지표: 실루엣 계수, 던 지수

### 2) 실루엣 계수

- -1< 실루엣 계수 < 1
- 분리도 > 응집도 => 1에 가까움 (자신의 클러스터에 잘 속해 있으며 다른 클러스터와는 멀리 떨어짐)
- 분리도 = 응집도 => 0에 가까움 (클러스터 경계에 위치해 군집 간 거리가 모호)
- 분리도 < 응집도 => -1에 가까움 (잘못된 클러스터에 할당되어 군집이 섞여있는 상태)

### 3) Dunn Index
클러스터 간 최소 거리와 클러스터 내 최대 거리의 비율로 클러스터링 품질을 평가하는 지표

- Dunn Index 해석:
    - 값이 클수록 좋은 군집화
    - 0 이상의 값을 가지며 상한 없음
    - 실루엣 계수와 함께 종합적으로 판단하는 것을 권장

- Dunn Index의 한계:
    - 군집 수가 많아질수록 계산 비용 증가
    - 극단적인 값에 큰 영향을 받음
    - 이상치에 민감하게 반응해 수치가 급격히 떨어질 수 있음

## <용어 정리>

- 응집도: 같은 군집 안의 데이터끼리 얼마나 가까운가
- 분리도: 서로 다른 군집끼리 얼마나 멀리 떨어져 있는가
- 센트로이드: 군집의 중심점. K-means가 이걸 옮겨가며 학습
- SSE: 군집 내 제곱 오차합. 작을수록 조밀함
- 덴드로그램: 계층적 군집화의 병합 과정을 그린 트리
- 차원의 저주: 차원이 높아질수록 모든 점이 비슷하게 멀어져 거리가 무의미해지는 현상
- 실루엣 계수: (분리도 - 응집도) 둘 중 큰 값. -1~1
