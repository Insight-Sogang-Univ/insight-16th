# 분류

## 1. 분류란?

### 1) 지도 학습: 회귀 vs 분류

- 회귀(데이터가 연속형 변수를 예측하기 위해 사용돰): 주어진 데이터(X)를 기반으로 정답 (Y)를 잘 맞추는 fit하는 함수
- 분류(데이터가 범주형 변수를 예측하기 위해 사용됨): 기존 데이터가 어떤 레이블에 속하는지 패턴을 알고리즘으로 인지한 후 새롭게 관측된 데이터에 대한 레이블을 판별하는 것

### 2) 이진 분류와 다중 분류

- 이진 분류: 예측하고자 하는 변수 어떤 기준에 대하여 참 또는 거짓의 값만을 가질 때
- 다중 분류: 예측하고자 하는 변수가 가질 수 있는 값이 3개 이상일 때



## 2. 분류 모델

### 1) 로지스틱 회귀

- 이진 분류 문제를 푸는 대표적인 알고리즘. 즉 독립 변수의 선형 조합에 로지특 함수를 적용하여 출력값을 0~1 사이로 변환해주는 것을 의미함
- 이진 분류를 단순 선형 회귀로 풀면 문제가 발생함. S자 형태인 함수에는 선형 회귀로 풀게 될 경우 범위를 벗어나는 값들을 가지게 됨
- 시그모이드 함수 ~ 로지스틱 함수:
    - 출력이 0과 1사이의 값을 가지면서 S자 형태로 그려지는 함수 (특징: 입력값이 커지면 1에 수렴, 작아지면 0에 수렴. 입력값이 (-무한대,무한대)일 때, 출력값은 0~1의 값을 가짐)
    - 시그모이드 함수의 가중치 값 w의 커질 수록 기울기가 가파르다
    - 시그모이드 함수의 가중치 값 b가 커질 수록 그래프가 왼쪽으로 이동함

### 2) 의사결정나무

- 조건에 따라 데이터를 분류하며, 최종적으로 데이터가 순수한 label의 집합으로 구성될 때까지 분류를 반복하는 분석 방법
- CART: 데이터셋을 임계값을 기준으로 두 child로 나누는 알고리즘 (임계값을 나누는 방법은 불순도(지니 계수)가 낮아지는 방향으로 함)
    - *불순도?: 분류하려는 데이터 집합에서 서로 다른 클래스가 섞여 있는 정도를 의미함 (지니 지수를 이용하여 불순도를 확인함)
    - *지니 지수: 불순도를 나타냄. 통계적 분산 정도를 정량화해서 표현한 값임.
- 불순도 감소 알고리즘: 현재 집단에 어느정도 다른 객체들이 섞여있는지 확인 -> 불순도가 낮은 쪽으로 가지 형성
- 학습시 고려해야 할 것:
    1. Hyperparameter 설정 (min_samples_split, min_samples_leaf, min_weight_fraction_leaf, max_leaf_nodes, max_features)
    2. 시각화 (분류가 잘 이루어졌는지 확인 가능)
    3. Pruning(가지치기 = 불필요한 노드 지우기)

### 3) SVM

- 클래스를 분류할 수 있는 다양한 경계선 중 최적의 라인을 찾아내는 알고리즘
- 구성:
    - support vector: 데이터 중 결정 경계와 가장 가까이에 있는 데이터들의 집합
    - Decision Boundary: 데이터 분류의 기준이 되는 경계
    - Margin: 결정 경계에서 서포트 벡터까지의 거리
    - Hyperplane: n차원 공간의 (n-1)차월 평면
    - Slack Variables: 완벽한 분리가 불가능할 때 분류를 위해 허용된 오차에 대한 변수

### 4) KNN

- 데이터로부터 거리가 가까운 K개의 다른 데이터 레이블을 참조하여 분류하는 알고리즘
- 순서:
    1. 데이터 준비 (학습 과정이 없기 때문)
    2. K 값 설정 (K는 가장 가까운 이웃의 개수 의미, 보통 홀수개로 설정)
    3. 거리 계산( 새로운 데이터 + 모든 기존 데이터 간의 거리 계산)
    4. 가장 가까운 K개의 이웃 선택( 가장 작은 거리 값을 가진 K개의 데이터 선택)
    5. 분류 (K개의 이웃 중 가장 자주 등장하는 클래스가 예측 결과가 됨)
- 특징: 어떠한 학습이 필요하지 않음. 새로운 데이터가 왔을 때 즉석에서 주변 데이터들을 이용하여 classify함



## 3. 분류 평가 지표

### 1) 혼동 행렬

- 분류 모델의 예측 결과를 정확한 예측과 잘못된 예측으로 구분하여 나타낸 표
- TP: 정답 참, 참으로 예측. TN: 정답 거짓, 거짓으로 예측. FP: 정답 거짓, 참으로 예측. FN: 정답 참, 거짓으로 예측
- 혼동 행렬을 이용한 분류 모델 평가 지표:
    1. 정확도(모든 가능한 예측 중 참인 비율): TP./TP + FP (관건은 FP)
    2. 정밀도(참이라고 예측한 경우 실제 참의 비율)
    3. 재현도(실제로 참인 경우 중 참으로 예측하는 비율)
- 정밀도 & 재현도: 임계값은 FP/FN의 비용을 따져서 정하며, PR 곡선은 그 트레이드 오프를 눈으로 보는 도구임. 임계값 설정이 매우 중요 (임계겂에 따라 포기하는 면적이 달라지기 때문)

### 2) F1 Score

- 정밀도와 재현율의 조화 평균
- 머신러닝 모델의 성능을 평가하는 주요 지표임

### 3) ROC/ AUC

- 분류 모델의 성능을 평가하는 중요한 도구
- ROC Curve: 얼마나 분류가 잘 되었는가를 보여주는 그래프
- AUC Curve: ROC와 x 축 사이의 면적(적분)



## 4. 하이퍼파라미터 최적화

### 1) 하이퍼파라미터 최적화

- 하이퍼파라미터: 학습 시작 전에 사용자가 직접 설정하는 변수
- 하이퍼파라미터 최적화: 적절한 하이퍼파라미터를 찾아 모델 성능을 향상시키는 것
- 하이퍼 파라미터 최적화 과정:
    1. 하이퍼파라미터 탐색
    2. 평가 지표 계산 정의
    3. 설정한 범위에서 값을 샘플링한 하이퍼파라미터 값을 사용하여 검증 데이터로 정확도 평가
    4. 위 반복, 정확도 결과를 보면서 범위 좁히기

### 2) 하이퍼파라미터 최적화 방법

- Grid Search:
    - 장점: 범위 넓음, step 작을 수록 꼼꼼하게 전 범위 탐색, 최적해를 정확히 찾을 수 있음.
    - 단점: 시간 오래 걸림
    - 적용: 큰범위 -> step 활용하여 범위 좁히기

- Random Search:
    - 장점: Grish Search 보다 속도가 빠름
    - 단점: 무작위라는 한계로 인해 정확도 감소

- Bayesian Optimization:
    - 특징: 여러개의 하이퍼 파라미터들에 대해서 "Acquisition Function”을 적용했을 때, "가장 큰 값"이 나올 확률이 높은 지점을 찾아냄
    - 방법: 초기 값 무작위로 실험하기, 어떤 파라미터의 조합이 좋을 지 확률 모델로 예측, 다음 실험시 조합이 가장 좋을 것 같은 위치 선택, 결과 반영하여 다시 반복
