## 1. Pandas란?

Pandas는 데이터 분석에 특화된 Python 라이브러리이다.
행과 열로 이루어진 테이블 형태의 데이터를 다루는 데 유용하며,
대표적으로 `Series`와 `DataFrame`이라는 자료구조를 사용한다.

### Pandas 불러오기

Pandas는 다음과 같이 불러올 수 있다.

`import pandas as pd`

`pd`는 pandas를 간단하게 사용하기 위한 별칭이다.



## 2. DataFrame이란?

DataFrame은 Pandas에서 사용하는 대표적인 **2차원 자료구조**이다.
행(Index)과 열(Column)로 이루어져 있으며, 여러 개의 Series가 모여 하나의 DataFrame을 구성한다.

DataFrame은 다음과 같이 만들 수 있다.

`df = pd.DataFrame(data)`


## 3. 데이터 확인하기

데이터 분석을 하기 전에는 데이터의 크기, 구조, 데이터 타입 등을 먼저 확인하는 것이 중요하다.

- `df.head()` : 앞부분의 데이터 확인
- `df.tail()` : 뒷부분의 데이터 확인
- `df.shape` : 행과 열의 개수 확인
- `df.info()` : 데이터의 전체적인 정보 확인
- `df.describe()` : 수치형 데이터의 기본 통계값 확인
- `df.isnull().sum()` : 각 열의 결측치 개수 확인
- `df['열 이름'].value_counts()` : 특정 열의 값별 개수 확인

## 4. 데이터 추출하기

필요한 데이터만 선택하기 위해 인덱싱과 슬라이싱을 사용할 수 있다.

### loc

`loc`는 행과 열의 **이름(label)**을 기준으로 데이터를 추출한다.

`df.loc[행 이름, 열 이름]`

`loc`로 범위를 지정하면 **마지막 값까지 포함**한다.

### iloc

`iloc`는 행과 열의 **정수형 위치(position)**를 기준으로 데이터를 추출한다.

`df.iloc[행 순서, 열 순서]`

`iloc`로 범위를 지정하면 **마지막 값은 포함하지 않는다.**

### Boolean Indexing

특정 조건을 만족하는 데이터만 추출할 수도 있다.

`df[df['age'] >= 20]`

위 코드는 `age`가 20 이상인 데이터만 추출한다.



## 5. 데이터 조작하기

Pandas에서는 데이터를 추가하거나 삭제하고, 정렬하거나 값을 변경할 수 있다.

- 새로운 열 추가 : `df['새로운 열'] = 리스트`
- 행/열 삭제 : `df.drop()`
- 인덱스 기준 정렬 : `df.sort_index()`
- 특정 열의 값 기준 정렬 : `df.sort_values(by='열 이름')`
- 값 변경 : `df.replace(바꿀 값, 새로운 값)`

## 6. 집계함수

집계함수는 여러 행의 데이터를 계산하여 하나의 대표적인 결과값을 구할 때 사용한다.

- `df['열 이름'].mean()` : 평균
- `df['열 이름'].median()` : 중앙값
- `df['열 이름'].std()` : 표준편차
- `df['열 이름'].sum()` : 합계
- `df['열 이름'].max()` : 최댓값
- `df['열 이름'].min()` : 최솟값

또한 `groupby()`를 집계함수와 함께 사용하면 특정 기준으로 데이터를 그룹화하여 계산할 수 있다.

`df.groupby('그룹 기준 열')['계산할 열'].mean()`

위 코드는 특정 열을 기준으로 데이터를 그룹화한 뒤 각 그룹의 평균을 구한다.
