# EDA 및 데이터 전처리

## 1. EDA 개념

EDA(Exploratory Data Analysis)는 탐색적 데이터 분석으로, 본격적인 분석이나 모델링 전에 데이터를 여러 방식으로 살펴보며 특징과 패턴을 파악하는 과정이다.

EDA를 통해 데이터의 분포, 특이한 값, 변수 간 관계 등을 확인하고 이후 전처리나 분석 방향을 정할 수 있다.

### EDA 대상

- 일변량(Univariate)
  - 하나의 변수를 분석
  - 데이터의 분포와 특징을 파악하는 것이 목적

- 다변량(Multivariate)
  - 여러 변수를 함께 분석
  - 변수들 사이의 관계를 파악하는 것이 목적

### EDA 종류

- 시각화(Graphic)
  - 그래프나 차트를 이용해 데이터를 확인
  - 데이터의 전체적인 형태와 패턴을 파악하기 좋음

- 비시각화(Non-Graphic)
  - 평균, 중앙값 등 통계값을 이용해 데이터를 확인
  - 정확한 수치를 파악하기 좋음

### EDA 유형

EDA는 대상과 방법에 따라 다음과 같이 나눌 수 있다.

- 일변량 비시각화: 하나의 변수의 분포를 통계값으로 확인
- 일변량 시각화: 하나의 변수의 분포를 그래프로 확인
- 다변량 비시각화: 여러 변수의 관계를 수치로 확인
- 다변량 시각화: 여러 변수의 관계를 그래프로 확인




- 순서형(Ordinal)
  - 범주 사이의 순서에 의미가 있음

#### 수치형(Numerical)

숫자로 표현되는 데이터

- 이산형(Discrete)
  - 각각의 값을 셀 수 있는 데이터

- 연속형(Continuous)
  - 일정 범위 안에서 연속적인 값을 가지는 데이터

이산형과 연속형은 단순히 정수와 소수로 구분하는 것이 아니라 값이 셀 수 있는지, 연속적인 범위를 가지는지를 기준으로 구분한다.







## 2. 데이터셋 파악하기

분석을 시작하기 전에 데이터의 전체적인 구조와 각 변수가 어떤 정보를 가지고 있는지 확인해야 한다.

### 데이터 확인

- `head(n)` : 데이터의 상위 n개 행 확인
- `info()` : 데이터 크기, 각 열의 자료형과 결측값 확인
- `describe()` : 평균, 표준편차, 최솟값, 최댓값 등 주요 통계값 확인

### 변수 유형

변수는 크게 범주형과 수치형으로 나눌 수 있다.

#### 범주형(Categorical)

몇 개의 범주로 구분되는 데이터

- 명목형(Nominal)
  - 범주 사이에 순서가 없음

- 순서형(Ordinal)
  - 범주 사이의 순서에 의미가 있음

#### 수치형(Numerical)

숫자로 표현되는 데이터

- 이산형(Discrete)
  - 각각의 값을 셀 수 있는 데이터

- 연속형(Continuous)
  - 일정 범위 안에서 연속적인 값을 가지는 데이터

이산형과 연속형은 단순히 정수와 소수로 구분하는 것이 아니라 값이 셀 수 있는지, 연속적인 범위를 가지는지를 기준으로 구분한다.





## 3. 데이터 전처리

데이터 전처리는 데이터를 분석에 적합한 형태로 만드는 과정이다.

데이터에 결측값이나 이상치가 포함되어 있으면 분석 결과가 왜곡될 수 있기 때문에 분석 전에 데이터를 확인하고 적절하게 처리해야 한다.

### 결측값 처리

결측값은 데이터 수집 과정에서 측정되지 않거나 누락된 데이터이다.

#### 결측값 확인

- `info()` : 각 열의 유효한 데이터 개수 확인
- `isnull()` : 결측값이면 True 반환
- `df.isnull().sum()` : 각 열의 결측값 개수 확인
- `notnull()` : 유효한 데이터이면 True 반환
- `replace()` : `?`, `0` 등으로 표시된 결측값을 NaN으로 변경

#### 결측값 처리 방법

1. 삭제
   - `dropna()` 사용
   - 결측값이 포함된 행이나 열을 삭제
   - 데이터가 너무 많이 삭제되면 정보가 손실될 수 있으므로 주의

2. 대체
   - `fillna()` 사용
   - 결측값을 평균값, 최빈값 등의 대표값으로 대체
   - 비슷한 특성을 가진 그룹의 대표값으로 대체할 수도 있음


### 이상치 처리

이상치는 일반적인 데이터 범위에서 크게 벗어난 값이다.

이상치는 평균이나 변수 간 관계에 영향을 주어 분석 결과를 왜곡할 수 있기 때문에 원인을 확인한 뒤 처리해야 한다.

#### 이상치 확인 방법

- `describe()` : 통계값을 이용해 확인
- Boxplot : 그래프를 이용해 확인
- Z-score : 평균에서 얼마나 떨어져 있는지 확인
- Tukey Fences : IQR을 이용해 이상치 판단

Tukey Fences에서는 일반적으로 다음 범위를 벗어난 값을 이상치로 판단한다.

- Q1 - 1.5 × IQR 미만
- Q3 + 1.5 × IQR 초과

#### 이상치 처리 방법

- 입력 오류 등으로 발생한 값은 삭제
- 필요한 경우 평균 등의 값으로 대체
- 자연적으로 발생한 이상치는 바로 삭제하지 않고 원인을 확인
- 필요하면 이상치를 따로 분리하여 분석



## 4. 피처 엔지니어링

피처 엔지니어링은 기존 변수를 분석에 더 적합한 형태로 변형하거나 가공하는 과정이다.

피처(feature)는 데이터프레임의 입력 변수 또는 column을 의미한다.

### 주요 방법

#### Label Encoding

범주형 데이터를 숫자로 변환한다.

#### One-Hot Encoding

범주형 데이터를 0과 1로 이루어진 형태로 변환한다.

`pd.get_dummies()`를 이용할 수 있다.

#### Binning

연속적인 값을 일정한 구간으로 나누는 방법이다.

`pd.cut()` 등을 이용하여 데이터를 여러 구간으로 구분할 수 있다.

#### 파생변수 생성

기존 변수를 활용하여 새로운 변수를 만드는 방법이다.

기존 데이터를 다른 방식으로 표현하여 분석에 필요한 정보를 추가할 수 있다.

#### Scaling

변수마다 값의 범위가 다른 경우 일정한 수준으로 맞추는 과정이다.

- StandardScaler
  - 평균을 0, 분산을 1로 변환

- MinMaxScaler
  - 데이터를 0~1 범위로 변환


## 5. 데이터 시각화

데이터 시각화는 데이터를 그래프, 차트 등의 형태로 표현하여 데이터의 분포, 패턴, 관계를 쉽게 파악하는 과정이다.

Python에서는 주로 Matplotlib과 Seaborn을 사용한다.

그래프를 선택하기 전 데이터가 범주형인지 수치형인지 확인하고 결측값과 이상치가 있는지도 확인해야 한다.

### 주요 그래프

- `boxplot`
  - 데이터의 분포와 이상치를 확인

- `countplot`
  - 범주형 변수의 빈도수를 확인

- `histplot`
  - 수치형 변수의 구간별 분포를 확인

- `kdeplot`
  - 데이터의 분포를 곡선 형태로 확인

- `barplot`
  - 범주에 따른 수치형 데이터의 평균값 비교

- `scatterplot`
  - 두 수치형 변수 사이의 관계와 패턴 확인

- `heatmap`
  - 변수 간 상관관계를 한눈에 확인

- `pairplot`
  - 여러 변수 사이의 관계를 한 번에 확인

### 상관계수

상관계수는 두 수치형 변수 사이의 관계 정도를 나타내며 -1에서 1 사이의 값을 가진다.

- 1에 가까움: 한 변수가 증가할 때 다른 변수도 증가하는 경향
- -1에 가까움: 한 변수가 증가할 때 다른 변수는 감소하는 경향
- 0에 가까움: 두 변수 사이의 관계가 약함
