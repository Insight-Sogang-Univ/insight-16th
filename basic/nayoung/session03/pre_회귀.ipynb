## 선형회귀
 <BR>

### 선형회귀 개념

_데이터를 가장 잘 설명하는 **회귀선(ex. 직선 y=wx+b)**을 찾는 과정_ <BR>
- 회귀선은 **현재 데이터를 요약**하는 동시에, 아직 관측되지 않은 **미래 값을 예측하는 도구** 
- 단순선형회귀(x가 한 개)와 다중선형회귀(x가 여러 개)로 나뉨

### 선형회귀 기본식

𝑦 = 𝑤𝑥 + 𝑏 <BR>

(1) 가중치 <BR>
𝑤(Weight)

(2) 편향 <BR> 
𝑏(Bias)

### 선형회귀분석 종류

#### **단순선형회귀분석(Simple linear Regression)**
 <BR>

- **하나의 독립변수(**$x$**)**만 존재하는 선형 회귀
- **데이터를** **직선 형태(하나의 독립 변수)**로 표현하는 경우
- 데이터를 가장 잘 설명하는 $y=wx+b$을 찾는 과정에 해당
- 잔차를 줄여나가며 직선을 결정짓는 적절한 **회귀계수(**$w$**)**와 **편향(**$b$**)**을 찾음.

#### **다중선형회귀분석(Multiple linear Regression)**
 <BR>

- **독립 변수가 두 개 이상**인 경우
- **독립 변수 $x_i$가 2개인 경우 위와 같이 3차원 공간**에 표현되고, **회귀선은 평면**이 됨 <BR>
변수가 2개보다 많은 경우에도 비슷한 느낌으로 차원이 확장됨
- 단순선형회귀와 마찬가지로 **최소제곱법/경사하강법**을 사용 가능, 이때 **잔차**는 관측치와 평면간의 차이 <BR>
→ 파악이 힘들 땐 경사하강법(Gradient Descent) 사용해야 함

##### 다중선형회귀분석 기본 가정
- 선형성
- 독립성
- 오차항의 평균은 0
- 등분산성
- 오차항은 자기상관되어 있지 않다
- 정규성

## 다중공선성

_독립 변수들 간의 강한 상관관계로 인해 회귀 모델이 불안정해지는 문제_ <BR>
다중공선성은 문제가 심각하면 **각 변수의 회귀계수 추정이 부정확해지고, 해석력이 떨어질 수 있기 때문에** 선형회귀에서 중요한 체크 사항

### 다중공선성 확인 방법
#### 1. 상관계수
- $-1 ≤ r ≤ 1$  사이의 값을 가진다.
- $r>0$: 양의 상관관계, 한 변수가 증가하면 다른 변수도 증가하는 경향.
- $r<0$: 음의 상관관계, 한 변수가 증가하면 다른 변수는 감소하는 경향.
- $r=0$: 두 변수 사이에 직접적인 연관성이 없다.

#### 2. 다중공선성 대처 방법
(1) 변수 제거 <BR>
(2) 변수 변환 <BR>
(3) 규제선형모델 활용 <BR>
(4) PCA(주성분 분석)

## 규제선형모델
_모델이 학습 데이터에 과적합(overfitting) 되지 않도록 규제를 가하고자 등장한 모델_ <BR>
**기존 선형 모델에서는 잔차 제곱합의 평균을 최소화하는 방식으로 비용 함수를 설정** <BR>
-> 오히려 회귀 계수가 쉽게 커지게 되어 과적합이 발생할 가능성이 커짐

### 규제선형모델 종류
1. L2규제: W(회귀계수)의 **제곱**에 대해 페널티를 부여하는 방식, **릿지(Ridge)회귀** <BR>
_기둥에 따라 다른 힘으로 누르는 망치_ <BR>
- 큰 기둥은 세게, 작은 기둥은 살살. 비율이 균등하지 않을 수는 있지만 전체적으로 작아짐
- 회귀계수의 제곱을 패널티로 주기 때문에, 큰 회귀계수일 수록 훨씬 강한 패널티를 받음
- 모든 계수의 크기가 작아지지만, 규제를 통해 변수가 제거되지는 않음(회귀계수가 0이 되지 않기 때문)

2. L1규제:  W(회귀계수)의 **절댓값**에 대해 페널티를 부여하는 방식, **라쏘(Lasso) 회귀** <BR>
_모든 기둥을 같은 힘으로 누르는 망치_ <BR>
- 작은 기둥들은 힘을 버티지 못하고 완전히 사라질 수도 있음의 절댓값을 패널티로 주기 때문에, 회귀계수의 크기와 상관없이 같은 패널티를 받음
- 작은 계수가 0이 되기도 하기 때문에 변수가 완전히 사라질 수 있음

3. L2규제 + L1규제**: 엘라스틱넷(Elastic Net) 회귀**
_L1, L2 두 개의 망치를 동시에 사용_ <BR>
**L1 망치:** 작은 기둥을 아예 없애 불필요한 변수 제거 <BR>
**L2 망치:** 남은 기둥을 균형 잡히게 눌러 크기 조절 <BR>
- 두 개의 망치를 함께 사용하면? → 불필요한 변수는 제거하고, 남은 변수들은 적절한 크기로 유지 
- L1의 변수 선택 기능(일부 계수 삭제)과 L2의 규제 기능(큰 계수를 줄임)을 동시에 활용하는 모델입니다.
- L1은 상관관계가 높은 변수 중 일부만 선택하는 경향이 있음 <BR>
→ L2를 함께 적용하면 상관관계가 높은 변수를 함께 고려하면서 규제 가능

## 모델 평가 방법

### 1. 시각화
 <BR>

- 회귀선이 데이터 전반을 잘 요약하는 모습
- 하지만 둘 중 어느 회귀선이 해당 데이터를 더 잘 요약하는지 객관적으로 비교하기는 어려움
- 두 집단의 차이의 정도를 객관적으로 비교하기 위해서 검정했듯이, 이 경우에도 통계지표를 활용하는 것이 좋음

### 2. 통계지표
 <BR>
<BR>
1) 모델 유의성 검정 <BR>
① F-statistic, Prob(F-statistic) <BR>
② T-statistic, Prob(T-statistic) <BR>
③ F-검정, T-검정 비교! <BR>

2) 모델의 성능 평가 <BR>
① 결정 계수 <BR>
② 조정된 결정 계수 <BR>
③ AIC, BIC(SC) <BR>

### 3. 성능 평가 지표
1) 평균 제곱 오차 (Mean Squared Error, MSE)
2) 평균 절대 오차 (Mean Absolute Error, MAE)
3) MSE와 MAE의 차이

### 4. 변수 유의성 평가
**T-검정** <BR>
t-검정은 독립 변수의 회귀계수가 유의미한지 검정하는 데 사용됩니다. <BR>
귀무가설 : 회귀계수(a)가 0이다 → 독립변수가 종속변수에 영향을 주지 않는다. <BR>
대립가설 : 회귀계수(a)가 0이 아니다 → 독립변수가 종속변수에 영향을 준다. <BR>
