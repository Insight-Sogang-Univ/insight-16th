# 분류

## 회귀와 분류의 차이
### 지도학습: 회귀 vs 분류
#### 회귀
**_주어진 데이터(X)를 기반으로 정답(Y)를 잘 맞추는(fit)하는 함수(function)_**
- 데이터가 연속형 변수를 예측하기 위해 사용될 때

#### 분류
**_기존 데이터가 어떤 레이블에 속하는지 패턴을 알고리즘으로 인지한 뒤에 새롭게 관측된 데이터에 대한 레이블을 판별하는 것_**
- 데이터가 범주형 변수를 예측하기 위해 사용될 때

## 분류 모델의 네 가지 종류와, 각 모델이 무엇인지

1. 로지스틱 회귀(Logistic Regression)

 <BR>

_이진 분류 문제를 푸는 대표적인 알고리즘으로, 샘플이 특정 클래스에 속할 확률을 추정하여 기준치에 따라 분류하는 것을 목표로 함_<BR>
- 출력이 0과 1사이의 값을 가지면서 S자 형태로 그려지는 함수를 이용해야 할 때 사용
- 대표 함수: 시그모이드 함수 (Sigmoid function) <BR>
특징: <BR>
(1) 입력값이 커지면 1에 수렴하고, 입력값이 작아지면 0에 수렴
(2) 입력값($x$)이 $(-\infin, \infin)$일 때, 출력값($y$)은 0~1의 값을 가짐

2. 의사결정나무 (Decision Tree)

<BR>

_조건에 따라 데이터를 분류하며, 최종적으로 데이터가 순수한 label의 집합으로 구성될 때까지 분류를 반복하는 분석 방법_ <BR>
- **Root Node**: 가장 상위에 위치한 노드
- **Parent Node**: 주어진 노드의 상위에 있는 노드
- **Child Node**: 하나의 노드로부터 분리된 2개 이상의 노드들
- **Binary Tree**(이진 트리): Tree 중에서 children이 최대 2개인 tree
- **Leaf Nodes**: 자식마디가 없는 노드, 모델에서 label에 해당
- **Edge**: Parent node와 child node를 연결하는 연결선
- **Height**: 특정 노드에서 가장 멀리 있는 leaf node까지의 경로에 있는 edge 개수
- **Depth:** Root node에서 특정 노드에 도달하기 위한 edge의 수 <BR>

- 대표 함수: CART(이진분할; Classification And Regression Tree) 알고리즘 <BR>
특징: <BR> 
(1) 임계값 설정 <BR>
(2) 불순도 감소 알고리즘 <BR>

- 실제 학습 시 고려해야 할 것들 <BR>

(1) Hyperparameter 설정 <BR>
min_samples_split: 분할되기 위해 노드가 가져야 하는 최소 샘플(데이터) 수 <BR>
min_samples_leaf: 리프 노드가 가지고 있어야 하는 최소 샘플 수 <BR>
min_weight_fraction_leaf: min_samples_leaf와 같지만 가중치가 부여된 전체 샘플 수에서의 비율 <BR>
min_samples_leaf와 같지만 가중치가 부여된 전체 샘플 수에서의 비율 <BR>
max_leaf_nodes: 리프 노드의 최대 개수 <BR>
max_features: 각 노드에서 분할에 사용할 특성의 최대 수 

(2) 시각화 <BR>

 <BR>
-> 왼쪽 그림의 경우 규제가 없어서 과적합이 이루어진 반면, 오른쪽 그림은 min_samples_leaf 개수를 제어하여 적절히 분류됨

(3) Pruning(가지치기) <BR>



3. SVM (Support Vector Machine) <BR>
**_클래스를 분류할 수 있는 다양한 경계선 중 최적의 라인을 찾아내는 알고리즘_**<BR>
**_데이터를 분리하는 초평면(Hyperplane) 중에서 데이터들과 가장 거리가 먼 초평면을 선택하여 분리하는 지도 학습 기반의 이진 선형 분류 모델_** <BR>

<BR>

[SVM의 장단점] <BR>

| **장점** | **단점** |
| --- | --- |
| 서포트 벡터만을 이용해서 결정 경계를 생성하므로 데이터가 적을 때 효과적 | 데이터 전처리 과정이 중요 |
| 새로운 데이터가 입력되면 전체 데이터 포인트와의 거리를 계산하지 않고 서포트 벡터와의 거리만 계산하면 되기 때문에 연산량 최소화 | 데이터가 많아질수록 최적화를 위한 평가 과정이 많아져서 속도가 느림 |
| 정확성이 뛰어나며, 비선형 모델 분류 가능 (커널 사용시) |  |
| 마진을 최대화하는 방식이 규제 역할을 하고 노이즈의 영향이 적음 |  |

4. KNN (K-Nearest Neighbor)

**_데이터로부터 거리가 가까운 k개의 다른 데이터 레이블을 참조하여 분류하는 알고리즘_**<BR>

<BR>

[KNN 알고리즘 계산 순서]
1. 데이터 준비
    
    KNN은 미리 학습하는 과정이 없기 때문에, 데이터를 준비
    
    각 데이터는 특징 벡터와 레이블(어떤 class인지)로 이루어짐
    
2. K 값 설정
    
    원하는 K 값을 설정. **K**는 **가장 가까운 이웃의 개수**를 나타내며, 보통 **홀수**개로 설정.
    (k를 짝수로 설정했을 때, 동점이 발생하는 문제를 방지하기 위해서 보통 홀수개로 설정)
    
3. 거리 계산
    
    새로운 데이터(예측하려는 데이터)가 주어지면, 이 값과 **기존 모든 데이터 간의 거리를 계산**.
    
    보통 이때 유클리드 거리, 맨해튼 거리 등을 사용.
    
4. 가장 가까운 K개의 이웃 선택
    
    계산된 거리 중에서 **가장 작은 거리 값을 가진 K개의 데이터를 선택**.
    
    이 데이터 포인트들이 “가장 가까운 이웃”.
    
5. 분류하기
    
    K개의 이웃 중 **가장 많이 등장하는 클래스가 예측 결과**가 됨. <BR>

[KNN의 장단점]<BR>

| **장점** | **단점** |
| --- | --- |
| 범주를 나눈 기준을 알지 못해도 데이터를 분류할 수 있음 | 다른 분류 모형과 달리 테스트 데이터의 개수에 따라 시간이 오래 걸림 |
| 입력 데이터만 주어지면 바로 예측값을 구할 수 있어서 다른 분류 모형과는 달리 학습 과정이 필요 없음 | 학습 데이터 모두를 거리 계산에 사용하기 때문에 학습 데이터의 양도 계산 시간에 영향을 미침 |
| 이해가 쉽고, 구현이 간단 | K의 값 결정이 어려움 |
| 추가된 데이터의 처리가 용이 | 수치형 데이터가 아니면 유사도를 정의하기 어려움 |
| 훈련 데이터를 요약하거나 일반화된 모델로 만들지 않고, 원본 데이터 전체를 그대로 저장 | 데이터 내 이상치가 존재하면 분류 성능에 큰 영향을 받음 |

## 분류 평가 지표

1. 혼동 행렬
**_분류 모델의 예측 결과를 정확한 예측과 잘못된 예측으로 구분하여 나타낸 표(행렬)_** <BR>

 <BR>

- **파란색 상자** : 예측을 올바르게 한 경우 (**예측과 실제 결과가 모두 일치**)
- **빨간색 상자** : 예측을 잘못한 경우 (**예측과 실제 결과가 불일치**)

[경우의 수] <BR>

| 구분 | 분류 값 | 설명 |
| --- | --- | --- |
| 예측이 정확한 경우 | **True Positive (TP)** | 정답이 참, 참으로 예측 |
|  | **True Negative (TN)** | 정답이 거짓, 거짓으로 예측 |
| 예측이 틀린 경우 | **False Positive (FP)** | 정답이 거짓, 참으로 예측 (예측이 틀림)  |
|  | **False Negative (FN)** | 정답이 참, 거짓으로 예측 (예측이 틀림)  |

[혼동 행렬을 이용한 분류 모델 평가 지표] <BR>
- 정확도 (Accuracy): 모든 가능한 예측 중 참인 비율
- 정밀도 (precision): 참이라고 예측한 경우 실제 참의 비율
- 재현도 (recall): 실제로 참인 경우 중 참으로 예측하는 비율

2. F1-Score

 <BR>

**_정밀도(precision)와 재현율(recall)의 조화 평균_**
- Precision(정밀도)과 Recall(재현율) 간의 균형을 효과적으로 평가하기 위해서 사용

3. ROC / AUC
(1) **ROC Curve** <BR>
: 얼마나 분류가 잘 되었는가를 보여주는 그래프

 <BR>

 <BR>

- True Positive Rate (TPR) : 참인 것들 중에 참이라고 예측한 비율 (=Recall)
- False Positive Rate (FPR) : 거짓인 것들 중에 참이라고 잘못 예측한 비율

(2) **AUC Curve** <BR>
: ROC와 x축 사이의 면적(적분값) <BR>

 <BR>

- 0.5에서 1사이의 값
- 1에 가까울수록 분류 성능이 좋음

## 하이퍼파라미터 최적화

<BR>

**하이퍼파라미터:** 학습 시작 전에 사용자가 직접 설정하는 변수 <BR>
- 하이퍼파라미터 최적화: 적절한 하이퍼파라미터를 찾아 모델 성능을 향상시키는 것<BR>

[하이퍼파라미터 최적화 과정]<BR>
1. Hyperparameter 탐색 범위 설정 (최적 값을 찾고 싶은 하이퍼파라미터의 범위를 설정)
2. 평가 지표 계산 함수(성능 평가 함수) 정의(탐색하려는 Hyperparameter를 인수로 받아 평가지표 값을 계산해주는 함수를 정의)
3. 설정한 범위에서 값을 샘플링한 Hyperparameter 값을 사용하여 검증 데이터로 정확도 평가
4. 위 단계를 특정 횟수 반복하며, 정확도 결과를 보고 하이퍼파라미터의 범위 좁힘

### 하이퍼파라미터 최적화 방법
1. Grid Search: 정해진 범위에서 Hyperparameter를 모두 순회

 <BR>

- **장점**: 범위가 넓고 step이 작을수록 꼼꼼하게 전 범위를 탐색해 최적해를 **정확히 찾을 수 있음**.
- **단점**: 시간이 너무 오래 걸림
- **적용**: 넓은 범위, 큰 step을 활용해 범위를 좁힘. <BR>

2. Random Search: 정해진 범위에서 Hyperparameter를 무작위로 탐색 <BR>

 <BR>

- **장점**: 속도가 Grid Search보다 빠르다.
- **단점**: 무작위라는 한계 때문에 **정확도가 떨어진다**. 따라서 Grid Search나 Bayesian Optimization에 비해 사용 빈도가 적다. <BR>

3. Bayesian Optimization: 사전 정보를 바탕으로 Hyperparameter 값을 확률적으로 추정하며 탐색 <BR>

 <BR>

- **특징** : "Gaussian Process"라는 통계학을 기반으로 만들어진 모델로, 여러 개의 하이퍼 파라미터들에 대해서, "Acquisition Function”을 적용했을 때, "가장 큰 값"이 나올 확률이 높은 지점을 찾아냄. <BR>
- **방법**: <BR>
    1. 초기 값 몇 개를 무작위로 실험 해봄
    2. 그 결과를 바탕으로 어떤 파라미터의 조합이 좋을 지를 확률 모델로 예측
    3. 다음 실험할 조합을 가장 좋을 것 같은 위치로 선택
    4. 결과를 반영하고 다시 반복
