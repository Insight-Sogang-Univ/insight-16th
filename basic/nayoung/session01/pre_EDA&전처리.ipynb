# EDA


## EDA 개념
_데이터 분석(모델링)을 위해 데이터를 **여러 가지 방식으로 파악하는 모든 과정**_
_Exploratory Data Analysis의 약자로, 탐색적 데이터 분석 과정_

### EDA 대상
#### Univariate(일변량)
-> EDA를 통해 한 번에 파악하려는 변수가 한 개
데이터를 설명하고 그 안에 존재하는 패턴을 찾는 것이 주요 목적

#### Multi-variate(다변량)
-> EDA를 통해 한 번에 파악하려는 **변수가 여러 개**
**여러 변수들 간의 관계**를 보는 것이 주요 목적

### EDA 종류
#### Graphic(시각화)


#### Non-Graphic(비시각화)



### EDA 유형

- **Uni-Non Graphic(일변량 비시각화) :** 주어진 데이터의 Distribution을 확인하는 것이 주목적
- **Uni-Graphic(일변량 시각화)** : 주어진 데이터를 전체적으로 살펴보는 것이 주목적
- **Multi-Non Graphic(다변량 비시각화)** : 주어진 둘 이상의 변수 간 관계를 확인하는 것이 주목적
- **Multi-Graphic(다변량 시각화)** : 주어진 둘 이상의 변수 간 관계를 전체적으로 살펴보는 것이 주목적


## 데이터셋 파악하기
### 1. 데이터 프레임 보기
**head(n):** 데이터 프레임 상위 n개 데이터 보여줌


### 2. 데이터 변수 확인
변수 = 데이터 프레임의 column = feature
#### 변수 정의 확인
- 어떤 정보를 가지는 변수인지 확인

#### 변수 유형 확인 
- 질적·범주형(Categorical Data)과 양적·수치형(Numerical Data)으로 구분

#### 변수 데이터 형식(자료형) 확인
- 날짜, 수치, 텍스트, 이미지 등의 구분


### 3. 데이터 분포 확인


_**단변수 분석:** 원시 데이터(raw data)의 평균값, 최빈값, 중간값 등 변수들의 분포를 산포도, 박스 플롯, 히스토그램 등의 그래프를 통해 하나의 데이터 분포를 확인해 분석할 수 있는 방법_


## 전처리


_데이터를 분석에 적합한 형태로 만드는 과정_

### 결측값 처리
**결측값 =** 데이터 수집 과정에서 측정되지 않거나 누락된 데이터

1. 결측값 확인

데이터프레임 상에서 NaN, ?, 0 등의 값으로 나타나 있는 결측값을 확인

2. 결측값 처리

- 삭제 : 데이터가 있는 행 또는 열을 삭제

 **삭제 시 주의할 점**
삭제는 결측값이 무작위로 발생한 경우에 사용합니다. 결측값이 무작위로 발생한 것이 아닌데 관측치를 삭제한 데이터를 사용할 경우, 왜곡된 모델이 생성될 가능성 있음

- 대체 : 결측값을 다른 값으로 대체

### 이상치 처리
**이상치 =* * 관측된 데이터의 범위에서 많이 벗어난 값

1. 이상치 확인

**통계**
describe() 활용

**시각화**
BoxPlot 활용

**Z-score**
데이터를 **평균(0)과 표준편차(1)로 정규화**하여, **평균으로부터 얼마나 떨어져**있는지 나타냄

**Tukey Fences**
사분위 범위(IQR, Interquartile Range)를 기반으로, 두 가지 경우에 이상치라고 판단



2. 이상치 제거 및 처리

- 전체 삭제
- 다른 값으로 대체
- 변수화
- 리샘플링


## 피처 엔지니어링
_해결하고자 하는 문제를 컴퓨터가 잘 이해할 수 있도록 피처(변수)들의 형태를 변형하거나 적절하게 처리하는 과정_

### 피처 엔지니어링 방식
1) 레이블인코딩, 원핫인코딩: 문자열 값들을 숫자형으로 변경하는 작업
2) 구간화(Binning): 연속적인 값을 일정한 구간(Bin)으로 구분 
3) 파생변수 생성: 기존의 피처를 변환해 새로운 피처 생성
4) 스케일링: 서로 다른 변수의 값 범위를 일정한 수준으로 맞추는 작업\

1) 레이블인코딩(Label Encoding) vs 원핫인코딩(One-hot Encoding)


2) 구간화(Binning)


3) 파생변수 생성


4) 스케일링
- 표준화
: 각 feature의 평균을 0, 분산을 1로 변경하여, 모든 피처들이 같은 스케일을 갖도록 함

- MinMaxScaler
: 모든 feature가 0과 1사이에 위치하게 만듦. 데이터가 2차원 셋일 경우, 모든 데이터는 x축의 0과 1 사이에, y축의 0과 1사이에 위치하도록 함.
