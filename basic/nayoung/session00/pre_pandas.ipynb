## 데이터 프레임
### 데이터 프레임 개념
**판다스에서 주로 사용하는 데이터 구조**로, 컴퓨터로 정보를 정리하고 저장하는 도구



### 데이터 프레임 구성 요소
#### 1. Index(행)와 Column(열)


- 함수의 (x,y) 좌표처럼, 행과 열 또한 데이터프레임 내에서 데이터의 위치를 나타내는 역할

#### 2. Series



- **‘값(value)’과 ‘인덱스(index)’**로 구성된, 가장 간단한 1차원 자료구조
- 시리즈를 구성하는 값들은 **모두 같은 데이터타입**이어야 함
#### 3. DataFrame



- 데이터프레임은 행과 열로 이루어진 **2차원 배열 구조**
- **여러 개의 시리즈(Series)가 모여** 하나의 **데이터프레임**을 구성
- **컬럼 단위로는 값들의 데이터 타입이 동일**해야 하며 **컬럼 간의 데이터타입은 달라도 무방**


## 데이터 확인, 추출, 조작 방법
### 확인
#### 1. head(), tail()
- **앞/뒤 일부 데이터를 미리 확인**하여 ****데이터가 **어떤 구조와 값을 가지고 있는지 개괄적으로 파악**
    - `df.head(n)` : 앞 n개의 행 반환 (기본값 n=5)
    - `df.tail(n)` : 뒤 n개의 행 반환

#### 2. 데이터 정보 확인
- 데이터프레임에 **몇 개의 행과 열이 있는지, 각 열의 데이터 타입, 결측치의 개수** 등을 확인
**df.shape** : (행 개수, 열 개수) 반환
**df.info()** : 여러 정보들을 종합적으로 반환
**df.describe()** : 수치형 열들의 기본적인 통계값 반환
**df.isnull()** : 데이터에 결측치(NaN)가 있는지의 여부를 반환
**df.dtypes** : 열들의 데이터타입 반환

#### 3. 고유값 및 개수 확인
- 데이터의 **전체 개수와 고유값**을 파악
**df.count()** : 결측치의 개수를 제외한 데이터의 행 수 반환
**df[열 이름].value_counts()** : 특정 열에 포함된 값들의 빈도수를 세어 반환
**df[열 이름].unique()** : 특정 열에 포함된 값들을 종류별로 1개씩만 반환


### 추출
#### 1. 인덱싱(Indexing) & 슬라이싱(Slicing)

- 인덱싱(Indexing)
: 단일 항목을 지목할 때 사용
→ 예: df['노트북 브랜드']

- 슬라이싱(Slicing)
: 범위를 지정해 여러 항목을 추출할 때 사용
df[start:end] 

#### 2. 행단위 추출 : loc과 iloc
- 데이터프레임의 **행 데이터**를 선택하기 위해서는 **`loc`**과 **`iloc`**를 사용
- **인덱스 이름(index label)**을 기준으로 행을 선택할 때는 **`loc`**을 이용
- **정수형 위치(integer position**)를 사용할 때는 **`iloc`**을 이용

#### 3. 열단위 추출 : 컬럼명
**df[열 이름]** 
**df[[열 이름]]**
**df[[열 이름1, 열 이름2...]]**
**df[열 이름][행 이름]**

#### 4. 불리언 인덱싱
- True 또는 False의 배열을 통해 데이터를 필터링하는 연산

#### 5. 인덱스 설정
- 기본값: 0부터 시작하는 정수형 인덱스

**df.set_index(열 이름)** : 기존의 열 중 하나를 인덱스로 설정
**df.reset_index()** : 인덱스를 리셋하고 기존 인덱스를 열에 추가
**df.reindex(인덱스 배열)** : 인덱스의 재배열에 따라 기존 행데이터도 인덱스를 따라감


### 조작

#### 1. 데이터 추가, 삭제
- 특정 데이터 셋을 열로 추가
**df['새로운 열 이름'] = 리스트** : 특정 리스트를 데이터프레임에 추가
- 열을 삭제
**df.drop(labels, axis)** 

#### 2. 데이터 정렬
**df.sort_index()** : 인덱스를 기준으로 정렬
**df.sort_values(by="열 이름")** : 특정 열의 값을 기준으로 정렬

#### 3. 데이터 변환
- 특정 컬럼 내 데이터의 자료형을 변환
**df[열 이름] = df[열 이름].astype(바꿀 자료형)**

- 데이터의 값 자체 변환
**df.replace(바꾸고 싶은 값, 바꿀 값)**

- 전치(transpose)를 통해 데이터프레임의 행과 열을 서로 맞바꾸기
**df.transpose()**
**df.T**

#### 4. 데이터프레임 병합
1. concat
- 단순히 두 데이터 프레임을 이어붙이는 방식

2. merge
- 공통 항목을 기준으로 병합하는 방식


## 집계함수
### 집계함수 개념
- 여러 행으로부터 하나의 결괏값을 반환하는 함수
- 각 컬럼의 평균, 표준편차, 합계 등.. 연산을 통해 대표값을 구할 때 주로 사용
- 결측치가 있을 경우 결측치는 무시하고 계산

**df['컬럼 명'].mean()** : 평균
**df['컬럼 명'].median()** : 중앙값
**df['컬럼 명'].std()** : 표준편차
**df['컬럼 명'].sum()** : 합계
**df['컬럼 명'].max()** : 최대값
**df['컬럼 명'].min()** : 최소값

### 그룹화


- groupby 메소드는 특정 기준으로 그룹을 묶고, 연산 처리를 가능하도록 함
- **집계함수와 함께 사용**

**df.groupby('column명1')[’column명2'].집계함수()**
 - column1로 그룹화를 하고, 그룹 별로 column2에 대해 집계함수를 적용해 결과를 시리즈 형태로 반환

**df.groupby('column명1')[[’column명2']].집계함수()**
 - 그룹 별로 column2에 대해 집계함수를 적용해 결과를 데이터프레임 형태로 반환
