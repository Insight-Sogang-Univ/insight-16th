### 0. 데이터 분석 프로세스


- 데이터 수집
    - 정형/비정형/반정형 데이터 수집

- 데이터 탐색(EDA)
    - 다양한 각도에서 수집한 데이터의 특징을 파악

- 데이터 전처리
    - 데이터를 분석에 적합한 형태로 손질.
    - ex. 이상치 제거, 결측치 처리, 노이즈 제거 => 데이터의 왜곡 피하고, 데이터의 신뢰도를 높임. 

- 데이터 모델링
    - 유용한 정보를 추출하기 위한 모델을 구축하는 단계
    - 예측, 분류, 군집화 등의 목적에 따라 모델을 선택하고 학습시키는 과정

### 1. EDA 개념

Exploratory Data Analysis의 약자. 데이터를 알아보는 모든 과정을 말한다.

- EDA 대상

    - 일변량: 파악하고자 하는 변수가 하나

    - 다변량: 여러 변수 간 관계를 보기 위해.

- EDA 종류

    - 시각화: 차트 또는 그림으로 파악
        - ex. 파이차트, 막대그래프, 히스토그램, 박스플롯,  산점도 등

    - 수치화: 비시각화. 그래픽적 요소 사용 X.
        - ex. 빈도표, 기술 통계량, 교차표, 상관계수

- EDA 유형

    - 대상과 종류에 따라 유형이 구분.

    - 주요 질문 통해 유형을 결정 가능.
        1. 데이터 변수가 몇 개인지?
        2. 결과를 어떻게 파악할 것인지?
        3. 데이터 유형은 무엇인지?


### 2. 데이터 확인하기
1. 데이터 읽기와 쓰기
- 데이터를 불러오는 것 = 읽기
    ex. read_csv, read_excel
    
- 데이터를 만드는 것 = 쓰기
    ex. to_csv, to_excel
    
2. 절대경로와 상대경로
- 절대: 처음부터 시작하여 목적지까지 전체적인 경로
- 상대: 현재 작업하고 있는 디렉터리 기준으로 상대적으로 표현되는 경로

3. CSV와 EXCEL
- CSV 파일: comma separated value의 약자로 데이터를 쉼표(,)로 구분하고 있는 텍스트 파일
    - 데이터의 크기가 작고 압축이 용이해 가장 널리 사용되는 데이터 형식!
- EXCEL 파일: 행과 열이 데이터 프레임의 행, 열로 일대일 대응
    - 특정 시트, 여러 시트를 불러올 수 있음

=> 두 형식 모두 판다스를 통해 데이터프레임 형식으로 읽어올 수 있음.

### 3. 데이터 셋 파악하기
1. 데이터 프레임 보기
- head(), tail() 등을 이용!

2. 데이터 변수 확인
- 변수 = DataFrame의 column = feature
- 변수가 범주형인지 수치형인지 구분

3. 데이터 분포 확인
- 날짜, 수치, 텍스트, 이미지 등의 구분
- 단변수 분석: 원시데이터의 평균값, 최빈값, 중간값 등의 변수 분포 -> 산포도, 박스 플롯, 히스토그램 등의 그래프로 데이터 분포를 확인&분석.

### 4. 전처리
- 데이터를 분석에 적합한 형태로 만드는 과정
- 머신러닝 등 데이터 분석의 정확도는 데이터의 품질에 의해 결정.
- 이상값, 결측값, 노이즈를 수정하고 목적에 맞게 변형하는 것이 중요.
- 데이터 정제, 변환, 필터링, 통합

### 5. 결측값 처리
데이터 수집 과정에서 측정되지 않거나 누락된 데이터인 결측값을 확인, 삭제, 대체

- 삭제 
- dropna(): 데이터프레임 내 결측값이 포함된 레이블을 제거하는 매서드
    - 기준: axis =0 & axis =1
    - 조건:
        - how = 'any': 결측값 하나라도 있으면 삭제 (default)
        - how = 'all': 모든 값이 결측값일때만 삭제
    - 특정 열을 기준으로 삭제: subset = ['column명']
    - 원본데이터 변경여부: inplace = False & inplace = True
    - 삭제 시 주의 점: 결측값이 무작위로 발생한 경우에 사용
- 대체: 결측값을 다른 값으로 대체
    - fillna()

### 6. 이상치 처리
관측 범위 내에서 많이 벗어난 값인 이상치를 확인, 삭제, 대체 

- 이상치 확인
    - 통계를 통해 확인: describe()
    - 시각화를 통해 확인: boxplot()
    - Z-score을 통해 확인
        - 평균(0)과 표준편차(1)로 정규화하여 평균으로부터 얼마나 떨어져 있는지
        - 평균에 가까울수록 0에 가깝고 멀어질수록 Z-score 커짐
        - 특정 값(일반적으로 2~4) 넘어가면 이상치로 간주
    - Tukey Fences: 4분위 범위를 기반으로, 두 가지 경우에 이상치라고 판단
- 이상치 제거 및 처리
    - 전체 삭제
    - 다른 값으로 대체
    - 변수화
    - 리샘플링

### 7. 피처 엔지니어링(변수 가공)
- 컴퓨터가 문제를 알아들을 수 있도록 변수를 가공(변형 or 적절히 처리)
- 도메인 지식 + 기존의 변수 사용: 도메인 지식 여부에 따라 결과와 유의성이 달라지기도 함
    - 레이블인코딩 vs 원핫인코딩: 문자열 값을 숫자형으로 변경하는 작업
    - 구간화: 연속적인 값을 일정한 구간으로 구분
    - 파생변수 생성: 기존의 피처를 변환해 새로운 피처 생성
    - 스케일링: 서로 다른 변수의 값 범위를 일정한 수준으로 맞추는 작업

### 8. 데이터 시각화
그림, 도형 등 그래픽적 요소 이용해 데이터를 표현하는 과정

- Matplotlib, Seaborn 라이브러리를 주로 사용
- 파라미터: 프로그래밍된 함수의 입력값
    ex. 코드 data = data, x = 'age'

    - data와 x는 매개변수
    - data와 'age'는 실제 전달되는 인자
    => data로서 data(데이터 프레임이 저장된 변수)를 활용하고 x축에 표현할 칼럼을 'age'로 지정.
    
    - 파라미터 hue: 범주형 변수를 넣어 내가 원하는 변수를 기준으로 데이터를 구분해 그래프에 표시

### 9. 다양한 그래프

1. boxplot
2. countplot(범주형 변수의 빈도수 확인하는 막대그래프), histplot(도수분포표를 그래프로 나타냄)
3. displot, kdeplot (커널밀도추정 그래프, 히스토그램을 연속적으로 곡선으로 연결)
4. barplot(x에 따른 y의 평균값 제시하는 막대그래프), pointplot(꺾은선 그래프) 
5. scatterplot(산점도 그래프), regplot(회귀선이 추가된 산점도 그래프)
6. catplot (수치형 데이터와 범주형 데이터의 관계를 보는 그래프)
7. pieplot(데이터 부분과 전체 간의 비율을 표현하는 원 그래프)
8. heatmap(변수 간 상관계수를 직관적으로 볼 수 있는 그래프)
9. violinplot(박스 플롯 + 커널밀도추정 그래프를 합친 그래프)
10. 여러 그래프 한번에 찍기, pairplot
