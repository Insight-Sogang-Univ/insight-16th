###  통계
- 숫자로 표현된 모든 사실
- 더 나은 의사결정의 바탕
- 통계학은 자료를 수집, 분석, 표현, 해석하는 기술이자 과학
- 모집단을 전부 조사하는 것은 비효율적/비현실적 -> 표본집단(샘플)만 추출해서 분석
- 샘플 자체에 대한 관심 -> 기술통계 ... EDA 
    - ex. 시험점수 분포 
- 모집단에 대한 관심, 모집단 추론 -> 추론통계 ... 회귀, 분류
    - ex. 시청률 조사 

### 자료의 척도
- 자료: 설명과 분석을 위해 수집되는 사실과 수치
    - 양적자료(수치로 표현) 
        - 구간(등간) 척도, 비율척도
    - 질적자료(수치로 표현 가능, 연산의 의미X)
        - 명목척도, 서열척도
- 측정척도는 자료 속 정보의 양, 요약 방법, 통계 방법을 결정  <BN>
- 자료의 종류: 데이터의 성격을 파악하는 큰 그림
- 척도: 그 데이터를 다루기 위한 더욱 구체적인 규칙과 기준

### 기술 통계
- 샘플에 관심 가지는 통계 방법
- 데이터를 있는 그대로 설명하고 이해하는 데 중점
    - ex. 우리 회사의 평균 매출, 제품 A의 불량률
- 3가지 주요 도구 집합
    - 중심화 경향 
        - 평균 : 이상치에 민감할 수 있음 
        - 중앙값: 평균보다 이상치에 덜 민감, 데이터의 중심경향성을 나타내는 대안 지표
        - 최빈값: 이상형 데이터, 데이터의 분포 이해하는 데 도움
    - 자료의 산포
        - 분산: 분산값이 크면 데이터가 많이 퍼져있음을 의미
        - 표준편차:  분산의 제곱근 취한 값으로 제곱 cm을 cm로 변환을 했다고 생각하자
        - 범위: 최대와 최소의 차이로, 이상치에 민감
        - 사분위수: 표본을 네 개의 동일한 크기로 나눈 값
        - 변동계수: 데이터의 상대적인 변동성을 나타냄
    - 자료의 정리 및 시각화
        - 히스토그램, 박스플롯, 파이차트...

### 추론통계(통계적 추론)
- 추론통계학은 의심과 도전을 견디며 정교한 논리체계로 발전
- 확률분포: 확률변수가 특정한 값을 가질 확률을 나타내는 함수
    - 확률 변수: 렌덤 세계에서 특정 사건이 발생할 경우의 수를 특정 값으로 대응시키는 함수
- 이산확률분포: 확률 변수가 이산적인 값을 가지는 경우
    - ex. 베르누이분포, 이항분포, 기하분포, 음이항분포, 초기하분포, 포아송분포
- 연속확률분포: 확률 변수가 연속적인 값을 가지는 경우
    - ex. 정규분포, T분포, F분포, 감마분포, 지수분포, 카이제곱분포, 균등분포


### 대표적인 분포 모형
- 정규분포 <BN>
평균 기준으로 대칭, 연속확률분포, 표준편차가 클수록 평평
    - 큰 수의 법칙: 표본 커질수록 표본평균이 모평균에 근사
    - 중심극한정리: 표본 30개 이상이면 표본평균의 분포는 근사적으로 정규분포를 따름
- Z 분포(표준정규분포) <BN>
평균이 0이고 표준편차가 1이 되도록 맞춘 표준적인 정규분포

T분포, 카이제곱분포, F분포는 정규분포를 따르는 모집단에서 추출한 표본 통계량들이 따르는 표본분포 <BN>
정규분포의 통계적 성질을 이용해 가설 검정, 구간 추청같은 통계적 추론을 하기위해 만들어진 도구

- T분포 <BN>
자유도가 커짐에 따라 표준정규분포에 수렴, 표본으로 모평균을 추정할 때 사용되는 분포. 
    - 자유도는 표본크기에 비례해 자유도 높을수록 정규분포에 가까워지고 통계분석 정확도 향상
    - 그러나, 너무 많은 자유도를 가지면 과적합 문제 발생(모델이 학습 데이터에 맞추어져 새로운 데이터에 대한 학습력 저하)
- T 분포 VS Z 분포
    - T분포는 표준정규분포보다는 더 중심으로부터 많이 퍼져있음.
    - 표본이 30미만이면 T분포, 30이상이면 Z분포
    - T분포는 분산을 모를때 표분으로부터 모분산을 추정하여 만들어지기 때문에, 표본이 작으면 추정의 불확실이 커짐. 
    - Z분포는 모집단의 평균과 분산을 정확히 알 때 사용. 모집단에서 직접 계산 -> 불확실 적음
- 카이제곱분포
    - 0에서 시작하여 오른쪽으로 긴 꼬리를 가진비대칭 분포
    - 독립적인 표준 정규 분포를 따르는 변수들의 제곱 합
    - 표본분산의 분포와 관련된 지표
- F분포
    - 두 독립적인 카이 제곱 분포의 비율
    - 두 집단 간의 분산 차이를 비교하는 데 사용
<BN>
- 분포 모형을 알아두어야 하는 이유
    - 가설검정을 위한 검정통계량 산출에 사용
    - 표본추출의 과정에서 인위적 조작의 여부를 확인

### 추론 통계 - 추정
수집한 데이터를 기반으로 미래의, 혹은 조사하지 못한 전체 정보를 추론하고 결론을 도출하는 방법을 다루는 분야. ex. 통계적 가설검정과 신뢰구간
- 추정: 표본으로 모수를 예측하는 과정 
    - 점추정: 어떤 모수를 표본에 기초하여 계산된 하나의 값으로 추정
    - 구간추정: 모수의 참 값이 포함되리라고 기대하는 추정치를 이렁한 범위와 신뢰도를 가지고 나타내는 방법
- 검정: 표본 통계량으로 모수에 대한 가설과 예상이 옳다고 판단할 수 있는지를 평가하는 과정


### 추론 통계 - 검정
- 가설 검정의 절차: 가설 검정 -> 유의 수준 설정 -> 검정 통계량 산출(Z,T,F값) -> 가설 기각/채택 판단(유의 확률 P-value)
- 데이터의 자료형에 따른 검정 방법:
    - 1) [수치형-수치형] 피어슨 상관분석
    - 2) [범주형-수치형] 범주 2개: F-test(ANOVA), 범주 3개 이상: T-test
    - 3) [범주형-범주형] Chi2-test
    - 4) [수치형-범주형]
