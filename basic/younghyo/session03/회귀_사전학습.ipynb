### 머신러닝

머신러닝은 인공지능의 한 분야.
데이터에서 패턴 발견 -> 예측 
머신러닝의 종류 
|종류| 정답 유무 | 특징 | 예시 |
| --- | --- | --- | --- |
|지도학습|정답 O|기계 예측 값 = 데이터의 정답|ex. 회귀, 분류|
|비지도 학습|정답X|데이터의 패턴 유사성 학습|ex. 군집화| 
|강화 학습|보상O|시행착오 반복화 보상 최대화, 벌 최소화|ex. 알파고, 자율주행|


모델: 학습의 결과물 <BR>
모델링: 공식을 만들어가는 전체적인 과정 <BR>
모델링의 과정: 설계 > 훈련 > 검증

회귀: 주어진 데이터 X를 기반으로 정답 Y를 잘 맞히는 함수(선)을 찾는 문제

분류: 기존 데이터가 어떤 레이블에 속하는지 패턴을 알고리즘으로 인지한 뒤 새롭게 관측된 데이터에 대한 레이브를 판별

회귀와 분류의 차이?
예측하고자 하는 값(종속변수)가 
- 연속형이면 회귀
- 범주형이면 분류

최소제곱법: 잔차의 제곱의 합이 최소가 되는 지점으로 최적의 회귀선을 구하는 방식
경사하강법: 함수의 값이 낮아지는 방향으로 이동하여 손실 함수의 최솟값을 찾는 방식

### 선형회귀(단순, 다중)
- 데이터를 잘 설명하는 선을 찾는 과정
- 현재 데이터 요약 + 미래 값 예측
- 단순선형회귀 (x값 한 개) / 다중선형회귀 (x값 여러 개)
    - 설명변수 개수에 차이가 있을 뿐, 두 회귀 분석은 본질적으로 같다

- 다중선형회귀의 경우 기본 가정 6가지 만족 하지 않으면 신뢰성 하락, 심각한 경우 분석 불가

1. 선형성: 종속변수와 설명변수 간의 관계가 선형적이어야 한다. 
2. 독립성: 각각의 설명변수가 서로 선형독립적이어야 한다. if not -> 다중공선성이 있다고 함
3. 오차항의 평균은 0에 가까워야 한다: 실제 값과 예측 값 사이의 평균은 0에 가까워야 한다.
4. 등분산성: 오차항의 분산이 일정해야 한다. 
5. 오차항은 자기상관되어 있지 않다 = 오차항의 공분산은 항상 0이어야 된다.
6. 정규성: 오차항이 정규분포를 따른다는 가정

### 다중공선성
- 독립 변수들 간의 강한 상관관계로 인해 회귀 모델이 불안정해지는 문제
- 회귀계수 추정이 부정확해지고 해석력이 떨어질 수 있어 선형회귀에서 중요한 체크 사항임.
    - ex. 주택의 크기와 방의 개수가 강한 상관관계를 가진다면, 두 변수들끼리 겹치는 상황이 발생함.
- **다중공선성이 높으면 어떤 독립 변수가 종속 변수에 얼마나 영항을 미치는지 구분하기 어렵기 때문에, 확인하고 처리하는 과정이 꼭 필요**

**<확인 방법>**
1. 상관계수 
2. VIF 지수 (분산 팽창 지수)

**<대체 방법>**
1. 변수 제거
2. 변수 변환: 변수들을 더하거나 빼서 새로운 변수
3. 규제선형모델 활용
4. PCA(주성분 분석) - 데이터의 차원을 축소하는 통계적 기법

### 규제선형모델

Problem: 과적합된 모델은 학습 데이터에는 완벽히 맞지만, 새로운 데이터에서는 성능이 크게 떨어짐.

=> 모델이 과적합되지 않도록 규제를 가하고자 등장한 모델
    - 회귀계수에 패널티를 부여하는 규제선형모델을 사용
    - 방식에 따라 3가지 모델이 존재

규제선형모델의 종류
L2 규제: 회귀계수의 제곱에 대해 패널티 부여, **릿지 회귀** 
- 회귀계수 제곱을 패널티로 줘, 큰 회귀계수일수록 훨씬 강한 패널티 부여. 모든 변수를 유지하면서 크기를 적절히 줄여 과적합 방지
L1 규제: 회귀계수의 절댓값에 대해 패널티 부여, **라쏘 회귀**
- 회귀계수의 크기와 상관없이 같은 패널티 부여
L2 규제 + L1규제: **엘라스틱넷 회귀**
- 불필요한 변수를 제거하고, 남은 변수들을 적절한 크기로 유지. 




### 모델 평가 방법

- <시각화>

    - x와 y에 대한 회귀선 시각화. but 어떤 회귀선이 해당 데이터를 잘 요약하는지 객관적인 비교가 어려운 경우가 있음 -> 통계지표 활용 

- <통계지표>

    - (1) 모델 유의성 검정
        - F 검정: 모델 전체에 대한 평가
            - F값이 클수록 통계적으로 유의미. Prob(F)이 0.05보다 작다면 회귀 모델 유의미 
        - T 검정: 개별 변수의 유의성에 대한 평가 
            - P> |t|가 0.05보다 작으면 해당 독립변수는 종속변수에 유의미한 영향을 미침

    - (2) 모델의 성능 평가
        - 결정 계수
        - 조정된 결정 계수
        - AIC< BIC

- <성능평가 지표>
    - 1) 평균 제곱 오차 (MSE)
        - 예측값과 실제값 사이의 오차 제곱의 평균
    - 2) 평균 절대 오차 (MAE)
        - 예측값과 실제값 사이의 절대값 오차 제곱의 평균
    - 3) MSE와 MAE의 차이 
        - MSE가 큰 오차에 민감하게 반응 -> 이상치에 민감
        - MAE는 모든 오차를 동일하게 취급 -> 이상치 존재에도 불구 전반적인 예측 성능 유지 가능

- <변수 유의성 평가>
    - T-검정: 독립변수와 회귀계수가 유의미한지 검정하는 데 사용
        - t값 계산 -> p값 확인 및 판단 

### 비선형회귀
- 다항식 회귀모델: 독립변수에 거듭제곱 항을 추가하여 데이터의 곡선적 특성을 모델링
- 지수 회귀모델: 종속 변수가 지수적으로 변화하는 관계를 모델링할 때 사용 가능
- 로그 회귀모델: 인구 성장, 감염병의 확산 같이 데이터가 초기에 빠르게 증가하고 점진적으로 감소를 보이는 데이터에 적합
- 스플라인 회귀모델: 데이터 패턴이 여러 구간에서 서로 다르게 나타날 때 유용 ex. 계절에 따라 판매량이 변하는 소매 데이터 분석 등에 적합
