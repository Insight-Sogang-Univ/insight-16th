### 6. 파이썬 변수
파이썬은 다음처럼 동일한 값에 여러 개의 변수를 선언할 수 있어요.  
다음과 같이 a, b 변수를 선언한 후 a의 두 번째 요솟값을 변경하면 b 값은 어떻게 되나요? 그리고 왜 이런 결과가 나오는지도 작성해봅시다! (검색 가능)
<br>
> a = b = [1,2,3]  
> a[1] = 4  
> print(b)  

```python
# 여기에 코드를 따라 쓰고 결괏값을 확인해 보세요.
a = b = [1,2,3]
a[1] = 4
print(b)
```

왜 위와 같은 결과가 나왔나요?<br>
답변: a와 b가 하나의 리스트를 가리키고 있기에 a의 리스트 내용이 바뀌면 b의 리스트도 바뀐다.
