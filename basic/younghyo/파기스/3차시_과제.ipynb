#### 1-(3) 메서드 만들기

```python
# 여기에 코드를 작성해 주세요.

class Membership:
    
    def __init__(self, name, tier, age):
        self.name = name
        self.tier = tier
        self.age = age
    
    def fare(self):
        if self.tier == "VIP" :
            return "10% 할인"
        elif self.tier == "Gold":
            return "5% 할인"
        elif self.tier == "Silver":
            return "3% 할인"
        else:
            return "1% 할인"
            
        
```
