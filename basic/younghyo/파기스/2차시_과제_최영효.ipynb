### 1. 시험 점수에 따라 등급 출력하기
시험 점수가 90점 이상이면 A, 80점 이상이면 B, 70점 이상이면 C, 60점 이상이면 D, 그 아래면 F를 출력하는 코드를 작성해 보세요.  
힌트: 강의자료를 꼼꼼히 다시 읽어볼까요~?

````python
score = 77 # 이 숫자는 여러분이 마음껏 바꾸셔도 됩니다!

if score >= 90:
    print('A')
elif score >= 80:
    print('B')
elif score >= 70:
    print('C')
elif score >= 60:
    print('D')
else:
    print('F')
````

여기에 답을 써주세요.
오름차순으로 60부터 조건문이 쓰여지면 score가 77이지만 C가 아닌 D가 출력된다. elif는 위에서부터 순서대로 검사하다가 처음 참이 되는 조건이 나오면 멈추기 때문이다. 따라서 먼저 걸리면 뒤 조건은 검사조차 안 하기 때문에, 위 조건문에서는 내림차순으로 써야 한다.

### 3. 3의 배수의 합 구하기
while 문을 사용해서 1부터 1000까지의 자연수 중 3의 배수의 합을 구해봅시다.

````python
result = 0
i = 1
while i <= 1000:
    if (i%3==0):
        result += i
    i += 1
    
print(result)
````

### 4. 별 표시하기
while문을 사용하여 출력셀과 같이 별(*)을 표시하는 코드를 작성해 봅시다.
<br>
아래와 같이 나오면 됩니다!



````python
i = 0
while True:
    i += 1
    if (i>5): # i 값이 5를 초과하면 while 문을 벗어납니다.
        break
    print('*'*i) # i 값의 개수만큼 *을 출력합니다.
````

### 5. 평균 점수 구하기
파기스에 총 14명의 학생이 있습니다. 이 학생들의 파이썬 점수는 다음과 같아요.<br>
`[70,60,55,80,45,90,100,70,90,60,75,80,90,85]`

for문을 사용하여 파기스의 평균 점수를 구해봅시다.

````python
pagis = [70,60,55,80,45,90,100,70,90,60,75,80,90,85]
result = 0

for i in pagis:
    result = result + i
average = result/len(pagis)
print("파기스의 평균 점수:",average)
````

### 6. 오류 찾기
1부터 5까지 출력하는 코드를 다음과 같이 작성해봤다고 해봅시다. 이 코드는 맞는 코드일까요? 만약 틀린 코드라면 어떤 문제가 있는지 쓰고 코드를 고쳐보세요.
```
count = 0
while count < 5:
    print(count)

```

답안 : 이 코드는 **[ 틀렸습니다.]**  
틀렸다면 이유는?

````python
# count를 증가시키는 코드가 없어서 count가 0으로 유지된다. 
# 0은 5보다 작기 때문에 항상 참이되어 무한루프에 빠진다.

count = 0
while count < 5:
    count += 1
    print(count)
    
# 1부터 5까지만 출력하고 싶다면 증가연산자를 통해 제한 조건을 마련해야 한다. 
````

### 7. 모든 입력의 평균값 구하는 함수 만들기
앞서 5번에서 파기스의 점수 평균을 구하는 for문을 작성해 보았죠? 이번에는 매번 코드를 새로 작성할 필요 없이, 원하는 만큼 숫자를 입력하면 자동으로 평균을 계산해 반환하는 함수를 직접 만들고 호출해 봅시다!

````python
def avg_score(*args):
    if len(args) == 0:
        return 0
    
    result = 0
    for i in args:
        result += i
                                       
    return (result/len(args))
````

````python
print(avg_score(1,3,5,8))
print(avg_score(10,40,50,90))

# 직접 함수를 호출해 보세요!
print(avg_score(2,3,4))
````
