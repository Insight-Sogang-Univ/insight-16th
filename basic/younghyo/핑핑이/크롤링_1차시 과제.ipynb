```python
from getpass import getpass

import requests

api_key = getpass("DART Open API 인증키를 입력하세요: ").strip()

endpoint = "https://opendart.fss.or.kr/api/fnlttSinglAcnt.json"

params = {
    "crtfc_key": api_key,
    "corp_code": "00164779",  # SK하이닉스 DART 고유번호
    "bsns_year": "2024",
    "reprt_code": "11011",    # 사업보고서(연간)
}

response = requests.get(endpoint, params=params, timeout=20)

print("HTTP 상태 코드:", response.status_code)
data = response.json()
print("status:", data.get("status"), "| message:", data.get("message"))

for item in data.get("list", []):
    print(f"{item['account_nm']:>10} | 당기: {item['thstrm_amount']:>15} | 전기: {item['frmtrm_amount']:>15}")
```
