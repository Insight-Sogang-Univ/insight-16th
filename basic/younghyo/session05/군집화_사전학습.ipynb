## <span style="background-color:#FFE6E6"> 1. 군집화 </span>

### <span style="color:violet"> 1-1 머신러닝 </span>
- 머신러닝 (지도학습-회귀, 분류 / 비지도학습-군집화, 연관분석)
    - 비지도학습: 정답이 없으므로 에이전트가 해야하는 작업이 명확히 정의되지 않는다. -> 모델의 성능을 명확히 측정할 수 없다.<BR>
    표현 학습을 통해 데이터셋의 고유 패턴을 식별할 수 있다 -> 목표가 명확하게 정의된 궤적이고 좁은 범위의 문제를 풀기에는 지도학습보다 불리, but 새로운 특성을 찾아내는 목적을 수행하기엔 적합. <BR>

    - 비지도 학습이 적합한 케이스
        - 패턴이 알려지지 않은 경우
        - 패턴이 계속 변하는 경우
        - 열린 문제를 해결하고 지식을 일반화해야 하는 경우

    - 실제 사용 경우:
        - 고객 세그먼트 분석: 고객의 성향 데이터를 바탕으로 비슷한 그룹끼리 묶기
        - 이상 거래 탐지: 정상적인 거래 패턴을 군집으로 정의 -> 이외는 이상치로 간주
        - 자연어 처리: 텍스트 데이터를 군집화해 분석하여, 유용한 인사이트를 도출

### <span style="color:violet"> 1-2 군집화 </span>
- 군집화: 데이터를 비슷한 특성을 가진 그룹으로 나누는 비지도 학습 기법 

![alt text](image.png)


- 지도학습과 다른 점: 
    - 지도학습은 독립변수와 종속변수 간의 관계에서 패턴을 찾아 종속변수를 예측하는 구조. but 군집화는 종속변수 설정 않고, 데이터 내부의 패턴을 인식해 유사성을 가진 그룹으로 나눔 -> 목표 없이 고유한 패턴을 발견하기에 비지도 학습으로 분류 됨.

- 군집화의 목표:
    - <span style='background-color: #f1f8ff'>응집도(cohesion) 최대화: 같은 군집에 속하는 데이터끼리 최대한 비슷하게.</span>
    - <span style='background-color: #f1f8ff'>분리도(separation) 최대화: 서로 다른 군집은 최대한 분리.</span>

### <span style="color:violet"> 1-3 군집화 과정 </span>

- 군집화 4단계
    - 1. 피처 선택 또는 추출: 군집화에 사용할 데이터의 특징 선택
    - 2. 군집화 알고리즘 선택: 데이터의 특성과 목표에 따라 적합한 알고리즘 선택
    - 3. 군집 유효성 검증: 군집화가 얼마나 잘 되었는지 평가
    - 4. 결과 해석: 각 군집이 지닌 특성 분석 및 해석
    => 마음에 안 들면 3(평가) -> 1(피처선택) -> 2(알고리즘선택) 반복 가능.

## <span style="background-color:#FFE6E6"> 2. 군집화를 위한 데이터 준비 </span>

### <span style="color:violet"> 2-1 군집화의 주요 고려사항 </span>

- 여러 번의 반복 시도를 통해 최적의 결과를 찾아내는 과정

- 1. 변수 유형 이해
    - 피처 종류/특성 명확히 이해
    - 변수 종류 연속형 or 명목형 -> 적합 알고리즘 변화
- 2. 거리 / 유사도 정의와 측정
    - 군집화는 거리 또는 유사도를 기반으로 그룹을 형성
    - 변수 유형에 따라 적합한 거리 측정 방식이 다르기 때문에. 정의가 중요.
- 3. 차원 축소
    - 변수가 많아질수록 모델 복잡. 효율/성능 저하.
    - 유사한 변수 통합. 변수 제거. => 변수 개수 줄이기.

### <span style="color:violet"> 2-2 변수 유형 이해 </span>

	
|유형|거리 측정|알고리즘|예시|
|---|---|---|---|
|연속형|유클리디안, 맨하탄|k-means, 계층적 군집화|키, 몸무게, 나이, 소득|
|명목형|해밍, 자카드|K-modes, 계층적 군집화|성별, 지역, 직업|
|혼합형|Gower distance|K-prototypes, 계층적 군집화|연속형+명목형|

**거리는 "빼기가 말이 되는가"가 정한다.** <Br>
**알고리즘은 "평균이 말이 되는가"가 정한다.** <BR>
**연속형은 스케일링이 필수**

### <span style="color:violet"> 2-3 거리 / 유사도 정의와 측정 </span>

- 거리가 가까울수록 유사하고, 멀수록 다르다고 판단.

1. 유클리디안 거리 (Euclidean Distance) <BR>
$ d(\mathbf{x}, \mathbf{y}) = \sqrt{\sum_{i=1}^n (x_i - y_i)^2} $
    - 가장 일반적인 거리 측정 방식
    - 직선거리를 의미. K-means에서 기본적 사용.

2. 맨하탄 거리 (Manhattan Distance) <BR>
$ d(\mathbf{x}, \mathbf{y}) = \sum_{i=1}^n |x_i - y_i|$
    - 이상치에 덜 민감
    - 각 차원 차이의 절댓값의 합
   - 참고) 유클리디안 vs 맨하탄
   

3. 코사인 유사도 (Cosine Similarity) <br>
$ \text{similarity} = \frac{\mathbf{x} \cdot \mathbf{y}}{||\mathbf{x}|| \times ||\mathbf{y}||} $
    - 벡터 간의 각도를 측정
    - 텍스트 데이터나 고차원 데이터에 효과적

### <span style="color:violet"> 2-4 차원 축소 </span>
- 군집화 알고리즘은 거리 기반 알고리즘인 경우가 많음.
- 데이터의 차원이 높아질수록(피처가 많아질수록) 차원의 저주 발생
- 차원의 저주란? 피처 공간이 너무 커서 알고리즘이 데이터를 효율적으로 훈련 X.
    -> 고차원일수록 모든 점이 비슷하게 멀어져, 거리 기반 알고리즘의 분별력 감소.
    **피처는 많을수록 좋은 게 아니다. -> 오히려 줄일수록 군집이 선명 -> 차원 축소**

- 차원 = 변수의 개수 
- 차원 축소 = 변수의 개수를 줄임
- 목표: 고차원 데이터 -> 저차원 공간에 투영. 핵심 정보는 지킴 -> 노이즈를 줄여 패턴 학습에 유리해짐.
    - 선형 투영(선형 차원 축소)
        - ex. 주성분 분석(PCA), 특잇값 분해(SVD), 랜덤 투영 등
    - 매니폴드 학습(비선형 차원 축소): 유클리드 거리가 아닌 데이터 포인트들 사이의 곡선 거리를 고려하며 학습
        - ex. t-SNE, UMAP, lsomap 등'

- PCA: 다차원 데이터에서 가장 정보량이 큰 방향을 찾고, 그 방향을 기준으로 데이터를 투영하여 차원을 줄이는 것.
    - 데이터의 분산을 최대한 보존하며 차원 줄임.
    - 상관관계가 높은 피처들을 결합해, 상관관계가 없는 더 적은 수의 피처로 데이터 표현.
    - 새로 파생된 성분 = 주성분
- PCA 주의사항:
    - 1. 표준화 필수: PCA 사용 전 데이터 표준화 필수
    - 2. 결측값 처리: 결측값 있으면 사용 불가
    - 3. 해석의 어려움: 주성분은 원본 특성들의 조합이라 직관적 해석 어렵
    - 4. 정보 손실: 차원 축소 과정에서 일부 정보 손실

## <span style="background-color:#FFE6E6"> 3. 군집화 알고리즘 </span>
- 알고리즘 선택 부분

### <span style="color:violet"> 3-1 계층적 군집화 </span>
- 가장 가까운 것부터 둘 씩 차례로 합쳐 나가면서, 무엇이 언제 뭉쳤는지를 나무 모양(덴드로그램)으로 기록하는 군집화 방법
- 반에서 친한 둘 -> 하나의 팀 -> 친한 팀 둘을 합침 -> 반 전체가 한 팀 될 때까지 반복.
- 기록: 누가 누구와, 몇번째로, 얼마나 친할 때 합쳐졌는지 기록이 남음 => 덴드로그램



- 트리 구조 (상향/하향)

- 상향식-응집형 계층적 군집화(agglomerative hierarchical clustering, 또는 병합 클러스터링)
    - 독립적 여러 클러스터들 -> 하나만 남을 때까지 가장 가까운 클러스터를 병함.
- 하향식-분리형 계층적 군집화(divisive hierarchical clustering, 또는 분할 클러스터링)
     - 전체포함 하나의 클러스터 -> 유사성이 낮은 데이터를 더 작은 클러스터로 나눔.

- 응집형 군집화 과정:
    - 거리 행렬 계산
    - 단일 클러스터 시작
    - 가장 가까운 클러스터 병합
    - 거리 행렬 업데이트
    - 반복

- [6가지 연결 방법]: 두 군집 간의 거리를 계산하는 방식.

1. Single linkage: 가장 가까운 두 점 기준. 고립 군집을 잘 찾음. but 이상치 취약.
2. complete linkage: 가장 먼 두 점 기준. 내부 응집성 중시. but 이상치 민감.
3. average: 모든 점 쌍의 평균 거리. 이상치에 덜 민감. but 계산량이 많음.
4. centroid: 두 군집 중심점 사이의 거리. 덴드로그램에서 inversion이 생길 수 있음.
5. median: 두 중심점의 정중앙을 대표점으로 씀. 군집 크기를 무시한다는 점이 centroid와 다름.
6. ward: 병합 후 SSE 증가분이 가장 작은 쌍부터 합침. 실무 기본값.

### <span style="color:violet"> 3-2 K-means </span>
- 정해진 개수(K)로 나누되, 각 군집 중심점(centroid)에 가장 가까운 데이터끼리 묶는 알고리즘. 군집을 중심점(평균) 하나로 대표하므로 평균을 못 구하는 데이터에는 쓸 수 없음.

- 주요 단계

1. 랜덤하게 k개의 중심점 선택
2. 각 샘플을 가장 가까운 중심점에 할당
3. 각 군집 샘플의 평균으로 중심점 이동
4. 할당이 안 변하거나 max_iter에 도달할 때까지 2~3 반복


- [ 장단점 ]

| 장점 | 단점 |
| --- | --- |
| 직관적, 구현 쉬움 | 초기 centroid에 민감 (→ k-means++) |
| 대용량 가능 | k 결정이 어려움 (→ 엘보우, 실루엣) |
| | 평균 기반이라 이상치 하나가 중심을 왜곡|
| | 길쭉하거나 초승달 모양 군집은 실패 |

- [ 엘보우 방법 ]: k를 바꿔가며 k-means를 돌리고 **SSE를 그래프**로 그려 최적 k를 추정하는 방법.

    - 급격히 꺾이는 지점(팔꿈치)이 최적 k. 그 이후로는 k를 늘려도 SSE 감소가 미미.
    - 장점: 직관적 / 단점: 완만하면 **주관 개입**, 팔꿈치가 안 보이기도 함 → **실루엣 계수와 함께** 보아야 함.


### <span style="color:violet"> 3-3 DBSCAN </span>

- 밀도가 높은 지역을 하나의 군집으로 묶고, 밀도 기준을 못 채우는 점은 군집에서 제외(이상치)하는 알고리즘.
- k-means의 약점(k 지정, 이상치 휘둘림, 원형만 잘 잡음)을 방어함. -> k 불필요, 이상치 분리, 모양 무관.

- 1단계: 샘플 분류

| 유형 | 조건 |
| --- | --- |
| **핵심(Core)** | 반경 ε 안 이웃이 MinPts 이상 |
| **경계(Border)** | 이웃은 부족하지만 다른 핵심 샘플의 ε 안 |
| **잡음(Noise)** | 나머지 전부 |

- 2단계: 핵심 샘플을 ε로 연결한 그룹을 클러스터로 생성 
- 3단계: 경계 샘플을 해당 핵심 샘플의 클러스터에 할당

- [ 장단점 ]

| 장점 | 단점 |
| --- | --- |
|임의의 기하학적 분포도 잘 처리| eps와 min_samples 설정에 많은 영향 받음(도메인 지식 필요) |
|이상치에 민감X| k 결정이 어려움 (→ 엘보우, 실루엣) |
| |연산량이 많아 속도 느림|
| | 밀도가 서로 다른 군집들은 잘 못 잡음|


### <span style="color:violet"> 3-4 Gaussian Mixture Model (GMM) </span>

- 데이터가 여러 가우시안 분포의 혼합이라고 가정하고, 각 분포를 하나의 클러스터로/로 보는 군집화 방법.

- k-means는 "1번 군집"이라고 딱 잘라 말하고, GMM은 "70% 1번, 30% 2번"처럼 확률로 답함.
- 각 군집을 **확률 분포**로 보고, 데이터 **생성 메커니즘**까지 모델링.

- 진행 과정: 전체 분포 확인 → 서로 다른 정규분포들로 구성됐다고 가정 → 정규분포 곡선들을 추출해 각 데이터의 소속을 결정 (각 분포가 하나의 군집)

- [ k-means vs GMM ]

| 구분 | k-means | GMM |
| --- | --- | --- |
| 기반 | 중심 | 확률 분포 |
| 할당 | 하나의 군집 (hard) | 확률 분산 (soft) |
| 장점 | 원형에 효과적, 빠름 | 길쭉하거나 겹친 분포도 처리 |
| 단점 | 길쭉한 분포에 약함 | 오래 걸림, 정규분포 가정 필요 |

### <span style="color:violet"> 3-5 알고리즘 선택 </span>
| 상황by상황| 선택 | 왜 |
| --- | --- | --- |
| 몇 개로 나눌지 감이 안 온다 | 계층적 | 덴드로그램 그리고 나중에 자름 |
| 대체로 동글동글하고 데이터가 크다 | k-means | 빠르고 직관적 |
| 길쭉하거나 서로 겹친다 | GMM | 타원형·중첩을 확률로 처리 |
| 이상치를 골라내는 게 목적 | DBSCAN | 이상치-1을 붙여 명시적 분리 |
| 초승달·소용돌이 모양 | DBSCAN | 밀도로만 판단 |
| 군집마다 밀도가 제각각 | HDBSCAN | 단일 밀도 한계 보완 |
| 경계 데이터의 애매함까지 알고 싶다 | GMM | predict_proba |


*주의사항: 이럴땐 쓰면 안됨

- **k-means**: 이상치가 많을 때 / 군집 크기가 심하게 불균형할 때
- **계층적**: 수만 건 이상
- **DBSCAN**: 피처가 수십 개 이상인 고차원
- **GMM**: 정규분포와 거리가 멀 때
- **전부**: 스케일링 안 한 데이터

## <span style="background-color:#FFE6E6"> 4. 군집화 평가 방법 </span>
### <span style="color:violet"> 4-1 외부 평가와 내부 평가 </span>

- 외부 평가: 
    - 데이터에 정답이 존재하는 경우 사용
    - 군집화 결과 vs 실제 레이블 -> 비교하여 얼마나 유사한지 평가
    - ARI: 군집화 결과와 실제 레이블 간의 일치도를 측정하는 대표 지표. 우연 배제.
- 내부 평가: 
    - 정답 레이블이 없는 경우에 사용 (주로 군집화)
    - 군집 내부 응집도와 군집 간 분리도를 기반으로 품질 정량화 -> 데이터 자체의 구조 기반으로 판단
    - 알고리즘별 내부 지표: k-means는 SSE 최소화, GMM은 로그 우도 최대화
    - 범용 지표: 실루엣 계수, 던 지수

### <span style="color:violet"> 4-2 실루엣 계수 </span>

- 클러스터 내 샘플들이 얼마나 조밀하게 모여 있는지로 군집 품질을 보는 지표.

$s^{(i)} = \frac{b^{(i)} - a^{(i)}}{\max\{b^{(i)}, a^{(i)}\}}$

- $a^{(i)}$ (**응집도**): 같은 클러스터 내 다른 점들과의 평균 거리
- $b^{(i)}$ (**분리도**): 가장 가까운 다른 군집 클러스터들과의 평균 거리

| 조건 | 값 | 의미 |
| --- | --- | --- |
| 분리도 > 응집도 | 1에 가까움 | 군집화 잘 됨 |
| 분리도 = 응집도 | 0에 가까움 | 경계에 위치, 모호 |
| 분리도 < 응집도 | -1에 가까움 | 잘못 할당됨 |
- 평균 0.5 이상이면 쓸만, 0.7 이상이면 잘 됨

### <span style="color:violet"> 4-3 Dunn Index </span>

- 클러스터 간 최소 거리(분리도)와 클러스터 내 최대 거리(응집도)의 비율

$\text{Dunn Index} = \frac{\min_{i \ne j} d(C_i, C_j)}{\max_{k} \delta(C_k)}$

- 클수록 좋음, 0 이상, 상한 없음
- 한계: 군집 수가 많으면 계산 비용 증가, 극단값(이상치)에 민감해 수치가 급락할 수 있음
=> 실루엣과 함께 종합 판단
