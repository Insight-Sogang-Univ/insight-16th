
### 회귀와 분류의 차이
- 지도학습: 회귀 vs 분류
- 회귀: 주어진 데이터를 기반으로 정답을 잘 맞추는 함수(선)를 찾는 문제
- 분류: 기존 데이터가 어떤 레이블에 속하는지 패턴을 알고리즘으로 인지한 뒤에 새 데이터를 예측 분류.
- 회귀는 연속형 변수를 예측, 분류는 범주형 변수를 예측

### 분류 모델의 네 가지 종류와, 각 모델이 무엇인지
- (1) 로지스틱 회귀: 독립 종속 변수 간의 선형 관계성에 기반해 분류. 
	- 이진 분류 문제
	- 분류 문제에는 직선보다 시그모이드 함수(S자 형태)가 효과적. (인공지능에 활용)
		- SVM: 클래스를 분류할 수 있는 다양한 경계선 중 최적의 라인을 찾아내는 알고리즘
- (2) 결정 트리: 데이터 균일도에 따른 규칙 기반의 분류
	- 의사결정 나무: 조건에 따라 데이터 분류-> 순수한 LABEL의 집합으로 구성될 때까지 분류 반복 like 스무고개
- (3) 서포트 벡터 머신: 개별 클래스 간의 최대 마진을 효과적으로 찾아 분류
	- 최적의 선을 찾는 법: 결정 경계는 거리가 가장 멀리 떨어진 경우를 선택함으로써 최적의 선을 찾음.
		- 장점: 데이터가 적을 때 효과적, but 단점: 데이터 전처리 과정이 중요
		- 장점: 연산량 최소화, but 단점: 데이터가 많아질수록 평과 과정이 많아져 속도가 느림.
		- 장점: 정확성 높음, 비선형 모델 분류 가능, 마진을 최대화는 방식이 규제역할을 하고 노이즈의 영향이 적음.
- (4) 최소 근접 알고리즘: 근접 거리 기준으로 분류
	- KNN: 비슷한 특성을 가진 데이터끼리 서로 가까이 있다는 점을 이용한 분류 알고리즘
		- 장점: 범주 나눈 기준 몰라도 분류 가능, but 단점: 다른 분류 모형과 달리 데이터 개수에 따라 시간이 오래 걸림
		- 장점: 학습 과정 필요없음, but 단점: 학습 데이터 모두를 거리 계산에 사용 -> 데이터 양이 시간에 영향을 미침.
 
### 분류 평가 지표
- (1) 혼동행렬
		-  분류 모델의 예측 결과를 정확한 예측과 잘못된 예측으로 구분하여 나타낸 행렬 <BR>
			- 예측결과X실제정답(True/False X True/False) 2X2 matrix <BR>
			- 결과 1: 예측 = 실제 <BR>
			- 결과 2: 예측 != 실제 <BR>
		- 혼동 행렬을 이용한 분류 모델 평가 지표 <BR>
			- 정확도: 모든 가능한 예측 중 참인 비율 <BR>
			- 정밀도: 참이라고 예측한 경우 실제 참의 비율 -> 스팸 메일 분류기 <BR>
			- 재현도: 실제로 참인 경우 중 참으로 예측하는 비율 -> 암환자 판단 <BR>
				- 정밀도와 재현도는 trade-off 관계: 경곗값을 조정하면 두 관계를 조정할 수 있다. 경곗값을 높이면 정밀도가 높아짐. 경곗값을 낮추면 재현도가 높아짐.

- (2) F1- Score
			- 정밀도와 재현율의 조화 평균
- (3) ROC/AUC
			- ROC Curve: 얼마나 분류가 잘 되었는가를 보여주는 그래프
			- AUC Curve: ROC와 x축 사이의 면적(적분값)
### 하이퍼파라미터 최적화
-  하이퍼 파라미터: 학습 시작 전에 사용자가 직접 설정하는 변수 
-  하이퍼 파라미터 최적화: 적절한 하이퍼파라미터를 찾아 모델 성능을 향상시키는 것
	-  최적화 과정: 파라미터 범위 설정 -> 평가 지표 계산 함수 정의 -> 샘플링한 파라미터 값을 사용하여 검증 데이터로 정확도 평가 -> 단계를 특정 횟수 반복하며 정확도 결과를 보고 파라미터의 범위를 좁힘
	-  최적화 방법:  
		- Grid Search: 정해진 범위에서 하이퍼파라미터를 모두 순회
			- 정확하지만 시간이 많이 걸림
		- Random Search: 정해진 범위에서 하이퍼파라미터를 무작위로 탐색
			- Grid보다 빠르지만, 정확성이 떨어짐
		- Bayseian Optimization: 사전정보를 바탕으로 하이퍼파라미터 값을 확률적으로 추정하며 탐색 
			- 가우시안 과정이라는 통계학 기반 모델로 가장 큰 값이 나올 확률이 높은 지점을 찾아냄
