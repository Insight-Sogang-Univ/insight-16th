### 분류
= 기존 데이터의 레이블 패턴을 알고리즘으로 인지 <br>
-> 새롭게 관측된 데이터의 레이블 판별<br>

= 범주형 변수 예측하기 위해 사용<br>

ML algorithm :  로지스틱회귀/결정트리/서포트벡터머신/최소근접알고리즘<br>

이진분류 : T/F <br>
ex) 5가 맞다 아니다<br>
다중분류 : 예측할 변수 값이 3개 이상<br>
ex) 1/2/3/4/5..<br>

### 로지스틱 회귀 (logistic regression)
- 이진 분류 -> 0~1의 값<br>
  ㄴ 이메일이 스팸인지 아닌지<br>
- 샘플이 특정 클래스에 속할 확률 추정 + 분류<br>

필요성) 이진분류를 직선(단순 선형회귀)으로 풀면 출력이 0~1을 벗어남 <br>
-> 로지스틱 회귀 사용<br>

사용 함수) 시그모이드 함수 <br>
- 출력: 0~1 -> 특정값 이상 :True / 이하: False
- 모양: S자
- 사용: sigmoid(wx+b) -> 적합한 w와 b 찾아야 됨


### 의사결정나무 (decision tree)
- 조건 -> 데이터 분류 -> label 집합

사용 알고리즘) 이진분할 알고리즘 (CART)<br>
- 임계값 기준 -> child 나눔 -> 지니지수 down (= 범주 섞임 down)
--> 분류 기준 잘 정립하는 게 중요 !! <br>

학습 시 고려할 점 <br>
- hyperparameter 설정
- 시각화 (-> 분류 잘 이뤄졌는 지)
- pruning (-> 불필요한 노드 지우기 -> 일반화 up)

### SVM (support vector machine)<br>
= 클래스 분류 경계선 중 최적의 선(초평면) 찾기<br>
- 고차원 공간에서도 효과적 
-> 활용) 패턴인식, 이미지 분석...<br>

구성)<br>
결졍 경계 : 데이터 분류 기준 경계<br>
서포트 벡터 : 결정 경계와 가장 가까운 데이터 집합<br>
마진 : 결정경계~서포트벡터 거리<br>
초평면 : n차원 공간의 n-1차원 평면<br>
슬랙 변수 : 완벽 분리 불가능 시 허용 오차 변수 for 분류<br>

최적의 초평면 찾는 법)<br>
마진이 가장 큰 경우 선택 <br>
-> 결정 경계까 데이터로부터 멀리 떨어진 거<br>

장점) <br>
데이터 적을 때 효과적<br>
새로운 데이터 입력 시 연산량 down<br>
정확도 up<br>
노이즈 영향 down<br>

단점) <br>
데이터 전처리 중요<br>
데이터 많아지면 속도 down<br>

### KNN(K-neareast neighbor)
- 비슷한 특성 가진 데이터끼리 가까이 있음 
-> k 개 이웃 설정 -> 그걸 기반으로 분류

과정)
1. 데이터 준비 (= 미리 학습 X)
2. k값 설정<br>
    ㄴ 보통 홀수
3. 거리 계산<br>
    ㄴ 새 데이터 & 기존 데이터간
4. 가장 가까운 k개 이웃 선택
5. 분류 <br>
    ㄴ 가장 많이 등장하는 클래스 = 예측 결과<br>

최적의 K : 너무 멀지도 작지도 않은.. 학습데이터 개수의 제곱근<br>

장점) <br>
범주의 기준 몰라도 분류 가능<br>
학습 필요 X<br>
원본 그대로 저장<br>

단점) <br>
개수 많으면 시간 up<br>
k 결정 어렵<br>
수치형 데이터 아니면 유사도 정의 어렵<br>
이상치 영향 up<br>

### 분류평가지표

### 혼동 행렬
= 정확한 예측/잘못된 예측으로 구분
(예측==실제 -> 파란색, 예측!=실제 -> 빨간색)

TP : 정답 참, 예측 참
TN : 정답 거짓, 예측 거짓
FP : 정답 거짓, 예측 참
FN : 정답 참, 예측 거짓

-> 분류평가지표 : 정확도/정밀도/재현도 
*precision-recall : trade-off*

- 정확도 : (TP+TN)/(TP+TN+FP+FN)
    ㄴ 단점: 정답 레이블 비율 불균형 -> 신뢰도 down
- 정밀도 : TP/TP+FP
    ㄴ 거짓을 참으로 판단한 정도 : 스팸 메일 분류기
- 재현도 : TP/TP+FN
    ㄴ 참을 거짓으로 판단한 정도 : 암 환자 판단

--> 정밀도&재현도 -> threshold 결정

### F1-score
= 정밀도&재현율 조화평균
f1-score= (2*p*r)/(p+r)

### ROC/AUC
ROC : 얼마나 분류가 잘되었는지
- TPR (=Recall) : 참 중 참으로 예측한 비율
- FPR : 거짓인 것 중 참이라고 예측한 비율

AUC : ROC와 x축 사이 면적
- 1에 가까울수록 분류 성능 good

### 하이퍼파라미터 최적화

### 하이퍼 파라미터<br>
= 학습 시작 전 + 사용자가 직접 설정하는 변수<br>
-> 최적화) 적절한 하.파 찾기 -> 모델 성능 향상<br>

1. 하.파 탐색 범위 설정<br>
2. 평가지표계산함수 정의<br>
3. 1에서 설정한 범위에서 하파값 사용 -> 정확도 평가<br>
4. 범위 좁히기<br>

방법) grid search / random search / bayesian optimization<br>

### grid search<br>
- 방법) 정해진 범위 모두 순회 -> 가장 좋은 거 택<br>
- 장점) 최적값을 정확히 찾음<br>
- 단점) 시간 마니..<br>
- 적용) 넓은 범위, 큰 step -> 범위 좁히기<br>

### random search <br>
- 방법) 범위 내 무자구이 탐색 -> 가장 좋은 거 택<br>
- 장점) 속도 : grid < random<br>
- 단점) 정확도 down : random < grid<br>

### bayesian optimization<br>
- 방법) 사전정보 -> 최적값 확률적으로 추정<br>
        무작위 실험 -> 좋은 조합 예측 -> 반영+예측 계속<br>
