### 연관규칙 조건
- 일정 이상의 지지도 (A, B 교집합) = 이 규칙이 현실에서 의미 있는 지<br>
ex: 품목 A와 B 가 함께 구매한 경우의 수가 일정 수준 이상<br>
지지도 low : 우연일 가능성.. + 실행 가능한 인사이트 도출 어려움<br>

- 일정 이상의 신뢰도 (A, B 교집합 / A) = A 일어났을 때, B도 일어날 것이라고 얼마나 확신할 수 있냐<br>
ex: 품목 A를 포함한 거래 중 품목 B를 구입하는 경우의 수가 일정 수준 이상<br>
신뢰도 low : 무의미 + 추천 실패 <br>

ex: <br>
나초-콜라 지지도 : 10% , 나초-콜라 신뢰도 : 75%<br>
지지도 꽤 있다 -> 분석할 가치 o<br>
나초사는 사람 중 75%가 콜라 산다거? -> 규칙 꽤 정확하네<br>

but, 걍 콜라 사는 사람이 85%고, 나초사면서 콜라사는 사람이 75%면 <br>
나초가 콜라 판매를 북돋는 게 아닌게 됨<br>

이런 경우엔? **향상도**가 필요!<br>

### 연관 분석의 평가 측도
- 향상도 (lift)<br>
: 신뢰도/B = (A교집합B)/(A)(B)<br>

A와 B가 독립이면 A살 때 B에 영향 안줌 = 향상도 1<br>
독립 아니면 A랑 B가 서로 영향 준 거 일 수도!<br>

P(A교집합B) < P(A)P(B) : A살 때 B도 잘 삼<br>
P(A교집합B) > P(A)P(B) : A사면 B는 잘 안삼<br>

향상도>1: 우연보다 높음<br>
향상도=1: 독립 (우연=노연관)<br>
향상도<1: 우연보다 낮음<br>

--> 규칙1이 규칙 2보다 효과적 : 지지도, 신뢰도, 향상도 모두 클 때<br>

### 연관 분석 알고리즘
**Apriori 알고리즘**
<br>
기존 문제 ) 모든 조합 -> 연관 규칙 찾기 : 비효율적 <br>
            ㄴ 상품 개수 많아지면 조합 수 upupupupup...<br>

설명 )<br>
상위 조합부터 스캔 -> 자주 발생 x 특정조합의 후속 조합은 배제<br>
= 말 안되는 조합 버리기 -> 계산량 down<br>
ex: {빵, 우유} 가 최소 지지도 만족 x -> {빵, 우유, 계란} 같은 건 검사할 필요 x<br>
ㄴ 최소지지도 = 하.파<br>

**연관 규칙을 찾기 위해 빈발 항목집합(최소지지도 넘는 항목들 조합)을 효율적으로 찾아줌**<br>
-> 더 이상 조합 생성 못할 때까지 반복하기<br>

장점) 많은 연관 패턴 발견 / 원리, 분석 easy <br>
단점) 데이터 커지면 속도 느려짐<br>
        ㄴ 후보 생성 -> 지지도 계산 -> 탈락 -> 후보 생성..<br>


**FP Growth 알고리즘**<br>

설명 )<br>

모든 거래 확인 - 아이템별 지지도 계산 - 최소 지지도 이상만 택<br>
