### **통계**
for 더 나은 의사결정

**통계**
<br>기술통계 -> 추리통계
<br>(표본 특성 분석) -> (일반화 여부 판단 -> 모집단 특성으로 추정)
<br>= 표본에서 모집단으로 확장해서 추론

**자료 척도**
**자료** (데이터의 본질적 성격 구분)
<br>= 요약되는 사실&수치 for 설명&해석


**측정척도** (데이터 측정 방식)
<br>-> 정보의 양 결정 -> 적합한 자료요약법 / 통계분석법 결정
- 질적 자료 (명목척도/서열척도) : 연산의미 X
<br>-- 명목척도 : 레이블/이름 -> 속성 식별 
<br>ex) 성별
<br>-- 서열척도 : 명목 + 순서/순위
<br>ex) 만족도

- 양적 자료 (구간척도/비율척도) : 연산의미 O
<br>-- 구간척도 : 값 간 순서+일정
<br>ex) 온도
<br>-- 비율척도 : 절대영점 + 임의의 측정단위 + 동간성
<br>ex) 거리 -> 0값이 의미를 가짐

### **기술통계**
: 데이터를 있는 그대로 설명 + 이해 -> 데이터 특성&패턴 보임
<br>ex) 올해 평균 매출


**주요 도구**
<br>1) 중심화 경향 
- 평균 : 이상치 민감 
<br>-> 절사 평균 : 평균의 단점 보완 (양끝 일정 비율 잘라냄)
- 중앙값 : 이상치 덜 민감
- 최빈값 : 가장 자주 나타나는 값 -> 데이터 분포 이해 도움

2) 자료의 산포
<br>= 데이터 퍼져있는 정도
- 분산 : 평균에서 멀리 떨어진 정도 -> 단위 제곱
- 표준편차 : 분산에 제곱근 -> 단위 일치
- 범위 : max-min -> 이상치에 매우 민감
- 사분위수 : 표본 사등분 (Q1, Q2, Q3) : quantile(f) 사용
- 사분위간 범위(IQR) : Q3-Q1제외 범위(tukey fence)
- 변동계수 : 상대적 변동성 
<br>ㄴex) 코끼리와 성인의 몸무게 숫자만 비교 X 비율 비교 O

3) 자료의 정리 & 시각화
- 히스토그램, 박스플롯, 파이차트

### **확률과 분포**
**추론통계**
<br>= 표본의 정보 -> 모수 추정/ 가설 검정
<br>표본 불확실성 - 표본분포
<br>모수추정 - 점추정
<br>추정의 현실성 - 구간추정
<br>구간추정의 오류고나리 - 신뢰수준
<br>적극적 주장검증 - 가설검정
<br>가설검정의 오류 관리 - 유의수준

**확률분포**
- 확률 변수 : 특정 사건에 발생할 수 있는 모든 경우의 수를 값에 대응
- 확률 분포 : 확률변수가 특정 값을 가질 확률 나타냄 (이산확률/연속확률)

**대표적인 분포**
- 정규분포 
<br>cf) 중심극한정리 (n>=30 일때)
- 표준정규분포(Z분포)
- T분포
<br>cf) 자유도, 자유도 증가 -> Z분포에 수렴, 과적합 주의
- 카이제곱분포
<br>cf) 자유도, 자유도 증가 -> 정규분포에 수렴
- F분포
<br>cf) 자유도, 자유도 증가 -> 정규분포에 수렴, 두 집단 간 분산 차이 비교

활용 ) 
- 검정통계량 산출 for 가설검정
- 표본 추출과정에서 인위적 조작 확인

### **추론통계**
<br>: 수집한 데이터 -> 전체 정보 추론+결론도출

**추론**
- 점추정 : 모수를 하나의 수치로 추정 (오차에 대한 정보고려X)
- 구간 추정 :모수 존재 예상 구간 추정
**추정량**
<br>: 표본정보에 의존, 표본통계량

**추정치**
<br>: 표본 -> 계산

**검정**
<br>: 표본통계량 -> 모수에 대한 예상/가설의 진위여부 판단 과정

**가설 검정**
<br>= 주장을 표본의 정보로 합당성 여부 판단

1) 가설 설정
<br>귀무가설(H0) = 기각하고자 하는 거 (~차이가 없다) 위험성 down
<br>대립가설(H1) = 주장하고자 하는 거 (~차이가 존재한다)

2) 유의 수준 설정
<br>: 허용가능 오류의 최대치
- 1종 오류 : 귀무 참인데 기각 확률 : 치명적 -> 유의수준
- 2종 오류 : 귀무 거짓인데 기각 못할 확률

3) 검정통계량 산출
<br>용도) 가설검정 결과 판단
- Z-score : 큰 표본 크기일때, 임계점 찾기 -> 임계점보다 크면 귀무 기각
- T-score : 작은 표본일 때, 표본평균 간 차이, 임계점 : T분포표 사용
- F-score: 두 표본 분산 차이, 1에 가까우면 두 분산 비슷

4) 가설 기각/채택 판단
- 유의 확률 : p<a(유의수준) -> 귀무기각


**데이터 자료형 - 검정법**
1) 수치-수치 : 피어슨 상관분석
<br>데이터 자유분방 = 시각화 효율 X 
<br>--> 상관분석 사용
<br>- 두 변수 간 피어슨 상관계수의 유의미 정도 검정
<br> **상관계수** != 인과계수
<br>= **선형적 관계**, 표준화됨, 수치적으로 연관되어있는 지
<br>ex) 아이스크림 판매량 - 날씨 - 바다 방문객수
<br>--> 상관계수는 날씨!
<br>+) 피어슨 상관계수+순위화 = 스피어만 상관계수

2) 범주-수치 : F-test, T-test
<br>시각화 :barplot + 신뢰구간
- T-test :범주 2개
<br>t 값 = 두 평균 차이 / 표준오차
<br>+) 두 그룹의 평균에 차이 없음 주장 불가
- F-test :범주 3개 이상 (ANOVA < F-test)
<br>= 집단 간 분산/집단 내 분산
<br>-> 평균 대비 그룹 간 차이 발견 가능

3) 범주-범주 : Chi2-test
<br>시각화 : 모자이크 플롯
<br>서로 연관되어 있는 지
<br>기대비도&관측빈도 : 차이큼 = 관련성 up
<br>+ 카이제곱분석 사용 : 자유도*2<= 면 유의미차이

4) 수치-범주 : X
<br>시각화 : kdeplot
