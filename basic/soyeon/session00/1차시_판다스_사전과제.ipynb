### **파이썬 라이브러리**

### NumPy

- 다차원 배열(ndarray)

### Pandas
- 데이터 분석 특화
- table (행, 열)
- 인덱싱, 그룹화 등
- 넘파이 확장판

### **시리즈**
- 1차원 구조 (index, value)
<br>= 데이터프레임의 특정 한 컬럼 추출
<br> ㄴ 사용 : df[열이름]
- default : 0 ~ 정수형 인덱스
- 시리즈 구성 값은 같은 데이터 타입
<br>    ㄴ object라고 무조건 타입 섞인 거 아님

사용 : 
<br>1. s = pd.Series(값, index=인덱스)
<br>2. s[열이름] -> 값
<br>3. s[[열이름]] -> 시리즈

### **데이터프레임**
-  2차원 배열 구조
<br>= 정보 정리&저장하는 도구
<br>= 데이터프레임 기본 구조
<br> = pandas 주요 데이터 구조

- 행(index), 열(index)
<br>= 데이터 위치 표시

- 시리즈들 = 데이터프레임
<br>컬럼 단위(시리즈) : 데이터 타입 동일
<br>컬럼 간(데이터프레임) : 달라도 ㄱㅊ

사용 : 
<br>1. 컬럼 별 데이터, 이름 담은 딕셔너리 생성
<br>2. df = pd.DataFrame(딕셔너리, index=인덱스)


### **데이터 확인**
데이터프레임 이해 -> 데이터 분석

**일부 미리 확인**
- head(), tail()
<br>앞/뒤 n개 데이터 확인 -> 데이터 구조/값 파악
<br>ex) df.head(n) : 앞 n개 행 반환 : default=5

**전체 특징 요약**
- df.shape -> (행개수, 열개수)
- df.info() -> 행개수, 열개수, non-null 개수, 데이터 타입..
- df.describe() -> 통계값 ex) mean, 25%, std, max..
- df.isnull() -> 데이터당 결측치 여부 반환 
<br>ㄴ df.isnull().sum() -> 컬럼 당 결측치 개수
- df.dtypes -> 열의 데이터 타입 반환

**고유값 및 개수 확인**
- df.count() -> 결측치 제외 데이터 행 수
- df[열이름].value_counts() -> 해당 열 속 값의 개수들 = 빈도수
- df[열이름].unique() -> 해당 열 속 값의 종류들



### **데이터 추출**
많은 양 데이터 -> 필요한 부분만 

**인덱싱(indexing)**
= 지목
<br>사용 : df[열이름] -> 시리즈
**슬라이싱(slicing)**
= 범위 지정
<br>사용 : df[start:end] -> start~end-1 행

**행단위 추출**
- loc
<br>기준 : index label (인덱스 이름)
<br>사용 : df.loc[행인덱스, 열이름] -> 행/값/열 추출
<br>끝 인덱스 포함 반환
<br>인덱싱&슬라이싱 가능

- iloc
<br>기준 : integer position (정수형 위치)
<br>사용 : df.iloc[행순서, 열순서] -> 행/값/열 추출
<br>빠름
<br>끝 인덱스 포함X 반환

**열단위 추출**
- df[열이름] -> 시리즈
- df[[열이름]] -> 데이터프레임+열 하나
- df[[열이름1, 열이름2..]] -> 데이터프레임
- df[열이름][행인덱스] -> 시리즈 속 값

**불리언 인덱싱**
특정 조건 만족 데이터 추출
<br>True / False 사용 필터링 연산
- df[조건문]
- df[(조건문1) 논리연산자 (조건문2)]
- df.loc[조건문, 열이름]
<br>*iloc은 정수형 순서 인덱스만 받음 = 불리언 인덱싱 불가*

**인덱스 설정**
- df.set_index(열이름) -> 기존 열 중 하나를 인덱스로
- df.reset_index() -> 인덱스 리셋 + 기존 인덱스는 열에 추가
- df.reindex(인덱스배열) -> 기존 인덱스 재배열+행도 / 새로운 인덱스면 값은 NaN
- df.index=인덱스배열 -> 인덱스 재배열+행은 그대로 
<br>ㄴ 조건 :기존 행 개수 = 새로운 행 개수 

### **데이터 조작**
**추가** 
- df['새로운열이름']=리스트
<br>ㄴ df에 추가됨

**삭제** 
- df.drop(labels, axis=방향, inplace=True/False)
<br>ㄴ default : axis=0, inplace=False
- del df[열이름]

**원본 바꾸기** 
- inplace 옵션 
- df=df.drop(~) : 객체 재할당

**데이터 정렬**
- df.sort_index(ascending=?) -> 기준: 인덱스
<br>ㄴ default: 오름차순
- df.sort_values(by="열이름) -> 기준: 해당 열

**데이터 변환**
- df[열이름]=df[열이름].astype(바꿀자료형)
<br>ㄴ재할당 해줘야 원본이 바뀜

- df.replace(바꾸고 싶은 값, 바꿀 값)
- df.transpose(), df.T -> 행, 열 맞바꿈
<br>ㄴ재할당 해줘야 원본이 바뀜

**데이터프레임 병합**
- concat
<br>두 데이터프레임 이어붙임
<br>행방향 결합 : df=pd.concat([df1,df2], axis=0)
<br>열방향 결합 : df=pd.concat([df1,df2], axis=1)

- merge
<br> 공통 항목을 기준으로 병합 (형태 달라도 가능)
<br> on : 기준될 열, how : 결합 방식
<br> 교집합 : df =pd.merge(df1,df2,how='inner', on=열이름)
<br> 합집합 : df =pd.merge(df1,df2,how='outer', on=열이름
)


### **집계함수**
결측치 무시
기본적으로 모든 행 대상 -> 대표값 구할때 사용

- 평균 : df[컬럼명].mean()
- 중앙값 : df[컬럼명].median()
- 표준편차 : df[컬럼명].std()
- 합계 : df[컬럼명].sum()
- 최대값 : df[컬럼명].max()
- 최소값 : df[컬럼명].min()


**그룹화(grouping)**
일부 행 대상 집계
ex) 노트북 브랜드 별 무게
- groupby() + 집계함수 : 특정 기준으로 그룹 묶고 연산 처리
- df.groupby(컬럼1)[컬럼2].집계함수()
<br>ㄴ컬럼1로 그룹화, 그룹별 컬럼2에 집계함수 적용, 시리즈로 반환

- df.groupby(컬럼1)[[컬럼2]].집계함수()
<br>ㄴ컬럼1로 그룹화, 그룹별 컬럼2에 집계함수 적용, df로 반환
