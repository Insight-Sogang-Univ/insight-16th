### **[1] EDA 개념**

**EDA**
<br>= Exploratory Data Analysis 
<br>= 탐색적 데이터 분석 과정

<br>데이터 파악 for 데이터 분석
- 시각화 -> 패턴 파악
- 특이성 확인
- 통계/그래프 등 -> 가설 검정

<br>**목적: 데이터 정확한 해석 + 올바른 문제 정의**

**EDA 대상**
- 일변량 (Univariate)
<br>한번에 파악 + 변수 1개
<br>**목적: 데이터 분포/패턴 파악**

- 다변량 (Multi-variate)
<br>한번에 파악 + 변수 n개
<br>**목적: 변수들간 관계 파악** (일변량 후 다변량 파악하기)

**EDA 종류**
- 시각화 (Graphic)
<br> ex) 차트, 그림..
<br> +) 한눈에 파악
- 비시각화 (Non-Graphic)
<br> ex) Summary Statistics(ex: .describe())
<br> +) 정확한 값 파악

**EDA 유형**
<br> EDA 대상 + 종류 --> 유형 구분
<br>ex) 데이터 변수 개수 / 결과 파악 방법 / 데이터 유형

- 일변량 비시각화 : 변수 1개 특징 with 숫자/통계량
- 일변량 시각화 : 변수 1개 분포 with 그래프
- 다변량 비시각화 : n개 변수 관계 with 숫자/통계량
- 다변량 시각화 : n개 변수 관계 with 그래프

### **데이터 셋 파악**
**1) 데이터 프레임 보기**
<br>ex) head(n)

**2) 데이터 변수 확인**
<br>변수= df의 열 = feature
- 변수 정의 : 어떤 정보 가짐?
- 변수 유형 : Categorical / Numerical
<br> 범주 (명목/순서), 수치 (이산/연속)
<br>*이산 = 일정 구간 잘랐을 때 셀 수 있는 지*

- 변수 자료형 확인 : 날짜, 수치, 이미지..
<br> ex) df.info()

**3) 데이터 분포 확인**
- 단변수 분석
<br>= 한번에 하나의 변수를 대상으로 그 변수의 분포/특징 확인

### **[3] 전처리**
<br>= 데이터를 분석에 적합한 형태로 만들기
<br>목적 : 
<br>이상값, 결측값, 노이즈 수정 + 목적에 맞게 변형
<br>-> 데이터 품질 up 
<br>-> 데이터 분석 정확도 up

**- 데이터 정제 (cleaning)**
<br>결측값 보완 + 이상값 제거
- 데이터 변환 (transformation)
<br>일관성 확보 + 중복 최소화 -> 분석 시간 down
- 데이터 필터링 (filtering)
<br>오류 발견 -> 삭제 + 보정 -> 데이터 품질 up
- 데이터 통합 (integration)
<br>유사 성질의 데이터 연계 등


**- 결측값 처리**
<br>= 데이터 수집 과정에서 측정 X / 누락된 데이터 처리
<br>목적) 변수 간 관계 왜곡 + 정확성 down + 시각적 표현 문제 발생 예방

**- 결측값 확인**
<br>NaN, ?, 0 등 으로 표현
<br>-> 진짜인지 아닌지 확인
<br>사용 함수) info, value_counts(), isnull(), notnull(), replace()

**- 결측값 처리**
- **삭제** : dropna()
<br>ㄴ 삭제 기준 axis=0/1
<br> ㄴ 삭제 조건 how='any'/'all'
<br>ㄴ 열 기준 subset=[컬럼명]
<br>ㄴ 원본 데이터 변경 여부 inplace=False/True

- 삭제 주의점
<br>패턴 있는 결측값은 섣불리 삭제 X
<br>= 정보 손실 가능성

- **대체** : fillna()
<br>대푯값 사용 (평균값, 최빈값 등)
<br>  ㄴ 일괄 대체 : 다 같은 값으로 대체
<br> ㄴ 유사 유형 대체 : 유사 범주에 따른 값으로 대체 
<br>(groupby사용하기)

**- 이상치 처리**
<br>= 관측 데이터 범위에서 많이 벗어난 값
<br>목적) 데이터 왜곡 예방

**1) 이상치 확인**
- 통계 : describe()
- 시각화 : boxplot
- z-score (=표준 점수) : 정규분포 따름 + 평균에서 멀수록 z-score up
- tukey fences : 사분위 범위(IQR) 기반 + 이상치 판단 기준
<br>ㄴ x< Q1-(1.5* IQR) || x>Q3+(1.5* IQR)

**2) 이상치 제거/처리**
- 전체 삭제 : human error의 경우
- 대체 : 절대적 관측치수가 적을 경우 -> human error여도 대체해주기
- 변수화 : 이상치 자연발생 -> 이상치 먼저 파악(모델에 포함 가능여부 등)
- 리샘플링 : 이상값 포함/제외 모델 만들기

<br>자연발생 이상치 접근법) 케이스 분리 + 분석

**- 피처 엔지니어링 (변수가공)**
<br>목적) 컴퓨터가 이해할 수 있게 + 피처 변형/처리

<br>도메인 지식 + 기존 변수 = 정보 추가

**1) 인코딩 (encoding)**
<br>문자열 -> 숫자형
- 레이블 인코딩
<br>범주형 변수 -> 0~N-1

- 원핫인코딩
<br>범주형 변수 -> 0/1 (이진벡터)
<br>사용) pd.get_dummies(df, columns=[..])

**2) 구간화 (binning)**
<br>연속, 수치형 -> 구간화 -> 범주형
<br>사용) pd.cut(data, bins=[구간 나누는 경계], labels=[각 구간 이름])

**3) 파생변수 생성**
<br>기존 피처들 + 도메인 지식 -> 새로운 피처
<br>논리적 타당성&기준 필요
<br>방법) 단위 변환, 표현형식변환, 요약 통계량 변환, 정보 추출, 변수 결합, 조건문 이용 등

**4) 스케일링** 
<br>변수 값의 범위 정규화 (이상치 제거 후 진행)
<br>sklearn 라이브러리 사용
<br>ex) from sklearn.preprocessing import StandardScaler

- StandardScaler (표준화)
<br>= z-score 표준화

- MinMaxScaler 
<br>= 선형 변환 (affin transformation) : 0~1 사이

### **[4] 데이터 시각화**
<br>데이터 -> 시각적요소 -> 패턴/관계/추세 파악

<br>**사용)** 
- Matplotlib : 수학적 확장
- Seaborn : 통계 데이터 시각화

<br>**순서)**
<br>1) 데이터 범주형/수치형
<br>2) 결측값/이상치 여부
<br>3) 어떤 그래프 사용?

<br>Parameter(매개변수) = 함수의 입력값
<br>Argument(인자) = 실제 전달값
<br>ex) sns.countplot(data=d1, x='survived')
<br>par- : data, x
<br>argu- : d1, 'survived'

<br>hue 파라미터
<br>= 범주형 변수를 인자로 받음 + 그걸 기준으로 데이터 구분 + 그래프에 표시

### [다양한 그래프]
**boxplot**
- IQR, 이상치
<br>


**countplot**
- 범주형 변수의 빈도수
<br>


**hisplot**
- 도수분포표, 수치형 변수의 구간별 빈도수
<br>


**displot/kdeplot**
- 히스토그램을 곡선으로 연결
<br>


barplot/pointplot
- 범주형 데이터 x , 수치형 데이터 y의 평균값
- 걈은선 : 신뢰구간(CI) 선 짧으면 변동 down
<br>

<br>


**scatterplot**
- 두 연속형 변수 간 관계 시각화 , 산점도 그래프
<br>=> 변수 간 패턴/상관관계/분포 파악
<br>


**regplot**
- 두개 연속 변수 사이 산점도 + 회귀선
<br>


**catplot**
- 수치형 데이터 & 범주형 데이터 관계
<br>


**pieplot**
- 부분과 전체 간의 비율 표현+강조
<br>


**heatmap**
- 변수 간의 상관계수 (corr())
- 상관계수 = 두 수치형 변수간 영향 주는 정도

<br>


**violinplot**
- boxplot+커널밀도추정함수
<br>


**pairplot**
