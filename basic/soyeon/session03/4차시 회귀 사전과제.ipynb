### 머신러닝
-> 데이터에서 패턴 발견 + 새로운 데이터 예측

종류)

**지도학습**
- 정답 있는 데이터 사용
- 목표 : 예측 = 정답을 위해 학습
- ex) 회귀, 분류..

**비지도학습**
- 정답 없는 데이터 사용
- 목표 : 데이터의 패턴/유사성 학습
- ex) 군집화 

**강화학습**
- 보상
- 목표 : 보상 최대화 학습
- ex) 자율주행..

### 모델링
**모델**<br>
= 데이터 학습 -> 패턴 찾음 -> 저장<br>
--> 새로운 데이터 입력 시 답 내놓음

**모델링**<br>
= 모델 만들어 가는 과정 (학습)
- 설계 : 데이터 특성 -> 어떤 알고리즘 사용?
- 훈련 : 데이터 제시하면서 정답 맞히게끔 수정
- 검증 : 훈련된 모델의 일반화 가능성 테스트

### 지도학습

**지도학습**<br>
= 정답 주고 학습 -> 예측값이 정답에 가까워지게<br>

**과정**
전제) 데이터 = labeled data (feature&taret)<br>
-> labeled data -> train set/test set<br>
-> train set으로 모델 학습<br>
-> test set으로 검증<br>
*분리해야 실제 성능 제대로 평가에 도움*<br>

- 데이터 준비 (labeled data)
- 예측
- 오차계산 (loss)
- 가중치 수정
- 반복
- 함수 획득
- 성능평가

**회귀/분류**
**회귀(연속형)**<br>
= 연속적인 숫자값 예측 : x기반 y 맞히는 함수 (선찾기)<br>
**분류(범주형)**<br>
= 어떤 레이블에 속하는 지 판별<br>

### 회귀 <br>
데이터(연속적) -> 패턴 찾기 -> 미래 값 예측<br>
종류) 선형회귀, 비선형회귀, 규제회귀...<br>

*w = 가중치 = 회귀계수*<br>
-> x가 y에 주는 영향력의 크기 (기울기)<br>

*b = 편향 = y절편*<br>
-> 데이터에 유연하게 최적화될 수 있게 함<br>

*최적의 w, b 찾는 법 (최소제곱법/경사하강법)*<br>
residual= Y-y= 예측-관측= 잔차<br>
RSS = 잔차 제곱 합<br>
--> RSS가 최소되는 지점으로 최적의 회귀선 구함!!<br>

*설명변수 = beta = 독립변수*<br>
*오차항 = 𝜖_i*<br>

*공분산=Cov(x,y)*<br>
= 두 변수 간의 상관된 정보 나타냄 -> 어떻게 함께 변하나<br>
-> Cov(x,y)=0 이면 독립적, Cov(x,y)!=0이면 자기상관<br>

*자기상관* <br>
= 한 변수의 현재와 과거가 상관관계 지님<br>

**선형회귀**


-> 데이터를 잘 설명하는 회귀선 찾기<br>
*회귀선 : 현재 데이터 요약 + 미래 값 예측*<br>

종류) 단순선형회귀 / 다중선형회귀<br>
단순선형회귀 <br>
- 하나의 독립변수 -> 직선 : y=wx+b 찾기<br>
- 잔차 줄여나가면서 적절한 w, b 찾기<br>

다중선형회귀<br>
- 독립변수 2개 이상 -> 회귀선은 주로 평면!<br>
- 식 :<br>

가정)<br>
*(전/후) 면 회귀 분석 전/후에 검증 가능!<br>
1. 선형성(전/후)<br>
종속변수와 설명변수 : 선형관계<br>
2. 독립성(전/후)<br>
설명변수간 선형독립성(<-> 다중공선성)<br>
3. 오차항의 평균은 0에 가깝기(후)<br>
오차항 = 실제-예측= 백색잡음= 랜덤변화 신호<br>
4. 등분산성(후)<br>
오차항의 분산이 일정해야 함 (<-> 이분산성)<br>
if 만족 x = 이분산성<br>
구간별로 오차가 달라져서 정확도가 떨어짐<br>
-> 확인) 잔차의 도표화/검정<br>
5. 오차항: 자기상관X = 공분산 0(후)<br>
-> 확인) 더빈-왓슨 검정/ ACF,PACF 함수<br>
6. 정규성 (후)<br>
오차항이 정규분포 따르기<br>
-> p-value를 계산할 때 사용되는 공식이 오차항이 정규분포 따른다고 가정하고 만듦<br>
-> 확인) 샤피로-월크 검정/ 자퀴-베라 검정/Q-Qplot 찍기<br>

### 다중공선성
: 독립변수 간 강한 상관관계 -> 회귀 모델 불안정<br>
-> 회귀계수 추정 부정확 + 해석력 down<br>
    = 어떤 독립변수가 종속변수에 얼마나 영향 미쳤는지<br>
확인법)<br>
1. 상관계수(r)<br>
= 변수 간의 통계적 관계(상관관계)의 정도를 수치로<br>
-1<=r<=1 사이 값 가짐<br>
r>0 : 증가 -> 증가<br>
r<0 : 증가 -> 감소<br>
r=0 : 연관성 X<br>

2. VIF 지수(분산 팽창 인수)<br>
VIF=1 :상관관계 X<br>
VIF<5 : 다중공선성 문제 X<br>
VIF>5 : 다중공선성 주의<br>
VIF>10 : 다중공선성 심각<br>

대처법) <br>
1. 변수 제거<br>
2. 변수 변환<br>
ex) 기존 변수 활용해서 새로운 변수 생성<br>
3. 규제선형모델 활용<br>
-> 모델의 복잡도 줄이기<br>
4. PCA(주성분 분석)<br>
-> 데이터의 차원 축소<br>
-> 분석/시각화 용이<br>

### 규제선형모델<br>
*너무 완벽해짐 -> 과적합의심*<br>
-> 과적합되지 않도록 규제가하는 모델<br>
= 기울기 크기 조정<br>
-> 오류 줄이기 + 특정 변수에 집착 X<br>

종류) <br>
**L2규제(릿지 회귀)**<br>
W^2에 패널티 부여 = 기둥에 따라 다른 힘으로 누름<br>
-> 큰 기둥은 세게! 작은 기둥은 살살!<br>
= 계수 크기 조절 -> 과적합 방지<br>

**L1규제(라쏘 회귀)**<br>
|W|에 패널티 부여 = 모든 기둥을 같은 힘으로 누름<br>
-> 작은 기둥은 사라질 수도... (변수 선택 기능)<br>

**L2+L1규제(엘라스틱넷 회귀)**<br>
L1: 불필요한 변수 제거<br>
L2: 남은 변수 적절한 크기로 유지<br>
-> 안정적 예측 + 희소성 달성<br>

### 모델 평가방법<br>

1. 시각화<br>
-> 어떤 회귀선이 데이터를 잘 요약하는 지<br>

2. 통계지표<br>
(시각화 보완 가능)<br>
- 모델 유의성 검정<br>
    1) F-statistic : 회귀 모델의 유의미성 (전체 모델)<br>
        ㄴ 0.05 > F : 유의미<br>
    2) T-statistic : 독립변수가 종속변수에 유의미한지 (개별 변수)<br>
        ㄴ |t| <0.05 : 유의미<br>
- 모델 성능평가<br>
    1) 결정계수(R**2)<br>
    : 회귀 모델이 종속변수의 전체 변동성 얼마나 잘 설명하는 지<br>
        ㄴ R**2 높으면 잘 설명하는 거<br>
    2) 조정된 결정계수<br>
    : 데이터와 관련 없는 변수 추가해도 높아지 수 있음<br>
    -> 변수 개수 증가에 덜 민감하도록 조정<br>
    3) AIC, BIC<br>
    : 정보 기준 <br>
        ㄴ 낮을 수록 좋음<br>

3. 성능 평가 지표<br>
- 평균제곱오차(MSE)<br>
    ㄴ 이상치에 민감<br>
- 평균절대 오차(MAE)<br>
    ㄴ 모든 오차를 동일 취급 -> 이상치 존재해도 적절히 유지<br>

4. 변수 유의성 평가<br>

- T검정<br>
: 독립변수의 회귀계수가 유의미한지 <br>
가설 - t값 계산 - p값 학인/판단<br>


### 비선형회귀 <br>
- 다항식 회귀 모델 <br>
Y=a+bX -> Y=a+bX+cX^2+dX^3 +...<br>
독립변수에 거듭제곱 항 추가 -> 비선형적 특성 모델링<br>
+) 회귀선이 데이터에 더 잘 맞게 그려짐<br>
-) feature가 넘 많으면 과적합<br>
--> 곡선 그리는 데이터 / 변화속도가 다른 경우<br>

- 지수 회귀 모델<br>
종속변수가 지수적으로 변화하는 관계 모델링<br>
Y=a*e^(bX) -log-> ln Y= ln a+ bX -선형회귀 사용<br>
--> 값이 시간에 따라 지수적으로 증가/감소<br>

- 로그 회귀 모델<br>
종속변수와 독립변수 간의 관계 : 로그함수로 표현 잘됨<br>
X -log-> 선형회귀 적용<br>
--> 초기에 빠르게 증가 + 점차율 점차 줄 때<br>

- 스플라인 회귀<br>
데이터 구간별로 나눔 + 선형/비선형함수 적용 + 예측<br>
구간 경계에서 연속성 유지하기<br>
--> 데이터 패턴이 구간별로 다르게 나타낼 때<br>
