- 지도 학습
- 비지도 학습 ex) 군집화

### 비지도 학습

특징) 
1. 정답 레이블 x = 작업 명확히 정의 x = 목표 명확 x
--> 모델 성능 측정 어렵
2. 피처학습 -> 데이터 내 고유 패턴 식별 = 새로운 특성 찾기

적합한 case)
1. 패턴 모름 = 정답 모름
2. 패턴 변함 = 정답 알아도 큰 의미 x
3. 열린 문제 해결 및 지식 일반화
ex) 이상 거래 탐지/자연어 처리/고객 성향 분석

### 군집화 (clustering)
= 비슷한 특성끼리 그룹화 (군집)<br>
-> 각 그룹의 구성 파악 -> 전체 데이터의 구조 이해<br>

지도학습 vs 군집화 <br>
| 지도학습 | 군집화 |
|---|---|
| 독립-종속 변수 속 패턴 찾기| 데이터 내부 패턴 인식|
| 종속변수 예측 | 종속변수 x, 유사성 가진 그룹 분류|


목표) 응집도 up / 분리도 up<br>
응집도(cohesion) : 같은 군집에 속하는 데이터끼리 비슷<br>
분리도(separation) : 다른 군집끼린 분리<br>
= 데이터 간 유사성 최대한 유지 + 서로 다른 그룹 구분<br>

과정)
1. 피처 선택/추출 
2. 군집화 알고리즘 선택
3. 군집 유효성 검증 (if 별로 : 1로 돌아감)
4. 결과 해석

## 과정 1 ) 피처 선택<br>

군집화 특징 : 여러번 반복 시도 -> 최적의 결과<br>
-  변수 유형 이해 (종류/특성) <br>
    ㄴ 유형에 따라 알고리즘 달라짐<br>

    |연속형|명목형|혼합형|
    |---|---|---|
    |수치로 측정 ex) 키, 소득| 범주로 구분 | 연속+범주|
    |유클리디안 거리, 맨하탄 거리..|해밍거리, 자카드 거리| - |
    |k-means, 계층적 군집화..|k-modes, 계층적 군집화..|k-prototypes, gower distance 기반 계층적 군집화..|
    |스케일링 필수 | 원핫인코딩, 더미변수 변환 | 변수 유형에 맞게 거리조합|

    ** <br>
    연속형 : 스케일링 필수 !<br>
    명목형의 거리 : 빼기 X 세기 O<br>
    혼합형 : gower distance 사용 + 가중평균<br>

    ** <br>
    k-means : 중심점=평균 -> 평균 정의 x 사용 x<br>
    k-modes : 중심점=최빈값 <br>
    k-prototypes : 연속형엔 평균, 명목형엔 최빈값<br>

    ** <br>
    계층적 군집화 ?<br>
    누가 누구랑 가장 가까운지 -> 거리 행렬 필요 (원본 ㄴㄴ)<br>
    ㄴ 거리 행렬 만드는 과정에서 시간 오래 걸림<br>



- 거리/유사도 정의, 측정<br>
    ㄴ 그룹 형성 기준 = 거리/유사도<br>
    ㄴ 유형에 따라 거리 측정 방식 다름 (잘 정의 해야 됨)<br>

    거리 측정
    : 군집/데이터 간의 비/유사성을 수치로 계산
    거리 down = 유사
    
    점-점 거리 측정 : 유클리디안 거리, 맨하탄 거리, 코사인 유사도


    |유클리디안 거리 | 맨하탄 거리 | 코사인 유사도|
    |---|---|---|
    |직선거리|각 차원 차이의 절댓값 합| 벡터 간 각도|
    | 일반적, k-means | 이상치에 덜 민감| 텍스트 데이터/고차원데이터|


- 차원 축소<br>
    ㄴ 변수 많으면 복잡 + 효율/성능 down<br>
    ㄴ 유사 변수 -> 통합 / 불필요 변수 -> 제거<br>

    필요성) <br>
    군집화 알고리즘 = 거리 기반 <br>
    -> 차원 up -> 알고리즘의 분별력 down<br>

    Why? <br>
    - 공간 늘어나는데 데이터는 그대로<br>
    -> 공백 많아짐 = 이웃 down<br>
    - 가깝나 머나 비슷해짐<br>
    -> 가까운 걸 묶어야 하는데 가깝나 머나 비슷해서 기준이 상실 = 분별력 감소<br>
    - 쓸데없는 피처 = 노이즈<br>
    
    차원 = 변수의 개수<br>
    목표) 고차원 데이터를 저차원 공간에 + 핵심 정보 지키기<br>
    -> 선형 투영 : 고차원 -> 저차원 투영 (PCA, SVD,..)<br>
    -> 비선형 차원 축소 : 데이터 점사이의 곡선 거리 고려<br>

    ** <br>
    PCA(principal component analysis)<br>
    목적) 다차원 데이터 -> 분산이 큰 방향 찾기 -> 데이터 투영 -> 차원down<br>
    특징) 분산 최대한 보존, 상관관계 높은 피처끼리 결합 -> 주성분<br>
    주의) 표준화(안하면 단위 큰 게 1등), 결측값 없애기, 해석 어려움, 정보 손실

## 과정 2) 알고리즘 선택 <br>
- 계층적 군집화(hierarchical clustering)<br>
: 가까운 것부터 2개씩 합쳐나감 <br>
-> 무엇이/언제 뭉쳤는 지 덴드로그램으로 기록<br>

덴드로그램의 수직축 = 두 군집 병합될 때의 유사성<br>
군집의 개수 사전 설정 X<br>
상향식 - 응집형 계층적 군집화 (병합 클러스터링)<br>
    ㄴ 여러개 -> 하나 남을 때까지 병합<br>
    1. 거리 행렬 계산 : 데이터 간 거리<br>
    2. 단일 클러스터 시작 <br>
    3. 가장 가까운 클러스터 병합 : 1+1 -> new<br>
    4. 거리 행렬 업데이트 : new + old <br>
    5. 반복 -> 하나 남음<br>

하향식 - 분리형 계층적 군집화 (분할 클러스터링)<br>
    ㄴ 하나 -> 여러 개될 때까지 나눔<br>

** 연결 방법(linkage)<br>
: 두 군집 간의 거리 계산<br>

1. 단일 연결법 (single linkage)<br>
: 군집 A와 B 사이 가장 가까운 두 점간의 거리 = 군집 간 거리<br>
고립된 군집 찾기 + 이상치 취약 (연쇄)<br>

2. 완전 연결법 (complete linkage)<br>
: 군집 A와 B 사이 가장 먼 두 점간의 거리 = 군집 간 거리<br>
군집들의 내부 응집성 + 이상치 민감<br>

3. 평균 연결법 (average linkage)<br>
: 군집 A/B의 모든 점 쌍 사이 거리 평균<br>
모든 항목 + 이상치 덜 민감 + 계산량 up<br>

4. 중심 연결법 (centroid method)<br>
: 각 군집의 중심점간의 거리<br>
계산 시간 up<br>

5. 중앙 연결법 (median)<br>
: 두 군집의 중앙값 사이 거리 = 대표값<br>
극단값 영향 down, 구조 파악 어렵<br>

6. Ward 연결법 (Ward's procedure)<br>
: 군집 내 SSE가 가장 적게 증가하도록 병합<br>

덴드로그램 읽기<br>
세로축 = 그때의 군집간 거리<br>
가까운 거 = 높이 작음<br>
막대 김 = 그만큼 멀리 있음+합칠 상대 없었음 = 경계<br>


장점)<br>
- k 미리 안정해도 됨
- 데이터 구조 한눈에
- 유형 구애 x
- 결과 같음 = random_state 필요 x<br>

단점) <br>
- 느림+메모리 마니
- 한번 합치면 끝
- 이상치 민감
- 높이는 사람이 결정

## K-means
: 데이터를 k개의 그룹으로 나눔 
+ 각 그룹의 중심점(centroid)과 거리가 가장 가까운 데이터끼리 묶기

1. k개의 중심점 정하기 + 그룹화
2. 그룹들의 중심 새로 잡기 (평균)
3. 새로운 중심으로 재그룹화
4. 반복 -> 중심점 이동했는데도 데이터의 소속 변화 없으면 완료

**
k는 사람이 정함

장점)
- 직관적
- 대용량 가능!

단점) 
- 초기 centroid 민감
- k 결정 어렵
- 이상치 민감
- 기하학적 모양은 파악 어렵


### 엘보우 방법 (elbow method)
: 클러스터 내 SSE 바탕으로 k 개수 추정<br>
-> SSE 감소 효과 있->미미 의 경계 = k

## DBSAN
: 밀도 높은 지역 데이터 -> 하나의 군집 <br>
+ 밀도 기준 만족 x = 군집에 포함 x

특징) 
- k 안정해도 됨
- 이상치는 따로 뺌
- 모양 안가림
- 특정 거리내 데이터는 가장 밀도 높은 군집과 그룹화

단계)
1. 샘플 분류 (핵심/경계/잡음)
2. 클러스터 생성
3. 경계 샘플 할당

장점) 
- 기하학적 분포도 처리
- 이상치 민감 x
- 모든 샘플 할당 x

단점)
- eps, min_samples 설정에 영향 up
- 연산 up 속도 down
- 밀도 다르면 군집 잘 못잡음..
- 차원...

## GMM (Gaussian Mixture Model)
: 가정) 데이터 = 가우시안 분포 , 각 분포 = 클러스터
+ 복잡한 확률 분포 -> 여러개의 가우시안 분포로 합쳐서 표현

특징) 
- 모델 기반 군집화 
- 군집 = 확률 분포 , 전체 = 확률 분포의 혼합
- 데이터의 생성 메커니즘 모델링 가능

과정)
1. 전체 데이터 셋 분포 확인 + 가정
2. 여러개 정규분포 곡선 추출 + 각 데이터가 어떤 분포에 속하는 지 결정 
-> 각 분포 = 하나의 군집 <br>
    ㄴ EM 알고리즘 사용 (기댓값-최대화 알고리즘) <br>
    -> 모델 파라미터 추정 (군집 중심, 모양, 크기, 방향)<br>


|k-means|GMM|
|---|---|
|중심 기반| 확률 분포 기반|
| 하나의 군집 | 확률로 분산 할당|
|원형에 효과적, fast | 유연|
| 길쭉이 분포 어렵| slow, 정규분포 가정 필요|

** k-means에게 바라지 말 것
: 확신의 정도 / 두번째로 가까운 군집 위치 / 임계값

-> k-means = 답 , GMM = 답+답 얼마나 믿을래

## 과정3) 유효성 검증 <br>
-> 정답 레이블 유무 <br>

외부 평가 (external evaluation) <br>
- 정답 레이블 존재
- 군집화 결과 & 실제 레이블 비교 -> 유사도 평가 (ARI : 1에 가까울 수록 일치) <br>

내부 평가 (Internal Evaluation) <br>
- 정답 레이블 없음
- 군집 내부 응집도 / 군집 간 분리도 기반 품질 정량화
    ㄴ 데이터 자체 구조를 기반으로 판단
- 범용 지표 : 실루엣 계수, 던 지수 <br>

### 실루엣 계수 <br>
: 클러스터 내 샘플들의 조밀성 = 군집의 품질 확인 <br>

1 : 분리도>응집도 : 군집화 great <br>
~ 0.5 이상 : ㄱㅊ, 0.7 이상: good <br>
0 : 분리도=응집도 : 분리 명확 x <br>
~ <br>
-1 : 분리도<응집도 : 군집화 bad <br>

### 던 지수  <br>
: 클러스터 간 분리도, 클러스터 내 응집도 비율 -> 클러스터링 품질 평가 <br>
값 up = 군집화 great  <br>
던지수>0  <br>
실루엣 계수와 종합적 판단하기 <br>

한계)  <br>
군집 수 많음 계산 up <br>
이상치에 민감 <br>
