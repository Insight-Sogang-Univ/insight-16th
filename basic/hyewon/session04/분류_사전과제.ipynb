주어진 데이터를 기반으로 정답을 잘 맞추는 함수

### 분류: 데이터가 범주형 변수를 예측하기 위해 사용

기존 데이터가 어떤 레이블에 속하는지 패턴을 알고리즘으로 인지한 뒤에 새롭게 관측된 데이터에 대한 레이블을 판별하는 것





# 분류모델의 종류

#### 1. 로지스틱 회귀
>**이진분류** 문제를 푸는 대표적인 알고리즘으로 샘플이 특정 클래스에 속할 확률을 추정하여 기준치에 따라 분류하는 것을 목표로 함. 

> 즉, 독립변수의 선형조합에 로지스틱 함수를 적용하여 출력값을 0에서 1사이로 변환해주는 것을 의미

- 시그모이드 함수: 출력이 0과 1사이의 값을 가지면서 S자의 형태로 그려지는 함수

![alt text](<스크린샷 2026-09-16 오전 1.58.19.png>)
#### 2. 의사결정나무

>조건에 따라 데이터를 분류하며, 최종적으로 데이터가 순수한 레이블의 집합으로 구성될 때까지 분류를 반복하는 분석 방법

![alt text](<스크린샷 2026-09-16 오전 1.59.18.png>)


- CART: 가장 널리 사용되는 의사결정나무 알고리즘. 데이터셋을 임계값을 기준으로 두 child로 나눔

임계값은 불순도(지니계수)가 낮아지는 방향으로 나눠야 함!

>CART 알고리즘의 주요 단계

(1) 임계값 설정..

(2) 불순도 감소 알고리즘 

<학습 시 고려해야 할 것>
1. 하이퍼파라미터 설정
2. 시각화
3. 가지치기: 불필요한 노드 지우기

#### 3. 서포트 벡터 머신
>클래스를 분류할 수 있는 다양한 경계선 중 최적의 라인을 찾아내는 알고리즘

> 데이터를 분리하는 초평면 중에서, 데이터들과 가장 거리가 먼 초평면을 선택하여 분리하는 지도학습 기반의 이진 선형 분류 모델

패턴인식, 이미지분석, 바이오인포매틱스 등 다양한 머신러닝 분야에서 활용

결정 경계는 데이터로부터 가장 멀리 떨어져 있는게 좋음.. 따라서 가장 거리가 큰 경우를 선택하며 최적의 선을 찾음

***장점***

데이터가 적을때 효과적, 연산량 최소화, 비선형 모델 분류 가능. 노이즈의 영향이 적음

***단점***

데이터 전처리 과정이 중요, 데이터가 많아질수록 평가과정이 많아져 속도가 느림


#### 4. KNN

>비슷한 특성을 가진 데이터끼리 서로 가까이 있는다는 점을 이용한 분류 알고리즘 

> 데이터로부터 거리가 가까운 k개의 다른 데이터 레이블을 참조하여 분류하는 알고리즘

=> **KNN의 가장 중요한 특징!**

어떠한 학습이 필요하지 않다는 것. 별도의 모델없이 데이터만을 이용해 새로운 데이터가 왔을 때 즉석에서 주변 데이터를 이용해 분류하는 것.

***장점***

범주를 나눈 기준을 알지 못해도 분류 가능, 학습과정 불필요, 이해가 쉽고 구현이 간단, 추가데이터 처리용이, 원본 데이터 전체를 그대로 저장

***단점***

테스트 데이터 개수에 따라 사간이 오래 걸림, 학습데이터 양도 계산 시간에 영향을 미침, K값 결정 어려움, 유사도 정의 어려움, 이상치 존재시 분류 성능에 큰 영향

# 분류 평가 지표

#### 1. 혼동 행렬

>분류 모델의 예측 결과를 정확한 예측과 잘못된 예측으로 구분하여 나타낸 표

![alt text](<스크린샷 2026-09-16 오전 2.16.35.png>)

이를 통해 알 수 있는 것
1. 정확도(모든 가능한 예츨 중 참인 비율): 모델이 입력된 데이터에 대해 얼마나 정확하게 예측하는지..
2. 정밀도(참이라고 예측한 경우 실제 참의 비율)
3. 재현도(실제로 참인 경우 참으로 예측하는 비율)

=> 정밀도와 재현도는 trade-off 관계

#### 2. F1-score
>정밀도와 재현율의 조화평균

>머신러닝 모델의 성능을 평가하는 주요지표
#### 3. ROC/AUC
> ROC Curve: 얼마나 분류가 잘 되었는가를 보여주는 그래프 

![alt text](<스크린샷 2026-09-16 오전 3.18.33.png>)

TPR: 참인 것들 중에 참이라고 예측한 비율 = recall
FPR: 거짓인 것들 중에 참이라고 잘못 예측한 비율 

>AUC Curve: ROC와 x축 사이의 면적

![alt text](<스크린샷 2026-09-16 오전 3.19.01.png>)

모델의 성능을 숫자로 표현 가능

0.5에서 1 사이의 값을 가지며 1에 가까울수록 분류 성능이 좋은 것.

# 하이퍼파라미터 최적화

#### 하이퍼파라미터란?
>학습시작 전에 사용자가 직접 설정하는 변수

하이퍼파라미터 최적화: 적절한 하이퍼파라미터를 찾아 모델 성능을 향상시키는 것

<최적화 과정>
1. 하이퍼파라미터 탐색 범위 설정
2. 평가지표계산 함수 정의
3. 설정한 범위에서 값을 샘플링한 하이퍼파라미터 값을 사용하여 검증 데이터로 정확도 평가
4. 위 단계를 반복하고 정확도 결과를 통해 하이퍼파라미터 범위 좁힘

#### 하이퍼파라미터 최적화 방법
>Grid Search: 정해진 범위에서 하이퍼파라미터 모두 순회
>Random Search: 정해진 범위에서 하이퍼파라미터를 무작위로 탐색
>Bayesian Optimization: 사전정보를 바탕으로 하이퍼파라미터 값을 확률적으로 추정하며 탐색
