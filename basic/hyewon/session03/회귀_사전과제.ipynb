# 회귀 사전과제

### 머신러닝이란?
> 인공지능의 한 분야로, 모델이라는 소프트웨어를 학습하여 새로운 데이터를 예측하거나 결정을 내릴 수 있도록 하는 기술

- 지도학습: 기계의 예측값이 데이터의 정답과 같아지도록 학습시키는 것 (회귀, 분류)
- 비지도학습: 데이터의 패턴 혹은 유사성을 학습시키는 것 (군집화)
- 강화학습: 시행착오를 반복하여 보상을 최대화하는 행동 학습

### 모델링
> 예측기(공식)을 만들어가는 전체적인 과정

### 모델
> 학습의 결과물, 즉 데이터를 학습하며 찾아낸 패턴이 저장된 상태

### 모델링의 과정
1. 설계
2. 훈련
3. 검증


### 지도학습이란?
> 정답레이블을 함께 주고 학습시키는 방법

데이터 준비 -> 예측 -> 오차계산 -> 가중치수정 -> 반복 -> 학습완료=함수획득 -> 성능평가


## 회귀
회귀란 주어진 데이터를 기반으로 정답을 잘 맞히는 함수 (연속적인 숫자값을 예측)

데이터를 가장 잘 설명하는 회귀선을 찾는 과정으로 회귀선은 현재 데이터를 요약하는 동시에 아직 관측되지 않은 미래 값을 예측하는 도구가 됨.

> 가중치 (w)

독립변수가 종속변수에 주는 **영향력의 크기**. 그래프 상 기울기, *회귀계수*라고 불림

> 편향 (b)

독립변수가 0일 때 나오는 기본값으로 그래프 상 y젉편

### 1. 단순선형회귀

하나의 독립변수만 존재하는 선형회귀, 직선 형태

잔차를 줄여나가며 직선을 결정짓는 적절한 회귀계수와 편향을 찾아감

=> 이 최적의 회귀계수와 편향을 찾는 방법으로 
1. 최소제곱법
2. 경사하강법




##### 최소제곱법
잔차의 제곱의 합이 최소가 되는 지점으로 최적의 회귀선을 구하는 방법

### 2. 다중회귀선형

독립변수가 두개 이상인 경우

회귀선은 평면이 됨

![alt text](<스크린샷 2026-09-13 오전 4.46.31.png>)

> 기본가정 6가지

1. 선형성
2. 독립성 (그렇지 않은 경우 다중공선성이 있다고 함)
3. 오차항의 평균은 0이다
4. 등분산성: 오차항의 분산이 일정
5. 오차항은 자기상관이 되어있지 않다: 공분산0
6. 정규성: 정규분포 따름



### 다중공선성
독립변수들 간의 강한 상관관계로 회귀모델이 불안정해지는 문제

>상관계수를 통한 확인

>VIF지수

회귀 모델의 R-square을 이용해서 계산

#### 대처방법

변수제거, 변수 변환, 규제선형모델 활용, 주성분 분석

### 규제선형모델
>모델이 학습데이터에 과적합되지 않도록 규제를 가하고자 등장한 모델

1. L2규제(릿지회귀): 회귀계수의 제곱에 대해 페널티를 부여하는 방식

회귀계수의 제곱을 패널티로 줘, 큰 회귀계수일수록 훨씬 강한 패널티를 받음 (변수가 제거되지는 않고 계수의 크기만 적절히 줄여 과적합 방지)
 
 => 예측변수가 많고 다중공선성 존재시 사용


2. L1규제(라쏘회귀): 회귀계수의 절댓값에 대해 페널티를 부여하는 방식

회귀계수의 크기와 상관없이 같은 패널티를 받음. 계수가 0이되기도 해 변수가 완전히 사라질 수 있음 (중요한 변수만 남음)

=> 예측 변수 수가 많고 그 중 일부만이 실제로 중요할 때, 모델의 해석을 간단하게 유지하고자 할 때 사용

3. L1규제 + L2규제(엘라스틱넷 회귀)

두 방법을 동시에 사용하는 방법으로 **불필요한 변수는 제거하고 남은 변수들은 적절한 크기로 유지**

-> 안정적인 예측과 희소성을 동시에 달성 가능

=> 예측변수 수가 많고, 그 중 중요한 변수를 선택하면서도 다중공선성을 관리할 때 사용




### 모델 평가방법

>1. 시각화

회귀선이 데이터 전반을 잘 요약하는 모습이지만, 어느 화귀선이 해당 데이터를 더 잘 요약하는지 비교는 어려움

>2. 통계지표

(1) 모델 유의성 검정
- F통계량: 회귀모델이 유의미한지 검정. 값이 클수록 모델이 통계적으로 유의미함
- T통계량: 각 독립변수가 종속변수에 유의미한 영향을 미치는지 판단
- F검정: 회귀모델이 유의미한지 검정
- T검정: 특정 독립변수가 유의미한지 검정

(2) 모델의 성능평가
- 결정계수(R스퀘어): 회귀모델이 종속변수의 전체 변동성을 얼마나 잘 설명하는지 나타냄. 모델을 활용했을 때 오차가 얼마나 줄어드는지, 데이터를 얼마나 잘 설명하는지 측정
- 조정된 결정계수: R은 독립변수의 개수가 증가하면 증가할수록 자연스레 증가. 따라서 변수 개수 증가에 덜 민감하도록 조정한 지표.
- AIC, BIC: 두 지표 모두 정보기준으로 값이 낮을수록 좋음


>3. 성능평가지표

(1) 평균 제곱 오차 (MSE)

예측값과 실제값 사이의 오차 제곱의 평균을 의미

(2) 평균 절대 오차 (MAE)

모델의 예측값과 실제값 사이의 절대값 오차의 평균을 의미

(3) MSE와 MAE의 차이
MSE는 큰 오차에 민감해서 이상치에 민감

MAE는 모든 오차를 동일하게 취급해서 이상치가 존재할 때도 전반적인 예측 성능 유지 가능

>4. 변수 유의성 평가

(1) T검정

독립 변수의 회귀계수가 유의미한지 검정하는데 사용

t 값 계산 후 p-value 확인 및 판단
