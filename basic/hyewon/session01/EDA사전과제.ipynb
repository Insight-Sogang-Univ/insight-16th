# EDA란?
데이터 분석을 위해 데이터를 다양한 각도에서 파악하는 모든 과정 

데이터의 특성을 파악하고 시각화 하면서 데이터를 이해하고 확인하는 단계

# EDA 대상
### 일변량 
- EDA를 통해 한번에 파악하려는 변수가 한개. 데이터를 설명하고 그 안에 존재하는 패턴을 찾는 것이 주 목적

### 다변량
- EDA를 통해 한번에 파악하려는 변수가 여러 개. 변수들 간의 관계를 보는 것이 주 목적.

# EDA 종류
### 시각화
- 차트 그림 등을 통해 데이터를 확인하는 방법

### 비시각화
- 그래픽적인 요소를 사용하지 않고 주로 summary statistics를 통해 데이터를 확인하는 방법

# EDA 유형
##### 일변량 비시각화 : Distribution을 확인
##### 일변량 시각화 : 전체적으로 확인
##### 다변량 비시각화 : 둘 이상의 변수 간 관계 확인
##### 다변량 시각화 : 변수 간 관계 전체적으로 살펴봄



# 데이터셋 파악하기
1. 데이터 프레임 보기 -> head(n) 이용
2. 데이터 변수 확인 -> 정의와 유형, 데이터 형식 
3. 데이터 분포 확인 -> 전처리 아이디어를 얻을 수 있음


# 전처리
### **전처리**: 데이터를 분석에 적합한 형태로 만드는 과정

데이터 품질은 분석의 정확도를 결정하기 때문에 이상값이나 결측값 등을 수정하고 변형시켜야 함

## 결측값 처리
#### **결측값**: 데이터 수집 과정에서 측정되지 않거나 누락된 데이터
- 결측값 확인: NaN, ?, 0 등의 값으로 나타남
- 결측값 처리: 삭제(dropna), 대체(fillna)

## 이상치 처리
#### **이상치**: 관측된 데이터의 범위에서 많이 벗어난 값
- 이상치 확인: 통계(describe), 시각화(boxflot), Z-score, Tukey Fences 
- 이상치 처리 및 제거: 전체삭제, 다른 값으로 대체, 변수화, 리샘플링 




## 엔지니어링 (변수가공)
#### 변수들의 형태를 변형하거나 적절하게 처리하는 과정
- 새로운 데이터 또는 변수의 추가 없이 기존의 데이터를 보다 유용하게 만드는 방법
1. 레이블인코딩/원핫인코딩
- 레이블 인코딩: 범주형 변수를 0부터 N-1까지의 숫자로 변환
- 원 핫코딩: 범주형 변수를 0과 1로 변환
2. 구간화
- 분석 목적이나 방법에 따라 연속 데이터를 일정한 구간으로 나눠서 분석
3. 파생변수 생성
- 기존의 변수를 다른 변수로 재정의하여 변수 추가
4. 스케일링
- 정규화를 통해 서로 다른 변수의 값 범위를 일정한 수준으로 맞추는 작업
- 표준화(평균 0, 분산1), Minmaxscaler(모든 feature가 0과 1 사이 위치)

- Matplotlib과 Seaborn 사용
- 시각화 전 체크
- 1. 범주형인지 수치형인지
- 2. 결측값 및 이상치 확인
