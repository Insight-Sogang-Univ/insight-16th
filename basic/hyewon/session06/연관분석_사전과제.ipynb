# 연관분석이란?
>상품이나 서비스를 구매하는 등 일련의 거래나 사건 안에 존재하는 항목 간의 일정한 연관규칙을 발견하는 분석

- 정답이 별도로 존재하지 않기에 비지도학습
- 연관분석 알고리즘으로 Apriori, FP-Growth가 있음
- 연관규칙의 형태
    - `if-then 형식`
- 연관규칙으로는 인과관계를 알 수 없음 only 상관관계만 알 수 있다
    - 관찰 데이터기 때문
    - 방향성은 그냥 수학적 비율일 뿐..
    - 공변요인 때문에 관계성을 보일 수 있음

# 연관규칙의 조건
##### 1. 두 품목이 함깨 구매한 경우의 수가 일정수준 이상일 것 

=> 일정 이상의 지지도
> 지지도: 전체 거래항목 중 품목 A와 품목 B가 동시에 포함하는 거래의 비율 

   - 지지도가 낮다면? 우연일 수 있는 규칙이라 신뢰도가 낮거나, 실행 가능한 인사이트 도출이 어려움
   - 이 규칙이 현실에서 의미가 있는가를 판단하는 1차적 기준


##### 2. 품목 A를 포함하는 거래 중 품목 B를 구입하는 경우이 수가 일정 수준 이상일 것 

=> 일정 이상의 신뢰도
> 신뢰도: 품목 A를 포함하는 거래 수 중 품목 A와 품목 B가 동시에 포함하는 거래의 비율

   - 신뢰도가 낮다면? 의미없는 연관규칙, 추천 실패로 이어짐


##### 3. A와 B가 우연히 함께 등장할 확률과 실제 함께 등장할 확률을 비교할 것

> 향상도: 품목 A가 주어지지 않았을 때의 품목 B의 확률 대비, 품목 A가 주어졌을때의 품목 B의 확률의 증가비율

   -  향상도가 1보다 크면 우연한 기회보다는 높은 확률, 즉 서로 양의 상관관계
   - 향상도가 1이면 독립
   - 향상도가 1보다 작으면 우연적 기회보다 낮은 확률, 즉 음의 상관관계


### 활용 사례

1. 지지도 측면
   - 허리케인 경보기간 중 발생한 전체 거래 중에서 재난 대비 용품과 딸기 팝타르트가 함께 담긴 비중이 크게 높아지는 걸 확인 ..
   - 월마트틑 재난 경보 하에서 유의미해지는 조합을 찾아냄

2. 신뢰도 측면
   - A: 허리케인 대비 물품 구매
   - B: 딸기 팝타르트 함깨 구매
   - 재난 대비 장바구니에 팝타르트가 따라오는 경향이 있음을 확인

3. 향상도 측면
   - A: 허리케인 경보발령상황
   - B: 딸기 팝타르트 구매
   - 평상시 구매확률 대비 7배.. 즉 1보가 훨씬 크며 양으ㅢ 상관이 존재하는 걸 확인




# Apriori 알고리즘
> 선험적 경험을 이용한다

무차별 탐색의 비효율을 개선하기 위해 상위 조합부터 차례로 스캔하며 특정 조합이 자주 발생하지 않는다면 이의 결과물로 탄생한 흐속 조합들까지 모두 후보에서 배제하는 방식의 알고리즘

이를 통해 하나의 조합만 검사하고도 이에서 파생된 다른 조합들까지 후보에서 배제할 수 있게 되어 시간과 연산량을 줄일 수 있다

![alt text](<스크린샷 2026-09-25 오전 2.34.05.png>)

##### **프로세스**
> 빈발 항목 집합 찾고 -> 단계적으로 조합을 생성해 필터링

1. 단일 항목 집합 생성: 최소 지지도 넘는 항목들만 다음단계러

2. 2개 항목 집합 생성: 1에서 선별된 걸로 2개 항목 조합 만듦.. 이 조합의 지지도 계산 후 최소 지지도를 넘지 않으면 그 조합 제거

3. 더 큰 항목 집합 생성: 이전 단계에서 통과된 조합을 바탕으로 3개 이상의 항목으로 구성된 조합 만듦. 여기도 최소지지도 넘지 않으면 그 조합 제거.. 이를 더이상 조합 생성 안될때까지 방복

4. 최종 빈발 항목 집합 선정: 남은 빈발 항ㅇ목 집합들을 바탕으로 연관규칙 생성. 각 규칙에 대해 지지도, 신회도, 향상도를 계산하여 가장 유용한 규칙 선택

##### **장단점**
- 장점
   - 다양한 상품 연관구매 패턴 발견
   - 원리 간단, 이해분석 용이
- 단점
   - 데이터 커질수록 속도 느려짐

# FP growth 알고리즘
> 위의 알고리즘의 연산 속도 문제 해결을 위해 자료구조를 이용함

- 빈발 푹목 집합을 더 효율적으로 찾아냄
- FP-Tree 구조를 이용해 Apriori를 효과적으로 구현
- 모든 거래를 확인해 아이템마다 지지도를 계산하고 최소 지지도 이상의 아이템만 선택 
   - 모든 거래에서 빈도가 높은 아이템 순서대로 순서 정렬
   - 부모 노드를 중심으로 거래를 자식노드로 추가해주며 트리 생성
   - 새 아이템 나오면 부모노트부터 시작하고, 아니면 기존노트 확장
   - 위 과정을 반복해서 FP-tree만들고 최소 지지도 이상의 패턴만 추출

##### **프로세스**
1. 데이터셋 확인

2. 빈도수 재정렬: 자주 구매된 항목이 먼저 오도록

3. FP-Tree 생성

4. 빈발 항목 집합 추출

5. 연관 규칙 도출

##### **장단점**
- 장점
   - 데이터를 두번만 스캔하는 트리구조라 apriori보다 훨씬 빠름
   - 후보 아이템리스트를 생성할 필요 없이 트리 구성만 하면 됨
- 단점 
   - 메모리를 효율적으로 사용하지 않음
   - apriori에 비해 설계 어려움.. 지지도 계산은 무조건 FP-tree가 만들어져야 가능
