# 사전과제

### 판다스 개념
 판다스란 데이터 분석에 특화된 파이썬 라이브러리로 행과 열로 이루어진 테이블 형태의 데이터 구조인 데이터 프레임을 기본 틀로 사용함. 
 
 수치연산이나 통계 계산에 최적화된 *넘파이의 확장판*이라고 볼 수 있음

### 판다스 라이브러리를 불러오는 법 
**라이브러리는 미리 만들어진 기능이 들어있는 상자**

- import pandas as pd 활용
- 데이터 불러오기 사용 (df=pd.read_csv('파일 경로'))





### 데이터 프레임이란?
컴퓨터로 정보를 처리하고 저장하는 도구
1. **index와 column** (행과 열) => 데이터의 좌표를 나타낸다
2. **Series** => 값과 인덱스로 구성된 1차원 자료구조. 시리즈 구성 값들은 모두 같은 데이터 타입이어야 함.
3. **Data Frame** => 행과 열로 이루어진 2차원 배열구조. 여러개의 시리즈가 모여 하나의 데이터 프레임이 된 것.


### 데이터 확인, 추출, 조작
1. **데이터 확인**
- head(), tail() 사용: 앞 뒤의 일부 데이터를 미리 확인하여 파악
- shape(), info(), describe(), isnull(), dtypes 사용
- count(), unique()등을 사용: 전체 개수와 고유값 파악

2. **데이터 추출**
- 인덱싱(단일 항목 지목)과 슬라이싱(범위지정)
- 행단위 추출: loc(인덱스이름사용), iloc(정수사용)
- 열단위 추출: 컬럼명
- 불리언 인덱싱(True 혹은 False의 배열으로 데이터 필터링)
- 인덱스 설정

3. **데이터 조작**
- 데이터 추가, 삭제: df['새로운 열 이름'] = 리스트, df.drop(labels, axis)
- 데이터 정렬: df.sort_index() -> 인덱스 기준, df.sort_values(by="열 이름") -> 특정 열값 기준
- 데이터 변환: 자료형 df[열 이름] = df[열 이름].astype(바꿀 자료형), 값  df.replace(바꾸고 싶은 값, 바꿀 값), 행과 열 df.transpose()
- 데이터 프레임 병합: concat(동일구조), merage(공통항목)

### 집계함수
여러 행으로부터 하나의 결괏값을 반환하는 함수 
- 대표적인 집계함수
1. 평균 .mean()
2. 중앙값 .median()
3. 표준편차 .std()
4. 합계 .sum()
5. 최대값 .max()
6. 최소값 .min()

- 그룹화: 전체 행이 아닌 일부 행을 대상으로 집계
df.groupby('column명1')[’column명2'].집계함수()를 사용해서 특정 그룹으로 묶고 집계함수 적용
