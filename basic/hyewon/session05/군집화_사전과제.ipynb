# 지도학습과 비지도학습

- 비지도학습은 목표가 명확하게 정의된 구체적이고 좁은 범위의 문제를 풀기에는 지도학습보다 불리하지만, 목표 자체가 명확하지 않고, 데이터에서 새로운 특성을 찾는 등의 목적을 수행하기에 적합하다

- **비지도학습이 적합한 case**
1. 패턴이 아직 알려지지 않은 경우 -> 정답을 알 수 없음
2. 패턴이 계속해서 변하는 경우
3. 열린 문제를 해결하고 지식을 일반화해야 하는 경우

- **사용되는 분야**
1. 고객 세그먼트 분석: 고객의 성향 데이터를 바탕으로 비슷한 성향의 고객을 그룹단위로 묶어서 분석
2. 이상 거래 탐지: 정상 거래 패턴을 군집으로 정의해, 이외의 데이터는 이상치로 간주해서 탐지
3. 지연어 처리: 텍스트 데이터를 군집화해 분석하여, 유용한 인사이트를 도출




# 군집화

>데이터를 비슷한 특성을 가진 그룹, 군집으로 나누는 비지도학습으로 각 그룹이 어떻게 구성되오 있는지 파악하면서 전체 데이터 구조를 이해할 수 있음. 

![alt text](<스크린샷 2026-09-19 오후 10.55.23.png>)

### 군집화의 목표
1. 응집도 최대화: 같은 군집끼리는 비슷하게
2. 분리도 최대화: 서로 다른 군집은 최대한 분리되도록


### 군집화 과정
1. 피처 선택 또는 추출: 군집화에 사용할 데이터의 특징을 선택
    
    -> 변수유형, 거리, 차원축소
2. 군집화 알고리즘 선택: 데이터의 특성과 목표에 따라 적합한 알고리즘 선택

    -> 계층적, k-means, DBSCAN, GMM, 선택가이드
3. 군집 유효성 검증: 얼마나 군집화가 잘 됐는지

    -> 실루엣 계수, 던 지수
4. 결과 해석: 각 군집이 지닌 특성을 분석, 해석



## 피처선택 

[군집화의 주요 고려사항]
1. 변수 유형 이해

   군집화에 사용될 피처의 종류와 특성을 명확하게 이해해야 함, 변수 종류가 연속형 or 명목형인지에 따라 적합한 군집화 알고리즘이 달라짐

   1. 연속형 변수

      - 예시: 키, 나이, 몸무게, 소득
      - 거리측정: 유클리디안, 맨하탄
      - 알고리즘: K-means, 계층적 군집화 

   2. 명목형 변수

      - 예시: 성별, 지역, 직업
      - 거리측정: 해밍, 자카드
      - 알고리즘: K-modes, 계층적 군집화 

   3. 혼합형

      - 예시: 연속+먕목
      - 거리측정: Gower distance
      - 알고리즘: K-prototypes, 계층적 군집화

2. 거리/유사도 정의와 측정

    군집화는 데이터간 거리나 유사도를 기반으로 그룹 형성

3. 차원 축소

    변수가 많아질수록 모델이 복잡해지고, 효율성과 성능이 떨어져 유사 변수 통합이나 필요없는 변수 제거를 통해 변수 개수를 줄이면 좋음


[유형마다 달라지는 이유]
- 거리: 순서가 있는지, 그 간격이 의미가 있다면 빼기가 가능한 거리방식 사용.. 순서가 없고 빼기가 불가한 명목형은 세는 방식 사용

- 알고리즘: k-means는 중심점을 평균으로 계산해서 평균이 정의되지 않는 변수에는 쓰지못함. 이 경우에는 중심점을 최빈값으로 사용하는 k-modes 사용

- 목표: 다차원 데이터에서 가장 정보량(분산)이 큰 방향을 찾고 그 방향을 기준으로 데이터를 투영하여 차원을 줄이는 것.

- 원본의 고차원 데이터에서 최대 분산 방향을 찾아 상관관계를 감소시키고, 저차원 공간에 투영하는 것. 이렇게 새로 파생된 성분을 주성분이라고 함!

# 군집화 알고리즘

### 1. 계층적 군집화
> 가장 가까운 것부터 둘씩 차례로 합쳐나가면서 무엇이 언제 뭉쳤는지를 덴드로그램으로 기록하는 군집화 방법 

**덴드로그램이란** 가까운 두 개체 또는 군집을 점차적으로 병합해가는 과정을 시각적으로 표현한 트리구조  

![alt text](<스크린샷 2026-09-20 오전 12.53.52.png>)

- 덴드로그램의 수직축(높이) = 두 군집이 병합될 때의 군집 간 거리 및 유사성

- 군집의 개수를 클러스터링이 종료된 후 선택해도 된다는 장점.
- 트리 구조 상향식: 응집형 계층적 군집화(병합 클러스터링)
    
       각 샘플이 독립적인 클러스터로 시작, 하나만 남을 때까지 가장 가까운 클러스터를 병합
- 트리 구조 하향식: 분리형 계층적 군집화(분할 클러스터링)

       전체를 포함하는 하나의 클라스터에서 시작해, 유사성이 낮은 데이터를 더 작은 클러스터로 나눔

![alt text](<스크린샷 2026-09-20 오전 12.59.28.png>)


##### [응집형 군집화 과정]
1. 거리 행렬 계산: 모든 데이터 포인트 간의 거리를 계산
2. 단일 클러스터 시작: 각 데이터 포인트를 개별 클러스터로 초기화
3. 가장 가까운 클러스터 병합: 정의된 기준에 따라 가장 가까운 두 클러스터를 병합
4. 거리 행렬 업데이트: 새 클러스터와 나머지 간 거리를 다시 계산
5. 반복

##### 연결 방법
1. 단일 연결법 / 최단 연결법
 - 군집 A와 B 사이의 가장 가까운 두 점 간 거리로 군집 간 거리를 정의
2. 완전 연결법 /  최장 연결법
 - A와 B 사이의 가장 먼 두 점 간 거리로 군집 간 거리 정의
3. 평균 연결법
 - 군집 A와 B의 모든 점 쌍 사이 거리의 평균
4. 중심 연결법
 - 각 군집의 중심점 간의 거리
5. 중앙 연결 법
 - 두 군집의 중앙값 사이의 거리
6. Ward 연결법
 - 군집 내 제곱합이 가장 적게 증가하도록 병합


##### 계층적 군집화 장단점
- 장점
    1. k를 미리 정하지 않아도 됨
    2. 데이터의 구조 전체가 눈에 보임
    3. 거리만 정의되면 어떤 유형의 데이터에도 적용 가능
    4. 결과가 항상 같음

- 단점
    1. 느리고 메모리를 많이 먹음
    2. 한범 합치면 되돌릴 수 없음
    3. 이상치 민감
    4. 자르는 높이를 사람이 정해야 됨.


# K-means

>데이터를 정해진 개수(K)의 그룹으로 나누되, 각 그룹의 중심점과의 거리가 가장 가까운 데이터끼리 묶는 군집화 알고리즘

- 군집화에서 가장 일반적으로 사용되는 알고리즘
- 중심점을 선택해, 그 중심에 가장 가까운 데이터들을 묶어나가는 기법
- 군집을 중심점 하나로 대표하기에 중심을 구할 수 없는 데이터에는 쓸 수 없음

[k-means 주요 단계]

![alt text](<스크린샷 2026-09-20 오전 1.37.48.png>)

1. 데이터 표본들 중 랜덤하게 k개의 중심점을 초기 클러스터 중심으로 선택
2. 각 표본들을 가장 가까운 중심점에 할당
3. 각 클러스터에 할당된 표본들의 평균을 계산하여 중심점 이동
4. 클러스터 할당이 변하지 않거나 최대 반복 횟수에 도달할 때까지 2,3 반복

### 엘보우 방법
> 클러스터 내 SSE를 바탕으로 그래프를 활용해 최적 클러스터 개수 k를 추정하는 방법

![alt text](<스크린샷 2026-09-20 오전 1.41.27.png>)
급격히 기울기가 변하는 지점이 최적의 k

# DBSCAN

>밀도가 높은 지역의 데이터를 하나의 군집으로 묶고, 밀도기준을 만족하지 못한ㄴ 점은 군집에 포함시키지 않는 군집화 알고리즘

k-means는 K를 정해줘여하고, 이상치에 잘 휘둘리고, 원형만 잘 잡지만 DBSCAN은 k를 안정해도 되고, 이상치를 따로 빼고, 모양을 안가림

- 명시적으로 이상치를 지정해서 클러스터링에서 제외
- 다른 알고리즘 대비 이상치로부터 자유로움
- 어떤 데이터가 여러 군집의 특정 거리 내에 있으면 가장 밀도가 높은 군집과 그룹화 

[주요 단계]
1. 샘플분류
2. 클러스터 생성
3. 경계 샘플 할당 

[장단점]
- 장점
   1. 클러스터 모양을 원형으로 가정하지 않음
   2. 이상치에 민감하지 않음
   3. 모든 샘플을 클러스터에 할당하지 않음

- 단점
   1. eps,min_samples 설정에 많은 영향을 받음
   2. 연산량이 많아 속도가 느림
   3. 밀도가 서로 다른 군집들은 잘 잡지 못함 
   4. 차원의 저주의 영향을 받음

# GMM
> 데이터가 여러 다른 모양의 가우시안 분포로 구성되었다고 가정하고, 각 분포를 클러스터로 인식하는 군집화 방법

[특징]
- 모델기반 군집화: 데이터를 생성하는 통계적 모델을 가정
- 각 군집을 확률분포로 간주, 전체를 여러 확률 분포의 혼합으로 모델링
- 데이터의 생성 메커니즘까지 모델링 가능

[진행 과정]
1. 전체 데이터 셋의 분포 확인
2. 전체 데이터셋은 서로 다른 정규분포 형태의 확률 분포 곡선으로 구성되어있다고 가정
3. 여러개의 정규분포 곡선을 추출하고, 개별 데이터가 어떤 분포에 속하는지 결정. 이때 각각의 분포가 하나의 군집이 됨

=> k-means는 답을 주고, GMM은 답과 함께 얼마나 믿을만한지를 줌

## 어떤 알고리즘을 고를까?

- 군집을 몇개로 나눠야할지 모른다 => 계층적 군집화
- 군집이 대체로 동글동글하고 데이터가 크다 => K-means
- 군집 모양이 길쭉하거나 서로 겹친다 => GMM
- 이상치를 골라내는 게 목적이다 => DBSCAN
- 군집 모양이 초승달, 소용돌이처럼 이상하다 => DBSCAN
- 군집마다 밀도가 제각각 => HDBSCAN
- 경계에 걸친 데이터의 애매함까지 알고 싶다 => GMM

#### 주의사항
- k-means: 이상치 많을 땐 평균 기반이라 중심이 통쨰로 끌려감, 군집 크기가 심하게 불균형한 데이터
- 계층적 군집화: 데이터가 지나치게 많으면 거리 행렬 계산량이 감당 어려움
- DBSCAN: 피처가 수십 개 이상인 고차원 데이터는 밀도 개념을 무너트림
- GMM: 데이터가 정규분포와 거리가 멀면, 성능이 나빠짐
- 전부: 스케일링 안한 데이터는 사용하면 안됨

# 군집화 평가 방법 

> 정답이 없는 모델의 성능평가 어떻게?

1. 외부평가
   - 데이터에 정답 레이블이 존재할 때 사용 
   - 군집화 결과와 실제 레이블 비교해서 유사도 평가
   - ARI라는 군집화 결과와 실제 레이블 간의 일치도를 측정하는 지표로 1에 가까우면 유사하다고 평가
2. 내부평가
   - 정답 레이블이 없는 경우 사용
   - 군집 내부 응집도와 군집 간 분리도를 기반으로 품질 정량화
   - K-means는 SSE최소화, GMM은 로그우도 최대화
   - 범용 지표: 실루엣 계수, 던 지수


### 실루엣 계수
> 클러스터 내 샘플들이 얼마나 조밀하게 모여있는지 측정 

![실루엣계수식](<스크린샷 2026-09-20 오전 2.30.18.png>)

- -1에서 1사이의 값을 가짐
- 1에 가까울수록 군집화가 잘 됨 
- 평균 숫자 하나가 아니라 그래프로 봐야함

### 던 지수
> 클러스터 간 최소거리(분리도)와 클러스터 내 최대거리(응집도)의 비율로 클러스터링 품질 평가

![던지수](<스크린샷 2026-09-20 오전 2.34.13.png>)

- 값이 클수록 좋은 군집화(분리도는 크고, 응집도는 높은 것)
- 0 이상의 값을 가지며 상한은 없음
- 실루엣 계수와 종합적으로 판단해야 함
- 군집이 많아지면 게신 비용 증가
- 극단적인 값에 영향을 받음 = 이상치에 민감하게 반응
