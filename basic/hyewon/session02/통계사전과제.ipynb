# 통계 사전과제

### 자료의 척도
> 질적자료 : 연산의 의미를 가지지 못함
- 명목척도: 레이블의 이름으로 요소의 속성을 식별하려고 하는 척도 (펀드유형, 성별)
- 서열척도: 항목의 순서 또는 순위를 나타내는 특성을 지닌 척도 (만족도, 직급체계)

> 양적자료 : 수치로 표현
- 구간척도: 값들 사이에 순서가 있고 연산이 가능 (곱셈 나눗셈 x)
- 비율척도: 절대 영점과 임의의 측정단위로 구성. (0은 아무것도 존재하지 않음을 나타냄)

### 기술통계

> **기술통계란** 데이터를 있는 그대로 설명하고 이해하는 데 중점을 둔 통계방법

#### 1. 중심화 경향
- 평균: 데이터 값의 총합을 데이터의 개수로 나눈 값 (이상치에 민감)
- 중앙값: 데이터를 크기 순서대로 일렬로 정렬했을 때 한가운데에 위치하는 값 (이상치 덜 민감)
- 최빈값: 데이터에서 가장 자주 나타나는 값 (주로 이산형 데이터)

#### 2. 자료의 산포
- 범위: 최대값과 최소값 차이
- 분산: 평균으로부터 얼마나 멀리 퍼져 있는지 나타냄
- 표준편차: 분산의 제곱근
- 사분위수: 데이터의 표본을 네 개의 동일한 크기로 나눈 값
- 변동계수: 데이터의 절대적인 변동성이 아닌 상대적인 변동성을 나타내는 통계적 지표
- 첨도와 왜도: 첨도는 확률분포의 꼬리가 두꺼운 정도를 나타내는 척도, 왜도는 비대칭도

#### 3. 지료의 정리 및 시각화

## 추론통계?
수집한 데이터를 기반으로 미래의, 혹은 조사하지 못한 전체 정보를 추론하고 결론을 도출하는 방법을 다루는 분야

### 추정
> 모집단을 대표할 수 있는 일부를 표본으로 추출하여 분석한 통계량을 이용하여 모집단의 모수를 예측하는 과정

### 검정
> 표본을 추출하여 얻은 표본 통계량으로 모집단 모수에 대한 새로운 예상, 가설 등이 옳다고 판단할 수 있는지를 평가하는 과정

### 가설검정 과정
> 1. 가설 설정: 귀무가설과 대립가설 설정

- 귀무가설: 통용되고 있는 가설 

- 대립가설: 새롭게 주장하고 싶은 가설

> 2. 유의수준 설정: 1종 오류의 확률

- 1종 오류: 귀무가설이 실제로 참이었는데 기각할 확률

- 2종 오류: 귀무가설이 실제로 거짓인데 기각하지 못할 확률

> 3. 검정 통계량 산출: 가설 검정의 결과를 판단하는 데 사용되는 값.. 

- Z값: 큰 표본 크기에 대한 가설검정에 주로 사용

- T값: 작은 표본 크기에 대한 가설검정에 주로 사용

- F값: 두 표본 분산 사이의 차이를 비교하는데 사용

> 4. 가설 기각/ 채택 판단: 유의확률이 유의수준보다 작으면 귀무가설 기각

### 데이터의 자료형에 따른 검정 방법
>1. [수치형-수치형] 피어슨 상관분석

- 두 변수간 피어슨 상관계수가 얼마나 유의미한지 검정

>2. [범주형-수치형] F-test, T-test

- T: 범주가 2개일 때 (t값은 절대값 2를 기준으로 그보다 크면 차이가 있다, p값은 0.05기준)
- F: 범주가 3개 이상일 때 (f값은 2~3이상이면 차이가 있다, p값은 f의 결과가 얼마나 유의미한지)

>3. [범주형-범주형] Chi2-test

- 범주가 많아질수록 카이제곱 통계량이 커짐
- 자유도의 2배 이상이면 차이가 존재
