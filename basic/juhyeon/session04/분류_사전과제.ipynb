### 회귀와 분류의 차이

지도학습에는 회귀와 분류방식으로 나눌 수 있다. 연속적인 숫자값을 **예측**하는 것을 회귀, 입력된 데이터를 주어진 항목들로 나누는 것을 분류라고 한다.
데이터가 연속형 변수를 예측하기 위해 사용될 때는 회귀, 범주형 변수를 예측하기 위해 사용될 때는 분류 방식으로 학습한다.

### 분류 모델

분류에 쓰이는 대표적인 머신러닝 알고리즘에는 로지스틱 회귀, 결정 트리, 서포트 벡터 머신, 최소 근접 알고리즘이 있다.

- **로지스틱 회귀**
    - 로지스틱 회귀는 시그모이드 함수를 이용하여 출력값을 0~1 사이로 변환하고 추정하여 기준치에 따라 분류하는 것을 목표로 한다. 이는 이진 분류 문제를 해결하는데에 자주 사용된다.

    - 필요성 : 이진 분류 문제를 보면 합격 = 1, 불합격 = 0여부를 결정할 때 직선을 사용할 경우 분류 작업이 제대로 동작하지 않는다. 따라서 0과 1사이의 값을 가지기 위해 s자 형태의 함수를 활용한 로지스틱 함수를 사용해야 한다.

- **결정 트리**
    - 스무고개처럼 질문을 던져서 데이터를 계속 반으로 쪼개고, 그룹 안이 같은 답(label)으로 채워질 때 멈추는 분류 방법이다.

    - 구성 : 트리는 가장 상위에 위치한 노드 = root node를 시작으로 노드와 노드를 연결시켜주는 간선 edge으로 구성되어 있다. 노드는 3가지 항위항목으로 나눌 수 있는데 주어진 노드의 상위에 있는 parent node, 하나의 노드로부터 분리된 2개 이상의 child node, 자식마디가 없는 leaf node가 있다. 이 노드와 간선으로 여러 이어진 구성을 트리라고 하며 특정노드에서 leaf node까지의 edge 개수를 height라고 한다.

    - 이진 분할 : 데이터셋을 불순도가 낮아지는 방향으로 임계값을 두개의 child로 나누는 알고리즘이다. 이때 이진 분할은 데이터를 "어떤 기준으로 분류했을 때 동일한 객체들로 잘 모아질 수 있을지" 분류기준을 잘 정하는 것이 중요하다.

- **SVM(서포트벡터머신)**
    - 데이터를 분리하는 초평면 중에서, 데이터들과의 거리가 가장 먼 초평면을 선택하여 분리하는 지도학습 기반의 모델이다.

    - 데이터를 나누는 선은 무수히 많을 수 있는데, 그 중 데이터로부터 가장 멀리 떨어진 선을 찾는다면 새로운 데이터가 들어와도 안정적으로 분류할 수 있다. 따라서 SVM은 Margin(거리)이 최대가 되는 결정 경계를 찾는다.

    - 결정 경계와 가장 가까이에 있는 데이터를 서포트 벡터라 하고, 두 평면 사이의 거리를 Margin이라고 한다. 완벽한 분리가 불가능할 때는 여유 변수를 두어 일부 오차를 허용한다. 데이터 분류의 기준이 되는 경계는 결정 경계라 한다.

    - 서포트 벡터만으로 결정 경계를 만들기 때문에 데이터가 적을 때 효과적이고 연산량이 적으며, 커널을 사용하면 비선형 분류도 가능하다. 다만 데이터 전처리가 중요하고 데이터가 많아질수록 최적화 평가 과정이 늘어나 속도가 느려진다는 단점이 있다.

- **KNN(최소근접)**
    - 데이터로부터 거리가 가까운 k개의 다른 데이터 레이블을 참조하여 분류하는 알고리즘이다. 비슷한 특성을 가진 데이터끼리 서로 가까이 있다는 가정을 이용한다.

    - KNN의 가장 큰 특징은 별도의 학습 과정이 필요 없다는 것이다. 미리 모델을 만들지 않고, 새로운 데이터가 들어왔을 때 그 즉시 기존 데이터와의 거리를 계산해서 분류한다. k는 참조할 이웃의 개수이며, 동점을 방지하기 위해 보통 홀수로 설정한다.

    - 우선 데이터를 준비하고 k값을 설정한 후 새 데이터와 기존 모든 데이터 간 거리를 계산한다. 그리고 거리가 가장 가까운 k개를 선택하고 k개 이웃 중 가장 많은 클래스로 분류하여 동점이 나오면 더 가까운 이웃 쪽으로 분류하거나 무작위로 선택한다.

    - 학습 과정이 없어 구현이 간단하고 이해하기 쉬우며 데이터가 추가되어도 처리가 용이하다. 하지만 얘측할 때마다 전체 데이터와 거리를 계산해야 해서 데이터가 많아지면 느려진다는 단점이 있다. 또한 적절한 k값 설정이 어려워, 이상치나 비수치형 데이터에 취약하다.



### 분류 평가 지표

- **혼 동행렬**
    - 분류 모델의 예측 결과를 정확한 예측과 잘못된 예측으로 구분하여 나타낸 표이다. 실제 정답과 예측값의 조합으로 TP(정답 참, 참으로 예측 - 맞음), FN(정답 참인데 거짓으로 예측 - 틀림), FP(정답 거짓인데 참으로 예측 - 틀림), TN(정답 거짓, 거짓으로 예측 - 맞음) 총 4가지 경우로 나뉜다.

    - 정확도 : 전체 예측 중 맞게 예측한 비율. (TP+TN)/(TP+TN+FP+FN). 단, 정답 레이블 비율이 불균형하면 신뢰하기 어렵다는 단점이 있다.

    - 정밀도 : 참이라고 예측한 것 중 실제로 참인 비율. TP/(TP+FP). 관건은 FP이며, "거짓을 참으로 잘못 판단한 정도"를 알 수 있다.

    - 재현도 : 실제 참인 것 중 참으로 예측한 비율. TP/(TP+FN). 관건은 FN이며, "참을 거짓으로 잘못 판단한 정도"를 알 수 있다.

- **F1-Score**
    - 정확도와 재현율의 조화평균이다. F1 = 2×Precision×Recall / (Precision+Recall)

    - 조화평균을 쓰는 이유는 정확도와 재현율 중 하나만 극단적으로 높아도 크게 오르지 않도록 하여, 둘 사이의 균형을 효과적으로 평가하기 위함이다.

- **ROC/AUC**
    - Roc Curve : 임계값을 바꿔가며 TPR과 FPR의 변화를 그린 그래프이다. 곡선이 좌상단에 가까울수록 더 좋은 분류기이다.

    - AUC : ROC 곡선과 x축 사이의 면적으로, 모델 성능을 하나의 숫자로 나타낸다. 0.5 ~ 1 사이 값이며 1에 가까울수록 성능이 좋다.

### 하이퍼파라미터 최적화

- **하이퍼파라미터 최적화**
    - 모델 학습 과정에 반영되는 값으로, 학습 시작 전에 사용자가 직접 설정하는 변수이다.

    - 하이퍼파라미터 최적화 : Tuning을 거쳐 적절한 하이퍼파라미터를 찾음으로써 모델 성능을 향상시키는 것이다.

    - **grid search**
        - 정해진 범위에서 하이퍼파라미터를 모두 순회하며 가장 좋은 성능을 내는 값을 찾는 기법이다. 시간이 너무 오래 걸리고 넓은 범위, 큰 step을 활용해 범위를 좁힌다.

    - **random search**
        - 정해진 범위에서 하이퍼파라미터를 무작위로 탐색해 가장 좋은 성능을 내는 값을 찾는 기법이다. 속도가 grid search보다 빠르고 무작위라는 한계 때문에 정확도가 떨어진다.

    - **bayesian optimization**
        - 사전 정보를 바탕으로 최적 하이퍼파라미터 값을 확률적으로 추정하며 탐색하는 기법이다.
        
        - gaussian process라는 통계학 기반 모델로, 여러 하이퍼파라미터 조합에 acquisition function을 적용했을 때 가장 큰 값이 나올 확률이 높은 지점을 찾아낸다.
