### 통계

통계는 기술통계, 추론통계의 영역으로 나뉜다.
기술통계는 수치와 시각화를 통해 샘플을 요약하여 표현하는 방법이고 추론통계는 추정과 검정을 통해 샘플을 바탕으로 모집단을 추론하는 방법이다.
EDA가 기술통계의 범주라면 머신러닝의 회귀, 분류가 추론통계의 범주이다.

### 자료의 척도

자료란 설명과 해석을 위해 수집되고 분석되며, 요약되는 사실과 수치를 말한다.
이러한 자료는 양적 자료와 질적 자료로 구분되며 양적 자료는 구간척도와 비율척도, 질적 자료는 명목척도 서열척도로 구분된다.

- **질적자료**
    - **명목척도** : 레이블이나 이름으로 요소의 속성을 식별하는 척도
    
    - **서열철도** : 항목 간의 순서나 순위를 나타내는 특징을 가진 척도

- **양적자료**
    - **구간척도** : 값들 사이에 순서가 있고 간격이 일정하여 덧셈과 뺄샘 또는 평균과 같은 연산이 가능한 척도

    - **비율척도** : 절대영점과 임의의 측정단위로 구성하는 척도 

### 기술 통계

기술통계는 데이터를 이해하고 있는 그대로 설명하고 이해하는 것에 중점을 둔 통계 방법이다.

- **중심화 경향**
    - **평균** : 데이터 값 총합을 데이터 개수로 나눈 값이며, 이상치에 민감할 수 있어 데이터의 양 끝을 일정 비율 잘라낸 후 평균을 계산하는 절산 평균을 대신 사용하기도 한다.

    - **중앙값** : 데이터를 크기 순서대로 정렬할 때, 한 가운데에 위치한 값이다. 중앙값은 평균에 비해 이상치에 덜 민감하며 데이터의 중심 경향성을 나타내는 대안적인 지표다.

    - **최빈값** : 데이터에서 나타나는 빈도가 높은 값을 말하며, 데이터의 분포를 이해하는 데 도움을 준다.

- **자료의 산포**
    - **범위** : 데이터의 최대값과 최소값의 차이를 나타내는 지표로 전체 변동 범위를 나타내며 이상치에 매우 민감하다.

    - **분산** : 데이터 값들이 평균으로부터 얼마나 멀리 분포되어 있는지를 나타내는 지표이다.

    - **표준편차** : 분산에 제곱근을 취한 값이다.

    - **사분위수** : 데이터의 표본을 동일한 크기로 네 개를 나눈 값으로, 하위 25%(Q1), 하위 50%(Q2), 하위 75%(Q3)에 위치한 값으로 나눈다.

    - **사분위간 범위** : 데이터의 중앙 50%가 퍼져있는 범위이고, Q1 - (1.5 * IQR) 보다 작은 값, Q3 + (1.5 * IQR)보다 큰 값의 범위 이상치를 탐지할 때 유용하다. 

    - **변동계수** : 데이터의 상대적인 변동성을 나타내는 지표이다. 변동계수 = (표준편차/평균)*100

- **자료의 정리 및 시각화**
    - 히스토그램, 박스플롯, 파이차트 등을 활용해 데이터의 분포나 특징을 한눈에 파악할 수 있다.

### 확률분포

***확률변수***는 무작위한 곳에서 어떤 특정 사건이 발생할 때 나타날 수 있는 모든 경우의 수를 특정 값으로 대응시키는 함수를 의미하며, ***확률분포***는 확률변수가 특정한 값을 가질 확률을 나타내는 함수를 말한다.

확률 분포는 확률 변수의 속성에 따라 이산확률분포, 연속확률분포로 나눌 수 있다.

- **이산확률분포** : 확률 변수가 유한하거나 셀 수 있는 이산적인 값을 가지는 경우를 의미한다.

- **연속확률분포** : 확률 변수가 무한한 범위를 가지는 연속적인 값을 가지는 경우를 의미한다.

- **대표적인 분포모형**
    1. **정규분포**
        - 종의 모양 형태를 띠우며 평균을 기준으로 대칭하고 표준편차가 클수록 곡선이 평평해지는 형태를 띠운다. 평군과 표준편차만 알면 모양이 완전히 정해진다.

        - 모집단의 분포가 정규분포를 따를 때 표본평균의 분포 또한 정규분포를 따른다.

        - **중심극한정리** : n개의 평균의 분포는 n이 커질수록 정규분포에 가까워진다는 정리이다. 즉, 표본의 크기가 커질수록 표본평균분포는 모집단의 분포와 상관없이 정규분포에 가까워지므로 모집단의 분포와 상관없이 정규분포에 가까워진다.

        - **큰 수의 법칙** : 표본 크기가 커지면, 표본평균의 값이 모평균에 가까워진다.

    2. **Z분포**
        - 평균이 0이고 표준편차가 1이 되도록 맞춘 표준적인 정규분포이다.

    3. **T분포**
        - 종의 모양과 평균을 기준으로 대칭 형태를 띠며, 자유도가 커질 수록 표준정규분포는 N(0,1)에 수렴한다.

        - 표본으로 모평균을 추정할 때 사용되며, 표본 크기가 작을수록 정규분포보다 더 정확한 결과를 제공한다.

        - T분포의 자유도는 표본 크기에 비례한다. 자유도가 높을수록 T분포는 정규분포에 가까워지고 통계적 분석의 정확도가 높아질 가능성이 있다. 그러나 많은 자유도를 가지면 과적합으로 문제가 발생할 수 있어 적절한 선택이 중요하다.

    4. **카이제곱분포**
        - 0에서 시작해서 오른쪽으로 꼬리를 그리는 비대칭한 분포의 형태를 띤다. 자유도가 증가할수록 중심이 오른쪽으로 이동하고 모양이 더 대칭적으로 변화하여 정규분포에 가까워진다.

        - 표본분산의 분포와 관련된 지표로, 주로 독립성을 검정하고 적합도 검정할때 사용된다.
    
    5. **F분포**
        - 0에서 시작해서 오른쪽으로 긴 꼬리를 가진 비대칭 분포의 형태를 가지며, 자유도 증가할수록 중심이 오른쪽으로 이동하고 정규분포에 가까워진다.

        - 두 집단 간의 분산 차이를 비교하는 데 사용된다.

### 추론통계

수집한 데이터를 기반으로 미래, 혹은 조사하지 못한 전체 정보를 추론하고 결론을 도출하는 분야이며, 통계적 가설검정과 신뢰구간 등이 여기에 속한다.

- **추정**
    - 모집단을 대표하는 표본을 추출해 분석한 통계량으로 모집단의 모수를 예측하는 과정이다.

    - **추정량** : 모수를 측정하는 데 사용되는 표본통계량

    - **추정치** : 추정량을 실제 표본자료로 계산한 특정한 값이다.

    - **점추정**
        - 모수를 표본에 기초해 계산된 하나의 값으로 추정하는 방법이며 표본오차를 고려하지 못한다는 한계가 있다.

    - **구간추정**
        - 모수의 참 값이 포함될 것으로 기대되는 값을 일정한 범위와 신뢰도를 가지고 추정하는 방법이다.
        - 오차를 반영할 수 있어 점추정보다 일반적으로 널리 사용된다. 

- **검정**
    - 가설검정은 모집단의 실제 값이 얼마가 된다는 주장과 관련해서 표본의 정보를 사용해서 가설의 합당성 여부를 판정하는 과정이다.

    - **절차**
        - **가설 설정**
            - 귀무가설과 대립가설을 설정한다.
            
            - 귀무가설은 주로 기각하고자 하는 명제를 설정하고 대립가설은 내가 중장하고자 하는 명제를 설정한다.

        - **유의 수준 설정**
            - 유의수준 : 연구자가 결과를 판단할 때 허용할 수 있는 오류의 최대치로 주로 5%로 설정한다.
            
            - 1종, 2종 오류
                - 1종 오류 : 귀무가설이 실제로 참인데 기각할 확률
                - 2종 오류 : 귀무가설이 실제로 거짓인데 기각하지 못할 확률

        - **검정통계량 산출**
            - 검정통계량 : 가설 검정의 결과를 판단하는 데 사용되는 값이다.

            - 데이터의 분포/특징에 따라 사용하는 통계량이 다르다.

        - **가설 기각/채택 판단**
            - 유의확률 : 검정통계량에서 유도할 수 있는 또 다른 판단의 척도이다.
            
            - p < a -> 귀무가설 기각

            - p > a -> 귀무가설 기각하지 못함

- **자료형에 따른 검정 방법**
    - **피어슨 상관분석**
        - 두 변수 간 피어슨 상관계수가 얼마나 유의미한지 검정하는 방법이다.

        - H0 : 두 변수 간 선형 상관관계가 존재한지 않는다. / Ha : 존재한다.

        - 결과는 상관계수, p-value의 튜플 형태로 나오며, |r| < 0.3은 약함, |r| < 0.5는 중간, |r| > 0.5는 강함으로 본다.

        - 절대적 기준은 아니며 선형 관계만 파악 가능하다는 한계가 있다.

    - **T-test**
        - H0 : 두 그룹의 평균에는 차이가 없다. / Ha : 차이가 있다.

        - t값은 두 평균 간 차이를 표준오차로 나눈 값이며, 보통 절대값 2를 기준으로 그보다 크면 차이가 있다고 판단한다.

    - **F-test/ANOVA**
       - t검정은 범주가 2개일 때만 쓸 수 있어, 범주가 3개 이상이면 분산분석을 사용한다.
       
       - H0 : 각 그룹별 평균은 차이가 없다. / Ha : 차이가 있다.

       - F값은 집단간 분산을 집단 내 분산으로 나눈 값이며 보통 2~3이상이면 차이가 있다고 판단한다.

    - **Chi2-test**
        - H0 : 두 변수는 관련이 없다 / Ha : 관련이 있다

        - 두 변수가 독립이라고 가정했을 때 기대되는 값과 실제 관측된 값의 차이를 이용해 계산한다. 차이가 클수록 두 변수가 독립적이지 않을 가능성이 높다고 판단한다.

        - 카이제곱 통계량은 범주 개수가 많아질수록 커지는 특성이 있어, 절대값이 아닌 자유도의 약 2배를 기준으로 판단한다.
