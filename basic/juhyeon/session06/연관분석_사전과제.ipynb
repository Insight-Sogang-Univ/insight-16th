# 연관분석

### 연관분석이란?
  상품이나 서비스를 구매하는 일 안에서 존재하는 항목 간의 일정한 연관 규칙을 발견하는 분석. 즉, "이 상품을 산 사람은 저 상품도 많이 산다" 같은 패턴을 데이터에서 찾아내는 분석 기법이다.
정답값이 별도로 존재하지 않기 때문에 비지도 학습으로 분류한다.
 
- ##### **연관규칙 분석** :
    **if-then**의 형식을 띠고 **A->B**로 표현한다. if인 **조건절**과 then에 해당하는 **결과절**로 구성되며 조건절 또는 결과절은 각각 상호 배반의 **아이템 집합**으로 이루어진다.
    <br>
    예를 들어,"삼겹살을 구매하는 사람들은 쌈장도 함께 산다." 라고 한다면 '삼겹살 구매'가 조건절, '쌈장 구매'가 결과절이며 각각의 아이템 집합은 '삼결살', '쌈장'이 된다.
      
    - <mark style="background-color: rgba(135, 206, 235, 0.7);">**주의**</mark>
    <br>
    연관규칙은 관찰 데이터이고 방향성은 수학적 비율일 뿐이기에 상관관계를 파악할 뿐 인과관계를 알 수는 없다.
    

### 개념


- <mark style="background-color: rgba(135, 206, 235, 0.7);">**지지도**</mark>
    - 전체 거래항목 중 두개의 품목이 동시에 포함하는 거래의 비율로 일정 수준 이상이어야 한다.
    지지도가 낮다면, 패턴 규칙이 충분하지 않아 의미가 있다고 하기에 부족하고 현실 적용이 불가능하다. 따라서 지지도는 "이 규칙이 현실에서 의미가 있는가"를 판단하는 1차적인 기준이 된다.

    $$지지도 = P(A \cap B)$$


- <mark style="background-color: rgba(135, 206, 235, 0.7);">**신뢰도**</mark>
    - 품목 A를 포함하는 거래 중 품목 B를 포함하는 거래의 비율로 일정 수준 이상이어야 한다.
    신뢰도가 낮다면 A를 사도 B를 잘 안 사는, 의미없는 연관규칙 발생하고 이는 비용만 들고 효과가 없는 추천 실패로 이어진다. 따라서 신뢰도는 결국 "A가 일어났을 때, B도 일어날 것이라고 얼마나 확신할 수 있는가"를 판단한는 기준이 된다.

    $$신뢰도 = \frac{P(A \cap B)}{P(A)} = \frac{지지도}{P(A)}$$

- <mark style="background-color: rgba(135, 206, 235, 0.7);">**향상도**</mark>
   <br>
   - 나초 -> 콜라 지지도 : 10%

   - 나초 -> 콜라 신뢰도 : 75%
   <br>

   예를 들어, 매점 전체 결제의 85%에 콜라가 포함되어 있다고 가정하면, 나초->콜라 신뢰도는 75%로 꽤 높아 보이지만 아무나 콜라를 살 확률 85%과 비교하면 오히려 낮다는 걸 알 수 있다. 즉, 나초는 콜라 판매를 끌어올리기 보단 깎아먹고 있었던 것이다.
   따라서 신뢰도민으로는 B 자체의 기존 인기도를 반영하지 못해, A 때문에 B가 잘 팔린 것인지 원래 B가 잘 팔리는 것인지 구분할 수 없다. 그래서 '향상도' 평가 척도가 필요하다.

   $$향상도 = \frac{P(A \cap B)}{P(A)P(B)} = \frac{신뢰도}{P(B)}$$

   1. <mark style="background-color: rgba(135, 206, 235, 0.7);">**독립일 때 기댓값**</mark>
       - 사건 A와 B가 독립일 때 :
         $$P(A \cap B) = P(A) \cdot P(B)$$
         <br>
         즉, A를 샀다는 사실이 B를 살 확률에 영향을 주지 않는다는 뜻으로 향상도는 1이 된다.

         $$향상도 = \frac{P(A \cap B)}{P(A) \cdot P(B)} = \frac{P(A) \cdot P(B)}{P(A) \cdot P(B)} = 1$$
       
       - 사건 A와 B가 독립이 아닐 때 :
         $$P(A \cap B) = P(A) \cdot P(B|A) = P(B) \cdot P(A|B)$$
         <br>
         즉, A를 샀다는 사실이 B를 살 확률에 영향을 주거나, B를 샀다는 사실이 A를 살 확률에 영향을 준다는 뜻으로, 실제로 관측한 P(A∩B)에 따라 향상도는 1보다 크거나, 작다.

    2. <mark style="background-color: rgba(135, 206, 235, 0.7);">**실제 동시 발생 확률**</mark>
       <br>
       실제로 관측한 P(A∩B)가, A와 B가 함께 등장할 확률의 기댓값 P(A)·P(B)보다 
       - 크면 "A사면 B도 잘 산다"는 양의 연관성을 갖고

       - 작으면 "A사면 B는 잘 안 산다"는 음의 연관성을 갖는다.

    3. **결론**
       <br>
       규칙의 효용성은 지지도, 신뢰도, 향상도 세 가지를 **모두 반영하여 평가해야 한다.

    4. **총규칙수 공식**
       - 총규칙수 = 3ⁿ - 2ⁿ⁺¹ + 1

       - 3ⁿ : 아이템 하나마다 갈 수 있는 곳 1.조건절 2.결과절 3.아예 규칠에 사용 안한 곳

       - 2ⁿ⁺¹ : 조걸절이 비는 경우 X 결과절이 비는 경우

       - 1 : 둘다 비는 경우 -> 중복 제거

       - => 쓸모있는 것만 걸려내려고 지지도, 신뢰도, 향상도 기준을 두는 것이다.



### 연관 분석 알고리즘

- ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**Apriori 알고리즘**</mark>
   연관규칙을 찾기 위해서는 A->B로 표현한 가능한 모든 조합을 확인해봐야 하는데 A와 B가 단일 품목이 아니라 여러 품목 묶음일수도 있어서 상품 개수가 늘어나는 만큼 조합수 또한 기하급수적으로 증가한다. 이 모든 것을 계산하기란 비효율적이기 때문에 하나의 조합만 검사하고도 이에 파생되는 다른 조합들까지 후보에서 배제하는, 시간과 연산량을 효과적으로 줄이는 Apriori 알고리즘을 이용한다.

   Apriori 알고리즘은 상위 조합에서부터 차례대로 특정 조합이 자주 발생하지 않을 시 그를 포함하는 후속 조합들까지 모두 후보에서 배제하는 방식이다.
   <br>

   - **알고리즘 프로세스**
      1. 단일 항목 집합을 생성한 후 모든 개별 상품에 대해 지지도를 계산한다. 최소 지지도를 넘는 항목들만 다음 단계로 넘어간다.

      2. 1단계에서 선별된 상품들로 2개 항목의 집합을 생성한다. 이 조합에 대해 지지도를 계산하고 최소 지지도를 넘지 않는 조합을 제거한다.

      3. 이전 단계에서 남은 조합들을 바탕으로 3개 이상의 항목으로 구성된 조합을 만든다. 위와 같이 지지도를 계산하고, 최소 지지도를 넘지 않는 조합을 제거한다. 더 이상 조합을 생성할 수 없을 때까지 위 과정을 반복하고

      4. 마지막으로 남은 빈발 항목 집합들을 바탕으로 연관 규칙을 생성한다. 각 규칙에 대해 지지도, 신뢰도, 향상도를 계산해서 가장 유용하고 적합한 규칙을 선택한다.

   - **장단점**
       - 장점 : 원리가 직관적이고 단순해서 이해하기 쉽고, 연산량도 무차별 탐색보다 훨씬 줄어든다.

       - 단점 : 데이터가 아주 커지면 여전히 느려진다.

- ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**FP Growth 알고리즘**</mark>
   Apriori 알고리즘은 여전히 데이터가 커질수록 속도가 느려지는 단점을 FP-Growth는 이 단점을 보완한 알고리즘이다. 빈발 품목 집합을 효율적으로 찾아낸다.
   FP-Tree라는 구조를 이용하는데, Tree, Array, Linked List를 합쳐놓은 형태로, 데이터를 단 두번만 스캔하여 트리를 구성하고 후보 아이템 셋을 따로 생성할 필요 없이 트리만 만들면 빈발 항목 집합을 뽑아낼 수 있다.
   <br>

   - **알고리즘 프로세스**
      1. **데이터셋 확인** :
       <br>
       모든 거래를 확인해 각 아이템마다 지지도를 계산하고, 최소 지지도 이상인 아이템만 선택한다.
      
      2. **빈도순 재정렬** :
       <br>
       남은 항목들을 거래별로 빈도순으로 재정렬하고 가장 자주 구매된 항목이 먼저 오도록 정렬한다. 이를 f-list라고 한다.

       3. **FP-트리 생성** :
       <br>
       재정렬된 데이터를 바탕으로 FP-트리를 생성한다. 맨 처음 0 노드를 만들고, 각 거래를 하나씩 추가하며 트리를 만들어간다. 이때 빈도가 높은 아이템일수록 0에 더 가까운 노드가 되며 새로운 아이템이 나오면 부모 노드부터 새 노드를 만들고 이미 있는 항목이면 기존 노드의 카운트를 증가시키는 방식으로 확장한다. 이 과정을 모든 거래에 반복한다.

       4. **빈발 항목 집합 추출** :
       <br>
       완성된 FP-트리를 바탕으로, 특정 항목을 기준으로 부분 트리를 만들어 그 항목과 연관된 빈발 항목 집합을 빠르게 도출한다.

       5. **연관 규칙 도출** :
       <br>
       추출된 빈발 항복 집합을 바탕으로 연관 규칙을 만든다. 각 규칙에 대해 지지도, 신뢰도, 향상도를 계산해서 가장 유용한 규칙을 선택한다.

   - **장단점**
       - 장점 : 데이터를 두 번만 스캔하는 트리 구조이기 때문에 Apriori보다 훨씬 속도가 빠르다. 후보 아이템 셋을 생성할 필요 없이 트리만 구성하면 끝이다.

       - 단점 : 대용량 데이터셋에서 메모리를 효율적으로 사용하지 못한다. Apriori에 비해 설계하기 어렵고 지지도 계산은 무조건 FP-Tree가 먼저 만들어진 후에 가능하다.
