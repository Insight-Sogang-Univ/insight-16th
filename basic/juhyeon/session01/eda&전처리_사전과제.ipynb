### EDA 개념

EDA는  데이터 분석을 위해 데이터를 **여러 가지 방식으로 파악하는 과정** 단계이다.

EDA는 Exploratory Data Analysis의 약자로, **탐색적 데이터 분석 과정**을 말하며 데이터에 대한 **탐색과 이해**를 기본으로 가져야 함을 뜻한다.

도표, 그래프, 통계 등 다양한 각도에서 수집한 데이터의 특징을 파악한다.

- ##### EDA 대상
    - **일변량** : EDA를 통해 **하나의 변수**를 살펴봄으로써 데이터를 설명하고 그 안에 존재하는 패턴을 찾는 것 주요 목적이다.

    - **다변량** : EDA를 통해 **여러 변수**들 간의 관계를 살펴보는 것이 주요 목적이다.

- ##### EDA 종류
    - **시각화** : 차트 또는 그림 등을 활용하여 데이터를 확인하는 방법으로, 데이터를 한눈에 파악할 수 있다.

    - **비시각화** : 주로 Summary Statistics(그래픽 요소X)를 통해 데이터를 확인하는 방법으로, 정확한 값이 필요할 때 쓰인다.

- ##### EDA 유형
    - **일변량 비시각화** : 주어진 데이터의 distribution을 확인한다.
    
    - **일변량 시각화** : 주어진 데이터를 전체적으로 살펴본다.

    - **다변량 비시각화** : 주어진 둘 이상의 변수 간 관계를 확인한다.

    - **다변량 시각화** : 주어진 둘 이상의 변수 간 관계를 전체적으로 살펴본다.


### 데이터셋 파악하기

- ##### 데이터 프레임 보기
    - head(n)을 활용하여 상위 n개의 데이터를 보고 데이터프레임의 흐름을 파악한다.

- ##### 데이터 변수 확인
    - 타이타닉 데이터셋의 변수 정의를 확인한다.
    
    - 타이타닉 데이터셋의 변수 유형을 확인한다.
       - 범주형 (몇 개의 카테고리로 나뉘는 데이터)
          - 명목형 : 순서가 의미 없다.
          - 순서형 : 순서에 의미가 있다.

       - 수치형 (숫자로 표현되는 데이터)
          - 이산형 : 셀 수 있는 값
          - 연속형 : 연속적인 값

    - 타이타닉 데이터셋의 변수 데이터 형식을 확인하다.

- ##### 데이터 분포 확인
    - 원시 데이터의 평균, 최빈값, 중간값 같은 특성을 파악하기 위해 그래프를 활용하여 하나의 데이터 분포를 확인하고 분석할 수 있는 방법이다.
        - 산포도 : 데이터 포인트들이 흩어진 모양
        - 박스플롯 : 사분위수와 이상치를 한눈에 파악
        - 히스토그램 : 값의 분포를 막대로 표현
    
    - 원시 데이터의 분포를 미리 확인해두어, 이상치와 결측치를 파악한 후 전처리 아이디어를 얻을 수 있다.

### 전처리

데이터를 분석에 적합한 형태로 만드는 과정이다.
머신러닝을 포함한 모든 데이터 분석의 정확도는 **데이터 품질**에 달려있기 때문에, 이상값, 결측값, 노이즈를 수정하고 분석 목적에 맞게 데이터를 다듬는 전처리 과정이 반드시 필요하다.

- 정체 : 결측값, 이상값 문제 해결

- 변환 : 형식, 구조 바꾸기

- 필터링 : 오류 찾아서 걸러내기

- 통합 : 여러 데이터 합치기

- ##### 결측값
    - 결측값은 데이터 수집 과정에서 측정되지 않거나 누락된 데이터이며, 결측값이 있는 상태로 데이터 분석 모델을 만들 경우 변수 간의 관계가 왜곡될 수 있다. 그렇다면 모델의 정확성은 떨어지고 시각적인 표현에도 문제가 발생할 수 있다.

    - **결측값 확인**
        - 데이터프레임에서 결측값은 NaN, ?, 0 등 다양한 형태로 나타날 수 있다. 특히 0으로 표시된 경우, 이게 측정값의 0값인지, 결측치의 0값인지 구분해서 보아야한다.

    - **결측값 처리**
        - **삭제**
          - dropna() : 데이터프레임 내의 결측값이 포함된 레이블을 제거하는 매서드

          - axis
              - axis = 0 : 결측값이 있는 행 삭제
              - axis = 1 : 결측값이 있는 열 삭제

          - how
              - how='any' : 결측값이 하나라도 있으면 삭제
              - how='all' : 모든 값이 결측값일 때 삭제

          - subset
              - subset=['column명'] : 특정 열에 결측값이 있을 때 삭제
              - 단, axis=0 일 때만 사용 가능
        
          - 주의
              - 삭제는 결측값이 무작위로 발생했을 때만 사용한다. 그러지 않을 경우, 왜곡된 모델이 생성될 수 있다.

        - **대체**
          - fillna() : 평균값, 최빈값 등을 활용하여 채운다.

          - 일괄 대체 : 모든 결측값을 동일한 하나의 값으로 채우는 방식

          - 유사 유형 대체 : 범주형 변수를 기준으로, 비슷한 그룹끼리 묶어서 그 그룹의 평균값으로 채우는 방식

- ##### 이상치
    - 이상치는 관측된 데이터의 범위에서 특출나게 벗어난 값을 말한다.

    - **이상치 확인**
        - **통계 확인**
           - describe()
           
           - 평균, 표준편차, 최소/최대값 등을 한 눈에 보여준다.
        
        - **시각화를 통한 확인**
           - 박스 밖으로 튀어나온 점들의 형태로 이상치를 확인한다.

        - **Z-score로 확인**
           - 데이터를 평균 0, 표준편차 1로 정규화해서, 평군에서 얼마나 멀리 떨어져 있는지를 하나의 숫자로 표현한다.
           
           - 평균에 가까울수록 Z-score는 0에 가까워진다.

           - 보통 Z-score가 2~3을 넘으면 이상치로 간주한다.

        - **Tukey Fences로 확인**
           - 가장 널리 쓰이는 통계적 방법으로 사분위 범위를 기반으로 이상치를 판단한다.
              - 하한 미만 : Q1 - (1.5 X IQR)
              - 상한 미만 : Q3 + (1.5 X IQR)
           - 1.5배가 기본값이지만, 데이터 특성에 따라 3.0배처럼 더 넓게 잡을 수 있다.

    - **이상치 처리**
        - **전체 삭제** : 이상치가 Human error로 생겼다면 삭제한다.

        - **다른 값으로 대체** : 관측치 개수 자체가 적은 경우, 삭제하면 데이터 양이 너무 줄어들어 신뢰성이 떨어질 수 있기 때문에 이럴 경우 삭제 대신 평균값 등으로 대체하거나, 다른 변수를 활용한 예측 모델로 값을 추정해서 채운다.

        - **변수화** : 이상치가 자연 발생한 경우에는 무작정 삭제하면 오히려 현상을 잘못 설명하게 될 수 있다. 따라서 바로 삭제하기 보다는 원인을 파악해서 새 변수를 만들면, 이상치를 지우지 않고도 모델에 포함시킬 수 있다.

        - **리샘플링** : 자연 발생한 이상치를 별도로 분리해서 모델을 따로 만드는 방법이다.

### 피처 엔지니어링 (변수 가공)

컴퓨터가 문제를 더 잘 이해할 수 있도록 피처의 형태를 변형하거나 처리하는 과정이다.
도메인 지식과 기존 변수를 활용해서 **기존 데이터에 정보를 추가**하는 과정으로, 데이터 전처리의 마지막 단계이다. 새로운 데이터를 외부에서 가져오는 게 아니라, 이미 가진 데이터를 더 유용하게 가공하는 거라서, **분석가가 도메인을 얼마나 잘 아느냐**에 따라 결과 품질이 크게 달라진다.

- ##### 레이블 인코딩 vs 원핫 인코딩
    - 기계는 숫자만 이해하고 텍스트는 이해하지 못하기 때문에 문자열 값을 숫자로 바꿔주는 작업이 필요하다.

    - **레이블 인코딩** : 범주형 변수를 0부터 N-1까지의 정수로 변환한다.

    - **원핫 인코딩** : 범주형 변수를 이진 벡터로 변환하여 표현하려는 값의 자리가만 1, 나머지는 0으로 만든다.

- ##### 구간화
    - 연속적인 데이터를 그대로 쓰기보다, 일정한 구간으로 나눠서 이산적인 값으로 표현하는 게 더 효율적일 때가 있다. 정해진 규칙은 없고, 분석 목적에 맞게 분석가가 직접 기준을 정한다.

- ##### 파생변수 생성
    - 기존 피처를 다른 피처로 **변환**해서 새로운 변수를 추가한다. 정해진 원칙이 없어서 분석가의 도메인 지식과 데이터에 대한 이해도에 따라 결과가 크게 달라진다.

- ##### 스케일링
    - 변수마다 값의 범위가 다르면 숫자가 큰 변수가 분석 결과를 과도하게 지배해버릴 수 있다. 그래서 모든 변수의 값 범위를 일정한 수준으로 맞춰주는 **정규화**작업이 필요하다.

    - **StandardScaler** : 각 요소의 평균을 0, 분산을 1로 바꾸어 데이터가 정규분포를 따른다고 가정하는 기법이다.

    - **MinMaxScaler** : 모든 요소값을 0과 1 사이로 압축하여 데이터가 서로 다른 비율의 속성으로 구성되어 있을 때 같은 비율로 맞춰주며, 연산 속도를 높이고 알고리즘을 최적화하는 데 효과적이다.

### 데이터 시각화

데이터 시각화는 숫자, 텍스트로 표현되는 정보를 그래프, 차트, 도표 같은 시각적 요소로 바꿔서 데이터의 패턴, 관계, 추세를 쉽게 파악할 수 있게 하는 과정이다.

주로 쓰는 라이브러리 : Maplotlib, Seaborn

시각화를 하기 전에 데이터가 범주형인지 수치형인지 파악하고, 결측값, 이상치가 있는지 확인해야 하며 데이터의 특성을 파악하고 난 뒤에는 어떤 그래프를 그려서 데이터를 바라볼지 고민해야 한다.

- ##### 그래프 종류
    1. **boxplot** : 사분위수와 이사치를 함께 보여주고 박스 안은 중각 50%, 박스 밖 점들은 이상치를 나타낸다.

    2. **countplot** : 범주형 변수의 빈도수를 세는 그래프

    3. **histplot** : 도수분포표를 그래프로 그린 것. 수치형 변수를 구간별 빈도수로 보여준다.

    4. **barplot, pointplot** : 범주형 x값에 따른 수치형 y값의 평균을 보여준다.

    5. **displot, kdeplot** : 히스토그램을 연속적인 곡선으로 매끄럽게 이어준 그래프. displot에 kde=True를 주면 막대 위에 곡선이 겹쳐 나오고, kdeplot은 곡선만 단독으로 보여준다.

    6. **scatterplot, regplot** : scatterplot는 두 연속형 변수 간의 관계를 점으로 흩뿌려서 보여주고, regplot는 scatterplot에 회귀선까지 추가로 그려주는 그래프이다.

    7. **catplot** : categoryplot의 줄임말로, 수치형 데이터와 범주형 데이터 사이의 관계를 볼 때 사용한다.

    8. **pieplot** : 부분과 전체의 비율을 표현하는 그래프이며 모든 데이터를 합치면 전체(가 되는 경우에 적합하다.

    9. **heatmap** : 변수 간 상관계수를 색으로 직관적으로 보여주는 그래프이며 corr()로 상관계수를 구한 뒤 히트맵으로 표현한다.

    10. **violinplot** : 박스플롯 + 커널밀도추정 그래프를 합친 형태로, 박스플롯보다 데이터의 **분포 모양**을 더 정확하게 볼 수 있다.
