### 머신러닝

머신러닝은 소프트웨어를 학습하여 데이터에서 패턴을 발견하고 새로운 데이터를 예측하거나 결정을 내릴 수 있도록 하는 기술이다.

학습에 따라, 정답이 있는 데이터를 사용하여 예측값과 정답이 같도록 학습하는 지도학습, 정답 없이 데이터의 패턴이나 유사성으로 학습하는 비지도학습으로 나뉜다. 또한 반복적인 시행착오를 통해 보상을 최대화하는 강화학습을 기반으로 하기도 한다.

- ##### 모델링
    - 모델링은 예측을 만들어가는 전체 과정을 의미하며 데이터 속에서 패턴을 찾아내고 오차가 적은 최적의 공식을 만들 때까지 스스로 학습하는 과정이다.
    
    - 설계 -> 훈련 -> 검증 과정을 통해 최종 모델이 생성된다.

- ##### 지도학습
    - 예측값과 정답이 같아지도록 정답을 주고 학습을 시키는 방법

    - 데이터가 독립변수와 종속변수가 모두 존재하는 labeled data의 형태여야 한다.

    - 학습 단계 : 데이터 준비 -> 예측 -> 오차계산 -> 가중치 수정 -> 반복(데이터준비~가중치수정) -> 학습 완료 -> 성능평가

### 선형회귀

선형 회귀는 데이터를 회귀선을 찾는 과정이며 현재 데이터를 요약하는 동시에 미래 값을 예측하는 도구이다.

- ##### 선형회귀분석의 종류
    - **단순선형회귀** 
        - 하나의 독립변수만 존재

        - 1차원으로 데이터를 직선 형태로 표현 / y = wx + b

        - 잔차를 줄여나가며 직선이 될 수 있도록 회귀계수와 편향을 찾는다.

    - **다중선형회귀**
        - 독립 변수가 두 개 이상 존재
        
        - 독립 변수가 n개인 경우, n차원 공간에 표현된다.

        - 각 독립 변수에 가중치를 곱하여 일차결합으로 선형성 표현 -> 종속변수를 어마나 잘 설명하는지를 나타낸다.

        - '종속변수에 대한 독립변수들의 가중평균'

        - **다중선형회귀를 만족하기 위한 조건**
            - 선형성
            - 독립성 -> 독립적이지 않을 경우 다중공선성 발생
            - 오차항의 평균 = 0 or 가까워야한다.
            - 등분산성 : 오차항의 분산이 일정해야 한다.
            - 오차항의 공분산 = 0 (자기상관X)
            - 정규성 : 오차항이 정규분포를 따른다는 가정으로 이 가정은 선택적이다.


데이터를 설명하는 올바른 회귀선을 판단하는 방법에는 단순, 다중 둘 다 최소제곱법과 경사하강법을 사용한다.

### 다중공선성

- **다중공선성 확인법**
    - **상관계수**
        - 변수 간 통계적 정도를 수치로 나타낸 지표

        - 범위 : -1 <= r <= 1
            - r > 0 : 한 변수 증가 -> 다른 변수도 증가
            - r < 0 : 한 변수 증가 -> 다른 변수는 감소
            - r = 0 : 직접적인 연관성 없다

    - **VIF지수**
        - 회귀 모델의 결정계수를 이용해서 계산
        - VIF가 높을수록 다중공선성 존재 가능성이 크다.

- **다중공선성 대처 방법**
    - 변수 제거

    - 변수 변환 : 독립변수를 더하거나 빼도 문제가 없는 경우

    - 규제선형모델 활용 : 릿지, 라쏘, 엘라스틱넷 등의 방법으로 모델 복잡도를 줄인다.

    - PCA : 데이터의 차원을 축소하기 위해 사용되는 통계적 기법으로, 고차원 데이터에서 중요한 정보를 최대 보존하면서 복잡성을 줄일 때 사용한다.

### 규제선형모델

모델이 학습 데이터에 과적합하지 않도록 규제를 하고자 등장한 모델로, 계수의 크기를 조정하여 복잡성을 줄이고 특정 독립 변수에 과하게 의존하는 것을 막는다.

- **규제선형모델의 종류**
    - L2규제 : 회귀계수의 제곱에 대해 페널티를 부여하는 방식 (릿지회귀)
    
    - L1규제 : 회귀계수의 절댓값에 대해 패널티를 부여하는 방식 (라쏘회귀)

    - L2규제 + L1규제 : 엘라스틱넷 회귀

- **릿지 회귀** : 큰 회귀계수일 수록 강한 패널티를 주어 기둥에 따라 다른 힘을 가한다. 모든 변수를 유지하면서 크기를 적절히 줄여 과적합을 방지한다.

- **라쏘 회귀** : 회귀계수의 크기와 상관없이 같은 패널티를 가하기 때문에 릿지 회귀와는 달리 작은 계수가 0이 되는, 완전히 사라지는 변수가 발생한다. 따라서 중요한 변수만 남는다.

- **엘라스틱넷** : 불필요한 변수는 없애고 중요한 남은 변수들만 적절한 계수의 크기로 유지한다.

### 모델 평가방법

- **시각화**

- **통계지표**
    - 모델 유의성 검정
        - F-검정 : 모델 전체에 대한 평가로, F-검정이 클수록 통계적으로 유의미하며, prob가 0.05보다 작다면 회귀 모델이 유의미한 영향을 미친다.

        - T-검정 : 개별 변수의 유의성에 대한 평가이고 prob가 t통계량의 유의확률에 비해 0.05보다 작으면 해당 독립변수는 종속변수에 유의미한 영향을 미친다.

    - 모델 성능 평가
        - 결정계수 : 모델이 종속변수 변동을 얼마나 설명하는지 나타내는 지표 (0~1, 1에 가까울수록 좋다)

        - Adjusted R^2 : 변수 개수가 늘어나면 R^2이 자동으로 커지는 문제를 보정한 지표

        - AIC/BIC : 정보 기준. 값이 낮을수록 좋은 모델. BIC이 AIC보다 변수가 많을 때 더 엄격한 패널티 부과

- **성능 평가 지표**
    - 평균제곱오차 : 예측값과 실제값 사이의 오차 제곱의 평균을 의미

    - 평균절대오차 : 모델의 예측값과 살제값 사이의 절대값 오차의 평균을 의미

    - 평균제곱오차는 큰 오차에 민감하게 반응하여 이상치에 민감한 반면, 평균절대오차는 모든 오차를 동일하게 취급하기 때문에 이상치가 존재할 경우 전반적인 예측 성능을 유지한다.

- **변수 유의성 평가**
    - T-검정 : 개별 변수가 종속변수에 영향을 주는지 확인하는 방법으로, p-value < 0.05 이면 해당 변수가 유의미하다고 판단한다.
