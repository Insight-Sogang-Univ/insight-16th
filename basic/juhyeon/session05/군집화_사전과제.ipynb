# 군집화

### 군집화 개념과 기본과정

- 정답이 없는 데이터를 사용하여 학습하는 비지도 학습 기법으로, 비슷한 특성을 가진 그룹으로 나누는 것

- 지도학습과 달리 종속변수를 설정하지 않기 때문에 목표 설정없이 데이터의 내부의 패텬을 발견하고 추출한다.

- 데이터 간의 유사성을 유지하면서 동시에 다른 그룹은 구분될 수 있도록 만드는 것을 목표한다.

- 데이터의 피처를 선택하고 데이터의 특성과 목표에 따라 적합한 알고리즘을 선택하여 군집화가 얼마나 잘 되었는지 평가한다. 이 평가 하는 과정에서 부족한 부분이 있다면 앞 과정들을 최적의 결과를 찾아내기까지 여러 번의 반복 후 결과를 분석하고 해석한다.

### 피처 선택

1. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**변수 유형 이해**</mark> : 
    변수 유형에 따라 군집화의 알고리즘 선택이 달라진다.

    | 유형 | 특징 | 거리측정 | 알고리즘 | 기타 |
    | --- | --- | --- | --- | --- |
    | 연속형 | 수치로 측정 가능한 변수 | 유클리이단, 맨하탄 등 | K-means, 계층적 군집화 | 단위가 다르면 큰 값의 변수에 영향을 받기 때문에 스케일링 필수 |
    | 명목형 | 범주로 구분되는 변수 | 해밍거리, 자카드 | K-modes, 계층적 군집화 | 컴퓨터가 이해할 수 있는 언어로 변환하기 위한 원-인코딩 변환 |
    | 혼합형 | 연속형과 명목형의 혼합 | Gower distance | K-prototypes, 계층적 군집화 | 각 변수 유형에 맞는 거리 함수를 조합하여 사용 |
    
    <br>

    ##### ***유형에 따라 달라지는 이유**

    1) **거리**
    <br> 
    변수 유형에 따라 순서, 간격이 의미 있는지가 다르기 때문에 빼기, 세기 기준으로 거리 재는 방법도 달라진다.

    - **연속형** : 순서, 간격 의미O -> 빼기 가능 -> 유클리디안 거리 사용

    - **명목형** : 순서, 간격 의미X -> 빼기 불가능 -> 세는 방식의 해밍, 자카드 거리 사용

    - **혼합형** : 자 하나로 못 재기 때문에 변수별로 다르게 잰 다음 가중평균 사용

    2) **알고리즘**
    <br>
    알고리즘이 중심점을 평균, 최빈값 즉, 어떻게 계산하느냐에 따라 쓸 수 있는 변수 유형이 나뉜다.

    - **K-means** : 중심점을 평균으로 계산한다. 평균이 안 되는 명목형에는 사용불가

    - **K-modes** : 중심점을 최빈값으로 계산한다. 명목형에 사용가능

    - **K-prototypes** : 연속형은 평균, 명목형은 최빈값을 각각 사용하여 둘이 합친다.

    - **계층적 군집화** : 평균을 계산하지 앟고 거리만 보기 때문에 연속형이든 명목형이든 상관없이 둘 다 사용가능하다. 그러나 거리 표를 통으로 만들어야 하기 때문에 시간복잡도가 느리게 되고 효율성이 떨어진다. 그래서 계층적 군집화는 소규모 데이터에서 구조를 볼때 유용하다.


2. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**거리/유사도 정의와 측정**</mark> :
    데이터 간의 유사성 또는 비유사성을 수치로 계산하는 방법으로, 거리가 가까울수록 유사하고 멀수록 유사와는 거리가 멀다.

    - **유클리디안 거리**
     <br>
     가장 일반적인 거리 측정방법으로 직선거리를 의미하며 K-means에서 사용된다.
     <br>

     $$d(\mathbf{x}, \mathbf{y}) = \sqrt{\sum_{i=1}^{n} (x_i - y_i)^2}$$

    -  **맨하탄 거리**
     <br>
     각 차원의 차이의 절댓값의 합으로 이상치에 덜 민감하다.
     <br>

     $$d(\mathbf{x}, \mathbf{y}) = \sum_{i=1}^{n} |x_i - y_i|$$

    - **코사인 유사도**
     <br>
     벡터 간의 각도를 측정하는 방법으로, 텍스트 데이터나 고차원 데이터에 유용하다. 
     <br>

     $$\text{similarity} = \frac{\mathbf{x} \cdot \mathbf{y}}{||\mathbf{x}|| \times ||\mathbf{y}||}$$


3. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**차원 축소**</mark> :
    데이터는 그대로인데 차원이 늘어난다면 공간은 또한 커져서 모든 점은 비슷하게 멀어진다. 즉, 가까운 것과 먼 것의 구분을 할 수 없게 되어 거리 기반 알고리즘의 분별력이 감소한다. 이때 필요한 것이 차원을 축소하는 것이다.

    ***why?: 왜 이런 일이 생기는 걸까?**

    - <u>공간이 데이터보다 훨씬 빨리 커진다.</u> 그 결과 점들이 텅 빈 공간에 뿔뿔이 흩어져 있어, 이웃이라는 개념이 사라진다.

    - <u>가까운 점과 먼 점의 차이가 없어진다.</u>

    - <u>그 결과 군집화 자체가 무의미해진다.</u> 군집화는 가까운 것끼리 묶는 작업이기 때문에 전부 비슷하게 멀어진다면 어떤 기준으로 묶어야할지 알 수 없게 된다. 이 때문에 거리 기반 알고리즘의 분별력이 감소하게 되는 것이다.

    - <u>쓸데없는 피처는 노이즈로 작용하여 문제를 악화시킨다.</u> 유클리디안 거리는 모든 차원의 차이를 다 더하기 때문에 정작 중요한 신호 2개가 있더라도 상관없는 피처 100개를 넣으면 무작위한 차이들이 전부 합해져서 진짜 신호를 덮어버린다. 따라서 피처는 줄이는 게 좋기 때문에 차원을 축소한다.

    ***차원 축소 & PCA** :

    - 차원 : 변수의 개수이며, 변수를 줄일 경우 이를 차원 축소라고 한다.

    - 고차원으로 인해 발생한 노이즈를, 고차원 데이터를 저차원 공간으로 투영하고 핵심 정보를 지킴으로써 학습에 유리하게 하는 것이 목표

    - 차원 축소에는 선형 투영, 매니폴드 학습(비선형) 이 두 갈래의 방법이 존재한다. 선형 투영은 고차원에서 저차원으로 직선적으로 투영하는 방식이며 매니폴드 학습은 데이터 사이의 곡선 거리를 고려하여 학습하는 방식이다.

    - PCA는 데이터 분산이 큰 방향을 기준으로 데이터를 투영하여 차원을 줄인다.
      상관관계가 높은 피처들을 결합해서, 상관관계 없는 더 적은 수의 피처로 데이터를 줄이고 재표현하고 이렇게 새로 만들어진 피처를 주성분이라 한다.
        - p.s. 주의사항 : PCA는 표준화된 완전한 데이터에만 쓸 수 있어서, 결측값이 있으면 사용 불가능하고 해석 또한 어려우며 차원 축소 과정에서도 일부 정보가 손실될 수도 있다.

### 알고리즘 선택


1. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**계층적 군집화**</mark>
    가장 가까운 것부터 두 개씩 차례로 합쳐나가면서, 무엇이 언제 뭉쳤는지를 덴드로그램 즉, 나무 모양의 형태를 띈 그래프를 활용하여 기록하는 군집화 방법이다.
  <br>
  <br>
  예를 들어, 반에서 팀을 짠다고 비유를 들어 가정해보았을 때

   1. 가장 친한 두 명을 짝지어준다
  
   2. 그 짝을 하나의 팀으로 본다

   3. 다시 가장 친한 팀 둘을 합친다
  
   4. 반 전체가 한 팀이 될 때까지 반복한다.
  <br>
  <br>
  => 이 과정으르 기록한 것이 덴드로그램이다.
  <br>
  계층적 군집화 알고리즘은 군집의 개수를 미리 설정하지 않고, 클러스터링이 종료된 후에 원하는 군집의 개수를 선택할 수 있다.
  <br>
  <br>
  ***클러스터링** - 비슷한 특성을 가진 데이터끼리 자동으로 그룹으로 묶는 비지도 학습기법
<br>
<br>

    ***응집형 vs 분리형**
    
   | 구분 | 방향 | 특징 |
   | --- | --- | --- |
   | 응집형 | 상향식 | 개별데이터를 시작으로 데이터가 하나의 클러스터만 남을 때까지 병합 |
   | 분리형 | 하향식 | 전체를 포함하는 하나의 클러스터에서 시작해서 개별 데이터로 쪼갬 | 
   
   <br>

    ***응집형 과정**
    <br>
    1. 모든 점 간의 거리 행렬을 계산한다.

    2. 각 점들을 개별 클러스터로 초기화한다.

    3. 가장 가까운 두 클러스터로 병합한다.

    4. 새 클러스터와 나머지 간 거리 다시 계산한다.

    5. 하나로 병합할 때까지 3~4 반복한다.

    ##### **6가지 연결법**

    | 연결법 | 기준 | 특징 |
   | --- | --- | --- |
   | single | 두 군집 간 가장 가까운 점들의 거리 | 고립된 군집 찾는데 중점이고 이상치에 취약 |
   | complete | 두 군집 간 가장 먼 점들의 거리 | 내부 응집성 중시, 이상치에 취약 |
   | averge | 모든 점 쌍 거리의 평균 | 이상치 덜 민감, 계산량 많음 |
   | centroid | 두 군집 중심점 간 거리 | 계산 오래 걸림 |
   | median | 두 군집 중앙값 간 거리 | 극단값에 덜 영향받고 기하학적 구조 해석이 어려움 |
   | ward 연결법 | 병합 후 군집 내 SSE 증가분이 최소인 방향 | - |

   <br>

   ***덴드로그램 읽는 법**
   - 세로축 = 합쳐진 높이 = 그때의 군집 간 거리

   - 낮게 붙은 것 = 가까운 것

   - 긴 세로 막대 = 그만큼의 멀어질 때까지 합칠 상대가 없었다는 것. -> 경계

   <br>

   ***장단점**
   - 장점 : k를 미리 정하지 않아도 되서 그려놓고 나중에 자를 수 있다. 데이터 구조 전체가 한눈에 잘 보이고 거리만 정의되면 어떤 유형의 데이터에도 적용이 가능하며 결과가 항상 같다.

   - 단점 : 메모리가 많이 할당되기 때문에 실행 속도가 느리고 한번 합친다면 되돌릴 수 없게 된다. 그리고 이상치에도 민감하다.

2. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**K-means**</mark>
   데이터를 k개 그룹으로 묶어, 각 그룹의 중심점과의 거리가 가까운 데이터, 점끼리 묶는 알고리즘 
   <br>

   ***주요 단계**
   - 무작위로 k개의 중심점을 초기 중심으로 선택하고 각 데이터를 가장 가까운 중심점에 할당한다. 그 다음 할당된 데이터들의 평균으로 중심점 이동시키고 할당이 변하지 않을 때까지 앞전처럼 각 데이터를 가까운 중심점에 할당 및 평균 중심점 이동 이 과정을 반복한다.

   ***장단점**
   - 장점 : 직관적인 동시에 구현이 쉽다.

   - 단점 : k 결정이 어렵고 평균 기반이라 중심이 왜곡되어 아웃라이어에 민감하며 길쭉하거나 초승달 모양 같이 원형이 아닌 군집은 해석하기 어렵다.

   ***k 정하는 법 : 엘보우 방법**
   - k 값을 바꾸어가며 SSE를 바탕으로 그래프를 그리고 기울기를 파악한다. 기울기가 급격히 꺾이는 지점이 최적의 k이고 엘보우 방법은 그래프가 완만할 때 주관이 개입하기 때문에 실루엣 계수와 함께 봐야하는 특징이 있다.

3. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**DBSCAN**</mark>
   밀도 높은 지역의 데이터를 하나의 군집으로 묶고, 밀도 기준을 만족 못하는 점은 군집에서 제외하는 알고리즘
   <br>

   ***주요 단계**
   - 핵심 샘플, 경계 샘플, 잡음 샘플 이 세가지를 기준으로 나눈다.
       - 핵심 샘플 - 샘플 범위 반경 ε 안에 있는 이웃이 지정된 개수 이상

       - 경계 샘플 - ε 이내 이웃은 부족하지만 다른 핵심 샘플의 ε안에 있는 것

       - 잡음 샘플 - 핵심 샘플, 경계 샘플이 아닌 나머지 

   - 핵심 샘플들을 ε로 연결한 그룹을 클리스터로 생성한다.

   - 경계 샘플을 해당 핵심 샘플의 클러스트에 할당한다.

   ***장단점**
   - 장점 : 모양을 원형으로 가정하지 않고 이상치에도 민감하지 않으며 모든 샘플을 억지로 할당하지 않아도 된다.

   - 단점 : 연산량이 많아서 속도가 느리고 밀도가 높은 곳에 집중하는 특성이 있어 낮은 곳은 노이즈로 취급해 버린다.

4. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**GMM**</mark>
   데이터가 여러 가우시안 분포로 구성되었다고 가정하고 각 분포를 클러스터로 인식하는 방법
   가우시안 혼합 모델은 k-means와 달리 확률로 답하는 특징을 갖으며 데이터의 생성 매커니즘까지 모델링이 가능하다. 또한 여러 개의 분포로 모델링도 가능하다.

   ***주요 단계**
   
   전체 데이터셋의 분포를 우선적으로 확인하고 여러 개의 정규분포로 구성되어 있다고 가정한다. 그 다음 각 정규분포를 추출하여 개별 데이터가 어떤 분포에 속하는지를 결정한다.

5. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**알고리즘 종류 선택**</mark>
   | 상황 | 도구 | 이유 |
   | --- | --- | --- |
   | 몇 개로 나눠야 할까 | 계층적 군집화 | 덴드로그램을 그리고 난 후 자를 위치 고르기 |
   | 군집이 동글거나 데이터가 클 경우 | k-means | 직관적이고 빠르며, 대용량에도 유용 |
   | 군집 모양이 길쭉하거나 겹친다 | GMM | 타원형, 중첩 구조를 확률로 처리 |
   | 이상치를 셀렉이 목적 | DBSCAN | 이상치에 -1을 붙여서 명시적으로 분리 |
   | 군집 모양이 초승달이거나 소용돌이 모양일 경우 | DBSCAN | 모양 상관없이 밀도로만 판단 |
   | 군집마다 밀도가 다른 경우 | HDBSCAN | DBSCAN의 단일 밀도 기준의 한계를 보완하기 |
   | 경계 위에 걸쳐진 데이터의 애매함 또한 알고 싶은 경우 | GMM | predict_proba를 사용해서 소속 확률을 반환 |

   <br>

   ***주의 할 점**
   - 이상치가 많은 데이터에 k-means을 사용하면 중심이 통째로 끌려가기 때문에 사용을 하지 않는다.

   - 군집 크기가 심하게 불균형한 데이터에 k-means는 큰 군집을 쪼개려는 특성이 있어 사용을 자제하여야 한다.

   - 계층적 군집화는 데이터가 수만 건 이상일 때 거리를 계산하기 위한 행렬 계산량이 매우 많기 때문에 사용하지 않는다.

   - GMM의 경우, 데이터가 정규분포와 거리가 멀때 가정이 틀리면 성능이 나빠질 수 있어 사용하지 않는다.


### 유효성 검증 & 결과 해석


지도 학습은 정답을 주고 예측값과 실제값을 비교하는 방식이자만 군집화는 비지도학습으로 정답이 없다. 따라서 군집화는 정답이 있느냐 없느냐에 따라 평가 방식이 갈린다.

1. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**외부 평가 vs 내부평가**</mark>
   | 구분 | 조건 | 방법 |
   | --- | --- | --- |
   | 외부 평가 | 정답이 있는 경우 | 군집화 결과와 실제 레이블을 비교하여 유사성을 평가 |
   | 내부 평가 | 정답이 없는 경우 | 데이터 자체의 구조를 기반으로 판단 -> 군집 내부의 응집도와 군집 간의 분리도를 기반 | 
   
   <br>
2. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**실루엣 계수**</mark>
   클러스터 내부 샘플들이 얼마나 조밀하게 모여 있는지를 측정하는 도구이다.

   $$s^{(i)} = \frac{b^{(i)} - a^{(i)}}{\max\{b^{(i)}, a^{(i)}\}}$$

   <br>
   - a^(i) = 샘플과 같은 클러스터 내의 다른 점들과의 평균 거리 (응집도)

   - b^(i) = 샘플과 가장 가까운 다른 클러스터의 점들과의 평균 거리 (분리도)

   - 계산 방식 : (분리도 - 응집도) / 둘 중 큰 값

   - 계수 해석 
       - 분리도가 응집도 보다 클 경우, 계수 값은 1에 가까워진다. -> 자기 클러스터에 잘 속하고 다른 클러스터와의 거리는 멀리 떨어져 있음을 의미하기 때문에 군집화가 잘 되어있는 상태이다.

       - 분리도와 응집도가 같을 경우, 계수 값은 0에 가까워진다. -> 클러스터 경계에 위치해서 거리가 애매모호함을 뜻한다. 따라서 분리가 명확하지 못한 상태이다.

       - 분리도가 응집도 보다 작을 경우, 계수 값은 -1에 가까워진다. -> 잘못된 클러스터에 할당되어 있어 군집이 섞여있음을 의미하기 때문에 군집화가 잘못되어 있는 상태이다.

       => 평균 실루엣 계수가 0.5 이상이면 보통, 0.7 이상이면 군집화가 잘된 상태이다.

       => 주의 : 실루엣 계수는 평균이 같더라도 내용은 다를 수 있기 때문에 평균만 반영하지 않고 군집별 분포와 함께 반영해서 봐야 한다.

3. ##### <mark style="background-color: rgba(135, 206, 235, 0.7);">**Dunn Index**</mark>
   클러스터 간 최소 거리와 클러스터 내 최대 거리의 비율로 품질을 평가하는 
   
   $$\text{Dunn Index} = \frac{\min_{i \neq j} d(C_i, C_j)}{\max_{1 \leq k \leq K} \delta(C_k)}$$

   <br>
   - d(C_i, C_j) = 클러스터 C_i와 C_j 사이 거리 (분리도)

   - δ(C_k) = 클러스터 C_k 내부에서 가장 멀리 떨어진 두 점 간 거리 (응집도)

   - 해석
       - 값이 커질 수록 군집화가 좋은 상태를 뜻하기 때문에 군집 간 거리는 크고 내부 거리는 작을 수록 좋다.

       - 0 이상의 값을 가지고 상한은 없다.

       - Dunn Index는 이상치 하나에 흔들리기 쉬운 반면, 실루엣 계수는 샘플 단위 평균으로 상대적 안정적이기 때문에, 서로의 약점을 보완하기 위해서 실루엣 계수와 함께 종합적으로 판단하는 것이 좋다. 

   - 단점 : 군집 수가 많아지면 계산의 정도도 증가하게 되어 비용이 클 수 커지고 극단적인 값에 영항을 많이 받는다. 또한 이상치에 민감하여 오히려 수치가 떨어질 수 있다.
