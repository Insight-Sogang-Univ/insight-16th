### 판다스의 개념

- pandas는 데이터 분석 활용에 특화된 파이썬 라이브러리로 행과 열로 구성된 테이블의 형태의 데이터 구조이다. DataFame을 기본적으로 사용하고 있다.

- 행렬이나 다차원 배열을 활용하여 통계 계산이나 수치 연산 등을 처리할 수 있도록 돕는 넘파이의 확장판이다.

### 판다스의 라이브러리를 불러오는 방법

- 라이브러리는 미리 만들어진 기능들이 들어 있는 도구 상자이며 일반적으로 "import pandas as pd"를 활용하여 판다스를 불러온다.

- 불러온 라이브러리 안에 필요한 기능을 꺼내 쓰기 위한 도구로 "pd.read_csv('파일 경로')"을 활용한다.

### 데이터프레임이란?

- 컴퓨터를 활용해 정보를 정리하고 저장하는 도구이며 데이터의 위치를 나타내는 행과 열로 구성된 테이블 형태로 2차원 배열 구조이다. 여러 개의 시리즈가 모여 하나의 데이터프레임을 구성하며 컬럼 단위로는 값들의 데이터 타입이 동일하여야 한다. 하지만 컬럼 간의 데이터 타입은 달라도 영향을 가지지 않는다.

- 시리즈는 값과 인덱스로 구성되어 있는 1차원 배열이며 인덱스 레이블을 따로 지정하지 않았을 때, 0부터 정수형 인덱스가 자동으로 부여된다.

- DataFrame에서 컬럼을 하나만 뽑아내면 그 결과가 바로 시리즈 객체가 된다. 다만 시리즈를 구성하는 값은 모두 같은 데이터 타입이어야 하고 숫자와 문자가 섞여 있으면, dtype이 object로 표시된다. 이때 object는 문자열 타입이 아니라 어떤 파이썬 객체든 담을 수 있는 상자 역할을 한다. 이렇게 타입이 통일되지 않은 컬럼은 산술 연산이 불가능하여 연산을 하려 할때, astype(int) 나 pd.to_number() 등으로 먼저 데이터 타입을 숫자로 변환해줘야 한다.

### 데이터를 확인, 추출, 조작하는 방법

1. 데이터 확인 
- 판다스에서는 방대한 데이터를 처음부터 전체 다 보는 대신 일부만 미리 확인하거나 요약함수로 빠르게 파악한다.
- head(n)과 tail(n)으로 앞뒤 n개 행을 확인하고, shape로 행과 열 개수, info()로 종합 정보, describe()로 수치형 열의 통계값, isnull()로 결측치 여부, dtype로 열별 데이터타입을 확인할 수 있다.
- count()로 결측치를 제외한 행 수, value_counts()로 특정 열 값들의 빈도수, unique()로 특정 열의 고유값을 종류별로 하나씩 확인할 수 있다. 

2. 데이터 추출
- 방대한 데이터에서 분석에 필요한 부분만 골라내는 작업으로, 단일 항목을 지목할 때는 인덱싱을, 범위를 지정해서 여러 항목을 뽑을 때는 슬라이싱을 사용한다.
- 행 단위 추출은 loc과 iloc으로 하는데, loc은 인덱스 이름을 기준으로 하고 iloc은 정수형 위치 기준으로 한다. 
- 열 단위로만 추출할 때는 컬럼명으로 바로 인덱싱하는 게 더 직관적이지만, 값을 수정할 때는 반드시 loc을 사용해야 한다.
- 특정 조건에 맞는 데이터만 걸러낼 때는 불리언 인덱싱을 사용하며 True/False 배열로 필터링하고 and, or로 조건을 여러 개 묶을 수 있다. loc은 불리언 인덱싱이 가능하지만 iloc은 정수형 위치만 받아 불가능하다.
- 인덱스는 기본적으로 0부터 시작하는 정수형이지만 set_index, reset_index, reindex 등으로 원하는 대로 설정하거나 변경할 수 있다.

3. 데이터 조작
- 특정 리스트를 열로 추가하거나 drop으로 열을 삭제할 수 있으며, sort_index()로 인덱스 기준 정렬, sort_values(by="열 이름")로 특정 열 값 기준 정렬이 가능하다.
- 특정 컬럼의 데이터 자료형을 astype()으로 변환할 수 있고, replace()로 값 자체를 바꿀 수 있으며, transpose()나. T로 행과 열을 서로 맞바꿀 수 있다.
- 데이터프레임을 병합하는 방법은 두 가지인데, concat은 동일한 구조의 데이터프레임을 단순히 행 또는 열 방향으로 이어붙이는 방식이고, merge는 공통 항목을 기준으로 병합하는 방식으로 on 인자로 기준 열을 지정하며 how 인자에 따라 결합 방식이 달라진다.

### 집계함수를 사용하는 방법

- 여러 행으로부터 하나의 결과값을 반환하는 함수로, 각 컬럼의 평균, 표준편차, 합계 등 대표값을 구할 때 주로 사용하며 결측치는 무시하고 계산한다.
- 대표적으로 mean(), median(), std(), sum(), max(), min()이 있다.
- 집계함수는 기본적으로 모든 행을 대상으로 연산하지만, 일부 행을 그룹으로 묶어 집계하면 더 유용한 인사이트를 얻을 수 있는데, 이때 사용하는 것이 groupby이다.
- groupby는 특정 기준으로 그룹을 묶어주는 메소드이지만 그 자체만으로는 아무 일도 일어나지 않고 집계함수와 함께 사용해야 하고 df.groupby('column명1')[’column명2'].집계함수()는 결과 시리즈로, df.groupby('column명1')[[’column명2']].집계함수()는 결과를 데이터프레임 형태로 반환한다.
