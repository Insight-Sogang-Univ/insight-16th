### **머신러닝**
- **머신러닝이란?**
    - 인공지능의 한 분야
    - 모델이라는 소프트웨얼르 학습하여 새로운 데이터를 예측하거나 결정을 내릴 수 있도록 하는 기술
    - 데이터에서 패턴을 발견, 이 패턴을 바탕으로 새로운 데이터에 대해 예측하는 것을 가능하게 함

- **머신러닝의 종류**
    - 지도 학습 : 정답(label)이 있는 데이터를 사용. 기계의 예측값이 데이터의 정답과 같아지도록 학습시키는 것. (회귀, 분류)    
    - 비지도 학습 : 정답(label)이 없는 데이터 사용. 데이터의 패턴 혹은 유사성을 학습시키는 것. (군집화)
    - 강화 학습 : 시행착오를 반복하여 보상을 통해 상은 최대화, 벌은 최소화하는 방향으로 행위를 강화하는 학습. (알파고, 자율 주행)

### **모델링**
- **모델이란??**
    - 학습의 결과물. 학습을 마친 인공지능 그 자체(예측기)
    - 데이터를 학습하며 찾아낸 패턴이 저장된 상태
    - 세로운 데이터를 넣었을 때 답을 내놓는 예측기 역할

- **모델링이란??**
    - 예측기(공식)를 만들어가는 전체적인 과정
    - 아무것도 모르는 기계가 주어진 데이터 속에서 스스로 패턴을 찾아내, 오차가 작은 최적의 공식을 완성할 때까지 학습하는 과정

- **모델링 과정**
1. 설계 : 데이터의 특성에 맞추어 어떤 알고리즘으로 학습할지 결정하는 단계
2. 훈련(Training) : 데이터를 기계에게 계속 보여주며 정답을 맞히도록 수정하는 단계
3. 검증(Evaluation) : 훈련시킨 모델이 새로운 데이터에서도 정답을 잘 맞히는지 테스트하는 단계    

### **지도 학습**
- **지도 학습이란??** : 정답 레이블을 함께 주고 학습을 시키는 방법. 예측값과 정답이 최대한 같아지는 방향으로 학습

- **지도 학습의 과정**
1. 데이터 준비 : 독립변수(입력)와 종속변수(정답)를 갖춘 데이터 마련
2. 예측 : 컴퓨터가 독립변수를 보고 종속변수를 맞춰봄
3. 오차 계산 : 예측값과 실제 정답의 차이 = 잔차. 이 잔차를 모아 만든 지표를 손실(loss) 이라고 함.
4. 가중치 수정 : 오차가 줄어드는 방향으로 내부 수치(가중치)를 스스로 조정
5. 반복 : 2 -> 3 -> 4 의 과정을 수만 번 되풀이하며 오차를 점점 줄여나감
6. 학습 완료 = 함수 획득 : 데이터 사이의 통계적 관계가 하나의 함수 형태로 남음. 덕분에 학습 때 본 적 없는 새 데이터가 들어와도 규칙에 따라 답을 낼 수 있음.(일반화)
7. 성능 평가 : 학습에 쓰지 않은 별도의 테스트 데이터로 얼마나 잘 맞히는지 측정

**기본적으로,** 지도 학습을 위해서는 반드시 데이터가 **labeled data**의 형태여야 한다
1. labeled data : 독립변수(feature), 종속변수(target/label)가 모두 존재하는 데이터. 즉, 입력 데이터와 그에 대응하는 출력값이 있는 데이터를 의미
- 독립변수 : 머신러닝에서 예측에 사용되는 입력 변수
- 종속변수 : 머신러닝에서 예측하고자 하는 대상 변수
2. labeled data 를 train set과 test set으로 나누어 줌
- train set : 모델을 학습(훈련)시킬 때 쓰이는 데이터
- test set: 학습된 모델을 검증하기 위해 사용하는 데이터 -> unseen data
3. train set을 통해 모델을 학습
- 독립변수를 통해 추정한 예측값이 주어진 종속변수와 같아지도록(가까워지도록) 모델을 훈련시킴

-> 기계의 예측이 우리가 의도한 정답이 되도록 지도(supervise)하는 과정

### **회귀와 분류** : 지도 학습의 대표적인 예시
- 회귀 : 연속적인 숫자값을 예측. 주어진 데이터(X)를 기반으로 정답(Y)을 잘 맞히는 함수를 찾는 문제 (ex, 주택 가격 예측)
- 분류 : 입력된 데이터를 주어진 항목들로 나눔. 기존 데이터가 어떤 레이블에 속하는지 패턴을 알고리즘으로 인지한 뒤에 새롭게 관측된 데이터에 대한 레이블을 판별하는 것 (ex, 대출상환여부 예측)

**회귀와 분류의 차이**
- 예측하고자 하는 값(=종속 변수)가 연속형이면 회귀, 범주형이면 분류를 사용

### **회귀(Regression)**
1. **회귀 개요**
- 데이터에서 패턴을 찾아 미래 값을 예측하는 것. 통계, 머신러닝의 주요 개념
- 종속변수가 연속적일 때 사용
- 선형회귀, 비선형회귀, 규제회귀(릿지, 라쏘, 엘라스틱), 다항회귀 등 다양한 기법 존재

2. **선형회귀**
- **선형회귀란??** : 데이터를 가장 잘 설명하는 회귀선을 찾는 과정
    - 회귀선은 현재 데이터를 요약하는 동시에, 아직 관측되지 않은 미래 값을 예측하는 도구
    - 단순선형회귀(x가 한 개)와 다중선형회귀(x가 여러 개)로 나누어 볼 수 있음

- **가중치와 편향** (y=wx+b 기준)
    - 가중치 w (Weight)
        - 독립변수 x가 종속변수 y에 주는 영향력의 크기. 그래프에서의 기울기
        - 회귀계수라고도 부름

    - 편향 b (Bias)
        - 독립변수 x가 0일 때 나오는 기본값. 그래프에서의 y절편
        - 모델이 데이터에 더 유연하게 최적화될 수 있도록 위아래로 움직여주는 역할

- **선형회귀분석의 종류**
    - **단순선형회귀** : 하나의 독립변수(x)만 존재하는 선형 회귀
        - 데이터를 직선 형태(하나의 독립변수)로 표현하는 경우
        - 데이터를 가장 잘 설명하는 y=wx+b을 찾는 과정
        - 잔차를 줄여나가며 직선을 결정짓는 적절한 회귀계수(w)와 편향(b)을 찾아감
            - 잔차? : 실제값과 예측값 간 차이

### 4. **다중선형회귀**



$$
y = W_1 x_1 + W_2 x_2 + ... + W_n x_n + b 
$$

- 독립 변수가 두 개 이상인 경우
- 독립 변수가 2개인 경우 3차원 공간에 표현 + 회귀선은 평면
- 최소제곱법, 경사하강법 사용가능. 잔차는 관측치와 평면간의 차이



1) **가장 일반적인 형태**

$$
y_i = \beta_0+\beta_1X_{1i}+\beta_2X_{2i}+...+\beta_kX_{ki}+\varepsilon_i
$$

여기서 상수항과 오차항을 제외할 경우,

$$
y=w_0x_0+w_1x_1+\cdot\cdot\cdot+w_mx_m=\sum_{i=0}^mw_ix_i=w^Tx
$$

- 선형회귀식은 결국 종속변수 y에 대한 독립변수들의 가중평균으로 표현 가능
- 설명변수에 가중치를 곱해 더함으로써 y를 얼마나 잘 설명하는지를 나타낼 수 있음


2. **다중선형회귀의 기본 가정 6가지**
    1) 선형성 : 종속변수와 설명변수 간의 관계는 선형적이어야 함.
    2) 독립성 : 각각의 설명변수가 서로 선형독립적이어야 함. 선형독립적이지 않은 경우 다중공선성이 있다고 함.
        - 다중공선성 : 독립 변수들이 서로 강한 상관관계를 가져, 어떤 변수가 종속 변수에 가장 큰 영향을 주는지를 알 수 없는 상태
    3) 오차항의 평균은 0이다(0에 가까워야 한다)
        - 오차항(백색 잡음) : 실제 값과 예측 값 사이의 차이   
    4) 등분산성 : 오차항의 분산은 일정햐야 함. 오차항의 분산이 일정하지 않은 경우, 이분산성이 있다고 함. 이 경우에는 회귀 분석의 결과가 정확하지 않을 수 있음.
    5) 오차항은 자기상관되어 있지 않음 : 오차항의 공분산은 항상 0이어야 함. 공분산이 0이 아닌경우, 자기상관이 있다고 함.
        - 자기상관: 한 변수의 현재 값이 과거 값과 상관관계를 가지는 현상. 시계열 자료에서 많이 나타남
        - 모델이 데이터를 랜덤하게 예측하는 것이 아닌, 이전 데이터의 패턴을 학습하여 반복적인 오류를 만들어 낼 가능성이 높음.
    6) 정규성 : 오차항이 정규분포를 따른다는 가정
        - 가정 6은 위배되어도 다중선형회귀분석의 결과에는 큰 영향을 주지 않음.

3. **회귀 분석 전 vs. 후 검증해야 할 가정 정리**

| 기본 가정 | 회귀 분석 전에 검증 가능? | 회귀 분석 후에 검증 가능? | 검증 방법 |
| --- | --- | --- | --- |
| 선형성 | ✅ | ✅ | scattor plot 찍어보기 |
| 독립성 (다중공선성) | ✅ | ✅ | VIF 지수, 상관계수 |
| 오차항 평균이 0이다 | ❌ | ✅ | np.mean(residuals) |
| 등분산성 | ❌ | ✅ | 잔차의 도표화 |
| 오차항 자기상관 없다 | 🔺 (회귀 분석 전에는 간접적으로 검증 가능, 분석 후에 최종 검증) | ✅ | Durbin-Watson 검정(분석 후) |
| 정규성 | ❌  | ✅ | Shapiro-Wilk 검증, Q-Q plot |

분석 전에는 '데이터'의 관계를 보고(가정 1,2), 분석 후에는 모델이 틀린 정도인 '오차항'의 성질(가정 3~6)을 확인

### 5. **다중공선성**
- 독립 변수들 간 강한 상관관계로 인해 회귀 모델이 불안정해지는 문제
- 문제가 심각하면 각 변수의 회귀계수 추정이 부정확해지고, 해석력이 떨어질 수 있음
- 다중공선성이 높은 경우 : 어떤 독립 변수가 종속 변수에 얼마나 영향을 미치는지 정확하게 구분하기 어렵기 때문에 확인하고 처리하는 과정 필요

    1) 다중공선성 확인 방법
    - 상관계수 : 변수 간의 통계적 관계인 상관관계의 정도를 수치로 나타낸 것
        - $-1 ≤ r ≤ 1$  사이의 값을 가진다.
        - $r>0$: 양의 상관관계, 한 변수가 증가하면 다른 변수도 증가하는 경향.
        - $r<0$: 음의 상관관계, 한 변수가 증가하면 다른 변수는 감소하는 경향.
        - $r=0$: 두 변수 사이에 직접적인 연관성이 없다.
    - VIF지수 (분산 팽창 인수)
        - VIF가 높은면 다중공선성이 존재한다 판단    
            - **VIF = 1**: 해당 독립 변수는 다른 변수와 상관관계가 전혀 없음을 의미
            - **VIF < 5**: 일반적으로 다중공선성 문제가 없다고 간주
            - **VIF > 5**: 다중공선성의 징후가 있을 수 있으며, 주의를 요함
            - **VIF > 10**: 다중공선성이 심각하다고 간주되며, 이 변수를 포함한 회귀 모델은 불안정할 가능성이 큼. 이 경우 해당 변수를 제거하거나, 변수 간의 상관관계를 줄이기 위한 조치가 필요할 수 있음.
            - 이는 절대적인 기준이 아닌, 일반적 가이드라인임.

    2) 다중공선성 대처 방법
        1. 변수 제거
        2. 변수 변환
            - 변수들을 더하거나 빼서 새로운 변수 생성
            - 독립변수를 더하거나 빼더라도 문제가 없는 경우
        3. 규제선형모델 활용
        4. PCA(주성분 분석)   
            - 데이터 차원 축소에 사용되는 통계적 기법. 고차원 데이터에서 중요한 정보를 최대한 보존하며 데이터의 복잡성을 줄이고자 함.
            - 상관 있는 변수들을 묶어서 더 적은 수의 대표 축으로 바꾸는 것이 PCA의 핵심 아이디어. 즉, 데이터의 방향성을 찾아 변환하고 중요한 정보만 남겨 차원을 축소     

### 6. **규제선형모델**
- **규제선형모델이란??**
    - 모델이 학습 데이터에 과적합 되지 않도록 규제를 가하고자 등장한 모델
    - 선형회귀 모델에서는 특성에 곱해지는 계수(기울기)의 크기를 조정하는 것
    - 잔차 제곱합의 평균을 최소화하는 방식 사용할 경우, 오히려 회귀 계수가 쉽게 커지게 되어 과적합 발생 가능성 커짐.
    - 다중회귀에서도 모델이 복잡해져 과적합 가능성 증가

- **규제선형모델의 종류**
    1) L2규제 : W(회귀계수)의 제곱에 대해 페널티를 부여하는 방식, 릿지(Ridge)회귀
    2) L1규제 : W(회귀계수)의 절댓값에 대해 페널티를 부여하는 방식, 라쏘(Lasso) 회귀
    3) L2규제ㅔ + L1규제 : 엘라스틱넷(Elastic Net)회귀

- **릿지 회귀(Ridge, L2 규제)** : 기둥에 따라 다른 힘으로 누르는 망치
    - 큰 기둥은 세게, 작은 기둥은 살살 비율은 균등하지 않을 수는 있지만, 전체적으로 작아짐.
    - 회귀계수의 제곱을 패널티로 주기 때문에, 큰 회귀계수일 수록 훨씬 강한 페널티를 받음.
    - 모든 계수의 크기가 작아지지만, 규제를 통해 변수가 제거되지는 않음.
    - 모든 변수를 유지하며, 크기를 적절히 줄여 과적합 방지    

$$

\text{Cost Function} = \text{RSS} + \alpha \sum_{j=1}^{p} w_j^2

$$

- **라쏘 회귀(Lasso, L1 규제)** : 모든 기둥을 같은 힘으로 누르는 망치
    - 작은 기둥들은 힘을 버티지 못해 완전히 사라질 수도 있음
    - 회귀계수의 절댓값을 패널티로 주기 때문에, 회귀계수의 크기와 상관없이 같은 패널티를 받음
    - 작은 계수가 0이 되기도 하기 때문에 변수가 완전히 사라질 수 잇음
    - 비중이 낮은 변수(작은 계순)는 아예 사라지고, 중요한 변수만 남음


$$
\text{Cost Function} = \text{RSS} + \alpha \sum_{j=1}^{p} |w_j|
$$

- **엘라스틱넷(Elastic Net, L1 + L2)** : L1, L2 두 개의 망치를 동시에 사용
    - L1 망치 : 작은 기둥을 아예 없애 불필요한 변수 제거
    - L2 망치 : 남은 기둥을 균형 잡히게 눌러 크기 조절
    - 두 개를 동시에 사용할 경우, 불필요한 변수는 제거하고, 남은 변수들은 적절한 크기로 유지
    - L1의 변수 선택 기능(일부 계수 삭제)과 L2의 규제 기능(큰 계수를 줄임)을 동시에 활용하는 모델
    - L1은 상관관계가 높은 변수 중 일부만 선택하는 경향이 있음. L2를 함께 적용하면 상관관계가 높은 변수를 함께 고려하면서 규제 가능
    - 안정적인 예측과 희소성을 동시에 달성할 수 있음

$$
\text{Cost Function} = \text{RSS} + \alpha_1 \sum_{j=1}^{p} |w_j| + \alpha_2 \sum_{j=1}^{p} w_j^2
$$

### **모델 평가방법**
- **시각화**
- **통계지표** 
    - 모델 유의성 검정
        - F-statistic, Prob(F-statistic) : 회귀 모델이 유의미한지 검정하는 지표 (종속 변수 y와 전체 독립 변수 X 간의 관계가 통계적으로 유의미한지)
            - 귀무가설 H0 : 모든 회귀 계수가 0이다(독립 변수들이 종속 변수들과 관계가 없다)
            - 대립가설 H1 : 적어도 하나의 회귀 계수는 0이 아니다(독립 변수 중 하나 이상이 종속 변수에 영향을 준다)
            - F-statistic의 값이 클수록 모델이 통계적으로 유의미함을 의미
            - Prob < 0.05 일 경우, 귀무가설 기각

        - T-statistic, Prob(T-statistic) : 각 독립변수가 종속변수 y에 유의미한 영향을 미치는지 판단하는 지표
            - 귀무가설 H0 : 해당 변수의 회귀 계수가 0이다(이 변수는 종속 변수에 영향을 미치지 않는다)
            - 대립가설 H1 : 해당 변수의 회귀 계수는 0이 아니다(이 변수는 종속 변수에 유의미한 영향을 준다)
            - P>|T-statistic| 기준으로 가설 검정 수행 

    - 모델의 성능 평가
        - 결정 계수 : 회귀 모델이 종속 변수의 전체 변동성을 얼마나 잘 설명하는지 나타내는 지표
            - 모델이 데이터를 얼마나 잘 설명하는지 측정
            - 평균값으로 데이터를 예측했을 때보다, 모델을 활용했을 때 오차가 얼마나 줄어드는지를 수치화한 것
            - SST : 평균으로 예측했을 때의 총 오차(우리가 극복해야 할 전체 변동성)
            - SSE : 모델이 예측하고도 여전히 남은 오차(모델의 한계)
            - SSR : 평균에 비해 Regression 모델이 얼마나 더 잘 맞혔는지(모델의 성과)
            - o <= 결정 계수 <= 1
                - 1에 가까울수록, SSR과 SST 값이 비슷해지고, SSE는 0에 가까워지므로 모델이 데이터를 잘 설명한다고 할 수 있음.
            - R-squared 값 = 0이면, 데이터들의 평균 = 모델로 예측 : 모델이 데이터를 전혀 설명하지 못함.    

        - 조정된 결정 계수 : 변수의 개수 증가에 덜 민감하도록 조정한 지표

        - AIC, BIC(SC) : 정보 기준이라 불리며, 값이 낮을수록 좋다고 평가
            - AIC : Akaike information creterion
                - BIC에 비해 복잡성에 대한 패널티가 비교적 작음
            - BIC(SC) : Schwarz Criterion, BIC : Bayesian Information Creterion
                - AIC 보다 더 엄격한 기준으로, 데이터의 양(n)에 따라 더 강한 패널티 부과

- **성능 평가 지표**
    - 평균 제곱 오차 (MSE) : 예측값과 실제값 사이의 오차 제곱의 평균 
    - 평균 절대 오차 (MAE) : 모델의 예측값과 실제값 사이의 절대값 오차의 평균
    - MSE와 MAE의 차이
        - MSE는 큰 오차에 대해 민감하게 반응. 따라서 이상치에 민감
        - MAE는 모든 오차를 동일하게 취급. 따라서 이상치가 존재하는 경우에도 전반적인 예측 성능을 유지할 수 있음

- **변수 유의성 평가**
    - T-검정 : 독립 변수의 회귀계수가 유의미한지 검정하는 데 사용
        - 귀무가설 H0 : 회귀계수(a)가 0이다(독립변수가 종속변수에 영향을 주지 않는다)
        - 대립가설 H1 : 회귀계수(a)가 0이 아니다(독립변수가 종속변수에 영향을 준다)


1. **t-값 계산**

$$
t = \frac{\hat{\beta} - 0}{SE(\hat{\beta})}
$$

- $\hat{\beta}$ : 추정된 회귀계수
- $SE(\hat{\beta})$ : 표준오차
- t-분포를 따름.

2. p-value 확인 및 판단
- p-value < 0.05 : 유의 수준 5%에서 H0 기각 = 해당 변수는 유의미하다
- p-value >= 0.05 : H0 기각 불가 = 해당 변수는 유의미하지 않다

- 회귀계수 = 0 이면 해당 독립변수는 의미 없음
- p-value가 작을수록 해당 변수는 종속변수에 더 큰 영향을 줄 확률이 높아짐
- t-값이 크면 귀무가설을 기각할 가능성이 높아짐

### 4.**비선형회귀**
- **다항식 회귀 모델(Polynomial Regression Model)**
    - 다항식 회귀 모델: Y = a + bX + cX^2 + dX^3 + ⋅⋅⋅의 형태로, 독립변수에 거듭제곱 항을 추가하여 데이터의 곡선적(비선형적) 특성을 모델링함.
    - 사용 방법
        - 기존의 선형 모델 Y = a + bX 에 대해 cX^2, dX^3 등을 포함하여 모델을 확장
        - 확장한 상태에 선형 회귀를 적용
    - 장점 : 선형 모델에 비해 회귀선을 데이터에 fit하게 그릴 수 있음
    - 단점 : 너무 많은 feature(독립변수)를 이용하면 결과가 overfitting 될 수 있음
    - 적합한 상황
        - 관계가 직선적이지 않고 곡선을 그리는 데이터 패턴에서 유용
        - 성장률, 감소율 등이 시간에 따라 변화하는 속도가 다른 경우에 좋음

- **지수 회귀 모델(Exponential model)**
    - 종속 변수가 지수적으로 변화하는 관계를 모델링할 때 사용할 수 있음
    - 사용 방법
        - 지수함수 형태의 데이터는 Y = a*e^bX의 수식을 따름
        - 양 변에 로그를 취하면 ln Y = ln a + bX가 됨 (선형회귀 식과 비슷)
        - 지수 모델은 종속 변수 Y에 로그를 취한 후 선형회귀를 적용하면 됨
            - 이렇게 계산된 값은 기존 데이터와 형태가 달라지므로, 원래의 데이터 형태로 복구하기 위해 예측 값에 지수함수를 적용함.
        - 흐름 : 종속 변수 Y에 로그함수 적용 -> 로그 변환된 Y와 기존 독립변수 X에 대해 선형 회귀 적용(모델이 ln Y와 X의 선형 관계 학습) -> Y값 예측 시, 모델이 예측한 Y값에 지수함수를 적용하여 로그 변환 전 Y로 변형
        - 적합한 상황
            - 값이 시간에 따라 지수적으로 증가하거나 감소하는 경우
            - 기술 발전, 투자 수익률 증가 등

- **로그 회귀모델(log regression model)**
    - 종속 변수와 독립 변수 간의 관계가 로그 함수를 통해 더 잘 표현될 때 사용
    - 사용 방법
        - 로그함수 형태의 데이터는 Y = a + b log(X)의 수식을 따름. 이는 선형식에서 X 자리에 log(X)가 들어간 것. 따라서 독립 변수 X에 로그 연산을 취한 후, 나머지 과정은 선형 회귀와 동일
    - 흐름 : 독립 변수 X에 로그함수를 적용 -> 선형 회귀를 적용
    - 이렇게 예측된 값은 Y이므로 예측 값에 별도의 처리를 하지 않고 그대로 사용 가능
    - 적합한 상황
        - 데이터가 초기에 빠르게 증가하고 점차 증가율이 줄어드는 패턴을 보일 때 효과적
        - 인구 성장, 감염병의 확산과 같이 초기 빠른 증가 후 점진적 감소를 보이는 데이터에 적합

- **스플라인 회귀(Spline Regression)**                                       
    - 데이터를 구간별로 나누고 각 구간에서 달른 선형 또는 비선형 함수를 적용하여 예측하는 방식
    - 사용 방법
        - 전체 데이터 범위를 여러 구간을 나누고, 각 구간에 대해 별도의 회귀모델(선형, 다항 등) 적용
        - 구간 경계에서의 연속성을 유지하는 것이 중요
    - 적합한 상황
        - 데이터 패턴이 여러 구간에서 서로 다르게 나타날 때 유용
        - 계절에 따라 판매량이 변화하는 소매 데이터 분석 등에서 적합
