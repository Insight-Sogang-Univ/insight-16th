1. **자료의 척도**
- 자료 : 설명과 해석을 위하여 수집, 분석, 요약되는 사실과 수치

- 자료의 종류
    - 양적 자료(수치로 표현) : 구간(등간)척도, 비율척도
    - 질적 자료(수치로 표현되더라도 연산의 의미를 가지지 못함) : 명목척도, 서열척도

이러한 측정척도는 자료 속에 포함된 정보의 양을 결정하며, 가장 적합한 자료요약 방법과 통계분석 방법을 결정함.    

2. **기술통계**
- 기술통계 : 데이터를 있는 그대로 설명하고 이해하는 데중점을 둔 통계 방법

- 주요 도구
    - 중심화 경향 : 데이터의 중심적인 경향을 나타냄
        - 평균 (series.mean()) : 데이터 값의 총합을 데이터의 개수로 나눈 값. 가장 대표적 지표. 이상치에 민감할 수 있음.
            - 절사평균 : 이상치에 민감한 점을 보완하기 위해 등장한 개념. 데이터의 양끝을 일정 비율 잘라낸 후 평균을 계산
        - 중앙값 (series.median()): 데이터를 크기 순으로 일렬로 정렬했을 때, 한가운데에 위치하는 값. 평균에 비해 이상치에 덜 민감. 대안적 지표
        - 최빈값 (series.mode()): 데이터에서 가장 자주 나타나는 값. 주로 이산형 데이터, 데이터의 분포를 이해하는 데 도움을 줌.
    
    - 자료의 산포 : 데이터가 퍼져있는 정도
        - 범위 (series.max() - series.min()) : 데이터의 최대, 최소값의 차이를 나타내는 지표. 데이터의 전체 변동 범위를 나타냄. 양쪽 극단값을 이용하기에, 이상치에 매우 민감.
        - 분산 (series.var()): 데이터 값들이 평균으로부터 얼마나 멀리 퍼져있는지를 나타내는 지표. 분산값이 크다 = 데이터가 많이 퍼져있다.
        - 표준편차 (series.std()) : 분산에 제곱근을 취한 값. 단위를 일치시키기 위해 사용.
        - 사분위수 : 데이터의 표본을 네 개의 동일한 크기로 나눈 값.
            - Q1 = 하위 25퍼센트에 위치한 값, 하위 사분위수 (series.quantile(0.25))
            - Q2 = 하위 50퍼센트에 위치한 값, 중앙값 (series.quantile(0.5))
            - Q3 = 하위 75퍼센트(상위25%)에 위치한 값, 상위 사분위수 (series.quantile(0.75))
        - 사분위간 범위(IQR) : Q3에서 Q1 값을 제외한 범위, 데이터의 중앙 50%가 퍼져있는 범위. 이상치 탐지에 유용 (Tukey Fence)
            - 아래쪽 이상치 : Q1 -1.5*IQR 보다 작은 값
            - 위쪽 이상치 : Q3 + 1.5*IQR보다 큰 값
        - 변동계수 : 데이터의 상대적인 변동성을 나타내는 통계적 지표
            - 변동계수 = (표준편차/평균) * 100
        - 첨도와 왜도
            - 첨도 (series.kurt()) : 확률분포의 꼬리가 두꺼운 정도를 나타내는 척도
            - 왜도 (series.skew()): 비대칭도, 실수 값 확률 변수의 확률 분포 비대칭성을 나타내는 지표. 평균 = 중앙값일 경우 왜도 : 0

    - 자료의 정리 및 시각화 : 히스토그램, 박스플롯, 파이차트

3. **확률과 분포**
- 추론 통계(통계적 추론) : 표본에서 얻은 정보를 바탕으로, 모집단의 특성(모수)을 확률적으로 추정하거나 가설을 검정하는 통계 분야.

- **확률분포**
    - 확률변수 : random 한 세계에서 어떤 특정 사건이 발생할 때 나타날 수 있는 모든 경우의 수를 특정 값으로 대응시키는 함수.
    - 확률분포 : 확률변수가 특정한 값을 가질 확률을 나타내는 함수
        - 이산확률분포 : 확률 변수가 이산적인 값을 가짐. 특정 값 선택 가능
        - 연속확률분포 : 확률 변수가 연속적인 값을 가짐. 특정 값 선택 불가

- **대표적 분포 모형**
    - 정규분포 : 가장 잘 알려진 연속확률분포. 많은 자연 현상, 사회 현상이 정규분포를 따르는 것으로 알려져있음.
        - 형태 : 종모양, 평균을 기준으로 대칭, 표준편차가 클수록 곡선이 평평해짐
        - 매개변수 : 평균, 표준편차   

    - Z분포(표준정규분포) : 평균=0, 표준편차=1 이 되도록 맞춘 표준적인 정규분포

    - T분포 : 표본으로 모평균(or 모평균 간의 차이)을 추정할 때 사용되는 분포. 표본 크기가 작을 때 정규분포보다 더 정화한 결과 제공.
        - 형태 : 종모양, 정규분포보다 꼬리가 두꺼움. 평균 기준으로 대칭. 자유도가 커짐에 따라 표준정규분포에 수렴
        - 매개변수 : 자유도(df)(통계에서 "얼마나 많은 정보가 독립적으로 변할 수 있는가"를 나타내는 숫자)

    - 카이제곱분포 : 독립적인 표준정규분포를 따르는 변수들의 제곱 합. 표본분산의 분포와 관련된 지표. 주로 독립성 검정, 적합도 검정 등에 사용
        - 형태 : 0에서 시작하여 오른쪽으로 긴 꼬리를 가진 비대칭 분포. 자유도가 증가함에 따라 중심이 오른쪽으로 이동하고 모양이 더 대칭적으로 변하며, 정규분포에 가까워짐.
        - 매개변수 : 자유도(df)

    - F분포 : 두 독립적인 카이제곱분포의 비율. 두 집단 간 분산 차이를 비교하는 데 사용
        - 형태 : 0에서 시작하여 오른쪽으로 긴 꼬리를 가진 비대칭 분포. 자유도가 증가함에 따라 중심이 오른쪽으로 이동하고 모양이 더 대칭적으로 변하며, 정규분포에 가까워짐.
        - 매개변수 : 자유도(df1, df2)       

- **분포모형을 알아두어야 하는 이유**
    - 가설검정을 위한 검정통계량 산출에 사용됨
    - 표본추출의 과정에서 인위적인 조작이 있었는지 확인 가능


4. **추론 통계 - 추정**
- 추론통계 : 수집한 데이터를 기반으로 미래의 , 혹은 조사하지 못한 전체 정보를 추론하고 결론을 도출하는 방법을 다루는 분야.
    - 추정 : 모집단을 대표할 수 있는 일부를 표본으로 추출하여 분석한 통계량을 이용해서 모집단의 모수를 예측하는 과정
    - 검정 : 표본을 추출하여 얻은 표본통계량으로 모집단 모수에 대한 새로운 예상·가설 등이 옳다고 판단할 수 있는지를 평가하는 과정
    - 추정량 : 표본정보에 의존하는 확률변수로서 모수를 추정하는 데 사용되는 표본통계량
    - 추정치 : 표본이 추출된 후 표본자료를 이용하여 계산한 통계량의 특정한 값

- 점추정 : 어떤 모수를 표본에 기초하여 계산된 하나의 값으로 추정하는 것

- 구간추정 : 모수의 참 값이 포함되리라고 기대하는 추정치를 일정한 범위와 신뢰도를 가지고 나타내는 방법

5. **추론 통계 - 검정**
- **가설 검정** : 모집단의 실제 값이 얼마가 된다는 주장과 관련해, 표본의 정보를 사용해서 가설의 합당성 여부를 판정하는 과정
    - **절차** :
        - 가설 설정 : 귀무가설 H0, 대립가설 H1 설정
            - 귀무가설 : 주로 기각하고자 하는 명제(~는 차이가 없다). 현재 통용되고 있는 가설로 지지될 때 위험성이 비교적 약한 가설
            - 대립가설 : 내가 주장하고자 하는 명제(~는 차이가 존재한다). 새롭게 주장하고 싶은 가설로 지지될 때 위험성이 비교적 강한 가설

        - 유의 수준 설정 : 유의수준이란 연구자가 결과를 판단할 때 허용할 수 있는 오류의 최대치로, 주로 0.05, 즉 5%로 설정. 유의 수준은 연구자가 어떤 결정이 얼마나 신뢰할 만한지를 판단하는 기준으로 사용
            - 1종 오류 : 귀무가설이 실제로 참이었는데 기각할 확률
            - 2종 오류 : 귀무가설이 실제로 거짓인데 기각하지 못할 확률
            - 가설검정은 1종 오류를 더 치명적인 것으로 파악. 따라서 유의수준은 1종 오류의 확률을 의미.    

        - 검정통계량 산출 : 검정통계량은 가설 검정의 결과를 판단하는 데 사용되는 값으로, 데이터의 분포나 특징에 따라 T값, F값 등 어떤 통계량을 사용할지 정해야 함.
            - Z값(Z-점수) : 표준 정규 분포를 사용하여 얻은 값. 주로 큰 표본 크기에 대한 가설 검정에 주로 사용. 
                - Z의 절댓값이 크면 해당 데이터가 평균으로부터 멀리 떨어져있으며, 때로는 비정상적이거나 이상치일 가능성이 있음을 뜻함.
                - 임계점 찾기
                    - Z = 1.96일 때 유의수준 0.05
                    - Z = 2.58일 때 유의수준 0.01
                    - Z값 > 임계점 : 귀무가설 기각

            - T값 : T분포를 사용하여 얻은 값으로, 작은 표본 크기에 대한 가설 검정에 주로 사용. 주로 모집단의 평균에 대한 가설을 검정하거나, 두 표본 평균 간의 차이를 검정할 때 사용.   
                - T값의 절댓값이 크면 해당 데이터가 모집단 평균으로부터 멀리 떨어져있으며, 관측된 표본 평균이 모집단 평균과 유의미한 차이가 있을 가능성이 높다는 것을 의미.
                - 임계점 찾기
                    - T분포는 자유도(df)에 따라 모양이 달라지므로 자주 쓰는 자유도와 유의수준에 따른 t분포표 사용.
                    - 같은 유의수준에 대해서 t값은 z값보다 커지며 자유도가 높아질수록 z값에 수렴함

            - F값 : F값은 F분포를 사용하여 얻은 값으로, 주로 두 표본 분산 사이의 차이를 비교하는 데 사용. F분포는 두 개의 독립적인 카이제곱 분포된 변수들의 비율로 정의. 주로 분산 분석(ANOVA)에서 두 개 이상의 그룹 간 평균의 차이를 검정하는 데 사용.
                - 수치 해석 : F값은 두 표본의 분산의 비이므로, 이 값이 1에 가까우면 두 표본의 분산이 서로 비슷하다는 것을 의미. 즉 두 그룹이 퍼진 정도가 비슷하다고 볼 수 있음.

        - 가설 기각/채택 판단 : 유의확률(p-value) :  가설 검정에 쓰일 수 있는 또 다른 척도. 유의확률은 검정통계량에서 유도할 수 있으며, 유의 확률(p)가 유의 수준(α)보다 작을 시에는 귀무가설을 기각, 클 시에는 귀무 가설을 기각하지 못함.

5-2 **데이터의 자료형에 따른 검정 방법**
1. [수치형 - 수치형] (피어슨)상관분석
- 두 변수간 피어슨 상관계수가 얼마나 유의미한지를 검정하는 방법
- 귀무 가설과 대립 가설
    - H0: 두 변수간 선형 상관관계가 존재하지 않는다.
    - Ha: 두 변수간 선형 상관관계가 존재한다.
- 결과 : (상관계수, p-value)의 튜플 형태로 나옴
    - (예시) 상관계수 : r 이라고 하면, 
        - |r| < 0.3 → 두 변수간 상관관계가 약함
        - |r| < 0.5 → 두 변수간 상관관계가 중간
        - |r| > 0.5 → 두 변수간 상관관계가 강함

2. [범주형 - 수치형] F-test(ANOVA), T-test
- 범주가 2개 : T-test 사용
    - T-test 수행시 t 값은 두 평균 간 차이를 표준 오차로 나눈 값
    - 가설
        - H0: 두 그룹의 평균에는 차이가 없다.
        - Ha: 두 그룹의 평균에는 차이가 있다.
    - 결과 : TtestResult(statistic = t-값, pvalue = p값, df = 자유도)의 형태로 나옴. t값은 보통 절대값 2를 기준으로, 그보다 크면 차이가 있으며 그보다 클수록 차이가 크다고 생각. p값은 이 결과가 얼마나 믿음직한지를 나타내며 보통 0.05가 기준.

- 범주가 3개 이상일 때 : F-test(ANOVA)
    - ANOVA 검정을 할 때엔, F값은 집단 간 분산 / 집단 내 분산으로 계산. 
    - 가설 (예시)
        - H0: 각 MBTI 그룹 별 수명 평균은 차이가 없다.
        - Ha: 각 MBTI 그룹 별 수명 평균은 차이가 있다.
    - 결과 : F_onewayResult(statistic = f값, p-value = p값)의 형태로 나옴. f값은 보통 2~3 이상이면 차이가 있으며 클수록 차이가 크고, p값은 역시 마찬가지로 f값 결과가 얼마나 유의미한지를 의미.
    - 주의
        - anova는 전체 평균 대비 각 그룹간 차이가 있는지만 알 수 있음
        - 즉, 모두가 서로 다르다가 아니라, 하나가 튀고 나머지가 비슷하더라도 귀무가설을 기각하는 결과가 나올 수 있음.

3. [범주형 - 범주형] : Chi2-test
- 가설
    - H0: 경영학과 여부와 갈색머리 여부는 관련이 없다
    - Ha: 경영학과 여부와 갈색머리 여부는 관련이 있다
- 기대빈도와 관측빈도간 관계를 알아봄.
- **[(관측빈도 - 기대빈도)^2/기대빈도]**를 쌍만 맞게, 순서 상관 없이 쭉 더해주면 됨.
- 카이제곱 분석은 범주 갯수에 제한이 없음.
- 카이제곱 통계량 해석은 어떠한 특정 숫자가 아닌, 자유도의 약 2배가량을 기준으로 잡음.
- 범주형 변수의 자유도 : 범주의 수 - 1
- 카이제곱 변수에서의 자유도 : (x변수의 자유도)* (y변수의 자유도)
