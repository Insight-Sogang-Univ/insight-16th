**판다스(Pandas)** :

- 미리 만들어진 기능들이 들어 있는 도구 상자

**판다스 라이브러리 불러오는법**: import pandas as pd

**데이터 불러오기**: df=pd.read_csv('파일 경로')

**데이터프레임**:
- 컴퓨터로 정보를 정리하고 저장하는 도구
- 판다스에서 사용하는 주요 데이터 구조
- 2차원 배열의 형태로 인덱스(행)와 컬럼(열)의 두 가지 기준에 의해 표 형태로 데이터가 저장됨
- 여러 개의 시리즈(Series)가 모여 하나의 데이터프레임을 구성
- 컬럼 단위로는 값들의 데이터 타입이 동일해야함
- 컬럼 간의 데이터타입은 달라도 무방

**데이터를 확인, 추출, 조작하는 방법**:
- **데이터 확인**
    - 앞/뒤 일부 데이터를 미리 확인하여 데이터가 어떤 구조와 값을 가지고 있는지 개괄적으로 파악
        - df.head(n): 앞 n개의 행 반환(기본값 n=5)
        - df.tail(n): 뒤 n개의 행 반환
    - 데이터에 대한 정보 확인
        - df.shape: (행 개수, 열 개수) 반환
        - df.info(): 여러 정보들을 종합적으로 반환
        - df.describe(): 여러 정보들을 종합적으로 반환
        - df.isnull(): 데이터에 결측치(NaN)가 있는지의 여부를 반환
        - df.dtypes: 열들의 데이터타입 반환
    - 고유값 및 개수 확인
        - df.count(): 결측치의 개수를 제외한 데이터의 행 수 반환
        - df[열 이름].value_counts(): 특정 열에 포함된 값들의 빈도수를 세어 반환
        - df[열 이름].unique(): 특정 열에 포함된 값들을 종류별로 1개씩만 반환
- **데이터 추출**
    - 인덱싱(Indexing) & 슬라이싱(Slicing)
        - 인덱스와 컬럼을 사용해 데이터를 추출하는 것
            - 인덱싱: 단일 항목 지목할 때 사용
            - 슬라이싱: 범위를 지정해 여러 항목을 추출할 때 사용
    - 행단위 추출: loc 과 iloc
        - 데이터프레임의 행 데이터 선택하기 위해 사용
            - loc: 인덱스 이름(index label)기준으로 행 선택할 때 이용, 범위의 끝 포함 지정 가능
                - df.loc[행 이름]: 행단위 추출
                - df.loc[행 이름, 열 이름]: 값단위 추출
                - df.loc[:, 열 이름]: 열단위 추출
            - iloc: 정수형 위치(integer position) 사용할 때 이용, 범위의 끝 제외 지정 가능
                - df.iloc[행 순서, 열 순서]
    - 열단위 추출: 컬럼명
        - loc&iloc 방식으로도 가능하나, 열 이름으로 인덱싱하는 것이 더 직관적
            - df[열 이름]
            - df[[열 이름]]
            - df[[열 이름1, 열 이름2...]]
            - df[열 이름][행 이름]
    - 불리언 인덱싱
        - 특정 조건 만족 데이터 추출할 떄 활용
        - True , False 배열을 통해 데이터 필터링
            - df[조건문]
            - df[(조건문1) 논리연산자 (조건문2)]
            - df.loc[조건문, 열 이름]
    - 인덱스 설정
        - 정수형 인덱스가 기본값이지만, 원하는 대로 변경 가능
            -df.set_index(열 이름): 기존의 열 중 하나를 인덱스로 설정
            - df.reset_index(): 인덱스를 리셋하고 기존 인덱스를 열에 추가
            - df.reindex(인덱스 배열): 인덱스 재배열에 따라 기존 행데이터도 인덱스를 따라감
- **데이터 조작**
    - 데이터 추가, 삭제
        - df['새로운 열 이름'] = 리스트: 추가
        - df.drop(labels, axis): 삭제
    - 데이터 정렬
        - df.sort_index(): 인덱스 기준 정렬
        - df.sort_values(by="열 이름"): 특정 열 값 기준 정렬
    - 데이터 변환
        - df[열 이름] = df[열 이름].astype(바꿀 자료형): 특정 컬럼 내 데이터 자료형 변환
        - df.replace(바꾸고 싶은 값, 바꿀 값): 데이터 값 자체 변환
        - df.transpose(), df.T: 전치를 통해 데이터프레임의 행과 열 변환
    - 데이터프레임 병합
        - concat: 두 데이터 프레임 이어붙이는 방식, 주로 동일한 구조의 데이터프레임 합칠 때 사용
            df = pd.concat([df1,df2], axis=0): 행방향 결합
            df = pd.concat([df1,df2], axis =1): 열방향 결합
        - merge: 공통항목 기준 병합(on 인자를 통해 기준이 될 열 지정, how 인자에 따라 어떤 방식으로 결합할지 결정)
            - df = pd.merge(df1, df2, how = 'inner', on = '~~'): 교집합 연산
            - df = pd.merge(df1, df2, how = 'outer', on = '~~'): 합집합 연산
                    







**집계함수**:
- 여러 행으로부터 하나의 결괏값을 반환하는 함수
- 각 컬럼의 평균, 표준편차, 합계 등 연산을 통해 대표값 구할 때 주로 사용
- 결측치는 무시하고 계산
    - 대표적 집계함수
        - df['컬럼 명'].mean() : 평균
        - df['컬럼 명'].median() : 중앙값
        - df['컬럼 명'].std() : 표준편차
        - df['컬럼 명'].sum() : 합계
        - df['컬럼 명'].max() : 최대값
        - df['컬럼 명'].min() : 최소값
    - 그룹화(Grouping):
        - groupby 메소드: 특정 기준으로 그룹을 묶고, 연산 처리
        - 보통 집계함수와 함께 사용
            - df.groupby('column명1')['column명2'].집계함수(): column1로 그룹화를 하고, 그룹 별로 column2에 대해 집계함수를 적용해 결과를 시리즈 형태로 반환
            - df.groupby('column명1')[[’column명2']].집계함수(): 그룹 별로 column2에 대해 집계함수를 적용해 결과를 데이터프레임 형태로 반환
