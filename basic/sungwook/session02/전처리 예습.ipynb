**데이터 탐색(EDA)**
- 다양한 각도(도표, 그래프, 통계)에서 수집한 데이터의 특징 파악
- 데이터의 특성, 패턴, 관계를 파악하고 시각화하며 데이터를 이해, 전처리, 모델링 등 후속 분석에 대한 방향성을 제시

**데이터 전처리**
- 데이터를 분석(모델링)에 적합한 형태로 손질하는 과정
- 이상치 제거, 결측치 처리, 노이즈 제거 등의 작업 수행
- 데이터 왜곡을 피하고 신뢰도를 높이는 과정으로 반복적으로 수행됨

**데이터 모델링**
- 데이터로부터 유용한 정보 추출을 위해 모델을 구축하는 단계
- 예측, 분류 등 목적에 따라 모델을 선택하고 학습시키는 과정

**EDA(Exploratory Data Analysis)**
- 탐색적 데이터 분석 과정
- 데이터 분석을 위해 데이터를 여러 방식으로 파악하는 모든 과정
- 시각화 같은 도구를 통해 패턴 발견
- 데이터의 특이성 확인
- 통계와 그래프로 가설을 검정하는 과정
등을 통해 주어진 데이터에 대해 알아보는 것

**EDA 대상(일변량/다변량)**
- Univariate(일변량)
    - EDA를 통해 한 번에 파악하려는 변수가 한 개
    - 데이터를 설명하고 그 안에 존재하는 패턴을 찾는 것이 주요 목적
- Multi-variate(다변량)
    - EDA를 통해 한 번에 파악하려는 변수가 여러 개
    - 여러 변수들 간의 관계를 보는 것이 주요 목적
    - 변수를 동시에 확인하기 전, 개별 데이터를 먼저 파악하는 것이 오류에 대처하기 용이         

**EDA 종류(시각화/비시각화)**
- Graphic(시각화)
    - 차트, 그림 등을 이용하여 데이터를 확인하는 방법
- Non-Graphic(비시각화)
    - 그래픽적 요소 사용하지 않고 주로 Summary Statistics를 통해 데이터를 확인하는 방법    

**EDA 유형**
- Uni-Non Graphic(일변량 비시각화) : 주어진 데이터의 Distribution을 확인하는 것이 주목적
- Uni-Graphic(일변량 시각화) : 주어진 데이터를 전체적으로 살펴보는 것이 주목적
- Multi-Non Graphic(다변량 비시각화) : 주어진 둘 이상의 변수 간 관계를 확인하는 것이 주목적
- Multi-Graphic(다변량 시각화) : 주어진 둘 이상의 변수 간 관계를 전체적으로 살펴보는 것이 주목적


**데이터 확인**
- 데이터 읽기/쓰기
    - 읽기 : 작업 공간에 데이터를 불러오는 것
    - 쓰기 : 새로운 데이터를 만드는 것
- 절대경로/상대령로
    - 절대 경로 : 처음(root 파일)부터 시작하여 목적지까지 전체적인경로(URL), 해당 파일까지의 모든 경로
    - 상대 경로 : 현재 작업하고 있는 디렉터리를 기준으로 상하위 디렉터리와 같이 상대적으로 표현되는 경로
        - ./ : 현재 디렉터리
        - ../ : 상위 디렉터리
        - / : 최상위(루트) 디렉터리

    - 여러 개의 시트로 구성된 데이터를 읽을 때 불러올 특정 시트 설정 가능

**데이터셋 파악하기**
- 데이터 프레임 보기
    - head(n) : 데이터 프레임 상위 n개 데이터 보여줌
- 데이터 변수 확인
    - 변수 = 데이터 프레임의 column = feature
        - 변수 정의 확인 : 어떤 정보를 가지는지
        - 변수 유형 확인 : 질적·범주형과 양적·수치형으로 구분
        - 변수 데이터 형식(자료형) 확인 : 날짜, 수치, 텍스트, 이미지 등의 구분

**데이터 분포 확인**
- 단변수 분석은 원시 데이터(raw data)의 평균값, 최빈값, 중간값 등 변수들의 분포를 산포도, 박스 플롯, 히스토그램 등의 그래프를 통해 단변수, 즉 하나의 데이터 분포를 확인해 분석할 수 있는 방법 
- 원시 데이터 분포의 확인을 통해 전처리 아이디어를 얻을 수 있음.

**전처리**
데이터 전처리 : 데이터를 분석에 적합한 형태로 만드는 과정
데이터 분석의 정확도는 데이터 품질에 의해 좌우된다. 따라서 데이터 품질을 높이기 위해 이상값, 결측값, 노이즈를 수정, 분석 목적에 맞게 변형하는 과정이 중요

- **전처리의 여러 기술**
    - 데이터 정제 : 데이터에서 누락된 결측값을 보완하고 튀는 값을 이상값으로 제거 해 데이터를 깨끗하게 만드는 기술
    - 데이터 변환 : 데이터 분석을 보다 쉽게 하기 위해 데이터를 변환해 일관성을 확보하고 데이터의 중복을 최소화해 데이터 분석 시간을 절약하는 기술
    - 데이터 필터링 : 데이터의 오류를 발견하고 삭제 및 보정을 통해 데이터의 품질을 향상시키는 기술
    - 데이터 통합 : 데이터 분석을 수월하게 하기 위해 유사한 성질의 데이터를 연계하는 등 데이터를 통합하는 기술

**결측값 처리**
결측값 : 데이터 수집 과정에서 측정되지 않거나 누락된 데이터
변수의 결측값이 있는 상태로 데이터 분석 모델을 만들면 변수 간의 관계가 왜곡될 수 있어 모델의 정확성이 떨어지게 되고, 데이터의 시각적 표현에도 문제를 일으킬 수 있음.
- **결측값 확인**
    - 데이터프레임 상 결측값 확인. 이때 0의 경우, 진짜 0의 값인지, 결측되어 0으로 표기된 것인지 확인 필요
    - 결측값 관련 pandas function
        - info() : 데이터 프레임 요약 정보 출력 -> 각 열에 속하는 유효한 값의 개수 보여줌
        - value_counts(dropna=false) : 각 열의 결측값을 포함한 전체 데이터 확인 가능
        - isnull() : 누락 데이터면 True, 유효한 데이터면 False 반환
            - df.isnull().sum() 으로 자주 사용
        - notnull() : 유효 데이터면 True, 누락 데이터면 False 반환
        - replace () : 결측값이 NaN이 아닌 '0'이나 '?' 등으로 입력되기도 하는데, 이때 replace를 활용해 NaN으로 변환 가능

**결측값 처리**
- 삭제 : 데이터가 있는 행 또는 열 삭제
    - dropna() : DataFrame내의 결측값 포함된 레이블을 제거하는 매서드
        - 삭제 기준 : axis = 0 / axis = 1
            - axis = 0 : 결측값이 있는 행 삭제. 분석 대상의 관측값(레코드) 제거 (디폴트)
            - axis = 1 : 결측값이 있는 열 삭제. 분석 대상이 갖는 특성(변수) 삭제
        - 삭제 조건 : how = 'any' / how = 'all'
            - how = 'any' : 하나라도 결측값 있으면 삭제 (디폴트)
            - how = 'all' : 모든 값이 결측값일때만 삭제
        - 특정 열(column) 가준 삭제
            - subset = ['column명'] (axis = 0 일때만 적용가능)
        - 원본 데이터 변경 여부 : inplace = False / inplace = True
            - inplace = False : 반영한 내용을 새로운 DataFrame으로 반환 (디폴트)            
            - inplace = True : 원본 DataFrame 자체를 변경하여 반영
    - **삭제 시 주의 할 점**
삭제는 결측값이 무작위로 발생한 경우에 사용. 결측값이 무작위로 발생한 것이 아닌데 관측치를 삭제한 데이터를 사용할 경우, 왜곡된 모델이 생성될 수 있음.


- 대체 : 결측값을 다른 값으로 대체 (평균값, 최빈값 등의 대표값 활용)
    - 일괄 대체 : 모든 변수들 일괄적으로 같은 값으로 대체
    - 유사 유형 대체 : 범주형 변수들을 활용해 유사한 범주에 따라 다른 값으로 대체
    - fillna() : 결측값 대체
        


**이상치 처리**
이상치 : 관측된 데이터의 범위에서 많이 벗어난 값

- 이상치 확인
    - 통계를 통해 확인 : describe ()
    - 시각화를 통해 확인 : BoxPlot
    - Z-score을 통해 확인
        - 데이터를 평균(0)과 표쥰편차(1)로 정규화하여, 평균으로부터 얼마나 떨어져있는지 나타냄
        - 평균에 가까울수록 0에 가깝고 멀어질수록 커짐
        - Z-score가 특정 기준값(일반적으로 2~3)을 넘어가는 데이터를 이상치로 간주
        - Tukey Fences을 통해 확인 : 사분위 범위를 기반으로, 두 가지 경우에 이상치라고 판단
            - Q1 - (1.5 * IQR) 미만
            - Q3 + (1.5 * IQR) 초과
            - Tukey Fences(튜키 펜스)는 이상치를 식별하는 통계적인 방법 중 하나로, 데이터의 사분위 범위(IQR, Interquartile Range)를 기반으로 함. IQR은 데이터의 중간 50% 범위를 측정하며, 이는 데이터의 상위 25%와 하위 25%를 제외한 중간 범위를 의미
            
- 이상치 제거 및 처리
    - 전체 삭제
        - 이상값이 Human error에 의해 발생한 경우 해당 관측치를 삭제
        - 단순 오타, 비현실적인 응답, 데이터 처리 과정에서의 오류 등의 경우에 사용
    - 다른 값으로 대체
        - 절대적 관측치의 숫자가 작은경우, 단순 삭제를 통해 이상치를 제거하면 관측치의 절대량이 작아져 신뢰성 문제 발생
        - 이런 경우 이상값이 Human error에 의해 발생하더라도 다른 값으로 대체, 결착값과 유사하게 다른 변수들을 사용해 예측 모델을 만들고, 이상값을 예측한 후 해당 값으로 대체하는 방법 사용
    - 변수화
        - 자연발생적인 이상값의 경우, 바로 삭제하는 것이 아닌 좀 더 찬찬히 이상값에 대해 파악하는 것이 중요
    - 리샘플링
        - 자연발생한 이상값을 처리하는 방식으로, 해당 이상값을 분리해서 모델을 만드는 방법            

**피처 엔지니어링(변수 가공)**
- 해결하고자 하는 문제를 컴퓨터가 잘 이해할 수 있도록 피쳐(변수)들의 형태를 변형하거나 적절하게 처리하는 과정
- 도메인 지식과 기존 변수를 사용해 기존의 데이터에 정보를 추가하는 일련의 과정
- 데이터 전처리의 마지막 단계. 새로운 데이터, 변수의 추가 없이 기존 데이터를 보다 유용하게 만드는 방법

**피처 엔지니어링 방식**
- 레이블인코딩(Label Encoding) vs 원핫인코딩(One-hot Encoding)
    - 텍스트로 주어지는 값들을 기계가 이해할 수 있도록 숫자로 바꾸는 작업
    - Label Encoding
        - 범주형 변수를 0부터 N-1까지의 숫자로 변환
    - One-Hot Encoding
        - 범주형 변수를 이진 벡터(0과 1)로 변환
        - 표현하고 싶은 단어 인덱스 = 1, 다른 인덱스 = 0

- 구간화(Binning)
    - 연속 데이터를 그대로 사용하기 보다 일정한 구간으로 나눠 분석하여 구간별 차이를 드러내는 것

- 파생변수 생성
    - 기존의 피처를 다른 피처로 변환(재정의)하여 변수를 추가. 기존 데이터를 다른 관점에서 분석해 새로운 특성, 정보를 얻기 위한 과정
    - 정해진 방법이 존재하지 않아 분석가의 도메인 지식과 데이터 특성에 대한 이해도에 따라 다양한 파생변수 생성 가능
    - 생성 방법으로는 단위 변화, 표현형식 변환, 요약 통계량 변환, 정보 추출, 변수 결합, 조건문 이용 등이 존재

- 스케일링
    - 서로 다른 변수의 값 범위를 일정한 수준으로 맞추는 작업. 숫자의 상대적 크기 차이로 인한 왜곡 방지를 위해 정규화 작업 수행
    - 모든 스케일러는 이상치에 민감하기 떄문에 이상치 제거가 선행되어야 함.
    - StandardScaler(표준화) : 각 feature 평균 = 0, 분산 = 1로 변경하여, 모든 피처들이 같은 스케일을 갖게 됨
    - MinMaxScaler : 모든 feature가 0과 1사이에 위치. 데이터가 서로 다른 비율의 속성으로 구성되어 있을 때, 같은 비율로 속성을 맞춤
        

**데이터 시각화**
- 데이터에 대한 이해를 돕기 위해 그림, 도형 등 그래픽 요소들을 이용해 데이터를 묘사하고 표현하는 과정
- Matplotlib / Seaborn
    - Matplotlib : Python 프로그래밍 언어 및 수학적 확장 Numpy 라이브러리를 활용한 플로팅 라이브러리
    - Seaborn : matplotlib을 기반으로 만들어져 통계 데이터 시각화에 최적화된 인기 라이브러리
- 시각화 하기 전 데이터가 범주형인지 수치형인지 파악, 결측값 및 이상치가 있는지 확인해야 함. 그 후, 어떤 그래프를 그려 데이터를 바라볼지 고민해야 함.

**파라미터**
프로그래밍된 함수의 입력값
- hue 파라미터
    - hue 파라미터에는 성별과 같은 범주형 변수를 넣어, 내가 원하는 변수를 기준으로 데이터를 구분하여 그래프에 표시할 수 있음.

**다양한 그래프**
- boxplot(상자 수염 그림) : 사분위수와 이상치를 보여주는 그래프
- countplot : 범주형 변수의 빈도수를 확인하는 그래프
- histplot : 도수분포표를 그래프로 나타낸 것. 수치형 변수의 구간별 빈도수를 보여줌
- displot/kdeplot : 히스토그램을 연속적으로 곡선으로 연결한 그래프
- barplot : 범주형 데이터 값 x에 따른 수치형 데이터 값 y의 평균 값 제공
- pointplot : 막대 그래프와 모양만 다르고 동일한 정보 제공
- scatterplot(산점도 그래프) : 두 연속형 변수 간의 관계를 시각화하기 위해 사용되는 그래프 유형. 각각의 데이터 포인트는 두 변수 값을 나타내며, x축과 y축 데이터 포인트를 분산하여 그림. **변수 간의 패턴, 상관관계, 분포 등을 쉽게 파악할 수 있다.**
- regplot(회귀선이 추가된 산점도 그래프) : 두 개의 연속 변수 사이의 산점도를 그리고 회귀선을 함께 나타내는 그래프
- catplot : categoryplot의 준말로, 수치형 데이터와 범주형 데이터의 관계를 볼 때 주로 사용
- pieplot : 데이터의 부분과 전체 간의 비율을 표현하는 그래프. 비율을 강조하기 위해 사용, 모든 데이터가 합쳐서 전체를 이루는 경우에 효과적으로 활용 가능
- heatmap : 변수간 상관계수를 직관적으로 볼 수 있는 그래프. corr() 매서드로 변수간의 상관계수를 구하고 이를 히트맵에 표현.
    - 음의 상관계수 : 서로 반비례 관계
    - 양의 상관계수 : 서로 비례 관계
- violinplot : 박스 플롯과 커널밀도추정 함수 그래프를 합쳐 놓은 그래프
- pairplot : 여러 그래프 한 번에 찍기
