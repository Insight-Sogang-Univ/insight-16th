# 1. EDA의 개념
- 0) EDA : 데이터 분석을 위해 여러 가지 방식으로 파악하는 모든 과정 / 탐색적 데이터 분석 과정
- 1) **EDA 대상** : 
      (1) 일변량 unvariate : 하나의 변수를 살펴봄 /데이터 설명하고 패턴 찾기
      (2) 다변량 multi-variate : 여러 개의 변수를 살펴봄 / 변수 간 관계 찾기
- 2) **EDA 종류** :
      (1) 시각화 : 차트 / 그림
      (2) 비시각화 : summary statistics
- 3) **EDA 유형** :

# 2. 데이터 확인
- 1) **데이터 읽기/쓰기** (데이터 불러오기/만들기) <br>
     (1) 절대경로와 상대경로 <br>
         절대경로: 처음부터 목적지까지 전체 URL <br>
         상대경로: 현재 디렉터리 (./) 기준으로 상(../)하위와 같이 상대적으로 표현되는 경로 (최상위: /)<br>

     (3) CSV와 EXCEL<br>
         CSV: 쉼표로 데이터를 구분하는 데이터 파일 / 가장 널리 사용됨/ pd.read_csv("파일경로/파일이름.csv") <br>

     (3) 데이터 분포 확인 <br>


# 3. 전처리
- 0) 데이터 전처리 : 데이터를 분석에 적합한 형태로 만드는 과정<br>
- 1) 결측값 처리 : 누락되거나 측정되지 않은 데이터 처리<br>
     (1) 결측값 확인 / NaN ? 0 ...<br>
     (2) 결측값 처리 /<br>
        - 삭제 :데이터가 있는 행 또는 열 삭제 (정보 손실 주의)
        - 대체 :결측값을 대체 (일괄 대체/유사 유형 대체 등)
- 2) 이상치 처리 : 관측된 데이터에서 많이 벗어난 값<br>
     (1) 이상치 확인 /통계, 시각화, Z-score,Tuckey Fences 통해 확인<br>
     (2) 이상치 제거 및 처리 / 전체 삭제, 대체, 변수화, 리샘플링<br>
- 3) 피처 엔지니어링 (변수 가공) : 해결하려는 문제를 컴퓨터가 잘 이해할 수 있도록 변수(피철)의 형태를 바꾸거나 처리하는 과정<br>
     (1) 레이블인코딩 / 원핫인코딩 : 기계가 텍스트를 이해하지 못하기에 기계가 이해하는 숫자로 바꾸는 작업 수행<br>
     (2) 구간화 : 일정한 구간으로 나눠서 분석<br>
     (3) 파생변수 생성 : 기존의 피처를 재정의해서 변수 추가<br>
     (4) 스케일링 : 변수의 값 범위를 일정한 수준으로 맞추는 작업. 정규화<br>


# 4. 데이터 시각화
: 데이터의 이해를 위해 그래픽 요소들을 이용하여 표현하는 과정
- 1) 파라미터 : 프로그래밍된 함수의 입력값
- 2) 다양한 그래프
    - boxplot (상자 수염 그림) : 사분위수와 이상치
    - countplot, histplot : 범주형 변수 빈도수 / 수치형 변수의 구간별 빈도수
    - displot, kdeplot : 히스토그램을 곡선으로 연결한 그래프
    - barplot, point plot : x에 따른 y의 평균값 제공 / 막대그래프와 모양만 다름
    - scatterplot, regplot : 두 연속형 변수 간의 관계 시각화 / 두 연속 변수 사이의 산점도 그리고 회귀선 나타냄
    - catplot: 수치형 데이터와 범주형 데이터의 관계
    - pieplot: 데이터의 부분과 전체 간의 비율 표현
    - heatmap : 변수의 상관관계를 볼 수 있음
    - violinplot : 박스플롯과 커널밀도추정 함수 그래프가 혼합된 그래프
    - pairplot : 여러 그래프 한번에 찍기
