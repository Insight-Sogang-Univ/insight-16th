# <span style="background-color: purple"> 1. 머신러닝과 모델링 </span>

### <span style="color: orange">1. **머신러닝이란?**</span>
- 인공지능의 한 분야로 데이터에서 패턴을 발견하고 이를 바탕으로 새로운 데이터에 대해 예측하는 것을 가능케한다.
1) 머신러닝의 종류 <br> <br> 1. 지도학습: 정답이 있는 데이터 사용 (회귀, 분류)  <br> 2. 비지도 학습 : 정답이 없는 데이터 사용 (군집화) <br> 3. 강화학습 : 보상 (알파고, 자율주행)

### <span style="color: orange"> 2. **모델링**</span>
1) 모델이란? <Br> 모델은 학습의 결과물이다. (학습을 마친 인공지능), 데이터를 학습하면 찾아낸 패턴이 저장된 상태이고, 새 데이터를 넣어도 예측하는 역할을 한다.
2) 모델링이란? <br> : 예측기(공식)를 만들어가는 전체적인 과정
3) 모델링의 과정 : <br> 1. 설계 : 데이터의 특성에 맞추어 어떤 알고리즘으로 학습할지 결정 <br> 2. 훈련 : 데이터를 기계에게 계속 보여주며 정답 맞추도록 계속 수정 <br> 3. 검증 : 훈련시킨 모델이 새로운 데이터에서도 정답을 잘 맞히는지 테스트 

### <span style="color: orange"> 3. **지도학습**</span>
1) 지도학습이란? <br> 정답 레이블을 함께 주고 학습을 시키는 방법 (예측값과 정답이 같아지도록)
2) 지도학습의 과정 <br> 0. 지도학습을 위한 데이터는 <span style="color: blue"> labeled data </span> 형태.(독립변수와 종속변수가 모두 존재하는 데이터) 이 중에서 train set으로 모델을 학습한다. <br> 1. 데이터 준비 > 2.예측 > 3. 오차계산 > 4. 가중치 수정 > 5.반복 > 6.학습 완료 = 함수 획득 > 7. 성능 평가
3) 회귀와 분류 <br> 회귀 : 연속적인 숫자값 예측, <br>분류 : 입력된 데이터를 주어진 항목들로 나눔 <br> 이 둘의 차이 : <span style="color: purple"> 연속형 - 회귀, 범주형 - 분류 </span>


# <span style="background-color: purple"> 2. 회귀 </span>
### <span style="color: orange"> 1. **회귀 개요** </span>
- 패턴을 찾아내어 미래 값을 예측
- 종속변수 연속적
### <span style="color: orange"> 2. **선형회귀** </span>
1) 선형회귀란? <br> : 데이터를 가장 잘 설명하는 회귀선을 찾는 과정 <br> 현재 데이터를 요약하는 동시에 아직 관측되지 않은 미래 값을 예측하는 도구가 된다.
2) 가중치와 편향의 이해 : y=wx+b <br> (1) 가중치 w :독립변수가 종속변수에 주는 영향력 크기, 기울기, 회귀계수 <br> (2) 편향 b : 독립변수가 0일 때 나오는 기본값, y절편
3) 선형회귀분석의 종류 : 단순선형회귀 분석, 다중선형회귀분석 (설명변수 개수 차이)
4) <span style="color: blue"> **단순선형회귀** </span> 
- 하나의 독립변수만 존재하는 선형회귀
- 데이터를 직선 형태로 표현
- 잔차를 줄여나가며 직선을 결정짓는 적절한 회귀계수와 편향 찾기
5) 잔차
- 실제값과 예측값 간의 차이
- 우리는 잔차를 줄여나가는 것을 목표로 한다.

### <span style="color: orange"> **3.최적의 w와 b를 찾는 두 가지 방법** </span>
1. 최소제곱법
- 잔차 = 회귀선-관측된 데이터 값
- 잔차의 제곱 합이 최소가 되는 지점으로 최적의 회귀선 구하는 방식

### <span style="color: orange"> **5. 다중공선성** </span>
0. 다중공선성
- 독립 변수 간 강한 상관관계로 회귀모델이 불안정해지는 문제
- 다중공선성이 높으면 : 어떤 독립 변수가 종속변수에 얼마나 영향을 미치는지 정확하게 구분 어려워짐  
1. 다중공선성 확인 방법
- 1. 상관계수 : 변수 간 통계적 관계인 상관관계의 정도를 수치화
- 2. VIF 지수 : 회귀 모델의 결정계수 R-square 를 사용하여 계산

2. 다중공선성 대처 방법
- 1. 변수 제거
- 2. 변수 변환
- 3. 규제 선형 모델 활용
- 4. PCA 주성분 분석 : 데이터의 차원을 축소하는 데에 사용되는 통계적 기법

### <span style="color: orange"> **6.규제선형모델** </span>
1. 규제선형모델이란?
- 모델이 학습 데이터에 과적합 되지 않도록 규제를 가하고자 등장 <br> : 기존 선형 모델에서는 잔차 제곱합 평균 최소화 함수를 사용했는데 이러면 오히려 회귀 계수가 쉽게 커져 과적합 발생 (특정 독립 변수에 과하게 의존)
- 최소제곱법과 회귀계수 값이 커지지 않도록 하는 방법의 균형이 중요

2. 규제선형모델의 종류
- 1. L2 규제 : 릿지 회귀 - W 제곱에 대해 페널티 부여
- 2. L1 규제 : 라쏘 회귀 - W의 절댓값에 페널티 부여
- 3. L1 + L2 규제 : 엘라스틱넷 회귀

3. **릿지 회귀** 
- 기둥에 따라 다른 힘으로 누르는 망치 : 전체적으로 작아진다
- 모든 변수를 유지하면서 크기를 적절히 줄여 과적합 방지 <br>
4. **라쏘 회귀** 
- 모든 기둥을 같은 힘으로 누르는 망치 : 완전히 사라질 수도 있다
- 비중이 낮은 변수는 사라지고 중요한 변수만 남는다.<br>
 5. **엘라스틱넷** 
- L1 + L2 : 불필요한 변수를 제거하고 남은 변수들 적절한 크기 유지
- 안정적 예측 + 희소성 달성
