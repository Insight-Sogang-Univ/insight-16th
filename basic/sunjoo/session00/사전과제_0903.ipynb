## **2. 데이터프레임의 구성**

### 1) 데이터프레임이란?

- **데이터프레임** : <br>컴퓨터로 정보를 정리하고 저장하는 도구 <br> 판다스에서 사용하는 주요 데이터 구조

### 2) 구성

1. **index & column**: 데이터의 위치 <br> - index : 행 / 가로 <br>  - column : 열 / 세로 

2. **Series** : 인덱스가 있는 1차원 배열 <br> - 만드는 법 : <br> 1) 리스트 생성 <br>   2) pd.Serise()로 시리즈 객체 생성 <br>   3) 필요하면 인덱스 추가

3. **DataFrame** : 행&열 / 여러 개의 시리즈로 구성<br> - 만드는 법 : <br> 1) 딕셔너리 생성 <br>   2) pd.DataFrame()로 데이터프레임 구성 <br>   3) 필요하면 인덱스 / 컬럼 지정

## **3. 데이터 확인**
  : 데이터에 대한 기본적인 이해

### 1) 방법
 1. **head(), tail()** : <br> 앞/ 뒤 데이터 파악해서 데이터의 구조와 값을 개괄적으로 파악
 
 2. **데이터에 대한 정보 확인** : <br> 행/ 열 개수, 데이터 타입, 결측치 개수 등 확인

 3. **고유값 및 개수 확인** : <br> 데이터 전체 개수와 고유값 파악

## **4. 데이터 추출**
  :필요한 부분만 골라내어 가공하는 능력

### 1) 방법
1. **indexing & slicing** : <br> 인덱스와 컬럼으로 데이터 추출 

2. **행단위 추출 (loc & iloc)** :<br> 행 데이터 선택을 위해 사용 <br>  - loc : 인덱스 이름 기준 <br> - iloc : 정수형 위치 기준

3. **열단위 추출 (컬럼명)** : <br> 열 데이터 선택을 위해 사용 <br> - df[열 이름] ,
df[[열 이름]] ,
df[[열 이름1, 열 이름2...]] ,
df[열 이름][행 이름] 

4. **불리언 인덱싱** : <br> 인덱스 / 컬럼명을 제외한 특정 조건에 만족하는 데이터 추출을 위해 사용

5. **인덱스 설정** : <br> 기본값은 0부터 시작하는 정수형 인덱스이지만 변경도 가능

## **5. 데이터 조작**

## **6. 집계함수**

### 1) 집계함수란?
- **집계함수**: 여러 행으로부터 하나의 결괏값을 반환하는 함수 
- 대표값 (평균, 중앙값, 표준편자 등..)을 구하기 위해 주로 사용
- 일부 행을 대상으로 집계할 때 더 유용한 경우가 있음

### 2) 그룹화 (grouping)
- **groupby** : <br> - 특정 기준으로 묶고 연산 처리 <br>-  집계함수와 함께 사용됨 : 그룹화를 하고 그룹별로 집계함수를 적용해 데이터프레임 형태로 반환하는 방식
