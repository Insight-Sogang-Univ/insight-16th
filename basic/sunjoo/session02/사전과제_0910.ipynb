# <span style="background-color:purple"> 1. 기술통계</span>

1. **기술통계란?**
기술통계 : 데이터를 있는 그대로 설명하고 이해하는 데 중점
2. **3가지 주요 도구**
- **중심화 경향 :**
1) 평균 = 데이터 총합 / 데이터 개수 , 이상치에 민감 (> 절사평균 등장)
2) 중앙값 = 한 가운데에 위치, 이상치에 덜 민감
3) 최빈값 = 가장 자주 나타나는 값, 이산형 데이터 분포 이해에 도움

- **자료의 분포** : 
1) 범위 = 최대값과 최소값의 차이, 이상치에 매우 민감
2) 분산 = 데이터가 평균으로부터 얼마나 멀리 퍼져있는지 나타냄
3) 표준편차 = 분산에 제곱근 취한 값 (단위 일치를 위해)

- **자료의 정리 및 시각화 :**
1)  사분위수 : 


    

      2) 사분위간 범위 (IQR)

      3) 변동계수 = 데이터의 상대적 변동성을 나타내는 통계적 지표
                           = (표준편차/평균)*100

* 첨도와 왜도

**첨도 : 확률분포의 꼬리가 두꺼운 정도를 나타내는 척도**

**왜도 : 비대칭도 실수 **값 확률 변수의 확률 분포 비대칭성을 나타내는 지표**

# 2. 확률과 분포

1. 도입
추론통계 : 표본에서 얻은 정보로 모수를 확률적으로 추정하거나 가설을 검정
2. **확률 분포**
1) 확률변수 : 어떤 특정 사건이 발생할 때 나타날 수 있는 모든 경우의 수를 특정 값으로 대응시키는 함수
2) 확률분포 : 확률변수가 특정한 값을 가질 확률을 나타내는 함수
- 이산확률분포 : 이산적인 값
- 연속확률분포 : 연속적인 값



1. 대표적인 분포 모형
1) 정규분포 : 종 모양으로 평균을 기준으로 대칭이다.



      2) Z 분포 (표준정규분포) :평균이 0이고 표준편차가 1이 되도록 맞춘 표준적인 정규분포



      3) T 분포 : 정규분포보다 꼬리가 더 두꺼움 분포. 표본으로 모평균을 추정할 때 사용된다.



      4) 카이제곱분포 : 독립적인 표준 정규 분포를 따르는 변수들의 제곱 합



      5)  F분포: 두 독립적인 카이 제곱 분포의 비율



# 4. 추론 통계 - 추정

1. 추론통계란?
수집한 데이터를 기반으로 미래의, 혹은 조사하지 못한 전체 정보를 추론하고 결론을 도출하는 방법을 다루는 분야로, 통계적 가설검정과 신뢰구간이 여기에 속한다.
- 추정 : 모집단을 대표할 수 있는 일부를 표본으로 추출하여 분석한 통계량을 이용해서 모집단의 **모수**를 예측하는 과정
- 검정: 표본을 추출하여 얻은 표본통계량으로 모집단 모수에 대한 새로운 예상·가설 등이 옳다고 판단할 수 있는지를 평가하는 과정
2. 점추정
: 어떤 모수를 표본에 기초하여 계산된 하나의 값으로 추정
3. 구간 추정
: 모수의 참 값이 포함되리라고 기대되는 추정치를 일정한 범위와 신뢰도를 가지고 나타내는 방법
