# <span style="color: purple"> 1. 군집화 <span>
### <span style="color: orange">1. 머신러닝 : 비지도 학습 <span>
- **머신러닝** : 컴퓨터가 스스로 학습할 수 있도록 도와주는 알고리즘이나 기술을 개발하는 분야<br>
      1) 지도학습 : 문제와 정답을 모두 알려주고 예측값이 정답과 같아지도록 학습시키는 것 (회귀, 분류)<br>
      2) 비지도학습: 정답을 알려주지 않고 스스로 데이터 속의 패턴이나 데이터 간 유사도를 학습시키는 것 (군집화, 연관분석)<br>

    <비지도학습의 특징> <br>
         - 데이터 자체에 내재된 구조를 학습해서 파악해서 학습<br>
         - 정답 레이블이 없기에 에이전트의 명확한 작업이 없음 > 모델의 성능을 명확히 측정 불가<br>
         - 표현학습으로 데이터셋의 고유 패턴 식별 > 목표 자체가 명확하지 않고 새로운 특성을 찾는 목적을 수행하기에 적합<br>

    <비지도학습에 적합한 케이스><BR>
         - 패턴이 알려지지 않은 경우 - 정답을 알 수 없음 <BR>
         - 패턴이 계속 변하는 경우 - 정답의 의미가 없음 <br>
         - 열린 문제를 해결하고 지식을 일반화해야 하는 경우<br>

    <비지도학습이 사용되는 예시><br>
         - 고객 세그먼트 분석, 이상 거래 탐지, 자연어 처리 <br>

### <span style="color: orange"> 2. 군집화란? <span>

- 군집화 : 데이터를 비슷한 특성을 가진 그룹으로 (군집으로) 나누는 비지도 학습 기법<br>
          * 군집 : 유사한 데이터들의 집합

    <지도학습과의 차이점><br>
         - 지도학습 : 종속변수를 예측하는 것 <-> 군집화 : 종속변수를 설정하지 않고 데이터 내부의 패턴을 인식해 유사성을 가진 그룹으로 나눔 (비지도학습)<br>
    <군집화의 목표><br>
         - **응집도**(같은 군집이라면 최대한 비슷하도록) 최대화
         - **분리도** (서로 다른 군집은 최대한 분리) 최대화
### <span style="color: orange"> 3. 군집화 과정 <span>
1. 피처 선택 / 추출 : 군집화에 사용할 데이터의 피처(특징)을 선택
2. 군집화 알고리즘 선택 : 데이터의 특성과 목표에 따라 적합한 알고리즘 선택
3. 군집 유효성 검증 : 군집화가 얼마나 잘 되었는지 평가 / 마음에 들지 않으면 다시 1
4. 결과 해석 : 각 군집이 지닌 특성 분석 및 해석

# <span style="color: purple"> 2. 군집화를 위한 데이터 준비 <span>
### <span style="color: orange"> 1. 군집화의 주요 고려사항

    - 군집화에 사용될 피처의 종류와 특성 명확히 이해
    - 변수 종류가 연속형, 명목형인지에 따라 적합한 군집화 알고리즘이 달라짐
2. 거리/ 유서도 정의와 측정
    - 데이터의 거리 / 유사도 를 기반으로 그룹 형성
    - 변수 유형에 따라 적합한 거리 측정 방식이 다르기에 정의가 중요
3. 차원 축소
    - 변수가 많아질수록 효율성과 성능이 떨어짐
    - 유사한 변수를 통합하거나 필요 없는 변수를 제거해 변수 개수를 줄일 수 있음

### <span style="color: orange"> 2. 변수 유형 이해 <span>
* 변수 유형에 따라 적합한 거리 측정 방법과 군집화 알고리즘이 달라진다
1. 연속형 변수 - 수치로 측정 가능한 변수 
    - 유클리디안 거리, 맨하탄 거리로 측정
    - k means, 계층적 군집화 등의 알고리즘
    - 스케일링 필수!!!
2. 명목형 변수 - 범주로 구분되는 변수
    - 해밍 거리, 자카드 거리 등
    - k modes, 계층적 군집화 등의 알고리즘
    - 원-핫 인코딩이나 더미 변수 변환으로 처리
3. 혼합형 변수
    - 연속형 + 명목형
    -  K-prototypes, Gower distance 기반 계층적 군집화 알고리즘 사용
    - 각 변수에 맞는 거리 함수를 조합하여 처리

    <왜 유형마다 다를까> <Br>
    1) 거리 : 두 값을 뺀 다음 차이를 확인. 
        * 명목형에서는 순서와 간격이 생기는 문제 발생 >> 해밍거리 (몇 개의 항목이 다른가), 자카드 거리 (얼마나 겹치는가)를 사용 (빼는 방식이 아닌 세는 방식)
        * 혼합형도 하나의 자로 잴 수 없는 문제 >> gower distance가 변수마다 다른 자를 써서 가중평균을 내줌
    2) 알고리즘 
        * 평균이 말이 안되는 경우 발생 ( 성별 0.6 )
        * 평균을 정의하는 k means는 평균이 정의되지 않는 변수에서는 사용 불가능
        * k modes는 최빈값을 사용하기에 말이 된다!
        * k prototypes는 연속엔 평균, 명목형엔 최빈값을 사용하여 합침
        * 계층적 군집화는 거리만 정의할 수 있다면 (누구와 누가 가장 가까운가) 어디서든 돈다.

### <span style="color: orange"> 4. 차원 축소 <span>


 <차원의 저주> <br>  
   * 피처 공간이 너무 커서 알고리즘이 데이터를 효과적으로 훈련할 수 없는 현상 : 고차원일수록 거리 기반 알고리즘의 분별력 감소
   * 왜 감소할까 ?
      1. 공간은 늘어나지만 데이터는 고정 - '이웃'이 사라지게 된다
      2. 그 결과 가까운 점과 먼 점과의 차이가 사라진다 : 묶는 기준이 사라진다
      3. 쓸데없는 피처가 노이즈를 만든다 - 유클리디안 거리는 모든 차원의 차이를 더하기에 무작위한 차이까지 더해진다
      : 따라서 줄일수록 군집이 선명해지기에 차원 축소를 진행한다 

<차원의 축소> <br>
   * 차원축소 : 차원= 변수의 개수를 줄이는 것 > 고차원 데이터를 저차원 공간에 투영(선형투영)하면서도 핵심 정보를 지켜낸다
   * 매니폴드 학습(비선형 차원 축소) : 유클리드 거리가 아닌 데이터 포인드들 사이의 곡선 거리를 고려하며 학습

<'PCA'> <br>
 
   * PCA의 핵심 목적 : 다차원 데이터에서 가장 정보량이 큰 (분산이 큰) 방향을 찾고 그 방향 기준으로 데이터를 투영해서 차원을 줄임 / 최대한 분산 보존!
   * 성관관계가 높은 피처들을 결합해서 상관관계가 없는 더 적은 수의 피처로 데이터 표현
   > 새로 파생된 성분 = 주성분
   * 주의사항 - 1) 표준화 필수 2) 결측값 처리 3) 해석의 어려움 4) 정보 손실

# <span style="color: purple"> 3. 군집화 알고리즘 <span>
### <span style="color: orange"> 1. 계층적 군집화 <span>
- 가장 가까운 것부터 둘씩 합쳐가면서 나무 모양으로 기록하는 (덴드로그램) 군집화 방법<br>
 <br>
<계층적 군집화의 특징> <br>
- 데이터셋의 관측치를 사용해서 덴드로그램을 만든다
- 군집의 개수는 클러스터링이 종료된 이후에 선택이 가능하다
- 상향식 : 응집형 계층적 군집화 <-> 하향식 : 분리형 계층적 군집화'<br>
<br>
- 응집형 : 각 샘플이 독립적인 클러스터로 시작해서 하나만 남을 대까지 가장 가까운 클러스터 병합
     * 과정
     1. 거리 형렬 계산
     2. 단일 클러스터 시작
     3. 가장 가까운 클러스터 병합
     4. 거리 행렬 업데이트
     5. 반복   
- 분할형 : 전체 포함하는 하나의 클러스터에서 유사성이 낮은 데이터를 더 작은 클러스터로 나눈다

<6가지 연결 방법><br>
1. 단일 연결법 : 두 군집에서 가장 가까운 두 점 간 거리로 군집 간 거리 정의
2. 완전 연결법 / 최장 연결법 : 두 군집에서 가장 먼 두 점 간 거리로 군집 간 거리 정의
3. 평균 연결법 : 모든 항목에 대한 거리 평균을 구함
4. 중심 연결법 : 두 군집의 중심 간 거리를 측정
5. 중앙 연결법 : 두 군집의 중앙값 사이의 거리 (중심 연결법 : 모든 개별 데이터 포인트의 산술 평균 <-> 중앙 연결법: 중심점 사이의 딱 중간 지점)
6. ward 연결법 : 군집 내 제곱합이 가장 적게 증가하도록 병합

<덴드로그램을 자를 때...><br>
* 가로선을 그어서 그 선이 끊는 세로선의 수가 곧 군집의 개수이다.

<계층적 군집화의 장단점> <br>
장점 : 
- k 를 미리 정하지 않아도 된다. 
- 데이터 구조 전체가 눈에 보인다
- 거리가 정의되면 데이터의 유형이 상관이 없다
- 결과가 항상 같다
단점 :
- 느리고 메모리를 많이 먹는다
- 한 번 합치면 되돌릴 수 없다
- 이상치에 민감하다
- 자르는 높이는 사람이 해야한다.

### <span style="color: orange"> 2. k-means <span>


- 데이터를 정해진 개수 k 의 그룹으로 나누되 각 그룹의 중심점과의 거리가 가장 가까운 데이터끼리 묶는 알고리즘이다.
- 군집화에서 가장 일반적 / 중심점으로 대표하기에 중심을 구할 수 없는 데이터에서는 사용이 어렵다.

< 주요 단계> <br>
1. 데이터 표본에서 래넘하게 k개의 중심점을 선택
2. 각 표본들을 가장 가까운 중심점에 할당
3. 각 클러스터에 할당된 표본들의 평균을 계산해서 중심점을 이동시킴
4. 클러스터 할당이 변하지 않거나 최대 반복 횟수에 도달할 때까지 2,3번 과정을 반복한다

<장단점> <br>
장점 : 
- 직관적이고 구현이 쉬움
- 대용량 데이터에서도 적용 가능
단점 :
- 초기 centroid 에 민감
- 군집 수 k 결정이 어려움
- 이상치에 민감
- 기하학적 모양 (길쭉, 초승달)의 군집은 파악하기 어려움

<엘보우 방법> <br>
 <br>
SSE를 바탕으로 그래프를 활용해서 최적 클러스특 개수를 추정
* 해당 그림에서는 3을 k로 지정 : 급격히 기울기가 변하는 지점

### <span style="color: orange"> 3. DBSCAN
- 밀도가 높은 지역의 데이터를 하나의 군집으로 묶고 밀도 기준을 만족시키지 못한다면 군집에 미포함 시키는 알고리즘
* kmeans의 단점을 방어 : k 안 정해도 되고, 이상치를 따로 제외하고, 모양을 가리지 않는다

<특징><br>
- 명시적으로 이상치를 지정해서 클러스터링에서 제외
- 다른 알고리즘 대비 이상치에서부터 자유로움
- 어떤 데이터가 여러 군집 특정 거리 내에 있으면 가장 밀도가 높은 군집과 그룹화

<단계><br>
1. 샘플 분류
<br>
2. 클러스터 생성
3. 경계 샘플 할당

<장단점> <br>
장점 :
- 임의의 기하학적 분포도 잘 처리
- 이상치에 민감하지 않음
- 모든 샘플을 클러스터에 할당하지 않음
단점:
- eps / min_samples 설정에 많은 영향을 받음
- 연산량이 많아서 속도가 느림
- 밀도가 서로 다르면 군집들을 잘 못 잡음

### <span style="color: orange"> 4. GMM <span>
- 데이터가 여러 다른 모양의 가우시안 분포로 구성되었다고 가정하고 각 분포를 클러스터로 인식
- 단일 분포로 표현하기 어려운 확률 분포를 여러 개의 가우시안 분포를 합쳐서 표현

<특징> <br>
- 모델 기반 군집화
- 각 군집을 확률 분포로 가정 /전체는 여러 확률 분포의 혼합으로 모델링
- 생성 메커니즘까지 모델링이 가능

<진행 과정><br>
1. 전체 데이터셋의 분포 확인
2. 전체 데이터셋은 서로 다른 정규 분포 형태의 확률 분포 곡선으로 구성되어 있다고 가정
3. 여러 개의 정규 분포 곡선 추출 > 개별 데이터의 분포를 결정하고 각 분포가 하나의 군집이 됨

      * 데이터가 너무 많을 때 (수만건이상)
2. k-means : 군집이 동글동글하고 데이터가 크다 -> 빠르고 직관적이며 대용량에 강함
      * 이상치가 많거나 군집 크기가 불균형하면 X
3. GMM : 군집 모양 길쭉하거나 서로 겹침/경계에 걸친 데이터의 애매함도 알고 싶다 -> 타원형 중첩 구조를 확률로 처리한다./ 소속확률을 반환한다
4. DBSCAN : 이상치 골라내기가 목적 / 군집 모양이 초승달 등 처럼 이상하다 -> 이상치에 -1 붙여 명시적으로 분리하고 모양을 가정하지 않으며 밀도로만 판단한다
     * 데이터가 정규분포와 거리가 멀 때
5. HDBSCAN : 군집마다 밀도가 다 다르다 -> DBSCAN의 단일 밀도 기준 한계를 보완

* 모두 스케일링 데이터는 필수!!!

# <span style="color: purple"> 4. 군집화 평가 방법 <span>

### <span style="color: orange"> 1. 외부 평가와 내부 평가 <span>
1. 외부 평가 - 정답 레이블이 있을 때 군집화 결과와 실제 레이블을 비교
2. 내부 평가 - 정답 레이블이 없을 때 군집 내부 응집도와 군집 가 분리도를 기반으로 품질 정량화
       * 실루엣 계수

### <span style="color: orange"> 2. 실루엣 계수 <span>
- 클러스터 내 샘플들이 얼마나 조밀하게 모여 있는지를 측정
<실루엣 계수 계산><br>
 <br>
ai= 응집도 (동일한 클러스터 내 다른 포인트들과 평균 거리) / <br> bi= 분리도 (가장 가까운 다른 클러스터의 샘플들과의 평균 거리)

<실루엣 계수 해석> <br>
* -1-1 사이의 값을 갖는다
- 분리도 > 응집도 : 계수는 1에 가까움 - 군집화가 잘 됨
- 분리도 = 응집도 : 0에 가까움 - 명확하지 않은 분리
- 분리도 < 응집도 : -1에 가까움 - 잘못된 군집화

주의 ! 평균만 보면 안된다! 평균 숫자와 그래프를 다 봐야한다


### <span style="color: orange"> 3. Dunn Index <span>
- 클러스터 간 최소 거리(분리도)와 클러스터 내 최대거리 (응집도)의 비율로 클러스터링 품질을 평가


<해석> <br>
- 값이 클수록 좋은 군집화이다
- 0 이상의 값! 상한은 없다
- 실루엣 계수와 함께 판단

<한계> <br>
- 군집 수가 많아질수록 계산 비용 증가
- 이상치에 큰 영향을 받는다
- 이상치에 민감하게 반응해서 수치가 급격히 떨어질 수 있음
