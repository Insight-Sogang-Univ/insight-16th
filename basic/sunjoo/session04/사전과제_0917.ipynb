# 사전과제 _0917_분류

# <span style="background-color: purple"> 1.분류란? </span>
### <span style="color: orange"> 1. 지도학습: 회귀 분류 </span>
- 회귀 : 연속적 숫자값을 예측 - 연속형 변수 예측
- 분류 : 입력된 데이터를 주어진 항목들로 나눔 - 범주형 변수 예측
### <span style="color: orange"> 2. 이진분류와 다중분류 </span>
- 이진분류 : 예측하고자 하는 변수가 어떤 기준에 대해 참/거짓 값 만을 가질 때
- 다중분류 : 예측하고자 하는 변수가 가질 수 있는 값이 3개 이상일 때


# <span style="background-color: purple"> 2. 분류모델 </span>
### <span style="color: orange"> 1. 로지스틱 회귀 </span>
- 이진 분류 문제 / 샘플이 특정 클래스에 속할 확률 추정 > 독립변수의 서녕조합에 로지스틱 함수를 적용해서 0-1 사이로 변환
- 필요성 : 이진 분류를 단순 선형 회귀로 풀면 문제 발생 (s자 형태로 표현 못함, 0-1 사이의 값을 넘는 값 가질 수 있음)
- 시그모이드 함수: 로지스틱함수 - 출력이 0-1 사이이면서 s자 형태로 그려지는 함수

ㄴ 입력값이 커지면 1, 작아지면 0 / 입력값이 무한대이면 출력값은 0-1 값
- 시그모이드 함수의 가중치 : 

1) w 에 따른 변화

2) b에 따른 변화


### <span style="color: orange"> 2. 의사결정나무 </span>
- 조건에 따라 데이터를 분류하고 최종적으로 데이터가 순수한 레이블의 집합으로 구성될 때까지 분류를 반복하는 분석 방법
- CART 알고리즘 : 데이터셋을 임계값 기준으로 두 child로 나누는 알고리즘 
                - 기준 : 불순도 (지니계수)가 낮아지는 방향으로
                - 불순도 : 데이터 집합에서 서로 다른 클래스가 섞여 있는 정도 의미
- CART의 주요 단계
                - 1) 임계값 설정
                - 2) 불순도 감소 알고리즘
- 실제 학습 시 고려해야할 것들 : 모수 설정, 차이점 시각화, pruning(가지치기=불필요한노드지우기)
 ### <span style="color: orange"> 3. SVM </span>
 - 클래스를 분류할 수 있는 다양한 경계선 중 최적의 라인을 찾아내는 알고리즘
 - 명확하게 분류 가능한 데이터 집단에서 뛰어난 성능을 보이며 고차원 공간에서 효과적으로 사용됨
 
 - 최적의 선 찾는 방법 : margin 즉, 거리가 가장 큰 경우를 선택함으로 최적의 선 찾음
 
### <span style="color: orange"> 4. KNN </span>

- 비슷한 특성을 가진 데이터끼리 서로 가까이 있다는 점을 이용한 분류 알고리즘: 데이터로부터 거리가 가까운 k개의 다른 데이터 레이블을 참조하여 분류하는 알고리즘
- 계산 순서
      - 1) 데이터 준비
      - 2) k 값 설정 : k는 가장 가까운 이웃의 개수 의미 / 보통 홀수
      - 3) 거리 계산: 새로운 데이터가 주어지면 기존 모든 데이터와의 거리 계산
      - 4) 가장 가까운 k개의 이웃 선택 
      - 5) 분류하기 : k개의 이웃 중 가장 많이 등장하는 클래스 예측 결과가 됨
- 최적의 k 선택 : 일반적으로 학습용 데이터 개수의 제곱근으로 설정
- 중요한 특징 : 어떠한 학습이 필요하지 않다.


# <span style="background-color: purple"> 3. 분류 평가 지표 </span>
### <span style="color: orange"> 1. 혼동 행렬 </span>
- 분류 모델의 예측 결과를 정확한 예측과 잘못된 예측으로 구분

- 정확도 : 모든 가능한 예측 중 참인 비율
- 정밀도 : 참이라고 예측했을 때 실제 참인 비율
- 재현도 : 실제로 참인 경우 중 참으로 예측하는 비율
### <span style="color: orange"> 2. F1-Score </span>
- 정밀도와 재현율의 조화 평균 (정밀도와 재현율 간의 균형을 효과적으로 평가하기 위함)

- 머신러닝 모델 성능을 평가하는 주요 지표
### <span style="color: orange"> 3. ROC / AUC </span>
- ROC 커브 : 얼마나 분류가 잘 되었는가를 보여주는 그래프

-AUC 커브 : ROC와 x축 사이의 면적 (적분값)

ㄴ 0.5-1 사이의 값을 가지면서 1에 가까울수록 분류 성능이 좋은 것에 해당

# <span style="background-color: purple"> 4. 하이퍼파라미터 최적화 </span>
### <span style="color: orange"> 1. 하이퍼파라미터 최적화 </span>
- 하이퍼파라미터 : 학습 시작 전, 사용자가 직접 설정하는 변수
- 하이퍼파라미터 최적화 : 적절한 하이퍼파라미터를 찾아 모델 성능 향상시킴
- 과정 :
     - 하이퍼파라미터 탐색 범위 설정
     - 평가 지표 계산 함수 정의
     - 설정한 범위에서 값을 샘플링한 하이퍼파라미터 값으로 검증 데이터로 정확도 평가
     - 위 단계를 반복하여, 정확도 결과를 보고 하이퍼파라미터 범위를 좁힘
### <span style="color: orange"> 2. 하이퍼파라미터 최적화 방법 </span>
1) grid search : 정해진 범위에서 하이퍼파라미터를 모두 순회
2) random search : 정해진 범위에서 하이퍼파라미터를 무작위로 탐색
3) Bayesian Optimization: 사전 정보를 바탕으로 최적 하이퍼파라미터 값을 확률적으로 추정하며 탐색
