## **1. 인공신경망의 시작: 퍼셉트론**

### **1-1. 인공신경망이란? (Artificial Neural Network)**
#### **1. 인공신경망**
**인공신경망**은 우리뇌의 **신경망(뇌를 구성하는 신경 세포끼리 서로 연결되어 정보를 처리하는 구조)** 을 모방한 모델

생물학적 뉴런이 정보를 처리하는 구조
1. 가지돌기를 통해 외부 전기신호를 받아들임
2. 모인 신호가 일정 이상의 크기를 가지면
3. 축삭돌기를 통해 신호를 전달


image.png

1. **입력(**$x_1, x_2, \dots, x_n$**)** 들을 받아 각각 **가중치**($w_1, w_2, \dots, w_n$)와 **편향**($b$)을 적용한다.
2. 종합한 값을 활성화 함수에 넘깁니다. 이때 활성화 함수는 임계값을 기준으로 삼는다.
    - 기준 미달이면? → **0**에 가깝게 죽여서 내보냄
    - 기준 통과면? → **유의미한 값**으로 살려서 내보냄
3. output이 출력

#### **2. 퍼셉트론**
**퍼셉트론**은 초기 형태의 인공신경망으로, **신경망(딥러닝)의 기원**이 되는 선형 이진 분류기(Linear Binary Classifier)

image.png

<퍼셉트론 출력값 구하는 절차>
= **입력 -> 가중합 -> 활성화 함수 -> 출력** 흐름을 따라감

1. 입력값과 가중치를 곱함 → $x_1 w_1$
2. 곱한 값들의 총합을 구함 → $x_1 w_1 + x_2 w_2 + \dots + x_n w_n$
(편향을 더해줄지는 옵션)
3. 총합이 임계값($\theta$)을 넘으면 $1$, 넘지 않으면 $0$을 출력
(이 경우의 활성화 함수는 Step 함수)

※ 활성화 함수: 임계값 초과 여부에 따라 출력값을 결정함. 사용하는 활성화 함수에 따라 출력값이 달라질 수 있음.

→ **Training(학습)**: 원하는 출력값을 내보내도록 **가중치를 조정**해가는 작업

#### **3. 용어 정리**
**1. 가중치(weight)** $w_1, w_2, \dots, w_n$

- 입력값이 출력값에 주는 영향력(중요도)을 결정
- 각 입력값에는 각각의 가중치가 존재
- **가중치의 값이 클수록 해당 입력 값이 중요함**

**2. 활성화 함수(Activation function)**

image.png

- 신호가 일정 이상의 값(임계값 $\theta$)을 넘으면 1, 넘지 못하면 0을 출력하는 퍼셉트론의 성질은 활성화 함수로 **계단 함수**를 사용
- 이후 등장한 발전된 신경망들은 계단 함수 외에도 다양한 활성화 함수를 사용함.
- Step 외의 다양한 활성화 함수
    - Sigmoid → 미분 불가 문제를 해결함(계단식에서는 미분 불가능)
    - Tanh (-1 ~ 1의 출력을 가짐) → 미분 불가 문제를 해결 함
    - **ReLU** (음수 부분을 0으로 만들어, 양수값을 주로 다룰 때 계산 속도상의 이점을 가짐) → 기울기 소실문제를 해결 함
    - leaky ReLU (ReLU에서 음수 부분에 미세한 기울기를 준 함수) → x<0일때의 Dying ReLU를 해결 - but 실무에서는 ReLU를 많이 씀
    
    image.png

**3. 임계값(Threshold,** $\theta$**)**

- 활성화 함수에 사용되는 값으로, 보통 세타($\theta$)로 표현.

$$
\begin{aligned}&\text{if } \sum_{i}^{n} w_i x_i \ge \theta \rightarrow y=1 \\&\text{if } \sum_{i}^{n} w_i x_i < \theta \rightarrow y=0\end{aligned}
$$

- 계단 함수에서는 가중합이 임계값을 넘으면 1, 넘지 못하면 0을 출력

**4. 편향(bias)**

- 위의 계단 함수를 구현한 식에서, 임계값을 좌변으로 넘긴것이 **편향($b$)**

$$
\begin{aligned}&\text{if } \sum_{i}^{n} w_i x_i + b \ge 0\rightarrow y=1 \\&\text{if } \sum_{i}^{n} w_i x_i + b < 0\rightarrow y=0\end{aligned}
$$

- 입력값 1에 편향 b가 곱해진 형태로 표현

    image.png

임계값과 편향은 **표현의 차이**이다.
오늘날 딥러닝에서는 임계값 대신 **편향(bias)** 개념을 주로 사용한다.    

### **1-2. 단층 퍼셉트론과 그 한계 (Single-Layer Perceptron)**
#### **1. 단층 퍼셉트론**
- 단순히 입력층과 출력층으로만 이루어진 퍼셉트론
- 값을 보내는 단계와 값을 받아서 출력하는 단계로만 이루어짐

#### **2. 논리 게이트 (Logic Gate) 구현**
단층 퍼셉트론을 이용하면 **AND, NAND, OR 논리 게이트**를 쉽게 구현할 수 있음

**논리 게이트란?**
- 디지털 회로에서 기본적인 논리 연산을 수행하는 데 사용되는 구성 요소
- 게이트 연산에 쓰이는 것은 **n개의 입력값**과 **하나의 출력값**
- AND, OR, NOT, NAND, NOR, XOR 등 다양한 종류가 존재, 이러한 논리 게이트를 조합하여 복잡한 논리 회로 구성 가능

#### **3. 대표적인 논리 게이트**
- **AND 게이트**
    - **입력값이 모두 1**인 경우에만 **출력값이 1**이 나오는 구조
    
        image.png
        
    - 가중치와 편향 값을 **w1, w2, b**라고 하였을때, AND 게이트를 만족하는 조합은?

        ⇒ `[0.5, 0.5, -0.7]`, `[0.5, 0.5, -0.8]`, `[1.0, 1.0, -1.0]` 등 다양한 조합이 존재.

$$
y = 0.5x_1 + 0.5x_2 + (-0.7)
$$    

나머지는 **원리가 똑같고 숫자만 바뀜**

| 게이트 | 조건 | 가중치·편향 예시 [w1, w2, b] |
| --- | --- | --- |
| **AND** | 둘 다 1일 때만 1 | `[0.5, 0.5, -0.7]` |
| **OR** | 둘 다 0일 때만 0 | `[0.6, 0.6, -0.5]` |
| **NAND** | 둘 다 1일 때만 0 | `[-0.5, -0.5, 0.7]` |
| **XOR** | 서로 다를 때만 1 | ❌ **가능한 조합이 없음** |
- **NAND** 는 **AND** 부호를 뒤집은 것!
- **XOR**은 조합 없음!



#### **4. 단층 퍼셉트론의 한계** -> XOR 게이트를 구현할 수 없다 !!
퍼셉트론은 **입력값과 가중치의 선형 결합**으로 이루어진 구조이기 떄문에, 직선 하나로 두 영역을 나눌 수 있는 선형 분류 문제에 대해서만 구현할 수 있음.

 **AND 게이트 → 하나의 직선으로 분리 가능**

 **XOR 게이트 → 하나의 직선으로 분리 불가능**

**XOR 게이트는 직선이 아닌 곡선/비선형 영역으로 분리해야만 구현할 수 있음!**

image.png

따라서 퍼셉트론 하나로는 XOR을 구현할 수 없고, 퍼셉트론을 여러 계층으로 사용할 경우에는 가능함

### **1-3. 다층 퍼셉트론 (Milti-Layer Perceptron, MLP)**
XOR 게이트는 **기존의 AND, NAND, OR 게이트를 조합하면(퍼셉트론의 층을 더 쌓으면)** 만들 수 있다!

하지만, 단순히 **층을 많이 쌓는 것만으로 비선형성이 생기지는 않음.** 활성화 함수 없이 가중합과 편향마 계산할 경우, 여러 층을 연결해도 결국 하나의 연산으로 합쳐지기 때문
<aside>

$y = W_2(W_1x+b_1)+b_2 = (W_2W_1)x+(W_2b_1+b_2) = Wx+b$

</aside>

따라서 마지막에 임계값으로 분류해도 결정 경계는 여전히 직선이므로, XOR을 구현할 수 없음.

다층 퍼셉트론이 XOR을 구현할 수 있는 이유는 **은닉층에서 비선형 활성화 함수를 적용해 새로운 특징을 만들고, 다음 층에서 그 특징들을 조합하기 때문**임. 예를 들어 은닉층의 두 뉴런이 각각 NAND와 OR를 계산하고, 출력층이 두 결과에 AND를 적용하면 XOR이 됨.

여기서 각 게이트는 **비선형인 계단 함수** 사용. 이 구성은 **뉴런 두 개의 은닉층 하나**로 충분

<aside>

$h = \sigma(W_1x+b_1), \qquad y = W_2h+b_2$

여기서 $\sigma$는 비선형 활성화 함수

</aside>

참고로 단층 퍼셉트론도 계단 함수라는 비선형 활성화 함수를 사용하지만, 가중합 한 번에 임계값을 적용하므로 결정 경계는 직선임. 다층 구조에서는 이런 중간 결과들을 다시 조합할 수 있다는 차이가 있음.

- 중간에 층을 추가하여 (**입력층** + 하나 이상의 **은닉층** + **출력층**)으로 구성
- XOR보다 더 복잡한 문제를 해결할 땐 **중간에 수많은 은닉층을 추가할 수 있음**

## **2. 딥러닝 모델은 어떻게 학습할까?**
### **2-1. 순전파와 역전파 (Forward & Backward Propagation)**
#### **1. 학습 알고리즘의 기본 구조**
1. **순전파**를 통해 **입력값으로부터 예측** 수행
2. **오차 계산**을 통해 **예측 정확도** 측정
3. **역전파**를 통해 출력층에서 다시 입력층 방향으로 **오차 전달**
4. **가중치 업데이트**
5. 1~4를 반복
-> 이 과정을 반복하다보면 오차값이 0에 수렴

- 순전파와 역전파
    1. **예측을 수행**: 문제를 푼다 → ***순전파***
    2. **오차 계산**: 내가 쓴 답과 실제 정답을 비교 → ***오차 계산***
    3. 오답 노트를 통해 **내가 틀린 문제를 다시 학습** → ***역전파***
    4. 나의 향상된 머리! → ***가중치 업데이트***
    5. 다시 시험 문제를 열심히 푼다


이러한 순전파/역전파의 과정은 **연쇄법칙**을 통해 일어남!

#### **2. 연쇄 법칙(Chain Rule)**
**합성함수의 미분** = **각 함수의 미분의 곱**으로 나타낼 수 있다

딥러닝에서는 입력값이 여러 은닉층을 거친 후 최종 출력값이 나옴. 이때, 입력값이 최종 출력에 얼마나 영향을 미쳤는지 계산할 때 **각 층이 다음 층에 미친 영향을 순서대로 곱함**.

딥러닝 모델을 학습시키기 위해서는 **오차를 줄여야 함.** 이를 위해 역전파 과정에서 **각 가중치가 최종 손실에 얼마나 기여했는지를 계산**해야 하는데, 이때 **연쇄법칙** 활용.

$$
\frac{\partial L}{\partial w_1}=\frac{\partial L}{\partial a_n}\times\cdots\times\frac{\partial a_2}{\partial a_1}\times\frac{\partial a_1}{\partial w_1}
$$

- 손실함수 = 예측값과 실제값의 차이를 수치로 표현
- 연쇄법칙을 통해 구하는 기울기 = '**손실 함수를 줄이기 위해 가중치를 어떻게 바꿀까?**'를 알려주는 신호

#### **3. 기울기 소실 문제 (Vanishing Gradient Problem)**
if 은닉층도 엄청 많고, 각 단계마다 미분값이 1보다 작은 값이라면?? -> 0에 수렴

$$
0.25\times0.25\times0.25\times\cdots\approx0
$$
(여기서 0.25 = 시그모이드 미분의 최댓값)

기울기 소실 문제가 있으면 앞쪽 층의 가중치가 거의 업데이트 되지 않아 학습이 제데로 이뤄지지 않는 문제 발생!

#### **4. 전체적인 흐름 정리**
**1. 순전파 = 예측하기**

입력값(x)이 **각 층(w)을 통과**하면 예측값(ŷ)이 나옴!

<aside>

입력 x → [가중치 w1] → [가중치 w2] → 예측값 ŷ

</aside>

**2. 손실 계산 = 얼마나 틀렸는지 측정하기**

예측값(ŷ)과 실제 정답(y)을 **비교해서 오차(손실)를 계산**!

<aside>

손실 L = (예측값 ŷ − 실제값 y)²

지금은 손실함수로 MSE를 사용

</aside>

**3. 역전파 = 책임 추적하기** (여기서 연쇄법칙이 사용)

손실이 계산되면, 출력층부터 거꾸로 거슬러 올라가며 **각 가중치가 손실에 얼마나 기여했는지 계산!**

<aside>

손실 L ← [가중치 w2의 책임은?] ← [가중치 w1의 책임은?]

</aside>

w1은 손실 L에 직접 연결되어 있지 않고 w1 → w2 → L 순서로 간접적으로 연결

따라서 w1의 책임은 **"w1이 w2에 준 영향 × w2가 L에 준 영향"**으로 곱해서 계산할 수밖에 없고, 이것이 **연쇄 법칙**!

**4. 가중치 업데이트 = 수정하기**

계산값을 바탕으로 **가중치를 조금씩 수정**. 이 과정을 계속 반복하면 모델의 정확도가 올라감!

<aside>

**이 "조금씩"이 얼마나 조금씩인지는 `2-4 경사하강법`에서 정함.**

순서 — 무엇으로 오차를 재나(`2-2`) → 층을 어떤 함수로 통과시키나(`2-3`) → 그래서 얼마씩 고치나(`2-4`)

</aside>

### **2-2. 손실 함수 (Loss Function)**
#### **0. 들어가기 전**
1. **목적함수 (Objective Function)**
    
    **우리가 달성하고 싶은 목표**. 최적화하려는 대상을 수학적으로 표현한 넓은 개념의 함수
    
    - 문제의 성격에 따라 최대화 or 최소화 가능.
        - 시험점수는 최대화, 실수는 최소화

2. **손실함수 (Loss Function)**
    
    **개별 데이터에 대한 오차를 측정하는 함수**.
    
    - 예측값 $\hat{y}$와 실제값 $y$의 차이를 계산해서 한 개의 데이터 포인트에 대해 모델이 얼마나 틀렸는지 평가.
        - 전체 시험 문제 중 1번 문제를 틀려서 깎인 점수

3. **비용함수 (Cost Function)**
    
    전체 데이터셋에 대한 손실을 합산하거나 평균 내어 **모델의 전체적인 성적을 평가하는 함수**.
    
    - 머신러닝의 목적함수는 보통 비용을 최소화하는 것이라, 학습할 때 **대부분은 비용함수를 최소화하는 것이 목표**
    - 시험 결과의 총점 혹은 감점 합
    
    $$
    \text{MSE} = \frac{1}{N} \sum_{i=1}^{N} (y_i - \hat{y}_i)^2
    $$
    

| **구분** | **정의** | **방향** |
| --- | --- | --- |
| 목적함수 | 최적화하려는 전체 대상 | 최대화/최소화 |
| 손실함수 | **데이터 한 개**에 대한 오차 | 최소화 |
| 비용함수 | **전체 데이터**에 대한 평균 오차 | 최소화 |

- **손실함수와 비용함수는 혼용**하는 경우가 많음! 엄밀히 말하면 다르지만, <u>최소화</u>를 목적으로 하는 함수들이라는 점!)

#### **1. 손실 함수의 종류**
**1. 평균 제곱 오차 (Mean Squared Error, MSE)**

$$
\frac{1}{N}\sum(y-\hat{y})^2
$$

예측 값이 **연속적**이고, 예측 값과 실제 값의 차이를 **직접 측정하는 `회귀 문제`**에 적합!

ex) 모델이 예측한 집 가격과 실제 집 가격의 차이를 최소화하는 집 가격 예측

**2. 이진 크로스 엔트로피 (Binary Cross-Entropy, BCE)**

image.png

$$
L= -(y \log \hat{y} + (1-y) \log(1-\hat{y}))
$$

**'이진 분류 문제(로지스틱 회귀)'** 에서 사용하는 손실 함수!

**이 식은 왜 이렇게 생겼을까?**

모델이 예측한 ‘정답이 1일 확률’을 p라고 하자

- 정답이 1이면 → p가 높아야 잘 예측한 것.
- 정답이 0이면 → 1 − p가 높아야 잘 예측한 것.

BCE는 **실제 정답에 부여한 확률에 −log를 적용한 손실**. 정답의 확률이 높으면 손실이 작고, 낮으면 손실이 커짐.

로그는 여러 데이터의 확률을 곱하는 계산을 합으로 바꿔줌. 앞의 마이너스는 ‘확률을 높이는 목표’를 ‘손실을 줄이는 목표’로 바꿔줌.

$$
P(y\mid x)=p^y(1-p)^{1-y}
$$

**여기에 −log를 적용하면 BCE(Binary Cross Entropy)**

$$
L=-\log\left[p^y(1-p)^{1-y}\right]=-\left[y\log p+(1-y)\log(1-p)\right]
$$

(1) 실제 정답이 1일 때 → $L=-\log{\hat{y}}$만 남음.

- 예측 확률이 1에 가까울수록 → loss가 0에 가까워짐 = 잘 맞힘
- 예측 확률이 0에 가까울수록 → loss가 급격히 증가 (확신했지만 틀림)
- ex) 실제 레이블이 고양이(1)인데 고양이일 예측 확률이 0.1이라면, 모델이 고양이가 아니라고 확신하며 틀린 것이므로 −log(0.1)로 계산되어 손실 값이 크다!

(2) 실제 정답이 0일 때는? 반대로 동작!

MSE는 단순히 loss의 크기만 보는 반면, BCE는 **얼마나 확신을 가지고 예측했는지**도 loss에 반영함.

- (참고) 왜 손실 함수 이름에 "엔트로피"가 들어갈까?
    
    엔트로피란 "정보의 불확실성"을 정량적으로 표현하는 것. 크로스 엔트로피 함수는 단순히 정답을 맞췄는지만 평가하는 게 아니라, 얼마나 확실성을 갖고 맞췄는지도 보기 때문에 이름에 엔트로피가 들어간다!

**3. 크로스 엔트로피 오차 (Cross Entropy Error, CEE)**

$$
-\sum y \log \hat{y}
$$

$$
y : 원-핫 인코딩\ 값 (1,0),\ \ \hat{y} : 예측값(확률)
$$

- **`다중 클래스 분류 문제`**에서 예측 확률이 실제 클래스와 얼마나 일치하는지 평가하는 데 적합.
- 2번과 거의 유사한 형태이지만, **다중 분류**에서 사용!
- BCE와 CEE는 왜 MSE와 달리 `(-)`가 붙을까?
    
    log 함수의 특성 때문! 확률값은 항상 0-1 사이인데, 0-1 사이의 수를 log에 넣으면 **결과가 항상 음수**가 나옴. 손실값은 양수여야 "오차가 크다/작다"를 표현할 수 있으니 앞에 마이너스를 붙여 양수로 바꿔주는 것!

**4. 손실 함수 정리**

| **손실 함수** | **사용 상황** | **예시** |
| --- | --- | --- |
| **MSE** | 회귀 문제 | 주택 가격 예측, 온도 예측, 주식 가격 예측 |
| **BCE** | 이진 분류 문제 | 이메일 스팸 필터링(스팸/정상), 질병 진단(양성/음성) |
| **CEE** | 다중 클래스 분류 문제 | 손글씨 숫자 인식(MNIST), 이미지 분류(CIFAR-10) |


### **2-3. 활성화 함수 (Activation Function)**
#### **1. 활성화 함수란?**

입력값 × 가중치 + 편향 → **활성화 함수** → 출력값

(wx+b를 생각!)

**왜 굳이 활성화 함수를 거칠까?**

**(1) 비선형성이 필요하다**

활성화 함수가 없으면 층을 아무리 쌓아도 결국 **선형 계산의 반복**. 1장에서 본 XOR 같은 문제를 못 푼다.

**(2) 미분이 되어야 한다** (**2-1과 이어지는 부분**)

2-1에서 역전파는 **미분값을 연쇄적으로 곱해 나간다**고 함. 그렇다면 활성화 함수는 **미분이 가능해야** 한다!

그런데 **계단 함수는 미분이 안 된다.** 0에서 끊기고, 나머지 구간은 기울기가 0. ⇒ **역전파를 할 수가 없음!**

그래서 활성화 함수는 이렇게 발전해 왔다

```
계단 함수      미분 불가              → 역전파 자체가 안 됨
   ↓
시그모이드     미분 가능! 하지만        → 미분값이 최대 0.25
   ↓                                   0.25를 5번 곱하면 0.00098… ⇒ 기울기 소실
ReLU          양수 구간 미분값 = 1     → 곱해도 기울기 소실 발생 X
```

**ReLU가 "진짜 중요"한 이유가 바로 이것!**

**가중치와 편향**

- **가중치(Weight):** 입력값이 결과에 얼마나 영향을 미치는 지를 결정하는 값
- **편향(Bias):** 활성화 함수의 출력값을 조정하는 값으로, 기준점 역할을 수행

활성화 함수는 **단층/다층 퍼셉트론**에 따라 다르게 사용된다!

|  | **단층 퍼셉트론** | **다층 퍼셉트론** |
| --- | --- | --- |
| **활성화 함수** | 계단 함수 | Sigmoid, ReLU 등 |
| **출력값** | 0 또는 1 (이진 출력) | 연속적인 값 (확률 또는 실수) |
| **특징** | 선형 문제만 풀 수 있음 | 복잡한 비선형 문제도 학습 가능 |


#### **2. 계단함수 (Heaviside Step Function)**

image.png

$$
h(x) = \begin{cases} 1 & (x \geq 0) \\ 0 & (x < 0) \end{cases}
$$

입력값과 가중치의 곱을 모두 더한 뒤, 그 합이 임계값을 넘으면 1, 아니면 0을 출력하는 함수! (주로 단층 퍼셉트론에서 사용)

- 입력 신호: 입력값과 가중치의 곱
- 출력 신호: 0 또는 1

**임계값**은 뉴런이 얼마나 쉽게 활성화되는지를 제어하고, **가중치**는 각 입력 신호가 결과에 미치는 영향력을 제어함.

즉, **가중치는 각 입력 신호가 결과에 미치는 영향력을 제어!**

계단 함수는 **미분값이 0**이라 역전파를 할 수 없기 때문에 다층 퍼셉트론에서는 사용하지 않음!


#### **3. 시그모이드 함수**

image.png

$$
f(x)=\frac{1}{1+e^{-x}}
$$

입력 신호를 **0과 1 사이의 연속적인 값**으로 변환하는 비선형 함수!

**계단 함수와의 차이점?** 출력값의 형태!!

계단 함수는 0 또는 1만 출력하는 반면, 시그모이드는 x값에 따라 출력값이 연속적으로 변화함. 덕분에 "확실히 맞다/틀리다"가 아니라 **얼마나 확실한지**를 표현할 수 있고, **미분도 가능.**

**하지만** 미분값이 **최대 0.25**. 층을 쌓아 곱해 나가면 → 2-1에서 본 **기울기 소실** 발생.


#### **4. ReLU 함수 (주로 사용)**

image.png

$$
f(x)=\max(0,x)
$$

입력이 0보다 크면 그대로, 0 이하면 0을 출력하는 비선형 함수! 단순하지만 딥러닝에서 가장 많이 사용!

**왜 ReLU를 가장 많이 쓸까?**

1. **계산이 간단하고 효율적** — 0보다 크냐 아니냐만 비교하므로 빠르다!
2. **비선형성 제공** — 음수는 0으로, 양수는 그대로 처리하면서 은닉층에 비선형성을 추가한다!
3. **기울기 소실 완화**  — **양수 구간에서 미분값이 항상 1**이므로, 아무리 곱해도 값이 죽지 않아서 역전파가 안정적으로 전달된다!

- 음수에서는 0이 돼서 다 죽는데요? (Dead Neuron)
    
    그래서 음수 구간에 미세한 기울기를 주는 **leaky ReLU** 같은 변형도 존재. 다만 실무에서는 **양수를 살리는 게 더 중요**해서 기본 ReLU를 많이 쓴다.
    

#### **5. Softmax 함수**

image.png

$$
f(x_i)=\frac{e^{x_i}}{\sum_j e^{x_j}}
$$

각 클래스에 대한 출력값을 **확률로 변환**하는 함수! **다중 클래스 분류 문제의 출력층**에서 사용!

출력값이 모두 0~1 사이이고 전체 합이 1이 되기 때문에, 각 클래스에 속할 확률을 직접 읽을 수 있다!

ex) 고양이/강아지/새 분류 → [고양이 0.7, 강아지 0.2, 새 0.1]

**Softmax + 크로스 엔트로피 조합**

Softmax는 **2-2에서 본 CEE**와 자연스럽게 연결됨. 예측 확률(softmax)과 정답 확률의 차이(크로스 엔트로피)를 바로 계산할 수 있어 학습 효율이 높음!

⇒ **가장 표준적인 다중 분류 조합!**

#### **6. Tanh(쌍곡탄젠트) 함수**
    
image.png
    
$$
f(x)=\tanh(x)=\frac{e^x-e^{-x}}{e^x+e^{-x}}
$$
    
**시그모이드의 변형**으로 출력 범위가 **-1 ~ 1**. 출력값의 중심이 0이라 가중치 업데이트가 더 안정적이라는 장점있음!
    
다만 시그모이드와 마찬가지로 입력값이 매우 크거나 작으면 미분값이 0에 가까워져 **기울기 소실**이 발생하여 실무에서는 잘 쓰지 않음.


**활성화 함수 정리**

| 활성화함수 | 출력 범위 | 주요 사용처 | 장점 | 단점 |
| --- | --- | --- | --- | --- |
| 계단 함수 | 0 또는 1 | 단층 퍼셉트론 | 구조가 단순 | **미분 불가** |
| Sigmoid | 0 ~ 1 | 이진 분류 출력층 | **확률**로 해석 가능 | **기울기 소실** |
| Tanh | -1 ~ 1 | 은닉층 (요즘은 ReLU에 밀림) | 출력 중심이 0으로 학습 안정적 | **기울기 소실** |
| ReLU | 0 ~ ∞ | **은닉층 전반** | **계산 빠름, 기울기 소실 완화** | 음수 입력 시 출력이 항상 0 |
| Softmax | 0 ~ 1 (합=1) | 다중 분류 출력층 | 클래스별 확률 직접 출력 | 출력층 전용, 은닉층엔 부적합 |

### **2-4. 경사하강법 (Gradient Descent)**
#### **1. 경사하강법의 원리**

**목표: 손실을 가장 작게 만드는 W를 찾는 것!**

손실함수를 W에 대한 함수로 보면 (편의상 b는 고정)

$$
E(W) = \frac{1}{n} \sum_{i=1}^{n} [t_i - (Wx_i + b)]^2
$$

x축을 **W**, y축을 **E**로 두면 아래로 볼록한 곡선. 우리가 찾는 건 이 곡선의 **바닥**이다!

image.png


**어떻게 바닥으로 가나? — 미분값의 부호 확인!**

| 지금 W의 위치 | 접선 기울기 | 현재값 − 기울기 | 결과 |
| --- | --- | --- | --- |
| 최솟값 **오른쪽** | **양수** | 현재값 − 양수 | W가 작아짐 → **왼쪽으로** |
| 최솟값 **왼쪽** | **음수** | 현재값 − 음수 | W가 커짐 → **오른쪽으로** |
| 최솟값 **도달** | **0** | 현재값 − 0 | **멈춤** |

**어느 쪽에서 출발하든 자동으로 바닥에 모임!**

$$
W \leftarrow W - \alpha \frac{\partial E}{\partial W}
$$

$\alpha$는 바로 다음에 볼 **학습률**!

image.png

b도 똑같음. x축을 b로 바꾸기만 하면 최적의 b도 같은 방법으로 찾을 수 있다!


**<상상해봅시다>**

짙은 **안개**(많은 데이터) 때문에 **산**(손실함수) 속에서 길을 잃었습니다. 발밑 지면의 기울기만 느낄 수 있어요. 골짜기로 빨리 내려가려면? **가장 가파른 쪽으로 한 걸음씩** 내려가면 됩니다. 이게 경사하강법의 원리!


#### **2. 학습률 (Learning Rate, α)**

방향은 알았을 때, **얼마나** 움직일까? 안개 낀 산 비유에서 **보폭**에 해당.

미분값을 그대로 빼면 문제가 생김. 한 번에 너무 많이 움직여서 최솟값을 지나치고, 반대편에서 또 지나치고 오히려 **멀어진다.**

image.png

그래서 미분값에 **α**를 곱해 이동량을 조절!

$$
W \leftarrow W - \boldsymbol{\alpha} \frac{\partial E}{\partial W}
$$

| 학습률 | 무슨 일이 벌어지나 |
| --- | --- |
| 너무 **작음** | 수렴은 하지만 **너무 오래 걸림** |
| 너무 **큼** | 수렴 못 하고 **발산 (손실이 계속 커짐)** |
| **적당함** | ✅ |

image.png

**얼마로 둘까?** → **`0.001`부터 시작!** (Adam의 기본값)

- (참고) 학습률은 누가 정해줄까?
    
    학습률은 개발자가 **직접 설정**하는 **하이퍼파라미터**
    
    **설정하는 방법**
    
    1. **초기에 작은 값부터 시도** — 보통 0.01, 0.001, 0.0001을 경험적으로 시도
    2. **실험을 통해 조정** — 학습이 너무 느리면 늘리고, 발산하면 줄이며 여러 번 실험
    
    **자동으로 찾는 방법도 존재!**
    
    - **Learning Rate Scheduler**: 학습 과정에서 학습률을 자동으로 줄이거나 늘려줌
    - **Grid Search, Random Search**: 여러 학습률을 자동으로 실험하며 최적값을 찾아줌


**이 α가 뒤에서 계속 나옵니다.** 2-5 옵티마이저는 결국 *이 α를 어떻게 더 똑똑하게 쓸까*에 대한 답이고, Adam 코드의 lr=0.001이 바로 이거예요.


#### **3. 배치·에포크·이터레이션**
위 식의 $\partial E/\partial W$를 구하려면 데이터를 어떻게 넣을까?

데이터가 100만 개라면 **한 번에 다 넣을까?** 메모리가 못 버틴다. 그래서 나눠서 넣는다. 이것이 **배치**

**1. 배치 (Batch)**

image.png

- 학습 과정에서 **매개변수를 조정**하기 위해 사용하는 **데이터의 묶음**. 사용자가 임의로 정할 수 있는 **하이퍼파라미터**.
- 배치 크기가 너무 크면 **학습이 느려**지고 **메모리가 부족**해질 수 있다.
- 배치 크기가 너무 작으면 가중치가 자주 업데이트되어 **학습이 불안정**해질 수 있다.


**2. 에포크 (Epoch)**

image.png

- 모든 데이터셋을 **학습하는 횟수**를 의미!
- 예시
    - 문제집이 **300페이지**라 가정
    - 배치를 나누는 것 = 문제집을 **30페이지씩 묶음**
    - 1 epoch의 학습 = 모델이 **전체 문제집을 한 번 풀어봄**


**3. 이터레이션(Iteration) or 스텝(Step)**

- **한 번의 에포크를 끝내기** 위해 필요한 배치의 수
- 1 Epoch에서 다루는 데이터의 수 = **배치 크기 × 이터레이션**
    
    → 전체 데이터가 2,000일 때 배치 크기를 200으로 하면 이터레이션은 총 10

image.png

- 한 번에 처리하므로 **메모리가 많이 필요**
- 항상 같은 데이터에 대해 경사를 구하므로 **수렴이 안정적**

매개변수를 조정할 때 전체가 아닌 **랜덤으로 선택된 하나의 데이터**에 대해서만 계산.

image.png

- 하나의 데이터만 메모리에 저장하면 되므로 **자원이 적은 컴퓨터에서도 사용 가능**
- 매개변수의 **변경폭이 불안정**하고 **정확도가 낮을** 수 있음

```python
model.fit(X_train, y_train, batch_size=1)
```


**3. 미니 배치 경사하강법 (Mini-Batch Gradient Descent)**

배치 크기가 1이면 정확도가 낮고, 전체 배치는 메모리가 부족함. ⇒ 그래서 나온 게 **미니 배치**

image.png

- **배치 크기를 적절히 지정**하여 해당 개수만큼 매개변수를 조정
- 전체 데이터보다 **빠르고**, 배치 크기 1보다 **안정적** ⇒ **가장 많이 사용**

### **2-5. 옵티마이저 (Optimizer)**
#### **0. 경사하강법만으로는 부족! — Local Minima**

앞에서 본 손실함수 그래프는 예쁘게 아래로 볼록한 모양 -> 경사하강법으로 **반드시** 바닥에 도달함!

실제로 MSE·RMSE는 볼록함수라는 게 증명되어 있어요.

- MSE, RMSE가 볼록함수인 이유
    
    MSE는 매개변수 w에 대해 이차 함수의 형태를 가지므로 볼록성을 유지. RMSE는 MSE의 제곱근이므로 루트가 적용되어도 전체 구조는 여전히 볼록성을 유지.
    

문제는 **딥러닝의 손실함수는 그렇게 안 생겼다!** 층이 여러 개라 울퉁불퉁하고, 어떤 모양인지 예측할 수도 없음.

진짜 바닥(**Global Minimum**)이 아닌 **작은 웅덩이(Local Minimum)에서도 기울기는 0**. (경사하강법의 규칙 : **기울기가 0이면 멈춘다!**)

image.png

⇒ **목적지가 아닌 곳에서도 갇힐 수 있음!**

기울기가 완만한 구간도 마찬가지. 조금씩만 움직이니 빠져나오는 데 한참 걸림.
(작은 기울기에서는 조금씩, 큰 기울기에서는 많이 움직임.)

**옵티마이저 : 경사하강법을 더 똑똑하게 굴릴 방법**

image.png


#### **1. 모멘텀(Momentum)**
"변수가 가던 방향으로 계속 가도록 속도를 추가!"

기존 경사하강법: **이전 기울기를 다 잊어버리는 기억상실증**. 매번 현재 기울기만 보고 움직임.

⇒ **이전에 가던 방향**을 기억하도록 **관성**을 부여!


**관성을 가진 공**처럼 이전 방향을 참고하면, 더 부드럽고 빠르게 **작은 언덕도 넘어갈 수 있음!**


image.png

- **빨간색 화살표** : 경사하강법이 이동하려는 방향 (**현재 기울기**)
- **초록색 화살표** : 이전 단계의 이동을 반영한 모멘텀 방향 (**관성**)
- **파란색 화살표** : 실제 이동 방향 (**둘의 합**)

**무엇이 좋아지나?**

- **방금 본 Local Minima를 관성의 힘으로 탈출**할 수 있음!
- 기울기가 매우 작은 구간도 빠르게 통과!
- (참고) 모멘텀을 적용한 경사하강법의 식
    
    이동량(속도) $v_t$를 업데이트한다 (이동량은 기울기 대신 사용).
    
    image.png
    
    - $v_t$ : 현재 단계의 이동량(속도)
    - $v_{t-1}$ : 이전 단계의 이동량
    - $\beta$ : 모멘텀 계수 (일반적으로 0.9)
    - $\nabla J(w)$ : 현재 단계에서의 비용 함수 J의 기울기
    
    image.png
    

```python
tf.keras.optimizers.SGD(learning_rate=0.01, momentum=0.9)
```

#### **2. 알엠에스프롭(RMSProp)**

**기존 경사하강법의 또 다른 문제?**

기울기가 **큰 방향에서는 너무 많이**, **작은 방향에서는 너무 조금** 이동.

⇒ 따라서 **각 방향마다 학습률을 따로 조정**해서 안정적이고 빠르게 학습!

**산을 내려가는 상황에서의 RMSProp**

기존 경사하강법은 무조건 현재 기울기만 보고 내려갑니다. 그러나 RMSProp은 **최근 기울기를 참고**해서

- **급한 경사에서는 속도를 줄이고**
- **완만한 경사에서는 속도를 유지하거나 증가**

안전하고 효율적으로 계곡으로 내려감!


#### **3. 아담 (Adam)**

**RMSProp**과 **Momentum**을 합친, **"방향과 스텝 사이즈를 적절하게 합친 방법!"**

⇒ **방향**과 **학습률** 두 가지를 모두 잡기 위한 방법!

각 파라미터의 **1차 모멘트(기울기의 평균)**와 **2차 모멘트(기울기의 제곱 평균)**를 사용하여 학습률을 조정.

- 이것들을 어떻게 활용할까?
    
    옵티마이저는 `compile()`의 `optimizer` 인자에서 지정. 단순히 문자열 `'adam'`으로 써도 작동.
    
    ```python
    adam = tf.keras.optimizers.Adam(learning_rate=0.001, beta_1=0.9, beta_2=0.999)
    model.compile(loss='categorical_crossentropy', optimizer=adam, metrics=['acc'])
    
    # 이렇게 짧게 써도 됩니다
    model.compile(loss='categorical_crossentropy', optimizer='adam', metrics=['acc'])
    ```
    
    `learning_rate=0.001` — **2-4 ②에서 본 그 α입니다.**

####  **4. 그래서 뭘 쓰나요?**

| 이럴 때 | 이걸 쓰세요 | 왜 |
| --- | --- | --- |
| **잘 모르겠다 / 대부분의 경우** | **Adam** | 방향과 보폭을 둘 다 잡음. 사실상 기본값 |
| 학습이 자꾸 한쪽으로 튄다 | 모멘텀 | 방향을 안정화 |
| 파라미터마다 기울기 크기 차이가 크다 | RMSProp | 방향별로 학습률을 따로 조정 |

**결론: 일단 Adam으로 시작!!** 잘 안 되면 그때 바꾸기.


**한 줄 정리**

- **모멘텀** = **방향**을 개선
- **RMSProp** = **보폭(학습률)** 을 개선
- **Adam** = **둘 다** = 모멘텀 + RMSProp


### **2-6. 정리 - 코드 한 세트로 다시 보기**

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

# ① 모델 구조
model = Sequential()
model.add(Dense(10, input_dim=5, activation='relu'))
model.add(Dense(3, activation='softmax'))

# ② 학습 방법 설정
model.compile(optimizer='adam',
              loss='categorical_crossentropy',
              metrics=['acc'])

# ③ 학습 실행
model.fit(X_train, y_train, batch_size=32, epochs=10)
```
